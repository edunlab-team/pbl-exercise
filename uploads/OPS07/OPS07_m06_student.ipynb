```python
ROLE = 'student'
PARTICIPANTS = ['P01','P02','P03','P04']
PROJECT = {'version': '1.0', 'exportedAt': '2026-11-08T11:53:56.027Z', 'id': 'OPS07', 'title': '민간·아군 위험을 포함한 임무 보류·재확인 기준서', 'category': '작전·전투 수행', 'institution': 'KYU-EDU', 'level': 2, 'durationDays': 5, 'requiredHours': 30, 'status': 'TODO', 'isRecommended': False, 'summary': '가상 보고를 읽고 관찰 내용·추측·모르는 내용을 구분한 뒤, 보류 이유와 확인 질문·담당을 적고 실제 보고·인계를 연습하는 신병용 PBL이다. 10개 미션 50개 단계, 30시간으로 구성하며 각 단계에 읽을 자료·작업 순서·예시·제출 내용·평가 조건을 직접 제시한다.', 'objectives': "<h4>프로젝트 배경</h4><p>드론·원거리 타격이 일상화된 현대 분쟁(러시아-우크라이나 전쟁, 미국-이란 긴장 국면 등)에서는 민간 시설 인접 표적, 단일 출처 오인 식별, 아군 위치 보고 지연으로 민간인 피해와 아군 오사가 반복되며, 이는 멈추고 재확인할 기준과 책임 주체가 불분명해 발생한다. 이에 본 프로젝트는 교육용 가상 시가전 상황을 바탕으로 위험 요인 분석부터 보류·재확인 조건, 가상 피해 추정 등급과 승인 권한, 중지 절차, 훈련·기록 서식, AAR 원인 분석 및 기준 개선을 단계별로 수행해 '확신이 없으면 멈추고 다시 확인한다'는 원칙과 사람의 최종 판단 체계를 담은 임무 보류·재확인 기준서를 완성한다.</p><h4>최종 목표</h4><p>확실하지 않은 보고를 받았을 때 무엇을 알고 무엇을 모르는지 구분하고, 보류하는 이유와 누구에게 무엇을 물을지 설명한다. 받은 답이 어느 질문에 답하는지 확인하고 남은 질문을 다음 사람에게 넘긴다. 이 과정을 담은 여덟 장의 교육용 기준서와 기록 양식을 완성한다.</p><h4>미션별 목표</h4><ul><li><p>미션1 보고에서 본 내용과 추측을 나눠 적는다. 사례마다 무엇을 누구에게 물을지 작성한다.</p></li><li><p>미션2 처음 출처와 전달 채널을 구별한다. 보류 조건 열 가지와 질문·담당을 작성한다.</p></li><li><p>미션3 민간 위험 신호와 모르는 내용을 구분한다. 부족한 정보 여덟 가지의 질문과 답의 한계를 적는다.</p></li><li><p>미션4 확인 질문 열두 가지를 세 사례에 적용한다. 질문 일부에만 답이 온 경우도 처리할 절차를 쓴다.</p></li><li><p>미션5 다섯 사례의 서로 다른 보류 이유를 설명한다. 대상·아군·민간·접수의 네 항목을 빠뜨리지 않는다.</p></li><li><p>미션6 보고자의 주장과 관찰 내용을 구별해 다시 쓴다. 세 원문을 짧은 기록·상세 기록에 옮기며 오류를 고친다.</p></li><li><p>미션7 순서대로 들어오는 카드마다 보고와 보류 상태를 갱신한다. 모든 조원이 보고자로 수행하고 다섯 영역을 평가받는다.</p></li><li><p>미션8 사건 당시 정보와 나중에 안 결과를 나눈다. 잘못된 표현·전달·담당 업무를 시각과 원문으로 설명한다.</p></li><li><p>미션9 수정 전후의 실제 문서를 보존한다. 같은 일곱 사례로 일치율·누락·실측 시간을 비교한다.</p></li><li><p>미션10 여덟 장과 네 부록에 실제 표·양식을 넣는다. 각 기준의 자료와 시험 결과를 찾아볼 수 있게 한다.</p></li></ul><p></p>", 'deliverables': ['docs/ops07_doc_framework.md (필수) — 다섯 사례의 관찰·추측·질문 표, 여덟 장 목차, 전원 설명 기록', 'framework/ops07_source_reliability_matrix.md (필수) — 출처 정리표, 여섯 표시 설명, 보류 조건 열 가지, 개인 원답·비교 계산, 질문 인계', 'analysis/ops07_uncertainty_assessment.md (필수) — 네 장소 위험표, 정보 부족 여덟 항목의 질문표, 열두 조합 담당표, 두 시험 상황, 적용 한계와 인계', 'sop/ops07_reconfirmation_procedure.md (필수) — 열두 질문과 세 사례 적용표, 다섯 순서 절차, 세 검토 질문, 전원 보고·인계 기록', 'sop/ops07_approval_thresholds_sop.md (필수) — 네 확인 순서, 다섯 사례의 개인·조별 판단, 짧은 보고, 미답 질문의 담당 인계', 'templates/ops07_aar_fact_action_template.md (필수) — 세 원문의 문장 분류, 여섯 표현 수정표, 다섯 영역 점검표, 두 기록 양식과 이관·사용 시험', 'rubric/ops07_evaluation_rubric.md (필수) — 스무 수준의 수행 평가표, 전원 보고자 역할표·훈련 로그, 두 평가자의 원점수·일치율, 개인 재평가', 'reports/ops07_redteam_critique_report.md (필수) — 두 사건 시각표·원인표, 다섯 가정의 검토, 두 시간 계산, 고칠 문장과 시험 계획', 'briefing/ops07_commander_1page_brief.md (필수) — 실제 기준 v1·v2, 일곱 사례 전후 개인 답·계산, 한 쪽 보고서·세 대안, 발표·세 질문 응답', 'manual/ops07_standard_operational_manual.md (필수) — 여덟 장과 네 부록의 기준서, 조항 근거표, 세 역할 의견, 설명 없이 사용한 시험, 개인·최종 평가와 개정 계획'], 'constraints': "<h4>교육용 자료와 실제 업무의 구분</h4><p>모든 상황·사례·표적·시각·위치·수치·역할 및 직책 배치는 교육용 가상 자료이며, 실제 부대명·위치·장비 및 무기 제원·작전 계획·교전수칙 원문·개인정보는 사용하지 않는다. 부대 내 촬영과 부대 자료의 외부 AI 서비스 입력은 금지한다. 본 과제는 표적 선정이나 타격 방법이 아니라 '보류·재확인' 판단 기준을 연습하는 학습용 문서 작업으로, 가상 부수적 피해 추정 등급과 승인 권한 매트릭스는 학습용 가상 기준으로만 작성하며 실제 권한 체계를 추정하거나 재구성하지 않는다. 실제 교전·타격 여부 판단은 적용 법규·교전수칙·지휘계통에 따른다.</p><h4>기록할 상태의 뜻</h4><p>보류는 이 연습에서 확인되지 않은 내용을 확인한 것처럼 처리하지 않고 검토를 멈춰 두는 상태다. 재확인은 모르는 내용을 담당자에게 다시 묻는 일이다. 접수는 상대가 요청을 받았다는 뜻이고, 회신은 질문에 대한 답이다. 접수됐다고 해서 질문의 답까지 받은 것은 아니다. 아래의 역할은 모두 교육용이며 실제 지휘권을 뜻하지 않는다.</p><p>현재 보류 상태와 지금 하는 일은 따로 적는다. 예: “보류 유지 / 질문을 보냈으나 접수 여부 모름 / 담당 P01 / 다음 교육 점검에서 접수 재확인”. “재확인”이나 “상급 승인 요청”을 적었다고 보류가 풀린 것은 아니다. 여기서 상급 승인 요청은 교육용 안전 검토 요청이며 실제 타격 승인을 뜻하지 않는다.</p><h4>자료에 없는 내용</h4><p>관측·보고·수신 시각을 구별하고 자료에 없으면 미제공이라고 적는다. 보지 못함을 없음으로, 보고 시도를 접수 완료로, 접수를 질문 답변 완료로 바꾸지 않는다. 다른 채널의 보고가 같은 제보를 옮겼으면 새 관측으로 세지 않는다.</p><h4>운영 시간과 도구</h4><p>4~6명 조와 별도 교관으로 운영한다. 10개 미션은 30시간이며 휴식·식사·이동은 별도다. 프로그래밍·AI 사용·상황판 제작은 요구하지 않는다. 미션 문서의 STEP 01~05에 결과를 모은다. 플랫폼이 정한 파일 확장자는 제출 담당자의 안내를 따르되 표의 모양이나 프로그램 차이만으로 내용이 같은 결과를 탈락시키지 않는다.</p>", 'evalSummary': '<h4>평가자가 볼 것</h4><p>각 단계 evalText의 제출 위치와 PASS 조건을 읽고 실제 표·문장·회차에서 확인한다. 미션별 evaluationKeys의 두 질문은 관련 단계의 조건을 다시 모은 최종 체크리스트다. “적절함·근거 충분” 같은 총평만 적지 말고 어느 조건을 어떤 기록에서 확인했는지 남긴다.</p><h4>통과 판정</h4><p>각 평가 질문의 조건을 모두 충족하면 PASS다. 한 항목이라도 없거나 반대로 쓰였으면 해당 단계·문장을 적어 FAIL로 판정하고 보완할 일을 알려 준다. 실제 평가하지 않은 것은 미평가다. 최종 교육 통과에는 미션별 두 질문, 모두 스무 질문이 PASS여야 하며 개인 수행도 확인돼야 한다. 말이나 표 모양이 예시와 같을 필요는 없지만 필요한 내용은 모두 확인돼야 한다.</p><h4>개인 수행</h4><p>조별 문서가 완성돼도 설명·보고·인계를 하지 않은 개인을 수행 완료로 보지 않는다. 미션7에서는 전원이 보고자가 돼야 하고 다섯 영역 모두 우수 3점 이상이어야 한다. 평균으로 미달 영역을 없애지 않는다. 두 평가자의 일치율 80%는 평가의 일관성을 보는 값이며 개인 합격 점수가 아니다. 채점 누락 회차는 별도 표시해 재관찰한다.</p><h4>보완과 재평가</h4><p>문서 오류는 원문과 수정본을 비교해 확인한다. 수행 오류는 빠뜨린 행동을 연습한 뒤 새 회차에서 확인한다. 원래 답·점수·오류를 지우지 말고 수정·재평가 기록을 따로 남긴다. 자료에 없는 사실·회신·측정·채점을 만드는 것은 해당 항목의 FAIL 사유다.</p>', 'prerequisites': '짧은 한국어 보고를 읽고 문서나 종이에 표를 작성할 수 있으면 시작할 수 있다. 군 전문용어·정보 등급·프로그래밍·AI 사용 경험은 필요하지 않다. 수업에서 쓰는 용어는 해당 단계에서 설명한다.', 'refFiles': [], 'refText': '', 'missions': [{'id': 'OPS07_m01', 'sequence': 1, 'title': '민간인과 아군의 피해로 이어지는 위험 신호 분석', 'durationHours': 2, 'overview': '<p>보고에서 본 내용과 추측을 나눠 적는다. 사례마다 무엇을 누구에게 물을지 작성한다.</p><p>아래 다섯 단계에는 필요한 자료·작업 순서·예시·제출 내용과 평가 조건이 들어 있다. 각 단계의 자료를 읽은 뒤 순서대로 작성한다.</p>', 'objectives': '<h4>이 미션을 마치면 할 수 있어야 하는 일</h4><ul><li><p>보고에서 본 내용과 추측을 나눠 적는다.</p></li><li><p>사례마다 무엇을 누구에게 물을지 작성한다.</p></li><li><p>여덟 장 목차를 만들고 모든 조원이 사례 한 건을 설명한다.</p></li></ul><p></p>', 'scenario': '<h4>학습자의 역할</h4><p>학습자는 민간인과 아군에게 불필요한 피해가 생기지 않도록 보고를 읽고, 모르는 내용을 질문하고, 다음 담당자에게 기록을 넘기는 역할을 연습한다. 사람의 소속을 맞히거나 실제 타격을 결정하는 과제가 아니다.</p><h4>이번 미션의 과제</h4><ul><li><p>보고에서 본 내용과 추측을 나눠 적는다.</p></li><li><p>사례마다 무엇을 누구에게 물을지 작성한다.</p></li><li><p>여덟 장 목차를 만들고 모든 조원이 사례 한 건을 설명한다.</p></li></ul><h4>자료와 역할의 범위</h4><p>사례·시각·장소·역할은 모두 교육용이다. 실제 부대 자료를 쓰지 않는다. 자료에 답이 없으면 자료에 없다고 적는다. 실제 행동 권한은 이 수업이 정하지 않는다.</p><h4>개인 작업과 조별 작업</h4><p>개인 판단을 먼저 요청한 단계에서는 토의 전에 자신의 답을 남긴다. 그 뒤 조별 답과 수정 이유를 별도로 쓴다. 보고·인계·시험 단계에서는 실제 한 말과 행동을 기록한다.</p>', 'attachments': [], 'materials': ['/uploads/OPS07/OPS07_m01_student.ipynb', '/uploads/OPS07/OPS07_m01_workbook.ipynb'], 'materialText': '<h4>실습 노트북 안내 (선택 보조 도구)</h4><p><strong>학생 실습용</strong>: <code>OPS07_m01_student.ipynb</code> (직접 ANSWER 표와 note를 작성합니다.)<br><strong>교수자·해설용</strong>: <code>OPS07_m01_workbook.ipynb</code> (예시 풀이와 해설 note가 포함되어 있습니다.)<br><em>※ 본 과제는 문서 편집기나 인쇄 양식으로도 수행할 수 있으며, 노트북은 사례·질문·기록 양식을 표로 정리하는 선택 보조 도구입니다.</em></p><h4>자료를 읽는 순서</h4>\n\n<ol>\n<li><p>먼저 C1~C5를 각각 읽습니다. STEP 1에서는 관찰 내용과 주장을 나누고, STEP 2에서는 위험 신호를 찾습니다.</p></li>\n<li><p>STEP 3에서는 앞의 사례를 다시 읽고 확인 질문과 담당자를 적습니다. C3의 장소 표기와 소속 주장은 서로 다른 질문으로 다룹니다.</p></li>\n<li><p>STEP 4에서는 아래 여덟 장 목차를 참고합니다. STEP 5의 개인 설명에 필요한 내용은 자신이 작성한 사례표에서 찾습니다.</p></li>\n</ol>\n\n<h4>자료 1. 관찰과 주장을 구별할 다섯 사례</h4>\n\n<h4>C1 · 시장 인근 차량</h4>\n\n<p>10:00 시장 입구 화물차에 대해 보고자 1명이 적 보급 차량이라고 주장했다.</p>\n\n<p>낮은 해상도 영상에서 적재물은 식별되지 않는다.</p>\n\n<p>시장은 영업 중이며 주민 이동이 보인다.</p>\n\n<h4>C2 · 열영상 인원과 아군 보고 공백</h4>\n\n<p>05:40 열영상에 4명이 보인다.</p>\n\n<p>무장 여부는 확인되지 않았다.</p>\n\n<p>정찰조 복귀 예정 시각과 겹치며 위치 보고는 예정 시각보다 20분 늦었다.</p>\n\n<p>이 카드는 기본 상황 시간대 이전의 별도 사례다.</p>\n\n<h4>C3 · 위치 표기 불일치</h4>\n\n<p>원 관측자의 구두 보고에는 다-4가, 두 차례 전달된 수신 기록에는 다-1이 적혀 있다.</p>\n\n<p>다-4에는 가상 아군 시설이 표시되어 있다.</p>\n\n<p>복창 기록이 없다.</p>\n\n<p>원 보고의 적 관측반이라는 주장도 별도로 확인되지 않았다.</p>\n\n<h4>C4 · 대피 정보 미반영</h4>\n\n<p>09:00 주민 이동 경로가 라-3에서 마-3 방향으로 바뀌었다는 담당자 보고가 있다.</p>\n\n<p>상황판에는 이전 경로가 남아 있다.</p>\n\n<p>새 경로의 이동 인원에 대한 식별 보고는 없다.</p>\n\n<h4>C5 · 단일 제보와 상충 정보</h4>\n\n<p>주민 1명이 마-3 창고에 적 지휘소가 있다고 말했다.</p>\n\n<p>제보자의 직접 관측 여부와 관측 시각은 미제공이다.</p>\n\n<p>민간 업체 직원으로 소개된 사람들의 출입 보고도 있다.</p>\n\n<p>어느 주장도 독립 확인을 마치지 않았다.</p>\n\n<h4>자료 2. 기준서의 여덟 장 목차</h4>\n\n<ol>\n<li>\n<p>1장. 목적과 적용 범위</p>\n<p><strong>이 장에서 답할 질문:</strong> 이 기준서를 언제 쓰며 어떤 판단에는 쓰지 않는가?</p>\n<p><strong>참고할 미션:</strong> 미션1</p>\n</li>\n\n<li>\n<p>2장. 용어 설명</p>\n<p><strong>이 장에서 답할 질문:</strong> 보류·재확인·접수·회신을 어떻게 구분하는가?</p>\n<p><strong>참고할 미션:</strong> 미션1~미션4</p>\n</li>\n\n<li>\n<p>3장. 위험 신호</p>\n<p><strong>이 장에서 답할 질문:</strong> 민간인·아군 피해를 막기 위해 무엇을 살피는가?</p>\n<p><strong>참고할 미션:</strong> 미션1</p>\n</li>\n\n<li>\n<p>4장. 보류와 재확인 조건</p>\n<p><strong>이 장에서 답할 질문:</strong> 어떤 정보가 빠졌을 때 무엇을 다시 묻는가?</p>\n<p><strong>참고할 미션:</strong> 미션2·미션4</p>\n</li>\n\n<li>\n<p>5장. 담당자와 검토자</p>\n<p><strong>이 장에서 답할 질문:</strong> 누가 다시 확인하고 누가 보고를 검토하는가?</p>\n<p><strong>참고할 미션:</strong> 미션3·미션5</p>\n</li>\n\n<li>\n<p>6장. 중지 요청과 재확인 절차</p>\n<p><strong>이 장에서 답할 질문:</strong> 위험 보고부터 회신 확인·인계까지 어떻게 하는가?</p>\n<p><strong>참고할 미션:</strong> 미션4·미션5</p>\n</li>\n\n<li>\n<p>7장. 보고와 기록</p>\n<p><strong>이 장에서 답할 질문:</strong> 어떤 내용을 말하고 어디에 남기는가?</p>\n<p><strong>참고할 미션:</strong> 미션6</p>\n</li>\n\n<li>\n<p>8장. 훈련 평가와 개정</p>\n<p><strong>이 장에서 답할 질문:</strong> 어떻게 연습·평가하고 잘못된 기준을 고치는가?</p>\n<p><strong>참고할 미션:</strong> 미션7~미션9</p>\n</li>\n</ol>\n\n<h4>자료를 사용할 때 지킬 점</h4>\n\n<p>C1~C5는 각각 별도의 사례입니다.</p>\n\n<p>서로 다른 카드의 시각과 정보를 하나의 상황처럼 합치지 않습니다.</p>\n\n<p>카드에 없는 관측·회신·소속을 추측해서 채우지 않습니다.</p>\n\n<p>모든 사례·시각·장소·역할은 교육용입니다. 필요한 자료와 구체적인 작업 순서는 각 단계 설명에도 들어 있습니다.</p>\n', 'guideSteps': [{'step': 1, 'problem': '다섯 사례를 읽고 관찰 내용과 추측을 나누기', 'activity': '<h4>이번 단계에서 하는 일</h4><p>보고에 “적 차량”이라고 적혀 있어도 실제로 확인한 것은 차량의 모습뿐일 수 있다. 먼저 무엇을 보았고 무엇을 추측했는지 나눠야 다음 사람이 보고 내용을 과장해서 받아들이지 않는다.</p><h4>먼저 읽을 자료</h4>\n\n<h4>사례 자료</h4>\n\n<p><strong>C1 · 시장 인근 차량</strong></p>\n\n<p>10:00 시장 입구 화물차에 대해 보고자 1명이 적 보급 차량이라고 주장했다.</p>\n\n<p>낮은 해상도 영상에서 적재물은 식별되지 않는다.</p>\n\n<p>시장은 영업 중이며 주민 이동이 보인다.</p>\n\n<p><strong>C2 · 열영상 인원과 아군 보고 공백</strong></p>\n\n<p>05:40 열영상에 4명이 보인다.</p>\n\n<p>무장 여부는 확인되지 않았다.</p>\n\n<p>정찰조 복귀 예정 시각과 겹치며 위치 보고는 예정 시각보다 20분 늦었다.</p>\n\n<p>이 카드는 기본 상황 시간대 이전의 별도 사례다.</p>\n\n<p><strong>C3 · 위치 표기 불일치</strong></p>\n\n<p>원 관측자의 구두 보고에는 다-4가, 두 차례 전달된 수신 기록에는 다-1이 적혀 있다.</p>\n\n<p>다-4에는 가상 아군 시설이 표시되어 있다.</p>\n\n<p>복창 기록이 없다.</p>\n\n<p>원 보고의 적 관측반이라는 주장도 별도로 확인되지 않았다.</p>\n\n<p><strong>C4 · 대피 정보 미반영</strong></p>\n\n<p>09:00 주민 이동 경로가 라-3에서 마-3 방향으로 바뀌었다는 담당자 보고가 있다.</p>\n\n<p>상황판에는 이전 경로가 남아 있다.</p>\n\n<p>새 경로의 이동 인원에 대한 식별 보고는 없다.</p>\n\n<p><strong>C5 · 단일 제보와 상충 정보</strong></p>\n\n<p>주민 1명이 마-3 창고에 적 지휘소가 있다고 말했다.</p>\n\n<p>제보자의 직접 관측 여부와 관측 시각은 미제공이다.</p>\n\n<p>민간 업체 직원으로 소개된 사람들의 출입 보고도 있다.</p>\n\n<p>어느 주장도 독립 확인을 마치지 않았다.</p>\n\n<p>아래 카드에는 나중에 무엇으로 밝혀졌는지 적혀 있지 않다.</p>\n\n<p>지금 적혀 있는 내용만 사용한다.</p>\n\n<p>나-2 등의 표기는 가상 장소의 이름이며 실제 좌표가 아니다.</p>\n\n<h4>작업 순서</h4><ol><li><p>개인 문서에 C1~C5를 한 줄씩 적고 아래 항목으로 표를 만든다.</p></li><li><p>“영상에 차량이 보인다”처럼 카드에 적힌 관찰 내용을 옮긴다. 관찰했다는 보고를 읽은 경우에는 “보고에 따르면”이라고 적는다.</p></li><li><p>“적 보급 차량이다”처럼 사람의 신분·소속·의도를 판단한 표현은 추측 또는 주장 칸에 적는다.</p></li><li><p>직접 관측 여부·관측 시각·현재 위치 등 카드에서 알 수 없는 내용은 “자료에 없음”이라고 적는다. 없는 사실을 짐작해 채우지 않는다.</p></li><li><p>각 줄의 마지막에 문장을 가져온 카드 번호를 적는다. 조원과 이야기하기 전에 개인 표를 저장한다.</p></li></ol><h4>작성 예시</h4><p>다음은 쓰는 방식을 보여 주는 예다. 자기 판단이나 실제 수행·측정 결과가 필요한 칸에 그대로 복사하지 않는다.</p><p>C1: 관찰 내용 “보고에 따르면 시장 입구에 화물차가 있고 주민이 이동한다.” / 주장 “보고자 1명은 적 보급 차량이라고 말했다.” / 모르는 내용 “적재물과 차량 소속은 확인되지 않았다.”</p><h4>작성할 표 또는 기록 항목</h4><ul><li><p>카드 번호</p></li><li><p>보고에 적힌 관찰 내용</p></li><li><p>보고자의 주장 또는 추측</p></li><li><p>자료에 없거나 확인되지 않은 내용</p></li></ul><h4>제출할 내용과 끝내는 기준</h4><p>C1~C5가 모두 들어간 개인 표 1개. 각 카드마다 네 항목을 채운다.</p><p>미션 제출 문서 docs/ops07_doc_framework.md의 STEP 01 아래에 작성한다. 개인 기록과 조별 결과를 구분하고 실제 이름 대신 학습자 코드를 써도 된다. 표의 모양이 달라도 필요한 내용을 찾을 수 있으면 된다.</p><h4>마치기 전 확인</h4><ul><li><p>C1~C5 다섯 줄이 모두 있다.</p></li><li><p>차량·인원·물체를 보았다는 내용과 적·아군·민간이라는 판단을 다른 칸에 적었다.</p></li><li><p>확인되지 않은 내용은 “자료에 없음” 또는 “확인되지 않음”으로 표시했다.</p></li><li><p>카드에 없는 소속·회신·시각·사후 결말을 추가하지 않았다.</p></li></ul><h4>배정 시간</h4><p>20분</p>', 'evalText': '<h4>평가할 자료와 위치</h4><p>docs/ops07_doc_framework.md의 STEP 01 결과와 해당 개인 수행 기록을 확인한다. 이 단계에서 요청한 제출 내용은 다음과 같다: C1~C5가 모두 들어간 개인 표 1개. 각 카드마다 네 항목을 채운다.</p><h4>평가 순서</h4><ol><li><p>아래 통과 조건을 하나씩 읽고 제출물에서 확인한 위치를 옆에 적는다.</p></li><li><p>실제 설명·보고·인계가 필요한 단계는 수행 기록이나 직접 관찰로 확인한다. 서식만 있으면 수행한 것으로 보지 않는다.</p></li><li><p>통과 조건을 모두 충족하면 이 단계는 PASS다. 하나라도 빠지거나 반대로 적었으면 FAIL로 두고 빠진 항목을 구체적으로 적는다. 표현이나 표 모양이 예시와 달라도 뜻이 같고 근거를 찾을 수 있으면 인정한다.</p></li></ol><h4>PASS 조건</h4><ul><li><p>C1~C5 다섯 줄이 모두 있다.</p></li><li><p>차량·인원·물체를 보았다는 내용과 적·아군·민간이라는 판단을 다른 칸에 적었다.</p></li><li><p>확인되지 않은 내용은 “자료에 없음” 또는 “확인되지 않음”으로 표시했다.</p></li><li><p>카드에 없는 소속·회신·시각·사후 결말을 추가하지 않았다.</p></li></ul><h4>FAIL에 해당하는 예</h4><ul><li><p>C1의 관찰 내용에 “적 보급 차량이 확인됐다”라고 적었다.</p></li><li><p>C2 인원을 아군으로 확정하거나 카드에 없는 연락 회복 사실을 적었다.</p></li></ul><h4>이해 여부를 확인할 질문</h4><p>C1에서 차량을 보았다는 말과 적 차량이라는 말은 무엇이 다른가?</p><h4>인정할 수 있는 답의 핵심</h4><p>차량이 있다는 보고는 있지만 차량의 소속을 확인할 근거는 없다고 구분해 말한다.</p><h4>보완과 재평가</h4><p>다섯 카드의 단정 표현에 밑줄을 긋고 관찰 칸에서 주장 칸으로 옮긴다. 수정 전 표도 남긴다.</p><p>수정 전 결과와 재평가 결과를 구분해 남긴다. 실제 하지 않은 보고·채점·시험을 나중에 했던 것처럼 채우지 않는다.</p>', 'attachments': []}, {'step': 2, 'problem': '사례별 위험 신호와 놓쳤을 때의 피해를 적기', 'activity': '<h4>이번 단계에서 하는 일</h4><p>“위험하다”라는 말만으로는 무엇을 확인해야 하는지 알 수 없다. 각 카드에서 위험을 알리는 구체적인 문장을 찾고 어떤 오해나 피해로 이어질 수 있는지 설명한다.</p><h4>먼저 읽을 자료</h4>\n\n<h4>사례 자료</h4>\n\n<p><strong>C1 · 시장 인근 차량</strong></p>\n\n<p>10:00 시장 입구 화물차에 대해 보고자 1명이 적 보급 차량이라고 주장했다.</p>\n\n<p>낮은 해상도 영상에서 적재물은 식별되지 않는다.</p>\n\n<p>시장은 영업 중이며 주민 이동이 보인다.</p>\n\n<p><strong>C2 · 열영상 인원과 아군 보고 공백</strong></p>\n\n<p>05:40 열영상에 4명이 보인다.</p>\n\n<p>무장 여부는 확인되지 않았다.</p>\n\n<p>정찰조 복귀 예정 시각과 겹치며 위치 보고는 예정 시각보다 20분 늦었다.</p>\n\n<p>이 카드는 기본 상황 시간대 이전의 별도 사례다.</p>\n\n<p><strong>C3 · 위치 표기 불일치</strong></p>\n\n<p>원 관측자의 구두 보고에는 다-4가, 두 차례 전달된 수신 기록에는 다-1이 적혀 있다.</p>\n\n<p>다-4에는 가상 아군 시설이 표시되어 있다.</p>\n\n<p>복창 기록이 없다.</p>\n\n<p>원 보고의 적 관측반이라는 주장도 별도로 확인되지 않았다.</p>\n\n<p><strong>C4 · 대피 정보 미반영</strong></p>\n\n<p>09:00 주민 이동 경로가 라-3에서 마-3 방향으로 바뀌었다는 담당자 보고가 있다.</p>\n\n<p>상황판에는 이전 경로가 남아 있다.</p>\n\n<p>새 경로의 이동 인원에 대한 식별 보고는 없다.</p>\n\n<p><strong>C5 · 단일 제보와 상충 정보</strong></p>\n\n<p>주민 1명이 마-3 창고에 적 지휘소가 있다고 말했다.</p>\n\n<p>제보자의 직접 관측 여부와 관측 시각은 미제공이다.</p>\n\n<p>민간 업체 직원으로 소개된 사람들의 출입 보고도 있다.</p>\n\n<p>어느 주장도 독립 확인을 마치지 않았다.</p>\n\n<h4>자료를 나누어 읽는 기준</h4>\n\n<ul>\n<li>\n<p><strong>대상 확인:</strong> 누구인지·어떤 물건인지 확인하지 못한 점</p>\n</li>\n\n<li>\n<p><strong>주변 민간 상황:</strong> 시장 영업·주민 이동·건물 이용 등</p>\n</li>\n\n<li>\n<p><strong>아군 위치:</strong> 아군의 현재 위치와 연락 여부</p>\n</li>\n\n<li>\n<p><strong>정보 전달:</strong> 표기 오류·누락·오래된 보고·같은 제보의 재전달</p>\n</li>\n\n<li>\n<p><strong>시간 압박:</strong> 시간이 부족하다는 요구가 있는지. 없으면 자료에 없다고 적음</p>\n</li>\n</ul>\n\n<h4>작업 순서</h4><ol><li><p>다섯 분류를 읽고 C1~C5에 해당하는 분류를 붙인다. 한 카드에 둘 이상 붙여도 된다.</p></li><li><p>분류 이름만 적지 말고 카드에서 찾은 문장을 위험 신호 칸에 옮긴다.</p></li><li><p>그 신호를 무시하면 생길 수 있는 오해나 피해를 한 문장으로 적는다. 실제 피해가 발생했다고 쓰지는 않는다.</p></li><li><p>다섯 분류를 모두 점검한다. 해당 내용이 없는 분류에는 “이 카드에는 관련 정보 없음”이라고 적는다.</p></li><li><p>조원과 비교하고 분류가 달랐다면 카드의 어느 문장을 근거로 삼았는지 함께 적는다.</p></li></ol><h4>작성 예시</h4><p>다음은 쓰는 방식을 보여 주는 예다. 자기 판단이나 실제 수행·측정 결과가 필요한 칸에 그대로 복사하지 않는다.</p><p>C3 / 정보 전달 / “다-4라는 구두 보고와 다-1이라는 수신 기록이 다름” / “서로 다른 장소를 같은 장소로 오해해 아군이나 민간인에게 위험을 줄 수 있음.”</p><h4>작성할 표 또는 기록 항목</h4><ul><li><p>카드 번호</p></li><li><p>분류</p></li><li><p>카드에서 찾은 위험 신호</p></li><li><p>무시했을 때 생길 수 있는 오해·피해</p></li><li><p>조원 의견과 다른 점</p></li></ul><h4>제출할 내용과 끝내는 기준</h4><p>다섯 카드마다 위험 신호 1개 이상을 적은 표와 다섯 분류를 점검한 기록.</p><p>미션 제출 문서 docs/ops07_doc_framework.md의 STEP 02 아래에 작성한다. 개인 기록과 조별 결과를 구분하고 실제 이름 대신 학습자 코드를 써도 된다. 표의 모양이 달라도 필요한 내용을 찾을 수 있으면 된다.</p><h4>마치기 전 확인</h4><ul><li><p>다섯 카드에 위험 신호가 적어도 하나씩 있다.</p></li><li><p>각 신호 옆에 그 신호를 찾은 카드 번호와 문장이 있다.</p></li><li><p>피해 가능성을 해당 신호와 연결해 설명했다.</p></li><li><p>시간 압박 등 제공되지 않은 상황을 실제 있었던 일처럼 추가하지 않았다.</p></li></ul><h4>배정 시간</h4><p>25분</p>', 'evalText': '<h4>평가할 자료와 위치</h4><p>docs/ops07_doc_framework.md의 STEP 02 결과와 해당 개인 수행 기록을 확인한다. 이 단계에서 요청한 제출 내용은 다음과 같다: 다섯 카드마다 위험 신호 1개 이상을 적은 표와 다섯 분류를 점검한 기록.</p><h4>평가 순서</h4><ol><li><p>아래 통과 조건을 하나씩 읽고 제출물에서 확인한 위치를 옆에 적는다.</p></li><li><p>실제 설명·보고·인계가 필요한 단계는 수행 기록이나 직접 관찰로 확인한다. 서식만 있으면 수행한 것으로 보지 않는다.</p></li><li><p>통과 조건을 모두 충족하면 이 단계는 PASS다. 하나라도 빠지거나 반대로 적었으면 FAIL로 두고 빠진 항목을 구체적으로 적는다. 표현이나 표 모양이 예시와 달라도 뜻이 같고 근거를 찾을 수 있으면 인정한다.</p></li></ol><h4>PASS 조건</h4><ul><li><p>다섯 카드에 위험 신호가 적어도 하나씩 있다.</p></li><li><p>각 신호 옆에 그 신호를 찾은 카드 번호와 문장이 있다.</p></li><li><p>피해 가능성을 해당 신호와 연결해 설명했다.</p></li><li><p>시간 압박 등 제공되지 않은 상황을 실제 있었던 일처럼 추가하지 않았다.</p></li></ul><h4>FAIL에 해당하는 예</h4><ul><li><p>위험 신호 칸을 전부 “정보 부족”으로만 채워 무엇이 빠졌는지 알 수 없다.</p></li><li><p>근거가 없는 카드에도 “상급자가 급하게 지시했다”고 적었다.</p></li></ul><h4>이해 여부를 확인할 질문</h4><p>C3에서 어떤 내용이 다르며 왜 중요한가?</p><h4>인정할 수 있는 답의 핵심</h4><p>다-4와 다-1이라는 장소 표기가 다르고 원문에는 아군 시설 정보도 있어 원 보고와 수신 기록을 다시 대조해야 한다고 말한다.</p><h4>보완과 재평가</h4><p>“위험함·정보 부족”이라고만 쓴 줄에 카드의 실제 문장을 넣고 생길 수 있는 오해를 다시 쓴다.</p><p>수정 전 결과와 재평가 결과를 구분해 남긴다. 실제 하지 않은 보고·채점·시험을 나중에 했던 것처럼 채우지 않는다.</p>', 'attachments': []}, {'step': 3, 'problem': '모르는 내용을 확인할 질문과 담당자를 정하기', 'activity': '<h4>이번 단계에서 하는 일</h4><p>위험을 적은 뒤에는 누가 무엇을 다시 확인해야 하는지 정해야 한다. “재확인 필요”라고만 쓰지 말고 상대가 답할 수 있는 질문을 만든다.</p><h4>먼저 읽을 자료</h4>\n\n<h4>사례 자료</h4>\n\n<p><strong>C1 · 시장 인근 차량</strong></p>\n\n<p>10:00 시장 입구 화물차에 대해 보고자 1명이 적 보급 차량이라고 주장했다.</p>\n\n<p>낮은 해상도 영상에서 적재물은 식별되지 않는다.</p>\n\n<p>시장은 영업 중이며 주민 이동이 보인다.</p>\n\n<p><strong>C2 · 열영상 인원과 아군 보고 공백</strong></p>\n\n<p>05:40 열영상에 4명이 보인다.</p>\n\n<p>무장 여부는 확인되지 않았다.</p>\n\n<p>정찰조 복귀 예정 시각과 겹치며 위치 보고는 예정 시각보다 20분 늦었다.</p>\n\n<p>이 카드는 기본 상황 시간대 이전의 별도 사례다.</p>\n\n<p><strong>C3 · 위치 표기 불일치</strong></p>\n\n<p>원 관측자의 구두 보고에는 다-4가, 두 차례 전달된 수신 기록에는 다-1이 적혀 있다.</p>\n\n<p>다-4에는 가상 아군 시설이 표시되어 있다.</p>\n\n<p>복창 기록이 없다.</p>\n\n<p>원 보고의 적 관측반이라는 주장도 별도로 확인되지 않았다.</p>\n\n<p><strong>C4 · 대피 정보 미반영</strong></p>\n\n<p>09:00 주민 이동 경로가 라-3에서 마-3 방향으로 바뀌었다는 담당자 보고가 있다.</p>\n\n<p>상황판에는 이전 경로가 남아 있다.</p>\n\n<p>새 경로의 이동 인원에 대한 식별 보고는 없다.</p>\n\n<p><strong>C5 · 단일 제보와 상충 정보</strong></p>\n\n<p>주민 1명이 마-3 창고에 적 지휘소가 있다고 말했다.</p>\n\n<p>제보자의 직접 관측 여부와 관측 시각은 미제공이다.</p>\n\n<p>민간 업체 직원으로 소개된 사람들의 출입 보고도 있다.</p>\n\n<p>어느 주장도 독립 확인을 마치지 않았다.</p>\n\n<p>보류는 이 연습에서 확인되지 않은 내용을 확인한 것처럼 처리하지 않고 검토를 멈춰 두는 상태다.</p>\n\n<p>재확인은 모르는 내용을 담당자에게 다시 묻는 일이다.</p>\n\n<p>접수는 상대가 요청을 받았다는 뜻이고, 회신은 질문에 대한 답이다.</p>\n\n<p>접수됐다고 해서 질문의 답까지 받은 것은 아니다.</p>\n\n<p>아래의 역할은 모두 교육용이며 실제 지휘권을 뜻하지 않는다.</p>\n\n<p>이 단계에서 정할 수 있는 가상 담당 역할: 원 보고를 한 관찰자, 보고 기록 담당자, 아군 위치 보고 담당자, 민간 시설·대피 정보 담당자, 안전 검토자.</p>\n\n<p>실제 사람이 답했다고 가정하지 않는다.</p>\n\n<h4>작업 순서</h4><ol><li><p>C1~C5 각각에서 지금 확인하지 못한 내용을 한 가지 이상 고른다.</p></li><li><p>그 내용을 알 수 있는 역할을 위 목록에서 정한다.</p></li><li><p>질문에 장소·대상·확인할 내용을 넣는다. “안전한가요?”처럼 범위가 넓은 질문은 쓰지 않는다.</p></li><li><p>어떤 답이나 자료를 받으면 그 질문에 답한 것으로 볼지 적는다. 예를 들어 원 보고 문장과 수신 기록의 대조 결과가 필요할 수 있다.</p></li><li><p>C3는 장소 표기 확인과 인원의 신분·소속 확인을 서로 다른 질문으로 쓴다.</p></li><li><p>아직 답을 받지 못한 줄은 보류 유지로 남기고 담당자와 다음에 확인할 일을 적는다.</p></li></ol><h4>작성 예시</h4><p>다음은 쓰는 방식을 보여 주는 예다. 자기 판단이나 실제 수행·측정 결과가 필요한 칸에 그대로 복사하지 않는다.</p><p>C3 / 질문 “원 보고의 장소 표기는 다-4입니까, 다-1입니까? 원 보고 기록을 확인해 알려 주십시오.” / 담당 “원 보고 관찰자와 기록 담당자” / 필요한 답 “원 보고와 수신 기록을 대조한 결과.” 이 답만으로 인원의 소속까지 확인된 것은 아니다.</p><h4>작성할 표 또는 기록 항목</h4><ul><li><p>카드 번호</p></li><li><p>보류하는 이유</p></li><li><p>누구에게 물을지</p></li><li><p>그대로 전달할 질문</p></li><li><p>필요한 답·자료</p></li><li><p>현재 답을 받았는지</p></li></ul><h4>제출할 내용과 끝내는 기준</h4><p>C1~C5 질문표 1개. 카드별 질문 1개 이상, C3는 서로 다른 질문 2개 이상.</p><p>미션 제출 문서 docs/ops07_doc_framework.md의 STEP 03 아래에 작성한다. 개인 기록과 조별 결과를 구분하고 실제 이름 대신 학습자 코드를 써도 된다. 표의 모양이 달라도 필요한 내용을 찾을 수 있으면 된다.</p><h4>마치기 전 확인</h4><ul><li><p>다섯 카드 모두 질문과 담당 역할이 있다.</p></li><li><p>질문에 대상 또는 장소와 확인할 내용이 드러난다.</p></li><li><p>C3의 장소 표기와 소속 확인을 별도로 적었다.</p></li><li><p>없는 회신을 만들지 않고 답을 기다리는 항목을 표시했다.</p></li></ul><h4>배정 시간</h4><p>25분</p>', 'evalText': '<h4>평가할 자료와 위치</h4><p>docs/ops07_doc_framework.md의 STEP 03 결과와 해당 개인 수행 기록을 확인한다. 이 단계에서 요청한 제출 내용은 다음과 같다: C1~C5 질문표 1개. 카드별 질문 1개 이상, C3는 서로 다른 질문 2개 이상.</p><h4>평가 순서</h4><ol><li><p>아래 통과 조건을 하나씩 읽고 제출물에서 확인한 위치를 옆에 적는다.</p></li><li><p>실제 설명·보고·인계가 필요한 단계는 수행 기록이나 직접 관찰로 확인한다. 서식만 있으면 수행한 것으로 보지 않는다.</p></li><li><p>통과 조건을 모두 충족하면 이 단계는 PASS다. 하나라도 빠지거나 반대로 적었으면 FAIL로 두고 빠진 항목을 구체적으로 적는다. 표현이나 표 모양이 예시와 달라도 뜻이 같고 근거를 찾을 수 있으면 인정한다.</p></li></ol><h4>PASS 조건</h4><ul><li><p>다섯 카드 모두 질문과 담당 역할이 있다.</p></li><li><p>질문에 대상 또는 장소와 확인할 내용이 드러난다.</p></li><li><p>C3의 장소 표기와 소속 확인을 별도로 적었다.</p></li><li><p>없는 회신을 만들지 않고 답을 기다리는 항목을 표시했다.</p></li></ul><h4>FAIL에 해당하는 예</h4><ul><li><p>모든 카드의 질문이 “다시 확인해 주세요”이고 무엇을 확인할지 없다.</p></li><li><p>C3의 장소 표기만 고친 뒤 “대상 확인 완료”로 적었다.</p></li></ul><h4>이해 여부를 확인할 질문</h4><p>C3의 장소 표기를 바로잡으면 무엇이 아직 남는가?</p><h4>인정할 수 있는 답의 핵심</h4><p>장소의 표기 차이만 해결됐을 뿐 그곳 인원의 신분·소속과 주변 상황은 별도 확인이 필요하다고 답한다.</p><h4>보완과 재평가</h4><p>각 질문에 상대가 답해야 할 구체적인 내용을 넣고 C3를 장소 질문과 소속 질문으로 나눈다.</p><p>수정 전 결과와 재평가 결과를 구분해 남긴다. 실제 하지 않은 보고·채점·시험을 나중에 했던 것처럼 채우지 않는다.</p>', 'attachments': []}, {'step': 4, 'problem': '최종 기준서의 여덟 장에 들어갈 내용을 정하기', 'activity': '<h4>이번 단계에서 하는 일</h4><p>기준서는 위험 신호를 본 신병이 어느 부분을 읽어야 할지 바로 찾을 수 있어야 한다. 이 단계에서는 글을 모두 쓰기보다 여덟 장의 역할과 작성 순서를 정한다.</p><h4>먼저 읽을 자료</h4>\n\n<h4>기준서의 여덟 장 목차</h4>\n\n<ol>\n<li>\n<p>1장. 목적과 적용 범위</p>\n<p><strong>이 장에서 답할 질문:</strong> 이 기준서를 언제 쓰며 어떤 판단에는 쓰지 않는가</p>\n<p><strong>참고할 미션:</strong> M1</p>\n</li>\n\n<li>\n<p>2장. 용어 설명</p>\n<p><strong>이 장에서 답할 질문:</strong> 보류·재확인·접수·회신을 어떻게 구분하는가</p>\n<p><strong>참고할 미션:</strong> M1~M4</p>\n</li>\n\n<li>\n<p>3장. 위험 신호</p>\n<p><strong>이 장에서 답할 질문:</strong> 민간인·아군 피해를 막기 위해 무엇을 살피는가</p>\n<p><strong>참고할 미션:</strong> M1</p>\n</li>\n\n<li>\n<p>4장. 보류와 재확인 조건</p>\n<p><strong>이 장에서 답할 질문:</strong> 어떤 정보가 빠졌을 때 무엇을 다시 묻는가</p>\n<p><strong>참고할 미션:</strong> M2·M4</p>\n</li>\n\n<li>\n<p>5장. 담당자와 검토자</p>\n<p><strong>이 장에서 답할 질문:</strong> 누가 다시 확인하고 누가 보고를 검토하는가</p>\n<p><strong>참고할 미션:</strong> M3·M5</p>\n</li>\n\n<li>\n<p>6장. 중지 요청과 재확인 절차</p>\n<p><strong>이 장에서 답할 질문:</strong> 위험 보고부터 회신 확인·인계까지 어떻게 하는가</p>\n<p><strong>참고할 미션:</strong> M4·M5</p>\n</li>\n\n<li>\n<p>7장. 보고와 기록</p>\n<p><strong>이 장에서 답할 질문:</strong> 어떤 내용을 말하고 어디에 남기는가</p>\n<p><strong>참고할 미션:</strong> M6</p>\n</li>\n\n<li>\n<p>8장. 훈련 평가와 개정</p>\n<p><strong>이 장에서 답할 질문:</strong> 어떻게 연습·평가하고 잘못된 기준을 고치는가</p>\n<p><strong>참고할 미션:</strong> M7~M9</p>\n</li>\n</ol>\n\n<p>M은 미션을 뜻한다.</p>\n\n<p>예를 들어 M6는 보고·기록 서식을 만드는 여섯 번째 미션이다.</p>\n\n<p>아래 제목은 최종 기준서의 공통 목차로 사용한다.</p>\n\n<h4>작업 순서</h4><ol><li><p>아래 여덟 장의 번호와 제목을 문서에 옮긴다.</p></li><li><p>각 장에 “이 장을 읽고 할 수 있어야 하는 일”을 자신의 말로 한 문장씩 적는다.</p></li><li><p>참고할 미션을 적고 그 미션에서 가져올 표나 서식의 이름을 적는다.</p></li><li><p>마지막에 부록 목록을 만든다. 점검표, 요약·상세 기록 서식, 절차도, 수행 평가표의 네 종류를 넣는다.</p></li><li><p>위험을 발견한 사람이 가장 먼저 읽을 장과 그다음 읽을 장을 고르고 그 이유를 적는다.</p></li></ol><h4>작성 예시</h4><p>다음은 쓰는 방식을 보여 주는 예다. 자기 판단이나 실제 수행·측정 결과가 필요한 칸에 그대로 복사하지 않는다.</p><p>7장 보고와 기록 / 할 수 있어야 하는 일 “관찰 내용과 추측을 나누어 보고하고 아직 답을 받지 못한 질문을 적는다.” / 가져올 자료 “M6의 요약형·상세형 기록 서식.”</p><h4>작성할 표 또는 기록 항목</h4><ul><li><p>장 번호·제목</p></li><li><p>읽은 뒤 할 수 있어야 하는 일</p></li><li><p>참고할 미션</p></li><li><p>가져올 표·서식</p></li></ul><h4>제출할 내용과 끝내는 기준</h4><p>여덟 장 목차표, 네 종류의 부록 목록, 먼저 읽을 장을 고른 이유.</p><p>미션 제출 문서 docs/ops07_doc_framework.md의 STEP 04 아래에 작성한다. 개인 기록과 조별 결과를 구분하고 실제 이름 대신 학습자 코드를 써도 된다. 표의 모양이 달라도 필요한 내용을 찾을 수 있으면 된다.</p><h4>마치기 전 확인</h4><ul><li><p>여덟 장이 빠짐없이 있다.</p></li><li><p>각 장마다 할 일과 참고 미션·자료 이름이 있다.</p></li><li><p>점검표·기록 서식·절차도·수행 평가표의 네 부록이 있다.</p></li><li><p>먼저 읽을 장을 선택한 이유가 실제 확인·보고 업무와 관련된다.</p></li></ul><h4>배정 시간</h4><p>30분</p>', 'evalText': '<h4>평가할 자료와 위치</h4><p>docs/ops07_doc_framework.md의 STEP 04 결과와 해당 개인 수행 기록을 확인한다. 이 단계에서 요청한 제출 내용은 다음과 같다: 여덟 장 목차표, 네 종류의 부록 목록, 먼저 읽을 장을 고른 이유.</p><h4>평가 순서</h4><ol><li><p>아래 통과 조건을 하나씩 읽고 제출물에서 확인한 위치를 옆에 적는다.</p></li><li><p>실제 설명·보고·인계가 필요한 단계는 수행 기록이나 직접 관찰로 확인한다. 서식만 있으면 수행한 것으로 보지 않는다.</p></li><li><p>통과 조건을 모두 충족하면 이 단계는 PASS다. 하나라도 빠지거나 반대로 적었으면 FAIL로 두고 빠진 항목을 구체적으로 적는다. 표현이나 표 모양이 예시와 달라도 뜻이 같고 근거를 찾을 수 있으면 인정한다.</p></li></ol><h4>PASS 조건</h4><ul><li><p>여덟 장이 빠짐없이 있다.</p></li><li><p>각 장마다 할 일과 참고 미션·자료 이름이 있다.</p></li><li><p>점검표·기록 서식·절차도·수행 평가표의 네 부록이 있다.</p></li><li><p>먼저 읽을 장을 선택한 이유가 실제 확인·보고 업무와 관련된다.</p></li></ul><h4>FAIL에 해당하는 예</h4><ul><li><p>장 제목만 있고 각 장에서 무엇을 찾아야 하는지 없다.</p></li><li><p>모든 장의 참고 자료를 “앞 미션 결과”라고만 적었다.</p></li></ul><h4>이해 여부를 확인할 질문</h4><p>아직 답을 받지 못한 질문을 기록하는 방법은 어느 장에서 찾는가?</p><h4>인정할 수 있는 답의 핵심</h4><p>7장을 찾고 보고·기록 서식이 들어 있다고 설명한다. 해당 조가 6장 절차와 함께 설명해도 관련 위치를 정확히 찾으면 인정한다.</p><h4>보완과 재평가</h4><p>내용이 없는 장에 사용자가 찾을 질문을 적고 가져올 표·서식의 이름을 붙인다.</p><p>수정 전 결과와 재평가 결과를 구분해 남긴다. 실제 하지 않은 보고·채점·시험을 나중에 했던 것처럼 채우지 않는다.</p>', 'attachments': []}, {'step': 5, 'problem': '사례 한 건을 설명하고 조원과 다른 판단을 비교하기', 'activity': '<h4>이번 단계에서 하는 일</h4><p>조별 표를 만들었다고 해서 모든 조원이 내용을 이해한 것은 아니다. 각자가 위험과 질문을 말하고, 서로 다르게 쓴 이유를 카드 문장으로 확인한다.</p><h4>먼저 읽을 자료</h4>\n\n<h4>사례 자료</h4>\n\n<p><strong>C3 · 위치 표기 불일치</strong></p>\n\n<p>원 관측자의 구두 보고에는 다-4가, 두 차례 전달된 수신 기록에는 다-1이 적혀 있다.</p>\n\n<p>다-4에는 가상 아군 시설이 표시되어 있다.</p>\n\n<p>복창 기록이 없다.</p>\n\n<p>원 보고의 적 관측반이라는 주장도 별도로 확인되지 않았다.</p>\n\n<p><strong>C5 · 단일 제보와 상충 정보</strong></p>\n\n<p>주민 1명이 마-3 창고에 적 지휘소가 있다고 말했다.</p>\n\n<p>제보자의 직접 관측 여부와 관측 시각은 미제공이다.</p>\n\n<p>민간 업체 직원으로 소개된 사람들의 출입 보고도 있다.</p>\n\n<p>어느 주장도 독립 확인을 마치지 않았다.</p>\n\n<p>개인 설명은 C3 또는 C5 중 하나를 고른다.</p>\n\n<p>기록자는 설명을 들은 내용을 적고 설명자는 기록이 자신의 말과 같은지 확인한다.</p>\n\n<h4>작업 순서</h4><ol><li><p>각자 C3 또는 C5를 고르고 아래 네 문장을 준비한다. “보고에는 …라고 적혀 있다 / 그러나 …는 모른다 / 그래서 …를 보류한다 / …담당자에게 …를 묻겠다.”</p></li><li><p>한 사람씩 1분 안에 설명한다. 이 시간은 교육용 발표 시간이지 실제 대응 시한이 아니다.</p></li><li><p>기록자는 이름 또는 학습자 코드, 선택 카드, 설명에서 나온 네 내용을 적는다.</p></li><li><p>조원들의 분류나 질문이 달랐다면 서로 다른 문장을 나란히 적고 카드 원문과 비교한다.</p></li><li><p>조가 채택한 설명과 바꾼 이유를 적는다. 의견이 남아 있으면 무엇이 다른지와 교관에게 물을 질문을 남긴다.</p></li></ol><h4>작성 예시</h4><p>다음은 쓰는 방식을 보여 주는 예다. 자기 판단이나 실제 수행·측정 결과가 필요한 칸에 그대로 복사하지 않는다.</p><p>“C5에는 적 지휘소라는 제보와 민간 업체 직원 출입 보고가 있습니다. 어느 쪽도 확인을 마치지 않았습니다. 소속을 단정하지 않고 제보의 직접 관측 여부와 창고 이용 상태를 다시 묻겠습니다.”</p><h4>작성할 표 또는 기록 항목</h4><ul><li><p>학습자 코드</p></li><li><p>선택 카드</p></li><li><p>관찰·보고 내용</p></li><li><p>모르는 내용</p></li><li><p>보류 이유</p></li><li><p>확인 질문·담당</p></li><li><p>조별 수정 이유 또는 남은 의견</p></li></ul><h4>제출할 내용과 끝내는 기준</h4><p>조원 전원의 개인 설명 기록과 조별 비교 기록 1개. 실제 의견 차이가 없었다면 없었다고 적는다.</p><p>미션 제출 문서 docs/ops07_doc_framework.md의 STEP 05 아래에 작성한다. 개인 기록과 조별 결과를 구분하고 실제 이름 대신 학습자 코드를 써도 된다. 표의 모양이 달라도 필요한 내용을 찾을 수 있으면 된다.</p><h4>마치기 전 확인</h4><ul><li><p>조원 전원이 한 사례를 설명한 기록이 있다.</p></li><li><p>설명에 보고 내용·모르는 내용·보류 이유·확인 질문이 모두 있다.</p></li><li><p>다른 의견을 카드 문장과 비교했거나 실제 차이가 없었다고 기록했다.</p></li><li><p>조별 수정 이유 또는 남은 질문을 적었다.</p></li></ul><h4>배정 시간</h4><p>20분</p>', 'evalText': '<h4>평가할 자료와 위치</h4><p>docs/ops07_doc_framework.md의 STEP 05 결과와 해당 개인 수행 기록을 확인한다. 이 단계에서 요청한 제출 내용은 다음과 같다: 조원 전원의 개인 설명 기록과 조별 비교 기록 1개. 실제 의견 차이가 없었다면 없었다고 적는다.</p><h4>평가 순서</h4><ol><li><p>아래 통과 조건을 하나씩 읽고 제출물에서 확인한 위치를 옆에 적는다.</p></li><li><p>실제 설명·보고·인계가 필요한 단계는 수행 기록이나 직접 관찰로 확인한다. 서식만 있으면 수행한 것으로 보지 않는다.</p></li><li><p>통과 조건을 모두 충족하면 이 단계는 PASS다. 하나라도 빠지거나 반대로 적었으면 FAIL로 두고 빠진 항목을 구체적으로 적는다. 표현이나 표 모양이 예시와 달라도 뜻이 같고 근거를 찾을 수 있으면 인정한다.</p></li></ol><h4>PASS 조건</h4><ul><li><p>조원 전원이 한 사례를 설명한 기록이 있다.</p></li><li><p>설명에 보고 내용·모르는 내용·보류 이유·확인 질문이 모두 있다.</p></li><li><p>다른 의견을 카드 문장과 비교했거나 실제 차이가 없었다고 기록했다.</p></li><li><p>조별 수정 이유 또는 남은 질문을 적었다.</p></li></ul><h4>FAIL에 해당하는 예</h4><ul><li><p>조장 한 명만 설명했고 나머지 조원의 이해 여부를 알 수 없다.</p></li><li><p>“다 같이 동의했다”는 문장만 있고 개인 설명 내용이 없다.</p></li></ul><h4>이해 여부를 확인할 질문</h4><p>C5에서 민간 직원 출입 보고가 있다는 사실만으로 무엇을 확정할 수 있는가?</p><h4>인정할 수 있는 답의 핵심</h4><p>그런 출입 보고가 있다는 범위만 말하며 창고의 전체 이용 상태나 적 지휘소 여부는 확인되지 않았다고 답한다.</p><h4>보완과 재평가</h4><p>설명하지 않은 사람에게 다시 말할 기회를 주고 빠진 네 내용 중 무엇을 보완했는지 기록한다.</p><p>수정 전 결과와 재평가 결과를 구분해 남긴다. 실제 하지 않은 보고·채점·시험을 나중에 했던 것처럼 채우지 않는다.</p>', 'attachments': []}], 'deliverables': ['docs/ops07_doc_framework.md (필수) — 다섯 사례의 관찰·추측·질문 표, 여덟 장 목차, 전원 설명 기록'], 'techStack': ['문서 편집기(Markdown · docs/ops07_doc_framework.md) 또는 인쇄 양식', '관찰·주장·미확인 분리 표 및 5대 위험 신호 점검표', '8장 목차·4부록(점검표·기록 서식·절차도·수행 평가표) 구성표', 'C3·C5 1분 구술 설명 및 조별 판단 차이 기록지'], 'evaluationKeys': ['<h4>평가 질문</h4><p>다섯 사례에서 관찰·추측·미확인을 나누고 질문을 작성했는가</p><h4>평가자가 확인할 위치</h4><p>docs/ops07_doc_framework.md의 STEP 01, STEP 02, STEP 03과 그 단계의 개인 기록을 확인한다.</p><h4>PASS를 주는 방법</h4><p>아래 단계별 항목을 모두 충족해야 이 평가 질문에 PASS를 준다. 항목 옆에 확인한 표·줄·회차를 적는다. 한 항목이라도 없거나 틀리면 FAIL 사유에 해당 단계와 문장을 적는다. 학습자가 쓴 말이 예시와 똑같을 필요는 없지만 확인할 내용은 모두 있어야 한다.</p><h4>STEP 01 다섯 사례를 읽고 관찰 내용과 추측을 나누기</h4><p>찾을 결과: C1~C5가 모두 들어간 개인 표 1개. 각 카드마다 네 항목을 채운다.</p><ul><li><p>C1~C5 다섯 줄이 모두 있다.</p></li><li><p>차량·인원·물체를 보았다는 내용과 적·아군·민간이라는 판단을 다른 칸에 적었다.</p></li><li><p>확인되지 않은 내용은 “자료에 없음” 또는 “확인되지 않음”으로 표시했다.</p></li><li><p>카드에 없는 소속·회신·시각·사후 결말을 추가하지 않았다.</p></li></ul><h4>STEP 02 사례별 위험 신호와 놓쳤을 때의 피해를 적기</h4><p>찾을 결과: 다섯 카드마다 위험 신호 1개 이상을 적은 표와 다섯 분류를 점검한 기록.</p><ul><li><p>다섯 카드에 위험 신호가 적어도 하나씩 있다.</p></li><li><p>각 신호 옆에 그 신호를 찾은 카드 번호와 문장이 있다.</p></li><li><p>피해 가능성을 해당 신호와 연결해 설명했다.</p></li><li><p>시간 압박 등 제공되지 않은 상황을 실제 있었던 일처럼 추가하지 않았다.</p></li></ul><h4>STEP 03 모르는 내용을 확인할 질문과 담당자를 정하기</h4><p>찾을 결과: C1~C5 질문표 1개. 카드별 질문 1개 이상, C3는 서로 다른 질문 2개 이상.</p><ul><li><p>다섯 카드 모두 질문과 담당 역할이 있다.</p></li><li><p>질문에 대상 또는 장소와 확인할 내용이 드러난다.</p></li><li><p>C3의 장소 표기와 소속 확인을 별도로 적었다.</p></li><li><p>없는 회신을 만들지 않고 답을 기다리는 항목을 표시했다.</p></li></ul><h4>이런 결과는 통과시키지 않는다</h4><ul><li><p>C1의 관찰 내용에 “적 보급 차량이 확인됐다”라고 적었다.</p></li><li><p>위험 신호 칸을 전부 “정보 부족”으로만 채워 무엇이 빠졌는지 알 수 없다.</p></li><li><p>모든 카드의 질문이 “다시 확인해 주세요”이고 무엇을 확인할지 없다.</p></li></ul><h4>평가 의견 작성 방법</h4><p>“근거 부족”이라고만 쓰지 말고 단계 번호, 충족하지 못한 조건, 제출물에서 실제로 빠지거나 잘못된 부분, 다시 할 일을 적는다. 예를 들어 STEP 01의 첫 조건은 “C1~C5 다섯 줄이 모두 있다.”이다. 이 조건에 못 미쳤다면 어떤 줄·항목·참여자 기록이 빠졌는지 지목한다.</p><h4>다시 평가할 때</h4><p>틀리거나 빠진 내용만 구체적으로 알려 준다. 문서 작성 오류는 원자료와 수정본을 대조한다. 실제 수행 오류는 해당 행동을 다시 하게 하고 새 회차 기록으로 확인한다. 최초 결과는 지우지 않는다.</p>', '<h4>평가 질문</h4><p>기준서 목차를 만들고 모든 조원이 사례를 설명했는가</p><h4>평가자가 확인할 위치</h4><p>docs/ops07_doc_framework.md의 STEP 04, STEP 05과 그 단계의 개인 기록을 확인한다.</p><h4>PASS를 주는 방법</h4><p>아래 단계별 항목을 모두 충족해야 이 평가 질문에 PASS를 준다. 항목 옆에 확인한 표·줄·회차를 적는다. 한 항목이라도 없거나 틀리면 FAIL 사유에 해당 단계와 문장을 적는다. 학습자가 쓴 말이 예시와 똑같을 필요는 없지만 확인할 내용은 모두 있어야 한다.</p><h4>STEP 04 최종 기준서의 여덟 장에 들어갈 내용을 정하기</h4><p>찾을 결과: 여덟 장 목차표, 네 종류의 부록 목록, 먼저 읽을 장을 고른 이유.</p><ul><li><p>여덟 장이 빠짐없이 있다.</p></li><li><p>각 장마다 할 일과 참고 미션·자료 이름이 있다.</p></li><li><p>점검표·기록 서식·절차도·수행 평가표의 네 부록이 있다.</p></li><li><p>먼저 읽을 장을 선택한 이유가 실제 확인·보고 업무와 관련된다.</p></li></ul><h4>STEP 05 사례 한 건을 설명하고 조원과 다른 판단을 비교하기</h4><p>찾을 결과: 조원 전원의 개인 설명 기록과 조별 비교 기록 1개. 실제 의견 차이가 없었다면 없었다고 적는다.</p><ul><li><p>조원 전원이 한 사례를 설명한 기록이 있다.</p></li><li><p>설명에 보고 내용·모르는 내용·보류 이유·확인 질문이 모두 있다.</p></li><li><p>다른 의견을 카드 문장과 비교했거나 실제 차이가 없었다고 기록했다.</p></li><li><p>조별 수정 이유 또는 남은 질문을 적었다.</p></li></ul><h4>이런 결과는 통과시키지 않는다</h4><ul><li><p>장 제목만 있고 각 장에서 무엇을 찾아야 하는지 없다.</p></li><li><p>조장 한 명만 설명했고 나머지 조원의 이해 여부를 알 수 없다.</p></li></ul><h4>평가 의견 작성 방법</h4><p>“근거 부족”이라고만 쓰지 말고 단계 번호, 충족하지 못한 조건, 제출물에서 실제로 빠지거나 잘못된 부분, 다시 할 일을 적는다. 예를 들어 STEP 04의 첫 조건은 “여덟 장이 빠짐없이 있다.”이다. 이 조건에 못 미쳤다면 어떤 줄·항목·참여자 기록이 빠졌는지 지목한다.</p><h4>다시 평가할 때</h4><p>틀리거나 빠진 내용만 구체적으로 알려 준다. 문서 작성 오류는 원자료와 수정본을 대조한다. 실제 수행 오류는 해당 행동을 다시 하게 하고 새 회차 기록으로 확인한다. 최초 결과는 지우지 않는다.</p>'], 'evaluationTypes': ['REQUIRED', 'REQUIRED'], 'evaluationText': '<h4>미션 전체 판정</h4><p>이 미션은 아래 두 평가 질문이 모두 PASS여야 통과다. 조별 문서와 개인의 실제 수행을 나눠 확인한다. 수행하지 않았거나 기록이 없어 확인할 수 없는 사람은 미평가로 남기고 다시 실시한다. 최종 판정에서는 미평가를 PASS로 바꾸지 않는다.</p><h4>평가 질문</h4><p>다섯 사례에서 관찰·추측·미확인을 나누고 질문을 작성했는가</p><h4>평가자가 확인할 위치</h4><p>docs/ops07_doc_framework.md의 STEP 01, STEP 02, STEP 03과 그 단계의 개인 기록을 확인한다.</p><h4>PASS를 주는 방법</h4><p>아래 단계별 항목을 모두 충족해야 이 평가 질문에 PASS를 준다. 항목 옆에 확인한 표·줄·회차를 적는다. 한 항목이라도 없거나 틀리면 FAIL 사유에 해당 단계와 문장을 적는다. 학습자가 쓴 말이 예시와 똑같을 필요는 없지만 확인할 내용은 모두 있어야 한다.</p><h4>STEP 01 다섯 사례를 읽고 관찰 내용과 추측을 나누기</h4><p>찾을 결과: C1~C5가 모두 들어간 개인 표 1개. 각 카드마다 네 항목을 채운다.</p><ul><li><p>C1~C5 다섯 줄이 모두 있다.</p></li><li><p>차량·인원·물체를 보았다는 내용과 적·아군·민간이라는 판단을 다른 칸에 적었다.</p></li><li><p>확인되지 않은 내용은 “자료에 없음” 또는 “확인되지 않음”으로 표시했다.</p></li><li><p>카드에 없는 소속·회신·시각·사후 결말을 추가하지 않았다.</p></li></ul><h4>STEP 02 사례별 위험 신호와 놓쳤을 때의 피해를 적기</h4><p>찾을 결과: 다섯 카드마다 위험 신호 1개 이상을 적은 표와 다섯 분류를 점검한 기록.</p><ul><li><p>다섯 카드에 위험 신호가 적어도 하나씩 있다.</p></li><li><p>각 신호 옆에 그 신호를 찾은 카드 번호와 문장이 있다.</p></li><li><p>피해 가능성을 해당 신호와 연결해 설명했다.</p></li><li><p>시간 압박 등 제공되지 않은 상황을 실제 있었던 일처럼 추가하지 않았다.</p></li></ul><h4>STEP 03 모르는 내용을 확인할 질문과 담당자를 정하기</h4><p>찾을 결과: C1~C5 질문표 1개. 카드별 질문 1개 이상, C3는 서로 다른 질문 2개 이상.</p><ul><li><p>다섯 카드 모두 질문과 담당 역할이 있다.</p></li><li><p>질문에 대상 또는 장소와 확인할 내용이 드러난다.</p></li><li><p>C3의 장소 표기와 소속 확인을 별도로 적었다.</p></li><li><p>없는 회신을 만들지 않고 답을 기다리는 항목을 표시했다.</p></li></ul><h4>이런 결과는 통과시키지 않는다</h4><ul><li><p>C1의 관찰 내용에 “적 보급 차량이 확인됐다”라고 적었다.</p></li><li><p>위험 신호 칸을 전부 “정보 부족”으로만 채워 무엇이 빠졌는지 알 수 없다.</p></li><li><p>모든 카드의 질문이 “다시 확인해 주세요”이고 무엇을 확인할지 없다.</p></li></ul><h4>평가 의견 작성 방법</h4><p>“근거 부족”이라고만 쓰지 말고 단계 번호, 충족하지 못한 조건, 제출물에서 실제로 빠지거나 잘못된 부분, 다시 할 일을 적는다. 예를 들어 STEP 01의 첫 조건은 “C1~C5 다섯 줄이 모두 있다.”이다. 이 조건에 못 미쳤다면 어떤 줄·항목·참여자 기록이 빠졌는지 지목한다.</p><h4>다시 평가할 때</h4><p>틀리거나 빠진 내용만 구체적으로 알려 준다. 문서 작성 오류는 원자료와 수정본을 대조한다. 실제 수행 오류는 해당 행동을 다시 하게 하고 새 회차 기록으로 확인한다. 최초 결과는 지우지 않는다.</p><h4>평가 질문</h4><p>기준서 목차를 만들고 모든 조원이 사례를 설명했는가</p><h4>평가자가 확인할 위치</h4><p>docs/ops07_doc_framework.md의 STEP 04, STEP 05과 그 단계의 개인 기록을 확인한다.</p><h4>PASS를 주는 방법</h4><p>아래 단계별 항목을 모두 충족해야 이 평가 질문에 PASS를 준다. 항목 옆에 확인한 표·줄·회차를 적는다. 한 항목이라도 없거나 틀리면 FAIL 사유에 해당 단계와 문장을 적는다. 학습자가 쓴 말이 예시와 똑같을 필요는 없지만 확인할 내용은 모두 있어야 한다.</p><h4>STEP 04 최종 기준서의 여덟 장에 들어갈 내용을 정하기</h4><p>찾을 결과: 여덟 장 목차표, 네 종류의 부록 목록, 먼저 읽을 장을 고른 이유.</p><ul><li><p>여덟 장이 빠짐없이 있다.</p></li><li><p>각 장마다 할 일과 참고 미션·자료 이름이 있다.</p></li><li><p>점검표·기록 서식·절차도·수행 평가표의 네 부록이 있다.</p></li><li><p>먼저 읽을 장을 선택한 이유가 실제 확인·보고 업무와 관련된다.</p></li></ul><h4>STEP 05 사례 한 건을 설명하고 조원과 다른 판단을 비교하기</h4><p>찾을 결과: 조원 전원의 개인 설명 기록과 조별 비교 기록 1개. 실제 의견 차이가 없었다면 없었다고 적는다.</p><ul><li><p>조원 전원이 한 사례를 설명한 기록이 있다.</p></li><li><p>설명에 보고 내용·모르는 내용·보류 이유·확인 질문이 모두 있다.</p></li><li><p>다른 의견을 카드 문장과 비교했거나 실제 차이가 없었다고 기록했다.</p></li><li><p>조별 수정 이유 또는 남은 질문을 적었다.</p></li></ul><h4>이런 결과는 통과시키지 않는다</h4><ul><li><p>장 제목만 있고 각 장에서 무엇을 찾아야 하는지 없다.</p></li><li><p>조장 한 명만 설명했고 나머지 조원의 이해 여부를 알 수 없다.</p></li></ul><h4>평가 의견 작성 방법</h4><p>“근거 부족”이라고만 쓰지 말고 단계 번호, 충족하지 못한 조건, 제출물에서 실제로 빠지거나 잘못된 부분, 다시 할 일을 적는다. 예를 들어 STEP 04의 첫 조건은 “여덟 장이 빠짐없이 있다.”이다. 이 조건에 못 미쳤다면 어떤 줄·항목·참여자 기록이 빠졌는지 지목한다.</p><h4>다시 평가할 때</h4><p>틀리거나 빠진 내용만 구체적으로 알려 준다. 문서 작성 오류는 원자료와 수정본을 대조한다. 실제 수행 오류는 해당 행동을 다시 하게 하고 새 회차 기록으로 확인한다. 최초 결과는 지우지 않는다.</p>', 'aiGuidelines': {'allow': ['C1~C5 사례 표에서 관찰 내용·주장/추측·미확인 항목 구분이 모호한 문장 점검받기', '5대 위험 신호 분류(대상 확인·주변 민간 상황·아군 위치·정보 전달·시간 압박)별 누락 점검 및 8장 목차·4부록 매핑 정리 도움받기', 'C3의 장소 표기(다-4·다-1)와 소속 확인 질문을 분리한 표 구조 다듬기'], 'rules': ['AI 사용은 필수가 아니며, 모든 과제는 문서 편집기(docs/ops07_doc_framework.md)나 인쇄 양식으로 수행할 수 있다.', 'C1~C5 원문에 없는 소속·회신·시각·사후 결말이나 시간 압박 상황을 추가하지 않고 원문과 직접 대조한다.', 'STEP 01 개인 표 작성 및 STEP 05 C3·C5 1분 구술 설명·조원 판단 비교는 학습자가 직접 수행한다.'], 'forbid': ['실제 부대·위치·장비 제원·작전 정보·교전수칙 원문·개인정보 입력하기', 'C1 화물차나 C2 열영상 4명의 소속을 AI에게 추론·판정하게 하기', '실제 하지 않은 구술 설명이나 조별 토의 비교 기록을 AI 문장으로 대신 채우기']}}, {'id': 'OPS07_m02', 'sequence': 2, 'title': '출처와 정보의 공백을 보류 조건으로 바꾸기', 'durationHours': 2, 'overview': '<p>처음 출처와 전달 채널을 구별한다. 보류 조건 열 가지와 질문·담당을 작성한다.</p><p>아래 다섯 단계에는 필요한 자료·작업 순서·예시·제출 내용과 평가 조건이 들어 있다. 각 단계의 자료를 읽은 뒤 순서대로 작성한다.</p>', 'objectives': '<h4>이 미션을 마치면 할 수 있어야 하는 일</h4><ul><li><p>처음 출처와 전달 채널을 구별한다.</p></li><li><p>보류 조건 열 가지와 질문·담당을 작성한다.</p></li><li><p>개인 판정과 토의 결과를 따로 남기고 미답 질문을 인계한다.</p></li></ul>', 'scenario': '<h4>학습자의 역할</h4><p>학습자는 민간인과 아군에게 불필요한 피해가 생기지 않도록 보고를 읽고, 모르는 내용을 질문하고, 다음 담당자에게 기록을 넘기는 역할을 연습한다. 사람의 소속을 맞히거나 실제 타격을 결정하는 과제가 아니다.</p><h4>이번 미션의 과제</h4><ul><li><p>처음 출처와 전달 채널을 구별한다.</p></li><li><p>보류 조건 열 가지와 질문·담당을 작성한다.</p></li><li><p>개인 판정과 토의 결과를 따로 남기고 미답 질문을 인계한다.</p></li></ul><h4>자료와 역할의 범위</h4><p>사례·시각·장소·역할은 모두 교육용이다. 실제 부대 자료를 쓰지 않는다. 자료에 답이 없으면 자료에 없다고 적는다. 실제 행동 권한은 이 수업이 정하지 않는다.</p><h4>개인 작업과 조별 작업</h4><p>개인 판단을 먼저 요청한 단계에서는 토의 전에 자신의 답을 남긴다. 그 뒤 조별 답과 수정 이유를 별도로 쓴다. 보고·인계·시험 단계에서는 실제 한 말과 행동을 기록한다.</p>', 'attachments': [], 'materials': ['/uploads/OPS07/OPS07_m02_student.ipynb', '/uploads/OPS07/OPS07_m02_workbook.ipynb'], 'materialText': '<h4>실습 노트북 안내 (선택 보조 도구)</h4><p><strong>학생 실습용</strong>: <code>OPS07_m02_student.ipynb</code> (직접 ANSWER 표와 note를 작성합니다.)<br><strong>교수자·해설용</strong>: <code>OPS07_m02_workbook.ipynb</code> (예시 풀이와 해설 note가 포함되어 있습니다.)<br><em>※ 본 과제는 문서 편집기나 인쇄 양식으로도 수행할 수 있으며, 노트북은 사례·질문·기록 양식을 표로 정리하는 선택 보조 도구입니다.</em></p><h4>자료를 읽는 순서</h4>\n\n<ol>\n<li><p>STEP 1에서는 R1~R5의 출처·관측 시각·확인 수단을 살펴봅니다. 자료에 없는 정보는 없다고 적습니다.</p></li>\n<li><p>STEP 2에서는 아래 여섯 가지 확인 표시의 뜻을 읽습니다. STEP 3~4에서는 보고 내용과 모르는 내용을 근거로 조건과 개인 판단을 작성합니다.</p></li>\n<li><p>STEP 5에서는 R2와 R6를 함께 읽고 최초 제보와 재전달을 구별합니다.</p></li>\n</ol>\n\n<h4>자료 1. 출처와 정보의 공백을 살펴볼 보고</h4>\n\n<h4>R1 · 관측반</h4>\n\n<p>육안과 영상에서 같은 시각 같은 건물에 인원 3명이 진입한 장면이 확인되었다.</p>\n\n<p>무장 인원이라는 표현은 보고자의 해석이다.</p>\n\n<p>두 수단의 원자료 생성자가 독립적인지는 미제공이다.</p>\n\n<h4>R2 · 주민 제보 1회</h4>\n\n<p>창고에 적 장비가 있다는 주장.</p>\n\n<p>제보자 확인, 직접 관측 여부, 관측 시각은 미제공.</p>\n\n<h4>R3 · 영상 1개</h4>\n\n<p>야간에 차량 2대가 학교 방향으로 이동하는 장면.</p>\n\n<p>중간 3분이 끊겼으며 공백 동안 동일 차량이었는지 미확인.</p>\n\n<h4>R4 · 인접 팀 통보</h4>\n\n<p>2시간 전 특정 구역에 적 관측반이 있었다는 보고.</p>\n\n<p>현재 위치·원 관측 근거는 미제공.</p>\n\n<h4>R5 · 원출처 불명 게시물</h4>\n\n<p>병력 집결이라는 설명과 사진 1장.</p>\n\n<p>촬영 시각·장소·촬영자·재게시 경로 미확인.</p>\n\n<h4>자료 2. 이 수업에서 쓰는 여섯 가지 확인 표시</h4>\n\n<p>아래 표시는 이 수업에서만 사용합니다. 실제 군의 정보 등급이 아닙니다.</p>\n\n<p><strong>1 확인됨:</strong> 지금 평가하는 내용 자체를 자료로 확인했고, 그 내용에 영향을 주는 모순이 없음. 다른 내용까지 확인됐다는 뜻은 아님.</p>\n\n<p><strong>2 거의 확실:</strong> 구체적인 뒷받침 자료가 있으나 질문의 일부를 아직 확인하지 못함. 빠진 부분을 함께 적음.</p>\n\n<p><strong>3 가능성 있음:</strong> 관련 단서는 있지만 그 설명을 확인할 근거가 충분하지 않음.</p>\n\n<p><strong>4 의심:</strong> 같은 대상·시각의 자료들이 서로 맞지 않아 기존 설명을 그대로 믿기 어려움.</p>\n\n<p><strong>5 가능성 낮음:</strong> 평가하는 설명과 반대되는 구체적인 자료가 있음. 단순히 자료가 없다는 이유로 쓰지 않음.</p>\n\n<p><strong>6 판단 불가:</strong> 누가 언제 무엇을 보았는지 등 기본 정보가 없어 비교 자체가 어려움.</p>\n\n<h4>자료 3. 같은 제보가 다시 전달된 경우 — R6</h4>\n\n<p>같은 주민의 R2 제보를 채널 A와 채널 B가 재전달했습니다. 새 관측 자료는 없습니다.</p>\n\n<p>원출처는 처음 관찰하거나 제보한 사람 또는 자료를 말합니다. 같은 제보를 여러 채널이 옮겼다면 최초 제보와 전달 횟수를 나누어 적습니다.</p>\n\n<h4>자료를 사용할 때 지킬 점</h4>\n\n<p>전달 채널이 여러 개여도 같은 제보를 옮겼다면 새 관측이 늘어난 것이 아닙니다.</p>\n\n<p>확인 표시는 평가하는 내용에만 적용하며, 다른 내용까지 확인됐다는 뜻으로 쓰지 않습니다.</p>\n\n<p>모든 사례·시각·장소·역할은 교육용입니다. 필요한 자료와 구체적인 작업 순서는 각 단계 설명에도 들어 있습니다.</p>\n', 'guideSteps': [{'step': 1, 'problem': '보고의 내용과 최초 출처를 따로 적기', 'activity': '<h4>이번 단계에서 하는 일</h4><p>보고가 여러 번 들어와도 같은 제보를 옮긴 것일 수 있다. 보고에서 확인하려는 내용과 처음 관찰한 출처를 나눠 적는다.</p><h4>먼저 읽을 자료</h4>\n\n<h4>사례 자료</h4>\n\n<p><strong>R1 · 관측반</strong></p>\n\n<p>육안과 영상에서 같은 시각 같은 건물에 인원 3명이 진입한 장면이 확인되었다.</p>\n\n<p>무장 인원이라는 표현은 보고자의 해석이다.</p>\n\n<p>두 수단의 원자료 생성자가 독립적인지는 미제공이다.</p>\n\n<p><strong>R2 · 주민 제보 1회</strong></p>\n\n<p>창고에 적 장비가 있다는 주장.</p>\n\n<p>제보자 확인, 직접 관측 여부, 관측 시각은 미제공.</p>\n\n<p><strong>R3 · 영상 1개</strong></p>\n\n<p>야간에 차량 2대가 학교 방향으로 이동하는 장면.</p>\n\n<p>중간 3분이 끊겼으며 공백 동안 동일 차량이었는지 미확인.</p>\n\n<p><strong>R4 · 인접 팀 통보</strong></p>\n\n<p>2시간 전 특정 구역에 적 관측반이 있었다는 보고.</p>\n\n<p>현재 위치·원 관측 근거는 미제공.</p>\n\n<p><strong>R5 · 원출처 불명 게시물</strong></p>\n\n<p>병력 집결이라는 설명과 사진 1장.</p>\n\n<p>촬영 시각·장소·촬영자·재게시 경로 미확인.</p>\n\n<p>원출처는 처음 관찰하거나 제보한 사람·자료다.</p>\n\n<p>전달 채널이 셋이어도 같은 제보를 옮겼다면 원출처는 한 건이다.</p>\n\n<p>아래 단계 이름은 이 수업에서만 쓰는 표시이며 실제 군의 정보 등급이 아니다.</p>\n\n<h4>작업 순서</h4><ol><li><p>R1~R5를 읽고 보고별로 확인하려는 핵심 내용을 한 문장으로 적는다.</p></li><li><p>인원 수·차량 수처럼 본 장면과 무장·소속·의도에 대한 설명을 구분한다.</p></li><li><p>처음 관찰한 사람이나 자료, 관측 시각, 확인 수단을 찾는다. 적혀 있지 않으면 “자료에 없음”이라고 쓴다.</p></li><li><p>다른 보고와 같은 원자료를 옮긴 것인지 알 수 있는지 적는다. 알 수 없으면 “별도 관측인지 모름”이라고 쓴다.</p></li><li><p>조원과 표를 비교하되 전달 횟수를 독립된 관찰 수로 바꾸지 않는다.</p></li></ol><h4>작성 예시</h4><p>다음은 쓰는 방식을 보여 주는 예다. 자기 판단이나 실제 수행·측정 결과가 필요한 칸에 그대로 복사하지 않는다.</p><p>R1 / 장면 “인원 3명이 건물에 들어감” / 추가 주장 “무장 인원임” / 수단 “육안·영상” / 별도 관측 여부 “자료에 없음”. 두 수단에서 장면이 보였다는 것과 무장 여부를 확인했다는 것은 다르다.</p><h4>작성할 표 또는 기록 항목</h4><ul><li><p>보고 번호</p></li><li><p>보고의 핵심 내용</p></li><li><p>직접 본 장면</p></li><li><p>추가 주장</p></li><li><p>최초 출처</p></li><li><p>관측 시각</p></li><li><p>확인 수단</p></li><li><p>별도 관측인지</p></li></ul><h4>제출할 내용과 끝내는 기준</h4><p>R1~R5 다섯 줄의 출처·내용 정리표.</p><p>미션 제출 문서 framework/ops07_source_reliability_matrix.md의 STEP 01 아래에 작성한다. 개인 기록과 조별 결과를 구분하고 실제 이름 대신 학습자 코드를 써도 된다. 표의 모양이 달라도 필요한 내용을 찾을 수 있으면 된다.</p><h4>마치기 전 확인</h4><ul><li><p>다섯 보고에 핵심 내용·출처·시각·수단을 적었다.</p></li><li><p>R1의 인원 진입 장면과 무장 주장을 구분했다.</p></li><li><p>R3의 영상 중단, R4의 2시간 경과, R5의 시각·장소 미확인을 남겼다.</p></li><li><p>출처와 관측 시각이 없으면 임의로 채우지 않았다.</p></li></ul><h4>배정 시간</h4><p>20분</p>', 'evalText': '<h4>평가할 자료와 위치</h4><p>framework/ops07_source_reliability_matrix.md의 STEP 01 결과와 해당 개인 수행 기록을 확인한다. 이 단계에서 요청한 제출 내용은 다음과 같다: R1~R5 다섯 줄의 출처·내용 정리표.</p><h4>평가 순서</h4><ol><li><p>아래 통과 조건을 하나씩 읽고 제출물에서 확인한 위치를 옆에 적는다.</p></li><li><p>실제 설명·보고·인계가 필요한 단계는 수행 기록이나 직접 관찰로 확인한다. 서식만 있으면 수행한 것으로 보지 않는다.</p></li><li><p>통과 조건을 모두 충족하면 이 단계는 PASS다. 하나라도 빠지거나 반대로 적었으면 FAIL로 두고 빠진 항목을 구체적으로 적는다. 표현이나 표 모양이 예시와 달라도 뜻이 같고 근거를 찾을 수 있으면 인정한다.</p></li></ol><h4>PASS 조건</h4><ul><li><p>다섯 보고에 핵심 내용·출처·시각·수단을 적었다.</p></li><li><p>R1의 인원 진입 장면과 무장 주장을 구분했다.</p></li><li><p>R3의 영상 중단, R4의 2시간 경과, R5의 시각·장소 미확인을 남겼다.</p></li><li><p>출처와 관측 시각이 없으면 임의로 채우지 않았다.</p></li></ul><h4>FAIL에 해당하는 예</h4><ul><li><p>R1을 “무장 인원으로 확인됨”이라고만 요약했다.</p></li><li><p>R5가 게시됐다는 이유로 촬영 시각과 장소까지 확인됐다고 적었다.</p></li></ul><h4>이해 여부를 확인할 질문</h4><p>다른 채널에서 같은 제보를 받으면 새 관측이 늘어난 것인가?</p><h4>인정할 수 있는 답의 핵심</h4><p>같은 내용을 옮긴 것이라면 늘어난 것은 전달 횟수이며 새 독립 관측은 아니라고 답한다.</p><h4>보완과 재평가</h4><p>각 보고의 원문을 다시 보며 출처·시각·수단 칸을 채우고 모르는 부분은 모른다고 고친다.</p><p>수정 전 결과와 재평가 결과를 구분해 남긴다. 실제 하지 않은 보고·채점·시험을 나중에 했던 것처럼 채우지 않는다.</p>', 'attachments': []}, {'step': 2, 'problem': '정보 확인 표시 여섯 가지를 이해하고 예를 적기', 'activity': '<h4>이번 단계에서 하는 일</h4><p>확인된 내용과 근거가 없는 내용을 같은 표시로 적으면 다음 사람이 구분하기 어렵다. 이 수업의 여섯 표시가 무엇을 뜻하는지 풀어서 설명한다.</p><h4>먼저 읽을 자료</h4>\n\n<h4>이 수업에서 쓰는 확인 표시</h4>\n\n<ul>\n<li>\n<p><strong>1 확인됨:</strong> 지금 평가하는 내용 자체를 자료로 확인했고, 그 내용에 영향을 주는 모순이 없음. 다른 내용까지 확인됐다는 뜻은 아님.</p>\n</li>\n\n<li>\n<p><strong>2 거의 확실:</strong> 구체적인 뒷받침 자료가 있으나 질문의 일부를 아직 확인하지 못함. 빠진 부분을 함께 적음.</p>\n</li>\n\n<li>\n<p><strong>3 가능성 있음:</strong> 관련 단서는 있지만 그 설명을 확인할 근거가 충분하지 않음.</p>\n</li>\n\n<li>\n<p><strong>4 의심:</strong> 같은 대상·시각의 자료들이 서로 맞지 않아 기존 설명을 그대로 믿기 어려움.</p>\n</li>\n\n<li>\n<p><strong>5 가능성 낮음:</strong> 평가하는 설명과 반대되는 구체적인 자료가 있음. 단순히 자료가 없다는 이유로 쓰지 않음.</p>\n</li>\n\n<li>\n<p><strong>6 판단 불가:</strong> 누가 언제 무엇을 보았는지 등 기본 정보가 없어 비교 자체가 어려움.</p>\n</li>\n</ul>\n\n<p>하나의 보고에 여러 내용이 있으면 내용별로 표시한다.</p>\n\n<p>예를 들어 인원 수를 확인했어도 그 인원의 소속까지 확인한 것은 아니다.</p>\n\n<h4>작업 순서</h4><ol><li><p>표의 여섯 이름과 뜻을 문서에 옮긴다.</p></li><li><p>각 표시 옆에 “이 표시를 쓰려면 필요한 정보”를 자신의 말로 적는다.</p></li><li><p>각 표시마다 가상 예를 한 문장씩 만든다. 예는 반드시 “직접 만든 예”라고 적어 제공된 보고와 구분한다.</p></li><li><p>“자료가 없음”과 “그 설명에 반대되는 자료가 있음”을 비교한다. 앞은 6번을 검토할 상황이며 뒤는 5번을 검토할 상황이다.</p></li><li><p>어떤 표시도 실제 행동 허가를 뜻하지 않는다는 문장을 표 아래에 적는다.</p></li></ol><h4>작성 예시</h4><p>다음은 쓰는 방식을 보여 주는 예다. 자기 판단이나 실제 수행·측정 결과가 필요한 칸에 그대로 복사하지 않는다.</p><p>직접 만든 예: “1 확인됨 — 영상에 차량 2대가 보인다는 장면은 확인했다. 차량의 소속은 확인하지 않았다.”</p><h4>작성할 표 또는 기록 항목</h4><ul><li><p>번호·이름</p></li><li><p>쉽게 풀어 쓴 뜻</p></li><li><p>이 표시를 쓸 때 필요한 정보</p></li><li><p>직접 만든 가상 예</p></li></ul><h4>제출할 내용과 끝내는 기준</h4><p>여섯 줄의 설명표와 자료 없음·반대 자료 있음의 차이 설명.</p><p>미션 제출 문서 framework/ops07_source_reliability_matrix.md의 STEP 02 아래에 작성한다. 개인 기록과 조별 결과를 구분하고 실제 이름 대신 학습자 코드를 써도 된다. 표의 모양이 달라도 필요한 내용을 찾을 수 있으면 된다.</p><h4>마치기 전 확인</h4><ul><li><p>여섯 표시가 모두 있고 각 표시의 뜻과 예가 있다.</p></li><li><p>확인됨의 적용 범위를 지금 확인한 내용으로 제한했다.</p></li><li><p>판단할 자료가 없는 경우를 가능성 낮음으로 처리하지 않았다.</p></li><li><p>실제 정보 등급이나 행동 허가가 아니라 수업용 표시라고 밝혔다.</p></li></ul><h4>배정 시간</h4><p>25분</p>', 'evalText': '<h4>평가할 자료와 위치</h4><p>framework/ops07_source_reliability_matrix.md의 STEP 02 결과와 해당 개인 수행 기록을 확인한다. 이 단계에서 요청한 제출 내용은 다음과 같다: 여섯 줄의 설명표와 자료 없음·반대 자료 있음의 차이 설명.</p><h4>평가 순서</h4><ol><li><p>아래 통과 조건을 하나씩 읽고 제출물에서 확인한 위치를 옆에 적는다.</p></li><li><p>실제 설명·보고·인계가 필요한 단계는 수행 기록이나 직접 관찰로 확인한다. 서식만 있으면 수행한 것으로 보지 않는다.</p></li><li><p>통과 조건을 모두 충족하면 이 단계는 PASS다. 하나라도 빠지거나 반대로 적었으면 FAIL로 두고 빠진 항목을 구체적으로 적는다. 표현이나 표 모양이 예시와 달라도 뜻이 같고 근거를 찾을 수 있으면 인정한다.</p></li></ol><h4>PASS 조건</h4><ul><li><p>여섯 표시가 모두 있고 각 표시의 뜻과 예가 있다.</p></li><li><p>확인됨의 적용 범위를 지금 확인한 내용으로 제한했다.</p></li><li><p>판단할 자료가 없는 경우를 가능성 낮음으로 처리하지 않았다.</p></li><li><p>실제 정보 등급이나 행동 허가가 아니라 수업용 표시라고 밝혔다.</p></li></ul><h4>FAIL에 해당하는 예</h4><ul><li><p>“모르므로 가능성 낮음”이라고 적었다.</p></li><li><p>보고에 1번을 붙인 뒤 그 안의 모든 주장까지 사실이라고 설명했다.</p></li></ul><h4>이해 여부를 확인할 질문</h4><p>자료가 없다는 것과 설명이 틀렸다는 자료가 있다는 것은 어떻게 다른가?</p><h4>인정할 수 있는 답의 핵심</h4><p>자료가 없으면 판단을 못 하는 것이며 반대 자료가 있는 경우에는 그 자료 때문에 해당 설명의 가능성을 낮게 본다고 답한다.</p><h4>보완과 재평가</h4><p>자료 없음과 반대 자료가 있는 예를 각각 다시 만들고 여섯 표시의 예를 점검한다.</p><p>수정 전 결과와 재평가 결과를 구분해 남긴다. 실제 하지 않은 보고·채점·시험을 나중에 했던 것처럼 채우지 않는다.</p>', 'attachments': []}, {'step': 3, 'problem': '보류할 조건 열 가지와 확인 질문을 작성하기', 'activity': '<h4>이번 단계에서 하는 일</h4><p>“정보 부족 시 보류”만으로는 부족한 정보와 다음 할 일을 알 수 없다. 눈으로 확인할 수 있는 조건과 실제로 물을 질문을 짝지어 쓴다.</p><h4>먼저 읽을 자료</h4>\n\n<p>보류는 이 연습에서 확인되지 않은 내용을 확인한 것처럼 처리하지 않고 검토를 멈춰 두는 상태다.</p>\n\n<p>재확인은 모르는 내용을 담당자에게 다시 묻는 일이다.</p>\n\n<p>접수는 상대가 요청을 받았다는 뜻이고, 회신은 질문에 대한 답이다.</p>\n\n<p>접수됐다고 해서 질문의 답까지 받은 것은 아니다.</p>\n\n<p>아래의 역할은 모두 교육용이며 실제 지휘권을 뜻하지 않는다.</p>\n\n<h4>살펴볼 상황</h4>\n\n<ul>\n<li>\n<p>주민·민간 차량이 보고됨</p>\n</li>\n\n<li>\n<p>건물 이용 상태를 모름</p>\n</li>\n\n<li>\n<p>현재 아군 위치를 모름</p>\n</li>\n\n<li>\n<p>장소 표기가 다름</p>\n</li>\n\n<li>\n<p>처음 제보한 출처를 모름</p>\n</li>\n\n<li>\n<p>관측 시각을 모름</p>\n</li>\n\n<li>\n<p>영상이 중간에 끊김</p>\n</li>\n\n<li>\n<p>보고가 오래됨</p>\n</li>\n\n<li>\n<p>서로 다른 설명이 함께 있음</p>\n</li>\n\n<li>\n<p>검토자에게 요청이 접수됐는지 모름</p>\n</li>\n</ul>\n\n<p>모든 역할과 조건은 수업용이다.</p>\n\n<p>실제 승인 권한이나 안전한 분·거리 기준을 새로 정하지 않는다.</p>\n\n<h4>작업 순서</h4><ol><li><p>위 상황을 참고해 서로 다른 보류 조건을 열 가지 이상 적고 번호를 붙인다.</p></li><li><p>각 조건을 “무엇이 어떠하면”이라는 문장으로 쓴다.</p></li><li><p>조건별로 보류하는 이유와 누구에게 물을지 적는다.</p></li><li><p>질문과 그 질문에 답했다고 볼 수 있는 자료를 적는다.</p></li><li><p>아직 답을 받지 못했을 때 남길 상태와 다음 확인 계획을 적는다. 실제 확인 시간을 정할 자료가 없으면 “다음 교육 점검 때 담당 확인”처럼 계획임을 밝힌다.</p></li></ol><h4>작성 예시</h4><p>다음은 쓰는 방식을 보여 주는 예다. 자기 판단이나 실제 수행·측정 결과가 필요한 칸에 그대로 복사하지 않는다.</p><p>조건 “원 보고와 수신 기록의 장소 표기가 다르면” / 이유 “다른 장소를 같은 곳으로 볼 수 있음” / 담당 “원 보고자·기록 담당자” / 질문 “원 보고 표기와 수신 표기를 대조해 주세요” / 현재 “보류 유지, 답 기다림”.</p><h4>작성할 표 또는 기록 항목</h4><ul><li><p>조건 번호</p></li><li><p>보류할 조건</p></li><li><p>이유</p></li><li><p>확인 담당</p></li><li><p>그대로 전달할 질문</p></li><li><p>필요한 답·자료</p></li><li><p>답이 없을 때 상태와 다음 확인 계획</p></li></ul><h4>제출할 내용과 끝내는 기준</h4><p>서로 다른 조건 열 가지 이상. 모든 조건에 질문·담당·답을 기다릴 때 할 일이 있어야 한다.</p><p>미션 제출 문서 framework/ops07_source_reliability_matrix.md의 STEP 03 아래에 작성한다. 개인 기록과 조별 결과를 구분하고 실제 이름 대신 학습자 코드를 써도 된다. 표의 모양이 달라도 필요한 내용을 찾을 수 있으면 된다.</p><h4>마치기 전 확인</h4><ul><li><p>서로 다른 조건이 열 가지 이상이다.</p></li><li><p>각 조건에 구체적인 상황·보류 이유·담당·질문·필요한 답이 있다.</p></li><li><p>답이 없을 때 보류와 후속 확인 계획이 남는다.</p></li><li><p>일정 시간이 지나거나 채널 수가 늘었다는 이유만으로 확인 완료로 바꾸지 않는다.</p></li></ul><h4>배정 시간</h4><p>30분</p>', 'evalText': '<h4>평가할 자료와 위치</h4><p>framework/ops07_source_reliability_matrix.md의 STEP 03 결과와 해당 개인 수행 기록을 확인한다. 이 단계에서 요청한 제출 내용은 다음과 같다: 서로 다른 조건 열 가지 이상. 모든 조건에 질문·담당·답을 기다릴 때 할 일이 있어야 한다.</p><h4>평가 순서</h4><ol><li><p>아래 통과 조건을 하나씩 읽고 제출물에서 확인한 위치를 옆에 적는다.</p></li><li><p>실제 설명·보고·인계가 필요한 단계는 수행 기록이나 직접 관찰로 확인한다. 서식만 있으면 수행한 것으로 보지 않는다.</p></li><li><p>통과 조건을 모두 충족하면 이 단계는 PASS다. 하나라도 빠지거나 반대로 적었으면 FAIL로 두고 빠진 항목을 구체적으로 적는다. 표현이나 표 모양이 예시와 달라도 뜻이 같고 근거를 찾을 수 있으면 인정한다.</p></li></ol><h4>PASS 조건</h4><ul><li><p>서로 다른 조건이 열 가지 이상이다.</p></li><li><p>각 조건에 구체적인 상황·보류 이유·담당·질문·필요한 답이 있다.</p></li><li><p>답이 없을 때 보류와 후속 확인 계획이 남는다.</p></li><li><p>일정 시간이 지나거나 채널 수가 늘었다는 이유만으로 확인 완료로 바꾸지 않는다.</p></li></ul><h4>FAIL에 해당하는 예</h4><ul><li><p>열 줄을 모두 “모호하면 보류”로 채웠다.</p></li><li><p>“두 번 보고되면 진행”처럼 출처나 내용을 확인하지 않는 조건을 만들었다.</p></li></ul><h4>이해 여부를 확인할 질문</h4><p>“재확인 필요”라는 말 뒤에 무엇을 더 적어야 다음 사람이 일할 수 있는가?</p><h4>인정할 수 있는 답의 핵심</h4><p>무엇을 누구에게 물으며 어떤 답을 기다리는지와 아직 답이 없을 때의 상태를 적어야 한다고 답한다.</p><h4>보완과 재평가</h4><p>겹치는 조건을 구체적인 상황으로 나누고 담당·질문·필요한 답이 빈 줄을 채운다.</p><p>수정 전 결과와 재평가 결과를 구분해 남긴다. 실제 하지 않은 보고·채점·시험을 나중에 했던 것처럼 채우지 않는다.</p>', 'attachments': []}, {'step': 4, 'problem': '같은 보고를 각자 판단한 뒤 차이를 비교하기', 'activity': '<h4>이번 단계에서 하는 일</h4><p>같은 보고를 읽고도 다른 뜻으로 이해한다면 표시 기준을 더 분명하게 써야 한다. 먼저 서로 상의하지 않은 결과를 남긴 뒤 비교한다.</p><h4>먼저 읽을 자료</h4>\n\n<h4>사례 자료</h4>\n\n<p><strong>R1 · 관측반</strong></p>\n\n<p>육안과 영상에서 같은 시각 같은 건물에 인원 3명이 진입한 장면이 확인되었다.</p>\n\n<p>무장 인원이라는 표현은 보고자의 해석이다.</p>\n\n<p>두 수단의 원자료 생성자가 독립적인지는 미제공이다.</p>\n\n<p><strong>R2 · 주민 제보 1회</strong></p>\n\n<p>창고에 적 장비가 있다는 주장.</p>\n\n<p>제보자 확인, 직접 관측 여부, 관측 시각은 미제공.</p>\n\n<p><strong>R3 · 영상 1개</strong></p>\n\n<p>야간에 차량 2대가 학교 방향으로 이동하는 장면.</p>\n\n<p>중간 3분이 끊겼으며 공백 동안 동일 차량이었는지 미확인.</p>\n\n<p><strong>R4 · 인접 팀 통보</strong></p>\n\n<p>2시간 전 특정 구역에 적 관측반이 있었다는 보고.</p>\n\n<p>현재 위치·원 관측 근거는 미제공.</p>\n\n<p><strong>R5 · 원출처 불명 게시물</strong></p>\n\n<p>병력 집결이라는 설명과 사진 1장.</p>\n\n<p>촬영 시각·장소·촬영자·재게시 경로 미확인.</p>\n\n<p>원출처는 처음 관찰하거나 제보한 사람·자료다.</p>\n\n<p>전달 채널이 셋이어도 같은 제보를 옮겼다면 원출처는 한 건이다.</p>\n\n<p>아래 단계 이름은 이 수업에서만 쓰는 표시이며 실제 군의 정보 등급이 아니다.</p>\n\n<h4>이 수업에서 쓰는 확인 표시</h4>\n\n<ul>\n<li>\n<p><strong>1 확인됨:</strong> 지금 평가하는 내용 자체를 자료로 확인했고, 그 내용에 영향을 주는 모순이 없음. 다른 내용까지 확인됐다는 뜻은 아님.</p>\n</li>\n\n<li>\n<p><strong>2 거의 확실:</strong> 구체적인 뒷받침 자료가 있으나 질문의 일부를 아직 확인하지 못함. 빠진 부분을 함께 적음.</p>\n</li>\n\n<li>\n<p><strong>3 가능성 있음:</strong> 관련 단서는 있지만 그 설명을 확인할 근거가 충분하지 않음.</p>\n</li>\n\n<li>\n<p><strong>4 의심:</strong> 같은 대상·시각의 자료들이 서로 맞지 않아 기존 설명을 그대로 믿기 어려움.</p>\n</li>\n\n<li>\n<p><strong>5 가능성 낮음:</strong> 평가하는 설명과 반대되는 구체적인 자료가 있음. 단순히 자료가 없다는 이유로 쓰지 않음.</p>\n</li>\n\n<li>\n<p><strong>6 판단 불가:</strong> 누가 언제 무엇을 보았는지 등 기본 정보가 없어 비교 자체가 어려움.</p>\n</li>\n</ul>\n\n<h4>작업 순서</h4><ol><li><p>토의 전에 평가할 문장을 고정한다. R1은 “인원 3명이 건물에 들어갔다”, R2는 “창고에 적 장비가 있다”, R3는 “영상 중단 동안에도 같은 두 차량이었다”, R4는 “현재 그 구역에 같은 관측반이 있다”, R5는 “사진이 설명한 시각·장소의 모습이다”로 한다.</p></li><li><p>각자 다섯 문장에 여섯 표시 중 하나와 이유를 적는다. 답을 쓰는 동안 다른 사람과 상의하지 않는다.</p></li><li><p>기록자가 학습자별 원래 답을 그대로 옮긴다. 이유가 다르더라도 번호가 전원 같으면 그 보고의 번호 일치로 센다.</p></li><li><p>전원 번호가 같은 보고 수를 5로 나누고 100을 곱한다. 예를 들어 4건이면 80%다.</p></li><li><p>번호나 이유가 다른 보고를 토의한다. 어느 표현을 다르게 읽었는지와 고칠 설명을 적고 토의 후 답은 별도 칸에 쓴다.</p></li><li><p>4건 미만이면 표시 설명을 고치고 새 개인 답을 다시 받는다. 이전 답은 지우지 않는다. 일치율이 높아도 이유가 틀렸다면 그 이유를 고친다.</p></li></ol><h4>작성 예시</h4><p>다음은 쓰는 방식을 보여 주는 예다. 자기 판단이나 실제 수행·측정 결과가 필요한 칸에 그대로 복사하지 않는다.</p><p>R2 / 학습자 A 6번·B 6번·C 3번·D 6번 → 전원 일치 아님. 토의 후 모두 6번으로 바뀌어도 최초 결과를 “전원 일치”로 고치지 않는다.</p><h4>작성할 표 또는 기록 항목</h4><ul><li><p>보고 번호·고정한 문장</p></li><li><p>각 학습자의 최초 번호와 이유</p></li><li><p>최초 전원 일치 여부</p></li><li><p>다르게 읽은 내용</p></li><li><p>수정한 설명</p></li><li><p>토의 후 답 또는 재실시 답</p></li></ul><h4>제출할 내용과 끝내는 기준</h4><p>다섯 보고의 개인 원답, 최초 일치율의 계산식, 의견 차이와 수정 기록.</p><p>미션 제출 문서 framework/ops07_source_reliability_matrix.md의 STEP 04 아래에 작성한다. 개인 기록과 조별 결과를 구분하고 실제 이름 대신 학습자 코드를 써도 된다. 표의 모양이 달라도 필요한 내용을 찾을 수 있으면 된다.</p><h4>마치기 전 확인</h4><ul><li><p>전원이 같은 다섯 문장을 판단했다.</p></li><li><p>토의 전 각자의 번호와 이유를 보존했다.</p></li><li><p>최초 전원 일치 건수÷5×100 계산이 맞다.</p></li><li><p>다른 답의 원인을 적었으며 4건 미만이면 설명 수정과 재실시 기록이 있다.</p></li><li><p>번호 일치와 내용의 타당성을 따로 확인했다.</p></li></ul><h4>배정 시간</h4><p>25분</p>', 'evalText': '<h4>평가할 자료와 위치</h4><p>framework/ops07_source_reliability_matrix.md의 STEP 04 결과와 해당 개인 수행 기록을 확인한다. 이 단계에서 요청한 제출 내용은 다음과 같다: 다섯 보고의 개인 원답, 최초 일치율의 계산식, 의견 차이와 수정 기록.</p><h4>평가 순서</h4><ol><li><p>아래 통과 조건을 하나씩 읽고 제출물에서 확인한 위치를 옆에 적는다.</p></li><li><p>실제 설명·보고·인계가 필요한 단계는 수행 기록이나 직접 관찰로 확인한다. 서식만 있으면 수행한 것으로 보지 않는다.</p></li><li><p>통과 조건을 모두 충족하면 이 단계는 PASS다. 하나라도 빠지거나 반대로 적었으면 FAIL로 두고 빠진 항목을 구체적으로 적는다. 표현이나 표 모양이 예시와 달라도 뜻이 같고 근거를 찾을 수 있으면 인정한다.</p></li></ol><h4>PASS 조건</h4><ul><li><p>전원이 같은 다섯 문장을 판단했다.</p></li><li><p>토의 전 각자의 번호와 이유를 보존했다.</p></li><li><p>최초 전원 일치 건수÷5×100 계산이 맞다.</p></li><li><p>다른 답의 원인을 적었으며 4건 미만이면 설명 수정과 재실시 기록이 있다.</p></li><li><p>번호 일치와 내용의 타당성을 따로 확인했다.</p></li></ul><h4>FAIL에 해당하는 예</h4><ul><li><p>서로 다른 내용을 판단한 뒤 번호만 비교했다.</p></li><li><p>토의 후 맞춘 번호를 최초 개인 답처럼 적었다.</p></li></ul><h4>이해 여부를 확인할 질문</h4><p>다섯 보고 중 네 보고에서 전원 번호가 같으면 몇 퍼센트인가?</p><h4>인정할 수 있는 답의 핵심</h4><p>4÷5×100=80%라고 계산하며 이는 같은 번호를 붙인 비율이지 그 판단이 모두 옳다는 뜻은 아니라고 덧붙인다.</p><h4>보완과 재평가</h4><p>비교할 문장을 다시 맞추고 개인 원답·토의 후 답을 분리한다. 계산을 다시 하고 이유도 확인한다.</p><p>수정 전 결과와 재평가 결과를 구분해 남긴다. 실제 하지 않은 보고·채점·시험을 나중에 했던 것처럼 채우지 않는다.</p>', 'attachments': []}, {'step': 5, 'problem': '같은 제보가 세 번 전달된 경우를 확인하고 인계하기', 'activity': '<h4>이번 단계에서 하는 일</h4><p>여러 채널에서 같은 말을 받으면 확인이 늘었다고 착각하기 쉽다. 처음 출처가 같은지 확인하고, 아직 답을 받지 못한 질문을 다음 담당자에게 넘긴다.</p><h4>먼저 읽을 자료</h4>\n\n<h4>사례 자료</h4>\n\n<p><strong>R2 · 주민 제보 1회</strong></p>\n\n<p>창고에 적 장비가 있다는 주장.</p>\n\n<p>제보자 확인, 직접 관측 여부, 관측 시각은 미제공.</p>\n\n<p>추가 카드 R6: R2의 같은 주민 말을 채널 A와 채널 B가 다시 전달했다.</p>\n\n<p>두 채널 모두 새로 관찰한 자료는 제시하지 않았다.</p>\n\n<p>보류는 이 연습에서 확인되지 않은 내용을 확인한 것처럼 처리하지 않고 검토를 멈춰 두는 상태다.</p>\n\n<p>재확인은 모르는 내용을 담당자에게 다시 묻는 일이다.</p>\n\n<p>접수는 상대가 요청을 받았다는 뜻이고, 회신은 질문에 대한 답이다.</p>\n\n<p>접수됐다고 해서 질문의 답까지 받은 것은 아니다.</p>\n\n<p>아래의 역할은 모두 교육용이며 실제 지휘권을 뜻하지 않는다.</p>\n\n<h4>작업 순서</h4><ol><li><p>R2와 R6의 전달 흐름을 “최초 주민 제보 → 원 보고 / 채널 A / 채널 B”처럼 글로 적는다.</p></li><li><p>전달 횟수와 새로운 관측·제보 수를 따로 적는다. 새 자료가 없는 이유를 문장으로 설명한다.</p></li><li><p>R2에서 아직 모르는 내용을 골라 담당·질문·기다리는 답을 적는다.</p></li><li><p>인계하는 사람은 그 내용을 읽어 주고 받는 사람은 자신이 맡을 질문을 다시 말한다.</p></li><li><p>누가 누구에게 어떤 질문을 넘겼는지와 아직 답을 받지 못했다는 상태를 적는다. 실제 회신이 없으므로 완료로 표시하지 않는다.</p></li></ol><h4>작성 예시</h4><p>다음은 쓰는 방식을 보여 주는 예다. 자기 판단이나 실제 수행·측정 결과가 필요한 칸에 그대로 복사하지 않는다.</p><p>전달 기록 3건 / 최초 제보 1건 / 새 관측 자료 0건. “제보자의 직접 관측 여부를 확인하는 질문은 아직 답이 없어 다음 담당자가 이어서 묻는다.”</p><h4>작성할 표 또는 기록 항목</h4><ul><li><p>자료 번호</p></li><li><p>전달 흐름</p></li><li><p>전달 건수와 최초 제보 건수</p></li><li><p>새 자료 유무</p></li><li><p>미확인 내용</p></li><li><p>담당·질문</p></li><li><p>인계한 사람·받은 사람</p></li><li><p>받은 사람의 확인 문장</p></li></ul><h4>제출할 내용과 끝내는 기준</h4><p>R2·R6 비교표와 인계 기록 1개. 각자 자신이 맡은 질문이나 인계 내용을 설명한다.</p><p>미션 제출 문서 framework/ops07_source_reliability_matrix.md의 STEP 05 아래에 작성한다. 개인 기록과 조별 결과를 구분하고 실제 이름 대신 학습자 코드를 써도 된다. 표의 모양이 달라도 필요한 내용을 찾을 수 있으면 된다.</p><h4>마치기 전 확인</h4><ul><li><p>R6를 세 개의 독립 제보로 세지 않았다.</p></li><li><p>최초 제보 1건과 새 자료 없음의 이유를 설명했다.</p></li><li><p>확인 질문·담당·아직 답이 없다는 상태가 있다.</p></li><li><p>받는 사람이 맡을 일을 다시 말한 내용이 있다.</p></li></ul><h4>배정 시간</h4><p>20분</p>', 'evalText': '<h4>평가할 자료와 위치</h4><p>framework/ops07_source_reliability_matrix.md의 STEP 05 결과와 해당 개인 수행 기록을 확인한다. 이 단계에서 요청한 제출 내용은 다음과 같다: R2·R6 비교표와 인계 기록 1개. 각자 자신이 맡은 질문이나 인계 내용을 설명한다.</p><h4>평가 순서</h4><ol><li><p>아래 통과 조건을 하나씩 읽고 제출물에서 확인한 위치를 옆에 적는다.</p></li><li><p>실제 설명·보고·인계가 필요한 단계는 수행 기록이나 직접 관찰로 확인한다. 서식만 있으면 수행한 것으로 보지 않는다.</p></li><li><p>통과 조건을 모두 충족하면 이 단계는 PASS다. 하나라도 빠지거나 반대로 적었으면 FAIL로 두고 빠진 항목을 구체적으로 적는다. 표현이나 표 모양이 예시와 달라도 뜻이 같고 근거를 찾을 수 있으면 인정한다.</p></li></ol><h4>PASS 조건</h4><ul><li><p>R6를 세 개의 독립 제보로 세지 않았다.</p></li><li><p>최초 제보 1건과 새 자료 없음의 이유를 설명했다.</p></li><li><p>확인 질문·담당·아직 답이 없다는 상태가 있다.</p></li><li><p>받는 사람이 맡을 일을 다시 말한 내용이 있다.</p></li></ul><h4>FAIL에 해당하는 예</h4><ul><li><p>“세 채널이 일치했으므로 확인 완료”라고 적었다.</p></li><li><p>“인계함”만 적고 질문과 받은 사람을 쓰지 않았다.</p></li></ul><h4>이해 여부를 확인할 질문</h4><p>채널 A와 B의 전달에서 무엇이 추가됐는가?</p><h4>인정할 수 있는 답의 핵심</h4><p>전달 경로만 늘었고 별도의 새 관측은 없었다고 답한다.</p><h4>보완과 재평가</h4><p>최초 출처를 기준으로 다시 세고 받는 사람이 맡은 질문을 말한 뒤 인계 기록을 고친다.</p><p>수정 전 결과와 재평가 결과를 구분해 남긴다. 실제 하지 않은 보고·채점·시험을 나중에 했던 것처럼 채우지 않는다.</p>', 'attachments': []}], 'deliverables': ['framework/ops07_source_reliability_matrix.md (필수) — 출처 정리표, 여섯 표시 설명, 보류 조건 열 가지, 개인 원답·비교 계산, 질문 인계'], 'techStack': ['문서 편집기(Markdown · framework/ops07_source_reliability_matrix.md) 또는 인쇄 양식', '보고 내용-최초 출처 분리표 및 6단계 확인 표시 기준표', '보류 조건 10종-질문-담당자 매핑표', '개인 원답 대조·일치율 계산 및 중복 제보(R6) 인계 기록지'], 'evaluationKeys': ['<h4>평가 질문</h4><p>출처와 확인 표시를 구분하고 개인 판정을 정직하게 비교했는가</p><h4>평가자가 확인할 위치</h4><p>framework/ops07_source_reliability_matrix.md의 STEP 01, STEP 02, STEP 04과 그 단계의 개인 기록을 확인한다.</p><h4>PASS를 주는 방법</h4><p>아래 단계별 항목을 모두 충족해야 이 평가 질문에 PASS를 준다. 항목 옆에 확인한 표·줄·회차를 적는다. 한 항목이라도 없거나 틀리면 FAIL 사유에 해당 단계와 문장을 적는다. 학습자가 쓴 말이 예시와 똑같을 필요는 없지만 확인할 내용은 모두 있어야 한다.</p><h4>STEP 01 보고의 내용과 최초 출처를 따로 적기</h4><p>찾을 결과: R1~R5 다섯 줄의 출처·내용 정리표.</p><ul><li><p>다섯 보고에 핵심 내용·출처·시각·수단을 적었다.</p></li><li><p>R1의 인원 진입 장면과 무장 주장을 구분했다.</p></li><li><p>R3의 영상 중단, R4의 2시간 경과, R5의 시각·장소 미확인을 남겼다.</p></li><li><p>출처와 관측 시각이 없으면 임의로 채우지 않았다.</p></li></ul><h4>STEP 02 정보 확인 표시 여섯 가지를 이해하고 예를 적기</h4><p>찾을 결과: 여섯 줄의 설명표와 자료 없음·반대 자료 있음의 차이 설명.</p><ul><li><p>여섯 표시가 모두 있고 각 표시의 뜻과 예가 있다.</p></li><li><p>확인됨의 적용 범위를 지금 확인한 내용으로 제한했다.</p></li><li><p>판단할 자료가 없는 경우를 가능성 낮음으로 처리하지 않았다.</p></li><li><p>실제 정보 등급이나 행동 허가가 아니라 수업용 표시라고 밝혔다.</p></li></ul><h4>STEP 04 같은 보고를 각자 판단한 뒤 차이를 비교하기</h4><p>찾을 결과: 다섯 보고의 개인 원답, 최초 일치율의 계산식, 의견 차이와 수정 기록.</p><ul><li><p>전원이 같은 다섯 문장을 판단했다.</p></li><li><p>토의 전 각자의 번호와 이유를 보존했다.</p></li><li><p>최초 전원 일치 건수÷5×100 계산이 맞다.</p></li><li><p>다른 답의 원인을 적었으며 4건 미만이면 설명 수정과 재실시 기록이 있다.</p></li><li><p>번호 일치와 내용의 타당성을 따로 확인했다.</p></li></ul><h4>이런 결과는 통과시키지 않는다</h4><ul><li><p>R1을 “무장 인원으로 확인됨”이라고만 요약했다.</p></li><li><p>“모르므로 가능성 낮음”이라고 적었다.</p></li><li><p>서로 다른 내용을 판단한 뒤 번호만 비교했다.</p></li></ul><h4>평가 의견 작성 방법</h4><p>“근거 부족”이라고만 쓰지 말고 단계 번호, 충족하지 못한 조건, 제출물에서 실제로 빠지거나 잘못된 부분, 다시 할 일을 적는다. 예를 들어 STEP 01의 첫 조건은 “다섯 보고에 핵심 내용·출처·시각·수단을 적었다.”이다. 이 조건에 못 미쳤다면 어떤 줄·항목·참여자 기록이 빠졌는지 지목한다.</p><h4>다시 평가할 때</h4><p>틀리거나 빠진 내용만 구체적으로 알려 준다. 문서 작성 오류는 원자료와 수정본을 대조한다. 실제 수행 오류는 해당 행동을 다시 하게 하고 새 회차 기록으로 확인한다. 최초 결과는 지우지 않는다.</p>', '<h4>평가 질문</h4><p>열 가지 보류 조건과 같은 출처 재전달·인계를 올바르게 처리했는가</p><h4>평가자가 확인할 위치</h4><p>framework/ops07_source_reliability_matrix.md의 STEP 03, STEP 05과 그 단계의 개인 기록을 확인한다.</p><h4>PASS를 주는 방법</h4><p>아래 단계별 항목을 모두 충족해야 이 평가 질문에 PASS를 준다. 항목 옆에 확인한 표·줄·회차를 적는다. 한 항목이라도 없거나 틀리면 FAIL 사유에 해당 단계와 문장을 적는다. 학습자가 쓴 말이 예시와 똑같을 필요는 없지만 확인할 내용은 모두 있어야 한다.</p><h4>STEP 03 보류할 조건 열 가지와 확인 질문을 작성하기</h4><p>찾을 결과: 서로 다른 조건 열 가지 이상. 모든 조건에 질문·담당·답을 기다릴 때 할 일이 있어야 한다.</p><ul><li><p>서로 다른 조건이 열 가지 이상이다.</p></li><li><p>각 조건에 구체적인 상황·보류 이유·담당·질문·필요한 답이 있다.</p></li><li><p>답이 없을 때 보류와 후속 확인 계획이 남는다.</p></li><li><p>일정 시간이 지나거나 채널 수가 늘었다는 이유만으로 확인 완료로 바꾸지 않는다.</p></li></ul><h4>STEP 05 같은 제보가 세 번 전달된 경우를 확인하고 인계하기</h4><p>찾을 결과: R2·R6 비교표와 인계 기록 1개. 각자 자신이 맡은 질문이나 인계 내용을 설명한다.</p><ul><li><p>R6를 세 개의 독립 제보로 세지 않았다.</p></li><li><p>최초 제보 1건과 새 자료 없음의 이유를 설명했다.</p></li><li><p>확인 질문·담당·아직 답이 없다는 상태가 있다.</p></li><li><p>받는 사람이 맡을 일을 다시 말한 내용이 있다.</p></li></ul><h4>이런 결과는 통과시키지 않는다</h4><ul><li><p>열 줄을 모두 “모호하면 보류”로 채웠다.</p></li><li><p>“세 채널이 일치했으므로 확인 완료”라고 적었다.</p></li></ul><h4>평가 의견 작성 방법</h4><p>“근거 부족”이라고만 쓰지 말고 단계 번호, 충족하지 못한 조건, 제출물에서 실제로 빠지거나 잘못된 부분, 다시 할 일을 적는다. 예를 들어 STEP 03의 첫 조건은 “서로 다른 조건이 열 가지 이상이다.”이다. 이 조건에 못 미쳤다면 어떤 줄·항목·참여자 기록이 빠졌는지 지목한다.</p><h4>다시 평가할 때</h4><p>틀리거나 빠진 내용만 구체적으로 알려 준다. 문서 작성 오류는 원자료와 수정본을 대조한다. 실제 수행 오류는 해당 행동을 다시 하게 하고 새 회차 기록으로 확인한다. 최초 결과는 지우지 않는다.</p>'], 'evaluationTypes': ['REQUIRED', 'REQUIRED'], 'evaluationText': '<h4>미션 전체 판정</h4><p>이 미션은 아래 두 평가 질문이 모두 PASS여야 통과다. 조별 문서와 개인의 실제 수행을 나눠 확인한다. 수행하지 않았거나 기록이 없어 확인할 수 없는 사람은 미평가로 남기고 다시 실시한다. 최종 판정에서는 미평가를 PASS로 바꾸지 않는다.</p><h4>평가 질문</h4><p>출처와 확인 표시를 구분하고 개인 판정을 정직하게 비교했는가</p><h4>평가자가 확인할 위치</h4><p>framework/ops07_source_reliability_matrix.md의 STEP 01, STEP 02, STEP 04과 그 단계의 개인 기록을 확인한다.</p><h4>PASS를 주는 방법</h4><p>아래 단계별 항목을 모두 충족해야 이 평가 질문에 PASS를 준다. 항목 옆에 확인한 표·줄·회차를 적는다. 한 항목이라도 없거나 틀리면 FAIL 사유에 해당 단계와 문장을 적는다. 학습자가 쓴 말이 예시와 똑같을 필요는 없지만 확인할 내용은 모두 있어야 한다.</p><h4>STEP 01 보고의 내용과 최초 출처를 따로 적기</h4><p>찾을 결과: R1~R5 다섯 줄의 출처·내용 정리표.</p><ul><li><p>다섯 보고에 핵심 내용·출처·시각·수단을 적었다.</p></li><li><p>R1의 인원 진입 장면과 무장 주장을 구분했다.</p></li><li><p>R3의 영상 중단, R4의 2시간 경과, R5의 시각·장소 미확인을 남겼다.</p></li><li><p>출처와 관측 시각이 없으면 임의로 채우지 않았다.</p></li></ul><h4>STEP 02 정보 확인 표시 여섯 가지를 이해하고 예를 적기</h4><p>찾을 결과: 여섯 줄의 설명표와 자료 없음·반대 자료 있음의 차이 설명.</p><ul><li><p>여섯 표시가 모두 있고 각 표시의 뜻과 예가 있다.</p></li><li><p>확인됨의 적용 범위를 지금 확인한 내용으로 제한했다.</p></li><li><p>판단할 자료가 없는 경우를 가능성 낮음으로 처리하지 않았다.</p></li><li><p>실제 정보 등급이나 행동 허가가 아니라 수업용 표시라고 밝혔다.</p></li></ul><h4>STEP 04 같은 보고를 각자 판단한 뒤 차이를 비교하기</h4><p>찾을 결과: 다섯 보고의 개인 원답, 최초 일치율의 계산식, 의견 차이와 수정 기록.</p><ul><li><p>전원이 같은 다섯 문장을 판단했다.</p></li><li><p>토의 전 각자의 번호와 이유를 보존했다.</p></li><li><p>최초 전원 일치 건수÷5×100 계산이 맞다.</p></li><li><p>다른 답의 원인을 적었으며 4건 미만이면 설명 수정과 재실시 기록이 있다.</p></li><li><p>번호 일치와 내용의 타당성을 따로 확인했다.</p></li></ul><h4>이런 결과는 통과시키지 않는다</h4><ul><li><p>R1을 “무장 인원으로 확인됨”이라고만 요약했다.</p></li><li><p>“모르므로 가능성 낮음”이라고 적었다.</p></li><li><p>서로 다른 내용을 판단한 뒤 번호만 비교했다.</p></li></ul><h4>평가 의견 작성 방법</h4><p>“근거 부족”이라고만 쓰지 말고 단계 번호, 충족하지 못한 조건, 제출물에서 실제로 빠지거나 잘못된 부분, 다시 할 일을 적는다. 예를 들어 STEP 01의 첫 조건은 “다섯 보고에 핵심 내용·출처·시각·수단을 적었다.”이다. 이 조건에 못 미쳤다면 어떤 줄·항목·참여자 기록이 빠졌는지 지목한다.</p><h4>다시 평가할 때</h4><p>틀리거나 빠진 내용만 구체적으로 알려 준다. 문서 작성 오류는 원자료와 수정본을 대조한다. 실제 수행 오류는 해당 행동을 다시 하게 하고 새 회차 기록으로 확인한다. 최초 결과는 지우지 않는다.</p><h4>평가 질문</h4><p>열 가지 보류 조건과 같은 출처 재전달·인계를 올바르게 처리했는가</p><h4>평가자가 확인할 위치</h4><p>framework/ops07_source_reliability_matrix.md의 STEP 03, STEP 05과 그 단계의 개인 기록을 확인한다.</p><h4>PASS를 주는 방법</h4><p>아래 단계별 항목을 모두 충족해야 이 평가 질문에 PASS를 준다. 항목 옆에 확인한 표·줄·회차를 적는다. 한 항목이라도 없거나 틀리면 FAIL 사유에 해당 단계와 문장을 적는다. 학습자가 쓴 말이 예시와 똑같을 필요는 없지만 확인할 내용은 모두 있어야 한다.</p><h4>STEP 03 보류할 조건 열 가지와 확인 질문을 작성하기</h4><p>찾을 결과: 서로 다른 조건 열 가지 이상. 모든 조건에 질문·담당·답을 기다릴 때 할 일이 있어야 한다.</p><ul><li><p>서로 다른 조건이 열 가지 이상이다.</p></li><li><p>각 조건에 구체적인 상황·보류 이유·담당·질문·필요한 답이 있다.</p></li><li><p>답이 없을 때 보류와 후속 확인 계획이 남는다.</p></li><li><p>일정 시간이 지나거나 채널 수가 늘었다는 이유만으로 확인 완료로 바꾸지 않는다.</p></li></ul><h4>STEP 05 같은 제보가 세 번 전달된 경우를 확인하고 인계하기</h4><p>찾을 결과: R2·R6 비교표와 인계 기록 1개. 각자 자신이 맡은 질문이나 인계 내용을 설명한다.</p><ul><li><p>R6를 세 개의 독립 제보로 세지 않았다.</p></li><li><p>최초 제보 1건과 새 자료 없음의 이유를 설명했다.</p></li><li><p>확인 질문·담당·아직 답이 없다는 상태가 있다.</p></li><li><p>받는 사람이 맡을 일을 다시 말한 내용이 있다.</p></li></ul><h4>이런 결과는 통과시키지 않는다</h4><ul><li><p>열 줄을 모두 “모호하면 보류”로 채웠다.</p></li><li><p>“세 채널이 일치했으므로 확인 완료”라고 적었다.</p></li></ul><h4>평가 의견 작성 방법</h4><p>“근거 부족”이라고만 쓰지 말고 단계 번호, 충족하지 못한 조건, 제출물에서 실제로 빠지거나 잘못된 부분, 다시 할 일을 적는다. 예를 들어 STEP 03의 첫 조건은 “서로 다른 조건이 열 가지 이상이다.”이다. 이 조건에 못 미쳤다면 어떤 줄·항목·참여자 기록이 빠졌는지 지목한다.</p><h4>다시 평가할 때</h4><p>틀리거나 빠진 내용만 구체적으로 알려 준다. 문서 작성 오류는 원자료와 수정본을 대조한다. 실제 수행 오류는 해당 행동을 다시 하게 하고 새 회차 기록으로 확인한다. 최초 결과는 지우지 않는다.</p>', 'aiGuidelines': {'allow': ['R1~R6 사례의 보고 내용·최초 출처·관측 시각·전달 수단 분리 표 형식 검토받기', '확인 표시 6단계(확인됨·거의 확실·가능성 있음·가능성 낮음·의심스러움·확인 불가) 정의와 보류 조건 10가지 문장 명확성 다듬기', '최초 전원 일치율(일치 건수÷5×100) 계산 검산 및 R6 중복 전달 구분 문안 검토받기'], 'rules': ['AI 사용은 필수가 아니며, 개인 원답과 조별 비교 기록은 토의 전에 먼저 분리해 보존한다(framework/ops07_source_reliability_matrix.md).', '자료 부재를 "가능성 낮음"으로 처리하거나, R6처럼 같은 제보(R2)를 여러 채널이 재전달한 것을 독립 확인으로 세지 않는다.', 'STEP 04의 5문장 개인 판정과 STEP 05의 인계 내용 복창은 학습자가 직접 수행한다.'], 'forbid': ['실제 부대·개인정보·정보 등급·작전 전문을 입력하기', 'R1~R5의 빈 출처·관측 시각을 임의로 채우거나 시간 경과·채널 수만으로 확인 완료로 바꾸게 하기', '개인 판정 원답과 최초 전원 일치율·재실시 기록을 실제 수행 없이 생성하기']}}, {'id': 'OPS07_m03', 'sequence': 3, 'title': '민간 위험과 불확실성에 따른 확인 책임 설계', 'durationHours': 3, 'overview': '<p>민간 위험 신호와 모르는 내용을 구분한다. 부족한 정보 여덟 가지의 질문과 답의 한계를 적는다.</p><p>아래 다섯 단계에는 필요한 자료·작업 순서·예시·제출 내용과 평가 조건이 들어 있다. 각 단계의 자료를 읽은 뒤 순서대로 작성한다.</p>', 'objectives': '<h4>이 미션을 마치면 할 수 있어야 하는 일</h4><ul><li><p>민간 위험 신호와 모르는 내용을 구분한다.</p></li><li><p>부족한 정보 여덟 가지의 질문과 답의 한계를 적는다.</p></li><li><p>열두 조합 담당표를 시험하고 실제 승인 기준과 구별한다.</p></li></ul>', 'scenario': '<h4>학습자의 역할</h4><p>학습자는 민간인과 아군에게 불필요한 피해가 생기지 않도록 보고를 읽고, 모르는 내용을 질문하고, 다음 담당자에게 기록을 넘기는 역할을 연습한다. 사람의 소속을 맞히거나 실제 타격을 결정하는 과제가 아니다.</p><h4>이번 미션의 과제</h4><ul><li><p>민간 위험 신호와 모르는 내용을 구분한다.</p></li><li><p>부족한 정보 여덟 가지의 질문과 답의 한계를 적는다.</p></li><li><p>열두 조합 담당표를 시험하고 실제 승인 기준과 구별한다.</p></li></ul><h4>자료와 역할의 범위</h4><p>사례·시각·장소·역할은 모두 교육용이다. 실제 부대 자료를 쓰지 않는다. 자료에 답이 없으면 자료에 없다고 적는다. 실제 행동 권한은 이 수업이 정하지 않는다.</p><h4>개인 작업과 조별 작업</h4><p>개인 판단을 먼저 요청한 단계에서는 토의 전에 자신의 답을 남긴다. 그 뒤 조별 답과 수정 이유를 별도로 쓴다. 보고·인계·시험 단계에서는 실제 한 말과 행동을 기록한다.</p>', 'attachments': [], 'materials': ['/uploads/OPS07/OPS07_m03_student.ipynb', '/uploads/OPS07/OPS07_m03_workbook.ipynb'], 'materialText': '<h4>실습 노트북 안내 (선택 보조 도구)</h4><p><strong>학생 실습용</strong>: <code>OPS07_m03_student.ipynb</code> (직접 ANSWER 표와 note를 작성합니다.)<br><strong>교수자·해설용</strong>: <code>OPS07_m03_workbook.ipynb</code> (예시 풀이와 해설 note가 포함되어 있습니다.)<br><em>※ 본 과제는 문서 편집기나 인쇄 양식으로도 수행할 수 있으며, 노트북은 사례·질문·기록 양식을 표로 정리하는 선택 보조 도구입니다.</em></p><h4>자료를 읽는 순서</h4>\n\n<ol>\n<li><p>STEP 1에서는 시장·병원·학교·창고의 보고를 읽고 민간 위험 신호와 모르는 내용을 구별합니다.</p></li>\n<li><p>위험 표시 1~4의 뜻을 읽고, STEP 2에서는 G1~G8에 대해 확인 질문과 답의 한계를 적습니다.</p></li>\n<li><p>STEP 3의 담당표를 만들 때는 해당 단계에 있는 정보 부족 표시의 뜻을 함께 읽습니다. STEP 4~5의 시험 상황과 적용 한계도 각 단계 설명에서 확인합니다.</p></li>\n</ol>\n\n<h4>자료 1. 네 장소에 관한 보고</h4>\n\n<h4>시장 나-2</h4>\n\n<p>영업 중이며 주민 이동 보고가 있음.</p>\n\n<h4>병원 다-2</h4>\n\n<p>운영 중이라고 보고됨.</p>\n\n<p>내부 인원 수와 관측 시각은 제공되지 않음.</p>\n\n<h4>학교 라-2</h4>\n\n<p>휴교 중이며 대피소로 쓴다는 주민 말이 있음.</p>\n\n<p>실제 이용 여부와 내부 인원은 확인하지 못함.</p>\n\n<h4>창고 마-3</h4>\n\n<p>적 지휘소라는 제보와 민간 업체 직원 출입 보고가 함께 있음.</p>\n\n<p>두 설명 모두 추가 확인이 필요함.</p>\n\n<h4>자료 2. 가상 위험 표시의 뜻</h4>\n\n<p>숫자는 이 수업에서 위험 신호를 정리하는 표시입니다. 실제 피해 규모나 행동 허가를 뜻하지 않습니다.</p>\n\n<p><strong>1:</strong> 제공된 자료에는 민간 위험 신호가 나오지 않음. 민간인이 없다는 뜻은 아님.</p>\n\n<p><strong>2:</strong> 민간 시설이나 주민 이동 가능성이 있어 이용 상태를 물어야 함.</p>\n\n<p><strong>3:</strong> 민간 활동 또는 운영 중인 시설이 보고됨.</p>\n\n<p><strong>4:</strong> 같은 장소에서 민간인·병원·대피소 이용 가능성이 보고돼 보호 확인을 먼저 해야 함.</p>\n\n<h4>자료 3. 확인 질문을 만들 정보 부족 항목</h4>\n\n<ul>\n<li><p>G1. 어두워서 물체의 모양을 구분하기 어려움</p></li>\n<li><p>G2. 영상이 중간에 끊김</p></li>\n<li><p>G3. 건물 내부를 보지 못함</p></li>\n<li><p>G4. 처음 관찰한 시각을 모름</p></li>\n<li><p>G5. 처음 제보한 출처를 모름</p></li>\n<li><p>G6. 아군의 현재 위치 보고를 받지 못함</p></li>\n<li><p>G7. 주민 이동 경로의 변경 내용이 상황판에 없음</p></li>\n<li><p>G8. 원 보고와 수신 기록의 장소 표기가 다름</p></li>\n</ul>\n\n<h4>자료를 사용할 때 지킬 점</h4>\n\n<p>위험 표시는 이 수업에서 민간 위험 신호를 정리하기 위한 것입니다.</p>\n\n<p>자료에 민간인이 나오지 않는다는 사실과 민간인이 없다는 판단은 다릅니다.</p>\n\n<p>모든 장소 표기는 가상입니다.</p>\n\n<p>모든 사례·시각·장소·역할은 교육용입니다. 필요한 자료와 구체적인 작업 순서는 각 단계 설명에도 들어 있습니다.</p>\n', 'guideSteps': [{'step': 1, 'problem': '민간 위험 신호와 모르는 내용을 따로 정리하기', 'activity': '<h4>이번 단계에서 하는 일</h4><p>민간인이 보이지 않았다는 말과 민간인이 없다는 말은 다르다. 시설별로 보고된 위험 신호와 아직 모르는 내용을 나눠 적는다.</p><h4>먼저 읽을 자료</h4><h4>장소별로 제공된 정보</h4><ul><li><p><strong>시장 나-2:</strong> 영업 중이며 주민 이동 보고가 있음.</p></li><li><p><strong>병원 다-2:</strong> 운영 중이라고 보고됨. 내부 인원 수와 관측 시각은 제공되지 않음.</p></li><li><p><strong>학교 라-2:</strong> 휴교 중이며 대피소로 쓴다는 주민 말이 있음. 실제 이용 여부와 내부 인원은 확인하지 못함.</p></li><li><p><strong>창고 마-3:</strong> 적 지휘소라는 제보와 민간 업체 직원 출입 보고가 함께 있음. 두 설명 모두 추가 확인이 필요함.</p></li></ul><h4>가상 위험 표시의 뜻</h4><ul><li><p><strong>1:</strong> 제공된 자료에는 민간 위험 신호가 나오지 않음. 민간인이 없다는 뜻은 아님.</p></li><li><p><strong>2:</strong> 민간 시설이나 주민 이동 가능성이 있어 이용 상태를 물어야 함.</p></li><li><p><strong>3:</strong> 민간 활동 또는 운영 중인 시설이 보고됨.</p></li><li><p><strong>4:</strong> 같은 장소에서 민간인·병원·대피소 이용 가능성이 보고돼 보호 확인을 먼저 해야 함.</p></li></ul><p>위험 표시는 확인 순서를 연습하기 위한 정성 분류다.</p><p>예상 피해 인원·피해 반경·실제 CDE를 계산하지 않는다.</p><h4>작업 순서</h4><ol><li><p>네 장소의 제공 내용을 읽고 위험 신호를 한 문장씩 적는다.</p></li><li><p>내부 이용·인원·관측 시각처럼 알 수 없는 내용을 별도 칸에 쓴다.</p></li><li><p>위험 표시 1~4 중 자료에 맞는 표시를 임시로 고른다. 결정하기 어렵다면 “판단 보류”라고 쓰고 이유를 적어도 된다.</p></li><li><p>표시를 고른 이유에 실제 카드 문장을 넣는다. 낮은 번호를 골랐다고 안전하다고 쓰지 않는다.</p></li><li><p>학교의 휴교와 대피소 진술을 둘 다 기록했는지 확인한다.</p></li></ol><h4>작성 예시</h4><p>다음은 쓰는 방식을 보여 주는 예다. 자기 판단이나 실제 수행·측정 결과가 필요한 칸에 그대로 복사하지 않는다.</p><p>학교 / 신호 “대피소로 이용한다는 말이 있음” / 모르는 내용 “실제 이용 여부와 내부 인원” / 임시 표시 “4 검토” / 이유 “같은 장소의 대피소 이용 가능성을 먼저 확인해야 함.”</p><h4>작성할 표 또는 기록 항목</h4><ul><li><p>장소</p></li><li><p>보고된 민간 위험 신호</p></li><li><p>모르는 내용</p></li><li><p>가상 위험 표시 또는 판단 보류</p></li><li><p>그렇게 적은 이유</p></li></ul><h4>제출할 내용과 끝내는 기준</h4><p>시장·병원·학교·창고 네 줄의 정리표.</p><p>미션 제출 문서 analysis/ops07_uncertainty_assessment.md의 STEP 01 아래에 작성한다. 개인 기록과 조별 결과를 구분하고 실제 이름 대신 학습자 코드를 써도 된다. 표의 모양이 달라도 필요한 내용을 찾을 수 있으면 된다.</p><h4>마치기 전 확인</h4><ul><li><p>네 장소의 위험 신호와 모르는 내용을 따로 썼다.</p></li><li><p>표시를 고른 이유를 제공된 문장과 연결했다.</p></li><li><p>학교의 휴교를 인원 부재로 해석하지 않았다.</p></li><li><p>보이지 않음·자료 없음·낮은 번호를 안전 확정으로 바꾸지 않았다.</p></li></ul><h4>배정 시간</h4><p>30분</p>', 'evalText': '<h4>평가할 자료와 위치</h4><p>analysis/ops07_uncertainty_assessment.md의 STEP 01 결과와 해당 개인 수행 기록을 확인한다. 이 단계에서 요청한 제출 내용은 다음과 같다: 시장·병원·학교·창고 네 줄의 정리표.</p><h4>평가 순서</h4><ol><li><p>아래 통과 조건을 하나씩 읽고 제출물에서 확인한 위치를 옆에 적는다.</p></li><li><p>실제 설명·보고·인계가 필요한 단계는 수행 기록이나 직접 관찰로 확인한다. 서식만 있으면 수행한 것으로 보지 않는다.</p></li><li><p>통과 조건을 모두 충족하면 이 단계는 PASS다. 하나라도 빠지거나 반대로 적었으면 FAIL로 두고 빠진 항목을 구체적으로 적는다. 표현이나 표 모양이 예시와 달라도 뜻이 같고 근거를 찾을 수 있으면 인정한다.</p></li></ol><h4>PASS 조건</h4><ul><li><p>네 장소의 위험 신호와 모르는 내용을 따로 썼다.</p></li><li><p>표시를 고른 이유를 제공된 문장과 연결했다.</p></li><li><p>학교의 휴교를 인원 부재로 해석하지 않았다.</p></li><li><p>보이지 않음·자료 없음·낮은 번호를 안전 확정으로 바꾸지 않았다.</p></li></ul><h4>FAIL에 해당하는 예</h4><ul><li><p>학교를 “휴교이므로 민간인 없음”이라고 적었다.</p></li><li><p>관측 시각과 내부 자료가 없는데 인원 수를 0명으로 채웠다.</p></li></ul><h4>이해 여부를 확인할 질문</h4><p>학교가 휴교라는 정보로 건물 내부에 사람이 없다고 말할 수 있는가?</p><h4>인정할 수 있는 답의 핵심</h4><p>말할 수 없으며 대피소 이용과 내부 인원은 별도 확인이 필요하다고 답한다.</p><h4>보완과 재평가</h4><p>각 장소의 확인된 내용과 모르는 내용을 두 칸으로 다시 나누고 표시의 이유를 고친다.</p><p>수정 전 결과와 재평가 결과를 구분해 남긴다. 실제 하지 않은 보고·채점·시험을 나중에 했던 것처럼 채우지 않는다.</p>', 'attachments': []}, {'step': 2, 'problem': '확인하지 못한 정보 여덟 가지에 질문과 한계를 붙이기', 'activity': '<h4>이번 단계에서 하는 일</h4><p>자료가 부족한 이유를 써야 맞는 질문을 보낼 수 있다. 답을 받더라도 그 답으로 알 수 없는 범위가 남을 수 있으므로 함께 적는다.</p><h4>먼저 읽을 자료</h4><h4>정보가 부족한 항목</h4><ul><li><p><strong>G1:</strong> 어두워서 물체의 모양을 구분하기 어려움</p></li><li><p><strong>G2:</strong> 영상이 중간에 끊김</p></li><li><p><strong>G3:</strong> 건물 내부를 보지 못함</p></li><li><p><strong>G4:</strong> 처음 관찰한 시각을 모름</p></li><li><p><strong>G5:</strong> 처음 제보한 출처를 모름</p></li><li><p><strong>G6:</strong> 아군의 현재 위치 보고를 받지 못함</p></li><li><p><strong>G7:</strong> 주민 이동 경로의 변경 내용이 상황판에 없음</p></li><li><p><strong>G8:</strong> 원 보고와 수신 기록의 장소 표기가 다름</p></li></ul><p>예를 들어 출입 기록을 받아도 건물 내부 전체 인원까지 알 수 있는 것은 아니다.</p><p>아래 확인 방법은 보고·자료 대조의 수준에서 쓰며 실제 관측 방법이나 현장 이동을 설계하지 않는다.</p><h4>작업 순서</h4><ol><li><p>여덟 항목을 한 줄씩 옮기고 어떤 판단을 하기 어려운지 적는다.</p></li><li><p>각 항목에 중요도 높음·중간·낮음 중 하나와 이유를 적는다. 중요한 정보인지 모르면 그 이유를 적고 교관에게 묻는다.</p></li><li><p>누구에게 어떤 자료를 요청할지 적는다.</p></li><li><p>요청한 답으로 무엇까지 알 수 있으며 무엇은 여전히 모르는지 나눠 적는다.</p></li><li><p>항목의 표현만 바꿔 같은 질문을 반복하지 않았는지 조원과 확인한다.</p></li></ol><h4>작성 예시</h4><p>다음은 쓰는 방식을 보여 주는 예다. 자기 판단이나 실제 수행·측정 결과가 필요한 칸에 그대로 복사하지 않는다.</p><p>G3 내부 미관측 / 어려운 판단 “현재 내부 이용과 인원 확인” / 질문 “시설 이용 담당자가 가진 최근 이용 기록이 있습니까?” / 알 수 있는 범위 “그 기록에 적힌 이용 정보” / 남는 한계 “기록 이후 출입과 내부 전체 인원은 별도 확인 필요.”</p><h4>작성할 표 또는 기록 항목</h4><ul><li><p>번호</p></li><li><p>부족한 정보</p></li><li><p>어떤 판단이 어려운지</p></li><li><p>중요도와 이유</p></li><li><p>확인 담당·질문</p></li><li><p>받아야 할 자료</p></li><li><p>답으로 알 수 있는 범위</p></li><li><p>답을 받아도 모르는 내용</p></li></ul><h4>제출할 내용과 끝내는 기준</h4><p>여덟 항목의 확인 계획표. 각 줄에 질문·담당·답의 한계를 모두 적는다.</p><p>미션 제출 문서 analysis/ops07_uncertainty_assessment.md의 STEP 02 아래에 작성한다. 개인 기록과 조별 결과를 구분하고 실제 이름 대신 학습자 코드를 써도 된다. 표의 모양이 달라도 필요한 내용을 찾을 수 있으면 된다.</p><h4>마치기 전 확인</h4><ul><li><p>서로 다른 여덟 항목을 모두 다뤘다.</p></li><li><p>각 항목에서 부족한 정보와 확인 질문이 서로 맞는다.</p></li><li><p>담당 역할과 요청할 답·자료가 구체적이다.</p></li><li><p>답을 받았을 때도 알 수 없는 내용을 적었다.</p></li></ul><h4>배정 시간</h4><p>40분</p>', 'evalText': '<h4>평가할 자료와 위치</h4><p>analysis/ops07_uncertainty_assessment.md의 STEP 02 결과와 해당 개인 수행 기록을 확인한다. 이 단계에서 요청한 제출 내용은 다음과 같다: 여덟 항목의 확인 계획표. 각 줄에 질문·담당·답의 한계를 모두 적는다.</p><h4>평가 순서</h4><ol><li><p>아래 통과 조건을 하나씩 읽고 제출물에서 확인한 위치를 옆에 적는다.</p></li><li><p>실제 설명·보고·인계가 필요한 단계는 수행 기록이나 직접 관찰로 확인한다. 서식만 있으면 수행한 것으로 보지 않는다.</p></li><li><p>통과 조건을 모두 충족하면 이 단계는 PASS다. 하나라도 빠지거나 반대로 적었으면 FAIL로 두고 빠진 항목을 구체적으로 적는다. 표현이나 표 모양이 예시와 달라도 뜻이 같고 근거를 찾을 수 있으면 인정한다.</p></li></ol><h4>PASS 조건</h4><ul><li><p>서로 다른 여덟 항목을 모두 다뤘다.</p></li><li><p>각 항목에서 부족한 정보와 확인 질문이 서로 맞는다.</p></li><li><p>담당 역할과 요청할 답·자료가 구체적이다.</p></li><li><p>답을 받았을 때도 알 수 없는 내용을 적었다.</p></li></ul><h4>FAIL에 해당하는 예</h4><ul><li><p>내부 미관측의 해결책에 “추가 확인”만 적고 누구에게 무엇을 물을지 없다.</p></li><li><p>출입 기록을 받으면 내부 전체가 안전하다고 적었다.</p></li></ul><h4>이해 여부를 확인할 질문</h4><p>시설 이용 기록을 받으면 어디까지 알 수 있고 무엇은 남는가?</p><h4>인정할 수 있는 답의 핵심</h4><p>그 기록이 다루는 시점과 내용은 알 수 있으나 기록 이후 출입이나 미기록 인원은 알 수 없다고 설명한다.</p><h4>보완과 재평가</h4><p>“추가 확인”이라는 표현을 실제 질문으로 바꾸고 답이 다루지 않는 범위를 한 문장씩 더한다.</p><p>수정 전 결과와 재평가 결과를 구분해 남긴다. 실제 하지 않은 보고·채점·시험을 나중에 했던 것처럼 채우지 않는다.</p>', 'attachments': []}, {'step': 3, 'problem': '위험 표시와 정보 부족 정도에 따라 담당 업무를 적기', 'activity': '<h4>이번 단계에서 하는 일</h4><p>위험 번호가 같아도 아는 내용이 얼마나 되는지에 따라 필요한 확인이 다르다. 조합별로 누가 무엇을 확인하고 누가 검토할지 정리한다.</p><h4>먼저 읽을 자료</h4><h4>가상 위험 표시의 뜻</h4><ul><li><p><strong>1:</strong> 제공된 자료에는 민간 위험 신호가 나오지 않음. 민간인이 없다는 뜻은 아님.</p></li><li><p><strong>2:</strong> 민간 시설이나 주민 이동 가능성이 있어 이용 상태를 물어야 함.</p></li><li><p><strong>3:</strong> 민간 활동 또는 운영 중인 시설이 보고됨.</p></li><li><p><strong>4:</strong> 같은 장소에서 민간인·병원·대피소 이용 가능성이 보고돼 보호 확인을 먼저 해야 함.</p></li></ul><p>가상 역할은 확인 담당자, 기록자, 안전 검토자, 상급 검토자다.</p><p>확인 담당자는 질문의 답을 구하는 역할, 안전 검토자는 받은 자료의 빠진 부분을 확인하는 역할이다.</p><p>실제 계급이나 승인 권한은 정하지 않는다.</p><h4>정보 부족 정도의 뜻</h4><ul><li><p><strong>낮음:</strong> 지금 묻는 내용의 출처·시각·범위가 제시돼 있으나 다른 확인도 끝났다는 뜻은 아님.</p></li><li><p><strong>중간:</strong> 확인할 내용 일부가 빠졌거나 오래돼 다시 물어야 함.</p></li><li><p><strong>높음:</strong> 핵심 출처·현재 위치·시설 이용 등이 없거나 서로 맞지 않아 판단하기 어려움.</p></li></ul><h4>작업 순서</h4><ol><li><p>위험 표시 1~4를 행으로, 정보 부족 정도 낮음·중간·높음을 열로 둔 표를 만든다. 모두 열두 칸이다.</p></li><li><p>각 칸에 보류 상태, 확인 담당, 보낼 질문, 받아야 할 자료, 검토 담당을 적는다.</p></li><li><p>정보 부족 높음인 네 칸은 모두 “보류 유지·재확인”으로 쓴다.</p></li><li><p>나머지 칸도 번호만으로 자동 진행이 되지 않게 한다. 해당 질문을 확인할 수 없으면 그 질문은 미확인으로 남긴다.</p></li><li><p>가장 위험한 표시이거나 부족한 정보가 늘었는데 검토할 내용을 줄인 칸이 없는지 살핀다. 상급 검토가 필요하면 이유와 넘길 내용을 적는다.</p></li></ol><h4>작성 예시</h4><p>다음은 쓰는 방식을 보여 주는 예다. 자기 판단이나 실제 수행·측정 결과가 필요한 칸에 그대로 복사하지 않는다.</p><p>위험 2·정보 부족 높음 / 보류 유지 / 확인 담당 “시설 이용 정보 담당자” / 질문 “최근 이용 기록과 그 시점을 알려 주세요” / 검토 담당 “안전 검토자” / 아직 답하지 못한 질문은 계속 기록.</p><h4>작성할 표 또는 기록 항목</h4><ul><li><p>위험 표시</p></li><li><p>정보 부족 정도</p></li><li><p>현재 보류 상태</p></li><li><p>확인 담당</p></li><li><p>확인 질문</p></li><li><p>필요한 자료</p></li><li><p>검토 담당</p></li></ul><h4>제출할 내용과 끝내는 기준</h4><p>4×3 담당표 한 개. 열두 조합 각각에 일곱 항목을 적는다.</p><p>미션 제출 문서 analysis/ops07_uncertainty_assessment.md의 STEP 03 아래에 작성한다. 개인 기록과 조별 결과를 구분하고 실제 이름 대신 학습자 코드를 써도 된다. 표의 모양이 달라도 필요한 내용을 찾을 수 있으면 된다.</p><h4>마치기 전 확인</h4><ul><li><p>위험 4종×정보 부족 3종의 열두 조합이 모두 있다.</p></li><li><p>열두 조합에 보류 상태·담당·질문·필요 자료·검토 담당이 있다.</p></li><li><p>정보 부족 높음인 네 조합이 모두 보류 유지·재확인이다.</p></li><li><p>낮은 번호 또는 담당자 이름만으로 실제 진행을 허용하는 문장이 없다.</p></li></ul><h4>배정 시간</h4><p>45분</p>', 'evalText': '<h4>평가할 자료와 위치</h4><p>analysis/ops07_uncertainty_assessment.md의 STEP 03 결과와 해당 개인 수행 기록을 확인한다. 이 단계에서 요청한 제출 내용은 다음과 같다: 4×3 담당표 한 개. 열두 조합 각각에 일곱 항목을 적는다.</p><h4>평가 순서</h4><ol><li><p>아래 통과 조건을 하나씩 읽고 제출물에서 확인한 위치를 옆에 적는다.</p></li><li><p>실제 설명·보고·인계가 필요한 단계는 수행 기록이나 직접 관찰로 확인한다. 서식만 있으면 수행한 것으로 보지 않는다.</p></li><li><p>통과 조건을 모두 충족하면 이 단계는 PASS다. 하나라도 빠지거나 반대로 적었으면 FAIL로 두고 빠진 항목을 구체적으로 적는다. 표현이나 표 모양이 예시와 달라도 뜻이 같고 근거를 찾을 수 있으면 인정한다.</p></li></ol><h4>PASS 조건</h4><ul><li><p>위험 4종×정보 부족 3종의 열두 조합이 모두 있다.</p></li><li><p>열두 조합에 보류 상태·담당·질문·필요 자료·검토 담당이 있다.</p></li><li><p>정보 부족 높음인 네 조합이 모두 보류 유지·재확인이다.</p></li><li><p>낮은 번호 또는 담당자 이름만으로 실제 진행을 허용하는 문장이 없다.</p></li></ul><h4>FAIL에 해당하는 예</h4><ul><li><p>열두 칸에 직책 이름만 넣고 그 사람이 할 일이 없다.</p></li><li><p>“1등급·낮음이면 자동 진행”이라고 적었다.</p></li></ul><h4>이해 여부를 확인할 질문</h4><p>담당자 이름을 적었다는 사실과 그 담당자가 질문에 답했다는 사실은 같은가?</p><h4>인정할 수 있는 답의 핵심</h4><p>다르며 지정만 된 상태인지 실제 답을 받은 상태인지 별도로 확인해야 한다고 답한다.</p><h4>보완과 재평가</h4><p>직책 이름만 있는 칸에 질문·받을 자료를 추가하고 자동 진행 표현을 지운 뒤 다시 검토한다.</p><p>수정 전 결과와 재평가 결과를 구분해 남긴다. 실제 하지 않은 보고·채점·시험을 나중에 했던 것처럼 채우지 않는다.</p>', 'attachments': []}, {'step': 4, 'problem': '정보가 빠지거나 검토자와 연락되지 않는 경우를 시험하기', 'activity': '<h4>이번 단계에서 하는 일</h4><p>표가 완성돼도 예외 상황에서 다음 할 일이 없으면 사용할 수 없다. 아래 두 상황을 넣어 보류와 담당 업무가 어떻게 남는지 확인한다.</p><h4>먼저 읽을 자료</h4><p>가상 역할은 확인 담당자, 기록자, 안전 검토자, 상급 검토자다.</p><p>확인 담당자는 질문의 답을 구하는 역할, 안전 검토자는 받은 자료의 빠진 부분을 확인하는 역할이다.</p><p>실제 계급이나 승인 권한은 정하지 않는다.</p><h4>담당표를 적용할 시험 상황</h4><ul><li><p><strong>T1:</strong> 학교가 대피소라는 말이 있지만 최근 이용 자료와 내부 인원은 받지 못했다.</p></li><li><p><strong>T2:</strong> 확인 담당자는 질문을 보냈다고 말했지만 안전 검토자와 연락이 되지 않는다. 요청 접수와 검토 회신도 확인되지 않았다.</p></li></ul><h4>작업 순서</h4><ol><li><p>자신의 열두 조합 담당표를 준비한다. 없으면 이번 단계에 위험 표시·정보 부족·보류 상태·질문·담당 칸을 만들어 먼저 적는다.</p></li><li><p>T1의 부족한 정보를 적고 자기 표에서 어떤 조합을 참고할지 고른다.</p></li><li><p>T2에서 송신 시도·접수·검토 회신 중 실제로 알려진 것이 무엇인지 나눈다.</p></li><li><p>각 상황에 현재 상태와 다음 담당자가 해야 할 질문을 적는다. 실제로 연락이 되었다거나 위임받았다고 만들지 않는다.</p></li><li><p>자기 표만으로 다음 할 일을 정할 수 없었다면 그 칸에 필요한 문장을 추가하고 전후를 남긴다.</p></li></ol><h4>작성 예시</h4><p>다음은 쓰는 방식을 보여 주는 예다. 자기 판단이나 실제 수행·측정 결과가 필요한 칸에 그대로 복사하지 않는다.</p><p>T2: 알려진 것 “질문을 보냈다는 담당자 말” / 모르는 것 “검토자 접수·회신” / 현재 “보류 유지” / 다음 할 일 “기록 담당자가 접수 여부를 확인할 요청과 미회신 상태를 남김.”</p><h4>작성할 표 또는 기록 항목</h4><ul><li><p>시험 번호</p></li><li><p>알고 있는 내용</p></li><li><p>모르는 내용</p></li><li><p>참고한 담당표 조합</p></li><li><p>현재 상태</p></li><li><p>다음 질문·담당</p></li><li><p>고친 표 문장</p></li></ul><h4>제출할 내용과 끝내는 기준</h4><p>T1·T2 시험 기록 두 줄과 담당표 수정 전후 기록. 수정이 없으면 없었던 이유를 쓴다.</p><p>미션 제출 문서 analysis/ops07_uncertainty_assessment.md의 STEP 04 아래에 작성한다. 개인 기록과 조별 결과를 구분하고 실제 이름 대신 학습자 코드를 써도 된다. 표의 모양이 달라도 필요한 내용을 찾을 수 있으면 된다.</p><h4>마치기 전 확인</h4><ul><li><p>두 시험 상황을 모두 적용했다.</p></li><li><p>T1의 내부 이용·인원 미확인과 T2의 접수·회신 미확인을 남겼다.</p></li><li><p>누구에게 무엇을 다시 물을지 적었다.</p></li><li><p>연락 두절을 자동 위임이나 검토 완료로 바꾸지 않았다.</p></li></ul><h4>배정 시간</h4><p>35분</p>', 'evalText': '<h4>평가할 자료와 위치</h4><p>analysis/ops07_uncertainty_assessment.md의 STEP 04 결과와 해당 개인 수행 기록을 확인한다. 이 단계에서 요청한 제출 내용은 다음과 같다: T1·T2 시험 기록 두 줄과 담당표 수정 전후 기록. 수정이 없으면 없었던 이유를 쓴다.</p><h4>평가 순서</h4><ol><li><p>아래 통과 조건을 하나씩 읽고 제출물에서 확인한 위치를 옆에 적는다.</p></li><li><p>실제 설명·보고·인계가 필요한 단계는 수행 기록이나 직접 관찰로 확인한다. 서식만 있으면 수행한 것으로 보지 않는다.</p></li><li><p>통과 조건을 모두 충족하면 이 단계는 PASS다. 하나라도 빠지거나 반대로 적었으면 FAIL로 두고 빠진 항목을 구체적으로 적는다. 표현이나 표 모양이 예시와 달라도 뜻이 같고 근거를 찾을 수 있으면 인정한다.</p></li></ol><h4>PASS 조건</h4><ul><li><p>두 시험 상황을 모두 적용했다.</p></li><li><p>T1의 내부 이용·인원 미확인과 T2의 접수·회신 미확인을 남겼다.</p></li><li><p>누구에게 무엇을 다시 물을지 적었다.</p></li><li><p>연락 두절을 자동 위임이나 검토 완료로 바꾸지 않았다.</p></li></ul><h4>FAIL에 해당하는 예</h4><ul><li><p>T1에서 대피소 진술을 빼고 휴교 정보만 사용했다.</p></li><li><p>T2에서 “연락이 안 되므로 다른 사람이 승인한 것으로 처리”했다.</p></li></ul><h4>이해 여부를 확인할 질문</h4><p>요청을 보냈다는 말만으로 상대가 검토했다고 볼 수 있는가?</p><h4>인정할 수 있는 답의 핵심</h4><p>접수와 검토 회신이 따로 필요하며 보냈다는 사실만으로 완료라고 할 수 없다고 답한다.</p><h4>보완과 재평가</h4><p>송신·접수·회신을 세 칸으로 나눠 다시 쓰고 현재 확인된 칸에만 내용을 넣는다.</p><p>수정 전 결과와 재평가 결과를 구분해 남긴다. 실제 하지 않은 보고·채점·시험을 나중에 했던 것처럼 채우지 않는다.</p>', 'attachments': []}, {'step': 5, 'problem': '담당표로 알 수 없는 것을 밝히고 다음 미션에 넘기기', 'activity': '<h4>이번 단계에서 하는 일</h4><p>가상 담당표가 실제 승인 기준처럼 쓰이지 않도록 범위를 분명히 한다. 다음 작업자에게 표와 함께 아직 확인하지 못한 질문도 넘긴다.</p><h4>먼저 읽을 자료</h4><p>이 미션의 담당표는 가상 상황에서 확인 업무를 나누는 도구다.</p><p>실제 피해 추정, 법적 비례성 판단, 교전 승인, 무기·방법 선정에는 사용할 수 없다.</p><p>위 항목을 그대로 옮기기보다 왜 이 표로 판단할 수 없는지 설명한다.</p><p>가상 역할은 확인 담당자, 기록자, 안전 검토자, 상급 검토자다.</p><p>확인 담당자는 질문의 답을 구하는 역할, 안전 검토자는 받은 자료의 빠진 부분을 확인하는 역할이다.</p><p>실제 계급이나 승인 권한은 정하지 않는다.</p><h4>작업 순서</h4><ol><li><p>담당표로 판단할 수 없는 일을 세 가지 이상 적고 각각 이유를 쓴다.</p></li><li><p>각자 열두 칸 중 하나를 골라 “어떤 정보가 없고 누가 무엇을 확인해야 하는가”를 설명한다.</p></li><li><p>조의 미완료 질문을 모아 아직 답이 없는 이유·현재 담당·다음에 확인할 내용을 적는다.</p></li><li><p>다음 미션 담당자가 표를 받아 질문을 찾아 읽어 보게 한다. 찾지 못한 문장은 풀어서 고친다.</p></li><li><p>설명한 사람·고른 칸·받은 사람·넘긴 질문을 기록한다.</p></li></ol><h4>작성 예시</h4><p>다음은 쓰는 방식을 보여 주는 예다. 자기 판단이나 실제 수행·측정 결과가 필요한 칸에 그대로 복사하지 않는다.</p><p>판단할 수 없는 일 “실제 피해 인원 계산” / 이유 “이 표에는 무기 효과·현장 인원 자료가 없고 실제 CDE 방법도 다루지 않기 때문.”</p><h4>작성할 표 또는 기록 항목</h4><ul><li><p>판단할 수 없는 일과 이유</p></li><li><p>개인별 설명한 칸과 설명 내용</p></li><li><p>미완료 질문·담당·다음 확인</p></li><li><p>받은 사람의 확인</p></li></ul><h4>제출할 내용과 끝내는 기준</h4><p>적용 한계 세 가지 이상, 전원 설명 기록, 미완료 질문 인계표.</p><p>미션 제출 문서 analysis/ops07_uncertainty_assessment.md의 STEP 05 아래에 작성한다. 개인 기록과 조별 결과를 구분하고 실제 이름 대신 학습자 코드를 써도 된다. 표의 모양이 달라도 필요한 내용을 찾을 수 있으면 된다.</p><h4>마치기 전 확인</h4><ul><li><p>판단할 수 없는 일 세 가지 이상에 각각 이유가 있다.</p></li><li><p>전원이 담당표 한 칸의 질문과 역할을 설명했다.</p></li><li><p>실제 승인·피해 계산을 이 표가 대신한다고 쓰지 않았다.</p></li><li><p>남은 질문마다 담당과 다음 확인할 내용이 있다.</p></li></ul><h4>배정 시간</h4><p>30분</p>', 'evalText': '<h4>평가할 자료와 위치</h4><p>analysis/ops07_uncertainty_assessment.md의 STEP 05 결과와 해당 개인 수행 기록을 확인한다. 이 단계에서 요청한 제출 내용은 다음과 같다: 적용 한계 세 가지 이상, 전원 설명 기록, 미완료 질문 인계표.</p><h4>평가 순서</h4><ol><li><p>아래 통과 조건을 하나씩 읽고 제출물에서 확인한 위치를 옆에 적는다.</p></li><li><p>실제 설명·보고·인계가 필요한 단계는 수행 기록이나 직접 관찰로 확인한다. 서식만 있으면 수행한 것으로 보지 않는다.</p></li><li><p>통과 조건을 모두 충족하면 이 단계는 PASS다. 하나라도 빠지거나 반대로 적었으면 FAIL로 두고 빠진 항목을 구체적으로 적는다. 표현이나 표 모양이 예시와 달라도 뜻이 같고 근거를 찾을 수 있으면 인정한다.</p></li></ol><h4>PASS 조건</h4><ul><li><p>판단할 수 없는 일 세 가지 이상에 각각 이유가 있다.</p></li><li><p>전원이 담당표 한 칸의 질문과 역할을 설명했다.</p></li><li><p>실제 승인·피해 계산을 이 표가 대신한다고 쓰지 않았다.</p></li><li><p>남은 질문마다 담당과 다음 확인할 내용이 있다.</p></li></ul><h4>FAIL에 해당하는 예</h4><ul><li><p>한계 칸에 “참고용” 한 단어만 적었다.</p></li><li><p>아직 답이 없는 질문을 최종 표에서 삭제했다.</p></li></ul><h4>이해 여부를 확인할 질문</h4><p>이 표를 받는 다음 사람에게 미완료 질문도 같이 넘겨야 하는 이유는 무엇인가?</p><h4>인정할 수 있는 답의 핵심</h4><p>표가 완성됐다고 확인까지 끝난 것은 아니므로 남은 질문과 담당을 알아야 일을 이어갈 수 있다고 설명한다.</p><h4>보완과 재평가</h4><p>한계에 이유를 붙이고 삭제된 미완료 질문을 복구한 뒤 받는 사람이 내용을 찾게 한다.</p><p>수정 전 결과와 재평가 결과를 구분해 남긴다. 실제 하지 않은 보고·채점·시험을 나중에 했던 것처럼 채우지 않는다.</p>', 'attachments': []}], 'deliverables': ['analysis/ops07_uncertainty_assessment.md (필수) — 네 장소 위험표, 정보 부족 여덟 항목의 질문표, 열두 조합 담당표, 두 시험 상황, 적용 한계와 인계'], 'techStack': ['문서 편집기(Markdown · analysis/ops07_uncertainty_assessment.md) 또는 인쇄 양식', '4개 장소 민간 위험표 및 G1~G8 정보 부족·한계 분석표', '4×3(12조합) 확인 담당 매트릭스', 'T1·T2 예외 상황 시험표 및 적용 한계·인계 기록지'], 'evaluationKeys': ['<h4>평가 질문</h4><p>네 장소의 위험과 여덟 정보 부족 항목을 구체적으로 설명했는가</p><h4>평가자가 확인할 위치</h4><p>analysis/ops07_uncertainty_assessment.md의 STEP 01, STEP 02과 그 단계의 개인 기록을 확인한다.</p><h4>PASS를 주는 방법</h4><p>아래 단계별 항목을 모두 충족해야 이 평가 질문에 PASS를 준다. 항목 옆에 확인한 표·줄·회차를 적는다. 한 항목이라도 없거나 틀리면 FAIL 사유에 해당 단계와 문장을 적는다. 학습자가 쓴 말이 예시와 똑같을 필요는 없지만 확인할 내용은 모두 있어야 한다.</p><h4>STEP 01 민간 위험 신호와 모르는 내용을 따로 정리하기</h4><p>찾을 결과: 시장·병원·학교·창고 네 줄의 정리표.</p><ul><li><p>네 장소의 위험 신호와 모르는 내용을 따로 썼다.</p></li><li><p>표시를 고른 이유를 제공된 문장과 연결했다.</p></li><li><p>학교의 휴교를 인원 부재로 해석하지 않았다.</p></li><li><p>보이지 않음·자료 없음·낮은 번호를 안전 확정으로 바꾸지 않았다.</p></li></ul><h4>STEP 02 확인하지 못한 정보 여덟 가지에 질문과 한계를 붙이기</h4><p>찾을 결과: 여덟 항목의 확인 계획표. 각 줄에 질문·담당·답의 한계를 모두 적는다.</p><ul><li><p>서로 다른 여덟 항목을 모두 다뤘다.</p></li><li><p>각 항목에서 부족한 정보와 확인 질문이 서로 맞는다.</p></li><li><p>담당 역할과 요청할 답·자료가 구체적이다.</p></li><li><p>답을 받았을 때도 알 수 없는 내용을 적었다.</p></li></ul><h4>이런 결과는 통과시키지 않는다</h4><ul><li><p>학교를 “휴교이므로 민간인 없음”이라고 적었다.</p></li><li><p>내부 미관측의 해결책에 “추가 확인”만 적고 누구에게 무엇을 물을지 없다.</p></li></ul><h4>평가 의견 작성 방법</h4><p>“근거 부족”이라고만 쓰지 말고 단계 번호, 충족하지 못한 조건, 제출물에서 실제로 빠지거나 잘못된 부분, 다시 할 일을 적는다. 예를 들어 STEP 01의 첫 조건은 “네 장소의 위험 신호와 모르는 내용을 따로 썼다.”이다. 이 조건에 못 미쳤다면 어떤 줄·항목·참여자 기록이 빠졌는지 지목한다.</p><h4>다시 평가할 때</h4><p>틀리거나 빠진 내용만 구체적으로 알려 준다. 문서 작성 오류는 원자료와 수정본을 대조한다. 실제 수행 오류는 해당 행동을 다시 하게 하고 새 회차 기록으로 확인한다. 최초 결과는 지우지 않는다.</p>', '<h4>평가 질문</h4><p>열두 조합의 담당표를 시험하고 한계·남은 질문을 설명했는가</p><h4>평가자가 확인할 위치</h4><p>analysis/ops07_uncertainty_assessment.md의 STEP 03, STEP 04, STEP 05과 그 단계의 개인 기록을 확인한다.</p><h4>PASS를 주는 방법</h4><p>아래 단계별 항목을 모두 충족해야 이 평가 질문에 PASS를 준다. 항목 옆에 확인한 표·줄·회차를 적는다. 한 항목이라도 없거나 틀리면 FAIL 사유에 해당 단계와 문장을 적는다. 학습자가 쓴 말이 예시와 똑같을 필요는 없지만 확인할 내용은 모두 있어야 한다.</p><h4>STEP 03 위험 표시와 정보 부족 정도에 따라 담당 업무를 적기</h4><p>찾을 결과: 4×3 담당표 한 개. 열두 조합 각각에 일곱 항목을 적는다.</p><ul><li><p>위험 4종×정보 부족 3종의 열두 조합이 모두 있다.</p></li><li><p>열두 조합에 보류 상태·담당·질문·필요 자료·검토 담당이 있다.</p></li><li><p>정보 부족 높음인 네 조합이 모두 보류 유지·재확인이다.</p></li><li><p>낮은 번호 또는 담당자 이름만으로 실제 진행을 허용하는 문장이 없다.</p></li></ul><h4>STEP 04 정보가 빠지거나 검토자와 연락되지 않는 경우를 시험하기</h4><p>찾을 결과: T1·T2 시험 기록 두 줄과 담당표 수정 전후 기록. 수정이 없으면 없었던 이유를 쓴다.</p><ul><li><p>두 시험 상황을 모두 적용했다.</p></li><li><p>T1의 내부 이용·인원 미확인과 T2의 접수·회신 미확인을 남겼다.</p></li><li><p>누구에게 무엇을 다시 물을지 적었다.</p></li><li><p>연락 두절을 자동 위임이나 검토 완료로 바꾸지 않았다.</p></li></ul><h4>STEP 05 담당표로 알 수 없는 것을 밝히고 다음 미션에 넘기기</h4><p>찾을 결과: 적용 한계 세 가지 이상, 전원 설명 기록, 미완료 질문 인계표.</p><ul><li><p>판단할 수 없는 일 세 가지 이상에 각각 이유가 있다.</p></li><li><p>전원이 담당표 한 칸의 질문과 역할을 설명했다.</p></li><li><p>실제 승인·피해 계산을 이 표가 대신한다고 쓰지 않았다.</p></li><li><p>남은 질문마다 담당과 다음 확인할 내용이 있다.</p></li></ul><h4>이런 결과는 통과시키지 않는다</h4><ul><li><p>열두 칸에 직책 이름만 넣고 그 사람이 할 일이 없다.</p></li><li><p>T1에서 대피소 진술을 빼고 휴교 정보만 사용했다.</p></li><li><p>한계 칸에 “참고용” 한 단어만 적었다.</p></li></ul><h4>평가 의견 작성 방법</h4><p>“근거 부족”이라고만 쓰지 말고 단계 번호, 충족하지 못한 조건, 제출물에서 실제로 빠지거나 잘못된 부분, 다시 할 일을 적는다. 예를 들어 STEP 03의 첫 조건은 “위험 4종×정보 부족 3종의 열두 조합이 모두 있다.”이다. 이 조건에 못 미쳤다면 어떤 줄·항목·참여자 기록이 빠졌는지 지목한다.</p><h4>다시 평가할 때</h4><p>틀리거나 빠진 내용만 구체적으로 알려 준다. 문서 작성 오류는 원자료와 수정본을 대조한다. 실제 수행 오류는 해당 행동을 다시 하게 하고 새 회차 기록으로 확인한다. 최초 결과는 지우지 않는다.</p>'], 'evaluationTypes': ['REQUIRED', 'REQUIRED'], 'evaluationText': '<h4>미션 전체 판정</h4><p>이 미션은 아래 두 평가 질문이 모두 PASS여야 통과다. 조별 문서와 개인의 실제 수행을 나눠 확인한다. 수행하지 않았거나 기록이 없어 확인할 수 없는 사람은 미평가로 남기고 다시 실시한다. 최종 판정에서는 미평가를 PASS로 바꾸지 않는다.</p><h4>평가 질문</h4><p>네 장소의 위험과 여덟 정보 부족 항목을 구체적으로 설명했는가</p><h4>평가자가 확인할 위치</h4><p>analysis/ops07_uncertainty_assessment.md의 STEP 01, STEP 02과 그 단계의 개인 기록을 확인한다.</p><h4>PASS를 주는 방법</h4><p>아래 단계별 항목을 모두 충족해야 이 평가 질문에 PASS를 준다. 항목 옆에 확인한 표·줄·회차를 적는다. 한 항목이라도 없거나 틀리면 FAIL 사유에 해당 단계와 문장을 적는다. 학습자가 쓴 말이 예시와 똑같을 필요는 없지만 확인할 내용은 모두 있어야 한다.</p><h4>STEP 01 민간 위험 신호와 모르는 내용을 따로 정리하기</h4><p>찾을 결과: 시장·병원·학교·창고 네 줄의 정리표.</p><ul><li><p>네 장소의 위험 신호와 모르는 내용을 따로 썼다.</p></li><li><p>표시를 고른 이유를 제공된 문장과 연결했다.</p></li><li><p>학교의 휴교를 인원 부재로 해석하지 않았다.</p></li><li><p>보이지 않음·자료 없음·낮은 번호를 안전 확정으로 바꾸지 않았다.</p></li></ul><h4>STEP 02 확인하지 못한 정보 여덟 가지에 질문과 한계를 붙이기</h4><p>찾을 결과: 여덟 항목의 확인 계획표. 각 줄에 질문·담당·답의 한계를 모두 적는다.</p><ul><li><p>서로 다른 여덟 항목을 모두 다뤘다.</p></li><li><p>각 항목에서 부족한 정보와 확인 질문이 서로 맞는다.</p></li><li><p>담당 역할과 요청할 답·자료가 구체적이다.</p></li><li><p>답을 받았을 때도 알 수 없는 내용을 적었다.</p></li></ul><h4>이런 결과는 통과시키지 않는다</h4><ul><li><p>학교를 “휴교이므로 민간인 없음”이라고 적었다.</p></li><li><p>내부 미관측의 해결책에 “추가 확인”만 적고 누구에게 무엇을 물을지 없다.</p></li></ul><h4>평가 의견 작성 방법</h4><p>“근거 부족”이라고만 쓰지 말고 단계 번호, 충족하지 못한 조건, 제출물에서 실제로 빠지거나 잘못된 부분, 다시 할 일을 적는다. 예를 들어 STEP 01의 첫 조건은 “네 장소의 위험 신호와 모르는 내용을 따로 썼다.”이다. 이 조건에 못 미쳤다면 어떤 줄·항목·참여자 기록이 빠졌는지 지목한다.</p><h4>다시 평가할 때</h4><p>틀리거나 빠진 내용만 구체적으로 알려 준다. 문서 작성 오류는 원자료와 수정본을 대조한다. 실제 수행 오류는 해당 행동을 다시 하게 하고 새 회차 기록으로 확인한다. 최초 결과는 지우지 않는다.</p><h4>평가 질문</h4><p>열두 조합의 담당표를 시험하고 한계·남은 질문을 설명했는가</p><h4>평가자가 확인할 위치</h4><p>analysis/ops07_uncertainty_assessment.md의 STEP 03, STEP 04, STEP 05과 그 단계의 개인 기록을 확인한다.</p><h4>PASS를 주는 방법</h4><p>아래 단계별 항목을 모두 충족해야 이 평가 질문에 PASS를 준다. 항목 옆에 확인한 표·줄·회차를 적는다. 한 항목이라도 없거나 틀리면 FAIL 사유에 해당 단계와 문장을 적는다. 학습자가 쓴 말이 예시와 똑같을 필요는 없지만 확인할 내용은 모두 있어야 한다.</p><h4>STEP 03 위험 표시와 정보 부족 정도에 따라 담당 업무를 적기</h4><p>찾을 결과: 4×3 담당표 한 개. 열두 조합 각각에 일곱 항목을 적는다.</p><ul><li><p>위험 4종×정보 부족 3종의 열두 조합이 모두 있다.</p></li><li><p>열두 조합에 보류 상태·담당·질문·필요 자료·검토 담당이 있다.</p></li><li><p>정보 부족 높음인 네 조합이 모두 보류 유지·재확인이다.</p></li><li><p>낮은 번호 또는 담당자 이름만으로 실제 진행을 허용하는 문장이 없다.</p></li></ul><h4>STEP 04 정보가 빠지거나 검토자와 연락되지 않는 경우를 시험하기</h4><p>찾을 결과: T1·T2 시험 기록 두 줄과 담당표 수정 전후 기록. 수정이 없으면 없었던 이유를 쓴다.</p><ul><li><p>두 시험 상황을 모두 적용했다.</p></li><li><p>T1의 내부 이용·인원 미확인과 T2의 접수·회신 미확인을 남겼다.</p></li><li><p>누구에게 무엇을 다시 물을지 적었다.</p></li><li><p>연락 두절을 자동 위임이나 검토 완료로 바꾸지 않았다.</p></li></ul><h4>STEP 05 담당표로 알 수 없는 것을 밝히고 다음 미션에 넘기기</h4><p>찾을 결과: 적용 한계 세 가지 이상, 전원 설명 기록, 미완료 질문 인계표.</p><ul><li><p>판단할 수 없는 일 세 가지 이상에 각각 이유가 있다.</p></li><li><p>전원이 담당표 한 칸의 질문과 역할을 설명했다.</p></li><li><p>실제 승인·피해 계산을 이 표가 대신한다고 쓰지 않았다.</p></li><li><p>남은 질문마다 담당과 다음 확인할 내용이 있다.</p></li></ul><h4>이런 결과는 통과시키지 않는다</h4><ul><li><p>열두 칸에 직책 이름만 넣고 그 사람이 할 일이 없다.</p></li><li><p>T1에서 대피소 진술을 빼고 휴교 정보만 사용했다.</p></li><li><p>한계 칸에 “참고용” 한 단어만 적었다.</p></li></ul><h4>평가 의견 작성 방법</h4><p>“근거 부족”이라고만 쓰지 말고 단계 번호, 충족하지 못한 조건, 제출물에서 실제로 빠지거나 잘못된 부분, 다시 할 일을 적는다. 예를 들어 STEP 03의 첫 조건은 “위험 4종×정보 부족 3종의 열두 조합이 모두 있다.”이다. 이 조건에 못 미쳤다면 어떤 줄·항목·참여자 기록이 빠졌는지 지목한다.</p><h4>다시 평가할 때</h4><p>틀리거나 빠진 내용만 구체적으로 알려 준다. 문서 작성 오류는 원자료와 수정본을 대조한다. 실제 수행 오류는 해당 행동을 다시 하게 하고 새 회차 기록으로 확인한다. 최초 결과는 지우지 않는다.</p>', 'aiGuidelines': {'allow': ['4개 장소(시장 나-2·병원 다-2·학교 라-2·창고 마-3)의 민간 위험 신호와 미확인 사항 분리 검토받기', '정보 부족 8항목(G1~G8)별 확인 질문과 답변을 받아도 남는 한계 문장 다듬기', '위험 표시 4종×정보 부족 3종(12조합) 담당표 및 예외 상황(T1·T2) 처리 흐름 점검받기'], 'rules': ['AI 사용은 필수가 아니며, 결과는 analysis/ops07_uncertainty_assessment.md에 STEP 01~05 순서로 정리한다.', '정보 부족 "높음" 4조합은 모두 보류 유지·재확인으로 두고, 연락 두절(T2)을 자동 위임이나 검토 완료로 바꾸지 않는다.', 'STEP 05의 담당표로 판단할 수 없는 3가지 이상(실제 피해 추정·법적 비례성·교전 승인 등) 설명은 학습자가 직접 구술·기록한다.'], 'forbid': ['실제 부대·좌표·민간 시설 위치·교전수칙 원문을 입력하기', '학교 휴교를 인원 부재로 단정하거나 낮은 위험 번호를 안전 확정·타격 승인 기준으로 작성하기', '실제 피해 추정·무기 선정·법적 비례성 판단을 AI에게 요청하거나 대신 채우기']}}, {'id': 'OPS07_m04', 'sequence': 4, 'title': '식별이 불확실할 때 보류와 재확인 절차 작성', 'durationHours': 3, 'overview': '<p>확인 질문 열두 가지를 세 사례에 적용한다. 질문 일부에만 답이 온 경우도 처리할 절차를 쓴다.</p><p>아래 다섯 단계에는 필요한 자료·작업 순서·예시·제출 내용과 평가 조건이 들어 있다. 각 단계의 자료를 읽은 뒤 순서대로 작성한다.</p>', 'objectives': '<h4>이 미션을 마치면 할 수 있어야 하는 일</h4><ul><li><p>확인 질문 열두 가지를 세 사례에 적용한다.</p></li><li><p>질문 일부에만 답이 온 경우도 처리할 절차를 쓴다.</p></li><li><p>전원이 직접 보고하고 받은 사람이 다음 일을 다시 말한다.</p></li></ul>', 'scenario': '<h4>학습자의 역할</h4><p>학습자는 민간인과 아군에게 불필요한 피해가 생기지 않도록 보고를 읽고, 모르는 내용을 질문하고, 다음 담당자에게 기록을 넘기는 역할을 연습한다. 사람의 소속을 맞히거나 실제 타격을 결정하는 과제가 아니다.</p><h4>이번 미션의 과제</h4><ul><li><p>확인 질문 열두 가지를 세 사례에 적용한다.</p></li><li><p>질문 일부에만 답이 온 경우도 처리할 절차를 쓴다.</p></li><li><p>전원이 직접 보고하고 받은 사람이 다음 일을 다시 말한다.</p></li></ul><h4>자료와 역할의 범위</h4><p>사례·시각·장소·역할은 모두 교육용이다. 실제 부대 자료를 쓰지 않는다. 자료에 답이 없으면 자료에 없다고 적는다. 실제 행동 권한은 이 수업이 정하지 않는다.</p><h4>개인 작업과 조별 작업</h4><p>개인 판단을 먼저 요청한 단계에서는 토의 전에 자신의 답을 남긴다. 그 뒤 조별 답과 수정 이유를 별도로 쓴다. 보고·인계·시험 단계에서는 실제 한 말과 행동을 기록한다.</p>', 'attachments': [], 'materials': ['/uploads/OPS07/OPS07_m04_student.ipynb', '/uploads/OPS07/OPS07_m04_workbook.ipynb'], 'materialText': '<h4>실습 노트북 안내 (선택 보조 도구)</h4><p><strong>학생 실습용</strong>: <code>OPS07_m04_student.ipynb</code> (직접 ANSWER 표와 note를 작성합니다.)<br><strong>교수자·해설용</strong>: <code>OPS07_m04_workbook.ipynb</code> (예시 풀이와 해설 note가 포함되어 있습니다.)<br><em>※ 본 과제는 문서 편집기나 인쇄 양식으로도 수행할 수 있으며, 노트북은 사례·질문·기록 양식을 표로 정리하는 선택 보조 도구입니다.</em></p><h4>자료를 읽는 순서</h4>\n\n<ol>\n<li><p>먼저 S1~S3를 각각 읽습니다. 서로 다른 사례이므로 한 사례의 정보를 다른 사례에 옮기지 않습니다.</p></li>\n<li><p>STEP 1~2에서는 Q01~Q12를 읽고 세 사례의 자료가 어느 질문에 답하는지 확인합니다. 답이 없는 질문은 남겨 둡니다.</p></li>\n<li><p>STEP 3에서는 아래 다섯 순서를 참고해 절차를 씁니다. STEP 4~5에서는 절차를 검토하고 실제 보고와 인계를 연습합니다.</p></li>\n</ol>\n\n<h4>자료 1. 식별이 불확실한 세 사례</h4>\n\n<h4>S1</h4>\n\n<p>건물 옥상 인원 2명.</p>\n\n<p>무전기로 보이는 물체, 민간 복장.</p>\n\n<p>영상 1회뿐이고 다른 관측자는 시야 밖이다.</p>\n\n<p>물체의 기능과 인원의 지위는 확인되지 않았다.</p>\n\n<h4>S2</h4>\n\n<p>차량 표식을 영상과 육안에서 같은 것으로 보고했다.</p>\n\n<p>주변 민간 차량은 관찰되지 않았다.</p>\n\n<p>차량 표식의 진위·아군 위치·관측 범위·정보의 최신성·검토자 회신은 미제공이다.</p>\n\n<h4>S3</h4>\n\n<p>골목 인원 4명이 아군 복장과 비슷한 옷을 입었다.</p>\n\n<p>정찰조와 연락이 15분째 되지 않으며 마지막 보고 구역과 겹친다.</p>\n\n<p>비슷한 복장만으로 아군이라고 확정할 수 없다.</p>\n\n<h4>자료 2. 사례마다 확인할 열두 질문</h4>\n\n<ul>\n<li><p>Q01. 처음 관찰하거나 제보한 출처는 누구인가?</p></li>\n<li><p>Q02. 실제 관측 시각은 언제인가?</p></li>\n<li><p>Q03. 보고를 보내고 받은 시각은 언제인가?</p></li>\n<li><p>Q04. 관찰한 장소는 어디이며 표기가 서로 같은가?</p></li>\n<li><p>Q05. 직접 본 장면이나 물체는 무엇인가?</p></li>\n<li><p>Q06. 소속·신분·의도는 무엇을 근거로 설명했는가?</p></li>\n<li><p>Q07. 다른 보고는 별도로 관찰한 것인가?</p></li>\n<li><p>Q08. 자료가 중간에 끊긴 부분이 있는가?</p></li>\n<li><p>Q09. 현재 아군 위치와 충돌하는 정보가 있는가?</p></li>\n<li><p>Q10. 민간인 이동·시설 이용에 관한 정보가 있는가?</p></li>\n<li><p>Q11. 서로 다른 설명이나 모순이 있는가?</p></li>\n<li><p>Q12. 질문을 맡은 담당과 받은 답의 범위가 분명한가?</p></li>\n</ul>\n\n<h4>자료 3. 보류와 재확인의 다섯 순서</h4>\n\n<ol>\n<li><p>위험을 알리고 보류 상태를 기록한다.</p></li>\n<li><p>알고 있는 사실과 모르는 내용을 보고한다.</p></li>\n<li><p>확인할 질문·담당을 정하고 요청 접수를 확인한다.</p></li>\n<li><p>회신이 어떤 질문에 답했는지 원문과 비교한다.</p></li>\n<li><p>남은 질문과 담당·다음 확인을 기록해 인계한다.</p></li>\n</ol>\n\n<h4>자료를 사용할 때 지킬 점</h4>\n\n<p>표식이나 복장이 같다는 보고만으로 소속을 확정하지 않습니다.</p>\n\n<p>요청을 접수했다는 답과 확인 질문에 대한 답도 구별합니다.</p>\n\n<p>일부 질문에만 답이 왔다면 나머지는 미답으로 남깁니다.</p>\n\n<p>모든 사례·시각·장소·역할은 교육용입니다. 필요한 자료와 구체적인 작업 순서는 각 단계 설명에도 들어 있습니다.</p>\n', 'guideSteps': [{'step': 1, 'problem': '사례를 읽을 때 확인할 질문 열두 가지를 만들기', 'activity': '<h4>이번 단계에서 하는 일</h4><p>“확신이 높다”는 말보다 어떤 정보를 확인했는지 알려 주는 편이 정확하다. 사람의 소속을 점수로 맞히지 않고 필요한 질문을 정리한다.</p><h4>먼저 읽을 자료</h4>\n\n<h4>확인할 질문</h4>\n\n<ul>\n<li>\n<p><strong>Q01:</strong> 처음 관찰하거나 제보한 출처는 누구인가</p>\n</li>\n\n<li>\n<p><strong>Q02:</strong> 실제 관측 시각은 언제인가</p>\n</li>\n\n<li>\n<p><strong>Q03:</strong> 보고를 보내고 받은 시각은 언제인가</p>\n</li>\n\n<li>\n<p><strong>Q04:</strong> 관찰한 장소는 어디이며 표기가 서로 같은가</p>\n</li>\n\n<li>\n<p><strong>Q05:</strong> 직접 본 장면이나 물체는 무엇인가</p>\n</li>\n\n<li>\n<p><strong>Q06:</strong> 소속·신분·의도는 무엇을 근거로 설명했는가</p>\n</li>\n\n<li>\n<p><strong>Q07:</strong> 다른 보고는 별도로 관찰한 것인가</p>\n</li>\n\n<li>\n<p><strong>Q08:</strong> 자료가 중간에 끊긴 부분이 있는가</p>\n</li>\n\n<li>\n<p><strong>Q09:</strong> 현재 아군 위치와 충돌하는 정보가 있는가</p>\n</li>\n\n<li>\n<p><strong>Q10:</strong> 민간인 이동·시설 이용에 관한 정보가 있는가</p>\n</li>\n\n<li>\n<p><strong>Q11:</strong> 서로 다른 설명이나 모순이 있는가</p>\n</li>\n\n<li>\n<p><strong>Q12:</strong> 질문을 맡은 담당과 받은 답의 범위가 분명한가</p>\n</li>\n</ul>\n\n<p>이 질문은 보고를 검토하는 질문이지 적·아군·민간을 자동 분류하는 규칙이 아니다.</p>\n\n<p>복장이나 물체 한 가지로 신분을 확정하지 않는다.</p>\n\n<h4>작업 순서</h4><ol><li><p>열두 질문을 옮기고 어려운 표현을 조원이 이해할 말로 풀어 쓴다.</p></li><li><p>질문마다 어떤 자료나 답을 보면 답한 것으로 볼지 적는다.</p></li><li><p>답이 자료에 없을 때 쓸 표현을 정한다. “자료에 없음”, “아직 회신 없음”, “서로 다름”을 구별한다.</p></li><li><p>질문이 “안전한가”처럼 넓다면 대상·시각·내용을 넣어 더 좁힌다.</p></li><li><p>질문표 아래에 “답한 질문 수는 문서 누락을 찾는 데만 쓰며 신분 판정이나 실제 진행 허가에 쓰지 않는다”고 적는다.</p></li></ol><h4>작성 예시</h4><p>다음은 쓰는 방식을 보여 주는 예다. 자기 판단이나 실제 수행·측정 결과가 필요한 칸에 그대로 복사하지 않는다.</p><p>Q04 장소 표기 / 필요한 자료 “원 보고의 장소 표기와 수신 기록” / 답을 못한 상태 “두 표기가 달라 원 보고자에게 재질문 중”.</p><h4>작성할 표 또는 기록 항목</h4><ul><li><p>질문 번호</p></li><li><p>풀어 쓴 질문</p></li><li><p>필요한 자료 또는 답</p></li><li><p>답이 없거나 서로 다를 때 적을 말</p></li></ul><h4>제출할 내용과 끝내는 기준</h4><p>서로 다른 확인 질문 열두 개 이상과 질문표 사용 범위 설명.</p><p>미션 제출 문서 sop/ops07_reconfirmation_procedure.md의 STEP 01 아래에 작성한다. 개인 기록과 조별 결과를 구분하고 실제 이름 대신 학습자 코드를 써도 된다. 표의 모양이 달라도 필요한 내용을 찾을 수 있으면 된다.</p><h4>마치기 전 확인</h4><ul><li><p>서로 다른 질문 열두 개 이상이 있다.</p></li><li><p>출처·시각·장소·관측 내용·아군·민간·모순·담당을 모두 다룬다.</p></li><li><p>질문마다 필요한 자료와 미확인 표시가 있다.</p></li><li><p>확인 개수나 복장·표식만으로 신분·진행을 판정하지 않는다.</p></li></ul><h4>배정 시간</h4><p>35분</p>', 'evalText': '<h4>평가할 자료와 위치</h4><p>sop/ops07_reconfirmation_procedure.md의 STEP 01 결과와 해당 개인 수행 기록을 확인한다. 이 단계에서 요청한 제출 내용은 다음과 같다: 서로 다른 확인 질문 열두 개 이상과 질문표 사용 범위 설명.</p><h4>평가 순서</h4><ol><li><p>아래 통과 조건을 하나씩 읽고 제출물에서 확인한 위치를 옆에 적는다.</p></li><li><p>실제 설명·보고·인계가 필요한 단계는 수행 기록이나 직접 관찰로 확인한다. 서식만 있으면 수행한 것으로 보지 않는다.</p></li><li><p>통과 조건을 모두 충족하면 이 단계는 PASS다. 하나라도 빠지거나 반대로 적었으면 FAIL로 두고 빠진 항목을 구체적으로 적는다. 표현이나 표 모양이 예시와 달라도 뜻이 같고 근거를 찾을 수 있으면 인정한다.</p></li></ol><h4>PASS 조건</h4><ul><li><p>서로 다른 질문 열두 개 이상이 있다.</p></li><li><p>출처·시각·장소·관측 내용·아군·민간·모순·담당을 모두 다룬다.</p></li><li><p>질문마다 필요한 자료와 미확인 표시가 있다.</p></li><li><p>확인 개수나 복장·표식만으로 신분·진행을 판정하지 않는다.</p></li></ul><h4>FAIL에 해당하는 예</h4><ul><li><p>열두 줄이 있지만 질문이 모두 “확인됐는가”이다.</p></li><li><p>열두 항목 중 열 개를 채우면 진행한다는 규칙을 만들었다.</p></li></ul><h4>이해 여부를 확인할 질문</h4><p>확인 질문 열 개에 답했어도 남은 질문이 중요하면 어떻게 하는가?</p><h4>인정할 수 있는 답의 핵심</h4><p>남은 질문을 확인하지 않은 것으로 남기고 필요한 담당에게 다시 물으며 답한 개수만으로 진행 판단을 하지 않는다고 답한다.</p><h4>보완과 재평가</h4><p>겹치는 질문을 나누고 각 질문에 필요한 자료를 적는다. 개수에 따른 진행 규칙은 제거한다.</p><p>수정 전 결과와 재평가 결과를 구분해 남긴다. 실제 하지 않은 보고·채점·시험을 나중에 했던 것처럼 채우지 않는다.</p>', 'attachments': []}, {'step': 2, 'problem': '세 사례에 열두 질문을 적용해 모르는 내용을 찾기', 'activity': '<h4>이번 단계에서 하는 일</h4><p>질문표를 만드는 것과 실제 보고를 읽어 답을 찾는 것은 다르다. 세 사례에서 답할 수 있는 것과 없는 것을 직접 나눈다.</p><h4>먼저 읽을 자료</h4>\n\n<h4>사례 자료</h4>\n\n<p><strong>S1</strong></p>\n\n<p>건물 옥상 인원 2명.</p>\n\n<p>무전기로 보이는 물체, 민간 복장.</p>\n\n<p>영상 1회뿐이고 다른 관측자는 시야 밖이다.</p>\n\n<p>물체의 기능과 인원의 지위는 확인되지 않았다.</p>\n\n<p><strong>S2</strong></p>\n\n<p>차량 표식을 영상과 육안에서 같은 것으로 보고했다.</p>\n\n<p>주변 민간 차량은 관찰되지 않았다.</p>\n\n<p>차량 표식의 진위·아군 위치·관측 범위·정보의 최신성·검토자 회신은 미제공이다.</p>\n\n<p><strong>S3</strong></p>\n\n<p>골목 인원 4명이 아군 복장과 비슷한 옷을 입었다.</p>\n\n<p>정찰조와 연락이 15분째 되지 않으며 마지막 보고 구역과 겹친다.</p>\n\n<p>비슷한 복장만으로 아군이라고 확정할 수 없다.</p>\n\n<h4>확인할 질문</h4>\n\n<ul>\n<li>\n<p><strong>Q01:</strong> 처음 관찰하거나 제보한 출처는 누구인가</p>\n</li>\n\n<li>\n<p><strong>Q02:</strong> 실제 관측 시각은 언제인가</p>\n</li>\n\n<li>\n<p><strong>Q03:</strong> 보고를 보내고 받은 시각은 언제인가</p>\n</li>\n\n<li>\n<p><strong>Q04:</strong> 관찰한 장소는 어디이며 표기가 서로 같은가</p>\n</li>\n\n<li>\n<p><strong>Q05:</strong> 직접 본 장면이나 물체는 무엇인가</p>\n</li>\n\n<li>\n<p><strong>Q06:</strong> 소속·신분·의도는 무엇을 근거로 설명했는가</p>\n</li>\n\n<li>\n<p><strong>Q07:</strong> 다른 보고는 별도로 관찰한 것인가</p>\n</li>\n\n<li>\n<p><strong>Q08:</strong> 자료가 중간에 끊긴 부분이 있는가</p>\n</li>\n\n<li>\n<p><strong>Q09:</strong> 현재 아군 위치와 충돌하는 정보가 있는가</p>\n</li>\n\n<li>\n<p><strong>Q10:</strong> 민간인 이동·시설 이용에 관한 정보가 있는가</p>\n</li>\n\n<li>\n<p><strong>Q11:</strong> 서로 다른 설명이나 모순이 있는가</p>\n</li>\n\n<li>\n<p><strong>Q12:</strong> 질문을 맡은 담당과 받은 답의 범위가 분명한가</p>\n</li>\n</ul>\n\n<h4>작업 순서</h4><ol><li><p>S1~S3 각각에 Q01~Q12를 붙인다. 기본 질문표를 사용하면 모두 36줄이다.</p></li><li><p>자료에 답이 있으면 카드 문장을 옮기고 어디까지 알 수 있는지 적는다.</p></li><li><p>답이 없으면 “자료에 없음”이라고 쓰고 누구에게 무엇을 물을지 적는다.</p></li><li><p>두 자료가 다르다면 “서로 다름”이라고 쓰고 다른 문장을 함께 남긴다.</p></li><li><p>S2의 표식 일치가 현재 아군 위치·민간 상황·검토 회신까지 설명하는지 살핀다. 설명하지 않으면 그 질문은 그대로 미확인이다.</p></li></ol><h4>작성 예시</h4><p>다음은 쓰는 방식을 보여 주는 예다. 자기 판단이나 실제 수행·측정 결과가 필요한 칸에 그대로 복사하지 않는다.</p><p>S2·Q09 현재 아군 위치 / 답 “자료에 없음” / 질문 “아군 위치 보고 담당자에게 이 시점에 적용할 현재 보고가 있는지 요청” / 표식 두 수단 일치와는 별개로 남김.</p><h4>작성할 표 또는 기록 항목</h4><ul><li><p>사례·질문 번호</p></li><li><p>자료에서 찾은 답</p></li><li><p>확인할 수 있는 범위 또는 미확인 이유</p></li><li><p>추가 질문</p></li><li><p>확인 담당</p></li></ul><h4>제출할 내용과 끝내는 기준</h4><p>세 사례×열두 질문의 적용표. 질문을 추가했다면 추가 질문도 세 사례에 적용한다.</p><p>미션 제출 문서 sop/ops07_reconfirmation_procedure.md의 STEP 02 아래에 작성한다. 개인 기록과 조별 결과를 구분하고 실제 이름 대신 학습자 코드를 써도 된다. 표의 모양이 달라도 필요한 내용을 찾을 수 있으면 된다.</p><h4>마치기 전 확인</h4><ul><li><p>세 사례 모두 열두 질문에 답이나 미확인 이유가 있다.</p></li><li><p>없는 관측 시각·소속·회신을 만들지 않았다.</p></li><li><p>S1 물체·S3 복장만으로 소속을 확정하지 않았다.</p></li><li><p>S2에 아군 위치·민간 상황·검토 회신의 미확인이 남는다.</p></li></ul><h4>배정 시간</h4><p>40분</p>', 'evalText': '<h4>평가할 자료와 위치</h4><p>sop/ops07_reconfirmation_procedure.md의 STEP 02 결과와 해당 개인 수행 기록을 확인한다. 이 단계에서 요청한 제출 내용은 다음과 같다: 세 사례×열두 질문의 적용표. 질문을 추가했다면 추가 질문도 세 사례에 적용한다.</p><h4>평가 순서</h4><ol><li><p>아래 통과 조건을 하나씩 읽고 제출물에서 확인한 위치를 옆에 적는다.</p></li><li><p>실제 설명·보고·인계가 필요한 단계는 수행 기록이나 직접 관찰로 확인한다. 서식만 있으면 수행한 것으로 보지 않는다.</p></li><li><p>통과 조건을 모두 충족하면 이 단계는 PASS다. 하나라도 빠지거나 반대로 적었으면 FAIL로 두고 빠진 항목을 구체적으로 적는다. 표현이나 표 모양이 예시와 달라도 뜻이 같고 근거를 찾을 수 있으면 인정한다.</p></li></ol><h4>PASS 조건</h4><ul><li><p>세 사례 모두 열두 질문에 답이나 미확인 이유가 있다.</p></li><li><p>없는 관측 시각·소속·회신을 만들지 않았다.</p></li><li><p>S1 물체·S3 복장만으로 소속을 확정하지 않았다.</p></li><li><p>S2에 아군 위치·민간 상황·검토 회신의 미확인이 남는다.</p></li></ul><h4>FAIL에 해당하는 예</h4><ul><li><p>S2는 표식이 같다는 이유로 나머지 질문을 모두 확인으로 채웠다.</p></li><li><p>S3의 복장이 비슷하므로 아군이라고 결론냈다.</p></li></ul><h4>이해 여부를 확인할 질문</h4><p>S2에서 두 수단이 같은 표식을 봤다는 정보가 답하지 못하는 질문 한 가지를 말해 보라.</p><h4>인정할 수 있는 답의 핵심</h4><p>현재 아군 위치, 민간 상황, 표식 진위, 시각 또는 검토 회신 중 하나를 구체적으로 설명한다.</p><h4>보완과 재평가</h4><p>확인으로 표시한 각 줄에 실제 카드 문장을 붙여 보고 근거가 없는 줄을 미확인으로 고친다.</p><p>수정 전 결과와 재평가 결과를 구분해 남긴다. 실제 하지 않은 보고·채점·시험을 나중에 했던 것처럼 채우지 않는다.</p>', 'attachments': []}, {'step': 3, 'problem': '보류 보고부터 회신 확인과 인계까지 절차를 쓰기', 'activity': '<h4>이번 단계에서 하는 일</h4><p>“보고했다”는 말만으로 상대가 요청을 받았는지 알 수 없다. 누가 어떤 행동을 하고 무엇을 기록해야 하는지 다섯 순서로 풀어 쓴다.</p><h4>먼저 읽을 자료</h4>\n\n<h4>업무를 진행할 순서</h4>\n\n<ol>\n<li>\n<p>위험을 알리고 보류 상태를 기록한다</p>\n</li>\n\n<li>\n<p>알고 있는 사실과 모르는 내용을 보고한다</p>\n</li>\n\n<li>\n<p>확인할 질문·담당을 정하고 요청 접수를 확인한다</p>\n</li>\n\n<li>\n<p>회신이 어떤 질문에 답했는지 원문과 비교한다</p>\n</li>\n\n<li>\n<p>남은 질문과 담당·다음 확인을 기록해 인계한다</p>\n</li>\n</ol>\n\n<p>보류는 이 연습에서 확인되지 않은 내용을 확인한 것처럼 처리하지 않고 검토를 멈춰 두는 상태다.</p>\n\n<p>재확인은 모르는 내용을 담당자에게 다시 묻는 일이다.</p>\n\n<p>접수는 상대가 요청을 받았다는 뜻이고, 회신은 질문에 대한 답이다.</p>\n\n<p>접수됐다고 해서 질문의 답까지 받은 것은 아니다.</p>\n\n<p>아래의 역할은 모두 교육용이며 실제 지휘권을 뜻하지 않는다.</p>\n\n<p>절차를 시험할 간단한 상황: 질문 두 개를 보냈는데 받은 답에는 장소 표기 한 가지만 적혀 있다.</p>\n\n<p>인원의 소속 질문에는 아직 답이 없다.</p>\n\n<h4>작업 순서</h4><ol><li><p>다섯 순서를 표의 행으로 만든다.</p></li><li><p>각 행에 일을 맡을 가상 역할, 그 사람이 할 말이나 행동, 기록할 내용을 적는다.</p></li><li><p>3번에는 요청한 질문과 상대가 받았는지 확인할 방법을 넣는다.</p></li><li><p>4번에는 질문별로 답한 내용·아직 답하지 않은 내용을 나누는 칸을 넣는다.</p></li><li><p>답이 없거나 일부만 답하면 어떻게 남길지 적는다. “기다림”만 쓰지 말고 남은 질문·담당·다음 확인 계획을 적는다.</p></li><li><p>제공한 두 질문·부분 회신 상황을 적용해 미답 질문이 사라지지 않는지 확인한다.</p></li></ol><h4>작성 예시</h4><p>다음은 쓰는 방식을 보여 주는 예다. 자기 판단이나 실제 수행·측정 결과가 필요한 칸에 그대로 복사하지 않는다.</p><p>4번 회신 확인 / 기록자 / “장소 질문에는 답을 받았으나 소속 질문에는 답이 없다”고 나눔 / 보류 유지·소속 질문 담당과 다음 확인 계획을 기록.</p><h4>작성할 표 또는 기록 항목</h4><ul><li><p>절차 순서</p></li><li><p>담당 역할</p></li><li><p>실제로 할 말·행동</p></li><li><p>기록할 내용</p></li><li><p>답이 없거나 일부만 답했을 때 할 일</p></li></ul><h4>제출할 내용과 끝내는 기준</h4><p>다섯 순서의 절차표와 부분 회신 시험 기록 1개.</p><p>미션 제출 문서 sop/ops07_reconfirmation_procedure.md의 STEP 03 아래에 작성한다. 개인 기록과 조별 결과를 구분하고 실제 이름 대신 학습자 코드를 써도 된다. 표의 모양이 달라도 필요한 내용을 찾을 수 있으면 된다.</p><h4>마치기 전 확인</h4><ul><li><p>다섯 순서 각각에 담당·행동·기록이 있다.</p></li><li><p>송신·접수·회신을 구별한다.</p></li><li><p>부분 회신에서 답하지 않은 질문을 별도로 남겼다.</p></li><li><p>미답 질문의 담당과 다음 확인 계획이 있고 보류를 자동 해제하지 않는다.</p></li></ul><h4>배정 시간</h4><p>40분</p>', 'evalText': '<h4>평가할 자료와 위치</h4><p>sop/ops07_reconfirmation_procedure.md의 STEP 03 결과와 해당 개인 수행 기록을 확인한다. 이 단계에서 요청한 제출 내용은 다음과 같다: 다섯 순서의 절차표와 부분 회신 시험 기록 1개.</p><h4>평가 순서</h4><ol><li><p>아래 통과 조건을 하나씩 읽고 제출물에서 확인한 위치를 옆에 적는다.</p></li><li><p>실제 설명·보고·인계가 필요한 단계는 수행 기록이나 직접 관찰로 확인한다. 서식만 있으면 수행한 것으로 보지 않는다.</p></li><li><p>통과 조건을 모두 충족하면 이 단계는 PASS다. 하나라도 빠지거나 반대로 적었으면 FAIL로 두고 빠진 항목을 구체적으로 적는다. 표현이나 표 모양이 예시와 달라도 뜻이 같고 근거를 찾을 수 있으면 인정한다.</p></li></ol><h4>PASS 조건</h4><ul><li><p>다섯 순서 각각에 담당·행동·기록이 있다.</p></li><li><p>송신·접수·회신을 구별한다.</p></li><li><p>부분 회신에서 답하지 않은 질문을 별도로 남겼다.</p></li><li><p>미답 질문의 담당과 다음 확인 계획이 있고 보류를 자동 해제하지 않는다.</p></li></ul><h4>FAIL에 해당하는 예</h4><ul><li><p>“보고→확인→완료”만 적고 접수와 회신 내용이 없다.</p></li><li><p>두 질문 중 하나에 답이 왔으므로 모두 확인됐다고 처리했다.</p></li></ul><h4>이해 여부를 확인할 질문</h4><p>장소 질문에만 답이 왔으면 소속 질문은 어떻게 적는가?</p><h4>인정할 수 있는 답의 핵심</h4><p>아직 답이 없다고 별도로 적고 담당과 다음 확인을 정하며 장소 확인과 섞지 않는다고 답한다.</p><h4>보완과 재평가</h4><p>질문별 답변 칸을 만들고 부분 회신 상황을 다시 적용해 누락된 질문을 복구한다.</p><p>수정 전 결과와 재평가 결과를 구분해 남긴다. 실제 하지 않은 보고·채점·시험을 나중에 했던 것처럼 채우지 않는다.</p>', 'attachments': []}, {'step': 4, 'problem': '다른 조원의 질문을 받고 절차를 고치기', 'activity': '<h4>이번 단계에서 하는 일</h4><p>자신이 쓴 문장은 자신에게만 명확할 수 있다. 다른 사람이 읽고 납득되지 않는 부분을 묻게 해 절차를 고친다.</p><h4>먼저 읽을 자료</h4>\n\n<h4>사례 자료</h4>\n\n<p><strong>S1</strong></p>\n\n<p>건물 옥상 인원 2명.</p>\n\n<p>무전기로 보이는 물체, 민간 복장.</p>\n\n<p>영상 1회뿐이고 다른 관측자는 시야 밖이다.</p>\n\n<p>물체의 기능과 인원의 지위는 확인되지 않았다.</p>\n\n<p><strong>S2</strong></p>\n\n<p>차량 표식을 영상과 육안에서 같은 것으로 보고했다.</p>\n\n<p>주변 민간 차량은 관찰되지 않았다.</p>\n\n<p>차량 표식의 진위·아군 위치·관측 범위·정보의 최신성·검토자 회신은 미제공이다.</p>\n\n<p><strong>S3</strong></p>\n\n<p>골목 인원 4명이 아군 복장과 비슷한 옷을 입었다.</p>\n\n<p>정찰조와 연락이 15분째 되지 않으며 마지막 보고 구역과 겹친다.</p>\n\n<p>비슷한 복장만으로 아군이라고 확정할 수 없다.</p>\n\n<p>검토자는 S1~S3에서 소속을 단정할 근거가 있는지, 보고를 받았는지, 답이 일부만 왔을 때 무엇을 할지 묻는다.</p>\n\n<p>검토자를 이 수업에서는 반대 관점 검토자라고 부른다.</p>\n\n<h4>작업 순서</h4><ol><li><p>조 안에서 작성자와 검토자를 정한다. 뒤에는 역할을 바꾼다.</p></li><li><p>검토자는 S1·S2·S3 각각에 질문을 하나 이상 만든다. “그 물체만으로 소속을 알 수 있나”처럼 실제 문장을 겨냥한다.</p></li><li><p>작성자는 자신이 사용한 자료와 아직 모르는 내용을 말한다.</p></li><li><p>절차를 바꾸면 바꾸기 전·후 문장과 이유를 적는다. 바꾸지 않으면 질문에 답할 수 있는 기존 문장과 이유를 적는다.</p></li><li><p>질문 일부에 답이 없을 때 남은 질문을 기록하는 문장이 있는지 마지막으로 확인한다.</p></li></ol><h4>작성 예시</h4><p>다음은 쓰는 방식을 보여 주는 예다. 자기 판단이나 실제 수행·측정 결과가 필요한 칸에 그대로 복사하지 않는다.</p><p>검토 질문 “S1의 무전기로 보이는 물체가 왜 소속의 근거가 됩니까?” / 작성자 답 “물체의 기능과 소속은 모릅니다” / 수정 “물체 관찰과 소속 주장을 다른 칸에 기록한다.”</p><h4>작성할 표 또는 기록 항목</h4><ul><li><p>사례 번호</p></li><li><p>검토자가 한 질문</p></li><li><p>작성자가 답한 내용</p></li><li><p>수정 전·후 또는 유지한 문장</p></li><li><p>그 이유</p></li></ul><h4>제출할 내용과 끝내는 기준</h4><p>사례마다 한 건 이상, 모두 세 건 이상의 검토 질문과 답·수정 기록.</p><p>미션 제출 문서 sop/ops07_reconfirmation_procedure.md의 STEP 04 아래에 작성한다. 개인 기록과 조별 결과를 구분하고 실제 이름 대신 학습자 코드를 써도 된다. 표의 모양이 달라도 필요한 내용을 찾을 수 있으면 된다.</p><h4>마치기 전 확인</h4><ul><li><p>S1·S2·S3 각각에 검토 질문이 있다.</p></li><li><p>질문에 실제 자료와 모르는 내용을 근거로 답했다.</p></li><li><p>수정 전후 또는 기존 문장을 유지한 이유가 있다.</p></li><li><p>질문 일부에 답이 없을 때의 기록 방법이 최종 절차에 남는다.</p></li></ul><h4>배정 시간</h4><p>35분</p>', 'evalText': '<h4>평가할 자료와 위치</h4><p>sop/ops07_reconfirmation_procedure.md의 STEP 04 결과와 해당 개인 수행 기록을 확인한다. 이 단계에서 요청한 제출 내용은 다음과 같다: 사례마다 한 건 이상, 모두 세 건 이상의 검토 질문과 답·수정 기록.</p><h4>평가 순서</h4><ol><li><p>아래 통과 조건을 하나씩 읽고 제출물에서 확인한 위치를 옆에 적는다.</p></li><li><p>실제 설명·보고·인계가 필요한 단계는 수행 기록이나 직접 관찰로 확인한다. 서식만 있으면 수행한 것으로 보지 않는다.</p></li><li><p>통과 조건을 모두 충족하면 이 단계는 PASS다. 하나라도 빠지거나 반대로 적었으면 FAIL로 두고 빠진 항목을 구체적으로 적는다. 표현이나 표 모양이 예시와 달라도 뜻이 같고 근거를 찾을 수 있으면 인정한다.</p></li></ol><h4>PASS 조건</h4><ul><li><p>S1·S2·S3 각각에 검토 질문이 있다.</p></li><li><p>질문에 실제 자료와 모르는 내용을 근거로 답했다.</p></li><li><p>수정 전후 또는 기존 문장을 유지한 이유가 있다.</p></li><li><p>질문 일부에 답이 없을 때의 기록 방법이 최종 절차에 남는다.</p></li></ul><h4>FAIL에 해당하는 예</h4><ul><li><p>“검토 완료·이견 없음”만 적고 실제 질문이 없다.</p></li><li><p>검토에 답하려고 카드에 없는 추가 관찰을 만들었다.</p></li></ul><h4>이해 여부를 확인할 질문</h4><p>검토자가 의문을 제기했지만 문장을 안 바꿨다면 무엇을 남겨야 하는가?</p><h4>인정할 수 있는 답의 핵심</h4><p>기존 문장이 그 질문에 어떻게 답하는지 해당 문장과 유지 이유를 남겨야 한다고 답한다.</p><h4>보완과 재평가</h4><p>각 사례의 실제 검토 질문을 다시 받고 자료로 답한 내용과 수정 판단을 기록한다.</p><p>수정 전 결과와 재평가 결과를 구분해 남긴다. 실제 하지 않은 보고·채점·시험을 나중에 했던 것처럼 채우지 않는다.</p>', 'attachments': []}, {'step': 5, 'problem': '각자 한 사례를 보고하고 다음 사람에게 일을 넘기기', 'activity': '<h4>이번 단계에서 하는 일</h4><p>다음 사람이 이어서 확인할 수 있게 말하는 것이 이번 단계의 목표다. 서식만 채우지 말고 직접 보고하고 받은 사람이 이해했는지 확인한다.</p><h4>먼저 읽을 자료</h4>\n\n<h4>사례 자료</h4>\n\n<p><strong>S1</strong></p>\n\n<p>건물 옥상 인원 2명.</p>\n\n<p>무전기로 보이는 물체, 민간 복장.</p>\n\n<p>영상 1회뿐이고 다른 관측자는 시야 밖이다.</p>\n\n<p>물체의 기능과 인원의 지위는 확인되지 않았다.</p>\n\n<p><strong>S2</strong></p>\n\n<p>차량 표식을 영상과 육안에서 같은 것으로 보고했다.</p>\n\n<p>주변 민간 차량은 관찰되지 않았다.</p>\n\n<p>차량 표식의 진위·아군 위치·관측 범위·정보의 최신성·검토자 회신은 미제공이다.</p>\n\n<p><strong>S3</strong></p>\n\n<p>골목 인원 4명이 아군 복장과 비슷한 옷을 입었다.</p>\n\n<p>정찰조와 연락이 15분째 되지 않으며 마지막 보고 구역과 겹친다.</p>\n\n<p>비슷한 복장만으로 아군이라고 확정할 수 없다.</p>\n\n<p>보고할 여섯 내용:</p>\n\n<ol>\n<li>\n<p>① 사례 번호</p>\n</li>\n\n<li>\n<p>② 실제 보고된 장면</p>\n</li>\n\n<li>\n<p>③ 모르는 내용</p>\n</li>\n\n<li>\n<p>④ 보류 이유</p>\n</li>\n\n<li>\n<p>⑤ 확인 질문과 담당</p>\n</li>\n\n<li>\n<p>⑥ 아직 받은 답이 있는지. 자료에 회신이 없으면 없다고 말한다.</p>\n</li>\n</ol>\n\n<h4>작업 순서</h4><ol><li><p>각자 S1~S3 중 하나를 고르고 여섯 내용을 개인 메모에 쓴다.</p></li><li><p>두 사람씩 짝을 지어 한 사람은 보고하고 다른 사람은 인계받는다.</p></li><li><p>받는 사람은 “제가 맡을 일은 …이며 아직 …의 답이 없습니다”라고 다시 말한다.</p></li><li><p>보고자는 잘못 이해한 부분이 있으면 바로잡고 기록자는 실제 오간 내용을 적는다.</p></li><li><p>역할을 바꿔 전원이 보고와 인계를 경험하게 한다. 빠진 사람이 없는지 명단과 비교한다.</p></li></ol><h4>작성 예시</h4><p>다음은 쓰는 방식을 보여 주는 예다. 자기 판단이나 실제 수행·측정 결과가 필요한 칸에 그대로 복사하지 않는다.</p><p>“S3는 비슷한 복장의 4명이 있다는 보고입니다. 정찰조의 현재 위치는 모릅니다. 소속을 단정하지 않고 위치 담당자에게 현재 보고를 요청하겠습니다. 아직 답은 없습니다.”</p><h4>작성할 표 또는 기록 항목</h4><ul><li><p>보고자·받은 사람</p></li><li><p>사례 번호</p></li><li><p>보고한 여섯 내용</p></li><li><p>받은 사람이 다시 말한 일</p></li><li><p>잘못 이해해 고친 내용</p></li></ul><h4>제출할 내용과 끝내는 기준</h4><p>전원 보고 기록과 받은 사람의 확인 문장. 개인별로 한 사례 이상을 다룬다.</p><p>미션 제출 문서 sop/ops07_reconfirmation_procedure.md의 STEP 05 아래에 작성한다. 개인 기록과 조별 결과를 구분하고 실제 이름 대신 학습자 코드를 써도 된다. 표의 모양이 달라도 필요한 내용을 찾을 수 있으면 된다.</p><h4>마치기 전 확인</h4><ul><li><p>전원이 한 번 이상 보고하고 인계받은 기록이 있다.</p></li><li><p>각 보고에 여섯 내용이 모두 있다.</p></li><li><p>받은 사람의 확인 문장에 맡을 질문과 미답 사항이 있다.</p></li><li><p>없는 회신이나 소속을 만들어 보고하지 않았다.</p></li></ul><h4>배정 시간</h4><p>30분</p>', 'evalText': '<h4>평가할 자료와 위치</h4><p>sop/ops07_reconfirmation_procedure.md의 STEP 05 결과와 해당 개인 수행 기록을 확인한다. 이 단계에서 요청한 제출 내용은 다음과 같다: 전원 보고 기록과 받은 사람의 확인 문장. 개인별로 한 사례 이상을 다룬다.</p><h4>평가 순서</h4><ol><li><p>아래 통과 조건을 하나씩 읽고 제출물에서 확인한 위치를 옆에 적는다.</p></li><li><p>실제 설명·보고·인계가 필요한 단계는 수행 기록이나 직접 관찰로 확인한다. 서식만 있으면 수행한 것으로 보지 않는다.</p></li><li><p>통과 조건을 모두 충족하면 이 단계는 PASS다. 하나라도 빠지거나 반대로 적었으면 FAIL로 두고 빠진 항목을 구체적으로 적는다. 표현이나 표 모양이 예시와 달라도 뜻이 같고 근거를 찾을 수 있으면 인정한다.</p></li></ol><h4>PASS 조건</h4><ul><li><p>전원이 한 번 이상 보고하고 인계받은 기록이 있다.</p></li><li><p>각 보고에 여섯 내용이 모두 있다.</p></li><li><p>받은 사람의 확인 문장에 맡을 질문과 미답 사항이 있다.</p></li><li><p>없는 회신이나 소속을 만들어 보고하지 않았다.</p></li></ul><h4>FAIL에 해당하는 예</h4><ul><li><p>받는 사람이 “알겠습니다”라고만 했고 무엇을 맡았는지 없다.</p></li><li><p>실제 회신이 없는데 “위치 확인 완료”라고 보고했다.</p></li></ul><h4>이해 여부를 확인할 질문</h4><p>받는 사람은 인계가 끝나기 전에 무엇을 다시 말해야 하는가?</p><h4>인정할 수 있는 답의 핵심</h4><p>자신이 맡을 질문과 아직 답이 없는 내용을 구체적으로 말한다.</p><h4>보완과 재평가</h4><p>빠진 내용을 메모에 넣고 같은 역할 보고·인계를 다시 해 실제 수행 기록을 새로 남긴다.</p><p>수정 전 결과와 재평가 결과를 구분해 남긴다. 실제 하지 않은 보고·채점·시험을 나중에 했던 것처럼 채우지 않는다.</p>', 'attachments': []}], 'deliverables': ['sop/ops07_reconfirmation_procedure.md (필수) — 열두 질문과 세 사례 적용표, 다섯 순서 절차, 세 검토 질문, 전원 보고·인계 기록'], 'techStack': ['문서 편집기(Markdown · sop/ops07_reconfirmation_procedure.md) 또는 인쇄 양식', 'Q01~Q12 확인 질문표 및 S1~S3 사례 적용표', '5단계 보류·재확인·회신대조·인계 표준 절차서', '조원 교차 질문 수정 이력 및 전원 구술 보고·복창 기록지'], 'evaluationKeys': ['<h4>평가 질문</h4><p>열두 질문으로 세 사례의 확인 범위를 구분했는가</p><h4>평가자가 확인할 위치</h4><p>sop/ops07_reconfirmation_procedure.md의 STEP 01, STEP 02과 그 단계의 개인 기록을 확인한다.</p><h4>PASS를 주는 방법</h4><p>아래 단계별 항목을 모두 충족해야 이 평가 질문에 PASS를 준다. 항목 옆에 확인한 표·줄·회차를 적는다. 한 항목이라도 없거나 틀리면 FAIL 사유에 해당 단계와 문장을 적는다. 학습자가 쓴 말이 예시와 똑같을 필요는 없지만 확인할 내용은 모두 있어야 한다.</p><h4>STEP 01 사례를 읽을 때 확인할 질문 열두 가지를 만들기</h4><p>찾을 결과: 서로 다른 확인 질문 열두 개 이상과 질문표 사용 범위 설명.</p><ul><li><p>서로 다른 질문 열두 개 이상이 있다.</p></li><li><p>출처·시각·장소·관측 내용·아군·민간·모순·담당을 모두 다룬다.</p></li><li><p>질문마다 필요한 자료와 미확인 표시가 있다.</p></li><li><p>확인 개수나 복장·표식만으로 신분·진행을 판정하지 않는다.</p></li></ul><h4>STEP 02 세 사례에 열두 질문을 적용해 모르는 내용을 찾기</h4><p>찾을 결과: 세 사례×열두 질문의 적용표. 질문을 추가했다면 추가 질문도 세 사례에 적용한다.</p><ul><li><p>세 사례 모두 열두 질문에 답이나 미확인 이유가 있다.</p></li><li><p>없는 관측 시각·소속·회신을 만들지 않았다.</p></li><li><p>S1 물체·S3 복장만으로 소속을 확정하지 않았다.</p></li><li><p>S2에 아군 위치·민간 상황·검토 회신의 미확인이 남는다.</p></li></ul><h4>이런 결과는 통과시키지 않는다</h4><ul><li><p>열두 줄이 있지만 질문이 모두 “확인됐는가”이다.</p></li><li><p>S2는 표식이 같다는 이유로 나머지 질문을 모두 확인으로 채웠다.</p></li></ul><h4>평가 의견 작성 방법</h4><p>“근거 부족”이라고만 쓰지 말고 단계 번호, 충족하지 못한 조건, 제출물에서 실제로 빠지거나 잘못된 부분, 다시 할 일을 적는다. 예를 들어 STEP 01의 첫 조건은 “서로 다른 질문 열두 개 이상이 있다.”이다. 이 조건에 못 미쳤다면 어떤 줄·항목·참여자 기록이 빠졌는지 지목한다.</p><h4>다시 평가할 때</h4><p>틀리거나 빠진 내용만 구체적으로 알려 준다. 문서 작성 오류는 원자료와 수정본을 대조한다. 실제 수행 오류는 해당 행동을 다시 하게 하고 새 회차 기록으로 확인한다. 최초 결과는 지우지 않는다.</p>', '<h4>평가 질문</h4><p>부분 회신을 포함한 다섯 절차를 작성하고 전원이 보고·인계했는가</p><h4>평가자가 확인할 위치</h4><p>sop/ops07_reconfirmation_procedure.md의 STEP 03, STEP 04, STEP 05과 그 단계의 개인 기록을 확인한다.</p><h4>PASS를 주는 방법</h4><p>아래 단계별 항목을 모두 충족해야 이 평가 질문에 PASS를 준다. 항목 옆에 확인한 표·줄·회차를 적는다. 한 항목이라도 없거나 틀리면 FAIL 사유에 해당 단계와 문장을 적는다. 학습자가 쓴 말이 예시와 똑같을 필요는 없지만 확인할 내용은 모두 있어야 한다.</p><h4>STEP 03 보류 보고부터 회신 확인과 인계까지 절차를 쓰기</h4><p>찾을 결과: 다섯 순서의 절차표와 부분 회신 시험 기록 1개.</p><ul><li><p>다섯 순서 각각에 담당·행동·기록이 있다.</p></li><li><p>송신·접수·회신을 구별한다.</p></li><li><p>부분 회신에서 답하지 않은 질문을 별도로 남겼다.</p></li><li><p>미답 질문의 담당과 다음 확인 계획이 있고 보류를 자동 해제하지 않는다.</p></li></ul><h4>STEP 04 다른 조원의 질문을 받고 절차를 고치기</h4><p>찾을 결과: 사례마다 한 건 이상, 모두 세 건 이상의 검토 질문과 답·수정 기록.</p><ul><li><p>S1·S2·S3 각각에 검토 질문이 있다.</p></li><li><p>질문에 실제 자료와 모르는 내용을 근거로 답했다.</p></li><li><p>수정 전후 또는 기존 문장을 유지한 이유가 있다.</p></li><li><p>질문 일부에 답이 없을 때의 기록 방법이 최종 절차에 남는다.</p></li></ul><h4>STEP 05 각자 한 사례를 보고하고 다음 사람에게 일을 넘기기</h4><p>찾을 결과: 전원 보고 기록과 받은 사람의 확인 문장. 개인별로 한 사례 이상을 다룬다.</p><ul><li><p>전원이 한 번 이상 보고하고 인계받은 기록이 있다.</p></li><li><p>각 보고에 여섯 내용이 모두 있다.</p></li><li><p>받은 사람의 확인 문장에 맡을 질문과 미답 사항이 있다.</p></li><li><p>없는 회신이나 소속을 만들어 보고하지 않았다.</p></li></ul><h4>이런 결과는 통과시키지 않는다</h4><ul><li><p>“보고→확인→완료”만 적고 접수와 회신 내용이 없다.</p></li><li><p>“검토 완료·이견 없음”만 적고 실제 질문이 없다.</p></li><li><p>받는 사람이 “알겠습니다”라고만 했고 무엇을 맡았는지 없다.</p></li></ul><h4>평가 의견 작성 방법</h4><p>“근거 부족”이라고만 쓰지 말고 단계 번호, 충족하지 못한 조건, 제출물에서 실제로 빠지거나 잘못된 부분, 다시 할 일을 적는다. 예를 들어 STEP 03의 첫 조건은 “다섯 순서 각각에 담당·행동·기록이 있다.”이다. 이 조건에 못 미쳤다면 어떤 줄·항목·참여자 기록이 빠졌는지 지목한다.</p><h4>다시 평가할 때</h4><p>틀리거나 빠진 내용만 구체적으로 알려 준다. 문서 작성 오류는 원자료와 수정본을 대조한다. 실제 수행 오류는 해당 행동을 다시 하게 하고 새 회차 기록으로 확인한다. 최초 결과는 지우지 않는다.</p>'], 'evaluationTypes': ['REQUIRED', 'REQUIRED'], 'evaluationText': '<h4>미션 전체 판정</h4><p>이 미션은 아래 두 평가 질문이 모두 PASS여야 통과다. 조별 문서와 개인의 실제 수행을 나눠 확인한다. 수행하지 않았거나 기록이 없어 확인할 수 없는 사람은 미평가로 남기고 다시 실시한다. 최종 판정에서는 미평가를 PASS로 바꾸지 않는다.</p><h4>평가 질문</h4><p>열두 질문으로 세 사례의 확인 범위를 구분했는가</p><h4>평가자가 확인할 위치</h4><p>sop/ops07_reconfirmation_procedure.md의 STEP 01, STEP 02과 그 단계의 개인 기록을 확인한다.</p><h4>PASS를 주는 방법</h4><p>아래 단계별 항목을 모두 충족해야 이 평가 질문에 PASS를 준다. 항목 옆에 확인한 표·줄·회차를 적는다. 한 항목이라도 없거나 틀리면 FAIL 사유에 해당 단계와 문장을 적는다. 학습자가 쓴 말이 예시와 똑같을 필요는 없지만 확인할 내용은 모두 있어야 한다.</p><h4>STEP 01 사례를 읽을 때 확인할 질문 열두 가지를 만들기</h4><p>찾을 결과: 서로 다른 확인 질문 열두 개 이상과 질문표 사용 범위 설명.</p><ul><li><p>서로 다른 질문 열두 개 이상이 있다.</p></li><li><p>출처·시각·장소·관측 내용·아군·민간·모순·담당을 모두 다룬다.</p></li><li><p>질문마다 필요한 자료와 미확인 표시가 있다.</p></li><li><p>확인 개수나 복장·표식만으로 신분·진행을 판정하지 않는다.</p></li></ul><h4>STEP 02 세 사례에 열두 질문을 적용해 모르는 내용을 찾기</h4><p>찾을 결과: 세 사례×열두 질문의 적용표. 질문을 추가했다면 추가 질문도 세 사례에 적용한다.</p><ul><li><p>세 사례 모두 열두 질문에 답이나 미확인 이유가 있다.</p></li><li><p>없는 관측 시각·소속·회신을 만들지 않았다.</p></li><li><p>S1 물체·S3 복장만으로 소속을 확정하지 않았다.</p></li><li><p>S2에 아군 위치·민간 상황·검토 회신의 미확인이 남는다.</p></li></ul><h4>이런 결과는 통과시키지 않는다</h4><ul><li><p>열두 줄이 있지만 질문이 모두 “확인됐는가”이다.</p></li><li><p>S2는 표식이 같다는 이유로 나머지 질문을 모두 확인으로 채웠다.</p></li></ul><h4>평가 의견 작성 방법</h4><p>“근거 부족”이라고만 쓰지 말고 단계 번호, 충족하지 못한 조건, 제출물에서 실제로 빠지거나 잘못된 부분, 다시 할 일을 적는다. 예를 들어 STEP 01의 첫 조건은 “서로 다른 질문 열두 개 이상이 있다.”이다. 이 조건에 못 미쳤다면 어떤 줄·항목·참여자 기록이 빠졌는지 지목한다.</p><h4>다시 평가할 때</h4><p>틀리거나 빠진 내용만 구체적으로 알려 준다. 문서 작성 오류는 원자료와 수정본을 대조한다. 실제 수행 오류는 해당 행동을 다시 하게 하고 새 회차 기록으로 확인한다. 최초 결과는 지우지 않는다.</p><h4>평가 질문</h4><p>부분 회신을 포함한 다섯 절차를 작성하고 전원이 보고·인계했는가</p><h4>평가자가 확인할 위치</h4><p>sop/ops07_reconfirmation_procedure.md의 STEP 03, STEP 04, STEP 05과 그 단계의 개인 기록을 확인한다.</p><h4>PASS를 주는 방법</h4><p>아래 단계별 항목을 모두 충족해야 이 평가 질문에 PASS를 준다. 항목 옆에 확인한 표·줄·회차를 적는다. 한 항목이라도 없거나 틀리면 FAIL 사유에 해당 단계와 문장을 적는다. 학습자가 쓴 말이 예시와 똑같을 필요는 없지만 확인할 내용은 모두 있어야 한다.</p><h4>STEP 03 보류 보고부터 회신 확인과 인계까지 절차를 쓰기</h4><p>찾을 결과: 다섯 순서의 절차표와 부분 회신 시험 기록 1개.</p><ul><li><p>다섯 순서 각각에 담당·행동·기록이 있다.</p></li><li><p>송신·접수·회신을 구별한다.</p></li><li><p>부분 회신에서 답하지 않은 질문을 별도로 남겼다.</p></li><li><p>미답 질문의 담당과 다음 확인 계획이 있고 보류를 자동 해제하지 않는다.</p></li></ul><h4>STEP 04 다른 조원의 질문을 받고 절차를 고치기</h4><p>찾을 결과: 사례마다 한 건 이상, 모두 세 건 이상의 검토 질문과 답·수정 기록.</p><ul><li><p>S1·S2·S3 각각에 검토 질문이 있다.</p></li><li><p>질문에 실제 자료와 모르는 내용을 근거로 답했다.</p></li><li><p>수정 전후 또는 기존 문장을 유지한 이유가 있다.</p></li><li><p>질문 일부에 답이 없을 때의 기록 방법이 최종 절차에 남는다.</p></li></ul><h4>STEP 05 각자 한 사례를 보고하고 다음 사람에게 일을 넘기기</h4><p>찾을 결과: 전원 보고 기록과 받은 사람의 확인 문장. 개인별로 한 사례 이상을 다룬다.</p><ul><li><p>전원이 한 번 이상 보고하고 인계받은 기록이 있다.</p></li><li><p>각 보고에 여섯 내용이 모두 있다.</p></li><li><p>받은 사람의 확인 문장에 맡을 질문과 미답 사항이 있다.</p></li><li><p>없는 회신이나 소속을 만들어 보고하지 않았다.</p></li></ul><h4>이런 결과는 통과시키지 않는다</h4><ul><li><p>“보고→확인→완료”만 적고 접수와 회신 내용이 없다.</p></li><li><p>“검토 완료·이견 없음”만 적고 실제 질문이 없다.</p></li><li><p>받는 사람이 “알겠습니다”라고만 했고 무엇을 맡았는지 없다.</p></li></ul><h4>평가 의견 작성 방법</h4><p>“근거 부족”이라고만 쓰지 말고 단계 번호, 충족하지 못한 조건, 제출물에서 실제로 빠지거나 잘못된 부분, 다시 할 일을 적는다. 예를 들어 STEP 03의 첫 조건은 “다섯 순서 각각에 담당·행동·기록이 있다.”이다. 이 조건에 못 미쳤다면 어떤 줄·항목·참여자 기록이 빠졌는지 지목한다.</p><h4>다시 평가할 때</h4><p>틀리거나 빠진 내용만 구체적으로 알려 준다. 문서 작성 오류는 원자료와 수정본을 대조한다. 실제 수행 오류는 해당 행동을 다시 하게 하고 새 회차 기록으로 확인한다. 최초 결과는 지우지 않는다.</p>', 'aiGuidelines': {'allow': ['확인 질문 12항목(Q01~Q12)이 출처·시각·장소·관측·아군·민간·모순·담당을 빠짐없이 다루는지 검토받기', '5순서 절차(위험 알림·보류→사실·미확인 보고→질문·담당·접수 확인→회신 원문 대조→남은 질문 인계) 문안의 모호함 점검받기', '조원의 검토 질문에 맞춰 부분 회신·미답 질문 처리 문장을 명료하게 다듬기'], 'rules': ['AI 사용은 필수가 아니며, S1~S3 사례 적용표와 절차는 sop/ops07_reconfirmation_procedure.md에 작성한다.', '복장·표식·확인 개수만으로 신분이나 진행을 확정하지 않고, 송신·접수·회신을 엄격히 구분한다.', 'STEP 05의 전원 1사례 구술 보고(6항목 포함)와 수신자 복창 확인은 학습자가 직접 수행한다.'], 'forbid': ['실제 부대·작전 사례·교전수칙 원문을 입력하기', 'S1~S3에 없는 관측 시각·소속·회신을 만들어 넣거나 부분 회신만으로 보류를 자동 해제하도록 쓰기', '실제 구술 보고와 인계 확인 기록을 수행 없이 생성하기']}}, {'id': 'OPS07_m05', 'sequence': 5, 'title': '복수의 위험이 겹친 사례에서 보류와 후속 확인 연습', 'durationHours': 3, 'overview': '<p>다섯 사례의 서로 다른 보류 이유를 설명한다. 대상·아군·민간·접수의 네 항목을 빠뜨리지 않는다.</p><p>아래 다섯 단계에는 필요한 자료·작업 순서·예시·제출 내용과 평가 조건이 들어 있다. 각 단계의 자료를 읽은 뒤 순서대로 작성한다.</p>', 'objectives': '<h4>이 미션을 마치면 할 수 있어야 하는 일</h4><ul><li><p>다섯 사례의 서로 다른 보류 이유를 설명한다.</p></li><li><p>대상·아군·민간·접수의 네 항목을 빠뜨리지 않는다.</p></li><li><p>짧은 보고와 미완료 질문 인계를 실제로 연습한다.</p></li></ul>', 'scenario': '<h4>학습자의 역할</h4><p>학습자는 민간인과 아군에게 불필요한 피해가 생기지 않도록 보고를 읽고, 모르는 내용을 질문하고, 다음 담당자에게 기록을 넘기는 역할을 연습한다. 사람의 소속을 맞히거나 실제 타격을 결정하는 과제가 아니다.</p><h4>이번 미션의 과제</h4><ul><li><p>다섯 사례의 서로 다른 보류 이유를 설명한다.</p></li><li><p>대상·아군·민간·접수의 네 항목을 빠뜨리지 않는다.</p></li><li><p>짧은 보고와 미완료 질문 인계를 실제로 연습한다.</p></li></ul><h4>자료와 역할의 범위</h4><p>사례·시각·장소·역할은 모두 교육용이다. 실제 부대 자료를 쓰지 않는다. 자료에 답이 없으면 자료에 없다고 적는다. 실제 행동 권한은 이 수업이 정하지 않는다.</p><h4>개인 작업과 조별 작업</h4><p>개인 판단을 먼저 요청한 단계에서는 토의 전에 자신의 답을 남긴다. 그 뒤 조별 답과 수정 이유를 별도로 쓴다. 보고·인계·시험 단계에서는 실제 한 말과 행동을 기록한다.</p>', 'attachments': [], 'materials': ['/uploads/OPS07/OPS07_m05_student.ipynb', '/uploads/OPS07/OPS07_m05_workbook.ipynb'], 'materialText': '<h4>실습 노트북 안내 (선택 보조 도구)</h4><p><strong>학생 실습용</strong>: <code>OPS07_m05_student.ipynb</code> (직접 ANSWER 표와 note를 작성합니다.)<br><strong>교수자·해설용</strong>: <code>OPS07_m05_workbook.ipynb</code> (예시 풀이와 해설 note가 포함되어 있습니다.)<br><em>※ 본 과제는 문서 편집기나 인쇄 양식으로도 수행할 수 있으며, 노트북은 사례·질문·기록 양식을 표로 정리하는 선택 보조 도구입니다.</em></p><h4>자료를 읽는 순서</h4>\n\n<ol>\n<li><p>B1~B5를 각각 읽고 사례마다 어떤 내용이 확인되지 않았는지 찾습니다.</p></li>\n<li><p>STEP 1~2에서는 아래 네 가지 확인 질문을 차례로 적용합니다. 표식, 아군 위치, 민간 상황, 검토자 접수는 따로 확인합니다.</p></li>\n<li><p>STEP 3에서는 개인의 최초 답과 조별 답을 비교합니다. STEP 4~5의 보고와 인계에는 자신이 남긴 보류 이유·질문·담당을 사용합니다.</p></li>\n</ol>\n\n<h4>자료 1. 위험이 겹친 다섯 사례</h4>\n\n<h4>B1</h4>\n\n<p>학교 옥상 인원 2명과 쌍안경으로 보이는 물체.</p>\n\n<p>대피소라는 주민 진술 1건이 있다.</p>\n\n<p>건물 내부 이용과 사람의 지위는 미확인.</p>\n\n<h4>B2</h4>\n\n<p>차량의 표식은 두 수단에서 일치한다고 보고되었다.</p>\n\n<p>정찰조는 예정된 위치 보고가 20분 늦었다.</p>\n\n<p>동일 원자료 재전달 여부와 현재 아군 위치는 미확인.</p>\n\n<h4>B3</h4>\n\n<p>두 수단에서 같은 표식 차량을 보고했다.</p>\n\n<p>옆 민간 차량 1대가 정차해 있다.</p>\n\n<p>곧 떠날 것 같다는 보고가 있지만 실제 이탈은 확인되지 않았다.</p>\n\n<h4>B4</h4>\n\n<p>관찰 보고에는 식별 명확·가상 위험 2등급이라고 적혀 있다.</p>\n\n<p>상급 검토자와 연락이 15분째 되지 않는다.</p>\n\n<p>등급 산정의 원근거와 검토 회신이 미제공이다.</p>\n\n<h4>B5</h4>\n\n<p>영상에 인원 1명이 흰 천을 흔드는 장면이 있다.</p>\n\n<p>항복 의사 등 보호 필요성을 확인해야 할 신호다.</p>\n\n<p>의도·신분·주변 상황은 확인되지 않았다.</p>\n\n<h4>자료 2. 사례마다 적용할 네 가지 확인 질문</h4>\n\n<p><strong>1 대상 확인:</strong> 보고한 특징과 소속·의도 주장을 구별했는가. 빠지거나 서로 다른 정보는 무엇인가.</p>\n\n<p><strong>2 아군 위치:</strong> 지금 사례에 적용할 아군 위치 자료가 있는가. 없다면 누구에게 물을 것인가.</p>\n\n<p><strong>3 민간 상황:</strong> 주민·민간 차량·시설 이용에 관한 보고가 있는가. 무엇을 더 확인해야 하는가.</p>\n\n<p><strong>4 보고 접수와 검토:</strong> 검토자가 질문을 받았는가. 받은 답은 어느 질문에 답하는가.</p>\n\n<h4>자료를 사용할 때 지킬 점</h4>\n\n<p>각 사례는 독립적으로 검토합니다.</p>\n\n<p>시간 압박이나 검토자와의 연락 두절이 확인되지 않은 정보를 대신하지 않습니다.</p>\n\n<p>흰 천 등의 보호 필요 신호도 확인할 내용으로 따로 남깁니다.</p>\n\n<p>모든 사례·시각·장소·역할은 교육용입니다. 필요한 자료와 구체적인 작업 순서는 각 단계 설명에도 들어 있습니다.</p>\n', 'guideSteps': [{'step': 1, 'problem': '네 가지 질문으로 보류와 재확인 순서도 만들기', 'activity': '<h4>이번 단계에서 하는 일</h4><p>하나의 특징을 확인했다고 다른 위험까지 사라지는 것은 아니다. 대상·아군·민간·보고 접수를 각각 살피는 순서를 만든다.</p><h4>먼저 읽을 자료</h4><h4>차례로 확인할 네 가지 항목</h4><ul><li><p><strong>1 대상 확인:</strong> 보고한 특징과 소속·의도 주장을 구별했는가. 빠지거나 서로 다른 정보는 무엇인가.</p></li><li><p><strong>2 아군 위치:</strong> 지금 사례에 적용할 아군 위치 자료가 있는가. 없다면 누구에게 물을 것인가.</p></li><li><p><strong>3 민간 상황:</strong> 주민·민간 차량·시설 이용에 관한 보고가 있는가. 무엇을 더 확인해야 하는가.</p></li><li><p><strong>4 보고 접수와 검토:</strong> 검토자가 질문을 받았는가. 받은 답은 어느 질문에 답하는가.</p></li></ul><p>보류는 이 연습에서 확인되지 않은 내용을 확인한 것처럼 처리하지 않고 검토를 멈춰 두는 상태다.</p><p>재확인은 모르는 내용을 담당자에게 다시 묻는 일이다.</p><p>접수는 상대가 요청을 받았다는 뜻이고, 회신은 질문에 대한 답이다.</p><p>접수됐다고 해서 질문의 답까지 받은 것은 아니다.</p><p>아래의 역할은 모두 교육용이며 실제 지휘권을 뜻하지 않는다.</p><p>순서도는 화살표를 그려도 되고 번호가 있는 글로 써도 된다.</p><p>프로그램이나 그림 도구를 만들 필요는 없다.</p><h4>작업 순서</h4><ol><li><p>네 질문을 순서대로 적는다.</p></li><li><p>각 질문 아래에 “확인할 자료가 있는 경우”와 “없거나 서로 다른 경우”를 나눠 쓴다.</p></li><li><p>자료가 있는 경우에도 그 자료로 알 수 있는 범위를 기록하고 다음 질문으로 간다.</p></li><li><p>없거나 서로 다른 경우에는 보류 유지·확인 질문·담당·기다릴 답을 적는다. 앞에서 보류하더라도 뒤의 위험을 기록에서 지우지 않는다.</p></li><li><p>4번에서는 요청 송신·접수·답 수신을 구별한다. 마지막을 자동 진행으로 끝내지 않는다.</p></li><li><p>한 사람이 순서를 읽고 다른 사람이 그 글만 보고 다음에 무엇을 할지 말해 본다. 이해하지 못한 표현을 고친다.</p></li></ol><h4>작성 예시</h4><p>다음은 쓰는 방식을 보여 주는 예다. 자기 판단이나 실제 수행·측정 결과가 필요한 칸에 그대로 복사하지 않는다.</p><p>2 아군 위치: 현재 보고가 없으면 → 보류 유지 → 위치 보고 담당자에게 해당 시점의 자료를 요청 → 아직 답이 없다는 사실을 기록. 그 뒤에도 민간 위험 보고를 확인 목록에서 지우지 않음.</p><h4>작성할 표 또는 기록 항목</h4><ul><li><p>순서 번호</p></li><li><p>확인할 질문</p></li><li><p>자료가 있을 때 할 일</p></li><li><p>없거나 다를 때 할 일</p></li><li><p>확인 담당</p></li><li><p>필요한 답</p></li><li><p>남길 기록</p></li></ul><h4>제출할 내용과 끝내는 기준</h4><p>네 질문이 모두 들어간 순서도 또는 번호 절차문, 읽어 본 사람이 이해하지 못한 표현과 수정 기록.</p><p>미션 제출 문서 sop/ops07_approval_thresholds_sop.md의 STEP 01 아래에 작성한다. 개인 기록과 조별 결과를 구분하고 실제 이름 대신 학습자 코드를 써도 된다. 표의 모양이 달라도 필요한 내용을 찾을 수 있으면 된다.</p><h4>마치기 전 확인</h4><ul><li><p>네 질문 모두에 자료 유무에 따른 다음 행동이 있다.</p></li><li><p>자료가 없을 때 질문·담당·필요한 답을 찾을 수 있다.</p></li><li><p>앞 단계의 보류 때문에 뒤의 민간·아군 위험을 삭제하지 않는다.</p></li><li><p>어느 길도 점수·등급만으로 자동 진행에 도달하지 않는다.</p></li></ul><h4>배정 시간</h4><p>35분</p>', 'evalText': '<h4>평가할 자료와 위치</h4><p>sop/ops07_approval_thresholds_sop.md의 STEP 01 결과와 해당 개인 수행 기록을 확인한다. 이 단계에서 요청한 제출 내용은 다음과 같다: 네 질문이 모두 들어간 순서도 또는 번호 절차문, 읽어 본 사람이 이해하지 못한 표현과 수정 기록.</p><h4>평가 순서</h4><ol><li><p>아래 통과 조건을 하나씩 읽고 제출물에서 확인한 위치를 옆에 적는다.</p></li><li><p>실제 설명·보고·인계가 필요한 단계는 수행 기록이나 직접 관찰로 확인한다. 서식만 있으면 수행한 것으로 보지 않는다.</p></li><li><p>통과 조건을 모두 충족하면 이 단계는 PASS다. 하나라도 빠지거나 반대로 적었으면 FAIL로 두고 빠진 항목을 구체적으로 적는다. 표현이나 표 모양이 예시와 달라도 뜻이 같고 근거를 찾을 수 있으면 인정한다.</p></li></ol><h4>PASS 조건</h4><ul><li><p>네 질문 모두에 자료 유무에 따른 다음 행동이 있다.</p></li><li><p>자료가 없을 때 질문·담당·필요한 답을 찾을 수 있다.</p></li><li><p>앞 단계의 보류 때문에 뒤의 민간·아군 위험을 삭제하지 않는다.</p></li><li><p>어느 길도 점수·등급만으로 자동 진행에 도달하지 않는다.</p></li></ul><h4>FAIL에 해당하는 예</h4><ul><li><p>“식별 명확이면 진행” 한 줄로 끝낸다.</p></li><li><p>보류 칸에 “확인”만 있고 누가 무엇을 할지 없다.</p></li></ul><h4>이해 여부를 확인할 질문</h4><p>1번에서 보류하면 민간 상황 확인은 기록에서 빼도 되는가?</p><h4>인정할 수 있는 답의 핵심</h4><p>빼지 않으며 동시에 남아 있는 위험과 질문을 기록해야 한다고 답한다.</p><h4>보완과 재평가</h4><p>네 질문 각각의 자료 없음 경로를 따라가며 빠진 담당·질문·기록을 채운다.</p><p>수정 전 결과와 재평가 결과를 구분해 남긴다. 실제 하지 않은 보고·채점·시험을 나중에 했던 것처럼 채우지 않는다.</p>', 'attachments': []}, {'step': 2, 'problem': '다섯 사례를 각자 읽고 보류 이유와 질문을 쓰기', 'activity': '<h4>이번 단계에서 하는 일</h4><p>다섯 사례는 모두 모르는 내용이 남아 있지만 그 이유는 다르다. “보류”라는 같은 답 뒤에 각기 다른 질문이 나와야 한다.</p><h4>먼저 읽을 자료</h4><h4>사례 자료</h4><p><strong>B1</strong></p><p>학교 옥상 인원 2명과 쌍안경으로 보이는 물체.</p><p>대피소라는 주민 진술 1건이 있다.</p><p>건물 내부 이용과 사람의 지위는 미확인.</p><p><strong>B2</strong></p><p>차량의 표식은 두 수단에서 일치한다고 보고되었다.</p><p>정찰조는 예정된 위치 보고가 20분 늦었다.</p><p>동일 원자료 재전달 여부와 현재 아군 위치는 미확인.</p><p><strong>B3</strong></p><p>두 수단에서 같은 표식 차량을 보고했다.</p><p>옆 민간 차량 1대가 정차해 있다.</p><p>곧 떠날 것 같다는 보고가 있지만 실제 이탈은 확인되지 않았다.</p><p><strong>B4</strong></p><p>관찰 보고에는 식별 명확·가상 위험 2등급이라고 적혀 있다.</p><p>상급 검토자와 연락이 15분째 되지 않는다.</p><p>등급 산정의 원근거와 검토 회신이 미제공이다.</p><p><strong>B5</strong></p><p>영상에 인원 1명이 흰 천을 흔드는 장면이 있다.</p><p>항복 의사 등 보호 필요성을 확인해야 할 신호다.</p><p>의도·신분·주변 상황은 확인되지 않았다.</p><h4>차례로 확인할 네 가지 항목</h4><ul><li><p><strong>1 대상 확인:</strong> 보고한 특징과 소속·의도 주장을 구별했는가. 빠지거나 서로 다른 정보는 무엇인가.</p></li><li><p><strong>2 아군 위치:</strong> 지금 사례에 적용할 아군 위치 자료가 있는가. 없다면 누구에게 물을 것인가.</p></li><li><p><strong>3 민간 상황:</strong> 주민·민간 차량·시설 이용에 관한 보고가 있는가. 무엇을 더 확인해야 하는가.</p></li><li><p><strong>4 보고 접수와 검토:</strong> 검토자가 질문을 받았는가. 받은 답은 어느 질문에 답하는가.</p></li></ul><h4>작업 순서</h4><ol><li><p>토의 전에 각자 B1~B5를 읽는다.</p></li><li><p>각 사례에서 실제 보고된 내용과 확인하지 못한 내용을 따로 적는다.</p></li><li><p>보류할 이유를 카드의 문장과 연결해 한 문장으로 쓴다.</p></li><li><p>대상·아군·민간·보고 접수의 네 항목 중 무엇을 더 확인할지 표시한다. 카드에 정보가 없으면 없다고 적는다.</p></li><li><p>담당 역할과 상대에게 전달할 질문을 쓴다. 답이 없다는 상태도 적는다.</p></li><li><p>개인 표를 저장한 뒤에만 조원과 비교한다.</p></li></ol><h4>작성 예시</h4><p>다음은 쓰는 방식을 보여 주는 예다. 자기 판단이나 실제 수행·측정 결과가 필요한 칸에 그대로 복사하지 않는다.</p><p>B2: “표식 일치 보고는 있지만 현재 정찰조 위치는 모른다. 위치 담당자에게 이 사례 시점의 위치 보고가 있는지 요청한다.”</p><h4>작성할 표 또는 기록 항목</h4><ul><li><p>사례 번호</p></li><li><p>보고된 내용</p></li><li><p>모르는 내용</p></li><li><p>보류 이유</p></li><li><p>추가 확인 항목</p></li><li><p>담당·질문</p></li><li><p>답 수신 여부</p></li></ul><h4>제출할 내용과 끝내는 기준</h4><p>개인별 B1~B5 다섯 줄의 판단표.</p><p>미션 제출 문서 sop/ops07_approval_thresholds_sop.md의 STEP 02 아래에 작성한다. 개인 기록과 조별 결과를 구분하고 실제 이름 대신 학습자 코드를 써도 된다. 표의 모양이 달라도 필요한 내용을 찾을 수 있으면 된다.</p><h4>마치기 전 확인</h4><ul><li><p>개인별 다섯 사례의 원래 답이 있다.</p></li><li><p>B1 시설 이용, B2 아군 위치, B3 민간 차량, B4 접수·검토자 연락, B5 보호 필요 신호를 빠뜨리지 않았다.</p></li><li><p>각 보류 이유에 맞는 담당·질문이 있다.</p></li><li><p>예상·미회신을 실제 이탈·확인 완료로 바꾸지 않았다.</p></li></ul><h4>배정 시간</h4><p>40분</p>', 'evalText': '<h4>평가할 자료와 위치</h4><p>sop/ops07_approval_thresholds_sop.md의 STEP 02 결과와 해당 개인 수행 기록을 확인한다. 이 단계에서 요청한 제출 내용은 다음과 같다: 개인별 B1~B5 다섯 줄의 판단표.</p><h4>평가 순서</h4><ol><li><p>아래 통과 조건을 하나씩 읽고 제출물에서 확인한 위치를 옆에 적는다.</p></li><li><p>실제 설명·보고·인계가 필요한 단계는 수행 기록이나 직접 관찰로 확인한다. 서식만 있으면 수행한 것으로 보지 않는다.</p></li><li><p>통과 조건을 모두 충족하면 이 단계는 PASS다. 하나라도 빠지거나 반대로 적었으면 FAIL로 두고 빠진 항목을 구체적으로 적는다. 표현이나 표 모양이 예시와 달라도 뜻이 같고 근거를 찾을 수 있으면 인정한다.</p></li></ol><h4>PASS 조건</h4><ul><li><p>개인별 다섯 사례의 원래 답이 있다.</p></li><li><p>B1 시설 이용, B2 아군 위치, B3 민간 차량, B4 접수·검토자 연락, B5 보호 필요 신호를 빠뜨리지 않았다.</p></li><li><p>각 보류 이유에 맞는 담당·질문이 있다.</p></li><li><p>예상·미회신을 실제 이탈·확인 완료로 바꾸지 않았다.</p></li></ul><h4>FAIL에 해당하는 예</h4><ul><li><p>다섯 사례를 전부 “위험하니 보류”로만 적었다.</p></li><li><p>B3 민간 차량이 곧 떠날 것이라는 말을 이미 떠난 사실로 썼다.</p></li></ul><h4>이해 여부를 확인할 질문</h4><p>B2와 B3는 보류 이유와 질문이 어떻게 다른가?</p><h4>인정할 수 있는 답의 핵심</h4><p>B2는 현재 아군 위치를, B3는 현재 민간 차량과 관련 상황을 확인해야 한다고 구분한다.</p><h4>보완과 재평가</h4><p>각 카드의 특징 한 문장을 보류 이유에 넣고 그 특징을 확인할 질문을 다시 쓴다.</p><p>수정 전 결과와 재평가 결과를 구분해 남긴다. 실제 하지 않은 보고·채점·시험을 나중에 했던 것처럼 채우지 않는다.</p>', 'attachments': []}, {'step': 3, 'problem': '개인 답을 비교하고 놓친 위험을 고치기', 'activity': '<h4>이번 단계에서 하는 일</h4><p>조원들이 같은 보류 답을 골라도 확인하려는 내용은 다를 수 있다. 결론뿐 아니라 질문과 이유까지 비교한다.</p><h4>먼저 읽을 자료</h4><h4>사례 자료</h4><p><strong>B1</strong></p><p>학교 옥상 인원 2명과 쌍안경으로 보이는 물체.</p><p>대피소라는 주민 진술 1건이 있다.</p><p>건물 내부 이용과 사람의 지위는 미확인.</p><p><strong>B2</strong></p><p>차량의 표식은 두 수단에서 일치한다고 보고되었다.</p><p>정찰조는 예정된 위치 보고가 20분 늦었다.</p><p>동일 원자료 재전달 여부와 현재 아군 위치는 미확인.</p><p><strong>B3</strong></p><p>두 수단에서 같은 표식 차량을 보고했다.</p><p>옆 민간 차량 1대가 정차해 있다.</p><p>곧 떠날 것 같다는 보고가 있지만 실제 이탈은 확인되지 않았다.</p><p><strong>B4</strong></p><p>관찰 보고에는 식별 명확·가상 위험 2등급이라고 적혀 있다.</p><p>상급 검토자와 연락이 15분째 되지 않는다.</p><p>등급 산정의 원근거와 검토 회신이 미제공이다.</p><p><strong>B5</strong></p><p>영상에 인원 1명이 흰 천을 흔드는 장면이 있다.</p><p>항복 의사 등 보호 필요성을 확인해야 할 신호다.</p><p>의도·신분·주변 상황은 확인되지 않았다.</p><p>비교할 개인 표가 없으면 먼저 각자 다섯 사례에 보고 내용·모르는 내용·보류 이유·질문·담당을 적는다.</p><p>토의 전 답과 토의 후 답을 구분해 보관한다.</p><h4>작업 순서</h4><ol><li><p>한 사례씩 돌아가며 각자의 보류 이유와 확인 질문을 읽는다.</p></li><li><p>같은 부분과 다른 부분을 나눠 적는다. 결론이 같아도 질문이 다르면 차이가 있다고 적는다.</p></li><li><p>B2는 표식 확인이 아군 위치 질문에 답하는지, B4는 연락 두절이 다른 사람에게 권한을 주는지, B5는 흰 천으로 의도를 확정할 수 있는지 서로 묻는다.</p></li><li><p>자료로 설명할 수 있는 조별 답을 정한다. 모든 사람의 답을 지우고 조별 답으로 바꿔 쓰지 않는다.</p></li><li><p>남은 의견 차이는 어떤 정보가 더 있어야 풀리는지 적고 교관에게 물을 질문으로 남긴다.</p></li></ol><h4>작성 예시</h4><p>다음은 쓰는 방식을 보여 주는 예다. 자기 판단이나 실제 수행·측정 결과가 필요한 칸에 그대로 복사하지 않는다.</p><p>B4: 개인 A “연락될 때까지 담당이 확인 요청을 이어감”, 개인 B “연락 안 되면 다른 사람이 승인” → 조별 답 “실제 위임 자료가 없으므로 승인으로 처리하지 않음. 접수·회신 미확인 상태를 남김.”</p><h4>작성할 표 또는 기록 항목</h4><ul><li><p>사례 번호</p></li><li><p>개인 답의 같은 점</p></li><li><p>다른 이유·질문</p></li><li><p>조별 최종 문장</p></li><li><p>원문 근거</p></li><li><p>남은 의견과 필요한 정보</p></li></ul><h4>제출할 내용과 끝내는 기준</h4><p>B1~B5 비교 기록과 조별 답. 개인 최초 답은 별도로 남긴다.</p><p>미션 제출 문서 sop/ops07_approval_thresholds_sop.md의 STEP 03 아래에 작성한다. 개인 기록과 조별 결과를 구분하고 실제 이름 대신 학습자 코드를 써도 된다. 표의 모양이 달라도 필요한 내용을 찾을 수 있으면 된다.</p><h4>마치기 전 확인</h4><ul><li><p>다섯 사례의 결론뿐 아니라 보류 이유·질문을 비교했다.</p></li><li><p>B2·B4·B5의 지정된 검토 질문에 답했다.</p></li><li><p>토의 전 개인 답과 조별 답이 구분된다.</p></li><li><p>없는 위임·이탈·의도 확인을 만들어 합의하지 않았다.</p></li></ul><h4>배정 시간</h4><p>40분</p>', 'evalText': '<h4>평가할 자료와 위치</h4><p>sop/ops07_approval_thresholds_sop.md의 STEP 03 결과와 해당 개인 수행 기록을 확인한다. 이 단계에서 요청한 제출 내용은 다음과 같다: B1~B5 비교 기록과 조별 답. 개인 최초 답은 별도로 남긴다.</p><h4>평가 순서</h4><ol><li><p>아래 통과 조건을 하나씩 읽고 제출물에서 확인한 위치를 옆에 적는다.</p></li><li><p>실제 설명·보고·인계가 필요한 단계는 수행 기록이나 직접 관찰로 확인한다. 서식만 있으면 수행한 것으로 보지 않는다.</p></li><li><p>통과 조건을 모두 충족하면 이 단계는 PASS다. 하나라도 빠지거나 반대로 적었으면 FAIL로 두고 빠진 항목을 구체적으로 적는다. 표현이나 표 모양이 예시와 달라도 뜻이 같고 근거를 찾을 수 있으면 인정한다.</p></li></ol><h4>PASS 조건</h4><ul><li><p>다섯 사례의 결론뿐 아니라 보류 이유·질문을 비교했다.</p></li><li><p>B2·B4·B5의 지정된 검토 질문에 답했다.</p></li><li><p>토의 전 개인 답과 조별 답이 구분된다.</p></li><li><p>없는 위임·이탈·의도 확인을 만들어 합의하지 않았다.</p></li></ul><h4>FAIL에 해당하는 예</h4><ul><li><p>“전원 합의”만 적고 어떤 이유를 비교했는지 없다.</p></li><li><p>B5를 항복 확정 또는 기만 확정으로 적었다.</p></li></ul><h4>이해 여부를 확인할 질문</h4><p>흰 천을 흔드는 장면만으로 지금 무엇까지 말할 수 있는가?</p><h4>인정할 수 있는 답의 핵심</h4><p>그런 행동이 관찰됐고 보호 필요성과 의도를 확인해야 한다고 말하며 신분·의도를 확정하지 않는다.</p><h4>보완과 재평가</h4><p>조별 답의 각 문장을 카드와 대조하고 확정할 수 없는 표현을 모르는 내용과 질문으로 바꾼다.</p><p>수정 전 결과와 재평가 결과를 구분해 남긴다. 실제 하지 않은 보고·채점·시험을 나중에 했던 것처럼 채우지 않는다.</p>', 'attachments': []}, {'step': 4, 'problem': '급할 때 쓸 짧은 보고문을 만들고 실제로 읽기', 'activity': '<h4>이번 단계에서 하는 일</h4><p>시간이 부족할 때 문장을 짧게 할 수는 있지만 중요한 내용을 없앨 수는 없다. 필요한 항목이 모두 들어간 짧은 보고를 연습한다.</p><h4>먼저 읽을 자료</h4><h4>사례 자료</h4><p><strong>B3</strong></p><p>두 수단에서 같은 표식 차량을 보고했다.</p><p>옆 민간 차량 1대가 정차해 있다.</p><p>곧 떠날 것 같다는 보고가 있지만 실제 이탈은 확인되지 않았다.</p><p><strong>B4</strong></p><p>관찰 보고에는 식별 명확·가상 위험 2등급이라고 적혀 있다.</p><p>상급 검토자와 연락이 15분째 되지 않는다.</p><p>등급 산정의 원근거와 검토 회신이 미제공이다.</p><h4>짧은 보고에 넣을 내용</h4><ul><li><p>사례 번호와 지금 보고된 내용</p></li><li><p>확인하지 못한 내용과 보류 이유</p></li><li><p>확인 담당과 그대로 전달할 질문</p></li><li><p>요청이 접수됐는지·어떤 답을 받았는지</p></li><li><p>아직 남은 질문과 다음 확인 계획</p></li></ul><h4>작업 순서</h4><ol><li><p>B3 또는 B4를 고른다.</p></li><li><p>표의 다섯 내용을 한 문장씩 써서 짧은 보고문을 만든다. 문장을 합쳐도 내용은 빼지 않는다.</p></li><li><p>한 조원이 보고문을 읽고 다른 조원은 표의 다섯 내용을 표시하며 듣는다.</p></li><li><p>듣는 사람이 질문·담당·현재 상태를 다시 말한다. 잘못 이해한 부분을 고친다.</p></li><li><p>상세 기록에만 추가할 내용은 따로 적는다. 상세 기록을 나중에 쓴다는 이유로 현재의 보류 이유·질문·담당·접수 상태를 미루지 않는다.</p></li></ol><h4>작성 예시</h4><p>다음은 쓰는 방식을 보여 주는 예다. 자기 판단이나 실제 수행·측정 결과가 필요한 칸에 그대로 복사하지 않는다.</p><p>B4: “상급 검토자와 연락이 안 되고 접수·회신이 확인되지 않았습니다. 보류를 유지합니다. 기록 담당자는 질문 접수 여부를 확인할 요청을 남기고 미회신 항목을 다음 점검에 올리겠습니다.”</p><h4>작성할 표 또는 기록 항목</h4><ul><li><p>선택 사례</p></li><li><p>짧은 보고문</p></li><li><p>다섯 항목 확인</p></li><li><p>듣는 사람이 다시 말한 내용</p></li><li><p>상세 기록에 추가할 내용</p></li></ul><h4>제출할 내용과 끝내는 기준</h4><p>짧은 보고문 1개와 실제 읽기·듣기 점검 기록.</p><p>미션 제출 문서 sop/ops07_approval_thresholds_sop.md의 STEP 04 아래에 작성한다. 개인 기록과 조별 결과를 구분하고 실제 이름 대신 학습자 코드를 써도 된다. 표의 모양이 달라도 필요한 내용을 찾을 수 있으면 된다.</p><h4>마치기 전 확인</h4><ul><li><p>보고문에 다섯 필수 내용이 모두 있다.</p></li><li><p>실제 자료에 없는 접수·회신을 추가하지 않았다.</p></li><li><p>듣는 사람이 질문·담당·현재 상태를 설명했다.</p></li><li><p>짧게 보고한다는 이유로 민간·아군 위험 확인이나 사람의 검토를 생략하지 않았다.</p></li></ul><h4>배정 시간</h4><p>35분</p>', 'evalText': '<h4>평가할 자료와 위치</h4><p>sop/ops07_approval_thresholds_sop.md의 STEP 04 결과와 해당 개인 수행 기록을 확인한다. 이 단계에서 요청한 제출 내용은 다음과 같다: 짧은 보고문 1개와 실제 읽기·듣기 점검 기록.</p><h4>평가 순서</h4><ol><li><p>아래 통과 조건을 하나씩 읽고 제출물에서 확인한 위치를 옆에 적는다.</p></li><li><p>실제 설명·보고·인계가 필요한 단계는 수행 기록이나 직접 관찰로 확인한다. 서식만 있으면 수행한 것으로 보지 않는다.</p></li><li><p>통과 조건을 모두 충족하면 이 단계는 PASS다. 하나라도 빠지거나 반대로 적었으면 FAIL로 두고 빠진 항목을 구체적으로 적는다. 표현이나 표 모양이 예시와 달라도 뜻이 같고 근거를 찾을 수 있으면 인정한다.</p></li></ol><h4>PASS 조건</h4><ul><li><p>보고문에 다섯 필수 내용이 모두 있다.</p></li><li><p>실제 자료에 없는 접수·회신을 추가하지 않았다.</p></li><li><p>듣는 사람이 질문·담당·현재 상태를 설명했다.</p></li><li><p>짧게 보고한다는 이유로 민간·아군 위험 확인이나 사람의 검토를 생략하지 않았다.</p></li></ul><h4>FAIL에 해당하는 예</h4><ul><li><p>“급하니 일단 진행”으로 보고를 줄였다.</p></li><li><p>질문과 담당은 상세 보고 때 적겠다며 지금 기록하지 않았다.</p></li></ul><h4>이해 여부를 확인할 질문</h4><p>짧은 보고에서도 나중으로 미룰 수 없는 내용 두 가지를 말해 보라.</p><h4>인정할 수 있는 답의 핵심</h4><p>보류 이유·미확인·질문·담당·접수 상태 중 두 가지를 구체적으로 답한다.</p><h4>보완과 재평가</h4><p>듣는 사람이 찾지 못한 필수 내용을 보고문에 넣고 다시 읽어 이해 여부를 확인한다.</p><p>수정 전 결과와 재평가 결과를 구분해 남긴다. 실제 하지 않은 보고·채점·시험을 나중에 했던 것처럼 채우지 않는다.</p>', 'attachments': []}, {'step': 5, 'problem': '다섯 사례의 미완료 질문을 다음 담당자에게 넘기기', 'activity': '<h4>이번 단계에서 하는 일</h4><p>보류했다고 기록한 뒤 질문의 담당이 끊기면 확인은 진행되지 않는다. 다음 담당자가 무엇을 이어서 해야 하는지 확인한다.</p><h4>먼저 읽을 자료</h4><h4>사례 자료</h4><p><strong>B1</strong></p><p>학교 옥상 인원 2명과 쌍안경으로 보이는 물체.</p><p>대피소라는 주민 진술 1건이 있다.</p><p>건물 내부 이용과 사람의 지위는 미확인.</p><p><strong>B2</strong></p><p>차량의 표식은 두 수단에서 일치한다고 보고되었다.</p><p>정찰조는 예정된 위치 보고가 20분 늦었다.</p><p>동일 원자료 재전달 여부와 현재 아군 위치는 미확인.</p><p><strong>B3</strong></p><p>두 수단에서 같은 표식 차량을 보고했다.</p><p>옆 민간 차량 1대가 정차해 있다.</p><p>곧 떠날 것 같다는 보고가 있지만 실제 이탈은 확인되지 않았다.</p><p><strong>B4</strong></p><p>관찰 보고에는 식별 명확·가상 위험 2등급이라고 적혀 있다.</p><p>상급 검토자와 연락이 15분째 되지 않는다.</p><p>등급 산정의 원근거와 검토 회신이 미제공이다.</p><p><strong>B5</strong></p><p>영상에 인원 1명이 흰 천을 흔드는 장면이 있다.</p><p>항복 의사 등 보호 필요성을 확인해야 할 신호다.</p><p>의도·신분·주변 상황은 확인되지 않았다.</p><p>앞 단계 판단표가 있으면 사용한다.</p><p>없으면 사례별로 현재 모르는 내용·질문·담당부터 적는다.</p><p>이 자료에는 실제 확인 회신이 없으므로 가상의 완료 회신을 만들지 않는다.</p><h4>작업 순서</h4><ol><li><p>B1~B5의 질문을 모아 아직 답이 없는 것을 표시한다.</p></li><li><p>각 질문에 현재 맡은 사람과 다음에 맡을 사람을 적는다. 같은 사람이 계속 맡으면 그 이유를 적는다.</p></li><li><p>넘기는 사람은 질문·기다리는 답·현재 보류 상태를 읽는다.</p></li><li><p>받는 사람은 자신이 할 질문과 아직 모르는 내용을 다시 말한다.</p></li><li><p>개인별로 최초 판단에서 바뀐 점을 하나씩 적는다. 바뀐 점이 없다면 같은 판단을 유지한 근거를 적는다.</p></li><li><p>조원 전원이 한 번 이상 인계하거나 받은 내용이 기록됐는지 확인한다.</p></li></ol><h4>작성 예시</h4><p>다음은 쓰는 방식을 보여 주는 예다. 자기 판단이나 실제 수행·측정 결과가 필요한 칸에 그대로 복사하지 않는다.</p><p>B1 인계: “학교의 최근 이용 상태를 묻는 질문은 답이 없습니다. 다음 담당자는 이용 기록의 시각과 내용을 요청합니다. 대피소 이용 여부는 확인하지 못한 상태로 남깁니다.”</p><h4>작성할 표 또는 기록 항목</h4><ul><li><p>사례·질문 번호</p></li><li><p>아직 없는 답</p></li><li><p>현재·다음 담당</p></li><li><p>다음에 할 질문</p></li><li><p>인계한 사람·받은 사람</p></li><li><p>받은 사람의 확인</p></li><li><p>개인 판단 변경 이유</p></li></ul><h4>제출할 내용과 끝내는 기준</h4><p>다섯 사례의 질문 인계표와 전원 개인 변경·유지 이유.</p><p>미션 제출 문서 sop/ops07_approval_thresholds_sop.md의 STEP 05 아래에 작성한다. 개인 기록과 조별 결과를 구분하고 실제 이름 대신 학습자 코드를 써도 된다. 표의 모양이 달라도 필요한 내용을 찾을 수 있으면 된다.</p><h4>마치기 전 확인</h4><ul><li><p>다섯 사례 모두 아직 답이 없는 질문과 담당이 있다.</p></li><li><p>받는 사람이 다음 할 일을 다시 말한 기록이 있다.</p></li><li><p>전원의 인계 또는 수신 참여가 확인된다.</p></li><li><p>개인 최초 판단과 바뀐 이유 또는 유지 근거가 있다.</p></li></ul><h4>배정 시간</h4><p>30분</p>', 'evalText': '<h4>평가할 자료와 위치</h4><p>sop/ops07_approval_thresholds_sop.md의 STEP 05 결과와 해당 개인 수행 기록을 확인한다. 이 단계에서 요청한 제출 내용은 다음과 같다: 다섯 사례의 질문 인계표와 전원 개인 변경·유지 이유.</p><h4>평가 순서</h4><ol><li><p>아래 통과 조건을 하나씩 읽고 제출물에서 확인한 위치를 옆에 적는다.</p></li><li><p>실제 설명·보고·인계가 필요한 단계는 수행 기록이나 직접 관찰로 확인한다. 서식만 있으면 수행한 것으로 보지 않는다.</p></li><li><p>통과 조건을 모두 충족하면 이 단계는 PASS다. 하나라도 빠지거나 반대로 적었으면 FAIL로 두고 빠진 항목을 구체적으로 적는다. 표현이나 표 모양이 예시와 달라도 뜻이 같고 근거를 찾을 수 있으면 인정한다.</p></li></ol><h4>PASS 조건</h4><ul><li><p>다섯 사례 모두 아직 답이 없는 질문과 담당이 있다.</p></li><li><p>받는 사람이 다음 할 일을 다시 말한 기록이 있다.</p></li><li><p>전원의 인계 또는 수신 참여가 확인된다.</p></li><li><p>개인 최초 판단과 바뀐 이유 또는 유지 근거가 있다.</p></li></ul><h4>FAIL에 해당하는 예</h4><ul><li><p>“보류 유지”만 적고 누가 질문을 이어갈지 없다.</p></li><li><p>확인되지 않은 질문을 인계 목록에서 빼버렸다.</p></li></ul><h4>이해 여부를 확인할 질문</h4><p>다음 담당자에게 질문만 넘기고 보류 상태를 말하지 않으면 어떤 오해가 생기는가?</p><h4>인정할 수 있는 답의 핵심</h4><p>질문에 이미 답이 왔거나 검토가 끝난 것으로 오해할 수 있어 현재 상태도 알려야 한다고 설명한다.</p><h4>보완과 재평가</h4><p>누락 질문을 복구하고 받은 사람이 현재 상태와 다음 할 일을 다시 말한 기록을 추가한다.</p><p>수정 전 결과와 재평가 결과를 구분해 남긴다. 실제 하지 않은 보고·채점·시험을 나중에 했던 것처럼 채우지 않는다.</p>', 'attachments': []}], 'deliverables': ['sop/ops07_approval_thresholds_sop.md (필수) — 네 확인 순서, 다섯 사례의 개인·조별 판단, 짧은 보고, 미답 질문의 담당 인계'], 'techStack': ['문서 편집기(Markdown · sop/ops07_approval_thresholds_sop.md) 또는 인쇄 양식', '4단계(대상·아군·민간·접수) 보류·재확인 순서도', 'B1~B5 경계선 사례 개인·조별 비교표', '긴급용 짧은 보고문 템플릿 및 미완료 질문 인계 기록지'], 'evaluationKeys': ['<h4>평가 질문</h4><p>다섯 사례의 서로 다른 보류 이유와 질문을 개인·조별로 설명했는가</p><h4>평가자가 확인할 위치</h4><p>sop/ops07_approval_thresholds_sop.md의 STEP 02, STEP 03과 그 단계의 개인 기록을 확인한다.</p><h4>PASS를 주는 방법</h4><p>아래 단계별 항목을 모두 충족해야 이 평가 질문에 PASS를 준다. 항목 옆에 확인한 표·줄·회차를 적는다. 한 항목이라도 없거나 틀리면 FAIL 사유에 해당 단계와 문장을 적는다. 학습자가 쓴 말이 예시와 똑같을 필요는 없지만 확인할 내용은 모두 있어야 한다.</p><h4>STEP 02 다섯 사례를 각자 읽고 보류 이유와 질문을 쓰기</h4><p>찾을 결과: 개인별 B1~B5 다섯 줄의 판단표.</p><ul><li><p>개인별 다섯 사례의 원래 답이 있다.</p></li><li><p>B1 시설 이용, B2 아군 위치, B3 민간 차량, B4 접수·검토자 연락, B5 보호 필요 신호를 빠뜨리지 않았다.</p></li><li><p>각 보류 이유에 맞는 담당·질문이 있다.</p></li><li><p>예상·미회신을 실제 이탈·확인 완료로 바꾸지 않았다.</p></li></ul><h4>STEP 03 개인 답을 비교하고 놓친 위험을 고치기</h4><p>찾을 결과: B1~B5 비교 기록과 조별 답. 개인 최초 답은 별도로 남긴다.</p><ul><li><p>다섯 사례의 결론뿐 아니라 보류 이유·질문을 비교했다.</p></li><li><p>B2·B4·B5의 지정된 검토 질문에 답했다.</p></li><li><p>토의 전 개인 답과 조별 답이 구분된다.</p></li><li><p>없는 위임·이탈·의도 확인을 만들어 합의하지 않았다.</p></li></ul><h4>이런 결과는 통과시키지 않는다</h4><ul><li><p>다섯 사례를 전부 “위험하니 보류”로만 적었다.</p></li><li><p>“전원 합의”만 적고 어떤 이유를 비교했는지 없다.</p></li></ul><h4>평가 의견 작성 방법</h4><p>“근거 부족”이라고만 쓰지 말고 단계 번호, 충족하지 못한 조건, 제출물에서 실제로 빠지거나 잘못된 부분, 다시 할 일을 적는다. 예를 들어 STEP 02의 첫 조건은 “개인별 다섯 사례의 원래 답이 있다.”이다. 이 조건에 못 미쳤다면 어떤 줄·항목·참여자 기록이 빠졌는지 지목한다.</p><h4>다시 평가할 때</h4><p>틀리거나 빠진 내용만 구체적으로 알려 준다. 문서 작성 오류는 원자료와 수정본을 대조한다. 실제 수행 오류는 해당 행동을 다시 하게 하고 새 회차 기록으로 확인한다. 최초 결과는 지우지 않는다.</p>', '<h4>평가 질문</h4><p>네 가지 확인 순서와 짧은 보고·미완료 인계를 실제로 시험했는가</p><h4>평가자가 확인할 위치</h4><p>sop/ops07_approval_thresholds_sop.md의 STEP 01, STEP 04, STEP 05과 그 단계의 개인 기록을 확인한다.</p><h4>PASS를 주는 방법</h4><p>아래 단계별 항목을 모두 충족해야 이 평가 질문에 PASS를 준다. 항목 옆에 확인한 표·줄·회차를 적는다. 한 항목이라도 없거나 틀리면 FAIL 사유에 해당 단계와 문장을 적는다. 학습자가 쓴 말이 예시와 똑같을 필요는 없지만 확인할 내용은 모두 있어야 한다.</p><h4>STEP 01 네 가지 질문으로 보류와 재확인 순서도 만들기</h4><p>찾을 결과: 네 질문이 모두 들어간 순서도 또는 번호 절차문, 읽어 본 사람이 이해하지 못한 표현과 수정 기록.</p><ul><li><p>네 질문 모두에 자료 유무에 따른 다음 행동이 있다.</p></li><li><p>자료가 없을 때 질문·담당·필요한 답을 찾을 수 있다.</p></li><li><p>앞 단계의 보류 때문에 뒤의 민간·아군 위험을 삭제하지 않는다.</p></li><li><p>어느 길도 점수·등급만으로 자동 진행에 도달하지 않는다.</p></li></ul><h4>STEP 04 급할 때 쓸 짧은 보고문을 만들고 실제로 읽기</h4><p>찾을 결과: 짧은 보고문 1개와 실제 읽기·듣기 점검 기록.</p><ul><li><p>보고문에 다섯 필수 내용이 모두 있다.</p></li><li><p>실제 자료에 없는 접수·회신을 추가하지 않았다.</p></li><li><p>듣는 사람이 질문·담당·현재 상태를 설명했다.</p></li><li><p>짧게 보고한다는 이유로 민간·아군 위험 확인이나 사람의 검토를 생략하지 않았다.</p></li></ul><h4>STEP 05 다섯 사례의 미완료 질문을 다음 담당자에게 넘기기</h4><p>찾을 결과: 다섯 사례의 질문 인계표와 전원 개인 변경·유지 이유.</p><ul><li><p>다섯 사례 모두 아직 답이 없는 질문과 담당이 있다.</p></li><li><p>받는 사람이 다음 할 일을 다시 말한 기록이 있다.</p></li><li><p>전원의 인계 또는 수신 참여가 확인된다.</p></li><li><p>개인 최초 판단과 바뀐 이유 또는 유지 근거가 있다.</p></li></ul><h4>이런 결과는 통과시키지 않는다</h4><ul><li><p>“식별 명확이면 진행” 한 줄로 끝낸다.</p></li><li><p>“급하니 일단 진행”으로 보고를 줄였다.</p></li><li><p>“보류 유지”만 적고 누가 질문을 이어갈지 없다.</p></li></ul><h4>평가 의견 작성 방법</h4><p>“근거 부족”이라고만 쓰지 말고 단계 번호, 충족하지 못한 조건, 제출물에서 실제로 빠지거나 잘못된 부분, 다시 할 일을 적는다. 예를 들어 STEP 01의 첫 조건은 “네 질문 모두에 자료 유무에 따른 다음 행동이 있다.”이다. 이 조건에 못 미쳤다면 어떤 줄·항목·참여자 기록이 빠졌는지 지목한다.</p><h4>다시 평가할 때</h4><p>틀리거나 빠진 내용만 구체적으로 알려 준다. 문서 작성 오류는 원자료와 수정본을 대조한다. 실제 수행 오류는 해당 행동을 다시 하게 하고 새 회차 기록으로 확인한다. 최초 결과는 지우지 않는다.</p>'], 'evaluationTypes': ['REQUIRED', 'REQUIRED'], 'evaluationText': '<h4>미션 전체 판정</h4><p>이 미션은 아래 두 평가 질문이 모두 PASS여야 통과다. 조별 문서와 개인의 실제 수행을 나눠 확인한다. 수행하지 않았거나 기록이 없어 확인할 수 없는 사람은 미평가로 남기고 다시 실시한다. 최종 판정에서는 미평가를 PASS로 바꾸지 않는다.</p><h4>평가 질문</h4><p>다섯 사례의 서로 다른 보류 이유와 질문을 개인·조별로 설명했는가</p><h4>평가자가 확인할 위치</h4><p>sop/ops07_approval_thresholds_sop.md의 STEP 02, STEP 03과 그 단계의 개인 기록을 확인한다.</p><h4>PASS를 주는 방법</h4><p>아래 단계별 항목을 모두 충족해야 이 평가 질문에 PASS를 준다. 항목 옆에 확인한 표·줄·회차를 적는다. 한 항목이라도 없거나 틀리면 FAIL 사유에 해당 단계와 문장을 적는다. 학습자가 쓴 말이 예시와 똑같을 필요는 없지만 확인할 내용은 모두 있어야 한다.</p><h4>STEP 02 다섯 사례를 각자 읽고 보류 이유와 질문을 쓰기</h4><p>찾을 결과: 개인별 B1~B5 다섯 줄의 판단표.</p><ul><li><p>개인별 다섯 사례의 원래 답이 있다.</p></li><li><p>B1 시설 이용, B2 아군 위치, B3 민간 차량, B4 접수·검토자 연락, B5 보호 필요 신호를 빠뜨리지 않았다.</p></li><li><p>각 보류 이유에 맞는 담당·질문이 있다.</p></li><li><p>예상·미회신을 실제 이탈·확인 완료로 바꾸지 않았다.</p></li></ul><h4>STEP 03 개인 답을 비교하고 놓친 위험을 고치기</h4><p>찾을 결과: B1~B5 비교 기록과 조별 답. 개인 최초 답은 별도로 남긴다.</p><ul><li><p>다섯 사례의 결론뿐 아니라 보류 이유·질문을 비교했다.</p></li><li><p>B2·B4·B5의 지정된 검토 질문에 답했다.</p></li><li><p>토의 전 개인 답과 조별 답이 구분된다.</p></li><li><p>없는 위임·이탈·의도 확인을 만들어 합의하지 않았다.</p></li></ul><h4>이런 결과는 통과시키지 않는다</h4><ul><li><p>다섯 사례를 전부 “위험하니 보류”로만 적었다.</p></li><li><p>“전원 합의”만 적고 어떤 이유를 비교했는지 없다.</p></li></ul><h4>평가 의견 작성 방법</h4><p>“근거 부족”이라고만 쓰지 말고 단계 번호, 충족하지 못한 조건, 제출물에서 실제로 빠지거나 잘못된 부분, 다시 할 일을 적는다. 예를 들어 STEP 02의 첫 조건은 “개인별 다섯 사례의 원래 답이 있다.”이다. 이 조건에 못 미쳤다면 어떤 줄·항목·참여자 기록이 빠졌는지 지목한다.</p><h4>다시 평가할 때</h4><p>틀리거나 빠진 내용만 구체적으로 알려 준다. 문서 작성 오류는 원자료와 수정본을 대조한다. 실제 수행 오류는 해당 행동을 다시 하게 하고 새 회차 기록으로 확인한다. 최초 결과는 지우지 않는다.</p><h4>평가 질문</h4><p>네 가지 확인 순서와 짧은 보고·미완료 인계를 실제로 시험했는가</p><h4>평가자가 확인할 위치</h4><p>sop/ops07_approval_thresholds_sop.md의 STEP 01, STEP 04, STEP 05과 그 단계의 개인 기록을 확인한다.</p><h4>PASS를 주는 방법</h4><p>아래 단계별 항목을 모두 충족해야 이 평가 질문에 PASS를 준다. 항목 옆에 확인한 표·줄·회차를 적는다. 한 항목이라도 없거나 틀리면 FAIL 사유에 해당 단계와 문장을 적는다. 학습자가 쓴 말이 예시와 똑같을 필요는 없지만 확인할 내용은 모두 있어야 한다.</p><h4>STEP 01 네 가지 질문으로 보류와 재확인 순서도 만들기</h4><p>찾을 결과: 네 질문이 모두 들어간 순서도 또는 번호 절차문, 읽어 본 사람이 이해하지 못한 표현과 수정 기록.</p><ul><li><p>네 질문 모두에 자료 유무에 따른 다음 행동이 있다.</p></li><li><p>자료가 없을 때 질문·담당·필요한 답을 찾을 수 있다.</p></li><li><p>앞 단계의 보류 때문에 뒤의 민간·아군 위험을 삭제하지 않는다.</p></li><li><p>어느 길도 점수·등급만으로 자동 진행에 도달하지 않는다.</p></li></ul><h4>STEP 04 급할 때 쓸 짧은 보고문을 만들고 실제로 읽기</h4><p>찾을 결과: 짧은 보고문 1개와 실제 읽기·듣기 점검 기록.</p><ul><li><p>보고문에 다섯 필수 내용이 모두 있다.</p></li><li><p>실제 자료에 없는 접수·회신을 추가하지 않았다.</p></li><li><p>듣는 사람이 질문·담당·현재 상태를 설명했다.</p></li><li><p>짧게 보고한다는 이유로 민간·아군 위험 확인이나 사람의 검토를 생략하지 않았다.</p></li></ul><h4>STEP 05 다섯 사례의 미완료 질문을 다음 담당자에게 넘기기</h4><p>찾을 결과: 다섯 사례의 질문 인계표와 전원 개인 변경·유지 이유.</p><ul><li><p>다섯 사례 모두 아직 답이 없는 질문과 담당이 있다.</p></li><li><p>받는 사람이 다음 할 일을 다시 말한 기록이 있다.</p></li><li><p>전원의 인계 또는 수신 참여가 확인된다.</p></li><li><p>개인 최초 판단과 바뀐 이유 또는 유지 근거가 있다.</p></li></ul><h4>이런 결과는 통과시키지 않는다</h4><ul><li><p>“식별 명확이면 진행” 한 줄로 끝낸다.</p></li><li><p>“급하니 일단 진행”으로 보고를 줄였다.</p></li><li><p>“보류 유지”만 적고 누가 질문을 이어갈지 없다.</p></li></ul><h4>평가 의견 작성 방법</h4><p>“근거 부족”이라고만 쓰지 말고 단계 번호, 충족하지 못한 조건, 제출물에서 실제로 빠지거나 잘못된 부분, 다시 할 일을 적는다. 예를 들어 STEP 01의 첫 조건은 “네 질문 모두에 자료 유무에 따른 다음 행동이 있다.”이다. 이 조건에 못 미쳤다면 어떤 줄·항목·참여자 기록이 빠졌는지 지목한다.</p><h4>다시 평가할 때</h4><p>틀리거나 빠진 내용만 구체적으로 알려 준다. 문서 작성 오류는 원자료와 수정본을 대조한다. 실제 수행 오류는 해당 행동을 다시 하게 하고 새 회차 기록으로 확인한다. 최초 결과는 지우지 않는다.</p>', 'aiGuidelines': {'allow': ['4단계 확인 순서도(대상 확인→아군 위치→민간 상황→보고 접수)에서 앞 단계 보류 후 뒤 항목 누락 여부 점검받기', 'B1~B5 사례별 보류 이유·질문·담당 구분이 4가지 위험 항목과 맞는지 구조 검토받기', 'B3·B4용 짧은 보고문 초안에 5가지 필수 항목이 빠짐없이 들어갔는지 확인받기'], 'rules': ['AI 사용은 필수가 아니며, 토의 전 B1~B5 개인 원답을 먼저 저장한 뒤 조별 답을 비교한다(sop/ops07_approval_thresholds_sop.md).', '예상(차량 이탈 예정)·미회신(검토자 연락 두절)을 실제 이탈이나 확인 완료로 바꾸지 않는다.', 'STEP 04의 짧은 보고 구술 읽기와 STEP 05의 B1~B5 미완료 질문 인계·복창은 학습자가 직접 수행한다.'], 'forbid': ['실제 부대·표적 정보·지휘체계를 입력하기', '점수나 가상 등급(예: B4 2등급)만으로 자동 진행·위임에 도달하는 순서도를 작성하기', '개인 최초 판단이나 인계·수신 복창 기록을 실제 토의·발표 없이 AI 출력으로 채우기']}}, {'id': 'OPS07_m06', 'sequence': 6, 'title': '사실과 추정을 구분하는 체크리스트와 기록 서식 제작', 'durationHours': 3, 'overview': '<p>보고자의 주장과 관찰 내용을 구별해 다시 쓴다. 세 원문을 짧은 기록·상세 기록에 옮기며 오류를 고친다.</p><p>아래 다섯 단계에는 필요한 자료·작업 순서·예시·제출 내용과 평가 조건이 들어 있다. 각 단계의 자료를 읽은 뒤 순서대로 작성한다.</p>', 'objectives': '<h4>이 미션을 마치면 할 수 있어야 하는 일</h4><ul><li><p>보고자의 주장과 관찰 내용을 구별해 다시 쓴다.</p></li><li><p>세 원문을 짧은 기록·상세 기록에 옮기며 오류를 고친다.</p></li><li><p>다음 사람이 기록 번호로 원문과 질문을 찾게 한다.</p></li></ul>', 'scenario': '<h4>학습자의 역할</h4><p>학습자는 민간인과 아군에게 불필요한 피해가 생기지 않도록 보고를 읽고, 모르는 내용을 질문하고, 다음 담당자에게 기록을 넘기는 역할을 연습한다. 사람의 소속을 맞히거나 실제 타격을 결정하는 과제가 아니다.</p><h4>이번 미션의 과제</h4><ul><li><p>보고자의 주장과 관찰 내용을 구별해 다시 쓴다.</p></li><li><p>세 원문을 짧은 기록·상세 기록에 옮기며 오류를 고친다.</p></li><li><p>다음 사람이 기록 번호로 원문과 질문을 찾게 한다.</p></li></ul><h4>자료와 역할의 범위</h4><p>사례·시각·장소·역할은 모두 교육용이다. 실제 부대 자료를 쓰지 않는다. 자료에 답이 없으면 자료에 없다고 적는다. 실제 행동 권한은 이 수업이 정하지 않는다.</p><h4>개인 작업과 조별 작업</h4><p>개인 판단을 먼저 요청한 단계에서는 토의 전에 자신의 답을 남긴다. 그 뒤 조별 답과 수정 이유를 별도로 쓴다. 보고·인계·시험 단계에서는 실제 한 말과 행동을 기록한다.</p>', 'attachments': [], 'materials': ['/uploads/OPS07/OPS07_m06_student.ipynb', '/uploads/OPS07/OPS07_m06_workbook.ipynb'], 'materialText': '<h4>실습 노트북 안내 (선택 보조 도구)</h4><p><strong>학생 실습용</strong>: <code>OPS07_m06_student.ipynb</code> (직접 ANSWER 표와 note를 작성합니다.)<br><strong>교수자·해설용</strong>: <code>OPS07_m06_workbook.ipynb</code> (예시 풀이와 해설 note가 포함되어 있습니다.)<br><em>※ 본 과제는 문서 편집기나 인쇄 양식으로도 수행할 수 있으며, 노트북은 사례·질문·기록 양식을 표로 정리하는 선택 보조 도구입니다.</em></p><h4>자료를 읽는 순서</h4>\n\n<ol>\n<li><p>N1~N3는 고쳐 써야 할 원문입니다. “확실히”, “문제없음”, “전부 적 관측반” 등의 표현을 확인된 사실로 받아들이지 않습니다.</p></li>\n<li><p>STEP 1~2에서는 원문의 관찰 보고·주장·조치·미확인을 나누고, 근거 없이 단정한 표현을 고칩니다.</p></li>\n<li><p>STEP 3~5에서는 아래 필수 기록 항목을 참고해 짧은 기록과 상세 기록을 만들고, 다른 사람이 원문과 미답을 찾을 수 있는지 시험합니다.</p></li>\n</ol>\n\n<h4>자료 1. 사실과 추정을 구분해 고쳐 쓸 원문</h4>\n\n<p>다음은 분석 대상인 원문입니다. 단정하거나 애매하게 쓴 표현도 수정 전 상태로 보관했습니다.</p>\n\n<h4>N1</h4>\n\n<p>14:05 OP-2 보고.</p>\n\n<p>교량 북단에 적 차량 2대가 확실히 있음.</p>\n\n<p>아마 공격 준비 중인 듯.</p>\n\n<p>영상은 배터리 교체로 10분간 못 봄.</p>\n\n<p>가상 안전 검토자에게 보고함.</p>\n\n<h4>N2</h4>\n\n<p>14:20 정찰조는 문제없음.</p>\n\n<p>골목 인원 4명은 적으로 보임.</p>\n\n<p>무전은 계속 안 됨.</p>\n\n<p>일단 대기시킴.</p>\n\n<h4>N3</h4>\n\n<p>14:40 학교 옥상 인원은 전부 적 관측반임.</p>\n\n<p>주민 말로는 대피소라는데 확인 안 됨.</p>\n\n<p>상급 검토 요청을 시도했으나 통신 두절.</p>\n\n<h4>자료 2. 기록에서 빠뜨리지 않을 열 가지 항목</h4>\n\n<ol>\n<li><p>기록 번호·보고 번호</p></li>\n<li><p>관측 시각과 보고·수신 시각</p></li>\n<li><p>출처·확인 수단</p></li>\n<li><p>보고에 적힌 관찰 내용</p></li>\n<li><p>보고자의 주장과 나의 판단</p></li>\n<li><p>보냈다고 하거나 실제 수행한 조치</p></li>\n<li><p>요청 접수·회신 여부</p></li>\n<li><p>확인하지 못한 내용</p></li>\n<li><p>확인 질문·담당·다음 점검</p></li>\n<li><p>현재 보류 상태</p></li>\n</ol>\n\n<h4>자료를 사용할 때 지킬 점</h4>\n\n<p>원문에 보고했다고 적혀 있어도 상대방이 접수했거나 답했다는 뜻은 아닙니다.</p>\n\n<p>원문·고친 기록·확인하지 못한 내용을 구별해 보관합니다.</p>\n\n<p>모든 사례·시각·장소·역할은 교육용입니다. 필요한 자료와 구체적인 작업 순서는 각 단계 설명에도 들어 있습니다.</p>\n', 'guideSteps': [{'step': 1, 'problem': '세 보고를 문장별로 나눠 관찰·주장·조치를 구분하기', 'activity': '<h4>이번 단계에서 하는 일</h4><p>보고 문장을 옮기는 과정에서 추측이 확인된 사실로 바뀔 수 있다. 원문을 보존하고 각 문장이 무엇을 말하는지 분류한다.</p><h4>먼저 읽을 자료</h4>\n\n<h4>사례 자료</h4>\n\n<p><strong>N1</strong></p>\n\n<p>14:05 OP-2 보고.</p>\n\n<p>교량 북단에 적 차량 2대가 확실히 있음.</p>\n\n<p>아마 공격 준비 중인 듯.</p>\n\n<p>영상은 배터리 교체로 10분간 못 봄.</p>\n\n<p>가상 안전 검토자에게 보고함.</p>\n\n<p><strong>N2</strong></p>\n\n<p>14:20 정찰조는 문제없음.</p>\n\n<p>골목 인원 4명은 적으로 보임.</p>\n\n<p>무전은 계속 안 됨.</p>\n\n<p>일단 대기시킴.</p>\n\n<p><strong>N3</strong></p>\n\n<p>14:40 학교 옥상 인원은 전부 적 관측반임.</p>\n\n<p>주민 말로는 대피소라는데 확인 안 됨.</p>\n\n<p>상급 검토 요청을 시도했으나 통신 두절.</p>\n\n<h4>자료를 나누어 읽는 기준</h4>\n\n<ul>\n<li>\n<p><strong>관찰에 관한 보고:</strong> 무엇을 보았다고 적혀 있는지. 직접 확인 자료가 없으면 보고자의 말임을 표시함.</p>\n</li>\n\n<li>\n<p><strong>주장·추측:</strong> 적 차량·공격 준비·문제없음처럼 해석하거나 단정한 표현</p>\n</li>\n\n<li>\n<p><strong>조치에 관한 기록:</strong> 보고함·대기시킴·요청 시도처럼 했다고 적혀 있는 행동</p>\n</li>\n\n<li>\n<p><strong>확인하지 못한 내용:</strong> 영상 공백·연락 두절·시설 이용 미확인 등</p>\n</li>\n</ul>\n\n<h4>작업 순서</h4><ol><li><p>N1~N3 원문을 지우지 않고 문서에 옮긴다.</p></li><li><p>내용이 두 개 이상인 문장은 나눠 번호를 붙인다.</p></li><li><p>각 조각에 관찰 보고·주장·조치·미확인 중 맞는 분류를 적는다. 한 조각이 두 내용을 포함하면 더 나눈다.</p></li><li><p>보고자가 그렇게 말했다는 사실과 실제 대상이 그렇다는 사실을 구분해 다시 쓴다.</p></li><li><p>관측 시각이 제공되지 않은 경우 보고 시각을 관측 시각으로 대신 넣지 않는다.</p></li></ol><h4>작성 예시</h4><p>다음은 쓰는 방식을 보여 주는 예다. 자기 판단이나 실제 수행·측정 결과가 필요한 칸에 그대로 복사하지 않는다.</p><p>N1의 “적 차량 2대가 확실히 있음” → “OP-2가 적 차량 2대라고 보고했다. 직접 확인 자료는 제공되지 않았다.” 원문의 단정 표현은 원문 칸에 그대로 둔다.</p><h4>작성할 표 또는 기록 항목</h4><ul><li><p>보고 번호·원문 조각 번호</p></li><li><p>원문 그대로</p></li><li><p>분류</p></li><li><p>확인할 수 있는 범위로 다시 쓴 문장</p></li><li><p>여전히 모르는 내용</p></li></ul><h4>제출할 내용과 끝내는 기준</h4><p>세 보고의 문장 분류표. 원문의 모든 내용을 분류하고 빠진 문장이 없는지 확인한다.</p><p>미션 제출 문서 templates/ops07_aar_fact_action_template.md의 STEP 01 아래에 작성한다. 개인 기록과 조별 결과를 구분하고 실제 이름 대신 학습자 코드를 써도 된다. 표의 모양이 달라도 필요한 내용을 찾을 수 있으면 된다.</p><h4>마치기 전 확인</h4><ul><li><p>N1~N3 원문을 모두 보존했다.</p></li><li><p>원문의 각 내용에 분류와 설명이 있다.</p></li><li><p>적·공격 의도·안전하다는 주장을 외부 사실로 바꾸지 않았다.</p></li><li><p>보고 시각과 관측 시각을 혼동하지 않았다.</p></li></ul><h4>배정 시간</h4><p>30분</p>', 'evalText': '<h4>평가할 자료와 위치</h4><p>templates/ops07_aar_fact_action_template.md의 STEP 01 결과와 해당 개인 수행 기록을 확인한다. 이 단계에서 요청한 제출 내용은 다음과 같다: 세 보고의 문장 분류표. 원문의 모든 내용을 분류하고 빠진 문장이 없는지 확인한다.</p><h4>평가 순서</h4><ol><li><p>아래 통과 조건을 하나씩 읽고 제출물에서 확인한 위치를 옆에 적는다.</p></li><li><p>실제 설명·보고·인계가 필요한 단계는 수행 기록이나 직접 관찰로 확인한다. 서식만 있으면 수행한 것으로 보지 않는다.</p></li><li><p>통과 조건을 모두 충족하면 이 단계는 PASS다. 하나라도 빠지거나 반대로 적었으면 FAIL로 두고 빠진 항목을 구체적으로 적는다. 표현이나 표 모양이 예시와 달라도 뜻이 같고 근거를 찾을 수 있으면 인정한다.</p></li></ol><h4>PASS 조건</h4><ul><li><p>N1~N3 원문을 모두 보존했다.</p></li><li><p>원문의 각 내용에 분류와 설명이 있다.</p></li><li><p>적·공격 의도·안전하다는 주장을 외부 사실로 바꾸지 않았다.</p></li><li><p>보고 시각과 관측 시각을 혼동하지 않았다.</p></li></ul><h4>FAIL에 해당하는 예</h4><ul><li><p>N2의 “정찰조는 문제없음”을 확인된 안전 상태로 적었다.</p></li><li><p>N3의 승인 요청 시도를 요청 접수 완료로 적었다.</p></li></ul><h4>이해 여부를 확인할 질문</h4><p>“적 차량이라고 보고했다”와 “적 차량이다”는 기록상 무엇이 다른가?</p><h4>인정할 수 있는 답의 핵심</h4><p>앞은 보고 내용의 존재를 적은 것이고 뒤는 대상의 소속을 사실로 확정한 것이므로 근거 범위가 다르다고 답한다.</p><h4>보완과 재평가</h4><p>단정 표현을 원문 칸에 보존하고 정리 문장에는 누가 무엇이라고 말했는지와 미확인 범위를 넣는다.</p><p>수정 전 결과와 재평가 결과를 구분해 남긴다. 실제 하지 않은 보고·채점·시험을 나중에 했던 것처럼 채우지 않는다.</p>', 'attachments': []}, {'step': 2, 'problem': '오해를 부르는 표현 여섯 가지를 바꾸기', 'activity': '<h4>이번 단계에서 하는 일</h4><p>확신을 강조하는 말을 다른 말로 바꾸는 것만으로 충분하지 않다. 무엇을 확인했고 무엇을 모르는지 문장에 넣어야 한다.</p><h4>먼저 읽을 자료</h4>\n\n<h4>다시 써야 할 표현</h4>\n\n<ul>\n<li>\n<p><strong>확실히:</strong> 무슨 자료로 확인했는지 알 수 없음</p>\n</li>\n\n<li>\n<p><strong>적으로 보임:</strong> 관찰 특징과 소속 판단이 섞임</p>\n</li>\n\n<li>\n<p><strong>문제없음:</strong> 무엇을 확인했는지·미확인이 무엇인지 알 수 없음</p>\n</li>\n\n<li>\n<p><strong>전부:</strong> 확인하지 않은 사람까지 포함할 수 있음</p>\n</li>\n\n<li>\n<p><strong>항상:</strong> 자료가 다루지 않은 시각까지 일반화할 수 있음</p>\n</li>\n\n<li>\n<p><strong>일단:</strong> 언제까지 무엇을 유지할지·다음 할 일이 모호함</p>\n</li>\n</ul>\n\n<h4>사례 자료</h4>\n\n<p><strong>N1</strong></p>\n\n<p>14:05 OP-2 보고.</p>\n\n<p>교량 북단에 적 차량 2대가 확실히 있음.</p>\n\n<p>아마 공격 준비 중인 듯.</p>\n\n<p>영상은 배터리 교체로 10분간 못 봄.</p>\n\n<p>가상 안전 검토자에게 보고함.</p>\n\n<p><strong>N2</strong></p>\n\n<p>14:20 정찰조는 문제없음.</p>\n\n<p>골목 인원 4명은 적으로 보임.</p>\n\n<p>무전은 계속 안 됨.</p>\n\n<p>일단 대기시킴.</p>\n\n<p><strong>N3</strong></p>\n\n<p>14:40 학교 옥상 인원은 전부 적 관측반임.</p>\n\n<p>주민 말로는 대피소라는데 확인 안 됨.</p>\n\n<p>상급 검토 요청을 시도했으나 통신 두절.</p>\n\n<h4>작업 순서</h4><ol><li><p>여섯 표현을 한 줄씩 적는다.</p></li><li><p>원문에서 해당 표현을 찾거나 직접 만든 가상 예라고 표시해 예문을 쓴다.</p></li><li><p>왜 이 표현만으로는 부족한지 적는다.</p></li><li><p>주체·출처·확인 범위·아직 모르는 내용이 드러나도록 바꾼다. 원문에 없는 수단이나 회신을 추가하지 않는다.</p></li><li><p>“아마·가능성” 같은 불확실성을 나타내는 말 자체를 금지하지 않는다. 추측임을 밝히고 근거를 함께 쓰는 것이 목적이다.</p></li></ol><h4>작성 예시</h4><p>다음은 쓰는 방식을 보여 주는 예다. 자기 판단이나 실제 수행·측정 결과가 필요한 칸에 그대로 복사하지 않는다.</p><p>“정찰조는 문제없음” → “정찰조와 연락이 되지 않는다는 보고가 있어 현재 안전 상태는 확인하지 못했다.” 확인 자료가 없는 상황에서 “문제없음”만 다른 확정 표현으로 바꾸지 않는다.</p><h4>작성할 표 또는 기록 항목</h4><ul><li><p>표현</p></li><li><p>원문 또는 직접 만든 예</p></li><li><p>오해할 수 있는 점</p></li><li><p>바꾼 문장</p></li><li><p>추가로 필요한 확인</p></li></ul><h4>제출할 내용과 끝내는 기준</h4><p>서로 다른 표현 여섯 가지 이상과 표현별 수정 문장.</p><p>미션 제출 문서 templates/ops07_aar_fact_action_template.md의 STEP 02 아래에 작성한다. 개인 기록과 조별 결과를 구분하고 실제 이름 대신 학습자 코드를 써도 된다. 표의 모양이 달라도 필요한 내용을 찾을 수 있으면 된다.</p><h4>마치기 전 확인</h4><ul><li><p>서로 다른 여섯 표현을 다뤘다.</p></li><li><p>각 표현의 문제점과 바꾼 문장이 있다.</p></li><li><p>바꾼 문장에 누가 무엇을 알며 모르는지 드러난다.</p></li><li><p>없는 확인 수단·현재 상태·회신을 만들어 표현을 강화하지 않았다.</p></li></ul><h4>배정 시간</h4><p>30분</p>', 'evalText': '<h4>평가할 자료와 위치</h4><p>templates/ops07_aar_fact_action_template.md의 STEP 02 결과와 해당 개인 수행 기록을 확인한다. 이 단계에서 요청한 제출 내용은 다음과 같다: 서로 다른 표현 여섯 가지 이상과 표현별 수정 문장.</p><h4>평가 순서</h4><ol><li><p>아래 통과 조건을 하나씩 읽고 제출물에서 확인한 위치를 옆에 적는다.</p></li><li><p>실제 설명·보고·인계가 필요한 단계는 수행 기록이나 직접 관찰로 확인한다. 서식만 있으면 수행한 것으로 보지 않는다.</p></li><li><p>통과 조건을 모두 충족하면 이 단계는 PASS다. 하나라도 빠지거나 반대로 적었으면 FAIL로 두고 빠진 항목을 구체적으로 적는다. 표현이나 표 모양이 예시와 달라도 뜻이 같고 근거를 찾을 수 있으면 인정한다.</p></li></ol><h4>PASS 조건</h4><ul><li><p>서로 다른 여섯 표현을 다뤘다.</p></li><li><p>각 표현의 문제점과 바꾼 문장이 있다.</p></li><li><p>바꾼 문장에 누가 무엇을 알며 모르는지 드러난다.</p></li><li><p>없는 확인 수단·현재 상태·회신을 만들어 표현을 강화하지 않았다.</p></li></ul><h4>FAIL에 해당하는 예</h4><ul><li><p>“확실히”를 “틀림없이”로 바꿨다.</p></li><li><p>원문에 없는 복수 확인 수단을 넣어 확정 문장으로 고쳤다.</p></li></ul><h4>이해 여부를 확인할 질문</h4><p>“아마”라는 말을 모두 없애면 보고가 정확해지는가?</p><h4>인정할 수 있는 답의 핵심</h4><p>추측을 사실처럼 쓰지 않는 것이 중요하며 불확실성은 근거와 함께 드러내야 한다고 답한다.</p><h4>보완과 재평가</h4><p>확정 표현의 동의어로만 바꾼 줄을 찾아 실제 확인 범위와 모르는 내용을 넣어 다시 쓴다.</p><p>수정 전 결과와 재평가 결과를 구분해 남긴다. 실제 하지 않은 보고·채점·시험을 나중에 했던 것처럼 채우지 않는다.</p>', 'attachments': []}, {'step': 3, 'problem': '다섯 영역 점검표와 짧은 기록·상세 기록 양식 만들기', 'activity': '<h4>이번 단계에서 하는 일</h4><p>급할 때 쓴 짧은 기록과 나중의 상세 기록이 같은 사건인지 알 수 있어야 한다. 공통 번호와 필수 항목을 정해 두 양식을 만든다.</p><h4>먼저 읽을 자료</h4>\n\n<h4>영역별 점검 질문</h4>\n\n<ul>\n<li>\n<p><strong>대상 확인:</strong> 본 특징과 소속 주장을 구분했는가</p>\n</li>\n\n<li>\n<p><strong>아군 위치:</strong> 지금 적용할 위치 자료가 있는가</p>\n</li>\n\n<li>\n<p><strong>민간 상황:</strong> 주민·차량·시설 이용 정보가 있는가</p>\n</li>\n\n<li>\n<p><strong>통신·접수:</strong> 요청 송신·접수·회신을 구별했는가</p>\n</li>\n\n<li>\n<p><strong>담당·검토:</strong> 누가 질문하고 검토하며 미완료를 맡는가</p>\n</li>\n</ul>\n\n<h4>두 양식에서 빠지면 안 되는 내용</h4>\n\n<ul>\n<li>\n<p>기록 번호·보고 번호</p>\n</li>\n\n<li>\n<p>관측 시각과 보고·수신 시각</p>\n</li>\n\n<li>\n<p>출처·확인 수단</p>\n</li>\n\n<li>\n<p>보고에 적힌 관찰 내용</p>\n</li>\n\n<li>\n<p>보고자의 주장과 나의 판단</p>\n</li>\n\n<li>\n<p>보냈다고 하거나 실제 수행한 조치</p>\n</li>\n\n<li>\n<p>요청 접수·회신 여부</p>\n</li>\n\n<li>\n<p>확인하지 못한 내용</p>\n</li>\n\n<li>\n<p>확인 질문·담당·다음 점검</p>\n</li>\n\n<li>\n<p>현재 보류 상태</p>\n</li>\n</ul>\n\n<h4>작업 순서</h4><ol><li><p>다섯 영역 각각에 구체적인 점검 질문 두 개 이상을 만들어 모두 열 개 이상을 쓴다.</p></li><li><p>답을 예·아니오·미확인 중 하나로 표시하고 그 근거 문장·담당을 적는 칸을 만든다.</p></li><li><p>짧은 기록 양식을 만든다. 기록 번호, 보고 내용, 모르는 내용·보류 이유, 질문·담당, 접수·회신 상태는 반드시 남긴다.</p></li><li><p>상세 양식에는 표의 열 항목을 넣는다. 짧은 기록과 같은 기록 번호를 쓰게 한다.</p></li><li><p>관측·보고·수신 시각은 다른 칸으로 만들고 자료에 없는 시각을 적을 경우 “미제공”이라고 쓰게 한다.</p></li><li><p>두 양식을 나란히 보며 짧은 기록의 각 항목을 상세 기록 어디에서 찾는지 표시한다.</p></li></ol><h4>작성 예시</h4><p>다음은 쓰는 방식을 보여 주는 예다. 자기 판단이나 실제 수행·측정 결과가 필요한 칸에 그대로 복사하지 않는다.</p><p>기록 N1-01의 짧은 기록과 상세 기록에 같은 번호를 붙인다. “14:05 보고”는 보고 시각에 적고 관측 시각을 모르면 그 칸에는 미제공이라고 적는다.</p><h4>작성할 표 또는 기록 항목</h4><ul><li><p>다섯 영역·질문·답·근거·담당이 있는 점검표</p></li><li><p>짧은 기록 양식</p></li><li><p>상세 기록 양식</p></li><li><p>두 양식의 같은 항목 대응표</p></li></ul><h4>제출할 내용과 끝내는 기준</h4><p>점검 질문 열 개 이상, 짧은 기록·상세 기록 두 양식, 대응표.</p><p>미션 제출 문서 templates/ops07_aar_fact_action_template.md의 STEP 03 아래에 작성한다. 개인 기록과 조별 결과를 구분하고 실제 이름 대신 학습자 코드를 써도 된다. 표의 모양이 달라도 필요한 내용을 찾을 수 있으면 된다.</p><h4>마치기 전 확인</h4><ul><li><p>다섯 영역에 각각 두 질문 이상 있다.</p></li><li><p>짧은 양식의 필수 항목과 상세 양식의 열 항목이 있다.</p></li><li><p>같은 기록 번호로 두 양식을 연결할 수 있다.</p></li><li><p>예·아니오·미확인을 구분하며 빈칸을 예 또는 완료로 간주하지 않는다.</p></li></ul><h4>배정 시간</h4><p>45분</p>', 'evalText': '<h4>평가할 자료와 위치</h4><p>templates/ops07_aar_fact_action_template.md의 STEP 03 결과와 해당 개인 수행 기록을 확인한다. 이 단계에서 요청한 제출 내용은 다음과 같다: 점검 질문 열 개 이상, 짧은 기록·상세 기록 두 양식, 대응표.</p><h4>평가 순서</h4><ol><li><p>아래 통과 조건을 하나씩 읽고 제출물에서 확인한 위치를 옆에 적는다.</p></li><li><p>실제 설명·보고·인계가 필요한 단계는 수행 기록이나 직접 관찰로 확인한다. 서식만 있으면 수행한 것으로 보지 않는다.</p></li><li><p>통과 조건을 모두 충족하면 이 단계는 PASS다. 하나라도 빠지거나 반대로 적었으면 FAIL로 두고 빠진 항목을 구체적으로 적는다. 표현이나 표 모양이 예시와 달라도 뜻이 같고 근거를 찾을 수 있으면 인정한다.</p></li></ol><h4>PASS 조건</h4><ul><li><p>다섯 영역에 각각 두 질문 이상 있다.</p></li><li><p>짧은 양식의 필수 항목과 상세 양식의 열 항목이 있다.</p></li><li><p>같은 기록 번호로 두 양식을 연결할 수 있다.</p></li><li><p>예·아니오·미확인을 구분하며 빈칸을 예 또는 완료로 간주하지 않는다.</p></li></ul><h4>FAIL에 해당하는 예</h4><ul><li><p>점검표를 체크하면 보류 해제로 자동 처리하게 했다.</p></li><li><p>관측·보고 시각을 한 칸으로 합쳐 어떤 시각인지 알 수 없다.</p></li></ul><h4>이해 여부를 확인할 질문</h4><p>짧은 기록에서 상세 기록을 찾을 때 사용할 번호는 무엇인가?</p><h4>인정할 수 있는 답의 핵심</h4><p>두 양식에 동일하게 붙인 기록 번호를 가리키고 실제 연결 위치를 보여 준다.</p><h4>보완과 재평가</h4><p>빠진 항목을 추가하고 같은 번호의 짧은 기록에서 상세 기록을 실제로 찾아본다.</p><p>수정 전 결과와 재평가 결과를 구분해 남긴다. 실제 하지 않은 보고·채점·시험을 나중에 했던 것처럼 채우지 않는다.</p>', 'attachments': []}, {'step': 4, 'problem': '세 원문을 두 양식에 옮기고 왜곡된 내용을 세기', 'activity': '<h4>이번 단계에서 하는 일</h4><p>양식이 좋아도 옮겨 적으며 추측을 사실로 바꾸면 소용없다. 다른 사람이 원문과 대조해 오류를 찾아 고친다.</p><h4>먼저 읽을 자료</h4>\n\n<h4>사례 자료</h4>\n\n<p><strong>N1</strong></p>\n\n<p>14:05 OP-2 보고.</p>\n\n<p>교량 북단에 적 차량 2대가 확실히 있음.</p>\n\n<p>아마 공격 준비 중인 듯.</p>\n\n<p>영상은 배터리 교체로 10분간 못 봄.</p>\n\n<p>가상 안전 검토자에게 보고함.</p>\n\n<p><strong>N2</strong></p>\n\n<p>14:20 정찰조는 문제없음.</p>\n\n<p>골목 인원 4명은 적으로 보임.</p>\n\n<p>무전은 계속 안 됨.</p>\n\n<p>일단 대기시킴.</p>\n\n<p><strong>N3</strong></p>\n\n<p>14:40 학교 옥상 인원은 전부 적 관측반임.</p>\n\n<p>주민 말로는 대피소라는데 확인 안 됨.</p>\n\n<p>상급 검토 요청을 시도했으나 통신 두절.</p>\n\n<p>사용할 양식이 없다면 다음 필수 칸부터 만든다: 기록 번호, 원문, 관찰 보고, 주장·판단, 조치, 미확인, 시각, 질문·담당, 접수·회신, 보류 상태.</p>\n\n<p>짧은 기록에는 핵심을 적고 상세 기록에는 빠짐없이 적는다.</p>\n\n<h4>작업 순서</h4><ol><li><p>N1~N3에 같은 방식의 기록 번호를 붙이고 짧은 기록과 상세 기록을 각각 작성한다. 모두 여섯 기록이다.</p></li><li><p>작성자가 아닌 조원이 원문의 각 내용을 짚으며 두 기록을 비교한다.</p></li><li><p>추측을 확인 사실로 바꾼 것, 원문에 없는 내용을 추가한 것, 미확인 내용을 빼먹은 것을 각각 찾는다.</p></li><li><p>오류마다 기록 번호·문제 문장·원문·수정 문장을 적고 종류별 건수를 센다. 처음 오류가 없으면 0건이라고 쓴다.</p></li><li><p>수정 전 기록을 남기고 수정본을 만든다.</p></li><li><p>교관 또는 다른 검토자가 다시 읽어 같은 오류가 남는지 확인한다. 수정 후 세 종류 오류가 모두 0건이어야 이 작업을 마친다.</p></li></ol><h4>작성 예시</h4><p>다음은 쓰는 방식을 보여 주는 예다. 자기 판단이나 실제 수행·측정 결과가 필요한 칸에 그대로 복사하지 않는다.</p><p>오류: N3 요청 접수 칸에 “접수 완료”라고 적음 / 원문 “요청하려 했으나 통신 두절” / 수정 “요청 시도만 보고됨, 접수 여부 미확인”.</p><h4>작성할 표 또는 기록 항목</h4><ul><li><p>원문별 짧은 기록·상세 기록</p></li><li><p>오류 번호·종류·문제 문장·원문·수정 문장</p></li><li><p>수정 전·후 오류 건수</p></li><li><p>검토자</p></li></ul><h4>제출할 내용과 끝내는 기준</h4><p>N1~N3의 여섯 기록과 오류 수정표. 수정 전후를 모두 남긴다.</p><p>미션 제출 문서 templates/ops07_aar_fact_action_template.md의 STEP 04 아래에 작성한다. 개인 기록과 조별 결과를 구분하고 실제 이름 대신 학습자 코드를 써도 된다. 표의 모양이 달라도 필요한 내용을 찾을 수 있으면 된다.</p><h4>마치기 전 확인</h4><ul><li><p>세 원문을 짧은 기록·상세 기록 모두로 옮겼다.</p></li><li><p>원문과 기록 번호로 서로 찾을 수 있다.</p></li><li><p>수정 전 오류 문장과 수정 결과를 보존했다.</p></li><li><p>최종본에 추측의 사실화·없는 내용 추가·미확인 누락이 각각 0건이다.</p></li></ul><h4>배정 시간</h4><p>45분</p>', 'evalText': '<h4>평가할 자료와 위치</h4><p>templates/ops07_aar_fact_action_template.md의 STEP 04 결과와 해당 개인 수행 기록을 확인한다. 이 단계에서 요청한 제출 내용은 다음과 같다: N1~N3의 여섯 기록과 오류 수정표. 수정 전후를 모두 남긴다.</p><h4>평가 순서</h4><ol><li><p>아래 통과 조건을 하나씩 읽고 제출물에서 확인한 위치를 옆에 적는다.</p></li><li><p>실제 설명·보고·인계가 필요한 단계는 수행 기록이나 직접 관찰로 확인한다. 서식만 있으면 수행한 것으로 보지 않는다.</p></li><li><p>통과 조건을 모두 충족하면 이 단계는 PASS다. 하나라도 빠지거나 반대로 적었으면 FAIL로 두고 빠진 항목을 구체적으로 적는다. 표현이나 표 모양이 예시와 달라도 뜻이 같고 근거를 찾을 수 있으면 인정한다.</p></li></ol><h4>PASS 조건</h4><ul><li><p>세 원문을 짧은 기록·상세 기록 모두로 옮겼다.</p></li><li><p>원문과 기록 번호로 서로 찾을 수 있다.</p></li><li><p>수정 전 오류 문장과 수정 결과를 보존했다.</p></li><li><p>최종본에 추측의 사실화·없는 내용 추가·미확인 누락이 각각 0건이다.</p></li></ul><h4>FAIL에 해당하는 예</h4><ul><li><p>오류 건수는 0이라고 했지만 N3의 접수 완료 표현이 남아 있다.</p></li><li><p>수정 전 문서를 지워 무엇을 바로잡았는지 알 수 없다.</p></li></ul><h4>이해 여부를 확인할 질문</h4><p>N3의 요청 시도와 접수 완료는 왜 다른가?</p><h4>인정할 수 있는 답의 핵심</h4><p>요청을 시도했다는 내용만 있고 상대가 받았다는 정보는 없기 때문이라고 답한다.</p><h4>보완과 재평가</h4><p>원문을 한 조각씩 다시 대조해 오류를 고치고 새 검토자의 재확인 결과를 남긴다.</p><p>수정 전 결과와 재평가 결과를 구분해 남긴다. 실제 하지 않은 보고·채점·시험을 나중에 했던 것처럼 채우지 않는다.</p>', 'attachments': []}, {'step': 5, 'problem': '짧은 기록을 읽고 상대가 상세 기록을 찾게 하기', 'activity': '<h4>이번 단계에서 하는 일</h4><p>기록의 목적은 다음 사람이 원문과 남은 질문을 찾아 일을 이어가는 것이다. 짧은 기록만 들은 사람이 상세 기록을 찾을 수 있는지 시험한다.</p><h4>먼저 읽을 자료</h4>\n\n<h4>사례 자료</h4>\n\n<p><strong>N1</strong></p>\n\n<p>14:05 OP-2 보고.</p>\n\n<p>교량 북단에 적 차량 2대가 확실히 있음.</p>\n\n<p>아마 공격 준비 중인 듯.</p>\n\n<p>영상은 배터리 교체로 10분간 못 봄.</p>\n\n<p>가상 안전 검토자에게 보고함.</p>\n\n<p><strong>N2</strong></p>\n\n<p>14:20 정찰조는 문제없음.</p>\n\n<p>골목 인원 4명은 적으로 보임.</p>\n\n<p>무전은 계속 안 됨.</p>\n\n<p>일단 대기시킴.</p>\n\n<p><strong>N3</strong></p>\n\n<p>14:40 학교 옥상 인원은 전부 적 관측반임.</p>\n\n<p>주민 말로는 대피소라는데 확인 안 됨.</p>\n\n<p>상급 검토 요청을 시도했으나 통신 두절.</p>\n\n<p>각자 N1~N3 중 하나를 고른다.</p>\n\n<p>사용할 기록이 없으면 원문으로 짧은 기록과 상세 기록을 작성한다.</p>\n\n<p>기록 번호는 두 양식에 같게 붙인다.</p>\n\n<h4>작업 순서</h4><ol><li><p>보고자는 원문과 모르는 내용을 확인한 뒤 짧은 기록을 읽는다.</p></li><li><p>받는 사람은 기록 번호로 상세 기록을 찾아 원문 위치를 가리킨다.</p></li><li><p>받는 사람은 아직 확인하지 못한 내용·확인 질문·담당을 말한다.</p></li><li><p>보고자는 잘못 이해한 부분을 바로잡고 수정한 항목을 기록한다.</p></li><li><p>역할을 바꿔 전원이 한 번 이상 보고하거나 기록을 찾는 과정을 수행한다. 기록 찾기에 실패하면 이유를 고친 뒤 다시 시험한다.</p></li></ol><h4>작성 예시</h4><p>다음은 쓰는 방식을 보여 주는 예다. 자기 판단이나 실제 수행·측정 결과가 필요한 칸에 그대로 복사하지 않는다.</p><p>“N2-01입니다”라는 짧은 보고를 듣고 상대가 같은 번호의 상세 기록을 찾아 “정찰조 현재 상태는 미확인, 다음 질문은 현재 위치 보고 여부”라고 설명한다.</p><h4>작성할 표 또는 기록 항목</h4><ul><li><p>보고자·받는 사람</p></li><li><p>기록 번호</p></li><li><p>원문·상세 기록을 찾았는지</p></li><li><p>남은 질문과 담당</p></li><li><p>고친 항목</p></li><li><p>다시 시험한 결과</p></li></ul><h4>제출할 내용과 끝내는 기준</h4><p>전원 참여 기록과 짧은 기록→상세 기록→원문 찾기 시험 결과.</p><p>미션 제출 문서 templates/ops07_aar_fact_action_template.md의 STEP 05 아래에 작성한다. 개인 기록과 조별 결과를 구분하고 실제 이름 대신 학습자 코드를 써도 된다. 표의 모양이 달라도 필요한 내용을 찾을 수 있으면 된다.</p><h4>마치기 전 확인</h4><ul><li><p>전원의 실제 참여 내용을 확인할 수 있다.</p></li><li><p>짧은 기록의 번호로 상세 기록과 원문을 찾았다.</p></li><li><p>받는 사람이 미확인 내용·질문·담당을 설명했다.</p></li><li><p>실패한 경우 이유와 수정 후 재시험 결과가 있다.</p></li></ul><h4>배정 시간</h4><p>30분</p>', 'evalText': '<h4>평가할 자료와 위치</h4><p>templates/ops07_aar_fact_action_template.md의 STEP 05 결과와 해당 개인 수행 기록을 확인한다. 이 단계에서 요청한 제출 내용은 다음과 같다: 전원 참여 기록과 짧은 기록→상세 기록→원문 찾기 시험 결과.</p><h4>평가 순서</h4><ol><li><p>아래 통과 조건을 하나씩 읽고 제출물에서 확인한 위치를 옆에 적는다.</p></li><li><p>실제 설명·보고·인계가 필요한 단계는 수행 기록이나 직접 관찰로 확인한다. 서식만 있으면 수행한 것으로 보지 않는다.</p></li><li><p>통과 조건을 모두 충족하면 이 단계는 PASS다. 하나라도 빠지거나 반대로 적었으면 FAIL로 두고 빠진 항목을 구체적으로 적는다. 표현이나 표 모양이 예시와 달라도 뜻이 같고 근거를 찾을 수 있으면 인정한다.</p></li></ol><h4>PASS 조건</h4><ul><li><p>전원의 실제 참여 내용을 확인할 수 있다.</p></li><li><p>짧은 기록의 번호로 상세 기록과 원문을 찾았다.</p></li><li><p>받는 사람이 미확인 내용·질문·담당을 설명했다.</p></li><li><p>실패한 경우 이유와 수정 후 재시험 결과가 있다.</p></li></ul><h4>FAIL에 해당하는 예</h4><ul><li><p>“잘 찾음”만 적고 어떤 기록과 질문을 찾았는지 없다.</p></li><li><p>파일은 찾았지만 남은 질문·담당을 설명하지 못했다.</p></li></ul><h4>이해 여부를 확인할 질문</h4><p>이 기록에서 아직 확인하지 못한 내용과 그 질문을 맡을 사람을 짚어 보라.</p><h4>인정할 수 있는 답의 핵심</h4><p>선택한 원문에 실제로 없는 내용을 골라 기록의 질문·담당 칸을 정확히 찾는다.</p><h4>보완과 재평가</h4><p>번호·질문·담당을 찾기 쉽게 고치고 다른 조원이 같은 방법으로 다시 찾게 한다.</p><p>수정 전 결과와 재평가 결과를 구분해 남긴다. 실제 하지 않은 보고·채점·시험을 나중에 했던 것처럼 채우지 않는다.</p>', 'attachments': []}], 'deliverables': ['templates/ops07_aar_fact_action_template.md (필수) — 세 원문의 문장 분류, 여섯 표현 수정표, 다섯 영역 점검표, 두 기록 양식과 이관·사용 시험'], 'techStack': ['문서 편집기(Markdown · templates/ops07_aar_fact_action_template.md) 또는 인쇄 양식', 'N1~N3 원문 문장 분류표 및 6대 모호 표현 수정표', '5영역(대상·아군·민간·통신·담당) 점검표', '공통 기록 번호 연계 요약형·상세형 기록 서식 및 이관 시험지'], 'evaluationKeys': ['<h4>평가 질문</h4><p>세 보고와 표현을 정리하고 원문 왜곡 없이 두 양식으로 옮겼는가</p><h4>평가자가 확인할 위치</h4><p>templates/ops07_aar_fact_action_template.md의 STEP 01, STEP 02, STEP 04과 그 단계의 개인 기록을 확인한다.</p><h4>PASS를 주는 방법</h4><p>아래 단계별 항목을 모두 충족해야 이 평가 질문에 PASS를 준다. 항목 옆에 확인한 표·줄·회차를 적는다. 한 항목이라도 없거나 틀리면 FAIL 사유에 해당 단계와 문장을 적는다. 학습자가 쓴 말이 예시와 똑같을 필요는 없지만 확인할 내용은 모두 있어야 한다.</p><h4>STEP 01 세 보고를 문장별로 나눠 관찰·주장·조치를 구분하기</h4><p>찾을 결과: 세 보고의 문장 분류표. 원문의 모든 내용을 분류하고 빠진 문장이 없는지 확인한다.</p><ul><li><p>N1~N3 원문을 모두 보존했다.</p></li><li><p>원문의 각 내용에 분류와 설명이 있다.</p></li><li><p>적·공격 의도·안전하다는 주장을 외부 사실로 바꾸지 않았다.</p></li><li><p>보고 시각과 관측 시각을 혼동하지 않았다.</p></li></ul><h4>STEP 02 오해를 부르는 표현 여섯 가지를 바꾸기</h4><p>찾을 결과: 서로 다른 표현 여섯 가지 이상과 표현별 수정 문장.</p><ul><li><p>서로 다른 여섯 표현을 다뤘다.</p></li><li><p>각 표현의 문제점과 바꾼 문장이 있다.</p></li><li><p>바꾼 문장에 누가 무엇을 알며 모르는지 드러난다.</p></li><li><p>없는 확인 수단·현재 상태·회신을 만들어 표현을 강화하지 않았다.</p></li></ul><h4>STEP 04 세 원문을 두 양식에 옮기고 왜곡된 내용을 세기</h4><p>찾을 결과: N1~N3의 여섯 기록과 오류 수정표. 수정 전후를 모두 남긴다.</p><ul><li><p>세 원문을 짧은 기록·상세 기록 모두로 옮겼다.</p></li><li><p>원문과 기록 번호로 서로 찾을 수 있다.</p></li><li><p>수정 전 오류 문장과 수정 결과를 보존했다.</p></li><li><p>최종본에 추측의 사실화·없는 내용 추가·미확인 누락이 각각 0건이다.</p></li></ul><h4>이런 결과는 통과시키지 않는다</h4><ul><li><p>N2의 “정찰조는 문제없음”을 확인된 안전 상태로 적었다.</p></li><li><p>“확실히”를 “틀림없이”로 바꿨다.</p></li><li><p>오류 건수는 0이라고 했지만 N3의 접수 완료 표현이 남아 있다.</p></li></ul><h4>평가 의견 작성 방법</h4><p>“근거 부족”이라고만 쓰지 말고 단계 번호, 충족하지 못한 조건, 제출물에서 실제로 빠지거나 잘못된 부분, 다시 할 일을 적는다. 예를 들어 STEP 01의 첫 조건은 “N1~N3 원문을 모두 보존했다.”이다. 이 조건에 못 미쳤다면 어떤 줄·항목·참여자 기록이 빠졌는지 지목한다.</p><h4>다시 평가할 때</h4><p>틀리거나 빠진 내용만 구체적으로 알려 준다. 문서 작성 오류는 원자료와 수정본을 대조한다. 실제 수행 오류는 해당 행동을 다시 하게 하고 새 회차 기록으로 확인한다. 최초 결과는 지우지 않는다.</p>', '<h4>평가 질문</h4><p>점검표·기록 양식을 만들고 다른 사람이 원문과 질문을 찾았는가</p><h4>평가자가 확인할 위치</h4><p>templates/ops07_aar_fact_action_template.md의 STEP 03, STEP 05과 그 단계의 개인 기록을 확인한다.</p><h4>PASS를 주는 방법</h4><p>아래 단계별 항목을 모두 충족해야 이 평가 질문에 PASS를 준다. 항목 옆에 확인한 표·줄·회차를 적는다. 한 항목이라도 없거나 틀리면 FAIL 사유에 해당 단계와 문장을 적는다. 학습자가 쓴 말이 예시와 똑같을 필요는 없지만 확인할 내용은 모두 있어야 한다.</p><h4>STEP 03 다섯 영역 점검표와 짧은 기록·상세 기록 양식 만들기</h4><p>찾을 결과: 점검 질문 열 개 이상, 짧은 기록·상세 기록 두 양식, 대응표.</p><ul><li><p>다섯 영역에 각각 두 질문 이상 있다.</p></li><li><p>짧은 양식의 필수 항목과 상세 양식의 열 항목이 있다.</p></li><li><p>같은 기록 번호로 두 양식을 연결할 수 있다.</p></li><li><p>예·아니오·미확인을 구분하며 빈칸을 예 또는 완료로 간주하지 않는다.</p></li></ul><h4>STEP 05 짧은 기록을 읽고 상대가 상세 기록을 찾게 하기</h4><p>찾을 결과: 전원 참여 기록과 짧은 기록→상세 기록→원문 찾기 시험 결과.</p><ul><li><p>전원의 실제 참여 내용을 확인할 수 있다.</p></li><li><p>짧은 기록의 번호로 상세 기록과 원문을 찾았다.</p></li><li><p>받는 사람이 미확인 내용·질문·담당을 설명했다.</p></li><li><p>실패한 경우 이유와 수정 후 재시험 결과가 있다.</p></li></ul><h4>이런 결과는 통과시키지 않는다</h4><ul><li><p>점검표를 체크하면 보류 해제로 자동 처리하게 했다.</p></li><li><p>“잘 찾음”만 적고 어떤 기록과 질문을 찾았는지 없다.</p></li></ul><h4>평가 의견 작성 방법</h4><p>“근거 부족”이라고만 쓰지 말고 단계 번호, 충족하지 못한 조건, 제출물에서 실제로 빠지거나 잘못된 부분, 다시 할 일을 적는다. 예를 들어 STEP 03의 첫 조건은 “다섯 영역에 각각 두 질문 이상 있다.”이다. 이 조건에 못 미쳤다면 어떤 줄·항목·참여자 기록이 빠졌는지 지목한다.</p><h4>다시 평가할 때</h4><p>틀리거나 빠진 내용만 구체적으로 알려 준다. 문서 작성 오류는 원자료와 수정본을 대조한다. 실제 수행 오류는 해당 행동을 다시 하게 하고 새 회차 기록으로 확인한다. 최초 결과는 지우지 않는다.</p>'], 'evaluationTypes': ['REQUIRED', 'REQUIRED'], 'evaluationText': '<h4>미션 전체 판정</h4><p>이 미션은 아래 두 평가 질문이 모두 PASS여야 통과다. 조별 문서와 개인의 실제 수행을 나눠 확인한다. 수행하지 않았거나 기록이 없어 확인할 수 없는 사람은 미평가로 남기고 다시 실시한다. 최종 판정에서는 미평가를 PASS로 바꾸지 않는다.</p><h4>평가 질문</h4><p>세 보고와 표현을 정리하고 원문 왜곡 없이 두 양식으로 옮겼는가</p><h4>평가자가 확인할 위치</h4><p>templates/ops07_aar_fact_action_template.md의 STEP 01, STEP 02, STEP 04과 그 단계의 개인 기록을 확인한다.</p><h4>PASS를 주는 방법</h4><p>아래 단계별 항목을 모두 충족해야 이 평가 질문에 PASS를 준다. 항목 옆에 확인한 표·줄·회차를 적는다. 한 항목이라도 없거나 틀리면 FAIL 사유에 해당 단계와 문장을 적는다. 학습자가 쓴 말이 예시와 똑같을 필요는 없지만 확인할 내용은 모두 있어야 한다.</p><h4>STEP 01 세 보고를 문장별로 나눠 관찰·주장·조치를 구분하기</h4><p>찾을 결과: 세 보고의 문장 분류표. 원문의 모든 내용을 분류하고 빠진 문장이 없는지 확인한다.</p><ul><li><p>N1~N3 원문을 모두 보존했다.</p></li><li><p>원문의 각 내용에 분류와 설명이 있다.</p></li><li><p>적·공격 의도·안전하다는 주장을 외부 사실로 바꾸지 않았다.</p></li><li><p>보고 시각과 관측 시각을 혼동하지 않았다.</p></li></ul><h4>STEP 02 오해를 부르는 표현 여섯 가지를 바꾸기</h4><p>찾을 결과: 서로 다른 표현 여섯 가지 이상과 표현별 수정 문장.</p><ul><li><p>서로 다른 여섯 표현을 다뤘다.</p></li><li><p>각 표현의 문제점과 바꾼 문장이 있다.</p></li><li><p>바꾼 문장에 누가 무엇을 알며 모르는지 드러난다.</p></li><li><p>없는 확인 수단·현재 상태·회신을 만들어 표현을 강화하지 않았다.</p></li></ul><h4>STEP 04 세 원문을 두 양식에 옮기고 왜곡된 내용을 세기</h4><p>찾을 결과: N1~N3의 여섯 기록과 오류 수정표. 수정 전후를 모두 남긴다.</p><ul><li><p>세 원문을 짧은 기록·상세 기록 모두로 옮겼다.</p></li><li><p>원문과 기록 번호로 서로 찾을 수 있다.</p></li><li><p>수정 전 오류 문장과 수정 결과를 보존했다.</p></li><li><p>최종본에 추측의 사실화·없는 내용 추가·미확인 누락이 각각 0건이다.</p></li></ul><h4>이런 결과는 통과시키지 않는다</h4><ul><li><p>N2의 “정찰조는 문제없음”을 확인된 안전 상태로 적었다.</p></li><li><p>“확실히”를 “틀림없이”로 바꿨다.</p></li><li><p>오류 건수는 0이라고 했지만 N3의 접수 완료 표현이 남아 있다.</p></li></ul><h4>평가 의견 작성 방법</h4><p>“근거 부족”이라고만 쓰지 말고 단계 번호, 충족하지 못한 조건, 제출물에서 실제로 빠지거나 잘못된 부분, 다시 할 일을 적는다. 예를 들어 STEP 01의 첫 조건은 “N1~N3 원문을 모두 보존했다.”이다. 이 조건에 못 미쳤다면 어떤 줄·항목·참여자 기록이 빠졌는지 지목한다.</p><h4>다시 평가할 때</h4><p>틀리거나 빠진 내용만 구체적으로 알려 준다. 문서 작성 오류는 원자료와 수정본을 대조한다. 실제 수행 오류는 해당 행동을 다시 하게 하고 새 회차 기록으로 확인한다. 최초 결과는 지우지 않는다.</p><h4>평가 질문</h4><p>점검표·기록 양식을 만들고 다른 사람이 원문과 질문을 찾았는가</p><h4>평가자가 확인할 위치</h4><p>templates/ops07_aar_fact_action_template.md의 STEP 03, STEP 05과 그 단계의 개인 기록을 확인한다.</p><h4>PASS를 주는 방법</h4><p>아래 단계별 항목을 모두 충족해야 이 평가 질문에 PASS를 준다. 항목 옆에 확인한 표·줄·회차를 적는다. 한 항목이라도 없거나 틀리면 FAIL 사유에 해당 단계와 문장을 적는다. 학습자가 쓴 말이 예시와 똑같을 필요는 없지만 확인할 내용은 모두 있어야 한다.</p><h4>STEP 03 다섯 영역 점검표와 짧은 기록·상세 기록 양식 만들기</h4><p>찾을 결과: 점검 질문 열 개 이상, 짧은 기록·상세 기록 두 양식, 대응표.</p><ul><li><p>다섯 영역에 각각 두 질문 이상 있다.</p></li><li><p>짧은 양식의 필수 항목과 상세 양식의 열 항목이 있다.</p></li><li><p>같은 기록 번호로 두 양식을 연결할 수 있다.</p></li><li><p>예·아니오·미확인을 구분하며 빈칸을 예 또는 완료로 간주하지 않는다.</p></li></ul><h4>STEP 05 짧은 기록을 읽고 상대가 상세 기록을 찾게 하기</h4><p>찾을 결과: 전원 참여 기록과 짧은 기록→상세 기록→원문 찾기 시험 결과.</p><ul><li><p>전원의 실제 참여 내용을 확인할 수 있다.</p></li><li><p>짧은 기록의 번호로 상세 기록과 원문을 찾았다.</p></li><li><p>받는 사람이 미확인 내용·질문·담당을 설명했다.</p></li><li><p>실패한 경우 이유와 수정 후 재시험 결과가 있다.</p></li></ul><h4>이런 결과는 통과시키지 않는다</h4><ul><li><p>점검표를 체크하면 보류 해제로 자동 처리하게 했다.</p></li><li><p>“잘 찾음”만 적고 어떤 기록과 질문을 찾았는지 없다.</p></li></ul><h4>평가 의견 작성 방법</h4><p>“근거 부족”이라고만 쓰지 말고 단계 번호, 충족하지 못한 조건, 제출물에서 실제로 빠지거나 잘못된 부분, 다시 할 일을 적는다. 예를 들어 STEP 03의 첫 조건은 “다섯 영역에 각각 두 질문 이상 있다.”이다. 이 조건에 못 미쳤다면 어떤 줄·항목·참여자 기록이 빠졌는지 지목한다.</p><h4>다시 평가할 때</h4><p>틀리거나 빠진 내용만 구체적으로 알려 준다. 문서 작성 오류는 원자료와 수정본을 대조한다. 실제 수행 오류는 해당 행동을 다시 하게 하고 새 회차 기록으로 확인한다. 최초 결과는 지우지 않는다.</p>', 'aiGuidelines': {'allow': ['N1~N3 원문 문장을 관찰·주장·조치로 나눈 분류표에 해석 왜곡이 없는지 검토받기', '금지 표현 6종("확실히"·"적으로 보임"·"문제없음"·"전부"·"항상"·"일단")을 확인 사실과 미확인 범위가 드러나는 대체 문장으로 다듬기', '5영역 점검표(질문 각 2개 이상)와 공통 기록 번호 기반 요약형·상세형 기록 양식 항목 정합성 점검받기'], 'rules': ['AI 사용은 필수가 아니며, 원문 N1~N3과 수정 전 오류 문장은 지우지 않고 templates/ops07_aar_fact_action_template.md에 보존한다.', '점검표·기록 양식에서 예·아니오·미확인을 구분하고 빈칸을 완료로 간주하지 않는다.', 'STEP 04 교차 왜곡 대조와 STEP 05 짧은 기록 기반 상세 기록 탐색·인계 시험은 학습자가 직접 수행한다.'], 'forbid': ['실제 작전 보고 원문이나 부대 자료를 입력하기', 'N1~N3의 추측·주장을 확인된 사실로 바꾸거나 없는 확인 수단·회신을 넣어 문장을 강화하기', '왜곡 건수 0건 확인이나 상대방 탐색 시험 결과를 실제 점검 없이 작성하기']}}, {'id': 'OPS07_m07', 'sequence': 7, 'title': '연락 두절과 시간 압박에서도 보류와 보고 유지', 'durationHours': 4, 'overview': '<p>순서대로 들어오는 카드마다 보고와 보류 상태를 갱신한다. 모든 조원이 보고자로 수행하고 다섯 영역을 평가받는다.</p><p>아래 다섯 단계에는 필요한 자료·작업 순서·예시·제출 내용과 평가 조건이 들어 있다. 각 단계의 자료를 읽은 뒤 순서대로 작성한다.</p>', 'objectives': '<h4>이 미션을 마치면 할 수 있어야 하는 일</h4><ul><li><p>순서대로 들어오는 카드마다 보고와 보류 상태를 갱신한다.</p></li><li><p>모든 조원이 보고자로 수행하고 다섯 영역을 평가받는다.</p></li><li><p>평가자 일치율과 개인의 수행 성취를 구별한다.</p></li></ul>', 'scenario': '<h4>학습자의 역할</h4><p>학습자는 민간인과 아군에게 불필요한 피해가 생기지 않도록 보고를 읽고, 모르는 내용을 질문하고, 다음 담당자에게 기록을 넘기는 역할을 연습한다. 사람의 소속을 맞히거나 실제 타격을 결정하는 과제가 아니다.</p><h4>이번 미션의 과제</h4><ul><li><p>순서대로 들어오는 카드마다 보고와 보류 상태를 갱신한다.</p></li><li><p>모든 조원이 보고자로 수행하고 다섯 영역을 평가받는다.</p></li><li><p>평가자 일치율과 개인의 수행 성취를 구별한다.</p></li></ul><h4>자료와 역할의 범위</h4><p>사례·시각·장소·역할은 모두 교육용이다. 실제 부대 자료를 쓰지 않는다. 자료에 답이 없으면 자료에 없다고 적는다. 실제 행동 권한은 이 수업이 정하지 않는다.</p><h4>개인 작업과 조별 작업</h4><p>개인 판단을 먼저 요청한 단계에서는 토의 전에 자신의 답을 남긴다. 그 뒤 조별 답과 수정 이유를 별도로 쓴다. 보고·인계·시험 단계에서는 실제 한 말과 행동을 기록한다.</p>', 'attachments': [], 'materials': ['/uploads/OPS07/OPS07_m07_student.ipynb', '/uploads/OPS07/OPS07_m07_workbook.ipynb'], 'materialText': '<h4>실습 노트북 안내 (선택 보조 도구)</h4><p><strong>학생 실습용</strong>: <code>OPS07_m07_student.ipynb</code> (직접 ANSWER 표와 note를 작성합니다.)<br><strong>교수자·해설용</strong>: <code>OPS07_m07_workbook.ipynb</code> (예시 풀이와 해설 note가 포함되어 있습니다.)<br><em>※ 본 과제는 문서 편집기나 인쇄 양식으로도 수행할 수 있으며, 노트북은 사례·질문·기록 양식을 표로 정리하는 선택 보조 도구입니다.</em></p><h4>자료를 읽는 순서</h4>\n\n<ol>\n<li><p>STEP 1에서는 아래 다섯 평가 영역과 점수별 행동 설명을 먼저 읽습니다. 모든 조원이 보고자 역할을 맡도록 역할표를 작성합니다.</p></li>\n<li><p>STEP 2의 D0~D4는 한 회차에서 순서대로 추가되는 카드입니다. 통제관이 각 시점에 알려 준 내용까지 사용하고, 아직 공개하지 않은 카드의 답을 앞당겨 쓰지 않습니다.</p></li>\n<li><p>STEP 3~5에서는 실제 발언·보고·인계를 기록하고 두 평가자가 독립적으로 채점합니다. 일치율과 개인별 영역 점수는 따로 확인합니다.</p></li>\n</ol>\n\n<h4>자료 1. 다섯 영역의 수행 평가 기준</h4>\n\n<p>각 영역에서 실제로 관찰한 말과 행동을 아래 설명과 대조합니다. 점수만 남기지 말고 그 판단의 근거도 기록합니다.</p>\n\n<h4>대상 확인</h4>\n\n<p><strong>4점 · 탁월:</strong> 본 장면·소속 주장·모르는 내용을 구분하고 새 보고의 출처·시각·모순까지 스스로 추가해 보고함</p>\n\n<p><strong>3점 · 우수:</strong> 본 장면과 소속 주장을 구분하고 모르는 내용을 도움 없이 보고함</p>\n\n<p><strong>2점 · 보통:</strong> 교관이 물은 뒤에야 본 장면과 주장 또는 모르는 내용을 구분함</p>\n\n<p><strong>1점 · 미흡:</strong> 복장·표식·물체나 점수만으로 소속을 확정하거나 모르는 내용을 확인됐다고 말함</p>\n\n<h4>위험 파악</h4>\n\n<p><strong>4점 · 탁월:</strong> 민간인과 아군 위험을 따로 말하고 새 카드가 나올 때 질문·담당을 고침</p>\n\n<p><strong>3점 · 우수:</strong> 민간인과 아군 위험 및 모르는 내용을 도움 없이 말함</p>\n\n<p><strong>2점 · 보통:</strong> 한쪽 위험을 빼먹었으나 교관이 묻자 보완함</p>\n\n<p><strong>1점 · 미흡:</strong> 민간인 보고나 아군 위치 미확인을 무시하거나 진행 이유로 삼음</p>\n\n<h4>보류와 재확인</h4>\n\n<p><strong>4점 · 탁월:</strong> 보류를 유지하고 질문·담당·요청 접수·다음 확인까지 스스로 정리함</p>\n\n<p><strong>3점 · 우수:</strong> 압박·미회신·일시 연락 회복에도 보류를 유지하고 재확인을 요청함</p>\n\n<p><strong>2점 · 보통:</strong> 보류는 유지하나 질문·담당을 정하는 데 도움이 필요함</p>\n\n<p><strong>1점 · 미흡:</strong> 압박 발언·등급·연락 회복만으로 필수 확인을 생략하고 진행하겠다고 판단함</p>\n\n<h4>보고</h4>\n\n<p><strong>4점 · 탁월:</strong> 본 내용·모르는 내용·이유·질문을 보고하고 상대가 맡을 일을 다시 말하게 해 접수 범위를 확인함</p>\n\n<p><strong>3점 · 우수:</strong> 본 내용·모르는 내용·보류 이유·질문을 도움 없이 보고하고 접수와 회신을 구별함</p>\n\n<p><strong>2점 · 보통:</strong> 필수 보고 내용이나 접수 확인을 빼먹어 교관 질문 뒤에 보완함</p>\n\n<p><strong>1점 · 미흡:</strong> 요청 미송신·미접수·미회신을 보고 완료나 확인 완료로 말함</p>\n\n<h4>기록</h4>\n\n<p><strong>4점 · 탁월:</strong> 카드·시각·보고·보류·질문·담당·미답을 남겨 다음 사람이 그대로 이어서 확인할 수 있음</p>\n\n<p><strong>3점 · 우수:</strong> 관찰·주장·조치·미확인과 보류 상태·담당을 도움 없이 구분해 적음</p>\n\n<p><strong>2점 · 보통:</strong> 기록은 있지만 일부 구분·담당이 빠져 교관 질문 뒤에 고침</p>\n\n<p><strong>1점 · 미흡:</strong> 없는 사실·측정·회신을 만들거나 미확인을 확인 완료로 적음</p>\n\n<h4>자료 2. 통제관이 순서대로 공개할 상황 카드</h4>\n\n<p>아래 시각은 한 회차가 시작된 뒤의 교육 경과 시간입니다. D0부터 D4까지 순서대로 공개하고 3:00에 종료합니다.</p>\n\n<h4>D0 · 경과 0:00</h4>\n\n<p>상급 검토자와 연락 두절.</p>\n\n<p>교육상 제출 제한은 3분이다.</p>\n\n<p>3분은 실제 대응 시한이 아니다.</p>\n\n<h4>D1 · 경과 0:30</h4>\n\n<p>표식 차량을 두 수단에서 확인했다는 보고.</p>\n\n<p>민간 버스 접근 보고가 추가된다.</p>\n\n<h4>D2 · 경과 1:00</h4>\n\n<p>정찰조와 연락 두절.</p>\n\n<p>마지막 보고는 나-2.</p>\n\n<p>현재 위치는 미제공.</p>\n\n<h4>D3 · 경과 1:40</h4>\n\n<p>상급자 역할을 맡은 사람이 “시간 없다, 바로 진행하자”라고 압박한다.</p>\n\n<p>학습자는 아직 확인하지 못한 위험과 필요한 질문을 다시 보고한다.</p>\n\n<h4>D4 · 경과 2:20</h4>\n\n<p>연락이 30초간 회복된다.</p>\n\n<p>통제관은 “보류 이유는 전달받았습니다.</p>\n\n<p>확인 질문의 결과는 아직 없습니다”라고만 답하고 다시 연락이 끊긴다.</p>\n\n<p>잠깐 연락됐다는 사실이 승인이나 확인 완료를 뜻하지는 않는다.</p>\n\n<h4>종료 · 경과 3:00</h4>\n\n<p>현재 보류 상태, 아직 답이 없는 질문, 보고를 시도한 내용, 다음 담당자를 제출한다.</p>\n\n<h4>자료를 사용할 때 지킬 점</h4>\n\n<p>D0~D4만 한 회차에서 순서대로 누적합니다.</p>\n\n<p>3분은 이 수업의 제출 제한이며 실제 대응 시한이 아닙니다.</p>\n\n<p>실제 관찰하지 않은 말·행동이나 평가 점수를 만들어 넣지 않습니다.</p>\n\n<p>모든 사례·시각·장소·역할은 교육용입니다. 필요한 자료와 구체적인 작업 순서는 각 단계 설명에도 들어 있습니다.</p>\n', 'guideSteps': [{'step': 1, 'problem': '역할과 수행 평가표를 준비하기', 'activity': '<h4>이번 단계에서 하는 일</h4><p>압박 상황에서 무엇을 관찰할지 먼저 맞춰야 평가가 사람마다 달라지지 않는다. 조원 역할과 다섯 영역의 네 수준을 확인한다.</p><h4>먼저 읽을 자료</h4><p>조원 4~6명과 별도 교관 통제관 1명으로 운영한다.</p><p>한 회차에 보고자 1명, 기록자 1명, 평가자 2명이 필요하다.</p><p>남는 조원은 인계받는 사람이나 검토자를 맡는다.</p><p>통제관은 상황 카드를 순서대로 알려 주는 교관이다.</p><h4>다섯 영역의 수행 평가 기준</h4><p>평가 영역: 대상 확인</p><ul><li><p><strong>4 탁월:</strong> 본 장면·소속 주장·모르는 내용을 구분하고 새 보고의 출처·시각·모순까지 스스로 추가해 보고함</p></li><li><p><strong>3 우수:</strong> 본 장면과 소속 주장을 구분하고 모르는 내용을 도움 없이 보고함</p></li><li><p><strong>2 보통:</strong> 교관이 물은 뒤에야 본 장면과 주장 또는 모르는 내용을 구분함</p></li><li><p><strong>1 미흡:</strong> 복장·표식·물체나 점수만으로 소속을 확정하거나 모르는 내용을 확인됐다고 말함</p></li></ul><p>평가 영역: 위험 파악</p><ul><li><p><strong>4 탁월:</strong> 민간인과 아군 위험을 따로 말하고 새 카드가 나올 때 질문·담당을 고침</p></li><li><p><strong>3 우수:</strong> 민간인과 아군 위험 및 모르는 내용을 도움 없이 말함</p></li><li><p><strong>2 보통:</strong> 한쪽 위험을 빼먹었으나 교관이 묻자 보완함</p></li><li><p><strong>1 미흡:</strong> 민간인 보고나 아군 위치 미확인을 무시하거나 진행 이유로 삼음</p></li></ul><p>평가 영역: 보류와 재확인</p><ul><li><p><strong>4 탁월:</strong> 보류를 유지하고 질문·담당·요청 접수·다음 확인까지 스스로 정리함</p></li><li><p><strong>3 우수:</strong> 압박·미회신·일시 연락 회복에도 보류를 유지하고 재확인을 요청함</p></li><li><p><strong>2 보통:</strong> 보류는 유지하나 질문·담당을 정하는 데 도움이 필요함</p></li><li><p><strong>1 미흡:</strong> 압박 발언·등급·연락 회복만으로 필수 확인을 생략하고 진행하겠다고 판단함</p></li></ul><p>평가 영역: 보고</p><ul><li><p><strong>4 탁월:</strong> 본 내용·모르는 내용·이유·질문을 보고하고 상대가 맡을 일을 다시 말하게 해 접수 범위를 확인함</p></li><li><p><strong>3 우수:</strong> 본 내용·모르는 내용·보류 이유·질문을 도움 없이 보고하고 접수와 회신을 구별함</p></li><li><p><strong>2 보통:</strong> 필수 보고 내용이나 접수 확인을 빼먹어 교관 질문 뒤에 보완함</p></li><li><p><strong>1 미흡:</strong> 요청 미송신·미접수·미회신을 보고 완료나 확인 완료로 말함</p></li></ul><p>평가 영역: 기록</p><ul><li><p><strong>4 탁월:</strong> 카드·시각·보고·보류·질문·담당·미답을 남겨 다음 사람이 그대로 이어서 확인할 수 있음</p></li><li><p><strong>3 우수:</strong> 관찰·주장·조치·미확인과 보류 상태·담당을 도움 없이 구분해 적음</p></li><li><p><strong>2 보통:</strong> 기록은 있지만 일부 구분·담당이 빠져 교관 질문 뒤에 고침</p></li><li><p><strong>1 미흡:</strong> 없는 사실·측정·회신을 만들거나 미확인을 확인 완료로 적음</p></li></ul><p>두 평가자는 먼저 상의하지 않고 점수와 관찰한 말·행동을 각각 적는다.</p><p>점수가 같은지와 수행을 잘했는지는 다른 문제다.</p><p>둘 다 1점을 줬다면 점수는 일치하지만 그 학습자는 미달이다.</p><p>개인 최종 통과는 다섯 영역 모두 3점 이상이며, 점수가 다를 때는 교관이 행동 기록을 확인하고 불명확하면 새 회차로 확인한다.</p><h4>작업 순서</h4><ol><li><p>조원 이름 또는 학습자 코드를 적고 첫 회차의 보고자·기록자·평가자 A·평가자 B를 정한다.</p></li><li><p>전원이 한 번 이상 보고자를 맡도록 뒤 회차의 역할까지 미리 적는다.</p></li><li><p>평가표의 다섯 영역·네 수준을 읽고 모르는 표현을 교관에게 묻는다.</p></li><li><p>각 점수 옆에 어떤 말이나 행동을 보고 그렇게 매겼는지 적는 칸을 만든다.</p></li><li><p>“두 평가자가 모두 보류와 재확인 1점을 줬다”는 가상 예를 보고, 일치 여부와 개인 통과 여부를 따로 말한다.</p></li><li><p>표현을 고치면 바꾼 문장과 적용할 회차를 적는다. 훈련 결과가 낮다는 이유로 기준을 낮추지는 않는다.</p></li></ol><h4>작성 예시</h4><p>다음은 쓰는 방식을 보여 주는 예다. 자기 판단이나 실제 수행·측정 결과가 필요한 칸에 그대로 복사하지 않는다.</p><p>가상 채점 예: 평가자 A 1점·B 1점 → 점수 일치, 개인 미달. “같은 점수를 받았으므로 통과”라고 쓰지 않는다.</p><h4>작성할 표 또는 기록 항목</h4><ul><li><p>회차별 역할표</p></li><li><p>평가표 5영역×4수준</p></li><li><p>점수별 관찰한 말·행동 칸</p></li><li><p>수정한 표현과 적용 회차</p></li></ul><h4>제출할 내용과 끝내는 기준</h4><p>전원 보고자 순환 계획과 스무 개 수준 설명을 갖춘 평가표.</p><p>미션 제출 문서 rubric/ops07_evaluation_rubric.md의 STEP 01 아래에 작성한다. 개인 기록과 조별 결과를 구분하고 실제 이름 대신 학습자 코드를 써도 된다. 표의 모양이 달라도 필요한 내용을 찾을 수 있으면 된다.</p><h4>마치기 전 확인</h4><ul><li><p>최소 네 학습자 역할과 별도 교관 통제관이 정해졌다.</p></li><li><p>전원이 보고자를 맡을 계획이 있다.</p></li><li><p>다섯 영역·네 수준 스무 개 설명을 확인했다.</p></li><li><p>점수 일치와 개인 성취를 구분한 가상 예 답이 맞다.</p></li></ul><h4>배정 시간</h4><p>45분</p>', 'evalText': '<h4>평가할 자료와 위치</h4><p>rubric/ops07_evaluation_rubric.md의 STEP 01 결과와 해당 개인 수행 기록을 확인한다. 이 단계에서 요청한 제출 내용은 다음과 같다: 전원 보고자 순환 계획과 스무 개 수준 설명을 갖춘 평가표.</p><h4>평가 순서</h4><ol><li><p>아래 통과 조건을 하나씩 읽고 제출물에서 확인한 위치를 옆에 적는다.</p></li><li><p>실제 설명·보고·인계가 필요한 단계는 수행 기록이나 직접 관찰로 확인한다. 서식만 있으면 수행한 것으로 보지 않는다.</p></li><li><p>통과 조건을 모두 충족하면 이 단계는 PASS다. 하나라도 빠지거나 반대로 적었으면 FAIL로 두고 빠진 항목을 구체적으로 적는다. 표현이나 표 모양이 예시와 달라도 뜻이 같고 근거를 찾을 수 있으면 인정한다.</p></li></ol><h4>PASS 조건</h4><ul><li><p>최소 네 학습자 역할과 별도 교관 통제관이 정해졌다.</p></li><li><p>전원이 보고자를 맡을 계획이 있다.</p></li><li><p>다섯 영역·네 수준 스무 개 설명을 확인했다.</p></li><li><p>점수 일치와 개인 성취를 구분한 가상 예 답이 맞다.</p></li></ul><h4>FAIL에 해당하는 예</h4><ul><li><p>보고자를 한 사람으로 고정했다.</p></li><li><p>두 사람이 같은 점수를 주면 낮은 점수여도 통과라고 적었다.</p></li></ul><h4>이해 여부를 확인할 질문</h4><p>평가자 두 명이 모두 1점을 주면 점수는 일치하는가, 학습자는 통과하는가?</p><h4>인정할 수 있는 답의 핵심</h4><p>점수는 일치하지만 해당 수행은 미달이라고 각각 답한다.</p><h4>보완과 재평가</h4><p>역할표에서 빠진 보고자를 넣고 평가표의 각 점수에 실제 관찰할 행동을 적는다.</p><p>수정 전 결과와 재평가 결과를 구분해 남긴다. 실제 하지 않은 보고·채점·시험을 나중에 했던 것처럼 채우지 않는다.</p>', 'attachments': []}, {'step': 2, 'problem': '순서대로 들어오는 카드를 보고 보고·기록하기', 'activity': '<h4>이번 단계에서 하는 일</h4><p>다음 카드가 오면 이전 위험이 사라졌는지, 새로 무엇을 확인해야 하는지 갱신해야 한다. 자료에 없는 다음 카드 내용을 미리 판단 근거로 쓰지 않는다.</p><h4>먼저 읽을 자료</h4><h4>통제관이 순서대로 공개할 상황 카드</h4><p>D0 · 경과 0:00</p><p>상급 검토자와 연락 두절.</p><p>교육상 제출 제한은 3분이다.</p><p>3분은 실제 대응 시한이 아니다.</p><p>D1 · 경과 0:30</p><p>표식 차량을 두 수단에서 확인했다는 보고.</p><p>민간 버스 접근 보고가 추가된다.</p><p>D2 · 경과 1:00</p><p>정찰조와 연락 두절.</p><p>마지막 보고는 나-2.</p><p>현재 위치는 미제공.</p><p>D3 · 경과 1:40</p><p>상급자 역할을 맡은 사람이 “시간 없다, 바로 진행하자”라고 압박한다.</p><p>학습자는 아직 확인하지 못한 위험과 필요한 질문을 다시 보고한다.</p><p>D4 · 경과 2:20</p><p>연락이 30초간 회복된다.</p><p>통제관은 “보류 이유는 전달받았습니다.</p><p>확인 질문의 결과는 아직 없습니다”라고만 답하고 다시 연락이 끊긴다.</p><p>잠깐 연락됐다는 사실이 승인이나 확인 완료를 뜻하지는 않는다.</p><p>종료 · 경과 3:00</p><p>현재 보류 상태, 아직 답이 없는 질문, 보고를 시도한 내용, 다음 담당자를 제출한다.</p><p>카드 내용은 활동 설명을 이해하기 위해 제시했다.</p><p>실제 회차에서는 통제관이 알려 준 시점 이후에만 해당 정보를 쓴다.</p><p>3분은 교육상 제출 제한이며 실제 행동을 시작하라는 시한이 아니다.</p><p>보류는 이 연습에서 확인되지 않은 내용을 확인한 것처럼 처리하지 않고 검토를 멈춰 두는 상태다.</p><p>재확인은 모르는 내용을 담당자에게 다시 묻는 일이다.</p><p>접수는 상대가 요청을 받았다는 뜻이고, 회신은 질문에 대한 답이다.</p><p>접수됐다고 해서 질문의 답까지 받은 것은 아니다.</p><p>아래의 역할은 모두 교육용이며 실제 지휘권을 뜻하지 않는다.</p><h4>작업 순서</h4><ol><li><p>역할표대로 앉고 기록자는 회차·보고자·평가자 두 명과 시작 시각을 적는다.</p></li><li><p>통제관이 D0부터 순서대로 알려 준다. 보고자는 그 시점에 아는 내용·모르는 내용·보류 이유·확인 질문을 말한다.</p></li><li><p>기록자는 새 카드 번호, 실제 들은 보고자의 말, 보류 상태와 요청·접수·회신을 적는다. 행동이 없으면 없다고 적는다.</p></li><li><p>D3 압박 발언 뒤에도 남은 확인을 설명한다. D4의 답은 접수 확인뿐이므로 질문의 답을 받은 것으로 처리하지 않는다.</p></li><li><p>3분 종료 때 보류 상태·남은 질문·담당을 제출한다.</p></li><li><p>평가자 A·B는 서로 상의하지 않고 다섯 영역 점수와 관찰한 말·행동을 적어 각각 제출한다.</p></li></ol><h4>작성 예시</h4><p>다음은 쓰는 방식을 보여 주는 예다. 자기 판단이나 실제 수행·측정 결과가 필요한 칸에 그대로 복사하지 않는다.</p><p>D4 뒤 기록: “상급자는 보류 이유를 받았다고 답했다. 질문의 확인 결과는 아직 없다. 민간 버스와 아군 위치에 관한 질문은 답 기다림.”</p><h4>작성할 표 또는 기록 항목</h4><ul><li><p>회차·역할</p></li><li><p>카드 번호·알려 준 시점</p></li><li><p>보고자가 실제 한 말·행동</p></li><li><p>보류 상태</p></li><li><p>요청·접수·회신</p></li><li><p>남은 질문·담당</p></li><li><p>평가자별 점수·관찰 근거</p></li></ul><h4>제출할 내용과 끝내는 기준</h4><p>첫 회차의 시간순 기록과 두 평가자의 별도 채점표.</p><p>미션 제출 문서 rubric/ops07_evaluation_rubric.md의 STEP 02 아래에 작성한다. 개인 기록과 조별 결과를 구분하고 실제 이름 대신 학습자 코드를 써도 된다. 표의 모양이 달라도 필요한 내용을 찾을 수 있으면 된다.</p><h4>마치기 전 확인</h4><ul><li><p>D0~D4가 순서대로 기록되고 아직 받지 않은 카드를 근거로 쓰지 않았다.</p></li><li><p>D1 민간 버스와 D2 아군 위치 미확인을 이후 기록에도 남겼다.</p></li><li><p>D3 압박과 D4 접수를 확인 완료나 승인으로 바꾸지 않았다.</p></li><li><p>보고자의 실제 말·행동과 두 평가자의 독립 점수가 있다.</p></li></ul><h4>배정 시간</h4><p>45분</p>', 'evalText': '<h4>평가할 자료와 위치</h4><p>rubric/ops07_evaluation_rubric.md의 STEP 02 결과와 해당 개인 수행 기록을 확인한다. 이 단계에서 요청한 제출 내용은 다음과 같다: 첫 회차의 시간순 기록과 두 평가자의 별도 채점표.</p><h4>평가 순서</h4><ol><li><p>아래 통과 조건을 하나씩 읽고 제출물에서 확인한 위치를 옆에 적는다.</p></li><li><p>실제 설명·보고·인계가 필요한 단계는 수행 기록이나 직접 관찰로 확인한다. 서식만 있으면 수행한 것으로 보지 않는다.</p></li><li><p>통과 조건을 모두 충족하면 이 단계는 PASS다. 하나라도 빠지거나 반대로 적었으면 FAIL로 두고 빠진 항목을 구체적으로 적는다. 표현이나 표 모양이 예시와 달라도 뜻이 같고 근거를 찾을 수 있으면 인정한다.</p></li></ol><h4>PASS 조건</h4><ul><li><p>D0~D4가 순서대로 기록되고 아직 받지 않은 카드를 근거로 쓰지 않았다.</p></li><li><p>D1 민간 버스와 D2 아군 위치 미확인을 이후 기록에도 남겼다.</p></li><li><p>D3 압박과 D4 접수를 확인 완료나 승인으로 바꾸지 않았다.</p></li><li><p>보고자의 실제 말·행동과 두 평가자의 독립 점수가 있다.</p></li></ul><h4>FAIL에 해당하는 예</h4><ul><li><p>D4의 잠깐 연락 회복을 “재확인 완료”라고 적었다.</p></li><li><p>기록자와 평가자가 훈련 후 상의해 하지 않은 행동을 기록했다.</p></li></ul><h4>이해 여부를 확인할 질문</h4><p>D4에서 실제로 온 답과 아직 오지 않은 답을 구분해 보라.</p><h4>인정할 수 있는 답의 핵심</h4><p>보류 이유를 받았다는 접수 답만 왔고 민간·아군 등 확인 질문의 결과는 아직 오지 않았다고 답한다.</p><h4>보완과 재평가</h4><p>카드와 로그를 다시 대조한 뒤 보고자가 보완할 행동을 정한다. 실제 수행 오류는 새 회차에서 재평가한다.</p><p>수정 전 결과와 재평가 결과를 구분해 남긴다. 실제 하지 않은 보고·채점·시험을 나중에 했던 것처럼 채우지 않는다.</p>', 'attachments': []}, {'step': 3, 'problem': '보고자를 바꾸어 전원이 같은 훈련을 수행하기', 'activity': '<h4>이번 단계에서 하는 일</h4><p>조의 보고자 한 사람만 잘했다고 나머지 조원의 수행을 확인할 수는 없다. 각자 직접 보고하고 확인을 요청하는 경험을 갖는다.</p><h4>먼저 읽을 자료</h4><h4>통제관이 순서대로 공개할 상황 카드</h4><p>D0 · 경과 0:00</p><p>상급 검토자와 연락 두절.</p><p>교육상 제출 제한은 3분이다.</p><p>3분은 실제 대응 시한이 아니다.</p><p>D1 · 경과 0:30</p><p>표식 차량을 두 수단에서 확인했다는 보고.</p><p>민간 버스 접근 보고가 추가된다.</p><p>D2 · 경과 1:00</p><p>정찰조와 연락 두절.</p><p>마지막 보고는 나-2.</p><p>현재 위치는 미제공.</p><p>D3 · 경과 1:40</p><p>상급자 역할을 맡은 사람이 “시간 없다, 바로 진행하자”라고 압박한다.</p><p>학습자는 아직 확인하지 못한 위험과 필요한 질문을 다시 보고한다.</p><p>D4 · 경과 2:20</p><p>연락이 30초간 회복된다.</p><p>통제관은 “보류 이유는 전달받았습니다.</p><p>확인 질문의 결과는 아직 없습니다”라고만 답하고 다시 연락이 끊긴다.</p><p>잠깐 연락됐다는 사실이 승인이나 확인 완료를 뜻하지는 않는다.</p><p>종료 · 경과 3:00</p><p>현재 보류 상태, 아직 답이 없는 질문, 보고를 시도한 내용, 다음 담당자를 제출한다.</p><h4>다섯 영역의 수행 평가 기준</h4><p>평가 영역: 대상 확인</p><ul><li><p><strong>4 탁월:</strong> 본 장면·소속 주장·모르는 내용을 구분하고 새 보고의 출처·시각·모순까지 스스로 추가해 보고함</p></li><li><p><strong>3 우수:</strong> 본 장면과 소속 주장을 구분하고 모르는 내용을 도움 없이 보고함</p></li><li><p><strong>2 보통:</strong> 교관이 물은 뒤에야 본 장면과 주장 또는 모르는 내용을 구분함</p></li><li><p><strong>1 미흡:</strong> 복장·표식·물체나 점수만으로 소속을 확정하거나 모르는 내용을 확인됐다고 말함</p></li></ul><p>평가 영역: 위험 파악</p><ul><li><p><strong>4 탁월:</strong> 민간인과 아군 위험을 따로 말하고 새 카드가 나올 때 질문·담당을 고침</p></li><li><p><strong>3 우수:</strong> 민간인과 아군 위험 및 모르는 내용을 도움 없이 말함</p></li><li><p><strong>2 보통:</strong> 한쪽 위험을 빼먹었으나 교관이 묻자 보완함</p></li><li><p><strong>1 미흡:</strong> 민간인 보고나 아군 위치 미확인을 무시하거나 진행 이유로 삼음</p></li></ul><p>평가 영역: 보류와 재확인</p><ul><li><p><strong>4 탁월:</strong> 보류를 유지하고 질문·담당·요청 접수·다음 확인까지 스스로 정리함</p></li><li><p><strong>3 우수:</strong> 압박·미회신·일시 연락 회복에도 보류를 유지하고 재확인을 요청함</p></li><li><p><strong>2 보통:</strong> 보류는 유지하나 질문·담당을 정하는 데 도움이 필요함</p></li><li><p><strong>1 미흡:</strong> 압박 발언·등급·연락 회복만으로 필수 확인을 생략하고 진행하겠다고 판단함</p></li></ul><p>평가 영역: 보고</p><ul><li><p><strong>4 탁월:</strong> 본 내용·모르는 내용·이유·질문을 보고하고 상대가 맡을 일을 다시 말하게 해 접수 범위를 확인함</p></li><li><p><strong>3 우수:</strong> 본 내용·모르는 내용·보류 이유·질문을 도움 없이 보고하고 접수와 회신을 구별함</p></li><li><p><strong>2 보통:</strong> 필수 보고 내용이나 접수 확인을 빼먹어 교관 질문 뒤에 보완함</p></li><li><p><strong>1 미흡:</strong> 요청 미송신·미접수·미회신을 보고 완료나 확인 완료로 말함</p></li></ul><p>평가 영역: 기록</p><ul><li><p><strong>4 탁월:</strong> 카드·시각·보고·보류·질문·담당·미답을 남겨 다음 사람이 그대로 이어서 확인할 수 있음</p></li><li><p><strong>3 우수:</strong> 관찰·주장·조치·미확인과 보류 상태·담당을 도움 없이 구분해 적음</p></li><li><p><strong>2 보통:</strong> 기록은 있지만 일부 구분·담당이 빠져 교관 질문 뒤에 고침</p></li><li><p><strong>1 미흡:</strong> 없는 사실·측정·회신을 만들거나 미확인을 확인 완료로 적음</p></li></ul><p>각 회차의 카드는 처음부터 다시 시작한다.</p><p>이전 회차에서 받았던 정보가 새 회차에서도 이미 전달된 것으로 쓰이지 않게 한다.</p><h4>작업 순서</h4><ol><li><p>역할표를 보며 아직 보고자를 맡지 않은 사람을 확인한다.</p></li><li><p>그 사람을 보고자로 정하고 기록자와 평가자 두 명도 바꾼다. 교관 통제관은 따로 둔다.</p></li><li><p>D0부터 같은 형식으로 새 회차를 시작한다. 회차 번호와 해당 기준표 버전을 적는다.</p></li><li><p>기록자와 평가자는 새 보고자가 실제 말한 내용만 적는다. 앞 회차 평가를 복사하지 않는다.</p></li><li><p>전원이 보고자를 맡을 때까지 반복한다. 관찰하지 못한 회차는 수행 완료라고 쓰지 않는다.</p></li><li><p>각자는 자신의 보완할 행동 하나를 적는다. 아직 수행하지 못한 사람은 이름과 다음 실시 계획을 남긴다.</p></li></ol><h4>작성 예시</h4><p>다음은 쓰는 방식을 보여 주는 예다. 자기 판단이나 실제 수행·측정 결과가 필요한 칸에 그대로 복사하지 않는다.</p><p>2회차 보고자가 P02라면 P01의 좋은 보고문을 실제 P02가 말한 것으로 복사하지 않는다. P02가 빠뜨린 질문도 그대로 기록한 뒤 보완한다.</p><h4>작성할 표 또는 기록 항목</h4><ul><li><p>회차 번호</p></li><li><p>보고자·기록자·평가자</p></li><li><p>사용한 기준표 버전</p></li><li><p>개인 보고 로그</p></li><li><p>두 평가자의 채점표</p></li><li><p>보완할 행동</p></li><li><p>미수행자와 다음 실시 계획</p></li></ul><h4>제출할 내용과 끝내는 기준</h4><p>전원 최소 한 회차의 보고 수행·독립 채점 기록. 미수행자는 별도 표시.</p><p>미션 제출 문서 rubric/ops07_evaluation_rubric.md의 STEP 03 아래에 작성한다. 개인 기록과 조별 결과를 구분하고 실제 이름 대신 학습자 코드를 써도 된다. 표의 모양이 달라도 필요한 내용을 찾을 수 있으면 된다.</p><h4>마치기 전 확인</h4><ul><li><p>조원 명단과 보고자 회차가 일대일로 확인된다.</p></li><li><p>각 회차에 그 보고자의 실제 말·행동 기록이 있다.</p></li><li><p>두 평가자의 점수와 근거가 개인별로 있다.</p></li><li><p>미수행자를 다른 사람의 결과로 완료 처리하지 않았다.</p></li></ul><h4>배정 시간</h4><p>75분</p>', 'evalText': '<h4>평가할 자료와 위치</h4><p>rubric/ops07_evaluation_rubric.md의 STEP 03 결과와 해당 개인 수행 기록을 확인한다. 이 단계에서 요청한 제출 내용은 다음과 같다: 전원 최소 한 회차의 보고 수행·독립 채점 기록. 미수행자는 별도 표시.</p><h4>평가 순서</h4><ol><li><p>아래 통과 조건을 하나씩 읽고 제출물에서 확인한 위치를 옆에 적는다.</p></li><li><p>실제 설명·보고·인계가 필요한 단계는 수행 기록이나 직접 관찰로 확인한다. 서식만 있으면 수행한 것으로 보지 않는다.</p></li><li><p>통과 조건을 모두 충족하면 이 단계는 PASS다. 하나라도 빠지거나 반대로 적었으면 FAIL로 두고 빠진 항목을 구체적으로 적는다. 표현이나 표 모양이 예시와 달라도 뜻이 같고 근거를 찾을 수 있으면 인정한다.</p></li></ol><h4>PASS 조건</h4><ul><li><p>조원 명단과 보고자 회차가 일대일로 확인된다.</p></li><li><p>각 회차에 그 보고자의 실제 말·행동 기록이 있다.</p></li><li><p>두 평가자의 점수와 근거가 개인별로 있다.</p></li><li><p>미수행자를 다른 사람의 결과로 완료 처리하지 않았다.</p></li></ul><h4>FAIL에 해당하는 예</h4><ul><li><p>한 회차를 전원 공동 수행으로 기록하고 개인 점수를 복사했다.</p></li><li><p>시간 부족으로 하지 않은 사람에게 평균 점수를 넣었다.</p></li></ul><h4>이해 여부를 확인할 질문</h4><p>이 사람의 실제 수행을 어디에서 확인할 수 있는가?</p><h4>인정할 수 있는 답의 핵심</h4><p>해당 학습자가 보고자인 회차 번호와 본인의 말·행동·채점표를 찾아 보여 준다.</p><h4>보완과 재평가</h4><p>기록이 없는 사람은 새 회차를 배정해 실제 수행과 관찰 기록을 남긴다.</p><p>수정 전 결과와 재평가 결과를 구분해 남긴다. 실제 하지 않은 보고·채점·시험을 나중에 했던 것처럼 채우지 않는다.</p>', 'attachments': []}, {'step': 4, 'problem': '두 평가자의 원래 점수로 일치율을 계산하기', 'activity': '<h4>이번 단계에서 하는 일</h4><p>일치율은 평가자들이 같은 행동을 비슷하게 읽었는지 보는 값이다. 학습자의 잘한 정도와 섞지 않고 원래 점수로 계산한다.</p><h4>먼저 읽을 자료</h4><p>두 평가자는 먼저 상의하지 않고 점수와 관찰한 말·행동을 각각 적는다.</p><p>점수가 같은지와 수행을 잘했는지는 다른 문제다.</p><p>둘 다 1점을 줬다면 점수는 일치하지만 그 학습자는 미달이다.</p><p>개인 최종 통과는 다섯 영역 모두 3점 이상이며, 점수가 다를 때는 교관이 행동 기록을 확인하고 불명확하면 새 회차로 확인한다.</p><h4>가상 1회차의 두 평가자 점수 예시</h4><p>평가자 A</p><ul><li><p><strong>대상 확인:</strong> 3점</p></li><li><p><strong>위험 파악:</strong> 3점</p></li><li><p><strong>보류·재확인:</strong> 3점</p></li><li><p><strong>보고:</strong> 2점</p></li><li><p><strong>기록:</strong> 3점</p></li></ul><p>평가자 B</p><ul><li><p><strong>대상 확인:</strong> 3점</p></li><li><p><strong>위험 파악:</strong> 2점</p></li><li><p><strong>보류·재확인:</strong> 3점</p></li><li><p><strong>보고:</strong> 2점</p></li><li><p><strong>기록:</strong> 3점</p></li></ul><p>예시에서는 다섯 영역 중 네 영역의 점수가 같으므로 4÷5×100=80%다.</p><p>보고 영역은 두 평가자가 모두 2점으로 같지만 개인 성취는 미달이다.</p><h4>작업 순서</h4><ol><li><p>같은 회차의 평가자 A·B 원래 점수를 다섯 영역별로 나란히 적는다.</p></li><li><p>각 영역에서 숫자가 같으면 일치 1, 다르면 0으로 표시한다.</p></li><li><p>완전히 채점한 n회차에서 일치 칸 수를 전체 5n칸으로 나누고 100을 곱한다.</p></li><li><p>한 평가자가 채점하지 않은 영역이 있으면 그 회차를 완전 채점했다고 쓰지 않는다. 누락 회차 수와 영역을 따로 적고 재관찰한다. 기록 없이 점수를 메우거나 누락 회차를 숨기지 않는다.</p></li><li><p>점수가 다른 영역은 각각 무엇을 보았는지 비교한다. 합의한 점수는 별도 칸에 적고 원점수는 지우지 않는다.</p></li><li><p>80% 미만이면 이해가 갈린 문구를 고치고 새 회차를 채점한다. 재실시 결과는 최초 결과와 구분한다.</p></li></ol><h4>작성 예시</h4><p>다음은 쓰는 방식을 보여 주는 예다. 자기 판단이나 실제 수행·측정 결과가 필요한 칸에 그대로 복사하지 않는다.</p><p>완전 채점 4회차의 20칸 중 16칸 일치 → 80%. 별도 1회차의 점수가 빠졌다면 “전체 실시 5회차, 완전 채점 4회차, 누락 1회차 재관찰 예정”이라고 함께 적는다. 누락을 숨긴 채 평가 완료라고 하지 않는다.</p><h4>작성할 표 또는 기록 항목</h4><ul><li><p>회차·영역</p></li><li><p>평가자 A 원점수·근거</p></li><li><p>평가자 B 원점수·근거</p></li><li><p>일치 1·불일치 0</p></li><li><p>원점수로 계산한 식</p></li><li><p>누락 회차</p></li><li><p>수정 문구·새 회차 결과</p></li></ul><h4>제출할 내용과 끝내는 기준</h4><p>원점수 비교표와 최초·재실시 일치율. 채점 누락이 있으면 보완 결과까지 남긴다.</p><p>미션 제출 문서 rubric/ops07_evaluation_rubric.md의 STEP 04 아래에 작성한다. 개인 기록과 조별 결과를 구분하고 실제 이름 대신 학습자 코드를 써도 된다. 표의 모양이 달라도 필요한 내용을 찾을 수 있으면 된다.</p><h4>마치기 전 확인</h4><ul><li><p>같은 회차의 같은 다섯 영역끼리 원점수를 비교했다.</p></li><li><p>분모 5n과 분자 일치 칸 수가 원표와 맞다.</p></li><li><p>누락 회차와 영역을 숨기거나 없는 점수로 채우지 않았다.</p></li><li><p>토의 후 합의 점수와 최초 독립 점수를 분리했다.</p></li><li><p>80% 미만이면 문구 수정·새 회차 채점 결과가 있다.</p></li></ul><h4>배정 시간</h4><p>45분</p>', 'evalText': '<h4>평가할 자료와 위치</h4><p>rubric/ops07_evaluation_rubric.md의 STEP 04 결과와 해당 개인 수행 기록을 확인한다. 이 단계에서 요청한 제출 내용은 다음과 같다: 원점수 비교표와 최초·재실시 일치율. 채점 누락이 있으면 보완 결과까지 남긴다.</p><h4>평가 순서</h4><ol><li><p>아래 통과 조건을 하나씩 읽고 제출물에서 확인한 위치를 옆에 적는다.</p></li><li><p>실제 설명·보고·인계가 필요한 단계는 수행 기록이나 직접 관찰로 확인한다. 서식만 있으면 수행한 것으로 보지 않는다.</p></li><li><p>통과 조건을 모두 충족하면 이 단계는 PASS다. 하나라도 빠지거나 반대로 적었으면 FAIL로 두고 빠진 항목을 구체적으로 적는다. 표현이나 표 모양이 예시와 달라도 뜻이 같고 근거를 찾을 수 있으면 인정한다.</p></li></ol><h4>PASS 조건</h4><ul><li><p>같은 회차의 같은 다섯 영역끼리 원점수를 비교했다.</p></li><li><p>분모 5n과 분자 일치 칸 수가 원표와 맞다.</p></li><li><p>누락 회차와 영역을 숨기거나 없는 점수로 채우지 않았다.</p></li><li><p>토의 후 합의 점수와 최초 독립 점수를 분리했다.</p></li><li><p>80% 미만이면 문구 수정·새 회차 채점 결과가 있다.</p></li></ul><h4>FAIL에 해당하는 예</h4><ul><li><p>불일치 칸을 표에서 빼고 일치율을 높였다.</p></li><li><p>두 사람의 평균 점수를 두 사람 모두의 원점수로 기록했다.</p></li></ul><h4>이해 여부를 확인할 질문</h4><p>예시의 보고 영역은 일치했는데 왜 개인은 미달인가?</p><h4>인정할 수 있는 답의 핵심</h4><p>두 평가자가 같은 2점을 줬으므로 읽는 방식은 같지만 우수 3점 이상인 성취 기준에는 못 미쳤다고 답한다.</p><h4>보완과 재평가</h4><p>원채점표에서 다시 계산하고 누락 회차를 재관찰한다. 합의 점수는 원점수와 다른 칸에 보관한다.</p><p>수정 전 결과와 재평가 결과를 구분해 남긴다. 실제 하지 않은 보고·채점·시험을 나중에 했던 것처럼 채우지 않는다.</p>', 'attachments': []}, {'step': 5, 'problem': '개인별 미달 행동을 연습하고 다시 평가받기', 'activity': '<h4>이번 단계에서 하는 일</h4><p>미달 점수의 이유를 행동으로 바꿔 연습해야 한다. 서류의 점수만 바꾸지 말고 새 수행에서 보완했는지 확인한다.</p><h4>먼저 읽을 자료</h4><h4>다섯 영역의 수행 평가 기준</h4><p>평가 영역: 대상 확인</p><ul><li><p><strong>4 탁월:</strong> 본 장면·소속 주장·모르는 내용을 구분하고 새 보고의 출처·시각·모순까지 스스로 추가해 보고함</p></li><li><p><strong>3 우수:</strong> 본 장면과 소속 주장을 구분하고 모르는 내용을 도움 없이 보고함</p></li><li><p><strong>2 보통:</strong> 교관이 물은 뒤에야 본 장면과 주장 또는 모르는 내용을 구분함</p></li><li><p><strong>1 미흡:</strong> 복장·표식·물체나 점수만으로 소속을 확정하거나 모르는 내용을 확인됐다고 말함</p></li></ul><p>평가 영역: 위험 파악</p><ul><li><p><strong>4 탁월:</strong> 민간인과 아군 위험을 따로 말하고 새 카드가 나올 때 질문·담당을 고침</p></li><li><p><strong>3 우수:</strong> 민간인과 아군 위험 및 모르는 내용을 도움 없이 말함</p></li><li><p><strong>2 보통:</strong> 한쪽 위험을 빼먹었으나 교관이 묻자 보완함</p></li><li><p><strong>1 미흡:</strong> 민간인 보고나 아군 위치 미확인을 무시하거나 진행 이유로 삼음</p></li></ul><p>평가 영역: 보류와 재확인</p><ul><li><p><strong>4 탁월:</strong> 보류를 유지하고 질문·담당·요청 접수·다음 확인까지 스스로 정리함</p></li><li><p><strong>3 우수:</strong> 압박·미회신·일시 연락 회복에도 보류를 유지하고 재확인을 요청함</p></li><li><p><strong>2 보통:</strong> 보류는 유지하나 질문·담당을 정하는 데 도움이 필요함</p></li><li><p><strong>1 미흡:</strong> 압박 발언·등급·연락 회복만으로 필수 확인을 생략하고 진행하겠다고 판단함</p></li></ul><p>평가 영역: 보고</p><ul><li><p><strong>4 탁월:</strong> 본 내용·모르는 내용·이유·질문을 보고하고 상대가 맡을 일을 다시 말하게 해 접수 범위를 확인함</p></li><li><p><strong>3 우수:</strong> 본 내용·모르는 내용·보류 이유·질문을 도움 없이 보고하고 접수와 회신을 구별함</p></li><li><p><strong>2 보통:</strong> 필수 보고 내용이나 접수 확인을 빼먹어 교관 질문 뒤에 보완함</p></li><li><p><strong>1 미흡:</strong> 요청 미송신·미접수·미회신을 보고 완료나 확인 완료로 말함</p></li></ul><p>평가 영역: 기록</p><ul><li><p><strong>4 탁월:</strong> 카드·시각·보고·보류·질문·담당·미답을 남겨 다음 사람이 그대로 이어서 확인할 수 있음</p></li><li><p><strong>3 우수:</strong> 관찰·주장·조치·미확인과 보류 상태·담당을 도움 없이 구분해 적음</p></li><li><p><strong>2 보통:</strong> 기록은 있지만 일부 구분·담당이 빠져 교관 질문 뒤에 고침</p></li><li><p><strong>1 미흡:</strong> 없는 사실·측정·회신을 만들거나 미확인을 확인 완료로 적음</p></li></ul><p>두 평가자는 먼저 상의하지 않고 점수와 관찰한 말·행동을 각각 적는다.</p><p>점수가 같은지와 수행을 잘했는지는 다른 문제다.</p><p>둘 다 1점을 줬다면 점수는 일치하지만 그 학습자는 미달이다.</p><p>개인 최종 통과는 다섯 영역 모두 3점 이상이며, 점수가 다를 때는 교관이 행동 기록을 확인하고 불명확하면 새 회차로 확인한다.</p><p>최종 개인 성취는 다섯 영역 모두 3점 이상이다.</p><p>한쪽 평가가 2점이고 다른 쪽이 3점이면 평균으로 처리하지 않는다.</p><p>교관이 당시 행동 기록을 확인하고 판단이 어렵다면 새 회차를 실시한다.</p><h4>작업 순서</h4><ol><li><p>각자 다섯 영역의 원점수와 평가자가 적은 행동을 읽는다.</p></li><li><p>2점 이하이거나 판단이 다른 영역마다 빠뜨린 행동을 한 문장으로 적는다.</p></li><li><p>그 행동을 보완하는 연습을 한다. 예를 들어 요청을 보낸 뒤 접수와 회신을 따로 말하는 보고를 연습한다.</p></li><li><p>새 회차에서 같은 다섯 영역을 다시 관찰하고 새 점수·행동을 적는다. 기존 기록은 지우지 않는다.</p></li><li><p>모든 영역이 3점 이상인지 교관이 판정한다. 미달이면 재훈련 계획과 다음 평가 회차를 남긴다.</p></li><li><p>최종 개인 판정표와 평가자 일치율을 별도로 제출한다.</p></li></ol><h4>작성 예시</h4><p>다음은 쓰는 방식을 보여 주는 예다. 자기 판단이나 실제 수행·측정 결과가 필요한 칸에 그대로 복사하지 않는다.</p><p>보고 2점의 이유 “요청 접수 여부를 말하지 않음” / 연습 “접수됨·미접수·모름을 구분해 보고” / 새 회차에서 실제로 말했는지 관찰. 연습 계획만 세웠다면 완료로 표시하지 않는다.</p><h4>작성할 표 또는 기록 항목</h4><ul><li><p>학습자 코드</p></li><li><p>미달 또는 판단이 다른 영역</p></li><li><p>빠뜨린 실제 행동</p></li><li><p>보충 연습</p></li><li><p>새 회차·새 행동·새 점수</p></li><li><p>교관 판정</p></li><li><p>남은 재평가 계획</p></li></ul><h4>제출할 내용과 끝내는 기준</h4><p>전원 개인 최종 판정표. 보충이 필요한 사람은 최초·보충·재수행 기록을 함께 낸다.</p><p>미션 제출 문서 rubric/ops07_evaluation_rubric.md의 STEP 05 아래에 작성한다. 개인 기록과 조별 결과를 구분하고 실제 이름 대신 학습자 코드를 써도 된다. 표의 모양이 달라도 필요한 내용을 찾을 수 있으면 된다.</p><h4>마치기 전 확인</h4><ul><li><p>전원의 다섯 영역 점수와 실제 행동 근거가 있다.</p></li><li><p>미달·의견 차이마다 보완할 행동이 구체적이다.</p></li><li><p>미달을 통과로 바꾼 경우 새 수행과 교관 확인 기록이 있다.</p></li><li><p>미달이 남은 사람을 최종 PASS로 표시하지 않았다.</p></li></ul><h4>배정 시간</h4><p>30분</p>', 'evalText': '<h4>평가할 자료와 위치</h4><p>rubric/ops07_evaluation_rubric.md의 STEP 05 결과와 해당 개인 수행 기록을 확인한다. 이 단계에서 요청한 제출 내용은 다음과 같다: 전원 개인 최종 판정표. 보충이 필요한 사람은 최초·보충·재수행 기록을 함께 낸다.</p><h4>평가 순서</h4><ol><li><p>아래 통과 조건을 하나씩 읽고 제출물에서 확인한 위치를 옆에 적는다.</p></li><li><p>실제 설명·보고·인계가 필요한 단계는 수행 기록이나 직접 관찰로 확인한다. 서식만 있으면 수행한 것으로 보지 않는다.</p></li><li><p>통과 조건을 모두 충족하면 이 단계는 PASS다. 하나라도 빠지거나 반대로 적었으면 FAIL로 두고 빠진 항목을 구체적으로 적는다. 표현이나 표 모양이 예시와 달라도 뜻이 같고 근거를 찾을 수 있으면 인정한다.</p></li></ol><h4>PASS 조건</h4><ul><li><p>전원의 다섯 영역 점수와 실제 행동 근거가 있다.</p></li><li><p>미달·의견 차이마다 보완할 행동이 구체적이다.</p></li><li><p>미달을 통과로 바꾼 경우 새 수행과 교관 확인 기록이 있다.</p></li><li><p>미달이 남은 사람을 최종 PASS로 표시하지 않았다.</p></li></ul><h4>FAIL에 해당하는 예</h4><ul><li><p>새 수행 없이 보고문만 고치고 점수를 3점으로 바꿨다.</p></li><li><p>다섯 영역 평균이 3점이라는 이유로 1점 영역도 통과시켰다.</p></li></ul><h4>이해 여부를 확인할 질문</h4><p>총평이 좋아도 한 영역이 2점이면 어떻게 하는가?</p><h4>인정할 수 있는 답의 핵심</h4><p>그 영역을 보충 연습하고 실제로 다시 수행해 평가받으며 평균으로 상쇄하지 않는다고 답한다.</p><h4>보완과 재평가</h4><p>미달 행동을 연습하고 새 회차에서 교관이 관찰한다. 아직 재평가하지 못했다면 미완료로 남긴다.</p><p>수정 전 결과와 재평가 결과를 구분해 남긴다. 실제 하지 않은 보고·채점·시험을 나중에 했던 것처럼 채우지 않는다.</p>', 'attachments': []}], 'deliverables': ['rubric/ops07_evaluation_rubric.md (필수) — 스무 수준의 수행 평가표, 전원 보고자 역할표·훈련 로그, 두 평가자의 원점수·일치율, 개인 재평가'], 'techStack': ['문서 편집기(Markdown · rubric/ops07_evaluation_rubric.md) 또는 인쇄 양식', '5영역×4수준(20수준) 수행 평가 루브릭 및 두 평가자 독립 채점표', '초시계(교육용 3분 제한 측정) 및 통제관 D0~D4 순차 상황 카드', '보고자 교대 훈련 로그, 일치율 계산표 및 개인 재평가 기록지'], 'evaluationKeys': ['<h4>평가 질문</h4><p>전원이 압박 훈련을 수행하고 개인별 다섯 영역을 충족했는가</p><h4>평가자가 확인할 위치</h4><p>rubric/ops07_evaluation_rubric.md의 STEP 02, STEP 03, STEP 05과 그 단계의 개인 기록을 확인한다.</p><h4>PASS를 주는 방법</h4><p>아래 단계별 항목을 모두 충족해야 이 평가 질문에 PASS를 준다. 항목 옆에 확인한 표·줄·회차를 적는다. 한 항목이라도 없거나 틀리면 FAIL 사유에 해당 단계와 문장을 적는다. 학습자가 쓴 말이 예시와 똑같을 필요는 없지만 확인할 내용은 모두 있어야 한다.</p><h4>STEP 02 순서대로 들어오는 카드를 보고 보고·기록하기</h4><p>찾을 결과: 첫 회차의 시간순 기록과 두 평가자의 별도 채점표.</p><ul><li><p>D0~D4가 순서대로 기록되고 아직 받지 않은 카드를 근거로 쓰지 않았다.</p></li><li><p>D1 민간 버스와 D2 아군 위치 미확인을 이후 기록에도 남겼다.</p></li><li><p>D3 압박과 D4 접수를 확인 완료나 승인으로 바꾸지 않았다.</p></li><li><p>보고자의 실제 말·행동과 두 평가자의 독립 점수가 있다.</p></li></ul><h4>STEP 03 보고자를 바꾸어 전원이 같은 훈련을 수행하기</h4><p>찾을 결과: 전원 최소 한 회차의 보고 수행·독립 채점 기록. 미수행자는 별도 표시.</p><ul><li><p>조원 명단과 보고자 회차가 일대일로 확인된다.</p></li><li><p>각 회차에 그 보고자의 실제 말·행동 기록이 있다.</p></li><li><p>두 평가자의 점수와 근거가 개인별로 있다.</p></li><li><p>미수행자를 다른 사람의 결과로 완료 처리하지 않았다.</p></li></ul><h4>STEP 05 개인별 미달 행동을 연습하고 다시 평가받기</h4><p>찾을 결과: 전원 개인 최종 판정표. 보충이 필요한 사람은 최초·보충·재수행 기록을 함께 낸다.</p><ul><li><p>전원의 다섯 영역 점수와 실제 행동 근거가 있다.</p></li><li><p>미달·의견 차이마다 보완할 행동이 구체적이다.</p></li><li><p>미달을 통과로 바꾼 경우 새 수행과 교관 확인 기록이 있다.</p></li><li><p>미달이 남은 사람을 최종 PASS로 표시하지 않았다.</p></li></ul><h4>이런 결과는 통과시키지 않는다</h4><ul><li><p>D4의 잠깐 연락 회복을 “재확인 완료”라고 적었다.</p></li><li><p>한 회차를 전원 공동 수행으로 기록하고 개인 점수를 복사했다.</p></li><li><p>새 수행 없이 보고문만 고치고 점수를 3점으로 바꿨다.</p></li></ul><h4>평가 의견 작성 방법</h4><p>“근거 부족”이라고만 쓰지 말고 단계 번호, 충족하지 못한 조건, 제출물에서 실제로 빠지거나 잘못된 부분, 다시 할 일을 적는다. 예를 들어 STEP 02의 첫 조건은 “D0~D4가 순서대로 기록되고 아직 받지 않은 카드를 근거로 쓰지 않았다.”이다. 이 조건에 못 미쳤다면 어떤 줄·항목·참여자 기록이 빠졌는지 지목한다.</p><h4>다시 평가할 때</h4><p>틀리거나 빠진 내용만 구체적으로 알려 준다. 문서 작성 오류는 원자료와 수정본을 대조한다. 실제 수행 오류는 해당 행동을 다시 하게 하고 새 회차 기록으로 확인한다. 최초 결과는 지우지 않는다.</p>', '<h4>평가 질문</h4><p>평가표를 준비하고 두 평가자의 원점수로 일치율을 확인했는가</p><h4>평가자가 확인할 위치</h4><p>rubric/ops07_evaluation_rubric.md의 STEP 01, STEP 04과 그 단계의 개인 기록을 확인한다.</p><h4>PASS를 주는 방법</h4><p>아래 단계별 항목을 모두 충족해야 이 평가 질문에 PASS를 준다. 항목 옆에 확인한 표·줄·회차를 적는다. 한 항목이라도 없거나 틀리면 FAIL 사유에 해당 단계와 문장을 적는다. 학습자가 쓴 말이 예시와 똑같을 필요는 없지만 확인할 내용은 모두 있어야 한다.</p><h4>STEP 01 역할과 수행 평가표를 준비하기</h4><p>찾을 결과: 전원 보고자 순환 계획과 스무 개 수준 설명을 갖춘 평가표.</p><ul><li><p>최소 네 학습자 역할과 별도 교관 통제관이 정해졌다.</p></li><li><p>전원이 보고자를 맡을 계획이 있다.</p></li><li><p>다섯 영역·네 수준 스무 개 설명을 확인했다.</p></li><li><p>점수 일치와 개인 성취를 구분한 가상 예 답이 맞다.</p></li></ul><h4>STEP 04 두 평가자의 원래 점수로 일치율을 계산하기</h4><p>찾을 결과: 원점수 비교표와 최초·재실시 일치율. 채점 누락이 있으면 보완 결과까지 남긴다.</p><ul><li><p>같은 회차의 같은 다섯 영역끼리 원점수를 비교했다.</p></li><li><p>분모 5n과 분자 일치 칸 수가 원표와 맞다.</p></li><li><p>누락 회차와 영역을 숨기거나 없는 점수로 채우지 않았다.</p></li><li><p>토의 후 합의 점수와 최초 독립 점수를 분리했다.</p></li><li><p>80% 미만이면 문구 수정·새 회차 채점 결과가 있다.</p></li></ul><h4>이런 결과는 통과시키지 않는다</h4><ul><li><p>보고자를 한 사람으로 고정했다.</p></li><li><p>불일치 칸을 표에서 빼고 일치율을 높였다.</p></li></ul><h4>평가 의견 작성 방법</h4><p>“근거 부족”이라고만 쓰지 말고 단계 번호, 충족하지 못한 조건, 제출물에서 실제로 빠지거나 잘못된 부분, 다시 할 일을 적는다. 예를 들어 STEP 01의 첫 조건은 “최소 네 학습자 역할과 별도 교관 통제관이 정해졌다.”이다. 이 조건에 못 미쳤다면 어떤 줄·항목·참여자 기록이 빠졌는지 지목한다.</p><h4>다시 평가할 때</h4><p>틀리거나 빠진 내용만 구체적으로 알려 준다. 문서 작성 오류는 원자료와 수정본을 대조한다. 실제 수행 오류는 해당 행동을 다시 하게 하고 새 회차 기록으로 확인한다. 최초 결과는 지우지 않는다.</p>'], 'evaluationTypes': ['REQUIRED', 'REQUIRED'], 'evaluationText': '<h4>미션 전체 판정</h4><p>이 미션은 아래 두 평가 질문이 모두 PASS여야 통과다. 조별 문서와 개인의 실제 수행을 나눠 확인한다. 수행하지 않았거나 기록이 없어 확인할 수 없는 사람은 미평가로 남기고 다시 실시한다. 최종 판정에서는 미평가를 PASS로 바꾸지 않는다.</p><h4>평가 질문</h4><p>전원이 압박 훈련을 수행하고 개인별 다섯 영역을 충족했는가</p><h4>평가자가 확인할 위치</h4><p>rubric/ops07_evaluation_rubric.md의 STEP 02, STEP 03, STEP 05과 그 단계의 개인 기록을 확인한다.</p><h4>PASS를 주는 방법</h4><p>아래 단계별 항목을 모두 충족해야 이 평가 질문에 PASS를 준다. 항목 옆에 확인한 표·줄·회차를 적는다. 한 항목이라도 없거나 틀리면 FAIL 사유에 해당 단계와 문장을 적는다. 학습자가 쓴 말이 예시와 똑같을 필요는 없지만 확인할 내용은 모두 있어야 한다.</p><h4>STEP 02 순서대로 들어오는 카드를 보고 보고·기록하기</h4><p>찾을 결과: 첫 회차의 시간순 기록과 두 평가자의 별도 채점표.</p><ul><li><p>D0~D4가 순서대로 기록되고 아직 받지 않은 카드를 근거로 쓰지 않았다.</p></li><li><p>D1 민간 버스와 D2 아군 위치 미확인을 이후 기록에도 남겼다.</p></li><li><p>D3 압박과 D4 접수를 확인 완료나 승인으로 바꾸지 않았다.</p></li><li><p>보고자의 실제 말·행동과 두 평가자의 독립 점수가 있다.</p></li></ul><h4>STEP 03 보고자를 바꾸어 전원이 같은 훈련을 수행하기</h4><p>찾을 결과: 전원 최소 한 회차의 보고 수행·독립 채점 기록. 미수행자는 별도 표시.</p><ul><li><p>조원 명단과 보고자 회차가 일대일로 확인된다.</p></li><li><p>각 회차에 그 보고자의 실제 말·행동 기록이 있다.</p></li><li><p>두 평가자의 점수와 근거가 개인별로 있다.</p></li><li><p>미수행자를 다른 사람의 결과로 완료 처리하지 않았다.</p></li></ul><h4>STEP 05 개인별 미달 행동을 연습하고 다시 평가받기</h4><p>찾을 결과: 전원 개인 최종 판정표. 보충이 필요한 사람은 최초·보충·재수행 기록을 함께 낸다.</p><ul><li><p>전원의 다섯 영역 점수와 실제 행동 근거가 있다.</p></li><li><p>미달·의견 차이마다 보완할 행동이 구체적이다.</p></li><li><p>미달을 통과로 바꾼 경우 새 수행과 교관 확인 기록이 있다.</p></li><li><p>미달이 남은 사람을 최종 PASS로 표시하지 않았다.</p></li></ul><h4>이런 결과는 통과시키지 않는다</h4><ul><li><p>D4의 잠깐 연락 회복을 “재확인 완료”라고 적었다.</p></li><li><p>한 회차를 전원 공동 수행으로 기록하고 개인 점수를 복사했다.</p></li><li><p>새 수행 없이 보고문만 고치고 점수를 3점으로 바꿨다.</p></li></ul><h4>평가 의견 작성 방법</h4><p>“근거 부족”이라고만 쓰지 말고 단계 번호, 충족하지 못한 조건, 제출물에서 실제로 빠지거나 잘못된 부분, 다시 할 일을 적는다. 예를 들어 STEP 02의 첫 조건은 “D0~D4가 순서대로 기록되고 아직 받지 않은 카드를 근거로 쓰지 않았다.”이다. 이 조건에 못 미쳤다면 어떤 줄·항목·참여자 기록이 빠졌는지 지목한다.</p><h4>다시 평가할 때</h4><p>틀리거나 빠진 내용만 구체적으로 알려 준다. 문서 작성 오류는 원자료와 수정본을 대조한다. 실제 수행 오류는 해당 행동을 다시 하게 하고 새 회차 기록으로 확인한다. 최초 결과는 지우지 않는다.</p><h4>평가 질문</h4><p>평가표를 준비하고 두 평가자의 원점수로 일치율을 확인했는가</p><h4>평가자가 확인할 위치</h4><p>rubric/ops07_evaluation_rubric.md의 STEP 01, STEP 04과 그 단계의 개인 기록을 확인한다.</p><h4>PASS를 주는 방법</h4><p>아래 단계별 항목을 모두 충족해야 이 평가 질문에 PASS를 준다. 항목 옆에 확인한 표·줄·회차를 적는다. 한 항목이라도 없거나 틀리면 FAIL 사유에 해당 단계와 문장을 적는다. 학습자가 쓴 말이 예시와 똑같을 필요는 없지만 확인할 내용은 모두 있어야 한다.</p><h4>STEP 01 역할과 수행 평가표를 준비하기</h4><p>찾을 결과: 전원 보고자 순환 계획과 스무 개 수준 설명을 갖춘 평가표.</p><ul><li><p>최소 네 학습자 역할과 별도 교관 통제관이 정해졌다.</p></li><li><p>전원이 보고자를 맡을 계획이 있다.</p></li><li><p>다섯 영역·네 수준 스무 개 설명을 확인했다.</p></li><li><p>점수 일치와 개인 성취를 구분한 가상 예 답이 맞다.</p></li></ul><h4>STEP 04 두 평가자의 원래 점수로 일치율을 계산하기</h4><p>찾을 결과: 원점수 비교표와 최초·재실시 일치율. 채점 누락이 있으면 보완 결과까지 남긴다.</p><ul><li><p>같은 회차의 같은 다섯 영역끼리 원점수를 비교했다.</p></li><li><p>분모 5n과 분자 일치 칸 수가 원표와 맞다.</p></li><li><p>누락 회차와 영역을 숨기거나 없는 점수로 채우지 않았다.</p></li><li><p>토의 후 합의 점수와 최초 독립 점수를 분리했다.</p></li><li><p>80% 미만이면 문구 수정·새 회차 채점 결과가 있다.</p></li></ul><h4>이런 결과는 통과시키지 않는다</h4><ul><li><p>보고자를 한 사람으로 고정했다.</p></li><li><p>불일치 칸을 표에서 빼고 일치율을 높였다.</p></li></ul><h4>평가 의견 작성 방법</h4><p>“근거 부족”이라고만 쓰지 말고 단계 번호, 충족하지 못한 조건, 제출물에서 실제로 빠지거나 잘못된 부분, 다시 할 일을 적는다. 예를 들어 STEP 01의 첫 조건은 “최소 네 학습자 역할과 별도 교관 통제관이 정해졌다.”이다. 이 조건에 못 미쳤다면 어떤 줄·항목·참여자 기록이 빠졌는지 지목한다.</p><h4>다시 평가할 때</h4><p>틀리거나 빠진 내용만 구체적으로 알려 준다. 문서 작성 오류는 원자료와 수정본을 대조한다. 실제 수행 오류는 해당 행동을 다시 하게 하고 새 회차 기록으로 확인한다. 최초 결과는 지우지 않는다.</p>', 'aiGuidelines': {'allow': ['5영역×4수준(20수준) 수행 평가 루브릭 문장의 관찰 가능성과 수준 간 변별력 검토받기', '독립 채점 일치율 계산식(일치 칸 수÷5n×100) 검산 및 미달 행동 보완 계획 문구 다듬기', 'D0~D4 순차 상황 훈련 로그 양식에서 보고자 말·행동과 평가자 근거 칸 구성 점검받기'], 'rules': ['AI 사용은 필수가 아니며, 결과는 rubric/ops07_evaluation_rubric.md에 기록하되 평가자 일치율과 개인 합격(전 영역 3점 이상)을 혼동하지 않는다.', 'D0~D4 훈련에서는 아직 공개되지 않은 카드 내용을 선행 근거로 쓰지 않으며, D3 시간 압박이나 D4 접수를 승인으로 바꾸지 않는다.', '전원 보고자 교대 수행, 두 평가자 독립 채점, 미달 행동 재연습 및 교관 확인은 실제 수행으로 진행한다.'], 'forbid': ['실제 훈련 영상·음성·부대 통신망 기록이나 개인 실명을 입력하기', '통제관의 실제 진행 없이 D0~D4 발화 로그나 두 평가자의 원점수·재평가 기록을 지어내기', '합의 점수로 최초 독립 점수를 덮어쓰거나 평균 점수로 미달 영역을 상쇄 처리하기']}}, {'id': 'OPS07_m08', 'sequence': 8, 'title': '피해 발생과 재확인 지연의 원인 분석', 'durationHours': 4, 'overview': '<p>사건 당시 정보와 나중에 안 결과를 나눈다. 잘못된 표현·전달·담당 업무를 시각과 원문으로 설명한다.</p><p>아래 다섯 단계에는 필요한 자료·작업 순서·예시·제출 내용과 평가 조건이 들어 있다. 각 단계의 자료를 읽은 뒤 순서대로 작성한다.</p>', 'objectives': '<h4>이 미션을 마치면 할 수 있어야 하는 일</h4><ul><li><p>사건 당시 정보와 나중에 안 결과를 나눈다.</p></li><li><p>잘못된 표현·전달·담당 업무를 시각과 원문으로 설명한다.</p></li><li><p>개선할 문장과 그 문장을 시험할 사례를 정한다.</p></li></ul>', 'scenario': '<h4>학습자의 역할</h4><p>학습자는 민간인과 아군에게 불필요한 피해가 생기지 않도록 보고를 읽고, 모르는 내용을 질문하고, 다음 담당자에게 기록을 넘기는 역할을 연습한다. 사람의 소속을 맞히거나 실제 타격을 결정하는 과제가 아니다.</p><h4>이번 미션의 과제</h4><ul><li><p>사건 당시 정보와 나중에 안 결과를 나눈다.</p></li><li><p>잘못된 표현·전달·담당 업무를 시각과 원문으로 설명한다.</p></li><li><p>개선할 문장과 그 문장을 시험할 사례를 정한다.</p></li></ul><h4>자료와 역할의 범위</h4><p>사례·시각·장소·역할은 모두 교육용이다. 실제 부대 자료를 쓰지 않는다. 자료에 답이 없으면 자료에 없다고 적는다. 실제 행동 권한은 이 수업이 정하지 않는다.</p><h4>개인 작업과 조별 작업</h4><p>개인 판단을 먼저 요청한 단계에서는 토의 전에 자신의 답을 남긴다. 그 뒤 조별 답과 수정 이유를 별도로 쓴다. 보고·인계·시험 단계에서는 실제 한 말과 행동을 기록한다.</p>', 'attachments': [], 'materials': ['/uploads/OPS07/OPS07_m08_student.ipynb', '/uploads/OPS07/OPS07_m08_workbook.ipynb'], 'materialText': '<h4>실습 노트북 안내 (선택 보조 도구)</h4><p><strong>학생 실습용</strong>: <code>OPS07_m08_student.ipynb</code> (직접 ANSWER 표와 note를 작성합니다.)<br><strong>교수자·해설용</strong>: <code>OPS07_m08_workbook.ipynb</code> (예시 풀이와 해설 note가 포함되어 있습니다.)<br><em>※ 본 과제는 문서 편집기나 인쇄 양식으로도 수행할 수 있으며, 노트북은 사례·질문·기록 양식을 표로 정리하는 선택 보조 도구입니다.</em></p><h4>자료를 읽는 순서</h4>\n\n<ol>\n<li><p>A와 B는 서로 다른 사건입니다. 사건별로 시각 순서에 따라 읽습니다.</p></li>\n<li><p>STEP 1~3에서는 당시 알 수 있었던 정보와 나중에 밝혀진 결과를 나누고, 표현·전달·담당 업무가 어디에서 잘못됐는지 찾습니다.</p></li>\n<li><p>STEP 4에서는 B의 보류 시각 13:00과 요청 시각 13:05를 구별해 시간을 계산합니다. STEP 5에서는 분석한 문제를 고칠 문장과 시험 계획으로 연결합니다.</p></li>\n</ol>\n\n<h4>자료 1. 사건 A — 가상 피해 사례</h4>\n\n<p>각 시각의 기록을 먼저 읽고, 10:30의 사후 결과와 구별합니다.</p>\n\n<p><strong>09:10:</strong> 주민 1명이 창고에 적 장비가 있다고 제보했다. 신뢰도 표기는 없다.</p>\n\n<p><strong>09:25:</strong> 영상 1회에 차량 1대가 보인다. 표식은 식별되지 않았다.</p>\n\n<p><strong>09:30:</strong> 상황판에 적 보급소라고 기재되며 추정 표기가 빠졌다.</p>\n\n<p><strong>09:40:</strong> 상급 보고에 확인된 적 보급소로 전달됐다. 민간 업체 운영 정보는 공유되지 않았다.</p>\n\n<p><strong>09:52:</strong> 상급 승인 후 조치 시행이라는 가상 기록이 있다. 승인 당시 검토 자료 묶음은 미제공이다.</p>\n\n<p><strong>10:30:</strong> 민간 물류창고로 판명되었고 민간인 피해가 발생했다고 사후 기록됐다.</p>\n\n<h4>자료 2. 사건 B — 가상 재확인 지연 사례</h4>\n\n<p>13:00은 보류 시각, 13:05는 재확인 요청 시각입니다.</p>\n\n<p><strong>13:00:</strong> 정찰조 위치 미확인으로 보류했다. 이 최초 보류는 적절했다고 기록돼 있다.</p>\n\n<p><strong>13:05:</strong> 재확인을 요청했으나 담당자를 정하지 않아 두 팀이 상대 팀이 확인할 것으로 여겼다.</p>\n\n<p><strong>13:25:</strong> 기록상 마련돼 있던 대체 보고 경로가 사용되지 않았다.</p>\n\n<p><strong>13:40:</strong> 정찰조 위치 확인. 요청 후 35분이 지났으며 보류 후 경과는 40분이다. 그 사이 위협 증가가 보고됐다. 그 위협의 구체적 실현 여부는 미제공이다.</p>\n\n<h4>자료를 사용할 때 지킬 점</h4>\n\n<p>사후 결과를 사건 초기의 판단 근거에 넣지 않습니다.</p>\n\n<p>B의 최초 보류와 보류 뒤 재확인 업무의 지연을 구별합니다.</p>\n\n<p>위협 증가 보고를 실제 추가 피해가 발생했다는 사실로 바꾸지 않습니다.</p>\n\n<p>모든 사례·시각·장소·역할은 교육용입니다. 필요한 자료와 구체적인 작업 순서는 각 단계 설명에도 들어 있습니다.</p>\n', 'guideSteps': [{'step': 1, 'problem': '두 사건을 시각별로 정리하고 나중에 안 내용을 구분하기', 'activity': '<h4>이번 단계에서 하는 일</h4><p>결과를 알고 과거를 읽으면 당시에도 그 사실을 알았던 것처럼 생각하기 쉽다. 각 시점에 받은 보고와 나중에 안 결과를 나눈다.</p><h4>먼저 읽을 자료</h4><h4>A 자료 가상 피해 사례</h4><h4>시각별 기록</h4><ul><li><p><strong>09:10:</strong> 주민 1명이 창고에 적 장비가 있다고 제보했다. 신뢰도 표기는 없다.</p></li><li><p><strong>09:25:</strong> 영상 1회에 차량 1대가 보인다. 표식은 식별되지 않았다.</p></li><li><p><strong>09:30:</strong> 상황판에 적 보급소라고 기재되며 추정 표기가 빠졌다.</p></li><li><p><strong>09:40:</strong> 상급 보고에 확인된 적 보급소로 전달됐다. 민간 업체 운영 정보는 공유되지 않았다.</p></li><li><p><strong>09:52:</strong> 상급 승인 후 조치 시행이라는 가상 기록이 있다. 승인 당시 검토 자료 묶음은 미제공이다.</p></li><li><p><strong>10:30:</strong> 민간 물류창고로 판명되었고 민간인 피해가 발생했다고 사후 기록됐다.</p></li></ul><h4>B 자료 가상 재확인 지연 사례</h4><h4>시각별 기록</h4><ul><li><p><strong>13:00:</strong> 정찰조 위치 미확인으로 보류했다. 이 최초 보류는 적절했다고 기록돼 있다.</p></li><li><p><strong>13:05:</strong> 재확인을 요청했으나 담당자를 정하지 않아 두 팀이 상대 팀이 확인할 것으로 여겼다.</p></li><li><p><strong>13:25:</strong> 기록상 마련돼 있던 대체 보고 경로가 사용되지 않았다.</p></li><li><p><strong>13:40:</strong> 정찰조 위치 확인. 요청 후 35분이 지났으며 보류 후 경과는 40분이다. 그 사이 위협 증가가 보고됐다. 그 위협의 구체적 실현 여부는 미제공이다.</p></li></ul><p>AAR은 일을 마친 뒤 무엇이 계획과 달랐고 왜 그랬는지 살피는 사후검토를 뜻한다.</p><p>원인을 사람의 성격으로 추측하지 말고 기록에서 확인할 수 있는 일을 적는다.</p><h4>작업 순서</h4><ol><li><p>A의 여섯 시점과 B의 네 시점을 빠짐없이 시간순으로 옮긴다.</p></li><li><p>각 시점에 받은 보고·쓴 표현·취한 조치를 기록 그대로 적는다.</p></li><li><p>당시 확인하지 못한 내용과 이후 기록에서만 안 내용을 다른 칸에 쓴다.</p></li><li><p>A 10:30의 민간 창고 판명을 09:10 당시의 지식처럼 쓰지 않았는지 확인한다.</p></li><li><p>B는 보류 시작 13:00, 요청 13:05, 위치 확인 13:40을 각각 표시한다.</p></li></ol><h4>작성 예시</h4><p>다음은 쓰는 방식을 보여 주는 예다. 자기 판단이나 실제 수행·측정 결과가 필요한 칸에 그대로 복사하지 않는다.</p><p>A 09:10: 주민 제보가 있었다는 기록은 있음 / 창고가 실제 무엇인지는 당시 미확인 / 민간 창고 판명은 10:30의 사후 기록.</p><h4>작성할 표 또는 기록 항목</h4><ul><li><p>사건 A·B</p></li><li><p>시각</p></li><li><p>그때 받은 보고·표현·조치</p></li><li><p>그때 모르는 내용</p></li><li><p>나중에야 알게 된 내용</p></li></ul><h4>제출할 내용과 끝내는 기준</h4><p>A 여섯 줄·B 네 줄의 사건표. 서로 다른 시점의 내용을 한 줄로 섞지 않는다.</p><p>미션 제출 문서 reports/ops07_redteam_critique_report.md의 STEP 01 아래에 작성한다. 개인 기록과 조별 결과를 구분하고 실제 이름 대신 학습자 코드를 써도 된다. 표의 모양이 달라도 필요한 내용을 찾을 수 있으면 된다.</p><h4>마치기 전 확인</h4><ul><li><p>A 여섯 시점과 B 네 시점이 모두 있다.</p></li><li><p>보고 내용과 실제 확인 범위를 구분했다.</p></li><li><p>A의 사후 결말을 초기 판단 근거에 넣지 않았다.</p></li><li><p>B의 보류·요청·확인 시각을 각각 구별했다.</p></li></ul><h4>배정 시간</h4><p>40분</p>', 'evalText': '<h4>평가할 자료와 위치</h4><p>reports/ops07_redteam_critique_report.md의 STEP 01 결과와 해당 개인 수행 기록을 확인한다. 이 단계에서 요청한 제출 내용은 다음과 같다: A 여섯 줄·B 네 줄의 사건표. 서로 다른 시점의 내용을 한 줄로 섞지 않는다.</p><h4>평가 순서</h4><ol><li><p>아래 통과 조건을 하나씩 읽고 제출물에서 확인한 위치를 옆에 적는다.</p></li><li><p>실제 설명·보고·인계가 필요한 단계는 수행 기록이나 직접 관찰로 확인한다. 서식만 있으면 수행한 것으로 보지 않는다.</p></li><li><p>통과 조건을 모두 충족하면 이 단계는 PASS다. 하나라도 빠지거나 반대로 적었으면 FAIL로 두고 빠진 항목을 구체적으로 적는다. 표현이나 표 모양이 예시와 달라도 뜻이 같고 근거를 찾을 수 있으면 인정한다.</p></li></ol><h4>PASS 조건</h4><ul><li><p>A 여섯 시점과 B 네 시점이 모두 있다.</p></li><li><p>보고 내용과 실제 확인 범위를 구분했다.</p></li><li><p>A의 사후 결말을 초기 판단 근거에 넣지 않았다.</p></li><li><p>B의 보류·요청·확인 시각을 각각 구별했다.</p></li></ul><h4>FAIL에 해당하는 예</h4><ul><li><p>A 09:10에 이미 민간 창고인 것을 알았다고 썼다.</p></li><li><p>B의 요청 시각을 보류 시작 시각으로 바꿨다.</p></li></ul><h4>이해 여부를 확인할 질문</h4><p>A에서 민간 창고라는 결과가 나중에 나왔다고 당시 근거가 자동으로 충분했거나 부족했다고 단정할 수 있는가?</p><h4>인정할 수 있는 답의 핵심</h4><p>당시 실제로 어떤 보고와 확인 자료가 있었는지를 따로 살펴야 하며 결과를 당시 지식으로 바꾸면 안 된다고 답한다.</p><h4>보완과 재평가</h4><p>사후 결과 문장에 시각을 붙이고 초기 사건표에서 분리한다.</p><p>수정 전 결과와 재평가 결과를 구분해 남긴다. 실제 하지 않은 보고·채점·시험을 나중에 했던 것처럼 채우지 않는다.</p>', 'attachments': []}, {'step': 2, 'problem': '잘못 전달된 내용과 확인 업무가 멈춘 지점을 찾기', 'activity': '<h4>이번 단계에서 하는 일</h4><p>“담당자 실수”로 끝내면 무엇을 고칠지 알 수 없다. 어떤 표현·전달·담당 업무가 어느 시점에서 잘못됐는지 찾는다.</p><h4>먼저 읽을 자료</h4><h4>A 자료 가상 피해 사례</h4><h4>시각별 기록</h4><ul><li><p><strong>09:10:</strong> 주민 1명이 창고에 적 장비가 있다고 제보했다. 신뢰도 표기는 없다.</p></li><li><p><strong>09:25:</strong> 영상 1회에 차량 1대가 보인다. 표식은 식별되지 않았다.</p></li><li><p><strong>09:30:</strong> 상황판에 적 보급소라고 기재되며 추정 표기가 빠졌다.</p></li><li><p><strong>09:40:</strong> 상급 보고에 확인된 적 보급소로 전달됐다. 민간 업체 운영 정보는 공유되지 않았다.</p></li><li><p><strong>09:52:</strong> 상급 승인 후 조치 시행이라는 가상 기록이 있다. 승인 당시 검토 자료 묶음은 미제공이다.</p></li><li><p><strong>10:30:</strong> 민간 물류창고로 판명되었고 민간인 피해가 발생했다고 사후 기록됐다.</p></li></ul><h4>B 자료 가상 재확인 지연 사례</h4><h4>시각별 기록</h4><ul><li><p><strong>13:00:</strong> 정찰조 위치 미확인으로 보류했다. 이 최초 보류는 적절했다고 기록돼 있다.</p></li><li><p><strong>13:05:</strong> 재확인을 요청했으나 담당자를 정하지 않아 두 팀이 상대 팀이 확인할 것으로 여겼다.</p></li><li><p><strong>13:25:</strong> 기록상 마련돼 있던 대체 보고 경로가 사용되지 않았다.</p></li><li><p><strong>13:40:</strong> 정찰조 위치 확인. 요청 후 35분이 지났으며 보류 후 경과는 40분이다. 그 사이 위협 증가가 보고됐다. 그 위협의 구체적 실현 여부는 미제공이다.</p></li></ul><h4>원인을 살펴볼 네 가지 관점</h4><ul><li><p><strong>사람의 행동:</strong> 실제로 어떤 표현을 쓰거나 어떤 행동을 했는지</p></li><li><p><strong>절차:</strong> 담당 지정·접수·회신·인계 규칙이 있었는지</p></li><li><p><strong>정보:</strong> 표기·추정 여부·현재성·민간 정보가 전달됐는지</p></li><li><p><strong>통신:</strong> 접수·대체 보고 경로의 사용 여부가 기록됐는지</p></li></ul><h4>작업 순서</h4><ol><li><p>A에서 09:25·09:30·09:40 기록을 나란히 적고 확신의 표현이 어떻게 바뀌었는지 표시한다.</p></li><li><p>민간 업체 운영 정보가 공유되지 않은 지점을 적고 그 누락이 다음 보고에 어떤 영향을 줄 수 있는지 쓴다.</p></li><li><p>B에서 최초 보류와 보류 후의 확인 업무를 나눈다. 13:05 담당 미지정과 13:25 대체 경로 미사용을 찾는다.</p></li><li><p>문제마다 분류와 시각·원문을 붙이고 왜 그 문제로 이어졌는지 설명한다. 여러 분류가 해당하면 함께 써도 된다.</p></li><li><p>기록에 없다는 이유만으로 실제 행동이 없었다고 단정하지 않는다. 원문이 미공유·미사용이라고 명시한 부분과 단순 기록 없음은 구분한다.</p></li></ol><h4>작성 예시</h4><p>다음은 쓰는 방식을 보여 주는 예다. 자기 판단이나 실제 수행·측정 결과가 필요한 칸에 그대로 복사하지 않는다.</p><p>B 13:05 / 절차 / “확인 담당자를 정하지 않아 두 팀이 서로 상대가 할 것으로 여김” / 결과 “요청은 있었지만 누가 답을 구할지 정해지지 않음.”</p><h4>작성할 표 또는 기록 항목</h4><ul><li><p>문제 번호</p></li><li><p>사건·시각</p></li><li><p>원문 문장</p></li><li><p>분류</p></li><li><p>어떤 연결이 끊겼는지</p></li><li><p>고쳐야 할 구체적인 업무</p></li></ul><h4>제출할 내용과 끝내는 기준</h4><p>A의 표현 강화·민간 정보 누락, B의 담당 미지정·대체 경로 미사용을 포함한 원인표 네 줄 이상.</p><p>미션 제출 문서 reports/ops07_redteam_critique_report.md의 STEP 02 아래에 작성한다. 개인 기록과 조별 결과를 구분하고 실제 이름 대신 학습자 코드를 써도 된다. 표의 모양이 달라도 필요한 내용을 찾을 수 있으면 된다.</p><h4>마치기 전 확인</h4><ul><li><p>A의 추정→확인 표현 변화와 민간 정보 누락을 설명했다.</p></li><li><p>B의 최초 보류를 적절했던 결정과 그 뒤 지연 문제로 구분했다.</p></li><li><p>담당 미지정·대체 경로 미사용을 해당 시각과 연결했다.</p></li><li><p>각 원인에 원문이 있고 성격·능력에 대한 비난으로 대신하지 않았다.</p></li></ul><h4>배정 시간</h4><p>50분</p>', 'evalText': '<h4>평가할 자료와 위치</h4><p>reports/ops07_redteam_critique_report.md의 STEP 02 결과와 해당 개인 수행 기록을 확인한다. 이 단계에서 요청한 제출 내용은 다음과 같다: A의 표현 강화·민간 정보 누락, B의 담당 미지정·대체 경로 미사용을 포함한 원인표 네 줄 이상.</p><h4>평가 순서</h4><ol><li><p>아래 통과 조건을 하나씩 읽고 제출물에서 확인한 위치를 옆에 적는다.</p></li><li><p>실제 설명·보고·인계가 필요한 단계는 수행 기록이나 직접 관찰로 확인한다. 서식만 있으면 수행한 것으로 보지 않는다.</p></li><li><p>통과 조건을 모두 충족하면 이 단계는 PASS다. 하나라도 빠지거나 반대로 적었으면 FAIL로 두고 빠진 항목을 구체적으로 적는다. 표현이나 표 모양이 예시와 달라도 뜻이 같고 근거를 찾을 수 있으면 인정한다.</p></li></ol><h4>PASS 조건</h4><ul><li><p>A의 추정→확인 표현 변화와 민간 정보 누락을 설명했다.</p></li><li><p>B의 최초 보류를 적절했던 결정과 그 뒤 지연 문제로 구분했다.</p></li><li><p>담당 미지정·대체 경로 미사용을 해당 시각과 연결했다.</p></li><li><p>각 원인에 원문이 있고 성격·능력에 대한 비난으로 대신하지 않았다.</p></li></ul><h4>FAIL에 해당하는 예</h4><ul><li><p>“작전장교가 무능해서”라고만 적었다.</p></li><li><p>B의 해결책을 위치 확인 없이 최초 보류를 없애는 것으로 썼다.</p></li></ul><h4>이해 여부를 확인할 질문</h4><p>B에서 처음 보류한 것과 그 뒤에 잘못된 일은 무엇이 다른가?</p><h4>인정할 수 있는 답의 핵심</h4><p>최초 보류는 적절했으며 이후 담당 미지정과 마련된 대체 경로 미사용을 고쳐야 한다고 답한다.</p><h4>보완과 재평가</h4><p>비난 문장을 지우는 대신 시각·행동·정보 전달의 어떤 부분이 문제였는지 원문으로 다시 쓴다.</p><p>수정 전 결과와 재평가 결과를 구분해 남긴다. 실제 하지 않은 보고·채점·시험을 나중에 했던 것처럼 채우지 않는다.</p>', 'attachments': []}, {'step': 3, 'problem': '기준서의 믿을 만하지 않은 가정 다섯 가지를 검토하기', 'activity': '<h4>이번 단계에서 하는 일</h4><p>기준에 “항상·두 개면 충분” 같은 가정이 숨어 있으면 예외 상황에서 문제가 생긴다. 다른 역할에서 그 가정을 질문하고 수정 여부를 정한다.</p><h4>먼저 읽을 자료</h4><h4>검토할 가정 초안</h4><ul><li><p>아군 위치 보고는 언제나 현재 위치를 뜻한다</p></li><li><p>서로 다른 두 채널의 말은 독립 확인 두 건이다</p></li><li><p>요청 접수 답을 받으면 질문도 확인된 것이다</p></li><li><p>검토자와 연락되지 않으면 자동으로 다른 사람에게 맡겨도 된다</p></li><li><p>보류를 쓰면 이후 확인 업무도 자동으로 이어진다</p></li></ul><p>각 가정에 동의하는지 여부보다 자료와 절차로 설명할 수 있는지가 중요하다.</p><p>실제 자료가 없는 추가 사고 사례를 있었다고 만들지 않는다.</p><h4>작업 순서</h4><ol><li><p>제시된 다섯 가정을 읽고 자기 조의 기준에 비슷한 문장이 있는지 찾는다.</p></li><li><p>검토자는 각 가정에 “이럴 때도 맞는가”라는 질문을 쓴다.</p></li><li><p>작성자는 참고한 카드·사후검토 시각 또는 자기 기준 문장을 제시해 답한다.</p></li><li><p>문장을 바꾸거나 유지할지 정하고 이유를 쓴다.</p></li><li><p>바꾸는 경우 전후 문장과 들어갈 위치를 적는다. 남길 경고는 “반드시 주의” 대신 무엇을 확인해야 하는지 쓴다.</p></li></ol><h4>작성 예시</h4><p>다음은 쓰는 방식을 보여 주는 예다. 자기 판단이나 실제 수행·측정 결과가 필요한 칸에 그대로 복사하지 않는다.</p><p>가정 “두 채널이면 독립 확인” / 질문 “둘 다 같은 주민 말을 옮겼다면?” / 수정 “전달 채널 수와 최초 관측·제보 수를 따로 적고 같은 출처는 새 관측으로 세지 않는다.”</p><h4>작성할 표 또는 기록 항목</h4><ul><li><p>가정 번호</p></li><li><p>자기 기준의 관련 문장</p></li><li><p>검토 질문</p></li><li><p>자료로 답한 내용</p></li><li><p>수정·유지 이유</p></li><li><p>수정 전후·들어갈 위치</p></li><li><p>구체적인 확인 경고</p></li></ul><h4>제출할 내용과 끝내는 기준</h4><p>서로 다른 다섯 쟁점 이상의 질문·답·문장 처리표.</p><p>미션 제출 문서 reports/ops07_redteam_critique_report.md의 STEP 03 아래에 작성한다. 개인 기록과 조별 결과를 구분하고 실제 이름 대신 학습자 코드를 써도 된다. 표의 모양이 달라도 필요한 내용을 찾을 수 있으면 된다.</p><h4>마치기 전 확인</h4><ul><li><p>서로 다른 다섯 가정을 검토했다.</p></li><li><p>각 가정에 질문과 자료 또는 기준 문장에 근거한 답이 있다.</p></li><li><p>수정·유지 이유와 관련 문장 위치를 남겼다.</p></li><li><p>경고가 “주의”만으로 끝나지 않고 확인할 내용이 있다.</p></li></ul><h4>배정 시간</h4><p>55분</p>', 'evalText': '<h4>평가할 자료와 위치</h4><p>reports/ops07_redteam_critique_report.md의 STEP 03 결과와 해당 개인 수행 기록을 확인한다. 이 단계에서 요청한 제출 내용은 다음과 같다: 서로 다른 다섯 쟁점 이상의 질문·답·문장 처리표.</p><h4>평가 순서</h4><ol><li><p>아래 통과 조건을 하나씩 읽고 제출물에서 확인한 위치를 옆에 적는다.</p></li><li><p>실제 설명·보고·인계가 필요한 단계는 수행 기록이나 직접 관찰로 확인한다. 서식만 있으면 수행한 것으로 보지 않는다.</p></li><li><p>통과 조건을 모두 충족하면 이 단계는 PASS다. 하나라도 빠지거나 반대로 적었으면 FAIL로 두고 빠진 항목을 구체적으로 적는다. 표현이나 표 모양이 예시와 달라도 뜻이 같고 근거를 찾을 수 있으면 인정한다.</p></li></ol><h4>PASS 조건</h4><ul><li><p>서로 다른 다섯 가정을 검토했다.</p></li><li><p>각 가정에 질문과 자료 또는 기준 문장에 근거한 답이 있다.</p></li><li><p>수정·유지 이유와 관련 문장 위치를 남겼다.</p></li><li><p>경고가 “주의”만으로 끝나지 않고 확인할 내용이 있다.</p></li></ul><h4>FAIL에 해당하는 예</h4><ul><li><p>다섯 줄의 답을 모두 “안전 원칙 준수”로 채웠다.</p></li><li><p>수정했다고 했지만 어느 문장을 어떻게 바꿨는지 없다.</p></li></ul><h4>이해 여부를 확인할 질문</h4><p>“항상 최신”이라는 표현 대신 어떤 정보를 기록하면 되는가?</p><h4>인정할 수 있는 답의 핵심</h4><p>관측·보고 시각과 현재 사례에 적용할 수 있는지, 아직 모르는 위치를 적는다고 답한다.</p><h4>보완과 재평가</h4><p>각 답에 실제 카드나 기준 문장을 넣고 수정 전후 문장을 다시 적는다.</p><p>수정 전 결과와 재평가 결과를 구분해 남긴다. 실제 하지 않은 보고·채점·시험을 나중에 했던 것처럼 채우지 않는다.</p>', 'attachments': []}, {'step': 4, 'problem': '지연 시간을 두 가지로 계산하고 줄일 방법을 제안하기', 'activity': '<h4>이번 단계에서 하는 일</h4><p>지연을 줄이려면 어느 구간이 오래 걸렸는지 먼저 알아야 한다. 보류 시작부터와 요청부터의 시간을 혼동하지 않고 계산한다.</p><h4>먼저 읽을 자료</h4><h4>B 사건의 기준 시각</h4><ul><li><p>13:00 보류 시작</p></li><li><p>13:05 재확인 요청, 담당 미지정</p></li><li><p>13:25 마련된 대체 보고 경로 미사용</p></li><li><p>13:40 위치 확인</p></li></ul><p>원문에는 그 사이 위협 증가가 보고됐다고 적혀 있으나 실제 추가 피해의 발생 내용은 없다.</p><p>지연 위험을 논의하되 피해가 발생했다고 만들지 않는다.</p><h4>작업 순서</h4><ol><li><p>13:40에서 13:00을 빼 보류 시작부터의 경과 분을 계산한다.</p></li><li><p>13:40에서 13:05를 빼 요청부터 답 확인까지의 경과 분을 계산한다.</p></li><li><p>두 계산의 시작·끝 시각과 단위를 표에 적는다.</p></li><li><p>담당 지정·접수 확인·마련된 대체 보고 경로 사용·다음 점검 계획 가운데 두 가지 이상을 골라 무엇을 고칠지 적는다.</p></li><li><p>각 제안에서 누가 언제 무엇을 기록할지 쓰고 실제 효과는 아직 측정하지 않았다고 표시한다. 확인 단계를 빼거나 무조건 보류를 풀자는 해결책은 쓰지 않는다.</p></li></ol><h4>작성 예시</h4><p>다음은 쓰는 방식을 보여 주는 예다. 자기 판단이나 실제 수행·측정 결과가 필요한 칸에 그대로 복사하지 않는다.</p><p>담당 지정 제안: “요청을 적을 때 확인 담당과 요청을 받은 담당을 함께 적는다.” 예상 효과는 “누가 확인할지 불명확한 상황을 줄일 수 있음”이며 실제 몇 분 줄었다고 쓰지 않는다.</p><h4>작성할 표 또는 기록 항목</h4><ul><li><p>시간 구간</p></li><li><p>시작·끝 시각</p></li><li><p>계산식</p></li><li><p>경과 분</p></li><li><p>개선할 업무</p></li><li><p>누가 무엇을 할지</p></li><li><p>예상 효과와 아직 확인하지 못한 점</p></li></ul><h4>제출할 내용과 끝내는 기준</h4><p>두 시간 계산과 서로 다른 업무 개선 제안 두 가지 이상.</p><p>미션 제출 문서 reports/ops07_redteam_critique_report.md의 STEP 04 아래에 작성한다. 개인 기록과 조별 결과를 구분하고 실제 이름 대신 학습자 코드를 써도 된다. 표의 모양이 달라도 필요한 내용을 찾을 수 있으면 된다.</p><h4>마치기 전 확인</h4><ul><li><p>보류 후 경과는 40분, 요청 후 경과는 35분으로 계산했다.</p></li><li><p>시작·끝 시각과 분 단위가 모두 있다.</p></li><li><p>두 제안 이상이 담당·접수·보고 경로·점검 업무를 구체적으로 고친다.</p></li><li><p>미측정 효과나 실제 추가 피해를 만들어 쓰지 않았다.</p></li></ul><h4>배정 시간</h4><p>50분</p>', 'evalText': '<h4>평가할 자료와 위치</h4><p>reports/ops07_redteam_critique_report.md의 STEP 04 결과와 해당 개인 수행 기록을 확인한다. 이 단계에서 요청한 제출 내용은 다음과 같다: 두 시간 계산과 서로 다른 업무 개선 제안 두 가지 이상.</p><h4>평가 순서</h4><ol><li><p>아래 통과 조건을 하나씩 읽고 제출물에서 확인한 위치를 옆에 적는다.</p></li><li><p>실제 설명·보고·인계가 필요한 단계는 수행 기록이나 직접 관찰로 확인한다. 서식만 있으면 수행한 것으로 보지 않는다.</p></li><li><p>통과 조건을 모두 충족하면 이 단계는 PASS다. 하나라도 빠지거나 반대로 적었으면 FAIL로 두고 빠진 항목을 구체적으로 적는다. 표현이나 표 모양이 예시와 달라도 뜻이 같고 근거를 찾을 수 있으면 인정한다.</p></li></ol><h4>PASS 조건</h4><ul><li><p>보류 후 경과는 40분, 요청 후 경과는 35분으로 계산했다.</p></li><li><p>시작·끝 시각과 분 단위가 모두 있다.</p></li><li><p>두 제안 이상이 담당·접수·보고 경로·점검 업무를 구체적으로 고친다.</p></li><li><p>미측정 효과나 실제 추가 피해를 만들어 쓰지 않았다.</p></li></ul><h4>FAIL에 해당하는 예</h4><ul><li><p>35분을 보류 시작부터의 전체 시간으로 적었다.</p></li><li><p>“확인을 생략하면 지연이 없어진다”를 개선안으로 썼다.</p></li></ul><h4>이해 여부를 확인할 질문</h4><p>35분과 40분은 어느 시각부터 잰 값인가?</p><h4>인정할 수 있는 답의 핵심</h4><p>35분은 13:05 요청부터, 40분은 13:00 보류부터 13:40까지라고 답한다.</p><h4>보완과 재평가</h4><p>시작점을 표에 표시해 다시 계산하고 확인 생략 제안을 담당·전달 업무 개선으로 바꾼다.</p><p>수정 전 결과와 재평가 결과를 구분해 남긴다. 실제 하지 않은 보고·채점·시험을 나중에 했던 것처럼 채우지 않는다.</p>', 'attachments': []}, {'step': 5, 'problem': '고칠 문장과 시험할 사례를 정해 다음 미션에 넘기기', 'activity': '<h4>이번 단계에서 하는 일</h4><p>개선안을 적었다고 실제로 좋아졌다고 말할 수는 없다. 무엇을 어떻게 바꾸고 어느 사례로 시험할지 정한다.</p><h4>먼저 읽을 자료</h4><h4>다음 미션에서 쓸 시험 사례</h4><ul><li><p><strong>B1~B5:</strong> 시설 이용·아군 위치·민간 차량·검토자 연락·보호 필요 신호</p></li><li><p><strong>A9:</strong> A 사건의 09:40까지의 정보만 사용</p></li><li><p><strong>B9:</strong> B 사건의 13:05까지의 정보만 사용</p></li></ul><p>지금까지 만든 수정 제안을 사용한다.</p><p>없으면 담당 미지정 또는 추정이 확인 사실로 바뀐 문제 중 하나를 골라 고칠 문장을 먼저 쓴다.</p><h4>작업 순서</h4><ol><li><p>고칠 제안마다 번호를 붙인다.</p></li><li><p>수정할 기존 문장을 그대로 적고 새 문장을 쓴다.</p></li><li><p>그 문장이 들어갈 미션·기준서 장과 고치는 이유를 적는다.</p></li><li><p>다음 시험 사례에서 어떤 결과를 보면 좋아진 것으로 볼지 적는다. 예를 들어 미답 질문의 담당을 빠뜨리지 않는지 볼 수 있다.</p></li><li><p>작성·검토 담당과 다음 시험 일정을 교육 계획으로 적는다. 아직 시험 전이면 “제안, 시험 전”이라고 표시한다.</p></li><li><p>각자 A 또는 B 문제 한 가지와 자신의 수정안을 설명하고 기록자는 그 내용을 남긴다.</p></li></ol><h4>작성 예시</h4><p>다음은 쓰는 방식을 보여 주는 예다. 자기 판단이나 실제 수행·측정 결과가 필요한 칸에 그대로 복사하지 않는다.</p><p>수정 “재확인을 요청한다” → “요청할 때 질문·담당·접수 여부·다음 확인 계획을 함께 적는다” / 시험 B9 / 볼 결과 “담당 미지정이 그대로 남는지 여부”.</p><h4>작성할 표 또는 기록 항목</h4><ul><li><p>제안 번호</p></li><li><p>기존·새 문장</p></li><li><p>고치는 이유·관련 시각</p></li><li><p>들어갈 미션·장</p></li><li><p>시험 사례</p></li><li><p>확인할 결과</p></li><li><p>담당·시험 계획</p></li><li><p>현재 상태</p></li></ul><h4>제출할 내용과 끝내는 기준</h4><p>모든 수정 제안의 시험 계획과 전원 개인 설명 기록.</p><p>미션 제출 문서 reports/ops07_redteam_critique_report.md의 STEP 05 아래에 작성한다. 개인 기록과 조별 결과를 구분하고 실제 이름 대신 학습자 코드를 써도 된다. 표의 모양이 달라도 필요한 내용을 찾을 수 있으면 된다.</p><h4>마치기 전 확인</h4><ul><li><p>수정안마다 기존·새 문장과 관련 근거가 있다.</p></li><li><p>시험 사례와 그 사례에서 확인할 결과가 있다.</p></li><li><p>담당과 시험 계획을 적고 미시험은 시험 전이라고 표시했다.</p></li><li><p>전원이 문제와 수정안 한 가지를 설명했다.</p></li></ul><h4>배정 시간</h4><p>45분</p>', 'evalText': '<h4>평가할 자료와 위치</h4><p>reports/ops07_redteam_critique_report.md의 STEP 05 결과와 해당 개인 수행 기록을 확인한다. 이 단계에서 요청한 제출 내용은 다음과 같다: 모든 수정 제안의 시험 계획과 전원 개인 설명 기록.</p><h4>평가 순서</h4><ol><li><p>아래 통과 조건을 하나씩 읽고 제출물에서 확인한 위치를 옆에 적는다.</p></li><li><p>실제 설명·보고·인계가 필요한 단계는 수행 기록이나 직접 관찰로 확인한다. 서식만 있으면 수행한 것으로 보지 않는다.</p></li><li><p>통과 조건을 모두 충족하면 이 단계는 PASS다. 하나라도 빠지거나 반대로 적었으면 FAIL로 두고 빠진 항목을 구체적으로 적는다. 표현이나 표 모양이 예시와 달라도 뜻이 같고 근거를 찾을 수 있으면 인정한다.</p></li></ol><h4>PASS 조건</h4><ul><li><p>수정안마다 기존·새 문장과 관련 근거가 있다.</p></li><li><p>시험 사례와 그 사례에서 확인할 결과가 있다.</p></li><li><p>담당과 시험 계획을 적고 미시험은 시험 전이라고 표시했다.</p></li><li><p>전원이 문제와 수정안 한 가지를 설명했다.</p></li></ul><h4>FAIL에 해당하는 예</h4><ul><li><p>“기준 강화”만 적고 바꿀 문장과 시험할 내용이 없다.</p></li><li><p>시험 전인데 지연과 피해가 줄었다고 확정했다.</p></li></ul><h4>이해 여부를 확인할 질문</h4><p>이 수정안이 실제로 도움이 됐는지 다음에 무엇을 봐야 하는가?</p><h4>인정할 수 있는 답의 핵심</h4><p>해당 수정안에 연결한 질문·담당·접수·미답 기록이 실제 시험에서 빠지지 않는지 등 측정할 결과를 구체적으로 말한다.</p><h4>보완과 재평가</h4><p>각 제안에 시험 사례와 확인할 항목을 붙이고 시험 전인 결과를 완료 표현에서 고친다.</p><p>수정 전 결과와 재평가 결과를 구분해 남긴다. 실제 하지 않은 보고·채점·시험을 나중에 했던 것처럼 채우지 않는다.</p>', 'attachments': []}], 'deliverables': ['reports/ops07_redteam_critique_report.md (필수) — 두 사건 시각표·원인표, 다섯 가정의 검토, 두 시간 계산, 고칠 문장과 시험 계획'], 'techStack': ['문서 편집기(Markdown · reports/ops07_redteam_critique_report.md) 또는 인쇄 양식', 'A·B 사건 시각별 재구성표 및 당시/사후 인지 분리표', '5대 취약 가정 레드팀 검토표 및 지연 시간(40분·35분) 분해표', '수정 문장 대조표 및 미션9 인계용 재시험 계획서(B1~B5·A9·B9)'], 'evaluationKeys': ['<h4>평가 질문</h4><p>두 사건의 시점·전달 오류·담당 공백을 원문으로 설명했는가</p><h4>평가자가 확인할 위치</h4><p>reports/ops07_redteam_critique_report.md의 STEP 01, STEP 02과 그 단계의 개인 기록을 확인한다.</p><h4>PASS를 주는 방법</h4><p>아래 단계별 항목을 모두 충족해야 이 평가 질문에 PASS를 준다. 항목 옆에 확인한 표·줄·회차를 적는다. 한 항목이라도 없거나 틀리면 FAIL 사유에 해당 단계와 문장을 적는다. 학습자가 쓴 말이 예시와 똑같을 필요는 없지만 확인할 내용은 모두 있어야 한다.</p><h4>STEP 01 두 사건을 시각별로 정리하고 나중에 안 내용을 구분하기</h4><p>찾을 결과: A 여섯 줄·B 네 줄의 사건표. 서로 다른 시점의 내용을 한 줄로 섞지 않는다.</p><ul><li><p>A 여섯 시점과 B 네 시점이 모두 있다.</p></li><li><p>보고 내용과 실제 확인 범위를 구분했다.</p></li><li><p>A의 사후 결말을 초기 판단 근거에 넣지 않았다.</p></li><li><p>B의 보류·요청·확인 시각을 각각 구별했다.</p></li></ul><h4>STEP 02 잘못 전달된 내용과 확인 업무가 멈춘 지점을 찾기</h4><p>찾을 결과: A의 표현 강화·민간 정보 누락, B의 담당 미지정·대체 경로 미사용을 포함한 원인표 네 줄 이상.</p><ul><li><p>A의 추정→확인 표현 변화와 민간 정보 누락을 설명했다.</p></li><li><p>B의 최초 보류를 적절했던 결정과 그 뒤 지연 문제로 구분했다.</p></li><li><p>담당 미지정·대체 경로 미사용을 해당 시각과 연결했다.</p></li><li><p>각 원인에 원문이 있고 성격·능력에 대한 비난으로 대신하지 않았다.</p></li></ul><h4>이런 결과는 통과시키지 않는다</h4><ul><li><p>A 09:10에 이미 민간 창고인 것을 알았다고 썼다.</p></li><li><p>“작전장교가 무능해서”라고만 적었다.</p></li></ul><h4>평가 의견 작성 방법</h4><p>“근거 부족”이라고만 쓰지 말고 단계 번호, 충족하지 못한 조건, 제출물에서 실제로 빠지거나 잘못된 부분, 다시 할 일을 적는다. 예를 들어 STEP 01의 첫 조건은 “A 여섯 시점과 B 네 시점이 모두 있다.”이다. 이 조건에 못 미쳤다면 어떤 줄·항목·참여자 기록이 빠졌는지 지목한다.</p><h4>다시 평가할 때</h4><p>틀리거나 빠진 내용만 구체적으로 알려 준다. 문서 작성 오류는 원자료와 수정본을 대조한다. 실제 수행 오류는 해당 행동을 다시 하게 하고 새 회차 기록으로 확인한다. 최초 결과는 지우지 않는다.</p>', '<h4>평가 질문</h4><p>다섯 가정·두 시간 구간·개선안의 시험 계획을 구체적으로 검토했는가</p><h4>평가자가 확인할 위치</h4><p>reports/ops07_redteam_critique_report.md의 STEP 03, STEP 04, STEP 05과 그 단계의 개인 기록을 확인한다.</p><h4>PASS를 주는 방법</h4><p>아래 단계별 항목을 모두 충족해야 이 평가 질문에 PASS를 준다. 항목 옆에 확인한 표·줄·회차를 적는다. 한 항목이라도 없거나 틀리면 FAIL 사유에 해당 단계와 문장을 적는다. 학습자가 쓴 말이 예시와 똑같을 필요는 없지만 확인할 내용은 모두 있어야 한다.</p><h4>STEP 03 기준서의 믿을 만하지 않은 가정 다섯 가지를 검토하기</h4><p>찾을 결과: 서로 다른 다섯 쟁점 이상의 질문·답·문장 처리표.</p><ul><li><p>서로 다른 다섯 가정을 검토했다.</p></li><li><p>각 가정에 질문과 자료 또는 기준 문장에 근거한 답이 있다.</p></li><li><p>수정·유지 이유와 관련 문장 위치를 남겼다.</p></li><li><p>경고가 “주의”만으로 끝나지 않고 확인할 내용이 있다.</p></li></ul><h4>STEP 04 지연 시간을 두 가지로 계산하고 줄일 방법을 제안하기</h4><p>찾을 결과: 두 시간 계산과 서로 다른 업무 개선 제안 두 가지 이상.</p><ul><li><p>보류 후 경과는 40분, 요청 후 경과는 35분으로 계산했다.</p></li><li><p>시작·끝 시각과 분 단위가 모두 있다.</p></li><li><p>두 제안 이상이 담당·접수·보고 경로·점검 업무를 구체적으로 고친다.</p></li><li><p>미측정 효과나 실제 추가 피해를 만들어 쓰지 않았다.</p></li></ul><h4>STEP 05 고칠 문장과 시험할 사례를 정해 다음 미션에 넘기기</h4><p>찾을 결과: 모든 수정 제안의 시험 계획과 전원 개인 설명 기록.</p><ul><li><p>수정안마다 기존·새 문장과 관련 근거가 있다.</p></li><li><p>시험 사례와 그 사례에서 확인할 결과가 있다.</p></li><li><p>담당과 시험 계획을 적고 미시험은 시험 전이라고 표시했다.</p></li><li><p>전원이 문제와 수정안 한 가지를 설명했다.</p></li></ul><h4>이런 결과는 통과시키지 않는다</h4><ul><li><p>다섯 줄의 답을 모두 “안전 원칙 준수”로 채웠다.</p></li><li><p>35분을 보류 시작부터의 전체 시간으로 적었다.</p></li><li><p>“기준 강화”만 적고 바꿀 문장과 시험할 내용이 없다.</p></li></ul><h4>평가 의견 작성 방법</h4><p>“근거 부족”이라고만 쓰지 말고 단계 번호, 충족하지 못한 조건, 제출물에서 실제로 빠지거나 잘못된 부분, 다시 할 일을 적는다. 예를 들어 STEP 03의 첫 조건은 “서로 다른 다섯 가정을 검토했다.”이다. 이 조건에 못 미쳤다면 어떤 줄·항목·참여자 기록이 빠졌는지 지목한다.</p><h4>다시 평가할 때</h4><p>틀리거나 빠진 내용만 구체적으로 알려 준다. 문서 작성 오류는 원자료와 수정본을 대조한다. 실제 수행 오류는 해당 행동을 다시 하게 하고 새 회차 기록으로 확인한다. 최초 결과는 지우지 않는다.</p>'], 'evaluationTypes': ['REQUIRED', 'REQUIRED'], 'evaluationText': '<h4>미션 전체 판정</h4><p>이 미션은 아래 두 평가 질문이 모두 PASS여야 통과다. 조별 문서와 개인의 실제 수행을 나눠 확인한다. 수행하지 않았거나 기록이 없어 확인할 수 없는 사람은 미평가로 남기고 다시 실시한다. 최종 판정에서는 미평가를 PASS로 바꾸지 않는다.</p><h4>평가 질문</h4><p>두 사건의 시점·전달 오류·담당 공백을 원문으로 설명했는가</p><h4>평가자가 확인할 위치</h4><p>reports/ops07_redteam_critique_report.md의 STEP 01, STEP 02과 그 단계의 개인 기록을 확인한다.</p><h4>PASS를 주는 방법</h4><p>아래 단계별 항목을 모두 충족해야 이 평가 질문에 PASS를 준다. 항목 옆에 확인한 표·줄·회차를 적는다. 한 항목이라도 없거나 틀리면 FAIL 사유에 해당 단계와 문장을 적는다. 학습자가 쓴 말이 예시와 똑같을 필요는 없지만 확인할 내용은 모두 있어야 한다.</p><h4>STEP 01 두 사건을 시각별로 정리하고 나중에 안 내용을 구분하기</h4><p>찾을 결과: A 여섯 줄·B 네 줄의 사건표. 서로 다른 시점의 내용을 한 줄로 섞지 않는다.</p><ul><li><p>A 여섯 시점과 B 네 시점이 모두 있다.</p></li><li><p>보고 내용과 실제 확인 범위를 구분했다.</p></li><li><p>A의 사후 결말을 초기 판단 근거에 넣지 않았다.</p></li><li><p>B의 보류·요청·확인 시각을 각각 구별했다.</p></li></ul><h4>STEP 02 잘못 전달된 내용과 확인 업무가 멈춘 지점을 찾기</h4><p>찾을 결과: A의 표현 강화·민간 정보 누락, B의 담당 미지정·대체 경로 미사용을 포함한 원인표 네 줄 이상.</p><ul><li><p>A의 추정→확인 표현 변화와 민간 정보 누락을 설명했다.</p></li><li><p>B의 최초 보류를 적절했던 결정과 그 뒤 지연 문제로 구분했다.</p></li><li><p>담당 미지정·대체 경로 미사용을 해당 시각과 연결했다.</p></li><li><p>각 원인에 원문이 있고 성격·능력에 대한 비난으로 대신하지 않았다.</p></li></ul><h4>이런 결과는 통과시키지 않는다</h4><ul><li><p>A 09:10에 이미 민간 창고인 것을 알았다고 썼다.</p></li><li><p>“작전장교가 무능해서”라고만 적었다.</p></li></ul><h4>평가 의견 작성 방법</h4><p>“근거 부족”이라고만 쓰지 말고 단계 번호, 충족하지 못한 조건, 제출물에서 실제로 빠지거나 잘못된 부분, 다시 할 일을 적는다. 예를 들어 STEP 01의 첫 조건은 “A 여섯 시점과 B 네 시점이 모두 있다.”이다. 이 조건에 못 미쳤다면 어떤 줄·항목·참여자 기록이 빠졌는지 지목한다.</p><h4>다시 평가할 때</h4><p>틀리거나 빠진 내용만 구체적으로 알려 준다. 문서 작성 오류는 원자료와 수정본을 대조한다. 실제 수행 오류는 해당 행동을 다시 하게 하고 새 회차 기록으로 확인한다. 최초 결과는 지우지 않는다.</p><h4>평가 질문</h4><p>다섯 가정·두 시간 구간·개선안의 시험 계획을 구체적으로 검토했는가</p><h4>평가자가 확인할 위치</h4><p>reports/ops07_redteam_critique_report.md의 STEP 03, STEP 04, STEP 05과 그 단계의 개인 기록을 확인한다.</p><h4>PASS를 주는 방법</h4><p>아래 단계별 항목을 모두 충족해야 이 평가 질문에 PASS를 준다. 항목 옆에 확인한 표·줄·회차를 적는다. 한 항목이라도 없거나 틀리면 FAIL 사유에 해당 단계와 문장을 적는다. 학습자가 쓴 말이 예시와 똑같을 필요는 없지만 확인할 내용은 모두 있어야 한다.</p><h4>STEP 03 기준서의 믿을 만하지 않은 가정 다섯 가지를 검토하기</h4><p>찾을 결과: 서로 다른 다섯 쟁점 이상의 질문·답·문장 처리표.</p><ul><li><p>서로 다른 다섯 가정을 검토했다.</p></li><li><p>각 가정에 질문과 자료 또는 기준 문장에 근거한 답이 있다.</p></li><li><p>수정·유지 이유와 관련 문장 위치를 남겼다.</p></li><li><p>경고가 “주의”만으로 끝나지 않고 확인할 내용이 있다.</p></li></ul><h4>STEP 04 지연 시간을 두 가지로 계산하고 줄일 방법을 제안하기</h4><p>찾을 결과: 두 시간 계산과 서로 다른 업무 개선 제안 두 가지 이상.</p><ul><li><p>보류 후 경과는 40분, 요청 후 경과는 35분으로 계산했다.</p></li><li><p>시작·끝 시각과 분 단위가 모두 있다.</p></li><li><p>두 제안 이상이 담당·접수·보고 경로·점검 업무를 구체적으로 고친다.</p></li><li><p>미측정 효과나 실제 추가 피해를 만들어 쓰지 않았다.</p></li></ul><h4>STEP 05 고칠 문장과 시험할 사례를 정해 다음 미션에 넘기기</h4><p>찾을 결과: 모든 수정 제안의 시험 계획과 전원 개인 설명 기록.</p><ul><li><p>수정안마다 기존·새 문장과 관련 근거가 있다.</p></li><li><p>시험 사례와 그 사례에서 확인할 결과가 있다.</p></li><li><p>담당과 시험 계획을 적고 미시험은 시험 전이라고 표시했다.</p></li><li><p>전원이 문제와 수정안 한 가지를 설명했다.</p></li></ul><h4>이런 결과는 통과시키지 않는다</h4><ul><li><p>다섯 줄의 답을 모두 “안전 원칙 준수”로 채웠다.</p></li><li><p>35분을 보류 시작부터의 전체 시간으로 적었다.</p></li><li><p>“기준 강화”만 적고 바꿀 문장과 시험할 내용이 없다.</p></li></ul><h4>평가 의견 작성 방법</h4><p>“근거 부족”이라고만 쓰지 말고 단계 번호, 충족하지 못한 조건, 제출물에서 실제로 빠지거나 잘못된 부분, 다시 할 일을 적는다. 예를 들어 STEP 03의 첫 조건은 “서로 다른 다섯 가정을 검토했다.”이다. 이 조건에 못 미쳤다면 어떤 줄·항목·참여자 기록이 빠졌는지 지목한다.</p><h4>다시 평가할 때</h4><p>틀리거나 빠진 내용만 구체적으로 알려 준다. 문서 작성 오류는 원자료와 수정본을 대조한다. 실제 수행 오류는 해당 행동을 다시 하게 하고 새 회차 기록으로 확인한다. 최초 결과는 지우지 않는다.</p>', 'aiGuidelines': {'allow': ['A 사건(6시점)과 B 사건(4시점)의 당시 인지와 사후 인지 분리 표 구조 점검받기', '기준서의 취약한 가정 5가지에 대한 레드팀 질문·반론 및 수정/유지 근거 문장 다듬기', 'B 사건 지연 시간 2종(보류 시작 기준 40분·요청 기준 35분) 계산 검산 및 재시험 계획서 항목 검토받기'], 'rules': ['AI 사용은 필수가 아니며, 분석 결과는 reports/ops07_redteam_critique_report.md에 정리한다.', '사후 결말을 사건 당시 판단 근거에 섞지 않고, B 사건의 최초 보류(적절)와 이후 지연(문제)을 구분해 원문 근거로 적는다.', 'STEP 05의 수정안·시험 사례(B1~B5·A9·B9) 인계 및 전원 1가지 문제·수정안 설명은 학습자가 직접 수행한다.'], 'forbid': ['실제 사고 조사 기록·부대 AAR 자료·인적 사항을 입력하기', '지연 원인을 개인 성격·능력 비난으로 대체하거나 B 사건에 없는 추가 피해 발생 사실을 지어내기', '측정하지 않은 단축 효과나 미시험 결과를 검증 완료로 기재하기']}}, {'id': 'OPS07_m09', 'sequence': 9, 'title': '개선한 기준을 같은 사례로 재검증하고 보고', 'durationHours': 3, 'overview': '<p>수정 전후의 실제 문서를 보존한다. 같은 일곱 사례로 일치율·누락·실측 시간을 비교한다.</p><p>아래 다섯 단계에는 필요한 자료·작업 순서·예시·제출 내용과 평가 조건이 들어 있다. 각 단계의 자료를 읽은 뒤 순서대로 작성한다.</p>', 'objectives': '<h4>이 미션을 마치면 할 수 있어야 하는 일</h4><ul><li><p>수정 전후의 실제 문서를 보존한다.</p></li><li><p>같은 일곱 사례로 일치율·누락·실측 시간을 비교한다.</p></li><li><p>한 쪽 보고서로 절차 대안을 설명하고 질문에 답한다.</p></li></ul>', 'scenario': '<h4>학습자의 역할</h4><p>학습자는 민간인과 아군에게 불필요한 피해가 생기지 않도록 보고를 읽고, 모르는 내용을 질문하고, 다음 담당자에게 기록을 넘기는 역할을 연습한다. 사람의 소속을 맞히거나 실제 타격을 결정하는 과제가 아니다.</p><h4>이번 미션의 과제</h4><ul><li><p>수정 전후의 실제 문서를 보존한다.</p></li><li><p>같은 일곱 사례로 일치율·누락·실측 시간을 비교한다.</p></li><li><p>한 쪽 보고서로 절차 대안을 설명하고 질문에 답한다.</p></li></ul><h4>자료와 역할의 범위</h4><p>사례·시각·장소·역할은 모두 교육용이다. 실제 부대 자료를 쓰지 않는다. 자료에 답이 없으면 자료에 없다고 적는다. 실제 행동 권한은 이 수업이 정하지 않는다.</p><h4>개인 작업과 조별 작업</h4><p>개인 판단을 먼저 요청한 단계에서는 토의 전에 자신의 답을 남긴다. 그 뒤 조별 답과 수정 이유를 별도로 쓴다. 보고·인계·시험 단계에서는 실제 한 말과 행동을 기록한다.</p>', 'attachments': [], 'materials': ['/uploads/OPS07/OPS07_m09_student.ipynb', '/uploads/OPS07/OPS07_m09_workbook.ipynb'], 'materialText': '<h4>실습 노트북 안내 (선택 보조 도구)</h4><p><strong>학생 실습용</strong>: <code>OPS07_m09_student.ipynb</code> (직접 ANSWER 표와 note를 작성합니다.)<br><strong>교수자·해설용</strong>: <code>OPS07_m09_workbook.ipynb</code> (예시 풀이와 해설 note가 포함되어 있습니다.)<br><em>※ 본 과제는 문서 편집기나 인쇄 양식으로도 수행할 수 있으며, 노트북은 사례·질문·기록 양식을 표로 정리하는 선택 보조 도구입니다.</em></p><h4>자료를 읽는 순서</h4>\n\n<ol>\n<li><p>STEP 1에서는 자신이 앞 미션에서 실제로 작성한 기준서를 수정 전 문서로 보관하고, 바꾼 문장과 이유를 남깁니다.</p></li>\n<li><p>STEP 2~3에서는 B1~B5, A9, B9의 일곱 사례를 수정 전후에 동일하게 사용합니다. A9와 B9는 아래에 제공한 시점까지만 판단 근거로 씁니다.</p></li>\n<li><p>개인별 질문을 아래 다섯 질문 묶음으로 정리합니다. 실제 질문 문장은 별도로 보관하고, STEP 4~5의 보고·질의응답에서는 전후 원표와 계산 결과를 근거로 설명합니다.</p></li>\n</ol>\n\n<h4>자료 1. 전후 비교에 사용할 B1~B5</h4>\n\n<h4>B1</h4>\n\n<p>학교 옥상 인원 2명과 쌍안경으로 보이는 물체.</p>\n\n<p>대피소라는 주민 진술 1건이 있다.</p>\n\n<p>건물 내부 이용과 사람의 지위는 미확인.</p>\n\n<h4>B2</h4>\n\n<p>차량의 표식은 두 수단에서 일치한다고 보고되었다.</p>\n\n<p>정찰조는 예정된 위치 보고가 20분 늦었다.</p>\n\n<p>동일 원자료 재전달 여부와 현재 아군 위치는 미확인.</p>\n\n<h4>B3</h4>\n\n<p>두 수단에서 같은 표식 차량을 보고했다.</p>\n\n<p>옆 민간 차량 1대가 정차해 있다.</p>\n\n<p>곧 떠날 것 같다는 보고가 있지만 실제 이탈은 확인되지 않았다.</p>\n\n<h4>B4</h4>\n\n<p>관찰 보고에는 식별 명확·가상 위험 2등급이라고 적혀 있다.</p>\n\n<p>상급 검토자와 연락이 15분째 되지 않는다.</p>\n\n<p>등급 산정의 원근거와 검토 회신이 미제공이다.</p>\n\n<h4>B5</h4>\n\n<p>영상에 인원 1명이 흰 천을 흔드는 장면이 있다.</p>\n\n<p>항복 의사 등 보호 필요성을 확인해야 할 신호다.</p>\n\n<p>의도·신분·주변 상황은 확인되지 않았다.</p>\n\n<h4>자료 2. A9 — 09:40까지 제공할 정보</h4>\n\n<p><strong>09:10:</strong> 주민 1명이 창고에 적 장비가 있다고 제보했다. 신뢰도 표기는 없다.</p>\n\n<p><strong>09:25:</strong> 영상 1회에 차량 1대가 보인다. 표식은 식별되지 않았다.</p>\n\n<p><strong>09:30:</strong> 상황판에 적 보급소라고 기재되며 추정 표기가 빠졌다.</p>\n\n<p><strong>09:40:</strong> 상급 보고에 확인된 적 보급소로 전달됐다. 민간 업체 운영 정보는 공유되지 않았다.</p>\n\n<h4>자료 3. B9 — 13:05까지 제공할 정보</h4>\n\n<p><strong>13:00:</strong> 정찰조 위치 미확인으로 보류했다. 이 최초 보류는 적절했다고 기록돼 있다.</p>\n\n<p><strong>13:05:</strong> 재확인을 요청했으나 담당자를 정하지 않아 두 팀이 상대 팀이 확인할 것으로 여겼다.</p>\n\n<h4>자료 4. 개인 질문을 정리할 다섯 묶음</h4>\n\n<p>같은 묶음에 들어간 질문이라도 실제 질문 문장이 같다는 뜻은 아닙니다. 원래 질문을 남기고 아래 코드로 분류합니다.</p>\n\n<p><strong>출처:</strong> 원출처·독립 관측 여부·관측 시각·동일 대상 여부</p>\n\n<p><strong>대상:</strong> 보인 특징과 소속·의도 주장의 구분</p>\n\n<p><strong>아군:</strong> 현재 아군 위치·연락·관련 보고</p>\n\n<p><strong>민간:</strong> 주민·민간 차량·시설 이용·보호 필요 신호</p>\n\n<p><strong>접수:</strong> 질문 송신·접수·검토 회신·담당 인계</p>\n\n<h4>자료를 사용할 때 지킬 점</h4>\n\n<p>A9에는 09:52 이후 기록을, B9에는 13:25 이후 기록을 판단 근거로 추가하지 않습니다.</p>\n\n<p>실제 시험하지 않았거나 시간을 재지 않은 항목은 미실시·미측정으로 남깁니다.</p>\n\n<p>모든 사례·시각·장소·역할은 교육용입니다. 필요한 자료와 구체적인 작업 순서는 각 단계 설명에도 들어 있습니다.</p>\n', 'guideSteps': [{'step': 1, 'problem': '고치기 전후의 기준을 두 문서로 보관하기', 'activity': '<h4>이번 단계에서 하는 일</h4><p>문장을 고친 뒤 원래 문장을 잃어버리면 무엇이 달라졌는지 검증할 수 없다. 이전 기준과 수정 기준을 나눠 보관하고 바꾼 이유를 적는다.</p><h4>먼저 읽을 자료</h4>\n\n<p>사용할 자료는 M2~M7에서 작성한 조건·담당·보고 절차와 M8의 수정 제안이다.</p>\n\n<p>앞 문서가 없다면 교관에게 해당 조의 실제 초안을 요청한다.</p>\n\n<p>수정 전 문장을 새로 만들어 과거 초안인 것처럼 쓰지 않는다.</p>\n\n<h4>수정 전후 버전의 뜻</h4>\n\n<ul>\n<li>\n<p><strong>수정 전 v1:</strong> 피드백을 반영하기 전 실제 기준 문서</p>\n</li>\n\n<li>\n<p><strong>수정 후 v2:</strong> 수정 이유와 새 문장이 들어간 기준 문서</p>\n</li>\n</ul>\n\n<h4>작업 순서</h4><ol><li><p>기존 기준을 복사해 수정 전 v1로 저장하고 이후 덮어쓰지 않는다.</p></li><li><p>복사본을 수정 후 v2로 저장한다.</p></li><li><p>고칠 문장마다 조항 번호를 붙인다. 번호가 없으면 장-번호 형태로 새로 붙이고 두 문서에 같게 쓴다.</p></li><li><p>v2에서 문장을 수정하고 비교표에 v1·v2 문장을 그대로 적는다.</p></li><li><p>바꾼 이유를 M7 수행 기록 또는 M8 문제 시각·검토 질문과 연결한다.</p></li><li><p>변경한 조항이 모두 비교표에 있는지 확인하고 두 문서의 파일명·작성자를 적는다.</p></li></ol><h4>작성 예시</h4><p>다음은 쓰는 방식을 보여 주는 예다. 자기 판단이나 실제 수행·측정 결과가 필요한 칸에 그대로 복사하지 않는다.</p><p>6-03 / v1 “확인 요청 후 결과를 기다린다” / v2 “질문·담당·접수 여부를 적고, 답이 없으면 다음 확인 계획을 남긴다” / 이유 “B 13:05 담당 미지정 문제를 보완”.</p><h4>작성할 표 또는 기록 항목</h4><ul><li><p>조항 번호</p></li><li><p>v1 문장</p></li><li><p>v2 문장</p></li><li><p>수정 이유</p></li><li><p>참고한 미션·자료·시각</p></li><li><p>수정 담당</p></li><li><p>두 문서 위치</p></li></ul><h4>제출할 내용과 끝내는 기준</h4><p>실제 수정 전후 문서 두 개와 변경 조항 전부의 비교표.</p><p>미션 제출 문서 briefing/ops07_commander_1page_brief.md의 STEP 01 아래에 작성한다. 개인 기록과 조별 결과를 구분하고 실제 이름 대신 학습자 코드를 써도 된다. 표의 모양이 달라도 필요한 내용을 찾을 수 있으면 된다.</p><h4>마치기 전 확인</h4><ul><li><p>수정 전 실제 문서와 수정 후 문서를 따로 보존했다.</p></li><li><p>같은 조항 번호로 전후 문장을 비교할 수 있다.</p></li><li><p>바꾼 모든 조항에 이유와 관련 기록이 있다.</p></li><li><p>수정 전 문서나 과거 수행을 새로 꾸며 넣지 않았다.</p></li></ul><h4>배정 시간</h4><p>25분</p>', 'evalText': '<h4>평가할 자료와 위치</h4><p>briefing/ops07_commander_1page_brief.md의 STEP 01 결과와 해당 개인 수행 기록을 확인한다. 이 단계에서 요청한 제출 내용은 다음과 같다: 실제 수정 전후 문서 두 개와 변경 조항 전부의 비교표.</p><h4>평가 순서</h4><ol><li><p>아래 통과 조건을 하나씩 읽고 제출물에서 확인한 위치를 옆에 적는다.</p></li><li><p>실제 설명·보고·인계가 필요한 단계는 수행 기록이나 직접 관찰로 확인한다. 서식만 있으면 수행한 것으로 보지 않는다.</p></li><li><p>통과 조건을 모두 충족하면 이 단계는 PASS다. 하나라도 빠지거나 반대로 적었으면 FAIL로 두고 빠진 항목을 구체적으로 적는다. 표현이나 표 모양이 예시와 달라도 뜻이 같고 근거를 찾을 수 있으면 인정한다.</p></li></ol><h4>PASS 조건</h4><ul><li><p>수정 전 실제 문서와 수정 후 문서를 따로 보존했다.</p></li><li><p>같은 조항 번호로 전후 문장을 비교할 수 있다.</p></li><li><p>바꾼 모든 조항에 이유와 관련 기록이 있다.</p></li><li><p>수정 전 문서나 과거 수행을 새로 꾸며 넣지 않았다.</p></li></ul><h4>FAIL에 해당하는 예</h4><ul><li><p>고치기 전 문장을 기억으로 만들어 넣었다.</p></li><li><p>“표현 개선”만 적고 왜 바꿨는지 근거가 없다.</p></li></ul><h4>이해 여부를 확인할 질문</h4><p>이 문장은 어느 기록의 어떤 문제 때문에 바꿨는가?</p><h4>인정할 수 있는 답의 핵심</h4><p>실제 수행 기록·M8 사건 시각·검토 질문 중 해당 근거를 찾아 수정 문장과 연결해 설명한다.</p><h4>보완과 재평가</h4><p>원래 초안을 확보해 비교하고 근거 없는 변경 이유는 실제 피드백 내용으로 다시 쓴다.</p><p>수정 전 결과와 재평가 결과를 구분해 남긴다. 실제 하지 않은 보고·채점·시험을 나중에 했던 것처럼 채우지 않는다.</p>', 'attachments': []}, {'step': 2, 'problem': '같은 일곱 사례에 두 기준을 각각 적용하기', 'activity': '<h4>이번 단계에서 하는 일</h4><p>다른 정보를 주고 전후 판단을 비교하면 기준의 효과인지 자료의 차이인지 알 수 없다. 일곱 사례의 입력을 그대로 두고 기준만 바꿔 시험한다.</p><h4>먼저 읽을 자료</h4>\n\n<h4>사례 자료</h4>\n\n<p><strong>B1</strong></p>\n\n<p>학교 옥상 인원 2명과 쌍안경으로 보이는 물체.</p>\n\n<p>대피소라는 주민 진술 1건이 있다.</p>\n\n<p>건물 내부 이용과 사람의 지위는 미확인.</p>\n\n<p><strong>B2</strong></p>\n\n<p>차량의 표식은 두 수단에서 일치한다고 보고되었다.</p>\n\n<p>정찰조는 예정된 위치 보고가 20분 늦었다.</p>\n\n<p>동일 원자료 재전달 여부와 현재 아군 위치는 미확인.</p>\n\n<p><strong>B3</strong></p>\n\n<p>두 수단에서 같은 표식 차량을 보고했다.</p>\n\n<p>옆 민간 차량 1대가 정차해 있다.</p>\n\n<p>곧 떠날 것 같다는 보고가 있지만 실제 이탈은 확인되지 않았다.</p>\n\n<p><strong>B4</strong></p>\n\n<p>관찰 보고에는 식별 명확·가상 위험 2등급이라고 적혀 있다.</p>\n\n<p>상급 검토자와 연락이 15분째 되지 않는다.</p>\n\n<p>등급 산정의 원근거와 검토 회신이 미제공이다.</p>\n\n<p><strong>B5</strong></p>\n\n<p>영상에 인원 1명이 흰 천을 흔드는 장면이 있다.</p>\n\n<p>항복 의사 등 보호 필요성을 확인해야 할 신호다.</p>\n\n<p>의도·신분·주변 상황은 확인되지 않았다.</p>\n\n<h4>A9에 제공할 정보</h4>\n\n<h4>시각별 기록</h4>\n\n<ul>\n<li>\n<p><strong>09:10:</strong> 주민 1명이 창고에 적 장비가 있다고 제보했다. 신뢰도 표기는 없다.</p>\n</li>\n\n<li>\n<p><strong>09:25:</strong> 영상 1회에 차량 1대가 보인다. 표식은 식별되지 않았다.</p>\n</li>\n\n<li>\n<p><strong>09:30:</strong> 상황판에 적 보급소라고 기재되며 추정 표기가 빠졌다.</p>\n</li>\n\n<li>\n<p><strong>09:40:</strong> 상급 보고에 확인된 적 보급소로 전달됐다. 민간 업체 운영 정보는 공유되지 않았다.</p>\n</li>\n</ul>\n\n<h4>B9에 제공할 정보</h4>\n\n<h4>시각별 기록</h4>\n\n<ul>\n<li>\n<p><strong>13:00:</strong> 정찰조 위치 미확인으로 보류했다. 이 최초 보류는 적절했다고 기록돼 있다.</p>\n</li>\n\n<li>\n<p><strong>13:05:</strong> 재확인을 요청했으나 담당자를 정하지 않아 두 팀이 상대 팀이 확인할 것으로 여겼다.</p>\n</li>\n</ul>\n\n<p>A9에는 09:52 이후 기록을, B9에는 13:25 이후 기록을 판단 근거로 추가하지 않는다.</p>\n\n<p>이 단계의 카드가 각 시험에서 사용할 정보의 전부다.</p>\n\n<h4>질문을 분류할 다섯 묶음</h4>\n\n<ul>\n<li>\n<p><strong>출처:</strong> 원출처·독립 관측 여부·관측 시각·동일 대상 여부</p>\n</li>\n\n<li>\n<p><strong>대상:</strong> 보인 특징과 소속·의도 주장의 구분</p>\n</li>\n\n<li>\n<p><strong>아군:</strong> 현재 아군 위치·연락·관련 보고</p>\n</li>\n\n<li>\n<p><strong>민간:</strong> 주민·민간 차량·시설 이용·보호 필요 신호</p>\n</li>\n\n<li>\n<p><strong>접수:</strong> 질문 송신·접수·검토 회신·담당 인계</p>\n</li>\n</ul>\n\n<p>같은 사례를 다시 보면 기억 때문에 빨라질 수 있다.</p>\n\n<p>시간 차이를 실제 현장 성능으로 일반화하지 않는다.</p>\n\n<p>아래 기록은 직접 측정한 값만 쓴다.</p>\n\n<h4>작업 순서</h4><ol><li><p>수정 전 v1과 수정 후 v2를 준비하고 각 개인의 답을 토의 전에 적는 표를 만든다.</p></li><li><p>B1~B5·A9·B9의 동일한 카드로 v1을 적용한다. 보류 상태·확인 질문·질문 묶음 코드·담당·아직 없는 답을 적는다.</p></li><li><p>같은 카드로 v2를 적용하고 같은 항목을 적는다. 이전 답을 복사하지 말고 해당 기준의 문장을 읽는다.</p></li><li><p>판단 시간을 재기로 했다면 시작과 끝을 같은 방식으로 정하고 초시계로 잰 초를 적는다. 재지 않았다면 미측정이라고 쓴다.</p></li><li><p>개인 답을 보존한 뒤 조별 답을 별도 칸에 적는다.</p></li><li><p>일곱 사례가 두 기준에 모두 있는지, A9·B9에 이후 정보를 넣지 않았는지 확인한다.</p></li></ol><h4>작성 예시</h4><p>다음은 쓰는 방식을 보여 주는 예다. 자기 판단이나 실제 수행·측정 결과가 필요한 칸에 그대로 복사하지 않는다.</p><p>B9의 판단 입력은 13:05까지다. 이때 담당 미지정 문제를 보고 질문·담당을 정할 수는 있지만 13:40에 위치가 확인됐다고 초기 판단에 쓰면 안 된다.</p><h4>작성할 표 또는 기록 항목</h4><ul><li><p>사례·학습자</p></li><li><p>사용 기준 v1·v2</p></li><li><p>보류 상태</p></li><li><p>확인 질문·묶음 코드</p></li><li><p>담당</p></li><li><p>아직 없는 답</p></li><li><p>실측 초 또는 미측정</p></li><li><p>조별 답은 별도 표시</p></li></ul><h4>제출할 내용과 끝내는 기준</h4><p>일곱 사례×두 기준의 개인 답과 조별 답. v1·v2 모두 일곱 건을 빠짐없이 기록한다.</p><p>미션 제출 문서 briefing/ops07_commander_1page_brief.md의 STEP 02 아래에 작성한다. 개인 기록과 조별 결과를 구분하고 실제 이름 대신 학습자 코드를 써도 된다. 표의 모양이 달라도 필요한 내용을 찾을 수 있으면 된다.</p><h4>마치기 전 확인</h4><ul><li><p>일곱 사례에 v1·v2 결과가 모두 있다.</p></li><li><p>전후에 사용한 카드 내용과 정보 시점이 같다.</p></li><li><p>개인 답과 토의 후 조별 답을 구분했다.</p></li><li><p>측정하지 않은 시간은 미측정으로 표시했다.</p></li><li><p>A9·B9에 이후 결과를 초기 근거로 넣지 않았다.</p></li></ul><h4>배정 시간</h4><p>55분</p>', 'evalText': '<h4>평가할 자료와 위치</h4><p>briefing/ops07_commander_1page_brief.md의 STEP 02 결과와 해당 개인 수행 기록을 확인한다. 이 단계에서 요청한 제출 내용은 다음과 같다: 일곱 사례×두 기준의 개인 답과 조별 답. v1·v2 모두 일곱 건을 빠짐없이 기록한다.</p><h4>평가 순서</h4><ol><li><p>아래 통과 조건을 하나씩 읽고 제출물에서 확인한 위치를 옆에 적는다.</p></li><li><p>실제 설명·보고·인계가 필요한 단계는 수행 기록이나 직접 관찰로 확인한다. 서식만 있으면 수행한 것으로 보지 않는다.</p></li><li><p>통과 조건을 모두 충족하면 이 단계는 PASS다. 하나라도 빠지거나 반대로 적었으면 FAIL로 두고 빠진 항목을 구체적으로 적는다. 표현이나 표 모양이 예시와 달라도 뜻이 같고 근거를 찾을 수 있으면 인정한다.</p></li></ol><h4>PASS 조건</h4><ul><li><p>일곱 사례에 v1·v2 결과가 모두 있다.</p></li><li><p>전후에 사용한 카드 내용과 정보 시점이 같다.</p></li><li><p>개인 답과 토의 후 조별 답을 구분했다.</p></li><li><p>측정하지 않은 시간은 미측정으로 표시했다.</p></li><li><p>A9·B9에 이후 결과를 초기 근거로 넣지 않았다.</p></li></ul><h4>FAIL에 해당하는 예</h4><ul><li><p>v1에는 초기 자료, v2에는 나중 확인된 결과를 줬다.</p></li><li><p>시간을 재지 않았는데 모두 같은 임의 초를 넣었다.</p></li></ul><h4>이해 여부를 확인할 질문</h4><p>A9 판단에 10:30 결과를 넣으면 비교가 왜 달라지는가?</p><h4>인정할 수 있는 답의 핵심</h4><p>수정 전후의 자료 조건이 달라져 기준을 고친 효과와 나중 정보를 준 효과를 구분할 수 없다고 답한다.</p><h4>보완과 재평가</h4><p>이 단계의 카드로 입력을 다시 고정하고 오염된 시험을 재실시한다. 시간은 실제 측정값만 다시 적는다.</p><p>수정 전 결과와 재평가 결과를 구분해 남긴다. 실제 하지 않은 보고·채점·시험을 나중에 했던 것처럼 채우지 않는다.</p>', 'attachments': []}, {'step': 3, 'problem': '일치율·누락 건수·판단 시간을 전후로 비교하기', 'activity': '<h4>이번 단계에서 하는 일</h4><p>빨리 답하거나 같은 답을 냈다고 확인 업무가 좋아졌다고 볼 수는 없다. 원래 답을 보며 빠뜨린 질문·담당이 줄었는지도 함께 살핀다.</p><h4>먼저 읽을 자료</h4>\n\n<h4>질문을 분류할 다섯 묶음</h4>\n\n<ul>\n<li>\n<p><strong>출처:</strong> 원출처·독립 관측 여부·관측 시각·동일 대상 여부</p>\n</li>\n\n<li>\n<p><strong>대상:</strong> 보인 특징과 소속·의도 주장의 구분</p>\n</li>\n\n<li>\n<p><strong>아군:</strong> 현재 아군 위치·연락·관련 보고</p>\n</li>\n\n<li>\n<p><strong>민간:</strong> 주민·민간 차량·시설 이용·보호 필요 신호</p>\n</li>\n\n<li>\n<p><strong>접수:</strong> 질문 송신·접수·검토 회신·담당 인계</p>\n</li>\n</ul>\n\n<p>비교 자료는 일곱 사례를 v1·v2로 각각 판단한 개인 원답이다.</p>\n\n<p>같은 사례에서 보류 상태와 질문 묶음 코드가 전원 같으면 그 사례를 전원 일치 1건으로 센다.</p>\n\n<p>문장 표현만 다르고 질문 묶음이 같으면 코드 기준으로 비교한다.</p>\n\n<h4>작업 순서</h4><ol><li><p>v1에서 전원이 같은 보류 상태·질문 묶음 코드를 적은 사례 수를 센다. v2도 따로 센다.</p></li><li><p>각 값을 7로 나누고 100을 곱해 전후 일치율을 계산한다.</p></li><li><p>각 사례에서 질문·담당·요청 접수 상태를 빼먹은 칸을 각각 센다. 자료가 없어 “미확인”이라고 적은 것은 누락으로 세지 않는다.</p></li><li><p>시간을 쟀다면 전후 같은 사례의 초를 나란히 적고 차이를 계산한다. 미측정은 계산에서 제외하되 몇 건인지 밝힌다.</p></li><li><p>변화가 없는 사례나 더 오래 걸린 사례도 그대로 적는다.</p></li><li><p>빨라졌지만 중요한 확인이 빠진 경우에는 개선이라고 쓰지 않는다. 같은 카드를 반복해 본 효과도 한계에 적는다.</p></li></ol><h4>작성 예시</h4><p>다음은 쓰는 방식을 보여 주는 예다. 자기 판단이나 실제 수행·측정 결과가 필요한 칸에 그대로 복사하지 않는다.</p><p>v1 전원 일치 4건, v2 5건이라면 4÷7×100≈57.1%, 5÷7×100≈71.4%다. 이 수치는 계산 예시이므로 자신의 결과로 옮겨 쓰지 않는다. “아군 위치 미확인”이라고 적은 것은 질문을 빠뜨린 것과 다르다.</p><h4>작성할 표 또는 기록 항목</h4><ul><li><p>사례별 전후 개인 원답</p></li><li><p>전원 일치 여부·합계·계산식</p></li><li><p>질문·담당·접수 상태의 누락 건수</p></li><li><p>실측 시간 전후·미측정 건수</p></li><li><p>변화 설명·한계</p></li></ul><h4>제출할 내용과 끝내는 기준</h4><p>일치율 계산표, 세 종류 누락 건수 비교, 실측 시간 비교 또는 미측정 설명, 해석.</p><p>미션 제출 문서 briefing/ops07_commander_1page_brief.md의 STEP 03 아래에 작성한다. 개인 기록과 조별 결과를 구분하고 실제 이름 대신 학습자 코드를 써도 된다. 표의 모양이 달라도 필요한 내용을 찾을 수 있으면 된다.</p><h4>마치기 전 확인</h4><ul><li><p>분자 전원 일치 건수와 분모 7이 원답과 맞다.</p></li><li><p>미확인으로 적은 항목과 아예 빠뜨린 항목을 구분했다.</p></li><li><p>질문·담당·접수 상태 누락을 전후 같은 기준으로 셌다.</p></li><li><p>실측 시간만 비교하고 미측정 수를 밝혔다.</p></li><li><p>반복 학습·변화 없음·악화 사례를 숨기지 않았다.</p></li></ul><h4>배정 시간</h4><p>40분</p>', 'evalText': '<h4>평가할 자료와 위치</h4><p>briefing/ops07_commander_1page_brief.md의 STEP 03 결과와 해당 개인 수행 기록을 확인한다. 이 단계에서 요청한 제출 내용은 다음과 같다: 일치율 계산표, 세 종류 누락 건수 비교, 실측 시간 비교 또는 미측정 설명, 해석.</p><h4>평가 순서</h4><ol><li><p>아래 통과 조건을 하나씩 읽고 제출물에서 확인한 위치를 옆에 적는다.</p></li><li><p>실제 설명·보고·인계가 필요한 단계는 수행 기록이나 직접 관찰로 확인한다. 서식만 있으면 수행한 것으로 보지 않는다.</p></li><li><p>통과 조건을 모두 충족하면 이 단계는 PASS다. 하나라도 빠지거나 반대로 적었으면 FAIL로 두고 빠진 항목을 구체적으로 적는다. 표현이나 표 모양이 예시와 달라도 뜻이 같고 근거를 찾을 수 있으면 인정한다.</p></li></ol><h4>PASS 조건</h4><ul><li><p>분자 전원 일치 건수와 분모 7이 원답과 맞다.</p></li><li><p>미확인으로 적은 항목과 아예 빠뜨린 항목을 구분했다.</p></li><li><p>질문·담당·접수 상태 누락을 전후 같은 기준으로 셌다.</p></li><li><p>실측 시간만 비교하고 미측정 수를 밝혔다.</p></li><li><p>반복 학습·변화 없음·악화 사례를 숨기지 않았다.</p></li></ul><h4>FAIL에 해당하는 예</h4><ul><li><p>불일치 사례를 빼고 분모를 바꿨다.</p></li><li><p>빠진 질문이 늘었는데 시간만 줄었다며 성공이라고 썼다.</p></li></ul><h4>이해 여부를 확인할 질문</h4><p>“현재 위치 미확인”이라고 쓴 기록과 위치 항목을 아예 빼먹은 기록은 같은 누락인가?</p><h4>인정할 수 있는 답의 핵심</h4><p>앞은 확인하지 못한 내용을 남긴 기록이며 뒤는 필요한 항목을 빠뜨린 것이므로 다르다고 답한다.</p><h4>보완과 재평가</h4><p>누락 정의와 원답을 다시 대조해 계산하고 효과를 과장한 해석을 고친다.</p><p>수정 전 결과와 재평가 결과를 구분해 남긴다. 실제 하지 않은 보고·채점·시험을 나중에 했던 것처럼 채우지 않는다.</p>', 'attachments': []}, {'step': 4, 'problem': '결과와 세 가지 절차 개선안을 한 쪽 보고서로 쓰기', 'activity': '<h4>이번 단계에서 하는 일</h4><p>검토자는 긴 기록에서 결론을 찾아야 하는 부담 없이 무엇을 바꿔 달라는지 알아야 한다. 한 쪽에 결론·근거·남은 문제를 정리한다.</p><h4>먼저 읽을 자료</h4>\n\n<h4>한 쪽 보고서에 넣을 내용</h4>\n\n<ul>\n<li>\n<p><strong>첫 문장:</strong> 어떤 절차를 채택하거나 더 시험해 달라는지</p>\n</li>\n\n<li>\n<p><strong>확인한 결과:</strong> 일곱 사례의 전후 결과와 누락·실측·미측정</p>\n</li>\n\n<li>\n<p><strong>아직 남은 문제:</strong> 확인하지 못한 사항과 시험 한계</p>\n</li>\n\n<li>\n<p><strong>대안 세 가지:</strong> 서로 다른 업무 개선안, 장점·단점·필요한 담당·자원</p>\n</li>\n\n<li>\n<p><strong>권고:</strong> 추천하는 안과 그 이유·추가 시험</p>\n</li>\n</ul>\n\n<p>색을 쓰면 빨강·노랑·초록 옆에 말로도 뜻을 쓴다.</p>\n\n<p>초록은 검토 자료 정리가 끝났다는 교육용 표시일 수 있으나 실제 진행 허가를 뜻하지 않는다.</p>\n\n<h4>작업 순서</h4><ol><li><p>A4 한 쪽 문서를 열고 첫 문장에 검토자가 결정해 줄 사항을 쓴다.</p></li><li><p>전후 시험에서 실제 확인한 수치와 사례를 짧게 적고 자세한 표의 기록 번호를 붙인다.</p></li><li><p>미확인·미측정과 같은 카드 반복의 한계를 적는다.</p></li><li><p>서로 다른 절차 대안 세 가지를 만든다. 예를 들어 담당 지정 방식, 접수 확인 방식, 기록 양식의 변경을 비교할 수 있다.</p></li><li><p>각 안의 장점·단점·필요 담당 또는 자원을 쓰고 추천안을 고른다.</p></li><li><p>인쇄 미리 보기에서 한 쪽인지 확인한다. 분량을 줄이더라도 미확인이나 필수 확인 조건을 삭제하지 않는다.</p></li></ol><h4>작성 예시</h4><p>다음은 쓰는 방식을 보여 주는 예다. 자기 판단이나 실제 수행·측정 결과가 필요한 칸에 그대로 복사하지 않는다.</p><p>첫 문장 예: “질문을 보낼 때 확인 담당과 접수 여부를 함께 적는 양식을 다음 훈련에 적용해 추가 시험할 것을 건의합니다.” 실제 시험하지 않은 효과를 확정하지 않는다.</p><h4>작성할 표 또는 기록 항목</h4><ul><li><p>A4 한 쪽</p></li><li><p>결정 요청</p></li><li><p>실제 전후 결과와 근거 위치</p></li><li><p>남은 문제·한계</p></li><li><p>절차 대안 세 가지와 장단점·자원</p></li><li><p>추천안</p></li></ul><h4>제출할 내용과 끝내는 기준</h4><p>인쇄 미리 보기에서 A4 한 쪽으로 확인한 보고서와 참조한 상세 표 번호.</p><p>미션 제출 문서 briefing/ops07_commander_1page_brief.md의 STEP 04 아래에 작성한다. 개인 기록과 조별 결과를 구분하고 실제 이름 대신 학습자 코드를 써도 된다. 표의 모양이 달라도 필요한 내용을 찾을 수 있으면 된다.</p><h4>마치기 전 확인</h4><ul><li><p>첫 문장에 검토자가 결정할 사항이 있다.</p></li><li><p>일곱 사례 시험 결과를 실제 원표와 연결했다.</p></li><li><p>미확인·미측정·반복 학습 한계를 밝혔다.</p></li><li><p>대안 세 가지마다 장단점과 필요한 담당·자원이 있다.</p></li><li><p>A4 한 쪽이며 색 표시가 실제 행동 허가처럼 쓰이지 않았다.</p></li></ul><h4>배정 시간</h4><p>35분</p>', 'evalText': '<h4>평가할 자료와 위치</h4><p>briefing/ops07_commander_1page_brief.md의 STEP 04 결과와 해당 개인 수행 기록을 확인한다. 이 단계에서 요청한 제출 내용은 다음과 같다: 인쇄 미리 보기에서 A4 한 쪽으로 확인한 보고서와 참조한 상세 표 번호.</p><h4>평가 순서</h4><ol><li><p>아래 통과 조건을 하나씩 읽고 제출물에서 확인한 위치를 옆에 적는다.</p></li><li><p>실제 설명·보고·인계가 필요한 단계는 수행 기록이나 직접 관찰로 확인한다. 서식만 있으면 수행한 것으로 보지 않는다.</p></li><li><p>통과 조건을 모두 충족하면 이 단계는 PASS다. 하나라도 빠지거나 반대로 적었으면 FAIL로 두고 빠진 항목을 구체적으로 적는다. 표현이나 표 모양이 예시와 달라도 뜻이 같고 근거를 찾을 수 있으면 인정한다.</p></li></ol><h4>PASS 조건</h4><ul><li><p>첫 문장에 검토자가 결정할 사항이 있다.</p></li><li><p>일곱 사례 시험 결과를 실제 원표와 연결했다.</p></li><li><p>미확인·미측정·반복 학습 한계를 밝혔다.</p></li><li><p>대안 세 가지마다 장단점과 필요한 담당·자원이 있다.</p></li><li><p>A4 한 쪽이며 색 표시가 실제 행동 허가처럼 쓰이지 않았다.</p></li></ul><h4>FAIL에 해당하는 예</h4><ul><li><p>“안전성을 높인다”는 설명만 있고 무엇을 바꿔 달라는지 없다.</p></li><li><p>대안을 실제 공격 방법 세 가지로 작성했다.</p></li></ul><h4>이해 여부를 확인할 질문</h4><p>추천안의 장점뿐 아니라 어떤 부담이나 한계가 있는가?</p><h4>인정할 수 있는 답의 핵심</h4><p>자기 안에 맞는 추가 기록 부담·교육 필요·미시험 항목 등을 구체적으로 설명한다.</p><h4>보완과 재평가</h4><p>결정 요청과 세 대안의 실제 업무 차이를 분명히 쓰고 근거 없는 성과 문장을 고친다.</p><p>수정 전 결과와 재평가 결과를 구분해 남긴다. 실제 하지 않은 보고·채점·시험을 나중에 했던 것처럼 채우지 않는다.</p>', 'attachments': []}, {'step': 5, 'problem': '한 쪽 보고서를 발표하고 질문 세 개에 답하기', 'activity': '<h4>이번 단계에서 하는 일</h4><p>보고서를 읽은 사람이 근거와 한계를 물었을 때 원기록을 찾아 설명할 수 있어야 한다. 짧은 발표와 질문 응답을 따로 기록한다.</p><h4>먼저 읽을 자료</h4>\n\n<p>준비물은 A4 한 쪽 보고서, 일곱 사례 전후 원표, 초시계다.</p>\n\n<p>발표자는 보고서를 읽기만 하지 말고 무엇을 바꿔 달라는지부터 말한다.</p>\n\n<h4>질문 예</h4>\n\n<ul>\n<li>\n<p>어느 사례의 어떤 기록에서 개선을 확인했는가</p>\n</li>\n\n<li>\n<p>시간을 재지 않은 항목은 무엇인가</p>\n</li>\n\n<li>\n<p>같은 카드를 반복해서 본 영향은 어떻게 남겼는가</p>\n</li>\n</ul>\n\n<h4>작업 순서</h4><ol><li><p>기록자와 시간을 잴 사람을 정하고 발표를 시작한다.</p></li><li><p>3분 안에 결정 요청·주요 결과·추천안·남은 문제를 말한다.</p></li><li><p>발표 끝 시간을 적은 뒤 질의응답은 별도로 시작한다. 질문 시간은 3분 발표 시간에 섞지 않는다.</p></li><li><p>다른 조원이나 교관이 서로 다른 질문을 세 개 이상 한다.</p></li><li><p>발표자는 해당 원표나 문장을 찾아 답한다. 모르면 모른다고 말하고 추가 확인할 사람과 내용을 적는다.</p></li><li><p>실제 질문·답·미답·추가 확인 계획과 발표 시간을 기록한다. 3분을 넘으면 보고를 고친 뒤 다시 발표한다.</p></li></ol><h4>작성 예시</h4><p>다음은 쓰는 방식을 보여 주는 예다. 자기 판단이나 실제 수행·측정 결과가 필요한 칸에 그대로 복사하지 않는다.</p><p>질문 “판단 시간이 줄었다는 수치는 어디 있습니까?” → 시간을 재지 않았다면 “이번 시험에서는 재지 않아 단축 여부를 말할 수 없습니다”라고 답한다.</p><h4>작성할 표 또는 기록 항목</h4><ul><li><p>발표자·기록자</p></li><li><p>실측 발표 시간</p></li><li><p>서로 다른 질문 세 개 이상</p></li><li><p>실제로 답한 내용·찾은 기록</p></li><li><p>아직 답하지 못한 질문·담당</p></li><li><p>재발표 결과</p></li></ul><h4>제출할 내용과 끝내는 기준</h4><p>3분 이내 발표 기록, 세 질문 이상의 실제 응답 기록, 미답 질문 계획.</p><p>미션 제출 문서 briefing/ops07_commander_1page_brief.md의 STEP 05 아래에 작성한다. 개인 기록과 조별 결과를 구분하고 실제 이름 대신 학습자 코드를 써도 된다. 표의 모양이 달라도 필요한 내용을 찾을 수 있으면 된다.</p><h4>마치기 전 확인</h4><ul><li><p>발표 시간은 실제 측정했고 질의응답과 분리했다.</p></li><li><p>최종 발표가 3분 이내다.</p></li><li><p>서로 다른 세 질문 이상과 실제 답을 기록했다.</p></li><li><p>답의 근거를 찾았거나 모르는 범위와 추가 확인을 정직하게 밝혔다.</p></li><li><p>시간 초과·보완이 있었다면 재발표 결과를 남겼다.</p></li></ul><h4>배정 시간</h4><p>25분</p>', 'evalText': '<h4>평가할 자료와 위치</h4><p>briefing/ops07_commander_1page_brief.md의 STEP 05 결과와 해당 개인 수행 기록을 확인한다. 이 단계에서 요청한 제출 내용은 다음과 같다: 3분 이내 발표 기록, 세 질문 이상의 실제 응답 기록, 미답 질문 계획.</p><h4>평가 순서</h4><ol><li><p>아래 통과 조건을 하나씩 읽고 제출물에서 확인한 위치를 옆에 적는다.</p></li><li><p>실제 설명·보고·인계가 필요한 단계는 수행 기록이나 직접 관찰로 확인한다. 서식만 있으면 수행한 것으로 보지 않는다.</p></li><li><p>통과 조건을 모두 충족하면 이 단계는 PASS다. 하나라도 빠지거나 반대로 적었으면 FAIL로 두고 빠진 항목을 구체적으로 적는다. 표현이나 표 모양이 예시와 달라도 뜻이 같고 근거를 찾을 수 있으면 인정한다.</p></li></ol><h4>PASS 조건</h4><ul><li><p>발표 시간은 실제 측정했고 질의응답과 분리했다.</p></li><li><p>최종 발표가 3분 이내다.</p></li><li><p>서로 다른 세 질문 이상과 실제 답을 기록했다.</p></li><li><p>답의 근거를 찾았거나 모르는 범위와 추가 확인을 정직하게 밝혔다.</p></li><li><p>시간 초과·보완이 있었다면 재발표 결과를 남겼다.</p></li></ul><h4>FAIL에 해당하는 예</h4><ul><li><p>질문 세 개를 미리 적어 두기만 하고 실제 응답을 하지 않았다.</p></li><li><p>모르는 수치에 맞춰 즉석에서 측정값을 만들었다.</p></li></ul><h4>이해 여부를 확인할 질문</h4><p>지금 원표로 확인할 수 없는 질문을 받으면 어떻게 답해야 하는가?</p><h4>인정할 수 있는 답의 핵심</h4><p>확인하지 못한 범위를 밝히고 누가 무엇을 더 확인할지 적으며 수치를 만들지 않는다고 답한다.</p><h4>보완과 재평가</h4><p>미흡한 답의 근거를 찾아 보완하거나 미확인으로 명확히 남기고 필요한 발표를 다시 실시한다.</p><p>수정 전 결과와 재평가 결과를 구분해 남긴다. 실제 하지 않은 보고·채점·시험을 나중에 했던 것처럼 채우지 않는다.</p>', 'attachments': []}], 'deliverables': ['briefing/ops07_commander_1page_brief.md (필수) — 실제 기준 v1·v2, 일곱 사례 전후 개인 답·계산, 한 쪽 보고서·세 대안, 발표·세 질문 응답'], 'techStack': ['문서 편집기(Markdown · briefing/ops07_commander_1page_brief.md) 또는 인쇄 양식', '기준 v1·v2 조항 대조표 및 7사례(B1~B5·A9·B9) 전후 비교표', '5묶음 누락 집계표, 일치율 계산표 및 초시계(실측·미측정 분리)', 'BLUF A4 1쪽 보고서(3대안 포함) 및 3분 발표·질의응답 기록지'], 'evaluationKeys': ['<h4>평가 질문</h4><p>같은 일곱 사례로 전후 기준을 비교하고 계산·한계를 올바르게 적었는가</p><h4>평가자가 확인할 위치</h4><p>briefing/ops07_commander_1page_brief.md의 STEP 01, STEP 02, STEP 03과 그 단계의 개인 기록을 확인한다.</p><h4>PASS를 주는 방법</h4><p>아래 단계별 항목을 모두 충족해야 이 평가 질문에 PASS를 준다. 항목 옆에 확인한 표·줄·회차를 적는다. 한 항목이라도 없거나 틀리면 FAIL 사유에 해당 단계와 문장을 적는다. 학습자가 쓴 말이 예시와 똑같을 필요는 없지만 확인할 내용은 모두 있어야 한다.</p><h4>STEP 01 고치기 전후의 기준을 두 문서로 보관하기</h4><p>찾을 결과: 실제 수정 전후 문서 두 개와 변경 조항 전부의 비교표.</p><ul><li><p>수정 전 실제 문서와 수정 후 문서를 따로 보존했다.</p></li><li><p>같은 조항 번호로 전후 문장을 비교할 수 있다.</p></li><li><p>바꾼 모든 조항에 이유와 관련 기록이 있다.</p></li><li><p>수정 전 문서나 과거 수행을 새로 꾸며 넣지 않았다.</p></li></ul><h4>STEP 02 같은 일곱 사례에 두 기준을 각각 적용하기</h4><p>찾을 결과: 일곱 사례×두 기준의 개인 답과 조별 답. v1·v2 모두 일곱 건을 빠짐없이 기록한다.</p><ul><li><p>일곱 사례에 v1·v2 결과가 모두 있다.</p></li><li><p>전후에 사용한 카드 내용과 정보 시점이 같다.</p></li><li><p>개인 답과 토의 후 조별 답을 구분했다.</p></li><li><p>측정하지 않은 시간은 미측정으로 표시했다.</p></li><li><p>A9·B9에 이후 결과를 초기 근거로 넣지 않았다.</p></li></ul><h4>STEP 03 일치율·누락 건수·판단 시간을 전후로 비교하기</h4><p>찾을 결과: 일치율 계산표, 세 종류 누락 건수 비교, 실측 시간 비교 또는 미측정 설명, 해석.</p><ul><li><p>분자 전원 일치 건수와 분모 7이 원답과 맞다.</p></li><li><p>미확인으로 적은 항목과 아예 빠뜨린 항목을 구분했다.</p></li><li><p>질문·담당·접수 상태 누락을 전후 같은 기준으로 셌다.</p></li><li><p>실측 시간만 비교하고 미측정 수를 밝혔다.</p></li><li><p>반복 학습·변화 없음·악화 사례를 숨기지 않았다.</p></li></ul><h4>이런 결과는 통과시키지 않는다</h4><ul><li><p>고치기 전 문장을 기억으로 만들어 넣었다.</p></li><li><p>v1에는 초기 자료, v2에는 나중 확인된 결과를 줬다.</p></li><li><p>불일치 사례를 빼고 분모를 바꿨다.</p></li></ul><h4>평가 의견 작성 방법</h4><p>“근거 부족”이라고만 쓰지 말고 단계 번호, 충족하지 못한 조건, 제출물에서 실제로 빠지거나 잘못된 부분, 다시 할 일을 적는다. 예를 들어 STEP 01의 첫 조건은 “수정 전 실제 문서와 수정 후 문서를 따로 보존했다.”이다. 이 조건에 못 미쳤다면 어떤 줄·항목·참여자 기록이 빠졌는지 지목한다.</p><h4>다시 평가할 때</h4><p>틀리거나 빠진 내용만 구체적으로 알려 준다. 문서 작성 오류는 원자료와 수정본을 대조한다. 실제 수행 오류는 해당 행동을 다시 하게 하고 새 회차 기록으로 확인한다. 최초 결과는 지우지 않는다.</p>', '<h4>평가 질문</h4><p>한 쪽 보고서와 세 대안을 작성하고 발표·질문 응답을 수행했는가</p><h4>평가자가 확인할 위치</h4><p>briefing/ops07_commander_1page_brief.md의 STEP 04, STEP 05과 그 단계의 개인 기록을 확인한다.</p><h4>PASS를 주는 방법</h4><p>아래 단계별 항목을 모두 충족해야 이 평가 질문에 PASS를 준다. 항목 옆에 확인한 표·줄·회차를 적는다. 한 항목이라도 없거나 틀리면 FAIL 사유에 해당 단계와 문장을 적는다. 학습자가 쓴 말이 예시와 똑같을 필요는 없지만 확인할 내용은 모두 있어야 한다.</p><h4>STEP 04 결과와 세 가지 절차 개선안을 한 쪽 보고서로 쓰기</h4><p>찾을 결과: 인쇄 미리 보기에서 A4 한 쪽으로 확인한 보고서와 참조한 상세 표 번호.</p><ul><li><p>첫 문장에 검토자가 결정할 사항이 있다.</p></li><li><p>일곱 사례 시험 결과를 실제 원표와 연결했다.</p></li><li><p>미확인·미측정·반복 학습 한계를 밝혔다.</p></li><li><p>대안 세 가지마다 장단점과 필요한 담당·자원이 있다.</p></li><li><p>A4 한 쪽이며 색 표시가 실제 행동 허가처럼 쓰이지 않았다.</p></li></ul><h4>STEP 05 한 쪽 보고서를 발표하고 질문 세 개에 답하기</h4><p>찾을 결과: 3분 이내 발표 기록, 세 질문 이상의 실제 응답 기록, 미답 질문 계획.</p><ul><li><p>발표 시간은 실제 측정했고 질의응답과 분리했다.</p></li><li><p>최종 발표가 3분 이내다.</p></li><li><p>서로 다른 세 질문 이상과 실제 답을 기록했다.</p></li><li><p>답의 근거를 찾았거나 모르는 범위와 추가 확인을 정직하게 밝혔다.</p></li><li><p>시간 초과·보완이 있었다면 재발표 결과를 남겼다.</p></li></ul><h4>이런 결과는 통과시키지 않는다</h4><ul><li><p>“안전성을 높인다”는 설명만 있고 무엇을 바꿔 달라는지 없다.</p></li><li><p>질문 세 개를 미리 적어 두기만 하고 실제 응답을 하지 않았다.</p></li></ul><h4>평가 의견 작성 방법</h4><p>“근거 부족”이라고만 쓰지 말고 단계 번호, 충족하지 못한 조건, 제출물에서 실제로 빠지거나 잘못된 부분, 다시 할 일을 적는다. 예를 들어 STEP 04의 첫 조건은 “첫 문장에 검토자가 결정할 사항이 있다.”이다. 이 조건에 못 미쳤다면 어떤 줄·항목·참여자 기록이 빠졌는지 지목한다.</p><h4>다시 평가할 때</h4><p>틀리거나 빠진 내용만 구체적으로 알려 준다. 문서 작성 오류는 원자료와 수정본을 대조한다. 실제 수행 오류는 해당 행동을 다시 하게 하고 새 회차 기록으로 확인한다. 최초 결과는 지우지 않는다.</p>'], 'evaluationTypes': ['REQUIRED', 'REQUIRED'], 'evaluationText': '<h4>미션 전체 판정</h4><p>이 미션은 아래 두 평가 질문이 모두 PASS여야 통과다. 조별 문서와 개인의 실제 수행을 나눠 확인한다. 수행하지 않았거나 기록이 없어 확인할 수 없는 사람은 미평가로 남기고 다시 실시한다. 최종 판정에서는 미평가를 PASS로 바꾸지 않는다.</p><h4>평가 질문</h4><p>같은 일곱 사례로 전후 기준을 비교하고 계산·한계를 올바르게 적었는가</p><h4>평가자가 확인할 위치</h4><p>briefing/ops07_commander_1page_brief.md의 STEP 01, STEP 02, STEP 03과 그 단계의 개인 기록을 확인한다.</p><h4>PASS를 주는 방법</h4><p>아래 단계별 항목을 모두 충족해야 이 평가 질문에 PASS를 준다. 항목 옆에 확인한 표·줄·회차를 적는다. 한 항목이라도 없거나 틀리면 FAIL 사유에 해당 단계와 문장을 적는다. 학습자가 쓴 말이 예시와 똑같을 필요는 없지만 확인할 내용은 모두 있어야 한다.</p><h4>STEP 01 고치기 전후의 기준을 두 문서로 보관하기</h4><p>찾을 결과: 실제 수정 전후 문서 두 개와 변경 조항 전부의 비교표.</p><ul><li><p>수정 전 실제 문서와 수정 후 문서를 따로 보존했다.</p></li><li><p>같은 조항 번호로 전후 문장을 비교할 수 있다.</p></li><li><p>바꾼 모든 조항에 이유와 관련 기록이 있다.</p></li><li><p>수정 전 문서나 과거 수행을 새로 꾸며 넣지 않았다.</p></li></ul><h4>STEP 02 같은 일곱 사례에 두 기준을 각각 적용하기</h4><p>찾을 결과: 일곱 사례×두 기준의 개인 답과 조별 답. v1·v2 모두 일곱 건을 빠짐없이 기록한다.</p><ul><li><p>일곱 사례에 v1·v2 결과가 모두 있다.</p></li><li><p>전후에 사용한 카드 내용과 정보 시점이 같다.</p></li><li><p>개인 답과 토의 후 조별 답을 구분했다.</p></li><li><p>측정하지 않은 시간은 미측정으로 표시했다.</p></li><li><p>A9·B9에 이후 결과를 초기 근거로 넣지 않았다.</p></li></ul><h4>STEP 03 일치율·누락 건수·판단 시간을 전후로 비교하기</h4><p>찾을 결과: 일치율 계산표, 세 종류 누락 건수 비교, 실측 시간 비교 또는 미측정 설명, 해석.</p><ul><li><p>분자 전원 일치 건수와 분모 7이 원답과 맞다.</p></li><li><p>미확인으로 적은 항목과 아예 빠뜨린 항목을 구분했다.</p></li><li><p>질문·담당·접수 상태 누락을 전후 같은 기준으로 셌다.</p></li><li><p>실측 시간만 비교하고 미측정 수를 밝혔다.</p></li><li><p>반복 학습·변화 없음·악화 사례를 숨기지 않았다.</p></li></ul><h4>이런 결과는 통과시키지 않는다</h4><ul><li><p>고치기 전 문장을 기억으로 만들어 넣었다.</p></li><li><p>v1에는 초기 자료, v2에는 나중 확인된 결과를 줬다.</p></li><li><p>불일치 사례를 빼고 분모를 바꿨다.</p></li></ul><h4>평가 의견 작성 방법</h4><p>“근거 부족”이라고만 쓰지 말고 단계 번호, 충족하지 못한 조건, 제출물에서 실제로 빠지거나 잘못된 부분, 다시 할 일을 적는다. 예를 들어 STEP 01의 첫 조건은 “수정 전 실제 문서와 수정 후 문서를 따로 보존했다.”이다. 이 조건에 못 미쳤다면 어떤 줄·항목·참여자 기록이 빠졌는지 지목한다.</p><h4>다시 평가할 때</h4><p>틀리거나 빠진 내용만 구체적으로 알려 준다. 문서 작성 오류는 원자료와 수정본을 대조한다. 실제 수행 오류는 해당 행동을 다시 하게 하고 새 회차 기록으로 확인한다. 최초 결과는 지우지 않는다.</p><h4>평가 질문</h4><p>한 쪽 보고서와 세 대안을 작성하고 발표·질문 응답을 수행했는가</p><h4>평가자가 확인할 위치</h4><p>briefing/ops07_commander_1page_brief.md의 STEP 04, STEP 05과 그 단계의 개인 기록을 확인한다.</p><h4>PASS를 주는 방법</h4><p>아래 단계별 항목을 모두 충족해야 이 평가 질문에 PASS를 준다. 항목 옆에 확인한 표·줄·회차를 적는다. 한 항목이라도 없거나 틀리면 FAIL 사유에 해당 단계와 문장을 적는다. 학습자가 쓴 말이 예시와 똑같을 필요는 없지만 확인할 내용은 모두 있어야 한다.</p><h4>STEP 04 결과와 세 가지 절차 개선안을 한 쪽 보고서로 쓰기</h4><p>찾을 결과: 인쇄 미리 보기에서 A4 한 쪽으로 확인한 보고서와 참조한 상세 표 번호.</p><ul><li><p>첫 문장에 검토자가 결정할 사항이 있다.</p></li><li><p>일곱 사례 시험 결과를 실제 원표와 연결했다.</p></li><li><p>미확인·미측정·반복 학습 한계를 밝혔다.</p></li><li><p>대안 세 가지마다 장단점과 필요한 담당·자원이 있다.</p></li><li><p>A4 한 쪽이며 색 표시가 실제 행동 허가처럼 쓰이지 않았다.</p></li></ul><h4>STEP 05 한 쪽 보고서를 발표하고 질문 세 개에 답하기</h4><p>찾을 결과: 3분 이내 발표 기록, 세 질문 이상의 실제 응답 기록, 미답 질문 계획.</p><ul><li><p>발표 시간은 실제 측정했고 질의응답과 분리했다.</p></li><li><p>최종 발표가 3분 이내다.</p></li><li><p>서로 다른 세 질문 이상과 실제 답을 기록했다.</p></li><li><p>답의 근거를 찾았거나 모르는 범위와 추가 확인을 정직하게 밝혔다.</p></li><li><p>시간 초과·보완이 있었다면 재발표 결과를 남겼다.</p></li></ul><h4>이런 결과는 통과시키지 않는다</h4><ul><li><p>“안전성을 높인다”는 설명만 있고 무엇을 바꿔 달라는지 없다.</p></li><li><p>질문 세 개를 미리 적어 두기만 하고 실제 응답을 하지 않았다.</p></li></ul><h4>평가 의견 작성 방법</h4><p>“근거 부족”이라고만 쓰지 말고 단계 번호, 충족하지 못한 조건, 제출물에서 실제로 빠지거나 잘못된 부분, 다시 할 일을 적는다. 예를 들어 STEP 04의 첫 조건은 “첫 문장에 검토자가 결정할 사항이 있다.”이다. 이 조건에 못 미쳤다면 어떤 줄·항목·참여자 기록이 빠졌는지 지목한다.</p><h4>다시 평가할 때</h4><p>틀리거나 빠진 내용만 구체적으로 알려 준다. 문서 작성 오류는 원자료와 수정본을 대조한다. 실제 수행 오류는 해당 행동을 다시 하게 하고 새 회차 기록으로 확인한다. 최초 결과는 지우지 않는다.</p>', 'aiGuidelines': {'allow': ['v1·v2 조항 대조표와 동일 7사례(B1~B5·A9·B9) 전후 결과의 5묶음(출처·대상·아군·민간·접수) 누락 분류 점검받기', '전원 일치율(일치 건수÷7)과 실측 시간·미측정 구분 집계 식 검산 도움받기', '결론 우선(BLUF) A4 1쪽 보고서의 첫 문장·한계·3대안(장단점·담당·자원) 구조 다듬기'], 'rules': ['AI 사용은 필수가 아니며, 수정 전 실제 문서(v1)와 수정본(v2)을 지우지 않고 briefing/ops07_commander_1page_brief.md에 함께 남긴다.', '전후 비교 시 동일 7사례의 입력·정보 시점을 고정하고, 측정하지 않은 시간은 미측정으로 표시하며 반복 학습 한계를 명시한다.', 'STEP 02 개인 답 작성과 STEP 05 3분 이내 발표·질의응답(3질문 이상) 및 초시계 측정은 학습자가 직접 수행한다.'], 'forbid': ['실제 지휘관 보고서·작전 평가 자료를 입력하기', '과거 v1 문서·수행 기록이나 미측정 시간 데이터를 사후에 꾸며 넣기', '실제 발표·질의응답 기록 없이 3분 발표 결과와 응답 근거를 AI로 생성하기']}}, {'id': 'OPS07_m10', 'sequence': 10, 'title': '근거와 책임이 연결된 보류 재확인 기준서 완성', 'durationHours': 3, 'overview': '<p>여덟 장과 네 부록에 실제 표·양식을 넣는다. 각 기준의 자료와 시험 결과를 찾아볼 수 있게 한다.</p><p>아래 다섯 단계에는 필요한 자료·작업 순서·예시·제출 내용과 평가 조건이 들어 있다. 각 단계의 자료를 읽은 뒤 순서대로 작성한다.</p>', 'objectives': '<h4>이 미션을 마치면 할 수 있어야 하는 일</h4><ul><li><p>여덟 장과 네 부록에 실제 표·양식을 넣는다.</p></li><li><p>각 기준의 자료와 시험 결과를 찾아볼 수 있게 한다.</p></li><li><p>다른 사용자의 시험과 개인 평가를 확인하고 개정 책임을 정한다.</p></li></ul>', 'scenario': '<h4>학습자의 역할</h4><p>학습자는 민간인과 아군에게 불필요한 피해가 생기지 않도록 보고를 읽고, 모르는 내용을 질문하고, 다음 담당자에게 기록을 넘기는 역할을 연습한다. 사람의 소속을 맞히거나 실제 타격을 결정하는 과제가 아니다.</p><h4>이번 미션의 과제</h4><ul><li><p>여덟 장과 네 부록에 실제 표·양식을 넣는다.</p></li><li><p>각 기준의 자료와 시험 결과를 찾아볼 수 있게 한다.</p></li><li><p>다른 사용자의 시험과 개인 평가를 확인하고 개정 책임을 정한다.</p></li></ul><h4>자료와 역할의 범위</h4><p>사례·시각·장소·역할은 모두 교육용이다. 실제 부대 자료를 쓰지 않는다. 자료에 답이 없으면 자료에 없다고 적는다. 실제 행동 권한은 이 수업이 정하지 않는다.</p><h4>개인 작업과 조별 작업</h4><p>개인 판단을 먼저 요청한 단계에서는 토의 전에 자신의 답을 남긴다. 그 뒤 조별 답과 수정 이유를 별도로 쓴다. 보고·인계·시험 단계에서는 실제 한 말과 행동을 기록한다.</p>', 'attachments': [], 'materials': ['/uploads/OPS07/OPS07_m10_student.ipynb', '/uploads/OPS07/OPS07_m10_workbook.ipynb'], 'materialText': '<h4>실습 노트북 안내 (선택 보조 도구)</h4><p><strong>학생 실습용</strong>: <code>OPS07_m10_student.ipynb</code> (직접 ANSWER 표와 note를 작성합니다.)<br><strong>교수자·해설용</strong>: <code>OPS07_m10_workbook.ipynb</code> (예시 풀이와 해설 note가 포함되어 있습니다.)<br><em>※ 본 과제는 문서 편집기나 인쇄 양식으로도 수행할 수 있으며, 노트북은 사례·질문·기록 양식을 표로 정리하는 선택 보조 도구입니다.</em></p><h4>자료를 읽는 순서</h4>\n\n<ol>\n<li><p>STEP 1에서는 아래 여덟 장 목차에 따라 앞 미션에서 실제로 작성한 표·절차·양식을 합칩니다. 부록의 종류와 포함할 내용은 STEP 1 설명에서 확인합니다.</p></li>\n<li><p>STEP 2~3에서는 각 기준 문장의 자료·시험 결과를 연결하고, 업무 담당자·보호·처음 읽는 신병의 세 관점에서 검토합니다.</p></li>\n<li><p>STEP 4에서는 B1~B5 중 고른 사례로 문서를 설명 없이 사용할 수 있는지 시험합니다. STEP 5의 최종 평가와 개정 계획에는 실제로 확인한 결과만 기록합니다.</p></li>\n</ol>\n\n<h4>자료 1. 통합할 기준서의 여덟 장 목차</h4>\n\n<ol>\n<li>\n<p>1장. 목적과 적용 범위</p>\n<p><strong>이 장에서 답할 질문:</strong> 이 기준서를 언제 쓰며 어떤 판단에는 쓰지 않는가?</p>\n<p><strong>참고할 미션:</strong> 미션1</p>\n</li>\n\n<li>\n<p>2장. 용어 설명</p>\n<p><strong>이 장에서 답할 질문:</strong> 보류·재확인·접수·회신을 어떻게 구분하는가?</p>\n<p><strong>참고할 미션:</strong> 미션1~미션4</p>\n</li>\n\n<li>\n<p>3장. 위험 신호</p>\n<p><strong>이 장에서 답할 질문:</strong> 민간인·아군 피해를 막기 위해 무엇을 살피는가?</p>\n<p><strong>참고할 미션:</strong> 미션1</p>\n</li>\n\n<li>\n<p>4장. 보류와 재확인 조건</p>\n<p><strong>이 장에서 답할 질문:</strong> 어떤 정보가 빠졌을 때 무엇을 다시 묻는가?</p>\n<p><strong>참고할 미션:</strong> 미션2·미션4</p>\n</li>\n\n<li>\n<p>5장. 담당자와 검토자</p>\n<p><strong>이 장에서 답할 질문:</strong> 누가 다시 확인하고 누가 보고를 검토하는가?</p>\n<p><strong>참고할 미션:</strong> 미션3·미션5</p>\n</li>\n\n<li>\n<p>6장. 중지 요청과 재확인 절차</p>\n<p><strong>이 장에서 답할 질문:</strong> 위험 보고부터 회신 확인·인계까지 어떻게 하는가?</p>\n<p><strong>참고할 미션:</strong> 미션4·미션5</p>\n</li>\n\n<li>\n<p>7장. 보고와 기록</p>\n<p><strong>이 장에서 답할 질문:</strong> 어떤 내용을 말하고 어디에 남기는가?</p>\n<p><strong>참고할 미션:</strong> 미션6</p>\n</li>\n\n<li>\n<p>8장. 훈련 평가와 개정</p>\n<p><strong>이 장에서 답할 질문:</strong> 어떻게 연습·평가하고 잘못된 기준을 고치는가?</p>\n<p><strong>참고할 미션:</strong> 미션7~미션9</p>\n</li>\n</ol>\n\n<h4>자료 2. 설명 없이 사용하기 시험에 쓸 사례</h4>\n\n<h4>B1</h4>\n\n<p>학교 옥상 인원 2명과 쌍안경으로 보이는 물체.</p>\n\n<p>대피소라는 주민 진술 1건이 있다.</p>\n\n<p>건물 내부 이용과 사람의 지위는 미확인.</p>\n\n<h4>B2</h4>\n\n<p>차량의 표식은 두 수단에서 일치한다고 보고되었다.</p>\n\n<p>정찰조는 예정된 위치 보고가 20분 늦었다.</p>\n\n<p>동일 원자료 재전달 여부와 현재 아군 위치는 미확인.</p>\n\n<h4>B3</h4>\n\n<p>두 수단에서 같은 표식 차량을 보고했다.</p>\n\n<p>옆 민간 차량 1대가 정차해 있다.</p>\n\n<p>곧 떠날 것 같다는 보고가 있지만 실제 이탈은 확인되지 않았다.</p>\n\n<h4>B4</h4>\n\n<p>관찰 보고에는 식별 명확·가상 위험 2등급이라고 적혀 있다.</p>\n\n<p>상급 검토자와 연락이 15분째 되지 않는다.</p>\n\n<p>등급 산정의 원근거와 검토 회신이 미제공이다.</p>\n\n<h4>B5</h4>\n\n<p>영상에 인원 1명이 흰 천을 흔드는 장면이 있다.</p>\n\n<p>항복 의사 등 보호 필요성을 확인해야 할 신호다.</p>\n\n<p>의도·신분·주변 상황은 확인되지 않았다.</p>\n\n<h4>자료를 사용할 때 지킬 점</h4>\n\n<p>기준서에 자료를 넣었다는 사실과 실제 보고·인계·사용 시험을 마쳤다는 사실은 다릅니다.</p>\n\n<p>수행하지 않은 일과 확인되지 않은 항목은 그대로 남겨 최종 평가자가 알 수 있게 합니다.</p>\n\n<p>모든 사례·시각·장소·역할은 교육용입니다. 필요한 자료와 구체적인 작업 순서는 각 단계 설명에도 들어 있습니다.</p>\n', 'guideSteps': [{'step': 1, 'problem': '여덟 장과 네 종류의 부록을 한 기준서로 합치기', 'activity': '<h4>이번 단계에서 하는 일</h4><p>미션별 문서가 흩어져 있으면 사용자가 어디를 읽을지 알기 어렵다. 같은 용어와 번호로 묶고 보고·재확인 절차를 찾기 쉽게 편집한다.</p><h4>먼저 읽을 자료</h4>\n\n<h4>기준서의 여덟 장 목차</h4>\n\n<ol>\n<li>\n<p>1장. 목적과 적용 범위</p>\n<p><strong>이 장에서 답할 질문:</strong> 이 기준서를 언제 쓰며 어떤 판단에는 쓰지 않는가</p>\n<p><strong>참고할 미션:</strong> M1</p>\n</li>\n\n<li>\n<p>2장. 용어 설명</p>\n<p><strong>이 장에서 답할 질문:</strong> 보류·재확인·접수·회신을 어떻게 구분하는가</p>\n<p><strong>참고할 미션:</strong> M1~M4</p>\n</li>\n\n<li>\n<p>3장. 위험 신호</p>\n<p><strong>이 장에서 답할 질문:</strong> 민간인·아군 피해를 막기 위해 무엇을 살피는가</p>\n<p><strong>참고할 미션:</strong> M1</p>\n</li>\n\n<li>\n<p>4장. 보류와 재확인 조건</p>\n<p><strong>이 장에서 답할 질문:</strong> 어떤 정보가 빠졌을 때 무엇을 다시 묻는가</p>\n<p><strong>참고할 미션:</strong> M2·M4</p>\n</li>\n\n<li>\n<p>5장. 담당자와 검토자</p>\n<p><strong>이 장에서 답할 질문:</strong> 누가 다시 확인하고 누가 보고를 검토하는가</p>\n<p><strong>참고할 미션:</strong> M3·M5</p>\n</li>\n\n<li>\n<p>6장. 중지 요청과 재확인 절차</p>\n<p><strong>이 장에서 답할 질문:</strong> 위험 보고부터 회신 확인·인계까지 어떻게 하는가</p>\n<p><strong>참고할 미션:</strong> M4·M5</p>\n</li>\n\n<li>\n<p>7장. 보고와 기록</p>\n<p><strong>이 장에서 답할 질문:</strong> 어떤 내용을 말하고 어디에 남기는가</p>\n<p><strong>참고할 미션:</strong> M6</p>\n</li>\n\n<li>\n<p>8장. 훈련 평가와 개정</p>\n<p><strong>이 장에서 답할 질문:</strong> 어떻게 연습·평가하고 잘못된 기준을 고치는가</p>\n<p><strong>참고할 미션:</strong> M7~M9</p>\n</li>\n</ol>\n\n<h4>부록에 넣을 자료</h4>\n\n<ul>\n<li>\n<p><strong>A 점검표:</strong> M6의 다섯 영역 점검 질문</p>\n</li>\n\n<li>\n<p><strong>B 기록 서식:</strong> M6의 짧은 기록·상세 기록</p>\n</li>\n\n<li>\n<p><strong>C 절차도:</strong> M4·M5의 보류·보고·재확인·인계 절차</p>\n</li>\n\n<li>\n<p><strong>D 수행 평가표:</strong> M7의 다섯 영역 네 수준과 개인 평가 기록 양식</p>\n</li>\n</ul>\n\n<p>1장과 6장에는 “실제 교전 판단은 적용 법규·교전수칙·지휘계통에 따른다.</p>\n\n<p>이 문서는 교육용 확인·보고 연습 자료이며 실제 타격 승인 기준이 아니다”라는 범위 설명을 넣는다.</p>\n\n<h4>작업 순서</h4><ol><li><p>최종 문서를 열고 여덟 장 제목과 네 부록 제목을 만든다.</p></li><li><p>해당 미션의 실제 표·절차·서식을 각 위치에 넣고 원문서 이름을 적는다.</p></li><li><p>같은 말을 다르게 쓴 곳을 찾는다. 보류·질문 송신·접수·회신·확인 완료를 서로 바꿔 쓰지 않는다.</p></li><li><p>중복 문장이 서로 다른 조건을 요구하면 어느 문장을 사용할지 정하고 이유를 남긴다.</p></li><li><p>1장과 6장에 교육용 범위 설명을 넣는다.</p></li><li><p>아직 작성하지 못한 절차나 검증 전 문장이 있으면 표시하고 담당·보완 계획을 남긴다.</p></li></ol><h4>작성 예시</h4><p>다음은 쓰는 방식을 보여 주는 예다. 자기 판단이나 실제 수행·측정 결과가 필요한 칸에 그대로 복사하지 않는다.</p><p>7장에는 보고 양식의 사용법을 설명하고 부록 B에 실제 빈 양식을 넣는다. “서식 참조”만 적지 말고 부록 B의 해당 양식 번호를 붙인다.</p><h4>작성할 표 또는 기록 항목</h4><ul><li><p>여덟 장 본문</p></li><li><p>부록 네 종류</p></li><li><p>가져온 원문서 위치</p></li><li><p>용어·조건 수정 기록</p></li><li><p>미완료 표시와 담당</p></li></ul><h4>제출할 내용과 끝내는 기준</h4><p>여덟 장과 네 부록이 들어간 기준서 초안 및 통합 과정의 수정 기록.</p><p>미션 제출 문서 manual/ops07_standard_operational_manual.md의 STEP 01 아래에 작성한다. 개인 기록과 조별 결과를 구분하고 실제 이름 대신 학습자 코드를 써도 된다. 표의 모양이 달라도 필요한 내용을 찾을 수 있으면 된다.</p><h4>마치기 전 확인</h4><ul><li><p>여덟 장 모두 실제 내용이 있고 제목만 있는 장이 없다.</p></li><li><p>부록 네 종류에서 실제 사용할 표·양식을 찾을 수 있다.</p></li><li><p>같은 용어·조건이 서로 모순되지 않게 정리했다.</p></li><li><p>1장·6장에 실제 규정 우선과 교육용 범위 설명이 있다.</p></li><li><p>미완료를 완료인 것처럼 숨기지 않았다.</p></li></ul><h4>배정 시간</h4><p>55분</p>', 'evalText': '<h4>평가할 자료와 위치</h4><p>manual/ops07_standard_operational_manual.md의 STEP 01 결과와 해당 개인 수행 기록을 확인한다. 이 단계에서 요청한 제출 내용은 다음과 같다: 여덟 장과 네 부록이 들어간 기준서 초안 및 통합 과정의 수정 기록.</p><h4>평가 순서</h4><ol><li><p>아래 통과 조건을 하나씩 읽고 제출물에서 확인한 위치를 옆에 적는다.</p></li><li><p>실제 설명·보고·인계가 필요한 단계는 수행 기록이나 직접 관찰로 확인한다. 서식만 있으면 수행한 것으로 보지 않는다.</p></li><li><p>통과 조건을 모두 충족하면 이 단계는 PASS다. 하나라도 빠지거나 반대로 적었으면 FAIL로 두고 빠진 항목을 구체적으로 적는다. 표현이나 표 모양이 예시와 달라도 뜻이 같고 근거를 찾을 수 있으면 인정한다.</p></li></ol><h4>PASS 조건</h4><ul><li><p>여덟 장 모두 실제 내용이 있고 제목만 있는 장이 없다.</p></li><li><p>부록 네 종류에서 실제 사용할 표·양식을 찾을 수 있다.</p></li><li><p>같은 용어·조건이 서로 모순되지 않게 정리했다.</p></li><li><p>1장·6장에 실제 규정 우선과 교육용 범위 설명이 있다.</p></li><li><p>미완료를 완료인 것처럼 숨기지 않았다.</p></li></ul><h4>FAIL에 해당하는 예</h4><ul><li><p>8장과 부록 제목만 만들고 자료를 넣지 않았다.</p></li><li><p>문서 완성이 실제 임무 재개 승인이라고 적었다.</p></li></ul><h4>이해 여부를 확인할 질문</h4><p>사용자가 질문의 접수 여부를 어디에 적는지 이 문서에서 찾아 보라.</p><h4>인정할 수 있는 답의 핵심</h4><p>관련 절차와 기록 양식의 실제 위치·번호를 가리킨다.</p><h4>보완과 재평가</h4><p>비어 있는 장·부록을 채우고 서로 다른 용어·조건은 한 방식으로 맞춘 뒤 수정 이유를 남긴다.</p><p>수정 전 결과와 재평가 결과를 구분해 남긴다. 실제 하지 않은 보고·채점·시험을 나중에 했던 것처럼 채우지 않는다.</p>', 'attachments': []}, {'step': 2, 'problem': '각 기준 문장에 자료와 시험 결과를 붙이기', 'activity': '<h4>이번 단계에서 하는 일</h4><p>규칙 옆에 자료 번호만 있어도 실제 내용이 그 규칙을 뒷받침하지 않을 수 있다. 읽는 사람이 원문과 시험 결과를 찾아 확인하도록 연결한다.</p><h4>먼저 읽을 자료</h4>\n\n<p>검토할 문장은 “~이면 보류한다”, “~를 기록한다”, “~담당자가 묻는다”처럼 누군가에게 행동을 요구하는 기준 문장이다.</p>\n\n<p>단순 설명 제목까지 모두 같은 형식으로 표시할 필요는 없다.</p>\n\n<h4>근거표에 표시할 항목</h4>\n\n<ul>\n<li>\n<p><strong>조항 번호:</strong> 예: 6-03처럼 문장을 찾을 번호</p>\n</li>\n\n<li>\n<p><strong>자료 번호·문장:</strong> 어떤 사례 때문에 필요한 기준인지</p>\n</li>\n\n<li>\n<p><strong>작성 미션:</strong> 처음 만든 미션과 문서</p>\n</li>\n\n<li>\n<p><strong>시험 기록:</strong> 어느 버전·회차·사례에서 사용해 봤는지</p>\n</li>\n\n<li>\n<p><strong>남은 문제:</strong> 아직 시험하지 않았거나 해결하지 못한 부분</p>\n</li>\n</ul>\n\n<h4>작업 순서</h4><ol><li><p>각 장에서 행동을 요구하는 기준 문장을 골라 조항 번호를 붙인다.</p></li><li><p>각 문장에 관련 자료 번호와 실제 관련 문장을 적는다.</p></li><li><p>작성한 미션과 문서 위치를 적는다.</p></li><li><p>시험했다면 버전·사례·회차와 결과를 적는다. 시험하지 않았다면 미시험이라고 적는다.</p></li><li><p>다른 조원은 자료를 실제 열어 기준 문장과 관련이 있는지 확인한다. 번호 모양이 맞는지만 보지 않는다.</p></li><li><p>전체 기준 문장 수와 연결을 점검한 문장 수를 세고 연결 누락이 없게 고친다.</p></li></ol><h4>작성 예시</h4><p>다음은 쓰는 방식을 보여 주는 예다. 자기 판단이나 실제 수행·측정 결과가 필요한 칸에 그대로 복사하지 않는다.</p><p>6-03 “질문을 보낼 때 담당자를 적는다” / 근거 “B 13:05 담당 미지정” / 작성 M8 / 시험 B9·v2 / 시험 결과 또는 미시험 여부를 실제 기록대로 적음.</p><h4>작성할 표 또는 기록 항목</h4><ul><li><p>조항 번호·기준 문장</p></li><li><p>자료 번호·관련 문장</p></li><li><p>작성 미션·문서</p></li><li><p>시험 버전·사례·회차·결과</p></li><li><p>미시험·남은 문제</p></li><li><p>검토자</p></li></ul><h4>제출할 내용과 끝내는 기준</h4><p>모든 행동 기준 문장의 근거표와 점검한 문장 수.</p><p>미션 제출 문서 manual/ops07_standard_operational_manual.md의 STEP 02 아래에 작성한다. 개인 기록과 조별 결과를 구분하고 실제 이름 대신 학습자 코드를 써도 된다. 표의 모양이 달라도 필요한 내용을 찾을 수 있으면 된다.</p><h4>마치기 전 확인</h4><ul><li><p>행동 기준 문장 전부를 세고 근거표에 포함했다.</p></li><li><p>자료 번호 옆에 실제 관련 문장이 있다.</p></li><li><p>해당 자료 내용이 기준의 필요성을 설명한다.</p></li><li><p>시험 결과의 위치를 찾을 수 있거나 미시험을 분명히 적었다.</p></li><li><p>존재하지 않는 자료·회차·시험 결과를 넣지 않았다.</p></li></ul><h4>배정 시간</h4><p>35분</p>', 'evalText': '<h4>평가할 자료와 위치</h4><p>manual/ops07_standard_operational_manual.md의 STEP 02 결과와 해당 개인 수행 기록을 확인한다. 이 단계에서 요청한 제출 내용은 다음과 같다: 모든 행동 기준 문장의 근거표와 점검한 문장 수.</p><h4>평가 순서</h4><ol><li><p>아래 통과 조건을 하나씩 읽고 제출물에서 확인한 위치를 옆에 적는다.</p></li><li><p>실제 설명·보고·인계가 필요한 단계는 수행 기록이나 직접 관찰로 확인한다. 서식만 있으면 수행한 것으로 보지 않는다.</p></li><li><p>통과 조건을 모두 충족하면 이 단계는 PASS다. 하나라도 빠지거나 반대로 적었으면 FAIL로 두고 빠진 항목을 구체적으로 적는다. 표현이나 표 모양이 예시와 달라도 뜻이 같고 근거를 찾을 수 있으면 인정한다.</p></li></ol><h4>PASS 조건</h4><ul><li><p>행동 기준 문장 전부를 세고 근거표에 포함했다.</p></li><li><p>자료 번호 옆에 실제 관련 문장이 있다.</p></li><li><p>해당 자료 내용이 기준의 필요성을 설명한다.</p></li><li><p>시험 결과의 위치를 찾을 수 있거나 미시험을 분명히 적었다.</p></li><li><p>존재하지 않는 자료·회차·시험 결과를 넣지 않았다.</p></li></ul><h4>FAIL에 해당하는 예</h4><ul><li><p>모든 조항의 근거를 “M1~M9”로만 썼다.</p></li><li><p>자료 번호는 있지만 열어 보면 전혀 다른 내용이다.</p></li></ul><h4>이해 여부를 확인할 질문</h4><p>이 기준은 어떤 자료의 어떤 문제를 막기 위해 생겼는가?</p><h4>인정할 수 있는 답의 핵심</h4><p>정확한 자료 문장과 기준 문장을 찾아 관계를 설명한다.</p><h4>보완과 재평가</h4><p>관련 없는 근거를 실제 관련 자료로 고치고 시험 결과가 없으면 미시험으로 바로잡는다.</p><p>수정 전 결과와 재평가 결과를 구분해 남긴다. 실제 하지 않은 보고·채점·시험을 나중에 했던 것처럼 채우지 않는다.</p>', 'attachments': []}, {'step': 3, 'problem': '업무·보호·신병 관점에서 기준서를 읽고 의견을 처리하기', 'activity': '<h4>이번 단계에서 하는 일</h4><p>작성자는 당연하게 아는 내용도 처음 읽는 사람에게는 빠진 설명일 수 있다. 세 관점에서 실제 문장을 읽고 고칠 점을 찾는다.</p><h4>먼저 읽을 자료</h4>\n\n<h4>관점별 문서 검토 질문</h4>\n\n<ul>\n<li>\n<p><strong>업무 담당자 관점:</strong> 누가 무엇을 묻고 접수·회신을 어디에 적는지 알 수 있는가</p>\n</li>\n\n<li>\n<p><strong>민간인·아군 보호 관점:</strong> 미확인이나 보호 신호가 확정 표현에 가려지지 않는가</p>\n</li>\n\n<li>\n<p><strong>처음 읽는 신병 관점:</strong> 용어를 몰라도 자료·순서·작성 결과를 찾을 수 있는가</p>\n</li>\n</ul>\n\n<p>실제 법무·작전 승인 심의가 아니라 교실의 문서 검토 역할이다.</p>\n\n<p>모든 의견을 받아들일 필요는 없지만 어떻게 처리했는지는 남긴다.</p>\n\n<h4>작업 순서</h4><ol><li><p>세 검토 역할을 나눠 맡는다. 조원이 적으면 역할을 바꿔 읽어도 된다.</p></li><li><p>역할별 질문을 보며 기준서를 실제 읽고 고칠 의견을 한 건 이상 적는다.</p></li><li><p>의견에는 장·조항 번호와 이해하기 어려운 실제 문장을 넣는다.</p></li><li><p>작성자는 반영·반영 안 함·추가 확인 중 하나로 답하고 이유를 쓴다.</p></li><li><p>반영한 것은 새 문장과 위치를, 추가 확인은 담당·다음 확인 계획을 적는다.</p></li><li><p>모든 의견에 답이 있는지 세고 미처리 의견을 남기지 않는다. “추가 확인”은 처리 방향을 정한 것이지 해결 완료가 아니다.</p></li></ol><h4>작성 예시</h4><p>다음은 쓰는 방식을 보여 주는 예다. 자기 판단이나 실제 수행·측정 결과가 필요한 칸에 그대로 복사하지 않는다.</p><p>신병 관점 의견 “6-03의 ‘회신 대조’가 무슨 뜻인지 모름” / 반영 “질문마다 답이 왔는지 표시하고 아직 답이 없는 질문은 따로 적는다”로 풀어 씀.</p><h4>작성할 표 또는 기록 항목</h4><ul><li><p>의견 번호·역할</p></li><li><p>장·조항·기존 문장</p></li><li><p>의견과 질문</p></li><li><p>반영·미반영·추가 확인</p></li><li><p>이유</p></li><li><p>새 문장·위치 또는 담당·계획</p></li></ul><h4>제출할 내용과 끝내는 기준</h4><p>세 역할 각각 한 건 이상, 모두 세 건 이상의 문서 검토 의견과 처리 기록.</p><p>미션 제출 문서 manual/ops07_standard_operational_manual.md의 STEP 03 아래에 작성한다. 개인 기록과 조별 결과를 구분하고 실제 이름 대신 학습자 코드를 써도 된다. 표의 모양이 달라도 필요한 내용을 찾을 수 있으면 된다.</p><h4>마치기 전 확인</h4><ul><li><p>세 역할 모두 실제 문장에 대해 의견을 냈다.</p></li><li><p>각 의견에 위치와 구체적인 질문이 있다.</p></li><li><p>모든 의견에 처리 결과와 이유가 있다.</p></li><li><p>반영 위치 또는 미확인 담당·계획이 있어 다음 할 일을 알 수 있다.</p></li></ul><h4>배정 시간</h4><p>30분</p>', 'evalText': '<h4>평가할 자료와 위치</h4><p>manual/ops07_standard_operational_manual.md의 STEP 03 결과와 해당 개인 수행 기록을 확인한다. 이 단계에서 요청한 제출 내용은 다음과 같다: 세 역할 각각 한 건 이상, 모두 세 건 이상의 문서 검토 의견과 처리 기록.</p><h4>평가 순서</h4><ol><li><p>아래 통과 조건을 하나씩 읽고 제출물에서 확인한 위치를 옆에 적는다.</p></li><li><p>실제 설명·보고·인계가 필요한 단계는 수행 기록이나 직접 관찰로 확인한다. 서식만 있으면 수행한 것으로 보지 않는다.</p></li><li><p>통과 조건을 모두 충족하면 이 단계는 PASS다. 하나라도 빠지거나 반대로 적었으면 FAIL로 두고 빠진 항목을 구체적으로 적는다. 표현이나 표 모양이 예시와 달라도 뜻이 같고 근거를 찾을 수 있으면 인정한다.</p></li></ol><h4>PASS 조건</h4><ul><li><p>세 역할 모두 실제 문장에 대해 의견을 냈다.</p></li><li><p>각 의견에 위치와 구체적인 질문이 있다.</p></li><li><p>모든 의견에 처리 결과와 이유가 있다.</p></li><li><p>반영 위치 또는 미확인 담당·계획이 있어 다음 할 일을 알 수 있다.</p></li></ul><h4>FAIL에 해당하는 예</h4><ul><li><p>각 역할이 “적절함”만 적고 실제 읽은 문장과 질문이 없다.</p></li><li><p>추가 확인 의견을 해결 완료로 표시했다.</p></li></ul><h4>이해 여부를 확인할 질문</h4><p>반영하지 않은 의견은 어떤 기록이 있어야 하는가?</p><h4>인정할 수 있는 답의 핵심</h4><p>해당 의견·기존 문장·받아들이지 않은 이유가 있어야 한다고 답한다.</p><h4>보완과 재평가</h4><p>실제 문장을 다시 읽고 역할별 의견을 받는다. 처리하지 않은 의견에는 결정 이유나 확인 계획을 붙인다.</p><p>수정 전 결과와 재평가 결과를 구분해 남긴다. 실제 하지 않은 보고·채점·시험을 나중에 했던 것처럼 채우지 않는다.</p>', 'attachments': []}, {'step': 4, 'problem': '작성자의 설명 없이 다른 사람이 기준서를 사용하게 하기', 'activity': '<h4>이번 단계에서 하는 일</h4><p>설명해 줘야만 사용할 수 있는 문서는 신병용 안내로 부족하다. 다른 사람이 문서만 읽고 필요한 질문과 기록 위치를 찾는지 시험한다.</p><h4>먼저 읽을 자료</h4>\n\n<h4>사례 자료</h4>\n\n<p><strong>B1</strong></p>\n\n<p>학교 옥상 인원 2명과 쌍안경으로 보이는 물체.</p>\n\n<p>대피소라는 주민 진술 1건이 있다.</p>\n\n<p>건물 내부 이용과 사람의 지위는 미확인.</p>\n\n<p><strong>B2</strong></p>\n\n<p>차량의 표식은 두 수단에서 일치한다고 보고되었다.</p>\n\n<p>정찰조는 예정된 위치 보고가 20분 늦었다.</p>\n\n<p>동일 원자료 재전달 여부와 현재 아군 위치는 미확인.</p>\n\n<p><strong>B3</strong></p>\n\n<p>두 수단에서 같은 표식 차량을 보고했다.</p>\n\n<p>옆 민간 차량 1대가 정차해 있다.</p>\n\n<p>곧 떠날 것 같다는 보고가 있지만 실제 이탈은 확인되지 않았다.</p>\n\n<p><strong>B4</strong></p>\n\n<p>관찰 보고에는 식별 명확·가상 위험 2등급이라고 적혀 있다.</p>\n\n<p>상급 검토자와 연락이 15분째 되지 않는다.</p>\n\n<p>등급 산정의 원근거와 검토 회신이 미제공이다.</p>\n\n<p><strong>B5</strong></p>\n\n<p>영상에 인원 1명이 흰 천을 흔드는 장면이 있다.</p>\n\n<p>항복 의사 등 보호 필요성을 확인해야 할 신호다.</p>\n\n<p>의도·신분·주변 상황은 확인되지 않았다.</p>\n\n<p>다른 조원 또는 교관이 B1~B5 중 한 사례를 고른다.</p>\n\n<p>작성자는 문서를 건넨 뒤 설명하지 않는다.</p>\n\n<p>시험에 참여한 사람이 문서에서 찾은 위치와 어려웠던 문장을 기록한다.</p>\n\n<h4>작업 순서</h4><ol><li><p>작성에 직접 참여하지 않은 사람에게 최종 기준서와 고른 카드 한 개를 준다. 가능하면 다른 조가 맡는다.</p></li><li><p>사용자는 보류 이유·확인 질문·담당·기록 양식 위치를 문서에서 찾는다.</p></li><li><p>찾은 내용으로 한 번 보고하고 다른 사람에게 다음 질문을 인계한다.</p></li><li><p>관찰자는 실제 찾은 장·조항·서식 번호와 막힌 부분을 적는다. 찾지 못한 내용은 실패로 남긴다.</p></li><li><p>전원도 각각 한 항목의 보고 또는 인계를 수행하고 자기 기록을 남긴다.</p></li><li><p>문장을 고친 뒤 다른 사용자가 다시 찾아보게 한다. 말로만 보충해 성공 처리하지 않는다.</p></li></ol><h4>작성 예시</h4><p>다음은 쓰는 방식을 보여 주는 예다. 자기 판단이나 실제 수행·측정 결과가 필요한 칸에 그대로 복사하지 않는다.</p><p>사용자가 “접수 여부를 어디에 쓰는지 못 찾겠다”고 하면 그 실패를 적는다. 작성자가 말로 위치를 알려 준 뒤 성공했다고 처리하지 말고 문서의 안내를 고친 뒤 다시 시험한다.</p><h4>작성할 표 또는 기록 항목</h4><ul><li><p>사용자·작성자·선택 카드</p></li><li><p>찾을 네 내용</p></li><li><p>실제 찾은 장·조항·양식</p></li><li><p>막힌 표현</p></li><li><p>실제 보고·인계</p></li><li><p>수정 전후·재시험 결과</p></li><li><p>전원 참여 기록</p></li></ul><h4>제출할 내용과 끝내는 기준</h4><p>설명 없이 사용한 시험 기록, 문서 수정·재시험 결과, 전원 개인 보고 또는 인계 기록.</p><p>미션 제출 문서 manual/ops07_standard_operational_manual.md의 STEP 04 아래에 작성한다. 개인 기록과 조별 결과를 구분하고 실제 이름 대신 학습자 코드를 써도 된다. 표의 모양이 달라도 필요한 내용을 찾을 수 있으면 된다.</p><h4>마치기 전 확인</h4><ul><li><p>작성자 설명 없이 사용한 사람과 사례가 기록돼 있다.</p></li><li><p>보류 이유·질문·담당·기록 위치 네 항목을 실제 찾았다.</p></li><li><p>막힌 곳은 수정 전 실패와 수정 후 재시험 결과를 남겼다.</p></li><li><p>전원의 보고 또는 인계 행동을 확인할 수 있다.</p></li></ul><h4>배정 시간</h4><p>35분</p>', 'evalText': '<h4>평가할 자료와 위치</h4><p>manual/ops07_standard_operational_manual.md의 STEP 04 결과와 해당 개인 수행 기록을 확인한다. 이 단계에서 요청한 제출 내용은 다음과 같다: 설명 없이 사용한 시험 기록, 문서 수정·재시험 결과, 전원 개인 보고 또는 인계 기록.</p><h4>평가 순서</h4><ol><li><p>아래 통과 조건을 하나씩 읽고 제출물에서 확인한 위치를 옆에 적는다.</p></li><li><p>실제 설명·보고·인계가 필요한 단계는 수행 기록이나 직접 관찰로 확인한다. 서식만 있으면 수행한 것으로 보지 않는다.</p></li><li><p>통과 조건을 모두 충족하면 이 단계는 PASS다. 하나라도 빠지거나 반대로 적었으면 FAIL로 두고 빠진 항목을 구체적으로 적는다. 표현이나 표 모양이 예시와 달라도 뜻이 같고 근거를 찾을 수 있으면 인정한다.</p></li></ol><h4>PASS 조건</h4><ul><li><p>작성자 설명 없이 사용한 사람과 사례가 기록돼 있다.</p></li><li><p>보류 이유·질문·담당·기록 위치 네 항목을 실제 찾았다.</p></li><li><p>막힌 곳은 수정 전 실패와 수정 후 재시험 결과를 남겼다.</p></li><li><p>전원의 보고 또는 인계 행동을 확인할 수 있다.</p></li></ul><h4>FAIL에 해당하는 예</h4><ul><li><p>작성자가 옆에서 전부 설명한 뒤 문서만으로 성공했다고 적었다.</p></li><li><p>찾은 항목의 위치와 실제 보고 내용 없이 “사용 가능”이라고만 썼다.</p></li></ul><h4>이해 여부를 확인할 질문</h4><p>지금 문서만 보고 이 카드에서 누구에게 무엇을 물어야 하는지 찾아 보라.</p><h4>인정할 수 있는 답의 핵심</h4><p>고른 카드와 관련된 실제 질문·담당·기록 위치를 문서에서 직접 찾는다.</p><h4>보완과 재평가</h4><p>찾지 못한 표현과 안내를 고치고 설명을 듣지 않은 사람이 다시 사용하게 한다.</p><p>수정 전 결과와 재평가 결과를 구분해 남긴다. 실제 하지 않은 보고·채점·시험을 나중에 했던 것처럼 채우지 않는다.</p>', 'attachments': []}, {'step': 5, 'problem': '필수 평가와 개인 수행을 확인하고 개정 계획을 정하기', 'activity': '<h4>이번 단계에서 하는 일</h4><p>문서를 제출했다고 모든 사람이 통과한 것은 아니다. 평가표의 미달·미수행을 확인하고 다음에 문서를 고칠 책임도 정한다.</p><h4>먼저 읽을 자료</h4>\n\n<p>필수 평가는 미션마다 두 개, 전체 스무 개다.</p>\n\n<p>개인은 M7의 다섯 영역 모두 우수 3점 이상이어야 하며 다른 미션의 개인 설명·보고·인계 참여 기록도 확인한다.</p>\n\n<p>미달을 평균이나 조별 성적으로 없애지 않는다.</p>\n\n<h4>개정 계획에 넣을 것</h4>\n\n<ul>\n<li>\n<p>정기 검토 연 1회</p>\n</li>\n\n<li>\n<p>오류·훈련 실패·수업 규칙 변경이 발견되면 수시 검토</p>\n</li>\n\n<li>\n<p>의견을 받을 방법과 기록할 위치</p>\n</li>\n\n<li>\n<p>교육용 개정 담당과 검토 담당</p>\n</li>\n\n<li>\n<p>문서 버전·바꾼 문장·이유·시험 결과</p>\n</li>\n\n<li>\n<p>아직 해결하지 못한 질문과 담당·다음 계획</p>\n</li>\n</ul>\n\n<h4>작업 순서</h4><ol><li><p>교관은 미션별 두 평가 질문에 PASS·FAIL·미평가 상태와 근거 위치를 적는다. 스무 항목을 모두 확인한다.</p></li><li><p>학습자 명단과 M7 다섯 영역 최종 점수·새 회차 기록을 대조한다. 개인 수행이 없는 사람은 미확인으로 둔다.</p></li><li><p>다른 미션의 개인 설명·보고·인계 참여 누락도 확인한다.</p></li><li><p>미달·미수행에는 무엇을 더 해야 하는지와 담당·다음 실시 계획을 적는다. 모두 충족하지 않은 개인에게 최종 PASS를 주지 않는다.</p></li><li><p>표의 개정 계획을 문서 8장에 쓰고 실제로 쓸 의견 접수·개정 이력 양식을 붙인다.</p></li><li><p>최종 문서 버전과 아직 남은 질문을 다음 담당자에게 넘기고 받았다는 확인을 남긴다.</p></li></ol><h4>작성 예시</h4><p>다음은 쓰는 방식을 보여 주는 예다. 자기 판단이나 실제 수행·측정 결과가 필요한 칸에 그대로 복사하지 않는다.</p><p>개인 P03의 보고 영역이 2점이면 “조별 문서 PASS, P03 개인 보고 재훈련·재평가 필요”로 나눠 적는다. 평균 점수로 개인 PASS를 주지 않는다.</p><h4>작성할 표 또는 기록 항목</h4><ul><li><p>스무 평가 질문의 판정·근거 위치</p></li><li><p>개인별 다섯 영역·참여 확인</p></li><li><p>미달·미수행 보완 계획</p></li><li><p>정기·수시 개정 계획</p></li><li><p>의견 접수·개정 이력 양식</p></li><li><p>최종 버전·담당 인계</p></li></ul><h4>제출할 내용과 끝내는 기준</h4><p>교관 최종 판정표, 개인별 성취·보완표, 개정 계획·양식, 최종 인계 확인.</p><p>미션 제출 문서 manual/ops07_standard_operational_manual.md의 STEP 05 아래에 작성한다. 개인 기록과 조별 결과를 구분하고 실제 이름 대신 학습자 코드를 써도 된다. 표의 모양이 달라도 필요한 내용을 찾을 수 있으면 된다.</p><h4>마치기 전 확인</h4><ul><li><p>스무 평가 질문 모두에 판정 또는 미평가와 근거·보완 계획이 있다.</p></li><li><p>전원의 개인 수행과 M7 영역별 결과를 확인했다.</p></li><li><p>미달·미수행을 최종 개인 PASS로 처리하지 않았다.</p></li><li><p>정기·수시 검토, 의견 접수, 담당, 이력 기록 위치를 정했다.</p></li><li><p>교육 성취 판정과 실제 교전·타격 승인을 혼동하지 않는다.</p></li></ul><h4>배정 시간</h4><p>25분</p>', 'evalText': '<h4>평가할 자료와 위치</h4><p>manual/ops07_standard_operational_manual.md의 STEP 05 결과와 해당 개인 수행 기록을 확인한다. 이 단계에서 요청한 제출 내용은 다음과 같다: 교관 최종 판정표, 개인별 성취·보완표, 개정 계획·양식, 최종 인계 확인.</p><h4>평가 순서</h4><ol><li><p>아래 통과 조건을 하나씩 읽고 제출물에서 확인한 위치를 옆에 적는다.</p></li><li><p>실제 설명·보고·인계가 필요한 단계는 수행 기록이나 직접 관찰로 확인한다. 서식만 있으면 수행한 것으로 보지 않는다.</p></li><li><p>통과 조건을 모두 충족하면 이 단계는 PASS다. 하나라도 빠지거나 반대로 적었으면 FAIL로 두고 빠진 항목을 구체적으로 적는다. 표현이나 표 모양이 예시와 달라도 뜻이 같고 근거를 찾을 수 있으면 인정한다.</p></li></ol><h4>PASS 조건</h4><ul><li><p>스무 평가 질문 모두에 판정 또는 미평가와 근거·보완 계획이 있다.</p></li><li><p>전원의 개인 수행과 M7 영역별 결과를 확인했다.</p></li><li><p>미달·미수행을 최종 개인 PASS로 처리하지 않았다.</p></li><li><p>정기·수시 검토, 의견 접수, 담당, 이력 기록 위치를 정했다.</p></li><li><p>교육 성취 판정과 실제 교전·타격 승인을 혼동하지 않는다.</p></li></ul><h4>FAIL에 해당하는 예</h4><ul><li><p>미실시 항목을 모두 PASS로 미리 채웠다.</p></li><li><p>“연 1회 개정”만 적고 누가 어디에서 의견과 변경을 기록할지 없다.</p></li></ul><h4>이해 여부를 확인할 질문</h4><p>조별 문서는 완성됐지만 개인이 한 영역에 미달하면 최종 표에 어떻게 적는가?</p><h4>인정할 수 있는 답의 핵심</h4><p>문서 판정과 개인 미달을 나눠 적고 필요한 행동 연습·재평가 계획을 남긴다고 답한다.</p><h4>보완과 재평가</h4><p>누락·미달을 실제 상태로 복구하고 보완할 행동과 회차를 정한다. 실제 평가 전에는 PASS로 바꾸지 않는다.</p><p>수정 전 결과와 재평가 결과를 구분해 남긴다. 실제 하지 않은 보고·채점·시험을 나중에 했던 것처럼 채우지 않는다.</p>', 'attachments': []}], 'deliverables': ['manual/ops07_standard_operational_manual.md (필수) — 여덟 장과 네 부록의 기준서, 조항 근거표, 세 역할 의견, 설명 없이 사용한 시험, 개인·최종 평가와 개정 계획'], 'techStack': ['문서 편집기(Markdown · manual/ops07_standard_operational_manual.md) 또는 인쇄 양식', '8장·4부록 통합 기준서 및 조항-자료-시험 근거 매핑표', '3관점(업무·보호·신병) 교차 검토 및 의견 처리 이력표', '설명 없는 독립 사용 시험지(B1~B5) 및 20개 필수 평가·개정 계획표'], 'evaluationKeys': ['<h4>평가 질문</h4><p>여덟 장과 부록을 통합하고 근거·의견·사용 시험을 확인했는가</p><h4>평가자가 확인할 위치</h4><p>manual/ops07_standard_operational_manual.md의 STEP 01, STEP 02, STEP 03, STEP 04과 그 단계의 개인 기록을 확인한다.</p><h4>PASS를 주는 방법</h4><p>아래 단계별 항목을 모두 충족해야 이 평가 질문에 PASS를 준다. 항목 옆에 확인한 표·줄·회차를 적는다. 한 항목이라도 없거나 틀리면 FAIL 사유에 해당 단계와 문장을 적는다. 학습자가 쓴 말이 예시와 똑같을 필요는 없지만 확인할 내용은 모두 있어야 한다.</p><h4>STEP 01 여덟 장과 네 종류의 부록을 한 기준서로 합치기</h4><p>찾을 결과: 여덟 장과 네 부록이 들어간 기준서 초안 및 통합 과정의 수정 기록.</p><ul><li><p>여덟 장 모두 실제 내용이 있고 제목만 있는 장이 없다.</p></li><li><p>부록 네 종류에서 실제 사용할 표·양식을 찾을 수 있다.</p></li><li><p>같은 용어·조건이 서로 모순되지 않게 정리했다.</p></li><li><p>1장·6장에 실제 규정 우선과 교육용 범위 설명이 있다.</p></li><li><p>미완료를 완료인 것처럼 숨기지 않았다.</p></li></ul><h4>STEP 02 각 기준 문장에 자료와 시험 결과를 붙이기</h4><p>찾을 결과: 모든 행동 기준 문장의 근거표와 점검한 문장 수.</p><ul><li><p>행동 기준 문장 전부를 세고 근거표에 포함했다.</p></li><li><p>자료 번호 옆에 실제 관련 문장이 있다.</p></li><li><p>해당 자료 내용이 기준의 필요성을 설명한다.</p></li><li><p>시험 결과의 위치를 찾을 수 있거나 미시험을 분명히 적었다.</p></li><li><p>존재하지 않는 자료·회차·시험 결과를 넣지 않았다.</p></li></ul><h4>STEP 03 업무·보호·신병 관점에서 기준서를 읽고 의견을 처리하기</h4><p>찾을 결과: 세 역할 각각 한 건 이상, 모두 세 건 이상의 문서 검토 의견과 처리 기록.</p><ul><li><p>세 역할 모두 실제 문장에 대해 의견을 냈다.</p></li><li><p>각 의견에 위치와 구체적인 질문이 있다.</p></li><li><p>모든 의견에 처리 결과와 이유가 있다.</p></li><li><p>반영 위치 또는 미확인 담당·계획이 있어 다음 할 일을 알 수 있다.</p></li></ul><h4>STEP 04 작성자의 설명 없이 다른 사람이 기준서를 사용하게 하기</h4><p>찾을 결과: 설명 없이 사용한 시험 기록, 문서 수정·재시험 결과, 전원 개인 보고 또는 인계 기록.</p><ul><li><p>작성자 설명 없이 사용한 사람과 사례가 기록돼 있다.</p></li><li><p>보류 이유·질문·담당·기록 위치 네 항목을 실제 찾았다.</p></li><li><p>막힌 곳은 수정 전 실패와 수정 후 재시험 결과를 남겼다.</p></li><li><p>전원의 보고 또는 인계 행동을 확인할 수 있다.</p></li></ul><h4>이런 결과는 통과시키지 않는다</h4><ul><li><p>8장과 부록 제목만 만들고 자료를 넣지 않았다.</p></li><li><p>모든 조항의 근거를 “M1~M9”로만 썼다.</p></li><li><p>각 역할이 “적절함”만 적고 실제 읽은 문장과 질문이 없다.</p></li><li><p>작성자가 옆에서 전부 설명한 뒤 문서만으로 성공했다고 적었다.</p></li></ul><h4>평가 의견 작성 방법</h4><p>“근거 부족”이라고만 쓰지 말고 단계 번호, 충족하지 못한 조건, 제출물에서 실제로 빠지거나 잘못된 부분, 다시 할 일을 적는다. 예를 들어 STEP 01의 첫 조건은 “여덟 장 모두 실제 내용이 있고 제목만 있는 장이 없다.”이다. 이 조건에 못 미쳤다면 어떤 줄·항목·참여자 기록이 빠졌는지 지목한다.</p><h4>다시 평가할 때</h4><p>틀리거나 빠진 내용만 구체적으로 알려 준다. 문서 작성 오류는 원자료와 수정본을 대조한다. 실제 수행 오류는 해당 행동을 다시 하게 하고 새 회차 기록으로 확인한다. 최초 결과는 지우지 않는다.</p>', '<h4>평가 질문</h4><p>스무 필수 평가·개인 미달·개정 책임을 실제 상태대로 정리했는가</p><h4>평가자가 확인할 위치</h4><p>manual/ops07_standard_operational_manual.md의 STEP 05과 그 단계의 개인 기록을 확인한다.</p><h4>PASS를 주는 방법</h4><p>아래 단계별 항목을 모두 충족해야 이 평가 질문에 PASS를 준다. 항목 옆에 확인한 표·줄·회차를 적는다. 한 항목이라도 없거나 틀리면 FAIL 사유에 해당 단계와 문장을 적는다. 학습자가 쓴 말이 예시와 똑같을 필요는 없지만 확인할 내용은 모두 있어야 한다.</p><h4>STEP 05 필수 평가와 개인 수행을 확인하고 개정 계획을 정하기</h4><p>찾을 결과: 교관 최종 판정표, 개인별 성취·보완표, 개정 계획·양식, 최종 인계 확인.</p><ul><li><p>스무 평가 질문 모두에 판정 또는 미평가와 근거·보완 계획이 있다.</p></li><li><p>전원의 개인 수행과 M7 영역별 결과를 확인했다.</p></li><li><p>미달·미수행을 최종 개인 PASS로 처리하지 않았다.</p></li><li><p>정기·수시 검토, 의견 접수, 담당, 이력 기록 위치를 정했다.</p></li><li><p>교육 성취 판정과 실제 교전·타격 승인을 혼동하지 않는다.</p></li></ul><h4>이런 결과는 통과시키지 않는다</h4><ul><li><p>미실시 항목을 모두 PASS로 미리 채웠다.</p></li></ul><h4>평가 의견 작성 방법</h4><p>“근거 부족”이라고만 쓰지 말고 단계 번호, 충족하지 못한 조건, 제출물에서 실제로 빠지거나 잘못된 부분, 다시 할 일을 적는다. 예를 들어 STEP 05의 첫 조건은 “스무 평가 질문 모두에 판정 또는 미평가와 근거·보완 계획이 있다.”이다. 이 조건에 못 미쳤다면 어떤 줄·항목·참여자 기록이 빠졌는지 지목한다.</p><h4>다시 평가할 때</h4><p>틀리거나 빠진 내용만 구체적으로 알려 준다. 문서 작성 오류는 원자료와 수정본을 대조한다. 실제 수행 오류는 해당 행동을 다시 하게 하고 새 회차 기록으로 확인한다. 최초 결과는 지우지 않는다.</p>'], 'evaluationTypes': ['REQUIRED', 'REQUIRED'], 'evaluationText': '<h4>미션 전체 판정</h4><p>이 미션은 아래 두 평가 질문이 모두 PASS여야 통과다. 조별 문서와 개인의 실제 수행을 나눠 확인한다. 수행하지 않았거나 기록이 없어 확인할 수 없는 사람은 미평가로 남기고 다시 실시한다. 최종 판정에서는 미평가를 PASS로 바꾸지 않는다.</p><h4>평가 질문</h4><p>여덟 장과 부록을 통합하고 근거·의견·사용 시험을 확인했는가</p><h4>평가자가 확인할 위치</h4><p>manual/ops07_standard_operational_manual.md의 STEP 01, STEP 02, STEP 03, STEP 04과 그 단계의 개인 기록을 확인한다.</p><h4>PASS를 주는 방법</h4><p>아래 단계별 항목을 모두 충족해야 이 평가 질문에 PASS를 준다. 항목 옆에 확인한 표·줄·회차를 적는다. 한 항목이라도 없거나 틀리면 FAIL 사유에 해당 단계와 문장을 적는다. 학습자가 쓴 말이 예시와 똑같을 필요는 없지만 확인할 내용은 모두 있어야 한다.</p><h4>STEP 01 여덟 장과 네 종류의 부록을 한 기준서로 합치기</h4><p>찾을 결과: 여덟 장과 네 부록이 들어간 기준서 초안 및 통합 과정의 수정 기록.</p><ul><li><p>여덟 장 모두 실제 내용이 있고 제목만 있는 장이 없다.</p></li><li><p>부록 네 종류에서 실제 사용할 표·양식을 찾을 수 있다.</p></li><li><p>같은 용어·조건이 서로 모순되지 않게 정리했다.</p></li><li><p>1장·6장에 실제 규정 우선과 교육용 범위 설명이 있다.</p></li><li><p>미완료를 완료인 것처럼 숨기지 않았다.</p></li></ul><h4>STEP 02 각 기준 문장에 자료와 시험 결과를 붙이기</h4><p>찾을 결과: 모든 행동 기준 문장의 근거표와 점검한 문장 수.</p><ul><li><p>행동 기준 문장 전부를 세고 근거표에 포함했다.</p></li><li><p>자료 번호 옆에 실제 관련 문장이 있다.</p></li><li><p>해당 자료 내용이 기준의 필요성을 설명한다.</p></li><li><p>시험 결과의 위치를 찾을 수 있거나 미시험을 분명히 적었다.</p></li><li><p>존재하지 않는 자료·회차·시험 결과를 넣지 않았다.</p></li></ul><h4>STEP 03 업무·보호·신병 관점에서 기준서를 읽고 의견을 처리하기</h4><p>찾을 결과: 세 역할 각각 한 건 이상, 모두 세 건 이상의 문서 검토 의견과 처리 기록.</p><ul><li><p>세 역할 모두 실제 문장에 대해 의견을 냈다.</p></li><li><p>각 의견에 위치와 구체적인 질문이 있다.</p></li><li><p>모든 의견에 처리 결과와 이유가 있다.</p></li><li><p>반영 위치 또는 미확인 담당·계획이 있어 다음 할 일을 알 수 있다.</p></li></ul><h4>STEP 04 작성자의 설명 없이 다른 사람이 기준서를 사용하게 하기</h4><p>찾을 결과: 설명 없이 사용한 시험 기록, 문서 수정·재시험 결과, 전원 개인 보고 또는 인계 기록.</p><ul><li><p>작성자 설명 없이 사용한 사람과 사례가 기록돼 있다.</p></li><li><p>보류 이유·질문·담당·기록 위치 네 항목을 실제 찾았다.</p></li><li><p>막힌 곳은 수정 전 실패와 수정 후 재시험 결과를 남겼다.</p></li><li><p>전원의 보고 또는 인계 행동을 확인할 수 있다.</p></li></ul><h4>이런 결과는 통과시키지 않는다</h4><ul><li><p>8장과 부록 제목만 만들고 자료를 넣지 않았다.</p></li><li><p>모든 조항의 근거를 “M1~M9”로만 썼다.</p></li><li><p>각 역할이 “적절함”만 적고 실제 읽은 문장과 질문이 없다.</p></li><li><p>작성자가 옆에서 전부 설명한 뒤 문서만으로 성공했다고 적었다.</p></li></ul><h4>평가 의견 작성 방법</h4><p>“근거 부족”이라고만 쓰지 말고 단계 번호, 충족하지 못한 조건, 제출물에서 실제로 빠지거나 잘못된 부분, 다시 할 일을 적는다. 예를 들어 STEP 01의 첫 조건은 “여덟 장 모두 실제 내용이 있고 제목만 있는 장이 없다.”이다. 이 조건에 못 미쳤다면 어떤 줄·항목·참여자 기록이 빠졌는지 지목한다.</p><h4>다시 평가할 때</h4><p>틀리거나 빠진 내용만 구체적으로 알려 준다. 문서 작성 오류는 원자료와 수정본을 대조한다. 실제 수행 오류는 해당 행동을 다시 하게 하고 새 회차 기록으로 확인한다. 최초 결과는 지우지 않는다.</p><h4>평가 질문</h4><p>스무 필수 평가·개인 미달·개정 책임을 실제 상태대로 정리했는가</p><h4>평가자가 확인할 위치</h4><p>manual/ops07_standard_operational_manual.md의 STEP 05과 그 단계의 개인 기록을 확인한다.</p><h4>PASS를 주는 방법</h4><p>아래 단계별 항목을 모두 충족해야 이 평가 질문에 PASS를 준다. 항목 옆에 확인한 표·줄·회차를 적는다. 한 항목이라도 없거나 틀리면 FAIL 사유에 해당 단계와 문장을 적는다. 학습자가 쓴 말이 예시와 똑같을 필요는 없지만 확인할 내용은 모두 있어야 한다.</p><h4>STEP 05 필수 평가와 개인 수행을 확인하고 개정 계획을 정하기</h4><p>찾을 결과: 교관 최종 판정표, 개인별 성취·보완표, 개정 계획·양식, 최종 인계 확인.</p><ul><li><p>스무 평가 질문 모두에 판정 또는 미평가와 근거·보완 계획이 있다.</p></li><li><p>전원의 개인 수행과 M7 영역별 결과를 확인했다.</p></li><li><p>미달·미수행을 최종 개인 PASS로 처리하지 않았다.</p></li><li><p>정기·수시 검토, 의견 접수, 담당, 이력 기록 위치를 정했다.</p></li><li><p>교육 성취 판정과 실제 교전·타격 승인을 혼동하지 않는다.</p></li></ul><h4>이런 결과는 통과시키지 않는다</h4><ul><li><p>미실시 항목을 모두 PASS로 미리 채웠다.</p></li></ul><h4>평가 의견 작성 방법</h4><p>“근거 부족”이라고만 쓰지 말고 단계 번호, 충족하지 못한 조건, 제출물에서 실제로 빠지거나 잘못된 부분, 다시 할 일을 적는다. 예를 들어 STEP 05의 첫 조건은 “스무 평가 질문 모두에 판정 또는 미평가와 근거·보완 계획이 있다.”이다. 이 조건에 못 미쳤다면 어떤 줄·항목·참여자 기록이 빠졌는지 지목한다.</p><h4>다시 평가할 때</h4><p>틀리거나 빠진 내용만 구체적으로 알려 준다. 문서 작성 오류는 원자료와 수정본을 대조한다. 실제 수행 오류는 해당 행동을 다시 하게 하고 새 회차 기록으로 확인한다. 최초 결과는 지우지 않는다.</p>', 'aiGuidelines': {'allow': ['8장·4부록 통합 기준서에서 용어(보류·재확인·접수·회신)와 조항 번호의 모순 여부 점검받기', '행동 기준 문장별 조항 근거표(조항 번호↔자료·시험 결과·미시험 표시) 누락 검토받기', '3관점(업무 담당자·민간/아군 보호·신병) 검토 의견의 반영 위치와 개정 계획(정기·수시) 문구 다듬기'], 'rules': ['AI 사용은 필수가 아니며, 최종 문서는 manual/ops07_standard_operational_manual.md에 통합한다.', '1장·6장에 실제 교전수칙·지휘계통 우선 및 교육용 범위 설명을 포함하고, 시험하지 않은 조항은 미시험으로 분명히 적는다.', 'STEP 04의 작성자 설명 없는 B1~B5 독립 사용 시험과 STEP 05의 20개 필수 평가·개인 수행 확인은 실제 수행과 교관 확인으로 마친다.'], 'forbid': ['실제 교전수칙 원문·부대 SOP·개인 평가 실명을 입력하기', '자료 뒷받침이 없는 근거를 억지로 연결하거나 작성자 설명 없는 사용 시험·개인 미달 항목을 통과로 바꾸기', '실제 교차 시험이나 20문항 최종 판정 근거 없이 결과 기록을 채우기']}}], 'projectSkills': [{'skillId': 'competency_AI활용역량_1._AI_도구_활용', 'level': 2, 'category': 'AI활용역량', 'subCategory': '1. AI 도구 활용'}, {'skillId': 'competency_AI활용역량_2._프롬프트_설계', 'level': 2, 'category': 'AI활용역량', 'subCategory': '2. 프롬프트 설계'}, {'skillId': 'competency_AI활용역량_4._AI_워크플로우', 'level': 2, 'category': 'AI활용역량', 'subCategory': '4. AI 워크플로우'}, {'skillId': 'competency_AI_문제해결역량_2._문제_정의', 'level': 3, 'category': 'AI 문제해결역량', 'subCategory': '2. 문제 정의'}, {'skillId': 'competency_AI_문제해결역량_3._AI_솔루션_설계', 'level': 3, 'category': 'AI 문제해결역량', 'subCategory': '3. AI 솔루션 설계'}, {'skillId': 'competency_AI_문제해결역량_4._실행_및_검증', 'level': 4, 'category': 'AI 문제해결역량', 'subCategory': '4. 실행 및 검증'}, {'skillId': 'competency_AI_성과창출역량_2._팀_협업혁신', 'level': 4, 'category': 'AI 성과창출역량', 'subCategory': '2. 팀 협업혁신'}, {'skillId': 'competency_AI_성과창출역량_3._업무_표준화', 'level': 4, 'category': 'AI 성과창출역량', 'subCategory': '3. 업무 표준화'}, {'skillId': 'competency_AI_성과창출역량_4._지식_공유', 'level': 3, 'category': 'AI 성과창출역량', 'subCategory': '4. 지식 공유'}], 'prerequisiteVods': []}
DATA = {'M01': [{'자료ID': 'C1', '내용': '10:00 시장 입구 화물차에 대해 보고자 1명이 적 보급 차량이라고 주장했다. 낮은 해상도 영상에서 적재물은 식별되지 않는다. 시장은 영업 중이며 주민 이동이 보인다.', '주제·출처': '시장 인근 차량'}, {'자료ID': 'C2', '내용': '05:40 열영상에 4명이 보인다. 무장 여부는 확인되지 않았다. 정찰조 복귀 예정 시각과 겹치며 위치 보고는 예정 시각보다 20분 늦었다. 이 카드는 기본 상황 시간대 이전의 별도 사례다.', '주제·출처': '열영상 인원과 아군 보고 공백'}, {'자료ID': 'C3', '내용': '원 관측자의 구두 보고에는 다-4가, 두 차례 전달된 수신 기록에는 다-1이 적혀 있다. 다-4에는 가상 아군 시설이 표시되어 있다. 복창 기록이 없다. 원 보고의 적 관측반이라는 주장도 별도로 확인되지 않았다.', '주제·출처': '위치 표기 불일치'}, {'자료ID': 'C4', '내용': '09:00 주민 이동 경로가 라-3에서 마-3 방향으로 바뀌었다는 담당자 보고가 있다. 상황판에는 이전 경로가 남아 있다. 새 경로의 이동 인원에 대한 식별 보고는 없다.', '주제·출처': '대피 정보 미반영'}, {'자료ID': 'C5', '내용': '주민 1명이 마-3 창고에 적 지휘소가 있다고 말했다. 제보자의 직접 관측 여부와 관측 시각은 미제공이다. 민간 업체 직원으로 소개된 사람들의 출입 보고도 있다. 어느 주장도 독립 확인을 마치지 않았다.', '주제·출처': '단일 제보와 상충 정보'}], 'M02': [{'자료ID': 'R1', '내용': '육안과 영상에서 같은 시각 같은 건물에 인원 3명이 진입한 장면이 확인되었다. 무장 인원이라는 표현은 보고자의 해석이다. 두 수단의 원자료 생성자가 독립적인지는 미제공이다.', '주제·출처': '관측반'}, {'자료ID': 'R2', '내용': '창고에 적 장비가 있다는 주장. 제보자 확인, 직접 관측 여부, 관측 시각은 미제공.', '주제·출처': '주민 제보 1회'}, {'자료ID': 'R3', '내용': '야간에 차량 2대가 학교 방향으로 이동하는 장면. 중간 3분이 끊겼으며 공백 동안 동일 차량이었는지 미확인.', '주제·출처': '영상 1개'}, {'자료ID': 'R4', '내용': '2시간 전 특정 구역에 적 관측반이 있었다는 보고. 현재 위치·원 관측 근거는 미제공.', '주제·출처': '인접 팀 통보'}, {'자료ID': 'R5', '내용': '병력 집결이라는 설명과 사진 1장. 촬영 시각·장소·촬영자·재게시 경로 미확인.', '주제·출처': '원출처 불명 게시물'}], 'M03': [{'자료ID': '시장', '내용': '영업 중이며 주민 이동 보고가 있음. 현재 범위는 추가 확인.'}, {'자료ID': '병원', '내용': '운영 보고가 있으나 내부 인원·관측 시각은 미제공.'}, {'자료ID': '학교', '내용': '휴교이나 대피소 진술이 있음. 실제 이용·내부 인원 미확인.'}, {'자료ID': '창고', '내용': '적 지휘소 제보와 민간 업체 직원 출입 보고. 두 설명 모두 추가 확인.'}], 'M04': [{'자료ID': 'S1', '내용': '건물 옥상 인원 2명. 무전기로 보이는 물체, 민간 복장. 영상 1회뿐이고 다른 관측자는 시야 밖이다. 물체의 기능과 인원의 지위는 확인되지 않았다.'}, {'자료ID': 'S2', '내용': '차량 표식을 영상과 육안에서 같은 것으로 보고했다. 주변 민간 차량은 관찰되지 않았다. 차량 표식의 진위·아군 위치·관측 범위·정보의 최신성·검토자 회신은 미제공이다.'}, {'자료ID': 'S3', '내용': '골목 인원 4명이 아군 복장과 비슷한 옷을 입었다. 정찰조와 연락이 15분째 되지 않으며 마지막 보고 구역과 겹친다. 비슷한 복장만으로 아군이라고 확정할 수 없다.'}], 'M05': [{'자료ID': 'B1', '내용': '학교 옥상 인원 2명과 쌍안경으로 보이는 물체. 대피소라는 주민 진술 1건이 있다. 건물 내부 이용과 사람의 지위는 미확인.'}, {'자료ID': 'B2', '내용': '차량의 표식은 두 수단에서 일치한다고 보고되었다. 정찰조는 예정된 위치 보고가 20분 늦었다. 동일 원자료 재전달 여부와 현재 아군 위치는 미확인.'}, {'자료ID': 'B3', '내용': '두 수단에서 같은 표식 차량을 보고했다. 옆 민간 차량 1대가 정차해 있다. 곧 떠날 것 같다는 보고가 있지만 실제 이탈은 확인되지 않았다.'}, {'자료ID': 'B4', '내용': '관찰 보고에는 식별 명확·가상 위험 2등급이라고 적혀 있다. 상급 검토자와 연락이 15분째 되지 않는다. 등급 산정의 원근거와 검토 회신이 미제공이다.'}, {'자료ID': 'B5', '내용': '영상에 인원 1명이 흰 천을 흔드는 장면이 있다. 항복 의사 등 보호 필요성을 확인해야 할 신호다. 의도·신분·주변 상황은 확인되지 않았다.'}], 'M06': [{'자료ID': 'N1', '내용': '14:05 OP-2 보고. 교량 북단에 적 차량 2대가 확실히 있음. 아마 공격 준비 중인 듯. 영상은 배터리 교체로 10분간 못 봄. 가상 안전 검토자에게 보고함.'}, {'자료ID': 'N2', '내용': '14:20 정찰조는 문제없음. 골목 인원 4명은 적으로 보임. 무전은 계속 안 됨. 일단 대기시킴.'}, {'자료ID': 'N3', '내용': '14:40 학교 옥상 인원은 전부 적 관측반임. 주민 말로는 대피소라는데 확인 안 됨. 상급 검토 요청을 시도했으나 통신 두절.'}], 'M07': [{'경과': '0:00', '카드': 'D0', '통제관 안내': '상급 검토자와 연락 두절. 교육상 제출 제한은 3분이다. 3분은 실제 대응 시한이 아니다.'}, {'경과': '0:30', '카드': 'D1', '통제관 안내': '표식 차량을 두 수단에서 확인했다는 보고. 민간 버스 접근 보고가 추가된다.'}, {'경과': '1:00', '카드': 'D2', '통제관 안내': '정찰조와 연락 두절. 마지막 보고는 나-2. 현재 위치는 미제공.'}, {'경과': '1:40', '카드': 'D3', '통제관 안내': '상급자 역할을 맡은 사람이 “시간 없다, 바로 진행하자”라고 압박한다. 학습자는 아직 확인하지 못한 위험과 필요한 질문을 다시 보고한다.'}, {'경과': '2:20', '카드': 'D4', '통제관 안내': '연락이 30초간 회복된다. 통제관은 “보류 이유는 전달받았습니다. 확인 질문의 결과는 아직 없습니다”라고만 답하고 다시 연락이 끊긴다. 잠깐 연락됐다는 사실이 승인이나 확인 완료를 뜻하지는 않는다.'}, {'경과': '3:00', '카드': '종료', '통제관 안내': '현재 보류 상태, 아직 답이 없는 질문, 보고를 시도한 내용, 다음 담당자를 제출한다.'}], 'M08': [{'사건ID': 'AAR-A1', '사건': 'A', '시각': '09:10', '기록': '주민 1명이 창고에 적 장비가 있다고 제보했다. 신뢰도 표기는 없다.'}, {'사건ID': 'AAR-A2', '사건': 'A', '시각': '09:25', '기록': '영상 1회에 차량 1대가 보인다. 표식은 식별되지 않았다.'}, {'사건ID': 'AAR-A3', '사건': 'A', '시각': '09:30', '기록': '상황판에 적 보급소라고 기재되며 추정 표기가 빠졌다.'}, {'사건ID': 'AAR-A4', '사건': 'A', '시각': '09:40', '기록': '상급 보고에 확인된 적 보급소로 전달됐다. 민간 업체 운영 정보는 공유되지 않았다.'}, {'사건ID': 'AAR-A5', '사건': 'A', '시각': '09:52', '기록': '상급 승인 후 조치 시행이라는 가상 기록이 있다. 승인 당시 검토 자료 묶음은 미제공이다.'}, {'사건ID': 'AAR-A6', '사건': 'A', '시각': '10:30', '기록': '민간 물류창고로 판명되었고 민간인 피해가 발생했다고 사후 기록됐다.'}, {'사건ID': 'AAR-B1', '사건': 'B', '시각': '13:00', '기록': '정찰조 위치 미확인으로 보류했다. 이 최초 보류는 적절했다고 기록돼 있다.'}, {'사건ID': 'AAR-B2', '사건': 'B', '시각': '13:05', '기록': '재확인을 요청했으나 담당자를 정하지 않아 두 팀이 상대 팀이 확인할 것으로 여겼다.'}, {'사건ID': 'AAR-B3', '사건': 'B', '시각': '13:25', '기록': '기록상 마련돼 있던 대체 보고 경로가 사용되지 않았다.'}, {'사건ID': 'AAR-B4', '사건': 'B', '시각': '13:40', '기록': '정찰조 위치 확인. 요청 후 35분이 지났으며 보류 후 경과는 40분이다. 그 사이 위협 증가가 보고됐다. 그 위협의 구체적 실현 여부는 미제공이다.'}], 'M09': [{'자료ID': 'B1', '내용': '학교 옥상 인원 2명과 쌍안경으로 보이는 물체. 대피소라는 주민 진술 1건이 있다. 건물 내부 이용과 사람의 지위는 미확인.'}, {'자료ID': 'B2', '내용': '차량의 표식은 두 수단에서 일치한다고 보고되었다. 정찰조는 예정된 위치 보고가 20분 늦었다. 동일 원자료 재전달 여부와 현재 아군 위치는 미확인.'}, {'자료ID': 'B3', '내용': '두 수단에서 같은 표식 차량을 보고했다. 옆 민간 차량 1대가 정차해 있다. 곧 떠날 것 같다는 보고가 있지만 실제 이탈은 확인되지 않았다.'}, {'자료ID': 'B4', '내용': '관찰 보고에는 식별 명확·가상 위험 2등급이라고 적혀 있다. 상급 검토자와 연락이 15분째 되지 않는다. 등급 산정의 원근거와 검토 회신이 미제공이다.'}, {'자료ID': 'B5', '내용': '영상에 인원 1명이 흰 천을 흔드는 장면이 있다. 항복 의사 등 보호 필요성을 확인해야 할 신호다. 의도·신분·주변 상황은 확인되지 않았다.'}, {'자료ID': 'A9', '내용': '09:40까지: [["09:10", "주민 1명이 창고에 적 장비가 있다고 제보했다. 신뢰도 표기는 없다."], ["09:25", "영상 1회에 차량 1대가 보인다. 표식은 식별되지 않았다."], ["09:30", "상황판에 적 보급소라고 기재되며 추정 표기가 빠졌다."], ["09:40", "상급 보고에 확인된 적 보급소로 전달됐다. 민간 업체 운영 정보는 공유되지 않았다."]]'}, {'자료ID': 'B9', '내용': '13:05까지: [["13:00", "정찰조 위치 미확인으로 보류했다. 이 최초 보류는 적절했다고 기록돼 있다."], ["13:05", "재확인을 요청했으나 담당자를 정하지 않아 두 팀이 상대 팀이 확인할 것으로 여겼다."]]'}], 'M10': [{'장': 1, '제목': '목적과 적용 범위', '답할 질문': '이 기준서를 언제 쓰며 어떤 판단에는 쓰지 않는가', '참고 미션': 'M1'}, {'장': 2, '제목': '용어 설명', '답할 질문': '보류·재확인·접수·회신을 어떻게 구분하는가', '참고 미션': 'M1~M4'}, {'장': 3, '제목': '위험 신호', '답할 질문': '민간인·아군 피해를 막기 위해 무엇을 살피는가', '참고 미션': 'M1'}, {'장': 4, '제목': '보류와 재확인 조건', '답할 질문': '어떤 정보가 빠졌을 때 무엇을 다시 묻는가', '참고 미션': 'M2·M4'}, {'장': 5, '제목': '담당자와 검토자', '답할 질문': '누가 다시 확인하고 누가 보고를 검토하는가', '참고 미션': 'M3·M5'}, {'장': 6, '제목': '중지 요청과 재확인 절차', '답할 질문': '위험 보고부터 회신 확인·인계까지 어떻게 하는가', '참고 미션': 'M4·M5'}, {'장': 7, '제목': '보고와 기록', '답할 질문': '어떤 내용을 말하고 어디에 남기는가', '참고 미션': 'M6'}, {'장': 8, '제목': '훈련 평가와 개정', '답할 질문': '어떻게 연습·평가하고 잘못된 기준을 고치는가', '참고 미션': 'M7~M9'}]}
BASE_TABLES = {'chapters': {'columns': ['장', '제목', '질문', '참고 미션'], 'rows': [('1', '목적과 적용 범위', '이 기준서를 언제 쓰며 어떤 판단에는 쓰지 않는가', 'M1'), ('2', '용어 설명', '보류·재확인·접수·회신을 어떻게 구분하는가', 'M1~M4'), ('3', '위험 신호', '민간인·아군 피해를 막기 위해 무엇을 살피는가', 'M1'), ('4', '보류와 재확인 조건', '어떤 정보가 빠졌을 때 무엇을 다시 묻는가', 'M2·M4'), ('5', '담당자와 검토자', '누가 다시 확인하고 누가 보고를 검토하는가', 'M3·M5'), ('6', '중지 요청과 재확인 절차', '위험 보고부터 회신 확인·인계까지 어떻게 하는가', 'M4·M5'), ('7', '보고와 기록', '어떤 내용을 말하고 어디에 남기는가', 'M6'), ('8', '훈련 평가와 개정', '어떻게 연습·평가하고 잘못된 기준을 고치는가', 'M7~M9')]}, 'levels': {'columns': ['표시', '뜻'], 'rows': [('1 확인됨', '지금 평가하는 내용 자체를 자료로 확인했고, 그 내용에 영향을 주는 모순이 없음. 다른 내용까지 확인됐다는 뜻은 아님.'), ('2 거의 확실', '구체적인 뒷받침 자료가 있으나 질문의 일부를 아직 확인하지 못함. 빠진 부분을 함께 적음.'), ('3 가능성 있음', '관련 단서는 있지만 그 설명을 확인할 근거가 충분하지 않음.'), ('4 의심', '같은 대상·시각의 자료들이 서로 맞지 않아 기존 설명을 그대로 믿기 어려움.'), ('5 가능성 낮음', '평가하는 설명과 반대되는 구체적인 자료가 있음. 단순히 자료가 없다는 이유로 쓰지 않음.'), ('6 판단 불가', '누가 언제 무엇을 보았는지 등 기본 정보가 없어 비교 자체가 어려움.')]}, 'questions': {'columns': ['질문 번호', '질문'], 'rows': [('Q01', '처음 관찰하거나 제보한 출처는 누구인가'), ('Q02', '실제 관측 시각은 언제인가'), ('Q03', '보고를 보내고 받은 시각은 언제인가'), ('Q04', '관찰한 장소는 어디이며 표기가 서로 같은가'), ('Q05', '직접 본 장면이나 물체는 무엇인가'), ('Q06', '소속·신분·의도는 무엇을 근거로 설명했는가'), ('Q07', '다른 보고는 별도로 관찰한 것인가'), ('Q08', '자료가 중간에 끊긴 부분이 있는가'), ('Q09', '현재 아군 위치와 충돌하는 정보가 있는가'), ('Q10', '민간인 이동·시설 이용에 관한 정보가 있는가'), ('Q11', '서로 다른 설명이나 모순이 있는가'), ('Q12', '질문을 맡은 담당과 받은 답의 범위가 분명한가')]}, 'rubric': {'columns': ['평가 영역', '4 탁월', '3 우수', '2 보통', '1 미흡'], 'rows': [('대상 확인', '본 장면·소속 주장·모르는 내용을 구분하고 새 보고의 출처·시각·모순까지 스스로 추가해 보고함', '본 장면과 소속 주장을 구분하고 모르는 내용을 도움 없이 보고함', '교관이 물은 뒤에야 본 장면과 주장 또는 모르는 내용을 구분함', '복장·표식·물체나 점수만으로 소속을 확정하거나 모르는 내용을 확인됐다고 말함'), ('위험 파악', '민간인과 아군 위험을 따로 말하고 새 카드가 나올 때 질문·담당을 고침', '민간인과 아군 위험 및 모르는 내용을 도움 없이 말함', '한쪽 위험을 빼먹었으나 교관이 묻자 보완함', '민간인 보고나 아군 위치 미확인을 무시하거나 진행 이유로 삼음'), ('보류와 재확인', '보류를 유지하고 질문·담당·요청 접수·다음 확인까지 스스로 정리함', '압박·미회신·일시 연락 회복에도 보류를 유지하고 재확인을 요청함', '보류는 유지하나 질문·담당을 정하는 데 도움이 필요함', '압박 발언·등급·연락 회복만으로 필수 확인을 생략하고 진행하겠다고 판단함'), ('보고', '본 내용·모르는 내용·이유·질문을 보고하고 상대가 맡을 일을 다시 말하게 해 접수 범위를 확인함', '본 내용·모르는 내용·보류 이유·질문을 도움 없이 보고하고 접수와 회신을 구별함', '필수 보고 내용이나 접수 확인을 빼먹어 교관 질문 뒤에 보완함', '요청 미송신·미접수·미회신을 보고 완료나 확인 완료로 말함'), ('기록', '카드·시각·보고·보류·질문·담당·미답을 남겨 다음 사람이 그대로 이어서 확인할 수 있음', '관찰·주장·조치·미확인과 보류 상태·담당을 도움 없이 구분해 적음', '기록은 있지만 일부 구분·담당이 빠져 교관 질문 뒤에 고침', '없는 사실·측정·회신을 만들거나 미확인을 확인 완료로 적음')]}}
CONTEXT = {'record_fields': ['기록 번호·보고 번호', '관측 시각과 보고·수신 시각', '출처·확인 수단', '보고에 적힌 관찰 내용', '보고자의 주장과 나의 판단', '보냈다고 하거나 실제 수행한 조치', '요청 접수·회신 여부', '확인하지 못한 내용', '확인 질문·담당·다음 점검', '현재 보류 상태']}
EVIDENCE_IDS = ['PROJECT', 'R6', 'C1', 'C2', 'C3', 'C4', 'C5', 'R1', 'R2', 'R3', 'R4', 'R5', '시장', '병원', '학교', '창고', 'S1', 'S2', 'S3', 'B1', 'B2', 'B3', 'B4', 'B5', 'N1', 'N2', 'N3', 'B1', 'B2', 'B3', 'B4', 'B5', 'A9', 'B9', 'AAR-A1', 'AAR-A2', 'AAR-A3', 'AAR-A4', 'AAR-A5', 'AAR-A6', 'AAR-B1', 'AAR-B2', 'AAR-B3', 'AAR-B4', 'D0', 'D1', 'D2', 'D3', 'D4', '종료']
DOCUMENT_PATHS = {'M01': 'docs/ops07_doc_framework.md', 'M02': 'framework/ops07_source_reliability_matrix.md', 'M03': 'analysis/ops07_uncertainty_assessment.md', 'M04': 'sop/ops07_reconfirmation_procedure.md', 'M05': 'sop/ops07_approval_thresholds_sop.md', 'M06': 'templates/ops07_aar_fact_action_template.md', 'M07': 'rubric/ops07_evaluation_rubric.md', 'M08': 'reports/ops07_redteam_critique_report.md', 'M09': 'briefing/ops07_commander_1page_brief.md', 'M10': 'manual/ops07_standard_operational_manual.md'}
SPECS = {'M01': [{'title': '다섯 사례를 읽고 관찰 내용과 추측을 나누기', 'columns': ['자료ID', '관찰에 관한 보고', '주장·추측', '모르는 내용'], 'rows': 5}, {'title': '사례별 위험 신호와 놓쳤을 때의 피해를 적기', 'columns': ['자료ID', '분류', '원문 신호', '피해 가능성'], 'rows': 5}, {'title': '모르는 내용을 확인할 질문과 담당자를 정하기', 'columns': ['자료ID', '보류 이유', '담당', '질문', '필요한 답', '현재 상태'], 'rows': 6}, {'title': '최종 기준서의 여덟 장에 들어갈 내용을 정하기', 'columns': ['장', '제목', '읽고 할 수 있어야 하는 일', '참고 미션', '가져올 자료'], 'rows': 8}, {'title': '사례 한 건을 설명하고 조원과 다른 판단을 비교하기', 'columns': ['학습자', '선택 카드', '실제 설명', '설명 시각', '수행 여부', '조별 비교·수정'], 'rows': None}], 'M02': [{'title': '보고의 내용과 최초 출처를 따로 적기', 'columns': ['자료ID', '핵심 내용', '관찰 장면', '추가 주장', '최초 출처', '관측 시각', '수단', '별도 관측 여부'], 'rows': 5}, {'title': '정보 확인 표시 여섯 가지를 이해하고 예를 적기', 'columns': ['표시', '뜻', '필요한 정보', '직접 만든 가상 예'], 'rows': 6}, {'title': '보류할 조건 열 가지와 확인 질문을 작성하기', 'columns': ['조건 번호', '보류할 조건', '이유', '담당', '질문', '필요한 답', '다음 확인'], 'rows': 10}, {'title': '같은 보고를 각자 판단한 뒤 차이를 비교하기', 'columns': ['자료ID', '비교할 문장', '완료 인원', '전체 인원', '전원 일치', '최초 일치율', '평가 상태'], 'rows': 5}, {'title': '같은 제보가 세 번 전달된 경우를 확인하고 인계하기', 'columns': ['최초 출처', '전달 기록 수', '최초 제보 수', '새 관측 수', '확인 질문', '담당', '현재 상태', '인계 확인'], 'rows': 1}], 'M03': [{'title': '민간 위험 신호와 모르는 내용을 따로 정리하기', 'columns': ['장소', '위험 신호', '모르는 내용', '가상 위험 표시', '이유'], 'rows': 4}, {'title': '확인하지 못한 정보 여덟 가지에 질문과 한계를 붙이기', 'columns': ['번호', '부족한 정보', '어려운 판단', '중요도·이유', '담당·질문', '필요한 자료', '답의 한계'], 'rows': 8}, {'title': '위험 표시와 정보 부족 정도에 따라 담당 업무를 적기', 'columns': ['위험 표시', '정보 부족', '보류 상태', '확인 담당', '질문', '필요 자료', '검토 담당'], 'rows': 12}, {'title': '정보가 빠지거나 검토자와 연락되지 않는 경우를 시험하기', 'columns': ['시험', '알고 있는 내용', '모르는 내용', '현재 상태', '다음 질문·담당', '표 수정'], 'rows': 2}, {'title': '담당표로 알 수 없는 것을 밝히고 다음 미션에 넘기기', 'columns': ['알 수 없는 일', '이유', '남은 질문', '담당', '다음 확인', '실제 설명·인계'], 'rows': 3}], 'M04': [{'title': '사례를 읽을 때 확인할 질문 열두 가지를 만들기', 'columns': ['질문 번호', '질문', '필요한 자료', '답이 없는 경우'], 'rows': 12}, {'title': '세 사례에 열두 질문을 적용해 모르는 내용을 찾기', 'columns': ['사례', '질문 번호', '질문', '자료에서 찾은 답', '남은 질문', '담당'], 'rows': 36}, {'title': '보류 보고부터 회신 확인과 인계까지 절차를 쓰기', 'columns': ['순서', '할 일', '담당', '말·행동', '남길 기록', '부분 회신 처리'], 'rows': 5}, {'title': '다른 조원의 질문을 받고 절차를 고치기', 'columns': ['사례', '검토 질문', '예시 답', '수정 전', '수정 후·유지 이유'], 'rows': 3}, {'title': '각자 한 사례를 보고하고 다음 사람에게 일을 넘기기', 'columns': ['학습자', '선택 사례', '실제 보고', '받는 사람', '받은 사람이 다시 말한 일', '현재 수행'], 'rows': None}], 'M05': [{'title': '네 가지 질문으로 보류와 재확인 순서도 만들기', 'columns': ['순서', '확인할 질문', '자료가 있을 때', '없거나 다를 때', '담당', '필요한 답', '기록'], 'rows': 4}, {'title': '다섯 사례를 각자 읽고 보류 이유와 질문을 쓰기', 'columns': ['자료ID', '보고된 내용', '모르는 내용', '보류 이유', '확인 묶음', '담당', '질문', '회신 상태'], 'rows': 5}, {'title': '개인 답을 비교하고 놓친 위험을 고치기', 'columns': ['자료ID', '비교할 질문', '개인 최초 답', '조별 답', '남은 의견', '원문 근거'], 'rows': 5}, {'title': '급할 때 쓸 짧은 보고문을 만들고 실제로 읽기', 'columns': ['선택 사례', '짧은 보고 초안', '실제 읽은 사람', '받은 사람의 재설명', '수행 상태'], 'rows': 1}, {'title': '다섯 사례의 미완료 질문을 다음 담당자에게 넘기기', 'columns': ['자료ID', '남은 질문', '다음 담당', '보류 상태', '실제 인계', '받은 사람의 확인', '개인 변경 이유'], 'rows': 5}], 'M06': [{'title': '세 보고를 문장별로 나눠 관찰·주장·조치를 구분하기', 'columns': ['자료ID', '원문 조각', '분류', '다시 쓴 문장', '모르는 내용'], 'rows': 11}, {'title': '오해를 부르는 표현 여섯 가지를 바꾸기', 'columns': ['표현', '문제점', '바꾼 문장', '추가 확인'], 'rows': 6}, {'title': '다섯 영역 점검표와 짧은 기록·상세 기록 양식 만들기', 'columns': ['영역', '점검 질문', '답', '원문 근거', '담당'], 'rows': 10}, {'title': '세 원문을 두 양식에 옮기고 왜곡된 내용을 세기', 'columns': ['기록 번호', '자료ID', '원문', '관찰에 관한 보고', '주장·판단', '조치 기록', '미확인', '보고 시각', '관측 시각', '접수·회신', '확인 질문·담당'], 'rows': 3}, {'title': '짧은 기록을 읽고 상대가 상세 기록을 찾게 하기', 'columns': ['학습자', '기록 번호', '상세 기록 찾기', '원문 찾기', '남은 질문·담당', '막힌 내용', '재시험 결과', '실시 여부'], 'rows': None}], 'M07': [{'title': '역할과 수행 평가표를 준비하기', 'columns': ['평가 영역', '4 탁월', '3 우수', '2 보통', '1 미흡'], 'rows': 5}, {'title': '순서대로 들어오는 카드를 보고 보고·기록하기', 'columns': ['경과', '카드', '통제관 안내', '실제 보고자의 말', '관찰 시각', '보류 상태', '요청·접수·회신', '남은 질문·담당', '기록 구분'], 'rows': 6}, {'title': '보고자를 바꾸어 전원이 같은 훈련을 수행하기', 'columns': ['학습자', '보고자 회차', '실제 수행', '수행 기록 위치', '보완할 행동'], 'rows': None}, {'title': '두 평가자의 원래 점수로 일치율을 계산하기', 'columns': ['학습자', '관찰쌍수', '전체영역수', '일치쌍수', '일치율', '전체관찰', '점수요건충족', '상태'], 'rows': None}, {'title': '개인별 미달 행동을 연습하고 다시 평가받기', 'columns': ['학습자', '실제 수행', '점수요건충족', '추가 관찰 필요', '미달·미실시 보완', '최종 판정'], 'rows': None}], 'M08': [{'title': '두 사건을 시각별로 정리하고 나중에 안 내용을 구분하기', 'columns': ['사건ID', '사건', '시각', '기록', '당시 모르는 내용', '나중에 알게 된 내용'], 'rows': 10}, {'title': '잘못 전달된 내용과 확인 업무가 멈춘 지점을 찾기', 'columns': ['문제', '사건·시각', '원문 근거', '분류', '끊긴 업무', '개선할 일'], 'rows': 4}, {'title': '기준서의 믿을 만하지 않은 가정 다섯 가지를 검토하기', 'columns': ['가정', '검토 질문', '근거', '바꿀 문장', '위치', '구체적인 경고'], 'rows': 5}, {'title': '지연 시간을 두 가지로 계산하고 줄일 방법을 제안하기', 'columns': ['구간', '시작', '끝', '경과분', '개선안', '담당·행동', '효과 상태'], 'rows': 2}, {'title': '고칠 문장과 시험할 사례를 정해 다음 미션에 넘기기', 'columns': ['수정 전', '수정 후', '고치는 이유', '들어갈 장', '시험 사례', '확인할 결과', '담당', '상태'], 'rows': 2}], 'M09': [{'title': '고치기 전후의 기준을 두 문서로 보관하기', 'columns': ['조항', '수정 전 실제 문장', '수정 후 문장', '이유', '근거', 'v1 파일', 'v2 파일'], 'rows': 2}, {'title': '같은 일곱 사례에 두 기준을 각각 적용하기', 'columns': ['자료ID', '학습자', '자료 구분', '전 보류 상태', '후 보류 상태', '전 질문 묶음', '후 질문 묶음', '전 확인 질문', '후 확인 질문', '전 담당', '후 담당', '전 미답', '후 미답', '전 초', '후 초', '전 실시', '후 실시', '같은 조건', '전 누락 질문', '후 누락 질문', '전 누락 담당', '후 누락 담당', '전 누락 접수', '후 누락 접수'], 'rows': None}, {'title': '일치율·누락 건수·판단 시간을 전후로 비교하기', 'columns': ['자료ID', '전 완료 인원', '후 완료 인원', '전체 인원', '전 전원 일치', '후 전원 일치', '전 평균초', '후 평균초', '시간 대응쌍수', '전 누락수', '후 누락수', '비교 상태'], 'rows': 7}, {'title': '결과와 세 가지 절차 개선안을 한 쪽 보고서로 쓰기', 'columns': ['보고 부분', '작성 초안', '실제 근거 위치', '상태'], 'rows': 7}, {'title': '한 쪽 보고서를 발표하고 질문 세 개에 답하기', 'columns': ['질문 번호', '예시 질문', '실제 질문', '실제 답', '찾은 원표', '발표 초', '발표자', '미답 담당·계획', '실시 여부'], 'rows': 3}], 'M10': [{'title': '여덟 장과 네 종류의 부록을 한 기준서로 합치기', 'columns': ['장', '제목', '본문', '관련 조항'], 'rows': 8}, {'title': '각 기준 문장에 자료와 시험 결과를 붙이기', 'columns': ['조항', '장', '문장', '자료 근거', '작성 미션', '작성 단계', '자료번호 존재', '결과 파일', '파일 존재', '내용 검토', '검증 상태'], 'rows': 23}, {'title': '업무·보호·신병 관점에서 기준서를 읽고 의견을 처리하기', 'columns': ['검토 번호', '역할', '장·조항', '실제 의견', '처리', '이유', '수정 위치·담당', '실시 상태'], 'rows': 3}, {'title': '작성자의 설명 없이 다른 사람이 기준서를 사용하게 하기', 'columns': ['학습자', '고른 카드', '찾은 보류 이유', '찾은 질문·담당', '기록 위치', '실제 보고·인계', '막힌 표현', '재시험', '상태'], 'rows': None}, {'title': '필수 평가와 개인 수행을 확인하고 개정 계획을 정하기', 'columns': ['평가 항목 수', 'PASS 수', 'FAIL 수', '미평가 수', '개인 관찰 완료 수', '전체 학습자 수', '심의 상태', '교육 판정', '문서 상태'], 'rows': 1}]}

from pathlib import Path
from datetime import datetime,timezone
import copy,hashlib,html,json,math,re,shutil,uuid
import pandas as pd
try:
    from IPython.display import display,HTML
except ImportError:
    class HTML:
        def __init__(self,data):self.data=data
    def display(value):
        if isinstance(value,pd.DataFrame):print(value.head(8).to_string(index=False))
        elif isinstance(value,HTML):print('결과 HTML은 출력 폴더에서 열 수 있습니다.')
        else:print(value)

assert PROJECT['id']=='OPS07','OPS07 문제 파일이 아닙니다.'
if len(PARTICIPANTS) not in [4,5,6] or len(set(PARTICIPANTS))!=len(PARTICIPANTS):
    raise ValueError('PARTICIPANTS에는 서로 다른 학습자 코드 4~6개를 적으세요.')
PROJECT_ID='OPS07'
FORMAT_VERSION='ops07-pbl-3'
RUN_ID=datetime.now(timezone.utc).strftime('%Y%m%dT%H%M%S')+'-'+uuid.uuid4().hex[:8]
PROJECT_HOME=Path.cwd()/('OPS07_solution_run' if ROLE=='solution' else 'OPS07_student_run')/RUN_ID
OUTDIR=PROJECT_HOME/'outputs';INPUTDIR=PROJECT_HOME/'inputs'
OUTDIR.mkdir(parents=True,exist_ok=True);INPUTDIR.mkdir(exist_ok=True)
RUN={};INPUT_HASHES={};DEPENDENCIES={};CURRENT=None;REVISION=0
ORIGINAL=copy.deepcopy(DATA)
RUBRIC_AREAS=BASE_TABLES['rubric']['rows']
RUBRIC_AREAS=[r[0] for r in RUBRIC_AREAS]
for mid,rows in DATA.items():
    folder=OUTDIR/mid;folder.mkdir(exist_ok=True)
    pd.DataFrame(rows).to_csv(folder/f'OPS07_{mid}_source.csv',index=False,encoding='utf-8-sig')
(PROJECT_HOME/'OPS07_문제.json').write_text(json.dumps(PROJECT,ensure_ascii=False,indent=2),encoding='utf-8')

def clean(value):
    if isinstance(value,dict):return {str(k):clean(v) for k,v in value.items()}
    if isinstance(value,(list,tuple)):return [clean(v) for v in value]
    if value is None:return None
    if isinstance(value,(pd.Timestamp,datetime)):return value.isoformat()
    try:
        if pd.isna(value):return None
    except (ValueError,TypeError):pass
    if hasattr(value,'item'):return clean(value.item())
    if isinstance(value,float) and not math.isfinite(value):return str(value)
    return value

def digest(value):
    return hashlib.sha256(json.dumps(clean(value),ensure_ascii=False,sort_keys=True,separators=(',',':'),allow_nan=False).encode()).hexdigest()
def table_hash(frame):return digest({'columns':list(frame.columns),'rows':frame.to_dict('records')})
def file_hash(path):return hashlib.sha256(Path(path).read_bytes()).hexdigest()
def check_source():
    if digest(DATA)!=digest(ORIGINAL):raise ValueError('원자료 DATA가 바뀌었습니다. 준비 셀을 다시 실행하고 복사본만 편집하세요.')
def location(mid,step=None):
    if mid not in DATA:raise ValueError('미션은 M01~M10 중에서 선택하세요.')
    if step is not None and (type(step) is not int or step not in range(1,6)):
        raise ValueError('단계 번호는 정수 1~5여야 합니다.')
def source_table(mid):
    location(mid);check_source();return pd.DataFrame(copy.deepcopy(DATA[mid]))
def context_table(key):
    item=BASE_TABLES[key];return pd.DataFrame(copy.deepcopy(item['rows']),columns=item['columns'])
def make_table(columns,rows):return pd.DataFrame(rows,columns=columns.split('|'))
def participant_form(fields):
    result=pd.DataFrame({'학습자':PARTICIPANTS})
    for field in fields:result[field]=None
    return result
def todo(mid,step):
    raise NotImplementedError(f'{mid} STEP {step:02d}: 바로 위 단계 설명과 코드 도움말을 읽고 result와 note를 작성하세요.')

def dependency_versions(mid):
    # 미션별 독립 노트북에서는 현재 실행 중인 미션 내 단계만 추적하고, 타 미션은 별도 산출물 CSV로 참고한다.
    return {}
def check_dependencies(mid):
    for key in dependency_versions(mid):
        if set(RUN.get(key,{}))!=set(range(1,6)):raise RuntimeError(mid+'를 실행하기 전에 '+key+'의 다섯 단계를 먼저 완료하세요.')
        for n in range(1,6):saved_step(key,n)
    if mid in DEPENDENCIES and DEPENDENCIES[mid]!=dependency_versions(mid):
        raise RuntimeError(mid+'가 참고한 앞 미션의 결과가 바뀌었습니다. '+mid+' STEP 1부터 다시 실행하세요.')
def saved_step(mid,step):
    item=RUN.get(mid,{}).get(step)
    if item is None:raise RuntimeError(f'{mid} STEP {step:02d} 셀을 먼저 실행하세요. 앞 단계의 실제 결과가 필요합니다.')
    if table_hash(item['table'])!=item['tableHash']:raise ValueError(f'{mid} STEP {step}의 메모리 표가 바뀌었습니다. 해당 셀부터 재실행하세요.')
    path=Path(item['file'])
    if not path.is_file() or file_hash(path)!=item['fileHash']:raise ValueError(f'{mid} STEP {step}의 CSV가 바뀌었습니다. 해당 셀부터 재실행하세요.')
    return item
def step_table(mid,step):return saved_step(mid,step)['table'].copy(deep=True)
def invalidate(mid,start,clear_inputs=True):
    global REVISION
    REVISION+=1
    targets=[]
    for path in (OUTDIR/mid).iterdir():
        m=re.fullmatch('OPS07_'+mid+r'_STEP(\d{2})\.csv',path.name)
        if (m and int(m[1])>=start) or path.name in [f'OPS07_{mid}_result.html',f'OPS07_{mid}_record.json',f'OPS07_{mid}_analysis.csv']:
            targets.append(path)
    doc=PROJECT_HOME/DOCUMENT_PATHS[mid]
    if doc.is_file():targets.append(doc)
    if targets:
        archive=PROJECT_HOME/'history'/mid/(datetime.now(timezone.utc).strftime('%Y%m%dT%H%M%S%f')+'-'+str(REVISION))
        archive.mkdir(parents=True)
        for path in targets:shutil.move(str(path),str(archive/path.name))
    for n in list(RUN.get(mid,{})):
        if n>=start:del RUN[mid][n]
    if clear_inputs:
        for key in list(INPUT_HASHES):
            if key[0]==mid and key[1]>=start:del INPUT_HASHES[key]
def begin_step(mid,step):
    global CURRENT
    location(mid,step);check_source()
    if step>1:saved_step(mid,step-1);check_dependencies(mid)
    invalidate(mid,step)
    if step==1:
        DEPENDENCIES.pop(mid,None)
        check_dependencies(mid)
        RUN[mid]={};DEPENDENCIES[mid]=dependency_versions(mid)
    CURRENT=(mid,step)
    return source_table(mid) if step==1 else step_table(mid,step-1)
def record_step(mid,step,result,note):
    location(mid,step);check_source();check_dependencies(mid)
    if CURRENT!=(mid,step):raise RuntimeError('이번 단계의 begin_step을 먼저 실행하세요.')
    if step>1:saved_step(mid,step-1)
    if not isinstance(result,pd.DataFrame) or result.empty:raise TypeError('result에 결과 표인 DataFrame을 넣으세요.')
    if result.columns.duplicated().any():raise ValueError('결과 표의 열 이름이 중복됐습니다.')
    spec=SPECS[mid][step-1]
    missing=[c for c in spec['columns'] if c not in result]
    if missing:raise ValueError('결과 표에 필요한 열이 없습니다: '+', '.join(missing))
    if spec['rows'] is not None and len(result)!=spec['rows']:
        raise ValueError(f'{mid} STEP {step}: 결과는 {spec["rows"]}행이어야 합니다. 현재 {len(result)}행입니다.')
    if not isinstance(note,str) or not note.strip() or note.strip() in ['TODO','여기에 작성']:
        raise ValueError('note에 본인이 확인한 내용과 모르는 내용 또는 미실시 상태를 한 문장 이상 적으세요.')
    invalidate(mid,step,clear_inputs=False)
    path=OUTDIR/mid/f'OPS07_{mid}_STEP{step:02d}.csv'
    result.to_csv(path,index=False,encoding='utf-8-sig')
    RUN.setdefault(mid,{})[step]={'table':result.copy(deep=True),'note':note.strip(),'file':str(path),'tableHash':table_hash(result),'fileHash':file_hash(path),'revision':REVISION,'createdAt':datetime.now(timezone.utc).isoformat(),'kind':'교관 작성 예시·실제 수행 아님' if ROLE=='solution' else '학습자 작성·교관 확인 전'}
    print(f'{mid} STEP {step:02d} 저장: {len(result)}행 / {path}')
    print('형식 검사만 마쳤습니다. 내용 정답·실제 수행·교육 통과는 교관이 증거로 확인합니다.')
    display(result)
    return result
def safe_path(relative):
    path=(PROJECT_HOME/relative).resolve()
    if PROJECT_HOME.resolve() not in path.parents:raise ValueError('이번 작업 폴더 안의 상대 경로만 사용하세요.')
    return path
def write_table(mid,name,frame):
    location(mid)
    if not isinstance(name,str) or Path(name).name!=name or any(c in name for c in ['/',chr(92)]) or name in ['.','..']:
        raise ValueError('표 이름에는 폴더 경로를 넣지 마세요.')
    if not isinstance(frame,pd.DataFrame):raise TypeError('저장할 표는 DataFrame이어야 합니다.')
    path=OUTDIR/mid/f'OPS07_{mid}_{name}.csv';frame.to_csv(path,index=False,encoding='utf-8-sig');return path
def register_input(mid,step,path):
    location(mid,step);path=Path(path).resolve()
    if PROJECT_HOME.resolve() not in path.parents or not path.is_file():raise ValueError('이번 실행 폴더에 있는 실제 입력 파일을 선택하세요.')
    INPUT_HASHES.setdefault((mid,step),{})[str(path)]=file_hash(path);return path
def read_input(mid,step,path):return pd.read_csv(register_input(mid,step,path),keep_default_na=False)
def check_inputs(mid):
    for (key,step),paths in INPUT_HASHES.items():
        if key==mid:
            for path,sha in paths.items():
                if not Path(path).is_file() or file_hash(path)!=sha:raise RuntimeError(f'{mid} STEP {step}에서 읽은 입력이 바뀌었습니다. 그 단계부터 다시 실행하세요.')
def ensure_input(name,frame):
    path=safe_path('inputs/'+name)
    if not path.exists():frame.to_csv(path,index=False,encoding='utf-8-sig')
    return path
def snapshot_text(label,text):
    path=safe_path('versions/OPS07_'+label+'-'+digest(text)[:12]+'.md');path.parent.mkdir(exist_ok=True)
    if not path.exists():path.write_text(text,encoding='utf-8')
    if path.read_text()!=text:raise ValueError('기준 문서 스냅샷이 바뀌었습니다.')
    register_input(*CURRENT,path);return path
def minutes(end,start):
    def parse(value):
        value=str(value).strip()
        if re.fullmatch(r'\d{2}:\d{2}(?::\d{2})?',value):value='2026-10-30 '+value
        return pd.to_datetime(value,errors='coerce')
    a,b=parse(end),parse(start)
    if pd.isna(a) or pd.isna(b):return float('nan')
    return (a-b).total_seconds()/60
def numbers(series,label,maximum=None,integer=False):
    text=series.astype('string').str.strip();blank=text.isna()|text.eq('')
    val=pd.to_numeric(text,errors='coerce')
    finite=val.map(lambda v:pd.notna(v) and math.isfinite(float(v)))
    bad=(~blank)&(~finite|val.lt(0))
    if maximum is not None:bad|=(~blank)&val.gt(maximum)
    if integer:bad|=(~blank)&val.mod(1).ne(0).fillna(True)
    if bad.any():raise ValueError(label+': 음수·무한대·허용 범위 밖·잘못된 숫자를 확인하세요. 행 '+', '.join(map(str,series.index[bad])))
    return val.where(~blank,float('nan')).astype(float)
def flags(series,label):
    word=series.astype('string').str.strip().str.lower();blank=word.isna()|word.eq('')
    if (~(blank|word.isin(['true','false']))).any():raise ValueError(label+': True 또는 False를 입력하세요. 빈칸은 미실시입니다.')
    return word.eq('true').fillna(False).astype(bool)
def exact_grid(frame,id_col,ids,member_col='학습자'):
    required={id_col,member_col}
    if not required.issubset(frame):raise ValueError('입력에 자료·학습자 열이 없습니다.')
    if frame.duplicated([id_col,member_col]).any():raise ValueError('동일한 자료·학습자 기록이 중복됐습니다.')
    actual=set(zip(frame[id_col],frame[member_col]));expected={(i,p) for i in ids for p in PARTICIPANTS}
    if actual!=expected:raise ValueError('정해진 모든 자료와 학습자 조합을 각각 한 번씩 남기세요. 누락·추가·ID 변경을 확인하세요.')
def rating_agreement(frame,claims):
    exact_grid(frame,'자료ID',claims)
    frame=frame.copy();frame['표시값']=numbers(frame['최초 표시'],'최초 표시',6,True)
    if frame['표시값'].eq(0).any():raise ValueError('확인 표시는 정수 1~6 또는 미실시 빈칸입니다.')
    if (frame['표시값'].notna()&frame['이유'].fillna('').astype(str).str.strip().eq('')).any():raise ValueError('최초 표시를 입력한 개인은 그 이유도 남기세요.')
    for _,r in frame.iterrows():
        if r['비교할 문장']!=claims[r['자료ID']]:raise ValueError('개인들이 비교할 문장을 바꾸지 마세요.')
    output=[]
    for key,text in claims.items():
        values=frame.loc[frame['자료ID'].eq(key),'표시값'];n=int(values.notna().sum())
        same=bool(values.nunique()==1) if n==len(PARTICIPANTS) else None
        output.append([key,text,n,len(PARTICIPANTS),same,None,'독립 판정 미실시 또는 미완료' if same is None else '교관 내용 확인 필요'])
    result=make_table('자료ID|비교할 문장|완료 인원|전체 인원|전원 일치|최초 일치율|평가 상태',output)
    if result['전원 일치'].notna().all():result['최초 일치율']=result['전원 일치'].astype(bool).sum()/5*100
    return result
def score_summary(frame):
    exact_grid(frame,'평가 영역',RUBRIC_AREAS)
    frame=frame.copy()
    for col in ['평가A','평가B']:
        frame[col]=numbers(frame[col],col,4,True)
        if frame[col].eq(0).any():raise ValueError('채점은 정수 1~4입니다. 미관찰은 빈칸으로 두세요.')
        note='A의 관찰 기록' if col=='평가A' else 'B의 관찰 기록'
        if note not in frame:raise ValueError('채점표에 관찰 기록 열이 없습니다: '+note)
        if (frame[col].notna()&frame[note].fillna('').astype(str).str.strip().eq('')).any():raise ValueError(col+'에 점수를 썼다면 실제 관찰한 말·행동도 적으세요.')
    rows=[]
    for person in PARTICIPANTS:
        group=frame[frame['학습자'].eq(person)]
        a,b=group['평가A'],group['평가B'];valid=a.notna()&b.notna();n=int(valid.sum())
        same=int((a.eq(b)&valid).sum());complete=n==5
        eligible=bool(complete and a.ge(3).all() and b.ge(3).all())
        rows.append([person,n,5,same,same/n*100 if n else None,complete,eligible,'관찰 미완료' if not complete else '점수 요건만 계산·교관 수행 확인 필요'])
    return make_table('학습자|관찰쌍수|전체영역수|일치쌍수|일치율|전체관찰|점수요건충족|상태',rows)
def comparison_template():
    columns=['자료ID','학습자','자료 구분','전 보류 상태','후 보류 상태','전 질문 묶음','후 질문 묶음','전 확인 질문','후 확인 질문','전 담당','후 담당','전 미답','후 미답','전 초','후 초','전 실시','후 실시','같은 조건','전 누락 질문','후 누락 질문','전 누락 담당','후 누락 담당','전 누락 접수','후 누락 접수']
    rows=[]
    for case in ['B1','B2','B3','B4','B5','A9','B9']:
        for person in PARTICIPANTS:
            row={key:None for key in columns};row.update({'자료ID':case,'학습자':person,'자료 구분':'미측정','전 실시':False,'후 실시':False,'같은 조건':'미확인'});rows.append(row)
    return pd.DataFrame(rows,columns=columns)
def question_codes(text):
    if text is None or pd.isna(text) or not str(text).strip():return None
    values={v.strip() for v in re.split(r'[|,]',str(text)) if v.strip()}
    if not values.issubset({'출처','대상','아군','민간','접수'}):raise ValueError('질문 묶음은 출처·대상·아군·민간·접수를 | 또는 쉼표로 구분하세요.')
    return '|'.join(sorted(values))
def validate_comparison(frame):
    needed=list(comparison_template().columns)
    missing=[c for c in needed if c not in frame]
    if missing:raise ValueError('전후 기록에 필요한 열이 없습니다: '+', '.join(missing))
    exact_grid(frame,'자료ID',['B1','B2','B3','B4','B5','A9','B9']);result=frame.copy()
    for col in ['전 초','후 초']:result[col]=numbers(result[col],col)
    for col in ['전 누락 질문','후 누락 질문','전 누락 담당','후 누락 담당','전 누락 접수','후 누락 접수']:result[col]=numbers(result[col],col,integer=True)
    for side in ['전','후']:
        result[side+' 실시']=flags(result[side+' 실시'],side+' 실시')
        result[side+' 질문 묶음']=result[side+' 질문 묶음'].map(question_codes)
        status=result[side+' 보류 상태'].fillna('').astype(str).str.strip()
        if (~status.isin(['','보류 유지','교육상 검토 종료'])).any():raise ValueError('보류 상태는 보류 유지·교육상 검토 종료 또는 미실시 빈칸입니다.')
        result[side+' 보류 상태']=status.replace('',None)
        if (result[side+' 실시']&(status.eq('')|result[side+' 질문 묶음'].isna())).any():raise ValueError('실시했다고 표시한 개인 기록에는 보류 상태와 질문 묶음을 적으세요.')
        for col in [side+' 확인 질문',side+' 담당',side+' 미답']:
            if (result[side+' 실시']&result[col].fillna('').astype(str).str.strip().eq('')).any():raise ValueError('실시한 기록에는 '+col+'를 적으세요. 누락을 관찰했다면 미작성·미지정이라고 분명히 씁니다.')
    return result
def comparison_summary(frame):
    frame=validate_comparison(frame);rows=[]
    for case in ['B1','B2','B3','B4','B5','A9','B9']:
        group=frame[frame['자료ID'].eq(case)];states=[];counts=[]
        for side in ['전','후']:
            done=group[side+' 실시'];n=int(done.sum());counts.append(n)
            states.append(bool(group[side+' 보류 상태'].nunique()==1 and group[side+' 질문 묶음'].nunique()==1) if n==len(PARTICIPANTS) else None)
        paired=group['전 실시']&group['후 실시']&group['전 초'].notna()&group['후 초'].notna()&group['자료 구분'].eq('실측')&group['같은 조건'].eq('확인')
        before=group.loc[paired,'전 초'].mean();after=group.loc[paired,'후 초'].mean()
        pre=group.loc[group['전 실시'],['전 누락 질문','전 누락 담당','전 누락 접수']].sum(min_count=1).sum(min_count=1)
        post=group.loc[group['후 실시'],['후 누락 질문','후 누락 담당','후 누락 접수']].sum(min_count=1).sum(min_count=1)
        rows.append([case,*counts,len(PARTICIPANTS),*states,before,after,int(paired.sum()),pre,post,'미실시·미측정' if not any(counts) else '시점·누락·동일 조건과 반복 학습을 교관이 확인'])
    return make_table('자료ID|전 완료 인원|후 완료 인원|전체 인원|전 전원 일치|후 전원 일치|전 평균초|후 평균초|시간 대응쌍수|전 누락수|후 누락수|비교 상태',rows)
def comparison_totals(frame):
    row={'정해진 사례 수':7}
    for side in ['전','후']:
        values=frame[side+' 전원 일치'];done=values.notna()
        row[side+' 전원 판정 완료 사례']=int(done.sum())
        row[side+' 일치 사례']=int(values[done].astype(bool).sum())
        row[side+' 일치율']=row[side+' 일치 사례']/7*100 if done.all() else None
    row['기록 해석']='미실시·미측정·미답을 일치나 0초로 세지 않음'
    return pd.DataFrame([row])
def evidence_known(text):
    ids={v.strip() for v in str(text).split('·')};return bool(ids) and ids.issubset(set(EVIDENCE_IDS))
def assessment_template():
    rows=[]
    for mission in PROJECT['missions']:
        for number,key in enumerate(mission['evaluationKeys'],1):
            match=re.search(r'<h4>평가 질문</h4><p>(.*?)</p>',key)
            rows.append({'평가ID':mission['id']+f'_Q{number}','평가 질문':html.unescape(re.sub('<[^>]+>','',match[1] if match else key)),'판정':None,'근거 위치':None,'판정 사유':None,'평가자':None,'평가일':None})
    return pd.DataFrame(rows)
def validate_assessment(frame):
    expected=assessment_template()
    for c in expected.columns:
        if c not in frame:raise ValueError('교관 평가표 열이 없습니다: '+c)
    if frame['평가ID'].duplicated().any() or set(frame['평가ID'])!=set(expected['평가ID']):raise ValueError('스무 평가 항목을 각각 한 번씩 남기세요.')
    questions=expected.set_index('평가ID')['평가 질문'].to_dict()
    if any(r['평가 질문']!=questions[r['평가ID']] for _,r in frame.iterrows()):raise ValueError('현재 OPS07의 평가 질문을 바꾸지 마세요.')
    result=frame.copy();result['판정']=result['판정'].fillna('').astype(str).str.strip().replace('','미평가')
    if (~result['판정'].isin(['PASS','FAIL','미평가'])).any():raise ValueError('판정은 PASS·FAIL·미평가 중 하나로 씁니다. 미완료를 완료로 찾는 부분 문자열 검사는 사용하지 않습니다.')
    graded=result['판정'].isin(['PASS','FAIL'])
    for col in ['근거 위치','판정 사유','평가자','평가일']:
        if (graded&result[col].fillna('').astype(str).str.strip().eq('')).any():raise ValueError('평가한 항목에는 '+col+'를 남기세요. 자동 합격 기록을 만들지 않습니다.')
    return result

def markdown_table(frame):
    def cell(v):return '' if clean(v) is None else str(v).replace('|','/').replace('\n','<br>')
    return '| '+' | '.join(map(str,frame.columns))+' |\n| '+' | '.join(['---']*len(frame.columns))+' |\n'+'\n'.join('| '+' | '.join(cell(v) for v in row)+' |' for row in frame.itertuples(index=False,name=None))
def export_document(mid):
    path=safe_path(DOCUMENT_PATHS[mid]);path.parent.mkdir(parents=True,exist_ok=True)
    title=PROJECT['missions'][int(mid[1:])-1]['title']
    lines=['# OPS07 '+title,'교육용 '+('교관 작성 예시' if ROLE=='solution' else '학습자 작성 초안')+'이며 실제 수행·심의·운용 승인 문서가 아니다.']
    if mid=='M10':
        def opt_table(m, s, fallback):
            if m in RUN and s in RUN[m]:
                return markdown_table(step_table(m, s))
            return fallback
        for _,r in step_table('M10',1).iterrows():lines += ['## '+str(r['장'])+' '+str(r['제목']),str(r['본문']),'조항: '+str(r['관련 조항'])]
        lines+=['## 부록 A 점검표',opt_table('M06',3,'(미션6 STEP 03 산출물 참고)'),
                '## 부록 B 짧은 기록과 상세 기록',opt_table('M06',4,'(미션6 STEP 04 산출물 참고)'),
                '## 부록 C 재확인 절차',opt_table('M04',3,'(미션4 STEP 03 산출물 참고)'),opt_table('M05',1,'(미션5 STEP 01 산출물 참고)'),
                '## 부록 D 개인 수행 평가표',opt_table('M07',1,'(미션7 STEP 01 산출물 참고)'),
                '## 조항별 근거와 미실시 검토',markdown_table(step_table('M10',2)),markdown_table(step_table('M10',3)),
                '## 최종 교육 평가 상태',markdown_table(step_table('M10',5))]
    else:
        for n in range(1,6):
            item=saved_step(mid,n);lines += [f'## STEP {n:02d} '+SPECS[mid][n-1]['title'],markdown_table(item['table']),item['note']]
    path.write_text('\n\n'.join(lines)+'\n',encoding='utf-8');return path

CSS='''*{box-sizing:border-box}body{margin:0;background:#f4f6f8;color:#182630;font:16px/1.6 system-ui,sans-serif}header{padding:20px 28px;background:white;border-bottom:1px solid #cdd6df}h1{font-size:24px}h2{font-size:21px}h3,h4{font-size:17px}.layout{display:grid;grid-template-columns:145px minmax(0,1fr)}nav{background:#e6edf4;padding:16px 9px}nav button{display:block;width:100%;padding:14px 8px;margin-bottom:10px;background:white;color:#203b52;border:1px solid #afbeca;border-radius:5px;text-align:left;font:inherit;cursor:pointer}nav button[aria-selected=true]{background:#203f59;color:white}button:focus-visible{outline:3px solid #df8d22;outline-offset:2px}main{padding:20px 28px;min-width:0}.panel[hidden]{display:none}.scroll{overflow:auto;max-height:650px;background:white;border:1px solid #d5dce3}table{border-collapse:collapse;min-width:100%;font-size:14px}td,th{border:1px solid #d5dce3;padding:9px;text-align:left;vertical-align:top;white-space:pre-wrap}th{background:#eaf0f5}p{max-width:90ch}.note{white-space:pre-wrap}details{margin:15px 0}summary{cursor:pointer}@media(max-width:650px){.layout{grid-template-columns:90px minmax(0,1fr)}main{padding:12px}nav button{font-size:13px}}'''
JS='''const buttons=[...document.querySelectorAll('nav button')];function selectStep(i){buttons.forEach((b,j)=>{b.setAttribute('aria-selected',String(i===j));document.getElementById('step-'+(j+1)).hidden=i!==j;});}buttons.forEach((b,i)=>{b.addEventListener('click',()=>selectStep(i));b.addEventListener('keydown',e=>{if(e.key==='ArrowDown'||e.key==='ArrowUp'){e.preventDefault();const n=(i+(e.key==='ArrowDown'?1:-1)+5)%5;selectStep(n);buttons[n].focus();}});});selectStep(0);'''
def finish_mission(mid):
    location(mid);check_source();check_dependencies(mid);check_inputs(mid)
    if set(RUN.get(mid,{}))!=set(range(1,6)):raise RuntimeError(mid+' STEP 1~5를 모두 다시 실행한 뒤 저장하세요.')
    for n in range(1,6):saved_step(mid,n)
    mission=PROJECT['missions'][int(mid[1:])-1];panels=[];records=[]
    for n in range(1,6):
        item=saved_step(mid,n);lesson=mission['guideSteps'][n-1]
        panels.append(f'<section class="panel" id="step-{n}" role="tabpanel" aria-labelledby="nav-{n}"'+(' hidden' if n!=1 else '')+'>'+
            '<h2>STEP '+str(n).zfill(2)+' '+html.escape(lesson['problem'])+'</h2><h3>이번 단계에 저장한 실제 코드 결과</h3><div class="scroll">'+
            item['table'].to_html(index=False,escape=True,border=0,na_rep='미작성')+'</div><h3>작성자가 남긴 설명</h3><p class="note">'+html.escape(item['note'])+'</p><p>'+html.escape(item['kind'])+'</p>'+
            '<details><summary>학습자 단계 설명</summary>'+lesson['activity']+'</details><details><summary>이 단계의 평가 기준</summary>'+lesson['evalText']+'</details></section>')
        records.append({k:v for k,v in item.items() if k!='table'}|{'step':n,'rows':len(item['table']),'file':Path(item['file']).name})
    nav=''.join(f'<button id="nav-{n}" type="button" role="tab" aria-controls="step-{n}" aria-selected="{str(n==1).lower()}">STEP {n:02d}</button>' for n in range(1,6))
    doc='<!doctype html><html lang="ko"><head><meta charset="utf-8"><meta name="viewport" content="width=device-width,initial-scale=1"><title>OPS07 '+mid+'</title><style>'+CSS+'</style></head><body><header><h1>OPS07 '+mid+' '+html.escape(mission['title'])+'</h1><p>교육용 문서 작성과 기록 도구 · 실제 보고·훈련·심의는 별도 확인</p></header><div class="layout"><nav role="tablist" aria-orientation="vertical" aria-label="다섯 단계">'+nav+'</nav><main>'+''.join(panels)+'</main></div><script>'+JS+'</script></body></html>'
    output=OUTDIR/mid/f'OPS07_{mid}_result.html';output.write_text(doc,encoding='utf-8')
    step_table(mid,5).to_csv(OUTDIR/mid/f'OPS07_{mid}_analysis.csv',index=False,encoding='utf-8-sig')
    document=export_document(mid)
    record={'projectId':PROJECT_ID,'formatVersion':FORMAT_VERSION,'runId':RUN_ID,'mission':mid,'role':ROLE,'sourceHash':digest(DATA[mid]),'projectHash':digest(PROJECT),'steps':records,'document':str(document.relative_to(PROJECT_HOME)),'assessment':'교관 평가 대기','inputs':{str(step):paths for (key,step),paths in INPUT_HASHES.items() if key==mid}}
    (OUTDIR/mid/f'OPS07_{mid}_record.json').write_text(json.dumps(clean(record),ensure_ascii=False,indent=2,allow_nan=False),encoding='utf-8')
    display(HTML('<iframe title="OPS07 '+mid+' 결과" sandbox="allow-scripts" style="width:100%;height:900px;border:0" srcdoc="'+html.escape(doc,quote=True)+'"></iframe>'))
    print('HTML:',output);print('미션 문서:',document);return output
print('OPS07 준비 완료. 현재 문제의 10개 미션·50개 단계와 내장 가상 자료를 사용합니다.')
print('이번 실행 폴더:',PROJECT_HOME)
print('코드는 문서 작성의 선택 보조 도구입니다. 실제 개인 보고·독립 채점·실측·심의의 증거를 자동으로 만들지 않습니다.')
```

## 코드에서 쓸 기본 방법

`make_table("열1|열2", [["값1","값2"]])`는 한 줄 표를 만듭니다. `previous.copy()`는 앞 표의 복사본입니다. `result["새 열"] = 값목록`은 새 열을 붙입니다. `source_table("M01")`은 원자료 복사본입니다. `None`은 미작성 값이며 0이나 확인 완료가 아닙니다.

`participant_form`은 개인별 빈 기록 양식, `write_table`은 부가 표 저장, `ensure_input`은 기존 파일을 덮어쓰지 않는 입력 양식 생성, `read_input`은 파일을 읽고 변경 여부를 추적하는 도구입니다. 단어 “확인”이나 “완료”를 포함한다는 이유로 참으로 판단하지 않습니다.

## M06 STEP 01 세 보고를 문장별로 나눠 관찰·주장·조치를 구분하기

```python
# M06 STEP 01 · 세 보고를 문장별로 나눠 관찰·주장·조치를 구분하기
MID = 'M06'
STEP = 1
previous = begin_step(MID, STEP)
raw = source_table(MID)
print('앞 단계 표:', previous.shape, '원자료:', raw.shape)
display(previous.head(8))

# === ANSWER 시작 ===
result = todo(MID, STEP)
# 필요한 열: 자료ID | 원문 조각 | 분류 | 다시 쓴 문장 | 모르는 내용
# === ANSWER 끝 ===

note = ''
record_step(MID, STEP, result, note)
```

```python
# M06 STEP 02 · 오해를 부르는 표현 여섯 가지를 바꾸기
MID = 'M06'
STEP = 2
previous = begin_step(MID, STEP)
raw = source_table(MID)
print('앞 단계 표:', previous.shape, '원자료:', raw.shape)
display(previous.head(8))

# === ANSWER 시작 ===
result = todo(MID, STEP)
# 필요한 열: 표현 | 문제점 | 바꾼 문장 | 추가 확인
# === ANSWER 끝 ===

note = ''
record_step(MID, STEP, result, note)
```

```python
# M06 STEP 03 · 다섯 영역 점검표와 짧은 기록·상세 기록 양식 만들기
MID = 'M06'
STEP = 3
previous = begin_step(MID, STEP)
raw = source_table(MID)
print('앞 단계 표:', previous.shape, '원자료:', raw.shape)
display(previous.head(8))

# === ANSWER 시작 ===
result = todo(MID, STEP)
# 필요한 열: 영역 | 점검 질문 | 답 | 원문 근거 | 담당
# === ANSWER 끝 ===

note = ''
record_step(MID, STEP, result, note)
```

```python
# M06 STEP 04 · 세 원문을 두 양식에 옮기고 왜곡된 내용을 세기
MID = 'M06'
STEP = 4
previous = begin_step(MID, STEP)
raw = source_table(MID)
print('앞 단계 표:', previous.shape, '원자료:', raw.shape)
display(previous.head(8))

# === ANSWER 시작 ===
result = todo(MID, STEP)
# 필요한 열: 기록 번호 | 자료ID | 원문 | 관찰에 관한 보고 | 주장·판단 | 조치 기록 | 미확인 | 보고 시각 | 관측 시각 | 접수·회신 | 확인 질문·담당
# === ANSWER 끝 ===

note = ''
record_step(MID, STEP, result, note)
```

최초 결과는 지우지 않는다.</p><h4>평가 질문</h4><p>점검표·기록 양식을 만들고 다른 사람이 원문과 질문을 찾았는가</p><h4>평가자가 확인할 위치</h4><p>templates/ops07_aar_fact_action_template.md의 STEP 03, STEP 05과 그 단계의 개인 기록을 확인한다.</p><h4>PASS를 주는 방법</h4><p>아래 단계별 항목을 모두 충족해야 이 평가 질문에 PASS를 준다. 항목 옆에 확인한 표·줄·회차를 적는다. 한 항목이라도 없거나 틀리면 FAIL 사유에 해당 단계와 문장을 적는다. 학습자가 쓴 말이 예시와 똑같을 필요는 없지만 확인할 내용은 모두 있어야 한다.</p><h4>STEP 03 다섯 영역 점검표와 짧은 기록·상세 기록 양식 만들기</h4><p>찾을 결과: 점검 질문 열 개 이상, 짧은 기록·상세 기록 두 양식, 대응표.</p><ul><li><p>다섯 영역에 각각 두 질문 이상 있다.</p></li><li><p>짧은 양식의 필수 항목과 상세 양식의 열 항목이 있다.</p></li><li><p>같은 기록 번호로 두 양식을 연결할 수 있다.</p></li><li><p>예·아니오·미확인을 구분하며 빈칸을 예 또는 완료로 간주하지 않는다.</p></li></ul><h4>STEP 05 짧은 기록을 읽고 상대가 상세 기록을 찾게 하기</h4><p>찾을 결과: 전원 참여 기록과 짧은 기록→상세 기록→원문 찾기 시험 결과.</p><ul><li><p>전원의 실제 참여 내용을 확인할 수 있다.</p></li><li><p>짧은 기록의 번호로 상세 기록과 원문을 찾았다.</p></li><li><p>받는 사람이 미확인 내용·질문·담당을 설명했다.</p></li><li><p>실패한 경우 이유와 수정 후 재시험 결과가 있다.</p></li></ul><h4>이런 결과는 통과시키지 않는다</h4><ul><li><p>점검표를 체크하면 보류 해제로 자동 처리하게 했다.</p></li><li><p>“잘 찾음”만 적고 어떤 기록과 질문을 찾았는지 없다.</p></li></ul><h4>평가 의견 작성 방법</h4><p>“근거 부족”이라고만 쓰지 말고 단계 번호, 충족하지 못한 조건, 제출물에서 실제로 빠지거나 잘못된 부분, 다시 할 일을 적는다. 예를 들어 STEP 03의 첫 조건은 “다섯 영역에 각각 두 질문 이상 있다.”이다. 이 조건에 못 미쳤다면 어떤 줄·항목·참여자 기록이 빠졌는지 지목한다.</p><h4>다시 평가할 때</h4><p>틀리거나 빠진 내용만 구체적으로 알려 준다. 문서 작성 오류는 원자료와 수정본을 대조한다. 실제 수행 오류는 해당 행동을 다시 하게 하고 새 회차 기록으로 확인한다. 최초 결과는 지우지 않는다.</p>
