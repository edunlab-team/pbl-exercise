```python
#@title 🚀 [먼저 실행] 실습 준비하기 { display-mode: "form" }
#@markdown ▶ 왼쪽의 실행 버튼을 누르거나, 메뉴에서 **런타임 → 모두 실행**을 누르세요. 아래 진행 막대가 **100%가** 되면 준비가 끝나요.
FILL_ANSWERS = False   # 학습자용: 빈 상태로 시작해요
from IPython.display import HTML, display

_SETUP_STEPS = ["메신저 · 진행 상태 · 가이드북", "미션 활동 · 가상 AI", "첫 화면"]


def _setup_html(pct, done, msg):
    color = "#16a34a" if pct >= 100 else "#0b57d0"
    items = "".join(f"<li style='margin:3px 0;color:#111827'>{'✅' if i < done else '⏳' if i == done else '▫️'} {name}</li>"
                    for i, name in enumerate(_SETUP_STEPS))
    return (f"<div style='font-family:sans-serif;max-width:640px;border:2px solid {color};border-radius:12px;padding:12px 16px;"
            f"background:#fff;color:#111827'><div style='font-weight:bold;font-size:16px;color:#111827'>🚀 실습 준비 {pct}%</div>"
            f"<div style='height:16px;background:#e3e8ef;border-radius:8px;overflow:hidden;margin:8px 0'>"
            f"<div style='height:100%;width:{pct}%;background:{color}'></div></div>"
            f"<div style='font-size:14px;color:#111827'>{msg}</div>"
            f"<ul style='list-style:none;padding:0;margin:8px 0 0;font-size:13.5px'>{items}</ul></div>")


_setup_bar = display(HTML(_setup_html(2, 0, "준비를 시작합니다...")), display_id=True)


def _setup(pct, done, msg):
    _setup_bar.update(HTML(_setup_html(pct, done, msg)))

# ===== 프로젝트 공통: 메신저 · 진행 상태 · 산출물 · 가이드북 =====
# 로그인이 없음: 개인 정보를 받지 않고 학습자는 모두 '전우님'으로 부름 (2026-11-05 피드백)
# 미션은 처음부터 차례로 하는 것이 원칙. 뒷부분을 미리 보려는 사람을 위해 잠그지 않고 열어 둠 (2026-11-05 피드백)
import base64, html, inspect, io, json, math, os, random, re, textwrap, threading, time, traceback
from datetime import date, datetime, timedelta, timezone
import numpy as np
from IPython.display import JSON
from google.colab import output

# FILL_ANSWERS(선생님용)는 이 셀 맨 위의 체크박스로 정함: True면 모든 활동의 값을 정답으로 채워 둔 상태로 시작함
BOOT_ID = datetime.now().strftime("%H%M%S%f")   # 이 셀을 실행할 때마다 바뀜: 화면이 예전 실행에 머물러 있으면 알아채고 처음 화면으로 돌림

SPEC = {'project': {'id': 'pbl_21b0fe0c',
             'emoji': '🛩️',
             'title': '드론·미사일 실루엣 이미지 분류 학습앱',
             'goal': '업무를 AI에게 말로 설명해 도구와 문서를 만드는 바이브 코딩을 익혀요. 비행체 실루엣을 익히는 학습앱을 만드는 일을 소재로, 프롬프트를 쓰고, AI가 만든 코드를 실행해 확인하고, 고쳐 달라고 다시 요청해요.',
             'welcome_lead': '가상의 실루엣 식별 교육개발반에 바이브 코딩 담당으로 합류했어요. 팀원들의 요청을 프롬프트로 바꿔 AI와 함께 도구와 문서를 만들어 봐요. M1부터 차례로 진행해 보세요.',
             'team_name': '실루엣 식별 교육개발반',
             'my_role': '바이브 코딩 담당',
             'use_label': '바이브 코딩에서는',
             'intro': '팀원들의 요청을 메신저로 받아, 가상 AI에게 프롬프트를 보내 실루엣 학습앱에 쓸 데이터 분석 코드와 문서를 만들어요. 비행체 식별은 소재일 뿐이고, 이번 프로젝트의 중심은 프롬프트를 쓰고 결과를 확인하는 방법이에요.',
             'banned_phrases': ['자동 위협 판정', 'AI가 위협을 판정'],
             'notes': ['데이터, 팀원, 학습 기록, 모델 측정값은 모두 교육용 가상 자료예요. 실제 조직·인물·장비 정보는 쓰지 않아요.',
                       '이번 과정의 AI는 전우님이 고른 문장에 맞춰 준비된 응답을 돌려주는 가상 AI예요. 다만 AI가 만든 코드는 실습 환경에서 실제로 실행돼요. 같은 방법을 실제 AI 도구에 그대로 쓸 수 있어요.',
                       '모델 학습이나 영상 내려받기는 하지 않아요. 가상 모델의 결과를 미리 만들어 두었고, 실루엣은 기본 도형으로 그린 교육용 형상이에요.',
                       'AI의 판정은 학습을 돕는 참고 정보예요. 실제 비행체 식별과 대응은 정식 교육과 조직의 절차를 따라요.'],
             'members': {'gyo': ['고교관', '실루엣 식별 교육개발반', '식별 교관', '#c2410c'],
                         'label': ['나라벨', '실루엣 식별 교육개발반', '데이터·라벨링 담당', '#1d4ed8'],
                         'model': ['한모델', '실루엣 식별 교육개발반', 'AI 개발 담당', '#0f766e'],
                         'jeon': ['전산호', '전산 담당 부서', '전산 담당', '#b91c1c'],
                         'boan': ['배보안', '보안 담당 부서', '보안 담당', '#334155'],
                         'hun': ['윤훈련', '교육 담당 부서', '교육 담당', '#7e22ce'],
                         'park': ['박대표', '시범 운영 참여자', '사용자 대표', '#15803d'],
                         'me': ['전우님', '실루엣 식별 교육개발반', '바이브 코딩 담당', '#334155']},
             'channels': {'notice': '# 프로젝트-공지',
                          'team': '# 개발반',
                          'dm_gyo': '✉ 고교관',
                          'dm_label': '✉ 나라벨',
                          'dm_model': '✉ 한모델',
                          'dm_jeon': '✉ 전산호 (전산 담당)',
                          'dm_boan': '✉ 배보안 (보안 담당)',
                          'dm_hun': '✉ 윤훈련 (교육 담당)',
                          'dm_park': '✉ 박대표 (사용자 대표)'},
             'always_channels': ['notice', 'team'],
             'welcome': [['notice', 'gyo', '{call}, 실루엣 식별 교육개발반에 오신 것을 환영합니다 👋 비행체 실루엣을 익히는 학습앱에 쓸 작은 도구와 문서를 AI와 함께 만들어요.'],
                         ['notice', 'gyo', '왼쪽 메뉴의 M1부터 차례로 진행해 주세요. 코드를 외울 필요는 없어요. AI에게 어떻게 설명할지가 중요해요.'],
                         ['team', 'label', '{call}, 데이터 담당 나라벨이에요. 제가 파일을 넘겨드리면 AI와 함께 표와 문서로 만들어 주세요 🙂']],
             'stages': [{'no': 1, 'name': '식별 단서와 실루엣·지형 데이터', 'subtitle': '프롬프트의 기본', 'missions': ['m1', 'm2']},
                        {'no': 2, 'name': '기준 모델과 딥러닝 분류 모델 비교', 'subtitle': '고치고 쪼개서 요청하기', 'missions': ['m3', 'm4']},
                        {'no': 3, 'name': '분류 점수 신뢰도와 학습 앱 화면', 'subtitle': '결과를 믿기 전에 검증하기', 'missions': ['m5', 'm6']},
                        {'no': 4, 'name': '교관 검토 절차와 배포 환경', 'subtitle': '표현 규칙과 안전한 활용', 'missions': ['m7', 'm8']},
                        {'no': 5, 'name': '시범 운영 사후검토와 운영 인계', 'subtitle': '검토하고 묶어서 인계하기', 'missions': ['m9', 'm10']}],
             'vfiles': {'m1_families.csv': 'family,wing,engine,body,tail,span_m,length_m\n'
                                           'multicopter,none,rotor_4,box,none,0.6,0.6\n'
                                           'fixed_wing,straight,pusher,slim,t_tail,3.0,2.0\n'
                                           'delta_wing,delta,pusher,slim,winglet,2.5,3.5\n'
                                           'cruise_missile,small,jet,tube,cross_fin,2.5,6.0\n'
                                           'ballistic_missile,none,rocket,long_tube,fin_4,1.0,12.0\n'
                                           'bird,flapping,none,oval,fan,1.2,0.5\n',
                        'm1_families_new.csv': 'family,wing,engine,body,tail,span_m,length_m\njet_delta,delta,jet,slim,winglet,3.2,3.8\n',
                        'm2_labels.csv': 'image_id,background,view,rater_a,rater_b\n'
                                         'IMG01,urban,top,Bird,multicopter\n'
                                         'IMG02,field,top,bird ,bird\n'
                                         'IMG03,sky,bottom,cruise_missile,Cruise_missile\n'
                                         'IMG04,field,top,Bird,bird\n'
                                         'IMG05,sky,side,delta_wing,Decoy\n'
                                         'IMG06,sky,top,Bird,multicopter\n'
                                         'IMG07,urban,side,multicopter ,bird \n'
                                         'IMG08,field,top,ballistic_missile,ballistic_missile\n'
                                         'IMG09,field,side,multicopter,multicopter\n'
                                         'IMG10,sky,bottom,cruise_missile,cruise_missile\n'
                                         'IMG11,field,top,Fixed_wing,fixed_wing\n'
                                         'IMG12,urban,bottom,fixed wing,fixed_wing\n'
                                         'IMG13,urban,side,delta_wing ,decoy \n'
                                         'IMG14,urban,bottom,ballistic_missile,ballistic_missile\n'
                                         'IMG15,field,bottom,ballistic_missile,ballistic_missile\n'
                                         'IMG16,sky,top,ballistic_missile,ballistic_missile \n'
                                         'IMG17,field,top,Fixed_wing,Fixed_wing\n'
                                         'IMG18,field,side,bird,multicopter\n'
                                         'IMG19,urban,bottom,cruise_missile,Ballistic_missile\n'
                                         'IMG20,sky,bottom,multicopter,bird\n'
                                         'IMG21,urban,bottom,multicopter,Multicopter\n'
                                         'IMG22,urban,top,fixed_wing,fixed_wing \n'
                                         'IMG23,sky,top,Cruise_missile,cruise_missile\n'
                                         'IMG24,sky,bottom,multicopter,multicopter \n'
                                         'IMG25,urban,top,delta_wing,delta_wing \n'
                                         'IMG26,sky,top,delta_wing,fixed_wing\n'
                                         'IMG27,urban,bottom,multicopter ,multicopter\n'
                                         'IMG28,urban,side,fixed_wing,fixed_wing\n'
                                         'IMG29,sky,top,ballistic_missile,ballistic missile\n'
                                         'IMG30,urban,side,bird,bird\n',
                        'm3_preds.csv': 'image_id,model,distance,polarity,truth,pred\n'
                                        'S001,decision_tree,near,white_hot,ballistic_missile,ballistic_missile\n'
                                        'S001,logistic,near,white_hot,ballistic_missile,ballistic_missile\n'
                                        'S001,knn,near,white_hot,ballistic_missile,ballistic_missile\n'
                                        'S002,decision_tree,near,white_hot,cruise_missile,cruise_missile\n'
                                        'S002,logistic,near,white_hot,cruise_missile,cruise_missile\n'
                                        'S002,knn,near,white_hot,cruise_missile,cruise_missile\n'
                                        'S003,decision_tree,near,white_hot,fixed_wing,delta_wing\n'
                                        'S003,logistic,near,white_hot,fixed_wing,fixed_wing\n'
                                        'S003,knn,near,white_hot,fixed_wing,fixed_wing\n'
                                        'S004,decision_tree,near,white_hot,delta_wing,fixed_wing\n'
                                        'S004,logistic,near,white_hot,delta_wing,delta_wing\n'
                                        'S004,knn,near,white_hot,delta_wing,delta_wing\n'
                                        'S005,decision_tree,near,white_hot,bird,bird\n'
                                        'S005,logistic,near,white_hot,bird,bird\n'
                                        'S005,knn,near,white_hot,bird,bird\n'
                                        'S006,decision_tree,near,white_hot,cruise_missile,cruise_missile\n'
                                        'S006,logistic,near,white_hot,cruise_missile,cruise_missile\n'
                                        'S006,knn,near,white_hot,cruise_missile,cruise_missile\n'
                                        'S007,decision_tree,near,white_hot,bird,bird\n'
                                        'S007,logistic,near,white_hot,bird,bird\n'
                                        'S007,knn,near,white_hot,bird,bird\n'
                                        'S008,decision_tree,near,white_hot,ballistic_missile,ballistic_missile\n'
                                        'S008,logistic,near,white_hot,ballistic_missile,ballistic_missile\n'
                                        'S008,knn,near,white_hot,ballistic_missile,ballistic_missile\n'
                                        'S009,decision_tree,near,white_hot,delta_wing,fixed_wing\n'
                                        'S009,logistic,near,white_hot,delta_wing,delta_wing\n'
                                        'S009,knn,near,white_hot,delta_wing,delta_wing\n'
                                        'S010,decision_tree,near,white_hot,multicopter,multicopter\n'
                                        'S010,logistic,near,white_hot,multicopter,multicopter\n'
                                        'S010,knn,near,white_hot,multicopter,multicopter\n'
                                        'S011,decision_tree,near,white_hot,fixed_wing,fixed_wing\n'
                                        'S011,logistic,near,white_hot,fixed_wing,fixed_wing\n'
                                        'S011,knn,near,white_hot,fixed_wing,fixed_wing\n'
                                        'S012,decision_tree,near,white_hot,bird,bird\n'
                                        'S012,logistic,near,white_hot,bird,bird\n'
                                        'S012,knn,near,white_hot,bird,bird\n'
                                        'S013,decision_tree,near,white_hot,multicopter,multicopter\n'
                                        'S013,logistic,near,white_hot,multicopter,multicopter\n'
                                        'S013,knn,near,white_hot,multicopter,multicopter\n'
                                        'S014,decision_tree,near,white_hot,bird,bird\n'
                                        'S014,logistic,near,white_hot,bird,bird\n'
                                        'S014,knn,near,white_hot,bird,bird\n'
                                        'S015,decision_tree,near,white_hot,multicopter,multicopter\n'
                                        'S015,logistic,near,white_hot,multicopter,multicopter\n'
                                        'S015,knn,near,white_hot,multicopter,multicopter\n'
                                        'S016,decision_tree,near,white_hot,delta_wing,delta_wing\n'
                                        'S016,logistic,near,white_hot,delta_wing,delta_wing\n'
                                        'S016,knn,near,white_hot,delta_wing,delta_wing\n'
                                        'S017,decision_tree,near,white_hot,cruise_missile,cruise_missile\n'
                                        'S017,logistic,near,white_hot,cruise_missile,cruise_missile\n'
                                        'S017,knn,near,white_hot,cruise_missile,cruise_missile\n'
                                        'S018,decision_tree,near,white_hot,cruise_missile,cruise_missile\n'
                                        'S018,logistic,near,white_hot,cruise_missile,cruise_missile\n'
                                        'S018,knn,near,white_hot,cruise_missile,cruise_missile\n'
                                        'S019,decision_tree,near,white_hot,cruise_missile,cruise_missile\n'
                                        'S019,logistic,near,white_hot,cruise_missile,cruise_missile\n'
                                        'S019,knn,near,white_hot,cruise_missile,cruise_missile\n'
                                        'S020,decision_tree,near,white_hot,bird,bird\n'
                                        'S020,logistic,near,white_hot,bird,bird\n'
                                        'S020,knn,near,white_hot,bird,bird\n'
                                        'S021,decision_tree,near,white_hot,cruise_missile,cruise_missile\n'
                                        'S021,logistic,near,white_hot,cruise_missile,cruise_missile\n'
                                        'S021,knn,near,white_hot,cruise_missile,cruise_missile\n'
                                        'S022,decision_tree,near,white_hot,fixed_wing,delta_wing\n'
                                        'S022,logistic,near,white_hot,fixed_wing,fixed_wing\n'
                                        'S022,knn,near,white_hot,fixed_wing,fixed_wing\n'
                                        'S023,decision_tree,near,white_hot,bird,bird\n'
                                        'S023,logistic,near,white_hot,bird,bird\n'
                                        'S023,knn,near,white_hot,bird,bird\n'
                                        'S024,decision_tree,near,white_hot,cruise_missile,cruise_missile\n'
                                        'S024,logistic,near,white_hot,cruise_missile,delta_wing\n'
                                        'S024,knn,near,white_hot,cruise_missile,cruise_missile\n'
                                        'S025,decision_tree,near,white_hot,cruise_missile,cruise_missile\n'
                                        'S025,logistic,near,white_hot,cruise_missile,delta_wing\n'
                                        'S025,knn,near,white_hot,cruise_missile,cruise_missile\n'
                                        'S026,decision_tree,near,white_hot,multicopter,multicopter\n'
                                        'S026,logistic,near,white_hot,multicopter,multicopter\n'
                                        'S026,knn,near,white_hot,multicopter,multicopter\n'
                                        'S027,decision_tree,near,white_hot,fixed_wing,fixed_wing\n'
                                        'S027,logistic,near,white_hot,fixed_wing,fixed_wing\n'
                                        'S027,knn,near,white_hot,fixed_wing,fixed_wing\n'
                                        'S028,decision_tree,near,white_hot,ballistic_missile,cruise_missile\n'
                                        'S028,logistic,near,white_hot,ballistic_missile,ballistic_missile\n'
                                        'S028,knn,near,white_hot,ballistic_missile,ballistic_missile\n'
                                        'S029,decision_tree,near,white_hot,bird,multicopter\n'
                                        'S029,logistic,near,white_hot,bird,bird\n'
                                        'S029,knn,near,white_hot,bird,bird\n'
                                        'S030,decision_tree,near,white_hot,delta_wing,fixed_wing\n'
                                        'S030,logistic,near,white_hot,delta_wing,delta_wing\n'
                                        'S030,knn,near,white_hot,delta_wing,delta_wing\n'
                                        'S031,decision_tree,near,white_hot,ballistic_missile,ballistic_missile\n'
                                        'S031,logistic,near,white_hot,ballistic_missile,ballistic_missile\n'
                                        'S031,knn,near,white_hot,ballistic_missile,ballistic_missile\n'
                                        'S032,decision_tree,near,white_hot,bird,bird\n'
                                        'S032,logistic,near,white_hot,bird,bird\n'
                                        'S032,knn,near,white_hot,bird,bird\n'
                                        'S033,decision_tree,near,white_hot,ballistic_missile,ballistic_missile\n'
                                        'S033,logistic,near,white_hot,ballistic_missile,ballistic_missile\n'
                                        'S033,knn,near,white_hot,ballistic_missile,ballistic_missile\n'
                                        'S034,decision_tree,near,white_hot,fixed_wing,delta_wing\n'
                                        'S034,logistic,near,white_hot,fixed_wing,fixed_wing\n'
                                        'S034,knn,near,white_hot,fixed_wing,fixed_wing\n'
                                        'S035,decision_tree,near,white_hot,multicopter,multicopter\n'
                                        'S035,logistic,near,white_hot,multicopter,bird\n'
                                        'S035,knn,near,white_hot,multicopter,multicopter\n'
                                        'S036,decision_tree,near,white_hot,multicopter,multicopter\n'
                                        'S036,logistic,near,white_hot,multicopter,multicopter\n'
                                        'S036,knn,near,white_hot,multicopter,multicopter\n'
                                        'S037,decision_tree,near,white_hot,multicopter,multicopter\n'
                                        'S037,logistic,near,white_hot,multicopter,multicopter\n'
                                        'S037,knn,near,white_hot,multicopter,multicopter\n'
                                        'S038,decision_tree,near,white_hot,multicopter,multicopter\n'
                                        'S038,logistic,near,white_hot,multicopter,multicopter\n'
                                        'S038,knn,near,white_hot,multicopter,multicopter\n'
                                        'S039,decision_tree,near,white_hot,bird,bird\n'
                                        'S039,logistic,near,white_hot,bird,bird\n'
                                        'S039,knn,near,white_hot,bird,bird\n'
                                        'S040,decision_tree,near,white_hot,bird,bird\n'
                                        'S040,logistic,near,white_hot,bird,bird\n'
                                        'S040,knn,near,white_hot,bird,bird\n'
                                        'S041,decision_tree,near,black_hot,bird,multicopter\n'
                                        'S041,logistic,near,black_hot,bird,bird\n'
                                        'S041,knn,near,black_hot,bird,bird\n'
                                        'S042,decision_tree,near,black_hot,delta_wing,delta_wing\n'
                                        'S042,logistic,near,black_hot,delta_wing,delta_wing\n'
                                        'S042,knn,near,black_hot,delta_wing,delta_wing\n'
                                        'S043,decision_tree,near,black_hot,delta_wing,delta_wing\n'
                                        'S043,logistic,near,black_hot,delta_wing,delta_wing\n'
                                        'S043,knn,near,black_hot,delta_wing,delta_wing\n'
                                        'S044,decision_tree,near,black_hot,delta_wing,delta_wing\n'
                                        'S044,logistic,near,black_hot,delta_wing,delta_wing\n'
                                        'S044,knn,near,black_hot,delta_wing,delta_wing\n'
                                        'S045,decision_tree,near,black_hot,fixed_wing,fixed_wing\n'
                                        'S045,logistic,near,black_hot,fixed_wing,fixed_wing\n'
                                        'S045,knn,near,black_hot,fixed_wing,fixed_wing\n'
                                        'S046,decision_tree,near,black_hot,fixed_wing,fixed_wing\n'
                                        'S046,logistic,near,black_hot,fixed_wing,fixed_wing\n'
                                        'S046,knn,near,black_hot,fixed_wing,fixed_wing\n'
                                        'S047,decision_tree,near,black_hot,bird,bird\n'
                                        'S047,logistic,near,black_hot,bird,bird\n'
                                        'S047,knn,near,black_hot,bird,bird\n'
                                        'S048,decision_tree,near,black_hot,fixed_wing,fixed_wing\n'
                                        'S048,logistic,near,black_hot,fixed_wing,fixed_wing\n'
                                        'S048,knn,near,black_hot,fixed_wing,fixed_wing\n'
                                        'S049,decision_tree,near,black_hot,delta_wing,delta_wing\n'
                                        'S049,logistic,near,black_hot,delta_wing,delta_wing\n'
                                        'S049,knn,near,black_hot,delta_wing,delta_wing\n'
                                        'S050,decision_tree,near,black_hot,ballistic_missile,ballistic_missile\n'
                                        'S050,logistic,near,black_hot,ballistic_missile,ballistic_missile\n'
                                        'S050,knn,near,black_hot,ballistic_missile,ballistic_missile\n'
                                        'S051,decision_tree,near,black_hot,fixed_wing,fixed_wing\n'
                                        'S051,logistic,near,black_hot,fixed_wing,fixed_wing\n'
                                        'S051,knn,near,black_hot,fixed_wing,fixed_wing\n'
                                        'S052,decision_tree,near,black_hot,multicopter,multicopter\n'
                                        'S052,logistic,near,black_hot,multicopter,multicopter\n'
                                        'S052,knn,near,black_hot,multicopter,bird\n'
                                        'S053,decision_tree,near,black_hot,bird,multicopter\n'
                                        'S053,logistic,near,black_hot,bird,bird\n'
                                        'S053,knn,near,black_hot,bird,bird\n'
                                        'S054,decision_tree,near,black_hot,cruise_missile,cruise_missile\n'
                                        'S054,logistic,near,black_hot,cruise_missile,cruise_missile\n'
                                        'S054,knn,near,black_hot,cruise_missile,cruise_missile\n'
                                        'S055,decision_tree,near,black_hot,ballistic_missile,ballistic_missile\n'
                                        'S055,logistic,near,black_hot,ballistic_missile,cruise_missile\n'
                                        'S055,knn,near,black_hot,ballistic_missile,ballistic_missile\n'
                                        'S056,decision_tree,near,black_hot,multicopter,bird\n'
                                        'S056,logistic,near,black_hot,multicopter,multicopter\n'
                                        'S056,knn,near,black_hot,multicopter,multicopter\n'
                                        'S057,decision_tree,near,black_hot,cruise_missile,delta_wing\n'
                                        'S057,logistic,near,black_hot,cruise_missile,cruise_missile\n'
                                        'S057,knn,near,black_hot,cruise_missile,cruise_missile\n'
                                        'S058,decision_tree,near,black_hot,cruise_missile,delta_wing\n'
                                        'S058,logistic,near,black_hot,cruise_missile,cruise_missile\n'
                                        'S058,knn,near,black_hot,cruise_missile,cruise_missile\n'
                                        'S059,decision_tree,near,black_hot,bird,bird\n'
                                        'S059,logistic,near,black_hot,bird,bird\n'
                                        'S059,knn,near,black_hot,bird,bird\n'
                                        'S060,decision_tree,near,black_hot,bird,bird\n'
                                        'S060,logistic,near,black_hot,bird,bird\n'
                                        'S060,knn,near,black_hot,bird,bird\n'
                                        'S061,decision_tree,mid,white_hot,fixed_wing,fixed_wing\n'
                                        'S061,logistic,mid,white_hot,fixed_wing,delta_wing\n'
                                        'S061,knn,mid,white_hot,fixed_wing,fixed_wing\n'
                                        'S062,decision_tree,mid,white_hot,fixed_wing,fixed_wing\n'
                                        'S062,logistic,mid,white_hot,fixed_wing,fixed_wing\n'
                                        'S062,knn,mid,white_hot,fixed_wing,fixed_wing\n'
                                        'S063,decision_tree,mid,white_hot,cruise_missile,cruise_missile\n'
                                        'S063,logistic,mid,white_hot,cruise_missile,cruise_missile\n'
                                        'S063,knn,mid,white_hot,cruise_missile,cruise_missile\n'
                                        'S064,decision_tree,mid,white_hot,ballistic_missile,ballistic_missile\n'
                                        'S064,logistic,mid,white_hot,ballistic_missile,ballistic_missile\n'
                                        'S064,knn,mid,white_hot,ballistic_missile,cruise_missile\n'
                                        'S065,decision_tree,mid,white_hot,ballistic_missile,ballistic_missile\n'
                                        'S065,logistic,mid,white_hot,ballistic_missile,cruise_missile\n'
                                        'S065,knn,mid,white_hot,ballistic_missile,cruise_missile\n'
                                        'S066,decision_tree,mid,white_hot,ballistic_missile,ballistic_missile\n'
                                        'S066,logistic,mid,white_hot,ballistic_missile,ballistic_missile\n'
                                        'S066,knn,mid,white_hot,ballistic_missile,ballistic_missile\n'
                                        'S067,decision_tree,mid,white_hot,cruise_missile,cruise_missile\n'
                                        'S067,logistic,mid,white_hot,cruise_missile,cruise_missile\n'
                                        'S067,knn,mid,white_hot,cruise_missile,cruise_missile\n'
                                        'S068,decision_tree,mid,white_hot,ballistic_missile,cruise_missile\n'
                                        'S068,logistic,mid,white_hot,ballistic_missile,ballistic_missile\n'
                                        'S068,knn,mid,white_hot,ballistic_missile,ballistic_missile\n'
                                        'S069,decision_tree,mid,white_hot,bird,bird\n'
                                        'S069,logistic,mid,white_hot,bird,bird\n'
                                        'S069,knn,mid,white_hot,bird,bird\n'
                                        'S070,decision_tree,mid,white_hot,cruise_missile,cruise_missile\n'
                                        'S070,logistic,mid,white_hot,cruise_missile,delta_wing\n'
                                        'S070,knn,mid,white_hot,cruise_missile,delta_wing\n'
                                        'S071,decision_tree,mid,white_hot,delta_wing,fixed_wing\n'
                                        'S071,logistic,mid,white_hot,delta_wing,delta_wing\n'
                                        'S071,knn,mid,white_hot,delta_wing,delta_wing\n'
                                        'S072,decision_tree,mid,white_hot,fixed_wing,fixed_wing\n'
                                        'S072,logistic,mid,white_hot,fixed_wing,fixed_wing\n'
                                        'S072,knn,mid,white_hot,fixed_wing,fixed_wing\n'
                                        'S073,decision_tree,mid,white_hot,delta_wing,delta_wing\n'
                                        'S073,logistic,mid,white_hot,delta_wing,delta_wing\n'
                                        'S073,knn,mid,white_hot,delta_wing,delta_wing\n'
                                        'S074,decision_tree,mid,white_hot,delta_wing,delta_wing\n'
                                        'S074,logistic,mid,white_hot,delta_wing,delta_wing\n'
                                        'S074,knn,mid,white_hot,delta_wing,delta_wing\n'
                                        'S075,decision_tree,mid,white_hot,bird,bird\n'
                                        'S075,logistic,mid,white_hot,bird,multicopter\n'
                                        'S075,knn,mid,white_hot,bird,bird\n'
                                        'S076,decision_tree,mid,white_hot,bird,bird\n'
                                        'S076,logistic,mid,white_hot,bird,bird\n'
                                        'S076,knn,mid,white_hot,bird,bird\n'
                                        'S077,decision_tree,mid,white_hot,fixed_wing,delta_wing\n'
                                        'S077,logistic,mid,white_hot,fixed_wing,delta_wing\n'
                                        'S077,knn,mid,white_hot,fixed_wing,fixed_wing\n'
                                        'S078,decision_tree,mid,white_hot,bird,bird\n'
                                        'S078,logistic,mid,white_hot,bird,multicopter\n'
                                        'S078,knn,mid,white_hot,bird,bird\n'
                                        'S079,decision_tree,mid,white_hot,delta_wing,fixed_wing\n'
                                        'S079,logistic,mid,white_hot,delta_wing,delta_wing\n'
                                        'S079,knn,mid,white_hot,delta_wing,delta_wing\n'
                                        'S080,decision_tree,mid,white_hot,delta_wing,fixed_wing\n'
                                        'S080,logistic,mid,white_hot,delta_wing,delta_wing\n'
                                        'S080,knn,mid,white_hot,delta_wing,fixed_wing\n'
                                        'S081,decision_tree,mid,white_hot,ballistic_missile,ballistic_missile\n'
                                        'S081,logistic,mid,white_hot,ballistic_missile,ballistic_missile\n'
                                        'S081,knn,mid,white_hot,ballistic_missile,ballistic_missile\n'
                                        'S082,decision_tree,mid,white_hot,bird,bird\n'
                                        'S082,logistic,mid,white_hot,bird,bird\n'
                                        'S082,knn,mid,white_hot,bird,bird\n'
                                        'S083,decision_tree,mid,white_hot,delta_wing,delta_wing\n'
                                        'S083,logistic,mid,white_hot,delta_wing,delta_wing\n'
                                        'S083,knn,mid,white_hot,delta_wing,delta_wing\n'
                                        'S084,decision_tree,mid,white_hot,fixed_wing,fixed_wing\n'
                                        'S084,logistic,mid,white_hot,fixed_wing,fixed_wing\n'
                                        'S084,knn,mid,white_hot,fixed_wing,fixed_wing\n'
                                        'S085,decision_tree,mid,white_hot,delta_wing,delta_wing\n'
                                        'S085,logistic,mid,white_hot,delta_wing,delta_wing\n'
                                        'S085,knn,mid,white_hot,delta_wing,delta_wing\n'
                                        'S086,decision_tree,mid,white_hot,bird,bird\n'
                                        'S086,logistic,mid,white_hot,bird,bird\n'
                                        'S086,knn,mid,white_hot,bird,bird\n'
                                        'S087,decision_tree,mid,white_hot,multicopter,multicopter\n'
                                        'S087,logistic,mid,white_hot,multicopter,multicopter\n'
                                        'S087,knn,mid,white_hot,multicopter,multicopter\n'
                                        'S088,decision_tree,mid,white_hot,fixed_wing,delta_wing\n'
                                        'S088,logistic,mid,white_hot,fixed_wing,delta_wing\n'
                                        'S088,knn,mid,white_hot,fixed_wing,fixed_wing\n'
                                        'S089,decision_tree,mid,white_hot,cruise_missile,cruise_missile\n'
                                        'S089,logistic,mid,white_hot,cruise_missile,delta_wing\n'
                                        'S089,knn,mid,white_hot,cruise_missile,delta_wing\n'
                                        'S090,decision_tree,mid,white_hot,ballistic_missile,ballistic_missile\n'
                                        'S090,logistic,mid,white_hot,ballistic_missile,ballistic_missile\n'
                                        'S090,knn,mid,white_hot,ballistic_missile,ballistic_missile\n'
                                        'S091,decision_tree,mid,white_hot,bird,bird\n'
                                        'S091,logistic,mid,white_hot,bird,bird\n'
                                        'S091,knn,mid,white_hot,bird,bird\n'
                                        'S092,decision_tree,mid,white_hot,ballistic_missile,ballistic_missile\n'
                                        'S092,logistic,mid,white_hot,ballistic_missile,ballistic_missile\n'
                                        'S092,knn,mid,white_hot,ballistic_missile,ballistic_missile\n'
                                        'S093,decision_tree,mid,white_hot,ballistic_missile,ballistic_missile\n'
                                        'S093,logistic,mid,white_hot,ballistic_missile,ballistic_missile\n'
                                        'S093,knn,mid,white_hot,ballistic_missile,ballistic_missile\n'
                                        'S094,decision_tree,mid,white_hot,delta_wing,delta_wing\n'
                                        'S094,logistic,mid,white_hot,delta_wing,delta_wing\n'
                                        'S094,knn,mid,white_hot,delta_wing,delta_wing\n'
                                        'S095,decision_tree,mid,white_hot,ballistic_missile,ballistic_missile\n'
                                        'S095,logistic,mid,white_hot,ballistic_missile,ballistic_missile\n'
                                        'S095,knn,mid,white_hot,ballistic_missile,ballistic_missile\n'
                                        'S096,decision_tree,mid,white_hot,fixed_wing,fixed_wing\n'
                                        'S096,logistic,mid,white_hot,fixed_wing,fixed_wing\n'
                                        'S096,knn,mid,white_hot,fixed_wing,fixed_wing\n'
                                        'S097,decision_tree,mid,white_hot,bird,bird\n'
                                        'S097,logistic,mid,white_hot,bird,bird\n'
                                        'S097,knn,mid,white_hot,bird,bird\n'
                                        'S098,decision_tree,mid,white_hot,ballistic_missile,ballistic_missile\n'
                                        'S098,logistic,mid,white_hot,ballistic_missile,ballistic_missile\n'
                                        'S098,knn,mid,white_hot,ballistic_missile,ballistic_missile\n'
                                        'S099,decision_tree,mid,white_hot,multicopter,bird\n'
                                        'S099,logistic,mid,white_hot,multicopter,multicopter\n'
                                        'S099,knn,mid,white_hot,multicopter,multicopter\n'
                                        'S100,decision_tree,mid,white_hot,multicopter,bird\n'
                                        'S100,logistic,mid,white_hot,multicopter,bird\n'
                                        'S100,knn,mid,white_hot,multicopter,multicopter\n'
                                        'S101,decision_tree,mid,black_hot,cruise_missile,delta_wing\n'
                                        'S101,logistic,mid,black_hot,cruise_missile,delta_wing\n'
                                        'S101,knn,mid,black_hot,cruise_missile,cruise_missile\n'
                                        'S102,decision_tree,mid,black_hot,delta_wing,fixed_wing\n'
                                        'S102,logistic,mid,black_hot,delta_wing,delta_wing\n'
                                        'S102,knn,mid,black_hot,delta_wing,delta_wing\n'
                                        'S103,decision_tree,mid,black_hot,fixed_wing,fixed_wing\n'
                                        'S103,logistic,mid,black_hot,fixed_wing,fixed_wing\n'
                                        'S103,knn,mid,black_hot,fixed_wing,fixed_wing\n'
                                        'S104,decision_tree,mid,black_hot,ballistic_missile,cruise_missile\n'
                                        'S104,logistic,mid,black_hot,ballistic_missile,cruise_missile\n'
                                        'S104,knn,mid,black_hot,ballistic_missile,cruise_missile\n'
                                        'S105,decision_tree,mid,black_hot,fixed_wing,fixed_wing\n'
                                        'S105,logistic,mid,black_hot,fixed_wing,fixed_wing\n'
                                        'S105,knn,mid,black_hot,fixed_wing,fixed_wing\n'
                                        'S106,decision_tree,mid,black_hot,cruise_missile,cruise_missile\n'
                                        'S106,logistic,mid,black_hot,cruise_missile,cruise_missile\n'
                                        'S106,knn,mid,black_hot,cruise_missile,cruise_missile\n'
                                        'S107,decision_tree,mid,black_hot,cruise_missile,cruise_missile\n'
                                        'S107,logistic,mid,black_hot,cruise_missile,cruise_missile\n'
                                        'S107,knn,mid,black_hot,cruise_missile,cruise_missile\n'
                                        'S108,decision_tree,mid,black_hot,delta_wing,delta_wing\n'
                                        'S108,logistic,mid,black_hot,delta_wing,delta_wing\n'
                                        'S108,knn,mid,black_hot,delta_wing,fixed_wing\n'
                                        'S109,decision_tree,mid,black_hot,multicopter,bird\n'
                                        'S109,logistic,mid,black_hot,multicopter,multicopter\n'
                                        'S109,knn,mid,black_hot,multicopter,multicopter\n'
                                        'S110,decision_tree,mid,black_hot,ballistic_missile,cruise_missile\n'
                                        'S110,logistic,mid,black_hot,ballistic_missile,cruise_missile\n'
                                        'S110,knn,mid,black_hot,ballistic_missile,ballistic_missile\n'
                                        'S111,decision_tree,mid,black_hot,ballistic_missile,cruise_missile\n'
                                        'S111,logistic,mid,black_hot,ballistic_missile,ballistic_missile\n'
                                        'S111,knn,mid,black_hot,ballistic_missile,ballistic_missile\n'
                                        'S112,decision_tree,mid,black_hot,delta_wing,delta_wing\n'
                                        'S112,logistic,mid,black_hot,delta_wing,delta_wing\n'
                                        'S112,knn,mid,black_hot,delta_wing,fixed_wing\n'
                                        'S113,decision_tree,mid,black_hot,fixed_wing,fixed_wing\n'
                                        'S113,logistic,mid,black_hot,fixed_wing,fixed_wing\n'
                                        'S113,knn,mid,black_hot,fixed_wing,fixed_wing\n'
                                        'S114,decision_tree,mid,black_hot,multicopter,multicopter\n'
                                        'S114,logistic,mid,black_hot,multicopter,bird\n'
                                        'S114,knn,mid,black_hot,multicopter,multicopter\n'
                                        'S115,decision_tree,mid,black_hot,multicopter,bird\n'
                                        'S115,logistic,mid,black_hot,multicopter,bird\n'
                                        'S115,knn,mid,black_hot,multicopter,multicopter\n'
                                        'S116,decision_tree,mid,black_hot,cruise_missile,delta_wing\n'
                                        'S116,logistic,mid,black_hot,cruise_missile,delta_wing\n'
                                        'S116,knn,mid,black_hot,cruise_missile,cruise_missile\n'
                                        'S117,decision_tree,mid,black_hot,cruise_missile,delta_wing\n'
                                        'S117,logistic,mid,black_hot,cruise_missile,cruise_missile\n'
                                        'S117,knn,mid,black_hot,cruise_missile,delta_wing\n'
                                        'S118,decision_tree,mid,black_hot,ballistic_missile,cruise_missile\n'
                                        'S118,logistic,mid,black_hot,ballistic_missile,ballistic_missile\n'
                                        'S118,knn,mid,black_hot,ballistic_missile,cruise_missile\n'
                                        'S119,decision_tree,mid,black_hot,cruise_missile,cruise_missile\n'
                                        'S119,logistic,mid,black_hot,cruise_missile,cruise_missile\n'
                                        'S119,knn,mid,black_hot,cruise_missile,cruise_missile\n'
                                        'S120,decision_tree,mid,black_hot,bird,multicopter\n'
                                        'S120,logistic,mid,black_hot,bird,multicopter\n'
                                        'S120,knn,mid,black_hot,bird,bird\n'
                                        'S121,decision_tree,far,white_hot,cruise_missile,cruise_missile\n'
                                        'S121,logistic,far,white_hot,cruise_missile,cruise_missile\n'
                                        'S121,knn,far,white_hot,cruise_missile,cruise_missile\n'
                                        'S122,decision_tree,far,white_hot,delta_wing,delta_wing\n'
                                        'S122,logistic,far,white_hot,delta_wing,delta_wing\n'
                                        'S122,knn,far,white_hot,delta_wing,delta_wing\n'
                                        'S123,decision_tree,far,white_hot,bird,multicopter\n'
                                        'S123,logistic,far,white_hot,bird,multicopter\n'
                                        'S123,knn,far,white_hot,bird,bird\n'
                                        'S124,decision_tree,far,white_hot,ballistic_missile,cruise_missile\n'
                                        'S124,logistic,far,white_hot,ballistic_missile,ballistic_missile\n'
                                        'S124,knn,far,white_hot,ballistic_missile,cruise_missile\n'
                                        'S125,decision_tree,far,white_hot,multicopter,bird\n'
                                        'S125,logistic,far,white_hot,multicopter,multicopter\n'
                                        'S125,knn,far,white_hot,multicopter,bird\n'
                                        'S126,decision_tree,far,white_hot,ballistic_missile,cruise_missile\n'
                                        'S126,logistic,far,white_hot,ballistic_missile,cruise_missile\n'
                                        'S126,knn,far,white_hot,ballistic_missile,cruise_missile\n'
                                        'S127,decision_tree,far,white_hot,delta_wing,fixed_wing\n'
                                        'S127,logistic,far,white_hot,delta_wing,delta_wing\n'
                                        'S127,knn,far,white_hot,delta_wing,delta_wing\n'
                                        'S128,decision_tree,far,white_hot,fixed_wing,delta_wing\n'
                                        'S128,logistic,far,white_hot,fixed_wing,fixed_wing\n'
                                        'S128,knn,far,white_hot,fixed_wing,delta_wing\n'
                                        'S129,decision_tree,far,white_hot,cruise_missile,delta_wing\n'
                                        'S129,logistic,far,white_hot,cruise_missile,delta_wing\n'
                                        'S129,knn,far,white_hot,cruise_missile,cruise_missile\n'
                                        'S130,decision_tree,far,white_hot,fixed_wing,delta_wing\n'
                                        'S130,logistic,far,white_hot,fixed_wing,delta_wing\n'
                                        'S130,knn,far,white_hot,fixed_wing,delta_wing\n'
                                        'S131,decision_tree,far,white_hot,delta_wing,delta_wing\n'
                                        'S131,logistic,far,white_hot,delta_wing,delta_wing\n'
                                        'S131,knn,far,white_hot,delta_wing,delta_wing\n'
                                        'S132,decision_tree,far,white_hot,ballistic_missile,cruise_missile\n'
                                        'S132,logistic,far,white_hot,ballistic_missile,cruise_missile\n'
                                        'S132,knn,far,white_hot,ballistic_missile,ballistic_missile\n'
                                        'S133,decision_tree,far,white_hot,fixed_wing,delta_wing\n'
                                        'S133,logistic,far,white_hot,fixed_wing,fixed_wing\n'
                                        'S133,knn,far,white_hot,fixed_wing,fixed_wing\n'
                                        'S134,decision_tree,far,white_hot,multicopter,bird\n'
                                        'S134,logistic,far,white_hot,multicopter,multicopter\n'
                                        'S134,knn,far,white_hot,multicopter,multicopter\n'
                                        'S135,decision_tree,far,white_hot,delta_wing,fixed_wing\n'
                                        'S135,logistic,far,white_hot,delta_wing,fixed_wing\n'
                                        'S135,knn,far,white_hot,delta_wing,delta_wing\n'
                                        'S136,decision_tree,far,white_hot,ballistic_missile,ballistic_missile\n'
                                        'S136,logistic,far,white_hot,ballistic_missile,cruise_missile\n'
                                        'S136,knn,far,white_hot,ballistic_missile,ballistic_missile\n'
                                        'S137,decision_tree,far,white_hot,bird,bird\n'
                                        'S137,logistic,far,white_hot,bird,multicopter\n'
                                        'S137,knn,far,white_hot,bird,multicopter\n'
                                        'S138,decision_tree,far,white_hot,delta_wing,fixed_wing\n'
                                        'S138,logistic,far,white_hot,delta_wing,fixed_wing\n'
                                        'S138,knn,far,white_hot,delta_wing,fixed_wing\n'
                                        'S139,decision_tree,far,white_hot,delta_wing,fixed_wing\n'
                                        'S139,logistic,far,white_hot,delta_wing,delta_wing\n'
                                        'S139,knn,far,white_hot,delta_wing,delta_wing\n'
                                        'S140,decision_tree,far,white_hot,ballistic_missile,ballistic_missile\n'
                                        'S140,logistic,far,white_hot,ballistic_missile,ballistic_missile\n'
                                        'S140,knn,far,white_hot,ballistic_missile,ballistic_missile\n'
                                        'S141,decision_tree,far,white_hot,bird,multicopter\n'
                                        'S141,logistic,far,white_hot,bird,multicopter\n'
                                        'S141,knn,far,white_hot,bird,bird\n'
                                        'S142,decision_tree,far,white_hot,delta_wing,delta_wing\n'
                                        'S142,logistic,far,white_hot,delta_wing,delta_wing\n'
                                        'S142,knn,far,white_hot,delta_wing,delta_wing\n'
                                        'S143,decision_tree,far,white_hot,bird,multicopter\n'
                                        'S143,logistic,far,white_hot,bird,bird\n'
                                        'S143,knn,far,white_hot,bird,bird\n'
                                        'S144,decision_tree,far,white_hot,fixed_wing,delta_wing\n'
                                        'S144,logistic,far,white_hot,fixed_wing,delta_wing\n'
                                        'S144,knn,far,white_hot,fixed_wing,delta_wing\n'
                                        'S145,decision_tree,far,white_hot,cruise_missile,delta_wing\n'
                                        'S145,logistic,far,white_hot,cruise_missile,delta_wing\n'
                                        'S145,knn,far,white_hot,cruise_missile,cruise_missile\n'
                                        'S146,decision_tree,far,white_hot,multicopter,multicopter\n'
                                        'S146,logistic,far,white_hot,multicopter,multicopter\n'
                                        'S146,knn,far,white_hot,multicopter,multicopter\n'
                                        'S147,decision_tree,far,white_hot,cruise_missile,cruise_missile\n'
                                        'S147,logistic,far,white_hot,cruise_missile,cruise_missile\n'
                                        'S147,knn,far,white_hot,cruise_missile,delta_wing\n'
                                        'S148,decision_tree,far,white_hot,delta_wing,fixed_wing\n'
                                        'S148,logistic,far,white_hot,delta_wing,fixed_wing\n'
                                        'S148,knn,far,white_hot,delta_wing,fixed_wing\n'
                                        'S149,decision_tree,far,white_hot,bird,bird\n'
                                        'S149,logistic,far,white_hot,bird,bird\n'
                                        'S149,knn,far,white_hot,bird,bird\n'
                                        'S150,decision_tree,far,white_hot,bird,bird\n'
                                        'S150,logistic,far,white_hot,bird,bird\n'
                                        'S150,knn,far,white_hot,bird,bird\n',
                        'm4_models.csv': 'model,threat_recall_pct,ms_per_image,size_mb,sends_outside,license_ok\n'
                                         'small_cnn,86.5,12,1.2,0,1\n'
                                         'mobilenet_v3,92.4,35,9.8,0,1\n'
                                         'dinov2_probe,95.1,180,86.0,0,1\n'
                                         'clip_zeroshot,78.3,210,340.0,0,1\n'
                                         'kan_classifier,84.0,25,0.8,0,1\n'
                                         'hopfield_net,71.2,8,0.4,0,1\n'
                                         'cloud_api_model,97.8,90,0.0,1,0\n',
                        'm5_scores.csv': 'image_id,split,top1_label,top1_score,top2_score,is_correct\n'
                                         'V001,val,ballistic_missile,0.638,0.218,0\n'
                                         'V002,val,ballistic_missile,0.34,0.335,0\n'
                                         'V003,val,cruise_missile,0.776,0.188,1\n'
                                         'V004,val,delta_wing,0.691,0.166,1\n'
                                         'V005,val,cruise_missile,0.896,0.077,0\n'
                                         'V006,val,ballistic_missile,0.962,0.033,1\n'
                                         'V007,val,cruise_missile,0.628,0.275,1\n'
                                         'V008,val,multicopter,0.866,0.072,0\n'
                                         'V009,val,multicopter,0.827,0.16,1\n'
                                         'V010,val,bird,0.673,0.24,0\n'
                                         'V011,val,fixed_wing,0.803,0.154,1\n'
                                         'V012,val,ballistic_missile,0.739,0.165,0\n'
                                         'V013,val,ballistic_missile,0.806,0.169,1\n'
                                         'V014,val,fixed_wing,0.705,0.191,0\n'
                                         'V015,val,ballistic_missile,0.964,0.017,1\n'
                                         'V016,val,ballistic_missile,0.629,0.264,0\n'
                                         'V017,val,fixed_wing,0.616,0.31,0\n'
                                         'V018,val,fixed_wing,0.526,0.451,0\n'
                                         'V019,val,multicopter,0.698,0.275,0\n'
                                         'V020,val,fixed_wing,0.818,0.107,0\n'
                                         'V021,val,multicopter,0.72,0.174,0\n'
                                         'V022,val,cruise_missile,0.446,0.265,1\n'
                                         'V023,val,multicopter,0.984,0.008,1\n'
                                         'V024,val,ballistic_missile,0.778,0.166,1\n'
                                         'V025,val,bird,0.662,0.212,0\n'
                                         'V026,val,multicopter,0.941,0.051,1\n'
                                         'V027,val,delta_wing,0.911,0.054,1\n'
                                         'V028,val,delta_wing,0.882,0.063,1\n'
                                         'V029,val,ballistic_missile,0.811,0.1,0\n'
                                         'V030,val,delta_wing,0.799,0.149,1\n'
                                         'V031,val,multicopter,0.572,0.265,0\n'
                                         'V032,val,ballistic_missile,0.863,0.105,1\n'
                                         'V033,val,fixed_wing,0.863,0.067,1\n'
                                         'V034,val,cruise_missile,0.71,0.147,1\n'
                                         'V035,val,bird,0.793,0.161,0\n'
                                         'V036,val,fixed_wing,0.868,0.061,1\n'
                                         'V037,val,fixed_wing,0.712,0.223,0\n'
                                         'V038,val,delta_wing,0.957,0.032,1\n'
                                         'V039,val,ballistic_missile,0.789,0.174,1\n'
                                         'V040,val,ballistic_missile,0.72,0.191,0\n'
                                         'V041,val,cruise_missile,0.814,0.179,0\n'
                                         'V042,val,cruise_missile,0.575,0.374,0\n'
                                         'V043,val,ballistic_missile,0.914,0.075,1\n'
                                         'V044,val,bird,0.772,0.155,1\n'
                                         'V045,val,ballistic_missile,0.48,0.324,1\n'
                                         'V046,val,delta_wing,0.926,0.058,1\n'
                                         'V047,val,ballistic_missile,0.902,0.064,1\n'
                                         'V048,val,ballistic_missile,0.491,0.486,0\n'
                                         'V049,val,fixed_wing,0.976,0.022,1\n'
                                         'V050,val,cruise_missile,0.837,0.126,0\n'
                                         'V051,val,multicopter,0.967,0.029,1\n'
                                         'V052,val,bird,0.803,0.123,0\n'
                                         'V053,val,multicopter,0.646,0.302,0\n'
                                         'V054,val,ballistic_missile,0.831,0.116,0\n'
                                         'V055,val,bird,0.862,0.106,1\n'
                                         'V056,val,delta_wing,0.852,0.088,1\n'
                                         'V057,val,multicopter,0.81,0.096,1\n'
                                         'V058,val,ballistic_missile,0.767,0.197,1\n'
                                         'V059,val,bird,0.869,0.106,1\n'
                                         'V060,val,fixed_wing,0.546,0.339,0\n'
                                         'V061,val,cruise_missile,0.797,0.147,0\n'
                                         'V062,val,multicopter,0.688,0.181,1\n'
                                         'V063,val,bird,0.88,0.069,1\n'
                                         'V064,val,delta_wing,0.709,0.279,1\n'
                                         'V065,val,cruise_missile,0.599,0.182,0\n'
                                         'V066,val,cruise_missile,0.929,0.051,0\n'
                                         'V067,val,delta_wing,0.629,0.267,0\n'
                                         'V068,val,multicopter,0.899,0.081,1\n'
                                         'V069,val,multicopter,0.852,0.095,1\n'
                                         'V070,val,multicopter,0.884,0.067,1\n'
                                         'V071,val,cruise_missile,0.688,0.148,0\n'
                                         'V072,val,ballistic_missile,0.822,0.117,0\n'
                                         'V073,val,multicopter,0.879,0.079,1\n'
                                         'V074,val,cruise_missile,0.925,0.074,1\n'
                                         'V075,val,delta_wing,0.637,0.247,1\n'
                                         'V076,val,ballistic_missile,0.685,0.299,0\n'
                                         'V077,val,multicopter,0.897,0.063,1\n'
                                         'V078,val,fixed_wing,0.714,0.208,1\n'
                                         'V079,val,multicopter,0.976,0.021,1\n'
                                         'V080,val,cruise_missile,0.976,0.012,1\n'
                                         'V081,val,delta_wing,0.524,0.321,1\n'
                                         'V082,val,delta_wing,0.769,0.119,1\n'
                                         'V083,val,fixed_wing,0.591,0.297,1\n'
                                         'V084,val,fixed_wing,0.931,0.032,1\n'
                                         'V085,val,bird,0.845,0.083,0\n'
                                         'V086,val,multicopter,0.906,0.052,0\n'
                                         'V087,val,multicopter,0.868,0.062,1\n'
                                         'V088,val,bird,0.573,0.388,0\n'
                                         'V089,val,fixed_wing,0.939,0.038,1\n'
                                         'V090,val,multicopter,0.913,0.048,1\n'
                                         'V091,val,multicopter,0.572,0.246,0\n'
                                         'V092,val,delta_wing,0.815,0.097,1\n'
                                         'V093,val,bird,0.899,0.071,1\n'
                                         'V094,val,bird,0.818,0.162,0\n'
                                         'V095,val,fixed_wing,0.825,0.094,1\n'
                                         'V096,val,fixed_wing,0.959,0.029,1\n'
                                         'V097,val,bird,0.724,0.241,1\n'
                                         'V098,val,multicopter,0.901,0.079,0\n'
                                         'V099,val,bird,0.696,0.253,1\n'
                                         'V100,val,bird,0.496,0.244,0\n'
                                         'V101,val,delta_wing,0.674,0.207,1\n'
                                         'V102,val,delta_wing,0.817,0.087,1\n'
                                         'V103,val,bird,0.944,0.04,1\n'
                                         'V104,val,cruise_missile,0.872,0.075,1\n'
                                         'V105,val,fixed_wing,0.735,0.131,0\n'
                                         'V106,val,multicopter,0.891,0.052,1\n'
                                         'V107,val,ballistic_missile,0.908,0.06,0\n'
                                         'V108,val,delta_wing,0.917,0.046,1\n'
                                         'V109,val,multicopter,0.51,0.357,1\n'
                                         'V110,val,bird,0.896,0.065,0\n'
                                         'V111,val,multicopter,0.852,0.068,1\n'
                                         'V112,val,delta_wing,0.574,0.223,1\n'
                                         'V113,val,cruise_missile,0.722,0.134,1\n'
                                         'V114,val,multicopter,0.697,0.144,0\n'
                                         'V115,val,ballistic_missile,0.426,0.421,0\n'
                                         'V116,val,multicopter,0.607,0.337,0\n'
                                         'V117,val,bird,0.914,0.07,1\n'
                                         'V118,val,cruise_missile,0.687,0.258,1\n'
                                         'V119,val,fixed_wing,0.836,0.089,1\n'
                                         'V120,val,cruise_missile,0.928,0.048,1\n'
                                         'V121,val,bird,0.815,0.126,0\n'
                                         'V122,val,multicopter,0.537,0.405,1\n'
                                         'V123,val,multicopter,0.717,0.158,1\n'
                                         'V124,val,cruise_missile,0.976,0.016,1\n'
                                         'V125,val,ballistic_missile,0.916,0.063,0\n'
                                         'V126,val,multicopter,0.898,0.099,1\n'
                                         'V127,val,multicopter,0.936,0.047,0\n'
                                         'V128,val,cruise_missile,0.34,0.335,0\n'
                                         'V129,val,ballistic_missile,0.845,0.126,1\n'
                                         'V130,val,bird,0.91,0.086,1\n'
                                         'V131,val,fixed_wing,0.666,0.202,0\n'
                                         'V132,val,cruise_missile,0.592,0.383,0\n'
                                         'V133,val,fixed_wing,0.864,0.119,1\n'
                                         'V134,val,multicopter,0.613,0.335,0\n'
                                         'V135,val,fixed_wing,0.34,0.335,0\n'
                                         'V136,val,cruise_missile,0.848,0.101,1\n'
                                         'V137,val,fixed_wing,0.813,0.134,1\n'
                                         'V138,val,ballistic_missile,0.907,0.066,1\n'
                                         'V139,val,multicopter,0.793,0.146,0\n'
                                         'V140,val,bird,0.95,0.035,1\n'
                                         'V141,val,fixed_wing,0.885,0.093,0\n'
                                         'V142,val,bird,0.958,0.022,1\n'
                                         'V143,val,fixed_wing,0.698,0.225,1\n'
                                         'V144,val,bird,0.594,0.29,0\n'
                                         'V145,val,delta_wing,0.771,0.19,0\n'
                                         'V146,val,bird,0.75,0.179,1\n'
                                         'V147,val,fixed_wing,0.656,0.282,0\n'
                                         'V148,val,ballistic_missile,0.72,0.161,1\n'
                                         'V149,val,bird,0.816,0.144,1\n'
                                         'V150,val,ballistic_missile,0.819,0.149,1\n'
                                         'V151,val,cruise_missile,0.61,0.252,1\n'
                                         'V152,val,cruise_missile,0.83,0.106,1\n'
                                         'V153,val,fixed_wing,0.819,0.172,1\n'
                                         'V154,val,ballistic_missile,0.617,0.18,1\n'
                                         'V155,val,fixed_wing,0.797,0.171,0\n'
                                         'V156,val,multicopter,0.76,0.116,1\n'
                                         'V157,val,multicopter,0.697,0.296,0\n'
                                         'V158,val,fixed_wing,0.89,0.082,1\n'
                                         'V159,val,ballistic_missile,0.749,0.205,0\n'
                                         'V160,val,ballistic_missile,0.808,0.103,1\n'
                                         'V161,val,fixed_wing,0.675,0.263,1\n'
                                         'V162,val,ballistic_missile,0.417,0.317,0\n'
                                         'V163,val,fixed_wing,0.918,0.08,1\n'
                                         'V164,val,multicopter,0.944,0.025,1\n'
                                         'V165,val,delta_wing,0.696,0.19,1\n'
                                         'V166,val,fixed_wing,0.973,0.022,0\n'
                                         'V167,val,delta_wing,0.651,0.263,1\n'
                                         'V168,val,delta_wing,0.489,0.258,1\n'
                                         'V169,val,multicopter,0.792,0.169,1\n'
                                         'V170,val,delta_wing,0.558,0.361,0\n'
                                         'V171,val,multicopter,0.907,0.068,1\n'
                                         'V172,val,fixed_wing,0.526,0.405,0\n'
                                         'V173,val,delta_wing,0.916,0.081,1\n'
                                         'V174,val,fixed_wing,0.481,0.369,0\n'
                                         'V175,val,delta_wing,0.432,0.427,0\n'
                                         'V176,val,ballistic_missile,0.661,0.329,0\n'
                                         'V177,val,fixed_wing,0.96,0.032,1\n'
                                         'V178,val,ballistic_missile,0.728,0.172,1\n'
                                         'V179,val,cruise_missile,0.367,0.362,0\n'
                                         'V180,val,cruise_missile,0.9,0.055,1\n'
                                         'V181,val,fixed_wing,0.633,0.245,1\n'
                                         'V182,val,fixed_wing,0.915,0.075,1\n'
                                         'V183,val,bird,0.99,0.01,1\n'
                                         'V184,val,cruise_missile,0.731,0.129,0\n'
                                         'V185,val,multicopter,0.716,0.163,0\n'
                                         'V186,val,fixed_wing,0.749,0.188,0\n'
                                         'V187,val,multicopter,0.599,0.366,1\n'
                                         'V188,val,multicopter,0.944,0.034,1\n'
                                         'V189,val,multicopter,0.887,0.099,1\n'
                                         'V190,val,bird,0.805,0.124,0\n'
                                         'V191,val,multicopter,0.646,0.177,1\n'
                                         'V192,val,delta_wing,0.598,0.355,0\n'
                                         'V193,val,bird,0.535,0.37,0\n'
                                         'V194,val,bird,0.77,0.137,0\n'
                                         'V195,val,ballistic_missile,0.414,0.401,0\n'
                                         'V196,val,delta_wing,0.428,0.423,0\n'
                                         'V197,val,multicopter,0.887,0.078,0\n'
                                         'V198,val,multicopter,0.816,0.113,0\n'
                                         'V199,val,delta_wing,0.72,0.175,0\n'
                                         'V200,val,cruise_missile,0.883,0.113,1\n'
                                         'V201,test,ballistic_missile,0.94,0.055,1\n'
                                         'V202,test,delta_wing,0.823,0.117,0\n'
                                         'V203,test,ballistic_missile,0.445,0.44,0\n'
                                         'V204,test,cruise_missile,0.556,0.27,1\n'
                                         'V205,test,ballistic_missile,0.808,0.15,0\n'
                                         'V206,test,multicopter,0.698,0.231,0\n'
                                         'V207,test,cruise_missile,0.928,0.046,1\n'
                                         'V208,test,cruise_missile,0.601,0.332,0\n'
                                         'V209,test,ballistic_missile,0.852,0.143,1\n'
                                         'V210,test,cruise_missile,0.895,0.098,1\n'
                                         'V211,test,delta_wing,0.703,0.175,0\n'
                                         'V212,test,bird,0.757,0.19,1\n'
                                         'V213,test,multicopter,0.93,0.067,1\n'
                                         'V214,test,cruise_missile,0.638,0.215,0\n'
                                         'V215,test,multicopter,0.736,0.236,1\n'
                                         'V216,test,bird,0.671,0.245,0\n'
                                         'V217,test,cruise_missile,0.683,0.243,0\n'
                                         'V218,test,fixed_wing,0.921,0.048,1\n'
                                         'V219,test,ballistic_missile,0.708,0.192,0\n'
                                         'V220,test,delta_wing,0.779,0.214,1\n'
                                         'V221,test,bird,0.601,0.214,0\n'
                                         'V222,test,ballistic_missile,0.811,0.184,1\n'
                                         'V223,test,delta_wing,0.948,0.043,1\n'
                                         'V224,test,ballistic_missile,0.928,0.041,1\n'
                                         'V225,test,multicopter,0.586,0.258,0\n'
                                         'V226,test,fixed_wing,0.959,0.03,1\n'
                                         'V227,test,multicopter,0.838,0.116,1\n'
                                         'V228,test,fixed_wing,0.826,0.132,1\n'
                                         'V229,test,ballistic_missile,0.607,0.361,0\n'
                                         'V230,test,fixed_wing,0.927,0.051,1\n'
                                         'V231,test,ballistic_missile,0.774,0.167,0\n'
                                         'V232,test,multicopter,0.867,0.115,1\n'
                                         'V233,test,ballistic_missile,0.82,0.116,0\n'
                                         'V234,test,fixed_wing,0.93,0.036,1\n'
                                         'V235,test,ballistic_missile,0.765,0.132,0\n'
                                         'V236,test,multicopter,0.357,0.352,0\n'
                                         'V237,test,fixed_wing,0.801,0.166,1\n'
                                         'V238,test,fixed_wing,0.961,0.038,1\n'
                                         'V239,test,delta_wing,0.6,0.314,0\n'
                                         'V240,test,multicopter,0.882,0.074,1\n'
                                         'V241,test,fixed_wing,0.901,0.061,1\n'
                                         'V242,test,cruise_missile,0.34,0.335,0\n'
                                         'V243,test,delta_wing,0.99,0.009,1\n'
                                         'V244,test,delta_wing,0.957,0.026,1\n'
                                         'V245,test,fixed_wing,0.79,0.129,0\n'
                                         'V246,test,multicopter,0.92,0.061,1\n'
                                         'V247,test,bird,0.932,0.057,1\n'
                                         'V248,test,multicopter,0.588,0.236,1\n'
                                         'V249,test,fixed_wing,0.762,0.159,1\n'
                                         'V250,test,fixed_wing,0.751,0.201,0\n'
                                         'V251,test,ballistic_missile,0.864,0.068,1\n'
                                         'V252,test,ballistic_missile,0.855,0.078,1\n'
                                         'V253,test,ballistic_missile,0.902,0.082,0\n'
                                         'V254,test,multicopter,0.786,0.199,1\n'
                                         'V255,test,fixed_wing,0.501,0.249,0\n'
                                         'V256,test,fixed_wing,0.903,0.052,1\n'
                                         'V257,test,delta_wing,0.638,0.334,1\n'
                                         'V258,test,bird,0.796,0.178,1\n'
                                         'V259,test,bird,0.836,0.141,1\n'
                                         'V260,test,multicopter,0.99,0.006,1\n'
                                         'V261,test,fixed_wing,0.927,0.038,1\n'
                                         'V262,test,cruise_missile,0.804,0.175,1\n'
                                         'V263,test,fixed_wing,0.69,0.241,1\n'
                                         'V264,test,bird,0.887,0.066,1\n'
                                         'V265,test,multicopter,0.791,0.14,1\n'
                                         'V266,test,fixed_wing,0.812,0.088,1\n'
                                         'V267,test,multicopter,0.851,0.105,1\n'
                                         'V268,test,bird,0.965,0.029,1\n'
                                         'V269,test,bird,0.679,0.166,0\n'
                                         'V270,test,ballistic_missile,0.788,0.123,1\n'
                                         'V271,test,fixed_wing,0.86,0.128,1\n'
                                         'V272,test,fixed_wing,0.826,0.165,1\n'
                                         'V273,test,ballistic_missile,0.617,0.344,1\n'
                                         'V274,test,multicopter,0.947,0.027,1\n'
                                         'V275,test,multicopter,0.596,0.363,0\n'
                                         'V276,test,fixed_wing,0.541,0.226,0\n'
                                         'V277,test,bird,0.76,0.113,1\n'
                                         'V278,test,multicopter,0.781,0.121,1\n'
                                         'V279,test,multicopter,0.951,0.047,1\n'
                                         'V280,test,bird,0.869,0.076,1\n'
                                         'V281,test,delta_wing,0.678,0.297,1\n'
                                         'V282,test,delta_wing,0.794,0.105,0\n'
                                         'V283,test,bird,0.742,0.17,1\n'
                                         'V284,test,ballistic_missile,0.95,0.048,1\n'
                                         'V285,test,ballistic_missile,0.825,0.162,0\n'
                                         'V286,test,multicopter,0.788,0.109,1\n'
                                         'V287,test,cruise_missile,0.805,0.117,0\n'
                                         'V288,test,cruise_missile,0.805,0.097,1\n'
                                         'V289,test,multicopter,0.642,0.178,1\n'
                                         'V290,test,fixed_wing,0.819,0.151,1\n'
                                         'V291,test,bird,0.668,0.15,1\n'
                                         'V292,test,ballistic_missile,0.34,0.335,0\n'
                                         'V293,test,delta_wing,0.837,0.075,0\n'
                                         'V294,test,ballistic_missile,0.491,0.413,0\n'
                                         'V295,test,fixed_wing,0.902,0.071,1\n'
                                         'V296,test,multicopter,0.99,0.006,1\n'
                                         'V297,test,multicopter,0.787,0.176,0\n'
                                         'V298,test,ballistic_missile,0.856,0.117,1\n'
                                         'V299,test,cruise_missile,0.933,0.046,1\n'
                                         'V300,test,ballistic_missile,0.696,0.29,0\n'
                                         'V301,test,fixed_wing,0.969,0.028,0\n'
                                         'V302,test,delta_wing,0.922,0.051,1\n'
                                         'V303,test,fixed_wing,0.861,0.128,1\n'
                                         'V304,test,multicopter,0.841,0.132,1\n'
                                         'V305,test,ballistic_missile,0.721,0.183,1\n'
                                         'V306,test,multicopter,0.82,0.113,1\n'
                                         'V307,test,multicopter,0.591,0.284,0\n'
                                         'V308,test,multicopter,0.877,0.102,1\n'
                                         'V309,test,ballistic_missile,0.734,0.257,1\n'
                                         'V310,test,bird,0.876,0.077,1\n'
                                         'V311,test,ballistic_missile,0.852,0.081,1\n'
                                         'V312,test,multicopter,0.826,0.096,1\n'
                                         'V313,test,cruise_missile,0.78,0.139,1\n'
                                         'V314,test,cruise_missile,0.585,0.371,0\n'
                                         'V315,test,multicopter,0.876,0.063,1\n'
                                         'V316,test,fixed_wing,0.719,0.155,1\n'
                                         'V317,test,ballistic_missile,0.868,0.111,1\n'
                                         'V318,test,fixed_wing,0.89,0.092,1\n'
                                         'V319,test,multicopter,0.705,0.186,1\n'
                                         'V320,test,bird,0.675,0.258,1\n'
                                         'V321,test,delta_wing,0.63,0.284,0\n'
                                         'V322,test,delta_wing,0.867,0.065,1\n'
                                         'V323,test,cruise_missile,0.783,0.167,0\n'
                                         'V324,test,multicopter,0.964,0.03,1\n'
                                         'V325,test,fixed_wing,0.806,0.166,0\n'
                                         'V326,test,bird,0.599,0.287,0\n'
                                         'V327,test,multicopter,0.848,0.09,1\n'
                                         'V328,test,delta_wing,0.634,0.33,0\n'
                                         'V329,test,multicopter,0.525,0.369,0\n'
                                         'V330,test,multicopter,0.961,0.019,1\n'
                                         'V331,test,multicopter,0.675,0.268,1\n'
                                         'V332,test,fixed_wing,0.962,0.032,0\n'
                                         'V333,test,multicopter,0.624,0.213,1\n'
                                         'V334,test,multicopter,0.836,0.083,1\n'
                                         'V335,test,ballistic_missile,0.821,0.091,0\n'
                                         'V336,test,bird,0.965,0.022,1\n'
                                         'V337,test,ballistic_missile,0.615,0.278,0\n'
                                         'V338,test,ballistic_missile,0.909,0.052,1\n'
                                         'V339,test,ballistic_missile,0.812,0.099,1\n'
                                         'V340,test,fixed_wing,0.774,0.215,1\n'
                                         'V341,test,bird,0.884,0.103,0\n'
                                         'V342,test,bird,0.844,0.148,0\n'
                                         'V343,test,bird,0.593,0.373,1\n'
                                         'V344,test,cruise_missile,0.843,0.127,1\n'
                                         'V345,test,delta_wing,0.554,0.393,1\n'
                                         'V346,test,fixed_wing,0.736,0.233,1\n'
                                         'V347,test,bird,0.444,0.376,0\n'
                                         'V348,test,ballistic_missile,0.762,0.205,0\n'
                                         'V349,test,delta_wing,0.803,0.111,1\n'
                                         'V350,test,cruise_missile,0.941,0.036,1\n'
                                         'V351,test,cruise_missile,0.777,0.191,1\n'
                                         'V352,test,delta_wing,0.812,0.17,1\n'
                                         'V353,test,ballistic_missile,0.83,0.158,0\n'
                                         'V354,test,ballistic_missile,0.893,0.084,1\n'
                                         'V355,test,multicopter,0.549,0.359,0\n'
                                         'V356,test,cruise_missile,0.833,0.14,0\n'
                                         'V357,test,fixed_wing,0.905,0.05,1\n'
                                         'V358,test,multicopter,0.99,0.006,1\n'
                                         'V359,test,ballistic_missile,0.913,0.085,1\n'
                                         'V360,test,cruise_missile,0.806,0.171,1\n'
                                         'V361,test,cruise_missile,0.794,0.122,1\n'
                                         'V362,test,bird,0.923,0.067,1\n'
                                         'V363,test,multicopter,0.597,0.255,0\n'
                                         'V364,test,bird,0.746,0.144,0\n'
                                         'V365,test,fixed_wing,0.795,0.135,1\n'
                                         'V366,test,fixed_wing,0.865,0.086,1\n'
                                         'V367,test,fixed_wing,0.979,0.012,1\n'
                                         'V368,test,ballistic_missile,0.969,0.03,1\n'
                                         'V369,test,cruise_missile,0.818,0.136,1\n'
                                         'V370,test,fixed_wing,0.7,0.189,1\n'
                                         'V371,test,ballistic_missile,0.53,0.379,1\n'
                                         'V372,test,delta_wing,0.933,0.046,1\n'
                                         'V373,test,bird,0.694,0.238,1\n'
                                         'V374,test,delta_wing,0.889,0.094,1\n'
                                         'V375,test,ballistic_missile,0.7,0.258,1\n'
                                         'V376,test,multicopter,0.701,0.25,1\n'
                                         'V377,test,multicopter,0.584,0.284,0\n'
                                         'V378,test,fixed_wing,0.825,0.082,1\n'
                                         'V379,test,bird,0.87,0.093,0\n'
                                         'V380,test,bird,0.87,0.077,1\n'
                                         'V381,test,ballistic_missile,0.966,0.016,0\n'
                                         'V382,test,multicopter,0.86,0.118,1\n'
                                         'V383,test,cruise_missile,0.986,0.01,1\n'
                                         'V384,test,cruise_missile,0.629,0.262,0\n'
                                         'V385,test,ballistic_missile,0.475,0.339,1\n'
                                         'V386,test,ballistic_missile,0.733,0.123,1\n'
                                         'V387,test,fixed_wing,0.934,0.054,1\n'
                                         'V388,test,multicopter,0.934,0.064,1\n'
                                         'V389,test,bird,0.927,0.043,0\n'
                                         'V390,test,delta_wing,0.94,0.028,1\n'
                                         'V391,test,bird,0.983,0.013,1\n'
                                         'V392,test,bird,0.932,0.041,1\n'
                                         'V393,test,ballistic_missile,0.762,0.126,0\n'
                                         'V394,test,bird,0.978,0.015,1\n'
                                         'V395,test,ballistic_missile,0.722,0.143,0\n'
                                         'V396,test,bird,0.778,0.148,0\n'
                                         'V397,test,cruise_missile,0.851,0.073,0\n'
                                         'V398,test,fixed_wing,0.933,0.056,1\n'
                                         'V399,test,multicopter,0.639,0.263,1\n'
                                         'V400,test,ballistic_missile,0.709,0.256,0\n',
                        'm6_items.csv': 'item_id,family,confusable_with,shown,wrong\n'
                                        'Q01,delta_wing,fixed_wing,40,19\n'
                                        'Q02,fixed_wing,delta_wing,40,11\n'
                                        'Q03,cruise_missile,delta_wing,38,17\n'
                                        'Q04,multicopter,bird,36,5\n'
                                        'Q05,bird,multicopter,36,8\n'
                                        'Q06,ballistic_missile,cruise_missile,30,6\n'
                                        'Q07,delta_wing,cruise_missile,3,2\n'
                                        'Q08,jet_delta,delta_wing,0,0\n'
                                        'Q09,multicopter,bird,22,6\n'
                                        'Q10,cruise_missile,fixed_wing,0,0\n',
                        'm7_facts.csv': 'family,wing,engine,tail\n'
                                        'multicopter,none,rotor_4,none\n'
                                        'fixed_wing,straight,pusher,t_tail\n'
                                        'delta_wing,delta,pusher,winglet\n'
                                        'cruise_missile,small,jet,cross_fin\n'
                                        'ballistic_missile,none,rocket,fin_4\n'
                                        'bird,flapping,none,fan\n',
                        'm7_drafts.csv': 'draft_id,family,field,claim,wait_hours\n'
                                         'D01,delta_wing,wing,delta,6\n'
                                         'D02,delta_wing,engine,jet,20\n'
                                         'D03,fixed_wing,tail,t_tail,52\n'
                                         'D04,cruise_missile,speed,900kmh,12\n'
                                         'D05,multicopter,engine,rotor_4,3\n'
                                         'D06,bird,tail,fan,60\n'
                                         'D07,ballistic_missile,wing,small,30\n'
                                         'D08,cruise_missile,maker,unknown_co,8\n',
                        'm8_lite.csv': 'variant,method,size_mb,ms_browser,threat_recall_pct,top1_acc_pct\n'
                                       'original,none,9.8,85,92.4,88.1\n'
                                       'weight_30,weight_prune,7.1,85,92.1,87.9\n'
                                       'weight_50,weight_prune,5.6,84,91.2,87.0\n'
                                       'channel_30,channel_prune,5.4,58,91.3,87.2\n'
                                       'channel_50,channel_prune,3.4,41,89.6,85.8\n'
                                       'channel_50_ft,channel_prune+fine_tune,3.4,41,91.6,87.4\n'
                                       'channel_70,channel_prune,1.9,24,84.7,80.5\n',
                        'm8_safety.csv': 'case_id,scenario,expected,output,score\n'
                                         'C01,clear,label,delta_wing,0.91\n'
                                         'C02,blur_mild,label,delta_wing,0.74\n'
                                         'C03,blur_heavy,unknown,delta_wing,0.81\n'
                                         'C04,blur_heavy,unknown,unknown,0.41\n'
                                         'C05,corrupt_file,unknown,unknown,0.0\n'
                                         'C06,empty_file,unknown,bird,0.52\n'
                                         'C07,burst_50,label,cruise_missile,0.88\n'
                                         'C08,clear,label,fixed_wing,0.86\n',
                        'm9_log.csv': 'learner_id,family,confusable_with,answered,correct\n'
                                      'L01,multicopter,bird,1,1\n'
                                      'L01,multicopter,bird,1,1\n'
                                      'L01,multicopter,bird,1,1\n'
                                      'L01,multicopter,bird,1,1\n'
                                      'L01,fixed_wing,delta_wing,1,1\n'
                                      'L01,fixed_wing,delta_wing,1,1\n'
                                      'L01,fixed_wing,delta_wing,0,\n'
                                      'L01,fixed_wing,delta_wing,0,\n'
                                      'L01,delta_wing,fixed_wing,1,0\n'
                                      'L01,delta_wing,fixed_wing,1,1\n'
                                      'L01,delta_wing,fixed_wing,1,0\n'
                                      'L01,delta_wing,fixed_wing,1,1\n'
                                      'L01,cruise_missile,delta_wing,1,0\n'
                                      'L01,cruise_missile,delta_wing,0,\n'
                                      'L01,cruise_missile,delta_wing,1,0\n'
                                      'L01,cruise_missile,delta_wing,1,1\n'
                                      'L01,ballistic_missile,cruise_missile,1,1\n'
                                      'L01,ballistic_missile,cruise_missile,1,1\n'
                                      'L01,ballistic_missile,cruise_missile,1,1\n'
                                      'L01,ballistic_missile,cruise_missile,1,1\n'
                                      'L01,bird,multicopter,1,1\n'
                                      'L01,bird,multicopter,1,1\n'
                                      'L01,bird,multicopter,1,1\n'
                                      'L01,bird,multicopter,1,1\n'
                                      'L02,multicopter,bird,1,1\n'
                                      'L02,multicopter,bird,1,1\n'
                                      'L02,multicopter,bird,1,0\n'
                                      'L02,multicopter,bird,0,\n'
                                      'L02,fixed_wing,delta_wing,1,1\n'
                                      'L02,fixed_wing,delta_wing,0,\n'
                                      'L02,fixed_wing,delta_wing,1,1\n'
                                      'L02,fixed_wing,delta_wing,1,0\n'
                                      'L02,delta_wing,fixed_wing,1,1\n'
                                      'L02,delta_wing,fixed_wing,1,0\n'
                                      'L02,delta_wing,fixed_wing,1,1\n'
                                      'L02,delta_wing,fixed_wing,1,1\n'
                                      'L02,cruise_missile,delta_wing,1,1\n'
                                      'L02,cruise_missile,delta_wing,1,0\n'
                                      'L02,cruise_missile,delta_wing,0,\n'
                                      'L02,cruise_missile,delta_wing,1,0\n'
                                      'L02,ballistic_missile,cruise_missile,1,1\n'
                                      'L02,ballistic_missile,cruise_missile,1,1\n'
                                      'L02,ballistic_missile,cruise_missile,1,1\n'
                                      'L02,ballistic_missile,cruise_missile,1,1\n'
                                      'L02,bird,multicopter,1,1\n'
                                      'L02,bird,multicopter,1,1\n'
                                      'L02,bird,multicopter,0,\n'
                                      'L02,bird,multicopter,1,0\n'
                                      'L03,multicopter,bird,1,1\n'
                                      'L03,multicopter,bird,1,0\n'
                                      'L03,multicopter,bird,1,1\n'
                                      'L03,multicopter,bird,1,1\n'
                                      'L03,fixed_wing,delta_wing,1,1\n'
                                      'L03,fixed_wing,delta_wing,1,0\n'
                                      'L03,fixed_wing,delta_wing,1,1\n'
                                      'L03,fixed_wing,delta_wing,1,1\n'
                                      'L03,delta_wing,fixed_wing,1,1\n'
                                      'L03,delta_wing,fixed_wing,0,\n'
                                      'L03,delta_wing,fixed_wing,0,\n'
                                      'L03,delta_wing,fixed_wing,1,0\n'
                                      'L03,cruise_missile,delta_wing,0,\n'
                                      'L03,cruise_missile,delta_wing,1,1\n'
                                      'L03,cruise_missile,delta_wing,1,0\n'
                                      'L03,cruise_missile,delta_wing,1,0\n'
                                      'L03,ballistic_missile,cruise_missile,1,1\n'
                                      'L03,ballistic_missile,cruise_missile,1,1\n'
                                      'L03,ballistic_missile,cruise_missile,0,\n'
                                      'L03,ballistic_missile,cruise_missile,1,1\n'
                                      'L03,bird,multicopter,1,0\n'
                                      'L03,bird,multicopter,1,1\n'
                                      'L03,bird,multicopter,1,1\n'
                                      'L03,bird,multicopter,1,1\n'
                                      'L04,multicopter,bird,0,\n'
                                      'L04,multicopter,bird,1,1\n'
                                      'L04,multicopter,bird,1,1\n'
                                      'L04,multicopter,bird,1,0\n'
                                      'L04,fixed_wing,delta_wing,1,0\n'
                                      'L04,fixed_wing,delta_wing,1,1\n'
                                      'L04,fixed_wing,delta_wing,1,1\n'
                                      'L04,fixed_wing,delta_wing,1,1\n'
                                      'L04,delta_wing,fixed_wing,1,1\n'
                                      'L04,delta_wing,fixed_wing,0,\n'
                                      'L04,delta_wing,fixed_wing,0,\n'
                                      'L04,delta_wing,fixed_wing,1,0\n'
                                      'L04,cruise_missile,delta_wing,0,\n'
                                      'L04,cruise_missile,delta_wing,1,0\n'
                                      'L04,cruise_missile,delta_wing,1,0\n'
                                      'L04,cruise_missile,delta_wing,1,1\n'
                                      'L04,ballistic_missile,cruise_missile,1,1\n'
                                      'L04,ballistic_missile,cruise_missile,1,1\n'
                                      'L04,ballistic_missile,cruise_missile,1,0\n'
                                      'L04,ballistic_missile,cruise_missile,0,\n'
                                      'L04,bird,multicopter,0,\n'
                                      'L04,bird,multicopter,1,1\n'
                                      'L04,bird,multicopter,1,1\n'
                                      'L04,bird,multicopter,1,1\n'
                                      'L05,multicopter,bird,1,0\n'
                                      'L05,multicopter,bird,1,1\n'
                                      'L05,multicopter,bird,1,0\n'
                                      'L05,multicopter,bird,1,1\n'
                                      'L05,fixed_wing,delta_wing,0,\n'
                                      'L05,fixed_wing,delta_wing,0,\n'
                                      'L05,fixed_wing,delta_wing,1,1\n'
                                      'L05,fixed_wing,delta_wing,1,1\n'
                                      'L05,delta_wing,fixed_wing,0,\n'
                                      'L05,delta_wing,fixed_wing,1,1\n'
                                      'L05,delta_wing,fixed_wing,1,0\n'
                                      'L05,delta_wing,fixed_wing,0,\n'
                                      'L05,cruise_missile,delta_wing,1,1\n'
                                      'L05,cruise_missile,delta_wing,1,1\n'
                                      'L05,cruise_missile,delta_wing,0,\n'
                                      'L05,cruise_missile,delta_wing,1,1\n'
                                      'L05,ballistic_missile,cruise_missile,1,1\n'
                                      'L05,ballistic_missile,cruise_missile,1,1\n'
                                      'L05,ballistic_missile,cruise_missile,1,1\n'
                                      'L05,ballistic_missile,cruise_missile,1,1\n'
                                      'L05,bird,multicopter,1,1\n'
                                      'L05,bird,multicopter,1,1\n'
                                      'L05,bird,multicopter,1,1\n'
                                      'L05,bird,multicopter,1,0\n'
                                      'L06,multicopter,bird,1,1\n'
                                      'L06,multicopter,bird,1,1\n'
                                      'L06,multicopter,bird,1,1\n'
                                      'L06,multicopter,bird,1,1\n'
                                      'L06,fixed_wing,delta_wing,1,1\n'
                                      'L06,fixed_wing,delta_wing,1,1\n'
                                      'L06,fixed_wing,delta_wing,1,1\n'
                                      'L06,fixed_wing,delta_wing,1,1\n'
                                      'L06,delta_wing,fixed_wing,1,1\n'
                                      'L06,delta_wing,fixed_wing,0,\n'
                                      'L06,delta_wing,fixed_wing,0,\n'
                                      'L06,delta_wing,fixed_wing,0,\n'
                                      'L06,cruise_missile,delta_wing,1,0\n'
                                      'L06,cruise_missile,delta_wing,1,1\n'
                                      'L06,cruise_missile,delta_wing,1,1\n'
                                      'L06,cruise_missile,delta_wing,1,0\n'
                                      'L06,ballistic_missile,cruise_missile,1,1\n'
                                      'L06,ballistic_missile,cruise_missile,1,1\n'
                                      'L06,ballistic_missile,cruise_missile,1,1\n'
                                      'L06,ballistic_missile,cruise_missile,1,1\n'
                                      'L06,bird,multicopter,1,1\n'
                                      'L06,bird,multicopter,1,1\n'
                                      'L06,bird,multicopter,1,0\n'
                                      'L06,bird,multicopter,1,0\n'
                                      'L07,multicopter,bird,1,1\n'
                                      'L07,multicopter,bird,1,0\n'
                                      'L07,multicopter,bird,1,1\n'
                                      'L07,multicopter,bird,1,1\n'
                                      'L07,fixed_wing,delta_wing,1,0\n'
                                      'L07,fixed_wing,delta_wing,1,1\n'
                                      'L07,fixed_wing,delta_wing,1,1\n'
                                      'L07,fixed_wing,delta_wing,0,\n'
                                      'L07,delta_wing,fixed_wing,0,\n'
                                      'L07,delta_wing,fixed_wing,1,1\n'
                                      'L07,delta_wing,fixed_wing,1,0\n'
                                      'L07,delta_wing,fixed_wing,1,1\n'
                                      'L07,cruise_missile,delta_wing,1,0\n'
                                      'L07,cruise_missile,delta_wing,0,\n'
                                      'L07,cruise_missile,delta_wing,1,0\n'
                                      'L07,cruise_missile,delta_wing,1,1\n'
                                      'L07,ballistic_missile,cruise_missile,1,1\n'
                                      'L07,ballistic_missile,cruise_missile,1,1\n'
                                      'L07,ballistic_missile,cruise_missile,1,1\n'
                                      'L07,ballistic_missile,cruise_missile,1,1\n'
                                      'L07,bird,multicopter,1,1\n'
                                      'L07,bird,multicopter,1,0\n'
                                      'L07,bird,multicopter,1,0\n'
                                      'L07,bird,multicopter,1,1\n'
                                      'L08,multicopter,bird,1,0\n'
                                      'L08,multicopter,bird,1,1\n'
                                      'L08,multicopter,bird,1,1\n'
                                      'L08,multicopter,bird,1,1\n'
                                      'L08,fixed_wing,delta_wing,1,1\n'
                                      'L08,fixed_wing,delta_wing,1,0\n'
                                      'L08,fixed_wing,delta_wing,1,1\n'
                                      'L08,fixed_wing,delta_wing,1,0\n'
                                      'L08,delta_wing,fixed_wing,0,\n'
                                      'L08,delta_wing,fixed_wing,1,1\n'
                                      'L08,delta_wing,fixed_wing,1,1\n'
                                      'L08,delta_wing,fixed_wing,1,0\n'
                                      'L08,cruise_missile,delta_wing,1,0\n'
                                      'L08,cruise_missile,delta_wing,1,1\n'
                                      'L08,cruise_missile,delta_wing,1,1\n'
                                      'L08,cruise_missile,delta_wing,1,1\n'
                                      'L08,ballistic_missile,cruise_missile,1,1\n'
                                      'L08,ballistic_missile,cruise_missile,1,1\n'
                                      'L08,ballistic_missile,cruise_missile,1,1\n'
                                      'L08,ballistic_missile,cruise_missile,1,1\n'
                                      'L08,bird,multicopter,1,1\n'
                                      'L08,bird,multicopter,0,\n'
                                      'L08,bird,multicopter,1,1\n'
                                      'L08,bird,multicopter,1,1\n'
                                      'L09,multicopter,bird,1,1\n'
                                      'L09,multicopter,bird,1,1\n'
                                      'L09,multicopter,bird,0,\n'
                                      'L09,multicopter,bird,1,1\n'
                                      'L09,fixed_wing,delta_wing,1,0\n'
                                      'L09,fixed_wing,delta_wing,1,1\n'
                                      'L09,fixed_wing,delta_wing,1,1\n'
                                      'L09,fixed_wing,delta_wing,1,1\n'
                                      'L09,delta_wing,fixed_wing,1,0\n'
                                      'L09,delta_wing,fixed_wing,0,\n'
                                      'L09,delta_wing,fixed_wing,0,\n'
                                      'L09,delta_wing,fixed_wing,1,1\n'
                                      'L09,cruise_missile,delta_wing,1,1\n'
                                      'L09,cruise_missile,delta_wing,0,\n'
                                      'L09,cruise_missile,delta_wing,1,1\n'
                                      'L09,cruise_missile,delta_wing,1,0\n'
                                      'L09,ballistic_missile,cruise_missile,1,1\n'
                                      'L09,ballistic_missile,cruise_missile,1,1\n'
                                      'L09,ballistic_missile,cruise_missile,1,0\n'
                                      'L09,ballistic_missile,cruise_missile,1,1\n'
                                      'L09,bird,multicopter,1,1\n'
                                      'L09,bird,multicopter,1,1\n'
                                      'L09,bird,multicopter,1,1\n'
                                      'L09,bird,multicopter,1,1\n'
                                      'L10,multicopter,bird,0,\n'
                                      'L10,multicopter,bird,1,1\n'
                                      'L10,multicopter,bird,0,\n'
                                      'L10,multicopter,bird,1,1\n'
                                      'L10,fixed_wing,delta_wing,1,1\n'
                                      'L10,fixed_wing,delta_wing,1,1\n'
                                      'L10,fixed_wing,delta_wing,1,1\n'
                                      'L10,fixed_wing,delta_wing,1,1\n'
                                      'L10,delta_wing,fixed_wing,1,1\n'
                                      'L10,delta_wing,fixed_wing,0,\n'
                                      'L10,delta_wing,fixed_wing,1,0\n'
                                      'L10,delta_wing,fixed_wing,0,\n'
                                      'L10,cruise_missile,delta_wing,1,1\n'
                                      'L10,cruise_missile,delta_wing,1,0\n'
                                      'L10,cruise_missile,delta_wing,1,1\n'
                                      'L10,cruise_missile,delta_wing,1,0\n'
                                      'L10,ballistic_missile,cruise_missile,1,1\n'
                                      'L10,ballistic_missile,cruise_missile,1,1\n'
                                      'L10,ballistic_missile,cruise_missile,1,1\n'
                                      'L10,ballistic_missile,cruise_missile,1,1\n'
                                      'L10,bird,multicopter,1,1\n'
                                      'L10,bird,multicopter,1,1\n'
                                      'L10,bird,multicopter,1,1\n'
                                      'L10,bird,multicopter,1,0\n'
                                      'L11,multicopter,bird,1,1\n'
                                      'L11,multicopter,bird,1,1\n'
                                      'L11,multicopter,bird,1,1\n'
                                      'L11,multicopter,bird,1,1\n'
                                      'L11,fixed_wing,delta_wing,1,1\n'
                                      'L11,fixed_wing,delta_wing,1,0\n'
                                      'L11,fixed_wing,delta_wing,1,1\n'
                                      'L11,fixed_wing,delta_wing,1,1\n'
                                      'L11,delta_wing,fixed_wing,1,0\n'
                                      'L11,delta_wing,fixed_wing,1,1\n'
                                      'L11,delta_wing,fixed_wing,1,1\n'
                                      'L11,delta_wing,fixed_wing,0,\n'
                                      'L11,cruise_missile,delta_wing,1,1\n'
                                      'L11,cruise_missile,delta_wing,0,\n'
                                      'L11,cruise_missile,delta_wing,0,\n'
                                      'L11,cruise_missile,delta_wing,1,0\n'
                                      'L11,ballistic_missile,cruise_missile,1,1\n'
                                      'L11,ballistic_missile,cruise_missile,1,1\n'
                                      'L11,ballistic_missile,cruise_missile,1,1\n'
                                      'L11,ballistic_missile,cruise_missile,1,1\n'
                                      'L11,bird,multicopter,1,1\n'
                                      'L11,bird,multicopter,1,1\n'
                                      'L11,bird,multicopter,1,1\n'
                                      'L11,bird,multicopter,1,1\n'
                                      'L12,multicopter,bird,1,1\n'
                                      'L12,multicopter,bird,1,1\n'
                                      'L12,multicopter,bird,1,1\n'
                                      'L12,multicopter,bird,1,0\n'
                                      'L12,fixed_wing,delta_wing,1,0\n'
                                      'L12,fixed_wing,delta_wing,1,1\n'
                                      'L12,fixed_wing,delta_wing,1,1\n'
                                      'L12,fixed_wing,delta_wing,1,1\n'
                                      'L12,delta_wing,fixed_wing,1,0\n'
                                      'L12,delta_wing,fixed_wing,1,0\n'
                                      'L12,delta_wing,fixed_wing,1,1\n'
                                      'L12,delta_wing,fixed_wing,0,\n'
                                      'L12,cruise_missile,delta_wing,0,\n'
                                      'L12,cruise_missile,delta_wing,0,\n'
                                      'L12,cruise_missile,delta_wing,1,0\n'
                                      'L12,cruise_missile,delta_wing,1,0\n'
                                      'L12,ballistic_missile,cruise_missile,1,1\n'
                                      'L12,ballistic_missile,cruise_missile,1,1\n'
                                      'L12,ballistic_missile,cruise_missile,1,0\n'
                                      'L12,ballistic_missile,cruise_missile,1,0\n'
                                      'L12,bird,multicopter,1,1\n'
                                      'L12,bird,multicopter,0,\n'
                                      'L12,bird,multicopter,1,1\n'
                                      'L12,bird,multicopter,1,1\n'
                                      'L13,multicopter,bird,1,1\n'
                                      'L13,multicopter,bird,1,1\n'
                                      'L13,multicopter,bird,1,1\n'
                                      'L13,multicopter,bird,1,1\n'
                                      'L13,fixed_wing,delta_wing,1,0\n'
                                      'L13,fixed_wing,delta_wing,1,0\n'
                                      'L13,fixed_wing,delta_wing,1,1\n'
                                      'L13,fixed_wing,delta_wing,1,1\n'
                                      'L13,delta_wing,fixed_wing,0,\n'
                                      'L13,delta_wing,fixed_wing,1,1\n'
                                      'L13,delta_wing,fixed_wing,1,1\n'
                                      'L13,delta_wing,fixed_wing,1,1\n'
                                      'L13,cruise_missile,delta_wing,0,\n'
                                      'L13,cruise_missile,delta_wing,0,\n'
                                      'L13,cruise_missile,delta_wing,0,\n'
                                      'L13,cruise_missile,delta_wing,1,1\n'
                                      'L13,ballistic_missile,cruise_missile,1,0\n'
                                      'L13,ballistic_missile,cruise_missile,1,1\n'
                                      'L13,ballistic_missile,cruise_missile,0,\n'
                                      'L13,ballistic_missile,cruise_missile,1,1\n'
                                      'L13,bird,multicopter,1,1\n'
                                      'L13,bird,multicopter,1,1\n'
                                      'L13,bird,multicopter,1,1\n'
                                      'L13,bird,multicopter,1,0\n'
                                      'L14,multicopter,bird,1,1\n'
                                      'L14,multicopter,bird,1,0\n'
                                      'L14,multicopter,bird,1,1\n'
                                      'L14,multicopter,bird,1,1\n'
                                      'L14,fixed_wing,delta_wing,1,1\n'
                                      'L14,fixed_wing,delta_wing,1,1\n'
                                      'L14,fixed_wing,delta_wing,1,0\n'
                                      'L14,fixed_wing,delta_wing,1,1\n'
                                      'L14,delta_wing,fixed_wing,1,0\n'
                                      'L14,delta_wing,fixed_wing,1,0\n'
                                      'L14,delta_wing,fixed_wing,1,0\n'
                                      'L14,delta_wing,fixed_wing,1,1\n'
                                      'L14,cruise_missile,delta_wing,1,1\n'
                                      'L14,cruise_missile,delta_wing,1,1\n'
                                      'L14,cruise_missile,delta_wing,1,0\n'
                                      'L14,cruise_missile,delta_wing,1,1\n'
                                      'L14,ballistic_missile,cruise_missile,1,1\n'
                                      'L14,ballistic_missile,cruise_missile,1,1\n'
                                      'L14,ballistic_missile,cruise_missile,1,1\n'
                                      'L14,ballistic_missile,cruise_missile,1,1\n'
                                      'L14,bird,multicopter,1,0\n'
                                      'L14,bird,multicopter,1,1\n'
                                      'L14,bird,multicopter,1,1\n'
                                      'L14,bird,multicopter,1,1\n'
                                      'L15,multicopter,bird,1,1\n'
                                      'L15,multicopter,bird,1,1\n'
                                      'L15,multicopter,bird,1,1\n'
                                      'L15,multicopter,bird,0,\n'
                                      'L15,fixed_wing,delta_wing,1,1\n'
                                      'L15,fixed_wing,delta_wing,1,1\n'
                                      'L15,fixed_wing,delta_wing,0,\n'
                                      'L15,fixed_wing,delta_wing,1,1\n'
                                      'L15,delta_wing,fixed_wing,0,\n'
                                      'L15,delta_wing,fixed_wing,1,0\n'
                                      'L15,delta_wing,fixed_wing,0,\n'
                                      'L15,delta_wing,fixed_wing,0,\n'
                                      'L15,cruise_missile,delta_wing,1,0\n'
                                      'L15,cruise_missile,delta_wing,1,1\n'
                                      'L15,cruise_missile,delta_wing,1,1\n'
                                      'L15,cruise_missile,delta_wing,1,0\n'
                                      'L15,ballistic_missile,cruise_missile,1,1\n'
                                      'L15,ballistic_missile,cruise_missile,1,1\n'
                                      'L15,ballistic_missile,cruise_missile,0,\n'
                                      'L15,ballistic_missile,cruise_missile,1,1\n'
                                      'L15,bird,multicopter,1,0\n'
                                      'L15,bird,multicopter,1,1\n'
                                      'L15,bird,multicopter,1,1\n'
                                      'L15,bird,multicopter,1,1\n'
                                      'L16,multicopter,bird,1,0\n'
                                      'L16,multicopter,bird,1,1\n'
                                      'L16,multicopter,bird,1,1\n'
                                      'L16,multicopter,bird,1,1\n'
                                      'L16,fixed_wing,delta_wing,1,1\n'
                                      'L16,fixed_wing,delta_wing,1,0\n'
                                      'L16,fixed_wing,delta_wing,1,1\n'
                                      'L16,fixed_wing,delta_wing,1,1\n'
                                      'L16,delta_wing,fixed_wing,0,\n'
                                      'L16,delta_wing,fixed_wing,0,\n'
                                      'L16,delta_wing,fixed_wing,1,1\n'
                                      'L16,delta_wing,fixed_wing,0,\n'
                                      'L16,cruise_missile,delta_wing,0,\n'
                                      'L16,cruise_missile,delta_wing,1,1\n'
                                      'L16,cruise_missile,delta_wing,1,0\n'
                                      'L16,cruise_missile,delta_wing,1,0\n'
                                      'L16,ballistic_missile,cruise_missile,1,0\n'
                                      'L16,ballistic_missile,cruise_missile,1,1\n'
                                      'L16,ballistic_missile,cruise_missile,1,1\n'
                                      'L16,ballistic_missile,cruise_missile,1,1\n'
                                      'L16,bird,multicopter,1,0\n'
                                      'L16,bird,multicopter,1,1\n'
                                      'L16,bird,multicopter,1,1\n'
                                      'L16,bird,multicopter,1,1\n'
                                      'L17,multicopter,bird,1,1\n'
                                      'L17,multicopter,bird,1,1\n'
                                      'L17,multicopter,bird,1,1\n'
                                      'L17,multicopter,bird,1,1\n'
                                      'L17,fixed_wing,delta_wing,1,1\n'
                                      'L17,fixed_wing,delta_wing,1,1\n'
                                      'L17,fixed_wing,delta_wing,1,0\n'
                                      'L17,fixed_wing,delta_wing,1,1\n'
                                      'L17,delta_wing,fixed_wing,0,\n'
                                      'L17,delta_wing,fixed_wing,1,0\n'
                                      'L17,delta_wing,fixed_wing,0,\n'
                                      'L17,delta_wing,fixed_wing,0,\n'
                                      'L17,cruise_missile,delta_wing,0,\n'
                                      'L17,cruise_missile,delta_wing,1,1\n'
                                      'L17,cruise_missile,delta_wing,1,1\n'
                                      'L17,cruise_missile,delta_wing,0,\n'
                                      'L17,ballistic_missile,cruise_missile,1,1\n'
                                      'L17,ballistic_missile,cruise_missile,1,1\n'
                                      'L17,ballistic_missile,cruise_missile,1,1\n'
                                      'L17,ballistic_missile,cruise_missile,1,1\n'
                                      'L17,bird,multicopter,1,1\n'
                                      'L17,bird,multicopter,1,1\n'
                                      'L17,bird,multicopter,0,\n'
                                      'L17,bird,multicopter,1,0\n'
                                      'L18,multicopter,bird,1,1\n'
                                      'L18,multicopter,bird,1,1\n'
                                      'L18,multicopter,bird,1,1\n'
                                      'L18,multicopter,bird,1,1\n'
                                      'L18,fixed_wing,delta_wing,1,0\n'
                                      'L18,fixed_wing,delta_wing,1,1\n'
                                      'L18,fixed_wing,delta_wing,1,0\n'
                                      'L18,fixed_wing,delta_wing,1,1\n'
                                      'L18,delta_wing,fixed_wing,0,\n'
                                      'L18,delta_wing,fixed_wing,1,1\n'
                                      'L18,delta_wing,fixed_wing,1,1\n'
                                      'L18,delta_wing,fixed_wing,1,0\n'
                                      'L18,cruise_missile,delta_wing,1,1\n'
                                      'L18,cruise_missile,delta_wing,1,1\n'
                                      'L18,cruise_missile,delta_wing,1,1\n'
                                      'L18,cruise_missile,delta_wing,1,0\n'
                                      'L18,ballistic_missile,cruise_missile,1,1\n'
                                      'L18,ballistic_missile,cruise_missile,1,0\n'
                                      'L18,ballistic_missile,cruise_missile,1,1\n'
                                      'L18,ballistic_missile,cruise_missile,1,1\n'
                                      'L18,bird,multicopter,1,0\n'
                                      'L18,bird,multicopter,1,0\n'
                                      'L18,bird,multicopter,1,1\n'
                                      'L18,bird,multicopter,1,1\n'
                                      'L19,multicopter,bird,1,1\n'
                                      'L19,multicopter,bird,1,1\n'
                                      'L19,multicopter,bird,1,1\n'
                                      'L19,multicopter,bird,1,1\n'
                                      'L19,fixed_wing,delta_wing,1,1\n'
                                      'L19,fixed_wing,delta_wing,1,0\n'
                                      'L19,fixed_wing,delta_wing,1,1\n'
                                      'L19,fixed_wing,delta_wing,1,1\n'
                                      'L19,delta_wing,fixed_wing,0,\n'
                                      'L19,delta_wing,fixed_wing,1,0\n'
                                      'L19,delta_wing,fixed_wing,1,0\n'
                                      'L19,delta_wing,fixed_wing,1,1\n'
                                      'L19,cruise_missile,delta_wing,1,1\n'
                                      'L19,cruise_missile,delta_wing,1,1\n'
                                      'L19,cruise_missile,delta_wing,1,1\n'
                                      'L19,cruise_missile,delta_wing,1,0\n'
                                      'L19,ballistic_missile,cruise_missile,1,1\n'
                                      'L19,ballistic_missile,cruise_missile,1,1\n'
                                      'L19,ballistic_missile,cruise_missile,1,1\n'
                                      'L19,ballistic_missile,cruise_missile,1,1\n'
                                      'L19,bird,multicopter,1,0\n'
                                      'L19,bird,multicopter,1,0\n'
                                      'L19,bird,multicopter,1,1\n'
                                      'L19,bird,multicopter,1,1\n'
                                      'L20,multicopter,bird,1,1\n'
                                      'L20,multicopter,bird,1,1\n'
                                      'L20,multicopter,bird,1,1\n'
                                      'L20,multicopter,bird,1,1\n'
                                      'L20,fixed_wing,delta_wing,0,\n'
                                      'L20,fixed_wing,delta_wing,1,1\n'
                                      'L20,fixed_wing,delta_wing,1,1\n'
                                      'L20,fixed_wing,delta_wing,1,1\n'
                                      'L20,delta_wing,fixed_wing,1,0\n'
                                      'L20,delta_wing,fixed_wing,1,0\n'
                                      'L20,delta_wing,fixed_wing,0,\n'
                                      'L20,delta_wing,fixed_wing,1,1\n'
                                      'L20,cruise_missile,delta_wing,1,1\n'
                                      'L20,cruise_missile,delta_wing,1,1\n'
                                      'L20,cruise_missile,delta_wing,1,0\n'
                                      'L20,cruise_missile,delta_wing,1,1\n'
                                      'L20,ballistic_missile,cruise_missile,1,1\n'
                                      'L20,ballistic_missile,cruise_missile,1,1\n'
                                      'L20,ballistic_missile,cruise_missile,1,1\n'
                                      'L20,ballistic_missile,cruise_missile,0,\n'
                                      'L20,bird,multicopter,1,1\n'
                                      'L20,bird,multicopter,1,1\n'
                                      'L20,bird,multicopter,1,0\n'
                                      'L20,bird,multicopter,1,1\n'
                                      'L21,multicopter,bird,1,1\n'
                                      'L21,multicopter,bird,1,1\n'
                                      'L21,multicopter,bird,1,1\n'
                                      'L21,multicopter,bird,1,1\n'
                                      'L21,fixed_wing,delta_wing,0,\n'
                                      'L21,fixed_wing,delta_wing,0,\n'
                                      'L21,fixed_wing,delta_wing,1,1\n'
                                      'L21,fixed_wing,delta_wing,1,1\n'
                                      'L21,delta_wing,fixed_wing,1,1\n'
                                      'L21,delta_wing,fixed_wing,0,\n'
                                      'L21,delta_wing,fixed_wing,0,\n'
                                      'L21,delta_wing,fixed_wing,0,\n'
                                      'L21,cruise_missile,delta_wing,1,1\n'
                                      'L21,cruise_missile,delta_wing,1,1\n'
                                      'L21,cruise_missile,delta_wing,1,1\n'
                                      'L21,cruise_missile,delta_wing,1,1\n'
                                      'L21,ballistic_missile,cruise_missile,1,1\n'
                                      'L21,ballistic_missile,cruise_missile,1,1\n'
                                      'L21,ballistic_missile,cruise_missile,1,1\n'
                                      'L21,ballistic_missile,cruise_missile,1,1\n'
                                      'L21,bird,multicopter,1,1\n'
                                      'L21,bird,multicopter,1,0\n'
                                      'L21,bird,multicopter,1,1\n'
                                      'L21,bird,multicopter,1,1\n'
                                      'L22,multicopter,bird,1,0\n'
                                      'L22,multicopter,bird,1,1\n'
                                      'L22,multicopter,bird,1,1\n'
                                      'L22,multicopter,bird,1,0\n'
                                      'L22,fixed_wing,delta_wing,1,1\n'
                                      'L22,fixed_wing,delta_wing,1,1\n'
                                      'L22,fixed_wing,delta_wing,1,0\n'
                                      'L22,fixed_wing,delta_wing,0,\n'
                                      'L22,delta_wing,fixed_wing,1,0\n'
                                      'L22,delta_wing,fixed_wing,1,1\n'
                                      'L22,delta_wing,fixed_wing,1,1\n'
                                      'L22,delta_wing,fixed_wing,1,0\n'
                                      'L22,cruise_missile,delta_wing,1,0\n'
                                      'L22,cruise_missile,delta_wing,1,0\n'
                                      'L22,cruise_missile,delta_wing,1,0\n'
                                      'L22,cruise_missile,delta_wing,1,1\n'
                                      'L22,ballistic_missile,cruise_missile,1,1\n'
                                      'L22,ballistic_missile,cruise_missile,1,1\n'
                                      'L22,ballistic_missile,cruise_missile,1,1\n'
                                      'L22,ballistic_missile,cruise_missile,1,1\n'
                                      'L22,bird,multicopter,1,1\n'
                                      'L22,bird,multicopter,1,1\n'
                                      'L22,bird,multicopter,1,1\n'
                                      'L22,bird,multicopter,1,1\n'
                                      'L23,multicopter,bird,1,1\n'
                                      'L23,multicopter,bird,1,1\n'
                                      'L23,multicopter,bird,1,1\n'
                                      'L23,multicopter,bird,0,\n'
                                      'L23,fixed_wing,delta_wing,1,1\n'
                                      'L23,fixed_wing,delta_wing,1,1\n'
                                      'L23,fixed_wing,delta_wing,1,1\n'
                                      'L23,fixed_wing,delta_wing,1,1\n'
                                      'L23,delta_wing,fixed_wing,1,1\n'
                                      'L23,delta_wing,fixed_wing,0,\n'
                                      'L23,delta_wing,fixed_wing,1,0\n'
                                      'L23,delta_wing,fixed_wing,0,\n'
                                      'L23,cruise_missile,delta_wing,0,\n'
                                      'L23,cruise_missile,delta_wing,1,0\n'
                                      'L23,cruise_missile,delta_wing,1,0\n'
                                      'L23,cruise_missile,delta_wing,1,1\n'
                                      'L23,ballistic_missile,cruise_missile,1,1\n'
                                      'L23,ballistic_missile,cruise_missile,1,1\n'
                                      'L23,ballistic_missile,cruise_missile,1,1\n'
                                      'L23,ballistic_missile,cruise_missile,1,1\n'
                                      'L23,bird,multicopter,1,1\n'
                                      'L23,bird,multicopter,1,1\n'
                                      'L23,bird,multicopter,1,0\n'
                                      'L23,bird,multicopter,1,1\n'
                                      'L24,multicopter,bird,0,\n'
                                      'L24,multicopter,bird,1,1\n'
                                      'L24,multicopter,bird,1,1\n'
                                      'L24,multicopter,bird,1,1\n'
                                      'L24,fixed_wing,delta_wing,1,1\n'
                                      'L24,fixed_wing,delta_wing,1,1\n'
                                      'L24,fixed_wing,delta_wing,1,1\n'
                                      'L24,fixed_wing,delta_wing,1,0\n'
                                      'L24,delta_wing,fixed_wing,0,\n'
                                      'L24,delta_wing,fixed_wing,0,\n'
                                      'L24,delta_wing,fixed_wing,1,0\n'
                                      'L24,delta_wing,fixed_wing,0,\n'
                                      'L24,cruise_missile,delta_wing,0,\n'
                                      'L24,cruise_missile,delta_wing,1,0\n'
                                      'L24,cruise_missile,delta_wing,1,1\n'
                                      'L24,cruise_missile,delta_wing,1,1\n'
                                      'L24,ballistic_missile,cruise_missile,1,1\n'
                                      'L24,ballistic_missile,cruise_missile,1,1\n'
                                      'L24,ballistic_missile,cruise_missile,1,1\n'
                                      'L24,ballistic_missile,cruise_missile,1,0\n'
                                      'L24,bird,multicopter,1,1\n'
                                      'L24,bird,multicopter,1,1\n'
                                      'L24,bird,multicopter,1,1\n'
                                      'L24,bird,multicopter,1,1\n'
                                      'L25,multicopter,bird,1,1\n'
                                      'L25,multicopter,bird,1,1\n'
                                      'L25,multicopter,bird,1,0\n'
                                      'L25,multicopter,bird,1,1\n'
                                      'L25,fixed_wing,delta_wing,1,1\n'
                                      'L25,fixed_wing,delta_wing,0,\n'
                                      'L25,fixed_wing,delta_wing,1,0\n'
                                      'L25,fixed_wing,delta_wing,1,0\n'
                                      'L25,delta_wing,fixed_wing,0,\n'
                                      'L25,delta_wing,fixed_wing,0,\n'
                                      'L25,delta_wing,fixed_wing,0,\n'
                                      'L25,delta_wing,fixed_wing,1,0\n'
                                      'L25,cruise_missile,delta_wing,1,1\n'
                                      'L25,cruise_missile,delta_wing,1,1\n'
                                      'L25,cruise_missile,delta_wing,1,1\n'
                                      'L25,cruise_missile,delta_wing,1,1\n'
                                      'L25,ballistic_missile,cruise_missile,1,1\n'
                                      'L25,ballistic_missile,cruise_missile,1,1\n'
                                      'L25,ballistic_missile,cruise_missile,1,1\n'
                                      'L25,ballistic_missile,cruise_missile,1,1\n'
                                      'L25,bird,multicopter,1,1\n'
                                      'L25,bird,multicopter,1,1\n'
                                      'L25,bird,multicopter,1,1\n'
                                      'L25,bird,multicopter,1,0\n'
                                      'L26,multicopter,bird,1,1\n'
                                      'L26,multicopter,bird,1,0\n'
                                      'L26,multicopter,bird,1,1\n'
                                      'L26,multicopter,bird,1,1\n'
                                      'L26,fixed_wing,delta_wing,1,0\n'
                                      'L26,fixed_wing,delta_wing,1,1\n'
                                      'L26,fixed_wing,delta_wing,1,1\n'
                                      'L26,fixed_wing,delta_wing,1,1\n'
                                      'L26,delta_wing,fixed_wing,1,1\n'
                                      'L26,delta_wing,fixed_wing,1,0\n'
                                      'L26,delta_wing,fixed_wing,0,\n'
                                      'L26,delta_wing,fixed_wing,1,0\n'
                                      'L26,cruise_missile,delta_wing,1,0\n'
                                      'L26,cruise_missile,delta_wing,1,1\n'
                                      'L26,cruise_missile,delta_wing,1,1\n'
                                      'L26,cruise_missile,delta_wing,1,1\n'
                                      'L26,ballistic_missile,cruise_missile,1,1\n'
                                      'L26,ballistic_missile,cruise_missile,1,1\n'
                                      'L26,ballistic_missile,cruise_missile,1,1\n'
                                      'L26,ballistic_missile,cruise_missile,1,1\n'
                                      'L26,bird,multicopter,1,1\n'
                                      'L26,bird,multicopter,1,1\n'
                                      'L26,bird,multicopter,1,1\n'
                                      'L26,bird,multicopter,1,1\n'
                                      'L27,multicopter,bird,1,1\n'
                                      'L27,multicopter,bird,1,0\n'
                                      'L27,multicopter,bird,1,1\n'
                                      'L27,multicopter,bird,1,1\n'
                                      'L27,fixed_wing,delta_wing,1,1\n'
                                      'L27,fixed_wing,delta_wing,1,1\n'
                                      'L27,fixed_wing,delta_wing,1,1\n'
                                      'L27,fixed_wing,delta_wing,1,1\n'
                                      'L27,delta_wing,fixed_wing,1,0\n'
                                      'L27,delta_wing,fixed_wing,0,\n'
                                      'L27,delta_wing,fixed_wing,0,\n'
                                      'L27,delta_wing,fixed_wing,1,0\n'
                                      'L27,cruise_missile,delta_wing,1,0\n'
                                      'L27,cruise_missile,delta_wing,0,\n'
                                      'L27,cruise_missile,delta_wing,1,1\n'
                                      'L27,cruise_missile,delta_wing,0,\n'
                                      'L27,ballistic_missile,cruise_missile,1,1\n'
                                      'L27,ballistic_missile,cruise_missile,1,1\n'
                                      'L27,ballistic_missile,cruise_missile,1,1\n'
                                      'L27,ballistic_missile,cruise_missile,1,1\n'
                                      'L27,bird,multicopter,1,1\n'
                                      'L27,bird,multicopter,1,1\n'
                                      'L27,bird,multicopter,1,1\n'
                                      'L27,bird,multicopter,1,0\n'
                                      'L28,multicopter,bird,1,1\n'
                                      'L28,multicopter,bird,1,1\n'
                                      'L28,multicopter,bird,1,1\n'
                                      'L28,multicopter,bird,1,1\n'
                                      'L28,fixed_wing,delta_wing,1,0\n'
                                      'L28,fixed_wing,delta_wing,1,1\n'
                                      'L28,fixed_wing,delta_wing,1,1\n'
                                      'L28,fixed_wing,delta_wing,1,0\n'
                                      'L28,delta_wing,fixed_wing,1,1\n'
                                      'L28,delta_wing,fixed_wing,1,1\n'
                                      'L28,delta_wing,fixed_wing,1,1\n'
                                      'L28,delta_wing,fixed_wing,1,0\n'
                                      'L28,cruise_missile,delta_wing,1,1\n'
                                      'L28,cruise_missile,delta_wing,1,1\n'
                                      'L28,cruise_missile,delta_wing,0,\n'
                                      'L28,cruise_missile,delta_wing,1,1\n'
                                      'L28,ballistic_missile,cruise_missile,1,1\n'
                                      'L28,ballistic_missile,cruise_missile,1,1\n'
                                      'L28,ballistic_missile,cruise_missile,1,1\n'
                                      'L28,ballistic_missile,cruise_missile,1,1\n'
                                      'L28,bird,multicopter,1,0\n'
                                      'L28,bird,multicopter,1,1\n'
                                      'L28,bird,multicopter,1,0\n'
                                      'L28,bird,multicopter,1,1\n'
                                      'L29,multicopter,bird,1,1\n'
                                      'L29,multicopter,bird,1,1\n'
                                      'L29,multicopter,bird,1,0\n'
                                      'L29,multicopter,bird,1,1\n'
                                      'L29,fixed_wing,delta_wing,1,1\n'
                                      'L29,fixed_wing,delta_wing,1,1\n'
                                      'L29,fixed_wing,delta_wing,1,1\n'
                                      'L29,fixed_wing,delta_wing,1,1\n'
                                      'L29,delta_wing,fixed_wing,0,\n'
                                      'L29,delta_wing,fixed_wing,1,1\n'
                                      'L29,delta_wing,fixed_wing,1,1\n'
                                      'L29,delta_wing,fixed_wing,0,\n'
                                      'L29,cruise_missile,delta_wing,1,1\n'
                                      'L29,cruise_missile,delta_wing,0,\n'
                                      'L29,cruise_missile,delta_wing,1,1\n'
                                      'L29,cruise_missile,delta_wing,1,1\n'
                                      'L29,ballistic_missile,cruise_missile,1,1\n'
                                      'L29,ballistic_missile,cruise_missile,1,1\n'
                                      'L29,ballistic_missile,cruise_missile,1,1\n'
                                      'L29,ballistic_missile,cruise_missile,1,1\n'
                                      'L29,bird,multicopter,1,1\n'
                                      'L29,bird,multicopter,1,1\n'
                                      'L29,bird,multicopter,1,1\n'
                                      'L29,bird,multicopter,1,1\n'
                                      'L30,multicopter,bird,1,1\n'
                                      'L30,multicopter,bird,1,0\n'
                                      'L30,multicopter,bird,1,1\n'
                                      'L30,multicopter,bird,1,1\n'
                                      'L30,fixed_wing,delta_wing,0,\n'
                                      'L30,fixed_wing,delta_wing,1,0\n'
                                      'L30,fixed_wing,delta_wing,0,\n'
                                      'L30,fixed_wing,delta_wing,1,0\n'
                                      'L30,delta_wing,fixed_wing,0,\n'
                                      'L30,delta_wing,fixed_wing,1,0\n'
                                      'L30,delta_wing,fixed_wing,1,1\n'
                                      'L30,delta_wing,fixed_wing,0,\n'
                                      'L30,cruise_missile,delta_wing,1,1\n'
                                      'L30,cruise_missile,delta_wing,1,1\n'
                                      'L30,cruise_missile,delta_wing,1,0\n'
                                      'L30,cruise_missile,delta_wing,0,\n'
                                      'L30,ballistic_missile,cruise_missile,1,0\n'
                                      'L30,ballistic_missile,cruise_missile,1,1\n'
                                      'L30,ballistic_missile,cruise_missile,1,1\n'
                                      'L30,ballistic_missile,cruise_missile,1,1\n'
                                      'L30,bird,multicopter,1,1\n'
                                      'L30,bird,multicopter,1,1\n'
                                      'L30,bird,multicopter,1,1\n'
                                      'L30,bird,multicopter,1,0\n'
                                      'L31,multicopter,bird,1,1\n'
                                      'L31,multicopter,bird,1,1\n'
                                      'L31,multicopter,bird,1,1\n'
                                      'L31,multicopter,bird,1,1\n'
                                      'L31,fixed_wing,delta_wing,1,0\n'
                                      'L31,fixed_wing,delta_wing,1,1\n'
                                      'L31,fixed_wing,delta_wing,1,1\n'
                                      'L31,fixed_wing,delta_wing,1,1\n'
                                      'L31,delta_wing,fixed_wing,1,1\n'
                                      'L31,delta_wing,fixed_wing,0,\n'
                                      'L31,delta_wing,fixed_wing,0,\n'
                                      'L31,delta_wing,fixed_wing,1,1\n'
                                      'L31,cruise_missile,delta_wing,1,1\n'
                                      'L31,cruise_missile,delta_wing,0,\n'
                                      'L31,cruise_missile,delta_wing,1,0\n'
                                      'L31,cruise_missile,delta_wing,1,0\n'
                                      'L31,ballistic_missile,cruise_missile,1,1\n'
                                      'L31,ballistic_missile,cruise_missile,1,1\n'
                                      'L31,ballistic_missile,cruise_missile,1,1\n'
                                      'L31,ballistic_missile,cruise_missile,1,1\n'
                                      'L31,bird,multicopter,1,1\n'
                                      'L31,bird,multicopter,1,1\n'
                                      'L31,bird,multicopter,1,1\n'
                                      'L31,bird,multicopter,0,\n'
                                      'L32,multicopter,bird,1,1\n'
                                      'L32,multicopter,bird,1,1\n'
                                      'L32,multicopter,bird,1,1\n'
                                      'L32,multicopter,bird,0,\n'
                                      'L32,fixed_wing,delta_wing,1,0\n'
                                      'L32,fixed_wing,delta_wing,0,\n'
                                      'L32,fixed_wing,delta_wing,1,1\n'
                                      'L32,fixed_wing,delta_wing,1,1\n'
                                      'L32,delta_wing,fixed_wing,0,\n'
                                      'L32,delta_wing,fixed_wing,0,\n'
                                      'L32,delta_wing,fixed_wing,1,1\n'
                                      'L32,delta_wing,fixed_wing,1,0\n'
                                      'L32,cruise_missile,delta_wing,1,1\n'
                                      'L32,cruise_missile,delta_wing,0,\n'
                                      'L32,cruise_missile,delta_wing,1,1\n'
                                      'L32,cruise_missile,delta_wing,1,1\n'
                                      'L32,ballistic_missile,cruise_missile,1,0\n'
                                      'L32,ballistic_missile,cruise_missile,1,1\n'
                                      'L32,ballistic_missile,cruise_missile,1,1\n'
                                      'L32,ballistic_missile,cruise_missile,1,1\n'
                                      'L32,bird,multicopter,1,1\n'
                                      'L32,bird,multicopter,1,1\n'
                                      'L32,bird,multicopter,1,1\n'
                                      'L32,bird,multicopter,1,1\n'
                                      'L33,multicopter,bird,1,1\n'
                                      'L33,multicopter,bird,1,1\n'
                                      'L33,multicopter,bird,1,0\n'
                                      'L33,multicopter,bird,1,0\n'
                                      'L33,fixed_wing,delta_wing,1,1\n'
                                      'L33,fixed_wing,delta_wing,1,0\n'
                                      'L33,fixed_wing,delta_wing,1,1\n'
                                      'L33,fixed_wing,delta_wing,0,\n'
                                      'L33,delta_wing,fixed_wing,1,1\n'
                                      'L33,delta_wing,fixed_wing,0,\n'
                                      'L33,delta_wing,fixed_wing,1,1\n'
                                      'L33,delta_wing,fixed_wing,1,0\n'
                                      'L33,cruise_missile,delta_wing,0,\n'
                                      'L33,cruise_missile,delta_wing,1,1\n'
                                      'L33,cruise_missile,delta_wing,1,1\n'
                                      'L33,cruise_missile,delta_wing,1,0\n'
                                      'L33,ballistic_missile,cruise_missile,1,1\n'
                                      'L33,ballistic_missile,cruise_missile,1,1\n'
                                      'L33,ballistic_missile,cruise_missile,1,1\n'
                                      'L33,ballistic_missile,cruise_missile,1,1\n'
                                      'L33,bird,multicopter,1,1\n'
                                      'L33,bird,multicopter,1,0\n'
                                      'L33,bird,multicopter,1,1\n'
                                      'L33,bird,multicopter,1,1\n'
                                      'L34,multicopter,bird,1,1\n'
                                      'L34,multicopter,bird,1,1\n'
                                      'L34,multicopter,bird,1,1\n'
                                      'L34,multicopter,bird,0,\n'
                                      'L34,fixed_wing,delta_wing,1,0\n'
                                      'L34,fixed_wing,delta_wing,1,0\n'
                                      'L34,fixed_wing,delta_wing,1,0\n'
                                      'L34,fixed_wing,delta_wing,0,\n'
                                      'L34,delta_wing,fixed_wing,1,0\n'
                                      'L34,delta_wing,fixed_wing,0,\n'
                                      'L34,delta_wing,fixed_wing,1,1\n'
                                      'L34,delta_wing,fixed_wing,1,1\n'
                                      'L34,cruise_missile,delta_wing,1,0\n'
                                      'L34,cruise_missile,delta_wing,1,1\n'
                                      'L34,cruise_missile,delta_wing,1,1\n'
                                      'L34,cruise_missile,delta_wing,0,\n'
                                      'L34,ballistic_missile,cruise_missile,1,1\n'
                                      'L34,ballistic_missile,cruise_missile,1,1\n'
                                      'L34,ballistic_missile,cruise_missile,1,1\n'
                                      'L34,ballistic_missile,cruise_missile,1,1\n'
                                      'L34,bird,multicopter,1,1\n'
                                      'L34,bird,multicopter,1,1\n'
                                      'L34,bird,multicopter,1,1\n'
                                      'L34,bird,multicopter,1,1\n'
                                      'L35,multicopter,bird,1,1\n'
                                      'L35,multicopter,bird,1,1\n'
                                      'L35,multicopter,bird,1,1\n'
                                      'L35,multicopter,bird,1,1\n'
                                      'L35,fixed_wing,delta_wing,1,1\n'
                                      'L35,fixed_wing,delta_wing,1,1\n'
                                      'L35,fixed_wing,delta_wing,1,0\n'
                                      'L35,fixed_wing,delta_wing,1,0\n'
                                      'L35,delta_wing,fixed_wing,1,0\n'
                                      'L35,delta_wing,fixed_wing,0,\n'
                                      'L35,delta_wing,fixed_wing,1,1\n'
                                      'L35,delta_wing,fixed_wing,1,1\n'
                                      'L35,cruise_missile,delta_wing,1,1\n'
                                      'L35,cruise_missile,delta_wing,1,1\n'
                                      'L35,cruise_missile,delta_wing,1,0\n'
                                      'L35,cruise_missile,delta_wing,1,1\n'
                                      'L35,ballistic_missile,cruise_missile,1,0\n'
                                      'L35,ballistic_missile,cruise_missile,1,1\n'
                                      'L35,ballistic_missile,cruise_missile,1,1\n'
                                      'L35,ballistic_missile,cruise_missile,1,1\n'
                                      'L35,bird,multicopter,1,1\n'
                                      'L35,bird,multicopter,1,1\n'
                                      'L35,bird,multicopter,0,\n'
                                      'L35,bird,multicopter,1,1\n'
                                      'L36,multicopter,bird,1,1\n'
                                      'L36,multicopter,bird,1,0\n'
                                      'L36,multicopter,bird,1,1\n'
                                      'L36,multicopter,bird,1,1\n'
                                      'L36,fixed_wing,delta_wing,1,1\n'
                                      'L36,fixed_wing,delta_wing,1,1\n'
                                      'L36,fixed_wing,delta_wing,1,0\n'
                                      'L36,fixed_wing,delta_wing,1,0\n'
                                      'L36,delta_wing,fixed_wing,0,\n'
                                      'L36,delta_wing,fixed_wing,1,0\n'
                                      'L36,delta_wing,fixed_wing,0,\n'
                                      'L36,delta_wing,fixed_wing,0,\n'
                                      'L36,cruise_missile,delta_wing,1,0\n'
                                      'L36,cruise_missile,delta_wing,1,1\n'
                                      'L36,cruise_missile,delta_wing,0,\n'
                                      'L36,cruise_missile,delta_wing,1,0\n'
                                      'L36,ballistic_missile,cruise_missile,1,1\n'
                                      'L36,ballistic_missile,cruise_missile,1,1\n'
                                      'L36,ballistic_missile,cruise_missile,1,1\n'
                                      'L36,ballistic_missile,cruise_missile,1,1\n'
                                      'L36,bird,multicopter,1,1\n'
                                      'L36,bird,multicopter,1,1\n'
                                      'L36,bird,multicopter,0,\n'
                                      'L36,bird,multicopter,1,0\n'
                                      'L37,multicopter,bird,1,1\n'
                                      'L37,multicopter,bird,1,0\n'
                                      'L37,multicopter,bird,1,1\n'
                                      'L37,multicopter,bird,1,1\n'
                                      'L37,fixed_wing,delta_wing,1,1\n'
                                      'L37,fixed_wing,delta_wing,1,1\n'
                                      'L37,fixed_wing,delta_wing,1,1\n'
                                      'L37,fixed_wing,delta_wing,1,1\n'
                                      'L37,delta_wing,fixed_wing,1,0\n'
                                      'L37,delta_wing,fixed_wing,0,\n'
                                      'L37,delta_wing,fixed_wing,0,\n'
                                      'L37,delta_wing,fixed_wing,1,1\n'
                                      'L37,cruise_missile,delta_wing,0,\n'
                                      'L37,cruise_missile,delta_wing,0,\n'
                                      'L37,cruise_missile,delta_wing,1,0\n'
                                      'L37,cruise_missile,delta_wing,1,1\n'
                                      'L37,ballistic_missile,cruise_missile,1,1\n'
                                      'L37,ballistic_missile,cruise_missile,1,1\n'
                                      'L37,ballistic_missile,cruise_missile,1,1\n'
                                      'L37,ballistic_missile,cruise_missile,1,1\n'
                                      'L37,bird,multicopter,1,0\n'
                                      'L37,bird,multicopter,1,1\n'
                                      'L37,bird,multicopter,1,0\n'
                                      'L37,bird,multicopter,1,0\n'
                                      'L38,multicopter,bird,1,1\n'
                                      'L38,multicopter,bird,1,1\n'
                                      'L38,multicopter,bird,1,1\n'
                                      'L38,multicopter,bird,1,1\n'
                                      'L38,fixed_wing,delta_wing,1,1\n'
                                      'L38,fixed_wing,delta_wing,1,1\n'
                                      'L38,fixed_wing,delta_wing,1,0\n'
                                      'L38,fixed_wing,delta_wing,1,1\n'
                                      'L38,delta_wing,fixed_wing,0,\n'
                                      'L38,delta_wing,fixed_wing,1,1\n'
                                      'L38,delta_wing,fixed_wing,1,0\n'
                                      'L38,delta_wing,fixed_wing,1,0\n'
                                      'L38,cruise_missile,delta_wing,1,1\n'
                                      'L38,cruise_missile,delta_wing,1,1\n'
                                      'L38,cruise_missile,delta_wing,1,1\n'
                                      'L38,cruise_missile,delta_wing,1,0\n'
                                      'L38,ballistic_missile,cruise_missile,1,1\n'
                                      'L38,ballistic_missile,cruise_missile,1,1\n'
                                      'L38,ballistic_missile,cruise_missile,1,1\n'
                                      'L38,ballistic_missile,cruise_missile,1,1\n'
                                      'L38,bird,multicopter,1,1\n'
                                      'L38,bird,multicopter,1,0\n'
                                      'L38,bird,multicopter,1,1\n'
                                      'L38,bird,multicopter,1,1\n'
                                      'L39,multicopter,bird,1,1\n'
                                      'L39,multicopter,bird,1,1\n'
                                      'L39,multicopter,bird,1,1\n'
                                      'L39,multicopter,bird,1,1\n'
                                      'L39,fixed_wing,delta_wing,1,0\n'
                                      'L39,fixed_wing,delta_wing,1,1\n'
                                      'L39,fixed_wing,delta_wing,1,1\n'
                                      'L39,fixed_wing,delta_wing,1,0\n'
                                      'L39,delta_wing,fixed_wing,0,\n'
                                      'L39,delta_wing,fixed_wing,1,1\n'
                                      'L39,delta_wing,fixed_wing,0,\n'
                                      'L39,delta_wing,fixed_wing,1,1\n'
                                      'L39,cruise_missile,delta_wing,0,\n'
                                      'L39,cruise_missile,delta_wing,1,0\n'
                                      'L39,cruise_missile,delta_wing,1,0\n'
                                      'L39,cruise_missile,delta_wing,1,0\n'
                                      'L39,ballistic_missile,cruise_missile,1,1\n'
                                      'L39,ballistic_missile,cruise_missile,1,1\n'
                                      'L39,ballistic_missile,cruise_missile,1,1\n'
                                      'L39,ballistic_missile,cruise_missile,1,1\n'
                                      'L39,bird,multicopter,1,1\n'
                                      'L39,bird,multicopter,1,1\n'
                                      'L39,bird,multicopter,1,1\n'
                                      'L39,bird,multicopter,1,0\n'
                                      'L40,multicopter,bird,1,1\n'
                                      'L40,multicopter,bird,1,1\n'
                                      'L40,multicopter,bird,1,1\n'
                                      'L40,multicopter,bird,1,1\n'
                                      'L40,fixed_wing,delta_wing,1,1\n'
                                      'L40,fixed_wing,delta_wing,1,1\n'
                                      'L40,fixed_wing,delta_wing,1,1\n'
                                      'L40,fixed_wing,delta_wing,1,1\n'
                                      'L40,delta_wing,fixed_wing,1,0\n'
                                      'L40,delta_wing,fixed_wing,0,\n'
                                      'L40,delta_wing,fixed_wing,0,\n'
                                      'L40,delta_wing,fixed_wing,1,1\n'
                                      'L40,cruise_missile,delta_wing,1,0\n'
                                      'L40,cruise_missile,delta_wing,1,0\n'
                                      'L40,cruise_missile,delta_wing,1,0\n'
                                      'L40,cruise_missile,delta_wing,1,1\n'
                                      'L40,ballistic_missile,cruise_missile,1,1\n'
                                      'L40,ballistic_missile,cruise_missile,1,1\n'
                                      'L40,ballistic_missile,cruise_missile,1,1\n'
                                      'L40,ballistic_missile,cruise_missile,1,1\n'
                                      'L40,bird,multicopter,1,1\n'
                                      'L40,bird,multicopter,1,0\n'
                                      'L40,bird,multicopter,1,1\n'
                                      'L40,bird,multicopter,1,1\n',
                        'm9_retrain.csv': 'condition,samples,before_recall_pct,after_recall_pct,fixed_errors,new_errors\n'
                                          'far_white_hot,60,58.3,71.7,11,3\n'
                                          'mid_black_hot,40,72.5,80.0,5,2\n'
                                          'near_white_hot,80,94.1,92.5,1,4\n'
                                          'far_black_hot,0,,,0,0\n',
                        'm10_run.csv': 'run_id,distance,grade,correct,reviewed,ms\n'
                                       'R001,mid,unknown,1,0,46\n'
                                       'R002,mid,check,0,1,45\n'
                                       'R003,mid,show,1,0,36\n'
                                       'R004,near,show,1,0,31\n'
                                       'R005,near,show,1,0,48\n'
                                       'R006,near,show,1,0,34\n'
                                       'R007,near,show,1,0,49\n'
                                       'R008,near,show,1,0,47\n'
                                       'R009,mid,show,1,0,38\n'
                                       'R010,far,check,1,1,65\n'
                                       'R011,near,check,1,1,47\n'
                                       'R012,near,check,1,1,49\n'
                                       'R013,mid,check,0,0,43\n'
                                       'R014,near,show,1,0,52\n'
                                       'R015,near,check,0,1,54\n'
                                       'R016,near,show,1,0,43\n'
                                       'R017,far,check,1,1,38\n'
                                       'R018,near,show,1,0,49\n'
                                       'R019,near,show,1,0,31\n'
                                       'R020,mid,show,1,0,55\n'
                                       'R021,far,check,1,1,53\n'
                                       'R022,far,show,1,0,38\n'
                                       'R023,near,check,1,1,55\n'
                                       'R024,near,show,1,0,61\n'
                                       'R025,near,show,1,0,62\n'
                                       'R026,near,check,1,1,51\n'
                                       'R027,near,check,1,1,64\n'
                                       'R028,far,check,1,1,39\n'
                                       'R029,mid,show,1,0,50\n'
                                       'R030,mid,show,1,0,58\n'
                                       'R031,near,show,1,0,58\n'
                                       'R032,near,check,1,0,58\n'
                                       'R033,near,show,1,0,63\n'
                                       'R034,far,unknown,1,0,58\n'
                                       'R035,near,show,1,0,37\n'
                                       'R036,far,check,1,1,65\n'
                                       'R037,near,show,1,0,62\n'
                                       'R038,mid,show,1,0,49\n'
                                       'R039,near,check,0,1,35\n'
                                       'R040,mid,show,1,0,64\n'
                                       'R041,near,show,1,0,33\n'
                                       'R042,near,show,1,0,42\n'
                                       'R043,mid,check,1,1,32\n'
                                       'R044,near,show,1,0,56\n'
                                       'R045,mid,check,0,1,35\n'
                                       'R046,near,show,1,0,58\n'
                                       'R047,near,show,1,0,50\n'
                                       'R048,near,show,1,0,32\n'
                                       'R049,mid,show,1,0,59\n'
                                       'R050,near,check,1,1,58\n'
                                       'R051,near,show,1,0,28\n'
                                       'R052,near,show,1,0,51\n'
                                       'R053,far,check,1,1,48\n'
                                       'R054,mid,unknown,0,0,36\n'
                                       'R055,near,show,1,0,45\n'
                                       'R056,near,show,1,0,41\n'
                                       'R057,near,show,1,0,55\n'
                                       'R058,near,show,1,0,63\n'
                                       'R059,near,show,1,0,41\n'
                                       'R060,mid,check,1,0,49\n'
                                       'R061,mid,check,0,1,56\n'
                                       'R062,mid,check,0,1,62\n'
                                       'R063,near,show,1,0,37\n'
                                       'R064,near,show,1,0,31\n'
                                       'R065,mid,show,1,0,38\n'
                                       'R066,mid,show,1,0,33\n'
                                       'R067,near,check,1,1,38\n'
                                       'R068,mid,show,1,0,34\n'
                                       'R069,far,check,1,1,29\n'
                                       'R070,near,check,0,1,63\n'
                                       'R071,mid,show,1,0,59\n'
                                       'R072,near,show,1,0,38\n'
                                       'R073,near,show,1,0,35\n'
                                       'R074,mid,check,1,1,62\n'
                                       'R075,mid,check,1,1,40\n'
                                       'R076,near,show,1,0,64\n'
                                       'R077,near,show,1,0,62\n'
                                       'R078,far,unknown,1,0,44\n'
                                       'R079,near,show,1,0,31\n'
                                       'R080,mid,show,1,0,54\n'
                                       'R081,mid,check,1,1,35\n'
                                       'R082,mid,check,1,1,57\n'
                                       'R083,near,show,1,0,64\n'
                                       'R084,mid,unknown,0,0,32\n'
                                       'R085,near,show,1,0,55\n'
                                       'R086,mid,unknown,1,0,50\n'
                                       'R087,near,show,0,0,57\n'
                                       'R088,mid,check,0,1,28\n'
                                       'R089,mid,unknown,0,0,36\n'
                                       'R090,mid,check,1,1,50\n'
                                       'R091,near,show,1,0,58\n'
                                       'R092,far,check,1,1,63\n'
                                       'R093,mid,show,1,0,61\n'
                                       'R094,near,show,1,0,42\n'
                                       'R095,mid,check,1,1,66\n'
                                       'R096,far,unknown,0,0,48\n'
                                       'R097,mid,show,1,0,37\n'
                                       'R098,mid,check,1,1,41\n'
                                       'R099,mid,show,1,0,62\n'
                                       'R100,mid,show,1,0,30\n'
                                       'R101,mid,show,1,0,30\n'
                                       'R102,near,show,1,0,36\n'
                                       'R103,mid,unknown,0,0,36\n'
                                       'R104,mid,show,1,0,29\n'
                                       'R105,mid,unknown,1,0,35\n'
                                       'R106,near,show,1,0,32\n'
                                       'R107,near,check,1,1,34\n'
                                       'R108,mid,check,1,1,57\n'
                                       'R109,mid,show,1,0,35\n'
                                       'R110,near,check,1,0,65\n'
                                       'R111,near,show,1,0,28\n'
                                       'R112,near,show,1,0,52\n'
                                       'R113,near,show,1,0,63\n'
                                       'R114,near,show,1,0,37\n'
                                       'R115,mid,show,1,0,56\n'
                                       'R116,mid,show,1,0,35\n'
                                       'R117,near,show,1,0,60\n'
                                       'R118,mid,show,1,0,39\n'
                                       'R119,mid,unknown,0,0,56\n'
                                       'R120,mid,show,1,0,59\n'
                                       'R121,near,unknown,0,0,34\n'
                                       'R122,far,show,1,0,45\n'
                                       'R123,mid,check,0,1,46\n'
                                       'R124,mid,show,1,0,40\n'
                                       'R125,mid,show,0,0,51\n'
                                       'R126,mid,show,1,0,48\n'
                                       'R127,near,show,1,0,41\n'
                                       'R128,mid,unknown,1,0,37\n'
                                       'R129,near,check,1,0,66\n'
                                       'R130,near,check,0,1,63\n'
                                       'R131,mid,unknown,0,0,52\n'
                                       'R132,mid,show,1,0,32\n'
                                       'R133,mid,check,1,1,33\n'
                                       'R134,mid,check,1,1,36\n'
                                       'R135,near,check,1,1,30\n'
                                       'R136,mid,check,0,1,47\n'
                                       'R137,mid,show,1,0,37\n'
                                       'R138,near,show,1,0,60\n'
                                       'R139,mid,show,1,0,61\n'
                                       'R140,near,unknown,0,0,33\n'
                                       'R141,mid,check,1,1,51\n'
                                       'R142,mid,unknown,0,0,28\n'
                                       'R143,near,show,1,0,53\n'
                                       'R144,near,show,1,0,50\n'
                                       'R145,mid,unknown,0,0,33\n'
                                       'R146,mid,show,1,0,45\n'
                                       'R147,near,show,1,0,35\n'
                                       'R148,mid,show,1,0,30\n'
                                       'R149,near,show,1,0,29\n'
                                       'R150,near,show,1,0,48\n'
                                       'R151,far,unknown,0,0,38\n'
                                       'R152,far,check,0,1,28\n'
                                       'R153,mid,unknown,1,0,49\n'
                                       'R154,near,show,1,0,66\n'
                                       'R155,mid,check,0,1,58\n'
                                       'R156,mid,check,1,1,53\n'
                                       'R157,near,show,1,0,39\n'
                                       'R158,mid,show,1,0,62\n'
                                       'R159,near,show,1,0,36\n'
                                       'R160,near,show,0,0,65\n'
                                       'R161,mid,show,1,0,35\n'
                                       'R162,mid,show,1,0,45\n'
                                       'R163,near,show,1,0,47\n'
                                       'R164,far,unknown,0,0,57\n'
                                       'R165,mid,check,0,1,54\n'
                                       'R166,mid,check,1,1,41\n'
                                       'R167,near,show,1,0,39\n'
                                       'R168,far,unknown,0,0,35\n'
                                       'R169,mid,check,0,1,55\n'
                                       'R170,near,check,1,1,33\n'
                                       'R171,mid,show,1,0,33\n'
                                       'R172,near,check,1,1,64\n'
                                       'R173,near,show,1,0,65\n'
                                       'R174,near,show,1,0,56\n'
                                       'R175,mid,unknown,1,0,39\n'
                                       'R176,mid,unknown,1,0,34\n'
                                       'R177,far,check,1,0,35\n'
                                       'R178,near,show,1,0,40\n'
                                       'R179,near,show,1,0,30\n'
                                       'R180,mid,check,1,1,28\n'
                                       'R181,near,show,1,0,31\n'
                                       'R182,mid,unknown,1,0,39\n'
                                       'R183,near,show,1,0,64\n'
                                       'R184,near,unknown,1,0,59\n'
                                       'R185,near,show,1,0,45\n'
                                       'R186,near,show,1,0,64\n'
                                       'R187,near,show,1,0,43\n'
                                       'R188,mid,show,1,0,53\n'
                                       'R189,mid,check,0,1,56\n'
                                       'R190,far,check,0,1,51\n'
                                       'R191,mid,check,1,1,53\n'
                                       'R192,mid,check,0,1,58\n'
                                       'R193,near,check,0,1,65\n'
                                       'R194,mid,unknown,1,0,44\n'
                                       'R195,mid,show,1,0,49\n'
                                       'R196,mid,show,1,0,29\n'
                                       'R197,mid,show,1,0,39\n'
                                       'R198,near,show,1,0,40\n'
                                       'R199,mid,show,1,0,48\n'
                                       'R200,near,show,1,0,50\n'
                                       'R201,near,show,1,0,57\n'
                                       'R202,near,show,1,0,32\n'
                                       'R203,near,show,1,0,30\n'
                                       'R204,near,unknown,1,0,44\n'
                                       'R205,far,unknown,0,0,36\n'
                                       'R206,mid,check,0,1,31\n'
                                       'R207,near,show,1,0,52\n'
                                       'R208,near,show,1,0,49\n'
                                       'R209,far,check,1,1,59\n'
                                       'R210,near,show,1,0,47\n'
                                       'R211,mid,show,1,0,65\n'
                                       'R212,mid,unknown,0,0,51\n'
                                       'R213,mid,check,0,1,32\n'
                                       'R214,mid,check,0,1,37\n'
                                       'R215,mid,check,1,1,62\n'
                                       'R216,near,unknown,0,0,53\n'
                                       'R217,near,show,1,0,39\n'
                                       'R218,near,show,1,0,49\n'
                                       'R219,near,unknown,0,0,61\n'
                                       'R220,near,show,1,0,32\n'
                                       'R221,near,show,1,0,64\n'
                                       'R222,mid,show,1,0,42\n'
                                       'R223,mid,check,1,1,29\n'
                                       'R224,far,unknown,1,0,39\n'
                                       'R225,mid,show,1,0,36\n'
                                       'R226,near,show,1,0,37\n'
                                       'R227,mid,unknown,1,0,49\n'
                                       'R228,mid,check,1,1,34\n'
                                       'R229,near,show,1,0,57\n'
                                       'R230,mid,check,0,0,30\n'
                                       'R231,near,show,1,0,61\n'
                                       'R232,mid,unknown,0,0,45\n'
                                       'R233,near,show,1,0,66\n'
                                       'R234,mid,check,1,0,64\n'
                                       'R235,near,show,1,0,64\n'
                                       'R236,near,unknown,0,0,62\n'
                                       'R237,near,check,1,1,56\n'
                                       'R238,near,check,1,1,59\n'
                                       'R239,mid,show,1,0,56\n'
                                       'R240,near,show,1,0,55\n'},
             'figures': {'f_loop': '<svg viewBox="0 0 320 156" xmlns="http://www.w3.org/2000/svg" font-family="sans-serif" font-size="10" '
                                   'fill="#111827"><defs><marker id="tfa" viewBox="0 0 10 10" refX="8" refY="5" markerWidth="6" markerHeight="6" '
                                   'orient="auto"><path d="M0,0 L10,5 L0,10 z" fill="#475569"/></marker></defs><ellipse cx="160.0" cy="70" rx="112" '
                                   'ry="48" fill="none" stroke="#e5e7eb" stroke-width="6"/><path d="M198.6,24.9 L248.1,40.3" stroke="#2563eb" '
                                   'stroke-width="2" marker-end="url(#tfa)" fill="none"/><path d="M271.9,71.8 L253.0,96.7" stroke="#2563eb" '
                                   'stroke-width="2" marker-end="url(#tfa)" fill="none"/><path d="M190.6,116.2 L129.4,116.2" stroke="#2563eb" '
                                   'stroke-width="2" marker-end="url(#tfa)" fill="none"/><path d="M67.0,96.7 L48.1,71.8" stroke="#2563eb" '
                                   'stroke-width="2" marker-end="url(#tfa)" fill="none"/><path d="M71.9,40.3 L121.4,24.9" stroke="#2563eb" '
                                   'stroke-width="2" marker-end="url(#tfa)" fill="none"/><rect x="116.0" y="10.0" width="88.0" height="24.0" rx="7" '
                                   'fill="#dbeafe" stroke="#94a3b8"/><text x="160.0" y="25.5" font-size="9" text-anchor="middle" fill="#111827" '
                                   'font-weight="bold">요청</text><rect x="222.5" y="43.2" width="88.0" height="24.0" rx="7" fill="#dbeafe" '
                                   'stroke="#94a3b8"/><text x="266.5" y="58.7" font-size="9" text-anchor="middle" fill="#111827" '
                                   'font-weight="bold">생성</text><rect x="181.8" y="96.8" width="88.0" height="24.0" rx="7" fill="#dbeafe" '
                                   'stroke="#94a3b8"/><text x="225.8" y="112.3" font-size="9" text-anchor="middle" fill="#111827" '
                                   'font-weight="bold">실행</text><rect x="50.2" y="96.8" width="88.0" height="24.0" rx="7" fill="#dbeafe" '
                                   'stroke="#94a3b8"/><text x="94.2" y="112.3" font-size="9" text-anchor="middle" fill="#111827" '
                                   'font-weight="bold">확인</text><rect x="9.5" y="43.2" width="88.0" height="24.0" rx="7" fill="#dbeafe" '
                                   'stroke="#94a3b8"/><text x="53.5" y="58.7" font-size="9" text-anchor="middle" fill="#111827" '
                                   'font-weight="bold">수정 요청</text><text x="160.0" y="74.0" font-size="10.5" text-anchor="middle" fill="#475569" '
                                   'font-weight="bold">바이브 코딩</text><text x="160.0" y="151.0" font-size="9" text-anchor="middle" fill="#475569">확인하고 '
                                   '다시 요청하는 일이 핵심이에요</text></svg>',
                         'f_four': '<svg viewBox="0 0 320 120" xmlns="http://www.w3.org/2000/svg" font-family="sans-serif" font-size="10" '
                                   'fill="#111827"><defs><marker id="tfa" viewBox="0 0 10 10" refX="8" refY="5" markerWidth="6" markerHeight="6" '
                                   'orient="auto"><path d="M0,0 L10,5 L0,10 z" fill="#475569"/></marker></defs><rect x="6.0" y="8" width="154.0" '
                                   'height="18" fill="#e5e7eb" stroke="#fff"/><text x="83.0" y="20.5" font-size="9" text-anchor="middle" '
                                   'fill="#111827" font-weight="bold">칸</text><rect x="160.0" y="8" width="154.0" height="18" fill="#e5e7eb" '
                                   'stroke="#fff"/><text x="237.0" y="20.5" font-size="9" text-anchor="middle" fill="#111827" font-weight="bold">무엇을 '
                                   '적나</text><rect x="6.0" y="26" width="154.0" height="18" fill="#fff" stroke="#e5e7eb"/><text x="83.0" y="38.5" '
                                   'font-size="9" text-anchor="middle" fill="#111827" font-weight="bold">맥락</text><rect x="160.0" y="26" '
                                   'width="154.0" height="18" fill="#fff" stroke="#e5e7eb"/><text x="237.0" y="38.5" font-size="9" '
                                   'text-anchor="middle" fill="#111827">왜 필요한지</text><rect x="6.0" y="44" width="154.0" height="18" fill="#fff" '
                                   'stroke="#e5e7eb"/><text x="83.0" y="56.5" font-size="9" text-anchor="middle" fill="#111827" '
                                   'font-weight="bold">입력</text><rect x="160.0" y="44" width="154.0" height="18" fill="#fff" stroke="#e5e7eb"/><text '
                                   'x="237.0" y="56.5" font-size="9" text-anchor="middle" fill="#111827">파일·열·값</text><rect x="6.0" y="62" '
                                   'width="154.0" height="18" fill="#fff" stroke="#e5e7eb"/><text x="83.0" y="74.5" font-size="9" '
                                   'text-anchor="middle" fill="#111827" font-weight="bold">할 일</text><rect x="160.0" y="62" width="154.0" '
                                   'height="18" fill="#fff" stroke="#e5e7eb"/><text x="237.0" y="74.5" font-size="9" text-anchor="middle" '
                                   'fill="#111827">규칙을 숫자로</text><rect x="6.0" y="80" width="154.0" height="18" fill="#fff" stroke="#e5e7eb"/><text '
                                   'x="83.0" y="92.5" font-size="9" text-anchor="middle" fill="#111827" font-weight="bold">출력 형식</text><rect '
                                   'x="160.0" y="80" width="154.0" height="18" fill="#fff" stroke="#e5e7eb"/><text x="237.0" y="92.5" font-size="9" '
                                   'text-anchor="middle" fill="#111827">받고 싶은 모양</text><text x="160.0" y="115.0" font-size="9" text-anchor="middle" '
                                   'fill="#475569">네 칸을 채우면 AI가 추측할 일이 줄어요</text></svg>',
                         'f_weft': '<svg viewBox="0 0 320 120" xmlns="http://www.w3.org/2000/svg" font-family="sans-serif" font-size="10" '
                                   'fill="#111827"><defs><marker id="tfa" viewBox="0 0 10 10" refX="8" refY="5" markerWidth="6" markerHeight="6" '
                                   'orient="auto"><path d="M0,0 L10,5 L0,10 z" fill="#475569"/></marker></defs><rect x="6.0" y="8" width="154.0" '
                                   'height="18" fill="#e5e7eb" stroke="#fff"/><text x="83.0" y="20.5" font-size="9" text-anchor="middle" '
                                   'fill="#111827" font-weight="bold">부분</text><rect x="160.0" y="8" width="154.0" height="18" fill="#e5e7eb" '
                                   'stroke="#fff"/><text x="237.0" y="20.5" font-size="9" text-anchor="middle" fill="#111827" font-weight="bold">보는 '
                                   '것</text><rect x="6.0" y="26" width="154.0" height="18" fill="#fff" stroke="#e5e7eb"/><text x="83.0" y="38.5" '
                                   'font-size="9" text-anchor="middle" fill="#111827" font-weight="bold">날개 W</text><rect x="160.0" y="26" '
                                   'width="154.0" height="18" fill="#fff" stroke="#e5e7eb"/><text x="237.0" y="38.5" font-size="9" '
                                   'text-anchor="middle" fill="#111827">모양과 위치</text><rect x="6.0" y="44" width="154.0" height="18" fill="#fff" '
                                   'stroke="#e5e7eb"/><text x="83.0" y="56.5" font-size="9" text-anchor="middle" fill="#111827" '
                                   'font-weight="bold">엔진 E</text><rect x="160.0" y="44" width="154.0" height="18" fill="#fff" '
                                   'stroke="#e5e7eb"/><text x="237.0" y="56.5" font-size="9" text-anchor="middle" fill="#111827">종류와 위치</text><rect '
                                   'x="6.0" y="62" width="154.0" height="18" fill="#fff" stroke="#e5e7eb"/><text x="83.0" y="74.5" font-size="9" '
                                   'text-anchor="middle" fill="#111827" font-weight="bold">동체 F</text><rect x="160.0" y="62" width="154.0" '
                                   'height="18" fill="#fff" stroke="#e5e7eb"/><text x="237.0" y="74.5" font-size="9" text-anchor="middle" '
                                   'fill="#111827">굵기와 길이</text><rect x="6.0" y="80" width="154.0" height="18" fill="#fff" stroke="#e5e7eb"/><text '
                                   'x="83.0" y="92.5" font-size="9" text-anchor="middle" fill="#111827" font-weight="bold">꼬리 T</text><rect '
                                   'x="160.0" y="80" width="154.0" height="18" fill="#fff" stroke="#e5e7eb"/><text x="237.0" y="92.5" font-size="9" '
                                   'text-anchor="middle" fill="#111827">모양</text><text x="160.0" y="115.0" font-size="9" text-anchor="middle" '
                                   'fill="#475569">네 부분으로 나눠 보면 단서가 생겨요</text></svg>',
                         'f_johnson': '<svg viewBox="0 0 320 96" xmlns="http://www.w3.org/2000/svg" font-family="sans-serif" font-size="10" '
                                      'fill="#111827"><defs><marker id="tfa" viewBox="0 0 10 10" refX="8" refY="5" markerWidth="6" markerHeight="6" '
                                      'orient="auto"><path d="M0,0 L10,5 L0,10 z" fill="#475569"/></marker></defs><rect x="8.0" y="18" width="99.3" '
                                      'height="30" rx="5" fill="rgb(37,99,235)" fill-opacity="0.35"/><text x="57.7" y="37.0" font-size="9.5" '
                                      'text-anchor="middle" fill="#111827">탐지</text><text x="57.7" y="62.0" font-size="8" text-anchor="middle" '
                                      'fill="#475569">2픽셀~</text><rect x="109.3" y="18" width="99.3" height="30" rx="5" fill="rgb(128,68,136)" '
                                      'fill-opacity="0.35"/><text x="159.0" y="37.0" font-size="9.5" text-anchor="middle" '
                                      'fill="#111827">인식</text><text x="159.0" y="62.0" font-size="8" text-anchor="middle" '
                                      'fill="#475569">8픽셀~</text><rect x="210.7" y="18" width="99.3" height="30" rx="5" fill="rgb(220,38,38)" '
                                      'fill-opacity="0.35"/><text x="260.3" y="37.0" font-size="9.5" text-anchor="middle" '
                                      'fill="#111827">식별</text><text x="260.3" y="62.0" font-size="8" text-anchor="middle" '
                                      'fill="#475569">13픽셀~</text><text x="160.0" y="91.0" font-size="9" text-anchor="middle" fill="#475569">크게 보일수록 '
                                      '더 자세히 알 수 있어요</text></svg>',
                         'f_schema': '<svg viewBox="0 0 320 102" xmlns="http://www.w3.org/2000/svg" font-family="sans-serif" font-size="10" '
                                     'fill="#111827"><defs><marker id="tfa" viewBox="0 0 10 10" refX="8" refY="5" markerWidth="6" markerHeight="6" '
                                     'orient="auto"><path d="M0,0 L10,5 L0,10 z" fill="#475569"/></marker></defs><rect x="6.0" y="8" width="154.0" '
                                     'height="18" fill="#e5e7eb" stroke="#fff"/><text x="83.0" y="20.5" font-size="9" text-anchor="middle" '
                                     'fill="#111827" font-weight="bold">칸</text><rect x="160.0" y="8" width="154.0" height="18" fill="#e5e7eb" '
                                     'stroke="#fff"/><text x="237.0" y="20.5" font-size="9" text-anchor="middle" fill="#111827" '
                                     'font-weight="bold">값</text><rect x="6.0" y="26" width="154.0" height="18" fill="#fff" stroke="#e5e7eb"/><text '
                                     'x="83.0" y="38.5" font-size="9" text-anchor="middle" fill="#111827" font-weight="bold">형상</text><rect '
                                     'x="160.0" y="26" width="154.0" height="18" fill="#fff" stroke="#e5e7eb"/><text x="237.0" y="38.5" '
                                     'font-size="9" text-anchor="middle" fill="#111827">여섯 가지 소문자</text><rect x="6.0" y="44" width="154.0" '
                                     'height="18" fill="#fff" stroke="#e5e7eb"/><text x="83.0" y="56.5" font-size="9" text-anchor="middle" '
                                     'fill="#111827" font-weight="bold">정체</text><rect x="160.0" y="44" width="154.0" height="18" fill="#fff" '
                                     'stroke="#e5e7eb"/><text x="237.0" y="56.5" font-size="9" text-anchor="middle" fill="#111827">미상이 '
                                     '기본값</text><rect x="6.0" y="62" width="154.0" height="18" fill="#fff" stroke="#e5e7eb"/><text x="83.0" y="74.5" '
                                     'font-size="9" text-anchor="middle" fill="#111827" font-weight="bold">배경</text><rect x="160.0" y="62" '
                                     'width="154.0" height="18" fill="#fff" stroke="#e5e7eb"/><text x="237.0" y="74.5" font-size="9" '
                                     'text-anchor="middle" fill="#111827">sky·field·urban</text><text x="160.0" y="97.0" font-size="9" '
                                     'text-anchor="middle" fill="#475569">모양과 정체를 다른 칸에 적어요</text></svg>',
                         'f_kappa': '<svg viewBox="0 0 320 92" xmlns="http://www.w3.org/2000/svg" font-family="sans-serif" font-size="10" '
                                    'fill="#111827"><defs><marker id="tfa" viewBox="0 0 10 10" refX="8" refY="5" markerWidth="6" markerHeight="6" '
                                    'orient="auto"><path d="M0,0 L10,5 L0,10 z" fill="#475569"/></marker></defs><rect x="8.0" y="14.0" width="92.0" '
                                    'height="46.0" rx="7" fill="#dbeafe" stroke="#94a3b8"/><text x="54.0" y="34.0" font-size="10" '
                                    'text-anchor="middle" fill="#111827" font-weight="bold">일치율</text><text x="54.0" y="47.0" font-size="8.5" '
                                    'text-anchor="middle" fill="#475569">실제로 같음</text><path d="M101.0,37.0 L113.0,37.0" stroke="#475569" '
                                    'stroke-width="1.6" marker-end="url(#tfa)" fill="none"/><rect x="114.0" y="14.0" width="92.0" height="46.0" '
                                    'rx="7" fill="#dbeafe" stroke="#94a3b8"/><text x="160.0" y="34.0" font-size="10" text-anchor="middle" '
                                    'fill="#111827" font-weight="bold">우연</text><text x="160.0" y="47.0" font-size="8.5" text-anchor="middle" '
                                    'fill="#475569">그냥 같을 확률</text><path d="M207.0,37.0 L219.0,37.0" stroke="#475569" stroke-width="1.6" '
                                    'marker-end="url(#tfa)" fill="none"/><rect x="220.0" y="14.0" width="92.0" height="46.0" rx="7" fill="#dbeafe" '
                                    'stroke="#94a3b8"/><text x="266.0" y="34.0" font-size="10" text-anchor="middle" fill="#111827" '
                                    'font-weight="bold">코헨 카파</text><text x="266.0" y="47.0" font-size="8.5" text-anchor="middle" fill="#475569">우연을 '
                                    '뺀 값</text><text x="160.0" y="87.0" font-size="9" text-anchor="middle" fill="#475569">우연히 같은 만큼을 빼고 '
                                    '봐요</text></svg>',
                         'f_clean': '<svg viewBox="0 0 320 92" xmlns="http://www.w3.org/2000/svg" font-family="sans-serif" font-size="10" '
                                    'fill="#111827"><defs><marker id="tfa" viewBox="0 0 10 10" refX="8" refY="5" markerWidth="6" markerHeight="6" '
                                    'orient="auto"><path d="M0,0 L10,5 L0,10 z" fill="#475569"/></marker></defs><rect x="8.0" y="14.0" width="92.0" '
                                    'height="46.0" rx="7" fill="#dbeafe" stroke="#94a3b8"/><text x="54.0" y="34.0" font-size="10" '
                                    'text-anchor="middle" fill="#111827" font-weight="bold">Bird</text><text x="54.0" y="47.0" font-size="8.5" '
                                    'text-anchor="middle" fill="#475569">대문자</text><path d="M101.0,37.0 L113.0,37.0" stroke="#475569" '
                                    'stroke-width="1.6" marker-end="url(#tfa)" fill="none"/><rect x="114.0" y="14.0" width="92.0" height="46.0" '
                                    'rx="7" fill="#dbeafe" stroke="#94a3b8"/><text x="160.0" y="34.0" font-size="10" text-anchor="middle" '
                                    'fill="#111827" font-weight="bold">bird </text><text x="160.0" y="47.0" font-size="8.5" text-anchor="middle" '
                                    'fill="#475569">빈칸</text><path d="M207.0,37.0 L219.0,37.0" stroke="#475569" stroke-width="1.6" '
                                    'marker-end="url(#tfa)" fill="none"/><rect x="220.0" y="14.0" width="92.0" height="46.0" rx="7" fill="#dbeafe" '
                                    'stroke="#94a3b8"/><text x="266.0" y="34.0" font-size="10" text-anchor="middle" fill="#111827" '
                                    'font-weight="bold">bird</text><text x="266.0" y="47.0" font-size="8.5" text-anchor="middle" fill="#475569">같은 '
                                    '값</text><text x="160.0" y="87.0" font-size="9" text-anchor="middle" fill="#475569">제각각인 값을 하나로 맞춰요</text></svg>',
                         'f_base': '<svg viewBox="0 0 320 92" xmlns="http://www.w3.org/2000/svg" font-family="sans-serif" font-size="10" '
                                   'fill="#111827"><defs><marker id="tfa" viewBox="0 0 10 10" refX="8" refY="5" markerWidth="6" markerHeight="6" '
                                   'orient="auto"><path d="M0,0 L10,5 L0,10 z" fill="#475569"/></marker></defs><rect x="8.0" y="14.0" width="65.5" '
                                   'height="46.0" rx="7" fill="#dbeafe" stroke="#94a3b8"/><text x="40.8" y="40.5" font-size="9.5" '
                                   'text-anchor="middle" fill="#111827" font-weight="bold">단순한 모델</text><path d="M74.5,37.0 L86.5,37.0" '
                                   'stroke="#475569" stroke-width="1.6" marker-end="url(#tfa)" fill="none"/><rect x="87.5" y="14.0" width="65.5" '
                                   'height="46.0" rx="7" fill="#dbeafe" stroke="#94a3b8"/><text x="120.2" y="40.5" font-size="10" '
                                   'text-anchor="middle" fill="#111827" font-weight="bold">성능 재기</text><path d="M154.0,37.0 L166.0,37.0" '
                                   'stroke="#475569" stroke-width="1.6" marker-end="url(#tfa)" fill="none"/><rect x="167.0" y="14.0" width="65.5" '
                                   'height="46.0" rx="7" fill="#dbeafe" stroke="#94a3b8"/><text x="199.8" y="40.5" font-size="9.5" '
                                   'text-anchor="middle" fill="#111827" font-weight="bold">복잡한 모델</text><path d="M233.5,37.0 L245.5,37.0" '
                                   'stroke="#475569" stroke-width="1.6" marker-end="url(#tfa)" fill="none"/><rect x="246.5" y="14.0" width="65.5" '
                                   'height="46.0" rx="7" fill="#dbeafe" stroke="#94a3b8"/><text x="279.2" y="40.5" font-size="10" '
                                   'text-anchor="middle" fill="#111827" font-weight="bold">차이 비교</text><text x="160.0" y="87.0" font-size="9" '
                                   'text-anchor="middle" fill="#475569">단순한 모델이 비교의 자가 돼요</text></svg>',
                         'f_pre': '<svg viewBox="0 0 320 92" xmlns="http://www.w3.org/2000/svg" font-family="sans-serif" font-size="10" '
                                  'fill="#111827"><defs><marker id="tfa" viewBox="0 0 10 10" refX="8" refY="5" markerWidth="6" markerHeight="6" '
                                  'orient="auto"><path d="M0,0 L10,5 L0,10 z" fill="#475569"/></marker></defs><rect x="8.0" y="14.0" width="92.0" '
                                  'height="46.0" rx="7" fill="#dbeafe" stroke="#94a3b8"/><text x="54.0" y="40.5" font-size="10" text-anchor="middle" '
                                  'fill="#111827" font-weight="bold">잡음 제거</text><path d="M101.0,37.0 L113.0,37.0" stroke="#475569" '
                                  'stroke-width="1.6" marker-end="url(#tfa)" fill="none"/><rect x="114.0" y="14.0" width="92.0" height="46.0" rx="7" '
                                  'fill="#dbeafe" stroke="#94a3b8"/><text x="160.0" y="40.5" font-size="10" text-anchor="middle" fill="#111827" '
                                  'font-weight="bold">크기 맞춤</text><path d="M207.0,37.0 L219.0,37.0" stroke="#475569" stroke-width="1.6" '
                                  'marker-end="url(#tfa)" fill="none"/><rect x="220.0" y="14.0" width="92.0" height="46.0" rx="7" fill="#dbeafe" '
                                  'stroke="#94a3b8"/><text x="266.0" y="40.5" font-size="10" text-anchor="middle" fill="#111827" '
                                  'font-weight="bold">특징 계산</text><text x="160.0" y="87.0" font-size="9" text-anchor="middle" fill="#475569">같은 기준으로 '
                                  '맞춘 뒤 숫자로 바꿔요</text></svg>',
                         'f_debug': '<svg viewBox="0 0 320 92" xmlns="http://www.w3.org/2000/svg" font-family="sans-serif" font-size="10" '
                                    'fill="#111827"><defs><marker id="tfa" viewBox="0 0 10 10" refX="8" refY="5" markerWidth="6" markerHeight="6" '
                                    'orient="auto"><path d="M0,0 L10,5 L0,10 z" fill="#475569"/></marker></defs><rect x="8.0" y="14.0" width="92.0" '
                                    'height="46.0" rx="7" fill="#dbeafe" stroke="#94a3b8"/><text x="54.0" y="40.5" font-size="9.5" '
                                    'text-anchor="middle" fill="#111827" font-weight="bold">오류 메시지</text><path d="M101.0,37.0 L113.0,37.0" '
                                    'stroke="#475569" stroke-width="1.6" marker-end="url(#tfa)" fill="none"/><rect x="114.0" y="14.0" width="92.0" '
                                    'height="46.0" rx="7" fill="#dbeafe" stroke="#94a3b8"/><text x="160.0" y="40.5" font-size="9.5" '
                                    'text-anchor="middle" fill="#111827" font-weight="bold">원하는 동작</text><path d="M207.0,37.0 L219.0,37.0" '
                                    'stroke="#475569" stroke-width="1.6" marker-end="url(#tfa)" fill="none"/><rect x="220.0" y="14.0" width="92.0" '
                                    'height="46.0" rx="7" fill="#dbeafe" stroke="#94a3b8"/><text x="266.0" y="40.5" font-size="10" '
                                    'text-anchor="middle" fill="#111827" font-weight="bold">고칠 범위</text><text x="160.0" y="87.0" font-size="9" '
                                    'text-anchor="middle" fill="#475569">이 세 가지를 담아 다시 요청해요</text></svg>',
                         'f_req': '<svg viewBox="0 0 320 120" xmlns="http://www.w3.org/2000/svg" font-family="sans-serif" font-size="10" '
                                  'fill="#111827"><defs><marker id="tfa" viewBox="0 0 10 10" refX="8" refY="5" markerWidth="6" markerHeight="6" '
                                  'orient="auto"><path d="M0,0 L10,5 L0,10 z" fill="#475569"/></marker></defs><rect x="6.0" y="8" width="154.0" '
                                  'height="18" fill="#e5e7eb" stroke="#fff"/><text x="83.0" y="20.5" font-size="9" text-anchor="middle" '
                                  'fill="#111827" font-weight="bold">항목</text><rect x="160.0" y="8" width="154.0" height="18" fill="#e5e7eb" '
                                  'stroke="#fff"/><text x="237.0" y="20.5" font-size="9" text-anchor="middle" fill="#111827" '
                                  'font-weight="bold">기준</text><rect x="6.0" y="26" width="154.0" height="18" fill="#fff" stroke="#e5e7eb"/><text '
                                  'x="83.0" y="38.5" font-size="9" text-anchor="middle" fill="#111827" font-weight="bold">재현율</text><rect x="160.0" '
                                  'y="26" width="154.0" height="18" fill="#fff" stroke="#e5e7eb"/><text x="237.0" y="38.5" font-size="9" '
                                  'text-anchor="middle" fill="#111827">90% 이상</text><rect x="6.0" y="44" width="154.0" height="18" fill="#fff" '
                                  'stroke="#e5e7eb"/><text x="83.0" y="56.5" font-size="9" text-anchor="middle" fill="#111827" '
                                  'font-weight="bold">속도</text><rect x="160.0" y="44" width="154.0" height="18" fill="#fff" stroke="#e5e7eb"/><text '
                                  'x="237.0" y="56.5" font-size="9" text-anchor="middle" fill="#111827">100ms 이하</text><rect x="6.0" y="62" '
                                  'width="154.0" height="18" fill="#fff" stroke="#e5e7eb"/><text x="83.0" y="74.5" font-size="9" '
                                  'text-anchor="middle" fill="#111827" font-weight="bold">크기</text><rect x="160.0" y="62" width="154.0" height="18" '
                                  'fill="#fff" stroke="#e5e7eb"/><text x="237.0" y="74.5" font-size="9" text-anchor="middle" fill="#111827">30MB '
                                  '이하</text><rect x="6.0" y="80" width="154.0" height="18" fill="#fff" stroke="#e5e7eb"/><text x="83.0" y="92.5" '
                                  'font-size="9" text-anchor="middle" fill="#111827" font-weight="bold">외부 전송</text><rect x="160.0" y="80" '
                                  'width="154.0" height="18" fill="#fff" stroke="#e5e7eb"/><text x="237.0" y="92.5" font-size="9" '
                                  'text-anchor="middle" fill="#111827">없음</text><text x="160.0" y="115.0" font-size="9" text-anchor="middle" '
                                  'fill="#475569">모델을 보기 전에 합격선을 정해요</text></svg>',
                         'f_tl': '<svg viewBox="0 0 320 92" xmlns="http://www.w3.org/2000/svg" font-family="sans-serif" font-size="10" '
                                 'fill="#111827"><defs><marker id="tfa" viewBox="0 0 10 10" refX="8" refY="5" markerWidth="6" markerHeight="6" '
                                 'orient="auto"><path d="M0,0 L10,5 L0,10 z" fill="#475569"/></marker></defs><rect x="8.0" y="14.0" width="92.0" '
                                 'height="46.0" rx="7" fill="#dbeafe" stroke="#94a3b8"/><text x="54.0" y="34.0" font-size="10" text-anchor="middle" '
                                 'fill="#111827" font-weight="bold">큰 모델</text><text x="54.0" y="47.0" font-size="8.5" text-anchor="middle" '
                                 'fill="#475569">이미 학습됨</text><path d="M101.0,37.0 L113.0,37.0" stroke="#475569" stroke-width="1.6" '
                                 'marker-end="url(#tfa)" fill="none"/><rect x="114.0" y="14.0" width="92.0" height="46.0" rx="7" fill="#dbeafe" '
                                 'stroke="#94a3b8"/><text x="160.0" y="34.0" font-size="10" text-anchor="middle" fill="#111827" '
                                 'font-weight="bold">앞부분</text><text x="160.0" y="47.0" font-size="8.5" text-anchor="middle" '
                                 'fill="#475569">그대로</text><path d="M207.0,37.0 L219.0,37.0" stroke="#475569" stroke-width="1.6" '
                                 'marker-end="url(#tfa)" fill="none"/><rect x="220.0" y="14.0" width="92.0" height="46.0" rx="7" fill="#dbeafe" '
                                 'stroke="#94a3b8"/><text x="266.0" y="34.0" font-size="10" text-anchor="middle" fill="#111827" '
                                 'font-weight="bold">끝부분</text><text x="266.0" y="47.0" font-size="8.5" text-anchor="middle" fill="#475569">새로 '
                                 '학습</text><text x="160.0" y="87.0" font-size="9" text-anchor="middle" fill="#475569">이미 배운 것을 빌려 와요</text></svg>',
                         'f_param': '<svg viewBox="0 0 320 92" xmlns="http://www.w3.org/2000/svg" font-family="sans-serif" font-size="10" '
                                    'fill="#111827"><defs><marker id="tfa" viewBox="0 0 10 10" refX="8" refY="5" markerWidth="6" markerHeight="6" '
                                    'orient="auto"><path d="M0,0 L10,5 L0,10 z" fill="#475569"/></marker></defs><rect x="8.0" y="14.0" width="65.5" '
                                    'height="46.0" rx="7" fill="#dbeafe" stroke="#94a3b8"/><text x="40.8" y="40.5" font-size="10" '
                                    'text-anchor="middle" fill="#111827" font-weight="bold">작은 함수</text><path d="M74.5,37.0 L86.5,37.0" '
                                    'stroke="#475569" stroke-width="1.6" marker-end="url(#tfa)" fill="none"/><rect x="87.5" y="14.0" width="65.5" '
                                    'height="46.0" rx="7" fill="#dbeafe" stroke="#94a3b8"/><text x="120.2" y="40.5" font-size="9.5" '
                                    'text-anchor="middle" fill="#111827" font-weight="bold">예시로 확인</text><path d="M154.0,37.0 L166.0,37.0" '
                                    'stroke="#475569" stroke-width="1.6" marker-end="url(#tfa)" fill="none"/><rect x="167.0" y="14.0" width="65.5" '
                                    'height="46.0" rx="7" fill="#dbeafe" stroke="#94a3b8"/><text x="199.8" y="40.5" font-size="10" '
                                    'text-anchor="middle" fill="#111827" font-weight="bold">전체 적용</text><path d="M233.5,37.0 L245.5,37.0" '
                                    'stroke="#475569" stroke-width="1.6" marker-end="url(#tfa)" fill="none"/><rect x="246.5" y="14.0" width="65.5" '
                                    'height="46.0" rx="7" fill="#dbeafe" stroke="#94a3b8"/><text x="279.2" y="40.5" font-size="10" '
                                    'text-anchor="middle" fill="#111827" font-weight="bold">비교표</text><text x="160.0" y="87.0" font-size="9" '
                                    'text-anchor="middle" fill="#475569">작게 만들고 확인한 뒤 넓혀요</text></svg>',
                         'f_rel': '<svg viewBox="0 0 320 150" xmlns="http://www.w3.org/2000/svg" font-family="sans-serif" font-size="10" '
                                  'fill="#111827"><defs><marker id="tfa" viewBox="0 0 10 10" refX="8" refY="5" markerWidth="6" markerHeight="6" '
                                  'orient="auto"><path d="M0,0 L10,5 L0,10 z" fill="#475569"/></marker></defs><rect x="34" y="12" width="274" '
                                  'height="108" fill="#fff" stroke="#94a3b8"/><line x1="34.0" y1="120.0" x2="308.0" y2="12.0" stroke="#94a3b8" '
                                  'stroke-dasharray="4 3"/><text x="304.0" y="24.0" font-size="8.5" text-anchor="end" '
                                  'fill="#475569">완벽</text><polyline points="157.3,98.4 184.7,87.6 212.1,76.8 239.5,60.6 266.9,44.4 294.3,25.0" '
                                  'fill="none" stroke="#2563eb" stroke-width="2"/><circle cx="157.3" cy="98.4" r="2.5" fill="#2563eb"/><circle '
                                  'cx="184.7" cy="87.6" r="2.5" fill="#2563eb"/><circle cx="212.1" cy="76.8" r="2.5" fill="#2563eb"/><circle '
                                  'cx="239.5" cy="60.6" r="2.5" fill="#2563eb"/><circle cx="266.9" cy="44.4" r="2.5" fill="#2563eb"/><circle '
                                  'cx="294.3" cy="25.0" r="2.5" fill="#2563eb"/><text x="171.0" y="142.0" font-size="9" text-anchor="middle" '
                                  'fill="#475569">mean score</text><text x="10" y="66.0" font-size="9" text-anchor="middle" fill="#475569" '
                                  'transform="rotate(-90 10 66.0)">actual accuracy</text><text x="40.0" y="24.0" font-size="8.5" text-anchor="start" '
                                  'fill="#475569">대각선 아래면 점수가 부풀어 있어요</text></svg>',
                         'f_ece': '<svg viewBox="0 0 320 84" xmlns="http://www.w3.org/2000/svg" font-family="sans-serif" font-size="10" '
                                  'fill="#111827"><defs><marker id="tfa" viewBox="0 0 10 10" refX="8" refY="5" markerWidth="6" markerHeight="6" '
                                  'orient="auto"><path d="M0,0 L10,5 L0,10 z" fill="#475569"/></marker></defs><rect x="6.0" y="8" width="77.0" '
                                  'height="18" fill="#e5e7eb" stroke="#fff"/><text x="44.5" y="20.5" font-size="9" text-anchor="middle" '
                                  'fill="#111827" font-weight="bold">구간</text><rect x="83.0" y="8" width="77.0" height="18" fill="#e5e7eb" '
                                  'stroke="#fff"/><text x="121.5" y="20.5" font-size="9" text-anchor="middle" fill="#111827" font-weight="bold">평균 '
                                  '점수</text><rect x="160.0" y="8" width="77.0" height="18" fill="#e5e7eb" stroke="#fff"/><text x="198.5" y="20.5" '
                                  'font-size="9" text-anchor="middle" fill="#111827" font-weight="bold">실제</text><rect x="237.0" y="8" width="77.0" '
                                  'height="18" fill="#e5e7eb" stroke="#fff"/><text x="275.5" y="20.5" font-size="9" text-anchor="middle" '
                                  'fill="#111827" font-weight="bold">차이</text><rect x="6.0" y="26" width="77.0" height="18" fill="#fff" '
                                  'stroke="#e5e7eb"/><text x="44.5" y="38.5" font-size="9" text-anchor="middle" fill="#111827" '
                                  'font-weight="bold">0.9~1.0</text><rect x="83.0" y="26" width="77.0" height="18" fill="#fff" '
                                  'stroke="#e5e7eb"/><text x="121.5" y="38.5" font-size="9" text-anchor="middle" fill="#111827">0.95</text><rect '
                                  'x="160.0" y="26" width="77.0" height="18" fill="#fff" stroke="#e5e7eb"/><text x="198.5" y="38.5" font-size="9" '
                                  'text-anchor="middle" fill="#111827">0.88</text><rect x="237.0" y="26" width="77.0" height="18" fill="#fff" '
                                  'stroke="#e5e7eb"/><text x="275.5" y="38.5" font-size="9" text-anchor="middle" fill="#111827">0.07</text><rect '
                                  'x="6.0" y="44" width="77.0" height="18" fill="#fff" stroke="#e5e7eb"/><text x="44.5" y="56.5" font-size="9" '
                                  'text-anchor="middle" fill="#111827" font-weight="bold">0.8~0.9</text><rect x="83.0" y="44" width="77.0" '
                                  'height="18" fill="#fff" stroke="#e5e7eb"/><text x="121.5" y="56.5" font-size="9" text-anchor="middle" '
                                  'fill="#111827">0.85</text><rect x="160.0" y="44" width="77.0" height="18" fill="#fff" stroke="#e5e7eb"/><text '
                                  'x="198.5" y="56.5" font-size="9" text-anchor="middle" fill="#111827">0.70</text><rect x="237.0" y="44" '
                                  'width="77.0" height="18" fill="#fff" stroke="#e5e7eb"/><text x="275.5" y="56.5" font-size="9" '
                                  'text-anchor="middle" fill="#111827">0.15</text><text x="160.0" y="79.0" font-size="9" text-anchor="middle" '
                                  'fill="#475569">차이를 표본 수로 가중 평균해요</text></svg>',
                         'f_grade': '<svg viewBox="0 0 320 92" xmlns="http://www.w3.org/2000/svg" font-family="sans-serif" font-size="10" '
                                    'fill="#111827"><defs><marker id="tfa" viewBox="0 0 10 10" refX="8" refY="5" markerWidth="6" markerHeight="6" '
                                    'orient="auto"><path d="M0,0 L10,5 L0,10 z" fill="#475569"/></marker></defs><rect x="8.0" y="14.0" width="92.0" '
                                    'height="46.0" rx="7" fill="#dbeafe" stroke="#94a3b8"/><text x="54.0" y="34.0" font-size="9.5" '
                                    'text-anchor="middle" fill="#111827" font-weight="bold">참고 판정 표시</text><text x="54.0" y="47.0" font-size="8.5" '
                                    'text-anchor="middle" fill="#475569">확신 높음</text><path d="M101.0,37.0 L113.0,37.0" stroke="#475569" '
                                    'stroke-width="1.6" marker-end="url(#tfa)" fill="none"/><rect x="114.0" y="14.0" width="92.0" height="46.0" '
                                    'rx="7" fill="#dbeafe" stroke="#94a3b8"/><text x="160.0" y="34.0" font-size="10" text-anchor="middle" '
                                    'fill="#111827" font-weight="bold">교관 확인</text><text x="160.0" y="47.0" font-size="8.5" text-anchor="middle" '
                                    'fill="#475569">중간</text><path d="M207.0,37.0 L219.0,37.0" stroke="#475569" stroke-width="1.6" '
                                    'marker-end="url(#tfa)" fill="none"/><rect x="220.0" y="14.0" width="92.0" height="46.0" rx="7" fill="#dbeafe" '
                                    'stroke="#94a3b8"/><text x="266.0" y="34.0" font-size="10" text-anchor="middle" fill="#111827" '
                                    'font-weight="bold">판별 불가</text><text x="266.0" y="47.0" font-size="8.5" text-anchor="middle" '
                                    'fill="#475569">1·2위 차이 작음</text><text x="160.0" y="87.0" font-size="9" text-anchor="middle" fill="#475569">모르면 '
                                    '사람에게 넘겨요</text></svg>',
                         'f_leak': '<svg viewBox="0 0 320 92" xmlns="http://www.w3.org/2000/svg" font-family="sans-serif" font-size="10" '
                                   'fill="#111827"><defs><marker id="tfa" viewBox="0 0 10 10" refX="8" refY="5" markerWidth="6" markerHeight="6" '
                                   'orient="auto"><path d="M0,0 L10,5 L0,10 z" fill="#475569"/></marker></defs><rect x="8.0" y="14.0" width="145.0" '
                                   'height="46.0" rx="7" fill="#dbeafe" stroke="#94a3b8"/><text x="80.5" y="34.0" font-size="9.5" '
                                   'text-anchor="middle" fill="#111827" font-weight="bold">검증 데이터</text><text x="80.5" y="47.0" font-size="8.5" '
                                   'text-anchor="middle" fill="#475569">기준 고르기</text><path d="M154.0,37.0 L166.0,37.0" stroke="#475569" '
                                   'stroke-width="1.6" marker-end="url(#tfa)" fill="none"/><rect x="167.0" y="14.0" width="145.0" height="46.0" '
                                   'rx="7" fill="#dbeafe" stroke="#94a3b8"/><text x="239.5" y="34.0" font-size="9.5" text-anchor="middle" '
                                   'fill="#111827" font-weight="bold">테스트 데이터</text><text x="239.5" y="47.0" font-size="8.5" text-anchor="middle" '
                                   'fill="#475569">한 번 확인</text><text x="160.0" y="87.0" font-size="9" text-anchor="middle" fill="#475569">고를 때 쓴 '
                                   '데이터로 성능을 재면 좋게 보여요</text></svg>',
                         'f_verify': '<svg viewBox="0 0 320 92" xmlns="http://www.w3.org/2000/svg" font-family="sans-serif" font-size="10" '
                                     'fill="#111827"><defs><marker id="tfa" viewBox="0 0 10 10" refX="8" refY="5" markerWidth="6" markerHeight="6" '
                                     'orient="auto"><path d="M0,0 L10,5 L0,10 z" fill="#475569"/></marker></defs><rect x="8.0" y="14.0" width="65.5" '
                                     'height="46.0" rx="7" fill="#dbeafe" stroke="#94a3b8"/><text x="40.8" y="40.5" font-size="10" '
                                     'text-anchor="middle" fill="#111827" font-weight="bold">AI 답변</text><path d="M74.5,37.0 L86.5,37.0" '
                                     'stroke="#475569" stroke-width="1.6" marker-end="url(#tfa)" fill="none"/><rect x="87.5" y="14.0" width="65.5" '
                                     'height="46.0" rx="7" fill="#dbeafe" stroke="#94a3b8"/><text x="120.2" y="40.5" font-size="10" '
                                     'text-anchor="middle" fill="#111827" font-weight="bold">근거 수치</text><path d="M154.0,37.0 L166.0,37.0" '
                                     'stroke="#475569" stroke-width="1.6" marker-end="url(#tfa)" fill="none"/><rect x="167.0" y="14.0" width="65.5" '
                                     'height="46.0" rx="7" fill="#dbeafe" stroke="#94a3b8"/><text x="199.8" y="40.5" font-size="10" '
                                     'text-anchor="middle" fill="#111827" font-weight="bold">작은 예시</text><path d="M233.5,37.0 L245.5,37.0" '
                                     'stroke="#475569" stroke-width="1.6" marker-end="url(#tfa)" fill="none"/><rect x="246.5" y="14.0" width="65.5" '
                                     'height="46.0" rx="7" fill="#dbeafe" stroke="#94a3b8"/><text x="279.2" y="40.5" font-size="10" '
                                     'text-anchor="middle" fill="#111827" font-weight="bold">판단</text><text x="160.0" y="87.0" font-size="9" '
                                     'text-anchor="middle" fill="#475569">답변만 믿지 말고 근거를 함께 받아요</text></svg>',
                         'f_pair': '<svg viewBox="0 0 320 84" xmlns="http://www.w3.org/2000/svg" font-family="sans-serif" font-size="10" '
                                   'fill="#111827"><defs><marker id="tfa" viewBox="0 0 10 10" refX="8" refY="5" markerWidth="6" markerHeight="6" '
                                   'orient="auto"><path d="M0,0 L10,5 L0,10 z" fill="#475569"/></marker></defs><rect x="6.0" y="8" width="102.7" '
                                   'height="18" fill="#e5e7eb" stroke="#fff"/><text x="57.3" y="20.5" font-size="9" text-anchor="middle" '
                                   'fill="#111827" font-weight="bold">형상 계열</text><rect x="108.7" y="8" width="102.7" height="18" fill="#e5e7eb" '
                                   'stroke="#fff"/><text x="160.0" y="20.5" font-size="9" text-anchor="middle" fill="#111827" '
                                   'font-weight="bold">헷갈리는 계열</text><rect x="211.3" y="8" width="102.7" height="18" fill="#e5e7eb" '
                                   'stroke="#fff"/><text x="262.7" y="20.5" font-size="9" text-anchor="middle" fill="#111827" '
                                   'font-weight="bold">오답률</text><rect x="6.0" y="26" width="102.7" height="18" fill="#fff" stroke="#e5e7eb"/><text '
                                   'x="57.3" y="38.5" font-size="9" text-anchor="middle" fill="#111827" font-weight="bold">delta_wing</text><rect '
                                   'x="108.7" y="26" width="102.7" height="18" fill="#fff" stroke="#e5e7eb"/><text x="160.0" y="38.5" font-size="9" '
                                   'text-anchor="middle" fill="#111827">fixed_wing</text><rect x="211.3" y="26" width="102.7" height="18" '
                                   'fill="#fff" stroke="#e5e7eb"/><text x="262.7" y="38.5" font-size="9" text-anchor="middle" '
                                   'fill="#111827">높음</text><rect x="6.0" y="44" width="102.7" height="18" fill="#fff" stroke="#e5e7eb"/><text '
                                   'x="57.3" y="56.5" font-size="9" text-anchor="middle" fill="#111827" font-weight="bold">multicopter</text><rect '
                                   'x="108.7" y="44" width="102.7" height="18" fill="#fff" stroke="#e5e7eb"/><text x="160.0" y="56.5" font-size="9" '
                                   'text-anchor="middle" fill="#111827">bird</text><rect x="211.3" y="44" width="102.7" height="18" fill="#fff" '
                                   'stroke="#e5e7eb"/><text x="262.7" y="56.5" font-size="9" text-anchor="middle" fill="#111827">중간</text><text '
                                   'x="160.0" y="79.0" font-size="9" text-anchor="middle" fill="#475569">자주 헷갈리는 둘을 묶어 내요</text></svg>',
                         'f_adapt': '<svg viewBox="0 0 320 92" xmlns="http://www.w3.org/2000/svg" font-family="sans-serif" font-size="10" '
                                    'fill="#111827"><defs><marker id="tfa" viewBox="0 0 10 10" refX="8" refY="5" markerWidth="6" markerHeight="6" '
                                    'orient="auto"><path d="M0,0 L10,5 L0,10 z" fill="#475569"/></marker></defs><rect x="8.0" y="14.0" width="65.5" '
                                    'height="46.0" rx="7" fill="#dbeafe" stroke="#94a3b8"/><text x="40.8" y="40.5" font-size="10" '
                                    'text-anchor="middle" fill="#111827" font-weight="bold">학습 기록</text><path d="M74.5,37.0 L86.5,37.0" '
                                    'stroke="#475569" stroke-width="1.6" marker-end="url(#tfa)" fill="none"/><rect x="87.5" y="14.0" width="65.5" '
                                    'height="46.0" rx="7" fill="#dbeafe" stroke="#94a3b8"/><text x="120.2" y="40.5" font-size="10" '
                                    'text-anchor="middle" fill="#111827" font-weight="bold">오답률</text><path d="M154.0,37.0 L166.0,37.0" '
                                    'stroke="#475569" stroke-width="1.6" marker-end="url(#tfa)" fill="none"/><rect x="167.0" y="14.0" width="65.5" '
                                    'height="46.0" rx="7" fill="#dbeafe" stroke="#94a3b8"/><text x="199.8" y="40.5" font-size="10" '
                                    'text-anchor="middle" fill="#111827" font-weight="bold">출제 배수</text><path d="M233.5,37.0 L245.5,37.0" '
                                    'stroke="#475569" stroke-width="1.6" marker-end="url(#tfa)" fill="none"/><rect x="246.5" y="14.0" width="65.5" '
                                    'height="46.0" rx="7" fill="#dbeafe" stroke="#94a3b8"/><text x="279.2" y="40.5" font-size="10" '
                                    'text-anchor="middle" fill="#111827" font-weight="bold">다음 문항</text><text x="160.0" y="87.0" font-size="9" '
                                    'text-anchor="middle" fill="#475569">약한 부분을 더 많이 연습해요</text></svg>',
                         'f_edge': '<svg viewBox="0 0 320 102" xmlns="http://www.w3.org/2000/svg" font-family="sans-serif" font-size="10" '
                                   'fill="#111827"><defs><marker id="tfa" viewBox="0 0 10 10" refX="8" refY="5" markerWidth="6" markerHeight="6" '
                                   'orient="auto"><path d="M0,0 L10,5 L0,10 z" fill="#475569"/></marker></defs><rect x="6.0" y="8" width="102.7" '
                                   'height="18" fill="#e5e7eb" stroke="#fff"/><text x="57.3" y="20.5" font-size="9" text-anchor="middle" '
                                   'fill="#111827" font-weight="bold">shown</text><rect x="108.7" y="8" width="102.7" height="18" fill="#e5e7eb" '
                                   'stroke="#fff"/><text x="160.0" y="20.5" font-size="9" text-anchor="middle" fill="#111827" '
                                   'font-weight="bold">wrong</text><rect x="211.3" y="8" width="102.7" height="18" fill="#e5e7eb" '
                                   'stroke="#fff"/><text x="262.7" y="20.5" font-size="9" text-anchor="middle" fill="#111827" '
                                   'font-weight="bold">처리</text><rect x="6.0" y="26" width="102.7" height="18" fill="#fff" stroke="#e5e7eb"/><text '
                                   'x="57.3" y="38.5" font-size="9" text-anchor="middle" fill="#111827" font-weight="bold">40</text><rect x="108.7" '
                                   'y="26" width="102.7" height="18" fill="#fff" stroke="#e5e7eb"/><text x="160.0" y="38.5" font-size="9" '
                                   'text-anchor="middle" fill="#111827">19</text><rect x="211.3" y="26" width="102.7" height="18" fill="#fff" '
                                   'stroke="#e5e7eb"/><text x="262.7" y="38.5" font-size="9" text-anchor="middle" fill="#111827">우선 출제</text><rect '
                                   'x="6.0" y="44" width="102.7" height="18" fill="#fff" stroke="#e5e7eb"/><text x="57.3" y="56.5" font-size="9" '
                                   'text-anchor="middle" fill="#111827" font-weight="bold">3</text><rect x="108.7" y="44" width="102.7" height="18" '
                                   'fill="#fff" stroke="#e5e7eb"/><text x="160.0" y="56.5" font-size="9" text-anchor="middle" '
                                   'fill="#111827">2</text><rect x="211.3" y="44" width="102.7" height="18" fill="#fff" stroke="#e5e7eb"/><text '
                                   'x="262.7" y="56.5" font-size="9" text-anchor="middle" fill="#111827">표본 적음</text><rect x="6.0" y="62" '
                                   'width="102.7" height="18" fill="#fff" stroke="#e5e7eb"/><text x="57.3" y="74.5" font-size="9" '
                                   'text-anchor="middle" fill="#111827" font-weight="bold">0</text><rect x="108.7" y="62" width="102.7" height="18" '
                                   'fill="#fff" stroke="#e5e7eb"/><text x="160.0" y="74.5" font-size="9" text-anchor="middle" '
                                   'fill="#111827">0</text><rect x="211.3" y="62" width="102.7" height="18" fill="#fef3c7" stroke="#e5e7eb"/><text '
                                   'x="262.7" y="74.5" font-size="9" text-anchor="middle" fill="#111827" font-weight="bold">새 문항</text><text '
                                   'x="160.0" y="97.0" font-size="9" text-anchor="middle" fill="#475569">0과 적은 표본을 어떻게 할지 미리 알려 주세요</text></svg>',
                         'f_hitl': '<svg viewBox="0 0 320 92" xmlns="http://www.w3.org/2000/svg" font-family="sans-serif" font-size="10" '
                                   'fill="#111827"><defs><marker id="tfa" viewBox="0 0 10 10" refX="8" refY="5" markerWidth="6" markerHeight="6" '
                                   'orient="auto"><path d="M0,0 L10,5 L0,10 z" fill="#475569"/></marker></defs><rect x="8.0" y="14.0" width="92.0" '
                                   'height="46.0" rx="7" fill="#dbeafe" stroke="#94a3b8"/><text x="54.0" y="40.5" font-size="10" '
                                   'text-anchor="middle" fill="#111827" font-weight="bold">AI 초안</text><path d="M101.0,37.0 L113.0,37.0" '
                                   'stroke="#475569" stroke-width="1.6" marker-end="url(#tfa)" fill="none"/><rect x="114.0" y="14.0" width="92.0" '
                                   'height="46.0" rx="7" fill="#dbeafe" stroke="#94a3b8"/><text x="160.0" y="40.5" font-size="10" '
                                   'text-anchor="middle" fill="#111827" font-weight="bold">사람 검토</text><path d="M207.0,37.0 L219.0,37.0" '
                                   'stroke="#475569" stroke-width="1.6" marker-end="url(#tfa)" fill="none"/><rect x="220.0" y="14.0" width="92.0" '
                                   'height="46.0" rx="7" fill="#dbeafe" stroke="#94a3b8"/><text x="266.0" y="40.5" font-size="9.5" '
                                   'text-anchor="middle" fill="#111827" font-weight="bold">승인 후 공개</text><text x="160.0" y="87.0" font-size="9" '
                                   'text-anchor="middle" fill="#475569">승인 전에는 공개하지 않아요</text></svg>',
                         'f_raci': '<svg viewBox="0 0 320 102" xmlns="http://www.w3.org/2000/svg" font-family="sans-serif" font-size="10" '
                                   'fill="#111827"><defs><marker id="tfa" viewBox="0 0 10 10" refX="8" refY="5" markerWidth="6" markerHeight="6" '
                                   'orient="auto"><path d="M0,0 L10,5 L0,10 z" fill="#475569"/></marker></defs><rect x="6.0" y="8" width="77.0" '
                                   'height="18" fill="#e5e7eb" stroke="#fff"/><text x="44.5" y="20.5" font-size="9" text-anchor="middle" '
                                   'fill="#111827" font-weight="bold">일</text><rect x="83.0" y="8" width="77.0" height="18" fill="#e5e7eb" '
                                   'stroke="#fff"/><text x="121.5" y="20.5" font-size="9" text-anchor="middle" fill="#111827" '
                                   'font-weight="bold">AI</text><rect x="160.0" y="8" width="77.0" height="18" fill="#e5e7eb" stroke="#fff"/><text '
                                   'x="198.5" y="20.5" font-size="9" text-anchor="middle" fill="#111827" font-weight="bold">교관</text><rect x="237.0" '
                                   'y="8" width="77.0" height="18" fill="#e5e7eb" stroke="#fff"/><text x="275.5" y="20.5" font-size="9" '
                                   'text-anchor="middle" fill="#111827" font-weight="bold">전산</text><rect x="6.0" y="26" width="77.0" height="18" '
                                   'fill="#fff" stroke="#e5e7eb"/><text x="44.5" y="38.5" font-size="9" text-anchor="middle" fill="#111827" '
                                   'font-weight="bold">초안</text><rect x="83.0" y="26" width="77.0" height="18" fill="#fff" stroke="#e5e7eb"/><text '
                                   'x="121.5" y="38.5" font-size="9" text-anchor="middle" fill="#111827">R</text><rect x="160.0" y="26" width="77.0" '
                                   'height="18" fill="#fff" stroke="#e5e7eb"/><text x="198.5" y="38.5" font-size="9" text-anchor="middle" '
                                   'fill="#111827">C</text><rect x="237.0" y="26" width="77.0" height="18" fill="#fff" stroke="#e5e7eb"/><text '
                                   'x="275.5" y="38.5" font-size="9" text-anchor="middle" fill="#111827">-</text><rect x="6.0" y="44" width="77.0" '
                                   'height="18" fill="#fff" stroke="#e5e7eb"/><text x="44.5" y="56.5" font-size="9" text-anchor="middle" '
                                   'fill="#111827" font-weight="bold">승인</text><rect x="83.0" y="44" width="77.0" height="18" fill="#fff" '
                                   'stroke="#e5e7eb"/><text x="121.5" y="56.5" font-size="9" text-anchor="middle" fill="#111827">-</text><rect '
                                   'x="160.0" y="44" width="77.0" height="18" fill="#fff" stroke="#e5e7eb"/><text x="198.5" y="56.5" font-size="9" '
                                   'text-anchor="middle" fill="#111827">A</text><rect x="237.0" y="44" width="77.0" height="18" fill="#fff" '
                                   'stroke="#e5e7eb"/><text x="275.5" y="56.5" font-size="9" text-anchor="middle" fill="#111827">I</text><rect '
                                   'x="6.0" y="62" width="77.0" height="18" fill="#fff" stroke="#e5e7eb"/><text x="44.5" y="74.5" font-size="9" '
                                   'text-anchor="middle" fill="#111827" font-weight="bold">반영</text><rect x="83.0" y="62" width="77.0" height="18" '
                                   'fill="#fff" stroke="#e5e7eb"/><text x="121.5" y="74.5" font-size="9" text-anchor="middle" '
                                   'fill="#111827">-</text><rect x="160.0" y="62" width="77.0" height="18" fill="#fff" stroke="#e5e7eb"/><text '
                                   'x="198.5" y="74.5" font-size="9" text-anchor="middle" fill="#111827">I</text><rect x="237.0" y="62" width="77.0" '
                                   'height="18" fill="#fff" stroke="#e5e7eb"/><text x="275.5" y="74.5" font-size="9" text-anchor="middle" '
                                   'fill="#111827">R</text><text x="160.0" y="97.0" font-size="9" text-anchor="middle" fill="#475569">R 수행, A 책임, C '
                                   '의논, I 공유</text></svg>',
                         'f_halluc': '<svg viewBox="0 0 320 92" xmlns="http://www.w3.org/2000/svg" font-family="sans-serif" font-size="10" '
                                     'fill="#111827"><defs><marker id="tfa" viewBox="0 0 10 10" refX="8" refY="5" markerWidth="6" markerHeight="6" '
                                     'orient="auto"><path d="M0,0 L10,5 L0,10 z" fill="#475569"/></marker></defs><rect x="8.0" y="14.0" width="92.0" '
                                     'height="46.0" rx="7" fill="#dbeafe" stroke="#94a3b8"/><text x="54.0" y="40.5" font-size="9.5" '
                                     'text-anchor="middle" fill="#111827" font-weight="bold">자료에 없음</text><path d="M101.0,37.0 L113.0,37.0" '
                                     'stroke="#475569" stroke-width="1.6" marker-end="url(#tfa)" fill="none"/><rect x="114.0" y="14.0" width="92.0" '
                                     'height="46.0" rx="7" fill="#dbeafe" stroke="#94a3b8"/><text x="160.0" y="34.0" font-size="9.5" '
                                     'text-anchor="middle" fill="#111827" font-weight="bold">AI가 지어냄</text><text x="160.0" y="47.0" font-size="8.5" '
                                     'text-anchor="middle" fill="#475569">그럴듯함</text><path d="M207.0,37.0 L219.0,37.0" stroke="#475569" '
                                     'stroke-width="1.6" marker-end="url(#tfa)" fill="none"/><rect x="220.0" y="14.0" width="92.0" height="46.0" '
                                     'rx="7" fill="#dbeafe" stroke="#94a3b8"/><text x="266.0" y="34.0" font-size="10" text-anchor="middle" '
                                     'fill="#111827" font-weight="bold">사실 확인</text><text x="266.0" y="47.0" font-size="8.5" text-anchor="middle" '
                                     'fill="#475569">자료와 대조</text><text x="160.0" y="87.0" font-size="9" text-anchor="middle" fill="#475569">없는 내용은 '
                                     '‘자료 없음’으로 쓰게 해요</text></svg>',
                         'f_prune': '<svg viewBox="0 0 320 92" xmlns="http://www.w3.org/2000/svg" font-family="sans-serif" font-size="10" '
                                    'fill="#111827"><defs><marker id="tfa" viewBox="0 0 10 10" refX="8" refY="5" markerWidth="6" markerHeight="6" '
                                    'orient="auto"><path d="M0,0 L10,5 L0,10 z" fill="#475569"/></marker></defs><rect x="8.0" y="14.0" width="92.0" '
                                    'height="46.0" rx="7" fill="#dbeafe" stroke="#94a3b8"/><text x="54.0" y="34.0" font-size="10" '
                                    'text-anchor="middle" fill="#111827" font-weight="bold">원래 모델</text><text x="54.0" y="47.0" font-size="8.5" '
                                    'text-anchor="middle" fill="#475569">크고 느림</text><path d="M101.0,37.0 L113.0,37.0" stroke="#475569" '
                                    'stroke-width="1.6" marker-end="url(#tfa)" fill="none"/><rect x="114.0" y="14.0" width="92.0" height="46.0" '
                                    'rx="7" fill="#dbeafe" stroke="#94a3b8"/><text x="160.0" y="34.0" font-size="10" text-anchor="middle" '
                                    'fill="#111827" font-weight="bold">가지치기</text><text x="160.0" y="47.0" font-size="8.5" text-anchor="middle" '
                                    'fill="#475569">작고 빠름</text><path d="M207.0,37.0 L219.0,37.0" stroke="#475569" stroke-width="1.6" '
                                    'marker-end="url(#tfa)" fill="none"/><rect x="220.0" y="14.0" width="92.0" height="46.0" rx="7" fill="#dbeafe" '
                                    'stroke="#94a3b8"/><text x="266.0" y="34.0" font-size="10" text-anchor="middle" fill="#111827" '
                                    'font-weight="bold">미세조정</text><text x="266.0" y="47.0" font-size="8.5" text-anchor="middle" fill="#475569">성능 '
                                    '회복</text><text x="160.0" y="87.0" font-size="9" text-anchor="middle" fill="#475569">줄인 뒤 다시 조금 '
                                    '학습해요</text></svg>',
                         'f_safe': '<svg viewBox="0 0 320 102" xmlns="http://www.w3.org/2000/svg" font-family="sans-serif" font-size="10" '
                                   'fill="#111827"><defs><marker id="tfa" viewBox="0 0 10 10" refX="8" refY="5" markerWidth="6" markerHeight="6" '
                                   'orient="auto"><path d="M0,0 L10,5 L0,10 z" fill="#475569"/></marker></defs><rect x="6.0" y="8" width="154.0" '
                                   'height="18" fill="#e5e7eb" stroke="#fff"/><text x="83.0" y="20.5" font-size="9" text-anchor="middle" '
                                   'fill="#111827" font-weight="bold">상황</text><rect x="160.0" y="8" width="154.0" height="18" fill="#e5e7eb" '
                                   'stroke="#fff"/><text x="237.0" y="20.5" font-size="9" text-anchor="middle" fill="#111827" font-weight="bold">나와야 '
                                   '할 것</text><rect x="6.0" y="26" width="154.0" height="18" fill="#fff" stroke="#e5e7eb"/><text x="83.0" y="38.5" '
                                   'font-size="9" text-anchor="middle" fill="#111827" font-weight="bold">심한 흐림</text><rect x="160.0" y="26" '
                                   'width="154.0" height="18" fill="#fff" stroke="#e5e7eb"/><text x="237.0" y="38.5" font-size="9" '
                                   'text-anchor="middle" fill="#111827">판별 불가</text><rect x="6.0" y="44" width="154.0" height="18" fill="#fff" '
                                   'stroke="#e5e7eb"/><text x="83.0" y="56.5" font-size="9" text-anchor="middle" fill="#111827" '
                                   'font-weight="bold">손상 파일</text><rect x="160.0" y="44" width="154.0" height="18" fill="#fff" '
                                   'stroke="#e5e7eb"/><text x="237.0" y="56.5" font-size="9" text-anchor="middle" fill="#111827">판별 불가</text><rect '
                                   'x="6.0" y="62" width="154.0" height="18" fill="#fff" stroke="#e5e7eb"/><text x="83.0" y="74.5" font-size="9" '
                                   'text-anchor="middle" fill="#111827" font-weight="bold">선명한 영상</text><rect x="160.0" y="62" width="154.0" '
                                   'height="18" fill="#fff" stroke="#e5e7eb"/><text x="237.0" y="74.5" font-size="9" text-anchor="middle" '
                                   'fill="#111827">참고 라벨</text><text x="160.0" y="97.0" font-size="9" text-anchor="middle" fill="#475569">모르면 답을 '
                                   '지어내지 않아요</text></svg>',
                         'f_secret': '<svg viewBox="0 0 320 102" xmlns="http://www.w3.org/2000/svg" font-family="sans-serif" font-size="10" '
                                     'fill="#111827"><defs><marker id="tfa" viewBox="0 0 10 10" refX="8" refY="5" markerWidth="6" markerHeight="6" '
                                     'orient="auto"><path d="M0,0 L10,5 L0,10 z" fill="#475569"/></marker></defs><rect x="6.0" y="8" width="154.0" '
                                     'height="18" fill="#e5e7eb" stroke="#fff"/><text x="83.0" y="20.5" font-size="9" text-anchor="middle" '
                                     'fill="#111827" font-weight="bold">넣어도 돼요</text><rect x="160.0" y="8" width="154.0" height="18" fill="#e5e7eb" '
                                     'stroke="#fff"/><text x="237.0" y="20.5" font-size="9" text-anchor="middle" fill="#111827" '
                                     'font-weight="bold">넣으면 안 돼요</text><rect x="6.0" y="26" width="154.0" height="18" fill="#fff" '
                                     'stroke="#e5e7eb"/><text x="83.0" y="38.5" font-size="9" text-anchor="middle" fill="#111827" '
                                     'font-weight="bold">열 이름</text><rect x="160.0" y="26" width="154.0" height="18" fill="#fff" '
                                     'stroke="#e5e7eb"/><text x="237.0" y="38.5" font-size="9" text-anchor="middle" fill="#111827">서버 주소</text><rect '
                                     'x="6.0" y="44" width="154.0" height="18" fill="#fff" stroke="#e5e7eb"/><text x="83.0" y="56.5" font-size="9" '
                                     'text-anchor="middle" fill="#111827" font-weight="bold">가상 샘플</text><rect x="160.0" y="44" width="154.0" '
                                     'height="18" fill="#fff" stroke="#e5e7eb"/><text x="237.0" y="56.5" font-size="9" text-anchor="middle" '
                                     'fill="#111827">계정</text><rect x="6.0" y="62" width="154.0" height="18" fill="#fff" stroke="#e5e7eb"/><text '
                                     'x="83.0" y="74.5" font-size="9" text-anchor="middle" fill="#111827" font-weight="bold">오류 메시지</text><rect '
                                     'x="160.0" y="62" width="154.0" height="18" fill="#fff" stroke="#e5e7eb"/><text x="237.0" y="74.5" '
                                     'font-size="9" text-anchor="middle" fill="#111827">실명 기록</text><text x="160.0" y="97.0" font-size="9" '
                                     'text-anchor="middle" fill="#475569">공개 AI에 넣기 전에 나눠 봐요</text></svg>',
                         'f_aar': '<svg viewBox="0 0 320 156" xmlns="http://www.w3.org/2000/svg" font-family="sans-serif" font-size="10" '
                                  'fill="#111827"><defs><marker id="tfa" viewBox="0 0 10 10" refX="8" refY="5" markerWidth="6" markerHeight="6" '
                                  'orient="auto"><path d="M0,0 L10,5 L0,10 z" fill="#475569"/></marker></defs><ellipse cx="160.0" cy="70" rx="112" '
                                  'ry="48" fill="none" stroke="#e5e7eb" stroke-width="6"/><path d="M207.7,26.6 L261.3,49.6" stroke="#2563eb" '
                                  'stroke-width="2" marker-end="url(#tfa)" fill="none"/><path d="M261.3,90.4 L207.7,113.4" stroke="#2563eb" '
                                  'stroke-width="2" marker-end="url(#tfa)" fill="none"/><path d="M112.3,113.4 L58.7,90.4" stroke="#2563eb" '
                                  'stroke-width="2" marker-end="url(#tfa)" fill="none"/><path d="M58.7,49.6 L112.3,26.6" stroke="#2563eb" '
                                  'stroke-width="2" marker-end="url(#tfa)" fill="none"/><rect x="116.0" y="10.0" width="88.0" height="24.0" rx="7" '
                                  'fill="#dbeafe" stroke="#94a3b8"/><text x="160.0" y="25.5" font-size="9" text-anchor="middle" fill="#111827" '
                                  'font-weight="bold">계획</text><rect x="228.0" y="58.0" width="88.0" height="24.0" rx="7" fill="#dbeafe" '
                                  'stroke="#94a3b8"/><text x="272.0" y="73.5" font-size="9" text-anchor="middle" fill="#111827" '
                                  'font-weight="bold">실제</text><rect x="116.0" y="106.0" width="88.0" height="24.0" rx="7" fill="#dbeafe" '
                                  'stroke="#94a3b8"/><text x="160.0" y="121.5" font-size="9" text-anchor="middle" fill="#111827" '
                                  'font-weight="bold">원인</text><rect x="4.0" y="58.0" width="88.0" height="24.0" rx="7" fill="#dbeafe" '
                                  'stroke="#94a3b8"/><text x="48.0" y="73.5" font-size="9" text-anchor="middle" fill="#111827" font-weight="bold">다음 '
                                  '조치</text><text x="160.0" y="74.0" font-size="10.5" text-anchor="middle" fill="#475569" '
                                  'font-weight="bold">AAR</text><text x="160.0" y="151.0" font-size="9" text-anchor="middle" fill="#475569">탓하지 않고 '
                                  '바꿀 것을 찾아요</text></svg>',
                         'f_reg': '<svg viewBox="0 0 320 84" xmlns="http://www.w3.org/2000/svg" font-family="sans-serif" font-size="10" '
                                  'fill="#111827"><defs><marker id="tfa" viewBox="0 0 10 10" refX="8" refY="5" markerWidth="6" markerHeight="6" '
                                  'orient="auto"><path d="M0,0 L10,5 L0,10 z" fill="#475569"/></marker></defs><rect x="6.0" y="8" width="102.7" '
                                  'height="18" fill="#e5e7eb" stroke="#fff"/><text x="57.3" y="20.5" font-size="9" text-anchor="middle" '
                                  'fill="#111827" font-weight="bold">조건</text><rect x="108.7" y="8" width="102.7" height="18" fill="#e5e7eb" '
                                  'stroke="#fff"/><text x="160.0" y="20.5" font-size="9" text-anchor="middle" fill="#111827" '
                                  'font-weight="bold">전</text><rect x="211.3" y="8" width="102.7" height="18" fill="#e5e7eb" stroke="#fff"/><text '
                                  'x="262.7" y="20.5" font-size="9" text-anchor="middle" fill="#111827" font-weight="bold">후</text><rect x="6.0" '
                                  'y="26" width="102.7" height="18" fill="#fff" stroke="#e5e7eb"/><text x="57.3" y="38.5" font-size="9" '
                                  'text-anchor="middle" fill="#111827" font-weight="bold">far</text><rect x="108.7" y="26" width="102.7" height="18" '
                                  'fill="#fff" stroke="#e5e7eb"/><text x="160.0" y="38.5" font-size="9" text-anchor="middle" '
                                  'fill="#111827">58</text><rect x="211.3" y="26" width="102.7" height="18" fill="#fff" stroke="#e5e7eb"/><text '
                                  'x="262.7" y="38.5" font-size="9" text-anchor="middle" fill="#111827">72</text><rect x="6.0" y="44" width="102.7" '
                                  'height="18" fill="#fff" stroke="#e5e7eb"/><text x="57.3" y="56.5" font-size="9" text-anchor="middle" '
                                  'fill="#111827" font-weight="bold">near</text><rect x="108.7" y="44" width="102.7" height="18" fill="#fff" '
                                  'stroke="#e5e7eb"/><text x="160.0" y="56.5" font-size="9" text-anchor="middle" fill="#111827">94</text><rect '
                                  'x="211.3" y="44" width="102.7" height="18" fill="#fff" stroke="#e5e7eb"/><text x="262.7" y="56.5" font-size="9" '
                                  'text-anchor="middle" fill="#111827">92</text><text x="160.0" y="79.0" font-size="9" text-anchor="middle" '
                                  'fill="#475569">평균이 올라도 떨어진 조건을 봐요</text></svg>',
                         'f_review': '<svg viewBox="0 0 320 156" xmlns="http://www.w3.org/2000/svg" font-family="sans-serif" font-size="10" '
                                     'fill="#111827"><defs><marker id="tfa" viewBox="0 0 10 10" refX="8" refY="5" markerWidth="6" markerHeight="6" '
                                     'orient="auto"><path d="M0,0 L10,5 L0,10 z" fill="#475569"/></marker></defs><ellipse cx="160.0" cy="70" '
                                     'rx="112" ry="48" fill="none" stroke="#e5e7eb" stroke-width="6"/><path d="M207.7,26.6 L261.3,49.6" '
                                     'stroke="#2563eb" stroke-width="2" marker-end="url(#tfa)" fill="none"/><path d="M261.3,90.4 L207.7,113.4" '
                                     'stroke="#2563eb" stroke-width="2" marker-end="url(#tfa)" fill="none"/><path d="M112.3,113.4 L58.7,90.4" '
                                     'stroke="#2563eb" stroke-width="2" marker-end="url(#tfa)" fill="none"/><path d="M58.7,49.6 L112.3,26.6" '
                                     'stroke="#2563eb" stroke-width="2" marker-end="url(#tfa)" fill="none"/><rect x="116.0" y="10.0" width="88.0" '
                                     'height="24.0" rx="7" fill="#dbeafe" stroke="#94a3b8"/><text x="160.0" y="25.5" font-size="9" '
                                     'text-anchor="middle" fill="#111827" font-weight="bold">설명</text><rect x="228.0" y="58.0" width="88.0" '
                                     'height="24.0" rx="7" fill="#dbeafe" stroke="#94a3b8"/><text x="272.0" y="73.5" font-size="9" '
                                     'text-anchor="middle" fill="#111827" font-weight="bold">테스트</text><rect x="116.0" y="106.0" width="88.0" '
                                     'height="24.0" rx="7" fill="#dbeafe" stroke="#94a3b8"/><text x="160.0" y="121.5" font-size="9" '
                                     'text-anchor="middle" fill="#111827" font-weight="bold">수정</text><rect x="4.0" y="58.0" width="88.0" '
                                     'height="24.0" rx="7" fill="#dbeafe" stroke="#94a3b8"/><text x="48.0" y="73.5" font-size="9" '
                                     'text-anchor="middle" fill="#111827" font-weight="bold">비교</text><text x="160.0" y="74.0" font-size="10.5" '
                                     'text-anchor="middle" fill="#475569" font-weight="bold">리뷰</text><text x="160.0" y="151.0" font-size="9" '
                                     'text-anchor="middle" fill="#475569">고치기 전에 먼저 설명하게 해요</text></svg>',
                         'f_scope': '<svg viewBox="0 0 320 92" xmlns="http://www.w3.org/2000/svg" font-family="sans-serif" font-size="10" '
                                    'fill="#111827"><defs><marker id="tfa" viewBox="0 0 10 10" refX="8" refY="5" markerWidth="6" markerHeight="6" '
                                    'orient="auto"><path d="M0,0 L10,5 L0,10 z" fill="#475569"/></marker></defs><rect x="8.0" y="14.0" width="92.0" '
                                    'height="46.0" rx="7" fill="#dbeafe" stroke="#94a3b8"/><text x="54.0" y="34.0" font-size="10" '
                                    'text-anchor="middle" fill="#111827" font-weight="bold">고칠 곳</text><text x="54.0" y="47.0" font-size="8.5" '
                                    'text-anchor="middle" fill="#475569">이 함수만</text><path d="M101.0,37.0 L113.0,37.0" stroke="#475569" '
                                    'stroke-width="1.6" marker-end="url(#tfa)" fill="none"/><rect x="114.0" y="14.0" width="92.0" height="46.0" '
                                    'rx="7" fill="#dbeafe" stroke="#94a3b8"/><text x="160.0" y="34.0" font-size="10" text-anchor="middle" '
                                    'fill="#111827" font-weight="bold">그대로</text><text x="160.0" y="47.0" font-size="8.5" text-anchor="middle" '
                                    'fill="#475569">원본 기록</text><path d="M207.0,37.0 L219.0,37.0" stroke="#475569" stroke-width="1.6" '
                                    'marker-end="url(#tfa)" fill="none"/><rect x="220.0" y="14.0" width="92.0" height="46.0" rx="7" fill="#dbeafe" '
                                    'stroke="#94a3b8"/><text x="266.0" y="34.0" font-size="10" text-anchor="middle" fill="#111827" '
                                    'font-weight="bold">그대로</text><text x="266.0" y="47.0" font-size="8.5" text-anchor="middle" fill="#475569">다른 '
                                    '코드</text><text x="160.0" y="87.0" font-size="9" text-anchor="middle" fill="#475569">고칠 범위를 말해 주세요</text></svg>',
                         'f_tool': '<svg viewBox="0 0 320 92" xmlns="http://www.w3.org/2000/svg" font-family="sans-serif" font-size="10" '
                                   'fill="#111827"><defs><marker id="tfa" viewBox="0 0 10 10" refX="8" refY="5" markerWidth="6" markerHeight="6" '
                                   'orient="auto"><path d="M0,0 L10,5 L0,10 z" fill="#475569"/></marker></defs><rect x="8.0" y="14.0" width="65.5" '
                                   'height="46.0" rx="7" fill="#dbeafe" stroke="#94a3b8"/><text x="40.8" y="40.5" font-size="10" '
                                   'text-anchor="middle" fill="#111827" font-weight="bold">설정값</text><path d="M74.5,37.0 L86.5,37.0" '
                                   'stroke="#475569" stroke-width="1.6" marker-end="url(#tfa)" fill="none"/><rect x="87.5" y="14.0" width="65.5" '
                                   'height="46.0" rx="7" fill="#dbeafe" stroke="#94a3b8"/><text x="120.2" y="40.5" font-size="10" '
                                   'text-anchor="middle" fill="#111827" font-weight="bold">계산 함수</text><path d="M154.0,37.0 L166.0,37.0" '
                                   'stroke="#475569" stroke-width="1.6" marker-end="url(#tfa)" fill="none"/><rect x="167.0" y="14.0" width="65.5" '
                                   'height="46.0" rx="7" fill="#dbeafe" stroke="#94a3b8"/><text x="199.8" y="40.5" font-size="10" '
                                   'text-anchor="middle" fill="#111827" font-weight="bold">표·그림</text><path d="M233.5,37.0 L245.5,37.0" '
                                   'stroke="#475569" stroke-width="1.6" marker-end="url(#tfa)" fill="none"/><rect x="246.5" y="14.0" width="65.5" '
                                   'height="46.0" rx="7" fill="#dbeafe" stroke="#94a3b8"/><text x="279.2" y="40.5" font-size="10" '
                                   'text-anchor="middle" fill="#111827" font-weight="bold">브리핑</text><text x="160.0" y="87.0" font-size="9" '
                                   'text-anchor="middle" fill="#475569">한 번에 다시 돌릴 수 있게 묶어요</text></svg>',
                         'f_hand': '<svg viewBox="0 0 320 120" xmlns="http://www.w3.org/2000/svg" font-family="sans-serif" font-size="10" '
                                   'fill="#111827"><defs><marker id="tfa" viewBox="0 0 10 10" refX="8" refY="5" markerWidth="6" markerHeight="6" '
                                   'orient="auto"><path d="M0,0 L10,5 L0,10 z" fill="#475569"/></marker></defs><rect x="6.0" y="8" width="102.7" '
                                   'height="18" fill="#e5e7eb" stroke="#fff"/><text x="57.3" y="20.5" font-size="9" text-anchor="middle" '
                                   'fill="#111827" font-weight="bold">일</text><rect x="108.7" y="8" width="102.7" height="18" fill="#e5e7eb" '
                                   'stroke="#fff"/><text x="160.0" y="20.5" font-size="9" text-anchor="middle" fill="#111827" '
                                   'font-weight="bold">담당</text><rect x="211.3" y="8" width="102.7" height="18" fill="#e5e7eb" stroke="#fff"/><text '
                                   'x="262.7" y="20.5" font-size="9" text-anchor="middle" fill="#111827" font-weight="bold">주기</text><rect x="6.0" '
                                   'y="26" width="102.7" height="18" fill="#fff" stroke="#e5e7eb"/><text x="57.3" y="38.5" font-size="9" '
                                   'text-anchor="middle" fill="#111827" font-weight="bold">새 형상 등록</text><rect x="108.7" y="26" width="102.7" '
                                   'height="18" fill="#fff" stroke="#e5e7eb"/><text x="160.0" y="38.5" font-size="9" text-anchor="middle" '
                                   'fill="#111827">데이터</text><rect x="211.3" y="26" width="102.7" height="18" fill="#fff" stroke="#e5e7eb"/><text '
                                   'x="262.7" y="38.5" font-size="9" text-anchor="middle" fill="#111827">확인 시</text><rect x="6.0" y="44" '
                                   'width="102.7" height="18" fill="#fff" stroke="#e5e7eb"/><text x="57.3" y="56.5" font-size="9" '
                                   'text-anchor="middle" fill="#111827" font-weight="bold">콘텐츠 검수</text><rect x="108.7" y="44" width="102.7" '
                                   'height="18" fill="#fff" stroke="#e5e7eb"/><text x="160.0" y="56.5" font-size="9" text-anchor="middle" '
                                   'fill="#111827">교관</text><rect x="211.3" y="44" width="102.7" height="18" fill="#fff" stroke="#e5e7eb"/><text '
                                   'x="262.7" y="56.5" font-size="9" text-anchor="middle" fill="#111827">월 1회</text><rect x="6.0" y="62" '
                                   'width="102.7" height="18" fill="#fff" stroke="#e5e7eb"/><text x="57.3" y="74.5" font-size="9" '
                                   'text-anchor="middle" fill="#111827" font-weight="bold">배포 승인</text><rect x="108.7" y="62" width="102.7" '
                                   'height="18" fill="#fff" stroke="#e5e7eb"/><text x="160.0" y="74.5" font-size="9" text-anchor="middle" '
                                   'fill="#111827">교육</text><rect x="211.3" y="62" width="102.7" height="18" fill="#fff" stroke="#e5e7eb"/><text '
                                   'x="262.7" y="74.5" font-size="9" text-anchor="middle" fill="#111827">배포 때</text><rect x="6.0" y="80" '
                                   'width="102.7" height="18" fill="#fff" stroke="#e5e7eb"/><text x="57.3" y="92.5" font-size="9" '
                                   'text-anchor="middle" fill="#111827" font-weight="bold">장애 대응</text><rect x="108.7" y="80" width="102.7" '
                                   'height="18" fill="#fff" stroke="#e5e7eb"/><text x="160.0" y="92.5" font-size="9" text-anchor="middle" '
                                   'fill="#111827">전산</text><rect x="211.3" y="80" width="102.7" height="18" fill="#fff" stroke="#e5e7eb"/><text '
                                   'x="262.7" y="92.5" font-size="9" text-anchor="middle" fill="#111827">발생 시</text><text x="160.0" y="115.0" '
                                   'font-size="9" text-anchor="middle" fill="#475569">누가 무엇을 언제 맡는지 적어요</text></svg>',
                         'f_lint': '<svg viewBox="0 0 320 102" xmlns="http://www.w3.org/2000/svg" font-family="sans-serif" font-size="10" '
                                   'fill="#111827"><defs><marker id="tfa" viewBox="0 0 10 10" refX="8" refY="5" markerWidth="6" markerHeight="6" '
                                   'orient="auto"><path d="M0,0 L10,5 L0,10 z" fill="#475569"/></marker></defs><rect x="6.0" y="8" width="154.0" '
                                   'height="18" fill="#e5e7eb" stroke="#fff"/><text x="83.0" y="20.5" font-size="9" text-anchor="middle" '
                                   'fill="#111827" font-weight="bold">항목</text><rect x="160.0" y="8" width="154.0" height="18" fill="#e5e7eb" '
                                   'stroke="#fff"/><text x="237.0" y="20.5" font-size="9" text-anchor="middle" fill="#111827" '
                                   'font-weight="bold">있음?</text><rect x="6.0" y="26" width="154.0" height="18" fill="#fff" stroke="#e5e7eb"/><text '
                                   'x="83.0" y="38.5" font-size="9" text-anchor="middle" fill="#111827" font-weight="bold">표본 수</text><rect '
                                   'x="160.0" y="26" width="154.0" height="18" fill="#fff" stroke="#e5e7eb"/><text x="237.0" y="38.5" font-size="9" '
                                   'text-anchor="middle" fill="#111827">O</text><rect x="6.0" y="44" width="154.0" height="18" fill="#fff" '
                                   'stroke="#e5e7eb"/><text x="83.0" y="56.5" font-size="9" text-anchor="middle" fill="#111827" '
                                   'font-weight="bold">한계</text><rect x="160.0" y="44" width="154.0" height="18" fill="#fff" stroke="#e5e7eb"/><text '
                                   'x="237.0" y="56.5" font-size="9" text-anchor="middle" fill="#111827">O</text><rect x="6.0" y="62" width="154.0" '
                                   'height="18" fill="#fff" stroke="#e5e7eb"/><text x="83.0" y="74.5" font-size="9" text-anchor="middle" '
                                   'fill="#111827" font-weight="bold">AI 산출물 표기</text><rect x="160.0" y="62" width="154.0" height="18" '
                                   'fill="#fef3c7" stroke="#e5e7eb"/><text x="237.0" y="74.5" font-size="9" text-anchor="middle" fill="#111827" '
                                   'font-weight="bold">X</text><text x="160.0" y="97.0" font-size="9" text-anchor="middle" fill="#475569">빠진 항목을 코드로 '
                                   '찾아요</text></svg>'},
             'workspace': '🛩️ 실루엣 식별 교육개발반',
             'call_default': '전우님',
             'stay_jobs': {}},
 'competencies': [{'id': 'D1',
                   'group': '디지털·AI 활용 역량',
                   'name': '프롬프트 구조화',
                   'criteria': '업무 요청을 맥락·입력·할 일·출력 형식으로 나눠 AI에게 분명하게 요청할 수 있다.',
                   'ncs': '디지털능력 (인공지능(AI)활용능력), 의사소통능력 (문서소통능력)'},
                  {'id': 'D2',
                   'group': '디지털·AI 활용 역량',
                   'name': '데이터 설명',
                   'criteria': '파일·열·값의 모양을 AI에게 정확히 알려 주어 추측 없이 동작하는 코드를 받을 수 있다.',
                   'ncs': '디지털능력 (디지털활용능력)'},
                  {'id': 'D3',
                   'group': '디지털·AI 활용 역량',
                   'name': '수정 요청(디버깅)',
                   'criteria': '오류 메시지와 기대 동작을 담아 수정을 요청하고, 고칠 범위를 정할 수 있다.',
                   'ncs': '문제해결능력 (문제분석능력), 디지털능력 (디지털활용능력)'},
                  {'id': 'D4',
                   'group': '디지털·AI 활용 역량',
                   'name': '나누어 만들기',
                   'criteria': '큰 일을 작은 단계로 쪼개 요청하고, 조건을 바꿔 돌릴 수 있는 함수로 만들어 달라고 할 수 있다.',
                   'ncs': '문제해결능력 (대안발굴능력), 디지털능력 (디지털활용능력)'},
                  {'id': 'D5',
                   'group': '디지털·AI 활용 역량',
                   'name': '결과 검증',
                   'criteria': 'AI의 결과를 근거 수치·경계 사례·작은 테스트로 확인하고, 믿어도 되는지 판단할 수 있다.',
                   'ncs': '수리능력 (통계활용능력), 문제해결능력 (문제분석능력)'},
                  {'id': 'D6',
                   'group': '디지털·AI 활용 역량',
                   'name': '안전한 AI 활용',
                   'criteria': 'AI에 넣으면 안 되는 정보를 구분하고, 사람이 확인할 일을 정할 수 있다.',
                   'ncs': '디지털능력 (디지털책임의식)'},
                  {'id': 'J1',
                   'group': '직무전문 역량',
                   'name': 'AI로 문서 만들기',
                   'criteria': '독자·양식·근거를 지정해 AI가 업무 문서 초안을 만들게 하고, 사실을 확인해 고칠 수 있다.',
                   'ncs': '의사소통능력 (문서소통능력)'},
                  {'id': 'C1',
                   'group': '협업·소통 역량',
                   'name': '결과로 조율하기',
                   'criteria': 'AI로 만든 비교 결과를 근거로 팀원의 다른 의견을 조율해 기준을 합의할 수 있다.',
                   'ncs': '대인관계능력 (협업능력·갈등관리능력)'}],
 'missions': [{'id': 'm1',
               'no': 1,
               'title': 'WEFT 식별 단서와 기본 도형 실루엣 만들기',
               'short': '프롬프트 네 칸',
               'kind': '개인',
               'competencies': ['D1'],
               'outline': 'AI에게 일을 시킬 때 꼭 넣어야 할 네 칸(맥락·입력·할 일·출력 형식)을 익히고, 형상 계열 6종을 기본 도형으로 그린 뒤 거리별로 몇 픽셀로 보이는지 표로 받아요.',
               'brief': {'subtitle': '거리별 식별 가능 수준표 만들기: 프롬프트의 네 칸',
                         'time': '약 25분',
                         'goal': '프롬프트의 네 칸을 채워 AI에게 요청하고, 형상 계열 실루엣과 거리별 식별 가능 수준표를 코드와 문서로 받아요.',
                         'why': '삼각형 날개면 모두 같은 비행체일까요? 날개·엔진·동체·꼬리로 나눠 보면 단서가 생기지만, 멀어지면 그 단서가 사라져요. 이런 계산을 AI에게 맡길 때, 무엇을 알려 주느냐에 따라 결과가 크게 달라져요.',
                         'case': {'title': '해외 사례: 항공기를 네 부분으로 나눠 보는 미 육군 교범',
                                  'source': '미 육군 교범 TC 3-01.80, FAS 소개 자료 (2016~2017)',
                                  'points': ['미 육군의 시각 항공기 식별 교범 TC 3-01.80은 항공기를 날개(Wings)·엔진(Engine)·동체(Fuselage)·꼬리(Tail)로 나눠 설명해요. 앞 글자를 따서 '
                                             'WEFT라고 불러요.',
                                             '2016년 2월 문서를 2017년 5월에 고쳐 냈고, 부록에는 무인 항공기 목록도 들어 있어요.',
                                             '무엇을 볼지 네 칸으로 나누면 설명이 쉬워져요. 프롬프트도 마찬가지로 칸을 나누면 AI가 추측할 일이 줄어요.']},
                         'team': '고교관(식별 교관)이 표를 부탁해요. 전우님이 AI에게 프롬프트를 보내 실루엣과 표, 문서를 받아요.',
                         'terms': [{'t': '바이브 코딩',
                                    'en': 'vibe coding',
                                    'fig': 'f_loop',
                                    'd': '만들고 싶은 것을 말로 설명해 AI가 코드를 쓰게 하고, 실행 결과를 보며 고쳐 가는 방식이에요.',
                                    'easy': '코드를 직접 쓰는 대신, AI에게 일을 설명하고 결과를 확인하는 역할을 맡는 거예요.',
                                    'use': '요청 → 생성 → 실행 → 확인 → 수정 요청을 반복해요.'},
                                   {'t': '프롬프트',
                                    'en': 'prompt',
                                    'fig': 'f_four',
                                    'd': 'AI에게 보내는 요청 글이에요.',
                                    'easy': '새로 온 동료에게 주는 업무 지시서라고 생각하면 돼요.',
                                    'use': '맥락·입력·할 일·출력 형식의 네 칸을 채우면 결과가 안정돼요.'},
                                   {'t': 'WEFT',
                                    'en': 'Wings·Engine·Fuselage·Tail',
                                    'fig': 'f_weft',
                                    'd': '항공기를 날개, 엔진, 동체, 꼬리 네 부분으로 나눠 식별하는 방법이에요.',
                                    'easy': '사람을 볼 때 머리, 옷, 키, 걸음걸이를 나눠 보는 것과 비슷해요.',
                                    'use': '데이터의 열 이름도 이 네 부분에 맞춰 두었어요.'},
                                   {'t': '존슨 기준',
                                    'en': 'Johnson criteria',
                                    'fig': 'f_johnson',
                                    'd': '물체가 몇 픽셀로 보이면 탐지, 인식, 식별이 가능한지 어림하는 기준이에요.',
                                    'easy': '‘뭔가 있다’ → ‘어떤 부류다’ → ‘구체적으로 무엇이다’ 순서로 더 많은 픽셀이 필요해요.',
                                    'use': '이번 실습은 2·8·13픽셀 기준을 써요.'}],
                         'steps': ['네 칸 익히기: 프롬프트의 네 칸', 'AI에게 시키기: 프롬프트를 만들어 AI에게 보내기', '내 말로 쓰기: 같은 일을 내 말로 요청하기'],
                         'glossary': [],
                         'outputs': ['M1 형상 계열 실루엣과 식별 가능 수준표']},
               'start_messages': [['dm_gyo', 'gyo', '{call}, 형상 계열 6종이 거리별로 몇 픽셀로 보이는지 표가 필요해요. 직접 계산하지 말고 AI에게 시켜 보세요. 파일은 m1_families.csv예요.']],
               'finish_messages': [['dm_gyo', 'gyo', '{call}, 표 잘 받았어요! 프롬프트도 다음에 그대로 쓸 수 있겠어요 👍']],
               'steps': [{'short': '네 칸 익히기',
                          'title': '프롬프트의 네 칸',
                          'lead': 'AI는 전우님 업무도, 파일도 몰라요. **네 칸**을 채워 알려 주면 돼요.',
                          'blocks': [{'id': 's1_info',
                                      'type': 'info',
                                      'title': '좋은 프롬프트의 네 칸',
                                      'icon': '🧩',
                                      'text': ['같은 일을 시켜도 무엇을 알려 주느냐에 따라 결과가 달라져요. 처음에는 아래 네 칸만 기억하세요.'],
                                      'table': {'head': ['칸', '무엇을 적나', '예'],
                                                'rows': [['맥락', '왜 필요한지, 어떤 일인지', '실루엣 학습앱에 넣을 표를 만들고 있어'],
                                                         ['입력', '파일 이름, 열 이름, 값의 뜻', 'm1_families.csv, 열은 family·span_m·length_m'],
                                                         ['할 일', '계산 규칙을 숫자로', '픽셀 수 = 대표 크기 ÷ (거리 × 0.001)'],
                                                         ['출력 형식', '받고 싶은 모양', '형상 계열은 행, 거리는 열인 표']]},
                                      'figure': 'f_four'},
                                     {'id': 's1_q',
                                      'type': 'quiz',
                                      'title': '어떤 프롬프트가 가장 좋을까요?',
                                      'text': '같은 일을 시키는 세 프롬프트예요. 다시 묻는 일이 가장 적을 것은 무엇일까요?',
                                      'options': ['드론 실루엣 코드 좀 짜 줘.',
                                                  '비행체가 잘 보이는지 알려 줘. 잘 부탁해.',
                                                  '실루엣 학습앱에 넣을 표를 만들고 있어. m1_families.csv의 크기로 거리마다 몇 픽셀로 보이는지 계산해 표로 보여 줘.'],
                                      'answer': 2,
                                      'explain': '맥락, 입력, 할 일, 출력 형식이 모두 들어 있어요. 정중한 말보다 구체적인 정보가 결과를 바꿔요.',
                                      'hints': ['네 칸 가운데 몇 칸이 채워졌는지 세어 보세요.', '파일 이름과 받고 싶은 모양까지 적힌 것을 찾아보세요.']}]},
                         {'short': 'AI에게 시키기',
                          'title': '프롬프트를 만들어 AI에게 보내기',
                          'lead': '칸마다 문장을 고르면 프롬프트가 만들어져요. 일부러 **아쉬운 문장**도 보내 보세요. 무엇이 달라지는지 보는 것이 이번 활동의 핵심이에요.',
                          'blocks': [{'id': 's2_vibe',
                                      'type': 'vibe',
                                      'title': '실루엣과 거리별 수준표 만들기',
                                      'ask': ['gyo', '형상 계열 파일을 드릴게요. 기본 도형 실루엣 6종과, 거리 100m~2,000m에서 각각 몇 픽셀로 보이는지 표로 받고 싶어요.'],
                                      'file': 'm1_families.csv',
                                      'slots': [{'id': 'ctx',
                                                 'label': '맥락',
                                                 'why': '왜 필요한지 알려요.',
                                                 'options': [{'text': '실루엣 식별 학습앱을 만들고 있어. 형상 계열마다 거리별로 화면에서 몇 픽셀로 보이는지 확인하려고 해.', 'good': True},
                                                             {'text': '코드 좀 짜 줘.', 'good': False},
                                                             {'text': '', 'good': False}],
                                                 'miss': {'say': '데이터를 읽어 요약 통계를 만들었어요.',
                                                          'code': 'import pandas as pd\n'
                                                                  "df = pd.read_csv('{VDIR}/m1_families.csv')\n"
                                                                  'print(df.describe())',
                                                          'problem': '평균, 최댓값 같은 요약만 나왔어요. 왜 필요한지 몰라서 AI가 가장 흔한 일을 했어요.',
                                                          'hint': '어떤 업무이고 무엇을 확인하려는지 한 문장으로 알려 주세요.'}},
                                                {'id': 'input',
                                                 'label': '입력',
                                                 'why': '파일과 열을 알려요.',
                                                 'options': [{'text': '데이터는 m1_families.csv이고, 열은 family(형상 계열), wing, engine, body, tail(WEFT 단서), '
                                                                      'span_m(폭, 미터), length_m(길이, 미터)이야.',
                                                              'good': True},
                                                             {'text': '드론 데이터 파일이 있어.', 'good': False},
                                                             {'text': '', 'good': False}],
                                                 'miss': {'say': '드론 파일을 읽는 코드를 만들었어요.',
                                                          'expect': 'error',
                                                          'code': 'import pandas as pd\n'
                                                                  "df = pd.read_csv('{VDIR}/drones.csv')\n"
                                                                  "print(df['size'] / df['distance'])",
                                                          'problem': 'AI가 파일 이름을 몰라 그럴듯한 이름을 지어냈고, 그래서 파일을 찾지 못했어요.',
                                                          'hint': '파일 이름과 열 이름을 그대로 적어 주세요. AI는 전우님 컴퓨터의 파일을 볼 수 없어요.'}},
                                                {'id': 'task',
                                                 'label': '할 일',
                                                 'why': '계산 규칙을 숫자로 적어요.',
                                                 'options': [{'text': '대표 크기 = √(span_m × length_m), 픽셀 수 = 대표 크기 ÷ (거리 × 0.001)로 거리 100, 300, 500, '
                                                                      '1000, 2000m마다 계산하고, 2픽셀 이상은 탐지, 8픽셀 이상은 인식, 13픽셀 이상은 식별로 표시해 줘.',
                                                              'good': True},
                                                             {'text': '거리별로 잘 보이는지 계산해 줘.', 'good': False},
                                                             {'text': '', 'good': False}],
                                                 'miss': {'say': '폭을 거리로 나눠 얼마나 잘 보이는지 계산했어요.',
                                                          'code': 'import pandas as pd\n'
                                                                  "df = pd.read_csv('{VDIR}/m1_families.csv')\n"
                                                                  'for d in [100, 300, 500, 1000, 2000]:\n'
                                                                  "    df[f'{d}m'] = (df['span_m'] / d).round(4)\n"
                                                                  "display(df[['family', '100m', '300m', '500m', '1000m', '2000m']])",
                                                          'problem': '오류는 없지만 폭을 거리로 나눈 작은 숫자만 나왔어요. 몇 픽셀인지, 탐지·인식·식별이 어디까지인지 알 수 없어요. 계산식을 몰라서 AI가 '
                                                                     '추측한 거예요. **오류가 없다고 맞는 결과는 아니에요.**',
                                                          'hint': '계산식과 기준값을 숫자로 적어 주세요.'}},
                                                {'id': 'out',
                                                 'label': '출력 형식',
                                                 'why': '받고 싶은 모양을 알려요.',
                                                 'options': [{'text': '형상 계열은 행, 거리는 열인 표로 보여 주고, 기본 도형으로 그린 실루엣 6종 그림도 만들어 줘. 표를 넣은 HTML 보고서는 '
                                                                      'report 변수에 담아 줘.',
                                                              'good': True},
                                                             {'text': '결과를 보여 줘.', 'good': False},
                                                             {'text': '', 'good': False}],
                                                 'miss': {'say': '계산 결과를 출력했어요.',
                                                          'code': 'import pandas as pd\n'
                                                                  "df = pd.read_csv('{VDIR}/m1_families.csv')\n"
                                                                  'for _, r in df.iterrows():\n'
                                                                  '    for d in [100, 300, 500, 1000, 2000]:\n'
                                                                  "        print(r['family'], d, round((r['span_m'] * r['length_m']) ** 0.5 / (d * "
                                                                  '0.001), 1))',
                                                          'problem': '숫자는 맞지만 한 줄씩 길게 나와 비교하기 어렵고, 실루엣 그림도 없어 학습앱에 넣을 수 없어요.',
                                                          'hint': '행과 열을 어떻게 놓을지, 그림과 문서도 받을지 적어 주세요.'}}],
                                      'done': {'say': '요청대로 실루엣 6종과 거리별 픽셀 수·수준표, 보고서를 만들었어요.',
                                               'code': 'import pandas as pd\n'
                                                       'import matplotlib.pyplot as plt\n'
                                                       'from matplotlib.patches import Polygon, Circle\n'
                                                       '\n'
                                                       "df = pd.read_csv('{VDIR}/m1_families.csv')\n"
                                                       'DIST = [100, 300, 500, 1000, 2000]\n'
                                                       '\n'
                                                       'def level(px):\n'
                                                       '    if px >= 13:\n'
                                                       "        return '식별'\n"
                                                       '    if px >= 8:\n'
                                                       "        return '인식'\n"
                                                       '    if px >= 2:\n'
                                                       "        return '탐지'\n"
                                                       "    return '어려움'\n"
                                                       '\n'
                                                       'rows = []\n'
                                                       'for _, r in df.iterrows():\n'
                                                       "    size = (r['span_m'] * r['length_m']) ** 0.5\n"
                                                       "    row = {'형상 계열': r['family']}\n"
                                                       '    for d in DIST:\n'
                                                       '        px = round(size / (d * 0.001), 1)\n'
                                                       "        row[f'{d}m'] = f'{px}px {level(px)}'\n"
                                                       '    rows.append(row)\n'
                                                       'table = pd.DataFrame(rows)\n'
                                                       'display(table)\n'
                                                       '\n'
                                                       'SIL = {\n'
                                                       "    'multicopter': [('c', (-.6, .6), .28), ('c', (.6, .6), .28), ('c', (-.6, -.6), .28), "
                                                       "('c', (.6, -.6), .28), ('c', (0, 0), .2),\n"
                                                       "                    ('p', [(-.65, -.55), (-.55, -.65), (.65, .55), (.55, .65)]), ('p', "
                                                       '[(-.65, .55), (-.55, .65), (.65, -.55), (.55, -.65)])],\n'
                                                       "    'fixed_wing': [('p', [(-.08, -.9), (.08, -.9), (.08, .9), (-.08, .9)]), ('p', [(-1, "
                                                       ".15), (1, .15), (1, -.1), (-1, -.1)]), ('p', [(-.35, -.9), (.35, -.9), (.35, -.72), (-.35, "
                                                       '-.72)])],\n'
                                                       "    'delta_wing': [('p', [(0, 1), (1, -.8), (0, -.5), (-1, -.8)])],\n"
                                                       "    'cruise_missile': [('p', [(-.09, -1), (.09, -1), (.09, .85), (0, 1), (-.09, .85)]), "
                                                       "('p', [(-.6, .05), (.6, .05), (.6, -.1), (-.6, -.1)]), ('p', [(-.28, -1), (.28, -1), (.09, "
                                                       '-.75), (-.09, -.75)])],\n'
                                                       "    'ballistic_missile': [('p', [(-.11, -1), (.11, -1), (.11, .7), (0, 1), (-.11, .7)]), "
                                                       "('p', [(-.3, -1), (.3, -1), (.11, -.7), (-.11, -.7)])],\n"
                                                       "    'bird': [('p', [(0, .55), (.12, .1), (.1, -.5), (0, -.75), (-.1, -.5), (-.12, .1)]), "
                                                       "('p', [(-1, .15), (-.5, .35), (0, .2), (.5, .35), (1, .15), (.5, .05), (0, -.1), (-.5, "
                                                       '.05)])],\n'
                                                       '}\n'
                                                       'fig, axes = plt.subplots(1, 6, figsize=(9, 1.9))\n'
                                                       'for ax, (name, parts) in zip(axes, SIL.items()):\n'
                                                       '    for p in parts:\n'
                                                       "        ax.add_patch(Circle(p[1], p[2], color='black') if p[0] == 'c' else Polygon(p[1], "
                                                       "color='black'))\n"
                                                       '    ax.set_xlim(-1.1, 1.1)\n'
                                                       '    ax.set_ylim(-1.1, 1.1)\n'
                                                       "    ax.set_aspect('equal')\n"
                                                       '    ax.set_title(name, fontsize=8)\n'
                                                       "    ax.axis('off')\n"
                                                       '\n'
                                                       "report = ('<h2>거리별 식별 가능 수준표</h2>'\n"
                                                       "          '<p>기준: 2픽셀 탐지, 8픽셀 인식, 13픽셀 식별. 대표 크기는 폭과 길이의 기하평균이고, 1픽셀은 거리의 0.001배로 "
                                                       "계산했어요.</p>'\n"
                                                       '          + table.to_html(index=False))',
                                               'lesson': '네 칸을 모두 채우니 AI가 추측할 일이 없어졌어요. 멀티콥터와 새는 300m만 돼도 탐지 수준이고, 큰 비행체도 500m부터는 형상을 구분하기 어렵다는 것이 '
                                                         '표에서 바로 보여요.'},
                                      'on_ok': [['dm_gyo', 'gyo', '오, 실루엣과 표가 바로 나왔네요. {call}, 이 프롬프트 저도 좀 빌려 쓸게요!']]}]},
                         {'short': '내 말로 쓰기',
                          'title': '같은 일을 내 말로 요청하기',
                          'lead': '고른 문장을 그대로 외울 필요는 없어요. **네 칸만 채우면** 말투는 자유예요.',
                          'blocks': [{'id': 's3_q',
                                      'type': 'quiz',
                                      'title': '결과를 받은 뒤에는',
                                      'text': 'AI가 만든 코드가 오류 없이 돌았어요. 다음으로 무엇을 하면 좋을까요?',
                                      'options': ['오류가 없으니 그대로 쓴다.', '하나만 손으로 계산해 맞는지 확인한다.', '더 어려운 말로 다시 요청한다.'],
                                      'answer': 1,
                                      'explain': '멀티콥터는 √(0.6 × 0.6) = 0.6m이니 100m에서 0.6 ÷ 0.1 = 6픽셀이에요. 표의 값과 같은지 한 칸만 봐도 큰 실수를 잡을 수 있어요.',
                                      'hints': ['앞 활동에서 오류 없이 이상한 숫자가 나온 경우를 떠올려 보세요.', '전부 확인할 필요는 없어요. 한 칸이면 충분해요.']},
                                     {'id': 's3_write',
                                      'type': 'write',
                                      'title': '내 프롬프트 쓰기',
                                      'prompt': '**새 요청**: 새 형상 계열 jet_delta(폭 3.2m, 길이 3.8m)가 추가됐어요. 파일은 m1_families_new.csv이고 열은 같아요. 거리 300m, '
                                                '1000m에서의 픽셀 수를 보고 싶어요. 네 칸을 채워 내 말로 프롬프트를 써 보세요.',
                                      'example': '새 형상 계열이 거리별로 몇 픽셀로 보이는지 학습앱에 넣으려고 해. 데이터는 m1_families_new.csv이고 열은 family, span_m, length_m이야. 대표 '
                                                 '크기 = √(span_m × length_m), 픽셀 수 = 대표 크기 ÷ (거리 × 0.001)로 거리 300m, 1000m에서 계산해 줘. 형상 계열은 행, 거리는 열인 '
                                                 '표로 보여 줘.',
                                      'min': 30,
                                      'checklist': ['맥락(왜 필요한지)을 적었어요', '파일 이름과 열 이름을 적었어요', '계산식이나 기준을 숫자로 적었어요', '받고 싶은 모양을 적었어요'],
                                      'model_label': '📝 모범 프롬프트',
                                      'placeholder': '꼭 입력해 주세요 · 내 말로 프롬프트를 써 보세요'}]}],
               'outputs': [{'id': 'm1_level_table',
                            'title': 'M1 형상 계열 실루엣과 식별 가능 수준표',
                            'sections': [{'h': '1. AI에게 보낸 프롬프트와 결과', 'blocks': ['s2_vibe']}, {'h': '2. 내가 쓴 프롬프트', 'blocks': ['s3_write']}]}]},
              {'id': 'm2',
               'no': 2,
               'title': '라벨링 규격 합의와 정사영상 배경 데이터셋 구성',
               'short': '데이터 설명하기',
               'kind': '협업',
               'competencies': ['D2', 'C1'],
               'outline': '두 사람이 같은 실루엣 30장에 붙인 라벨이 얼마나 같은지 계산하면서, 파일과 값의 모양을 AI에게 알려 주는 법을 익혀요. 결과로 라벨링 규격 문서도 요청해요.',
               'brief': {'subtitle': '라벨 일치율 계산과 규격 문서: 데이터를 설명하기',
                         'time': '약 25분',
                         'goal': '값이 지저분한 라벨 데이터를 AI에게 설명해 일치율과 코헨 카파를 계산하고, 그 결과를 근거로 라벨링 규격 문서를 요청해요.',
                         'why': '두 사람이 같은 실루엣에 다른 라벨을 붙였다면 누구 말이 맞을까요? 먼저 값이 어떻게 적혀 있는지부터 봐야 해요. AI는 전우님의 데이터가 지저분하다는 걸 모를 수 있거든요.',
                         'case': {'title': '해외 사례: 같은 모양이어도 다른 물체일 수 있어요',
                                  'source': 'Kyiv Independent, Ukrainska Pravda 보도 (2024.11, 2025.2)',
                                  'points': ['2024년 11월 우크라이나 국방정보총국은 러시아가 쏘는 드론의 절반가량이 샤헤드를 흉내 낸 값싼 기만용 드론이라고 밝혔어요. 합판과 해외 부품으로 만들어 방공망에 부담을 준다는 '
                                             '내용이에요.',
                                             '2025년 2월에는 우크라이나 공군도 공중 위협의 절반은 실제, 절반은 기만용이라고 설명했어요.',
                                             '모양이 같아도 정체는 다를 수 있어요. 라벨 하나에 모양과 정체를 섞으면 데이터 전체가 흔들려요.']},
                         'team': '나라벨(데이터 담당)과 고교관(교관)의 라벨이 서로 달라요. 전우님이 AI로 일치율을 계산해 이야기의 근거를 만들어요.',
                         'terms': [{'t': '라벨링 규격',
                                    'en': 'annotation schema',
                                    'fig': 'f_schema',
                                    'd': '라벨을 어떤 값으로, 어떤 기준으로 붙일지 정한 약속이에요.',
                                    'easy': '모두가 같은 양식으로 적게 하는 작성 규칙이에요.',
                                    'use': 'AI에게 값의 목록을 알려 주면 규격 위반도 찾아 줘요.'},
                                   {'t': '코헨 카파',
                                    'en': "Cohen's kappa",
                                    'fig': 'f_kappa',
                                    'd': '두 사람의 일치율에서 우연히 같을 확률을 뺀 지표예요.',
                                    'easy': '동전 던지기로도 어느 정도는 같아지니, 그만큼을 빼고 보는 거예요.',
                                    'use': '1에 가까울수록 잘 맞아요. 0.6 이상이면 대체로 괜찮다고 봐요.'},
                                   {'t': '데이터 정리',
                                    'en': 'data cleaning',
                                    'fig': 'f_clean',
                                    'd': '대소문자, 빈칸처럼 값이 제각각인 것을 하나로 맞추는 일이에요.',
                                    'easy': '‘Bird’와 ‘bird ’를 같은 값으로 맞추는 일이에요.',
                                    'use': '프롬프트에 ‘값이 지저분하니 먼저 정리해 줘’라고 적어요.'}],
                         'steps': ['데이터 보여 주기: AI에게 데이터를 설명하는 법', '일치율 계산: 일치율을 계산하고 갈린 라벨 찾기', '기준 합의: 결과로 기준을 합의하고 내 말로 요청하기'],
                         'glossary': [],
                         'outputs': ['M2 라벨 일치율과 라벨링 규격 요청']},
               'start_messages': [['team', 'label', '{call}, 같은 실루엣 30장을 고교관님과 제가 라벨링했는데 서로 많이 다른 것 같아요.'],
                                  ['team', 'gyo', '먼저 숫자로 봅시다. 파일은 m2_labels.csv예요. 그런데 값을 제각각으로 적었어요.']],
               'finish_messages': [['team', 'gyo', '{call}, 일치율과 규격 요청까지 나왔네요. 다음 회의는 이걸로 시작합시다.']],
               'steps': [{'short': '데이터 보여 주기',
                          'title': 'AI에게 데이터를 설명하는 법',
                          'lead': 'AI는 파일을 볼 수 없어요. **앞부분 몇 줄과 값의 종류**를 알려 주는 것이 가장 빠른 방법이에요.',
                          'blocks': [{'id': 's1_info',
                                      'type': 'info',
                                      'title': '데이터를 설명할 때 넣을 것',
                                      'icon': '🗂',
                                      'text': ['열 이름만으로는 부족할 때가 많아요. 값이 어떻게 적혀 있는지까지 알려 주세요. 이 파일의 실루엣은 하늘과 정사영상(지형 사진을 지도처럼 편 영상) 배경에 올린 것이라 '
                                               'background 열도 있어요.'],
                                      'bullets': ['**파일 이름과 열 이름**: m2_labels.csv, image_id, background, view, rater_a, rater_b',
                                                  '**값의 종류**: multicopter, fixed_wing, delta_wing, cruise_missile, ballistic_missile, bird 여섯 가지',
                                                  '**주의할 점**: 대소문자, 앞뒤 빈칸, 밑줄 대신 띄어쓰기가 섞여 있음',
                                                  '**샘플 몇 줄**: 실제 값 대신 가상 샘플로 보여 줘도 돼요'],
                                      'figure': 'f_clean'},
                                     {'id': 's1_q',
                                      'type': 'quiz',
                                      'title': '먼저 알려 줄 것',
                                      'text': '값에 ‘Bird’, ‘bird ’가 섞여 있어요. 이 사실을 알려 주지 않으면 어떻게 될까요?',
                                      'options': ['AI가 스스로 알아채 항상 정리해 준다.', '같은 값을 다른 값으로 세어 일치율이 실제보다 낮게 나올 수 있다.', '파일을 읽지 못해 반드시 오류가 난다.'],
                                      'answer': 1,
                                      'explain': '오류 없이 틀린 숫자가 나와서 더 위험해요. 다음 활동에서 직접 확인해 봐요.',
                                      'hints': ['컴퓨터는 ‘Bird’와 ‘bird’를 같은 글자로 볼까요?', '오류가 나는 경우와 조용히 틀리는 경우를 나눠 생각해 보세요.']}]},
                         {'short': '일치율 계산',
                          'title': '일치율을 계산하고 갈린 라벨 찾기',
                          'lead': '**값 정리** 문장을 빼고도 보내 보세요. 숫자가 어떻게 달라지는지 비교해 보면 좋아요.',
                          'blocks': [{'id': 's2_vibe',
                                      'type': 'vibe',
                                      'title': '라벨 일치율 계산',
                                      'ask': ['label', '두 사람이 얼마나 같게 라벨을 붙였는지, 어떤 실루엣에서 갈렸는지 알고 싶어요.'],
                                      'file': 'm2_labels.csv',
                                      'slots': [{'id': 'input',
                                                 'label': '입력',
                                                 'why': '값의 모양까지 알려요.',
                                                 'options': [{'text': 'm2_labels.csv의 열은 image_id, background, view, rater_a, rater_b이고, 라벨 값은 '
                                                                      'multicopter, fixed_wing, delta_wing, cruise_missile, ballistic_missile, '
                                                                      'bird야. 대소문자와 빈칸이 섞여 있으니 먼저 소문자로 바꾸고, 앞뒤 빈칸을 지우고, 띄어쓰기는 밑줄로 바꿔 줘.',
                                                              'good': True},
                                                             {'text': 'm2_labels.csv에 두 사람의 라벨이 있어.', 'good': False},
                                                             {'text': '', 'good': False}],
                                                 'miss': {'say': '일치율을 계산했어요.',
                                                          'code': 'import pandas as pd\n'
                                                                  "df = pd.read_csv('{VDIR}/m2_labels.csv')\n"
                                                                  "print('일치율:', round((df['rater_a'] == df['rater_b']).mean() * 100, 1), '%')",
                                                          'problem': '오류는 없지만 ‘Bird’와 ‘bird ’를 다른 값으로 세어 일치율이 실제보다 낮아요. 값을 정리한 결과와 비교해 보세요.',
                                                          'hint': '값이 어떻게 적혀 있는지, 무엇을 먼저 정리할지 적어 주세요.'}},
                                                {'id': 'task',
                                                 'label': '할 일',
                                                 'why': '무엇을 세는지 정해요.',
                                                 'options': [{'text': '여섯 값에 없는 라벨은 규격 위반으로 따로 뽑고, 나머지로 일치율과 코헨 카파(일치율에서 우연히 같을 확률을 뺀 값)를 계산해 줘. 두 '
                                                                      '사람이 다른 실루엣 목록도 뽑아 줘.',
                                                              'good': True},
                                                             {'text': '일치도를 분석해 줘.', 'good': False},
                                                             {'text': '', 'good': False}],
                                                 'miss': {'say': 'rater_a를 정답으로 두고 정확도를 계산했어요.',
                                                          'code': 'import pandas as pd\n'
                                                                  '\n'
                                                                  "df = pd.read_csv('{VDIR}/m2_labels.csv')\n"
                                                                  "ALLOWED = ['multicopter', 'fixed_wing', 'delta_wing', 'cruise_missile', "
                                                                  "'ballistic_missile', 'bird']\n"
                                                                  "for c in ['rater_a', 'rater_b']:\n"
                                                                  "    df[c] = df[c].str.strip().str.lower().str.replace(' ', '_')\n"
                                                                  "print('정확도:', round((df['rater_b'] == df['rater_a']).mean(), 2))",
                                                          'problem': '규격에 없는 decoy까지 그대로 비교했고, 한 사람을 정답처럼 다뤘어요. 어떤 비교인지 말하지 않아 AI가 추측했어요.',
                                                          'hint': '규격 위반을 어떻게 다룰지, 어떤 지표와 목록이 필요한지 적어 주세요.'}},
                                                {'id': 'out',
                                                 'label': '출력 형식',
                                                 'why': '회의에서 볼 모양으로 받아요.',
                                                 'options': [{'text': '요약 표, 규격 위반 표, 갈린 실루엣 표를 보여 주고, 세 표를 넣은 HTML 문서를 report 변수에 담아 줘.',
                                                              'good': True},
                                                             {'text': '결과를 정리해 줘.', 'good': False},
                                                             {'text': '', 'good': False}],
                                                 'miss': {'say': '결과를 문장으로 정리했어요.',
                                                          'code': 'import pandas as pd\n'
                                                                  '\n'
                                                                  "df = pd.read_csv('{VDIR}/m2_labels.csv')\n"
                                                                  "ALLOWED = ['multicopter', 'fixed_wing', 'delta_wing', 'cruise_missile', "
                                                                  "'ballistic_missile', 'bird']\n"
                                                                  "for c in ['rater_a', 'rater_b']:\n"
                                                                  "    df[c] = df[c].str.strip().str.lower().str.replace(' ', '_')\n"
                                                                  "print('두 사람의 라벨은 대체로 양호합니다.')",
                                                          'problem': '‘대체로 양호’라는 말뿐이에요. 숫자가 없어 회의에서 근거로 쓸 수 없어요.',
                                                          'hint': '표로 받고 싶다고, 무엇을 넣을지 적어 주세요.'}}],
                                      'done': {'say': '값을 정리한 뒤 일치율, 코헨 카파, 규격 위반과 갈린 실루엣 목록을 만들었어요.',
                                               'code': 'import pandas as pd\n'
                                                       '\n'
                                                       "df = pd.read_csv('{VDIR}/m2_labels.csv')\n"
                                                       "ALLOWED = ['multicopter', 'fixed_wing', 'delta_wing', 'cruise_missile', 'ballistic_missile', "
                                                       "'bird']\n"
                                                       "for c in ['rater_a', 'rater_b']:\n"
                                                       "    df[c] = df[c].str.strip().str.lower().str.replace(' ', '_')\n"
                                                       '\n'
                                                       "bad = df[~df['rater_a'].isin(ALLOWED) | ~df['rater_b'].isin(ALLOWED)]\n"
                                                       'ok = df.drop(bad.index)\n'
                                                       "po = (ok['rater_a'] == ok['rater_b']).mean()\n"
                                                       "pe = sum((ok['rater_a'] == k).mean() * (ok['rater_b'] == k).mean() for k in ALLOWED)\n"
                                                       'kappa = (po - pe) / (1 - pe)\n'
                                                       "summary = pd.DataFrame([{'전체': len(df), '규격 위반': len(bad), '비교한 수': len(ok), '일치율(%)': "
                                                       "round(po * 100, 1), '코헨 카파': round(kappa, 2)}])\n"
                                                       'display(summary)\n'
                                                       '\n'
                                                       "cols = ['image_id', 'view', 'rater_a', 'rater_b']\n"
                                                       'display(bad[cols])\n'
                                                       "split = ok[ok['rater_a'] != ok['rater_b']]\n"
                                                       'display(split[cols])\n'
                                                       '\n'
                                                       "report = ('<h2>라벨 일치율</h2>' + summary.to_html(index=False)\n"
                                                       "          + f'<h3>규격 위반 {len(bad)}건</h3>' + bad[cols].to_html(index=False)\n"
                                                       "          + f'<h3>갈린 실루엣 {len(split)}건</h3>' + split[cols].to_html(index=False))",
                                               'lesson': '값의 모양을 알려 주니 조용히 틀리는 일이 사라졌어요. 규격 위반 칸에는 모양이 아니라 정체인 decoy가 나왔어요. 라벨에 모양과 정체가 섞였다는 뜻이에요.'},
                                      'on_ok': [['team', 'gyo', '{call}, decoy가 형상 칸에 들어가 있었군요. 표로 보니 기준을 어떻게 고칠지 보이네요.']]}]},
                         {'short': '기준 합의',
                          'title': '결과로 기준을 합의하고 내 말로 요청하기',
                          'lead': 'AI는 계산을 도와줄 뿐이에요. 어떤 기준을 쓸지는 **사람이 근거를 보고 정해요.**',
                          'blocks': [{'id': 's3_decide',
                                      'type': 'decide',
                                      'title': '팀원의 제안 판단하기',
                                      'text': '일치율 결과를 본 팀원들의 제안이에요. 받아들일지 정하고 이유를 적어 주세요.',
                                      'reason_min': 8,
                                      'items': [{'id': 'a',
                                                 'who': 'label',
                                                 'text': 'decoy도 형상 라벨 값에 넣어요. 어차피 구분해야 하잖아요.',
                                                 'answer': 'reject',
                                                 'hint': '실루엣만 보고 진짜와 기만용을 가를 수 있을까요?',
                                                 'explain': '모양이 같으면 실루엣으로 가를 수 없어요. 형상 칸에는 보이는 모양만, 정체는 다른 칸에 따로 적어요.',
                                                 'reason': '모양과 정체를 나눠야 해요. 정체는 실루엣만으로 알 수 없어 다른 칸에 적어요.'},
                                                {'id': 'b',
                                                 'who': 'gyo',
                                                 'text': '형상 값은 여섯 가지 소문자로만 입력하게 하고, 정체는 ‘미상’을 기본값으로 하는 칸을 따로 둬요.',
                                                 'answer': 'accept',
                                                 'hint': '앞에서 데이터를 정리하느라 무엇이 번거로웠는지 떠올려 보세요.',
                                                 'explain': '입력 규칙을 정하면 정리할 일이 줄고, AI에게 설명하기도 쉬워져요.',
                                                 'reason': '입력 규칙과 칸을 나누면 값이 흔들리지 않아 계산과 설명이 쉬워져요.'}]},
                                     {'id': 's3_write',
                                      'type': 'write',
                                      'title': '규격 문서를 요청하는 프롬프트 쓰기',
                                      'prompt': '방금 합의한 내용으로 **라벨링 규격 문서**를 AI에게 요청해 보세요. 누가 읽는지, 어떤 항목이 들어가야 하는지, 근거 데이터를 적어 주세요.',
                                      'example': '새로 온 라벨링 담당이 읽을 라벨링 규격 문서를 만들어 줘. 항목은 형상 값 여섯 가지와 값마다의 WEFT 기준, 입력 규칙(소문자, 빈칸 없음, 밑줄), 정체 칸의 기본값 '
                                                 '‘미상’, 애매할 때 처리 방법이야. 근거로 m2_labels.csv의 일치율 표와 규격 위반 목록을 넣어 줘. 한 쪽 분량의 HTML로 만들어 줘.',
                                      'min': 30,
                                      'checklist': ['누가 읽는 문서인지 적었어요', '들어갈 항목을 적었어요', '근거 데이터를 적었어요', '분량이나 형식을 적었어요'],
                                      'model_label': '📝 모범 프롬프트',
                                      'placeholder': '꼭 입력해 주세요 · 내 말로 프롬프트를 써 보세요'}]}],
               'outputs': [{'id': 'm2_label_doc',
                            'title': 'M2 라벨 일치율과 라벨링 규격 요청',
                            'sections': [{'h': '1. 일치율 계산', 'blocks': ['s2_vibe']},
                                         {'h': '2. 팀원 제안에 대한 결정', 'blocks': ['s3_decide']},
                                         {'h': '3. 규격 문서를 요청하는 프롬프트', 'blocks': ['s3_write']}]}]},
              {'id': 'm3',
               'no': 3,
               'title': '전처리와 형태 특징으로 만드는 기준 모델 3종',
               'short': '오류 고쳐 달라고 하기',
               'kind': '개인',
               'competencies': ['D3'],
               'outline': '기준 모델 3종의 예측을 조건별로 비교하는 표를 받고, 표본이 없는 조건에서 난 오류를 메시지 그대로 알려 고쳐 달라고 요청해요.',
               'brief': {'subtitle': '조건별 성능표와 오류 수정: 다시 요청하는 법',
                         'time': '약 25분',
                         'goal': '기준 모델의 조건별 성능표를 요청하고, 오류가 나면 메시지와 원하는 동작, 고칠 범위를 담아 수정을 요청해요.',
                         'why': '모양을 숫자로 바꾸면 멀리 있는 실루엣도 가를 수 있을까요? 조건을 나눠 봐야 알 수 있어요. 그런데 조건을 나누다 보면 오류가 자주 나요. 오류를 만났을 때 어떻게 다시 요청하는지가 바이브 코딩의 '
                                '절반이에요.',
                         'case': {'title': '해외 사례: 드론과 새를 가리는 대회의 어려움',
                                  'source': 'Laroca 외, arXiv 2504.19347 (2025.4)',
                                  'points': ['소형 드론을 새와 구별해 찾는 국제 대회(Drone-vs-Bird)에서 상위 3위 안에 든 연구팀의 보고서예요.',
                                             '연구팀은 작은 드론이 새와 구별하기 어려운 경우가 많다고 보고, 프레임 하나만 보지 않고 앞뒤 프레임을 함께 살펴 놓친 탐지를 줄였어요.',
                                             '조건이 나빠지면 성능이 어떻게 무너지는지 조건별로 나눠 보는 일이 먼저예요. 이번 미션의 표가 그 시작이에요.']},
                         'team': '한모델(AI 개발 담당)이 기준 모델의 예측 파일을 넘겨 줘요. 전우님이 AI로 표를 받고, 오류가 나면 고쳐 달라고 요청해요.',
                         'terms': [{'t': '기준 모델',
                                    'en': 'baseline model',
                                    'fig': 'f_base',
                                    'd': '복잡한 모델과 비교할 출발점이 되는 단순한 모델이에요.',
                                    'easy': '새 방법이 정말 나아졌는지 재는 자예요.',
                                    'use': '결정 트리, 로지스틱 회귀, k-NN 세 가지로 비교해요.'},
                                   {'t': '형태 특징과 전처리',
                                    'en': 'shape feature, preprocessing',
                                    'fig': 'f_pre',
                                    'd': '실루엣을 같은 기준으로 맞추고(잡음 제거, 크기 맞춤), 면적·길쭉함 같은 숫자로 바꾸는 일이에요.',
                                    'easy': '사진을 키, 몸무게 같은 숫자 카드로 바꾸는 거예요.',
                                    'use': '이번에는 그 결과로 나온 예측만 다뤄요. 미디언 필터는 점잡음을 지우는 전처리예요.'},
                                   {'t': '오류 메시지',
                                    'en': 'error message',
                                    'fig': 'f_debug',
                                    'd': '코드가 멈춘 이유를 컴퓨터가 알려 주는 글이에요.',
                                    'easy': '맨 아래 줄이 가장 중요해요. 그대로 복사해 AI에게 주면 돼요.',
                                    'use': 'ZeroDivisionError는 0으로 나눴다는 뜻이에요.'}],
                         'steps': ['성능표 받기: 모델과 조건을 나눠 표로 받기', '오류 고치기: 오류 메시지로 수정 요청하기', '내 말로 쓰기: 다른 오류를 내 말로 요청하기'],
                         'glossary': [],
                         'outputs': ['M3 기준 모델 조건별 성능표와 수정 기록']},
               'start_messages': [['dm_model', 'model', '{call}, 기준 모델 3종의 예측이에요. 파일은 m3_preds.csv예요. 거리별로 어떤 모델이 나은지 표를 받아 볼래요?']],
               'finish_messages': [['dm_model', 'model', '{call}, 고칠 곳만 고쳐 주셨네요. 표본이 없는 조건도 바로 보여서 좋아요.']],
               'steps': [{'short': '성능표 받기',
                          'title': '모델과 조건을 나눠 표로 받기',
                          'lead': '전체 정확도 하나로는 어디서 무너지는지 알 수 없어요. **무엇으로 나눌지** 적어 주세요.',
                          'blocks': [{'id': 's1_vibe',
                                      'type': 'vibe',
                                      'title': '기준 모델 조건별 성능표',
                                      'ask': ['model', '모델과 거리별로 정확도를 비교하고 싶어요.'],
                                      'file': 'm3_preds.csv',
                                      'slots': [{'id': 'input',
                                                 'label': '입력',
                                                 'why': '열의 뜻을 알려요.',
                                                 'options': [{'text': 'm3_preds.csv의 열은 image_id, model(decision_tree·logistic·knn), '
                                                                      'distance(near·mid·far), polarity(white_hot·black_hot), truth(실제 라벨), pred(예측 '
                                                                      '라벨)야.',
                                                              'good': True},
                                                             {'text': '모델 결과 CSV가 있어.', 'good': False},
                                                             {'text': '', 'good': False}],
                                                 'miss': {'say': '모델 결과로 정확도를 계산했어요.',
                                                          'expect': 'error',
                                                          'code': 'import pandas as pd\n'
                                                                  "df = pd.read_csv('{VDIR}/m3_preds.csv')\n"
                                                                  "print(df.groupby('classifier')['accuracy'].mean())",
                                                          'problem': 'AI가 열 이름을 classifier, accuracy로 추측해서 KeyError가 났어요.',
                                                          'hint': '열 이름과 뜻을 그대로 적어 주세요.'}},
                                                {'id': 'task',
                                                 'label': '할 일',
                                                 'why': '무엇으로 나눌지 정해요.',
                                                 'options': [{'text': 'truth와 pred가 같으면 정답으로 보고, 모델과 거리별 정확도(%)를 소수 첫째 자리까지 계산해 줘.', 'good': True},
                                                             {'text': '모델 성능을 알려 줘.', 'good': False},
                                                             {'text': '', 'good': False}],
                                                 'miss': {'say': '모델별 전체 정확도를 계산했어요.',
                                                          'code': 'import pandas as pd\n'
                                                                  "df = pd.read_csv('{VDIR}/m3_preds.csv')\n"
                                                                  "df['ok'] = df['truth'] == df['pred']\n"
                                                                  "display(df.groupby('model')['ok'].mean().round(3).reset_index())",
                                                          'problem': '오류는 없지만 숫자 하나로 합쳐져서 먼 거리에서 무너지는 것이 가려졌어요.',
                                                          'hint': '거리별로 나눠 달라고 적어 주세요.'}},
                                                {'id': 'out',
                                                 'label': '출력 형식',
                                                 'why': '표의 모양을 정해요.',
                                                 'options': [{'text': '모델은 행, 거리는 near·mid·far 순서의 열인 표로 보여 주고, 같은 값을 막대그래프로도 그려 줘.', 'good': True},
                                                             {'text': '보기 좋게 보여 줘.', 'good': False},
                                                             {'text': '', 'good': False}],
                                                 'miss': {'say': '보기 좋게 문장으로 정리했어요.',
                                                          'code': 'import pandas as pd\n'
                                                                  "df = pd.read_csv('{VDIR}/m3_preds.csv')\n"
                                                                  "for m in df['model'].unique():\n"
                                                                  "    print(f'{m} 모델의 평가를 마쳤습니다!')",
                                                          'problem': '‘보기 좋게’는 사람마다 달라요. 결과 없이 평가했다는 말만 나왔어요.',
                                                          'hint': '행과 열, 그림까지 적어 주세요.'}}],
                                      'done': {'say': '모델과 거리별 성능표와 그래프를 만들었어요.',
                                               'code': 'import pandas as pd\n'
                                                       'import matplotlib.pyplot as plt\n'
                                                       '\n'
                                                       "df = pd.read_csv('{VDIR}/m3_preds.csv')\n"
                                                       "df['ok'] = df['truth'] == df['pred']\n"
                                                       "table = df.pivot_table(index='model', columns='distance', values='ok', "
                                                       "aggfunc='mean')[['near', 'mid', 'far']]\n"
                                                       'table = (table * 100).round(1)\n'
                                                       'display(table.reset_index())\n'
                                                       '\n'
                                                       "ax = table.plot(kind='bar', figsize=(6, 3), rot=0)\n"
                                                       "ax.set_ylabel('accuracy (%)')\n"
                                                       'ax.set_ylim(0, 100)',
                                               'lesson': '거리로 나누니 세 모델 모두 far에서 크게 떨어지는 것이 보여요. 나눠 보지 않았다면 놓쳤을 거예요.'},
                                      'on_ok': [['dm_model',
                                                 'model',
                                                 '{call}, 표 좋아요. 그런데 야간 열화상 극성(white_hot·black_hot)까지 나눠 보고 싶어요. AI에게 조건을 더해 달라고 해 볼래요?']]}]},
                         {'short': '오류 고치기',
                          'title': '오류 메시지로 수정 요청하기',
                          'lead': '극성까지 나눴더니 **오류**가 났어요. 어떻게 다시 요청할까요?',
                          'blocks': [{'id': 's2_vibe',
                                      'type': 'vibe',
                                      'title': '오류 고쳐 달라고 하기',
                                      'ask': ['model', '극성까지 나눴더니 오류가 나요. 저도 원인을 잘 모르겠어요… 부탁해요!'],
                                      'file': 'm3_preds.csv',
                                      'slots': [{'id': 'what',
                                                 'label': '무슨 일이 있었는지',
                                                 'why': '오류 메시지를 그대로 붙여 넣어요.',
                                                 'options': [{'text': '모델·거리·극성별 정확도를 계산하다가 이 오류가 났어: ZeroDivisionError: division by zero',
                                                              'good': True},
                                                             {'text': '안 돼. 다시 해 줘.', 'good': False},
                                                             {'text': '', 'good': False}],
                                                 'miss': {'say': '다시 만들었어요. 이번에는 극성을 빼 봤어요.',
                                                          'code': 'import pandas as pd\n'
                                                                  "df = pd.read_csv('{VDIR}/m3_preds.csv')\n"
                                                                  "df['ok'] = df['truth'] == df['pred']\n"
                                                                  "display(df.groupby(['model', 'distance'])['ok'].mean().round(3).reset_index())",
                                                          'problem': '오류는 사라졌지만 극성으로 나누는 기능도 함께 사라졌어요. 무엇이 안 되는지 몰라서 AI가 조건을 줄였어요.',
                                                          'hint': '오류 메시지를 그대로 알려 주세요.'}},
                                                {'id': 'fix',
                                                 'label': '어떻게 고칠지',
                                                 'why': '원하는 동작과 범위를 정해요.',
                                                 'options': [{'text': 'far에 black_hot 표본이 없어서 0으로 나눈 것 같아. 표본이 없으면 ‘자료 없음’으로 표시하게 acc 함수만 고치고, 나머지 '
                                                                      '계산은 그대로 둬. 조건별 표본 수도 함께 보여 주고, 표를 넣은 HTML 성능표 문서를 report 변수에 담아 줘.',
                                                              'good': True},
                                                             {'text': '전부 새로 짜 줘.', 'good': False},
                                                             {'text': '', 'good': False}],
                                                 'miss': {'say': '처음부터 새로 짰어요. 더 간단하게 정리했어요.',
                                                          'code': 'import pandas as pd\n'
                                                                  "df = pd.read_csv('{VDIR}/m3_preds.csv')\n"
                                                                  "df['ok'] = df['truth'] == df['pred']\n"
                                                                  "print(round(df['ok'].mean() * 100, 1))",
                                                          'problem': '오류는 사라졌지만 조건별 표까지 함께 사라졌어요. 범위를 정하지 않으면 잘 되던 부분도 바뀌어요.',
                                                          'hint': '무엇은 그대로 두고 어디만 어떻게 바꿀지 적어 주세요.'}}],
                                      'done': {'say': 'acc 함수만 고치고, 표본 수와 문서를 추가했어요.',
                                               'code': 'import pandas as pd\n'
                                                       'import matplotlib.pyplot as plt\n'
                                                       '\n'
                                                       "df = pd.read_csv('{VDIR}/m3_preds.csv')\n"
                                                       '\n'
                                                       'def acc(df, model, dist, pol):\n'
                                                       "    s = df[(df['model'] == model) & (df['distance'] == dist) & (df['polarity'] == pol)]\n"
                                                       '    if len(s) == 0:\n'
                                                       "        return '자료 없음'\n"
                                                       "    return round(int((s['truth'] == s['pred']).sum()) / len(s) * 100, 1)\n"
                                                       '\n'
                                                       'rows = []\n'
                                                       "for model in ['decision_tree', 'logistic', 'knn']:\n"
                                                       "    for dist in ['near', 'mid', 'far']:\n"
                                                       "        for pol in ['white_hot', 'black_hot']:\n"
                                                       "            n = len(df[(df['model'] == model) & (df['distance'] == dist) & (df['polarity'] "
                                                       '== pol)])\n'
                                                       "            rows.append({'모델': model, '거리': dist, '극성': pol, '정확도(%)': acc(df, model, dist, "
                                                       "pol), '표본': n})\n"
                                                       'table = pd.DataFrame(rows)\n'
                                                       'display(table)\n'
                                                       '\n'
                                                       "report = ('<h2>기준 모델 조건별 성능표</h2><p>가상 예측 기록 기준이에요. ‘자료 없음’은 표본이 없는 조건이에요.</p>' + "
                                                       'table.to_html(index=False))',
                                               'lesson': '오류 메시지와 원하는 동작, 고칠 범위를 함께 적으니 한 번에 고쳐졌어요. 잘 되던 계산은 그대로이고, 표본이 없는 조건이 ‘자료 없음’으로 드러났어요.'},
                                      'start': {'say': '극성으로 나누는 조건을 추가했어요.',
                                                'expect': 'error',
                                                'code': 'import pandas as pd\n'
                                                        '\n'
                                                        "df = pd.read_csv('{VDIR}/m3_preds.csv')\n"
                                                        '\n'
                                                        'def acc(df, model, dist, pol):\n'
                                                        "    s = df[(df['model'] == model) & (df['distance'] == dist) & (df['polarity'] == pol)]\n"
                                                        "    return round(int((s['truth'] == s['pred']).sum()) / len(s) * 100, 1)\n"
                                                        '\n'
                                                        'rows = []\n'
                                                        "for model in ['decision_tree', 'logistic', 'knn']:\n"
                                                        "    for dist in ['near', 'mid', 'far']:\n"
                                                        "        for pol in ['white_hot', 'black_hot']:\n"
                                                        "            rows.append({'모델': model, '거리': dist, '극성': pol, '정확도(%)': acc(df, model, dist, "
                                                        'pol)})\n'
                                                        'display(pd.DataFrame(rows))'}}]},
                         {'short': '내 말로 쓰기',
                          'title': '다른 오류를 내 말로 요청하기',
                          'lead': '오류를 만나면 **겁먹지 말고 그대로 붙여 넣으세요.** 코드를 다 읽지 못해도 수정을 요청할 수 있어요.',
                          'blocks': [{'id': 's3_write',
                                      'type': 'write',
                                      'title': '수정 프롬프트 쓰기',
                                      'prompt': "**새 상황**: 새 파일을 넣었더니 `KeyError: 'polarity'`가 났어요. 이 파일의 실제 열 이름은 thermal_mode예요. 잘 되던 정확도 계산은 그대로 "
                                                '두고 싶어요. 어떻게 요청할까요?',
                                      'example': "새 파일을 넣었더니 KeyError: 'polarity'가 났어. 이 파일의 실제 열 이름은 thermal_mode야. 파일을 읽은 뒤 thermal_mode를 "
                                                 'polarity로 바꾸는 줄만 추가하고, 정확도 계산과 나머지 코드는 그대로 둬.',
                                      'min': 30,
                                      'checklist': ['오류 메시지를 그대로 적었어요', '원인이 될 만한 사실(실제 열 이름)을 적었어요', '원하는 동작을 적었어요', '그대로 둘 부분을 적었어요'],
                                      'model_label': '📝 모범 프롬프트',
                                      'placeholder': '꼭 입력해 주세요 · 내 말로 프롬프트를 써 보세요'}]}],
               'outputs': [{'id': 'm3_baseline',
                            'title': 'M3 기준 모델 조건별 성능표와 수정 기록',
                            'sections': [{'h': '1. 조건별 성능표', 'blocks': ['s1_vibe']},
                                         {'h': '2. 오류 수정과 문서', 'blocks': ['s2_vibe']},
                                         {'h': '3. 내가 쓴 수정 프롬프트', 'blocks': ['s3_write']}]}]},
              {'id': 'm4',
               'no': 4,
               'title': '딥러닝 분류 모델 비교와 모델 선정 회의',
               'short': '쪼개서 요청하기',
               'kind': '협업',
               'competencies': ['D4', 'C1'],
               'outline': '요구 성능을 먼저 정하고, 모델 하나를 판정하는 작은 함수부터 만들어 달라고 요청해요. 기준을 바꿔 다시 돌리면서 모델 선정 회의의 근거를 만들어요.',
               'brief': {'subtitle': '요구 성능 판정 함수와 모델 비교: 쪼개서 요청하기',
                         'time': '약 30분',
                         'goal': '큰 일을 작은 단계로 쪼개 요청하고, 조건을 바꿔 돌릴 수 있는 함수로 후보 모델을 비교해 선정 근거를 만들어요.',
                         'why': '가장 정확한 모델이 학습앱에 가장 알맞은 모델일까요? 속도, 크기, 외부 전송, 라이선스까지 봐야 해요. 이렇게 조건이 많을 때 한 번에 전부 요청하면 어디가 틀렸는지 찾기 어려워요. 작게 만들고 확인한 뒤 '
                                '넓혀 가요.',
                         'case': {'title': '해외 사례: 위성영상까지 학습한 큰 모델, 그래도 따져 볼 조건',
                                  'source': 'Meta AI 발표, Hugging Face 모델 카드 (2025.8)',
                                  'points': ['2025년 8월 Meta는 라벨 없이 학습한 영상 모델 DINOv3를 공개했어요. Maxar 위성영상으로 학습한 버전도 함께 내놓았어요.',
                                             '모델 카드를 보면 위성 쪽 데이터만 약 4억 9천만 장이에요. 이전 버전과 달리 자체 라이선스를 쓰고, 조건에 동의해야 가중치를 받을 수 있어요.',
                                             '성능이 좋다고 바로 쓸 수 있는 건 아니에요. 기기에서 돌아가는지, 라이선스가 맞는지 조건별로 따져 봐야 해요.']},
                         'team': '고교관은 재현율, 전산호는 속도와 크기, 배보안은 외부 전송을 먼저 봐요. 전우님이 같은 표로 근거를 만들어 조율해요.',
                         'terms': [{'t': '요구 성능',
                                    'en': 'requirements',
                                    'fig': 'f_req',
                                    'd': '무엇을, 얼마나 잘, 얼마나 빨리 해야 하는지 숫자로 정한 기준이에요.',
                                    'easy': '모델을 고르기 전에 정하는 합격선이에요.',
                                    'use': '먼저 정해야 결과를 보고 기준을 바꾸는 일을 막을 수 있어요.'},
                                   {'t': '전이학습',
                                    'en': 'transfer learning',
                                    'fig': 'f_tl',
                                    'd': '다른 일에 이미 학습된 모델을 가져와 일부만 새로 학습하는 방법이에요.',
                                    'easy': '다른 언어를 배워 본 사람이 새 언어를 더 빨리 배우는 것과 비슷해요.',
                                    'use': '이번에는 학습은 하지 않고 측정값만 비교해요.'},
                                   {'t': '함수로 나누기',
                                    'en': 'function',
                                    'fig': 'f_param',
                                    'd': '한 가지 일을 하는 코드를 이름 붙여 묶고, 값만 바꿔 다시 쓰게 하는 것이에요.',
                                    'easy': '숫자만 바꿔 넣으면 되는 계산기를 만드는 거예요.',
                                    'use': '프롬프트에 ‘기준값을 인자로 받는 함수로 만들어 줘’라고 적어요.'}],
                         'steps': ['쪼개기: 큰 일을 작은 단계로 쪼개기', '함수로 만들기: 기준을 바꿔 돌릴 수 있는 함수 요청하기', '모델 선정: 결과로 조율하고 내 말로 요청하기'],
                         'glossary': [],
                         'outputs': ['M4 요구 성능 판정과 모델 선정 기록']},
               'start_messages': [['team', 'gyo', '{call}, 어떤 모델을 학습앱에 넣을지 정해야 해요. 저는 위협 형상을 놓치면 안 된다고 봐요.'],
                                  ['team', 'jeon', '저는 휴대 기기에서 빨리 돌아가야 한다고 봐요. 측정값은 m4_models.csv에 있어요.']],
               'finish_messages': [['team', 'gyo', '{call}, 같은 표를 보니 이야기가 빨라졌네요. 기준은 이대로 갑시다.']],
               'steps': [{'short': '쪼개기',
                          'title': '큰 일을 작은 단계로 쪼개기',
                          'lead': '한 번에 전부 요청하면 어디가 틀렸는지 찾기 어려워요. **작게 만들고 확인한 뒤 넓혀요.**',
                          'blocks': [{'id': 's1_info',
                                      'type': 'info',
                                      'title': '이번 요구 성능',
                                      'icon': '📐',
                                      'text': ['고교관, 전산호, 배보안의 요구를 한 표로 모았어요. 먼저 이 기준을 정한 뒤 모델을 봐요.'],
                                      'table': {'head': ['담당', '요구', '숫자'],
                                                'rows': [['고교관', '위협 형상을 놓치지 않기', '재현율 90% 이상'],
                                                         ['전산호', '휴대 기기에서 빠르게', '한 장 100ms 이하, 30MB 이하'],
                                                         ['배보안', '외부로 보내지 않기', '외부 전송 없음, 라이선스 확인']]},
                                      'figure': 'f_req'},
                                     {'id': 's1_order',
                                      'type': 'order',
                                      'title': '요청 순서 정하기',
                                      'text': '후보 모델을 판정하는 도구를 만들려고 해요. 어떤 순서로 요청하면 좋을까요?',
                                      'items': [{'id': 'a', 'text': '데이터 모양을 알려 주고 파일을 읽게 하기', 'hint': 'AI가 파일을 먼저 알아야 해요.'},
                                                {'id': 'b', 'text': '모델 하나를 판정하는 함수 만들기', 'hint': '작은 단위부터 시작해요.'},
                                                {'id': 'c', 'text': '답을 아는 모델 하나로 함수 확인하기', 'hint': '넓히기 전에 맞는지 봐요.'},
                                                {'id': 'd', 'text': '전체 후보에 적용해 비교표 만들기', 'hint': '확인한 함수를 넓혀요.'}],
                                      'answer': ['a', 'b', 'c', 'd'],
                                      'explain': '작은 단위로 만들고 확인한 뒤 넓히면, 틀렸을 때 어디가 문제인지 바로 알 수 있어요.'}]},
                         {'short': '함수로 만들기',
                          'title': '기준을 바꿔 돌릴 수 있는 함수 요청하기',
                          'lead': '기준값을 코드 속에 박아 두면 회의 때마다 고쳐야 해요. **인자로 받는 함수**로 요청해요.',
                          'blocks': [{'id': 's2_vibe',
                                      'type': 'vibe',
                                      'title': '요구 성능 판정 함수',
                                      'ask': ['jeon', '재현율, 속도, 크기, 외부 전송을 한 번에 판정하고, 왜 떨어졌는지도 알고 싶어요.'],
                                      'file': 'm4_models.csv',
                                      'slots': [{'id': 'small',
                                                 'label': '작게 시작',
                                                 'why': '함수 하나부터 만들어요.',
                                                 'options': [{'text': 'm4_models.csv의 열은 model, threat_recall_pct, ms_per_image, size_mb, '
                                                                      'sends_outside(1이면 외부 전송), license_ok(1이면 확인됨)야. 모델 한 행을 받아 기준을 넘는지와 떨어진 이유를 '
                                                                      '돌려주는 meets 함수를 먼저 만들고, mobilenet_v3로 한 번 확인해 줘.',
                                                              'good': True},
                                                             {'text': '모델 비교 프로그램을 한 번에 다 만들어 줘.', 'good': False},
                                                             {'text': '', 'good': False}],
                                                 'miss': {'say': '한 번에 다 만들었어요. 결과는 이래요.',
                                                          'code': 'import pandas as pd\n'
                                                                  "df = pd.read_csv('{VDIR}/m4_models.csv')\n"
                                                                  "best = df.sort_values('threat_recall_pct', ascending=False).iloc[0]\n"
                                                                  "print('추천 모델:', best['model'], best['threat_recall_pct'])",
                                                          'problem': '재현율만 보고 cloud_api_model을 추천했어요. 외부 전송과 라이선스 조건은 빠졌는데, 한 번에 받아서 어디가 빠졌는지 알아채기 '
                                                                     '어려워요.',
                                                          'hint': '모델 하나를 판정하는 함수부터 요청해 주세요.'}},
                                                {'id': 'param',
                                                 'label': '조건 바꾸기',
                                                 'why': '기준값을 인자로 받아요.',
                                                 'options': [{'text': '기준값은 min_recall, max_ms, max_mb 인자로 받게 해 줘. 기본값은 90, 100, 30이고, 외부 전송이 있거나 '
                                                                      '라이선스가 확인되지 않으면 항상 탈락이야.',
                                                              'good': True},
                                                             {'text': '기준은 코드에 적당히 넣어 줘.', 'good': False},
                                                             {'text': '', 'good': False}],
                                                 'miss': {'say': '기준을 코드 안에 넣었어요.',
                                                          'code': 'import pandas as pd\n'
                                                                  "df = pd.read_csv('{VDIR}/m4_models.csv')\n"
                                                                  "ok = df[(df['threat_recall_pct'] >= 90) & (df['ms_per_image'] <= 100) & "
                                                                  "(df['size_mb'] <= 30)]\n"
                                                                  "display(ok[['model', 'threat_recall_pct']])",
                                                          'problem': '기준이 코드에 박혀 있어 회의 때마다 고쳐야 하고, 외부 전송 모델도 그대로 통과했어요. 왜 떨어졌는지도 알 수 없어요.',
                                                          'hint': '기준을 인자로 받게 하고, 항상 탈락인 조건도 적어 주세요.'}},
                                                {'id': 'out',
                                                 'label': '비교표',
                                                 'why': '함수를 전체에 적용해요.',
                                                 'options': [{'text': '전체 모델에 기준 A(기본값)와 기준 B(재현율 85, 50ms, 10MB)를 적용한 비교표를 보여 주고, 속도와 재현율을 점으로 찍어 '
                                                                      '그려 줘. 표를 넣은 HTML 비교 문서를 report 변수에 담아 줘.',
                                                              'good': True},
                                                             {'text': '결과를 정리해 줘.', 'good': False},
                                                             {'text': '', 'good': False}],
                                                 'miss': {'say': '통과한 모델 이름을 정리했어요.',
                                                          'code': 'import pandas as pd\n'
                                                                  "df = pd.read_csv('{VDIR}/m4_models.csv')\n"
                                                                  "print(', '.join(df['model'].tolist()), '모델을 검토했습니다.')",
                                                          'problem': '어떤 기준으로 무엇이 통과했는지 알 수 없어요. 회의에서 쓸 근거가 안 돼요.',
                                                          'hint': '어떤 기준을 어떤 모양으로 비교할지 적어 주세요.'}}],
                                      'done': {'say': 'meets 함수를 mobilenet_v3로 확인한 뒤, 기준 두 가지를 전체에 적용했어요.',
                                               'code': 'import pandas as pd\n'
                                                       'import matplotlib.pyplot as plt\n'
                                                       '\n'
                                                       "df = pd.read_csv('{VDIR}/m4_models.csv')\n"
                                                       '\n'
                                                       'def meets(r, min_recall=90, max_ms=100, max_mb=30):\n'
                                                       '    why = []\n'
                                                       "    if r['threat_recall_pct'] < min_recall:\n"
                                                       "        why.append('재현율')\n"
                                                       "    if r['ms_per_image'] > max_ms:\n"
                                                       "        why.append('속도')\n"
                                                       "    if r['size_mb'] > max_mb:\n"
                                                       "        why.append('크기')\n"
                                                       "    if r['sends_outside'] == 1:\n"
                                                       "        why.append('외부 전송')\n"
                                                       "    if r['license_ok'] != 1:\n"
                                                       "        why.append('라이선스')\n"
                                                       "    return '통과' if not why else '탈락: ' + ', '.join(why)\n"
                                                       '\n'
                                                       '# 작은 확인: mobilenet_v3는 기본 기준을 통과해야 해요\n'
                                                       "print('확인:', meets(df[df['model'] == 'mobilenet_v3'].iloc[0]))\n"
                                                       '\n'
                                                       "table = df[['model', 'threat_recall_pct', 'ms_per_image', 'size_mb']].copy()\n"
                                                       "table['기준 A'] = df.apply(meets, axis=1)\n"
                                                       "table['기준 B'] = df.apply(lambda r: meets(r, 85, 50, 10), axis=1)\n"
                                                       'display(table)\n'
                                                       '\n'
                                                       'fig, ax = plt.subplots(figsize=(6, 3))\n'
                                                       "ax.scatter(df['ms_per_image'], df['threat_recall_pct'])\n"
                                                       'for _, r in df.iterrows():\n'
                                                       "    ax.annotate(r['model'], (r['ms_per_image'], r['threat_recall_pct']), fontsize=7)\n"
                                                       "ax.set_xlabel('ms per image')\n"
                                                       "ax.set_ylabel('threat recall (%)')\n"
                                                       '\n'
                                                       "report = ('<h2>모델 후보 비교</h2><p>기준 A: 재현율 90, 100ms, 30MB. 기준 B: 재현율 85, 50ms, 10MB. 외부 전송과 "
                                                       "라이선스 미확인은 항상 탈락이에요. 측정값은 가상 자료예요.</p>' + table.to_html(index=False))",
                                               'lesson': '재현율이 가장 높은 cloud_api_model은 외부 전송과 라이선스로 탈락했어요. 함수로 나누니 기준을 바꿔도 한 줄이면 다시 비교할 수 있고, 왜 '
                                                         '떨어졌는지도 바로 보여요.'},
                                      'on_ok': [['team', 'boan', '{call}, 탈락 이유까지 나오니 좋네요. 재현율 1등이 외부 전송 모델이었군요.']]}]},
                         {'short': '모델 선정',
                          'title': '결과로 조율하고 내 말로 요청하기',
                          'lead': 'AI의 표는 근거일 뿐이에요. 어떤 기준으로 갈지는 **사람이 합의해요.**',
                          'blocks': [{'id': 's3_decide',
                                      'type': 'decide',
                                      'title': '모델 선정 회의',
                                      'text': '비교표를 본 팀원들의 의견이에요. 받아들일지 정하고 이유를 적어 주세요.',
                                      'reason_min': 8,
                                      'items': [{'id': 'a',
                                                 'who': 'model',
                                                 'text': '재현율이 제일 높은 cloud_api_model로 가요. 외부 전송은 나중에 조건을 풀어 달라고 하면 되잖아요.',
                                                 'answer': 'reject',
                                                 'hint': '기준을 언제 정했는지 떠올려 보세요.',
                                                 'explain': '결과를 보고 기준을 바꾸면 합격선의 의미가 없어져요. 외부 전송 금지는 먼저 합의한 조건이에요.',
                                                 'reason': '먼저 합의한 기준을 결과에 맞춰 바꾸면 안 되고 외부 전송 조건은 지켜야 해요.'},
                                                {'id': 'b',
                                                 'who': 'gyo',
                                                 'text': '기준 A를 통과한 모델로 가되, 선정 기록에 탈락한 모델과 이유도 함께 남겨요.',
                                                 'answer': 'accept',
                                                 'hint': '나중에 다시 볼 사람에게 무엇이 필요할까요?',
                                                 'explain': '왜 그 모델을 골랐는지 남겨 두면 다음 검토가 빨라요. 표가 그 근거가 돼요.',
                                                 'reason': '선정과 탈락 이유를 남겨야 다음 검토 때 근거로 쓸 수 있어요.'}]},
                                     {'id': 's3_write',
                                      'type': 'write',
                                      'title': '쪼개서 요청하는 프롬프트 쓰기',
                                      'prompt': '**새 요청**: 새 후보 모델이 들어온 m4_models_new.csv에 같은 판정을 하고 싶어요. 기준에 ‘배터리 사용량 5% 이하’(battery_pct 열)를 더하려고 '
                                                '해요. 어떤 순서로, 무엇을 고쳐 달라고 할까요?',
                                      'example': 'm4_models_new.csv에 battery_pct(배터리 사용량, %) 열이 더 있어. 먼저 meets 함수에 max_battery 인자(기본값 5)만 추가하고, 다른 '
                                                 '기준은 그대로 둬. mobilenet_v3 한 행으로 확인한 뒤 전체에 적용한 비교표를 보여 줘. 탈락 이유도 함께 적어 줘.',
                                      'min': 30,
                                      'checklist': ['작은 단계부터 요청했어요', '기준을 인자로 받게 했어요', '한 행으로 확인하게 했어요', '전체에 적용한 결과의 모양을 적었어요'],
                                      'model_label': '📝 모범 프롬프트',
                                      'placeholder': '꼭 입력해 주세요 · 내 말로 프롬프트를 써 보세요'}]}],
               'outputs': [{'id': 'm4_selection',
                            'title': 'M4 요구 성능 판정과 모델 선정 기록',
                            'sections': [{'h': '1. 판정 함수와 비교표', 'blocks': ['s2_vibe']},
                                         {'h': '2. 모델 선정 회의 결정', 'blocks': ['s3_decide']},
                                         {'h': '3. 내가 쓴 프롬프트', 'blocks': ['s3_write']}]}]},
              {'id': 'm5',
               'no': 5,
               'title': '분류 점수 캘리브레이션과 판별 불가 기준',
               'short': '근거 받고 검증하기',
               'kind': '개인',
               'competencies': ['D5'],
               'outline': 'AI가 ‘점수 0.9면 90%가 맞아요’라고 답하면 믿어도 될까요? 근거 수치를 함께 달라고 요청하고, 검증 데이터로 정한 기준을 테스트 데이터로 한 번만 확인해요.',
               'brief': {'subtitle': '점수 믿어도 될까 확인하기: 근거를 함께 받기',
                         'time': '약 30분',
                         'goal': 'AI의 답을 그대로 믿지 않고 근거 수치와 표본 수를 함께 요청해, 점수와 실제 정답률을 비교하고 판별 불가 기준을 확인해요.',
                         'why': '점수 0.9는 열에 아홉은 맞는다는 뜻일까요? AI는 틀린 답도 자신 있게 말해요. 답만 받지 말고 다시 셀 수 있는 숫자를 함께 받아야 믿어도 되는지 판단할 수 있어요.',
                         'case': {'title': '해외 사례: 처음에는 구별할 수 없는 샤헤드와 기만용 드론',
                                  'source': 'NV, Ukrinform 보도 (2025.5)',
                                  'points': ['2025년 5월 우크라이나 공군 대변인 이나트는 샤헤드와 기만용 드론을 지상에서 소리로 구별할 수 있게 됐다고 말했어요. 눈으로 보거나 비행 정보를 봐도 알 수 있지만, 그러려면 '
                                             '시간과 경험이 필요하다고 했어요.',
                                             '다른 인터뷰에서는 처음부터 바로 구별하기는 어렵고, 가까워진 뒤에야 소리나 눈으로 구별할 수 있다고 설명했어요.',
                                             '모양만으로 가를 수 없는 경우가 있어요. 그럴 때는 답을 지어내지 않고 ‘판별 불가’라고 표시하는 기준이 필요해요.']},
                         'team': '한모델(AI 개발 담당)이 점수 파일을 넘겨 줘요. 고교관은 점수를 학습자에게 그대로 보여 줘도 되는지 궁금해해요.',
                         'terms': [{'t': '캘리브레이션',
                                    'en': 'calibration',
                                    'fig': 'f_rel',
                                    'd': '모델 점수가 실제 정답 비율과 맞는지 확인하고 맞추는 일이에요.',
                                    'easy': '‘자신감’과 ‘실력’이 같은지 보는 거예요.',
                                    'use': '점수 구간별로 실제 정답 비율을 비교해요. 그 구간 비율로 점수를 바꿔 쓰기도 해요.'},
                                   {'t': 'ECE',
                                    'en': 'Expected Calibration Error',
                                    'fig': 'f_ece',
                                    'd': '구간별 평균 점수와 실제 정답률의 차이를 표본 수로 가중 평균한 값이에요.',
                                    'easy': '0에 가까울수록 점수를 믿을 만해요.',
                                    'use': '숫자 하나로 비교하되, 구간별 표도 함께 봐요.'},
                                   {'t': '판별 불가',
                                    'en': 'reject option',
                                    'fig': 'f_grade',
                                    'd': '확신이 낮을 때 답을 내지 않고 사람에게 넘기는 선택지예요.',
                                    'easy': '모르면 모른다고 말하는 규칙이에요.',
                                    'use': '1위와 2위 점수 차이가 작으면 판별 불가로 둬요.'},
                                   {'t': '검증 데이터와 테스트 데이터',
                                    'en': 'validation, test',
                                    'fig': 'f_leak',
                                    'd': '검증 데이터로 기준을 고르고, 테스트 데이터로 마지막에 한 번만 확인해요.',
                                    'easy': '연습 문제로 공부하고, 본시험은 한 번만 보는 거예요.',
                                    'use': '고를 때 쓴 데이터로 성능을 재면 실제보다 좋게 나와요.'}],
                         'steps': ['답만 받으면: 근거 없는 답은 위험해요', '근거까지 받기: 근거 수치와 함께 요청하기', '모양과 정체: 모양과 정체를 나누고 내 말로 요청하기'],
                         'glossary': [],
                         'outputs': ['M5 점수 근거표와 행동 등급 기준']},
               'start_messages': [['dm_model', 'model', '{call}, 모델 점수가 대부분 0.8이 넘어요. 잘 되는 것 같은데, 학습자에게 그대로 보여 줘도 될까요? 파일은 m5_scores.csv예요.']],
               'finish_messages': [['dm_model', 'model', '{call}, 점수를 그대로 보여 주면 안 되겠네요. 근거 표 덕분에 바로 알았어요.']],
               'steps': [{'short': '답만 받으면',
                          'title': '근거 없는 답은 위험해요',
                          'lead': 'AI가 자신 있게 답해도 **다시 셀 수 있는 숫자**가 없으면 믿을지 판단할 수 없어요.',
                          'blocks': [{'id': 's1_info',
                                      'type': 'info',
                                      'title': 'AI 답변에 함께 받을 것',
                                      'icon': '🔍',
                                      'text': ['답을 받을 때 아래를 함께 요청하면, 코드를 다 읽지 못해도 믿을지 판단할 수 있어요.'],
                                      'bullets': ['**근거 수치**: 구간별 표본 수, 평균 점수, 실제 정답률',
                                                  '**작은 예시**: 손으로 계산해 볼 수 있는 한 구간',
                                                  '**나눈 방법**: 어떤 데이터로 고르고 어떤 데이터로 확인했는지'],
                                      'figure': 'f_verify'},
                                     {'id': 's1_q',
                                      'type': 'quiz',
                                      'title': '어떤 답을 믿을까요?',
                                      'text': '두 AI 답변 가운데 어느 쪽을 더 믿을 수 있을까요?',
                                      'options': ['‘이 모델은 점수가 잘 맞아요. 0.9면 90%가 맞다고 보면 돼요.’',
                                                  '‘0.9~1.0 구간 표본은 N건, 평균 점수는 A, 실제 정답률은 B예요. 표를 첨부해요.’',
                                                  '‘최신 기법을 썼으니 안심하고 쓰셔도 돼요.’'],
                                      'answer': 1,
                                      'explain': '다시 셀 수 있는 숫자가 있어야 해요. 다른 답변은 맞는지 틀린지 확인할 방법이 없어요.',
                                      'hints': ['어느 쪽이 직접 확인할 방법을 주나요?', '자신 있는 말투와 맞는 내용은 별개예요.']}]},
                         {'short': '근거까지 받기',
                          'title': '근거 수치와 함께 요청하기',
                          'lead': '**근거**와 **나눈 방법** 문장을 빼고도 보내 보세요. AI의 답이 어떻게 달라지는지 비교해 봐요.',
                          'blocks': [{'id': 's2_vibe',
                                      'type': 'vibe',
                                      'title': '점수와 실제 정답률 비교',
                                      'ask': ['model', '점수가 믿을 만한지, 어디부터 교관이 확인해야 하는지 숫자로 보고 싶어요.'],
                                      'file': 'm5_scores.csv',
                                      'slots': [{'id': 'task',
                                                 'label': '할 일',
                                                 'why': '무엇을 비교할지 정해요.',
                                                 'options': [{'text': 'm5_scores.csv의 열은 image_id, split(val·test), top1_label, top1_score, '
                                                                      'top2_score, is_correct야. top1_score를 0.3~0.5, 0.5~0.6, 0.6~0.7, 0.7~0.8, '
                                                                      '0.8~0.9, 0.9~1.0 구간으로 나눠 구간별 평균 점수와 실제 정답률을 비교해 줘. ECE도 계산해 줘.',
                                                              'good': True},
                                                             {'text': '점수가 정확한지 봐 줘.', 'good': False},
                                                             {'text': '', 'good': False}],
                                                 'miss': {'say': '전체 평균을 비교했어요.',
                                                          'code': 'import pandas as pd\n'
                                                                  "df = pd.read_csv('{VDIR}/m5_scores.csv')\n"
                                                                  "print('평균 점수', round(df['top1_score'].mean(), 3), '실제 정답률', "
                                                                  "round(df['is_correct'].mean(), 3))",
                                                          'problem': '평균 두 개만으로는 어느 구간이 어긋나는지 알 수 없어요. 무엇을 비교할지 구체적으로 말해야 해요.',
                                                          'hint': '어떤 구간으로 나눠 무엇을 비교할지 적어 주세요.'}},
                                                {'id': 'evidence',
                                                 'label': '근거',
                                                 'why': '다시 셀 수 있는 숫자를 받아요.',
                                                 'options': [{'text': '결론만 쓰지 말고 구간별 표본 수, 평균 점수, 실제 정답률을 표로 보여 줘. 점수와 실제 정답률을 비교하는 그림도 그려 줘.',
                                                              'good': True},
                                                             {'text': '믿어도 되는지 결론만 말해 줘.', 'good': False},
                                                             {'text': '', 'good': False}],
                                                 'miss': {'say': '결론입니다. 이 모델은 점수가 잘 맞아요. 점수 0.9면 90% 맞는다고 보면 돼요!',
                                                          'code': "print('결론: 점수를 그대로 믿어도 됩니다. 0.9면 90%가 맞아요.')",
                                                          'problem': '자신 있게 답했지만 숫자가 하나도 없어요. **근거를 요청하지 않으면 AI는 그럴듯한 결론만 줘요.** 실제 정답률과 비교해 보면 다를 수 '
                                                                     '있어요.',
                                                          'hint': '표본 수와 구간별 수치를 표로 달라고 적어 주세요.'}},
                                                {'id': 'split',
                                                 'label': '나눈 방법',
                                                 'why': '고르는 데이터와 확인하는 데이터를 나눠요.',
                                                 'options': [{'text': '등급 기준은 top1_score 0.8 이상이고 1·2위 차이 0.3 이상이면 ‘참고 판정 표시’, 차이 0.1 이상이면 ‘교관 확인’, '
                                                                      '나머지는 ‘판별 불가’야. val에서 등급별 정답률을 보고, 정한 기준을 test에는 마지막에 한 번만 적용해 줘. 표를 넣은 HTML '
                                                                      '문서를 report 변수에 담아 줘.',
                                                              'good': True},
                                                             {'text': '전체 데이터로 제일 좋아 보이는 기준을 찾아 줘.', 'good': False},
                                                             {'text': '', 'good': False}],
                                                 'miss': {'say': '전체 데이터에서 정답률이 가장 높게 나오는 점수 기준을 찾았어요.',
                                                          'code': 'import pandas as pd\n'
                                                                  "df = pd.read_csv('{VDIR}/m5_scores.csv')\n"
                                                                  "best = max([x / 100 for x in range(50, 96, 5)], key=lambda t: df[df['top1_score'] "
                                                                  ">= t]['is_correct'].mean())\n"
                                                                  "print('최적 기준:', best, '정답률:', round(df[df['top1_score'] >= "
                                                                  "best]['is_correct'].mean(), 3))",
                                                          'problem': 'val과 test를 섞어 기준을 골랐어요. 고를 때 쓴 데이터로 성능을 재면 실제보다 좋게 나와요. 이 숫자로는 처음 보는 데이터에서의 '
                                                                     '성능을 말할 수 없어요.',
                                                          'hint': '어디서 고르고 어디서 한 번만 확인할지 적어 주세요.'}}],
                                      'done': {'say': '구간별 근거 표와 그림을 만들고, val에서 본 등급 기준을 test에 한 번 적용했어요.',
                                               'code': 'import pandas as pd\n'
                                                       'import matplotlib.pyplot as plt\n'
                                                       '\n'
                                                       "df = pd.read_csv('{VDIR}/m5_scores.csv')\n"
                                                       "val = df[df['split'] == 'val'].copy()\n"
                                                       "test = df[df['split'] == 'test'].copy()\n"
                                                       'BINS = [0.3, 0.5, 0.6, 0.7, 0.8, 0.9, 1.0]\n'
                                                       '\n'
                                                       'def reliability(d):\n'
                                                       "    g = d.groupby(pd.cut(d['top1_score'], BINS), observed=True)\n"
                                                       "    t = g.agg(표본=('is_correct', 'size'), 평균점수=('top1_score', 'mean'), 실제정답률=('is_correct', "
                                                       "'mean')).reset_index(drop=True)\n"
                                                       "    t.insert(0, '구간', [str(i) for i in g.groups.keys()])\n"
                                                       '    return t.round(3)\n'
                                                       '\n'
                                                       'rel = reliability(val)\n'
                                                       "ece = round(((rel['평균점수'] - rel['실제정답률']).abs() * rel['표본'] / rel['표본'].sum()).sum(), 3)\n"
                                                       'display(rel)\n'
                                                       "print('val ECE:', ece)\n"
                                                       '\n'
                                                       'def grade(r):\n'
                                                       "    margin = r['top1_score'] - r['top2_score']\n"
                                                       "    if r['top1_score'] >= 0.8 and margin >= 0.3:\n"
                                                       "        return '참고 판정 표시'\n"
                                                       '    if margin >= 0.1:\n'
                                                       "        return '교관 확인'\n"
                                                       "    return '판별 불가'\n"
                                                       '\n'
                                                       "val['등급'] = val.apply(grade, axis=1)\n"
                                                       "test['등급'] = test.apply(grade, axis=1)\n"
                                                       'def summary(d):\n'
                                                       "    return d.groupby('등급').agg(표본=('is_correct', 'size'), 실제정답률=('is_correct', "
                                                       "'mean')).round(3).reset_index()\n"
                                                       'g_val = summary(val)\n'
                                                       'g_test = summary(test)\n'
                                                       'display(g_val)\n'
                                                       'display(g_test)\n'
                                                       '\n'
                                                       'fig, ax = plt.subplots(figsize=(4, 3.2))\n'
                                                       "ax.plot([0.3, 1], [0.3, 1], '--', color='gray')\n"
                                                       "ax.plot(rel['평균점수'], rel['실제정답률'], 'o-')\n"
                                                       "ax.set_xlabel('mean score')\n"
                                                       "ax.set_ylabel('actual accuracy')\n"
                                                       '\n'
                                                       "report = ('<h2>분류 점수 근거표</h2>' + f'<p>val 기준 ECE {ece}. 점수가 실제 정답률보다 높게 나와요. 점수를 그대로 보여 주지 "
                                                       "않아요.</p>'\n"
                                                       "          + rel.to_html(index=False) + '<h3>등급별 정답률 (val)</h3>' + "
                                                       'g_val.to_html(index=False)\n'
                                                       "          + '<h3>등급별 정답률 (test, 한 번 확인)</h3>' + g_test.to_html(index=False))",
                                               'lesson': '근거를 함께 받으니 점수가 실제 정답률보다 높다는 것이 보였어요. 처음의 ‘0.9면 90%’라는 답이 틀렸다는 것도요. 등급을 나누면 어디부터 교관이 확인해야 '
                                                         '할지도 숫자로 보여요.'},
                                      'on_ok': [['dm_model', 'model', '{call}, 점수가 생각보다 부풀어 있네요. 등급으로 보여 주는 게 낫겠어요.']]}]},
                         {'short': '모양과 정체',
                          'title': '모양과 정체를 나누고 내 말로 요청하기',
                          'lead': '모델이 알려 주는 것은 **모양**이에요. 정체는 다른 근거가 필요해요.',
                          'blocks': [{'id': 's3_cls',
                                      'type': 'classify',
                                      'title': '모델이 말해도 될까요?',
                                      'text': '실루엣 분류 모델의 화면에 나와도 되는 말과 사람이 다른 근거로 판단해야 하는 말을 나눠 보세요.',
                                      'primary_label': '판단',
                                      'item_label': '문장',
                                      'contrib': False,
                                      'categories': [{'id': 'ok', 'name': '모델이 말해도 돼요'}, {'id': 'no', 'name': '사람이 다른 근거로 판단해요'}],
                                      'items': [{'id': 'a',
                                                 'title': '이 실루엣은 delta_wing과 비슷해요. (참고)',
                                                 'accept': ['ok'],
                                                 'explain': '보이는 모양을 말하는 것이고, 참고라고 표시했어요.',
                                                 'hints': ['모양에 대한 말인가요, 정체에 대한 말인가요?', '참고라고 밝혔나요?']},
                                                {'id': 'b',
                                                 'title': '이것은 기만용 드론이에요.',
                                                 'accept': ['no'],
                                                 'explain': '모양이 같아도 정체는 다를 수 있어요. 실루엣만으로 정할 수 없어요.',
                                                 'hints': ['해외 사례를 떠올려 보세요.', '실루엣만 보고 진짜와 기만용을 가를 수 있었나요?']},
                                                {'id': 'c',
                                                 'title': '1·2위 점수 차이가 작아 판별 불가예요.',
                                                 'accept': ['ok'],
                                                 'explain': '모르면 모른다고 말하는 것은 모델이 해야 할 말이에요.',
                                                 'hints': ['답을 지어냈나요?', '사람에게 넘기는 말인가요?']},
                                                {'id': 'd',
                                                 'title': '위협이 확실하니 바로 대응하세요.',
                                                 'accept': ['no'],
                                                 'explain': '대응은 정해진 절차로 사람이 결정해요. 학습앱이 할 말이 아니에요.',
                                                 'hints': ['학습앱이 결정할 일인가요?', '누가 책임지는 일인가요?']}]},
                                     {'id': 's3_write',
                                      'type': 'write',
                                      'title': '근거를 요청하는 프롬프트 쓰기',
                                      'prompt': '**새 요청**: 다른 모델의 점수 파일 m5_scores_b.csv도 믿어도 되는지 보고 싶어요. 열은 같아요. 결론만 받지 않도록 어떻게 요청할까요?',
                                      'example': 'm5_scores_b.csv의 점수가 믿을 만한지 확인하고 싶어. 열은 split, top1_score, top2_score, is_correct야. val에서 '
                                                 'top1_score 구간별 표본 수, 평균 점수, 실제 정답률을 표로 보여 주고 ECE도 계산해 줘. 결론만 쓰지 말고 표를 함께 줘. 정한 기준은 test에 마지막에 한 번만 '
                                                 '적용해 줘.',
                                      'min': 30,
                                      'checklist': ['근거 수치를 요청했어요', '표본 수를 요청했어요', '고르는 데이터와 확인하는 데이터를 나눴어요', '결론만 받지 않게 했어요'],
                                      'model_label': '📝 모범 프롬프트',
                                      'placeholder': '꼭 입력해 주세요 · 내 말로 프롬프트를 써 보세요'}]}],
               'outputs': [{'id': 'm5_calibration',
                            'title': 'M5 점수 근거표와 행동 등급 기준',
                            'sections': [{'h': '1. 근거 표와 등급', 'blocks': ['s2_vibe']},
                                         {'h': '2. 모양과 정체 구분', 'blocks': ['s3_cls']},
                                         {'h': '3. 내가 쓴 프롬프트', 'blocks': ['s3_write']}]}]},
              {'id': 'm6',
               'no': 6,
               'title': '실루엣 식별 학습 앱 화면 설계와 사용성 점검',
               'short': '숫자와 예외',
               'kind': '협업',
               'competencies': ['D1', 'C1'],
               'outline': '‘자주 틀리는 문항을 많이 내 줘’ 같은 모호한 말을 숫자 규칙과 예외 처리로 바꾸고, 출제 계획표와 학습 앱 화면 시안을 받아요.',
               'brief': {'subtitle': '출제 규칙과 앱 화면 시안: 숫자와 예외로 말하기',
                         'time': '약 30분',
                         'goal': '모호한 요청을 숫자 규칙과 예외 처리로 바꿔 출제 계획표와 화면 시안을 받고, 사용성 의견을 판단해요.',
                         'why': 'AI 참고 판정은 언제 보여 줘야 할까요? ‘적당히’, ‘자주’ 같은 말은 사람마다 달라요. AI도 마음대로 정해 버려요. 빈칸처럼 예외가 있는 데이터는 어떻게 할지까지 알려 줘야 해요.',
                         'case': {'title': '해외 사례: 한 번 누르면 보고되는 ePPO 앱',
                                  'source': 'Mezha, Kyiv Post, Media Center Ukraine 보도 (2022.10~2023.3)',
                                  'points': ['2022년 10월 우크라이나에서 시민이 순항미사일과 자폭 드론, 적 항공기를 휴대폰으로 방공 부서에 알리는 ePPO 앱이 나왔어요.',
                                             '2023년 3월 개발 그룹의 공동 창업자는 33만 명이 내려받았다고 밝혔어요. 개발자는 이 앱이 놓칠 뻔한 샤헤드와 칼리브르를 몇 차례 찾는 데 도움이 됐다고 말했지만, 정확한 수치는 '
                                             '내놓지 않았어요.',
                                             '짧은 순간에 쓰는 화면일수록 단순해야 해요. 무엇을 언제 보여 줄지 숫자와 순서로 정해 두어야 해요.']},
                         'team': '고교관은 자주 틀리는 문항을 더 내고 싶어 하고, 박대표(사용자 대표)는 화면을 직접 써 본 의견을 줘요.',
                         'terms': [{'t': '혼동 쌍',
                                    'en': 'confusion pair',
                                    'fig': 'f_pair',
                                    'd': '학습자가 자주 서로 헷갈리는 두 형상 계열이에요.',
                                    'easy': '쌍둥이처럼 자주 헷갈리는 둘을 함께 묶은 거예요.',
                                    'use': '오답률이 높은 쌍을 먼저 출제해요.'},
                                   {'t': '적응형 출제',
                                    'en': 'adaptive testing',
                                    'fig': 'f_adapt',
                                    'd': '학습 기록에 맞춰 낼 문항과 횟수를 바꾸는 방식이에요.',
                                    'easy': '약한 부분을 더 많이 연습시키는 거예요.',
                                    'use': '오답률 기준을 숫자로 적어 AI에게 요청해요.'},
                                   {'t': '예외 처리',
                                    'en': 'edge case',
                                    'fig': 'f_edge',
                                    'd': '빈칸, 0, 표본이 적은 경우처럼 평소와 다른 값을 어떻게 할지 정하는 것이에요.',
                                    'easy': '‘이럴 땐 이렇게’를 미리 말해 두는 거예요.',
                                    'use': '말하지 않으면 AI가 조용히 아무렇게나 처리해요.'}],
                         'steps': ['숫자로 바꾸기: 모호한 말을 숫자 규칙으로', '규칙으로 요청: 숫자 규칙과 예외를 담아 요청하기', '사용성 점검: 사용성 의견을 판단하고 내 말로 요청하기'],
                         'glossary': [],
                         'outputs': ['M6 출제 계획표와 학습 앱 화면 시안']},
               'start_messages': [['dm_gyo', 'gyo', '{call}, 자주 틀리는 문항을 많이 내고 싶어요. 그런데 ‘자주’를 어떻게 정할지가 문제예요. 파일은 m6_items.csv예요.']],
               'finish_messages': [['dm_park', 'park', '{call}, 먼저 풀어 보고 AI 판정을 보니 훨씬 공부가 돼요. 이대로 가면 좋겠어요.']],
               'steps': [{'short': '숫자로 바꾸기',
                          'title': '모호한 말을 숫자 규칙으로',
                          'lead': '‘자주’, ‘적당히’는 사람마다 달라요. **숫자로 바꾸면** AI도 사람도 같은 뜻으로 읽어요.',
                          'blocks': [{'id': 's1_match',
                                      'type': 'match',
                                      'title': '모호한 말을 숫자 규칙과 짝짓기',
                                      'text': '왼쪽 말을 프롬프트에 쓸 수 있게 바꾼 문장을 골라 주세요.',
                                      'left_label': '모호한 말',
                                      'right_label': '프롬프트에 쓸 문장',
                                      'pairs': [{'left': '자주 틀리는 문항', 'right': '오답률 40% 이상', 'hint': '가장 높은 기준이에요.'},
                                                {'left': '가끔 틀리는 문항', 'right': '오답률 20% 이상 40% 미만', 'hint': '중간 구간이에요.'},
                                                {'left': '아직 모르는 문항', 'right': 'shown이 0', 'hint': '한 번도 내지 않았어요.'},
                                                {'left': '믿기 어려운 기록', 'right': 'shown이 5 미만', 'hint': '표본이 적어요.'}],
                                      'hints': ['숫자가 적힌 쪽이 어느 말을 가리키는지 찾아보세요.', 'shown은 낸 횟수, wrong은 틀린 횟수예요.'],
                                      'explain': '숫자로 바꾸면 결과가 왜 그렇게 나왔는지도 설명할 수 있어요.'}]},
                         {'short': '규칙으로 요청',
                          'title': '숫자 규칙과 예외를 담아 요청하기',
                          'lead': '**예외** 문장을 빼고도 보내 보세요. 빈칸이 어떻게 처리되는지 확인해 봐요.',
                          'blocks': [{'id': 's2_vibe',
                                      'type': 'vibe',
                                      'title': '출제 계획표와 화면 시안',
                                      'ask': ['gyo', '문항별로 몇 배로 낼지 정하고, 학습 흐름이 보이는 화면 시안까지 받고 싶어요.'],
                                      'file': 'm6_items.csv',
                                      'slots': [{'id': 'rule',
                                                 'label': '숫자 규칙',
                                                 'why': '기준을 숫자로 적어요.',
                                                 'options': [{'text': 'm6_items.csv의 열은 item_id, family, confusable_with, shown, wrong이야. 오답률 = '
                                                                      'wrong ÷ shown × 100으로 계산하고, 40% 이상이면 ‘우선 출제’(3배), 20% 이상이면 ‘자주 출제’(2배), 그 아래는 '
                                                                      '‘기본’(1배)으로 표시해 줘.',
                                                              'good': True},
                                                             {'text': '자주 틀리는 문항을 많이 내게 해 줘.', 'good': False},
                                                             {'text': '', 'good': False}],
                                                 'miss': {'say': '많이 틀린 순서로 정렬했어요.',
                                                          'code': 'import pandas as pd\n'
                                                                  "df = pd.read_csv('{VDIR}/m6_items.csv')\n"
                                                                  "display(df.sort_values('wrong', ascending=False))",
                                                          'problem': '많이 낸 문항이 위로 올라왔을 뿐, 몇 배로 낼지 정해지지 않았어요. ‘많이’가 숫자가 아니어서 AI가 정렬로 때웠어요.',
                                                          'hint': '오답률 계산식과 기준값, 배수를 적어 주세요.'}},
                                                {'id': 'edge',
                                                 'label': '예외',
                                                 'why': '빈칸과 적은 표본을 어떻게 할지 정해요.',
                                                 'options': [{'text': 'shown이 0이면 오답률을 비우고 ‘새 문항’(1배)으로, shown이 5 미만이면 오답률은 보여 주되 ‘표본 적음’(1배)으로 표시해 '
                                                                      '줘.',
                                                              'good': True},
                                                             {'text': '나머지는 알아서 해 줘.', 'good': False},
                                                             {'text': '', 'good': False}],
                                                 'miss': {'say': '기준대로 표시했어요. 예외는 따로 처리하지 않았어요.',
                                                          'code': 'import pandas as pd\n'
                                                                  "df = pd.read_csv('{VDIR}/m6_items.csv')\n"
                                                                  "df['오답률(%)'] = (df['wrong'] / df['shown'] * 100).round(1)\n"
                                                                  "df['계획'] = df['오답률(%)'].apply(lambda v: '우선 출제' if v >= 40 else ('자주 출제' if v >= "
                                                                  "20 else '기본'))\n"
                                                                  "display(df[['item_id', 'family', 'shown', '오답률(%)', '계획']])",
                                                          'problem': '오류는 없지만 Q07은 세 번 낸 기록만으로 ‘우선 출제’가 됐고, 한 번도 안 낸 Q08·Q10은 ‘기본’으로 섞였어요. **말하지 않은 '
                                                                     '예외는 AI가 조용히 처리해요.**',
                                                          'hint': 'shown이 0이거나 적을 때 어떻게 할지 적어 주세요.'}},
                                                {'id': 'screen',
                                                 'label': '화면',
                                                 'why': '순서와 조건을 정해요.',
                                                 'options': [{'text': '학습 흐름은 문항 → 응답 → 즉시 피드백 → AI 참고 판정 → 해설 순서야. AI 참고 판정은 응답 전에는 숨기고, 등급과 ‘교관 확인 '
                                                                      '전’ 표시를 함께 보여 주는 HTML 화면 시안을 report 변수에 담아 줘. 아래에 출제 계획표도 넣어 줘.',
                                                              'good': True},
                                                             {'text': '화면을 멋지게 만들어 줘.', 'good': False},
                                                             {'text': '', 'good': False}],
                                                 'miss': {'say': '멋진 화면을 만들었어요!',
                                                          'code': 'report = (\'<h2>🚀 AI 실루엣 판독기</h2><p style="color:#2563eb;font-size:20px"><b>AI '
                                                                  "판정: delta_wing 99%</b></p>'\n"
                                                                  "          '<p>이 실루엣의 형상 계열은 "
                                                                  "무엇일까요?</p><ol><li>fixed_wing</li><li>delta_wing</li></ol>')\n"
                                                                  "print('화면을 만들었어요.')",
                                                          'problem': '문항보다 AI 판정이 먼저 보여서 학습자가 스스로 생각할 기회가 사라졌어요. 근거 없는 99%도 들어갔고요. ‘멋지게’로는 순서를 알 수 '
                                                                     '없어요.',
                                                          'hint': '화면에 무엇이 어떤 순서로 나오는지 적어 주세요.'}}],
                                      'done': {'say': '규칙과 예외대로 출제 계획표를 만들고, 응답 후에 AI 판정이 열리는 화면 시안을 만들었어요.',
                                               'code': 'import pandas as pd\n'
                                                       '\n'
                                                       "df = pd.read_csv('{VDIR}/m6_items.csv')\n"
                                                       '\n'
                                                       'def plan(r):\n'
                                                       "    if r['shown'] == 0:\n"
                                                       "        return pd.Series(['', '새 문항', 1])\n"
                                                       "    rate = round(r['wrong'] / r['shown'] * 100, 1)\n"
                                                       "    if r['shown'] < 5:\n"
                                                       "        return pd.Series([rate, '표본 적음', 1])\n"
                                                       '    if rate >= 40:\n'
                                                       "        return pd.Series([rate, '우선 출제', 3])\n"
                                                       '    if rate >= 20:\n'
                                                       "        return pd.Series([rate, '자주 출제', 2])\n"
                                                       "    return pd.Series([rate, '기본', 1])\n"
                                                       '\n'
                                                       "df[['오답률(%)', '계획', '배수']] = df.apply(plan, axis=1)\n"
                                                       "table = df[['item_id', 'family', 'confusable_with', 'shown', '오답률(%)', '계획', '배수']]\n"
                                                       'display(table)\n'
                                                       '\n'
                                                       "report = ('<h2>실루엣 식별 학습앱 시안</h2>'\n"
                                                       '    \'<div style="border:1px solid '
                                                       '#cbd5e1;border-radius:12px;padding:14px;max-width:360px">\'\n'
                                                       '    \'<svg viewBox="-110 -110 220 220" width="120"><polygon points="0,-100 100,80 0,50 '
                                                       '-100,80" fill="#111"/></svg>\'\n'
                                                       "    '<p><b>문항.</b> 이 실루엣의 형상 계열은 무엇일까요?</p>'\n"
                                                       "    '<ol><li>fixed_wing</li><li>delta_wing</li><li>cruise_missile</li><li>bird</li></ol>'\n"
                                                       "    '<details><summary>응답 후 열기</summary>'\n"
                                                       "    '<p><b>즉시 피드백.</b> 정답은 delta_wing이에요.</p>'\n"
                                                       "    '<p><b>AI 참고 판정.</b> delta_wing, 등급: 교관 확인, 교관 확인 전</p>'\n"
                                                       "    '<p><b>해설.</b> 삼각형 날개와 뒤쪽 엔진을 보세요. 정체는 모양만으로 정하지 않아요.</p>'\n"
                                                       "    '</details></div><h3>출제 계획표</h3>' + table.to_html(index=False))",
                                               'lesson': '숫자와 예외를 적으니 Q07, Q08, Q10처럼 조심할 문항이 제대로 드러났어요. 순서를 적으니 AI 판정도 응답 뒤에 나와요. 20%에 딱 걸린 Q06도 '
                                                         '기준대로 ‘자주 출제’가 됐어요.'},
                                      'on_ok': [['dm_gyo', 'gyo', '{call}, 표본이 적은 문항까지 구분됐네요. 이제 박대표님 의견을 받아 볼까요?']]}]},
                         {'short': '사용성 점검',
                          'title': '사용성 의견을 판단하고 내 말로 요청하기',
                          'lead': 'AI로 만든 화면은 **쓰는 사람의 의견**으로 고쳐요. 다만 의견을 전부 따르지는 않아요.',
                          'blocks': [{'id': 's3_decide',
                                      'type': 'decide',
                                      'title': '사용성 의견 판단하기',
                                      'text': '시범 사용자와 교관의 의견이에요. 받아들일지 정하고 이유를 적어 주세요.',
                                      'reason_min': 8,
                                      'items': [{'id': 'a',
                                                 'who': 'park',
                                                 'text': '문항마다 AI 판정을 먼저 보여 주세요. 그러면 빨리 끝나요.',
                                                 'answer': 'reject',
                                                 'hint': '이 앱의 목적이 빨리 끝내는 일일까요?',
                                                 'explain': '먼저 답을 보면 스스로 익힐 기회가 사라져요. 응답 후에 보여 주는 순서를 지켜요.',
                                                 'reason': '먼저 판정을 보면 스스로 익힐 기회가 없어지니 응답 후에 보여 줘요.'},
                                                {'id': 'b',
                                                 'who': 'park',
                                                 'text': '틀린 뒤에 헷갈린 두 실루엣을 나란히 보여 주세요.',
                                                 'answer': 'accept',
                                                 'hint': '혼동 쌍을 익히는 데 도움이 될까요?',
                                                 'explain': '혼동 쌍을 나란히 보면 차이가 바로 보여요. 프롬프트에 화면 요소 하나만 더 적으면 돼요.',
                                                 'reason': '혼동 쌍을 나란히 보면 차이를 익히기 쉬워요.'}]},
                                     {'id': 's3_write',
                                      'type': 'write',
                                      'title': '숫자와 예외를 담은 프롬프트 쓰기',
                                      'prompt': '**새 요청**: ‘최근에 많이 틀린 사람에게 복습 문항을 좀 더 내 줘’라는 요청을 받았어요. 파일은 m6_learners.csv이고, 열은 learner_id, shown, '
                                                'wrong이에요. 숫자 규칙과 예외를 정해 프롬프트를 써 보세요.',
                                      'example': 'm6_learners.csv의 열은 learner_id, shown, wrong이야. 오답률 = wrong ÷ shown × 100으로 계산하고, 40% 이상이면 복습 5문항, '
                                                 '20% 이상이면 복습 3문항, 그 아래는 복습 1문항으로 정해 줘. shown이 0이면 ‘기록 없음’, 5 미만이면 ‘표본 적음’으로 표시하고 복습 1문항으로 해 줘. '
                                                 'learner_id, 오답률, 복습 수, 표시 순서의 표로 보여 줘.',
                                      'min': 30,
                                      'checklist': ['모호한 말을 숫자로 바꿨어요', '경계값을 분명히 적었어요', '빈칸이나 0, 적은 표본을 어떻게 할지 적었어요', '받고 싶은 모양을 적었어요'],
                                      'model_label': '📝 모범 프롬프트',
                                      'placeholder': '꼭 입력해 주세요 · 내 말로 프롬프트를 써 보세요'}]}],
               'outputs': [{'id': 'm6_app',
                            'title': 'M6 출제 계획표와 학습 앱 화면 시안',
                            'sections': [{'h': '1. 숫자 규칙 짝짓기', 'blocks': ['s1_match']},
                                         {'h': '2. 출제 계획표와 화면 시안', 'blocks': ['s2_vibe']},
                                         {'h': '3. 사용성 의견 판단', 'blocks': ['s3_decide']},
                                         {'h': '4. 내가 쓴 프롬프트', 'blocks': ['s3_write']}]}]},
              {'id': 'm7',
               'no': 7,
               'title': '교관 해설 검토(HITL) 절차와 역할·권한 정하기',
               'short': '문서 요청과 환각',
               'kind': '협업',
               'competencies': ['J1', 'D5'],
               'outline': 'AI 해설 초안을 근거표와 대조하고, 독자·양식·근거를 지정해 교관 검토 절차서 초안을 요청해요. AI가 없는 내용을 지어내지 않게 하는 법을 익혀요.',
               'brief': {'subtitle': '해설 대조표와 검토 절차서: 환각 없는 문서 요청하기',
                         'time': '약 30분',
                         'goal': '독자·양식·근거를 지정해 문서를 요청하고, 자료에 없는 내용을 ‘자료 없음’으로 쓰게 해 환각을 막아요.',
                         'why': 'AI가 쓴 해설, 누가 언제 확인하고 공개할까요? AI 문장은 틀려도 매끄러워요. 해설 하나가 틀리면 많은 학습자에게 같은 오류가 퍼져요. 자료 밖의 말을 쓰지 못하게 하고, 사람이 승인하는 절차를 문서로 '
                                '남겨요.',
                         'case': {'title': '해외 사례: 교육 인원을 크게 늘리려는 미 합동 대드론 교육기관',
                                  'source': 'KSWO, DVIDS 보도 (2025.4~5)',
                                  'points': ['2025년 5월 미 합동 소형무인기 대응 대학(JCU)의 책임자는 연간 교육 인원을 약 1,000명에서 앞으로 3,000~5,000명으로 늘릴 것으로 내다봤어요. 비슷한 시기의 다른 '
                                             '자료에는 연 최대 1,200명으로 적혀 있어요.',
                                             '교육 인원이 몇 배로 늘면 교재의 작은 오류도 그만큼 많은 사람에게 전해져요.',
                                             'AI가 해설 초안을 쓸수록 공개 전에 누가 확인하는지 정해 둬야 해요.']},
                         'team': '고교관이 해설을 승인하고, 나라벨이 근거표를 검수해요. 윤훈련(교육 담당)이 기한을 넘긴 건을 처리해요.',
                         'terms': [{'t': 'HITL (사람 참여형)',
                                    'en': 'Human-in-the-Loop',
                                    'fig': 'f_hitl',
                                    'd': 'AI가 만든 결과를 사람이 확인하고 결정하는 절차를 일에 넣는 방식이에요.',
                                    'easy': 'AI는 초안, 결정은 사람이에요.',
                                    'use': '승인 전에는 공개되지 않게 절차를 정해요.'},
                                   {'t': 'RACI 표',
                                    'en': 'Responsible·Accountable·Consulted·Informed',
                                    'fig': 'f_raci',
                                    'd': '일마다 누가 하고, 누가 책임지고, 누구와 의논하고, 누구에게 알리는지 정리한 표예요.',
                                    'easy': '‘이 일은 누구 몫?’을 한눈에 보는 표예요.',
                                    'use': 'AI에게 표 양식을 지정해 초안을 받아요.'},
                                   {'t': '환각',
                                    'en': 'hallucination',
                                    'fig': 'f_halluc',
                                    'd': 'AI가 자료에 없는 내용을 사실처럼 지어내는 현상이에요.',
                                    'easy': '모르는 것도 그럴듯하게 말해 버리는 거예요.',
                                    'use': '‘자료에 없으면 자료 없음으로 써 줘’라고 적어요.'}],
                         'steps': ['역할 정하기: AI와 사람의 역할 나누기', '문서 요청: 독자·근거·표현을 정해 문서 요청하기', '내 말로 쓰기: 문서를 받은 뒤 확인하고 내 말로 요청하기'],
                         'glossary': [],
                         'outputs': ['M7 해설 대조표와 교관 검토 절차서']},
               'start_messages': [['dm_gyo', 'gyo', '{call}, AI가 쓴 해설 초안이 쌓였어요. 근거표와 맞는지 대조하고, 검토 절차서 초안도 필요해요. 파일은 m7_drafts.csv와 m7_facts.csv예요.']],
               'finish_messages': [['dm_hun', 'hun', '{call}, 절차서 초안 잘 받았어요. 표시까지 분명해서 교관님 검토로 바로 넘길게요.']],
               'steps': [{'short': '역할 정하기',
                          'title': 'AI와 사람의 역할 나누기',
                          'lead': 'AI는 초안을 쓰고, **결정은 사람이** 해요. 문서를 요청하기 전에 역할부터 정해요.',
                          'blocks': [{'id': 's1_match',
                                      'type': 'match',
                                      'title': '누가 무엇을 맡나요?',
                                      'text': '담당과 맡을 일을 짝지어 주세요.',
                                      'left_label': '담당',
                                      'right_label': '맡을 일',
                                      'pairs': [{'left': 'AI', 'right': '해설 초안 작성', 'hint': '결정이 아니라 초안이에요.'},
                                                {'left': '고교관', 'right': '해설 승인·수정·반려', 'hint': '내용에 책임을 지는 사람이에요.'},
                                                {'left': '나라벨', 'right': '형상 근거표 검수', 'hint': '데이터 담당이에요.'},
                                                {'left': '전산호', 'right': '승인된 해설만 앱에 반영', 'hint': '앱을 다루는 담당이에요.'},
                                                {'left': '윤훈련', 'right': '기한 넘긴 건 처리', 'hint': '교육 일정을 관리해요.'}],
                                      'hints': ['AI가 결정까지 맡는 짝은 없어요.', '내용, 데이터, 앱, 일정 가운데 누가 무엇을 맡을지 떠올려 보세요.'],
                                      'explain': '역할을 먼저 정하면 프롬프트에 ‘승인 전에는 공개하지 않음’을 분명히 적을 수 있어요.'}]},
                         {'short': '문서 요청',
                          'title': '독자·근거·표현을 정해 문서 요청하기',
                          'lead': '**자료에 없으면** 문장을 빼고도 보내 보세요. AI가 무엇을 지어내는지 확인해 봐요.',
                          'blocks': [{'id': 's2_vibe',
                                      'type': 'vibe',
                                      'title': '해설 대조표와 검토 절차서',
                                      'ask': ['gyo', '초안의 주장이 근거표와 맞는지 대조하고, 교관이 읽을 검토 절차서 초안까지 받고 싶어요.'],
                                      'file': 'm7_drafts.csv',
                                      'slots': [{'id': 'reader',
                                                 'label': '독자와 양식',
                                                 'why': '누가 읽는지, 어떤 순서인지 정해요.',
                                                 'options': [{'text': 'm7_drafts.csv(draft_id, family, field, claim, wait_hours)의 주장을 '
                                                                      'm7_facts.csv(family, wing, engine, tail)와 대조한 표를 만들어 줘. 교관이 읽을 검토 절차서를 HTML로 '
                                                                      'report 변수에 담아 줘. 순서는 목적, 절차 표(단계·담당·기한), 대조 결과, 기한 초과 목록이야.',
                                                              'good': True},
                                                             {'text': '검토 절차에 대해 써 줘.', 'good': False},
                                                             {'text': '', 'good': False}],
                                                 'miss': {'say': '검토 절차의 중요성을 정리했어요.',
                                                          'code': "report = ('<h2>HITL의 중요성</h2><p>인공지능 시대에 사람의 검토는 매우 중요합니다. 교육 품질을 높이기 위해 모두가 노력해야 "
                                                                  "합니다.</p>')\n"
                                                                  "print('문서를 만들었어요.')",
                                                          'problem': '누구나 쓸 수 있는 일반론이에요. 표도, 담당도, 기한도 없어 교관이 바로 쓸 수 없어요.',
                                                          'hint': '독자와 문서 순서, 넣을 표를 적어 주세요.'}},
                                                {'id': 'source',
                                                 'label': '근거',
                                                 'why': '자료 밖의 말을 막아요.',
                                                 'options': [{'text': '근거는 이 두 파일뿐이야. 근거표에 없는 항목은 지어내지 말고 ‘자료 없음’으로 적어 줘. 기한은 48시간이고, 넘으면 ‘초과: 교육 '
                                                                      '담당에게 올림’으로 표시해 줘.',
                                                              'good': True},
                                                             {'text': '참고할 만한 내용도 알아서 넣어 줘.', 'good': False},
                                                             {'text': '', 'good': False}],
                                                 'miss': {'say': '참고 내용까지 풍부하게 넣었어요.',
                                                          'code': "report = ('<h2>교관 해설 검토 절차서</h2><p>순항미사일은 시속 900km로 비행하며, 해외 교육기관의 승인율은 평균 "
                                                                  "98%입니다.</p>'\n"
                                                                  "          '<p>검토 기한은 국제 표준에 따라 24시간으로 정합니다.</p>')\n"
                                                                  "print('문서를 만들었어요.')",
                                                          'problem': '속도, 승인율 98%, 국제 표준 모두 파일에 없는 내용이에요. **AI가 그럴듯하게 지어낸 환각이에요.** 문장이 매끄러워서 더 위험해요.',
                                                          'hint': '근거 파일을 한정하고, 없으면 ‘자료 없음’으로 쓰게 해 주세요.'}},
                                                {'id': 'tone',
                                                 'label': '표현',
                                                 'why': '상태와 책임을 분명히 적어요.',
                                                 'options': [{'text': '아직 아무도 승인하지 않았으니 문서 맨 위에 ‘AI 초안, 교관 확인 전’이라고 표시하고, 승인 전에는 공개하지 않는다고 적어 줘. 대조 '
                                                                      '결과는 ‘일치’, ‘불일치’, ‘자료 없음’으로만 써 줘.',
                                                              'good': True},
                                                             {'text': '자신 있게 확정된 문서처럼 써 줘.', 'good': False},
                                                             {'text': '', 'good': False}],
                                                 'miss': {'say': '확정 문서처럼 썼어요.',
                                                          'code': 'import pandas as pd\n'
                                                                  "d = pd.read_csv('{VDIR}/m7_drafts.csv')\n"
                                                                  "report = '<h2>검토 완료 보고</h2><p>모든 해설을 승인 완료했으며 오류가 없음을 확인했습니다.</p>' + "
                                                                  "d[['draft_id', 'family']].to_html(index=False)",
                                                          'problem': '아직 아무도 보지 않았는데 ‘승인 완료’, ‘오류 없음’으로 나왔어요. 이 문장이 퍼지면 틀린 해설이 그대로 공개돼요.',
                                                          'hint': '초안이라는 표시와 쓸 수 있는 낱말을 적어 주세요.'}}],
                                      'done': {'say': '대조표를 만들고, 근거 파일만으로 검토 절차서 초안을 만들었어요. 교관 확인이 필요해요.',
                                               'code': 'import pandas as pd\n'
                                                       '\n'
                                                       "facts = pd.read_csv('{VDIR}/m7_facts.csv').set_index('family')\n"
                                                       "drafts = pd.read_csv('{VDIR}/m7_drafts.csv')\n"
                                                       'DUE = 48\n'
                                                       '\n'
                                                       'def check(r):\n'
                                                       "    if r['field'] not in facts.columns:\n"
                                                       "        return '자료 없음'\n"
                                                       "    if facts.loc[r['family'], r['field']] == r['claim']:\n"
                                                       "        return '일치'\n"
                                                       "    return '불일치 (근거표: ' + facts.loc[r['family'], r['field']] + ')'\n"
                                                       '\n'
                                                       "drafts['대조'] = drafts.apply(check, axis=1)\n"
                                                       "drafts['기한'] = drafts['wait_hours'].apply(lambda h: '초과: 교육 담당에게 올림' if h > DUE else '기한 "
                                                       "내')\n"
                                                       "table = drafts[['draft_id', 'family', 'field', 'claim', '대조', 'wait_hours', '기한']]\n"
                                                       'display(table)\n'
                                                       '\n'
                                                       "steps = pd.DataFrame([{'단계': 'AI 초안 작성', '담당': 'AI', '기한': '-'}, {'단계': '근거표 대조', '담당': "
                                                       "'나라벨', '기한': '24시간'},\n"
                                                       "                      {'단계': '승인·수정·반려', '담당': '고교관', '기한': '48시간'}, {'단계': '승인된 해설 반영', "
                                                       "'담당': '전산호', '기한': '승인 후'},\n"
                                                       "                      {'단계': '기한 초과 건 처리', '담당': '윤훈련', '기한': '초과 시'}])\n"
                                                       "late = table[table['wait_hours'] > DUE]\n"
                                                       "report = ('<p><b>AI 초안, 교관 확인 전.</b> 승인 전에는 공개하지 않아요.</p><h2>교관 해설 검토 절차서</h2>'\n"
                                                       "          '<h3>목적</h3><p>AI 해설 초안이 교관 승인 없이 공개되지 않게 해요.</p>'\n"
                                                       "          '<h3>절차</h3>' + steps.to_html(index=False)\n"
                                                       "          + '<h3>대조 결과</h3>' + table.to_html(index=False)\n"
                                                       "          + f'<h3>기한 초과 {len(late)}건</h3>' + late[['draft_id', 'family', "
                                                       "'wait_hours']].to_html(index=False))",
                                               'lesson': '근거를 파일로 한정하니 속도나 제조사 주장은 ‘자료 없음’으로 나왔어요. D02와 D07의 불일치도 대조로 잡혔어요. 그래도 이 문서는 초안이에요. 교관이 '
                                                         '확인해야 끝나요.'},
                                      'on_ok': [['dm_gyo', 'gyo', '{call}, 불일치 건이 바로 보이네요. 절차서는 제가 읽고 고칠게요.']]}]},
                         {'short': '내 말로 쓰기',
                          'title': '문서를 받은 뒤 확인하고 내 말로 요청하기',
                          'lead': '문서를 받았다고 끝이 아니에요. **자료와 대조**하는 일은 사람의 몫이에요.',
                          'blocks': [{'id': 's3_q',
                                      'type': 'quiz',
                                      'title': '문서를 받은 뒤에는',
                                      'text': 'AI가 쓴 절차서 초안이 매끄럽게 나왔어요. 가장 먼저 무엇을 할까요?',
                                      'options': ['문장이 매끄러우니 바로 공유한다.', '수치와 고유명사를 파일과 대조한다.', '더 길게 써 달라고 한다.'],
                                      'answer': 1,
                                      'explain': '매끄러운 문장과 맞는 내용은 별개예요. 숫자와 이름부터 대조하면 환각을 빨리 잡을 수 있어요.',
                                      'hints': ['환각은 어디에 잘 숨어 있을까요?', '다시 셀 수 있는 것부터 확인해 보세요.']},
                                     {'id': 's3_write',
                                      'type': 'write',
                                      'title': '문서를 요청하는 프롬프트 쓰기',
                                      'prompt': '**새 요청**: 새로 온 교관이 읽을 ‘혼동 쌍 해설 안내문’이 필요해요. 근거는 m7_facts.csv와 m6_items.csv뿐이에요. 독자, 순서, 근거, 없는 내용 '
                                                '처리, 초안 표시를 담아 프롬프트를 써 보세요.',
                                      'example': '새로 온 교관이 읽을 혼동 쌍 해설 안내문을 HTML 한 쪽으로 만들어 줘. 순서는 안내문의 목적, 혼동 쌍 표(형상 계열, 헷갈리는 계열, 오답률), 쌍마다의 WEFT 차이, '
                                                 '주의할 점이야. 근거는 m7_facts.csv와 m6_items.csv뿐이고, 없는 내용은 ‘자료 없음’으로 적어 줘. 맨 위에 ‘AI 초안, 교관 확인 전’이라고 표시해 줘.',
                                      'min': 30,
                                      'checklist': ['독자를 적었어요', '문서 순서를 적었어요', '근거 자료를 한정했어요', '없는 내용을 어떻게 쓸지 적었어요', '초안이라는 표시를 요청했어요'],
                                      'model_label': '📝 모범 프롬프트',
                                      'placeholder': '꼭 입력해 주세요 · 내 말로 프롬프트를 써 보세요'}]}],
               'outputs': [{'id': 'm7_review',
                            'title': 'M7 해설 대조표와 교관 검토 절차서',
                            'sections': [{'h': '1. 역할 정하기', 'blocks': ['s1_match']},
                                         {'h': '2. 대조표와 절차서', 'blocks': ['s2_vibe']},
                                         {'h': '3. 내가 쓴 프롬프트', 'blocks': ['s3_write']}]}]},
              {'id': 'm8',
               'no': 8,
               'title': '배포 환경에 맞춘 모델 경량화와 오프라인 점검',
               'short': '민감 정보와 사람 확인',
               'kind': '개인',
               'competencies': ['D6'],
               'outline': 'AI에게 무엇을 넘기면 안 되는지 나눠 보고, 값은 빼고 구조만 알려 경량화 후보 비교와 안전 실패 점검표를 받아요.',
               'brief': {'subtitle': '경량화 비교와 안전 점검: 넣으면 안 되는 정보 가리기',
                         'time': '약 25분',
                         'goal': 'AI에 넣으면 안 되는 정보를 가려 프롬프트를 쓰고, 경량화 후보 비교표와 오프라인 안전 점검표를 받아 사람이 확인할 일을 정해요.',
                         'why': '모델을 줄이면 무엇을 얻고 무엇을 잃을까요? 그리고 그 비교를 AI에게 맡길 때 서버 주소나 로그를 그대로 붙여 넣어도 될까요? 인터넷이 안 되는 환경에 넣을 앱일수록, 공개 AI에 무엇을 보내는지 먼저 '
                                '따져 봐야 해요.',
                         'case': {'title': '해외 사례: 사이버 보안 기관 국장 대행이 공개 AI에 문서를 올렸다는 보도',
                                  'source': 'Politico 보도를 다룬 TechCrunch, Computing 등 (2026.1)',
                                  'points': ['2026년 1월 Politico는 미국 사이버보안·인프라보안국(CISA) 국장 대행이 ‘공무용’ 표시가 있는 계약 문서를 공개 버전의 ChatGPT에 올렸다고 보도했어요. 비밀 문서는 '
                                             '아니었다고 해요.',
                                             '자동 보안 감시가 이 일을 잡아내 내부 검토로 이어졌다고 전해졌어요. CISA는 승인된 예외 범위에서 쓴 것이라고 설명했어요.',
                                             '규칙을 만드는 곳에서도 일어나는 일이에요. 넣기 전에 한 번 나눠 보는 습관이 필요해요.']},
                         'team': '전산호(전산 담당)가 경량화 후보와 점검 기록을 줘요. 배보안(보안 담당)이 무엇을 AI에 넣는지 확인해요.',
                         'terms': [{'t': '가지치기',
                                    'en': 'pruning',
                                    'fig': 'f_prune',
                                    'd': '모델에서 덜 중요한 연결이나 채널을 없애 크기와 계산을 줄이는 방법이에요.',
                                    'easy': '나무의 잔가지를 쳐서 가볍게 만드는 거예요.',
                                    'use': '채널을 줄이면 속도까지 빨라지지만 성능이 떨어질 수 있어요. 미세조정으로 일부 되찾아요.'},
                                   {'t': '안전 실패',
                                    'en': 'fail-safe',
                                    'fig': 'f_safe',
                                    'd': '문제가 생겼을 때 틀린 답을 내지 않고 안전한 쪽으로 멈추는 것이에요.',
                                    'easy': '모르면 ‘판별 불가’라고 말하고 멈추는 거예요.',
                                    'use': '심한 흐림, 손상 파일, 빈 파일에서 라벨이 나오면 실패로 봐요.'},
                                   {'t': '넣으면 안 되는 정보',
                                    'en': 'sensitive data',
                                    'fig': 'f_secret',
                                    'd': '조직 밖으로 나가면 안 되는 주소, 계정, 인명, 위치 같은 정보예요.',
                                    'easy': '게시판에 올리면 안 되는 것은 공개 AI에도 넣지 않아요.',
                                    'use': '열 이름과 가상 샘플만으로도 코드를 받을 수 있어요.'}],
                         'steps': ['나눠 보기: AI에 넣어도 되는 것과 안 되는 것', '구조만 알리기: 값은 빼고 구조만 알려 요청하기', '사람이 할 일: 사람이 확인할 일과 내 프롬프트'],
                         'glossary': [],
                         'outputs': ['M8 경량화 비교와 오프라인 점검표']},
               'start_messages': [['dm_jeon', 'jeon', '{call}, 줄인 모델 후보와 점검 기록이에요. 파일은 m8_lite.csv와 m8_safety.csv예요. 어떤 후보가 기준을 넘는지 AI에게 시켜 봐 주세요.'],
                                  ['dm_boan', 'boan', '{call}, 잠깐요. 서버 주소나 점검 로그를 공개 AI에 그대로 넣지 마세요.']],
               'finish_messages': [['dm_boan', 'boan', '{call}, 구조만 알려 주고 코드는 내부에서 돌렸군요. 이렇게 해 주세요 👍']],
               'steps': [{'short': '나눠 보기',
                          'title': 'AI에 넣어도 되는 것과 안 되는 것',
                          'lead': '프롬프트에 넣기 전에 **한 번 나눠 보세요.** 값이 없어도 구조만 알면 코드를 받을 수 있어요.',
                          'blocks': [{'id': 's1_cls',
                                      'type': 'classify',
                                      'title': '공개 AI에 넣어도 될까요?',
                                      'text': '조직이 승인한 환경이 아니라 일반 공개 AI라고 생각하고 나눠 보세요.',
                                      'primary_label': '판단',
                                      'item_label': '내용',
                                      'contrib': False,
                                      'categories': [{'id': 'ok', 'name': '넣어도 돼요'}, {'id': 'no', 'name': '넣으면 안 돼요'}],
                                      'items': [{'id': 'a',
                                                 'title': '열 이름: variant, size_mb, ms_browser',
                                                 'accept': ['ok'],
                                                 'explain': '열 이름만으로는 실제 환경을 알 수 없어요. 코드를 받는 데 꼭 필요한 정보예요.',
                                                 'hints': ['실제 값이 들어 있나요?', '이 정보만으로 시스템을 알 수 있나요?']},
                                                {'id': 'b',
                                                 'title': '내부 서버 주소와 관리자 계정',
                                                 'accept': ['no'],
                                                 'explain': '외부로 나가면 되돌릴 수 없어요. 코드에는 자리표시자만 쓰고 값은 내 환경에서 넣어요.',
                                                 'hints': ['이 값으로 시스템에 들어갈 수 있나요?', '조직 밖으로 나가도 되는 내용인가요?']},
                                                {'id': 'c',
                                                 'title': '직접 지어낸 가상 샘플 2줄',
                                                 'accept': ['ok'],
                                                 'explain': '값의 모양을 보여 주는 가장 안전한 방법이에요.',
                                                 'hints': ['실제로 있었던 값인가요?', '지어낸 값이면 무엇이 문제일까요?']},
                                                {'id': 'd',
                                                 'title': '학습자 이름이 든 시범 운영 기록 원본',
                                                 'accept': ['no'],
                                                 'explain': '개인 정보가 그대로 나가요. 원본은 내 환경에서만 돌려요.',
                                                 'hints': ['사람을 알아볼 수 있나요?', '원본에는 무엇이 들어 있을까요?']}]}]},
                         {'short': '구조만 알리기',
                          'title': '값은 빼고 구조만 알려 요청하기',
                          'lead': '**민감 정보를 넣은 문장**도 일부러 골라 보세요. 결과는 나와도 무엇이 문제인지 볼 수 있어요.',
                          'blocks': [{'id': 's2_vibe',
                                      'type': 'vibe',
                                      'title': '경량화 비교와 안전 점검표',
                                      'ask': ['jeon', '크기 5MB 이하, 브라우저에서 60ms 이하, 재현율 90% 이상인 후보를 찾고, 안전 실패도 확인하고 싶어요.'],
                                      'file': 'm8_lite.csv',
                                      'slots': [{'id': 'safe',
                                                 'label': '넘길 정보',
                                                 'why': '값은 빼고 구조만 알려요.',
                                                 'options': [{'text': '실제 값은 넣지 않을게. m8_lite.csv의 열은 variant, method, size_mb, ms_browser, '
                                                                      'threat_recall_pct, top1_acc_pct이고, m8_safety.csv의 열은 case_id, scenario, '
                                                                      'expected, output, score야. 코드는 내 환경에서 돌릴게.',
                                                              'good': True},
                                                             {'text': '내부 서버 주소와 관리자 계정, 점검 로그 전체를 그대로 붙여 넣을게. 여기서 알아서 찾아 줘.', 'good': False},
                                                             {'text': '', 'good': False}],
                                                 'miss': {'say': '붙여 주신 내용으로 후보 표를 만들었어요.',
                                                          'code': 'import pandas as pd\n'
                                                                  "df = pd.read_csv('{VDIR}/m8_lite.csv')\n"
                                                                  "display(df[['variant', 'size_mb', 'ms_browser']])",
                                                          'problem': '코드는 돌았지만 **내부 서버 주소와 관리자 계정이 이미 외부 AI로 나갔어요.** 한 번 보낸 내용은 되돌리기 어려워요. 코드를 받는 데 '
                                                                     '값은 필요 없었어요.',
                                                          'hint': '열 이름만 알려 주고, 코드는 내 환경에서 돌린다고 적어 주세요.'}},
                                                {'id': 'rule',
                                                 'label': '기준',
                                                 'why': '기준값을 숫자로 적어요.',
                                                 'options': [{'text': 'size_mb 5 이하, ms_browser 60 이하, threat_recall_pct 90 이상이면 통과, 아니면 어느 기준에서 '
                                                                      '탈락했는지 적어 줘. 기준값은 맨 위에 설정값으로 둬 줘.',
                                                              'good': True},
                                                             {'text': '가볍고 빠른 걸로 골라 줘.', 'good': False},
                                                             {'text': '', 'good': False}],
                                                 'miss': {'say': '가장 가벼운 모델을 골랐어요.',
                                                          'code': 'import pandas as pd\n'
                                                                  "df = pd.read_csv('{VDIR}/m8_lite.csv')\n"
                                                                  "best = df.sort_values('size_mb').iloc[0]\n"
                                                                  "print('추천:', best['variant'], best['size_mb'], 'MB')",
                                                          'problem': '가장 작은 channel_70을 골랐지만 재현율이 기준 아래예요. ‘가볍고 빠른’만 말해서 재현율을 보지 않았어요.',
                                                          'hint': '세 기준을 숫자로 적어 주세요.'}},
                                                {'id': 'fail',
                                                 'label': '안전 점검',
                                                 'why': '실패할 때의 동작을 확인해요.',
                                                 'options': [{'text': 'm8_safety.csv에서 expected가 unknown인데 output이 unknown이 아니면 ‘안전 실패’로 표시하고 목록을 뽑아 '
                                                                      '줘. 두 표를 넣은 오프라인 점검표를 HTML로 report 변수에 담고, 맨 위에 ‘AI 초안, 전산·보안 담당 확인 전’이라고 적어 '
                                                                      '줘.',
                                                              'good': True},
                                                             {'text': '문제 없는지 확인해 줘.', 'good': False},
                                                             {'text': '', 'good': False}],
                                                 'miss': {'say': '확인했어요. 대부분 잘 동작해요.',
                                                          'code': 'import pandas as pd\n'
                                                                  "s = pd.read_csv('{VDIR}/m8_safety.csv')\n"
                                                                  "print('점검', len(s), '건 완료. 큰 문제 없어 보입니다.')",
                                                          'problem': '‘큰 문제 없어 보입니다’만 있어요. 심한 흐림에서 라벨을 낸 C03과 빈 파일에서 라벨을 낸 C06이 가려졌어요.',
                                                          'hint': '어떤 경우를 실패로 볼지 규칙을 적어 주세요.'}}],
                                      'done': {'say': '값 없이 구조만으로 코드를 만들었어요. 내 환경에서 돌린 결과와 점검표예요. 배포 결정은 담당자 확인이 필요해요.',
                                               'code': 'import pandas as pd\n'
                                                       'import matplotlib.pyplot as plt\n'
                                                       '\n'
                                                       'MAX_MB, MAX_MS, MIN_RECALL = 5, 60, 90\n'
                                                       "lite = pd.read_csv('{VDIR}/m8_lite.csv')\n"
                                                       "safety = pd.read_csv('{VDIR}/m8_safety.csv')\n"
                                                       '\n'
                                                       'def verdict(r):\n'
                                                       '    why = []\n'
                                                       "    if r['size_mb'] > MAX_MB:\n"
                                                       "        why.append('크기')\n"
                                                       "    if r['ms_browser'] > MAX_MS:\n"
                                                       "        why.append('속도')\n"
                                                       "    if r['threat_recall_pct'] < MIN_RECALL:\n"
                                                       "        why.append('재현율')\n"
                                                       "    return '통과' if not why else '탈락: ' + ', '.join(why)\n"
                                                       '\n'
                                                       "lite['판정'] = lite.apply(verdict, axis=1)\n"
                                                       "display(lite[['variant', 'method', 'size_mb', 'ms_browser', 'threat_recall_pct', '판정']])\n"
                                                       '\n'
                                                       "safety['결과'] = safety.apply(lambda r: '안전 실패' if r['expected'] == 'unknown' and r['output'] "
                                                       "!= 'unknown' else '정상', axis=1)\n"
                                                       "bad = safety[safety['결과'] == '안전 실패']\n"
                                                       "display(bad[['case_id', 'scenario', 'output', 'score', '결과']])\n"
                                                       '\n'
                                                       'fig, ax = plt.subplots(figsize=(6, 3))\n'
                                                       "ax.scatter(lite['ms_browser'], lite['threat_recall_pct'])\n"
                                                       'for _, r in lite.iterrows():\n'
                                                       "    ax.annotate(r['variant'], (r['ms_browser'], r['threat_recall_pct']), fontsize=7)\n"
                                                       "ax.axhline(MIN_RECALL, color='gray', linestyle='--')\n"
                                                       "ax.axvline(MAX_MS, color='gray', linestyle='--')\n"
                                                       "ax.set_xlabel('ms in browser')\n"
                                                       "ax.set_ylabel('threat recall (%)')\n"
                                                       '\n'
                                                       "report = ('<p><b>AI 초안, 전산·보안 담당 확인 전.</b></p><h2>오프라인 점검표</h2>'\n"
                                                       "          + f'<p>기준: {MAX_MB}MB 이하, {MAX_MS}ms 이하, 재현율 {MIN_RECALL}% 이상. 측정값은 가상 자료예요.</p>'\n"
                                                       "          + lite[['variant', 'size_mb', 'ms_browser', 'threat_recall_pct', "
                                                       "'판정']].to_html(index=False)\n"
                                                       "          + f'<h3>안전 실패 {len(bad)}건</h3>' + bad[['case_id', 'scenario', 'output', "
                                                       "'score']].to_html(index=False))",
                                               'lesson': '값을 넘기지 않아도 코드는 똑같이 받을 수 있어요. 미세조정한 channel_50_ft만 세 기준을 모두 넘었고, 안전 실패 두 건은 배포 전에 사람이 확인하고 '
                                                         '고쳐야 해요.'},
                                      'on_ok': [['dm_jeon', 'jeon', '{call}, 안전 실패 두 건이 바로 보이네요. 이건 고친 뒤에 다시 점검할게요.']]}]},
                         {'short': '사람이 할 일',
                          'title': '사람이 확인할 일과 내 프롬프트',
                          'lead': 'AI가 ‘통과’라고 해도 **배포를 결정하는 것은 사람이에요.**',
                          'blocks': [{'id': 's3_multi',
                                      'type': 'multi',
                                      'title': '사람이 확인할 일',
                                      'text': 'AI가 만든 점검표를 받은 뒤 사람이 꼭 할 일을 모두 골라 주세요.',
                                      'items': [{'text': '안전 실패로 나온 경우를 실제로 다시 재현해 본다.',
                                                 'correct': True,
                                                 'explain': '표의 값이 맞는지, 고친 뒤에 사라졌는지 사람이 확인해요.',
                                                 'hint': '표만 보고 끝내도 될까요?'},
                                                {'text': '전산·보안 담당의 확인을 받은 뒤 배포를 결정한다.',
                                                 'correct': True,
                                                 'explain': '배포는 책임이 따르는 결정이에요. AI 표는 근거일 뿐이에요.',
                                                 'hint': '누가 책임지는 일일까요?'},
                                                {'text': 'AI가 통과라고 했으니 바로 배포한다.',
                                                 'correct': False,
                                                 'explain': '통과는 표에 적은 기준을 넘었다는 뜻일 뿐이에요.',
                                                 'hint': '기준에 없는 위험도 있을까요?'},
                                                {'text': '프롬프트에 민감 정보가 들어가지 않았는지 다시 본다.',
                                                 'correct': True,
                                                 'explain': '보내기 전에 한 번 더 보는 습관이 가장 싸고 확실한 방법이에요.',
                                                 'hint': '보낸 뒤에는 되돌릴 수 있을까요?'}]},
                                     {'id': 's3_write',
                                      'type': 'write',
                                      'title': '민감 정보를 뺀 프롬프트 쓰기',
                                      'prompt': '**새 요청**: 실제 점검 로그에서 응답 시간이 200ms를 넘는 요청을 세어 보고 싶어요. 로그에는 기기 번호와 내부 주소가 들어 있어요. 값을 넘기지 않고 어떻게 '
                                                '요청할까요?',
                                      'example': '실제 값은 넣지 않고 구조만 알려 줄게. 파일은 log.csv이고 열은 device_id, server_addr, response_ms야. 샘플은 D001, ADDR1, '
                                                 '120처럼 지어낸 값이야. response_ms가 200을 넘는 건수와 비율을 계산하는 코드를 만들어 줘. device_id와 server_addr는 결과에 넣지 말아 줘. '
                                                 '코드는 내 환경에서 돌릴게.',
                                      'min': 30,
                                      'checklist': ['실제 값을 넣지 않았어요', '열 이름과 가상 샘플로 설명했어요', '기준을 숫자로 적었어요', '결과에 민감 정보가 나오지 않게 했어요'],
                                      'model_label': '📝 모범 프롬프트',
                                      'placeholder': '꼭 입력해 주세요 · 내 말로 프롬프트를 써 보세요'}]}],
               'outputs': [{'id': 'm8_offline',
                            'title': 'M8 경량화 비교와 오프라인 점검표',
                            'sections': [{'h': '1. 공개 AI에 넣어도 될까요?', 'blocks': ['s1_cls']},
                                         {'h': '2. 비교표와 안전 점검표', 'blocks': ['s2_vibe']},
                                         {'h': '3. 사람이 확인할 일', 'blocks': ['s3_multi']},
                                         {'h': '4. 내가 쓴 프롬프트', 'blocks': ['s3_write']}]}]},
              {'id': 'm9',
               'no': 9,
               'title': '시범 운영 학습 기록 사후검토(AAR)와 재학습',
               'short': '리뷰와 테스트',
               'kind': '개인',
               'competencies': ['D3', 'D5'],
               'outline': '남이 만든 오답률 코드를 바로 고치지 않고, 먼저 설명하게 하고 답을 아는 테스트로 확인해요. 고칠 범위를 정해 고친 뒤, 재학습 전후를 비교하는 사후검토 문서를 받아요.',
               'brief': {'subtitle': '오답률 코드 리뷰와 AAR 문서: 테스트로 확인하기',
                         'time': '약 30분',
                         'goal': 'AI에게 코드를 먼저 설명하게 하고, 답을 아는 테스트와 수정 범위를 요청해 오답률 계산을 바로잡은 뒤 AAR 문서를 받아요.',
                         'why': '재학습했더니 정확도가 올랐다면 그걸로 끝일까요? 새 형상은 계속 나오고, 도구는 계속 고쳐야 해요. 그런데 숫자가 이상하다고 바로 ‘고쳐 줘’라고 하면 AI가 다른 곳까지 바꾸기 쉬워요. 먼저 설명하게 '
                                '하고, 답을 아는 값으로 확인해요.',
                         'case': {'title': '해외 사례: 제트 엔진을 단 샤헤드로 추정되는 드론의 등장',
                                  'source': 'Euromaidan Press, NV, Kyiv Post 보도 (2025.6)',
                                  'points': ['2025년 6월 키이우 공격에서 제트 엔진을 단 드론이 쓰인 것으로 보도됐어요. 주민들은 평소와 다른 소리를 들었고, 격추된 잔해에서 작은 제트 엔진이 나왔다고 해요.',
                                             '다만 정확한 기종은 확인되지 않았고, 처음 쓰인 날짜도 매체마다 조금씩 달라요. 게란-3로 추정한다는 보도가 많아요.',
                                             '새 형상이나 개량형이 나오면 교육 자료와 모델을 다시 봐야 해요. 고칠 때마다 도구가 맞는지 확인하는 습관이 필요해요.']},
                         'team': '한모델(AI 개발 담당)이 받아 온 코드예요. 윤훈련(교육 담당)이 AAR 문서를 기다려요. 사람을 탓하지 않고 원인을 찾아요.',
                         'terms': [{'t': '사후검토 (AAR)',
                                    'en': 'After Action Review',
                                    'fig': 'f_aar',
                                    'd': '일이 끝난 뒤 계획과 실제를 비교해 원인과 다음 조치를 찾는 검토예요.',
                                    'easy': '누구 잘못인지가 아니라 무엇을 바꿀지 찾는 시간이에요.',
                                    'use': 'AI에게 양식과 근거 파일을 주면 초안을 빠르게 받을 수 있어요.'},
                                   {'t': '회귀 오류',
                                    'en': 'regression',
                                    'fig': 'f_reg',
                                    'd': '고치고 나니 전에 맞던 것이 틀리게 되는 오류예요.',
                                    'easy': '한쪽을 고치니 다른 쪽이 망가지는 거예요.',
                                    'use': '재학습 뒤에는 새로 생긴 오류 수도 함께 봐요.'},
                                   {'t': '코드 리뷰',
                                    'en': 'code review',
                                    'fig': 'f_review',
                                    'd': '코드를 고치기 전에 읽고 의심되는 곳을 찾는 일이에요.',
                                    'easy': 'AI에게 ‘먼저 설명해 봐’라고 시키는 것도 리뷰예요.',
                                    'use': '설명, 테스트, 수정, 비교 순서로 요청해요.'},
                                   {'t': '수정 범위',
                                    'en': 'scope',
                                    'fig': 'f_scope',
                                    'd': '고쳐도 되는 곳과 건드리면 안 되는 곳을 정한 것이에요.',
                                    'easy': '‘이 함수만 고쳐’라고 말하는 거예요.',
                                    'use': '범위가 없으면 잘 되던 부분까지 바뀔 수 있어요.'}],
                         'steps': ['먼저 읽기: 고치기 전에 먼저 설명하게 하기', '테스트로 고치기: 설명·테스트·범위를 담아 요청하기', '내 말로 쓰기: 후속 제안 판단과 내 프롬프트'],
                         'glossary': [],
                         'outputs': ['M9 사후검토 문서와 수정 기록']},
               'start_messages': [['dm_model',
                                   'model',
                                   '{call}, 시범 운영 기록으로 오답률을 냈는데 delta_wing이 이상하게 낮아 보여요. 시간 초과가 많았다는 말을 들었거든요. 파일은 m9_log.csv예요.']],
               'finish_messages': [['dm_hun', 'hun', '{call}, 전후 비교와 표본 없는 조건까지 적혀 있네요. 다음 교육 자료에 반영할게요.']],
               'steps': [{'short': '먼저 읽기',
                          'title': '고치기 전에 먼저 설명하게 하기',
                          'lead': '숫자가 이상할 때 바로 ‘고쳐 줘’라고 하면 **무엇이 왜 바뀌었는지 모르게 돼요.**',
                          'blocks': [{'id': 's1_info',
                                      'type': 'info',
                                      'title': '남의 코드를 고칠 때의 네 단계',
                                      'icon': '🧪',
                                      'text': ['AI가 만든 코드도, 다른 사람이 만든 코드도 같아요. 아래 순서로 요청해요.'],
                                      'table': {'head': ['단계', '프롬프트에 적을 말', '왜'],
                                                'rows': [['설명', '고치기 전에 무엇을 계산하는지 설명해 줘', '내가 이해한 뜻과 같은지 확인해요'],
                                                         ['테스트', '응답 3건 중 1건 오답이면 33.3%가 나오는 assert를 만들어 줘', '답을 아는 값으로 확인해요'],
                                                         ['수정', 'wrong_rate 함수만 고쳐 줘', '다른 곳이 바뀌지 않아요'],
                                                         ['비교', '고치기 전후를 표로 보여 줘', '무엇이 달라졌는지 봐요']]},
                                      'figure': 'f_review'},
                                     {'id': 's1_q',
                                      'type': 'quiz',
                                      'title': '처음 무엇을 요청할까요?',
                                      'text': '오답률 코드가 이상해 보여요. 가장 먼저 AI에게 무엇을 요청할까요?',
                                      'options': ['전부 새로 짜 줘.', '이 코드가 무엇을 어떻게 계산하는지 먼저 설명하고, 의심되는 곳을 알려 줘.', '오답률을 더 높게 나오게 해 줘.'],
                                      'answer': 1,
                                      'explain': '먼저 이해해야 무엇을 고칠지 정할 수 있어요. 셋째처럼 원하는 답을 정해 주면 AI는 거기에 맞춰 버려요.',
                                      'hints': ['어느 요청이 원인을 찾게 할까요?', '결과를 정해 놓고 요청하면 어떻게 될까요?']}]},
                         {'short': '테스트로 고치기',
                          'title': '설명·테스트·범위를 담아 요청하기',
                          'lead': '지금 코드는 **오류 없이 돌아요.** 그래서 더 찾기 어려워요.',
                          'blocks': [{'id': 's2_vibe',
                                      'type': 'vibe',
                                      'title': '오답률 코드 리뷰와 AAR 문서',
                                      'ask': ['model', '이 코드로 낸 오답률이 맞는지 봐 주세요.'],
                                      'file': 'm9_log.csv',
                                      'slots': [{'id': 'explain',
                                                 'label': '먼저 설명',
                                                 'why': '고치기 전에 읽게 해요.',
                                                 'options': [{'text': 'm9_log.csv의 열은 learner_id, family, confusable_with, answered(응답 1, 미응답 0), '
                                                                      'correct(정답 1, 오답 0, 미응답은 빈칸)야. 고치기 전에 wrong_rate가 무엇을 무엇으로 나누는지 설명하고, 의심되는 곳을 '
                                                                      '알려 줘.',
                                                              'good': True},
                                                             {'text': '그냥 바로 고쳐 줘.', 'good': False},
                                                             {'text': '', 'good': False}],
                                                 'miss': {'say': '고쳤어요! 분모를 사람 수로 바꿨더니 숫자가 커졌어요.',
                                                          'code': 'import pandas as pd\n'
                                                                  "df = pd.read_csv('{VDIR}/m9_log.csv')\n"
                                                                  'rows = []\n'
                                                                  "for fam, s in df.groupby('family'):\n"
                                                                  "    wrong = ((s['answered'] == 1) & (s['correct'] == 0)).sum()\n"
                                                                  "    rows.append({'형상 계열': fam, '오답률(%)': round(wrong / 40 * 100, 1)})\n"
                                                                  'display(pd.DataFrame(rows))',
                                                          'problem': '오답 수를 사람 수(40)로 나눠서 100%에 가깝거나 넘을 수 있는 숫자가 됐어요. 무엇이 문제인지 설명 없이 고치니 **다른 잘못으로 '
                                                                     '바뀐 줄도 모르게 돼요.**',
                                                          'hint': '고치기 전에 설명부터 하게 해 주세요.'}},
                                                {'id': 'test',
                                                 'label': '테스트',
                                                 'why': '답을 아는 값으로 확인해요.',
                                                 'options': [{'text': '응답 3건 중 오답 1건, 미응답 1건인 작은 표로 33.3%가 나오는 assert 테스트를 먼저 만들어 줘.', 'good': True},
                                                             {'text': '잘 되는지 알아서 확인해 줘.', 'good': False},
                                                             {'text': '', 'good': False}],
                                                 'miss': {'say': '확인했어요. 잘 동작해요!',
                                                          'code': 'import pandas as pd\n'
                                                                  "df = pd.read_csv('{VDIR}/m9_log.csv')\n"
                                                                  "a = df[df['answered'] == 1]\n"
                                                                  "print('전체 오답률', round((a['correct'] == 0).mean() * 100, 1), '%')\n"
                                                                  "print('잘 동작해요!')",
                                                          'problem': '‘잘 동작해요’라는 말뿐이고, 무엇으로 확인했는지 없어요. 전체 평균만 냈고 형상 계열별 표도 사라졌어요.',
                                                          'hint': '손으로 계산할 수 있는 작은 예시를 적어 주세요.'}},
                                                {'id': 'scope',
                                                 'label': '범위와 문서',
                                                 'why': '고칠 곳과 문서를 정해요.',
                                                 'options': [{'text': 'wrong_rate 함수만 고치고 미응답 기록은 지우지 말아 줘. 전후 비교표와 미응답 수를 보여 주고, '
                                                                      'm9_retrain.csv(condition, samples, before_recall_pct, after_recall_pct, '
                                                                      'fixed_errors, new_errors)의 재학습 전후도 함께 넣은 AAR 문서를 HTML로 report 변수에 담아 줘. 사람을 '
                                                                      '탓하지 말고 원인과 다음 조치를 적어 줘.',
                                                              'good': True},
                                                             {'text': '알아서 깔끔하게 전부 정리해 줘.', 'good': False},
                                                             {'text': '', 'good': False}],
                                                 'miss': {'say': '깔끔하게 정리했어요. 미응답 기록은 지웠어요.',
                                                          'code': 'import pandas as pd\n'
                                                                  "df = pd.read_csv('{VDIR}/m9_log.csv')\n"
                                                                  "clean = df[df['answered'] == 1].copy()\n"
                                                                  "display(clean.groupby('family')['correct'].apply(lambda s: round((s == 0).mean() "
                                                                  '* 100, 1)).reset_index())\n'
                                                                  "print('남은 행', len(clean), '/ 원래 행', len(df))",
                                                          'problem': '오답률은 맞아졌지만 미응답 기록이 사라져, 시간 초과가 많은 문항을 알 수 없게 됐어요. 부탁하지 않은 일까지 했어요.',
                                                          'hint': '고칠 곳과 건드리면 안 되는 것을 적어 주세요.'}}],
                                      'done': {'say': '분모가 미응답까지 포함한 것이 원인이에요. 분모는 응답 수여야 해요. 테스트가 통과한 뒤 wrong_rate만 고쳤고, 미응답 기록은 그대로 뒀어요.',
                                               'code': 'import pandas as pd\n'
                                                       'import matplotlib.pyplot as plt\n'
                                                       '\n'
                                                       "df = pd.read_csv('{VDIR}/m9_log.csv')\n"
                                                       '\n'
                                                       '# 설명: 예전 코드는 오답 수를 전체 행(미응답 포함)으로 나눠 오답률이 낮게 나왔어요.\n'
                                                       'def old_rate(s):\n'
                                                       "    wrong = ((s['answered'] == 1) & (s['correct'] == 0)).sum()\n"
                                                       '    return round(wrong / len(s) * 100, 1)\n'
                                                       '\n'
                                                       'def wrong_rate(s):\n'
                                                       "    a = s[s['answered'] == 1]\n"
                                                       "    return round((a['correct'] == 0).sum() / len(a) * 100, 1)\n"
                                                       '\n'
                                                       '# 테스트: 응답 3건 중 오답 1건, 미응답 1건\n'
                                                       "small = pd.DataFrame({'answered': [1, 1, 1, 0], 'correct': [1, 0, 1, None]})\n"
                                                       'assert wrong_rate(small) == 33.3\n'
                                                       "print('테스트 통과')\n"
                                                       '\n'
                                                       'rows = []\n'
                                                       "for fam, s in df.groupby('family'):\n"
                                                       "    rows.append({'형상 계열': fam, '고치기 전(%)': old_rate(s), '고친 뒤(%)': wrong_rate(s), '응답': "
                                                       "int((s['answered'] == 1).sum()), '미응답': int((s['answered'] == 0).sum())})\n"
                                                       "table = pd.DataFrame(rows).sort_values('고친 뒤(%)', ascending=False)\n"
                                                       'display(table)\n'
                                                       '\n'
                                                       "re = pd.read_csv('{VDIR}/m9_retrain.csv').fillna('자료 없음')\n"
                                                       'display(re)\n'
                                                       '\n'
                                                       "plot_df = table.rename(columns={'고치기 전(%)': 'before', '고친 뒤(%)': 'after'})\n"
                                                       "ax = plot_df.plot(x='형상 계열', y=['before', 'after'], kind='bar', figsize=(6.5, 3), rot=20)\n"
                                                       "ax.set_xlabel('')\n"
                                                       "ax.set_ylabel('wrong rate (%)')\n"
                                                       '\n'
                                                       'top = table.iloc[0]\n'
                                                       "report = ('<p><b>AI 초안, 교관 확인 전.</b></p><h2>시범 운영 사후검토(AAR)</h2>'\n"
                                                       '          + f"<p><b>결론:</b> 오답률이 가장 높은 계열은 {top[\'형상 계열\']}({top[\'고친 뒤(%)\']}%)이고, 미응답은 '
                                                       '{top[\'미응답\']}건이에요.</p>"\n'
                                                       "          + '<h3>오답률 전후</h3>' + table.to_html(index=False)\n"
                                                       "          + '<h3>재학습 전후</h3>' + re.to_html(index=False)\n"
                                                       "          + '<h3>원인과 다음 조치</h3><ul><li>분모에 미응답이 포함돼 오답률이 낮게 보였어요. 계산 함수를 고쳤어요.</li>'\n"
                                                       "          + '<li>near_white_hot은 재현율이 떨어지고 새 오류가 늘었어요. 회귀를 다시 확인해요.</li>'\n"
                                                       "          + '<li>far_black_hot은 표본이 없어 재학습 효과를 말할 수 없어요. 자료를 더 모아요.</li></ul>')",
                                               'lesson': '설명을 먼저 받으니 분모 문제가 드러났고, 테스트가 있어 맞게 고쳤다는 것을 확인했어요. 재학습 표에서는 평균이 올라도 near_white_hot은 떨어졌다는 '
                                                         '것도 보였어요.'},
                                      'start': {'say': '지금 쓰고 있는 코드예요. delta_wing 오답률이 이상하게 낮아요.',
                                                'code': 'import pandas as pd\n'
                                                        '\n'
                                                        "df = pd.read_csv('{VDIR}/m9_log.csv')\n"
                                                        '\n'
                                                        'def wrong_rate(s):\n'
                                                        "    wrong = ((s['answered'] == 1) & (s['correct'] == 0)).sum()\n"
                                                        '    return round(wrong / len(s) * 100, 1)\n'
                                                        '\n'
                                                        "rows = [{'형상 계열': fam, '오답률(%)': wrong_rate(s)} for fam, s in df.groupby('family')]\n"
                                                        "display(pd.DataFrame(rows).sort_values('오답률(%)', ascending=False))"},
                                      'on_ok': [['dm_model', 'model', '{call}, 분모가 문제였군요. 미응답도 그대로 남아서 시간 초과 문제도 함께 볼 수 있겠어요.']]}]},
                         {'short': '내 말로 쓰기',
                          'title': '후속 제안 판단과 내 프롬프트',
                          'lead': '문서를 받은 뒤에는 **다음 조치를 사람이** 정해요.',
                          'blocks': [{'id': 's3_decide',
                                      'type': 'decide',
                                      'title': '후속 제안 판단하기',
                                      'text': 'AAR 문서를 본 팀원들의 제안이에요. 받아들일지 정하고 이유를 적어 주세요.',
                                      'reason_min': 8,
                                      'items': [{'id': 'a',
                                                 'who': 'model',
                                                 'text': '평균 재현율이 올랐으니 이번 재학습은 성공으로 보고하고 near 조건은 넘어가요.',
                                                 'answer': 'reject',
                                                 'hint': '표에서 near_white_hot은 어떻게 됐나요?',
                                                 'explain': '평균이 올라도 잘 되던 조건이 떨어지면 회귀 오류예요. 함께 적어야 해요.',
                                                 'reason': '평균만 보면 회귀를 놓쳐요. 떨어진 조건도 함께 적어야 해요.'},
                                                {'id': 'b',
                                                 'who': 'hun',
                                                 'text': '미응답이 많은 delta_wing 문항은 제한 시간과 보기 수를 먼저 봐요. 학습자를 탓할 일은 아니에요.',
                                                 'answer': 'accept',
                                                 'hint': '미응답은 누구 탓일까요, 무엇을 바꿔 볼 수 있을까요?',
                                                 'explain': '사람이 아니라 조건을 바꿔 보는 것이 AAR이에요. 같은 도구로 다시 비교할 수 있어요.',
                                                 'reason': '탓하지 않고 조건을 바꿔 다시 비교하는 것이 AAR이에요.'}]},
                                     {'id': 's3_write',
                                      'type': 'write',
                                      'title': '리뷰 프롬프트 쓰기',
                                      'prompt': '**새 요청**: 다른 팀에서 받은 평균 응답 시간 계산 코드가 이상해 보여요. AI에게 어떻게 요청할까요? 고치기 전 설명, 테스트, 범위를 담아 보세요.',
                                      'example': '이 코드를 고치기 전에 평균 응답 시간을 어떻게 계산하는지 먼저 설명하고, 의심되는 곳을 알려 줘. 응답 시간이 10초, 20초, 30초면 평균 20초가 나오는 assert '
                                                 '테스트도 만들어 줘. 고치는 건 계산 함수만 하고, 원본 파일과 다른 코드는 건드리지 마. 고치기 전후 결과를 표로 보여 줘.',
                                      'min': 30,
                                      'checklist': ['먼저 설명하게 했어요', '답을 아는 테스트를 요청했어요', '고칠 곳과 건드리면 안 되는 곳을 적었어요', '전후를 비교하게 했어요'],
                                      'model_label': '📝 모범 프롬프트',
                                      'placeholder': '꼭 입력해 주세요 · 내 말로 프롬프트를 써 보세요'}]}],
               'outputs': [{'id': 'm9_aar',
                            'title': 'M9 사후검토 문서와 수정 기록',
                            'sections': [{'h': '1. AI에게 보낸 프롬프트와 결과', 'blocks': ['s2_vibe']},
                                         {'h': '2. 후속 제안 판단', 'blocks': ['s3_decide']},
                                         {'h': '3. 내가 쓴 리뷰 프롬프트', 'blocks': ['s3_write']}]}]},
              {'id': 'm10',
               'no': 10,
               'title': '배포 점검 회의와 운영 인계 브리핑',
               'short': '묶어서 인계하기',
               'kind': '협업',
               'competencies': ['D4', 'J1'],
               'outline': '앞에서 배운 프롬프트를 모아 조건별 결과와 인계표를 만들고 문서를 스스로 점검하는 미니 도구를 만든 뒤, 나만의 프롬프트 틀을 정리해요.',
               'brief': {'subtitle': '인계 도구와 1쪽 브리핑: 묶어서 돌리고 조건까지 적기',
                         'time': '약 30분',
                         'goal': '다시 돌릴 수 있는 미니 도구를 요청하고, 수치마다 조건을 밝힌 브리핑과 운영 인계표, 자동 점검, 내 프롬프트 틀을 만들어요.',
                         'why': '학습앱을 넘기는 날, 무엇을 확인하고 누구에게 무엇을 맡길까요? 자동화할수록 틀린 내용도 빠르게 퍼져요. 수치에는 조건을 붙이고, 담당과 주기를 적고, 올리기 전에 사람이 확인해야 해요.',
                         'case': {'title': '해외 사례: 모든 분대에 소형 드론을 두겠다는 미 국방부 메모',
                                  'source': 'Breaking Defense, DefenseScoop 보도 (2025.7)',
                                  'points': ['2025년 7월 미국 국방장관은 ‘미군 드론 우위 강화’ 메모에 서명했어요. 소모품처럼 쓰는 소형 드론을 분대마다 두는 것을 목표로 했고, 목표 시점은 매체에 따라 2026 회계연도 '
                                             '말 또는 2026년 말로 전해졌어요.',
                                             '드론을 관련 훈련 전반에 넣으라는 내용도 들어 있어요. 다만 이 메모만으로 충분하지 않을 수 있다는 전문가 의견도 함께 보도됐어요.',
                                             '교육 대상이 넓어질수록 콘텐츠를 누가 고치고 승인해 배포하는지 먼저 정해 둬야 해요. 인계 문서가 그 역할을 해요.']},
                         'team': '윤훈련(교육 담당)이 인계 문서를 받아요. 전산호와 배보안이 점검 의견을 줘요.',
                         'terms': [{'t': '미니 도구',
                                    'en': 'small tool',
                                    'fig': 'f_tool',
                                    'd': '설정값, 계산, 표와 그림, 문서까지 한 번에 돌아가는 작은 코드 묶음이에요.',
                                    'easy': '데이터만 바꾸면 다시 쓸 수 있는 계산기예요.',
                                    'use': '설정값을 맨 위에 모아 달라고 요청해요.'},
                                   {'t': '운영 인계서',
                                    'en': 'handover document',
                                    'fig': 'f_hand',
                                    'd': '무엇을 누가 어떤 주기로 맡는지 적어 넘기는 문서예요.',
                                    'easy': '내가 없어도 일이 돌아가게 하는 안내서예요.',
                                    'use': '새 형상 등록, 콘텐츠 검수, 배포 승인, 장애 대응을 빠뜨리지 않아요.'},
                                   {'t': '문서 자동 점검 (린트)',
                                    'en': 'document lint',
                                    'fig': 'f_lint',
                                    'd': '문서에 꼭 들어갈 항목이 빠졌는지 코드로 검사하는 것이에요.',
                                    'easy': '제출 전 체크리스트를 코드로 만든 거예요.',
                                    'use': '표본 수, 한계, 담당, AI 산출물 표기가 있는지 검사하게 해요.'},
                                   {'t': '프롬프트 모음',
                                    'en': 'prompt library',
                                    'fig': 'f_four',
                                    'd': '잘 통한 프롬프트를 틀로 모아 두고 다시 쓰는 개인 자료예요.',
                                    'easy': '자주 쓰는 업무 양식함이에요.',
                                    'use': '파일 이름과 숫자 자리를 빈칸으로 두면 다음에 바로 채워 쓸 수 있어요.'}],
                         'steps': ['도구로 묶기: 미니 도구와 브리핑 요청하기', '배포 점검: 담당별 의견과 올리기 전 점검', '프롬프트 모음: 나만의 프롬프트 틀 만들기'],
                         'glossary': [],
                         'outputs': ['M10 1쪽 브리핑·운영 인계표와 프롬프트 틀']},
               'start_messages': [['dm_hun',
                                   'hun',
                                   '{call}, 마지막이에요. 통합 실행 기록으로 조건별 결과와 인계표가 든 1쪽 브리핑을 만들어 주세요. 파일은 m10_run.csv예요. 다음 달에도 같은 방식으로 돌릴 수 있으면 좋겠어요.']],
               'finish_messages': [['dm_hun', 'hun', '{call}, 인계 잘 받았어요. 어떤 조건에서 조심할지, 누가 무엇을 맡는지 분명하네요. 고생 많았어요.']],
               'steps': [{'short': '도구로 묶기',
                          'title': '미니 도구와 브리핑 요청하기',
                          'lead': '지금까지 배운 것을 모두 써 보세요. **네 칸, 숫자 규칙, 근거 수치, 범위**가 모두 필요해요.',
                          'blocks': [{'id': 's1_vibe',
                                      'type': 'vibe',
                                      'title': '인계 도구와 1쪽 브리핑',
                                      'ask': ['hun', '거리별 결과, 인계표, 자동 점검까지 한 번에 나오게 해 주세요.'],
                                      'file': 'm10_run.csv',
                                      'slots': [{'id': 'tool',
                                                 'label': '구성',
                                                 'why': '다시 돌릴 수 있게 해요.',
                                                 'options': [{'text': 'm10_run.csv의 열은 run_id, distance(near·mid·far), grade(show·check·unknown), '
                                                                      'correct, reviewed(교관이 확인했으면 1), ms야. 파일 이름과 표본 기준을 맨 위 설정값으로 두고, 계산은 함수로 나눠 '
                                                                      '줘. 다음 달 파일로도 바로 돌릴 수 있게 해 줘.',
                                                              'good': True},
                                                             {'text': '분석 코드 짜 줘.', 'good': False},
                                                             {'text': '', 'good': False}],
                                                 'miss': {'say': '전체 결과를 계산했어요. 정답률이 꽤 높아요!',
                                                          'code': 'import pandas as pd\n'
                                                                  "df = pd.read_csv('{VDIR}/m10_run.csv')\n"
                                                                  "print('정답률:', round(df['correct'].mean() * 100, 1), '%')",
                                                          'problem': '구간을 나누지 않은 숫자 하나뿐이에요. 파일 이름도 코드 속에 박혀 있어 다음 달에 다시 쓰기 어려워요.',
                                                          'hint': '무엇을 설정값으로 둘지, 어떻게 나눌지 적어 주세요.'}},
                                                {'id': 'metric',
                                                 'label': '지표',
                                                 'why': '무엇을 어떻게 나눌지 정해요.',
                                                 'options': [{'text': 'distance별로 표본 수, grade가 show인 비율, show의 정답률, grade가 check인 건 중 reviewed 비율, '
                                                                      '평균 ms를 표로 만들어 줘. 표본이 30건 미만이면 ‘표본 적음’으로 표시해 줘.',
                                                              'good': True},
                                                             {'text': '조건별 결과를 보여 줘.', 'good': False},
                                                             {'text': '', 'good': False}],
                                                 'miss': {'say': '거리별 정답률을 표로 만들었어요.',
                                                          'code': 'import pandas as pd\n'
                                                                  "df = pd.read_csv('{VDIR}/m10_run.csv')\n"
                                                                  "display(df.groupby('distance')['correct'].mean().round(2).reset_index())",
                                                          'problem': 'far 구간의 값이 얼마나 믿을 만한지 알 수 없어요. 표본 수와 교관 확인 비율이 빠졌어요.',
                                                          'hint': '지표와 표본 수, 표본이 적을 때의 표시를 적어 주세요.'}},
                                                {'id': 'report',
                                                 'label': '브리핑',
                                                 'why': '독자와 조건 표시, 점검을 정해요.',
                                                 'options': [{'text': '교육 담당이 읽을 1쪽 HTML 브리핑을 report 변수에 담아 줘. 순서는 결론 한 줄, 조건별 표, 그림, 한계, 운영 인계표(새 '
                                                                      '형상 등록·콘텐츠 검수·배포 승인·장애 대응의 담당과 주기), AI 산출물 표기야. 수치에는 가상 기록 몇 건 기준인지 적고, 표본 수와 '
                                                                      '담당, AI 산출물 표기가 빠졌는지 검사하는 lint_report 함수도 만들어 줘.',
                                                              'good': True},
                                                             {'text': '배포해도 된다고 브리핑 써 줘.', 'good': False},
                                                             {'text': '', 'good': False}],
                                                 'miss': {'say': '배포해도 된다는 브리핑을 썼어요.',
                                                          'code': "report = '<h2>배포 승인 요청</h2><p>본 학습앱은 정확도가 매우 우수하여 모든 상황에서 안심하고 쓸 수 있습니다.</p>'\n"
                                                                  "print('브리핑을 만들었어요.')",
                                                          'problem': '조건도 숫자도 없이 ‘모든 상황’이라고 단정했어요. 원하는 결론을 프롬프트에 적으면 AI는 거기에 맞춰 줘요. far 구간은 실제로 조심해야 '
                                                                     '하는데도요.',
                                                          'hint': '결론을 정해 주지 말고, 독자·순서·조건 표시와 점검을 요청해 주세요.'}}],
                                      'done': {'say': '설정값과 함수로 나눈 미니 도구예요. 파일 이름만 바꾸면 다음 달에도 쓸 수 있어요.',
                                               'code': 'import pandas as pd\n'
                                                       'import matplotlib.pyplot as plt\n'
                                                       '\n'
                                                       '# ===== 설정값 =====\n'
                                                       "FILE = '{VDIR}/m10_run.csv'\n"
                                                       'MIN_SAMPLES = 30\n'
                                                       "DIST = ['near', 'mid', 'far']\n"
                                                       "HANDOVER = [{'일': '새 형상 등록', '담당': '나라벨', '주기': '새 형상 확인 시'}, {'일': '콘텐츠 검수', '담당': '고교관', "
                                                       "'주기': '월 1회'},\n"
                                                       "            {'일': '배포 승인', '담당': '윤훈련', '주기': '배포 때마다'}, {'일': '장애 대응', '담당': '전산호', '주기': "
                                                       "'발생 시'}]\n"
                                                       '\n'
                                                       'def result_table(df):\n'
                                                       '    rows = []\n'
                                                       '    for d in DIST:\n'
                                                       "        s = df[df['distance'] == d]\n"
                                                       "        show = s[s['grade'] == 'show']\n"
                                                       "        check = s[s['grade'] == 'check']\n"
                                                       "        rows.append({'거리': d, '표본': len(s), '참고 판정 표시(%)': round(len(show) / len(s) * 100, "
                                                       '1),\n'
                                                       "                     '표시 건 정답률(%)': round(show['correct'].mean() * 100, 1) if len(show) else "
                                                       "'자료 없음',\n"
                                                       "                     '교관 확인 비율(%)': round(check['reviewed'].mean() * 100, 1) if len(check) "
                                                       "else '자료 없음',\n"
                                                       "                     '평균 ms': round(s['ms'].mean(), 1), '비고': '표본 적음' if len(s) < "
                                                       "MIN_SAMPLES else ''})\n"
                                                       '    return pd.DataFrame(rows)\n'
                                                       '\n'
                                                       'def make_report(t, n):\n'
                                                       "    low = t[t['비고'] != '']['거리'].tolist()\n"
                                                       '    limit = f"표본이 적은 구간: {\', \'.join(low)}. 이 구간의 수치는 참고만 해요." if low else \'표본이 적은 구간은 '
                                                       "없어요.'\n"
                                                       "    far = t[t['거리'] == 'far'].iloc[0]\n"
                                                       '    return (f"<h2>배포 점검 1쪽 브리핑</h2><p><b>결론:</b> far에서는 참고 판정 표시가 {far[\'참고 판정 표시(%)\']}%로 '
                                                       '적어 교관 확인을 전제로 운영해요.</p>"\n'
                                                       "            + f'<p>가상 기록 {n}건 기준이에요.</p>' + t.to_html(index=False)\n"
                                                       "            + f'<h3>한계</h3><p>{limit} 시범 운영 범위 밖의 조건은 확인하지 못했어요.</p>'\n"
                                                       "            + '<h3>운영 인계표</h3>' + pd.DataFrame(HANDOVER).to_html(index=False)\n"
                                                       "            + '<p><i>AI 산출물이에요. 담당자가 확인한 뒤에 배포해요.</i></p>')\n"
                                                       '\n'
                                                       'def lint_report(html):\n'
                                                       "    need = {'표본 수': '건 기준', '한계': '한계', '담당': '담당', 'AI 산출물 표기': 'AI 산출물'}\n"
                                                       '    return [k for k, v in need.items() if v not in html]\n'
                                                       '\n'
                                                       'df = pd.read_csv(FILE)\n'
                                                       'table = result_table(df)\n'
                                                       'display(table)\n'
                                                       '\n'
                                                       "ax = table.plot(x='거리', y='참고 판정 표시(%)', kind='bar', legend=False, figsize=(5, 2.6), rot=0)\n"
                                                       "ax.set_xlabel('distance')\n"
                                                       "ax.set_ylabel('shown (%)')\n"
                                                       '\n'
                                                       'report = make_report(table, len(df))\n'
                                                       'missing = lint_report(report)\n'
                                                       "print('자동 점검:', '빠진 항목 없음' if not missing else missing)",
                                               'lesson': '결론을 정해 주지 않고 조건을 요구하니, 어떤 구간을 조심해야 하는지 드러났어요. 인계표로 누가 무엇을 맡는지 남았고, 자동 점검이 있어 빠뜨림도 줄어요.'},
                                      'on_ok': [['dm_hun', 'hun', '{call}, 조건과 담당까지 나왔네요. 이제 전산·보안 담당 의견을 볼까요?']]}]},
                         {'short': '배포 점검',
                          'title': '담당별 의견과 올리기 전 점검',
                          'lead': '브리핑을 올리기 전에 **사람이 마지막으로 확인**해요.',
                          'blocks': [{'id': 's2_decide',
                                      'type': 'decide',
                                      'title': '배포 점검 의견 판단하기',
                                      'text': '브리핑을 받은 사람들의 의견이에요. 받아들일지 정하고 이유를 적어 주세요.',
                                      'reason_min': 8,
                                      'items': [{'id': 'a',
                                                 'who': 'hun',
                                                 'text': '‘모든 거리에서 안전함’이라고 단정해 주세요. 읽는 사람이 헷갈려요.',
                                                 'answer': 'reject',
                                                 'hint': '표의 수치가 단정을 뒷받침하나요?',
                                                 'explain': '근거 없이 단정하면 확신이 부풀어요. 쉬운 말로 쓰되 조건은 남겨요.',
                                                 'reason': '수치가 단정을 뒷받침하지 않아 쉬운 말로 쓰되 조건을 남겨요.'},
                                                {'id': 'b',
                                                 'who': 'boan',
                                                 'text': '앱 업데이트는 승인받은 파일만 인터넷 없이 넣게 하고, 담당과 경로를 인계표에 적어 주세요.',
                                                 'answer': 'accept',
                                                 'hint': '오프라인 환경에서 파일은 어떻게 들어갈까요?',
                                                 'explain': '경로와 담당을 적어 두면 다음 사람도 같은 방식으로 할 수 있어요. 프롬프트에 항목만 더하면 돼요.',
                                                 'reason': '경로와 담당을 남겨야 다음 사람도 같은 절차로 할 수 있어요.'}]},
                                     {'id': 's2_multi',
                                      'type': 'multi',
                                      'title': '올리기 전에 확인할 것',
                                      'text': 'AI가 만든 브리핑을 올리기 전에 꼭 확인할 것을 모두 골라 주세요.',
                                      'items': [{'text': '표의 숫자가 파일에서 다시 계산한 값과 같은지 한두 개 대조한다.',
                                                 'correct': True,
                                                 'explain': '한두 칸만 대조해도 큰 실수를 잡을 수 있어요.',
                                                 'hint': '숫자가 틀리면 어떻게 될까요?'},
                                                {'text': '수치마다 데이터와 표본 수, 조건이 적혀 있는지 본다.',
                                                 'correct': True,
                                                 'explain': '조건이 없으면 숫자가 오해되기 쉬워요.',
                                                 'hint': '표본이 적은 구간을 떠올려 보세요.'},
                                                {'text': '문장이 매끄러우면 내용도 맞으니 건너뛴다.',
                                                 'correct': False,
                                                 'explain': '매끄러운 문장과 맞는 내용은 별개예요.',
                                                 'hint': '표현이 매끄러운 것과 사실이 맞는 것은 같을까요?'},
                                                {'text': '실루엣과 자료의 출처·라이선스 표기가 빠지지 않았는지 본다.',
                                                 'correct': True,
                                                 'explain': '배포할 때 꼭 확인할 항목이에요. 인계표에도 적어 두면 좋아요.',
                                                 'hint': '다른 사람의 자료를 썼다면 무엇을 적어야 할까요?'},
                                                {'text': 'AI가 만든 문서이니 작성 책임은 AI에게 있다고 적는다.',
                                                 'correct': False,
                                                 'explain': '문서의 책임은 사람에게 있어요. AI는 도구예요.',
                                                 'hint': '브리핑을 올리는 사람은 누구인가요?'}]}]},
                         {'short': '프롬프트 모음',
                          'title': '나만의 프롬프트 틀 만들기',
                          'lead': '이번 프로젝트의 **가장 큰 산출물**이에요. 다음 업무에서 바로 쓸 나만의 틀을 남겨요.',
                          'blocks': [{'id': 's3_write',
                                      'type': 'write',
                                      'title': '내 프롬프트 틀 쓰기',
                                      'prompt': '앞으로 자주 쓸 **프롬프트 틀**을 써 보세요. 파일 이름이나 숫자처럼 바뀌는 자리는 [대괄호]로 비워 두세요. 네 칸에 더해 검증, 범위, 민감 정보 내용도 넣어 보세요.',
                                      'example': '[업무 맥락]을 위해 [무엇]을 확인하려고 해. 값은 보여 주지 않을게. 파일은 [파일 이름]이고 열은 [열 이름과 값의 종류]야. [계산 규칙과 기준값]으로 계산해 줘. '
                                                 '[예외]는 [처리 방법]으로 해 줘. 결과는 [표의 모양]으로 보여 주고, 다시 셀 수 있는 숫자와 표본 수를 함께 적어 줘. 먼저 [작은 예시]로 확인한 뒤 전체에 적용해 '
                                                 '줘. 고치는 건 [범위]만 하고, 원본 파일은 건드리지 마. 문서는 [독자]가 읽을 [순서]로 만들고 없는 내용은 ‘자료 없음’으로 적어 줘. AI 산출물이라는 표시도 넣어 '
                                                 '줘.',
                                      'min': 40,
                                      'checklist': ['맥락, 입력, 할 일, 출력 형식이 있어요',
                                                    '숫자 규칙과 예외 자리가 있어요',
                                                    '근거 수치와 검증을 요청했어요',
                                                    '수정 범위와 민감 정보 내용이 있어요',
                                                    '다음 업무에 바로 쓸 수 있어요'],
                                      'model_label': '📝 모범 프롬프트',
                                      'placeholder': '꼭 입력해 주세요 · 내 말로 프롬프트를 써 보세요',
                                      'explain': '잘 통한 프롬프트는 저장해 두세요. 쓸수록 나에게 맞는 틀이 돼요.'}]}],
               'outputs': [{'id': 'm10_handover',
                            'title': 'M10 1쪽 브리핑·운영 인계표와 프롬프트 틀',
                            'sections': [{'h': '1. 인계 도구와 브리핑', 'blocks': ['s1_vibe']},
                                         {'h': '2. 배포 점검 의견 판단', 'blocks': ['s2_decide']},
                                         {'h': '3. 올리기 전 점검', 'blocks': ['s2_multi']},
                                         {'h': '4. 나만의 프롬프트 틀', 'blocks': ['s3_write']}]}]}]}   # 프로젝트·미션 명세 (만드는 도구가 넣음). 정답·힌트도 여기에 있어 화면(브라우저)으로는 보내지 않음
_P = SPEC["project"]
PROJECT = {"title": _P["title"], "goal": _P["goal"], "stages": _P["stages"]}
TEAM_NAME = _P["team_name"]            # 학습자가 합류하는 가상 조직 이름 (예: 드론분석 TF)
MY_ROLE = _P["my_role"]                # 학습자의 역할 (예: 신입 분석관)
MISSION_INFO = {m["id"]: (m["no"], m["title"], m["short"], m["kind"], m["outline"]) for m in SPEC["missions"]}   # (번호, 제목, 짧은 이름, 형태, 소개)
IMPLEMENTED = [m["id"] for m in SPEC["missions"] if m.get("steps")]

# ---------------------------------------------------------------
# 학습자 호칭: 모두 '전우님' (개인 정보를 묻지 않음. 군 체계를 추정할 정보도 남지 않음)
# ---------------------------------------------------------------
CALL_NAME = _P.get("call_default") or "전우님"


def call_name():
    """학습자를 부르는 말."""
    return CALL_NAME


def me_label():
    return CALL_NAME


MEMBERS = {k: tuple(v) for k, v in _P["members"].items()}   # 아이디: (이름, 소속, 역할, 아바타 색)
CHANNELS = dict(_P["channels"])
ALWAYS_CHANNELS = tuple(_P["always_channels"])

# ---------------------------------------------------------------
# 미션 등록부 (미션 파일들이 채움)
# ---------------------------------------------------------------
BRIEFINGS = {}      # 미션 소개
VIEWS = {}          # 화면에 보낼 미션별 상태
RESETTERS = {}      # 미션별 상태 초기화
STARTERS = {}       # 미션 시작 때 오는 메시지
CALLBACKS = {}      # 화면 버튼 → 파이썬 함수
CODE_TOPICS = {}    # '본 과정의 코드 보기'에 보여 줄 함수

# ---------------------------------------------------------------
# 진행 상태
# ---------------------------------------------------------------
S = {}


def _reset_state():
    S.clear()
    S.update(msgs=[], minute=0, files=[],
             progress={m: {"started": False, "step": 0, "done": False, "issued": False} for m in MISSION_INFO})
    for reset in RESETTERS.values():
        reset()
    _welcome()
    if FILL_ANSWERS:                 # 선생님용: 빈 값 없이 모든 활동을 정답으로 채우고 산출물까지 만들어 둠
        _prefill_all()


def _welcome():
    for ch, who, text in _P["welcome"]:
        post(ch, who, text)


def post(ch, who, text):
    """메신저에 메시지 남기기. {call}은 학습자를 부르는 말(전우님)로 바뀝니다."""
    t = 9 * 60 + S["minute"]
    S["minute"] += 2
    S["msgs"].append({"id": len(S["msgs"]), "ch": ch, "who": who, "time": f"{t // 60:02d}:{t % 60:02d}", "text": text})


def _fmt(text):
    return text.replace("{call}", call_name())


def _advance(mid, step):
    p = S["progress"][mid]
    p["step"] = max(p["step"], step)


def _status(mid):
    """잠금 없음: 준비된 미션은 언제나 열려 있음. 필수 활동을 다 하면 done."""
    if mid not in IMPLEMENTED:
        return "soon"
    return "done" if S["progress"][mid]["done"] else "open"


def overall_progress():
    """전체 진행률: 미션마다 끝낸 필수 활동의 비율을 평균."""
    units = sum(_mission_fraction(mid) for mid in MISSION_INFO if mid in BRIEFINGS)
    done = sum(S["progress"][m]["done"] for m in MISSION_INFO)
    return {"pct": round(100 * units / len(MISSION_INFO)), "done": done, "total": len(MISSION_INFO)}


def _add_file(fid, title, mission, doc, blanks=0):
    S["files"] = [f for f in S["files"] if f["id"] != fid]
    S["files"].append({"id": fid, "name": f"{fid}.html", "title": title, "mission": mission, "html": doc, "blanks": blanks})


DOC_CSS = ("body{font-family:sans-serif;max-width:860px;margin:24px auto;color:#111827;background:#fff;line-height:1.6;padding:0 16px}"
           "h1{font-size:22px;border-bottom:2px solid #111827;padding-bottom:6px}h2{font-size:16px;margin-top:22px}"
           "table{border-collapse:collapse;margin:8px 0}td,th{border:1px solid #c8ced6;padding:5px 10px;font-size:13px;text-align:left;vertical-align:top;color:#111827}"
           "th{background:#eef2f7}.meta{color:#111827;font-size:13px;background:#f3f6fa;padding:6px 10px;border-radius:6px}"
           ".note{color:#111827;font-size:13px}.ok{color:#146c2e;font-weight:bold}.bad{color:#b3261e;font-weight:bold}"
           ".cards{display:grid;grid-template-columns:repeat(auto-fill,minmax(240px,1fr));gap:12px}"
           ".card{border:1px solid #c8ced6;border-radius:8px;padding:10px}.card img{width:48%;border-radius:4px}"
           ".imgs img{max-width:100%;border:1px solid #c8ced6;border-radius:6px}.chg{background:#fff4ce}"
           "pre{background:#f3f6fa;padding:8px 10px;border-radius:6px;white-space:pre-wrap}"
           ".blank{display:inline-block;min-width:180px;height:1.1em;border-bottom:1px solid #6b7280}"
           ".blankbox{min-height:54px;border:1px dashed #6b7280;border-radius:6px;margin:6px 0}"
           "td.empty{min-width:120px;height:1.6em}img{max-width:100%}")


def _doc(title, body):
    return (f"<!doctype html><html><head><meta charset='utf-8'><title>{html.escape(title)}</title><style>{DOC_CSS}</style></head>"
            f"<body><h1>{html.escape(title)}</h1><p class='meta'>프로젝트: {html.escape(PROJECT['title'])}<br>"
            f"{html.escape(TEAM_NAME)} · 작성일: {date.today():%Y-%m-%d}</p>{body}</body></html>")


# ---------------------------------------------------------------
# 본 과정의 코드 보기
# ---------------------------------------------------------------
def code_topic(key, title, desc, funcs, marks=()):
    """화면의 한 단계에서 쓰이는 함수 목록을 등록. marks에 든 글자가 있는 줄은 강조 표시됩니다."""
    CODE_TOPICS[key] = {"title": title, "desc": desc, "funcs": list(funcs), "marks": list(marks)}


def call_str(name, *args, **kw):
    """함수 호출을 코드 한 줄로 적기 (예: simulate('m1', level=0.5))."""
    parts = [repr(a) for a in args] + [f"{k}={v!r}" for k, v in kw.items()]
    return f"{name}({', '.join(parts)})"


def app_code(topic):
    t = CODE_TOPICS.get(topic)
    if not t:
        return JSON({"ok": False, "blocks": []})
    blocks = []
    for name in t["funcs"]:
        fn = globals().get(name)
        try:
            src = textwrap.dedent(inspect.getsource(fn)).rstrip()
            doc = (inspect.getdoc(fn) or "").split("\n")[0]
        except Exception:
            src, doc = f"# {name}: 코드를 불러오지 못했어요.", ""
        lines = src.splitlines()
        hl = [i for i, line in enumerate(lines) if any(m in line for m in t["marks"])]
        blocks.append({"name": name, "doc": doc, "code": src, "hl": hl})
    return JSON({"ok": True, "topic": topic, "title": t["title"], "desc": t["desc"], "blocks": blocks})


# ---------------------------------------------------------------
# 화면에 보낼 상태
# ---------------------------------------------------------------
def _mission_meta(mid):
    no, title, short, kind, outline = MISSION_INFO[mid]
    stage = next(s["no"] for s in PROJECT["stages"] if mid in s["missions"])
    meta = {"id": mid, "no": no, "stage": stage, "title": title, "short": short, "kind": kind,
            "outline": outline, "status": _status(mid), "progress": S["progress"][mid]}
    if mid in BRIEFINGS and mid in IMPLEMENTED:
        meta["brief"] = BRIEFINGS[mid]
    return meta


def app_state():
    members = {}
    for k, (n, o, r, c) in MEMBERS.items():
        members[k] = {"name": n, "org": o, "role": r, "color": c, "ext": k != "me" and o != TEAM_NAME}
    used = {m["ch"] for m in S["msgs"]}
    return {
        "call": call_name(),
        "project": {"title": PROJECT["title"], "goal": PROJECT["goal"],
                    "stages": [{"no": s["no"], "name": s["name"], "subtitle": s.get("subtitle", ""), "missions": s["missions"]} for s in PROJECT["stages"]]},
        "missions": {mid: _mission_meta(mid) for mid in MISSION_INFO},
        "views": {mid: fn() for mid, fn in VIEWS.items()},
        "members": members,
        "channels": {k: v for k, v in CHANNELS.items() if k in ALWAYS_CHANNELS or k in used},
        "msgs": [{**m, "text": m["text"] if m["who"] == "me" else _fmt(m["text"])} for m in S["msgs"]],
        "files": [{k: f.get(k) for k in ("id", "name", "title", "mission", "blanks")} for f in S["files"]],
        "overall": overall_progress(), "boot": BOOT_ID,
        "teacher": bool(FILL_ANSWERS),        # 선생님용: 값이 모두 채워진 상태 (화면에 표시만 함)
        "assets": assets_status(),            # 참고자료(사진)가 필요한 프로젝트면 폴더와 빠진 파일을 알려 줌
    }


def _resp(**kw):
    return JSON({"ok": True, **kw, "app": app_state()})


def app_get():
    return JSON({"ok": True, "app": app_state()})


def app_start(mid):
    if _status(mid) == "soon":
        return _resp(ok=False, msg="아직 준비 중인 미션이에요.")
    _ensure_started(mid)
    return _resp(ok=True)


def _ensure_started(mid):
    """미션을 처음 열 때 한 번: 시작 메시지를 보냄."""
    p = S["progress"][mid]
    if not p["started"]:
        p["started"] = True
        _advance(mid, 1)
        if mid in STARTERS:
            STARTERS[mid]()


def app_send(ch, text):
    text = (text or "").strip()
    if text:
        post(ch, "me", text)
    return _resp(ok=True)


def app_file(fid):
    f = next((f for f in S["files"] if f["id"] == fid), None)
    return JSON({"ok": bool(f), "html": f["html"] if f else "", "name": f["name"] if f else ""})


def app_reset():
    _reset_state()
    return _resp(ok=True, msg="처음 상태로 돌아갔어요.")


CALLBACKS.update({"app.state": app_get, "app.start": app_start, "app.send": app_send,
                  "app.file": app_file, "app.reset": app_reset, "app.code": app_code})
code_topic("app.chat", "메신저는 이렇게 동작해요", "팀원 메시지는 미션 진행에 맞춰 파이썬 함수 post()가 남깁니다. "
           "화면의 버튼을 누르면 google.colab의 invokeFunction이 파이썬 함수를 부르고, 결과(JSON)로 화면을 다시 그려요.",
           ["post", "app_send"], marks=["S[\"msgs\"].append", "post(ch"])
code_topic("app.mission", "미션 시작은 이렇게 동작해요", "[미션 시작]을 누르면 app_start()가 진행 상태를 바꾸고, "
           "미션마다 등록된 시작 메시지(STARTERS)를 보냅니다.", ["app_start", "_ensure_started"], marks=["STARTERS"])
code_topic("app.files", "산출물은 이렇게 만들어져요", "산출물을 받으면 입력한 내용으로 HTML 문서를 만들어 파일 목록에 저장합니다. "
           "하지 않은 활동은 빈칸으로 남아요.", ["_doc", "_add_file"], marks=["S[\"files\"]"])


# ---------------------------------------------------------------
# 뒤에서 도는 작업 (학습 · 계산 · 내려받기): 화면이 1.5초마다 app.job으로 진행을 물어봄
# 오래 걸리는 작업 중 다른 화면을 열면 Colab 런타임이 끊기는 일이 있어서, 명세의 stay_jobs에 적은 작업은 화면 이동을 잠금
# ---------------------------------------------------------------
JOBS = {}


def start_job(name, fn, *args):
    """작업을 뒤에서(스레드) 실행. 이미 같은 작업이 돌고 있으면 False."""
    if JOBS.get(name, {}).get("state") == "running":
        return False
    job = {"state": "running", "pct": 1, "msg": "시작하는 중...", "t0": time.time(), "error": None}
    JOBS[name] = job

    def run():
        try:
            fn(job, *args)
            job.update(state="done", pct=100)
        except Exception as e:
            job.update(state="error", error=f"{type(e).__name__}: {e}", trace=traceback.format_exc()[-2500:])
    threading.Thread(target=run, daemon=True).start()
    return True


def _say(job, pct, msg):
    job["pct"], job["msg"] = max(job.get("pct", 0), int(pct)), msg


def app_job(name):
    """화면이 2초마다 묻는 작업 진행 상황. 끝나면 새 화면 상태를 함께 보냄."""
    j = JOBS.get(name)
    if not j:
        return JSON({"ok": True, "state": "none"})
    out = {"ok": True, "state": j["state"], "pct": j["pct"], "msg": j["msg"], "sec": int(time.time() - j["t0"]), "error": j.get("error"),
           "data": j.get("data")}
    if j["state"] != "running":
        out["app"] = app_state()
    return JSON(out)


# ---------------------------------------------------------------
# 그림 도우미: 화면에 보낼 영상을 데이터 주소로
# ---------------------------------------------------------------
def to_data_url(img, fmt="JPEG"):
    """PIL 영상 → 데이터 주소 (JPEG는 RGB로 바꿔 저장)."""
    buf = io.BytesIO()
    (img.convert("RGB") if fmt.upper() == "JPEG" else img).save(buf, format=fmt, quality=90)
    return f"data:image/{fmt.lower()};base64," + base64.b64encode(buf.getvalue()).decode()


def _img_url(img, width=None):
    """화면에 보낼 영상: 폭을 맞춰(픽셀이 보이게 NEAREST) JPEG 데이터 주소로."""
    if width and img.size[0] != width:
        from PIL import Image
        img = img.resize((width, max(1, round(width * img.size[1] / img.size[0]))), Image.NEAREST)
    return to_data_url(img)


CALLBACKS["app.job"] = app_job
# ===== 참고자료(사진): 인터넷에서 받지 않음. 제작자가 따로 준 사진을 학습자가 폴더에 넣고, 블록의 "asset:번호"로 씀 =====
# 명세의 project.assets: {"folder"(사진을 넣을 폴더 이름), "items"(번호·파일·제목·저작자·라이선스)}
# 정답은 여기에 없음. 사진은 노트북에 담지 않고 빌드 도구가 <코드>_images 폴더·zip으로 따로 내줌.
import shutil, zipfile

ASSET_CFG = _P.get("assets") or {}
ASSET_ITEMS = {it["id"]: it for it in ASSET_CFG.get("items", [])}
ASSET_STATE = {"ok": not ASSET_ITEMS, "msg": ""}
_ASSET_URL_CACHE = {}


def asset_folder():
    """사진을 넣는 폴더: Colab은 /content/<folder>, 점검 도구는 PBL_ASSET_LOCAL(자료 폴더)이나 PBL_ASSET_ROOT(임시 폴더)."""
    local = os.environ.get("PBL_ASSET_LOCAL")
    if local:
        return local
    root = os.environ.get("PBL_ASSET_ROOT") or ("/content" if os.path.isdir("/content") else os.getcwd())
    return os.path.join(root, ASSET_CFG.get("folder") or "pbl_photos")


def _asset_missing():
    folder = asset_folder()
    return [it["file"] for it in ASSET_ITEMS.values() if not os.path.exists(os.path.join(folder, it["file"]))]


def _extract_zips(folder):
    """폴더에 넣은 zip을 풀어 사진을 폴더 바로 아래로 모음 (zip 안에 폴더가 한 겹 있어도 됨)."""
    for name in os.listdir(folder):
        if not name.lower().endswith(".zip"):
            continue
        tmp = os.path.join(folder, "_unzip_tmp")
        shutil.rmtree(tmp, ignore_errors=True)
        with zipfile.ZipFile(os.path.join(folder, name)) as z:
            z.extractall(tmp)
        for dirpath, _, files in os.walk(tmp):
            for f in files:
                shutil.move(os.path.join(dirpath, f), os.path.join(folder, f))
        shutil.rmtree(tmp, ignore_errors=True)


def assets_prepare():
    """사진 확인: 폴더를 만들어 두고, 넣어 둔 zip이 있으면 풀고, 빠진 사진을 ASSET_STATE에 남김. 다시 불러도 안전함."""
    if not ASSET_ITEMS:
        return True
    folder = asset_folder()
    os.makedirs(folder, exist_ok=True)
    try:
        _extract_zips(folder)
    except Exception as e:
        ASSET_STATE.update(ok=False, msg=f"zip을 풀지 못했어요({type(e).__name__}). 사진 파일을 그대로 넣어 주세요.")
        return False
    missing = _asset_missing()
    _ASSET_URL_CACHE.clear()
    if missing:
        ASSET_STATE.update(ok=False, msg=f"참고자료(사진) {len(missing)}개가 아직 없어요. 왼쪽 📁 파일 패널에서 {folder} 폴더에 넣어 주세요.")
        return False
    ASSET_STATE.update(ok=True, msg="참고자료(사진)를 확인했어요.")
    return True


def assets_status():
    """화면에 보낼 사진 상태: 필요 없는 프로젝트면 None."""
    if not ASSET_ITEMS:
        return None
    return {"ok": ASSET_STATE["ok"], "folder": asset_folder(), "missing": _asset_missing(), "total": len(ASSET_ITEMS), "msg": ASSET_STATE["msg"]}


def app_assets_check():
    """화면의 [다시 확인] 버튼: 사진을 넣은 뒤 다시 찾음."""
    ok = assets_prepare()
    return _resp(ok=ok, msg=ASSET_STATE["msg"])


def asset_path(aid):
    """참고자료 파일 경로 (없으면 None). 직접 만든 활동에서 원본 파일을 읽을 때 씀 (예: 출처 정보 확인)."""
    it = ASSET_ITEMS.get(str(aid))
    if not it:
        return None
    p = os.path.join(asset_folder(), it["file"])
    return p if os.path.exists(p) else None


def asset_data_url(aid, width=960):
    """화면에 보낼 이미지: 폭을 맞춘 JPEG 데이터 주소 (한 번 만들면 캐시). PIL이 없으면 파일을 그대로 보냄."""
    key = (str(aid), int(width))
    if key not in _ASSET_URL_CACHE:
        path = asset_path(aid)
        try:
            from PIL import Image
            im = Image.open(path).convert("RGB")
            if im.size[0] > width:
                im = im.resize((width, max(1, round(width * im.size[1] / im.size[0]))), Image.LANCZOS)
            _ASSET_URL_CACHE[key] = to_data_url(im)
        except ImportError:
            mime = "png" if path.lower().endswith(".png") else "jpeg"
            with open(path, "rb") as f:
                _ASSET_URL_CACHE[key] = f"data:image/{mime};base64," + base64.b64encode(f.read()).decode()
    return _ASSET_URL_CACHE[key]


def app_asset(aid, width=960):
    """화면의 이미지 요청 ("asset:번호"). 화면 상태 전체는 보내지 않음 (가볍게)."""
    it = ASSET_ITEMS.get(str(aid))
    if not it:
        return JSON({"ok": False, "msg": f"없는 참고자료예요: {aid}"})
    if not asset_path(aid):
        return JSON({"ok": False, "missing": True, "file": it["file"], "folder": asset_folder(),
                     "msg": f"사진을 찾지 못했어요: {os.path.join(asset_folder(), it['file'])}"})
    try:
        return JSON({"ok": True, "url": asset_data_url(aid, max(240, min(int(width), 2400))), "caption": ""})
    except Exception as e:
        return JSON({"ok": False, "msg": f"이미지를 열지 못했어요: {type(e).__name__}"})


def _asset_refs(obj, out=None):
    """명세 안에서 "asset:번호"를 모두 찾음 (산출물의 출처 목록용)."""
    out = [] if out is None else out
    if isinstance(obj, str) and obj.startswith("asset:"):
        if obj[6:] not in out:
            out.append(obj[6:])
    elif isinstance(obj, (list, tuple)):
        for x in obj:
            _asset_refs(x, out)
    elif isinstance(obj, dict):
        for x in obj.values():
            _asset_refs(x, out)
    return out


def asset_credits_html(ids):
    """산출물 문서에 붙는 '사용한 자료와 출처' 표 (주소 없이 이름만: 답이 새지 않게, 링크를 달지 않게)."""
    rows = []
    for aid in ids:
        it = ASSET_ITEMS.get(aid)
        if not it:
            continue
        rows.append(f"<tr><td>{html.escape(aid)}</td><td>{html.escape(it.get('title', ''))}</td><td>{html.escape(it.get('credit', ''))}</td>"
                    f"<td>{html.escape(it.get('license', ''))}</td><td>{html.escape(it.get('changes', ''))}</td></tr>")
    if not rows:
        return ""
    return ("<h2>사용한 자료와 출처</h2><table><tr><th>번호</th><th>자료</th><th>저작자·제작</th><th>라이선스</th><th>변경</th></tr>"
            + "".join(rows) + "</table>")


CALLBACKS.update({"app.asset": app_asset, "app.assets_check": app_assets_check})
_setup(30, 1, "메신저 · 진행 상태 · 가이드북 준비 완료")

# ===== 미션 단계 엔진: 명세(SPEC)의 블록으로 미션을 진행하고, 채점·힌트·산출물을 만듦 =====
# 블록 종류: info(읽기) · quiz(확인 질문) · multi(여러 개 고르기) · match(짝짓기) · order(순서 맞추기)
#            classify(사례 분류) · write(글쓰기) · decide(받아들이기/거절 결정) · vibe(바이브 코딩 실습) · custom(직접 만든 활동)
# 원칙 (시행착오에서 정한 것)
#  - 정답·풀이·힌트는 파이썬에만 둠. 화면에는 학습자가 맞힌 뒤에만 풀이를 보냄
#  - 틀리면 힌트를 한 단계씩 자세하게. 맞혀도 화면이 저절로 넘어가지 않음 (학습자가 [다음 단계]를 누름)
#  - 처음부터 차례로 하는 것이 원칙이지만, 뒷부분을 미리 보려는 사람을 위해 잠그지 않음 (2026-11-05 피드백)
#  - 산출물은 언제든 받을 수 있음: 안 한 활동이 있으면 먼저 알려 주고 물어본 뒤, 그래도 받으면 그 부분은 빈칸으로 만듦
#  - 선생님용(FILL_ANSWERS)은 모든 활동의 값이 채워진 상태로 시작함
#  - 글쓰기(write)는 정답을 판정하지 않음: 최소 글자 수만 보고 제출을 받은 뒤 모범답안(example)을 보여 줌
CUSTOM_VIEWS = {}     # 미션 아이디 → 화면에 더 보낼 상태를 돌려주는 함수 (직접 만든 활동용)
CUSTOM_RESET = {}     # 미션 아이디 → 처음 상태를 만드는 함수
CUSTOM_DOCS = {}      # (미션, 블록) → 산출물 문서에 넣을 HTML을 돌려주는 함수
AUTOPLAY = {}         # (미션, 블록) → 점검 도구가 이 활동을 끝까지 해 보는 함수 (직접 만든 활동용)
DECIDE_OPTS = [["accept", "받아들임"], ["reject", "거절하고 대안 제시"]]
MSPEC = {m["id"]: m for m in SPEC["missions"]}
BSPEC = {}            # (미션, 블록) → (단계 번호, 블록 명세)
for _m in SPEC["missions"]:
    for _si, _st in enumerate(_m.get("steps", []), 1):
        for _b in _st.get("blocks", []):
            BSPEC[(_m["id"], _b["id"])] = (_si, _b)


def _required(b):
    """필수 블록인가: 읽기(info)는 confirm이 있을 때만, 나머지는 required가 False가 아니면 필수."""
    if b["type"] == "info":
        return bool(b.get("confirm"))
    return b.get("required", True)


def _bst(mid, bid):
    return S[mid]["blk"].setdefault(bid, {"ok": False, "tries": 0, "value": None, "item_tries": {}})


def _step_blocks(mid, si):
    st = MSPEC[mid]["steps"][si - 1]
    return st.get("blocks", [])


def _step_done(mid, si):
    """단계 완료: 필수 블록을 모두 마침 (블록 없이 직접 만든 화면(page)은 그 코드가 step_ok로 알림)."""
    st = MSPEC[mid]["steps"][si - 1]
    if st.get("page"):
        return bool(S[mid].get("pages", {}).get(si))
    return all(_bst(mid, b["id"])["ok"] for b in _step_blocks(mid, si) if _required(b))


def _todo_list(mid):
    """아직 안 한 필수 활동: [{"step": 단계 번호, "short": 단계 이름, "blocks": [활동 이름…]}]."""
    out = []
    for si, st in enumerate(MSPEC[mid]["steps"], 1):
        if st.get("page"):
            if not _step_done(mid, si):
                out.append({"step": si, "short": st["short"], "blocks": [st.get("title") or st["short"]]})
            continue
        left = [b.get("title") or BLOCK_NAMES.get(b["type"], "활동") for b in st.get("blocks", []) if _required(b) and not _bst(mid, b["id"])["ok"]]
        if left:
            out.append({"step": si, "short": st["short"], "blocks": left})
    return out


def _mission_fraction(mid):
    """끝낸 필수 활동의 비율 (0~1)."""
    done = total = 0
    for si, st in enumerate(MSPEC[mid]["steps"], 1):
        if st.get("page"):
            total += 1
            done += _step_done(mid, si)
            continue
        for b in st.get("blocks", []):
            if _required(b):
                total += 1
                done += _bst(mid, b["id"])["ok"]
    return done / total if total else 0.0


def _update_done(mid):
    """필수 활동을 다 하면 미션 완료 표시."""
    S["progress"][mid]["done"] = not _todo_list(mid)


def step_ok(mid, si):
    """직접 만든 화면(page) 단계를 마쳤다고 알림."""
    S[mid].setdefault("pages", {})[si] = True
    _advance(mid, si + 1)
    _update_done(mid)


def block_ok(mid, bid, value=None):
    """블록을 마쳤다고 기록: 처음 마쳤을 때 팀원 메시지(on_ok)를 보내고, 진행 상태를 고침."""
    si, b = BSPEC[(mid, bid)]
    st = _bst(mid, bid)
    first = not st["ok"]
    st["ok"] = True
    if value is not None:
        st["value"] = value
    if first:
        for ch, who, text in b.get("on_ok", []):
            post(ch, who, text)
    if _step_done(mid, si):
        _advance(mid, si + 1)
    _update_done(mid)


WRONG_MSG = "❌ 현재 선택을 다시 고민해 보세요."      # 틀린 선택 안내: 화면에서 빨간색으로 보임 (2026-11-05 피드백 "주황색 말고 빨간색으로")


def _hint(spec, tries, key="hints"):
    hs = spec.get(key) or ["다시 생각해 보세요."]
    k = max(1, min(tries, len(hs)))
    return f"{WRONG_MSG} 힌트 {k}/{len(hs)}: {hs[k - 1]}"


# ---- 블록마다 채점: (맞음?, 저장할 값, 화면에 보낼 것) ----
def _chk_info(mid, b, st, payload):
    return True, True, {"text": "✅ 확인했어요."}


def _chk_quiz(mid, b, st, payload):
    try:
        pick = int(payload)
    except (TypeError, ValueError):
        pick = -1
    if pick == b["answer"]:
        return True, {"pick": pick}, {"text": "✅ 맞아요. " + b["explain"]}
    st["tries"] += 1
    return False, None, {"text": _hint(b, st["tries"])}


def _chk_multi(mid, b, st, payload):
    picked = {int(i) for i in (payload or []) if str(i).lstrip("-").isdigit()}
    res = []
    for i, it in enumerate(b["items"]):
        ok = (i in picked) == bool(it["correct"])
        res.append({"i": i, "ok": ok, "text": it.get("explain", "") if ok else "❌ " + it.get("hint", "다시 생각해 보세요.")})
    wrong = sum(not r["ok"] for r in res)
    if wrong:
        st["tries"] += 1
        return False, None, {"items": res, "text": f"{WRONG_MSG} 다시 볼 항목이 {wrong}개 있어요."}
    return True, {"picked": sorted(picked)}, {"items": res, "text": "✅ 모두 맞게 골랐어요."}


def _chk_match(mid, b, st, payload):
    payload = payload or {}
    res = {p["left"]: str(payload.get(p["left"])) == str(p["right"]) for p in b["pairs"]}
    wrong = [p for p in b["pairs"] if not res[p["left"]]]
    if not wrong:
        return True, {p["left"]: p["right"] for p in b["pairs"]}, {"result": res, "hints": [], "text": "✅ 모두 맞게 짝지었어요."}
    st["tries"] += 1
    n, gen = st["tries"], b.get("hints") or ["틀린 칸을 다시 살펴보세요."]
    hints = [gen[0]]
    if n >= 2:
        hints += [f"‘{p['left']}’: {p['hint']}" for p in wrong if p.get("hint")]
    if n >= 3:
        hints += gen[1:]
    return False, None, {"result": res, "hints": hints, "text": f"{WRONG_MSG} 빨간 칸 {len(wrong)}개가 맞지 않아요."}


def _chk_order(mid, b, st, payload):
    right = [it["id"] for it in b["items"]]
    order = [o for o in (payload or []) if o in right]
    if order == right:
        return True, {"order": order}, {"text": "✅ 순서가 맞아요.", "wrong": None}
    if len(order) != len(right):
        return False, None, {"text": "모든 항목을 한 번씩 놓아 주세요.", "wrong": None}
    st["tries"] += 1
    n = st["tries"]
    i = next(k for k, (a, c) in enumerate(zip(order, right)) if a != c)
    it = b["items"][i]
    text = f"❌ 현재 순서를 다시 고민해 보세요. {i + 1}번째 자리가 맞지 않아요. " + (b.get("hints") or ["앞 단계의 결과가 다음 단계의 재료가 되는지 따져 보세요."])[0]
    if n >= 2 and it.get("hint"):
        text += f" 힌트: {it['hint']}"
    if n >= 3:
        text += f" ({i + 1}번째 자리: ‘{it['text']}’)"
    return False, None, {"text": text, "wrong": i}


def _chk_classify(mid, b, st, payload):
    payload = payload or {}
    it = next((x for x in b["items"] if x["id"] == payload.get("item")), None)
    if not it:
        return False, None, {"text": "항목을 찾지 못했어요.", "item": None}
    names = {c["id"]: c["name"] for c in b["categories"]}
    val = dict(st["value"] or {})
    prim = payload.get("primary")
    contrib = [c for c in (payload.get("contrib") or []) if c in names and c != prim]
    if prim in it["accept"]:
        val[it["id"]] = {"primary": prim, "contrib": contrib}
        miss = [names[c] for c in it.get("contrib", []) if c not in contrib and c != prim]
        text = "✅ " + it["explain"] + (f" (함께 생각해 볼 것: {', '.join(miss)})" if miss else "")
        return all(x["id"] in val for x in b["items"]), val, {"text": text, "item": it["id"], "item_ok": True}
    t = st["item_tries"][it["id"]] = st["item_tries"].get(it["id"], 0) + 1
    return False, val, {"text": _hint(it, t), "item": it["id"], "item_ok": False}


def _chk_write(mid, b, st, payload):
    """글쓰기: 정답을 판정하지 않음. 최소 글자 수를 채우면 제출을 받고 모범답안(example)을 보여 줌.
    (학습자가 뜻은 맞게 써도 규칙(정규식)에는 틀릴 수 있어서, 맞고 틀림을 따지지 않기로 함: 2026-10-19 피드백)"""
    text = (payload or "").strip()
    n = b.get("min", 10)
    if len(text) < n:
        return False, None, {"text": f"✏ {n}자 이상 써 주세요. (지금 {len(text)}자)"}
    return True, text, {"text": "✅ 제출했어요.", "model": b.get("example", ""), "explain": b.get("explain", ""), "checklist": b.get("checklist", [])}


def _chk_decide(mid, b, st, payload):
    payload = payload or {}
    ch, rs = payload.get("choices") or {}, payload.get("reasons") or {}
    opts = [o[0] for o in b.get("options", DECIDE_OPTS)]
    if any(ch.get(it["id"]) not in opts for it in b["items"]):
        return False, None, {"text": "항목마다 하나를 골라 주세요.", "items": []}
    rmin = b.get("reason_min", 10)
    short = [it["id"] for it in b["items"] if len((rs.get(it["id"]) or "").strip()) < rmin]
    if short:
        return False, None, {"text": f"이유를 {rmin}자 이상 적어 주세요. (남은 항목 {len(short)}개)", "items": []}
    wrong = {it["id"] for it in b["items"] if ch[it["id"]] != it["answer"]}
    items = [{"id": it["id"], "ok": it["id"] not in wrong,
              "text": it.get("explain", "") if it["id"] not in wrong else "❌ " + it.get("hint", "다시 생각해 보세요.")} for it in b["items"]]
    if wrong:
        st["tries"] += 1
        return False, None, {"text": f"{WRONG_MSG} 다시 볼 항목이 {len(wrong)}개 있어요.", "items": items}
    return True, {it["id"]: {"choice": ch[it["id"]], "reason": rs[it["id"]].strip()} for it in b["items"]}, {"text": "✅ 모두 알맞게 결정했어요.", "items": items}


CHECKERS = {"info": _chk_info, "quiz": _chk_quiz, "multi": _chk_multi, "match": _chk_match, "order": _chk_order,
            "classify": _chk_classify, "write": _chk_write, "decide": _chk_decide}


def blk_check(mid, bid, payload=None):
    """화면의 [확인] 버튼: 블록 하나를 채점. correct=True면 그 블록을 마친 것 (화면은 저절로 넘어가지 않음)."""
    if (mid, bid) not in BSPEC:
        return _resp(ok=False, msg="없는 활동이에요.")
    si, b = BSPEC[(mid, bid)]
    _ensure_started(mid)                                            # 잠그지 않으므로 미리 열어 본 단계에서도 동작함
    fn = CHECKERS.get(b["type"])
    if not fn:
        return _resp(ok=False, msg="이 활동은 화면의 안내대로 진행해요.")
    st = _bst(mid, bid)
    ok, value, extra = fn(mid, b, st, payload)
    if value is not None:
        st["value"] = value
    if ok:
        block_ok(mid, bid)
    return _resp(ok=True, correct=bool(ok), **extra)


# ---- 산출물 문서 ----
def _esc_br(t):
    return html.escape(str(t)).replace("\n", "<br>")


def _info_html(b):
    out = "".join(f"<p>{_md(p)}</p>" for p in b.get("text", []))
    if b.get("bullets"):
        out += "<ul>" + "".join(f"<li>{_md(x)}</li>" for x in b["bullets"]) + "</ul>"
    if b.get("table"):
        t = b["table"]
        out += "<table><tr>" + "".join(f"<th>{_md(c)}</th>" for c in t["head"]) + "</tr>" + "".join(
            "<tr>" + "".join(f"<td>{_md(c)}</td>" for c in row) + "</tr>" for row in t["rows"]) + "</table>"
    if b.get("sources"):
        out += "<p class='note'>출처: " + " · ".join(f"<a href='{html.escape(u)}'>{html.escape(n)}</a>" for n, u in b["sources"]) + "</p>"
    return out


def _md(t):
    """명세 글에서 **굵게**만 지원 (나머지는 그대로 이스케이프)."""
    return re.sub(r"\*\*(.+?)\*\*", r"<b>\1</b>", html.escape(str(t)))


BLANK = "<span class='blank'></span>"
BLANKBOX = "<div class='blankbox'></div>"
BLOCK_NAMES = {"quiz": "확인 질문", "multi": "여러 개 고르기", "match": "짝짓기", "order": "순서 맞추기", "classify": "분류하기",
               "write": "글쓰기", "decide": "결정하기", "vibe": "바이브 코딩 실습", "custom": "활동", "info": "읽기"}


def _block_html(mid, b):
    """산출물에 들어가는 블록 하나. 하지 않은 활동은 답·풀이 없이 빈칸으로 (정답이 산출물로 새지 않게)."""
    st, t = _bst(mid, b["id"]), b["type"]
    v = st["value"]
    done = st["ok"]
    head = f"<h3>{html.escape(b.get('title') or '')}</h3>" if b.get("title") else ""
    if t == "info":
        return head + _info_html(b)
    if t == "quiz":
        pick = (v or {}).get("pick")
        if not done or pick is None:
            return head + f"<p><b>Q.</b> {_md(b['text'])}</p><p>고른 답: {BLANK}</p>"
        return (head + f"<p><b>Q.</b> {_md(b['text'])}</p><p>고른 답: {_md(b['options'][pick])}</p>"
                f"<p class='note'>풀이: {_md(b['explain'])}</p>")
    if t == "multi":
        if not done:
            return head + f"<p>{_md(b['text'])}</p><p>고른 항목: {BLANK}</p>"
        return head + f"<p>{_md(b['text'])}</p><ul>" + "".join(
            f"<li>{'✅ 해당' if it['correct'] else '⬜ 해당 없음'} · {_md(it['text'])}<br><span class='note'>{_md(it.get('explain', ''))}</span></li>" for it in b["items"]) + "</ul>"
    if t == "match":
        return head + "<table>" + "".join(f"<tr><th>{_md(p['left'])}</th>" + (f"<td>{_md(p.get('right_label', p['right']))}</td>" if done else "<td class='empty'></td>") + "</tr>" for p in b["pairs"]) + "</table>"
    if t == "order":
        if not done:
            return head + "<ol>" + "".join(f"<li>{BLANK}</li>" for _ in b["items"]) + "</ol>"
        return head + "<ol>" + "".join(f"<li><b>{_md(it['text'])}</b>{' (' + _md(it['desc']) + ')' if it.get('desc') else ''}</li>" for it in b["items"]) + "</ol>"
    if t == "classify":
        names = {c["id"]: c["name"] for c in b["categories"]}
        rows = ""
        for it in b["items"]:
            z = (v or {}).get(it["id"])
            if z:
                rows += (f"<tr><td>{_md(it['title'])}</td><td><b>{names.get(z.get('primary'), '')}</b></td>"
                         + ("" if b.get("contrib") is False else f"<td>{', '.join(names[c] for c in z.get('contrib', []))}</td>")
                         + f"<td>{_md(it['explain'])}</td></tr>")
            else:
                rows += (f"<tr><td>{_md(it['title'])}</td><td class='empty'></td>" + ("" if b.get("contrib") is False else "<td class='empty'></td>") + "<td class='empty'></td></tr>")
        return head + (f"<table><tr><th>{_md(b.get('item_label', '사례'))}</th><th>{_md(b.get('primary_label', '직접 원인'))}</th>"
                       + ("" if b.get("contrib") is False else "<th>함께 작용한 것</th>") + f"<th>풀이</th></tr>{rows}</table>")
    if t == "write":
        return head + (f"<p class='note'>{_md(b['prompt'])}</p>" if b.get("prompt") else "") + (f"<p>{_esc_br(v)}</p>" if done and v else BLANKBOX)
    if t == "decide":
        opts = dict(b.get("options", DECIDE_OPTS))
        rows = ""
        for it in b["items"]:
            z = (v or {}).get(it["id"]) if done else None
            rows += f"<tr><td>{_md(it.get('who', ''))}</td><td>{_md(it['text'])}</td>" + (
                f"<td><b>{opts.get(z.get('choice'), '')}</b></td><td>{_esc_br(z.get('reason', ''))}</td>" if z else "<td class='empty'></td><td class='empty'></td>") + "</tr>"
        return head + f"<table><tr><th>누가</th><th>요청</th><th>결정</th><th>이유 · 대안</th></tr>{rows}</table>"
    if t == "vibe":
        return head + _vibe_html(mid, b, st)
    if not done:
        return head + BLANKBOX
    fn = CUSTOM_DOCS.get((mid, b["id"]))
    if fn:
        return head + fn(st)
    return head + f"<pre>{html.escape(json.dumps(v, ensure_ascii=False, indent=1))}</pre>"


def _mission_docs(mid, blanks=0):
    """산출물 문서: 명세의 outputs대로 (없으면 미션 전체를 문서 하나로). blanks는 비어 있는 활동 수."""
    m = MSPEC[mid]
    outs = m.get("outputs") or [{"id": f"{mid}_report", "title": f"M{m['no']} {m['short']} 결과",
                                 "sections": [{"h": f"{si}. {st['short']}", "blocks": [b["id"] for b in st.get("blocks", [])]}
                                              for si, st in enumerate(m["steps"], 1)]}]
    credits = asset_credits_html(_asset_refs(m)) if ASSET_ITEMS and not blanks else ""      # 이 미션에 쓴 사진의 출처 (미션을 마친 뒤 공개)
    for k, o in enumerate(outs):
        body = f"<p class='meta'>{_md(o['intro'])}</p>" if o.get("intro") else ""
        for sec in o.get("sections", []):
            body += f"<h2>{html.escape(sec['h'])}</h2>" if sec.get("h") else ""
            body += "".join(f"<p>{_md(p)}</p>" for p in sec.get("text", []))
            for bid in sec.get("blocks", []):
                body += _block_html(mid, BSPEC[(mid, bid)][1])
        if k == len(outs) - 1:
            body += credits
        _add_file(o["id"], o["title"], mid, _doc(o["title"], body), blanks)


def blk_todo(mid):
    """[산출물] 화면: 아직 안 한 활동 목록."""
    return JSON({"ok": True, "todo": _todo_list(mid)})


def blk_finish(mid, force=False):
    """[산출물 받기]: 안 한 활동이 있으면 먼저 알려 주고 물어봄(need_confirm). 그래도 받으면(force) 안 한 부분은 빈칸으로 만듦."""
    m = MSPEC[mid]
    _ensure_started(mid)
    todo = _todo_list(mid)
    if todo and not force:
        return _resp(ok=False, need_confirm=True, todo=todo)
    n_blank = sum(len(t["blocks"]) for t in todo)
    _mission_docs(mid, n_blank)
    p = S["progress"][mid]
    first = not p["issued"] and not todo
    if not todo:
        p["issued"] = True
    p["done"] = not todo
    if first:
        for ch, who, text in m.get("finish_messages", []):
            post(ch, who, text)
    msg = (f"M{m['no']} 산출물이 📁 파일에 저장됐어요." if not todo
           else f"M{m['no']} 산출물을 저장했어요. 하지 않은 활동 {n_blank}개는 빈칸이에요. 마저 하고 다시 받으면 채워져요.")
    return _resp(ok=True, msg=msg, blanks=n_blank)


def _prefill_block(mid, b):
    """선생님용: 블록 하나를 정답으로 채움."""
    t, st = b["type"], _bst(mid, b["id"])
    if t == "info":
        if b.get("confirm"):
            st["value"] = True
            block_ok(mid, b["id"])
        return
    if t == "quiz":
        st["value"] = {"pick": b["answer"]}
    elif t == "multi":
        st["value"] = {"picked": [i for i, it in enumerate(b["items"]) if it["correct"]]}
    elif t == "match":
        st["value"] = {p["left"]: str(p["right"]) for p in b["pairs"]}
    elif t == "order":
        st["value"] = {"order": [it["id"] for it in b["items"]]}
    elif t == "classify":
        st["value"] = {it["id"]: {"primary": it["accept"][0], "contrib": list(it.get("contrib", []))} for it in b["items"]}
    elif t == "write":
        st["value"] = b.get("example", "")
    elif t == "decide":
        st["value"] = {it["id"]: {"choice": it["answer"], "reason": it.get("reason") or it.get("explain") or ""} for it in b["items"]}
    elif t == "vibe":
        _vibe_prefill(mid, b)
    elif t == "custom":
        fn = AUTOPLAY.get((mid, b["id"]))
        if fn:
            fn()
        return
    block_ok(mid, b["id"])


def _prefill_all():
    """선생님용: 모든 미션의 모든 활동(심화 포함)을 채우고 산출물까지 만들어 둠. 빈 값이 없는 상태."""
    for mid in IMPLEMENTED:
        _ensure_started(mid)
        for si, st in enumerate(MSPEC[mid]["steps"], 1):
            if st.get("page") and (mid, f"page{si}") in AUTOPLAY:
                AUTOPLAY[(mid, f"page{si}")]()
            for b in st.get("blocks", []):
                _prefill_block(mid, b)
        blk_finish(mid, True)


# ---- 미션 등록: 소개 · 처음 상태 · 시작 메시지 · 화면 상태 ----
def _pub_state(mid, b):
    """화면에 보낼 블록 상태: 맞힌 뒤에만 풀이를 함께 보냄 (정답은 보내지 않음)."""
    st = _bst(mid, b["id"])
    out = {"ok": st["ok"], "tries": st["tries"], "value": st["value"]}
    if st["ok"]:
        if b["type"] == "quiz":
            out["explain"] = b["explain"]
        elif b["type"] == "multi":
            out["explains"] = [it.get("explain", "") for it in b["items"]]
        elif b["type"] == "decide":
            out["explains"] = {it["id"]: it.get("explain", "") for it in b["items"]}
        elif b["type"] in ("write", "match", "order") and b.get("explain"):
            out["explain"] = b["explain"]
        if b["type"] == "write":
            out["model"] = b.get("example", "")      # 모범답안: 제출한 뒤에만 보냄
    if b["type"] == "classify":
        out["explains"] = {it["id"]: it["explain"] for it in b["items"] if it["id"] in (st["value"] or {})}
    if b["type"] == "write" and st["ok"]:
        out["checklist"] = b.get("checklist", [])
    return out


def _mission_reset(mid):
    S[mid] = {"blk": {}, "pages": {}, "vibe": {}}
    if mid in CUSTOM_RESET:
        CUSTOM_RESET[mid]()


def _mission_view(mid):
    m = MSPEC[mid]
    out = {"blk": {b["id"]: _pub_state(mid, b) for st in m["steps"] for b in st.get("blocks", [])},
           "complete": [_step_done(mid, si) for si in range(1, len(m["steps"]) + 1)]}
    if mid in CUSTOM_VIEWS:
        out["custom"] = CUSTOM_VIEWS[mid]()
    return out


def _mission_start(mid):
    for ch, who, text in MSPEC[mid].get("start_messages", []):
        post(ch, who, text)


for _m in SPEC["missions"]:
    if not _m.get("steps"):
        continue
    BRIEFINGS[_m["id"]] = dict(_m["brief"], steps_short=[st["short"] for st in _m["steps"]],
                               comps=[{"name": c["name"], "criteria": c.get("criteria", "")} for c in SPEC.get("competencies", [])
                                      if c.get("id") in _m.get("competencies", [])])
    RESETTERS[_m["id"]] = (lambda mid: lambda: _mission_reset(mid))(_m["id"])
    VIEWS[_m["id"]] = (lambda mid: lambda: _mission_view(mid))(_m["id"])
    STARTERS[_m["id"]] = (lambda mid: lambda: _mission_start(mid))(_m["id"])

CALLBACKS.update({"blk.check": blk_check, "blk.finish": blk_finish, "blk.todo": blk_todo})
code_topic("app.blocks", "미션 활동은 이렇게 채점해요", "[확인]을 누르면 blk_check()가 명세의 정답과 비교해요. 틀리면 힌트를 한 단계씩 자세하게 보내고, "
           "맞히면 풀이를 보낸 뒤 단계를 다 마쳤는지 확인해요. 글쓰기는 맞고 틀림을 따지지 않고, 제출하면 모범답안을 보내요. 정답은 파이썬에만 있고 화면으로는 보내지 않아요.",
           ["blk_check", "_chk_quiz", "_chk_match", "_chk_write", "block_ok"], marks=["_hint(", "block_ok(", "\"model\"", "_advance("])
code_topic("app.docs", "산출물 문서는 이렇게 만들어요", "산출물을 받으면 명세의 outputs 순서대로, 학습자가 답한 내용을 모아 HTML 문서를 만들어요. 하지 않은 활동은 빈칸이에요.",
           ["blk_finish", "_mission_docs", "_block_html"], marks=["_add_file(", "_doc(", "finish_messages"])
# ===== 바이브 코딩 실습: 가상 데이터 파일 · 가상 AI (인터넷·외부 AI 없이 동작) =====
# 가상 데이터: 명세의 project.vfiles {파일 이름: 글}을 첫 실행 때 폴더에 저장함. 내려받는 것이 없음
# 가상 AI: 학습자가 고른 프롬프트 문장에 따라 명세에 적힌 응답을 돌려줌. 응답의 코드는 실제로 실행해 결과·오류를 그대로 보여 줌
#  - 정답(어느 문장이 좋은지)·응답은 파이썬에만 둠. 화면에는 보낸 뒤의 응답만 감
#  - 시도 기록이 남아 산출물의 '프롬프트 개선 기록'이 됨
import contextlib

VFILES = _P.get("vfiles") or {}
_VIBE_CACHE = {}


def vdata_folder():
    """가상 데이터 폴더: Colab은 /content/pbl_data, 점검 도구는 PBL_VDATA_ROOT(임시 폴더)."""
    root = os.environ.get("PBL_VDATA_ROOT") or ("/content" if os.path.isdir("/content") else os.getcwd())
    return os.path.join(root, _P.get("vfolder") or "pbl_data")


def vdata_prepare():
    """가상 데이터 파일을 폴더에 저장 (코드를 복사해 새 셀에서 돌려도 같은 파일을 읽게)."""
    if not VFILES:
        return
    folder = vdata_folder()
    os.makedirs(folder, exist_ok=True)
    for name, text in VFILES.items():
        with open(os.path.join(folder, name), "w", encoding="utf-8") as f:
            f.write(text)


def _vibe_st(mid, bid):
    return S[mid].setdefault("vibe", {}).setdefault(bid, {"picks": {}, "attempts": [], "last": None})


def _vibe_prompt(b, picks):
    """고른 문장을 이어 붙인 프롬프트."""
    lines = []
    for sl in b["slots"]:
        i = picks.get(sl["id"])
        if i is not None and 0 <= int(i) < len(sl["options"]) and sl["options"][int(i)].get("text"):
            lines.append(sl["options"][int(i)]["text"])
    return "\n".join(lines)


def _vibe_code(code):
    return (code or "").replace("{VDIR}", vdata_folder())


def _vibe_run(code):
    """AI가 만든 코드를 실제로 실행: 출력·표·그림·오류·문서(report 변수)를 모아 돌려줌. 같은 코드는 다시 돌리지 않음."""
    code = _vibe_code(code)
    if not code.strip():
        return {}
    if code in _VIBE_CACHE:
        return _VIBE_CACHE[code]
    tables, figs = [], []

    def display(obj=None, *a, **k):               # Colab의 display와 같은 이름: 복사해 새 셀에서 돌려도 그대로 동작함
        if hasattr(obj, "to_html"):
            tables.append(obj.to_html(index=False, border=0))
        else:
            print(obj)

    env = {"__name__": "__vibe__", "display": display}
    out, err = io.StringIO(), None
    cwd = os.getcwd()
    plt, prev = None, None
    if "matplotlib" in code:
        try:
            import matplotlib.pyplot as plt
            prev = plt.get_backend()
            plt.switch_backend("Agg")
        except Exception:
            plt = None
    try:
        os.makedirs(vdata_folder(), exist_ok=True)
        os.chdir(vdata_folder())
        with contextlib.redirect_stdout(out):
            exec(compile(code, "AI가 만든 코드", "exec"), env)
    except Exception as e:
        tb = traceback.extract_tb(e.__traceback__)
        frame = next((f for f in reversed(tb) if f.filename == "AI가 만든 코드"), None)
        err = "Traceback (most recent call last):\n"
        if frame:
            src = code.splitlines()[frame.lineno - 1].strip()
            err += f'  File "AI가 만든 코드", line {frame.lineno}\n    {src}\n'
        err += f"{type(e).__name__}: {e}"
    finally:
        os.chdir(cwd)
        if plt is not None:
            for n in plt.get_fignums():
                buf = io.BytesIO()
                plt.figure(n).savefig(buf, format="png", dpi=110, bbox_inches="tight")
                figs.append("data:image/png;base64," + base64.b64encode(buf.getvalue()).decode())
            plt.close("all")
            try:
                plt.switch_backend(prev)
            except Exception:
                pass
    report = env.get("report") if isinstance(env.get("report"), str) else None
    res = {"stdout": out.getvalue()[-6000:], "tables": tables, "figs": figs, "report": report, "error": err}
    _VIBE_CACHE[code] = res
    return res


def _vibe_reply(rep, ok, slot_label=None, slot_id=None):
    """명세의 응답 + 실제 실행 결과 → 화면에 보낼 것."""
    out = {k: rep.get(k, "") for k in ("say", "problem", "hint", "lesson")}
    out.update(code=_vibe_code(rep.get("code", "")), ok=ok, slot=slot_label, slot_id=slot_id, open_code=bool(rep.get("open_code")))
    out.update(_vibe_run(rep.get("code", "")))
    return out


def _chk_vibe(mid, b, st, payload):
    """프롬프트 보내기: 고른 문장을 앞 칸부터 보고, 부족한 칸이 있으면 그 칸의 응답을, 모두 좋으면 done 응답을 돌려줌."""
    picks = {k: int(v) for k, v in ((payload or {}).get("picks") or {}).items() if str(v).lstrip("-").isdigit()}
    if any(sl["id"] not in picks for sl in b["slots"]):
        return False, None, {"text": "칸마다 문장을 하나씩 골라 주세요."}
    v = _vibe_st(mid, b["id"])
    v["picks"] = picks
    miss = None
    for sl in b["slots"]:
        opt = sl["options"][picks[sl["id"]]] if 0 <= picks[sl["id"]] < len(sl["options"]) else {}
        if not opt.get("good"):
            miss = (sl, opt)
            break
    prompt = _vibe_prompt(b, picks)
    if miss:
        sl, opt = miss
        rep = _vibe_reply(opt.get("miss") or sl["miss"], False, sl["label"], sl["id"])
        st["tries"] += 1
    else:
        rep = _vibe_reply(b["done"], True)
    v["last"] = rep
    v["attempts"].append({"no": len(v["attempts"]) + 1, "prompt": prompt, "ok": rep["ok"], "slot": rep["slot"],
                          "result": re.sub(r"\*\*", "", rep["problem"]) or "요청대로 동작함"})
    resp = {"reply": rep, "attempts": [{k: a[k] for k in ("no", "ok", "slot", "result")} for a in v["attempts"]], "text": ""}
    if rep["ok"]:
        return True, {"picks": picks, "prompt": prompt}, resp
    return False, None, resp


def app_vibe(mid, bid):
    """화면이 블록을 다시 그릴 때 마지막 응답을 받아 감 (상태 전체에 넣으면 그림 때문에 무거워져서 따로 보냄)."""
    v = _vibe_st(mid, bid)
    b = BSPEC[(mid, bid)][1]
    start = _vibe_reply(b["start"], False) if b.get("start") else None      # 지금까지의 코드와 결과: 블록 맨 위에 보여 줌
    return JSON({"ok": True, "reply": v["last"], "picks": v["picks"], "start": start,
                 "attempts": [{k: a[k] for k in ("no", "ok", "slot", "result")} for a in v["attempts"]]})


def _vibe_prefill(mid, b):
    """선생님용: 좋은 문장을 고른 상태 + done 응답."""
    picks = {sl["id"]: next(i for i, o in enumerate(sl["options"]) if o.get("good")) for sl in b["slots"]}
    st = _bst(mid, b["id"])
    ok, value, _ = _chk_vibe(mid, b, st, {"picks": picks})
    st["value"] = value
    return ok


def _vibe_html(mid, b, st):
    """산출물 문서: 최종 프롬프트 · 개선 기록 · 실행 결과 · AI가 만든 문서. 하지 않았으면 빈칸."""
    v = _vibe_st(mid, b["id"])
    if not st["ok"]:
        out = "<h4>최종 프롬프트</h4><div class='blankbox'></div><h4>실행 결과</h4><div class='blankbox'></div>"
    else:
        rep = v["last"] if (v["last"] or {}).get("ok") else _vibe_reply(b["done"], True)
        out = f"<h4>최종 프롬프트</h4><pre>{html.escape((st['value'] or {}).get('prompt', ''))}</pre>"
        if rep.get("stdout"):
            out += f"<h4>실행 결과</h4><pre>{html.escape(rep['stdout'])}</pre>"
        elif rep.get("tables") or rep.get("figs"):
            out += "<h4>실행 결과</h4>"
        out += "".join(rep.get("tables") or [])
        out += "".join(f"<p><img src='{u}'></p>" for u in rep.get("figs") or [])
        if rep.get("report"):
            out += f"<h4>AI가 만든 문서</h4><div class='card'>{rep['report']}</div>"
        if rep.get("code"):
            out += f"<details><summary>AI가 만든 코드 보기</summary><pre>{html.escape(rep['code'])}</pre></details>"
    if v["attempts"]:
        rows = "".join(f"<tr><td>{a['no']}</td><td>{'성공' if a['ok'] else html.escape(a['slot'] or '')}</td><td>{html.escape(a['result'])}</td></tr>" for a in v["attempts"])
        out += f"<h4>프롬프트 개선 기록</h4><table><tr><th>시도</th><th>부족했던 칸</th><th>결과</th></tr>{rows}</table>"
    return out


CHECKERS["vibe"] = _chk_vibe
CALLBACKS["app.vibe"] = app_vibe
vdata_prepare()
_setup(60, 2, "미션 활동 · 가상 AI 준비 완료")

# ===== 시작 준비: 화면 버튼과 파이썬 함수 연결 =====
_reset_state()
for _name, _fn in CALLBACKS.items():
    output.register_callback(_name, _fn)

_setup(100, 3, "✅ 준비 완료! 아래 <b>💬 실루엣 식별 교육개발반 열기</b> 셀을 실행하세요. (‘모두 실행’을 눌렀다면 곧 자동으로 열려요)")
```

## 💬 실루엣 식별 교육개발반 메신저

```python
#@title 💬 실루엣 식별 교육개발반 열기 { display-mode: "form" }
#@markdown 위의 **🚀 [먼저 실행] 실습 준비하기** 셀의 진행 막대가 100%가 된 뒤 이 셀을 실행하면, 아래에 메신저 화면이 열립니다.
from IPython.display import HTML, display

WORKSPACE_NAME = "🛩️ 실루엣 식별 교육개발반"

APP_HTML = r"""
<style>
  .ws { --ink:#111827; --line:#c3cad5; --soft:#f2f5f9; --blue:#0b57d0; --blue-bg:#e8f0fe; --green:#146c2e; --green-bg:#e6f4ea;
        --red:#b3261e; --red-bg:#fce8e6; --amber:#8a4b00; --amber-bg:#fff4ce; --violet:#3f0e40;
        color-scheme: light; font-family: 'Noto Sans KR', 'Malgun Gothic', 'Apple SD Gothic Neo', sans-serif;
        border: 1px solid var(--line); border-radius: 10px; box-sizing: border-box; overflow: hidden; background: #fff; color: var(--ink);
        font-size: 14px; display: flex; flex-direction: column; height: 900px; width: 100%; max-width: 100%;
        min-width: 360px; min-height: 480px; resize: both; position: relative; }
  .ws * { box-sizing: border-box; color: inherit; }
  .ws.w-base { width: 1100px; } .ws.w-wide { width: 1400px; } .ws.w-full { width: 100%; }
  .ws button, .ws input, .ws select, .ws textarea { font-family: inherit; }
  .ws a { color: var(--blue); }
  .ws h2, .ws h3, .ws h4, .ws h5, .ws p, .ws li, .ws label, .ws td, .ws th, .ws dt, .ws dd, .ws summary { color: var(--ink); }
  .ws ::placeholder { color: #1e40af; opacity: .85; font-style: italic; }
  /* 상단 바 */
  .ws .topbar { display: flex; align-items: center; gap: 8px; padding: 6px 10px; background: var(--violet); color: #fff; font-size: 13px; flex-wrap: wrap; }
  .ws .topbar * { color: #fff; }
  .ws .topbar .name { font-weight: bold; margin: 0 4px; font-size: 14px; }
  .ws .topbar .spacer { flex: 1; }
  .ws .tb { background: transparent; border: 1px solid #b89bb9; border-radius: 6px; padding: 2px 9px; cursor: pointer; font-size: 13px; position: relative; }
  .ws .tb:hover { background: #5a2a5b; }
  .ws .tb.on { background: #fff; border-color: #fff; } .ws .tb.on, .ws .tb.on * { color: var(--violet); }
  .ws .tb .count { position: absolute; top: -6px; right: -8px; background: #e01e5a; border-radius: 9px; padding: 0 5px; font-size: 11px; display: none; }
  .ws.side-off .tb .count.has { display: inline; }
  .ws .tprog { display: flex; align-items: center; gap: 6px; font-size: 12px; }
  .ws .tprog .pb { width: 120px; height: 9px; background: #6b3d6c; border-radius: 6px; overflow: hidden; }
  .ws .tprog .pb i { display: block; height: 100%; background: #6ee7a8; width: 0; transition: width .4s; }
  .ws .me-chip { background: #5a2a5b; border: 1px solid #b89bb9; border-radius: 14px; padding: 2px 10px; cursor: pointer; font-size: 12.5px; }
  .ws .me-chip:hover { background: #6d3a6e; }
  /* 본문 */
  .ws .body { display: flex; flex: 1; min-height: 0; position: relative; }
  .ws .side { width: 236px; background: #3f0e40; color: #fff; padding: 8px 0 14px; overflow-y: auto; flex-shrink: 0; transition: width .2s, padding .2s; }
  .ws .side * { color: #fff; }
  .ws.side-off .side, .ws.onboard .side { width: 0; padding: 0; }
  .ws .side h4 { margin: 14px 14px 4px; font-size: 12.5px; font-weight: normal; line-height: 1.4; border-top: 1px solid #6d3a6e; padding-top: 10px; }
  .ws .side h4 b { font-weight: bold; }
  .ws .item { padding: 5px 14px; cursor: pointer; display: flex; align-items: center; gap: 6px; white-space: nowrap; font-size: 13.5px; }
  .ws .item span.t { overflow: hidden; text-overflow: ellipsis; flex: 1; }
  .ws .item:hover { background: #5a2a5b; }
  .ws .item.active { background: #1164a3; }
  .ws .item.unread { font-weight: bold; }
  .ws .item.soon span.t { font-style: italic; }
  .ws .item .ic { width: 18px; text-align: center; flex-shrink: 0; }
  .ws .item .mini { font-size: 11px; background: #1164a3; border-radius: 8px; padding: 0 6px; }
  .ws .badge { background: #e01e5a; color: #fff; border-radius: 9px; padding: 0 7px; font-size: 12px; }
  .ws .main { flex: 1; display: flex; flex-direction: column; min-width: 0; }
  .ws .title { padding: 10px 16px; font-weight: bold; border-bottom: 1px solid var(--line); font-size: 15px; }
  .ws .title small { font-weight: normal; margin-right: 6px; }
  .ws .scroll { flex: 1; overflow-y: auto; padding: 14px 18px 30px; min-height: 0; }
  .ws .footer { display: flex; gap: 10px; align-items: center; padding: 7px 14px; background: var(--soft); border-top: 1px solid var(--line); }
  .ws .note { flex: 1; font-size: 13px; } .ws .note.ok { color: var(--green); } .ws .note.bad { color: var(--red); font-weight: bold; }
  .ws.busy .note::before { content: '⏳ '; }
  /* 코드 보기 */
  .ws .codep { width: 0; overflow: hidden; border-left: 1px solid var(--line); background: #fbfcfe; display: flex; flex-direction: column; flex-shrink: 0; }
  .ws.code-on .codep { width: 410px; }
  .ws.compact.code-on .codep { position: absolute; right: 0; top: 0; bottom: 0; width: min(430px, 94%); z-index: 20; box-shadow: -8px 0 18px rgba(0,0,0,.22); }
  .ws.onboard .codep, .ws.onboard .code-tab, .ws.no-code .codep, .ws.no-code .code-tab, .ws.no-code #tb-code { display: none; }
  .ws .code-tab { position: absolute; right: 0; top: 30%; z-index: 10; writing-mode: vertical-rl; background: #1e293b; color: #fff; border: 0;
                  border-radius: 8px 0 0 8px; padding: 12px 6px; cursor: pointer; font-size: 13px; letter-spacing: 1px; box-shadow: -2px 2px 8px rgba(0,0,0,.2); }
  .ws .code-tab * { color: #fff; }
  .ws.code-on .code-tab { display: none; }
  .ws .codep .ch { display: flex; align-items: center; gap: 6px; padding: 9px 12px; background: #1e293b; }
  .ws .codep .ch, .ws .codep .ch * { color: #fff; }
  .ws .codep .ch b { flex: 1; font-size: 14px; }
  .ws .codep .ch button { background: transparent; border: 1px solid #94a3b8; border-radius: 6px; cursor: pointer; padding: 1px 8px; }
  .ws .codep .cs { flex: 1; overflow-y: auto; padding: 10px 12px 24px; }
  .ws .codep .intro { font-size: 12.5px; line-height: 1.55; background: var(--blue-bg); border-radius: 8px; padding: 8px 10px; margin-bottom: 10px; }
  .ws .codep h6 { font-size: 13px; margin: 12px 0 4px; display: flex; align-items: center; gap: 6px; }
  .ws .codep h6 .sp { flex: 1; }
  .ws .codep .desc { font-size: 12.5px; line-height: 1.5; margin: 0 0 6px; }
  .ws .codep details { border: 1px solid var(--line); border-radius: 8px; margin: 6px 0; background: #fff; }
  .ws .codep details > summary { padding: 6px 8px; cursor: pointer; font-size: 12.5px; color: var(--ink); }
  .ws .codep details > summary code { font-weight: bold; color: #6d28d9; }
  .ws .cb { display: flex; border-top: 1px solid #e5e9f0; font: 12px/18px Consolas, 'D2Coding', Menlo, monospace; }
  .ws .cb .ln { margin: 0; padding: 8px 6px; text-align: right; background: #eef2f7; user-select: none; min-width: 30px; }
  .ws .cb .cw { position: relative; overflow-x: auto; flex: 1; }
  .ws .cb pre { margin: 0; padding: 8px 10px; white-space: pre; position: relative; z-index: 1; font: inherit; }
  .ws .cb .hlb { position: absolute; left: 0; right: 0; height: 18px; background: #fff3b0; z-index: 0; }
  .ws .code .k { color: #7c3aed; font-weight: bold; } .ws .code .s { color: #b3261e; } .ws .code .c { color: #146c2e; font-style: italic; }
  .ws .code .n { color: #9a3412; } .ws .code .f { color: #0b57d0; }
  .ws .live pre { background: #0f172a; border-radius: 8px; padding: 10px; margin: 0; white-space: pre-wrap; word-break: break-all; font: 12px/1.6 Consolas, 'D2Coding', Menlo, monospace; }
  .ws .live pre, .ws .live pre * { color: #f8fafc; }
  .ws .live pre .c { color: #86efac; font-style: italic; } .ws .live pre .s { color: #fca5a5; } .ws .live pre .k { color: #c4b5fd; font-weight: bold; }
  .ws .live pre .n { color: #fdba74; } .ws .live pre .f { color: #93c5fd; }
  .ws .live .empty { font-size: 12.5px; padding: 8px 10px; border: 1px dashed var(--line); border-radius: 8px; }
  .ws .cpy { font-size: 11.5px; padding: 1px 8px; border-radius: 6px; border: 1px solid var(--line); background: #fff; cursor: pointer; }
  /* 공통 요소 */
  .ws .btn { padding: 7px 14px; border-radius: 7px; border: 1px solid #007a5a; background: #007a5a; color: #fff; cursor: pointer; font-size: 14px; font-weight: bold; }
  .ws .btn * { color: #fff; }
  .ws .btn:hover { filter: brightness(1.08); }
  .ws .btn.ghost { background: #fff; } .ws .btn.ghost, .ws .btn.ghost * { color: #006048; }
  .ws .btn.small { padding: 3px 10px; font-size: 12.5px; }
  .ws .btn.big { padding: 10px 20px; font-size: 15px; }
  .ws .btn.blue { background: var(--blue); border-color: var(--blue); }
  .ws .btn:disabled { background: #fff; border: 1px dashed #6b7280; cursor: not-allowed; font-weight: normal; }
  .ws .btn:disabled, .ws .btn:disabled * { color: var(--ink); }
  .ws .btn.pulse { animation: pulse 1.6s infinite; }
  @keyframes pulse { 0% { box-shadow: 0 0 0 0 rgba(11,87,208,.55); } 70% { box-shadow: 0 0 0 9px rgba(11,87,208,0); } 100% { box-shadow: 0 0 0 0 rgba(11,87,208,0); } }
  .ws .row { display: flex; flex-wrap: wrap; gap: 8px; align-items: center; margin: 8px 0; }
  .ws .card { border: 1px solid var(--line); border-radius: 10px; padding: 12px 14px; margin: 10px 0; background: #fff; }
  .ws .card.focus { border: 2px solid var(--blue); box-shadow: 0 0 0 3px var(--blue-bg); }
  .ws .card h4 { margin: 0 0 6px; font-size: 15px; }
  .ws .ok { color: var(--green); } .ws .bad { color: var(--red); }
  .ws .fb { margin-top: 8px; padding: 8px 10px; border-radius: 7px; font-size: 13.5px; line-height: 1.6; }
  .ws .fb.ok { background: var(--green-bg); color: #0b3d1a; border: 1px solid #9fd5ad; }
  .ws .fb.bad { background: var(--red-bg); color: #5c0f0a; border: 1px solid #efa9a3; }
  .ws .fb.info { background: var(--blue-bg); color: #0a2a66; border: 1px solid #a8c3f0; }
  .ws .fb.ok *, .ws .fb.bad *, .ws .fb.info * { color: inherit; }
  .ws .err { color: var(--red); font-size: 13px; font-weight: bold; margin-top: 4px; }
  .ws .err:empty { display: none; }
  .ws .opt-tag { display: inline-block; background: var(--blue-bg); border-radius: 4px; font-size: 11px; padding: 0 5px; margin-left: 4px; }
  .ws .missing { outline: 3px solid var(--red); outline-offset: 2px; border-radius: 6px; }
  .ws select, .ws input[type=text], .ws input[type=date], .ws input[type=time], .ws input[type=number], .ws textarea {
    padding: 6px 8px; font-size: 13.5px; background: #fff; color: var(--ink); border: 1px solid #8a94a6; border-radius: 6px; max-width: 100%; }
  .ws textarea { width: 100%; min-height: 56px; line-height: 1.5; }
  .ws select option, .ws select optgroup { color: var(--ink); background: #fff; }
  .ws .count-line { font-size: 12.5px; margin-top: 2px; }
  .ws .count-line.short { color: var(--red); font-weight: bold; } .ws .count-line.enough { color: var(--green); font-weight: bold; }
  .ws table.t { border-collapse: collapse; width: 100%; font-size: 13px; margin: 6px 0; }
  .ws table.t th, .ws table.t td { border-bottom: 1px solid #dfe4ea; padding: 6px 8px; text-align: left; vertical-align: middle; }
  .ws table.t th { font-weight: bold; background: var(--soft); }
  .ws table.t tr.split td { background: var(--amber-bg); }
  .ws .lbl-unk { font-style: italic; text-decoration: underline dotted; } .ws .lbl-wrong { color: var(--red); font-weight: bold; }
  .ws details > summary { cursor: pointer; color: var(--blue); margin: 6px 0; font-weight: bold; }
  .ws .hint { font-size: 13px; line-height: 1.55; }
  .ws .tip { background: var(--soft); border-left: 4px solid var(--blue); border-radius: 6px; padding: 8px 12px; font-size: 13.5px; line-height: 1.6; margin: 8px 0; }
  /* 할 일 상자 */
  /* 단계 안내 */
  .ws .sub { border-left: 4px solid #cbd5e1; padding: 6px 0 6px 12px; margin: 10px 0; }
  .ws .sub.now { border-left-color: var(--blue); background: linear-gradient(90deg, var(--blue-bg), #fff 70%); border-radius: 0 8px 8px 0; }
  .ws .sub.done { border-left-color: var(--green); }
  .ws .sub-h { font-weight: bold; margin-bottom: 6px; font-size: 14px; }
  .ws .look { background: #fff; border: 1px dashed var(--blue); border-radius: 8px; padding: 7px 10px; line-height: 1.6; }
  /* 선택지 */
  .ws .choices { display: flex; flex-wrap: wrap; gap: 6px; }
  .ws .choices.col { flex-direction: column; align-items: stretch; }
  .ws .choice { display: flex; align-items: flex-start; gap: 7px; border: 1px solid #8a94a6; border-radius: 8px; padding: 6px 10px; cursor: pointer; background: #fff; font-size: 13.5px; line-height: 1.45; }
  .ws .choice:hover { border-color: var(--blue); background: #f5f8ff; }
  .ws .choice.on { border: 2px solid var(--blue); background: var(--blue-bg); font-weight: bold; padding: 5px 9px; }
  .ws .choice input { margin: 2px 0 0; accent-color: var(--blue); }
  .ws .choice small { display: block; font-weight: normal; font-size: 12px; }
  .ws .choice.good { border-color: var(--green); background: var(--green-bg); } .ws .choice.wrong { border-color: var(--red); background: var(--red-bg); }
  .ws .checks { display: grid; grid-template-columns: repeat(auto-fill, minmax(250px, 1fr)); gap: 6px; }
  .ws .fieldrow { margin: 10px 0; } .ws .fieldrow > .fl { font-weight: bold; margin-bottom: 4px; display: block; }
  /* 메신저 */
  .ws .msgs { flex: 1; overflow-y: auto; padding: 8px 16px; min-height: 0; }
  .ws .msg { display: flex; gap: 10px; padding: 7px 6px; border-radius: 6px; }
  .ws .msg:hover { background: var(--soft); }
  .ws .msg.mention { background: #fff4ce; border-left: 3px solid #f0b429; }
  .ws .avatar { width: 34px; height: 34px; border-radius: 6px; display: flex; align-items: center; justify-content: center; font-weight: bold; flex-shrink: 0; }
  .ws .avatar, .ws .avatar * { color: #fff; }
  .ws .meta { font-size: 13px; } .ws .meta b { margin-right: 4px; }
  .ws .time { font-size: 12px; margin-left: 6px; }
  .ws .text { white-space: pre-wrap; word-break: break-word; margin-top: 2px; line-height: 1.55; }
  .ws .tag { font-size: 11.5px; border-radius: 4px; padding: 1px 6px; margin-left: 4px; white-space: nowrap; background: #e9edf3; border: 1px solid var(--line); }
  .ws .tag.ext { background: #f3e8fd; border-color: #d7b8f5; }
  .ws .compose { display: flex; gap: 8px; padding: 10px 14px; border-top: 1px solid var(--line); }
  .ws .compose input { flex: 1; padding: 8px; font-size: 14px; }
  /* 첫 화면 */
  .ws .onb { max-width: 560px; margin: 26px auto; border: 2px solid var(--blue); border-radius: 14px; padding: 22px 24px; background: #fff; }
  .ws .onb h2 { margin: 0 0 6px; font-size: 21px; }
  .ws .onb .lead { line-height: 1.6; margin: 0 0 14px; }
  .ws .onb select, .ws .onb input { width: 100%; font-size: 15px; padding: 8px 10px; }
  .ws .preview-call { background: var(--blue-bg); border-radius: 8px; padding: 8px 12px; margin-top: 12px; }
  /* 홈 */
  .ws .hero { background: linear-gradient(135deg, #3f0e40, #1164a3); border-radius: 12px; padding: 18px 20px; }
  .ws .hero, .ws .hero * { color: #fff; }
  .ws .hero h2 { margin: 4px 0 8px; font-size: 21px; }
  .ws .hero p { margin: 0; line-height: 1.65; font-size: 13.5px; }
  .ws .greet { font-size: 16px; font-weight: bold; margin: 14px 0 6px; }
  .ws .bigprog { display: flex; align-items: center; gap: 10px; }
  .ws .bigprog .pb { flex: 1; height: 14px; background: #e3e8ef; border-radius: 8px; overflow: hidden; }
  .ws .bigprog .pb i { display: block; height: 100%; background: linear-gradient(90deg, #16a34a, #22c55e); }
  .ws .stages { display: grid; grid-template-columns: repeat(auto-fill, minmax(250px, 1fr)); gap: 12px; margin-top: 10px; }
  .ws .stage { border: 1px solid var(--line); border-radius: 10px; padding: 10px 12px; }
  .ws .stage h5 { margin: 0 0 8px; font-size: 13.5px; }
  .ws .mcard { display: flex; gap: 8px; align-items: flex-start; padding: 7px 8px; border-radius: 8px; cursor: pointer; border: 1px solid transparent; }
  .ws .mcard:hover { background: var(--soft); border-color: var(--line); }
  .ws .pill { font-size: 11.5px; padding: 1px 8px; border-radius: 10px; white-space: nowrap; border: 1px solid transparent; }
  .ws .pill.done { background: var(--green-bg); border-color: #9fd5ad; } .ws .pill.open { background: var(--blue-bg); border-color: #a8c3f0; }
  .ws .pill.locked { background: #fff; border: 1px dashed #6b7280; } .ws .pill.soon { background: var(--amber-bg); border-color: #e3c96b; }
  /* 미션 */
  .ws .stepper { display: flex; flex-wrap: wrap; gap: 6px; padding: 8px 16px; border-bottom: 1px solid var(--line); background: #fafbfd; }
  .ws .stp { font-size: 12.5px; padding: 4px 10px; border-radius: 14px; border: 1px solid #8a94a6; background: #fff; cursor: pointer; }
  .ws .stp.done { border-color: var(--green); background: var(--green-bg); }
  .ws .stp.cur, .ws .stp.cur.done { background: var(--blue); border-color: var(--blue); font-weight: bold; }
  .ws .stp.cur, .ws .stp.cur * { color: #fff; }
  .ws .stp:disabled { cursor: not-allowed; border-style: dashed; background: #fff; }
  .ws .brief h2 { margin: 6px 0 2px; font-size: 21px; }
  .ws .brief .sub-t { margin: 0 0 10px; font-size: 15px; }
  .ws .brief section { margin: 16px 0; }
  .ws .brief h4 { margin: 0 0 6px; font-size: 15px; }
  .ws .brief p, .ws .brief li { line-height: 1.7; }
  .ws .why { background: var(--soft); border-left: 4px solid #1164a3; border-radius: 6px; padding: 10px 14px; }
  .ws .terms { display: grid; grid-template-columns: repeat(auto-fill, minmax(300px, 1fr)); gap: 12px; }
  .ws .term { border: 1px solid var(--line); border-radius: 10px; padding: 10px 12px; background: #fff; }
  .ws .term h5 { margin: 0 0 6px; font-size: 14.5px; } .ws .term h5 small { font-weight: normal; font-size: 11.5px; margin-left: 4px; }
  .ws .term dl { margin: 0; display: grid; grid-template-columns: 74px 1fr; gap: 4px 8px; font-size: 13px; line-height: 1.55; }
  .ws .term dt { font-weight: bold; } .ws .term dd { margin: 0; }
  .ws .term dt.k1 { color: var(--blue); } .ws .term dt.k2 { color: var(--green); } .ws .term dt.k3 { color: #9a3412; }
  .ws .fig { background: #f8fafc; border: 1px solid #e3e8ef; border-radius: 8px; margin: 4px 0 8px; padding: 4px; }
  .ws .fig svg { display: block; width: 100%; height: auto; max-height: 170px; }
  /* M1 실험실 */
  .ws .lab2 { display: grid; grid-template-columns: minmax(0, 1fr) minmax(0, 1fr); gap: 12px; }
  .ws .main.narrow .lab2 { grid-template-columns: minmax(0, 1fr); }
  .ws .labcol { border: 1px solid var(--line); border-radius: 10px; padding: 10px; min-width: 0; background: #fff; }
  .ws .labcol.focus { border: 2px solid var(--blue); box-shadow: 0 0 0 3px var(--blue-bg); }
  .ws .labcol h5 { margin: 0; font-size: 14px; } .ws .labcol .cap { font-size: 12.5px; margin: 2px 0 6px; }
  .ws .ctl { display: grid; grid-template-columns: 92px minmax(0, 1fr) 54px; align-items: center; gap: 6px; margin: 3px 0; font-size: 13px; border-radius: 6px; padding: 1px 3px; }
  .ws .ctl input[type=range], .ws .ctl select { width: 100%; min-width: 0; }
  .ws .ctl output { text-align: right; font-weight: bold; }
  .ws .ctl.flash { animation: flash 1.8s ease-out; }
  @keyframes flash { 0% { background: #fde047; } 100% { background: transparent; } }
  .ws .imgbox { position: relative; display: block; line-height: 0; margin-top: 8px; }
  .ws .imgbox img { width: 100%; display: block; border-radius: 6px; border: 1px solid var(--line); }
  .ws .imgbox.flash img { animation: imgflash 1.4s ease-out; }
  @keyframes imgflash { 0% { box-shadow: 0 0 0 4px #fde047; } 100% { box-shadow: 0 0 0 0 transparent; } }
  .ws .mk { position: absolute; transform: translate(-50%, -50%); background: rgba(255,255,255,.92); border: 1px solid #111827;
            border-radius: 10px; font-size: 11px; line-height: 16px; padding: 0 5px; pointer-events: none; font-weight: bold; }
  .ws .mk.cur { background: #fde047; border: 2px solid #b3261e; font-size: 13px; line-height: 18px; z-index: 2; }
  .ws .north { position: absolute; right: 6px; top: 6px; background: rgba(255,255,255,.92); border-radius: 4px; font-size: 11px; line-height: 15px; padding: 1px 4px; font-weight: bold; }
  .ws .info { margin-top: 8px; font-size: 13px; background: var(--soft); border-radius: 6px; padding: 6px 10px; line-height: 1.65; }
  .ws .legend { font-size: 12.5px; }
  .ws .qtabs { display: flex; flex-wrap: wrap; gap: 6px; margin: 6px 0 10px; }
  .ws .qhead { display: flex; align-items: center; gap: 8px; } .ws .qhead .sp { flex: 1; }
  .ws .qtext { font-size: 15px; line-height: 1.6; margin: 4px 0 8px; }
  .ws .split { display: grid; grid-template-columns: minmax(220px, 300px) minmax(0, 1fr); gap: 14px; align-items: start; }
  .ws .main.narrow .split { grid-template-columns: minmax(0, 1fr); }
  .ws .gsdlist .gr { display: flex; align-items: center; gap: 8px; padding: 7px 10px; border: 1px solid var(--line); border-radius: 8px; margin-bottom: 6px; cursor: pointer; background: #fff; }
  .ws .gsdlist .gr.cur { border: 2px solid var(--blue); background: var(--blue-bg); font-weight: bold; }
  .ws .gsdlist .gr.locked { cursor: not-allowed; border-style: dashed; }
  .ws .gsdlist .gr .sp { flex: 1; }
  .ws figure.ch { margin: 0; text-align: center; font-size: 12.5px; }
  .ws figure.ch img { width: 200px; max-width: 100%; border-radius: 6px; border: 1px solid var(--line); display: block; }
  .ws .ecard .imgs { display: flex; gap: 10px; flex-wrap: wrap; }
  .ws .ecard .imgs figure { margin: 0; text-align: center; font-size: 12.5px; }
  .ws .ecard .imgs img { width: 220px; max-width: 100%; border-radius: 6px; border: 1px solid var(--line); display: block; }
  /* M2 */
  .ws .strip { display: flex; flex-wrap: wrap; gap: 5px; margin: 6px 0 10px; }
  .ws .strip button { min-width: 38px; padding: 3px 7px; border-radius: 8px; border: 1px solid #8a94a6; background: #fff; cursor: pointer; font-size: 13px; }
  .ws .strip button.done { background: var(--green-bg); border-color: var(--green); }
  .ws .strip button.cur { background: var(--blue); border-color: var(--blue); font-weight: bold; } .ws .strip button.cur, .ws .strip button.cur * { color: #fff; }
  .ws .lab-l { display: grid; grid-template-columns: minmax(0, 1.3fr) minmax(260px, 1fr); gap: 14px; align-items: start; }
  .ws .main.narrow .lab-l { grid-template-columns: minmax(0, 1fr); }
  .ws .lab-l > .stick { position: sticky; top: 0; z-index: 3; background: #fff; padding-bottom: 4px; }
  .ws .main.narrow .lab-l > .stick { position: static; }
  .ws .ov { position: relative; display: block; line-height: 0; }
  .ws .ov img { width: 100%; border-radius: 8px; border: 1px solid var(--line); }
  .ws .chipbig img { width: 100%; max-width: 300px; border-radius: 8px; border: 1px solid var(--line); display: block; }
  .ws .tiles { display: grid; grid-template-columns: repeat(auto-fill, minmax(170px, 1fr)); gap: 10px; }
  .ws .tile { border: 1px solid var(--line); border-radius: 8px; padding: 10px 12px; }
  .ws .tile .v { font-size: 24px; font-weight: bold; margin: 2px 0; }
  .ws .tile .d { font-size: 12.5px; line-height: 1.45; }
  .ws .acc { border: 1px solid var(--line); border-radius: 10px; margin: 8px 0; background: #fff; }
  .ws .acc .acc-h { display: flex; align-items: center; gap: 8px; padding: 9px 12px; font-weight: bold; }
  .ws .acc .acc-h .sp { flex: 1; }
  .ws .acc.locked .acc-h { border-style: dashed; }
  .ws .acc.open { border: 2px solid var(--blue); box-shadow: 0 0 0 3px var(--blue-bg); }
  .ws .acc .acc-b { padding: 0 12px 12px; }
  .ws .acc .acc-s { padding: 0 12px 10px; font-size: 13px; line-height: 1.55; }
  /* M3 · M4 */
  .ws .post { background: #fff; border: 1px solid var(--line); border-left: 4px solid #e01e5a; border-radius: 8px; padding: 9px 12px; font-size: 14.5px; line-height: 1.6; }
  .ws .kv { display: grid; grid-template-columns: auto 1fr; gap: 3px 12px; font-size: 13.5px; }
  .ws .kv b { white-space: nowrap; }
  .ws .raw { font: 12.5px/1.7 Consolas, 'D2Coding', Menlo, monospace; background: #0f172a; border-radius: 8px; padding: 10px 12px; overflow-x: auto; }
  .ws .raw, .ws .raw * { color: #e2e8f0; } .ws .raw .rk { color: #93c5fd; }
  .ws .jsonbox { font: 12px/1.6 Consolas, 'D2Coding', Menlo, monospace; background: #f8fafc; border: 1px solid var(--line); border-radius: 8px; padding: 10px; white-space: pre; overflow-x: auto; }
  .ws .sunres { background: #fff7ed; border: 1px solid #fdba74; border-radius: 8px; padding: 8px 12px; line-height: 1.7; }
  .ws .calc { display: flex; flex-wrap: wrap; gap: 8px; align-items: flex-end; }
  .ws .calc label { display: flex; flex-direction: column; font-size: 12.5px; font-weight: bold; gap: 2px; }
  .ws .calc input { width: 130px; }
  .ws .casebox { position: relative; line-height: 0; }
  .ws .casebox img { width: 100%; border-radius: 8px; border: 1px solid var(--line); }
  .ws .casebox svg { position: absolute; left: 0; top: 0; width: 100%; height: 100%; pointer-events: none; }
  .ws .casebox.small { width: 58%; flex-shrink: 0; } .ws .casebox.small .mk { font-size: 10px; line-height: 14px; }
  .ws .imgcap { font-size: 12.5px; font-weight: bold; margin: 4px 0; line-height: 1.4; }
  .ws .track { border: 1px solid var(--line); border-radius: 10px; padding: 10px; min-width: 0; }
  .ws .track h5 { margin: 0 0 6px; font-size: 14.5px; }
  .ws .stage-r { border: 1px solid #dfe4ea; border-radius: 8px; padding: 8px; margin: 6px 0; }
  .ws .stage-r.next { border: 2px solid var(--blue); background: #f7faff; }
  .ws .stage-r .sh { display: flex; align-items: center; gap: 8px; font-weight: bold; } .ws .stage-r .sh .sp { flex: 1; }
  .ws .stage-r img { width: 100%; max-height: 300px; object-fit: contain; background: #0f172a; border-radius: 6px; border: 1px solid var(--line); margin-top: 6px; display: block; }
  .ws .stage-r .st { font-size: 13px; line-height: 1.55; margin-top: 4px; }
  .ws .pipe { display: flex; flex-wrap: wrap; align-items: center; gap: 6px; font-size: 13px; margin: 6px 0; }
  .ws .pipe span.box { border: 1px solid var(--blue); background: var(--blue-bg); border-radius: 6px; padding: 3px 8px; font-weight: bold; }
  .ws .pipe span.x { border: 1px dashed var(--red); background: var(--red-bg); border-radius: 6px; padding: 3px 8px; text-decoration: line-through; }
  .ws .ordrow { display: grid; grid-template-columns: minmax(0, 1fr) 130px; gap: 8px; align-items: center; padding: 7px 0; border-bottom: 1px solid #e5e9f0; }
  .ws .ordrow b { display: block; } .ws .ordrow .d { font-size: 12.5px; }
  .ws .chk-chip { display: inline-block; font-size: 12px; border-radius: 10px; padding: 1px 7px; margin: 2px 3px 2px 0; border: 1px solid; }
  .ws .chk-chip.y { background: var(--green-bg); border-color: #9fd5ad; } .ws .chk-chip.n { background: var(--red-bg); border-color: #efa9a3; font-weight: bold; }
  .ws .tilerow { display: grid; grid-template-columns: 250px minmax(0, 1fr) 150px; gap: 12px; align-items: start; padding: 10px 0; border-bottom: 1px solid #e5e9f0; }
  .ws .main.narrow .tilerow { grid-template-columns: minmax(0, 1fr); }
  .ws .tilerow img { width: 100%; max-width: 250px; border-radius: 6px; border: 1px solid var(--line); display: block; cursor: zoom-in; image-rendering: pixelated; }
  .ws .file { border: 1px solid var(--line); border-radius: 8px; padding: 10px 12px; margin-bottom: 10px; }
  .ws .preview { width: 100%; height: 480px; border: 1px solid var(--line); border-radius: 8px; background: #fff; }
  /* 3단계 · 힌트 · 작업 진행 · 그래프 */
  .ws .jobbar { border: 1px solid var(--line); border-radius: 8px; padding: 8px 10px; margin: 8px 0; background: #f8fafc; }
  .ws .jobbar.err { border-color: #efa9a3; background: var(--red-bg); }
  .ws .jobbar .jb-top { display: flex; gap: 8px; align-items: center; font-size: 13px; flex-wrap: wrap; } .ws .jobbar .sp { flex: 1; }
  .ws .jobbar .pb { height: 10px; background: #e3e8ef; border-radius: 6px; overflow: hidden; margin-top: 6px; }
  .ws .jobbar .pb i { display: block; height: 100%; background: linear-gradient(90deg, #0b57d0, #38bdf8); transition: width .5s; }
  .ws .hintbox { border: 1px dashed #d4a72c; background: #fffbea; border-radius: 8px; padding: 6px 10px; margin: 6px 0; font-size: 13px; }
  .ws .hintbox .hb-h { display: flex; align-items: center; gap: 8px; } .ws .hintbox .sp { flex: 1; }
  .ws .hintbox .hb-i { margin-top: 5px; line-height: 1.55; }
  .ws .raw .rl { border-radius: 4px; padding: 0 3px; } .ws .raw .rl.hl { background: #a16207; } .ws .raw .rl.hl, .ws .raw .rl.hl * { color: #fffbea; }
  .ws .srcline { font-size: 12.5px; margin: 2px 0 4px; }
  .ws .srcchip { font: 12px Consolas, 'D2Coding', Menlo, monospace; border: 1px solid var(--blue); background: var(--blue-bg); border-radius: 5px; padding: 1px 6px; cursor: pointer; }
  .ws .convrow { border-top: 1px solid #eef2f7; padding-top: 8px; }
  .ws table.t.conv td, .ws table.t.conv th { font-size: 12.5px; padding: 4px 8px; }
  .ws .params { display: flex; flex-wrap: wrap; gap: 8px 14px; align-items: center; margin: 6px 0; font-size: 13px; }
  .ws .params label { display: flex; gap: 6px; align-items: center; flex-wrap: wrap; }
  .ws .metric { margin-top: 4px; font-size: 13px; background: #ecfdf5; border: 1px solid #a7f3d0; border-radius: 6px; padding: 4px 8px; line-height: 1.5; }
  .ws .chart { margin: 6px 0; flex: 1; min-width: 260px; max-width: 440px; } .ws .chart svg { width: 100%; height: auto; display: block; }
  .ws .card.warnc { border-color: #e3c96b; background: #fffdf3; }
  .ws .grid2 { display: grid; grid-template-columns: repeat(auto-fill, minmax(330px, 1fr)); gap: 12px; }
  .ws .bar { display: inline-block; height: 9px; background: var(--blue); border-radius: 4px; vertical-align: middle; }
  .ws .stress { border-top: 1px solid #e5e9f0; padding-top: 8px; margin-top: 10px; } .ws .stress h5 { margin: 0 0 4px; font-size: 14px; }
  .ws .zoombox { position: absolute; inset: 0; background: rgba(15, 23, 42, .84); z-index: 50; display: flex; align-items: center; justify-content: center; cursor: zoom-out; }
  .ws .zoombox .zb-in { max-width: 94%; max-height: 94%; text-align: center; }
  .ws .zoombox img { max-width: 100%; max-height: 78vh; image-rendering: pixelated; border-radius: 8px; background: #fff; }
  .ws .zoombox .zb-cap, .ws .zoombox .zb-cap * { color: #fff; font-size: 13px; margin-top: 6px; }
  .ws .gal { display: grid; grid-template-columns: repeat(auto-fill, minmax(300px, 1fr)); gap: 10px; margin-top: 6px; }
  .ws .gal-i { display: grid; grid-template-columns: 1fr 1fr; gap: 4px; align-content: start; } .ws .gal-i.one { grid-template-columns: 1fr; }
  .ws .gal-i img { width: 100%; border-radius: 6px; border: 1px solid var(--line); cursor: zoom-in; display: block; }
  .ws .gal-i .hint { grid-column: 1 / -1; }
  .ws .roles { display: grid; grid-template-columns: repeat(auto-fill, minmax(150px, 1fr)); gap: 8px; margin: 6px 0; }
  .ws .role { border-radius: 8px; padding: 8px 10px; border: 1px solid; font-size: 13px; }
  .ws .role.red { background: #fef2f2; border-color: #fca5a5; } .ws .role.blue { background: #eff6ff; border-color: #93c5fd; }
  .ws .cues { display: grid; grid-template-columns: repeat(auto-fill, minmax(260px, 1fr)); gap: 8px; margin: 6px 0; }
  .ws .cue { border: 1px solid var(--line); border-radius: 8px; padding: 8px; background: #fff; }
  .ws .cue .ctl { margin-top: 4px; }
  .ws .cardgrid { display: grid; grid-template-columns: repeat(auto-fill, minmax(210px, 1fr)); gap: 8px; margin-top: 8px; }
  .ws .ccard { border: 1px solid var(--line); border-radius: 8px; padding: 6px; background: #fff; } .ws .ccard.flag { border: 2px solid #dc2626; }
  .ws .ccard img { width: 96px; height: 96px; image-rendering: pixelated; border-radius: 4px; }
  .ws tr.locked-row td { opacity: .55; }
  .ws pre.jsonbox { white-space: pre; }
  /* 모델 들여다보기 · 4단계 */
  .ws .archwrap { overflow-x: auto; border: 1px solid var(--line); border-radius: 8px; background: #f8fafc; padding: 6px; margin: 6px 0; }
  .ws .archwrap svg { display: block; }
  .ws .vnav { position: sticky; top: 0; background: #fff; z-index: 2; padding: 4px 0; } .ws .vnav .sp { flex: 1; }
  .ws .vstep { border-top: 1px solid #e5e9f0; padding-top: 8px; } .ws .vstep h5 { font-size: 15px; margin: 4px 0; }
  .ws pre.formula { background: #0f172a; border-radius: 8px; padding: 8px 10px; white-space: pre-wrap; font: 12.5px/1.6 Consolas, 'D2Coding', Menlo, monospace; }
  .ws pre.formula, .ws pre.formula * { color: #e2e8f0; }
  .ws .example { background: #f0f9ff; border: 1px solid #bae6fd; border-radius: 8px; padding: 8px 10px; margin: 8px 0; }
  .ws table.num3 { border-collapse: collapse; font: 12px Consolas, 'D2Coding', Menlo, monospace; } .ws table.num3 td { border: 1px solid #cbd5e1; padding: 3px 6px; text-align: right; background: #fff; }
  .ws table.num3.w td { background: #fef3c7; }
  .ws .example .big { font-size: 18px; font-weight: bold; } .ws .calcline { margin-top: 6px; font: 13px/1.6 Consolas, 'D2Coding', Menlo, monospace; }
  .ws .fmaps { display: flex; flex-wrap: wrap; gap: 6px; } .ws .fmaps img { width: 64px; image-rendering: pixelated; border-radius: 4px; border: 1px solid var(--line); }
  .ws .legend7 { display: flex; flex-wrap: wrap; gap: 10px; font-size: 12.5px; margin: 6px 0; } .ws .legend7 i { display: inline-block; width: 14px; height: 10px; margin-right: 4px; border-radius: 2px; }
  .ws .maptile { display: grid; grid-template-columns: minmax(0, 300px) minmax(0, 1fr); gap: 10px; border-top: 1px solid #e5e9f0; padding: 8px 0; }
  .ws .main.narrow .maptile { grid-template-columns: minmax(0, 1fr); }
  .ws .maptile img { width: 100%; border-radius: 6px; cursor: zoom-in; } .ws .maptile li { margin: 3px 0; font-size: 13px; line-height: 1.5; }
  .ws .sw { display: inline-block; width: 10px; height: 10px; border-radius: 2px; margin-right: 4px; }
  .ws .adjcard { display: grid; grid-template-columns: 170px minmax(0, 1fr); gap: 10px; border-top: 1px solid #e5e9f0; padding: 10px 0; }
  .ws .main.narrow .adjcard { grid-template-columns: minmax(0, 1fr); }
  .ws .adjcard img, .ws .biascard img { width: 160px; max-width: 100%; border-radius: 6px; image-rendering: pixelated; cursor: zoom-in; }
  .ws .biasgrid { display: grid; grid-template-columns: repeat(auto-fill, minmax(170px, 1fr)); gap: 10px; }
  .ws .biascard { border: 1px solid var(--line); border-radius: 8px; padding: 6px; }
  .ws .flow { display: flex; flex-wrap: wrap; gap: 6px; align-items: center; margin: 6px 0; font-size: 13px; }
  .ws .flow .box { border: 1px solid var(--blue); background: var(--blue-bg); border-radius: 6px; padding: 3px 8px; font-weight: bold; }
  .ws .flow .box.auto { border-color: #9ca3af; background: #f3f4f6; }
  /* 무거운 작업 중 안내 · 가이드북 · 4·5단계 */
  .ws .stay { border: 2px solid #d97706; background: #fffbeb; border-radius: 10px; padding: 10px 12px; margin: 8px 0; font-size: 13.5px; line-height: 1.55; }
  .ws .stay b { color: #92400e; display: block; margin-bottom: 2px; }
  .ws .stay.shake { animation: wsShake .5s; } @keyframes wsShake { 0%, 100% { transform: translateX(0); } 25% { transform: translateX(-6px); } 75% { transform: translateX(6px); } }
  .ws .gtabs { display: flex; gap: 4px; padding: 6px 10px 0; background: #e2e8f0; }
  .ws .gtabs button { flex: 1; border: 1px solid var(--line); border-bottom: none; border-radius: 8px 8px 0 0; background: #f1f5f9; padding: 6px 4px; font-size: 13px; cursor: pointer; color: var(--ink); }
  .ws .gtabs button.on { background: #fff; font-weight: bold; color: var(--blue); }
  .ws .codep .term { margin: 8px 0; } .ws .term.now { border: 2px solid #f59e0b; background: #fffbeb; } .ws .term.flash { box-shadow: 0 0 0 3px #fde68a; }
  .ws .term dt.k4 { color: #6d28d9; } .ws .term dt.k5 { color: #0f766e; } .ws .term dd.fx { font: 12px/1.6 Consolas, 'D2Coding', Menlo, monospace; background: #f5f3ff; border-radius: 6px; padding: 2px 6px; }
  .ws .codep .term dl { grid-template-columns: 70px 1fr; font-size: 12.5px; }
  .ws tr.wrongrow td { background: #fef2f2; }
  .ws .heatwrap { display: grid; grid-template-columns: minmax(0, 1.25fr) minmax(0, 1fr); gap: 12px; align-items: start; margin-top: 8px; }
  .ws .main.narrow .heatwrap { grid-template-columns: minmax(0, 1fr); }
  .ws .heatimg { width: 100%; border-radius: 8px; border: 1px solid var(--line); cursor: zoom-in; }
  .ws details.case { border: 1px solid var(--line); border-radius: 8px; padding: 4px 10px; margin: 8px 0; background: #fbfcfe; }
  .ws details.case > summary { color: var(--ink); }
  .ws .casefig { margin: 4px 8px 4px 0; } .ws .casefig img { width: 200px; max-width: 100%; border-radius: 6px; border: 1px solid var(--line); cursor: zoom-in; image-rendering: pixelated; }
  .ws .casefig figcaption { font-size: 12px; }
  .ws .timeline { display: grid; gap: 3px; margin: 6px 0; font-size: 13px; } .ws .timeline > div { display: grid; grid-template-columns: 70px minmax(0, 1fr); gap: 6px; }
  .ws .voices { border-top: 1px dashed var(--line); margin-top: 6px; padding-top: 6px; }
  .ws .voice { display: flex; gap: 8px; align-items: flex-start; margin: 5px 0; font-size: 13px; line-height: 1.5; }
  .ws .voice .avatar { width: 24px; height: 24px; border-radius: 6px; color: #fff; font-size: 12px; display: flex; align-items: center; justify-content: center; flex-shrink: 0; }
  .ws .actgrp { border-top: 1px solid #e5e9f0; padding: 8px 0; } .ws .actgrp small, .ws .secgrid small { display: block; color: #475569; }
  .ws .flow .fbox { border: 1px solid var(--blue); background: var(--blue-bg); border-radius: 8px; padding: 5px 8px; max-width: 150px; }
  .ws .flow .fbox small { display: block; color: #475569; } .ws .flow .fbox .hint { font-size: 11.5px; } .ws .flow .arr { font-weight: bold; color: var(--blue); }
  .ws .orditem { display: flex; gap: 8px; align-items: center; border: 1px solid var(--line); border-radius: 8px; padding: 6px 8px; margin: 4px 0; background: #fff; }
  .ws .orditem small { color: #475569; } .ws .orditem.wrong { border: 2px solid var(--red); background: #fef2f2; }
  .ws .secgrid { display: grid; grid-template-columns: repeat(auto-fill, minmax(260px, 1fr)); gap: 4px 10px; }
  .ws .lint { display: grid; gap: 4px; margin: 6px 0; } .ws .qa { border-top: 1px dashed var(--line); padding-top: 6px; margin-top: 6px; }

  .ws .gstep { margin: 8px 0 4px; } .ws .gstep-h { font-size: 12px; font-weight: bold; color: #6b21a8; background: #f5f3ff; border-radius: 6px; padding: 3px 8px; margin: 6px 0 4px; }
  .ws .term .fig { margin: 4px 0 6px; } .ws .term .fig svg { width: 100%; max-width: 360px; height: auto; display: block; }
  /* 미션 블록 (명세로 만든 단계 화면) */
  .ws .stepfoot { margin: 14px 0 8px; }
  .ws .blkimg { max-width: 100%; border-radius: 8px; border: 1px solid var(--line); cursor: zoom-in; display: block; margin: 6px 0; }
  /* 글쓰기 블록: 제출한 뒤에 보이는 모범답안 (정답 판정 없음) */
  .ws .model { margin: 10px 0 2px; padding: 10px 12px; border: 1px solid #a8c3f0; border-radius: 8px; background: var(--blue-bg); }
  .ws .model-h { font-weight: bold; color: #0a2a66; margin-bottom: 2px; }
  .ws .model p { margin: 4px 0; line-height: 1.65; }
  /* 첫 화면 (입력 없이 인사만) */
  .ws .onb.welcome { text-align: center; max-width: 560px; margin: 8vh auto 0; }
  .ws .onb.welcome .wave { font-size: 44px; }
  .ws .onb.welcome h2 { font-size: 26px; margin: 8px 0 10px; }
  .ws .onb.welcome .lead { line-height: 1.7; }
  .ws .me-chip { cursor: default; }
  /* 해외 사례 토글 */
  .ws details.case { margin-top: 8px; border: 1px solid var(--line); border-radius: 8px; background: #fff; }
  .ws details.case > summary { cursor: pointer; padding: 7px 10px; font-weight: bold; font-size: 13px; }
  .ws details.case .case-b { padding: 2px 12px 10px; }
  .ws details.case .case-b p { margin: 6px 0; line-height: 1.65; font-size: 13.5px; }
  /* 산출물 단계 */
  .ws ul.todo { margin: 6px 0; padding-left: 20px; }
  .ws ul.todo li { margin: 4px 0; line-height: 1.55; }
  /* 바이브 코딩 실습 */
  .ws .vb-slot { margin: 10px 0; padding: 8px 10px; border: 1px solid var(--line); border-radius: 8px; background: var(--soft); }
  .ws .vb-label { margin: 8px 0 4px; font-size: 13.5px; }
  .ws .vb-slot .vb-label { margin-top: 0; }
  .ws .vb-prompt { background: #fff; border: 2px solid var(--blue); border-radius: 8px; padding: 9px 11px; white-space: pre-wrap; line-height: 1.6; font-family: inherit; font-size: 13.5px; margin: 4px 0 8px; }
  .ws .vb-prompt.empty { border-style: dashed; border-color: #8a94a6; }
  .ws .vb-replybox { scroll-margin-top: 8px; }
  .ws .vb-reply { margin-top: 12px; padding: 10px 12px; border: 1px solid #a8c3f0; border-left: 5px solid var(--blue); border-radius: 8px; background: #f8fbff; }
  .ws .vb-reply.ok { border-color: #9fd5ad; border-left-color: var(--green); background: #f6fbf7; }
  .ws .vb-who { font-weight: bold; margin-bottom: 4px; }
  .ws .vb-say { line-height: 1.6; margin-bottom: 6px; }
  .ws details.vb-codebox { margin: 6px 0; }
  .ws details.vb-codebox > summary { cursor: pointer; font-size: 13px; font-weight: bold; padding: 2px 0; }
  .ws .vb-code { background: #1e2430; color: #e6edf3; border-radius: 8px; padding: 9px 11px; margin: 4px 0; overflow-x: auto; font-family: Menlo, Consolas, monospace; font-size: 12.5px; line-height: 1.5; white-space: pre; }
  .ws .vb-code span { color: #e6edf3; }
  .ws .vb-code .c { color: #8b949e; } .ws .vb-code .s { color: #a5d6ff; } .ws .vb-code .k { color: #ff9e8e; } .ws .vb-code .n { color: #79c0ff; } .ws .vb-code .f { color: #d2a8ff; }
  .ws .vb-out { background: #fff; border: 1px solid var(--line); border-radius: 8px; padding: 8px 10px; margin: 4px 0; overflow-x: auto; font-family: Menlo, Consolas, monospace; font-size: 12.5px; white-space: pre-wrap; }
  .ws .vb-err { background: var(--red-bg); border: 1px solid #e6a8a2; border-radius: 8px; padding: 8px 10px; margin: 4px 0; overflow-x: auto; font-family: Menlo, Consolas, monospace; font-size: 12.5px; white-space: pre-wrap; }
  .ws .vb-tbl { overflow-x: auto; margin: 4px 0; }
  .ws .vb-tbl table { border-collapse: collapse; background: #fff; }
  .ws .vb-tbl th, .ws .vb-tbl td { border: 1px solid var(--line); padding: 3px 8px; font-size: 12.5px; text-align: left; white-space: nowrap; }
  .ws .vb-tbl th { background: #eef2f7; }
  .ws iframe.vb-doc { width: 100%; height: 300px; border: 1px solid var(--line); border-radius: 8px; background: #fff; }
  .ws .vb-problem { margin-top: 8px; padding: 7px 10px; border-radius: 8px; background: var(--red-bg); border: 1px solid #efa9a3; color: #5c0f0a; line-height: 1.6; }
  .ws .vb-problem * { color: inherit; }
  .ws .vb-reply.miss { border-color: #efa9a3; border-left-color: var(--red); background: #fffafa; }
  .ws details.vb-data { margin: 8px 0; }
  .ws details.vb-data > summary, .ws details.vb-log > summary { cursor: pointer; font-size: 13px; font-weight: bold; padding: 2px 0; }
  .ws details.vb-log { margin-top: 10px; }
  .ws details.vb-log th { white-space: nowrap; }
  .ws details.vb-log td:first-child, .ws details.vb-log td:nth-child(2) { white-space: nowrap; }
  .ws .model-t { white-space: pre-wrap; line-height: 1.6; }
  .ws .selfcheck { margin-top: 10px; }
  .ws .selfcheck .choice { display: flex; margin: 4px 0; }
  /* 틀린 선택: 주황색 대신 빨간색으로 (2026-11-05 피드백) */
  .ws .choice.wrong, .ws .choice.on.wrong { border: 2px solid var(--red); background: var(--red-bg); padding: 5px 9px; }
  .ws .choice.wrong input { accent-color: var(--red); }
  .ws .hintbox.wrongbox { border: 1px solid #efa9a3; background: var(--red-bg); color: #5c0f0a; }
  .ws .hintbox.wrongbox * { color: inherit; }
  .ws tr.wrongrow td { background: var(--red-bg); }
  .ws tr.wrongrow select { border: 2px solid var(--red); }
  /* 부제목 */
  .ws .stage-sub { margin: -4px 0 8px; }
  .ws .msub { margin-top: 2px; }
</style>

<div class="ws" id="ws-root">
  <div class="topbar">
    <button class="tb" id="tb-side" onclick="toggleSide()" title="왼쪽 메뉴 접기/펴기">☰<span class="count" id="tb-count"></span></button>
    <span class="name">__WORKSPACE__</span>
    <span class="tprog" id="tb-prog" title="전체 미션 진행률"></span>
    <span class="spacer"></span>
    <span class="me-chip" id="tb-me"></span>
    <button class="tb" id="tb-code" onclick="toggleCode()" title="핵심 용어와 지금 화면의 코드 보기">📘 가이드북</button>
    <span>화면 폭</span>
    <button class="tb" data-w="base" onclick="setWidth('base')">기본</button>
    <button class="tb" data-w="wide" onclick="setWidth('wide')">넓게</button>
    <button class="tb" data-w="full" onclick="setWidth('full')">가득</button>
    <span>높이</span>
    <button class="tb" onclick="setHeight(-120)" title="낮게">−</button>
    <button class="tb" onclick="setHeight(120)" title="높게">+</button>
  </div>
  <div class="body">
    <div class="side" id="side"></div>
    <div class="main" id="main"></div>
    <div class="codep" id="codep"></div>
    <button class="code-tab" id="code-tab" onclick="toggleCode(true)">📘 본 과정의 가이드북 ◀</button>
  </div>
  <div class="footer">
    <span class="note" id="note"></span>
    <button class="btn ghost small" onclick="resetAll()">↺ 처음부터</button>
  </div>
</div>

<script>
// ===================== 프로젝트 설정 (만드는 도구가 넣음) =====================
const PROJ = {"id": "pbl_21b0fe0c", "team_name": "실루엣 식별 교육개발반", "ui_key": "pbl-ui-pbl_21b0fe0c", "boot_note": "🔄 파이썬이 새로 시작돼 진행 상황이 처음으로 돌아갔어요.", "welcome_lead": "가상의 실루엣 식별 교육개발반에 바이브 코딩 담당으로 합류했어요. 팀원들의 요청을 프롬프트로 바꿔 AI와 함께 도구와 문서를 만들어 봐요. M1부터 차례로 진행해 보세요.", "use_label": "바이브 코딩에서는", "stay_jobs": {}};   // 팀 이름 · 저장 키 · 안내 문구 · 작업 중 잠글 작업
const PROJ_FIGS = {"f_loop": "<svg viewBox=\"0 0 320 156\" xmlns=\"http://www.w3.org/2000/svg\" font-family=\"sans-serif\" font-size=\"10\" fill=\"#111827\"><defs><marker id=\"tfa\" viewBox=\"0 0 10 10\" refX=\"8\" refY=\"5\" markerWidth=\"6\" markerHeight=\"6\" orient=\"auto\"><path d=\"M0,0 L10,5 L0,10 z\" fill=\"#475569\"/><\/marker><\/defs><ellipse cx=\"160.0\" cy=\"70\" rx=\"112\" ry=\"48\" fill=\"none\" stroke=\"#e5e7eb\" stroke-width=\"6\"/><path d=\"M198.6,24.9 L248.1,40.3\" stroke=\"#2563eb\" stroke-width=\"2\" marker-end=\"url(#tfa)\" fill=\"none\"/><path d=\"M271.9,71.8 L253.0,96.7\" stroke=\"#2563eb\" stroke-width=\"2\" marker-end=\"url(#tfa)\" fill=\"none\"/><path d=\"M190.6,116.2 L129.4,116.2\" stroke=\"#2563eb\" stroke-width=\"2\" marker-end=\"url(#tfa)\" fill=\"none\"/><path d=\"M67.0,96.7 L48.1,71.8\" stroke=\"#2563eb\" stroke-width=\"2\" marker-end=\"url(#tfa)\" fill=\"none\"/><path d=\"M71.9,40.3 L121.4,24.9\" stroke=\"#2563eb\" stroke-width=\"2\" marker-end=\"url(#tfa)\" fill=\"none\"/><rect x=\"116.0\" y=\"10.0\" width=\"88.0\" height=\"24.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"160.0\" y=\"25.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">요청<\/text><rect x=\"222.5\" y=\"43.2\" width=\"88.0\" height=\"24.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"266.5\" y=\"58.7\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">생성<\/text><rect x=\"181.8\" y=\"96.8\" width=\"88.0\" height=\"24.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"225.8\" y=\"112.3\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">실행<\/text><rect x=\"50.2\" y=\"96.8\" width=\"88.0\" height=\"24.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"94.2\" y=\"112.3\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">확인<\/text><rect x=\"9.5\" y=\"43.2\" width=\"88.0\" height=\"24.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"53.5\" y=\"58.7\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">수정 요청<\/text><text x=\"160.0\" y=\"74.0\" font-size=\"10.5\" text-anchor=\"middle\" fill=\"#475569\" font-weight=\"bold\">바이브 코딩<\/text><text x=\"160.0\" y=\"151.0\" font-size=\"9\" text-anchor=\"middle\" fill=\"#475569\">확인하고 다시 요청하는 일이 핵심이에요<\/text><\/svg>", "f_four": "<svg viewBox=\"0 0 320 120\" xmlns=\"http://www.w3.org/2000/svg\" font-family=\"sans-serif\" font-size=\"10\" fill=\"#111827\"><defs><marker id=\"tfa\" viewBox=\"0 0 10 10\" refX=\"8\" refY=\"5\" markerWidth=\"6\" markerHeight=\"6\" orient=\"auto\"><path d=\"M0,0 L10,5 L0,10 z\" fill=\"#475569\"/><\/marker><\/defs><rect x=\"6.0\" y=\"8\" width=\"154.0\" height=\"18\" fill=\"#e5e7eb\" stroke=\"#fff\"/><text x=\"83.0\" y=\"20.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">칸<\/text><rect x=\"160.0\" y=\"8\" width=\"154.0\" height=\"18\" fill=\"#e5e7eb\" stroke=\"#fff\"/><text x=\"237.0\" y=\"20.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">무엇을 적나<\/text><rect x=\"6.0\" y=\"26\" width=\"154.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"83.0\" y=\"38.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">맥락<\/text><rect x=\"160.0\" y=\"26\" width=\"154.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"237.0\" y=\"38.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">왜 필요한지<\/text><rect x=\"6.0\" y=\"44\" width=\"154.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"83.0\" y=\"56.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">입력<\/text><rect x=\"160.0\" y=\"44\" width=\"154.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"237.0\" y=\"56.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">파일·열·값<\/text><rect x=\"6.0\" y=\"62\" width=\"154.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"83.0\" y=\"74.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">할 일<\/text><rect x=\"160.0\" y=\"62\" width=\"154.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"237.0\" y=\"74.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">규칙을 숫자로<\/text><rect x=\"6.0\" y=\"80\" width=\"154.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"83.0\" y=\"92.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">출력 형식<\/text><rect x=\"160.0\" y=\"80\" width=\"154.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"237.0\" y=\"92.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">받고 싶은 모양<\/text><text x=\"160.0\" y=\"115.0\" font-size=\"9\" text-anchor=\"middle\" fill=\"#475569\">네 칸을 채우면 AI가 추측할 일이 줄어요<\/text><\/svg>", "f_weft": "<svg viewBox=\"0 0 320 120\" xmlns=\"http://www.w3.org/2000/svg\" font-family=\"sans-serif\" font-size=\"10\" fill=\"#111827\"><defs><marker id=\"tfa\" viewBox=\"0 0 10 10\" refX=\"8\" refY=\"5\" markerWidth=\"6\" markerHeight=\"6\" orient=\"auto\"><path d=\"M0,0 L10,5 L0,10 z\" fill=\"#475569\"/><\/marker><\/defs><rect x=\"6.0\" y=\"8\" width=\"154.0\" height=\"18\" fill=\"#e5e7eb\" stroke=\"#fff\"/><text x=\"83.0\" y=\"20.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">부분<\/text><rect x=\"160.0\" y=\"8\" width=\"154.0\" height=\"18\" fill=\"#e5e7eb\" stroke=\"#fff\"/><text x=\"237.0\" y=\"20.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">보는 것<\/text><rect x=\"6.0\" y=\"26\" width=\"154.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"83.0\" y=\"38.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">날개 W<\/text><rect x=\"160.0\" y=\"26\" width=\"154.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"237.0\" y=\"38.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">모양과 위치<\/text><rect x=\"6.0\" y=\"44\" width=\"154.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"83.0\" y=\"56.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">엔진 E<\/text><rect x=\"160.0\" y=\"44\" width=\"154.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"237.0\" y=\"56.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">종류와 위치<\/text><rect x=\"6.0\" y=\"62\" width=\"154.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"83.0\" y=\"74.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">동체 F<\/text><rect x=\"160.0\" y=\"62\" width=\"154.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"237.0\" y=\"74.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">굵기와 길이<\/text><rect x=\"6.0\" y=\"80\" width=\"154.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"83.0\" y=\"92.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">꼬리 T<\/text><rect x=\"160.0\" y=\"80\" width=\"154.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"237.0\" y=\"92.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">모양<\/text><text x=\"160.0\" y=\"115.0\" font-size=\"9\" text-anchor=\"middle\" fill=\"#475569\">네 부분으로 나눠 보면 단서가 생겨요<\/text><\/svg>", "f_johnson": "<svg viewBox=\"0 0 320 96\" xmlns=\"http://www.w3.org/2000/svg\" font-family=\"sans-serif\" font-size=\"10\" fill=\"#111827\"><defs><marker id=\"tfa\" viewBox=\"0 0 10 10\" refX=\"8\" refY=\"5\" markerWidth=\"6\" markerHeight=\"6\" orient=\"auto\"><path d=\"M0,0 L10,5 L0,10 z\" fill=\"#475569\"/><\/marker><\/defs><rect x=\"8.0\" y=\"18\" width=\"99.3\" height=\"30\" rx=\"5\" fill=\"rgb(37,99,235)\" fill-opacity=\"0.35\"/><text x=\"57.7\" y=\"37.0\" font-size=\"9.5\" text-anchor=\"middle\" fill=\"#111827\">탐지<\/text><text x=\"57.7\" y=\"62.0\" font-size=\"8\" text-anchor=\"middle\" fill=\"#475569\">2픽셀~<\/text><rect x=\"109.3\" y=\"18\" width=\"99.3\" height=\"30\" rx=\"5\" fill=\"rgb(128,68,136)\" fill-opacity=\"0.35\"/><text x=\"159.0\" y=\"37.0\" font-size=\"9.5\" text-anchor=\"middle\" fill=\"#111827\">인식<\/text><text x=\"159.0\" y=\"62.0\" font-size=\"8\" text-anchor=\"middle\" fill=\"#475569\">8픽셀~<\/text><rect x=\"210.7\" y=\"18\" width=\"99.3\" height=\"30\" rx=\"5\" fill=\"rgb(220,38,38)\" fill-opacity=\"0.35\"/><text x=\"260.3\" y=\"37.0\" font-size=\"9.5\" text-anchor=\"middle\" fill=\"#111827\">식별<\/text><text x=\"260.3\" y=\"62.0\" font-size=\"8\" text-anchor=\"middle\" fill=\"#475569\">13픽셀~<\/text><text x=\"160.0\" y=\"91.0\" font-size=\"9\" text-anchor=\"middle\" fill=\"#475569\">크게 보일수록 더 자세히 알 수 있어요<\/text><\/svg>", "f_schema": "<svg viewBox=\"0 0 320 102\" xmlns=\"http://www.w3.org/2000/svg\" font-family=\"sans-serif\" font-size=\"10\" fill=\"#111827\"><defs><marker id=\"tfa\" viewBox=\"0 0 10 10\" refX=\"8\" refY=\"5\" markerWidth=\"6\" markerHeight=\"6\" orient=\"auto\"><path d=\"M0,0 L10,5 L0,10 z\" fill=\"#475569\"/><\/marker><\/defs><rect x=\"6.0\" y=\"8\" width=\"154.0\" height=\"18\" fill=\"#e5e7eb\" stroke=\"#fff\"/><text x=\"83.0\" y=\"20.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">칸<\/text><rect x=\"160.0\" y=\"8\" width=\"154.0\" height=\"18\" fill=\"#e5e7eb\" stroke=\"#fff\"/><text x=\"237.0\" y=\"20.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">값<\/text><rect x=\"6.0\" y=\"26\" width=\"154.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"83.0\" y=\"38.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">형상<\/text><rect x=\"160.0\" y=\"26\" width=\"154.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"237.0\" y=\"38.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">여섯 가지 소문자<\/text><rect x=\"6.0\" y=\"44\" width=\"154.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"83.0\" y=\"56.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">정체<\/text><rect x=\"160.0\" y=\"44\" width=\"154.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"237.0\" y=\"56.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">미상이 기본값<\/text><rect x=\"6.0\" y=\"62\" width=\"154.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"83.0\" y=\"74.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">배경<\/text><rect x=\"160.0\" y=\"62\" width=\"154.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"237.0\" y=\"74.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">sky·field·urban<\/text><text x=\"160.0\" y=\"97.0\" font-size=\"9\" text-anchor=\"middle\" fill=\"#475569\">모양과 정체를 다른 칸에 적어요<\/text><\/svg>", "f_kappa": "<svg viewBox=\"0 0 320 92\" xmlns=\"http://www.w3.org/2000/svg\" font-family=\"sans-serif\" font-size=\"10\" fill=\"#111827\"><defs><marker id=\"tfa\" viewBox=\"0 0 10 10\" refX=\"8\" refY=\"5\" markerWidth=\"6\" markerHeight=\"6\" orient=\"auto\"><path d=\"M0,0 L10,5 L0,10 z\" fill=\"#475569\"/><\/marker><\/defs><rect x=\"8.0\" y=\"14.0\" width=\"92.0\" height=\"46.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"54.0\" y=\"34.0\" font-size=\"10\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">일치율<\/text><text x=\"54.0\" y=\"47.0\" font-size=\"8.5\" text-anchor=\"middle\" fill=\"#475569\">실제로 같음<\/text><path d=\"M101.0,37.0 L113.0,37.0\" stroke=\"#475569\" stroke-width=\"1.6\" marker-end=\"url(#tfa)\" fill=\"none\"/><rect x=\"114.0\" y=\"14.0\" width=\"92.0\" height=\"46.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"160.0\" y=\"34.0\" font-size=\"10\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">우연<\/text><text x=\"160.0\" y=\"47.0\" font-size=\"8.5\" text-anchor=\"middle\" fill=\"#475569\">그냥 같을 확률<\/text><path d=\"M207.0,37.0 L219.0,37.0\" stroke=\"#475569\" stroke-width=\"1.6\" marker-end=\"url(#tfa)\" fill=\"none\"/><rect x=\"220.0\" y=\"14.0\" width=\"92.0\" height=\"46.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"266.0\" y=\"34.0\" font-size=\"10\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">코헨 카파<\/text><text x=\"266.0\" y=\"47.0\" font-size=\"8.5\" text-anchor=\"middle\" fill=\"#475569\">우연을 뺀 값<\/text><text x=\"160.0\" y=\"87.0\" font-size=\"9\" text-anchor=\"middle\" fill=\"#475569\">우연히 같은 만큼을 빼고 봐요<\/text><\/svg>", "f_clean": "<svg viewBox=\"0 0 320 92\" xmlns=\"http://www.w3.org/2000/svg\" font-family=\"sans-serif\" font-size=\"10\" fill=\"#111827\"><defs><marker id=\"tfa\" viewBox=\"0 0 10 10\" refX=\"8\" refY=\"5\" markerWidth=\"6\" markerHeight=\"6\" orient=\"auto\"><path d=\"M0,0 L10,5 L0,10 z\" fill=\"#475569\"/><\/marker><\/defs><rect x=\"8.0\" y=\"14.0\" width=\"92.0\" height=\"46.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"54.0\" y=\"34.0\" font-size=\"10\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">Bird<\/text><text x=\"54.0\" y=\"47.0\" font-size=\"8.5\" text-anchor=\"middle\" fill=\"#475569\">대문자<\/text><path d=\"M101.0,37.0 L113.0,37.0\" stroke=\"#475569\" stroke-width=\"1.6\" marker-end=\"url(#tfa)\" fill=\"none\"/><rect x=\"114.0\" y=\"14.0\" width=\"92.0\" height=\"46.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"160.0\" y=\"34.0\" font-size=\"10\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">bird <\/text><text x=\"160.0\" y=\"47.0\" font-size=\"8.5\" text-anchor=\"middle\" fill=\"#475569\">빈칸<\/text><path d=\"M207.0,37.0 L219.0,37.0\" stroke=\"#475569\" stroke-width=\"1.6\" marker-end=\"url(#tfa)\" fill=\"none\"/><rect x=\"220.0\" y=\"14.0\" width=\"92.0\" height=\"46.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"266.0\" y=\"34.0\" font-size=\"10\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">bird<\/text><text x=\"266.0\" y=\"47.0\" font-size=\"8.5\" text-anchor=\"middle\" fill=\"#475569\">같은 값<\/text><text x=\"160.0\" y=\"87.0\" font-size=\"9\" text-anchor=\"middle\" fill=\"#475569\">제각각인 값을 하나로 맞춰요<\/text><\/svg>", "f_base": "<svg viewBox=\"0 0 320 92\" xmlns=\"http://www.w3.org/2000/svg\" font-family=\"sans-serif\" font-size=\"10\" fill=\"#111827\"><defs><marker id=\"tfa\" viewBox=\"0 0 10 10\" refX=\"8\" refY=\"5\" markerWidth=\"6\" markerHeight=\"6\" orient=\"auto\"><path d=\"M0,0 L10,5 L0,10 z\" fill=\"#475569\"/><\/marker><\/defs><rect x=\"8.0\" y=\"14.0\" width=\"65.5\" height=\"46.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"40.8\" y=\"40.5\" font-size=\"9.5\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">단순한 모델<\/text><path d=\"M74.5,37.0 L86.5,37.0\" stroke=\"#475569\" stroke-width=\"1.6\" marker-end=\"url(#tfa)\" fill=\"none\"/><rect x=\"87.5\" y=\"14.0\" width=\"65.5\" height=\"46.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"120.2\" y=\"40.5\" font-size=\"10\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">성능 재기<\/text><path d=\"M154.0,37.0 L166.0,37.0\" stroke=\"#475569\" stroke-width=\"1.6\" marker-end=\"url(#tfa)\" fill=\"none\"/><rect x=\"167.0\" y=\"14.0\" width=\"65.5\" height=\"46.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"199.8\" y=\"40.5\" font-size=\"9.5\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">복잡한 모델<\/text><path d=\"M233.5,37.0 L245.5,37.0\" stroke=\"#475569\" stroke-width=\"1.6\" marker-end=\"url(#tfa)\" fill=\"none\"/><rect x=\"246.5\" y=\"14.0\" width=\"65.5\" height=\"46.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"279.2\" y=\"40.5\" font-size=\"10\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">차이 비교<\/text><text x=\"160.0\" y=\"87.0\" font-size=\"9\" text-anchor=\"middle\" fill=\"#475569\">단순한 모델이 비교의 자가 돼요<\/text><\/svg>", "f_pre": "<svg viewBox=\"0 0 320 92\" xmlns=\"http://www.w3.org/2000/svg\" font-family=\"sans-serif\" font-size=\"10\" fill=\"#111827\"><defs><marker id=\"tfa\" viewBox=\"0 0 10 10\" refX=\"8\" refY=\"5\" markerWidth=\"6\" markerHeight=\"6\" orient=\"auto\"><path d=\"M0,0 L10,5 L0,10 z\" fill=\"#475569\"/><\/marker><\/defs><rect x=\"8.0\" y=\"14.0\" width=\"92.0\" height=\"46.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"54.0\" y=\"40.5\" font-size=\"10\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">잡음 제거<\/text><path d=\"M101.0,37.0 L113.0,37.0\" stroke=\"#475569\" stroke-width=\"1.6\" marker-end=\"url(#tfa)\" fill=\"none\"/><rect x=\"114.0\" y=\"14.0\" width=\"92.0\" height=\"46.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"160.0\" y=\"40.5\" font-size=\"10\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">크기 맞춤<\/text><path d=\"M207.0,37.0 L219.0,37.0\" stroke=\"#475569\" stroke-width=\"1.6\" marker-end=\"url(#tfa)\" fill=\"none\"/><rect x=\"220.0\" y=\"14.0\" width=\"92.0\" height=\"46.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"266.0\" y=\"40.5\" font-size=\"10\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">특징 계산<\/text><text x=\"160.0\" y=\"87.0\" font-size=\"9\" text-anchor=\"middle\" fill=\"#475569\">같은 기준으로 맞춘 뒤 숫자로 바꿔요<\/text><\/svg>", "f_debug": "<svg viewBox=\"0 0 320 92\" xmlns=\"http://www.w3.org/2000/svg\" font-family=\"sans-serif\" font-size=\"10\" fill=\"#111827\"><defs><marker id=\"tfa\" viewBox=\"0 0 10 10\" refX=\"8\" refY=\"5\" markerWidth=\"6\" markerHeight=\"6\" orient=\"auto\"><path d=\"M0,0 L10,5 L0,10 z\" fill=\"#475569\"/><\/marker><\/defs><rect x=\"8.0\" y=\"14.0\" width=\"92.0\" height=\"46.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"54.0\" y=\"40.5\" font-size=\"9.5\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">오류 메시지<\/text><path d=\"M101.0,37.0 L113.0,37.0\" stroke=\"#475569\" stroke-width=\"1.6\" marker-end=\"url(#tfa)\" fill=\"none\"/><rect x=\"114.0\" y=\"14.0\" width=\"92.0\" height=\"46.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"160.0\" y=\"40.5\" font-size=\"9.5\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">원하는 동작<\/text><path d=\"M207.0,37.0 L219.0,37.0\" stroke=\"#475569\" stroke-width=\"1.6\" marker-end=\"url(#tfa)\" fill=\"none\"/><rect x=\"220.0\" y=\"14.0\" width=\"92.0\" height=\"46.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"266.0\" y=\"40.5\" font-size=\"10\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">고칠 범위<\/text><text x=\"160.0\" y=\"87.0\" font-size=\"9\" text-anchor=\"middle\" fill=\"#475569\">이 세 가지를 담아 다시 요청해요<\/text><\/svg>", "f_req": "<svg viewBox=\"0 0 320 120\" xmlns=\"http://www.w3.org/2000/svg\" font-family=\"sans-serif\" font-size=\"10\" fill=\"#111827\"><defs><marker id=\"tfa\" viewBox=\"0 0 10 10\" refX=\"8\" refY=\"5\" markerWidth=\"6\" markerHeight=\"6\" orient=\"auto\"><path d=\"M0,0 L10,5 L0,10 z\" fill=\"#475569\"/><\/marker><\/defs><rect x=\"6.0\" y=\"8\" width=\"154.0\" height=\"18\" fill=\"#e5e7eb\" stroke=\"#fff\"/><text x=\"83.0\" y=\"20.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">항목<\/text><rect x=\"160.0\" y=\"8\" width=\"154.0\" height=\"18\" fill=\"#e5e7eb\" stroke=\"#fff\"/><text x=\"237.0\" y=\"20.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">기준<\/text><rect x=\"6.0\" y=\"26\" width=\"154.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"83.0\" y=\"38.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">재현율<\/text><rect x=\"160.0\" y=\"26\" width=\"154.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"237.0\" y=\"38.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">90% 이상<\/text><rect x=\"6.0\" y=\"44\" width=\"154.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"83.0\" y=\"56.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">속도<\/text><rect x=\"160.0\" y=\"44\" width=\"154.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"237.0\" y=\"56.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">100ms 이하<\/text><rect x=\"6.0\" y=\"62\" width=\"154.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"83.0\" y=\"74.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">크기<\/text><rect x=\"160.0\" y=\"62\" width=\"154.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"237.0\" y=\"74.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">30MB 이하<\/text><rect x=\"6.0\" y=\"80\" width=\"154.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"83.0\" y=\"92.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">외부 전송<\/text><rect x=\"160.0\" y=\"80\" width=\"154.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"237.0\" y=\"92.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">없음<\/text><text x=\"160.0\" y=\"115.0\" font-size=\"9\" text-anchor=\"middle\" fill=\"#475569\">모델을 보기 전에 합격선을 정해요<\/text><\/svg>", "f_tl": "<svg viewBox=\"0 0 320 92\" xmlns=\"http://www.w3.org/2000/svg\" font-family=\"sans-serif\" font-size=\"10\" fill=\"#111827\"><defs><marker id=\"tfa\" viewBox=\"0 0 10 10\" refX=\"8\" refY=\"5\" markerWidth=\"6\" markerHeight=\"6\" orient=\"auto\"><path d=\"M0,0 L10,5 L0,10 z\" fill=\"#475569\"/><\/marker><\/defs><rect x=\"8.0\" y=\"14.0\" width=\"92.0\" height=\"46.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"54.0\" y=\"34.0\" font-size=\"10\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">큰 모델<\/text><text x=\"54.0\" y=\"47.0\" font-size=\"8.5\" text-anchor=\"middle\" fill=\"#475569\">이미 학습됨<\/text><path d=\"M101.0,37.0 L113.0,37.0\" stroke=\"#475569\" stroke-width=\"1.6\" marker-end=\"url(#tfa)\" fill=\"none\"/><rect x=\"114.0\" y=\"14.0\" width=\"92.0\" height=\"46.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"160.0\" y=\"34.0\" font-size=\"10\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">앞부분<\/text><text x=\"160.0\" y=\"47.0\" font-size=\"8.5\" text-anchor=\"middle\" fill=\"#475569\">그대로<\/text><path d=\"M207.0,37.0 L219.0,37.0\" stroke=\"#475569\" stroke-width=\"1.6\" marker-end=\"url(#tfa)\" fill=\"none\"/><rect x=\"220.0\" y=\"14.0\" width=\"92.0\" height=\"46.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"266.0\" y=\"34.0\" font-size=\"10\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">끝부분<\/text><text x=\"266.0\" y=\"47.0\" font-size=\"8.5\" text-anchor=\"middle\" fill=\"#475569\">새로 학습<\/text><text x=\"160.0\" y=\"87.0\" font-size=\"9\" text-anchor=\"middle\" fill=\"#475569\">이미 배운 것을 빌려 와요<\/text><\/svg>", "f_param": "<svg viewBox=\"0 0 320 92\" xmlns=\"http://www.w3.org/2000/svg\" font-family=\"sans-serif\" font-size=\"10\" fill=\"#111827\"><defs><marker id=\"tfa\" viewBox=\"0 0 10 10\" refX=\"8\" refY=\"5\" markerWidth=\"6\" markerHeight=\"6\" orient=\"auto\"><path d=\"M0,0 L10,5 L0,10 z\" fill=\"#475569\"/><\/marker><\/defs><rect x=\"8.0\" y=\"14.0\" width=\"65.5\" height=\"46.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"40.8\" y=\"40.5\" font-size=\"10\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">작은 함수<\/text><path d=\"M74.5,37.0 L86.5,37.0\" stroke=\"#475569\" stroke-width=\"1.6\" marker-end=\"url(#tfa)\" fill=\"none\"/><rect x=\"87.5\" y=\"14.0\" width=\"65.5\" height=\"46.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"120.2\" y=\"40.5\" font-size=\"9.5\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">예시로 확인<\/text><path d=\"M154.0,37.0 L166.0,37.0\" stroke=\"#475569\" stroke-width=\"1.6\" marker-end=\"url(#tfa)\" fill=\"none\"/><rect x=\"167.0\" y=\"14.0\" width=\"65.5\" height=\"46.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"199.8\" y=\"40.5\" font-size=\"10\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">전체 적용<\/text><path d=\"M233.5,37.0 L245.5,37.0\" stroke=\"#475569\" stroke-width=\"1.6\" marker-end=\"url(#tfa)\" fill=\"none\"/><rect x=\"246.5\" y=\"14.0\" width=\"65.5\" height=\"46.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"279.2\" y=\"40.5\" font-size=\"10\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">비교표<\/text><text x=\"160.0\" y=\"87.0\" font-size=\"9\" text-anchor=\"middle\" fill=\"#475569\">작게 만들고 확인한 뒤 넓혀요<\/text><\/svg>", "f_rel": "<svg viewBox=\"0 0 320 150\" xmlns=\"http://www.w3.org/2000/svg\" font-family=\"sans-serif\" font-size=\"10\" fill=\"#111827\"><defs><marker id=\"tfa\" viewBox=\"0 0 10 10\" refX=\"8\" refY=\"5\" markerWidth=\"6\" markerHeight=\"6\" orient=\"auto\"><path d=\"M0,0 L10,5 L0,10 z\" fill=\"#475569\"/><\/marker><\/defs><rect x=\"34\" y=\"12\" width=\"274\" height=\"108\" fill=\"#fff\" stroke=\"#94a3b8\"/><line x1=\"34.0\" y1=\"120.0\" x2=\"308.0\" y2=\"12.0\" stroke=\"#94a3b8\" stroke-dasharray=\"4 3\"/><text x=\"304.0\" y=\"24.0\" font-size=\"8.5\" text-anchor=\"end\" fill=\"#475569\">완벽<\/text><polyline points=\"157.3,98.4 184.7,87.6 212.1,76.8 239.5,60.6 266.9,44.4 294.3,25.0\" fill=\"none\" stroke=\"#2563eb\" stroke-width=\"2\"/><circle cx=\"157.3\" cy=\"98.4\" r=\"2.5\" fill=\"#2563eb\"/><circle cx=\"184.7\" cy=\"87.6\" r=\"2.5\" fill=\"#2563eb\"/><circle cx=\"212.1\" cy=\"76.8\" r=\"2.5\" fill=\"#2563eb\"/><circle cx=\"239.5\" cy=\"60.6\" r=\"2.5\" fill=\"#2563eb\"/><circle cx=\"266.9\" cy=\"44.4\" r=\"2.5\" fill=\"#2563eb\"/><circle cx=\"294.3\" cy=\"25.0\" r=\"2.5\" fill=\"#2563eb\"/><text x=\"171.0\" y=\"142.0\" font-size=\"9\" text-anchor=\"middle\" fill=\"#475569\">mean score<\/text><text x=\"10\" y=\"66.0\" font-size=\"9\" text-anchor=\"middle\" fill=\"#475569\" transform=\"rotate(-90 10 66.0)\">actual accuracy<\/text><text x=\"40.0\" y=\"24.0\" font-size=\"8.5\" text-anchor=\"start\" fill=\"#475569\">대각선 아래면 점수가 부풀어 있어요<\/text><\/svg>", "f_ece": "<svg viewBox=\"0 0 320 84\" xmlns=\"http://www.w3.org/2000/svg\" font-family=\"sans-serif\" font-size=\"10\" fill=\"#111827\"><defs><marker id=\"tfa\" viewBox=\"0 0 10 10\" refX=\"8\" refY=\"5\" markerWidth=\"6\" markerHeight=\"6\" orient=\"auto\"><path d=\"M0,0 L10,5 L0,10 z\" fill=\"#475569\"/><\/marker><\/defs><rect x=\"6.0\" y=\"8\" width=\"77.0\" height=\"18\" fill=\"#e5e7eb\" stroke=\"#fff\"/><text x=\"44.5\" y=\"20.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">구간<\/text><rect x=\"83.0\" y=\"8\" width=\"77.0\" height=\"18\" fill=\"#e5e7eb\" stroke=\"#fff\"/><text x=\"121.5\" y=\"20.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">평균 점수<\/text><rect x=\"160.0\" y=\"8\" width=\"77.0\" height=\"18\" fill=\"#e5e7eb\" stroke=\"#fff\"/><text x=\"198.5\" y=\"20.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">실제<\/text><rect x=\"237.0\" y=\"8\" width=\"77.0\" height=\"18\" fill=\"#e5e7eb\" stroke=\"#fff\"/><text x=\"275.5\" y=\"20.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">차이<\/text><rect x=\"6.0\" y=\"26\" width=\"77.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"44.5\" y=\"38.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">0.9~1.0<\/text><rect x=\"83.0\" y=\"26\" width=\"77.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"121.5\" y=\"38.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">0.95<\/text><rect x=\"160.0\" y=\"26\" width=\"77.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"198.5\" y=\"38.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">0.88<\/text><rect x=\"237.0\" y=\"26\" width=\"77.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"275.5\" y=\"38.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">0.07<\/text><rect x=\"6.0\" y=\"44\" width=\"77.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"44.5\" y=\"56.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">0.8~0.9<\/text><rect x=\"83.0\" y=\"44\" width=\"77.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"121.5\" y=\"56.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">0.85<\/text><rect x=\"160.0\" y=\"44\" width=\"77.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"198.5\" y=\"56.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">0.70<\/text><rect x=\"237.0\" y=\"44\" width=\"77.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"275.5\" y=\"56.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">0.15<\/text><text x=\"160.0\" y=\"79.0\" font-size=\"9\" text-anchor=\"middle\" fill=\"#475569\">차이를 표본 수로 가중 평균해요<\/text><\/svg>", "f_grade": "<svg viewBox=\"0 0 320 92\" xmlns=\"http://www.w3.org/2000/svg\" font-family=\"sans-serif\" font-size=\"10\" fill=\"#111827\"><defs><marker id=\"tfa\" viewBox=\"0 0 10 10\" refX=\"8\" refY=\"5\" markerWidth=\"6\" markerHeight=\"6\" orient=\"auto\"><path d=\"M0,0 L10,5 L0,10 z\" fill=\"#475569\"/><\/marker><\/defs><rect x=\"8.0\" y=\"14.0\" width=\"92.0\" height=\"46.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"54.0\" y=\"34.0\" font-size=\"9.5\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">참고 판정 표시<\/text><text x=\"54.0\" y=\"47.0\" font-size=\"8.5\" text-anchor=\"middle\" fill=\"#475569\">확신 높음<\/text><path d=\"M101.0,37.0 L113.0,37.0\" stroke=\"#475569\" stroke-width=\"1.6\" marker-end=\"url(#tfa)\" fill=\"none\"/><rect x=\"114.0\" y=\"14.0\" width=\"92.0\" height=\"46.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"160.0\" y=\"34.0\" font-size=\"10\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">교관 확인<\/text><text x=\"160.0\" y=\"47.0\" font-size=\"8.5\" text-anchor=\"middle\" fill=\"#475569\">중간<\/text><path d=\"M207.0,37.0 L219.0,37.0\" stroke=\"#475569\" stroke-width=\"1.6\" marker-end=\"url(#tfa)\" fill=\"none\"/><rect x=\"220.0\" y=\"14.0\" width=\"92.0\" height=\"46.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"266.0\" y=\"34.0\" font-size=\"10\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">판별 불가<\/text><text x=\"266.0\" y=\"47.0\" font-size=\"8.5\" text-anchor=\"middle\" fill=\"#475569\">1·2위 차이 작음<\/text><text x=\"160.0\" y=\"87.0\" font-size=\"9\" text-anchor=\"middle\" fill=\"#475569\">모르면 사람에게 넘겨요<\/text><\/svg>", "f_leak": "<svg viewBox=\"0 0 320 92\" xmlns=\"http://www.w3.org/2000/svg\" font-family=\"sans-serif\" font-size=\"10\" fill=\"#111827\"><defs><marker id=\"tfa\" viewBox=\"0 0 10 10\" refX=\"8\" refY=\"5\" markerWidth=\"6\" markerHeight=\"6\" orient=\"auto\"><path d=\"M0,0 L10,5 L0,10 z\" fill=\"#475569\"/><\/marker><\/defs><rect x=\"8.0\" y=\"14.0\" width=\"145.0\" height=\"46.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"80.5\" y=\"34.0\" font-size=\"9.5\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">검증 데이터<\/text><text x=\"80.5\" y=\"47.0\" font-size=\"8.5\" text-anchor=\"middle\" fill=\"#475569\">기준 고르기<\/text><path d=\"M154.0,37.0 L166.0,37.0\" stroke=\"#475569\" stroke-width=\"1.6\" marker-end=\"url(#tfa)\" fill=\"none\"/><rect x=\"167.0\" y=\"14.0\" width=\"145.0\" height=\"46.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"239.5\" y=\"34.0\" font-size=\"9.5\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">테스트 데이터<\/text><text x=\"239.5\" y=\"47.0\" font-size=\"8.5\" text-anchor=\"middle\" fill=\"#475569\">한 번 확인<\/text><text x=\"160.0\" y=\"87.0\" font-size=\"9\" text-anchor=\"middle\" fill=\"#475569\">고를 때 쓴 데이터로 성능을 재면 좋게 보여요<\/text><\/svg>", "f_verify": "<svg viewBox=\"0 0 320 92\" xmlns=\"http://www.w3.org/2000/svg\" font-family=\"sans-serif\" font-size=\"10\" fill=\"#111827\"><defs><marker id=\"tfa\" viewBox=\"0 0 10 10\" refX=\"8\" refY=\"5\" markerWidth=\"6\" markerHeight=\"6\" orient=\"auto\"><path d=\"M0,0 L10,5 L0,10 z\" fill=\"#475569\"/><\/marker><\/defs><rect x=\"8.0\" y=\"14.0\" width=\"65.5\" height=\"46.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"40.8\" y=\"40.5\" font-size=\"10\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">AI 답변<\/text><path d=\"M74.5,37.0 L86.5,37.0\" stroke=\"#475569\" stroke-width=\"1.6\" marker-end=\"url(#tfa)\" fill=\"none\"/><rect x=\"87.5\" y=\"14.0\" width=\"65.5\" height=\"46.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"120.2\" y=\"40.5\" font-size=\"10\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">근거 수치<\/text><path d=\"M154.0,37.0 L166.0,37.0\" stroke=\"#475569\" stroke-width=\"1.6\" marker-end=\"url(#tfa)\" fill=\"none\"/><rect x=\"167.0\" y=\"14.0\" width=\"65.5\" height=\"46.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"199.8\" y=\"40.5\" font-size=\"10\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">작은 예시<\/text><path d=\"M233.5,37.0 L245.5,37.0\" stroke=\"#475569\" stroke-width=\"1.6\" marker-end=\"url(#tfa)\" fill=\"none\"/><rect x=\"246.5\" y=\"14.0\" width=\"65.5\" height=\"46.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"279.2\" y=\"40.5\" font-size=\"10\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">판단<\/text><text x=\"160.0\" y=\"87.0\" font-size=\"9\" text-anchor=\"middle\" fill=\"#475569\">답변만 믿지 말고 근거를 함께 받아요<\/text><\/svg>", "f_pair": "<svg viewBox=\"0 0 320 84\" xmlns=\"http://www.w3.org/2000/svg\" font-family=\"sans-serif\" font-size=\"10\" fill=\"#111827\"><defs><marker id=\"tfa\" viewBox=\"0 0 10 10\" refX=\"8\" refY=\"5\" markerWidth=\"6\" markerHeight=\"6\" orient=\"auto\"><path d=\"M0,0 L10,5 L0,10 z\" fill=\"#475569\"/><\/marker><\/defs><rect x=\"6.0\" y=\"8\" width=\"102.7\" height=\"18\" fill=\"#e5e7eb\" stroke=\"#fff\"/><text x=\"57.3\" y=\"20.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">형상 계열<\/text><rect x=\"108.7\" y=\"8\" width=\"102.7\" height=\"18\" fill=\"#e5e7eb\" stroke=\"#fff\"/><text x=\"160.0\" y=\"20.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">헷갈리는 계열<\/text><rect x=\"211.3\" y=\"8\" width=\"102.7\" height=\"18\" fill=\"#e5e7eb\" stroke=\"#fff\"/><text x=\"262.7\" y=\"20.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">오답률<\/text><rect x=\"6.0\" y=\"26\" width=\"102.7\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"57.3\" y=\"38.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">delta_wing<\/text><rect x=\"108.7\" y=\"26\" width=\"102.7\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"160.0\" y=\"38.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">fixed_wing<\/text><rect x=\"211.3\" y=\"26\" width=\"102.7\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"262.7\" y=\"38.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">높음<\/text><rect x=\"6.0\" y=\"44\" width=\"102.7\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"57.3\" y=\"56.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">multicopter<\/text><rect x=\"108.7\" y=\"44\" width=\"102.7\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"160.0\" y=\"56.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">bird<\/text><rect x=\"211.3\" y=\"44\" width=\"102.7\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"262.7\" y=\"56.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">중간<\/text><text x=\"160.0\" y=\"79.0\" font-size=\"9\" text-anchor=\"middle\" fill=\"#475569\">자주 헷갈리는 둘을 묶어 내요<\/text><\/svg>", "f_adapt": "<svg viewBox=\"0 0 320 92\" xmlns=\"http://www.w3.org/2000/svg\" font-family=\"sans-serif\" font-size=\"10\" fill=\"#111827\"><defs><marker id=\"tfa\" viewBox=\"0 0 10 10\" refX=\"8\" refY=\"5\" markerWidth=\"6\" markerHeight=\"6\" orient=\"auto\"><path d=\"M0,0 L10,5 L0,10 z\" fill=\"#475569\"/><\/marker><\/defs><rect x=\"8.0\" y=\"14.0\" width=\"65.5\" height=\"46.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"40.8\" y=\"40.5\" font-size=\"10\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">학습 기록<\/text><path d=\"M74.5,37.0 L86.5,37.0\" stroke=\"#475569\" stroke-width=\"1.6\" marker-end=\"url(#tfa)\" fill=\"none\"/><rect x=\"87.5\" y=\"14.0\" width=\"65.5\" height=\"46.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"120.2\" y=\"40.5\" font-size=\"10\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">오답률<\/text><path d=\"M154.0,37.0 L166.0,37.0\" stroke=\"#475569\" stroke-width=\"1.6\" marker-end=\"url(#tfa)\" fill=\"none\"/><rect x=\"167.0\" y=\"14.0\" width=\"65.5\" height=\"46.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"199.8\" y=\"40.5\" font-size=\"10\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">출제 배수<\/text><path d=\"M233.5,37.0 L245.5,37.0\" stroke=\"#475569\" stroke-width=\"1.6\" marker-end=\"url(#tfa)\" fill=\"none\"/><rect x=\"246.5\" y=\"14.0\" width=\"65.5\" height=\"46.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"279.2\" y=\"40.5\" font-size=\"10\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">다음 문항<\/text><text x=\"160.0\" y=\"87.0\" font-size=\"9\" text-anchor=\"middle\" fill=\"#475569\">약한 부분을 더 많이 연습해요<\/text><\/svg>", "f_edge": "<svg viewBox=\"0 0 320 102\" xmlns=\"http://www.w3.org/2000/svg\" font-family=\"sans-serif\" font-size=\"10\" fill=\"#111827\"><defs><marker id=\"tfa\" viewBox=\"0 0 10 10\" refX=\"8\" refY=\"5\" markerWidth=\"6\" markerHeight=\"6\" orient=\"auto\"><path d=\"M0,0 L10,5 L0,10 z\" fill=\"#475569\"/><\/marker><\/defs><rect x=\"6.0\" y=\"8\" width=\"102.7\" height=\"18\" fill=\"#e5e7eb\" stroke=\"#fff\"/><text x=\"57.3\" y=\"20.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">shown<\/text><rect x=\"108.7\" y=\"8\" width=\"102.7\" height=\"18\" fill=\"#e5e7eb\" stroke=\"#fff\"/><text x=\"160.0\" y=\"20.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">wrong<\/text><rect x=\"211.3\" y=\"8\" width=\"102.7\" height=\"18\" fill=\"#e5e7eb\" stroke=\"#fff\"/><text x=\"262.7\" y=\"20.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">처리<\/text><rect x=\"6.0\" y=\"26\" width=\"102.7\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"57.3\" y=\"38.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">40<\/text><rect x=\"108.7\" y=\"26\" width=\"102.7\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"160.0\" y=\"38.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">19<\/text><rect x=\"211.3\" y=\"26\" width=\"102.7\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"262.7\" y=\"38.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">우선 출제<\/text><rect x=\"6.0\" y=\"44\" width=\"102.7\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"57.3\" y=\"56.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">3<\/text><rect x=\"108.7\" y=\"44\" width=\"102.7\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"160.0\" y=\"56.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">2<\/text><rect x=\"211.3\" y=\"44\" width=\"102.7\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"262.7\" y=\"56.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">표본 적음<\/text><rect x=\"6.0\" y=\"62\" width=\"102.7\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"57.3\" y=\"74.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">0<\/text><rect x=\"108.7\" y=\"62\" width=\"102.7\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"160.0\" y=\"74.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">0<\/text><rect x=\"211.3\" y=\"62\" width=\"102.7\" height=\"18\" fill=\"#fef3c7\" stroke=\"#e5e7eb\"/><text x=\"262.7\" y=\"74.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">새 문항<\/text><text x=\"160.0\" y=\"97.0\" font-size=\"9\" text-anchor=\"middle\" fill=\"#475569\">0과 적은 표본을 어떻게 할지 미리 알려 주세요<\/text><\/svg>", "f_hitl": "<svg viewBox=\"0 0 320 92\" xmlns=\"http://www.w3.org/2000/svg\" font-family=\"sans-serif\" font-size=\"10\" fill=\"#111827\"><defs><marker id=\"tfa\" viewBox=\"0 0 10 10\" refX=\"8\" refY=\"5\" markerWidth=\"6\" markerHeight=\"6\" orient=\"auto\"><path d=\"M0,0 L10,5 L0,10 z\" fill=\"#475569\"/><\/marker><\/defs><rect x=\"8.0\" y=\"14.0\" width=\"92.0\" height=\"46.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"54.0\" y=\"40.5\" font-size=\"10\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">AI 초안<\/text><path d=\"M101.0,37.0 L113.0,37.0\" stroke=\"#475569\" stroke-width=\"1.6\" marker-end=\"url(#tfa)\" fill=\"none\"/><rect x=\"114.0\" y=\"14.0\" width=\"92.0\" height=\"46.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"160.0\" y=\"40.5\" font-size=\"10\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">사람 검토<\/text><path d=\"M207.0,37.0 L219.0,37.0\" stroke=\"#475569\" stroke-width=\"1.6\" marker-end=\"url(#tfa)\" fill=\"none\"/><rect x=\"220.0\" y=\"14.0\" width=\"92.0\" height=\"46.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"266.0\" y=\"40.5\" font-size=\"9.5\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">승인 후 공개<\/text><text x=\"160.0\" y=\"87.0\" font-size=\"9\" text-anchor=\"middle\" fill=\"#475569\">승인 전에는 공개하지 않아요<\/text><\/svg>", "f_raci": "<svg viewBox=\"0 0 320 102\" xmlns=\"http://www.w3.org/2000/svg\" font-family=\"sans-serif\" font-size=\"10\" fill=\"#111827\"><defs><marker id=\"tfa\" viewBox=\"0 0 10 10\" refX=\"8\" refY=\"5\" markerWidth=\"6\" markerHeight=\"6\" orient=\"auto\"><path d=\"M0,0 L10,5 L0,10 z\" fill=\"#475569\"/><\/marker><\/defs><rect x=\"6.0\" y=\"8\" width=\"77.0\" height=\"18\" fill=\"#e5e7eb\" stroke=\"#fff\"/><text x=\"44.5\" y=\"20.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">일<\/text><rect x=\"83.0\" y=\"8\" width=\"77.0\" height=\"18\" fill=\"#e5e7eb\" stroke=\"#fff\"/><text x=\"121.5\" y=\"20.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">AI<\/text><rect x=\"160.0\" y=\"8\" width=\"77.0\" height=\"18\" fill=\"#e5e7eb\" stroke=\"#fff\"/><text x=\"198.5\" y=\"20.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">교관<\/text><rect x=\"237.0\" y=\"8\" width=\"77.0\" height=\"18\" fill=\"#e5e7eb\" stroke=\"#fff\"/><text x=\"275.5\" y=\"20.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">전산<\/text><rect x=\"6.0\" y=\"26\" width=\"77.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"44.5\" y=\"38.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">초안<\/text><rect x=\"83.0\" y=\"26\" width=\"77.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"121.5\" y=\"38.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">R<\/text><rect x=\"160.0\" y=\"26\" width=\"77.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"198.5\" y=\"38.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">C<\/text><rect x=\"237.0\" y=\"26\" width=\"77.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"275.5\" y=\"38.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">-<\/text><rect x=\"6.0\" y=\"44\" width=\"77.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"44.5\" y=\"56.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">승인<\/text><rect x=\"83.0\" y=\"44\" width=\"77.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"121.5\" y=\"56.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">-<\/text><rect x=\"160.0\" y=\"44\" width=\"77.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"198.5\" y=\"56.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">A<\/text><rect x=\"237.0\" y=\"44\" width=\"77.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"275.5\" y=\"56.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">I<\/text><rect x=\"6.0\" y=\"62\" width=\"77.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"44.5\" y=\"74.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">반영<\/text><rect x=\"83.0\" y=\"62\" width=\"77.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"121.5\" y=\"74.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">-<\/text><rect x=\"160.0\" y=\"62\" width=\"77.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"198.5\" y=\"74.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">I<\/text><rect x=\"237.0\" y=\"62\" width=\"77.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"275.5\" y=\"74.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">R<\/text><text x=\"160.0\" y=\"97.0\" font-size=\"9\" text-anchor=\"middle\" fill=\"#475569\">R 수행, A 책임, C 의논, I 공유<\/text><\/svg>", "f_halluc": "<svg viewBox=\"0 0 320 92\" xmlns=\"http://www.w3.org/2000/svg\" font-family=\"sans-serif\" font-size=\"10\" fill=\"#111827\"><defs><marker id=\"tfa\" viewBox=\"0 0 10 10\" refX=\"8\" refY=\"5\" markerWidth=\"6\" markerHeight=\"6\" orient=\"auto\"><path d=\"M0,0 L10,5 L0,10 z\" fill=\"#475569\"/><\/marker><\/defs><rect x=\"8.0\" y=\"14.0\" width=\"92.0\" height=\"46.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"54.0\" y=\"40.5\" font-size=\"9.5\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">자료에 없음<\/text><path d=\"M101.0,37.0 L113.0,37.0\" stroke=\"#475569\" stroke-width=\"1.6\" marker-end=\"url(#tfa)\" fill=\"none\"/><rect x=\"114.0\" y=\"14.0\" width=\"92.0\" height=\"46.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"160.0\" y=\"34.0\" font-size=\"9.5\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">AI가 지어냄<\/text><text x=\"160.0\" y=\"47.0\" font-size=\"8.5\" text-anchor=\"middle\" fill=\"#475569\">그럴듯함<\/text><path d=\"M207.0,37.0 L219.0,37.0\" stroke=\"#475569\" stroke-width=\"1.6\" marker-end=\"url(#tfa)\" fill=\"none\"/><rect x=\"220.0\" y=\"14.0\" width=\"92.0\" height=\"46.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"266.0\" y=\"34.0\" font-size=\"10\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">사실 확인<\/text><text x=\"266.0\" y=\"47.0\" font-size=\"8.5\" text-anchor=\"middle\" fill=\"#475569\">자료와 대조<\/text><text x=\"160.0\" y=\"87.0\" font-size=\"9\" text-anchor=\"middle\" fill=\"#475569\">없는 내용은 ‘자료 없음’으로 쓰게 해요<\/text><\/svg>", "f_prune": "<svg viewBox=\"0 0 320 92\" xmlns=\"http://www.w3.org/2000/svg\" font-family=\"sans-serif\" font-size=\"10\" fill=\"#111827\"><defs><marker id=\"tfa\" viewBox=\"0 0 10 10\" refX=\"8\" refY=\"5\" markerWidth=\"6\" markerHeight=\"6\" orient=\"auto\"><path d=\"M0,0 L10,5 L0,10 z\" fill=\"#475569\"/><\/marker><\/defs><rect x=\"8.0\" y=\"14.0\" width=\"92.0\" height=\"46.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"54.0\" y=\"34.0\" font-size=\"10\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">원래 모델<\/text><text x=\"54.0\" y=\"47.0\" font-size=\"8.5\" text-anchor=\"middle\" fill=\"#475569\">크고 느림<\/text><path d=\"M101.0,37.0 L113.0,37.0\" stroke=\"#475569\" stroke-width=\"1.6\" marker-end=\"url(#tfa)\" fill=\"none\"/><rect x=\"114.0\" y=\"14.0\" width=\"92.0\" height=\"46.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"160.0\" y=\"34.0\" font-size=\"10\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">가지치기<\/text><text x=\"160.0\" y=\"47.0\" font-size=\"8.5\" text-anchor=\"middle\" fill=\"#475569\">작고 빠름<\/text><path d=\"M207.0,37.0 L219.0,37.0\" stroke=\"#475569\" stroke-width=\"1.6\" marker-end=\"url(#tfa)\" fill=\"none\"/><rect x=\"220.0\" y=\"14.0\" width=\"92.0\" height=\"46.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"266.0\" y=\"34.0\" font-size=\"10\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">미세조정<\/text><text x=\"266.0\" y=\"47.0\" font-size=\"8.5\" text-anchor=\"middle\" fill=\"#475569\">성능 회복<\/text><text x=\"160.0\" y=\"87.0\" font-size=\"9\" text-anchor=\"middle\" fill=\"#475569\">줄인 뒤 다시 조금 학습해요<\/text><\/svg>", "f_safe": "<svg viewBox=\"0 0 320 102\" xmlns=\"http://www.w3.org/2000/svg\" font-family=\"sans-serif\" font-size=\"10\" fill=\"#111827\"><defs><marker id=\"tfa\" viewBox=\"0 0 10 10\" refX=\"8\" refY=\"5\" markerWidth=\"6\" markerHeight=\"6\" orient=\"auto\"><path d=\"M0,0 L10,5 L0,10 z\" fill=\"#475569\"/><\/marker><\/defs><rect x=\"6.0\" y=\"8\" width=\"154.0\" height=\"18\" fill=\"#e5e7eb\" stroke=\"#fff\"/><text x=\"83.0\" y=\"20.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">상황<\/text><rect x=\"160.0\" y=\"8\" width=\"154.0\" height=\"18\" fill=\"#e5e7eb\" stroke=\"#fff\"/><text x=\"237.0\" y=\"20.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">나와야 할 것<\/text><rect x=\"6.0\" y=\"26\" width=\"154.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"83.0\" y=\"38.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">심한 흐림<\/text><rect x=\"160.0\" y=\"26\" width=\"154.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"237.0\" y=\"38.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">판별 불가<\/text><rect x=\"6.0\" y=\"44\" width=\"154.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"83.0\" y=\"56.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">손상 파일<\/text><rect x=\"160.0\" y=\"44\" width=\"154.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"237.0\" y=\"56.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">판별 불가<\/text><rect x=\"6.0\" y=\"62\" width=\"154.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"83.0\" y=\"74.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">선명한 영상<\/text><rect x=\"160.0\" y=\"62\" width=\"154.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"237.0\" y=\"74.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">참고 라벨<\/text><text x=\"160.0\" y=\"97.0\" font-size=\"9\" text-anchor=\"middle\" fill=\"#475569\">모르면 답을 지어내지 않아요<\/text><\/svg>", "f_secret": "<svg viewBox=\"0 0 320 102\" xmlns=\"http://www.w3.org/2000/svg\" font-family=\"sans-serif\" font-size=\"10\" fill=\"#111827\"><defs><marker id=\"tfa\" viewBox=\"0 0 10 10\" refX=\"8\" refY=\"5\" markerWidth=\"6\" markerHeight=\"6\" orient=\"auto\"><path d=\"M0,0 L10,5 L0,10 z\" fill=\"#475569\"/><\/marker><\/defs><rect x=\"6.0\" y=\"8\" width=\"154.0\" height=\"18\" fill=\"#e5e7eb\" stroke=\"#fff\"/><text x=\"83.0\" y=\"20.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">넣어도 돼요<\/text><rect x=\"160.0\" y=\"8\" width=\"154.0\" height=\"18\" fill=\"#e5e7eb\" stroke=\"#fff\"/><text x=\"237.0\" y=\"20.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">넣으면 안 돼요<\/text><rect x=\"6.0\" y=\"26\" width=\"154.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"83.0\" y=\"38.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">열 이름<\/text><rect x=\"160.0\" y=\"26\" width=\"154.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"237.0\" y=\"38.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">서버 주소<\/text><rect x=\"6.0\" y=\"44\" width=\"154.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"83.0\" y=\"56.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">가상 샘플<\/text><rect x=\"160.0\" y=\"44\" width=\"154.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"237.0\" y=\"56.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">계정<\/text><rect x=\"6.0\" y=\"62\" width=\"154.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"83.0\" y=\"74.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">오류 메시지<\/text><rect x=\"160.0\" y=\"62\" width=\"154.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"237.0\" y=\"74.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">실명 기록<\/text><text x=\"160.0\" y=\"97.0\" font-size=\"9\" text-anchor=\"middle\" fill=\"#475569\">공개 AI에 넣기 전에 나눠 봐요<\/text><\/svg>", "f_aar": "<svg viewBox=\"0 0 320 156\" xmlns=\"http://www.w3.org/2000/svg\" font-family=\"sans-serif\" font-size=\"10\" fill=\"#111827\"><defs><marker id=\"tfa\" viewBox=\"0 0 10 10\" refX=\"8\" refY=\"5\" markerWidth=\"6\" markerHeight=\"6\" orient=\"auto\"><path d=\"M0,0 L10,5 L0,10 z\" fill=\"#475569\"/><\/marker><\/defs><ellipse cx=\"160.0\" cy=\"70\" rx=\"112\" ry=\"48\" fill=\"none\" stroke=\"#e5e7eb\" stroke-width=\"6\"/><path d=\"M207.7,26.6 L261.3,49.6\" stroke=\"#2563eb\" stroke-width=\"2\" marker-end=\"url(#tfa)\" fill=\"none\"/><path d=\"M261.3,90.4 L207.7,113.4\" stroke=\"#2563eb\" stroke-width=\"2\" marker-end=\"url(#tfa)\" fill=\"none\"/><path d=\"M112.3,113.4 L58.7,90.4\" stroke=\"#2563eb\" stroke-width=\"2\" marker-end=\"url(#tfa)\" fill=\"none\"/><path d=\"M58.7,49.6 L112.3,26.6\" stroke=\"#2563eb\" stroke-width=\"2\" marker-end=\"url(#tfa)\" fill=\"none\"/><rect x=\"116.0\" y=\"10.0\" width=\"88.0\" height=\"24.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"160.0\" y=\"25.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">계획<\/text><rect x=\"228.0\" y=\"58.0\" width=\"88.0\" height=\"24.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"272.0\" y=\"73.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">실제<\/text><rect x=\"116.0\" y=\"106.0\" width=\"88.0\" height=\"24.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"160.0\" y=\"121.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">원인<\/text><rect x=\"4.0\" y=\"58.0\" width=\"88.0\" height=\"24.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"48.0\" y=\"73.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">다음 조치<\/text><text x=\"160.0\" y=\"74.0\" font-size=\"10.5\" text-anchor=\"middle\" fill=\"#475569\" font-weight=\"bold\">AAR<\/text><text x=\"160.0\" y=\"151.0\" font-size=\"9\" text-anchor=\"middle\" fill=\"#475569\">탓하지 않고 바꿀 것을 찾아요<\/text><\/svg>", "f_reg": "<svg viewBox=\"0 0 320 84\" xmlns=\"http://www.w3.org/2000/svg\" font-family=\"sans-serif\" font-size=\"10\" fill=\"#111827\"><defs><marker id=\"tfa\" viewBox=\"0 0 10 10\" refX=\"8\" refY=\"5\" markerWidth=\"6\" markerHeight=\"6\" orient=\"auto\"><path d=\"M0,0 L10,5 L0,10 z\" fill=\"#475569\"/><\/marker><\/defs><rect x=\"6.0\" y=\"8\" width=\"102.7\" height=\"18\" fill=\"#e5e7eb\" stroke=\"#fff\"/><text x=\"57.3\" y=\"20.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">조건<\/text><rect x=\"108.7\" y=\"8\" width=\"102.7\" height=\"18\" fill=\"#e5e7eb\" stroke=\"#fff\"/><text x=\"160.0\" y=\"20.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">전<\/text><rect x=\"211.3\" y=\"8\" width=\"102.7\" height=\"18\" fill=\"#e5e7eb\" stroke=\"#fff\"/><text x=\"262.7\" y=\"20.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">후<\/text><rect x=\"6.0\" y=\"26\" width=\"102.7\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"57.3\" y=\"38.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">far<\/text><rect x=\"108.7\" y=\"26\" width=\"102.7\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"160.0\" y=\"38.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">58<\/text><rect x=\"211.3\" y=\"26\" width=\"102.7\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"262.7\" y=\"38.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">72<\/text><rect x=\"6.0\" y=\"44\" width=\"102.7\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"57.3\" y=\"56.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">near<\/text><rect x=\"108.7\" y=\"44\" width=\"102.7\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"160.0\" y=\"56.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">94<\/text><rect x=\"211.3\" y=\"44\" width=\"102.7\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"262.7\" y=\"56.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">92<\/text><text x=\"160.0\" y=\"79.0\" font-size=\"9\" text-anchor=\"middle\" fill=\"#475569\">평균이 올라도 떨어진 조건을 봐요<\/text><\/svg>", "f_review": "<svg viewBox=\"0 0 320 156\" xmlns=\"http://www.w3.org/2000/svg\" font-family=\"sans-serif\" font-size=\"10\" fill=\"#111827\"><defs><marker id=\"tfa\" viewBox=\"0 0 10 10\" refX=\"8\" refY=\"5\" markerWidth=\"6\" markerHeight=\"6\" orient=\"auto\"><path d=\"M0,0 L10,5 L0,10 z\" fill=\"#475569\"/><\/marker><\/defs><ellipse cx=\"160.0\" cy=\"70\" rx=\"112\" ry=\"48\" fill=\"none\" stroke=\"#e5e7eb\" stroke-width=\"6\"/><path d=\"M207.7,26.6 L261.3,49.6\" stroke=\"#2563eb\" stroke-width=\"2\" marker-end=\"url(#tfa)\" fill=\"none\"/><path d=\"M261.3,90.4 L207.7,113.4\" stroke=\"#2563eb\" stroke-width=\"2\" marker-end=\"url(#tfa)\" fill=\"none\"/><path d=\"M112.3,113.4 L58.7,90.4\" stroke=\"#2563eb\" stroke-width=\"2\" marker-end=\"url(#tfa)\" fill=\"none\"/><path d=\"M58.7,49.6 L112.3,26.6\" stroke=\"#2563eb\" stroke-width=\"2\" marker-end=\"url(#tfa)\" fill=\"none\"/><rect x=\"116.0\" y=\"10.0\" width=\"88.0\" height=\"24.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"160.0\" y=\"25.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">설명<\/text><rect x=\"228.0\" y=\"58.0\" width=\"88.0\" height=\"24.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"272.0\" y=\"73.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">테스트<\/text><rect x=\"116.0\" y=\"106.0\" width=\"88.0\" height=\"24.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"160.0\" y=\"121.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">수정<\/text><rect x=\"4.0\" y=\"58.0\" width=\"88.0\" height=\"24.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"48.0\" y=\"73.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">비교<\/text><text x=\"160.0\" y=\"74.0\" font-size=\"10.5\" text-anchor=\"middle\" fill=\"#475569\" font-weight=\"bold\">리뷰<\/text><text x=\"160.0\" y=\"151.0\" font-size=\"9\" text-anchor=\"middle\" fill=\"#475569\">고치기 전에 먼저 설명하게 해요<\/text><\/svg>", "f_scope": "<svg viewBox=\"0 0 320 92\" xmlns=\"http://www.w3.org/2000/svg\" font-family=\"sans-serif\" font-size=\"10\" fill=\"#111827\"><defs><marker id=\"tfa\" viewBox=\"0 0 10 10\" refX=\"8\" refY=\"5\" markerWidth=\"6\" markerHeight=\"6\" orient=\"auto\"><path d=\"M0,0 L10,5 L0,10 z\" fill=\"#475569\"/><\/marker><\/defs><rect x=\"8.0\" y=\"14.0\" width=\"92.0\" height=\"46.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"54.0\" y=\"34.0\" font-size=\"10\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">고칠 곳<\/text><text x=\"54.0\" y=\"47.0\" font-size=\"8.5\" text-anchor=\"middle\" fill=\"#475569\">이 함수만<\/text><path d=\"M101.0,37.0 L113.0,37.0\" stroke=\"#475569\" stroke-width=\"1.6\" marker-end=\"url(#tfa)\" fill=\"none\"/><rect x=\"114.0\" y=\"14.0\" width=\"92.0\" height=\"46.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"160.0\" y=\"34.0\" font-size=\"10\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">그대로<\/text><text x=\"160.0\" y=\"47.0\" font-size=\"8.5\" text-anchor=\"middle\" fill=\"#475569\">원본 기록<\/text><path d=\"M207.0,37.0 L219.0,37.0\" stroke=\"#475569\" stroke-width=\"1.6\" marker-end=\"url(#tfa)\" fill=\"none\"/><rect x=\"220.0\" y=\"14.0\" width=\"92.0\" height=\"46.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"266.0\" y=\"34.0\" font-size=\"10\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">그대로<\/text><text x=\"266.0\" y=\"47.0\" font-size=\"8.5\" text-anchor=\"middle\" fill=\"#475569\">다른 코드<\/text><text x=\"160.0\" y=\"87.0\" font-size=\"9\" text-anchor=\"middle\" fill=\"#475569\">고칠 범위를 말해 주세요<\/text><\/svg>", "f_tool": "<svg viewBox=\"0 0 320 92\" xmlns=\"http://www.w3.org/2000/svg\" font-family=\"sans-serif\" font-size=\"10\" fill=\"#111827\"><defs><marker id=\"tfa\" viewBox=\"0 0 10 10\" refX=\"8\" refY=\"5\" markerWidth=\"6\" markerHeight=\"6\" orient=\"auto\"><path d=\"M0,0 L10,5 L0,10 z\" fill=\"#475569\"/><\/marker><\/defs><rect x=\"8.0\" y=\"14.0\" width=\"65.5\" height=\"46.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"40.8\" y=\"40.5\" font-size=\"10\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">설정값<\/text><path d=\"M74.5,37.0 L86.5,37.0\" stroke=\"#475569\" stroke-width=\"1.6\" marker-end=\"url(#tfa)\" fill=\"none\"/><rect x=\"87.5\" y=\"14.0\" width=\"65.5\" height=\"46.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"120.2\" y=\"40.5\" font-size=\"10\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">계산 함수<\/text><path d=\"M154.0,37.0 L166.0,37.0\" stroke=\"#475569\" stroke-width=\"1.6\" marker-end=\"url(#tfa)\" fill=\"none\"/><rect x=\"167.0\" y=\"14.0\" width=\"65.5\" height=\"46.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"199.8\" y=\"40.5\" font-size=\"10\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">표·그림<\/text><path d=\"M233.5,37.0 L245.5,37.0\" stroke=\"#475569\" stroke-width=\"1.6\" marker-end=\"url(#tfa)\" fill=\"none\"/><rect x=\"246.5\" y=\"14.0\" width=\"65.5\" height=\"46.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"279.2\" y=\"40.5\" font-size=\"10\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">브리핑<\/text><text x=\"160.0\" y=\"87.0\" font-size=\"9\" text-anchor=\"middle\" fill=\"#475569\">한 번에 다시 돌릴 수 있게 묶어요<\/text><\/svg>", "f_hand": "<svg viewBox=\"0 0 320 120\" xmlns=\"http://www.w3.org/2000/svg\" font-family=\"sans-serif\" font-size=\"10\" fill=\"#111827\"><defs><marker id=\"tfa\" viewBox=\"0 0 10 10\" refX=\"8\" refY=\"5\" markerWidth=\"6\" markerHeight=\"6\" orient=\"auto\"><path d=\"M0,0 L10,5 L0,10 z\" fill=\"#475569\"/><\/marker><\/defs><rect x=\"6.0\" y=\"8\" width=\"102.7\" height=\"18\" fill=\"#e5e7eb\" stroke=\"#fff\"/><text x=\"57.3\" y=\"20.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">일<\/text><rect x=\"108.7\" y=\"8\" width=\"102.7\" height=\"18\" fill=\"#e5e7eb\" stroke=\"#fff\"/><text x=\"160.0\" y=\"20.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">담당<\/text><rect x=\"211.3\" y=\"8\" width=\"102.7\" height=\"18\" fill=\"#e5e7eb\" stroke=\"#fff\"/><text x=\"262.7\" y=\"20.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">주기<\/text><rect x=\"6.0\" y=\"26\" width=\"102.7\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"57.3\" y=\"38.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">새 형상 등록<\/text><rect x=\"108.7\" y=\"26\" width=\"102.7\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"160.0\" y=\"38.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">데이터<\/text><rect x=\"211.3\" y=\"26\" width=\"102.7\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"262.7\" y=\"38.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">확인 시<\/text><rect x=\"6.0\" y=\"44\" width=\"102.7\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"57.3\" y=\"56.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">콘텐츠 검수<\/text><rect x=\"108.7\" y=\"44\" width=\"102.7\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"160.0\" y=\"56.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">교관<\/text><rect x=\"211.3\" y=\"44\" width=\"102.7\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"262.7\" y=\"56.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">월 1회<\/text><rect x=\"6.0\" y=\"62\" width=\"102.7\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"57.3\" y=\"74.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">배포 승인<\/text><rect x=\"108.7\" y=\"62\" width=\"102.7\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"160.0\" y=\"74.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">교육<\/text><rect x=\"211.3\" y=\"62\" width=\"102.7\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"262.7\" y=\"74.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">배포 때<\/text><rect x=\"6.0\" y=\"80\" width=\"102.7\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"57.3\" y=\"92.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">장애 대응<\/text><rect x=\"108.7\" y=\"80\" width=\"102.7\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"160.0\" y=\"92.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">전산<\/text><rect x=\"211.3\" y=\"80\" width=\"102.7\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"262.7\" y=\"92.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">발생 시<\/text><text x=\"160.0\" y=\"115.0\" font-size=\"9\" text-anchor=\"middle\" fill=\"#475569\">누가 무엇을 언제 맡는지 적어요<\/text><\/svg>", "f_lint": "<svg viewBox=\"0 0 320 102\" xmlns=\"http://www.w3.org/2000/svg\" font-family=\"sans-serif\" font-size=\"10\" fill=\"#111827\"><defs><marker id=\"tfa\" viewBox=\"0 0 10 10\" refX=\"8\" refY=\"5\" markerWidth=\"6\" markerHeight=\"6\" orient=\"auto\"><path d=\"M0,0 L10,5 L0,10 z\" fill=\"#475569\"/><\/marker><\/defs><rect x=\"6.0\" y=\"8\" width=\"154.0\" height=\"18\" fill=\"#e5e7eb\" stroke=\"#fff\"/><text x=\"83.0\" y=\"20.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">항목<\/text><rect x=\"160.0\" y=\"8\" width=\"154.0\" height=\"18\" fill=\"#e5e7eb\" stroke=\"#fff\"/><text x=\"237.0\" y=\"20.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">있음?<\/text><rect x=\"6.0\" y=\"26\" width=\"154.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"83.0\" y=\"38.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">표본 수<\/text><rect x=\"160.0\" y=\"26\" width=\"154.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"237.0\" y=\"38.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">O<\/text><rect x=\"6.0\" y=\"44\" width=\"154.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"83.0\" y=\"56.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">한계<\/text><rect x=\"160.0\" y=\"44\" width=\"154.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"237.0\" y=\"56.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">O<\/text><rect x=\"6.0\" y=\"62\" width=\"154.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"83.0\" y=\"74.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">AI 산출물 표기<\/text><rect x=\"160.0\" y=\"62\" width=\"154.0\" height=\"18\" fill=\"#fef3c7\" stroke=\"#e5e7eb\"/><text x=\"237.0\" y=\"74.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">X<\/text><text x=\"160.0\" y=\"97.0\" font-size=\"9\" text-anchor=\"middle\" fill=\"#475569\">빠진 항목을 코드로 찾아요<\/text><\/svg>"};   // 핵심 용어 카드에 붙는 그림(SVG)
const MISSION_UI = {"m1": {"steps": [{"short": "네 칸 익히기", "title": "프롬프트의 네 칸", "lead": "AI는 전우님 업무도, 파일도 몰라요. **네 칸**을 채워 알려 주면 돼요.", "page": null, "blocks": [{"id": "s1_info", "type": "info", "title": "좋은 프롬프트의 네 칸", "icon": "🧩", "req": false, "text": ["같은 일을 시켜도 무엇을 알려 주느냐에 따라 결과가 달라져요. 처음에는 아래 네 칸만 기억하세요."], "table": {"head": ["칸", "무엇을 적나", "예"], "rows": [["맥락", "왜 필요한지, 어떤 일인지", "실루엣 학습앱에 넣을 표를 만들고 있어"], ["입력", "파일 이름, 열 이름, 값의 뜻", "m1_families.csv, 열은 family·span_m·length_m"], ["할 일", "계산 규칙을 숫자로", "픽셀 수 = 대표 크기 ÷ (거리 × 0.001)"], ["출력 형식", "받고 싶은 모양", "형상 계열은 행, 거리는 열인 표"]]}, "figure": "f_four"}, {"id": "s1_q", "type": "quiz", "title": "어떤 프롬프트가 가장 좋을까요?", "req": true, "text": "같은 일을 시키는 세 프롬프트예요. 다시 묻는 일이 가장 적을 것은 무엇일까요?", "options": ["드론 실루엣 코드 좀 짜 줘.", "비행체가 잘 보이는지 알려 줘. 잘 부탁해.", "실루엣 학습앱에 넣을 표를 만들고 있어. m1_families.csv의 크기로 거리마다 몇 픽셀로 보이는지 계산해 표로 보여 줘."]}]}, {"short": "AI에게 시키기", "title": "프롬프트를 만들어 AI에게 보내기", "lead": "칸마다 문장을 고르면 프롬프트가 만들어져요. 일부러 **아쉬운 문장**도 보내 보세요. 무엇이 달라지는지 보는 것이 이번 활동의 핵심이에요.", "page": null, "blocks": [{"id": "s2_vibe", "type": "vibe", "title": "실루엣과 거리별 수준표 만들기", "req": true, "ask": ["gyo", "형상 계열 파일을 드릴게요. 기본 도형 실루엣 6종과, 거리 100m~2,000m에서 각각 몇 픽셀로 보이는지 표로 받고 싶어요."], "has_start": false, "files": [{"name": "m1_families.csv", "head": ["family", "wing", "engine", "body", "tail", "span_m", "length_m"], "rows": [["multicopter", "none", "rotor_4", "box", "none", "0.6", "0.6"], ["fixed_wing", "straight", "pusher", "slim", "t_tail", "3.0", "2.0"], ["delta_wing", "delta", "pusher", "slim", "winglet", "2.5", "3.5"], ["cruise_missile", "small", "jet", "tube", "cross_fin", "2.5", "6.0"], ["ballistic_missile", "none", "rocket", "long_tube", "fin_4", "1.0", "12.0"]], "n": 6}], "slots": [{"id": "ctx", "label": "맥락", "why": "왜 필요한지 알려요.", "options": [{"i": 0, "text": "실루엣 식별 학습앱을 만들고 있어. 형상 계열마다 거리별로 화면에서 몇 픽셀로 보이는지 확인하려고 해."}, {"i": 1, "text": "코드 좀 짜 줘."}, {"i": 2, "text": ""}]}, {"id": "input", "label": "입력", "why": "파일과 열을 알려요.", "options": [{"i": 0, "text": "데이터는 m1_families.csv이고, 열은 family(형상 계열), wing, engine, body, tail(WEFT 단서), span_m(폭, 미터), length_m(길이, 미터)이야."}, {"i": 1, "text": "드론 데이터 파일이 있어."}, {"i": 2, "text": ""}]}, {"id": "task", "label": "할 일", "why": "계산 규칙을 숫자로 적어요.", "options": [{"i": 1, "text": "거리별로 잘 보이는지 계산해 줘."}, {"i": 0, "text": "대표 크기 = √(span_m × length_m), 픽셀 수 = 대표 크기 ÷ (거리 × 0.001)로 거리 100, 300, 500, 1000, 2000m마다 계산하고, 2픽셀 이상은 탐지, 8픽셀 이상은 인식, 13픽셀 이상은 식별로 표시해 줘."}, {"i": 2, "text": ""}]}, {"id": "out", "label": "출력 형식", "why": "받고 싶은 모양을 알려요.", "options": [{"i": 1, "text": "결과를 보여 줘."}, {"i": 0, "text": "형상 계열은 행, 거리는 열인 표로 보여 주고, 기본 도형으로 그린 실루엣 6종 그림도 만들어 줘. 표를 넣은 HTML 보고서는 report 변수에 담아 줘."}, {"i": 2, "text": ""}]}]}]}, {"short": "내 말로 쓰기", "title": "같은 일을 내 말로 요청하기", "lead": "고른 문장을 그대로 외울 필요는 없어요. **네 칸만 채우면** 말투는 자유예요.", "page": null, "blocks": [{"id": "s3_q", "type": "quiz", "title": "결과를 받은 뒤에는", "req": true, "text": "AI가 만든 코드가 오류 없이 돌았어요. 다음으로 무엇을 하면 좋을까요?", "options": ["오류가 없으니 그대로 쓴다.", "하나만 손으로 계산해 맞는지 확인한다.", "더 어려운 말로 다시 요청한다."]}, {"id": "s3_write", "type": "write", "title": "내 프롬프트 쓰기", "req": true, "prompt": "**새 요청**: 새 형상 계열 jet_delta(폭 3.2m, 길이 3.8m)가 추가됐어요. 파일은 m1_families_new.csv이고 열은 같아요. 거리 300m, 1000m에서의 픽셀 수를 보고 싶어요. 네 칸을 채워 내 말로 프롬프트를 써 보세요.", "placeholder": "꼭 입력해 주세요 · 내 말로 프롬프트를 써 보세요", "model_label": "📝 모범 프롬프트", "min": 30}]}]}, "m2": {"steps": [{"short": "데이터 보여 주기", "title": "AI에게 데이터를 설명하는 법", "lead": "AI는 파일을 볼 수 없어요. **앞부분 몇 줄과 값의 종류**를 알려 주는 것이 가장 빠른 방법이에요.", "page": null, "blocks": [{"id": "s1_info", "type": "info", "title": "데이터를 설명할 때 넣을 것", "icon": "🗂", "req": false, "text": ["열 이름만으로는 부족할 때가 많아요. 값이 어떻게 적혀 있는지까지 알려 주세요. 이 파일의 실루엣은 하늘과 정사영상(지형 사진을 지도처럼 편 영상) 배경에 올린 것이라 background 열도 있어요."], "bullets": ["**파일 이름과 열 이름**: m2_labels.csv, image_id, background, view, rater_a, rater_b", "**값의 종류**: multicopter, fixed_wing, delta_wing, cruise_missile, ballistic_missile, bird 여섯 가지", "**주의할 점**: 대소문자, 앞뒤 빈칸, 밑줄 대신 띄어쓰기가 섞여 있음", "**샘플 몇 줄**: 실제 값 대신 가상 샘플로 보여 줘도 돼요"], "figure": "f_clean"}, {"id": "s1_q", "type": "quiz", "title": "먼저 알려 줄 것", "req": true, "text": "값에 ‘Bird’, ‘bird ’가 섞여 있어요. 이 사실을 알려 주지 않으면 어떻게 될까요?", "options": ["AI가 스스로 알아채 항상 정리해 준다.", "같은 값을 다른 값으로 세어 일치율이 실제보다 낮게 나올 수 있다.", "파일을 읽지 못해 반드시 오류가 난다."]}]}, {"short": "일치율 계산", "title": "일치율을 계산하고 갈린 라벨 찾기", "lead": "**값 정리** 문장을 빼고도 보내 보세요. 숫자가 어떻게 달라지는지 비교해 보면 좋아요.", "page": null, "blocks": [{"id": "s2_vibe", "type": "vibe", "title": "라벨 일치율 계산", "req": true, "ask": ["label", "두 사람이 얼마나 같게 라벨을 붙였는지, 어떤 실루엣에서 갈렸는지 알고 싶어요."], "has_start": false, "files": [{"name": "m2_labels.csv", "head": ["image_id", "background", "view", "rater_a", "rater_b"], "rows": [["IMG01", "urban", "top", "Bird", "multicopter"], ["IMG02", "field", "top", "bird ", "bird"], ["IMG03", "sky", "bottom", "cruise_missile", "Cruise_missile"], ["IMG04", "field", "top", "Bird", "bird"], ["IMG05", "sky", "side", "delta_wing", "Decoy"]], "n": 30}], "slots": [{"id": "input", "label": "입력", "why": "값의 모양까지 알려요.", "options": [{"i": 1, "text": "m2_labels.csv에 두 사람의 라벨이 있어."}, {"i": 0, "text": "m2_labels.csv의 열은 image_id, background, view, rater_a, rater_b이고, 라벨 값은 multicopter, fixed_wing, delta_wing, cruise_missile, ballistic_missile, bird야. 대소문자와 빈칸이 섞여 있으니 먼저 소문자로 바꾸고, 앞뒤 빈칸을 지우고, 띄어쓰기는 밑줄로 바꿔 줘."}, {"i": 2, "text": ""}]}, {"id": "task", "label": "할 일", "why": "무엇을 세는지 정해요.", "options": [{"i": 0, "text": "여섯 값에 없는 라벨은 규격 위반으로 따로 뽑고, 나머지로 일치율과 코헨 카파(일치율에서 우연히 같을 확률을 뺀 값)를 계산해 줘. 두 사람이 다른 실루엣 목록도 뽑아 줘."}, {"i": 1, "text": "일치도를 분석해 줘."}, {"i": 2, "text": ""}]}, {"id": "out", "label": "출력 형식", "why": "회의에서 볼 모양으로 받아요.", "options": [{"i": 1, "text": "결과를 정리해 줘."}, {"i": 0, "text": "요약 표, 규격 위반 표, 갈린 실루엣 표를 보여 주고, 세 표를 넣은 HTML 문서를 report 변수에 담아 줘."}, {"i": 2, "text": ""}]}]}]}, {"short": "기준 합의", "title": "결과로 기준을 합의하고 내 말로 요청하기", "lead": "AI는 계산을 도와줄 뿐이에요. 어떤 기준을 쓸지는 **사람이 근거를 보고 정해요.**", "page": null, "blocks": [{"id": "s3_decide", "type": "decide", "title": "팀원의 제안 판단하기", "req": true, "text": "일치율 결과를 본 팀원들의 제안이에요. 받아들일지 정하고 이유를 적어 주세요.", "reason_min": 8, "items": [{"id": "a", "who": "label", "text": "decoy도 형상 라벨 값에 넣어요. 어차피 구분해야 하잖아요."}, {"id": "b", "who": "gyo", "text": "형상 값은 여섯 가지 소문자로만 입력하게 하고, 정체는 ‘미상’을 기본값으로 하는 칸을 따로 둬요."}]}, {"id": "s3_write", "type": "write", "title": "규격 문서를 요청하는 프롬프트 쓰기", "req": true, "prompt": "방금 합의한 내용으로 **라벨링 규격 문서**를 AI에게 요청해 보세요. 누가 읽는지, 어떤 항목이 들어가야 하는지, 근거 데이터를 적어 주세요.", "placeholder": "꼭 입력해 주세요 · 내 말로 프롬프트를 써 보세요", "model_label": "📝 모범 프롬프트", "min": 30}]}]}, "m3": {"steps": [{"short": "성능표 받기", "title": "모델과 조건을 나눠 표로 받기", "lead": "전체 정확도 하나로는 어디서 무너지는지 알 수 없어요. **무엇으로 나눌지** 적어 주세요.", "page": null, "blocks": [{"id": "s1_vibe", "type": "vibe", "title": "기준 모델 조건별 성능표", "req": true, "ask": ["model", "모델과 거리별로 정확도를 비교하고 싶어요."], "has_start": false, "files": [{"name": "m3_preds.csv", "head": ["image_id", "model", "distance", "polarity", "truth", "pred"], "rows": [["S001", "decision_tree", "near", "white_hot", "ballistic_missile", "ballistic_missile"], ["S001", "logistic", "near", "white_hot", "ballistic_missile", "ballistic_missile"], ["S001", "knn", "near", "white_hot", "ballistic_missile", "ballistic_missile"], ["S002", "decision_tree", "near", "white_hot", "cruise_missile", "cruise_missile"], ["S002", "logistic", "near", "white_hot", "cruise_missile", "cruise_missile"]], "n": 450}], "slots": [{"id": "input", "label": "입력", "why": "열의 뜻을 알려요.", "options": [{"i": 1, "text": "모델 결과 CSV가 있어."}, {"i": 0, "text": "m3_preds.csv의 열은 image_id, model(decision_tree·logistic·knn), distance(near·mid·far), polarity(white_hot·black_hot), truth(실제 라벨), pred(예측 라벨)야."}, {"i": 2, "text": ""}]}, {"id": "task", "label": "할 일", "why": "무엇으로 나눌지 정해요.", "options": [{"i": 1, "text": "모델 성능을 알려 줘."}, {"i": 0, "text": "truth와 pred가 같으면 정답으로 보고, 모델과 거리별 정확도(%)를 소수 첫째 자리까지 계산해 줘."}, {"i": 2, "text": ""}]}, {"id": "out", "label": "출력 형식", "why": "표의 모양을 정해요.", "options": [{"i": 1, "text": "보기 좋게 보여 줘."}, {"i": 0, "text": "모델은 행, 거리는 near·mid·far 순서의 열인 표로 보여 주고, 같은 값을 막대그래프로도 그려 줘."}, {"i": 2, "text": ""}]}]}]}, {"short": "오류 고치기", "title": "오류 메시지로 수정 요청하기", "lead": "극성까지 나눴더니 **오류**가 났어요. 어떻게 다시 요청할까요?", "page": null, "blocks": [{"id": "s2_vibe", "type": "vibe", "title": "오류 고쳐 달라고 하기", "req": true, "ask": ["model", "극성까지 나눴더니 오류가 나요. 저도 원인을 잘 모르겠어요… 부탁해요!"], "has_start": true, "files": [{"name": "m3_preds.csv", "head": ["image_id", "model", "distance", "polarity", "truth", "pred"], "rows": [["S001", "decision_tree", "near", "white_hot", "ballistic_missile", "ballistic_missile"], ["S001", "logistic", "near", "white_hot", "ballistic_missile", "ballistic_missile"], ["S001", "knn", "near", "white_hot", "ballistic_missile", "ballistic_missile"], ["S002", "decision_tree", "near", "white_hot", "cruise_missile", "cruise_missile"], ["S002", "logistic", "near", "white_hot", "cruise_missile", "cruise_missile"]], "n": 450}], "slots": [{"id": "what", "label": "무슨 일이 있었는지", "why": "오류 메시지를 그대로 붙여 넣어요.", "options": [{"i": 1, "text": "안 돼. 다시 해 줘."}, {"i": 0, "text": "모델·거리·극성별 정확도를 계산하다가 이 오류가 났어: ZeroDivisionError: division by zero"}, {"i": 2, "text": ""}]}, {"id": "fix", "label": "어떻게 고칠지", "why": "원하는 동작과 범위를 정해요.", "options": [{"i": 0, "text": "far에 black_hot 표본이 없어서 0으로 나눈 것 같아. 표본이 없으면 ‘자료 없음’으로 표시하게 acc 함수만 고치고, 나머지 계산은 그대로 둬. 조건별 표본 수도 함께 보여 주고, 표를 넣은 HTML 성능표 문서를 report 변수에 담아 줘."}, {"i": 1, "text": "전부 새로 짜 줘."}, {"i": 2, "text": ""}]}]}]}, {"short": "내 말로 쓰기", "title": "다른 오류를 내 말로 요청하기", "lead": "오류를 만나면 **겁먹지 말고 그대로 붙여 넣으세요.** 코드를 다 읽지 못해도 수정을 요청할 수 있어요.", "page": null, "blocks": [{"id": "s3_write", "type": "write", "title": "수정 프롬프트 쓰기", "req": true, "prompt": "**새 상황**: 새 파일을 넣었더니 `KeyError: 'polarity'`가 났어요. 이 파일의 실제 열 이름은 thermal_mode예요. 잘 되던 정확도 계산은 그대로 두고 싶어요. 어떻게 요청할까요?", "placeholder": "꼭 입력해 주세요 · 내 말로 프롬프트를 써 보세요", "model_label": "📝 모범 프롬프트", "min": 30}]}]}, "m4": {"steps": [{"short": "쪼개기", "title": "큰 일을 작은 단계로 쪼개기", "lead": "한 번에 전부 요청하면 어디가 틀렸는지 찾기 어려워요. **작게 만들고 확인한 뒤 넓혀요.**", "page": null, "blocks": [{"id": "s1_info", "type": "info", "title": "이번 요구 성능", "icon": "📐", "req": false, "text": ["고교관, 전산호, 배보안의 요구를 한 표로 모았어요. 먼저 이 기준을 정한 뒤 모델을 봐요."], "table": {"head": ["담당", "요구", "숫자"], "rows": [["고교관", "위협 형상을 놓치지 않기", "재현율 90% 이상"], ["전산호", "휴대 기기에서 빠르게", "한 장 100ms 이하, 30MB 이하"], ["배보안", "외부로 보내지 않기", "외부 전송 없음, 라이선스 확인"]]}, "figure": "f_req"}, {"id": "s1_order", "type": "order", "title": "요청 순서 정하기", "req": true, "text": "후보 모델을 판정하는 도구를 만들려고 해요. 어떤 순서로 요청하면 좋을까요?", "items": [{"id": "c", "text": "답을 아는 모델 하나로 함수 확인하기"}, {"id": "d", "text": "전체 후보에 적용해 비교표 만들기"}, {"id": "b", "text": "모델 하나를 판정하는 함수 만들기"}, {"id": "a", "text": "데이터 모양을 알려 주고 파일을 읽게 하기"}]}]}, {"short": "함수로 만들기", "title": "기준을 바꿔 돌릴 수 있는 함수 요청하기", "lead": "기준값을 코드 속에 박아 두면 회의 때마다 고쳐야 해요. **인자로 받는 함수**로 요청해요.", "page": null, "blocks": [{"id": "s2_vibe", "type": "vibe", "title": "요구 성능 판정 함수", "req": true, "ask": ["jeon", "재현율, 속도, 크기, 외부 전송을 한 번에 판정하고, 왜 떨어졌는지도 알고 싶어요."], "has_start": false, "files": [{"name": "m4_models.csv", "head": ["model", "threat_recall_pct", "ms_per_image", "size_mb", "sends_outside", "license_ok"], "rows": [["small_cnn", "86.5", "12", "1.2", "0", "1"], ["mobilenet_v3", "92.4", "35", "9.8", "0", "1"], ["dinov2_probe", "95.1", "180", "86.0", "0", "1"], ["clip_zeroshot", "78.3", "210", "340.0", "0", "1"], ["kan_classifier", "84.0", "25", "0.8", "0", "1"]], "n": 7}], "slots": [{"id": "small", "label": "작게 시작", "why": "함수 하나부터 만들어요.", "options": [{"i": 0, "text": "m4_models.csv의 열은 model, threat_recall_pct, ms_per_image, size_mb, sends_outside(1이면 외부 전송), license_ok(1이면 확인됨)야. 모델 한 행을 받아 기준을 넘는지와 떨어진 이유를 돌려주는 meets 함수를 먼저 만들고, mobilenet_v3로 한 번 확인해 줘."}, {"i": 1, "text": "모델 비교 프로그램을 한 번에 다 만들어 줘."}, {"i": 2, "text": ""}]}, {"id": "param", "label": "조건 바꾸기", "why": "기준값을 인자로 받아요.", "options": [{"i": 0, "text": "기준값은 min_recall, max_ms, max_mb 인자로 받게 해 줘. 기본값은 90, 100, 30이고, 외부 전송이 있거나 라이선스가 확인되지 않으면 항상 탈락이야."}, {"i": 1, "text": "기준은 코드에 적당히 넣어 줘."}, {"i": 2, "text": ""}]}, {"id": "out", "label": "비교표", "why": "함수를 전체에 적용해요.", "options": [{"i": 1, "text": "결과를 정리해 줘."}, {"i": 0, "text": "전체 모델에 기준 A(기본값)와 기준 B(재현율 85, 50ms, 10MB)를 적용한 비교표를 보여 주고, 속도와 재현율을 점으로 찍어 그려 줘. 표를 넣은 HTML 비교 문서를 report 변수에 담아 줘."}, {"i": 2, "text": ""}]}]}]}, {"short": "모델 선정", "title": "결과로 조율하고 내 말로 요청하기", "lead": "AI의 표는 근거일 뿐이에요. 어떤 기준으로 갈지는 **사람이 합의해요.**", "page": null, "blocks": [{"id": "s3_decide", "type": "decide", "title": "모델 선정 회의", "req": true, "text": "비교표를 본 팀원들의 의견이에요. 받아들일지 정하고 이유를 적어 주세요.", "reason_min": 8, "items": [{"id": "a", "who": "model", "text": "재현율이 제일 높은 cloud_api_model로 가요. 외부 전송은 나중에 조건을 풀어 달라고 하면 되잖아요."}, {"id": "b", "who": "gyo", "text": "기준 A를 통과한 모델로 가되, 선정 기록에 탈락한 모델과 이유도 함께 남겨요."}]}, {"id": "s3_write", "type": "write", "title": "쪼개서 요청하는 프롬프트 쓰기", "req": true, "prompt": "**새 요청**: 새 후보 모델이 들어온 m4_models_new.csv에 같은 판정을 하고 싶어요. 기준에 ‘배터리 사용량 5% 이하’(battery_pct 열)를 더하려고 해요. 어떤 순서로, 무엇을 고쳐 달라고 할까요?", "placeholder": "꼭 입력해 주세요 · 내 말로 프롬프트를 써 보세요", "model_label": "📝 모범 프롬프트", "min": 30}]}]}, "m5": {"steps": [{"short": "답만 받으면", "title": "근거 없는 답은 위험해요", "lead": "AI가 자신 있게 답해도 **다시 셀 수 있는 숫자**가 없으면 믿을지 판단할 수 없어요.", "page": null, "blocks": [{"id": "s1_info", "type": "info", "title": "AI 답변에 함께 받을 것", "icon": "🔍", "req": false, "text": ["답을 받을 때 아래를 함께 요청하면, 코드를 다 읽지 못해도 믿을지 판단할 수 있어요."], "bullets": ["**근거 수치**: 구간별 표본 수, 평균 점수, 실제 정답률", "**작은 예시**: 손으로 계산해 볼 수 있는 한 구간", "**나눈 방법**: 어떤 데이터로 고르고 어떤 데이터로 확인했는지"], "figure": "f_verify"}, {"id": "s1_q", "type": "quiz", "title": "어떤 답을 믿을까요?", "req": true, "text": "두 AI 답변 가운데 어느 쪽을 더 믿을 수 있을까요?", "options": ["‘이 모델은 점수가 잘 맞아요. 0.9면 90%가 맞다고 보면 돼요.’", "‘0.9~1.0 구간 표본은 N건, 평균 점수는 A, 실제 정답률은 B예요. 표를 첨부해요.’", "‘최신 기법을 썼으니 안심하고 쓰셔도 돼요.’"]}]}, {"short": "근거까지 받기", "title": "근거 수치와 함께 요청하기", "lead": "**근거**와 **나눈 방법** 문장을 빼고도 보내 보세요. AI의 답이 어떻게 달라지는지 비교해 봐요.", "page": null, "blocks": [{"id": "s2_vibe", "type": "vibe", "title": "점수와 실제 정답률 비교", "req": true, "ask": ["model", "점수가 믿을 만한지, 어디부터 교관이 확인해야 하는지 숫자로 보고 싶어요."], "has_start": false, "files": [{"name": "m5_scores.csv", "head": ["image_id", "split", "top1_label", "top1_score", "top2_score", "is_correct"], "rows": [["V001", "val", "ballistic_missile", "0.638", "0.218", "0"], ["V002", "val", "ballistic_missile", "0.34", "0.335", "0"], ["V003", "val", "cruise_missile", "0.776", "0.188", "1"], ["V004", "val", "delta_wing", "0.691", "0.166", "1"], ["V005", "val", "cruise_missile", "0.896", "0.077", "0"]], "n": 400}], "slots": [{"id": "task", "label": "할 일", "why": "무엇을 비교할지 정해요.", "options": [{"i": 0, "text": "m5_scores.csv의 열은 image_id, split(val·test), top1_label, top1_score, top2_score, is_correct야. top1_score를 0.3~0.5, 0.5~0.6, 0.6~0.7, 0.7~0.8, 0.8~0.9, 0.9~1.0 구간으로 나눠 구간별 평균 점수와 실제 정답률을 비교해 줘. ECE도 계산해 줘."}, {"i": 1, "text": "점수가 정확한지 봐 줘."}, {"i": 2, "text": ""}]}, {"id": "evidence", "label": "근거", "why": "다시 셀 수 있는 숫자를 받아요.", "options": [{"i": 0, "text": "결론만 쓰지 말고 구간별 표본 수, 평균 점수, 실제 정답률을 표로 보여 줘. 점수와 실제 정답률을 비교하는 그림도 그려 줘."}, {"i": 1, "text": "믿어도 되는지 결론만 말해 줘."}, {"i": 2, "text": ""}]}, {"id": "split", "label": "나눈 방법", "why": "고르는 데이터와 확인하는 데이터를 나눠요.", "options": [{"i": 0, "text": "등급 기준은 top1_score 0.8 이상이고 1·2위 차이 0.3 이상이면 ‘참고 판정 표시’, 차이 0.1 이상이면 ‘교관 확인’, 나머지는 ‘판별 불가’야. val에서 등급별 정답률을 보고, 정한 기준을 test에는 마지막에 한 번만 적용해 줘. 표를 넣은 HTML 문서를 report 변수에 담아 줘."}, {"i": 1, "text": "전체 데이터로 제일 좋아 보이는 기준을 찾아 줘."}, {"i": 2, "text": ""}]}]}]}, {"short": "모양과 정체", "title": "모양과 정체를 나누고 내 말로 요청하기", "lead": "모델이 알려 주는 것은 **모양**이에요. 정체는 다른 근거가 필요해요.", "page": null, "blocks": [{"id": "s3_cls", "type": "classify", "title": "모델이 말해도 될까요?", "req": true, "text": "실루엣 분류 모델의 화면에 나와도 되는 말과 사람이 다른 근거로 판단해야 하는 말을 나눠 보세요.", "primary_label": "판단", "item_label": "문장", "contrib": false, "categories": [["ok", "모델이 말해도 돼요", ""], ["no", "사람이 다른 근거로 판단해요", ""]], "items": [{"id": "a", "title": "이 실루엣은 delta_wing과 비슷해요. (참고)"}, {"id": "b", "title": "이것은 기만용 드론이에요."}, {"id": "c", "title": "1·2위 점수 차이가 작아 판별 불가예요."}, {"id": "d", "title": "위협이 확실하니 바로 대응하세요."}]}, {"id": "s3_write", "type": "write", "title": "근거를 요청하는 프롬프트 쓰기", "req": true, "prompt": "**새 요청**: 다른 모델의 점수 파일 m5_scores_b.csv도 믿어도 되는지 보고 싶어요. 열은 같아요. 결론만 받지 않도록 어떻게 요청할까요?", "placeholder": "꼭 입력해 주세요 · 내 말로 프롬프트를 써 보세요", "model_label": "📝 모범 프롬프트", "min": 30}]}]}, "m6": {"steps": [{"short": "숫자로 바꾸기", "title": "모호한 말을 숫자 규칙으로", "lead": "‘자주’, ‘적당히’는 사람마다 달라요. **숫자로 바꾸면** AI도 사람도 같은 뜻으로 읽어요.", "page": null, "blocks": [{"id": "s1_match", "type": "match", "title": "모호한 말을 숫자 규칙과 짝짓기", "req": true, "text": "왼쪽 말을 프롬프트에 쓸 수 있게 바꾼 문장을 골라 주세요.", "left_label": "모호한 말", "right_label": "프롬프트에 쓸 문장", "lefts": ["가끔 틀리는 문항", "믿기 어려운 기록", "자주 틀리는 문항", "아직 모르는 문항"], "rights": ["오답률 40% 이상", "오답률 20% 이상 40% 미만", "shown이 5 미만", "shown이 0"]}]}, {"short": "규칙으로 요청", "title": "숫자 규칙과 예외를 담아 요청하기", "lead": "**예외** 문장을 빼고도 보내 보세요. 빈칸이 어떻게 처리되는지 확인해 봐요.", "page": null, "blocks": [{"id": "s2_vibe", "type": "vibe", "title": "출제 계획표와 화면 시안", "req": true, "ask": ["gyo", "문항별로 몇 배로 낼지 정하고, 학습 흐름이 보이는 화면 시안까지 받고 싶어요."], "has_start": false, "files": [{"name": "m6_items.csv", "head": ["item_id", "family", "confusable_with", "shown", "wrong"], "rows": [["Q01", "delta_wing", "fixed_wing", "40", "19"], ["Q02", "fixed_wing", "delta_wing", "40", "11"], ["Q03", "cruise_missile", "delta_wing", "38", "17"], ["Q04", "multicopter", "bird", "36", "5"], ["Q05", "bird", "multicopter", "36", "8"]], "n": 10}], "slots": [{"id": "rule", "label": "숫자 규칙", "why": "기준을 숫자로 적어요.", "options": [{"i": 1, "text": "자주 틀리는 문항을 많이 내게 해 줘."}, {"i": 0, "text": "m6_items.csv의 열은 item_id, family, confusable_with, shown, wrong이야. 오답률 = wrong ÷ shown × 100으로 계산하고, 40% 이상이면 ‘우선 출제’(3배), 20% 이상이면 ‘자주 출제’(2배), 그 아래는 ‘기본’(1배)으로 표시해 줘."}, {"i": 2, "text": ""}]}, {"id": "edge", "label": "예외", "why": "빈칸과 적은 표본을 어떻게 할지 정해요.", "options": [{"i": 1, "text": "나머지는 알아서 해 줘."}, {"i": 0, "text": "shown이 0이면 오답률을 비우고 ‘새 문항’(1배)으로, shown이 5 미만이면 오답률은 보여 주되 ‘표본 적음’(1배)으로 표시해 줘."}, {"i": 2, "text": ""}]}, {"id": "screen", "label": "화면", "why": "순서와 조건을 정해요.", "options": [{"i": 0, "text": "학습 흐름은 문항 → 응답 → 즉시 피드백 → AI 참고 판정 → 해설 순서야. AI 참고 판정은 응답 전에는 숨기고, 등급과 ‘교관 확인 전’ 표시를 함께 보여 주는 HTML 화면 시안을 report 변수에 담아 줘. 아래에 출제 계획표도 넣어 줘."}, {"i": 1, "text": "화면을 멋지게 만들어 줘."}, {"i": 2, "text": ""}]}]}]}, {"short": "사용성 점검", "title": "사용성 의견을 판단하고 내 말로 요청하기", "lead": "AI로 만든 화면은 **쓰는 사람의 의견**으로 고쳐요. 다만 의견을 전부 따르지는 않아요.", "page": null, "blocks": [{"id": "s3_decide", "type": "decide", "title": "사용성 의견 판단하기", "req": true, "text": "시범 사용자와 교관의 의견이에요. 받아들일지 정하고 이유를 적어 주세요.", "reason_min": 8, "items": [{"id": "a", "who": "park", "text": "문항마다 AI 판정을 먼저 보여 주세요. 그러면 빨리 끝나요."}, {"id": "b", "who": "park", "text": "틀린 뒤에 헷갈린 두 실루엣을 나란히 보여 주세요."}]}, {"id": "s3_write", "type": "write", "title": "숫자와 예외를 담은 프롬프트 쓰기", "req": true, "prompt": "**새 요청**: ‘최근에 많이 틀린 사람에게 복습 문항을 좀 더 내 줘’라는 요청을 받았어요. 파일은 m6_learners.csv이고, 열은 learner_id, shown, wrong이에요. 숫자 규칙과 예외를 정해 프롬프트를 써 보세요.", "placeholder": "꼭 입력해 주세요 · 내 말로 프롬프트를 써 보세요", "model_label": "📝 모범 프롬프트", "min": 30}]}]}, "m7": {"steps": [{"short": "역할 정하기", "title": "AI와 사람의 역할 나누기", "lead": "AI는 초안을 쓰고, **결정은 사람이** 해요. 문서를 요청하기 전에 역할부터 정해요.", "page": null, "blocks": [{"id": "s1_match", "type": "match", "title": "누가 무엇을 맡나요?", "req": true, "text": "담당과 맡을 일을 짝지어 주세요.", "left_label": "담당", "right_label": "맡을 일", "lefts": ["나라벨", "윤훈련", "전산호", "AI", "고교관"], "rights": ["기한 넘긴 건 처리", "해설 승인·수정·반려", "승인된 해설만 앱에 반영", "형상 근거표 검수", "해설 초안 작성"]}]}, {"short": "문서 요청", "title": "독자·근거·표현을 정해 문서 요청하기", "lead": "**자료에 없으면** 문장을 빼고도 보내 보세요. AI가 무엇을 지어내는지 확인해 봐요.", "page": null, "blocks": [{"id": "s2_vibe", "type": "vibe", "title": "해설 대조표와 검토 절차서", "req": true, "ask": ["gyo", "초안의 주장이 근거표와 맞는지 대조하고, 교관이 읽을 검토 절차서 초안까지 받고 싶어요."], "has_start": false, "files": [{"name": "m7_drafts.csv", "head": ["draft_id", "family", "field", "claim", "wait_hours"], "rows": [["D01", "delta_wing", "wing", "delta", "6"], ["D02", "delta_wing", "engine", "jet", "20"], ["D03", "fixed_wing", "tail", "t_tail", "52"], ["D04", "cruise_missile", "speed", "900kmh", "12"], ["D05", "multicopter", "engine", "rotor_4", "3"]], "n": 8}], "slots": [{"id": "reader", "label": "독자와 양식", "why": "누가 읽는지, 어떤 순서인지 정해요.", "options": [{"i": 0, "text": "m7_drafts.csv(draft_id, family, field, claim, wait_hours)의 주장을 m7_facts.csv(family, wing, engine, tail)와 대조한 표를 만들어 줘. 교관이 읽을 검토 절차서를 HTML로 report 변수에 담아 줘. 순서는 목적, 절차 표(단계·담당·기한), 대조 결과, 기한 초과 목록이야."}, {"i": 1, "text": "검토 절차에 대해 써 줘."}, {"i": 2, "text": ""}]}, {"id": "source", "label": "근거", "why": "자료 밖의 말을 막아요.", "options": [{"i": 1, "text": "참고할 만한 내용도 알아서 넣어 줘."}, {"i": 0, "text": "근거는 이 두 파일뿐이야. 근거표에 없는 항목은 지어내지 말고 ‘자료 없음’으로 적어 줘. 기한은 48시간이고, 넘으면 ‘초과: 교육 담당에게 올림’으로 표시해 줘."}, {"i": 2, "text": ""}]}, {"id": "tone", "label": "표현", "why": "상태와 책임을 분명히 적어요.", "options": [{"i": 1, "text": "자신 있게 확정된 문서처럼 써 줘."}, {"i": 0, "text": "아직 아무도 승인하지 않았으니 문서 맨 위에 ‘AI 초안, 교관 확인 전’이라고 표시하고, 승인 전에는 공개하지 않는다고 적어 줘. 대조 결과는 ‘일치’, ‘불일치’, ‘자료 없음’으로만 써 줘."}, {"i": 2, "text": ""}]}]}]}, {"short": "내 말로 쓰기", "title": "문서를 받은 뒤 확인하고 내 말로 요청하기", "lead": "문서를 받았다고 끝이 아니에요. **자료와 대조**하는 일은 사람의 몫이에요.", "page": null, "blocks": [{"id": "s3_q", "type": "quiz", "title": "문서를 받은 뒤에는", "req": true, "text": "AI가 쓴 절차서 초안이 매끄럽게 나왔어요. 가장 먼저 무엇을 할까요?", "options": ["문장이 매끄러우니 바로 공유한다.", "수치와 고유명사를 파일과 대조한다.", "더 길게 써 달라고 한다."]}, {"id": "s3_write", "type": "write", "title": "문서를 요청하는 프롬프트 쓰기", "req": true, "prompt": "**새 요청**: 새로 온 교관이 읽을 ‘혼동 쌍 해설 안내문’이 필요해요. 근거는 m7_facts.csv와 m6_items.csv뿐이에요. 독자, 순서, 근거, 없는 내용 처리, 초안 표시를 담아 프롬프트를 써 보세요.", "placeholder": "꼭 입력해 주세요 · 내 말로 프롬프트를 써 보세요", "model_label": "📝 모범 프롬프트", "min": 30}]}]}, "m8": {"steps": [{"short": "나눠 보기", "title": "AI에 넣어도 되는 것과 안 되는 것", "lead": "프롬프트에 넣기 전에 **한 번 나눠 보세요.** 값이 없어도 구조만 알면 코드를 받을 수 있어요.", "page": null, "blocks": [{"id": "s1_cls", "type": "classify", "title": "공개 AI에 넣어도 될까요?", "req": true, "text": "조직이 승인한 환경이 아니라 일반 공개 AI라고 생각하고 나눠 보세요.", "primary_label": "판단", "item_label": "내용", "contrib": false, "categories": [["ok", "넣어도 돼요", ""], ["no", "넣으면 안 돼요", ""]], "items": [{"id": "a", "title": "열 이름: variant, size_mb, ms_browser"}, {"id": "b", "title": "내부 서버 주소와 관리자 계정"}, {"id": "c", "title": "직접 지어낸 가상 샘플 2줄"}, {"id": "d", "title": "학습자 이름이 든 시범 운영 기록 원본"}]}]}, {"short": "구조만 알리기", "title": "값은 빼고 구조만 알려 요청하기", "lead": "**민감 정보를 넣은 문장**도 일부러 골라 보세요. 결과는 나와도 무엇이 문제인지 볼 수 있어요.", "page": null, "blocks": [{"id": "s2_vibe", "type": "vibe", "title": "경량화 비교와 안전 점검표", "req": true, "ask": ["jeon", "크기 5MB 이하, 브라우저에서 60ms 이하, 재현율 90% 이상인 후보를 찾고, 안전 실패도 확인하고 싶어요."], "has_start": false, "files": [{"name": "m8_lite.csv", "head": ["variant", "method", "size_mb", "ms_browser", "threat_recall_pct", "top1_acc_pct"], "rows": [["original", "none", "9.8", "85", "92.4", "88.1"], ["weight_30", "weight_prune", "7.1", "85", "92.1", "87.9"], ["weight_50", "weight_prune", "5.6", "84", "91.2", "87.0"], ["channel_30", "channel_prune", "5.4", "58", "91.3", "87.2"], ["channel_50", "channel_prune", "3.4", "41", "89.6", "85.8"]], "n": 7}], "slots": [{"id": "safe", "label": "넘길 정보", "why": "값은 빼고 구조만 알려요.", "options": [{"i": 1, "text": "내부 서버 주소와 관리자 계정, 점검 로그 전체를 그대로 붙여 넣을게. 여기서 알아서 찾아 줘."}, {"i": 0, "text": "실제 값은 넣지 않을게. m8_lite.csv의 열은 variant, method, size_mb, ms_browser, threat_recall_pct, top1_acc_pct이고, m8_safety.csv의 열은 case_id, scenario, expected, output, score야. 코드는 내 환경에서 돌릴게."}, {"i": 2, "text": ""}]}, {"id": "rule", "label": "기준", "why": "기준값을 숫자로 적어요.", "options": [{"i": 0, "text": "size_mb 5 이하, ms_browser 60 이하, threat_recall_pct 90 이상이면 통과, 아니면 어느 기준에서 탈락했는지 적어 줘. 기준값은 맨 위에 설정값으로 둬 줘."}, {"i": 1, "text": "가볍고 빠른 걸로 골라 줘."}, {"i": 2, "text": ""}]}, {"id": "fail", "label": "안전 점검", "why": "실패할 때의 동작을 확인해요.", "options": [{"i": 0, "text": "m8_safety.csv에서 expected가 unknown인데 output이 unknown이 아니면 ‘안전 실패’로 표시하고 목록을 뽑아 줘. 두 표를 넣은 오프라인 점검표를 HTML로 report 변수에 담고, 맨 위에 ‘AI 초안, 전산·보안 담당 확인 전’이라고 적어 줘."}, {"i": 1, "text": "문제 없는지 확인해 줘."}, {"i": 2, "text": ""}]}]}]}, {"short": "사람이 할 일", "title": "사람이 확인할 일과 내 프롬프트", "lead": "AI가 ‘통과’라고 해도 **배포를 결정하는 것은 사람이에요.**", "page": null, "blocks": [{"id": "s3_multi", "type": "multi", "title": "사람이 확인할 일", "req": true, "text": "AI가 만든 점검표를 받은 뒤 사람이 꼭 할 일을 모두 골라 주세요.", "items": ["안전 실패로 나온 경우를 실제로 다시 재현해 본다.", "전산·보안 담당의 확인을 받은 뒤 배포를 결정한다.", "AI가 통과라고 했으니 바로 배포한다.", "프롬프트에 민감 정보가 들어가지 않았는지 다시 본다."]}, {"id": "s3_write", "type": "write", "title": "민감 정보를 뺀 프롬프트 쓰기", "req": true, "prompt": "**새 요청**: 실제 점검 로그에서 응답 시간이 200ms를 넘는 요청을 세어 보고 싶어요. 로그에는 기기 번호와 내부 주소가 들어 있어요. 값을 넘기지 않고 어떻게 요청할까요?", "placeholder": "꼭 입력해 주세요 · 내 말로 프롬프트를 써 보세요", "model_label": "📝 모범 프롬프트", "min": 30}]}]}, "m9": {"steps": [{"short": "먼저 읽기", "title": "고치기 전에 먼저 설명하게 하기", "lead": "숫자가 이상할 때 바로 ‘고쳐 줘’라고 하면 **무엇이 왜 바뀌었는지 모르게 돼요.**", "page": null, "blocks": [{"id": "s1_info", "type": "info", "title": "남의 코드를 고칠 때의 네 단계", "icon": "🧪", "req": false, "text": ["AI가 만든 코드도, 다른 사람이 만든 코드도 같아요. 아래 순서로 요청해요."], "table": {"head": ["단계", "프롬프트에 적을 말", "왜"], "rows": [["설명", "고치기 전에 무엇을 계산하는지 설명해 줘", "내가 이해한 뜻과 같은지 확인해요"], ["테스트", "응답 3건 중 1건 오답이면 33.3%가 나오는 assert를 만들어 줘", "답을 아는 값으로 확인해요"], ["수정", "wrong_rate 함수만 고쳐 줘", "다른 곳이 바뀌지 않아요"], ["비교", "고치기 전후를 표로 보여 줘", "무엇이 달라졌는지 봐요"]]}, "figure": "f_review"}, {"id": "s1_q", "type": "quiz", "title": "처음 무엇을 요청할까요?", "req": true, "text": "오답률 코드가 이상해 보여요. 가장 먼저 AI에게 무엇을 요청할까요?", "options": ["전부 새로 짜 줘.", "이 코드가 무엇을 어떻게 계산하는지 먼저 설명하고, 의심되는 곳을 알려 줘.", "오답률을 더 높게 나오게 해 줘."]}]}, {"short": "테스트로 고치기", "title": "설명·테스트·범위를 담아 요청하기", "lead": "지금 코드는 **오류 없이 돌아요.** 그래서 더 찾기 어려워요.", "page": null, "blocks": [{"id": "s2_vibe", "type": "vibe", "title": "오답률 코드 리뷰와 AAR 문서", "req": true, "ask": ["model", "이 코드로 낸 오답률이 맞는지 봐 주세요."], "has_start": true, "files": [{"name": "m9_log.csv", "head": ["learner_id", "family", "confusable_with", "answered", "correct"], "rows": [["L01", "multicopter", "bird", "1", "1"], ["L01", "multicopter", "bird", "1", "1"], ["L01", "multicopter", "bird", "1", "1"], ["L01", "multicopter", "bird", "1", "1"], ["L01", "fixed_wing", "delta_wing", "1", "1"]], "n": 960}], "slots": [{"id": "explain", "label": "먼저 설명", "why": "고치기 전에 읽게 해요.", "options": [{"i": 0, "text": "m9_log.csv의 열은 learner_id, family, confusable_with, answered(응답 1, 미응답 0), correct(정답 1, 오답 0, 미응답은 빈칸)야. 고치기 전에 wrong_rate가 무엇을 무엇으로 나누는지 설명하고, 의심되는 곳을 알려 줘."}, {"i": 1, "text": "그냥 바로 고쳐 줘."}, {"i": 2, "text": ""}]}, {"id": "test", "label": "테스트", "why": "답을 아는 값으로 확인해요.", "options": [{"i": 0, "text": "응답 3건 중 오답 1건, 미응답 1건인 작은 표로 33.3%가 나오는 assert 테스트를 먼저 만들어 줘."}, {"i": 1, "text": "잘 되는지 알아서 확인해 줘."}, {"i": 2, "text": ""}]}, {"id": "scope", "label": "범위와 문서", "why": "고칠 곳과 문서를 정해요.", "options": [{"i": 1, "text": "알아서 깔끔하게 전부 정리해 줘."}, {"i": 0, "text": "wrong_rate 함수만 고치고 미응답 기록은 지우지 말아 줘. 전후 비교표와 미응답 수를 보여 주고, m9_retrain.csv(condition, samples, before_recall_pct, after_recall_pct, fixed_errors, new_errors)의 재학습 전후도 함께 넣은 AAR 문서를 HTML로 report 변수에 담아 줘. 사람을 탓하지 말고 원인과 다음 조치를 적어 줘."}, {"i": 2, "text": ""}]}]}]}, {"short": "내 말로 쓰기", "title": "후속 제안 판단과 내 프롬프트", "lead": "문서를 받은 뒤에는 **다음 조치를 사람이** 정해요.", "page": null, "blocks": [{"id": "s3_decide", "type": "decide", "title": "후속 제안 판단하기", "req": true, "text": "AAR 문서를 본 팀원들의 제안이에요. 받아들일지 정하고 이유를 적어 주세요.", "reason_min": 8, "items": [{"id": "a", "who": "model", "text": "평균 재현율이 올랐으니 이번 재학습은 성공으로 보고하고 near 조건은 넘어가요."}, {"id": "b", "who": "hun", "text": "미응답이 많은 delta_wing 문항은 제한 시간과 보기 수를 먼저 봐요. 학습자를 탓할 일은 아니에요."}]}, {"id": "s3_write", "type": "write", "title": "리뷰 프롬프트 쓰기", "req": true, "prompt": "**새 요청**: 다른 팀에서 받은 평균 응답 시간 계산 코드가 이상해 보여요. AI에게 어떻게 요청할까요? 고치기 전 설명, 테스트, 범위를 담아 보세요.", "placeholder": "꼭 입력해 주세요 · 내 말로 프롬프트를 써 보세요", "model_label": "📝 모범 프롬프트", "min": 30}]}]}, "m10": {"steps": [{"short": "도구로 묶기", "title": "미니 도구와 브리핑 요청하기", "lead": "지금까지 배운 것을 모두 써 보세요. **네 칸, 숫자 규칙, 근거 수치, 범위**가 모두 필요해요.", "page": null, "blocks": [{"id": "s1_vibe", "type": "vibe", "title": "인계 도구와 1쪽 브리핑", "req": true, "ask": ["hun", "거리별 결과, 인계표, 자동 점검까지 한 번에 나오게 해 주세요."], "has_start": false, "files": [{"name": "m10_run.csv", "head": ["run_id", "distance", "grade", "correct", "reviewed", "ms"], "rows": [["R001", "mid", "unknown", "1", "0", "46"], ["R002", "mid", "check", "0", "1", "45"], ["R003", "mid", "show", "1", "0", "36"], ["R004", "near", "show", "1", "0", "31"], ["R005", "near", "show", "1", "0", "48"]], "n": 240}], "slots": [{"id": "tool", "label": "구성", "why": "다시 돌릴 수 있게 해요.", "options": [{"i": 1, "text": "분석 코드 짜 줘."}, {"i": 0, "text": "m10_run.csv의 열은 run_id, distance(near·mid·far), grade(show·check·unknown), correct, reviewed(교관이 확인했으면 1), ms야. 파일 이름과 표본 기준을 맨 위 설정값으로 두고, 계산은 함수로 나눠 줘. 다음 달 파일로도 바로 돌릴 수 있게 해 줘."}, {"i": 2, "text": ""}]}, {"id": "metric", "label": "지표", "why": "무엇을 어떻게 나눌지 정해요.", "options": [{"i": 0, "text": "distance별로 표본 수, grade가 show인 비율, show의 정답률, grade가 check인 건 중 reviewed 비율, 평균 ms를 표로 만들어 줘. 표본이 30건 미만이면 ‘표본 적음’으로 표시해 줘."}, {"i": 1, "text": "조건별 결과를 보여 줘."}, {"i": 2, "text": ""}]}, {"id": "report", "label": "브리핑", "why": "독자와 조건 표시, 점검을 정해요.", "options": [{"i": 1, "text": "배포해도 된다고 브리핑 써 줘."}, {"i": 0, "text": "교육 담당이 읽을 1쪽 HTML 브리핑을 report 변수에 담아 줘. 순서는 결론 한 줄, 조건별 표, 그림, 한계, 운영 인계표(새 형상 등록·콘텐츠 검수·배포 승인·장애 대응의 담당과 주기), AI 산출물 표기야. 수치에는 가상 기록 몇 건 기준인지 적고, 표본 수와 담당, AI 산출물 표기가 빠졌는지 검사하는 lint_report 함수도 만들어 줘."}, {"i": 2, "text": ""}]}]}]}, {"short": "배포 점검", "title": "담당별 의견과 올리기 전 점검", "lead": "브리핑을 올리기 전에 **사람이 마지막으로 확인**해요.", "page": null, "blocks": [{"id": "s2_decide", "type": "decide", "title": "배포 점검 의견 판단하기", "req": true, "text": "브리핑을 받은 사람들의 의견이에요. 받아들일지 정하고 이유를 적어 주세요.", "reason_min": 8, "items": [{"id": "a", "who": "hun", "text": "‘모든 거리에서 안전함’이라고 단정해 주세요. 읽는 사람이 헷갈려요."}, {"id": "b", "who": "boan", "text": "앱 업데이트는 승인받은 파일만 인터넷 없이 넣게 하고, 담당과 경로를 인계표에 적어 주세요."}]}, {"id": "s2_multi", "type": "multi", "title": "올리기 전에 확인할 것", "req": true, "text": "AI가 만든 브리핑을 올리기 전에 꼭 확인할 것을 모두 골라 주세요.", "items": ["표의 숫자가 파일에서 다시 계산한 값과 같은지 한두 개 대조한다.", "수치마다 데이터와 표본 수, 조건이 적혀 있는지 본다.", "문장이 매끄러우면 내용도 맞으니 건너뛴다.", "실루엣과 자료의 출처·라이선스 표기가 빠지지 않았는지 본다.", "AI가 만든 문서이니 작성 책임은 AI에게 있다고 적는다."]}]}, {"short": "프롬프트 모음", "title": "나만의 프롬프트 틀 만들기", "lead": "이번 프로젝트의 **가장 큰 산출물**이에요. 다음 업무에서 바로 쓸 나만의 틀을 남겨요.", "page": null, "blocks": [{"id": "s3_write", "type": "write", "title": "내 프롬프트 틀 쓰기", "req": true, "prompt": "앞으로 자주 쓸 **프롬프트 틀**을 써 보세요. 파일 이름이나 숫자처럼 바뀌는 자리는 [대괄호]로 비워 두세요. 네 칸에 더해 검증, 범위, 민감 정보 내용도 넣어 보세요.", "placeholder": "꼭 입력해 주세요 · 내 말로 프롬프트를 써 보세요", "model_label": "📝 모범 프롬프트", "min": 40}]}]}};   // 미션 단계와 블록 (정답·힌트·풀이는 빠져 있음)
// ===================== 상태 =====================
let A = null, view = 'welcome', seen = {}, mstep = {}, cache = {}, drafts = {};
let ui = { w: 'full', h: 900, side: true, code: null };
try { Object.assign(ui, JSON.parse(localStorage.getItem(PROJ.ui_key) || '{}')); } catch (e) {}
try { google.colab.output.setIframeHeight(0, true, { maxHeight: 5000 }); } catch (e) {}
const $ = id => document.getElementById(id);
const NUM = n => '①②③④⑤⑥⑦⑧⑨⑩⑪⑫'[n - 1] || n;

function h(tag, props, ...kids) {
  const e = document.createElement(tag);
  for (const [k, v] of Object.entries(props || {})) {
    if (v === undefined || v === null || v === false) continue;
    if (k === 'class') e.className = v;
    else if (k === 'html') e.innerHTML = v;            // 엔진·화면에 적힌 신뢰할 수 있는 문구에만 사용
    else if (k.startsWith('on')) e[k] = v;
    else if (k === 'style') e.style.cssText = v;
    else if (k === 'value') e.value = v;
    else e.setAttribute(k, v === true ? '' : v);
  }
  for (const c of kids.flat(9)) if (c !== null && c !== undefined && c !== false) e.append(c instanceof Node ? c : document.createTextNode(String(c)));
  return e;
}
const esc = s => String(s).replace(/&/g, '&amp;').replace(/</g, '&lt;').replace(/>/g, '&gt;');

let busy = 0;
async function invoke(name, ...args) {
  busy++; $('ws-root').classList.add('busy');
  try {
    const res = await google.colab.kernel.invokeFunction(name, args, {});
    const d = res.data['application/json'];
    if (d.app) refresh(d.app);
    if (d.msg) note(d.msg, d.ok === false ? 'bad' : 'ok');
    if (d.code) setLive(d.code, d.fn);
    return d;
  } catch (e) {
    note('⚠️ 파이썬과 연결되지 않았어요. 위의 🚀 [먼저 실행] 실습 준비하기 셀과 이 셀을 차례로 다시 실행해 주세요.', 'bad');
    throw e;
  } finally { busy--; if (!busy) $('ws-root').classList.remove('busy'); }
}
function note(t, kind) { const n = $('note'); n.textContent = t || ''; n.className = 'note' + (kind ? ' ' + kind : ''); }

// ===================== 화면 틀 =====================
function codeOn() { return ui.code === null ? $('ws-root').offsetWidth >= 1250 : ui.code; }
function applyUi() {
  const root = $('ws-root');
  root.classList.remove('w-base', 'w-wide', 'w-full');
  root.classList.add('w-' + ui.w);
  root.style.width = '';
  root.style.height = ui.h + 'px';
  root.classList.toggle('side-off', !ui.side);
  root.classList.toggle('code-on', codeOn());
  root.classList.toggle('compact', root.offsetWidth < 1000);
  document.querySelectorAll('.ws [data-w]').forEach(b => b.classList.toggle('on', b.dataset.w === ui.w));
  $('tb-side').classList.toggle('on', !ui.side);
  $('tb-code').classList.toggle('on', codeOn());
  try { localStorage.setItem(PROJ.ui_key, JSON.stringify(ui)); } catch (e) {}
}
function setWidth(w) { ui.w = w; applyUi(); }
function setHeight(d) { ui.h = Math.max(480, Math.min(2000, $('ws-root').offsetHeight + d)); applyUi(); }
function toggleSide() { ui.side = !ui.side; applyUi(); }
function toggleCode(force) { ui.code = force === true ? true : !codeOn(); applyUi(); if (codeOn()) renderCode(); }
new ResizeObserver(() => {
  const root = $('ws-root');
  root.classList.toggle('compact', root.offsetWidth < 1000);
  root.classList.toggle('code-on', codeOn());
  $('tb-code').classList.toggle('on', codeOn());
  const hh = root.offsetHeight;
  if (hh && Math.abs(hh - ui.h) > 2) { ui.h = hh; try { localStorage.setItem(PROJ.ui_key, JSON.stringify(ui)); } catch (e) {} }
}).observe($('ws-root'));
new ResizeObserver(() => { const m = $('main'); m.classList.toggle('narrow', m.offsetWidth < 760); }).observe($('main'));

// ===================== 상태 갱신 =====================
function refresh(app) {
  const first = !A;
  if (A && app.boot && A.boot && app.boot !== A.boot) {   // 파이썬이 새로 시작됨(런타임 재시작·셀 재실행) → 화면도 처음 상태로
    mstep = {}; cache = {}; drafts = {}; BUSY = null; for (const k in POLL) clearTimeout(POLL[k]);
    A = app; view = 'welcome'; renderTop(); renderSide(); renderMain();
    note(PROJ.boot_note, 'bad');
    return;
  }
  A = app;
  renderTop(); renderSide();
  if (first || !view.startsWith('mission:')) renderMain();
}
function openView(v) { if (v !== view && busyBlock()) return; view = v; renderSide(); renderMain(); }

// ===================== 상단: 진행률 · 내 정보 =====================
function renderTop() {
  const o = A.overall;
  $('tb-prog').innerHTML = '';
  $('tb-prog').append(h('span', { class: 'pb' }, h('i', { style: `width:${o.pct}%` })), h('b', {}, `전체 진행 ${o.pct}%`), h('span', {}, `· 미션 ${o.done}/${o.total} 완료`));
  $('tb-me').textContent = `👤 ${A.call}`;
}

// ===================== 첫 화면: 입력 없이 인사만 (로그인 없음) =====================
function renderWelcome(main) {
  main.append(h('div', { class: 'scroll' }, h('div', { class: 'onb welcome' },
    h('div', { class: 'wave' }, '👋'), h('h2', {}, `어서오세요, ${A.call}`),
    h('p', { class: 'lead' }, PROJ.welcome_lead),
    h('div', { class: 'row', style: 'justify-content:center;margin-top:18px' }, h('button', { class: 'btn big pulse', id: 'go-home', onclick: () => openView('home') }, '시작하기 ▶')))));
}

// ===================== 사이드바 =====================
const ICON = { done: '✅', open: '▶', soon: '⏳' };
function renderSide() {
  const side = $('side'); side.innerHTML = '';
  $('ws-root').classList.toggle('onboard', view === 'welcome');
  if (view === 'welcome') return;
  const item = (key, icon, label, extra, cls) => side.append(h('div', { class: 'item' + (view === key ? ' active' : '') + (cls ? ' ' + cls : ''), title: label, onclick: () => openView(key) },
    h('span', { class: 'ic' }, icon), h('span', { class: 't' }, label), extra));
  item('home', '🏠', '프로젝트 개요');
  side.append(h('h4', {}, h('b', {}, '채널')));
  let total = 0;
  for (const [id, name] of Object.entries(A.channels)) {
    const n = A.msgs.filter(m => m.ch === id && m.id >= (seen[id] || 0)).length;
    total += n;
    item(id, '', name, n ? h('span', { class: 'badge' }, n) : null, n ? 'unread' : '');
  }
  for (const s of A.project.stages) {
    side.append(h('h4', {}, h('b', {}, s.no + '단계'), ' · ' + s.name));
    for (const mid of s.missions) {
      const m = A.missions[mid];
      const comp = (A.views[mid] || {}).complete || [];
      const mini = m.status === 'open' && m.progress.started ? h('span', { class: 'mini' }, `${comp.filter(Boolean).length}/${comp.length}`) : null;
      item('mission:' + mid, ICON[m.status], `M${m.no} ${m.short}`, mini, m.status === 'soon' ? 'soon' : '');
    }
  }
  side.append(h('h4', {}, h('b', {}, '앱')));
  item('files', '📁', '파일', A.files.length ? h('span', { class: 'badge', style: 'background:#1164a3' }, A.files.length) : null);
  item('roster', '👥', '팀원');
  $('tb-count').textContent = total;
  $('tb-count').classList.toggle('has', total > 0);
}

// ===================== 안내 띠 =====================
function nextTodo() {
  for (const s of A.project.stages) for (const mid of s.missions) { const m = A.missions[mid]; if (m.status === 'open') return m; }
  return null;
}
// ===================== 본 과정의 코드 보기 =====================
const CODE = { topic: null, data: {}, live: null, liveFn: null, open: null };
function hl(code) {
  const re = /(#[^\n]*)|("{3}[\s\S]*?"{3}|'{3}[\s\S]*?'{3}|\b[rbfu]{0,2}"(?:\\.|[^"\\\n])*"|\b[rbfu]{0,2}'(?:\\.|[^'\\\n])*'|"(?:\\.|[^"\\\n])*"|'(?:\\.|[^'\\\n])*')|\b(def|return|if|elif|else|for|in|while|import|from|as|with|not|and|or|is|None|True|False|lambda|try|except|class|yield|global|continue|break|pass)\b|\b(\d+\.?\d*(?:e[-+]?\d+)?)\b|\b([A-Za-z_]\w*)(?=\()/g;
  let out = '', last = 0, m;
  while ((m = re.exec(code))) {
    out += esc(code.slice(last, m.index));
    const cls = m[1] ? 'c' : m[2] ? 's' : m[3] ? 'k' : m[4] ? 'n' : 'f';
    out += `<span class="${cls}">${esc(m[0])}</span>`;
    last = re.lastIndex;
  }
  return out + esc(code.slice(last));
}
function copyText(t) {
  const done = () => note('📋 코드를 복사했어요. 노트북에 새 코드 셀을 만들고 붙여 넣어 실행해 보세요.', 'ok');
  const fallback = () => {          // Colab 출력 창에서 클립보드 권한이 막혀 있을 때
    const ta = h('textarea', { style: 'position:fixed;left:-9999px;top:0' }); ta.value = t; document.body.append(ta); ta.select();
    let ok = false; try { ok = document.execCommand('copy'); } catch (e) {}
    ta.remove();
    if (ok) done(); else note('복사하지 못했어요. 코드를 드래그해서 복사해 주세요.', 'bad');
  };
  try { navigator.clipboard.writeText(t).then(done, fallback); } catch (e) { fallback(); }
}
async function setTopic(topic) {
  CODE.topic = topic;
  if (topic && !CODE.data[topic]) { try { CODE.data[topic] = await invoke('app.code', topic); } catch (e) {} }
  renderCode();
}
function setLive(code, fn) { CODE.live = code; CODE.liveFn = fn || null; if (fn) CODE.open = fn; renderCode(); }
function codeBlock(b, open) {
  const lines = b.code.split('\n');
  const cw = h('div', { class: 'cw' });
  for (const i of b.hl) cw.append(h('div', { class: 'hlb', style: `top:${8 + i * 18}px` }));
  cw.append(h('pre', { class: 'code', html: hl(b.code) }));
  return h('details', { open: open, 'data-fn': b.name }, h('summary', {}, h('code', {}, b.name + '()'), b.doc ? ': ' + b.doc : '', ' ',
      h('button', { class: 'cpy', onclick: e => { e.preventDefault(); copyText(b.code); } }, '📋 복사')),
    h('div', { class: 'cb' }, h('pre', { class: 'ln' }, lines.map((_, i) => i + 1).join('\n')), cw));
}
// 📘 본 과정의 가이드북: [핵심 용어] 탭(미션 소개의 용어 + 단계마다 나오는 어려운 말) · [코드 보기] 탭(지금 화면의 파이썬 코드)
const GUIDE = { tab: 'terms', focus: null };
function openGuide(tab, focus) { GUIDE.tab = tab || 'terms'; GUIDE.focus = focus || null; toggleCode(true); renderCode(); }
function guideTerms(sc) {
  const mid = view.startsWith('mission:') ? view.slice(8) : null, m = mid ? A.missions[mid] : null;
  if (!m || !m.brief) { sc.append(h('p', { class: 'desc' }, '미션을 열면 그 미션의 핵심 용어가 여기에 나와요.')); return; }
  const st = curStep(mid), g = m.brief.glossary || [];
  sc.append(h('div', { class: 'intro' }, '미션 소개에 나온 ', h('b', {}, '핵심 용어'), '와, 단계마다 나오는 어려운 말을 다시 볼 수 있어요. ',
    h('b', {}, '노란 테두리'), '는 지금 단계에 나오는 말이에요. 코드는 위의 ', h('b', {}, '💻 코드 보기'), ' 탭에 있어요.'));
  // 단계 이름은 머리글로 한 번만: 용어 이름과 같아도 헷갈리지 않게 ‘N단계 「이름」’으로 써요 (2026-10-20 제작자 피드백)
  const stepName = k => `${k}단계 「${m.brief.steps_short[k - 1] || ''}」`;
  const now = g.filter(t => t.step === st), rest = g.filter(t => t.step !== st);
  if (now.length) sc.append(h('h6', {}, `🔎 지금 ${stepName(st)}에 나오는 말`), ...now.map(t => termCard(t, true)));
  sc.append(h('h6', {}, `📖 M${m.no} 핵심 용어 (미션 소개)`), ...m.brief.terms.map(t => termCard(t)));
  if (rest.length) {
    const byStep = [...new Set(rest.map(t => t.step))].sort((a, b) => a - b);
    sc.append(h('details', { open: !!(GUIDE.focus && rest.some(t => t.t === GUIDE.focus)) }, h('summary', {}, `M${m.no}의 다른 단계에 나오는 말 (${rest.length}개)`),
      ...byStep.map(k => h('div', { class: 'gstep' }, h('div', { class: 'gstep-h' }, `🧭 ${stepName(k)}에 나오는 말`), ...rest.filter(t => t.step === k).map(t => termCard(t))))));
  }
  const others = Object.values(A.missions).filter(x => x.brief && x.id !== mid);
  if (others.length) sc.append(h('details', {}, h('summary', {}, '다른 미션의 핵심 용어'),
    ...others.map(x => h('div', {}, h('h6', {}, `M${x.no}. ${x.title}`), ...x.brief.terms.map(t => termCard(t))))));
  if (GUIDE.focus) {
    const el = [...sc.querySelectorAll('.term')].find(e => e.dataset.t === GUIDE.focus);
    if (el) { el.classList.add('flash'); setTimeout(() => el.scrollIntoView({ block: 'start', behavior: 'smooth' }), 30); }
    GUIDE.focus = null;
  }
}
function guideCode(sc) {
  const d = CODE.data[CODE.topic];
  if (!(d && d.ok) && !CODE.live) { sc.append(h('p', { class: 'desc' }, '이 화면에는 따로 보여 줄 코드가 없어요. 직접 계산하거나 파일을 읽는 활동에서 그 코드를 보여 줘요.')); return; }
  sc.append(h('div', { class: 'intro' }, '지금 화면이 파이썬 코드로 어떻게 동작하는지 보여 줘요. 고칠 필요는 없어요. ',
    h('b', {}, '📋 복사'), '를 눌러 노트북의 새 코드 셀에 붙여 넣고 실행하면 같은 결과를 직접 볼 수 있어요. ', h('b', {}, '노란 줄'), '은 이 단계의 핵심 부분이에요.'));
  sc.append(h('h6', {}, '▶ 방금 실행된 코드', h('span', { class: 'sp' }), CODE.live ? h('button', { class: 'cpy', onclick: () => copyText(CODE.live) }, '📋 복사') : null),
    h('div', { class: 'live' }, CODE.live ? h('pre', { html: hl(CODE.live) }) : h('div', { class: 'empty' }, '버튼을 누르거나 조건을 바꾸면, 그때 실행된 코드가 여기에 나타나요.')));
  if (d && d.ok) {
    sc.append(h('h6', {}, '📄 ' + d.title), h('p', { class: 'desc' }, d.desc));
    const openName = d.blocks.some(b => b.name === CODE.open) ? CODE.open : d.blocks[0].name;
    for (const b of d.blocks) sc.append(codeBlock(b, b.name === openName));
  } else sc.append(h('p', { class: 'desc' }, '이 화면에는 따로 보여 줄 코드가 없어요.'));
}
function renderCode() {
  const p = $('codep');
  if (!codeOn()) return;
  p.innerHTML = '';
  const sc = h('div', { class: 'cs' });
  const tabs = h('div', { class: 'gtabs' }, [['terms', '📖 핵심 용어'], ['code', '💻 코드 보기']].map(([k, t]) =>
    h('button', { class: GUIDE.tab === k ? 'on' : '', onclick: () => { GUIDE.tab = k; renderCode(); } }, t)));
  if (GUIDE.tab === 'terms') guideTerms(sc); else guideCode(sc);
  p.append(h('div', { class: 'ch' }, h('b', {}, '📘 본 과정의 가이드북'), h('button', { onclick: () => toggleCode() }, '닫기 ✕')), tabs, sc);
  if (GUIDE.tab === 'code') {
    const opened = sc.querySelector(`details[data-fn="${CODE.open}"]`);
    if (opened && CODE.liveFn) opened.scrollIntoView({ block: 'nearest' });
  }
}

// ===================== 공통 위젯 =====================
function taskBox() { return { el: document.createComment('task'), set() {} }; }   // 할 일 상자는 화면에서 뺐어요
function choiceGroup(name, options, value, onchange, opts) {
  opts = opts || {};
  const wrap = h('div', { class: 'choices' + (opts.col ? ' col' : '') });
  options.forEach((o, i) => {
    const [v, label, small] = Array.isArray(o) ? o : [i, o];
    const inp = h('input', { type: 'radio', name, value: v, checked: value !== null && value !== undefined && String(value) === String(v) });
    inp.onchange = () => { wrap.querySelectorAll('.choice').forEach(l => { l.classList.toggle('on', l.querySelector('input').checked); l.classList.remove('wrong'); }); if (onchange) onchange(String(v)); };
    wrap.append(h('label', { class: 'choice' + (inp.checked ? ' on' : '') }, inp, h('span', {}, label, small ? h('small', {}, small) : null)));
  });
  wrap.value = () => { const s = wrap.querySelector('input:checked'); return s ? s.value : ''; };
  return wrap;
}
function textCounter(ta, min) {
  const line = h('div', { class: 'count-line' });
  const upd = () => { const n = ta.value.trim().length; line.textContent = n >= min ? `✔ ${n}자 (충분해요)` : `지금 ${n}자 · ${min}자 이상 적어야 해요`; line.className = 'count-line ' + (n >= min ? 'enough' : 'short'); };
  ta.addEventListener('input', upd); upd();
  return line;
}
function flagMissing(el, msgEl, msg) {
  if (msgEl) msgEl.textContent = msg ? '⚠ ' + msg : '';
  if (el) el.classList.toggle('missing', !!msg);
}
function scrollTo(el) { if (el) el.scrollIntoView({ block: 'center', behavior: 'smooth' }); }
function nextBtn(mid, k, ready, why, label) {
  const msg = h('div', { class: 'err' });
  const btn = h('button', { class: 'btn', onclick: () => { const r = ready ? ready() : true; if (r) goStep(mid, k); else { msg.textContent = '⚠ ' + why(); } } }, label || '다음 단계 ▶');
  return { el: h('div', {}, h('div', { class: 'row' }, btn), msg), btn, msg };
}
// 틀린 선택 표시: 고른 보기를 빨간색으로 (다른 보기를 고르면 choiceGroup이 지움)
function markWrong(group) { const inp = group && group.querySelector('input:checked'); if (inp) inp.closest('.choice').classList.add('wrong'); }
function figure(key) { return FIG[key] ? h('div', { class: 'fig', html: FIG[key] }) : null; }

// ===================== 그림 (용어 설명) =====================
// 핵심 용어 카드에 붙는 작은 SVG 그림. 프로젝트 명세의 figures가 여기에 들어가요 (termCard의 fig 이름으로 찾음).
const T = 'font-family="sans-serif" font-size="10" fill="#111827"';
const FIG = Object.assign({}, PROJ_FIGS);

// ===================== 홈 · 채팅 · 팀원 · 파일 =====================
function renderMain() {
  if (!A) return;
  $('ws-root').classList.add('no-code');          // 코드 보기는 미션 실습 단계에서만 보여요
  const main = $('main'); main.innerHTML = '';
  if (view === 'welcome') return renderWelcome(main);
  if (view === 'home') { return renderHome(main); }
  if (view === 'files') { return renderFiles(main); }
  if (view === 'roster') { return renderTeam(main); }
  if (view.startsWith('mission:')) return renderMission(main, view.slice(8));
  if (A.channels[view]) { return renderChat(main, view); }
  view = 'home'; renderHome(main);
}
function titleBar(main, text, small) { main.append(h('div', { class: 'title' }, small ? h('small', {}, small) : null, text)); }

function renderHome(main) {
  titleBar(main, '프로젝트 개요', '🏠');
  const sc = h('div', { class: 'scroll' });
  const o = A.overall;
  sc.append(h('div', { class: 'hero' }, h('div', { style: 'font-size:12px' }, '프로젝트'), h('h2', {}, A.project.title), h('p', {}, A.project.goal)));
  if (A.teacher) sc.append(h('div', { class: 'tip', style: 'margin-top:12px' }, '🧑‍🏫 선생님용 파일이에요. 모든 활동의 값이 채워져 있고, 산출물도 📁 파일에 만들어져 있어요.'));
  if (A.assets && !A.assets.ok) sc.append(assetNotice());
  sc.append(h('div', { class: 'greet' }, `👋 어서오세요, ${A.call}.`),
    h('div', { class: 'card' }, h('div', { class: 'bigprog' }, h('b', {}, '전체 진행'), h('span', { class: 'pb' }, h('i', { style: `width:${o.pct}%` })), h('b', {}, o.pct + '%')),
      h('div', { class: 'hint', style: 'margin-top:6px' }, `미션 ${o.total}개 중 ${o.done}개 완료 · 끝낸 활동만큼 올라가요.`)));
  const nx = nextTodo();
  if (nx) sc.append(h('div', { class: 'row' }, h('button', { class: 'btn big pulse', onclick: () => openView('mission:' + nx.id) },
    `▶ ${nx.progress.started ? '이어서 하기' : '시작하기'}: M${nx.no} ${nx.title}`)));
  const grid = h('div', { class: 'stages' });
  for (const s of A.project.stages) {
    const box = h('div', { class: 'stage' }, h('h5', {}, `${s.no}단계 · ${s.name}`), s.subtitle ? h('div', { class: 'hint stage-sub' }, s.subtitle) : null);
    for (const mid of s.missions) {
      const m = A.missions[mid];
      const label = { done: '완료', open: '진행 가능', soon: '준비 중' }[m.status];
      box.append(h('div', { class: 'mcard', onclick: () => openView('mission:' + mid) }, h('span', {}, ICON[m.status]),
        h('div', { style: 'flex:1' }, h('div', {}, h('b', {}, `M${m.no} `), m.title), m.brief && m.brief.subtitle ? h('div', { class: 'hint msub' }, m.brief.subtitle) : null, h('div', { class: 'hint' }, m.kind === '협업' ? '👥 협업' : '👤 개인')),
        h('span', { class: 'pill ' + m.status }, label)));
    }
    grid.append(box);
  }
  sc.append(h('h4', { style: 'margin:18px 0 0' }, '단계와 미션'), grid);
  main.append(sc);
}

function renderChat(main, ch) {
  seen[ch] = A.msgs.length;
  titleBar(main, A.channels[ch]);
  const box = h('div', { class: 'msgs' });
  for (const m of A.msgs.filter(m => m.ch === ch)) {
    const who = A.members[m.who];
    box.append(h('div', { class: 'msg' + (m.who !== 'me' && m.text.includes(A.call) ? ' mention' : '') },
      h('div', { class: 'avatar', style: 'background:' + who.color }, who.name[0]),
      h('div', {}, h('div', { class: 'meta' }, h('b', {}, who.name), who.role, h('span', { class: 'tag' + (who.ext ? ' ext' : '') }, who.org), h('span', { class: 'time' }, m.time)),
        h('div', { class: 'text' }, m.text))));
  }
  const input = h('input', { type: 'text', placeholder: '메시지 보내기 (Enter) · 가상 팀원은 미션 진행에 맞춰 메시지를 보내요' });
  const send = () => { if (input.value.trim()) { invoke('app.send', ch, input.value); input.value = ''; } };
  input.addEventListener('keydown', e => { if (e.key === 'Enter' && !e.isComposing) send(); });
  main.append(box, h('div', { class: 'compose' }, input, h('button', { class: 'btn', onclick: send }, '보내기')));
  box.scrollTop = box.scrollHeight;
  renderSide();
}

function renderTeam(main) {
  titleBar(main, '팀원', '👥');
  const sc = h('div', { class: 'scroll' });
  for (const m of Object.values(A.members)) sc.append(h('div', { class: 'msg' }, h('div', { class: 'avatar', style: 'background:' + m.color }, m.name[0]),
    h('div', {}, h('b', {}, m.name), ' ', m.role, h('span', { class: 'tag' + (m.ext ? ' ext' : '') }, m.org))));
  main.append(sc);
}

function renderFiles(main, openId) {
  titleBar(main, '파일', '📁');
  const sc = h('div', { class: 'scroll' });
  main.append(sc);
  if (!A.files.length) { sc.append(h('div', { class: 'tip' }, '아직 산출물이 없어요. 미션의 [산출물] 단계에서 받으면 여기에 저장돼요.')); return; }
  sc.append(h('div', { class: 'tip' }, '💾 산출물은 노트북이 켜져 있는 동안만 보관돼요. 받을 때마다 [⬇ 내려받기]로 내 컴퓨터에 저장해 두세요.'));
  const prev = h('div');
  for (const f of A.files) sc.append(h('div', { class: 'file' }, h('b', {}, '📄 ' + f.title), h('span', { class: 'hint' }, `  ${f.name} · M${A.missions[f.mission].no}`), f.blanks ? h('span', { class: 'pill soon', style: 'margin-left:6px' }, `안 한 활동 ${f.blanks}개`) : null,
    h('div', { class: 'row' }, h('button', { class: 'btn small ghost', onclick: () => showFile(f, prev) }, '미리보기'), h('button', { class: 'btn small', onclick: () => downloadFile(f) }, '⬇ 내려받기'))));
  sc.append(prev);
  showFile(A.files.find(f => f.id === openId) || A.files[A.files.length - 1], prev);
}
async function showFile(f, box) { const d = await invoke('app.file', f.id); box.innerHTML = ''; box.append(h('h4', {}, '미리보기: ' + f.title), h('iframe', { class: 'preview', srcdoc: d.html })); }
async function downloadFile(f) {
  const d = await invoke('app.file', f.id);
  const a = h('a', { href: URL.createObjectURL(new Blob([d.html], { type: 'text/html' })), download: d.name });
  document.body.append(a); a.click(); a.remove();
}
function openFile(id) { if (busyBlock()) return; view = 'files'; renderSide(); const main = $('main'); main.innerHTML = ''; renderFiles(main, id); }

// ===================== 미션 공통 =====================
function nSteps(m) { return m.brief ? m.brief.steps_short.length : 0; }
function curStep(mid) { return mstep[mid] !== undefined ? mstep[mid] : 0; }
function goStep(mid, k) { if (busyBlock()) return; mstep[mid] = k; renderSide(); renderMain(); }
function renderMission(main, mid) {
  const m = A.missions[mid];
  if (m.status === 'soon' || !m.brief) {
    setTopic(null);
    main.append(h('div', { class: 'scroll brief' }, h('span', { class: 'pill soon' }, '준비 중'), h('h2', {}, `M${m.no}. ${m.title}`),
      h('p', { class: 'sub-t' }, (m.kind === '협업' ? '👥 협업' : '👤 개인') + ' 미션'),
      h('section', {}, h('h4', {}, '🎯 미션 소개'), h('p', {}, m.outline)), h('div', { class: 'tip' }, '이 미션은 아직 준비 중이에요. 열려 있는 미션을 먼저 진행해 주세요.')));
    return;
  }
  const n = nSteps(m), st = curStep(mid);
  const bar = h('div', { class: 'stepper' });
  const comp = (A.views[mid] || {}).complete || [];
  // 잠그지 않음: 뒷부분을 미리 볼 수 있게 모든 단계를 열어 둠. 처음 들어갈 때 시작 메시지만 보냄
  ['미션 소개', ...m.brief.steps_short.map((s, i) => `${i + 1} ${s}`), '📦 산출물'].forEach((name, i) => {
    const ok = i >= 1 && i <= n && comp[i - 1];
    bar.append(h('button', { class: 'stp' + (i === st ? ' cur' : '') + (ok ? ' done' : ''), onclick: async () => {
      if (i >= 1 && !m.progress.started) await invoke('app.start', mid);
      goStep(mid, i);
    } }, (ok ? '✅ ' : '') + name));
  });
  const body = h('div', { class: 'scroll' });
  main.append(bar, body);
  const practice = st >= 1 && st <= n;
  $('ws-root').classList.toggle('no-code', !practice);
  setTopic(practice ? TOPICS[mid][st] : null);
  STEPS[mid][Math.min(st, n + 1)](body, m);
}

function termCard(t, now) {   // 핵심 용어 카드: 미션 소개와 📘 가이드북이 함께 씀 (가이드북 용어에는 수식·더 알아보기가 붙음)
  const rows = [['k1', '한 줄 정의', t.d], ['k2', '쉽게 말하면', t.easy], ['k3', PROJ.use_label, t.use], ['k4', '수식', t.formula], ['k5', '더 알아보기', t.more]].filter(r => r[2]);
  return h('div', { class: 'term' + (now ? ' now' : ''), 'data-t': t.t }, h('h5', {}, '📖 ' + t.t, t.en ? h('small', {}, ' ' + t.en) : null), figure(t.fig),
    h('dl', {}, rows.map(([k, a, b]) => [h('dt', { class: k }, a), h('dd', { class: k === 'k4' ? 'fx' : '' }, b)])));
}
function startButtons(m) {
  const go = async () => { if (!m.progress.started) await invoke('app.start', m.id); goStep(m.id, firstTodoStep(m)); };
  return h('button', { class: 'btn big' + (m.progress.started ? '' : ' pulse'), onclick: go }, m.progress.started ? '이어서 하기 ▶' : '미션 시작 ▶');
}
function firstTodoStep(m) { const c = (A.views[m.id] || {}).complete || []; const k = c.findIndex(x => !x); return k < 0 ? 1 : k + 1; }
// 해외 사례: 기사 주소를 달지 않고, 기사의 주요 내용을 눌러서 열어 보는 토글 (2026-11-05 피드백)
function caseToggle(cs) {
  if (!cs) return null;
  return h('details', { class: 'case' }, h('summary', {}, '📰 ' + cs.title),
    h('div', { class: 'case-b' }, (cs.points || []).map(t => h('p', { html: md(t) })), h('div', { class: 'hint' }, '출처: ' + cs.source)));
}
function assetNotice(extra) {
  const S = A.assets || {};
  return h('div', { class: 'card focus' }, h('h4', {}, '📷 참고자료(사진)를 넣어 주세요'),
    h('p', {}, '사진이 필요한 활동이 있어요. 왼쪽 📁 파일 패널에서 아래 폴더에 받은 사진(또는 zip)을 넣은 뒤 [다시 확인]을 눌러 주세요.'),
    h('pre', { class: 'vb-out' }, S.folder || ''), extra || null,
    S.missing && S.missing.length ? h('div', { class: 'hint' }, `아직 없는 사진 ${S.missing.length}개: ${S.missing.slice(0, 8).join(', ')}${S.missing.length > 8 ? ' …' : ''}`) : null,
    h('div', { class: 'row' }, h('button', { class: 'btn', onclick: async () => { await invoke('app.assets_check'); cache.asset = {}; renderMain(); } }, '다시 확인')));
}
function renderBrief(body, m) {
  const b = m.brief;
  const sec = (title, ...kids) => h('section', {}, h('h4', {}, title), ...kids);
  body.append(h('div', { class: 'brief' },
    h('h2', {}, `M${m.no}. ${m.title}`), h('p', { class: 'sub-t' }, b.subtitle),
    sec('🎯 미션 목표', h('p', {}, b.goal)),
    h('div', { class: 'row' }, startButtons(m)),
    h('section', { class: 'why' }, h('h4', {}, '📰 왜 중요할까요?'), h('p', { style: 'margin:0' }, b.why),
      caseToggle(b.case)),
    sec('🧭 수행 단계', h('ol', {}, b.steps.map(s => h('li', {}, s)))),
    sec('📦 산출물', h('ul', {}, b.outputs.map(s => h('li', {}, s)))),
    b.comps && b.comps.length ? sec('🧩 이 미션에서 기르는 역량', h('ul', {}, b.comps.map(c => h('li', {}, h('b', {}, c.name), c.criteria ? ': ' + c.criteria : '')))) : null,
    b.team ? sec('👥 함께하는 사람', h('p', {}, b.team)) : null,
    sec('📖 핵심 용어', h('p', { class: 'hint', style: 'margin:0 0 8px' }, `용어마다 ‘한 줄 정의 → 쉽게 말하면 → ${PROJ.use_label}’ 순서로 설명했어요. 미션 중에도 이 화면으로 돌아와 다시 볼 수 있어요.`),
      h('div', { class: 'terms' }, b.terms.map(termCard))),
    h('div', { class: 'row', style: 'margin-top:16px' }, startButtons(m))));
}
// 산출물 단계: 언제든 받을 수 있음. 안 한 활동이 있으면 먼저 알려 주고 물어본 뒤, 그래도 받으면 그 부분은 빈칸으로 만듦 (2026-11-05 피드백)
function todoList(m, todo) {
  return h('ul', { class: 'todo' }, todo.map(t => h('li', {}, h('a', { href: '#', onclick: e => { e.preventDefault(); goStep(m.id, t.step); } }, `${t.step}단계 「${t.short}」`), ': ', t.blocks.join(', '))));
}
function doneView(body, m, files, nextMid) {
  const area = h('div'), box = h('div', { class: 'brief' }, h('h2', {}, `📦 M${m.no} 산출물`), h('p', { class: 'sub-t' }, m.title), area);
  body.append(box);
  const draw = async () => {
    const r = await invoke('blk.todo', m.id), todo = r.todo || [];
    area.innerHTML = '';
    const mine = A.files.filter(f => f.mission === m.id);
    area.append(h('section', {}, h('h4', {}, '📋 이 미션의 진행 상황'),
      todo.length ? todoList(m, todo) : h('div', { class: 'fb ok' }, '✅ 필수 활동을 모두 마쳤어요.')));
    const ask = h('div');
    area.append(h('section', {}, h('h4', {}, '📄 받을 문서'), h('ul', {}, files.map(([, t]) => h('li', {}, t))),
      h('div', { class: 'row' }, h('button', { class: 'btn big' + (todo.length ? '' : ' pulse'), id: 'get-output', onclick: async () => {
        const d = await invoke('blk.finish', m.id);
        if (!d.need_confirm) return draw();
        ask.innerHTML = '';
        ask.append(h('div', { class: 'card focus', id: 'confirm-box' }, h('h4', {}, '⚠ 아직 하지 않은 활동이 있어요'), todoList(m, d.todo),
          h('p', {}, '그래도 산출물을 받을까요? 하지 않은 부분은 빈칸으로 나와요.'),
          h('div', { class: 'row' }, h('button', { class: 'btn ghost', onclick: () => goStep(m.id, d.todo[0].step) }, '마저 하러 가기'),
            h('button', { class: 'btn', id: 'force-output', onclick: async () => { await invoke('blk.finish', m.id, true); draw(); } }, '그래도 받기'))));
      } }, mine.length ? '📦 산출물 다시 받기' : '📦 산출물 받기')), ask));
    if (mine.length) area.append(h('section', {}, h('h4', {}, '📁 저장된 산출물'), mine.map(f => h('div', { class: 'row' }, h('span', {}, '📄 ' + f.title),
      f.blanks ? h('span', { class: 'pill soon' }, `안 한 활동 ${f.blanks}개`) : null, h('button', { class: 'btn small ghost', onclick: () => openFile(f.id) }, '열어 보기'))),
      h('div', { class: 'tip' }, '💾 산출물은 노트북이 켜져 있는 동안만 보관돼요. 📁 파일에서 [⬇ 내려받기]로 저장해 두세요.')));
    area.append(h('div', { class: 'row', style: 'margin-top:16px' }, nextMid && A.missions[nextMid].brief
      ? h('button', { class: 'btn ghost', onclick: () => openView('mission:' + nextMid) }, `다음 미션: M${A.missions[nextMid].no} ${A.missions[nextMid].title} ▶`)
      : h('button', { class: 'btn ghost', onclick: () => openView('home') }, '🏠 프로젝트 개요로')));
  };
  draw();
}

// ===================== 공통 위젯: 작업 진행 · 그래프 · 힌트 · 크게 보기 · 작업 중 화면 잠금 · 확인 질문 =====================
const POLL = {};
const pct = v => v === null || v === undefined ? '없음' : (v * 100).toFixed(1) + '%';
// 받침에 맞는 조사: josa('판정', '을', '를') → '을'
function josa(w, a, b) { const c = String(w || '').trim().slice(-1).charCodeAt(0); return c >= 0xAC00 && c <= 0xD7A3 && (c - 0xAC00) % 28 ? a : b; }
const ci = c => c ? `[${Math.round(c[0] * 100)}~${Math.round(c[1] * 100)}%]` : '';
function jobBar(d) {
  const p = d.state === 'done' ? 100 : d.pct || 0;
  return h('div', { class: 'jobbar' + (d.state === 'error' ? ' err' : '') },
    h('div', { class: 'jb-top' }, h('b', {}, d.state === 'error' ? '⚠ 문제가 생겼어요' : d.state === 'done' ? '✅ 끝났어요' : `⏳ ${p}%`), h('span', {}, d.msg || ''),
      h('span', { class: 'sp' }), d.sec ? h('span', { class: 'hint' }, `${Math.floor(d.sec / 60)}분 ${d.sec % 60}초 지남`) : null),
    h('div', { class: 'pb' }, h('i', { style: `width:${p}%` })));
}
async function jobState(name) {
  try { const r = await google.colab.kernel.invokeFunction('app.job', [name], {}); return r.data['application/json']; } catch (e) { return null; }
}
// 뒤에서 도는 작업(내려받기·학습·평가)을 1.5초마다 확인해 진행 막대를 그림. 끝나면 onDone (이 화면에서 진행을 본 경우만)
function watchJob(name, el, onDone, onTick) {
  clearTimeout(POLL[name]);
  let seen = false;
  const tick = async () => {
    const d = await jobState(name);
    if (d && d.state !== 'running' && BUSY === name) BUSY = null;
    if (!d || !el.isConnected) return;
    if (d.state === 'running') {
      if (STAY_JOBS[name]) { BUSY = name; BUSY_T = Date.now(); }
      seen = true; el.innerHTML = ''; el.append(jobBar(d)); if (onTick) onTick(d, el);
      POLL[name] = setTimeout(tick, 1500); return;
    }
    if (d.app) refresh(d.app);
    if (d.state === 'error') {
      if (!seen) return;                 // 예전에 끝난 작업의 오류는 다시 보여 주지 않음 (버튼을 다시 누르면 새로 시작)
      el.innerHTML = '';
      el.append(jobBar(d), h('div', { class: 'fb bad' }, '오류 내용: ' + (d.error || '') + '. 버튼을 다시 눌러 보세요. 계속되면 런타임 → 모두 실행으로 다시 시작해 주세요.'));
      return;
    }
    if (d.state === 'done' && seen) { el.innerHTML = ''; if (onDone) onDone(d); }
  };
  tick();
}
async function startJob(fn, args, name, el, onDone, onTick) {
  const d = await invoke(fn, ...args);
  if (d && d.ok !== false) watchJob(name, el, onDone, onTick);
  return d;
}
// 무거운 준비 작업이 도는 동안에는 다른 화면으로 못 가게 막음: 작업 중 다른 화면을 열면 Colab 런타임이 끊기는 일이 있었음
let BUSY = null, BUSY_T = 0;
const STAY_JOBS = PROJ.stay_jobs;          // 작업 이름 → 화면에 보일 이름 (프로젝트 명세의 stay_jobs)
function stayCard(what, extra) {
  return h('div', { class: 'stay', id: 'stay-card' }, h('b', {}, '⚠ 이 화면에서 기다려 주세요'),
    h('div', {}, `${what} 작업이 도는 동안(약 1~3분)에는 왼쪽 메뉴나 위쪽 단계 버튼 등 다른 화면을 누르지 마세요. 다른 화면을 열면 Colab 런타임이 끊길 수 있어요.`),
    extra ? h('div', { class: 'hint' }, extra) : null);
}
function busyBlock() {
  if (!BUSY) return false;
  if (Date.now() - BUSY_T > 15 * 60 * 1000) { BUSY = null; return false; }     // 15분이 지나도 소식이 없으면 풀어 줌
  note(`⏳ ${STAY_JOBS[BUSY] || '준비'} 작업이 끝날 때까지 이 화면에서 기다려 주세요. 다른 화면을 열면 Colab 런타임이 끊길 수 있어요.`, 'bad');
  const c = document.getElementById('stay-card');
  if (c) { c.classList.remove('shake'); void c.offsetWidth; c.classList.add('shake'); c.scrollIntoView({ block: 'center', behavior: 'smooth' }); }
  return true;
}
// 확인 질문 카드 (4·5단계 공통): 틀리면 힌트가 한 단계씩 자세해지고, 맞혀도 화면이 저절로 넘어가지 않음 → 풀이를 읽고 [다음 단계]를 직접 누름
// (이 동작은 화면 글로 따로 알리지 않음: 2026-10-19 피드백)
function quizCard(mid, qid, fnName, onSolved, title) {
  const box = h('div', { class: 'card' });
  const draw = () => {
    const q = A.views[mid].quiz[qid];
    box.className = 'card' + (q.ok ? '' : ' focus');
    box.innerHTML = '';
    box.append(h('h4', {}, (q.ok ? '✅ ' : '❓ ') + (title || '확인 질문')), h('div', { class: 'qtext' }, q.text));
    if (q.ok) {
      box.append(h('div', { class: 'choices col' }, q.options.map((o, i) => h('div', { class: 'choice' + (i === q.pick ? ' on' : '') }, h('span', {}, (i === q.pick ? '✔ ' : '') + o)))),
        h('div', { class: 'fb ok' }, '✅ 맞아요. ' + q.explain));
      return;
    }
    const ch = choiceGroup(`${mid}-${qid}`, q.options, null, null, { col: true }), fb = h('div');
    box.append(ch, h('div', { class: 'row' }, h('button', { class: 'btn', onclick: async () => {
      const v = ch.value();
      if (v === '') { fb.className = 'fb bad'; fb.textContent = '⚠ 보기를 하나 골라 주세요.'; return; }
      const r = await invoke(fnName, qid, Number(v));
      if (r.correct) { draw(); if (onSolved) onSolved(); } else { fb.className = 'fb bad'; fb.textContent = r.text; }
    } }, '확인')), fb);
  };
  draw();
  return box;
}
// 📘 가이드북(오른쪽)의 핵심 용어 탭을 열고 그 용어로 이동
function guideBtn(label, focus) { return h('button', { class: 'btn small ghost', onclick: () => openGuide('terms', focus) }, label || '📘 가이드북에서 용어 자세히 보기'); }
// 단계별 힌트: 버튼을 누를 때마다 한 줄씩 더 보여 줌 (스스로 생각할 기회를 먼저 줌)
function hintBox(key, hints, title) {
  const st = cache.hints || (cache.hints = {});
  const box = h('div', { class: 'hintbox' });
  const draw = () => {
    box.innerHTML = '';
    const n = st[key] || 0;
    box.append(h('div', { class: 'hb-h' }, h('b', {}, '💡 ' + (title || '힌트')), h('span', { class: 'sp' }),
      n < hints.length ? h('button', { class: 'btn small ghost', onclick: () => { st[key] = n + 1; draw(); } }, n ? `힌트 더 보기 (${n + 1}/${hints.length})` : `힌트 보기 (1/${hints.length})`)
        : h('span', { class: 'hint' }, '힌트를 모두 봤어요')));
    hints.slice(0, n).forEach((t, i) => box.append(h('div', { class: 'hb-i' }, h('b', {}, `${i + 1}. `), t)));
  };
  draw();
  return box;
}
// 영상 크게 보기
function zoomImg(src, cap) {
  const root = $('ws-root');
  const box = h('div', { class: 'zoombox', onclick: () => box.remove() }, h('div', { class: 'zb-in' }, h('img', { src }), h('div', { class: 'zb-cap' }, (cap || '') + '  · 아무 곳이나 누르면 닫혀요')));
  root.append(box);
}
// 작은 SVG 그래프: o = {x:[최소,최대], y:[최소,최대], xl, yl, series:[{pts, color, dash, dots, label}], marks:[{x,y,color,label}], vlines, hlines, w, h}
function chart(o) {
  const W = o.w || 380, H = o.h || 230, L = 44, R = 14, T = 14, B = 38;
  const sx = v => L + (v - o.x[0]) / (o.x[1] - o.x[0]) * (W - L - R), sy = v => H - B - (v - o.y[0]) / (o.y[1] - o.y[0]) * (H - T - B);
  const f = v => Math.abs(v) >= 10 || Number.isInteger(v) ? String(Math.round(v * 10) / 10) : v.toFixed(2);
  let s = `<svg viewBox="0 0 ${W} ${H}" xmlns="http://www.w3.org/2000/svg" font-family="sans-serif" font-size="10" fill="#111827">`;
  s += `<rect x="${L}" y="${T}" width="${W - L - R}" height="${H - T - B}" fill="#fff" stroke="#c3cad5"/>`;
  for (let i = 0; i <= 4; i++) {
    const xv = o.x[0] + (o.x[1] - o.x[0]) * i / 4, yv = o.y[0] + (o.y[1] - o.y[0]) * i / 4;
    s += `<line x1="${sx(xv)}" y1="${T}" x2="${sx(xv)}" y2="${H - B}" stroke="#eef2f7"/><text x="${sx(xv)}" y="${H - B + 13}" text-anchor="middle">${f(xv)}</text>`;
    s += `<line x1="${L}" y1="${sy(yv)}" x2="${W - R}" y2="${sy(yv)}" stroke="#eef2f7"/><text x="${L - 4}" y="${sy(yv) + 3}" text-anchor="end">${f(yv)}</text>`;
  }
  s += `<text x="${(L + W - R) / 2}" y="${H - 6}" text-anchor="middle" font-weight="bold">${esc(o.xl || '')}</text>`;
  s += `<text x="12" y="${(T + H - B) / 2}" text-anchor="middle" font-weight="bold" transform="rotate(-90 12 ${(T + H - B) / 2})">${esc(o.yl || '')}</text>`;
  for (const v of o.vlines || []) s += `<line x1="${sx(v.x ?? v)}" y1="${T}" x2="${sx(v.x ?? v)}" y2="${H - B}" stroke="${v.color || '#8a4b00'}" stroke-dasharray="4 3"/>` + (v.label ? `<text x="${sx(v.x) + 3}" y="${T + 10}" fill="${v.color || '#8a4b00'}">${esc(v.label)}</text>` : '');
  for (const v of o.hlines || []) s += `<line x1="${L}" y1="${sy(v.y ?? v)}" x2="${W - R}" y2="${sy(v.y ?? v)}" stroke="${v.color || '#8a4b00'}" stroke-dasharray="4 3"/>` + (v.label ? `<text x="${W - R - 3}" y="${sy(v.y) - 3}" text-anchor="end" fill="${v.color || '#8a4b00'}">${esc(v.label)}</text>` : '');
  let ly = T + 12;
  for (const se of o.series || []) {
    const pts = se.pts.filter(p => p[1] !== null && p[1] !== undefined);
    if (se.line !== false && pts.length > 1) s += `<polyline fill="none" stroke="${se.color}" stroke-width="2" ${se.dash ? 'stroke-dasharray="5 3"' : ''} points="${pts.map(p => sx(p[0]) + ',' + sy(p[1])).join(' ')}"/>`;
    if (se.dots) for (const p of pts) s += `<circle cx="${sx(p[0])}" cy="${sy(p[1])}" r="${se.r || 2.6}" fill="${p[2] || se.color}" ${p[3] ? `stroke="${p[3]}" stroke-width="1.5"` : ''}/>`;
    if (se.label) { s += `<rect x="${W - R - 118}" y="${ly - 8}" width="10" height="10" fill="${se.color}"/><text x="${W - R - 104}" y="${ly + 1}">${esc(se.label)}</text>`; ly += 14; }
  }
  for (const mk of o.marks || []) s += `<circle cx="${sx(mk.x)}" cy="${sy(mk.y)}" r="5.5" fill="none" stroke="${mk.color || '#dc2626'}" stroke-width="2.5"/>` + (mk.label ? `<text x="${sx(mk.x) + 8}" y="${sy(mk.y) - 6}" font-weight="bold" fill="${mk.color || '#dc2626'}">${esc(mk.label)}</text>` : '');
  return h('div', { class: 'chart', html: s + '</svg>' });
}
function reqChip(text, ok) { return h('span', { class: 'chk-chip ' + (ok ? 'y' : 'n') }, (ok ? '✅ ' : '⚠ ') + text); }
function statTile(label, value, desc) { return h('div', { class: 'tile' }, h('div', { class: 'd' }, label), h('div', { class: 'v' }, value), desc ? h('div', { class: 'd' }, desc) : null); }
function selectBox(opts, value, onchange) {
  const s = h('select', {}, opts.map(([v, t]) => h('option', { value: v, selected: String(value) === String(v) }, t)));
  s.onchange = () => onchange(s.value);
  return s;
}
function textBox(value, min, placeholder, oninput) {
  const ta = h('textarea', { placeholder });
  ta.value = value || '';
  ta.addEventListener('input', () => oninput(ta.value));
  return { ta, line: textCounter(ta, min) };
}

// ===================== 결과 보여 주기 도우미 (2026-10-20 제작자 피드백) =====================
// put: el.append는 null·undefined를 "null" 글자로 그려요(‘null’이 보이는 문제). 조건부 자식이 있으면 늘 put으로 붙여요.
function put(el, ...kids) { el.append(...kids.flat(9).filter(k => k !== null && k !== undefined && k !== false && k !== '')); return el; }
// 혼동행렬 표: o = { rows: [실제 이름…], cols: [예측 이름…], cells: [[개수…]…], good: (i, j) => 맞은 칸인가, rowHead, colHead }
// 행(실제)마다 비율로 칸을 칠해요: 맞은 칸은 초록, 틀린 칸은 빨강.
function confusionTable(o) {
  const good = o.good || ((i, j) => i === j);
  const tot = o.cells.map(r => r.reduce((a, b) => a + b, 0));
  const head = h('tr', {}, h('th', { colspan: 2 }, ''), h('th', { colspan: o.cols.length, style: 'text-align:center' }, (o.colHead || '예측') + ' →'));
  const head2 = h('tr', {}, h('th', {}, ''), h('th', {}, ''), o.cols.map(c => h('th', { style: 'text-align:center' }, c)));
  const rows = o.rows.map((r, i) => h('tr', {},
    i === 0 ? h('th', { rowspan: o.rows.length, style: 'text-align:center;vertical-align:middle' }, (o.rowHead || '실제') + ' ↓') : null, h('th', {}, r),
    o.cells[i].map((v, j) => {
      const f = tot[i] ? v / tot[i] : 0, ok = good(i, j);
      return h('td', { style: `text-align:center;font-weight:bold;min-width:72px;background:${ok ? `rgba(22,163,74,${0.08 + 0.55 * f})` : `rgba(220,38,38,${0.06 + 0.55 * f})`}` },
        String(v), h('div', { class: 'hint' }, tot[i] ? Math.round(f * 100) + '%' : '0%'), o.tags && o.tags[i] && o.tags[i][j] ? h('div', { class: 'hint' }, o.tags[i][j]) : null);
    })));
  return h('table', { class: 't cm', style: 'width:auto' }, head, head2, rows);
}
// 지표 표: rows = [[지표, 값, 뜻], …]
function metricTable(rows) {
  return h('table', { class: 't', style: 'width:auto' }, h('tr', {}, h('th', {}, '지표'), h('th', {}, '값'), h('th', {}, '뜻')),
    rows.map(r => h('tr', {}, h('td', {}, h('b', {}, r[0])), h('td', { style: 'text-align:right;font-weight:bold' }, r[1]), h('td', { class: 'hint' }, r[2] || ''))));
}
// 가로 막대: items = [[이름, 값, 색?], …], o = { max, unit, signed(음수 허용), fmt }
function barChart(items, o) {
  o = o || {};
  const max = o.max || Math.max(1e-9, ...items.map(x => Math.abs(x[1])));
  const fmt = o.fmt || (v => (Math.abs(v) >= 10 ? Math.round(v) : Math.round(v * 100) / 100) + (o.unit || ''));
  return h('div', { class: 'barchart', style: 'display:grid;grid-template-columns:max-content 1fr max-content;gap:4px 8px;align-items:center;font-size:12.5px;margin:6px 0' },
    items.map(([name, v, color]) => {
      const w = Math.min(100, Math.abs(v) / max * (o.signed ? 50 : 100));
      const bar = o.signed
        ? h('div', { style: 'position:relative;height:14px;background:#f1f5f9;border-radius:3px' }, h('div', { style: 'position:absolute;left:50%;top:0;bottom:0;width:1px;background:#94a3b8' }),
          h('div', { style: `position:absolute;top:0;bottom:0;${v >= 0 ? 'left:50%' : `left:${50 - w}%`};width:${w}%;background:${color || (v >= 0 ? '#dc2626' : '#2563eb')};border-radius:3px` }))
        : h('div', { style: 'height:14px;background:#f1f5f9;border-radius:3px' }, h('div', { style: `height:100%;width:${w}%;background:${color || '#2563eb'};border-radius:3px` }));
      return [h('span', {}, name), bar, h('b', {}, fmt(v))];
    }));
}
// 분포(히스토그램, 겹쳐 쌓기): o = { edges: [경계…], series: [{ data: [개수…], color, label }], vline: 값, vlabel, xl, yl }
function histChart(o) {
  const W = 520, H = 220, L = 42, R = 12, T = 12, B = 36, n = o.edges.length - 1;
  const tot = Array.from({ length: n }, (_, i) => o.series.reduce((a, s) => a + s.data[i], 0)), max = Math.max(1, ...tot);
  const x0 = o.edges[0], x1 = o.edges[n], sx = v => L + (v - x0) / (x1 - x0) * (W - L - R), sy = v => H - B - v / max * (H - T - B);
  let s = `<svg viewBox="0 0 ${W} ${H}" xmlns="http://www.w3.org/2000/svg" font-family="sans-serif" font-size="10" fill="#111827"><rect x="${L}" y="${T}" width="${W - L - R}" height="${H - T - B}" fill="#fff" stroke="#c3cad5"/>`;
  for (let i = 0; i <= 4; i++) { const v = max * i / 4; s += `<text x="${L - 4}" y="${sy(v) + 3}" text-anchor="end">${Math.round(v)}</text><line x1="${L}" x2="${W - R}" y1="${sy(v)}" y2="${sy(v)}" stroke="#eef2f7"/>`; }
  for (let i = 0; i <= 4; i++) { const v = x0 + (x1 - x0) * i / 4; s += `<text x="${sx(v)}" y="${H - B + 13}" text-anchor="middle">${Math.round(v * 100) / 100}</text>`; }
  for (let i = 0; i < n; i++) {
    let acc = 0;
    for (const se of o.series) { const v = se.data[i]; if (v) s += `<rect x="${sx(o.edges[i]) + 1}" y="${sy(acc + v)}" width="${sx(o.edges[i + 1]) - sx(o.edges[i]) - 2}" height="${sy(acc) - sy(acc + v)}" fill="${se.color}" fill-opacity="0.85"/>`; acc += v; }
  }
  if (o.vline !== undefined && o.vline !== null) s += `<line x1="${sx(o.vline)}" x2="${sx(o.vline)}" y1="${T}" y2="${H - B}" stroke="#111827" stroke-width="2" stroke-dasharray="5 3"/><text x="${sx(o.vline) + 4}" y="${T + 11}" font-weight="bold">${esc(o.vlabel || '')}</text>`;
  let ly = T + 12;
  for (const se of o.series) { s += `<rect x="${W - R - 120}" y="${ly - 8}" width="10" height="10" fill="${se.color}"/><text x="${W - R - 106}" y="${ly + 1}">${esc(se.label)}</text>`; ly += 14; }
  s += `<text x="${(L + W - R) / 2}" y="${H - 6}" text-anchor="middle" font-weight="bold">${esc(o.xl || '')}</text><text x="11" y="${(T + H - B) / 2}" text-anchor="middle" font-weight="bold" transform="rotate(-90 11 ${(T + H - B) / 2})">${esc(o.yl || '')}</text>`;
  return h('div', { class: 'chart', html: s + '</svg>' });
}
// 결정 영역 지도: o = { x: [최소, 최대], y: [최소, 최대], n, p: [[점수…]…] (아래쪽 줄부터), pts: [[x, y, 종류]…], thr, xl, yl, kinds: { 종류: [색, 모양('o'|'^'|'s'|'d'), 이름] } }
// 칸 색: 임계값 이상이면 빨강 쪽(양성), 아래면 파랑 쪽. 경계가 굵은 선으로 보여요.
function regionMap(o) {
  const W = 460, H = 380, L = 46, R = 12, T = 12, B = 40, n = o.n;
  const cw = (W - L - R) / n, ch = (H - T - B) / n;
  const sx = v => L + (v - o.x[0]) / (o.x[1] - o.x[0]) * (W - L - R), sy = v => H - B - (v - o.y[0]) / (o.y[1] - o.y[0]) * (H - T - B);
  let s = `<svg viewBox="0 0 ${W} ${H}" xmlns="http://www.w3.org/2000/svg" font-family="sans-serif" font-size="10" fill="#111827">`;
  for (let r = 0; r < n; r++) for (let c = 0; c < n; c++) {
    const p = o.p[r][c], pos = p >= o.thr, a = pos ? 0.15 + 0.55 * (p - o.thr) / Math.max(1e-6, 1 - o.thr) : 0.15 + 0.55 * (o.thr - p) / Math.max(1e-6, o.thr);
    const base = pos ? [220, 38, 38] : [37, 99, 235], mix = v => Math.round(255 - (255 - v) * a);   // 흰 바탕에 섞은 불투명 색 (칸 사이 틈이 비치지 않게)
    s += `<rect x="${(L + c * cw).toFixed(2)}" y="${(H - B - (r + 1) * ch).toFixed(2)}" width="${(cw + 1).toFixed(2)}" height="${(ch + 1).toFixed(2)}" fill="rgb(${base.map(mix).join(',')})" shape-rendering="crispEdges"/>`;
  }
  for (let r = 0; r < n; r++) for (let c = 0; c < n; c++) {          // 경계선: 이웃 칸과 판정이 다르면 선을 그음
    const pos = o.p[r][c] >= o.thr, x = L + c * cw, y = H - B - (r + 1) * ch;
    if (c + 1 < n && (o.p[r][c + 1] >= o.thr) !== pos) s += `<line x1="${(x + cw).toFixed(1)}" x2="${(x + cw).toFixed(1)}" y1="${y.toFixed(1)}" y2="${(y + ch).toFixed(1)}" stroke="#111827" stroke-width="1.6"/>`;
    if (r + 1 < n && (o.p[r + 1][c] >= o.thr) !== pos) s += `<line x1="${x.toFixed(1)}" x2="${(x + cw).toFixed(1)}" y1="${y.toFixed(1)}" y2="${y.toFixed(1)}" stroke="#111827" stroke-width="1.6"/>`;
  }
  const mark = (x, y, [color, shape]) => shape === '^' ? `<path d="M${x},${y - 4.5} L${x + 4},${y + 3} L${x - 4},${y + 3} Z" fill="${color}" stroke="#fff" stroke-width="0.8"/>`
    : shape === 's' ? `<rect x="${x - 3.5}" y="${y - 3.5}" width="7" height="7" fill="${color}" stroke="#fff" stroke-width="0.8"/>`
      : shape === 'd' ? `<path d="M${x},${y - 4.5} L${x + 4.5},${y} L${x},${y + 4.5} L${x - 4.5},${y} Z" fill="${color}" stroke="#fff" stroke-width="0.8"/>`
        : `<circle cx="${x}" cy="${y}" r="3.6" fill="${color}" stroke="#fff" stroke-width="0.8"/>`;
  for (const [x, y, k] of o.pts) if (x >= o.x[0] && x <= o.x[1] && y >= o.y[0] && y <= o.y[1]) s += mark(sx(x).toFixed(1), sy(y).toFixed(1), o.kinds[k] || ['#111827', 'o']);
  s += `<rect x="${L}" y="${T}" width="${W - L - R}" height="${H - T - B}" fill="none" stroke="#64748b"/>`;
  for (let i = 0; i <= 4; i++) {
    const xv = o.x[0] + (o.x[1] - o.x[0]) * i / 4, yv = o.y[0] + (o.y[1] - o.y[0]) * i / 4;
    s += `<text x="${sx(xv)}" y="${H - B + 13}" text-anchor="middle">${Math.round(xv * 100) / 100}</text><text x="${L - 4}" y="${sy(yv) + 3}" text-anchor="end">${Math.round(yv * 100) / 100}</text>`;
  }
  s += `<text x="${(L + W - R) / 2}" y="${H - 6}" text-anchor="middle" font-weight="bold">${esc(o.xl || '')}</text><text x="11" y="${(T + H - B) / 2}" text-anchor="middle" font-weight="bold" transform="rotate(-90 11 ${(T + H - B) / 2})">${esc(o.yl || '')}</text>`;
  const legend = h('div', { class: 'row', style: 'gap:10px;flex-wrap:wrap;font-size:12px' }, Object.values(o.kinds).map(([color, shape, name]) =>
    h('span', { html: `<svg width="12" height="12" viewBox="-6 -6 12 12">${mark(0, 0, [color, shape])}</svg> ${esc(name)}` })),
    h('span', { class: 'hint' }, `빨간 영역: 점수 ${o.thr} 이상 (양성으로 판정) · 파란 영역: 그 아래 · 검은 선: 결정 경계`));
  return h('div', {}, h('div', { class: 'chart', html: s + '</svg>' }), legend);
}

// ===================== 미션 단계: 명세(MISSION_UI)의 블록으로 화면 그리기 =====================
// 블록 종류마다 그리는 함수(BLOCK_VIEW)가 있고, 단계 화면(stepPage)이 블록을 차례로 그린 뒤 맨 아래에 [다음 단계]나 [완료하기]를 둬요.
// 원칙: 맞혀도 저절로 넘어가지 않음 · 틀리면 힌트가 한 단계씩 자세해짐 · 정답은 파이썬에만 있음(화면에는 맞힌 뒤의 풀이만 옴)
// 이런 동작 원칙은 화면 글로 따로 알리지 않아요 ("틀리면 힌트가 나와요" 같은 안내는 과한 정보: 2026-10-19 피드백)
function bst(mid, bid) { return ((A.views[mid] || {}).blk || {})[bid] || { ok: false, tries: 0, value: null }; }
const md = t => esc(t === undefined || t === null ? '' : t).replace(/\*\*(.+?)\*\*/g, '<b>$1</b>');
function mdp(t, tag) { return h(tag || 'p', { html: md(t) }); }
function stepPage(mid, si) { return (body, m) => renderStep(body, m, si); }
function renderStep(body, m, si) {
  const st = MISSION_UI[m.id].steps[si - 1], n = MISSION_UI[m.id].steps.length;
  const foot = h('div', { class: 'stepfoot' });
  const drawFoot = () => {      // 잠그지 않음: 뒷부분을 미리 볼 수 있게 [다음 단계]를 늘 보여 줌
    foot.innerHTML = '';
    const req = st.blocks.filter(b => b.req), left = req.filter(b => !bst(m.id, b.id).ok).length;
    if (left) foot.append(h('div', { class: 'hint' }, `이 단계의 필수 활동 ${req.length}개 중 ${req.length - left}개를 마쳤어요.`));
    foot.append(si < n ? nextBtn(m.id, si + 1).el : nextBtn(m.id, n + 1, null, null, '📦 산출물 받으러 가기 ▶').el);
  };
  if (st.lead) body.append(h('div', { class: 'tip', html: md(st.lead) }));
  for (const b of st.blocks) body.append(blockView(m, si, b, drawFoot));
  body.append(foot);
  drawFoot();
}
// 블록 하나: ctx.changed()를 부르면 그 블록과 단계 아래쪽 버튼을 다시 그려요 (직접 만든 활동도 같은 방식)
function blockView(m, si, b, changed) {
  const wrap = h('div');
  const ctx = { m, si, b, wrap, st: () => bst(m.id, b.id), redraw: null, changed: () => { ctx.redraw(); changed(); } };
  const fn = BLOCK_VIEW[b.type];
  ctx.redraw = () => { wrap.innerHTML = ''; fn ? fn(wrap, ctx) : wrap.append(h('div', { class: 'fb bad' }, `알 수 없는 활동 종류: ${b.type}`)); };
  ctx.redraw();
  return wrap;
}
function voicesView(vs, title) {
  return h('div', { class: 'voices' }, title ? h('div', { class: 'hint' }, title) : null, vs.map(([who, text]) => {
    const p = A.members[who] || { name: who, color: '#64748b' };
    return h('div', { class: 'voice' }, h('span', { class: 'avatar', style: `background:${p.color}` }, p.name[0]), h('div', {}, h('b', {}, p.name), ' ', h('span', { html: md(text) })));
  }));
}
// 실습 자료 이미지: "asset:번호"는 파이썬에서 받아 오고(한 번 받으면 캐시), 그 밖의 주소는 그대로 씀. 누르면 크게 보기
function assetImg(ref, cap, cls) {
  const img = h('img', { class: cls || '', alt: cap || '', title: '눌러서 크게 보기' });
  if (typeof ref === 'string' && ref.startsWith('asset:')) {
    const id = ref.slice(6), c = cache.asset || (cache.asset = {});
    const fail = r => {
      if (!img.isConnected) return;
      img.replaceWith(r && r.missing ? assetNotice(h('div', { class: 'fb bad' }, r.msg)) : h('div', { class: 'fb bad' }, (r && r.msg) || '사진을 불러오지 못했어요.'));
    };
    if (c[id]) img.src = c[id];
    else invoke('app.asset', id, 960).then(r => { if (r && r.ok) { c[id] = r.url; img.src = r.url; } else fail(r); }).catch(fail);
    img.onclick = () => invoke('app.asset', id, 1800).then(r => { if (r && r.ok) zoomImg(r.url, cap || ''); });
  } else {
    img.src = ref;
    img.onclick = () => zoomImg(ref, cap || '');
  }
  return img;
}
function assetCaption(ref) { return null; }   // 이미지마다 같은 안내를 반복하지 않아요 (출처는 미션을 마친 뒤 산출물에 정리)
function caseBody(it) {
  return h('div', {}, (it.body || []).map(t => mdp(t)),
    it.facts ? h('table', { class: 't' }, it.facts.map(([k, v]) => h('tr', {}, h('th', { style: 'width:170px', html: md(k) }), h('td', { html: md(v) })))) : null,
    it.images ? h('div', { class: 'row', style: 'align-items:flex-start' }, it.images.map(([cap, src]) => h('figure', { class: 'casefig' },
      assetImg(src, cap), h('figcaption', {}, cap)))) : null,
    it.timeline ? h('div', { class: 'timeline' }, it.timeline.map(([t, x]) => h('div', {}, h('b', {}, t), h('span', { html: md(x) })))) : null,
    it.voices ? voicesView(it.voices, '나온 말') : null);
}
function hintList(hints, tries) {
  return hints && hints.length ? h('div', { class: 'hintbox wrongbox' }, h('div', { class: 'hb-h' }, h('b', {}, `❌ 힌트 (틀린 횟수 ${tries}번)`)),
    hints.map((t, i) => h('div', { class: 'hb-i' }, h('b', {}, `${i + 1}. `), t))) : null;
}
const guideRow = b => b.guide ? h('div', { class: 'row' }, guideBtn(b.guide_label || '📘 가이드북에서 용어 자세히 보기', b.guide)) : null;

const BLOCK_VIEW = {
  info(w, c) {
    const b = c.b, ok = c.st().ok;
    w.append(h('div', { class: 'card' + (b.confirm && !ok ? ' focus' : '') }, b.title ? h('h4', {}, (b.icon || '📄') + ' ' + b.title) : null,
      (b.text || []).map(t => mdp(t)), b.figure ? figure(b.figure) : null,
      b.image ? assetImg(b.image, b.title || '', 'blkimg') : null, b.image ? assetCaption(b.image) : null,
      b.bullets ? h('ul', {}, b.bullets.map(x => mdp(x, 'li'))) : null,
      b.table ? h('table', { class: 't' }, h('tr', {}, b.table.head.map(x => h('th', { html: md(x) }))), b.table.rows.map(r => h('tr', {}, r.map(x => h('td', { html: md(x) }))))) : null,
      b.voices ? voicesView(b.voices) : null, b.tip ? h('div', { class: 'tip', html: md(b.tip) }) : null,
      b.case ? caseToggle(b.case) : null,
      b.sources ? h('div', { class: 'hint' }, '출처: ' + b.sources.join(' · ')) : null,
      guideRow(b),
      b.confirm ? (ok ? h('div', { class: 'fb ok' }, '✅ 확인했어요.') : h('div', { class: 'row' }, h('button', { class: 'btn', onclick: async () => {
        const r = await invoke('blk.check', c.m.id, b.id, true); if (r.correct) c.changed(); } }, b.confirm === true ? '읽었어요 ✔' : b.confirm))) : null));
  },
  quiz(w, c) {
    const b = c.b, s = c.st();
    const box = h('div', { class: 'card' + (s.ok ? '' : ' focus') }, h('h4', {}, (s.ok ? '✅ ' : '❓ ') + (b.title || '확인 질문')), h('div', { class: 'qtext', html: md(b.text) }));
    if (s.ok) {
      const pick = (s.value || {}).pick;
      box.append(h('div', { class: 'choices col' }, b.options.map((o, i) => h('div', { class: 'choice' + (i === pick ? ' on' : '') }, h('span', {}, (i === pick ? '✔ ' : '') + o)))),
        h('div', { class: 'fb ok', html: '✅ 맞아요. ' + md(s.explain) }));
    } else {
      const ch = choiceGroup(`${c.m.id}-${b.id}`, b.options, null, null, { col: true }), fb = h('div');
      box.append(ch, h('div', { class: 'row' }, h('button', { class: 'btn', onclick: async () => {
        const v = ch.value();
        if (v === '') { fb.className = 'fb bad'; fb.textContent = '⚠ 보기를 하나 골라 주세요.'; return; }
        const r = await invoke('blk.check', c.m.id, b.id, Number(v));
        if (r.correct) c.changed(); else { fb.className = 'fb bad'; fb.textContent = r.text; markWrong(ch); }
      } }, '확인')), fb);
    }
    box.append(guideRow(b) || '');
    w.append(box);
  },
  multi(w, c) {
    const b = c.b, s = c.st(), key = 'mp' + c.m.id + b.id;
    const pick = drafts[key] || (drafts[key] = new Set((s.value || {}).picked || []));
    const res = h('div');
    const box = h('div', { class: 'card' + (s.ok ? '' : ' focus') }, h('h4', {}, (s.ok ? '✅ ' : '☑ ') + (b.title || '해당하는 것 모두 고르기')), h('div', { class: 'qtext', html: md(b.text) }));
    b.items.forEach((it, i) => {
      const cb = h('input', { type: 'checkbox', checked: pick.has(i), disabled: s.ok, onchange: () => { cb.checked ? pick.add(i) : pick.delete(i); cb.closest('.choice').classList.remove('wrong'); } });
      box.append(h('label', { class: 'choice', 'data-i': i, style: 'margin:5px 0' }, cb, h('span', { html: md(it) })));
      if (s.ok && (s.explains || [])[i]) box.append(h('div', { class: 'hint', style: 'margin:-2px 0 6px 28px', html: md(s.explains[i]) }));
    });
    if (s.ok) box.append(h('div', { class: 'fb ok' }, '✅ 모두 맞게 골랐어요.'));
    else box.append(h('div', { class: 'row' }, h('button', { class: 'btn', onclick: async () => {
      const r = await invoke('blk.check', c.m.id, b.id, [...pick]);
      if (r.correct) { c.changed(); return; }
      res.innerHTML = '';
      box.querySelectorAll('.choice').forEach(l => l.classList.remove('wrong'));
      (r.items || []).filter(x => !x.ok).forEach(x => { const l = box.querySelector(`.choice[data-i="${x.i}"]`); if (l) l.classList.add('wrong'); });
      res.append(h('div', { class: 'fb bad' }, r.text), ...(r.items || []).filter(x => !x.ok).map(x => h('div', { class: 'fb bad' }, `${x.i + 1}번: ${x.text}`)));
    } }, '확인')), res);
    box.append(guideRow(b) || '');
    w.append(box);
  },
  match(w, c) {
    const b = c.b, key = 'mt' + c.m.id + b.id;
    const mt = drafts[key] || (drafts[key] = Object.assign({}, c.st().value || {}));
    const box = h('div');
    const draw = (res, hints, msg) => {
      box.innerHTML = '';
      const ok = c.st().ok;
      const opts = [['', '고르세요'], ...b.rights.map(r => [r, r])];
      const tbl = h('table', { class: 't' }, h('tr', {}, h('th', {}, b.left_label || '항목'), h('th', {}, b.right_label || '짝'), h('th', {}, '채점')),
        b.lefts.map(l => {
          const r = res ? res[l] : (ok ? true : null);
          return h('tr', { class: r === false ? 'wrongrow' : '' }, h('td', { html: '<b>' + md(l) + '</b>' }),
            h('td', {}, ok ? (mt[l] || '') : selectBox(opts, mt[l] || '', v => { mt[l] = v; })),
            h('td', { class: r === null ? '' : r ? 'ok' : 'bad' }, r === null ? '' : r ? '✅' : '❌ 다시'));
        }));
      box.append(h('div', { class: 'card' + (ok ? '' : ' focus') }, h('h4', {}, (ok ? '✅ ' : '🔗 ') + (b.title || '짝짓기')), b.text ? mdp(b.text) : null, tbl,
        ok ? h('div', { class: 'fb ok', html: '✅ 모두 맞게 짝지었어요.' + (c.st().explain ? ' ' + md(c.st().explain) : '') }) : h('div', { class: 'row' }, h('button', { class: 'btn', onclick: async () => {
          const r = await invoke('blk.check', c.m.id, b.id, mt);
          if (r.correct) c.changed(); else draw(r.result, r.hints, r.text);
        } }, '확인')),
        msg ? h('div', { class: 'fb bad' }, msg) : null, hintList(hints, c.st().tries), guideRow(b)));
    };
    draw(null, null);
    w.append(box);
  },
  order(w, c) {
    const b = c.b, s = c.st(), key = 'od' + c.m.id + b.id;
    const ord = drafts[key] || (drafts[key] = (s.value || {}).order ? [...s.value.order] : b.items.map(x => x.id));
    const nm = Object.fromEntries(b.items.map(x => [x.id, x]));
    const list = h('div'), fb = h('div');
    const draw = wrong => {
      list.innerHTML = '';
      const ok = c.st().ok;
      ord.forEach((id, i) => list.append(h('div', { class: 'orditem' + (wrong === i ? ' wrong' : '') }, h('b', {}, `${i + 1}.`),
        h('span', { style: 'flex:1', html: md(nm[id].text) + (nm[id].desc ? ' <small>(' + md(nm[id].desc) + ')</small>' : '') }),
        ok ? null : h('button', { class: 'btn small ghost', disabled: i === 0, onclick: () => { [ord[i - 1], ord[i]] = [ord[i], ord[i - 1]]; draw(); } }, '▲'),
        ok ? null : h('button', { class: 'btn small ghost', disabled: i === ord.length - 1, onclick: () => { [ord[i + 1], ord[i]] = [ord[i], ord[i + 1]]; draw(); } }, '▼'))));
    };
    draw();
    w.append(h('div', { class: 'card' + (s.ok ? '' : ' focus') }, h('h4', {}, (s.ok ? '✅ ' : '🧩 ') + (b.title || '순서 맞추기')), b.text ? mdp(b.text) : null, list,
      s.ok ? h('div', { class: 'fb ok', html: '✅ 순서가 맞아요.' + (s.explain ? ' ' + md(s.explain) : '') }) : h('div', { class: 'row' }, h('button', { class: 'btn', onclick: async () => {
        const r = await invoke('blk.check', c.m.id, b.id, ord);
        if (r.correct) c.changed(); else { fb.className = 'fb bad'; fb.textContent = r.text; draw(r.wrong); }
      } }, '순서 확인'), h('span', { class: 'hint' }, '▲▼ 버튼으로 순서를 바꿔요.')), fb, guideRow(b)));
  },
  classify(w, c) {
    const b = c.b, C = b.categories, PL = b.primary_label || '직접 원인';
    w.append(h('div', { class: 'card' }, h('h4', {}, (c.st().ok ? '✅ ' : '🔎 ') + (b.title || '분류하기')), b.text ? mdp(b.text) : null,
      h('div', {}, C.map(([k, n, d]) => h('span', { class: 'chk-chip', title: d || '' }, n))), guideRow(b)));
    const name = k => (C.find(x => x[0] === k) || [k, k])[1];
    for (const it of b.items) {
      const box = h('div');
      const draw = () => {
        box.innerHTML = '';
        const st = c.st(), z = (st.value || {})[it.id], ex = (st.explains || {})[it.id];
        const card = h('div', { class: 'card' + (z ? '' : ' focus') }, h('h4', {}, (z ? '✅ ' : '📂 ') + it.title), caseBody(it));
        if (z) {
          const cs = b.contrib === false ? '' : (z.contrib.length ? ` · 함께 작용한 것: ${z.contrib.map(name).join(', ')}` : '');
          card.append(h('div', { class: 'fb ok' }, `✅ ${PL}: ${name(z.primary)}${cs}`));
          if (ex) card.append(h('div', { class: 'hint', html: md(ex) }));
          box.append(card);
          return;
        }
        const key = 'cl' + c.m.id + b.id + it.id, d = drafts[key] || (drafts[key] = { p: null, k: new Set() });
        const pr = choiceGroup(key, C.map(([k, n, ds]) => [k, n, ds]), d.p, v => { d.p = v; });
        const ks = b.contrib === false ? null : h('div', { class: 'row' }, C.map(([k, n]) => {
          const cb = h('input', { type: 'checkbox', checked: d.k.has(k), onchange: () => { cb.checked ? d.k.add(k) : d.k.delete(k); } });
          return h('label', { class: 'choice' }, cb, h('span', {}, n));
        }));
        const fb = h('div');
        card.append(...[h('div', { class: 'qtext' }, `${PL}${josa(PL, '을', '를')} ${b.contrib === false ? '' : '하나 '}고르세요`), pr,
          ks ? h('div', { class: 'qtext' }, '함께 작용한 것도 있으면 골라 주세요 (여러 개 골라도 돼요)') : null, ks,
          h('div', { class: 'row' }, h('button', { class: 'btn', onclick: async () => {
            if (!d.p) { fb.className = 'fb bad'; fb.textContent = '⚠ 하나를 골라 주세요.'; return; }
            const r = await invoke('blk.check', c.m.id, b.id, { item: it.id, primary: d.p, contrib: [...d.k] });
            if (r.correct) c.changed(); else if (r.item_ok) draw(); else { fb.className = 'fb bad'; fb.textContent = r.text || r.msg || '다시 확인해 주세요.'; markWrong(pr); }
          } }, '확인')), fb].filter(Boolean));
        box.append(card);
      };
      draw();
      w.append(box);
    }
  },
  // 글쓰기: 정답을 판정하지 않아요. 최소 글자 수만 채우면 제출되고, 제출한 뒤에 모범답안이 보여요 (2026-10-19 피드백)
  write(w, c) {
    const b = c.b, s = c.st(), key = 'wr' + c.m.id + b.id;
    const d = drafts[key] || (drafts[key] = { t: s.value || '' });
    const tb = textBox(d.t, b.min || 10, b.placeholder || '꼭 입력해 주세요', v => { d.t = v; });
    const fb = h('div');
    if (s.ok) fb.append(h('div', { class: 'fb ok' }, '✅ 제출했어요.'));
    w.append(h('div', { class: 'card' + (s.ok ? '' : ' focus') }, h('h4', {}, (s.ok ? '✅ ' : '✍ ') + (b.title || '글쓰기')), b.prompt ? mdp(b.prompt) : null,
      h('div', { class: 'fieldrow' }, tb.ta, tb.line),
      h('div', { class: 'row' }, h('button', { class: 'btn', onclick: async () => {
        const r = await invoke('blk.check', c.m.id, b.id, tb.ta.value);
        if (r.correct) c.changed(); else { fb.className = 'fb bad'; fb.textContent = r.text; }
      } }, s.ok ? '고쳐서 다시 제출' : '제출하기'), b.guide ? guideBtn(b.guide_label || '📘 가이드북', b.guide) : null), fb,
      s.ok && s.model ? h('div', { class: 'model' }, h('div', { class: 'model-h' }, b.model_label || '📝 모범답안'), h('div', { class: 'model-t' }, s.model),
        s.explain ? h('div', { class: 'hint', html: md(s.explain) }) : null) : null,
      s.ok && s.checklist && s.checklist.length ? h('div', { class: 'selfcheck' }, h('div', { class: 'model-h' }, '🔍 내 글 점검'),
        s.checklist.map(t => h('label', { class: 'choice' }, h('input', { type: 'checkbox' }), h('span', {}, t)))) : null));
  },
  decide(w, c) {
    const b = c.b, s = c.st(), key = 'dc' + c.m.id + b.id;
    const d = drafts[key] || (drafts[key] = { ch: {}, rs: {} });
    if (s.value) for (const k in s.value) { d.ch[k] = d.ch[k] || s.value[k].choice; d.rs[k] = d.rs[k] || s.value[k].reason; }
    const opts = b.options || [['accept', '받아들임'], ['reject', '거절하고 대안 제시']];
    const res = {}, box = h('div'), fbAll = h('div');
    const draw = () => {
      box.innerHTML = '';
      const ok = c.st().ok;
      for (const it of b.items) {
        const p = A.members[it.who];
        const ch = choiceGroup(key + it.id, opts, d.ch[it.id] || null, v => { d.ch[it.id] = v; });
        const tb = textBox(d.rs[it.id] || '', b.reason_min || 10, '꼭 입력해 주세요 · 결정한 이유 또는 대안', v => { d.rs[it.id] = v; });
        const r = res[it.id], ex = ok ? (c.st().explains || {})[it.id] : null;
        if (r && !r.ok) markWrong(ch);
        box.append(h('div', { class: 'card' + (r && !r.ok ? ' focus' : '') },
          h('div', { class: 'voice' }, p ? h('span', { class: 'avatar', style: `background:${p.color}` }, p.name[0]) : null,
            h('div', {}, p ? h('b', {}, `${p.name} (${p.role})`) : null, ' ', h('span', { html: '“' + md(it.text) + '”' }))),
          ch, tb.ta, tb.line, r && !r.ok ? h('div', { class: 'fb bad' }, r.text) : null, ex ? h('div', { class: 'hint', html: md(ex) }) : null));
      }
    };
    draw();
    w.append(h('div', { class: 'card' + (s.ok ? '' : ' focus') }, h('h4', {}, (s.ok ? '✅ ' : '🗣 ') + (b.title || '결정하기')), b.text ? mdp(b.text) : null, box,
      h('div', { class: 'row' }, h('button', { class: 'btn', onclick: async () => {
        const r = await invoke('blk.check', c.m.id, b.id, { choices: d.ch, reasons: d.rs });
        for (const k in res) delete res[k];
        (r.items || []).forEach(x => { res[x.id] = x; });
        if (r.correct) c.changed(); else { fbAll.className = 'fb bad'; fbAll.textContent = r.text; draw(); }
      } }, s.ok ? '다시 저장' : '결정 저장')), fbAll, guideRow(b)));
  },
  // 바이브 코딩 실습: 칸마다 문장을 골라 프롬프트를 만들고, 가상 AI에게 보내고, AI가 만든 코드의 실제 실행 결과를 보고 고쳐 다시 보냄
  vibe(w, c) {
    const b = c.b, s = c.st(), key = 'vb' + c.m.id + b.id;
    const d = drafts[key] || (drafts[key] = { picks: {}, reply: null, attempts: [], loaded: false });
    const replyBox = h('div', { class: 'vb-replybox' }), fb = h('div'), promptBox = h('pre', { class: 'vb-prompt' });
    const promptText = () => b.slots.map(sl => { const o = sl.options.find(x => String(x.i) === String(d.picks[sl.id])); return o && o.text ? o.text : ''; }).filter(Boolean).join('\n');
    const drawPrompt = () => { const t = promptText(); promptBox.textContent = t || '칸마다 문장을 고르면 여기에 프롬프트가 만들어져요.'; promptBox.classList.toggle('empty', !t); };
    const drawReply = () => {
      replyBox.innerHTML = '';
      const r = d.reply;
      if (r) replyBox.append(vibeReply(r));
      if (d.attempts.length > 1 || (d.attempts.length && !d.attempts[0].ok)) replyBox.append(h('details', { class: 'vb-log', open: !(r && r.ok) }, h('summary', {}, `📝 프롬프트 개선 기록 (${d.attempts.length}번)`),
        h('table', { class: 't' }, h('tr', {}, h('th', {}, '시도'), h('th', {}, '부족했던 칸'), h('th', {}, '결과')),
          d.attempts.map(a => h('tr', {}, h('td', {}, a.no), h('td', {}, a.ok ? '✅ 성공' : (a.slot || '')), h('td', {}, a.result))))));
    };
    if (!d.loaded && (s.ok || s.tries || b.has_start)) {      // 다시 그릴 때 파이썬에 남은 마지막 응답과 ‘지금까지의 코드’를 받아 옴 (선생님용은 처음부터 채워져 있음)
      d.loaded = true;
      invoke('app.vibe', c.m.id, b.id).then(r => {
        d.reply = r.reply; d.attempts = r.attempts || []; d.start = r.start;
        for (const k in r.picks || {}) d.picks[k] = String(r.picks[k]);
        c.redraw();
      });
    }
    const send = async () => {
      if (b.slots.some(sl => d.picks[sl.id] === undefined)) { fb.className = 'fb bad'; fb.textContent = '⚠ 칸마다 문장을 하나씩 골라 주세요.'; return; }
      fb.className = 'fb'; fb.textContent = '⏳ 가상 AI가 코드를 만들고 실행하는 중이에요…';
      const r = await invoke('blk.check', c.m.id, b.id, { picks: d.picks });
      fb.className = ''; fb.textContent = '';
      d.loaded = true;
      if (r.reply) { d.reply = r.reply; d.attempts = r.attempts || []; }
      if (r.correct) c.changed(); else drawReply();
      w.querySelectorAll('.vb-slot .choice').forEach(l => l.classList.remove('wrong'));
      if (!r.correct && r.reply && r.reply.slot_id) markWrong(w.querySelector(`.vb-slot[data-slot="${r.reply.slot_id}"]`));
      const el = wrapEl().querySelector('.vb-replybox'), sc = el && el.closest('.scroll');      // 바깥 페이지는 움직이지 않고 미션 화면만 응답 위치로
      if (sc) sc.scrollTop += el.getBoundingClientRect().top - sc.getBoundingClientRect().top - 8;
    };
    const wrapEl = () => w;
    w.append(h('div', { class: 'card' + (s.ok ? '' : ' focus') }, h('h4', {}, (s.ok ? '✅ ' : '🤖 ') + (b.title || '바이브 코딩 실습')),
      b.ask ? voicesView([b.ask]) : null, b.text ? mdp(b.text) : null,
      d.start ? h('div', { class: 'vb-start' }, h('div', { class: 'vb-label' }, h('b', {}, '📌 지금까지의 코드와 결과')), vibeReply(Object.assign({}, d.start, { open_code: true }))) : null,
      (b.files || []).map(f => h('details', { class: 'vb-data', open: true }, h('summary', {}, `📄 ${f.name} (${f.n}행, 앞부분만 보여요)`),
        h('div', { class: 'vb-tbl' }, h('table', { class: 't' }, h('tr', {}, f.head.map(x => h('th', {}, x))), f.rows.map(r => h('tr', {}, r.map(x => h('td', {}, x)))))))),
      b.slots.map(sl => h('div', { class: 'vb-slot', 'data-slot': sl.id }, h('div', { class: 'vb-label' }, h('b', {}, sl.label), sl.why ? h('span', { class: 'hint' }, ' ' + sl.why) : null),
        choiceGroup(key + sl.id, sl.options.map(o => [o.i, o.text || '(이 칸은 비워 두기)']), d.picks[sl.id] === undefined ? null : d.picks[sl.id], v => { d.picks[sl.id] = v; drawPrompt(); }, { col: true }))),
      h('div', { class: 'vb-label' }, h('b', {}, '✍ 내 프롬프트')), promptBox,
      h('div', { class: 'row' }, h('button', { class: 'btn', onclick: send }, s.ok || d.attempts.length ? '🤖 다시 보내기' : '🤖 가상 AI에게 보내기'), guideRow(b)), fb, replyBox));
    drawPrompt();
    drawReply();
  },
  custom(w, c) {
    const fn = window[c.b.renderer];
    if (typeof fn !== 'function') { w.append(h('div', { class: 'fb bad' }, `화면 함수 ${c.b.renderer}를 찾지 못했어요.`)); return; }
    fn(w, c);
  },
};
// 가상 AI의 응답 한 건: 말 → 코드 → 실제 실행 결과(출력·표·그림·오류) → 만들어진 문서 → 짚어 볼 점
function vibeReply(r) {
  const out = h('div', { class: 'vb-reply' + (r.ok ? ' ok' : '') + (r.slot ? ' miss' : '') }, h('div', { class: 'vb-who' }, '🤖 가상 AI'), r.say ? h('div', { class: 'vb-say', html: md(r.say) }) : null);
  if (r.code) out.append(h('details', { class: 'vb-codebox', open: !!r.open_code }, h('summary', {}, '💻 AI가 만든 코드 (눌러서 보기) ', h('button', { class: 'cpy', onclick: e => { e.preventDefault(); copyText(r.code); } }, '📋 복사')),
    h('pre', { class: 'vb-code', html: hl(r.code) })));
  if (r.code) out.append(h('div', { class: 'vb-label' }, h('b', {}, '▶ 실행 결과')));
  if (r.stdout) out.append(h('pre', { class: 'vb-out' }, r.stdout));
  (r.tables || []).forEach(t => out.append(h('div', { class: 'vb-tbl', html: t })));
  (r.figs || []).forEach(u => out.append(h('img', { class: 'blkimg', src: u, title: '눌러서 크게 보기', onclick: () => zoomImg(u, '') })));
  if (r.error) out.append(h('pre', { class: 'vb-err' }, r.error));
  if (r.report) out.append(h('div', { class: 'vb-label' }, h('b', {}, '📄 AI가 만든 문서')), h('iframe', { class: 'preview vb-doc', srcdoc: '<meta charset="utf-8"><style>' + VB_DOC_CSS + '</style>' + r.report }));
  if (r.problem) out.append(h('div', { class: 'vb-problem' }, h('b', {}, '🔎 결과를 보니: '), h('span', { html: md(r.problem) })));
  if (r.hint) out.append(h('div', { class: 'vb-problem' }, h('b', {}, `❌ ${r.slot ? '‘' + r.slot + '’ 칸의 ' : ''}현재 선택을 다시 고민해 보세요: `), h('span', { html: md(r.hint) })));
  if (r.lesson) out.append(h('div', { class: 'fb ok', html: '✅ ' + md(r.lesson) }));
  return out;
}
const VB_DOC_CSS = 'body{font-family:sans-serif;color:#111827;background:#fff;line-height:1.6;font-size:14px;padding:6px 10px}table{border-collapse:collapse;margin:6px 0}td,th{border:1px solid #c8ced6;padding:4px 8px;font-size:13px;text-align:left}th{background:#eef2f7}h1,h2,h3{margin:8px 0 6px}';

// ===================== 단계 등록 =====================
const STEPS = {
  m1: [renderBrief, stepPage('m1', 1), stepPage('m1', 2), stepPage('m1', 3), (b, m) => doneView(b, m, [["m1_level_table", "M1 형상 계열 실루엣과 식별 가능 수준표"]], 'm2')],
  m2: [renderBrief, stepPage('m2', 1), stepPage('m2', 2), stepPage('m2', 3), (b, m) => doneView(b, m, [["m2_label_doc", "M2 라벨 일치율과 라벨링 규격 요청"]], 'm3')],
  m3: [renderBrief, stepPage('m3', 1), stepPage('m3', 2), stepPage('m3', 3), (b, m) => doneView(b, m, [["m3_baseline", "M3 기준 모델 조건별 성능표와 수정 기록"]], 'm4')],
  m4: [renderBrief, stepPage('m4', 1), stepPage('m4', 2), stepPage('m4', 3), (b, m) => doneView(b, m, [["m4_selection", "M4 요구 성능 판정과 모델 선정 기록"]], 'm5')],
  m5: [renderBrief, stepPage('m5', 1), stepPage('m5', 2), stepPage('m5', 3), (b, m) => doneView(b, m, [["m5_calibration", "M5 점수 근거표와 행동 등급 기준"]], 'm6')],
  m6: [renderBrief, stepPage('m6', 1), stepPage('m6', 2), stepPage('m6', 3), (b, m) => doneView(b, m, [["m6_app", "M6 출제 계획표와 학습 앱 화면 시안"]], 'm7')],
  m7: [renderBrief, stepPage('m7', 1), stepPage('m7', 2), stepPage('m7', 3), (b, m) => doneView(b, m, [["m7_review", "M7 해설 대조표와 교관 검토 절차서"]], 'm8')],
  m8: [renderBrief, stepPage('m8', 1), stepPage('m8', 2), stepPage('m8', 3), (b, m) => doneView(b, m, [["m8_offline", "M8 경량화 비교와 오프라인 점검표"]], 'm9')],
  m9: [renderBrief, stepPage('m9', 1), stepPage('m9', 2), stepPage('m9', 3), (b, m) => doneView(b, m, [["m9_aar", "M9 사후검토 문서와 수정 기록"]], 'm10')],
  m10: [renderBrief, stepPage('m10', 1), stepPage('m10', 2), stepPage('m10', 3), (b, m) => doneView(b, m, [["m10_handover", "M10 1쪽 브리핑·운영 인계표와 프롬프트 틀"]], null)],
};
const TOPICS = {
  m1: [null, null, null, null, null],
  m2: [null, null, null, null, null],
  m3: [null, null, null, null, null],
  m4: [null, null, null, null, null],
  m5: [null, null, null, null, null],
  m6: [null, null, null, null, null],
  m7: [null, null, null, null, null],
  m8: [null, null, null, null, null],
  m9: [null, null, null, null, null],
  m10: [null, null, null, null, null],
};

async function resetAll() {
  if (!confirm('모든 진행 상황을 지우고 처음부터 할까요?')) return;
  mstep = {}; seen = {}; cache = {}; drafts = {}; CODE.live = null; BUSY = null;
  for (const k in POLL) clearTimeout(POLL[k]);
  await invoke('app.reset');
  openView('home');
}

applyUi();
invoke('app.state');
</script>
"""

if "app_state" not in globals():
    display(HTML("<div style='padding:14px 16px;border:2px solid #b3261e;border-radius:10px;background:#fce8e6;color:#111827;font-size:15px'>"
                 "⚠️ 먼저 위의 <b>🚀 [먼저 실행] 실습 준비하기</b> 셀을 실행해 주세요. 진행 막대가 100%가 된 뒤 이 셀을 다시 실행하면 메신저가 열려요.</div>"))
else:
    display(HTML(APP_HTML.replace("__WORKSPACE__", WORKSPACE_NAME)))
```
