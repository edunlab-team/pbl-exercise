```python
#@title ② M01 작성 화면 (실행하세요) { display-mode: "form" }
import ipywidgets as w
from IPython.display import display, HTML
from datetime import datetime, timezone
import json, zipfile, io, time

OFFICIAL_TITLE = '야전 군수 보급 프로세스(요청-승인-불출-수령) 및 필수 증빙 문서 체계 분석'
FORM_PATH=ROOT/'forms/LOG07_M01_forms.md'
ANSWER_PATH=ROOT/'docs/log07_baseline_assessment.md'
DIALOGUE_PATH=ROOT/'ai_dialogues'
FORM_TEXT=FORM_PATH.read_text(encoding='utf-8')
BASE_TEXT=ANSWER_PATH.read_text(encoding='utf-8') if ANSWER_PATH.exists() else (INSTRUCTOR_SAMPLE if globals().get('INSTRUCTOR_SAMPLE') is not None else FORM_TEXT)

def _escape(s):
    import html
    return html.escape(s)

items=[('미션 안내','materials/LOG07_M01_lesson.md'),('제공 자료 E01~E06','materials/LOG07_M01_data.md'),('공통 증빙 예시','materials/LOG07_보급기록_예시.md'),('빈 작성 양식','forms/LOG07_M01_forms.md')]
reference=w.Dropdown(options=[k for k,v in items],description='자료 선택',layout=w.Layout(width='95%'))
refout=w.Output(layout=w.Layout(border='1px solid #ddd',max_height='380px',overflow='auto'))
def show_ref(change=None):
    key=dict(items)[reference.value]
    with refout:
        refout.clear_output(wait=True)
        display(Markdown((ROOT/key).read_text(encoding='utf-8')))
reference.observe(show_ref,names='value');show_ref()

name=w.Text(description='작성자',placeholder='학습자 이름 또는 실습 식별명',layout=w.Layout(width='90%'))
team=w.Text(description='조/역할',placeholder='조 또는 맡은 역할 (없으면 개인 실습)',layout=w.Layout(width='90%'))
answer=w.Textarea(value=BASE_TEXT,description='실습 기록',layout=w.Layout(width='98%',height='460px'))
purpose=w.Text(value='자료 분석·검토',description='AI 목적',layout=w.Layout(width='92%'))
request=w.Textarea(description='보낸 요청',placeholder='AI에게 실제로 보낸 질문 원문 전체',layout=w.Layout(width='98%',height='115px'))
response=w.Textarea(description='받은 답변',placeholder='AI에게 실제로 받은 답변 원문 전체',layout=w.Layout(width='98%',height='150px'))
status=w.Output();save=w.Button(description='작성 내용 저장',button_style='success');store=w.Button(description='AI 대화 원문 저장');backup=w.Button(description='JSON 백업',button_style='info');restore=w.FileUpload(accept='.json',multiple=False);restore_btn=w.Button(description='백업 JSON 복원');export=w.Button(description='검토용 ZIP 만들기',button_style='primary')

def save_answer(_):
    with status:
        status.clear_output(wait=True)
        if not answer.value.strip() or answer.value.strip()==FORM_TEXT.strip():
            print('빈 작성 양식 그대로는 저장하지 않습니다. 실제 작성한 부분이나 미실시 이유를 적으세요.');return
        ANSWER_PATH.parent.mkdir(parents=True,exist_ok=True)
        if ANSWER_PATH.exists():
            history=ROOT/'history';history.mkdir(exist_ok=True)
            (history/f'm01_{time.time_ns()}.md').write_bytes(ANSWER_PATH.read_bytes())
        metadata='작성자: '+name.value+'\n조/역할: '+team.value+'\n\n'
        ANSWER_PATH.write_text(metadata+answer.value,encoding='utf-8')
        print('저장 완료:',ANSWER_PATH)

def save_dialogue(_):
    with status:
        status.clear_output(wait=True)
        if not all(x.strip() for x in [request.value,response.value,purpose.value]):
            print('실제 요청·답변·대화 목적을 모두 입력하세요.');return
        DIALOGUE_PATH.mkdir(exist_ok=True)
        p=DIALOGUE_PATH/f'M01_{time.time_ns()}.json'
        p.write_text(json.dumps({'mission':1,'purpose':purpose.value,'request':request.value,'response':response.value,'saved_at':datetime.now(timezone.utc).isoformat()},ensure_ascii=False,indent=2),encoding='utf-8')
        print('AI 원문 저장:',p.name)

def payload():
    return {'version':'LOG07-M01-1','mission':1,'title':OFFICIAL_TITLE,'name':name.value,'team':team.value,'answer':answer.value,'purpose':purpose.value,'request':request.value,'response':response.value,'saved_at':datetime.now(timezone.utc).isoformat(),'note':'본 JSON은 입력 중인 초안 백업이며 학습 통과 증거가 아닙니다'}

def download_or_link(path):
    try:
        from google.colab import files as colab_files
        colab_files.download(str(path))
    except ImportError:
        display(FileLink(str(path)))

def save_backup(_):
    with status:
        status.clear_output(wait=True)
        p=ROOT.parent/f'LOG07_m01_작성백업_{time.time_ns()}.json'
        p.write_text(json.dumps(payload(),ensure_ascii=False,indent=2),encoding='utf-8')
        print('작성 중인 내용 백업 생성:',p.name)
        download_or_link(p)

def do_restore(_):
    with status:
        status.clear_output(wait=True)
        if not restore.value:
            print('복원할 JSON을 먼저 선택하세요.');return
        value=restore.value
        f=next(iter(value.values())) if isinstance(value,dict) else value[0]
        content=f['content'] if isinstance(f,dict) else f.content
        data=json.loads(bytes(content).decode('utf-8'))
        if data.get('version')!='LOG07-M01-1' or data.get('mission')!=1:raise ValueError('M01 작성 백업만 복원할 수 있습니다')
        for control,key in [(name,'name'),(team,'team'),(answer,'answer'),(purpose,'purpose'),(request,'request'),(response,'response')]: control.value=str(data.get(key,''))
        print('복원 완료. 입력란을 확인한 다음 저장하세요.')

def make_zip(_):
    with status:
        status.clear_output(wait=True)
        p=ROOT.parent/f'LOG07_m01_검토용_{time.time_ns()}.zip'
        with zipfile.ZipFile(p,'w',zipfile.ZIP_DEFLATED) as zout:
            for file in sorted(ROOT.rglob('*')):
                if file.is_file() and not any(q in file.parts for q in ['__pycache__','.ipynb_checkpoints']):
                    zout.write(file,file.relative_to(ROOT))
            zout.writestr('작성중_백업.json',json.dumps(payload(),ensure_ascii=False,indent=2))
        print('검토용 ZIP 생성:',p.name,'· 학습 PASS/교관 승인이 아닙니다')
        download_or_link(p)

save.on_click(save_answer);store.on_click(save_dialogue);backup.on_click(save_backup);restore_btn.on_click(do_restore);export.on_click(make_zip)
display(w.HTML(f'<h3>M01 · {_escape(OFFICIAL_TITLE)}</h3><p>위의 활동별 설명을 읽고, 양식의 해당 제목 아래에 직접 기록하세요. 예시와 대조한 이유를 반드시 남깁니다.</p><p>공식 과제: 6단계 흐름·RACI 30칸·기본 확인 15개 이상·환경별 확인 2개 이상. 학생이 작성한 결과와 AI 대화 원문을 구분해 저장합니다.</p>'))
display(reference,refout,name,team,answer,save,w.HTML('<h4>실제 AI 대화 기록</h4>'),purpose,request,response,store,w.HBox([backup,export]),w.HTML('<h4>백업 복원</h4>'),restore,restore_btn,status)
```
