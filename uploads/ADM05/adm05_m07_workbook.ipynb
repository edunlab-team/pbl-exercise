- **검증 완료 지표:** 국방 표준 문서 규범 준수, 주관적 편향 배제, 인간 최종 승인(HITL) 원칙 완비


```python
%%writefile rubric/adm05_evaluation_rubric.md
# [AD미션 5_미션 7] 고부하 스트레스 검증 및 평가 루브릭 명세서 (모범 답안)

## 1. 관찰 가능한 행동 지표 기반 4단계 평가 루브릭
| 평가 영역 | 탁월 (4점) | 우수 (3점) | 보통 (2점) | 미흡 (1점) |
| :--- | :--- | :--- | :--- | :--- |
| **1. 분류 정확성** | 3대 요소 분류 오류 0건 | 오류 1건 이하 (즉시 교정) | 오류 2~3건 발생 | 오류 4건 이상 발생 (분류 붕괴) |
| **2. 시각·출처 기재** | 시각/출처 결측 0건 | 결측 1건 (추적 가능) | 결측 2~3건 발생 | 결측 4건 이상 (타임라인 상실) |
| **3. 중복 처리** | 동일 표적 중복 100% 필터링 | 중복 1건 경미 계상 | 중복 2건 계상 | 전과/피해 2배 이상 중복 부풀림 |
| **4. 미확인 관리** | 책임자·기한 기재율 100% | 기재율 90% 이상 | 기재율 70~89% | 기재율 70% 미만 (추적 포기) |
| **5. 보고 적시성** | 10분 이내 보고서 상신 완결 | 12분 이내 상신 완결 | 15분 이내 상신 완결 | 15분 초과 (보고 골든타임 상실) |

## 2. 과부하 붕괴 임계점 분석 및 비상 패스트트랙 프로토콜
- **붕괴 임계점:** 1인당 분당 일지 인입량이 **15건을 초과**하는 시점부터 미확인 사항 책임자 누락률이 40% 이상으로 급증함.
- **비상 패스트트랙 전환 규칙:**
  1. 인입량 분당 15건 초과 시 '비상 간이 모드' 자동 선포.
  2. 비핵심 행정 조치 기술을 일시 중단하고 **[적 돌파 / 아군 사상 / 탄약 결핍] 3대 핵심 위험**만을 3줄로 즉시 상신.
  3. 전투 종료 후 2시간 이내에 보관된 녹취록을 통해 상세 일지 사후 소급 복원 의무화.
```
