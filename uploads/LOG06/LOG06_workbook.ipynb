```python
%%writefile docs/log06_doc_framework.md
# [LOG06_미션 01] 야전 긴급 상황 임시 물자 취급 및 격리 지침서 골격 (모범 답안)

## 1. 6단계 라이프사이클별 실태 분석표
| 단계 | 일반 물자(L)에서 생기는 문제 | 출처불명 장비(U)에서 생기는 문제 | 사고·정산 오류 경로 | 필수 증빙 |
| :---: | :--- | :--- | :--- | :--- |
| **발견** | 보급 소요 식별 지연, 발견 좌표 누락 | 호기심에 의한 임의 접촉·전원 조작 | 부비트랩 폭발 및 전파 역탐지 노출 | 발견 보고 일지, 비접촉 사진 |
| **인수** | 간이 수기 메모 누락, 불출 대장 차이 | 인수자 단독 서명, 직무 분리 위반 | 횡령 의혹 및 단독 조작 책임 공방 | 임시 영수증(이중 서명) |
| **격리** | 물자 혼적, 보관 구역 식별 불명 | 격리 표지 미부착, 접근 차단선 부재 | 타 부대원 무단 접촉, 2차 오염 | 격리 확인서, 격리 표지판 |
| **보고** | 교전 중 보고 순연, 수량 오기 | 이상 징후(팽창/누출) 지연 보고 | 화재/폭발 위험 방치, 지휘 공백 | 상황 보고 일지, 긴급 전언 |
| **인계** | 인계 수령증 분실, 구두 인계 | 전문기관 인수관 서명 누락 | 물자 유실 및 관할권 분쟁 발생 | 인계 확인서(쌍방 서명) |
| **처리결과기록** | 폐기·소모 증빙 미비, 사후 구두 처리 | 자체 무단 분해 또는 임의 폐기 | 군수 감사 지적 및 징계 회부 | 사후 정산 보고서, 파기 사진 |

## 2. 가상 사례 C1~C5 결함 분석표
| 사례 | 대상 | 있었던 조치·기록 | 빠진 조치·기록 | 식별된 결함 및 개선 방향 |
| :---: | :---: | :--- | :--- | :--- |
| **C1** | U-2 | 구두 보고 1건 | 비접촉 수칙 준수, 전원 조작 차단 | 비접촉·전원차단 기본 수칙 미전파. 현장 안전수칙 카드 지급 필요 |
| **C2** | U-5 | 인수 메모 1장 | 격리 구역 표지 설치, 접근 통제선 | 안전 격리 구역 미지정 방치. 차폐 보관함 및 경고 표지 부착 의무화 |
| **C3** | U-4 | 시각 미기재 메모 | 분단위 보고 타임스탬프, 긴급 EOD 요청 | 위험 징후 보고 기준 부재. 팽창 식별 즉시 15분 내 전파 기준 수립 |
| **C4** | L-3 | 구두 전언만 있음 | 정비 확인관 입회 소모 확인서 | 비공식 구두 소모. 현장 간이 소모 확인서(카본 복사식) 즉시 발행 |
| **C5** | U-1 | 단독 서명 인계증 | 인수관(정보사/EOD) 공식 서명 날인 | 일방적 인계 처리. 쌍방 대면 이중 서명 없이는 인계 불인정 |

## 3. 표준 업무 지침서 8개 장 목차 골격
| 장 번호 | 장 제목 | 핵심 규정 내용 | 지휘관 결심 정보 |
| :---: | :--- | :--- | :--- |
| 제1장 | 총칙 및 적용 범위 | 비접촉·조작금지·전문기관 인계 원칙 | 부대 방호 및 통제 책임선 확립 |
| 제2장 | 용어의 정의 | L물자 vs U장비, 격리, 증빙 신뢰도 정의 | 위험물과 일반 보급품의 취급 분리 결심 |
| 제3장 | 야전 임시 취급 6단계 수칙 | 발견-인수-격리-보고-인계-기록 절차 | 현장 초동 조치 적정성 평가 |
| 제4장 | 필수 증빙 및 신뢰도 기준 | 영수증·확인서 양식, G1~G6 판정 기준 | 정산 증빙 채택 및 사후 인정 여부 결정 |
| 제5장 | 사후 정산 표준 보고 절차 | 타임라인 충돌 해소 및 불일치 소명 | 전투 손실 인정 및 감사 대비 완료도 |
| 제6장 | 지휘관 확인 서명 및 감사 체크리스트 | 직무 분리, 결재 임계치, 의사결정 트리 | 지휘관 최종 법적·행정적 결재 승인 |
| 제7장 | 대량 유실 대비 스트레스 훈련 | 다량 유입, 기록 분실 시 재구성 루브릭 | 극한 전장 상황 부대 회복 탄력성 점검 |
| 제8장 | 지침 개정 관리 및 인간 책임 | 피드백 반영 절차, 인간 최종 검증(HITL) | 군수 규정 지속 가능성 및 감사 무결성 |
```
