4. `reports/재개기록.md`와 마지막 기록지를 열어 **마지막 미션·실행일시·작성 문장**이 저장 사본과 같은지 확인합니다. outputs의 execution.json 시간도 대조합니다. 없으면 진행을 멈추고 저장 사본부터 확인합니다.

### 저장과 새 작업 환경 복원 (계속)

5. 이어 할 미션의 코드 저장·실행 셀을 선택합니다. 복원한 work 코드를 다시 쓸 때는 노트북 셀에도 같은 수정 내용을 반영하고 백업 셀을 먼저 실행합니다. 처음부터 모두 실행해 복원 코드를 기본값으로 덮지 않습니다. 같은 분석을 재실행하면 이전 로그는 outputs/history에 남고 새 로그의 실행일시만 새로 찍힙니다. 재개기록에 직접 적은 이전 일시·마지막 미션은 자동으로 바꾸지 않습니다.

### 링크가 열리지 않으면(노트북만 따로 연 경우) 핵심 안내 (계속)

- 보안: 제공된 교육용 가상자료만 사용합니다. 비공개 업무자료, 실습에는 제공된 가상자료만 사용하고, 실제 이름·연락처 같은 개인정보는 AI 도구나 노트북에 넣지 않습니다.

- 저장: 노트북 사본과 reports·직접 작성한 docs·work·outputs를 원래 폴더 구조대로 저장합니다.
- 제출: 제출처·마감 일시·시간대는 miliai.kr 해당 과제 공지를 확인하고, 표시가 없으면 제출 전에 교육 담당자에게 확인합니다. 별도 공지가 없으면 미션마다 그 미션 산출물을 ZIP 하나로 묶어 해당 미션 제출란에 첨부합니다.
- 파일명: 파일명은 과제ID_미션번호_산출물명으로 작성하고 실제 형식의 확장자를 붙입니다. 실명·개인 식별번호는 파일명에 넣지 않습니다.

````python
import sys, subprocess, importlib.metadata as metadata
from pathlib import Path
import shutil, datetime
if not ((3,9) <= sys.version_info[:2] <= (3,12)):
    print("⚠️ 이 패키지는 Python 3.9~3.12에서 검증했습니다. 현재 Python " + sys.version.split()[0] + "에서도 계속 진행하지만 결과 수치가 다를 수 있습니다.")
required = {"numpy":"2.0.2","pandas":"2.3.3","scipy":"1.13.1","scikit-learn":"1.6.1","matplotlib":"3.9.4"}
missing=[]
for name,version in required.items():
    try:
        if metadata.version(name)!=version: missing.append(f"{name}=={version}")
    except metadata.PackageNotFoundError: missing.append(f"{name}=={version}")
try:
    if metadata.version("pytest").split(".")[0] != "8": missing.append("pytest>=8,<9")
except metadata.PackageNotFoundError: missing.append("pytest>=8,<9")
if missing: print("⚠️ 검증 기준과 다른 패키지 버전: " + ", ".join(missing) + ". 계속 진행하지만 결과 수치가 다를 수 있습니다. 같은 결과가 필요하면 requirements.txt 기준 환경을 사용하세요. 자동 설치하지 않습니다.")
ROOT=Path.cwd()
payload = {'src/learning_support.py': '# coding: utf-8\n"""Distributed runtime: actual execution, plot export and evidence checks; no teacher answers."""\nfrom pathlib import Path\nimport contextlib, io, json, re, runpy, sys, hashlib, platform, datetime, shutil, traceback\nROOT = Path(__file__).resolve().parent.parent\n\ndef execute(mission, variant=False):\n    import matplotlib\n    matplotlib.use(\'Agg\')\n    import matplotlib.pyplot as plt\n    stem = f\'m{mission:02d}\'\n    source = ROOT / (\'work\' if variant else \'src\') / (stem + (\'_challenge.py\' if variant else \'.py\'))\n    folder = ROOT / \'outputs\' / stem / (\'challenge\' if variant else \'baseline\')\n    if folder.exists():\n        archive=ROOT/\'outputs/history\'/stem/(folder.name+\'-\'+datetime.datetime.now().strftime(\'%Y%m%d-%H%M%S-%f\'))\n        archive.parent.mkdir(parents=True,exist_ok=True)\n        shutil.move(str(folder),str(archive))\n    folder.mkdir(parents=True, exist_ok=True)\n    if not variant:\n        checks=json.loads((ROOT/\'configs/checks.json\').read_text(encoding=\'utf-8\'))[str(mission)]\n        for alias in checks.get(\'aliases\',[]):\n            dest=ROOT/alias\n            dest.parent.mkdir(parents=True,exist_ok=True)\n            if not dest.exists():\n                dest.write_bytes(source.read_bytes())\n            elif dest.read_bytes()!=source.read_bytes():\n                print(\'기존 원문 산출물 파일을 보존했습니다:\',alias,\'; 제출 전 baseline/work 중 의도한 코드와 대조하세요.\')\n    figures = []\n    def save_figures(*args, **kwargs):\n        for number in plt.get_fignums():\n            dest = folder / f\'figure_{len(figures)+1:02d}.png\'\n            plt.figure(number).savefig(dest, dpi=110, bbox_inches=\'tight\')\n            figures.append(str(dest.relative_to(ROOT)))\n        plt.close(\'all\')\n    old_show = plt.show\n    plt.show = save_figures\n    out, err = io.StringIO(), io.StringIO()\n    ok = False\n    try:\n        with contextlib.redirect_stdout(out), contextlib.redirect_stderr(err):\n            runpy.run_path(str(source), run_name=\'__main__\')\n            save_figures()\n        ok = True\n    except BaseException:\n        traceback.print_exc(file=err)\n        raise\n    finally:\n        plt.show = old_show\n        plt.close(\'all\')\n        (folder/\'stdout.txt\').write_text(out.getvalue(), encoding=\'utf-8\')\n        (folder/\'stderr.txt\').write_text(err.getvalue(), encoding=\'utf-8\')\n        meta = dict(completed=ok, source=str(source.relative_to(ROOT)), source_sha256=hashlib.sha256(source.read_bytes()).hexdigest(), python=platform.python_version(), timestamp=datetime.datetime.now(datetime.timezone.utc).isoformat(), figures=figures)\n        (folder/\'execution.json\').write_text(json.dumps(meta,ensure_ascii=False,indent=2),encoding=\'utf-8\')\n    print(out.getvalue())\n    for f in figures: print(\'그래프 파일:\', f)\n    return folder\n\ndef check_baseline(mission):\n    spec = json.loads((ROOT/\'configs/checks.json\').read_text(encoding=\'utf-8\'))[str(mission)]\n    folder=ROOT/\'outputs\'/f\'m{mission:02d}\'/\'baseline\'\n    meta=json.loads((folder/\'execution.json\').read_text(encoding=\'utf-8\'))\n    assert meta[\'completed\'], \'실행이 완료되지 않았습니다. 오류를 먼저 확인하세요.\'\n    source=ROOT/meta[\'source\']\n    assert meta[\'source_sha256\']==hashlib.sha256(source.read_bytes()).hexdigest(), \'실행 후 코드가 변경되었습니다. 다시 실행하세요.\'\n    stdout=(folder/\'stdout.txt\').read_text(encoding=\'utf-8\')\n    missing=[p for p in spec[\'patterns\'] if not re.search(p,stdout)]\n    assert not missing, \'기본 실행 확인 지표 불일치: \'+repr(missing)+\'; 원문 예시/환경/수정 이력을 대조하세요.\'\n    print(\'기본 실행 확인 PASS — 도전 과제 및 해석 완료 판정은 별도입니다.\')\n    return stdout\n\ndef check_action(mission):\n    spec=json.loads((ROOT/\'configs/checks.json\').read_text(encoding=\'utf-8\'))[str(mission)]\n    if spec[\'explanation_only\']:\n        print(\'설명형 도전: 코드 변경 대신 기록지의 근거와 해석을 확인하세요.\')\n        return\n    folder=ROOT/\'outputs\'/f\'m{mission:02d}\'/\'challenge\'\n    assert (folder/\'execution.json\').exists(), \'도전 코드를 수정하고 실행하세요.\'\n    meta=json.loads((folder/\'execution.json\').read_text(encoding=\'utf-8\'))\n    assert meta[\'completed\'], \'도전 실행이 완료되지 않았습니다.\'\n    target=ROOT/meta[\'source\']\n    assert hashlib.sha256(target.read_bytes()).hexdigest()==meta[\'source_sha256\'], \'수정 후 재실행이 필요합니다.\'\n    base=ROOT/\'src\'/f\'m{mission:02d}.py\'\n    # AST comparison ignores cosmetic comments/whitespace while accepting valid independent solutions.\n    import ast\n    assert ast.dump(ast.parse(target.read_text(encoding=\'utf-8\')))!=ast.dump(ast.parse(base.read_text(encoding=\'utf-8\'))), \'주석만 바꾼 상태입니다. 도전 과제의 설정 또는 로직을 수정하세요.\'\n    print(\'수정·재실행 증빙 확인 — 요구한 변경의 적합성과 해석은 채점자가 확인합니다. 출력 동일도 허용합니다.\')\n\nif __name__==\'__main__\':\n    mission=int(sys.argv[1]); mode=sys.argv[2] if len(sys.argv)>2 else \'run\'\n    if mode==\'run\':execute(mission)\n    elif mode==\'challenge\':execute(mission, True)\n    elif mode==\'check\':check_baseline(mission)\n    elif mode==\'action\':check_action(mission)\n    else:raise SystemExit(\'unknown mode\')\n', 'configs/checks.json': '{\n  "1": {\n    "patterns": [\n      "정상 샘플 20건 중 검증 통과 20건, 오류 0건",\n      "데이터 딕셔너리\\\\] 피처 24개",\n      "제원 불일치 시험:.*2단계 이상 차이",\n      "6\\\\.검사관 보정.*7\\\\.정비 등급 승인"\n    ],\n    "explanation_only": false,\n    "aliases": []\n  },\n  "2": {\n    "patterns": [\n      "원천 사진 158장",\n      "R7 조명 보정·정규화\\\\(평균 밝기 0\\\\.3 미만 사진 수\\\\)\\\\s+\\\\d+건",\n      "정합성 검증 12/12 통과, 통과율 100%",\n      "정제 후 사진 150장.*긴급 점검 6장 유지.*IoU 0\\\\.99"\n    ],\n    "explanation_only": false,\n    "aliases": [\n      "src/etl/qlt03_data_cleansing.py"\n    ]\n  },\n  "3": {\n    "patterns": [\n      "학습 400장 / 평가 300장",\n      "정확도 [\\\\d.]+%, 정밀도 [\\\\d.]+%, 재현율 [\\\\d.]+%, F1 [\\\\d.]+",\n      "면적 2% 미만 작은 결함 31장 재현율 \\\\d+%",\n      "유의미한 피처 7개",\n      "EDA 차트 8종 생성"\n    ],\n    "explanation_only": false,\n    "aliases": []\n  },\n  "4": {\n    "patterns": [\n      "파생 피처 24개",\n      "분할 원측정 MAPE [\\\\d.]+% -> 유형별 선형 보정 후 [\\\\d.]+%",\n      "깊이 추정.*MAPE [\\\\d.]+%, MAE [\\\\d.]+mm",\n      "면적만 쓰는 기준 MAPE [\\\\d.]+% -> 제안 모델 MAPE [\\\\d.]+%",\n      "제거 시 오차가 가장 크게 늘어난 변수군은"\n    ],\n    "explanation_only": true,\n    "aliases": [\n      "src/features/qlt03_feature_engineering.py"\n    ]\n  },\n  "5": {\n    "patterns": [\n      "MAE 개선 [\\\\d.]+%, RMSE 개선 [\\\\d.]+%.*R2 [\\\\d.]+",\n      "과소 판정\\\\(실제보다 낮게\\\\) \\\\d+장",\n      "신뢰도 80 이상 자동 통과 \\\\d+장\\\\(\\\\d+%\\\\): 일치율 [\\\\d.]+%"\n    ],\n    "explanation_only": false,\n    "aliases": [\n      "src/models/qlt03_advanced_model.py"\n    ]\n  },\n  "6": {\n    "patterns": [\n      "업로드 판정 결과\\\\(8장",\n      "평가 300장 중 기여도 상위 3개 요인 자동 산출 300건",\n      "손상 부위 표시와 정답 마스크의 IoU: U01 [\\\\d.]+",\n      "U06: 재촬영 요구"\n    ],\n    "explanation_only": false,\n    "aliases": [\n      "src/explainability/qlt03_xai_explainer.py"\n    ]\n  },\n  "7": {\n    "patterns": [\n      "평상 절감 [\\\\d.]+%",\n      "AI 판정 8건 전부 \'검사대기\' 등록, 자동 확정 0건",\n      "권한 우회 시도 10건 중 차단 10건",\n      "감사 로그 25건 해시 체인 무결성 100%",\n      "위변조 검출\\\\(4번\\\\)"\n    ],\n    "explanation_only": false,\n    "aliases": [\n      "src/optimization/qlt03_simulator.py"\n    ]\n  },\n  "8": {\n    "patterns": [\n      "진흙 혼입 비율별 판정.*비율당 120장",\n      "경보 정확도가 95% 미만으로 처음 떨어지는 진흙 비율: 5%",\n      "3000건 기준 로딩 [\\\\d.]+초",\n      "선형 외삽"\n    ],\n    "explanation_only": false,\n    "aliases": [\n      "src/dashboard/qlt03_app.py"\n    ]\n  },\n  "9": {\n    "patterns": [\n      "보류 구간 B 검증",\n      "ROI 3대 지표.*연 20000장",\n      "연간 비용 절감액 [\\\\d,]+만원.*판독 시간 단축률 [\\\\d.]+%.*투자 회수 기간 [\\\\d.]+개월",\n      "장수 민감도: 5000장"\n    ],\n    "explanation_only": false,\n    "aliases": []\n  },\n  "10": {\n    "patterns": [\n      "신규 240장 포함 840장 정합성 검증 7/7 통과",\n      "검증 80장.*챔피언 MAE [\\\\d.]+.*챌린저 MAE [\\\\d.]+",\n      "처리 구분:.*112.*86.*42.*자동 확정 0건",\n      "운영 문서 12개 항목 중 작성 완료 12개 \\\\(100%\\\\)"\n    ],\n    "explanation_only": false,\n    "aliases": [\n      "pipeline/qlt03_retrain.py"\n    ]\n  }\n}', 'tests/test_missions.py': 'from pathlib import Path\nimport sys\nimport pytest\nsys.path.insert(0,str(Path(__file__).resolve().parents[1]/"src"))\nfrom learning_support import check_baseline\n@pytest.mark.parametrize("mission",range(1,11))\ndef test_baseline_execution_evidence(mission):\n    check_baseline(mission)\n', 'requirements.txt': 'numpy==2.0.2\npandas==2.3.3\nscipy==1.13.1\nscikit-learn==1.6.1\nmatplotlib==3.9.4\npytest>=8,<9\n\n# 선택 앱 환경: requirements-dashboard.txt\n', 'docs/templates/m01_record.md': "# QLT03 미션 1 수행 기록 [교육용가상자료]\n\n원문 산출물: docs/qlt03_logistics_domain_spec.md 및 데이터 딕셔너리\n실습가이드 제출물: 데이터 사전, 라벨링 기준, 업무 프로세스 맵\n이 문서는 학습자가 작성하는 양식입니다. 빈 양식은 완료 산출물이 아닙니다.\n\n## 기본 실행\n- 실행 환경/일시: (작성)\n- 실행한 코드와 출력 파일: outputs/m01/baseline/stdout.txt\n- 확인할 결과(원문 예시):\n데이터 사전 피처 24개(목표 20개 이상), 정상 샘플 20건 검증 오류 0건\n예측 대상 4종(결함 면적, 깊이, 손상도, 정비 등급)\n- 직접 관찰한 지표/단위/분모: (작성)\n- 원문 예시와 차이 및 원문 Q1 달성 여부: (작성)\n- 표/그래프에서 발견한 근거: (작성)\n\n## 도전 과제\n데이터 사전에 '촬영 거리' 같은 필드를 하나 추가하고 실제 데이터와 대조해 누락으로 표시되는지 확인한다.\n- 변경 전/후 코드 또는 설명형 과제의 근거: (작성)\n- **확인할 것**: 데이터 딕셔너리 피처 수 · 정상 샘플 검증 통과·오류 건수 · photo.shot_distance_m 누락 메시지 · 검사관 보정·등급 승인 단계\n- 변경 전/후 결과와 단위: (작성)\n- 결과가 같으면 변경이 적용된 증거와 같은 이유: (작성)\n- **생각해 볼 질문**: 촬영 거리 필드를 스키마에만 추가하면 기존 정상 샘플의 검증 결과가 어떻게 바뀌는가? 가짜 측정값으로 누락을 채우는 대신 스키마와 수집 기록을 어떻게 일치시킬지 설명한다.\n- 내 답과 적용 한계: (작성)\n\n## AI 활용 및 최종 산출물\n- 사용한 AI/질문/응답 중 판단에 사용한 부분/직접 확인한 점: (작성, 미사용이면 미사용 이유)\n- 원문이 요구한 산출물 본문 또는 파일 연결: (작성)\n- 제출할 최종 경로·선택본(기본/도전)·선택 이유: (작성; 원문_제출경로_연결표.md 참조)\n- 최종 판단을 담당하는 가상 역할/사람 확인 절차: (작성)\n", 'docs/templates/m07_record.md': '# QLT03 미션 7 수행 기록 [교육용가상자료]\n\n원문 산출물: src/optimization/qlt03_simulator.py\n실습가이드 제출물: 최적화 코드, 비용 비교표, 민감도, 결재 이력, 차단 로그\n이 문서는 학습자가 작성하는 양식입니다. 빈 양식은 완료 산출물이 아닙니다.\n\n## 기본 실행\n- 실행 환경/일시: (작성)\n- 실행한 코드와 출력 파일: outputs/m07/baseline/stdout.txt\n- 확인할 결과(원문 예시):\n평상 절감 59.5%·저조도 28.2%(목표 15% 이상), 진흙 20%는 -7.9%로 미달\n자동 확정 0건, 권한 우회 차단 10/10(100%), 감사 로그 25건 위·변조 검출\n- 직접 관찰한 지표/단위/분모: (작성)\n- 원문 예시와 차이 및 원문 Q1 달성 여부: (작성)\n- 표/그래프에서 발견한 근거: (작성)\n\n## 도전 과제\n검사관 오류율 P_HUMAN을 8%에서 3%로 낮춰 관행 비용과 절감률이 어떻게 달라지는지 확인한다.\n- 변경 전/후 코드 또는 설명형 과제의 근거: (작성)\n- **확인할 것**: 관행·HITL 비용과 절감률 · 평상·저조도·진흙 시나리오 · 자동 확정 건수 · 권한 우회 차단과 감사 로그\n- 변경 전/후 결과와 단위: (작성)\n- 결과가 같으면 변경이 적용된 증거와 같은 이유: (작성)\n- **생각해 볼 질문**: 검사관 오류율을 8%에서 3%로 낮추면 비교 기준인 전수 정밀 판독 비용과 HITL 절감률은 어떻게 달라지는가? 진흙 시나리오의 미달을 숨기지 않고 어느 조건에서 경제성이 성립하는지 설명한다.\n- 내 답과 적용 한계: (작성)\n\n## AI 활용 및 최종 산출물\n- 사용한 AI/질문/응답 중 판단에 사용한 부분/직접 확인한 점: (작성, 미사용이면 미사용 이유)\n- 원문이 요구한 산출물 본문 또는 파일 연결: (작성)\n- 제출할 최종 경로·선택본(기본/도전)·선택 이유: (작성; 원문_제출경로_연결표.md 참조)\n- 최종 판단을 담당하는 가상 역할/사람 확인 절차: (작성)\n', 'docs/templates/m06_record.md': '# QLT03 미션 6 수행 기록 [교육용가상자료]\n\n원문 산출물: src/explainability/qlt03_xai_explainer.py 및 XAI 리포트\n실습가이드 제출물: 업로드 판정표, 시각화 그래프, 기여도 표, 자연어 설명\n이 문서는 학습자가 작성하는 양식입니다. 빈 양식은 완료 산출물이 아닙니다.\n\n## 기본 실행\n- 실행 환경/일시: (작성)\n- 실행한 코드와 출력 파일: outputs/m06/baseline/stdout.txt\n- 확인할 결과(원문 예시):\n기여도 상위 3개 자동 산출 300건 중 300건(100%)\n업로드 8장 중 유형 판정 7장 일치, 손상 부위 IoU 0.64~0.98, 저조도 사진은 재촬영 요구\n- 직접 관찰한 지표/단위/분모: (작성)\n- 원문 예시와 차이 및 원문 Q1 달성 여부: (작성)\n- 표/그래프에서 발견한 근거: (작성)\n\n## 도전 과제\n재촬영 기준 mud 0.03을 0.01로 바꿔 재촬영 요구 사진이 어떻게 달라지는지 확인한다.\n- 변경 전/후 코드 또는 설명형 과제의 근거: (작성)\n- **확인할 것**: U01~U08의 오염 추정치·처리 상태 · 재촬영 요구 사진 목록 · 기여도 상위 3개 산출 건수 · 표시 부위 IoU\n- 변경 전/후 결과와 단위: (작성)\n- 결과가 같으면 변경이 적용된 증거와 같은 이유: (작성)\n- **생각해 볼 질문**: mud 기준을 0.03에서 0.01로 낮추면 어떤 사진이 새로 재촬영 대상으로 바뀌는가? 바뀌지 않았다면 8장의 mud 값과 저조도 조건을 근거로 이유를 설명하고 판정 기여도의 부호를 해석한다.\n- 내 답과 적용 한계: (작성)\n\n## AI 활용 및 최종 산출물\n- 사용한 AI/질문/응답 중 판단에 사용한 부분/직접 확인한 점: (작성, 미사용이면 미사용 이유)\n- 원문이 요구한 산출물 본문 또는 파일 연결: (작성)\n- 제출할 최종 경로·선택본(기본/도전)·선택 이유: (작성; 원문_제출경로_연결표.md 참조)\n- 최종 판단을 담당하는 가상 역할/사람 확인 절차: (작성)\n', 'docs/templates/m03_record.md': '# QLT03 미션 3 수행 기록 [교육용가상자료]\n\n원문 산출물: notebooks/qlt03_eda_analysis.ipynb 및 EDA 요약 리포트\n실습가이드 제출물: Baseline 성능표, EDA 차트 8종, 상관표, 제원 불일치 표, 파레토 결과\n이 문서는 학습자가 작성하는 양식입니다. 빈 양식은 완료 산출물이 아닙니다.\n\n## 기본 실행\n- 실행 환경/일시: (작성)\n- 실행한 코드와 출력 파일: outputs/m03/baseline/stdout.txt\n- 확인할 결과(원문 예시):\nBaseline 정확도 86.0%·재현율 92.2%, 면적 2% 미만 작은 결함 재현율 74%\n유의 피처 7개(목표 5개 이상), EDA 차트 8종, 제원 불일치 후보 47대\n- 직접 관찰한 지표/단위/분모: (작성)\n- 원문 예시와 차이 및 원문 Q1 달성 여부: (작성)\n- 표/그래프에서 발견한 근거: (작성)\n\n## 도전 과제\nBaseline 로지스틱 회귀의 규제 강도 C를 0.5에서 5로 바꿔 성능 변화를 확인하고 한계가 해소되는지 설명한다.\n- 변경 전/후 코드 또는 설명형 과제의 근거: (작성)\n- **확인할 것**: 정확도·정밀도·재현율·F1 · 작은 결함 재현율 · 저조도와 밝은 사진 정확도 · 제원 불일치 후보 수\n- 변경 전/후 결과와 단위: (작성)\n- 결과가 같으면 변경이 적용된 증거와 같은 이유: (작성)\n- **생각해 볼 질문**: C를 0.5에서 5로 바꿨을 때 전체 성능과 작은 결함·저조도 성능은 함께 좋아졌는가? C가 커질수록 규제가 약해지는 점과 위치·모양 정보를 쓰지 않는 입력의 한계를 연결해 설명한다.\n- 내 답과 적용 한계: (작성)\n\n## AI 활용 및 최종 산출물\n- 사용한 AI/질문/응답 중 판단에 사용한 부분/직접 확인한 점: (작성, 미사용이면 미사용 이유)\n- 원문이 요구한 산출물 본문 또는 파일 연결: (작성)\n- 제출할 최종 경로·선택본(기본/도전)·선택 이유: (작성; 원문_제출경로_연결표.md 참조)\n- 최종 판단을 담당하는 가상 역할/사람 확인 절차: (작성)\n', 'docs/templates/m08_record.md': "# QLT03 미션 8 수행 기록 [교육용가상자료]\n\n원문 산출물: src/dashboard/qlt03_app.py (Streamlit/Dash 등)\n실습가이드 제출물: 진흙 스윕 표, 복합 열화 표, 처리 시간·로딩 시간, 한계치 요약\n이 문서는 학습자가 작성하는 양식입니다. 빈 양식은 완료 산출물이 아닙니다.\n\n## 기본 실행\n- 실행 환경/일시: (작성)\n- 실행 경로(앱/시뮬레이터): (작성)\n- 앱 증거: 내려받은 qlt03_m08_app_stdout.txt·qlt03_m08_sweep.csv와 화면 캡처 경로: (작성)\n- 시뮬레이터 증거: outputs/m08/baseline/stdout.txt 및 figure 파일: (작성)\n- 실제 사용한 경로의 증거를 연결한다. 앱 로그 파일을 baseline 로그로 가장하지 않는다.\n- 확인할 결과(원문 예시):\n진흙 0%: 자동 처리 75%·경보 정확도 100%, 5%부터 93%로 95% 미달\n시뮬레이터의 요약 집계·목록·차트 렌더링 예시 시간은 0.1초 안팎. 브라우저를 포함한 웹앱 응답은 별도 측정하며 이 값으로 2초 목표 달성을 판정하지 않음, 흐림 사진 자동 처리 경보 정확도 91%\n- 직접 관찰한 지표/단위/분모: (작성)\n- 원문 예시와 차이 및 원문 Q1 달성 여부: (작성)\n- 표/그래프에서 발견한 근거: (작성)\n\n## 도전 과제\n경보를 등급 2 이상에서 등급 3(3단계 정비)만으로 바꿔 경보 정확도와 자동 처리 비율이 어떻게 달라지는지 확인한다.\n- 변경 전/후 코드 또는 설명형 과제의 근거: (작성)\n- **확인할 것**: 진흙 비율별 경보 정확도 · 품질 게이트 통과 자동 처리 비율 · 흐림·잡음·저조도 복합 조건 · 처리시간 실측과 대량 처리 외삽\n- 변경 전/후 결과와 단위: (작성)\n- 결과가 같으면 변경이 적용된 증거와 같은 이유: (작성)\n- **생각해 볼 질문**: 경보 정의만 등급 3으로 바꿨을 때 자동 처리 비율이 그대로인 이유는 무엇인가? 게이트 통과 사진을 분모로 삼아 경보 정확도를 비교하고 자동 처리 0건의 '-'를 100%로 해석할 수 없는 이유를 적는다.\n- 내 답과 적용 한계: (작성)\n\n## AI 활용 및 최종 산출물\n- 사용한 AI/질문/응답 중 판단에 사용한 부분/직접 확인한 점: (작성, 미사용이면 미사용 이유)\n- 원문이 요구한 산출물 본문 또는 파일 연결: (작성)\n- 제출할 최종 경로·선택본(기본/도전)·선택 이유: (작성; 원문_제출경로_연결표.md 참조)\n- 최종 판단을 담당하는 가상 역할/사람 확인 절차: (작성)\n", 'docs/templates/m05_record.md': '# QLT03 미션 5 수행 기록 [교육용가상자료]\n\n원문 산출물: src/models/qlt03_advanced_model.py\n실습가이드 제출물: 성능 비교표, 혼동행렬, 신뢰도 보정표, 상황별 성능표\n이 문서는 학습자가 작성하는 양식입니다. 빈 양식은 완료 산출물이 아닙니다.\n\n## 기본 실행\n- 실행 환경/일시: (작성)\n- 실행한 코드와 출력 파일: outputs/m05/baseline/stdout.txt\n- 확인할 결과(원문 예시):\nMAE 개선 73.4%, RMSE 개선 55.1%(목표 20% 이상), R2 0.946(목표 0.85 이상)\n등급 일치율 93.3%, 신뢰도 80 이상 234장(78%) 일치율 100%, 진흙 가림 사진은 80 이상 0%\n- 직접 관찰한 지표/단위/분모: (작성)\n- 원문 예시와 차이 및 원문 Q1 달성 여부: (작성)\n- 표/그래프에서 발견한 근거: (작성)\n\n## 도전 과제\n자동 통과 기준 신뢰도를 80에서 70으로 낮추면 자동 통과 비율과 일치율이 어떻게 달라지는지 확인한다.\n- 변경 전/후 코드 또는 설명형 과제의 근거: (작성)\n- **확인할 것**: 자동 통과 후보 수·비율·실제 일치율 · 저조도·진흙 조건의 문턱 이상 비율 · 등급 혼동행렬의 과소 판정 수 · MAE·RMSE·R2\n- 변경 전/후 결과와 단위: (작성)\n- 결과가 같으면 변경이 적용된 증거와 같은 이유: (작성)\n- **생각해 볼 질문**: 신뢰도 80→70 하향 후 후보 비율과 실제 일치율은 어떻게 달라졌는가? 혼동행렬과 진흙 조건을 근거로 범위를 넓히는 대가를 설명하고 자동 통과가 최종 정비 등급 확정을 뜻하는지 구분한다.\n- 내 답과 적용 한계: (작성)\n\n## AI 활용 및 최종 산출물\n- 사용한 AI/질문/응답 중 판단에 사용한 부분/직접 확인한 점: (작성, 미사용이면 미사용 이유)\n- 원문이 요구한 산출물 본문 또는 파일 연결: (작성)\n- 제출할 최종 경로·선택본(기본/도전)·선택 이유: (작성; 원문_제출경로_연결표.md 참조)\n- 최종 판단을 담당하는 가상 역할/사람 확인 절차: (작성)\n', 'docs/templates/m02_record.md': '# QLT03 미션 2 수행 기록 [교육용가상자료]\n\n원문 산출물: src/etl/qlt03_data_cleansing.py 및 정제 품질 보고서\n실습가이드 제출물: 정제 모듈 코드, 정제 규칙 로그, 정합성 검증 결과, 전후 통계 비교표\n이 문서는 학습자가 작성하는 양식입니다. 빈 양식은 완료 산출물이 아닙니다.\n\n## 기본 실행\n- 실행 환경/일시: (작성)\n- 실행한 코드와 출력 파일: outputs/m02/baseline/stdout.txt\n- 확인할 결과(원문 예시):\n정제 규칙 11종(목표 10종 이상), 정합성 검증 12/12 통과(100%)\n정제 후 사진 150장, 긴급 점검 6장 유지, 정제 박스 평균 IoU 0.99\n- 직접 관찰한 지표/단위/분모: (작성)\n- 원문 예시와 차이 및 원문 Q1 달성 여부: (작성)\n- 표/그래프에서 발견한 근거: (작성)\n\n## 도전 과제\nR7 조명 보정 기준(평균 밝기 0.3)을 0.4로 바꿔 보정 대상 사진 수가 어떻게 달라지는지 확인한다.\n- 변경 전/후 코드 또는 설명형 과제의 근거: (작성)\n- **확인할 것**: R7 문턱 미만 사진 수 · 정합성 12항목 · 정제 후 사진·긴급 점검 보존 수 · 전후 평균 밝기와 박스 IoU\n- 변경 전/후 결과와 단위: (작성)\n- 결과가 같으면 변경이 적용된 증거와 같은 이유: (작성)\n- **생각해 볼 질문**: 밝기 0.3→0.4 변경 후 R7의 집계 건수와 정제 후 평균 밝기는 각각 어떻게 달라졌는가? 현재 구현이 모든 사진을 정규화한다는 점을 코드에서 찾아 집계 문턱과 실제 적용 조건을 구분한다.\n- 내 답과 적용 한계: (작성)\n\n## AI 활용 및 최종 산출물\n- 사용한 AI/질문/응답 중 판단에 사용한 부분/직접 확인한 점: (작성, 미사용이면 미사용 이유)\n- 원문이 요구한 산출물 본문 또는 파일 연결: (작성)\n- 제출할 최종 경로·선택본(기본/도전)·선택 이유: (작성; 원문_제출경로_연결표.md 참조)\n- 최종 판단을 담당하는 가상 역할/사람 확인 절차: (작성)\n', 'docs/templates/m09_record.md': '# QLT03 미션 9 수행 기록 [교육용가상자료]\n\n원문 산출물: reports/qlt03_roi_analysis.md\n실습가이드 제출물: 재보정 코드, 원인별 오판표, 전후 비교표, 3대 지표와 민감도\n이 문서는 학습자가 작성하는 양식입니다. 빈 양식은 완료 산출물이 아닙니다.\n\n## 기본 실행\n- 실행 환경/일시: (작성)\n- 원문 재현 이력: outputs/m09/baseline/stdout.txt (최종 성능·경제성 기준 아님)\n- 수정 기본 실행: outputs/m09/revised/stdout.txt (최종 보고·도전 비교 기준)\n- 확인할 결과(실습 코드 채점 기준):\n실습 코드 채점 기준: 연간 순절감액 2,751만원, 판독 시간 단축률 0.3%, 투자 회수 10.9개월.\n반올림 전 2,751.111111만원/년·0.292475%·10.904685개월. 기존 AI→개선 AI, 동일 B 360장, 연 20,000장, 시급 1.2만원, 증분 운영비 150만원/년, 증분 투자비 2,500만원이라는 교육용 가정이다. outputs/m09/revised/stdout.txt와 전후 총비용 차이를 확인한다. 환경에 따른 차이는 실제 로그와 산식으로 설명하며 예시 숫자를 맞추려고 출력을 고치지 않는다.\n- 직접 관찰한 지표/단위/분모: (작성)\n- 실습 코드 기준과 차이 및 원문 Q1 달성 여부: (작성)\n- 표/그래프에서 발견한 근거: (작성)\n\n## 도전 과제\n연 판정 사진 수 N_YR을 20000장에서 10000장으로 바꿔 절감액과 회수 기간이 어떻게 달라지는지 확인한다.\n- 변경 전/후 코드 또는 설명형 과제의 근거: (작성)\n- **확인할 것**: 보류 구간 B의 등급 일치율·오판 손실 · 연간 비용 절감액과 고정 운영비 · 판독 시간 단축률 · 투자 회수 기간\n- 변경 전/후 결과와 단위: (작성)\n- 결과가 같으면 변경이 적용된 증거와 같은 이유: (작성)\n- **생각해 볼 질문**: 연 사진 수 20000→10000 변경 후 연 절감액이 정확히 절반이 되는가? 고정 운영비 150만원을 포함한 산식으로 확인하고 보류 구간의 모델 성능과 시간 단축률이 유지되는 이유를 설명한다.\n- 내 답과 적용 한계: (작성)\n\n## AI 활용 및 최종 산출물\n- 사용한 AI/질문/응답 중 판단에 사용한 부분/직접 확인한 점: (작성, 미사용이면 미사용 이유)\n- 원문이 요구한 산출물 본문 또는 파일 연결: (작성)\n- 제출할 최종 경로·선택본(기본/도전)·선택 이유: (작성; 원문_제출경로_연결표.md 참조)\n- 최종 판단을 담당하는 가상 역할/사람 확인 절차: (작성)\n', 'docs/templates/m04_record.md': '# QLT03 미션 4 수행 기록 [교육용가상자료]\n\n원문 산출물: src/features/qlt03_feature_engineering.py 및 기준 모델\n실습가이드 제출물: 분할 성능표, 면적 보정 결과, 깊이·손상도 오차표, 변수군 제거 결과\n이 문서는 학습자가 작성하는 양식입니다. 빈 양식은 완료 산출물이 아닙니다.\n\n## 기본 실행\n- 실행 환경/일시: (작성)\n- 실행한 코드와 출력 파일: outputs/m04/baseline/stdout.txt\n- 확인할 결과(원문 예시):\n분할 IoU 균열 0.89·부식 0.74·파손 0.61·누유 0.94, 면적 MAPE 20.8% -> 10.2%\n손상도 MAPE 8.8%(목표 25% 이하, 기준 30.8%), 깊이 MAPE 28.0%\n- 직접 관찰한 지표/단위/분모: (작성)\n- 원문 예시와 차이 및 원문 Q1 달성 여부: (작성)\n- 표/그래프에서 발견한 근거: (작성)\n\n## 도전 과제\n깊이 추정 MAPE가 손상도보다 큰 이유를 사진 한 장으로는 깊이가 직접 보이지 않는 점과 연결해 설명한다.\n- 변경 전/후 코드 또는 설명형 과제의 근거: (작성)\n- **확인할 것**: 유형별 IoU·Dice · 면적 원측정·보정 MAPE · 깊이 MAPE와 MAE · 손상도 기준·제안 MAPE · 변수군 제거 시 오차 증가\n- 변경 전/후 결과와 단위: (작성)\n- 결과가 같으면 변경이 적용된 증거와 같은 이유: (작성)\n- **생각해 볼 질문**: 기본 실행에서 깊이 MAPE와 손상도 MAPE를 각각 인용하고, 한 장의 사진으로 깊이를 직접 관측할 수 없다는 점을 들어 차이를 설명한다. 임계값 분할과 실제 Mask R-CNN 학습을 구분하고 깊이 게이지 등 추가 확인 방법을 한 가지 제시한다.\n- 내 답과 적용 한계: (작성)\n\n## AI 활용 및 최종 산출물\n- 사용한 AI/질문/응답 중 판단에 사용한 부분/직접 확인한 점: (작성, 미사용이면 미사용 이유)\n- 원문이 요구한 산출물 본문 또는 파일 연결: (작성)\n- 제출할 최종 경로·선택본(기본/도전)·선택 이유: (작성; 원문_제출경로_연결표.md 참조)\n- 최종 판단을 담당하는 가상 역할/사람 확인 절차: (작성)\n', 'docs/templates/m10_record.md': '# QLT03 미션 10 수행 기록 [교육용가상자료]\n\n원문 산출물: docs/qlt03_operations_guide.md 및 pipeline/qlt03_retrain.py\n실습가이드 제출물: 통합 파이프라인 코드, 처리 구분 표, 자체 점검표, 인터페이스 명세, 브리핑 요약\n이 문서는 학습자가 작성하는 양식입니다. 빈 양식은 완료 산출물이 아닙니다.\n\n## 기본 실행\n- 실행 환경/일시: (작성)\n- 원문 재현 이력: outputs/m10/baseline/stdout.txt (최종 성능·경제성 기준 아님)\n- 수정 기본 실행: outputs/m10/revised/stdout.txt (최종 보고·도전 비교 기준)\n- 확인할 결과(실습 코드 채점 기준):\n실습 코드 최종시험 MAE 6.43, 등급 정확도 87.5%(시험 240장). 빈 운영지침은 0/12이며 문서 미완료 점검은 FAIL이 정상이다. 직접 작성 후 완료 수와 자체 점검 결과를 다시 기록한다.\n처리 구분 간이 승인 후보 112·정밀 판독 86·재촬영 42장, 자동 확정 0건\n- 직접 관찰한 지표/단위/분모: (작성)\n- 원문 예시와 차이 및 원문 Q1 달성 여부: (작성)\n- 표/그래프에서 발견한 근거: (작성)\n\n## 도전 과제\n코드의 승격 기준 0.03(MAE 3% 개선)을 0.15로 높여 승격 판정이 어떻게 달라지는지 확인한다.\n- 변경 전/후 코드 또는 설명형 과제의 근거: (작성)\n- **확인할 것**: 챔피언·챌린저 검증 MAE와 개선율 · 승격 판정 · 간이 승인 후보·정밀 판독·재촬영 수 · 자동 확정 수·자체 점검·문서 완비\n- 변경 전/후 결과와 단위: (작성)\n- 결과가 같으면 변경이 적용된 증거와 같은 이유: (작성)\n- **생각해 볼 질문**: 검증 MAE로 계산한 개선율은 15%를 넘는가? 문턱 변경 후 승격 판정과 처리 구분이 같더라도 오류라고 할 수 없는 이유를 적고 세 처리 구분이 신규 240장과 맞는지 검산한다.\n- 내 답과 적용 한계: (작성)\n\n## AI 활용 및 최종 산출물\n- 사용한 AI/질문/응답 중 판단에 사용한 부분/직접 확인한 점: (작성, 미사용이면 미사용 이유)\n- 원문이 요구한 산출물 본문 또는 파일 연결: (작성)\n- 제출할 최종 경로·선택본(기본/도전)·선택 이유: (작성; 원문_제출경로_연결표.md 참조)\n- 최종 판단을 담당하는 가상 역할/사람 확인 절차: (작성)\n', 'notebooks/qlt03_eda_analysis.ipynb': '{\n  "cells": [\n    {\n      "cell_type": "markdown",\n      "metadata": {},\n      "source": "## 운영자 준비 사항\\n\\n운영자 준비 사항: Python 3.12와 표준 Jupyter를 준비합니다. pandas, numpy, matplotlib, scipy, scikit-learn, pytest와 Jupyter의 IPython을 미리 준비합니다. streamlit은 미션 8의 웹앱을 사용할 때만 필요합니다. 인터넷 연결은 필요하지 않습니다. 교육장에서 제공하는 생성형 AI를 사용할 수 있도록 준비합니다.",\n      "id": "p1b-054c4610b74c1c6a"\n    },\n    {\n      "cell_type": "markdown",\n      "metadata": {},\n      "source": "# QLT03 미션 3 EDA 실행 예제 [교육용가상자료]\\n\\n기본 실행 예제입니다. 학습자 해석과 도전 과제는 assignment 노트북 또는 미션 3 기록지에 작성하세요.\\n\\n보안·저장·제출 방법은 바로 아래 핵심 안내에 있습니다.\\n\\n### 링크가 열리지 않으면(노트북만 따로 연 경우) 핵심 안내\\n\\n- 보안: 제공된 교육용 가상자료만 사용합니다. 비공개 업무자료, 실습에는 제공된 가상자료만 사용하고, 실제 이름·연락처 같은 개인정보는 AI 도구나 노트북에 넣지 않습니다.\\n- 공용 PC: 종료 전 허용된 저장 위치에 저장하고 다시 열어 확인한 뒤 사용한 교육장 계정에서 로그아웃합니다. 공용 PC에 비밀번호를 저장하지 않으며 로컬 사본은 조직 규정에 따라 정리합니다.\\n- 저장: 노트북 사본과 reports·직접 작성한 docs·work·outputs를 원래 폴더 구조대로 저장합니다.\\n- 제출: 제출처·마감 일시·시간대는 miliai.kr 해당 과제 공지를 확인하고, 표시가 없으면 제출 전에 교육 담당자에게 확인합니다. 별도 공지가 없으면 미션마다 그 미션 산출물을 ZIP 하나로 묶어 해당 미션 제출란에 첨부합니다.\\n- 파일명: 파일명은 과제ID_미션번호_산출물명으로 작성하고 실제 형식의 확장자를 붙입니다. 실명·개인 식별번호는 파일명에 넣지 않습니다.",\n      "id": "p1b-481570f40712ed01"\n    },\n    {\n      "cell_type": "code",\n      "execution_count": null,\n      "id": "e9e4d810535d",\n      "metadata": {},\n      "outputs": [],\n      "source": "import sys, subprocess\\nimport importlib.metadata as metadata\\nrequired = [\'numpy==2.0.2\', \'pandas==2.3.3\', \'scipy==1.13.1\', \'scikit-learn==1.6.1\', \'matplotlib==3.9.4\']\\nmissing = []\\nfor spec in required:\\n    name, version = spec.split(\\"==\\")\\n    try:\\n        if metadata.version(name) != version: missing.append(spec)\\n    except metadata.PackageNotFoundError: missing.append(spec)\\nif missing: raise RuntimeError(\\"필요 패키지: \\" + \\", \\".join(missing) + \\". 담당자에게 준비된 환경을 요청하세요. 자동 설치하지 않습니다.\\")\\nprint(\\"준비된 환경의 패키지 버전 확인 완료\\")"\n    },\n    {\n      "cell_type": "code",\n      "execution_count": null,\n      "id": "7190ac4e8e5f",\n      "metadata": {},\n      "outputs": [],\n      "source": "\\"\\"\\"밝기 히스토그램과 평균 색만 쓰는 외관 결함 유무 1차 판별 Baseline 비전 모델(로지스틱 회귀)을 구현해 성능과 한계(작은 균열·조명 의존)를 확인하고, 탐색적 분석(EDA) 차트 8종·상관분석(p-value)·정비 공수 파레토(ABC)·제원 기록과의 불일치 분석으로 핵심 관리 대상을 찾는다.\\"\\"\\"\\nimport numpy as np\\nimport pandas as pd\\nimport matplotlib.pyplot as plt\\nfrom scipy import stats\\nfrom sklearn.linear_model import LogisticRegression\\nfrom sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score\\nfrom sklearn.preprocessing import StandardScaler\\nfrom sklearn.pipeline import make_pipeline\\n\\n# ---------------------------------------------------------------------------\\n# [교육용가상자료] 장비 외관 사진 생성기 (실제 장비·조직·위치 사진이 아니라 난수로 그린 64x64 합성 이미지)\\n# 도장된 금속 표면에 균열(crack)·부식(corrosion)·파손(damage)·누유(leak)를 그리고, 조명 편차와 진흙(mud) 오염을 섞는다.\\n# 이미지와 함께 장비 관리 기록(제원 요약)과 정답 결함 마스크·손상도·정비 등급을 만든다.\\n# ---------------------------------------------------------------------------\\nfrom scipy import ndimage as ndi\\n\\nSZ = 64\\nCLASSES = [\\"none\\", \\"crack\\", \\"corrosion\\", \\"damage\\", \\"leak\\"]\\nKO_CLS = {\\"none\\": \\"결함 없음\\", \\"crack\\": \\"균열\\", \\"corrosion\\": \\"부식\\", \\"damage\\": \\"파손\\", \\"leak\\": \\"누유\\"}\\nCLS_W = {\\"none\\": 0.0, \\"crack\\": 1.4, \\"corrosion\\": 1.0, \\"damage\\": 1.6, \\"leak\\": 0.6}      # 손상도 계산용 결함 유형 가중치\\nGRADE_NAME = {1: \\"1단계 정비\\", 2: \\"2단계 정비\\", 3: \\"3단계 정비\\"}\\nYY, XX = np.mgrid[0:SZ, 0:SZ]\\n\\n\\ndef grade_of(idx):\\n    return np.where(np.asarray(idx) >= 65, 3, np.where(np.asarray(idx) >= 35, 2, 1))\\n\\n\\ndef blob(rng, cy, cx, r, rough=0.35):\\n    \\"\\"\\"불규칙한 원형 영역 마스크.\\"\\"\\"\\n    noise = ndi.gaussian_filter(rng.normal(size=(SZ, SZ)), 3)\\n    noise = noise / (np.abs(noise).max() + 1e-9)\\n    return (np.hypot(YY - cy, XX - cx) + rough * r * 2.2 * noise) < r\\n\\n\\ndef draw(rng, cls, sev):\\n    \\"\\"\\"결함 유형·심각도 sev(0~1)에 맞는 정답 마스크와 색 정보를 만든다.\\"\\"\\"\\n    m = np.zeros((SZ, SZ), bool)\\n    if cls == \\"crack\\":\\n        y, x, ang = rng.uniform(12, 52), rng.uniform(12, 52), rng.uniform(0, np.pi)\\n        for _ in range(int(20 + 40 * sev)):\\n            ang += rng.normal(0, 0.35); y += np.sin(ang) * 1.2; x += np.cos(ang) * 1.2\\n            if 0 <= int(y) < SZ and 0 <= int(x) < SZ:\\n                m[int(y), int(x)] = True\\n        m = ndi.binary_dilation(m, iterations=1 + int(sev > 0.6))\\n    elif cls == \\"corrosion\\":\\n        for _ in range(int(rng.integers(1, 4))):\\n            m |= blob(rng, rng.uniform(14, 50), rng.uniform(14, 50), 3 + 9 * sev * rng.uniform(0.6, 1.1))\\n    elif cls == \\"damage\\":\\n        m = blob(rng, rng.uniform(20, 44), rng.uniform(20, 44), 6 + 12 * sev, 0.3)\\n    elif cls == \\"leak\\":\\n        x0, y0 = rng.uniform(16, 48), rng.uniform(8, 24)\\n        w, ln = 2 + 4 * sev, 12 + 26 * sev\\n        m = (np.abs(XX - x0 - 0.15 * (YY - y0)) < w * (1 - 0.4 * (YY - y0) / ln)) & (YY > y0) & (YY < y0 + ln)\\n    return m\\n\\n\\ndef make_images(n=600, seed=0, mud=0.0, p_none=0.25, dark=False):\\n    \\"\\"\\"가상 외관 사진 n장(+진흙 오염 mud 비율)과 관리 기록·정답. 반환: meta(DataFrame), imgs(n,64,64,3), masks(n,64,64).\\"\\"\\"\\n    rng = np.random.default_rng(seed)\\n    imgs, masks, rows = np.zeros((n, SZ, SZ, 3), np.float32), np.zeros((n, SZ, SZ), bool), []\\n    for i in range(n):\\n        cls = \\"none\\" if rng.random() < p_none else str(rng.choice(CLASSES[1:], p=[0.25, 0.3, 0.25, 0.2]))\\n        sev = 0.0 if cls == \\"none\\" else float(rng.beta(1.6, 2.0))\\n        base = np.array([0.42, 0.46, 0.40]) * (1 + 0.5 * ndi.gaussian_filter(rng.normal(size=(SZ, SZ)), 8)[..., None] * 3)\\n        img = base + rng.normal(0, 0.012, (SZ, SZ, 3))\\n        m = draw(rng, cls, sev)\\n        a = ndi.gaussian_filter(m.astype(float), 0.8)[..., None]                        # 경계를 부드럽게 섞는다\\n        if cls == \\"crack\\":\\n            col = img * rng.uniform(0.2, 0.45)\\n        elif cls == \\"corrosion\\":\\n            col = np.array([0.55, 0.30, 0.12]) * rng.uniform(0.75, 1.15) * (0.85 + 0.3 * rng.random((SZ, SZ, 1)))\\n        elif cls == \\"damage\\":\\n            col = np.ones_like(img) * np.array([0.34, 0.34, 0.36]) * (1.0 - 0.45 * sev) * rng.uniform(0.75, 1.0)\\n            rim = m & ~ndi.binary_erosion(m, iterations=2)\\n            a = np.maximum(a, 0.0); img = np.where(rim[..., None], 0.7, img)\\n        elif cls == \\"leak\\":\\n            col = np.ones_like(img) * np.array([0.10, 0.09, 0.08]) * rng.uniform(0.8, 2.6)\\n        else:\\n            col = img\\n        img = img * (1 - a) + col * a\\n        if rng.random() < 0.35:                                   # 무해한 얼룩(결함 아님): 옅은 갈색 번짐\\n            st = blob(rng, rng.uniform(5, 58), rng.uniform(5, 58), rng.uniform(6, 14), 0.5)\\n            al = ndi.gaussian_filter(st.astype(float), 2.5)[..., None] * rng.uniform(0.15, 0.4)\\n            img = img * (1 - al) + np.array([0.50, 0.40, 0.30]) * al\\n        if rng.random() < 0.3:                                    # 무해한 긴 스크래치(밝은 선)\\n            y0, x0, ang = rng.uniform(5, 58), rng.uniform(5, 58), rng.uniform(0, np.pi)\\n            for t in np.linspace(0, 25, 40):\\n                yy, xx = int(y0 + t * np.sin(ang)), int(x0 + t * np.cos(ang))\\n                if 0 <= yy < SZ and 0 <= xx < SZ:\\n                    img[yy, xx] = np.clip(img[yy, xx] * 1.5, 0, 1)\\n        if mud > 0:                                           # 진흙 오염: 결함 위를 덮을 수 있다(정답 마스크는 그대로)\\n            mm = np.zeros((SZ, SZ), bool)\\n            while mm.mean() < mud:\\n                mm |= blob(rng, rng.uniform(0, SZ), rng.uniform(0, SZ), rng.uniform(5, 11), 0.5)\\n            img[mm] = np.array([0.38, 0.30, 0.22]) * (0.85 + 0.3 * rng.random((mm.sum(), 1)))\\n        gain = rng.uniform(0.3, 0.55) if dark else rng.uniform(0.5, 1.15)\\n        ang = rng.uniform(0, 2 * np.pi)\\n        light = gain * (1 + 0.3 * (np.cos(ang) * (XX - 32) + np.sin(ang) * (YY - 32)) / 32)\\n        imgs[i] = np.clip(img * light[..., None] + rng.normal(0, 0.025, (SZ, SZ, 3)), 0, 1)\\n        masks[i] = m\\n        area = float(m.mean())\\n        depth = {\\"none\\": 0.0, \\"crack\\": 0.3 + 3.0 * sev, \\"corrosion\\": 0.1 + 1.5 * sev, \\"damage\\": 1.0 + 8.0 * sev, \\"leak\\": 0.0}[cls] * rng.lognormal(0, 0.1)\\n        idx = float(np.clip(100 * (1 - np.exp(-(CLS_W[cls] * area * 9 + 0.05 * depth))) + rng.normal(0, 1.5), 0, 100))\\n        rows.append({\\"img_id\\": f\\"IMG{seed:02d}{i:04d}\\", \\"cls\\": cls, \\"sev\\": sev, \\"area_frac\\": area, \\"depth_mm\\": depth, \\"damage_idx\\": idx,\\n                     \\"grade\\": int(grade_of(idx)), \\"gain\\": gain, \\"mud\\": mud})\\n    meta = pd.DataFrame(rows)\\n    # 장비 관리 기록(제원 요약): 사진 판정과 대조할 값\\n    age = rng.integers(1, 25, n)\\n    meta[\\"day\\"] = np.sort(rng.integers(0, 90, n))\\n    meta[\\"equip_type\\"] = rng.choice([\\"vehicle\\", \\"generator\\", \\"comm\\", \\"trailer\\"], n, p=[0.4, 0.2, 0.2, 0.2])\\n    meta[\\"mfg_year\\"] = 2026 - age\\n    meta[\\"op_hours\\"] = np.clip(rng.normal(180 * age, 400, n) + 8 * meta[\\"damage_idx\\"], 10, 20000)\\n    meta[\\"env\\"] = rng.choice([\\"coastal\\", \\"inland\\"], n, p=[0.3, 0.7])\\n    reg = np.clip(meta[\\"grade\\"] + rng.choice([-1, 0, 0, 0, 0, 0, 1, 2], n), 1, 3)           # 장비 관리 기록상 직전 정비 등급\\n    meta[\\"reg_grade\\"] = reg\\n    meta[\\"emergency\\"] = rng.random(n) < 0.04                                               # 긴급 점검 요청(화질 불량이어도 보존)\\n    return meta, imgs, masks\\n\\n# ---- 전처리(조명 보정·표준화)와 결함 후보 분할, 특징 추출 (cv2 대신 numpy·scipy 사용) ----\\ndef correct(img):\\n    \\"\\"\\"조명 보정: 밝기에 평면(조명 기울기)을 적합해 나눈다. 어두운 결함 화소는 제외하고 두 번 반복 적합한다. 반환: 보정 밝기(정상 면 약 1), 색 비율.\\"\\"\\"\\n    g = img.mean(axis=2)\\n    A = np.c_[np.ones(SZ * SZ), XX.ravel() / SZ, YY.ravel() / SZ]\\n    keep = np.ones(SZ * SZ, bool)\\n    for _ in range(3):\\n        coef, *_ = np.linalg.lstsq(A[keep], g.ravel()[keep], rcond=None)\\n        plane = (A @ coef).reshape(SZ, SZ)\\n        keep = (np.abs(g - plane) < 0.08 * np.maximum(plane, 1e-3)).ravel()\\n        if keep.sum() < 200:\\n            keep[:] = True\\n    c = g / np.maximum(plane, 1e-3)\\n    chroma = img[..., 0] / (img.sum(axis=2) + 1e-6)\\n    return c, chroma\\n\\n\\nSEGP = {\\"dark\\": 0.7, \\"rust\\": 0.5, \\"rim\\": 1.45}            # 분할 임계값(어두움 보정 밝기, 붉은 비율, 밝은 테두리): 재보정 대상\\n\\n\\ndef segment(img):\\n    \\"\\"\\"결함 후보 마스크: 어두운 영역(균열·누유·파손) + 붉은 갈색 영역(부식) + 밝은 테두리(파손).\\"\\"\\"\\n    c, chroma = correct(img)\\n    dark = c < SEGP[\\"dark\\"]\\n    rust = (chroma > SEGP[\\"rust\\"]) & (c > 0.45)\\n    rim = c > SEGP[\\"rim\\"]\\n    m = dark | rust | ndi.binary_dilation(rim, iterations=1) & ndi.binary_dilation(dark, iterations=3)\\n    m = ndi.binary_opening(m, structure=np.ones((2, 2))) if m.sum() > 12 else m\\n    return m, c, chroma, dark, rust, rim\\n\\n\\ndef extract(img):\\n    \\"\\"\\"분할 결과와 색·모양에서 24개 특징을 만든다.\\"\\"\\"\\n    m, c, chroma, dark, rust, rim = segment(img)\\n    lab, n = ndi.label(m)\\n    area = float(m.sum())\\n    if n:\\n        sizes = ndi.sum(m, lab, range(1, n + 1))\\n        k = int(np.argmax(sizes)) + 1\\n        sl = ndi.find_objects(lab)[k - 1]\\n        h, w = sl[0].stop - sl[0].start, sl[1].stop - sl[1].start\\n        big = float(sizes.max())\\n        elong, fill = max(h, w) / max(min(h, w), 1), big / (h * w)\\n    else:\\n        big, elong, fill = 0.0, 1.0, 0.0\\n    gy, gx = np.gradient(c)\\n    mag = np.hypot(gx, gy)\\n    inside = m.sum() > 0\\n    return {\\"area_px\\": area, \\"area_frac\\": area / SZ ** 2, \\"n_comp\\": float(n), \\"big_px\\": big, \\"elong\\": elong, \\"fill\\": fill,\\n            \\"dark_frac\\": float(dark.mean()), \\"rust_frac\\": float(rust.mean()), \\"rim_frac\\": float(rim.mean()),\\n            \\"c_mean\\": float(c.mean()), \\"c_std\\": float(c.std()), \\"c_min\\": float(c.min()), \\"edge_mean\\": float(mag.mean()), \\"edge_p95\\": float(np.quantile(mag, 0.95)),\\n            \\"chroma_mean\\": float(chroma.mean()), \\"chroma_max\\": float(chroma.max()),\\n            \\"in_dark\\": float(c[m].mean()) if inside else 1.0, \\"in_chroma\\": float(chroma[m].mean()) if inside else float(chroma.mean()),\\n            \\"ring_contrast\\": float(c[ndi.binary_dilation(m, iterations=3) & ~m].mean() - c[m].mean()) if inside and (~m).any() else 0.0,\\n            \\"thin\\": float(1 - fill), \\"gain\\": float(img.mean()), \\"sat\\": float((img.max(axis=2) - img.min(axis=2)).mean()),\\n            \\"cy\\": float(ndi.center_of_mass(m)[0] / SZ) if inside else 0.5, \\"cx\\": float(ndi.center_of_mass(m)[1] / SZ) if inside else 0.5}\\n\\n\\ndef feature_table(imgs):\\n    return pd.DataFrame([extract(im) for im in imgs])\\n\\nmtr, itr, _ = make_images(400, seed=1)\\nmte, ite, kte = make_images(300, seed=2)\\n\\n\\ndef hist_feat(imgs):\\n    \\"\\"\\"Baseline 특징: 밝기 히스토그램 16칸 + 채널 평균 3 + 밝기 표준편차 1.\\"\\"\\"\\n    out = []\\n    for im in imgs:\\n        g = im.mean(axis=2)\\n        h, _ = np.histogram(g / max(g.max(), 1e-3), bins=16, range=(0, 1))\\n        out.append(np.r_[h / g.size, im.mean(axis=(0, 1)) / max(g.mean(), 1e-3), g.std() / max(g.mean(), 1e-3)])\\n    return np.array(out)\\n\\n\\nHtr, Hte = hist_feat(itr), hist_feat(ite)\\nytr, yte = (mtr[\\"cls\\"] != \\"none\\").values, (mte[\\"cls\\"] != \\"none\\").values\\nbase = make_pipeline(StandardScaler(), LogisticRegression(max_iter=2000, C=0.5)).fit(Htr, ytr)\\npr = base.predict(Hte)\\nprint(f\\"[Baseline 결과] 학습 {len(mtr)}장 / 평가 {len(mte)}장, 결함 사진 {yte.mean():.1%}\\")\\nprint(f\\"정확도 {accuracy_score(yte, pr):.1%}, 정밀도 {precision_score(yte, pr):.1%}, 재현율 {recall_score(yte, pr):.1%}, F1 {f1_score(yte, pr):.3f}\\")\\nrec = pd.DataFrame({\\"cls\\": mte[\\"cls\\"], \\"hit\\": pr == yte}).query(\\"cls != \'none\'\\").groupby(\\"cls\\")[\\"hit\\"].mean().round(2)\\nprint(\\"결함 유형별 재현율:\\", \\", \\".join(f\\"{KO_CLS[k]} {v:.0%}\\" for k, v in rec.items()))\\nsmall = (mte[\\"area_frac\\"] > 0) & (mte[\\"area_frac\\"] < 0.02)\\nprint(f\\"면적 2% 미만 작은 결함 {int(small.sum())}장 재현율 {pr[small.values].mean():.0%} (큰 결함 {pr[(mte[\'area_frac\'] >= 0.02).values].mean():.0%})\\")\\ndk = mte[\\"gain\\"] < 0.6\\nprint(f\\"조명이 어두운 사진(gain<0.6) 정확도 {(pr == yte)[dk.values].mean():.0%} vs 밝은 사진 {(pr == yte)[~dk.values].mean():.0%}\\")\\nprint(\\"-> 전체 통계만 보는 Baseline은 작은 균열을 놓치고 조명에 흔들린다. 결함 위치·모양을 보는 분할 특징이 필요하다(미션 4~5의 동기).\\")\\n\\n# 1) 상관분석: 결함 유무와의 관계(점이연 상관, 유의수준 0.05) - 사진 특징 + 장비 제원 기록\\nFall = pd.concat([feature_table(ite), feature_table(itr[:200])], ignore_index=True)\\nmall = pd.concat([mte, mtr[:200]], ignore_index=True)\\nFall[\\"op_hours\\"], Fall[\\"reg_grade\\"], Fall[\\"mfg_year\\"] = mall[\\"op_hours\\"], mall[\\"reg_grade\\"], mall[\\"mfg_year\\"]\\nFall[\\"coastal\\"] = (mall[\\"env\\"] == \\"coastal\\").astype(int)\\nyall = (mall[\\"cls\\"] != \\"none\\").astype(float)\\nrows = []\\nfor f in [\\"c_std\\", \\"c_min\\", \\"edge_p95\\", \\"dark_frac\\", \\"rust_frac\\", \\"sat\\", \\"op_hours\\", \\"reg_grade\\", \\"mfg_year\\", \\"coastal\\"]:\\n    r, p = stats.pearsonr(Fall[f].astype(float), yall)\\n    rows.append({\\"피처\\": f, \\"상관계수\\": round(r, 3), \\"p-value\\": f\\"{p:.1e}\\", \\"유의\\": p < 0.05})\\ncorr = pd.DataFrame(rows).sort_values(\\"상관계수\\", key=abs, ascending=False)\\nprint(\\"\\\\n[결함 유무와의 상관분석]\\")\\nprint(corr.to_string(index=False))\\nn_sig = int(corr[\\"유의\\"].sum())\\nprint(f\\"유의미한 피처 {n_sig}개 (목표 5개 이상: {\'달성\' if n_sig >= 5 else \'미달\'})\\")\\n\\n# 2) 제원 기록과의 불일치: 사진 판정 등급 vs 관리 기록 등급\\nct = pd.crosstab(mall[\\"grade\\"], mall[\\"reg_grade\\"]).rename(index=GRADE_NAME, columns=GRADE_NAME)\\ngap = (mall[\\"grade\\"] - mall[\\"reg_grade\\"]).abs() >= 2\\nprint(f\\"\\\\n[제원 대조] 사진 등급과 관리 기록 등급이 2단계 이상 다른 장비 {int(gap.sum())}대({gap.mean():.1%}) -> 제원 이상 신호 후보\\")\\nprint(ct.to_string())\\n\\n# 3) 파레토(ABC): 정비 공수(가상)가 큰 (장비 구분, 결함 유형) 조합\\nHOURS = {1: 2, 2: 8, 3: 30}\\nmall[\\"hours\\"] = mall[\\"grade\\"].map(HOURS)\\ng = mall[mall[\\"cls\\"] != \\"none\\"].groupby([\\"equip_type\\", \\"cls\\"])[\\"hours\\"].sum().sort_values(ascending=False)\\ncum = g.cumsum() / g.sum()\\nabc = pd.cut(cum, [0, 0.8, 0.95, 1.0], labels=[\\"A\\", \\"B\\", \\"C\\"])\\nprint(\\"\\\\n[파레토 A등급(누적 정비 공수 80%)]\\", \\", \\".join(f\\"{e}/{KO_CLS[c]}\\" for (e, c), v in abc.items() if v == \\"A\\"))\\nprint(f\\"A등급은 전체 조합 {len(g)}개 중 {int((abc == \'A\').sum())}개({(abc == \'A\').mean():.0%})\\")\\n\\n# 4) EDA 차트 8종\\nfig, ax = plt.subplots(2, 4, figsize=(18, 8))\\na = ax.ravel()\\na[0].imshow(np.concatenate([ite[int(np.where(mte[\'cls\'] == c)[0][0])] for c in CLASSES], axis=1)); a[0].set_title(\\"1. Samples: none/crack/corrosion/damage/leak\\"); a[0].axis(\\"off\\")\\nmall[\\"cls\\"].value_counts().plot.bar(ax=a[1]); a[1].set_title(\\"2. Photos by defect class\\")\\na[2].hist(mall.loc[yall == 0, \\"gain\\"], bins=20, alpha=0.6, label=\\"none\\"); a[2].hist(mall.loc[yall == 1, \\"gain\\"], bins=20, alpha=0.6, label=\\"defect\\"); a[2].legend(); a[2].set_title(\\"3. Lighting gain\\")\\nmall.boxplot(column=\\"damage_idx\\", by=\\"cls\\", ax=a[3], rot=30); a[3].set_title(\\"4. Damage index by class\\")\\na[4].imshow(ct.values, cmap=\\"Blues\\"); a[4].set_xticks(range(3), [1, 2, 3]); a[4].set_yticks(range(3), [1, 2, 3]); a[4].set_xlabel(\\"registered grade\\"); a[4].set_ylabel(\\"photo grade\\"); a[4].set_title(\\"5. Photo vs registered grade\\")\\na[5].scatter(mall[\\"op_hours\\"], mall[\\"damage_idx\\"], s=6, c=mall[\\"env\\"].map({\\"coastal\\": \\"tab:red\\", \\"inland\\": \\"tab:blue\\"})); a[5].set_title(\\"6. Op hours vs damage (red = coastal)\\")\\na[6].barh(corr[\\"피처\\"], corr[\\"상관계수\\"]); a[6].set_title(\\"7. Correlation with defect\\")\\na[7].plot(range(1, len(cum) + 1), cum.values, \\"o-\\"); a[7].axhline(0.8, ls=\\"--\\", c=\\"gray\\"); a[7].set_title(\\"8. Pareto of repair hours\\")\\nplt.suptitle(\\"QLT03 EDA - baseline defect screening (synthetic photos)\\"); plt.tight_layout(); plt.show()\\nprint(\\"\\\\nEDA 차트 8종 생성 (목표 8종 이상: 달성)\\")\\n"\n    },\n    {\n      "cell_type": "markdown",\n      "metadata": {},\n      "source": "## 내 관찰\\n\\n- 핵심 발견점 3개: (작성)\\n- 그래프와 수치 근거: (작성)\\n- 한계: (작성)",\n      "id": "p1b-7677b791585e497a"\n    }\n  ],\n  "metadata": {\n    "kernelspec": {\n      "display_name": "Python 3",\n      "language": "python",\n      "name": "python3"\n    }\n  },\n  "nbformat": 4,\n  "nbformat_minor": 5\n}\n', 'reports/m01_record.md': "# QLT03 미션 1 수행 기록 [교육용가상자료]\n\n원문 산출물: docs/qlt03_logistics_domain_spec.md 및 데이터 딕셔너리\n실습가이드 제출물: 데이터 사전, 라벨링 기준, 업무 프로세스 맵\n이 문서는 학습자가 작성하는 양식입니다. 빈 양식은 완료 산출물이 아닙니다.\n\n## 기본 실행\n- 실행 환경/일시: (작성)\n- 실행한 코드와 출력 파일: outputs/m01/baseline/stdout.txt\n- 확인할 결과(원문 예시):\n데이터 사전 피처 24개(목표 20개 이상), 정상 샘플 20건 검증 오류 0건\n예측 대상 4종(결함 면적, 깊이, 손상도, 정비 등급)\n- 직접 관찰한 지표/단위/분모: (작성)\n- 원문 예시와 차이 및 원문 Q1 달성 여부: (작성)\n- 표/그래프에서 발견한 근거: (작성)\n\n## 도전 과제\n데이터 사전에 '촬영 거리' 같은 필드를 하나 추가하고 실제 데이터와 대조해 누락으로 표시되는지 확인한다.\n- 변경 전/후 코드 또는 설명형 과제의 근거: (작성)\n- **확인할 것**: 데이터 딕셔너리 피처 수 · 정상 샘플 검증 통과·오류 건수 · photo.shot_distance_m 누락 메시지 · 검사관 보정·등급 승인 단계\n- 변경 전/후 결과와 단위: (작성)\n- 결과가 같으면 변경이 적용된 증거와 같은 이유: (작성)\n- **생각해 볼 질문**: 촬영 거리 필드를 스키마에만 추가하면 기존 정상 샘플의 검증 결과가 어떻게 바뀌는가? 가짜 측정값으로 누락을 채우는 대신 스키마와 수집 기록을 어떻게 일치시킬지 설명한다.\n- 내 답과 적용 한계: (작성)\n\n## AI 활용 및 최종 산출물\n- 사용한 AI/질문/응답 중 판단에 사용한 부분/직접 확인한 점: (작성, 미사용이면 미사용 이유)\n- 원문이 요구한 산출물 본문 또는 파일 연결: (작성)\n- 제출할 최종 경로·선택본(기본/도전)·선택 이유: (작성; 원문_제출경로_연결표.md 참조)\n- 최종 판단을 담당하는 가상 역할/사람 확인 절차: (작성)\n", 'reports/m07_record.md': '# QLT03 미션 7 수행 기록 [교육용가상자료]\n\n원문 산출물: src/optimization/qlt03_simulator.py\n실습가이드 제출물: 최적화 코드, 비용 비교표, 민감도, 결재 이력, 차단 로그\n이 문서는 학습자가 작성하는 양식입니다. 빈 양식은 완료 산출물이 아닙니다.\n\n## 기본 실행\n- 실행 환경/일시: (작성)\n- 실행한 코드와 출력 파일: outputs/m07/baseline/stdout.txt\n- 확인할 결과(원문 예시):\n평상 절감 59.5%·저조도 28.2%(목표 15% 이상), 진흙 20%는 -7.9%로 미달\n자동 확정 0건, 권한 우회 차단 10/10(100%), 감사 로그 25건 위·변조 검출\n- 직접 관찰한 지표/단위/분모: (작성)\n- 원문 예시와 차이 및 원문 Q1 달성 여부: (작성)\n- 표/그래프에서 발견한 근거: (작성)\n\n## 도전 과제\n검사관 오류율 P_HUMAN을 8%에서 3%로 낮춰 관행 비용과 절감률이 어떻게 달라지는지 확인한다.\n- 변경 전/후 코드 또는 설명형 과제의 근거: (작성)\n- **확인할 것**: 관행·HITL 비용과 절감률 · 평상·저조도·진흙 시나리오 · 자동 확정 건수 · 권한 우회 차단과 감사 로그\n- 변경 전/후 결과와 단위: (작성)\n- 결과가 같으면 변경이 적용된 증거와 같은 이유: (작성)\n- **생각해 볼 질문**: 검사관 오류율을 8%에서 3%로 낮추면 비교 기준인 전수 정밀 판독 비용과 HITL 절감률은 어떻게 달라지는가? 진흙 시나리오의 미달을 숨기지 않고 어느 조건에서 경제성이 성립하는지 설명한다.\n- 내 답과 적용 한계: (작성)\n\n## AI 활용 및 최종 산출물\n- 사용한 AI/질문/응답 중 판단에 사용한 부분/직접 확인한 점: (작성, 미사용이면 미사용 이유)\n- 원문이 요구한 산출물 본문 또는 파일 연결: (작성)\n- 제출할 최종 경로·선택본(기본/도전)·선택 이유: (작성; 원문_제출경로_연결표.md 참조)\n- 최종 판단을 담당하는 가상 역할/사람 확인 절차: (작성)\n', 'reports/m06_record.md': '# QLT03 미션 6 수행 기록 [교육용가상자료]\n\n원문 산출물: src/explainability/qlt03_xai_explainer.py 및 XAI 리포트\n실습가이드 제출물: 업로드 판정표, 시각화 그래프, 기여도 표, 자연어 설명\n이 문서는 학습자가 작성하는 양식입니다. 빈 양식은 완료 산출물이 아닙니다.\n\n## 기본 실행\n- 실행 환경/일시: (작성)\n- 실행한 코드와 출력 파일: outputs/m06/baseline/stdout.txt\n- 확인할 결과(원문 예시):\n기여도 상위 3개 자동 산출 300건 중 300건(100%)\n업로드 8장 중 유형 판정 7장 일치, 손상 부위 IoU 0.64~0.98, 저조도 사진은 재촬영 요구\n- 직접 관찰한 지표/단위/분모: (작성)\n- 원문 예시와 차이 및 원문 Q1 달성 여부: (작성)\n- 표/그래프에서 발견한 근거: (작성)\n\n## 도전 과제\n재촬영 기준 mud 0.03을 0.01로 바꿔 재촬영 요구 사진이 어떻게 달라지는지 확인한다.\n- 변경 전/후 코드 또는 설명형 과제의 근거: (작성)\n- **확인할 것**: U01~U08의 오염 추정치·처리 상태 · 재촬영 요구 사진 목록 · 기여도 상위 3개 산출 건수 · 표시 부위 IoU\n- 변경 전/후 결과와 단위: (작성)\n- 결과가 같으면 변경이 적용된 증거와 같은 이유: (작성)\n- **생각해 볼 질문**: mud 기준을 0.03에서 0.01로 낮추면 어떤 사진이 새로 재촬영 대상으로 바뀌는가? 바뀌지 않았다면 8장의 mud 값과 저조도 조건을 근거로 이유를 설명하고 판정 기여도의 부호를 해석한다.\n- 내 답과 적용 한계: (작성)\n\n## AI 활용 및 최종 산출물\n- 사용한 AI/질문/응답 중 판단에 사용한 부분/직접 확인한 점: (작성, 미사용이면 미사용 이유)\n- 원문이 요구한 산출물 본문 또는 파일 연결: (작성)\n- 제출할 최종 경로·선택본(기본/도전)·선택 이유: (작성; 원문_제출경로_연결표.md 참조)\n- 최종 판단을 담당하는 가상 역할/사람 확인 절차: (작성)\n', 'reports/qlt03_roi_analysis.md': '# QLT03 원문 산출물 작성 틀 — qlt03_roi_analysis\n\n[교육용가상자료] 미션 9 원문 지정 경로: `reports/qlt03_roi_analysis.md`\n수행 기록 reports/m09_record.md와 별도인 최종 문서입니다. 아래 모든 절에 본문과 해당 실행 근거를 직접 작성하세요. 적용되지 않는 절에는 이유를 적습니다. 빈 틀은 완료 산출물이 아닙니다.\n\n## 비교 대상과 동일 분모\n(작성)\n\n## 교육용 비용·물량·기간 가정\n(작성)\n\n## 기준안과 개선안의 실측 결과\n(작성)\n\n## 비용·시간 산식과 단위\n(작성)\n\n## 순절감과 회수기간 검산\n(작성)\n\n## 민감도·불확실성·적용 한계\n(작성)\n\n## 결론과 근거 로그 연결\n(작성)\n', 'reports/m03_record.md': '# QLT03 미션 3 수행 기록 [교육용가상자료]\n\n원문 산출물: notebooks/qlt03_eda_analysis.ipynb 및 EDA 요약 리포트\n실습가이드 제출물: Baseline 성능표, EDA 차트 8종, 상관표, 제원 불일치 표, 파레토 결과\n이 문서는 학습자가 작성하는 양식입니다. 빈 양식은 완료 산출물이 아닙니다.\n\n## 기본 실행\n- 실행 환경/일시: (작성)\n- 실행한 코드와 출력 파일: outputs/m03/baseline/stdout.txt\n- 확인할 결과(원문 예시):\nBaseline 정확도 86.0%·재현율 92.2%, 면적 2% 미만 작은 결함 재현율 74%\n유의 피처 7개(목표 5개 이상), EDA 차트 8종, 제원 불일치 후보 47대\n- 직접 관찰한 지표/단위/분모: (작성)\n- 원문 예시와 차이 및 원문 Q1 달성 여부: (작성)\n- 표/그래프에서 발견한 근거: (작성)\n\n## 도전 과제\nBaseline 로지스틱 회귀의 규제 강도 C를 0.5에서 5로 바꿔 성능 변화를 확인하고 한계가 해소되는지 설명한다.\n- 변경 전/후 코드 또는 설명형 과제의 근거: (작성)\n- **확인할 것**: 정확도·정밀도·재현율·F1 · 작은 결함 재현율 · 저조도와 밝은 사진 정확도 · 제원 불일치 후보 수\n- 변경 전/후 결과와 단위: (작성)\n- 결과가 같으면 변경이 적용된 증거와 같은 이유: (작성)\n- **생각해 볼 질문**: C를 0.5에서 5로 바꿨을 때 전체 성능과 작은 결함·저조도 성능은 함께 좋아졌는가? C가 커질수록 규제가 약해지는 점과 위치·모양 정보를 쓰지 않는 입력의 한계를 연결해 설명한다.\n- 내 답과 적용 한계: (작성)\n\n## AI 활용 및 최종 산출물\n- 사용한 AI/질문/응답 중 판단에 사용한 부분/직접 확인한 점: (작성, 미사용이면 미사용 이유)\n- 원문이 요구한 산출물 본문 또는 파일 연결: (작성)\n- 제출할 최종 경로·선택본(기본/도전)·선택 이유: (작성; 원문_제출경로_연결표.md 참조)\n- 최종 판단을 담당하는 가상 역할/사람 확인 절차: (작성)\n', 'reports/m08_record.md': "# QLT03 미션 8 수행 기록 [교육용가상자료]\n\n원문 산출물: src/dashboard/qlt03_app.py (Streamlit/Dash 등)\n실습가이드 제출물: 진흙 스윕 표, 복합 열화 표, 처리 시간·로딩 시간, 한계치 요약\n이 문서는 학습자가 작성하는 양식입니다. 빈 양식은 완료 산출물이 아닙니다.\n\n## 기본 실행\n- 실행 환경/일시: (작성)\n- 실행 경로(앱/시뮬레이터): (작성)\n- 앱 증거: 내려받은 qlt03_m08_app_stdout.txt·qlt03_m08_sweep.csv와 화면 캡처 경로: (작성)\n- 시뮬레이터 증거: outputs/m08/baseline/stdout.txt 및 figure 파일: (작성)\n- 실제 사용한 경로의 증거를 연결한다. 앱 로그 파일을 baseline 로그로 가장하지 않는다.\n- 확인할 결과(원문 예시):\n진흙 0%: 자동 처리 75%·경보 정확도 100%, 5%부터 93%로 95% 미달\n시뮬레이터의 요약 집계·목록·차트 렌더링 예시 시간은 0.1초 안팎. 브라우저를 포함한 웹앱 응답은 별도 측정하며 이 값으로 2초 목표 달성을 판정하지 않음, 흐림 사진 자동 처리 경보 정확도 91%\n- 직접 관찰한 지표/단위/분모: (작성)\n- 원문 예시와 차이 및 원문 Q1 달성 여부: (작성)\n- 표/그래프에서 발견한 근거: (작성)\n\n## 도전 과제\n경보를 등급 2 이상에서 등급 3(3단계 정비)만으로 바꿔 경보 정확도와 자동 처리 비율이 어떻게 달라지는지 확인한다.\n- 변경 전/후 코드 또는 설명형 과제의 근거: (작성)\n- **확인할 것**: 진흙 비율별 경보 정확도 · 품질 게이트 통과 자동 처리 비율 · 흐림·잡음·저조도 복합 조건 · 처리시간 실측과 대량 처리 외삽\n- 변경 전/후 결과와 단위: (작성)\n- 결과가 같으면 변경이 적용된 증거와 같은 이유: (작성)\n- **생각해 볼 질문**: 경보 정의만 등급 3으로 바꿨을 때 자동 처리 비율이 그대로인 이유는 무엇인가? 게이트 통과 사진을 분모로 삼아 경보 정확도를 비교하고 자동 처리 0건의 '-'를 100%로 해석할 수 없는 이유를 적는다.\n- 내 답과 적용 한계: (작성)\n\n## AI 활용 및 최종 산출물\n- 사용한 AI/질문/응답 중 판단에 사용한 부분/직접 확인한 점: (작성, 미사용이면 미사용 이유)\n- 원문이 요구한 산출물 본문 또는 파일 연결: (작성)\n- 제출할 최종 경로·선택본(기본/도전)·선택 이유: (작성; 원문_제출경로_연결표.md 참조)\n- 최종 판단을 담당하는 가상 역할/사람 확인 절차: (작성)\n", 'reports/m05_record.md': '# QLT03 미션 5 수행 기록 [교육용가상자료]\n\n원문 산출물: src/models/qlt03_advanced_model.py\n실습가이드 제출물: 성능 비교표, 혼동행렬, 신뢰도 보정표, 상황별 성능표\n이 문서는 학습자가 작성하는 양식입니다. 빈 양식은 완료 산출물이 아닙니다.\n\n## 기본 실행\n- 실행 환경/일시: (작성)\n- 실행한 코드와 출력 파일: outputs/m05/baseline/stdout.txt\n- 확인할 결과(원문 예시):\nMAE 개선 73.4%, RMSE 개선 55.1%(목표 20% 이상), R2 0.946(목표 0.85 이상)\n등급 일치율 93.3%, 신뢰도 80 이상 234장(78%) 일치율 100%, 진흙 가림 사진은 80 이상 0%\n- 직접 관찰한 지표/단위/분모: (작성)\n- 원문 예시와 차이 및 원문 Q1 달성 여부: (작성)\n- 표/그래프에서 발견한 근거: (작성)\n\n## 도전 과제\n자동 통과 기준 신뢰도를 80에서 70으로 낮추면 자동 통과 비율과 일치율이 어떻게 달라지는지 확인한다.\n- 변경 전/후 코드 또는 설명형 과제의 근거: (작성)\n- **확인할 것**: 자동 통과 후보 수·비율·실제 일치율 · 저조도·진흙 조건의 문턱 이상 비율 · 등급 혼동행렬의 과소 판정 수 · MAE·RMSE·R2\n- 변경 전/후 결과와 단위: (작성)\n- 결과가 같으면 변경이 적용된 증거와 같은 이유: (작성)\n- **생각해 볼 질문**: 신뢰도 80→70 하향 후 후보 비율과 실제 일치율은 어떻게 달라졌는가? 혼동행렬과 진흙 조건을 근거로 범위를 넓히는 대가를 설명하고 자동 통과가 최종 정비 등급 확정을 뜻하는지 구분한다.\n- 내 답과 적용 한계: (작성)\n\n## AI 활용 및 최종 산출물\n- 사용한 AI/질문/응답 중 판단에 사용한 부분/직접 확인한 점: (작성, 미사용이면 미사용 이유)\n- 원문이 요구한 산출물 본문 또는 파일 연결: (작성)\n- 제출할 최종 경로·선택본(기본/도전)·선택 이유: (작성; 원문_제출경로_연결표.md 참조)\n- 최종 판단을 담당하는 가상 역할/사람 확인 절차: (작성)\n', 'reports/m02_record.md': '# QLT03 미션 2 수행 기록 [교육용가상자료]\n\n원문 산출물: src/etl/qlt03_data_cleansing.py 및 정제 품질 보고서\n실습가이드 제출물: 정제 모듈 코드, 정제 규칙 로그, 정합성 검증 결과, 전후 통계 비교표\n이 문서는 학습자가 작성하는 양식입니다. 빈 양식은 완료 산출물이 아닙니다.\n\n## 기본 실행\n- 실행 환경/일시: (작성)\n- 실행한 코드와 출력 파일: outputs/m02/baseline/stdout.txt\n- 확인할 결과(원문 예시):\n정제 규칙 11종(목표 10종 이상), 정합성 검증 12/12 통과(100%)\n정제 후 사진 150장, 긴급 점검 6장 유지, 정제 박스 평균 IoU 0.99\n- 직접 관찰한 지표/단위/분모: (작성)\n- 원문 예시와 차이 및 원문 Q1 달성 여부: (작성)\n- 표/그래프에서 발견한 근거: (작성)\n\n## 도전 과제\nR7 조명 보정 기준(평균 밝기 0.3)을 0.4로 바꿔 보정 대상 사진 수가 어떻게 달라지는지 확인한다.\n- 변경 전/후 코드 또는 설명형 과제의 근거: (작성)\n- **확인할 것**: R7 문턱 미만 사진 수 · 정합성 12항목 · 정제 후 사진·긴급 점검 보존 수 · 전후 평균 밝기와 박스 IoU\n- 변경 전/후 결과와 단위: (작성)\n- 결과가 같으면 변경이 적용된 증거와 같은 이유: (작성)\n- **생각해 볼 질문**: 밝기 0.3→0.4 변경 후 R7의 집계 건수와 정제 후 평균 밝기는 각각 어떻게 달라졌는가? 현재 구현이 모든 사진을 정규화한다는 점을 코드에서 찾아 집계 문턱과 실제 적용 조건을 구분한다.\n- 내 답과 적용 한계: (작성)\n\n## AI 활용 및 최종 산출물\n- 사용한 AI/질문/응답 중 판단에 사용한 부분/직접 확인한 점: (작성, 미사용이면 미사용 이유)\n- 원문이 요구한 산출물 본문 또는 파일 연결: (작성)\n- 제출할 최종 경로·선택본(기본/도전)·선택 이유: (작성; 원문_제출경로_연결표.md 참조)\n- 최종 판단을 담당하는 가상 역할/사람 확인 절차: (작성)\n', 'reports/m09_record.md': '# QLT03 미션 9 수행 기록 [교육용가상자료]\n\n원문 산출물: reports/qlt03_roi_analysis.md\n실습가이드 제출물: 재보정 코드, 원인별 오판표, 전후 비교표, 3대 지표와 민감도\n이 문서는 학습자가 작성하는 양식입니다. 빈 양식은 완료 산출물이 아닙니다.\n\n## 기본 실행\n- 실행 환경/일시: (작성)\n- 원문 재현 이력: outputs/m09/baseline/stdout.txt (최종 성능·경제성 기준 아님)\n- 수정 기본 실행: outputs/m09/revised/stdout.txt (최종 보고·도전 비교 기준)\n- 확인할 결과(실습 코드 채점 기준):\n실습 코드 채점 기준: 연간 순절감액 2,751만원, 판독 시간 단축률 0.3%, 투자 회수 10.9개월.\n반올림 전 2,751.111111만원/년·0.292475%·10.904685개월. 기존 AI→개선 AI, 동일 B 360장, 연 20,000장, 시급 1.2만원, 증분 운영비 150만원/년, 증분 투자비 2,500만원이라는 교육용 가정이다. outputs/m09/revised/stdout.txt와 전후 총비용 차이를 확인한다. 환경에 따른 차이는 실제 로그와 산식으로 설명하며 예시 숫자를 맞추려고 출력을 고치지 않는다.\n- 직접 관찰한 지표/단위/분모: (작성)\n- 실습 코드 기준과 차이 및 원문 Q1 달성 여부: (작성)\n- 표/그래프에서 발견한 근거: (작성)\n\n## 도전 과제\n연 판정 사진 수 N_YR을 20000장에서 10000장으로 바꿔 절감액과 회수 기간이 어떻게 달라지는지 확인한다.\n- 변경 전/후 코드 또는 설명형 과제의 근거: (작성)\n- **확인할 것**: 보류 구간 B의 등급 일치율·오판 손실 · 연간 비용 절감액과 고정 운영비 · 판독 시간 단축률 · 투자 회수 기간\n- 변경 전/후 결과와 단위: (작성)\n- 결과가 같으면 변경이 적용된 증거와 같은 이유: (작성)\n- **생각해 볼 질문**: 연 사진 수 20000→10000 변경 후 연 절감액이 정확히 절반이 되는가? 고정 운영비 150만원을 포함한 산식으로 확인하고 보류 구간의 모델 성능과 시간 단축률이 유지되는 이유를 설명한다.\n- 내 답과 적용 한계: (작성)\n\n## AI 활용 및 최종 산출물\n- 사용한 AI/질문/응답 중 판단에 사용한 부분/직접 확인한 점: (작성, 미사용이면 미사용 이유)\n- 원문이 요구한 산출물 본문 또는 파일 연결: (작성)\n- 제출할 최종 경로·선택본(기본/도전)·선택 이유: (작성; 원문_제출경로_연결표.md 참조)\n- 최종 판단을 담당하는 가상 역할/사람 확인 절차: (작성)\n', 'reports/m04_record.md': '# QLT03 미션 4 수행 기록 [교육용가상자료]\n\n원문 산출물: src/features/qlt03_feature_engineering.py 및 기준 모델\n실습가이드 제출물: 분할 성능표, 면적 보정 결과, 깊이·손상도 오차표, 변수군 제거 결과\n이 문서는 학습자가 작성하는 양식입니다. 빈 양식은 완료 산출물이 아닙니다.\n\n## 기본 실행\n- 실행 환경/일시: (작성)\n- 실행한 코드와 출력 파일: outputs/m04/baseline/stdout.txt\n- 확인할 결과(원문 예시):\n분할 IoU 균열 0.89·부식 0.74·파손 0.61·누유 0.94, 면적 MAPE 20.8% -> 10.2%\n손상도 MAPE 8.8%(목표 25% 이하, 기준 30.8%), 깊이 MAPE 28.0%\n- 직접 관찰한 지표/단위/분모: (작성)\n- 원문 예시와 차이 및 원문 Q1 달성 여부: (작성)\n- 표/그래프에서 발견한 근거: (작성)\n\n## 도전 과제\n깊이 추정 MAPE가 손상도보다 큰 이유를 사진 한 장으로는 깊이가 직접 보이지 않는 점과 연결해 설명한다.\n- 변경 전/후 코드 또는 설명형 과제의 근거: (작성)\n- **확인할 것**: 유형별 IoU·Dice · 면적 원측정·보정 MAPE · 깊이 MAPE와 MAE · 손상도 기준·제안 MAPE · 변수군 제거 시 오차 증가\n- 변경 전/후 결과와 단위: (작성)\n- 결과가 같으면 변경이 적용된 증거와 같은 이유: (작성)\n- **생각해 볼 질문**: 기본 실행에서 깊이 MAPE와 손상도 MAPE를 각각 인용하고, 한 장의 사진으로 깊이를 직접 관측할 수 없다는 점을 들어 차이를 설명한다. 임계값 분할과 실제 Mask R-CNN 학습을 구분하고 깊이 게이지 등 추가 확인 방법을 한 가지 제시한다.\n- 내 답과 적용 한계: (작성)\n\n## AI 활용 및 최종 산출물\n- 사용한 AI/질문/응답 중 판단에 사용한 부분/직접 확인한 점: (작성, 미사용이면 미사용 이유)\n- 원문이 요구한 산출물 본문 또는 파일 연결: (작성)\n- 제출할 최종 경로·선택본(기본/도전)·선택 이유: (작성; 원문_제출경로_연결표.md 참조)\n- 최종 판단을 담당하는 가상 역할/사람 확인 절차: (작성)\n', 'reports/m10_record.md': '# QLT03 미션 10 수행 기록 [교육용가상자료]\n\n원문 산출물: docs/qlt03_operations_guide.md 및 pipeline/qlt03_retrain.py\n실습가이드 제출물: 통합 파이프라인 코드, 처리 구분 표, 자체 점검표, 인터페이스 명세, 브리핑 요약\n이 문서는 학습자가 작성하는 양식입니다. 빈 양식은 완료 산출물이 아닙니다.\n\n## 기본 실행\n- 실행 환경/일시: (작성)\n- 원문 재현 이력: outputs/m10/baseline/stdout.txt (최종 성능·경제성 기준 아님)\n- 수정 기본 실행: outputs/m10/revised/stdout.txt (최종 보고·도전 비교 기준)\n- 확인할 결과(실습 코드 채점 기준):\n실습 코드 최종시험 MAE 6.43, 등급 정확도 87.5%(시험 240장). 빈 운영지침은 0/12이며 문서 미완료 점검은 FAIL이 정상이다. 직접 작성 후 완료 수와 자체 점검 결과를 다시 기록한다.\n처리 구분 간이 승인 후보 112·정밀 판독 86·재촬영 42장, 자동 확정 0건\n- 직접 관찰한 지표/단위/분모: (작성)\n- 원문 예시와 차이 및 원문 Q1 달성 여부: (작성)\n- 표/그래프에서 발견한 근거: (작성)\n\n## 도전 과제\n코드의 승격 기준 0.03(MAE 3% 개선)을 0.15로 높여 승격 판정이 어떻게 달라지는지 확인한다.\n- 변경 전/후 코드 또는 설명형 과제의 근거: (작성)\n- **확인할 것**: 챔피언·챌린저 검증 MAE와 개선율 · 승격 판정 · 간이 승인 후보·정밀 판독·재촬영 수 · 자동 확정 수·자체 점검·문서 완비\n- 변경 전/후 결과와 단위: (작성)\n- 결과가 같으면 변경이 적용된 증거와 같은 이유: (작성)\n- **생각해 볼 질문**: 검증 MAE로 계산한 개선율은 15%를 넘는가? 문턱 변경 후 승격 판정과 처리 구분이 같더라도 오류라고 할 수 없는 이유를 적고 세 처리 구분이 신규 240장과 맞는지 검산한다.\n- 내 답과 적용 한계: (작성)\n\n## AI 활용 및 최종 산출물\n- 사용한 AI/질문/응답 중 판단에 사용한 부분/직접 확인한 점: (작성, 미사용이면 미사용 이유)\n- 원문이 요구한 산출물 본문 또는 파일 연결: (작성)\n- 제출할 최종 경로·선택본(기본/도전)·선택 이유: (작성; 원문_제출경로_연결표.md 참조)\n- 최종 판단을 담당하는 가상 역할/사람 확인 절차: (작성)\n', 'reports/재개기록.md': '# 재개 기록 (직접 작성)\n\n- 마지막 미션: (작성)\n- 마지막 실행일시: (작성)\n- 마지막 작성 문장: (작성)\n- 이어 할 일: (작성)\n', 'docs/원문_제출경로_연결표.md': '# 제출 경로 연결표\n\n현재 실행 열의 파일이 기본 예제 코드입니다. 원문 이름의 Python 파일은 비교용 예제이며 제출 기본값이 아닙니다. 원문 파일을 덮지 말고 현재 실행 파일을 제출하고 아래 산출물 이름과 연결하세요.\n\n| 미션 | 원문 산출물 | 기본 예제(현재 실행) | 도전 작업 | 비교용 예제 |\n|---|---|---|---|---|\n| 미션 1 | docs/qlt03_logistics_domain_spec.md 및 데이터 딕셔너리 | src/m01.py | work/m01_challenge.py | src/m01.py |\n| 미션 2 | src/etl/qlt03_data_cleansing.py 및 정제 품질 보고서 | src/m02.py | work/m02_challenge.py | src/m02.py |\n| 미션 3 | notebooks/qlt03_eda_analysis.ipynb 및 EDA 요약 리포트 | src/m03.py | work/m03_challenge.py | src/m03.py |\n| 미션 4 | src/features/qlt03_feature_engineering.py 및 기준 모델 | src/m04.py | 설명 기록 | src/m04.py |\n| 미션 5 | src/models/qlt03_advanced_model.py | src/m05.py | work/m05_challenge.py | src/m05.py |\n| 미션 6 | src/explainability/qlt03_xai_explainer.py 및 XAI 리포트 | src/m06.py | work/m06_challenge.py | src/m06.py |\n| 미션 7 | src/optimization/qlt03_simulator.py | src/m07.py | work/m07_challenge.py | src/m07.py |\n| 미션 8 | src/dashboard/qlt03_app.py (Streamlit/Dash 등) | src/m08.py | work/m08_challenge.py | src/m08.py |\n| 미션 9 | reports/qlt03_roi_analysis.md | work/revised/m09.py | work/m09_challenge.py | src/m09.py |\n| 미션 10 | docs/qlt03_operations_guide.md 및 pipeline/qlt03_retrain.py | work/revised/m10.py | work/m10_challenge.py | src/m10.py |\n', 'docs/파일_정본_안내.md': '# 파일 정본과 작성 위치\n\n- 02_데이터셋: 배포된 빈 기록양식·자료의 정본.\n- docs/templates: 환경 초기화용 양식 사본.\n- reports/mNN_record.md: 학습자가 직접 작성하고 제출할 기록. 빈 배포 상태에서는 양식과 같으며 작성 뒤에는 이 파일을 우선합니다.\n- data/samples: 생성 데이터 구조를 확인하는 미리보기 사본. 기본 실습 데이터는 각 미션 코드에서 생성하므로 이 파일을 수정해도 기본 실행 데이터는 바뀌지 않습니다. 02_데이터셋과 동일 이름의 자료는 같은 배포 입력이며 실험할 때 사본의 변경 경로를 기록합니다.\n- src/mNN.py 및 원문 이름 Python 사본: 비교용 예제. 현재 실행·제출은 제출 경로 연결표의 수정 파일을 사용합니다. 앱 파일은 실행용입니다.\n\n`logistics`가 포함된 파일명은 제공 원문의 산출물 이름을 유지한 것입니다. 이번 실습에서는 외관 검사 자료와 판정 결과를 담으며 다른 과제 자료를 뜻하지 않습니다.\n', 'docs/학습시간과_선택미션.md': '# 학습시간과 선택 미션\n\n초보 소요시간은 원문의 약 1.3~1.7배를 준비 계획용 추정 범위로 참고하세요. 파일 조작·개념 학습·대기·재시도에 따라 달라지며 실제 학습자 실측이나 확정 시간이 아닙니다.\n\n선택 미션 4·미션 6·미션 8은 최종 PASS에서 제외하며 필수 7미션 21문항으로 판정합니다.\n\n| 미션 | 원문 학습계획 | 화면 표시 시간(참고) |\n|---|---|---|\n| 미션 1 | 1.5시간 | 1시간 |\n| 미션 2 | 2시간 | 2시간 |\n| 미션 3 | 2.5시간 | 2시간 |\n| 미션 4 | 3시간 | 3시간 |\n| 미션 5 | 3.5시간 | 4시간 |\n| 미션 6 | 3시간 | 3시간 |\n| 미션 7 | 3.5시간 | 4시간 |\n| 미션 8 | 3시간 | 3시간 |\n| 미션 9 | 3시간 | 3시간 |\n| 미션 10 | 3.5시간 | 3시간 |\n\n플랫폼은 시간을 정수로 표시하므로 미션별 표시와 원문 시간이 다를 수 있습니다. 학습 계획과 시간 기록은 원문 열을 기준으로 합니다.\n\n플랫폼 시간 표시는 정수 합계 28시간이며 원문 계획은 28.5시간입니다. 0.5시간 차이는 표시 방식의 차이이므로 학습 계획은 위 원문 시간을 따르세요.\n', 'docs/M08_실측정오표.md': '# 미션 8 실측 실습 적용 기준\n\n| 원문 예시 | 현재 고정 환경 실측 | 이유와 적용 범위 |\n|---|---|---|\n| 진흙 0% 자동 처리 73% | 120장 중 자동 처리 후보 90장, 75% | 원문 예시와 실행 관측의 차이. 원문 코드·기대 파일은 보존하며 현재 기록은 실제 출력과 분모를 사용합니다. 경보 정확도는 100%이고, 진흙 비율이 5%이면 93%로 낮아져 목표인 95%에 미달합니다. |\n\n다른 환경의 수치는 로그와 환경을 기록해 설명하세요. 예시에 맞추려고 결과를 조작하지 않습니다.\n', 'docs/M09_M10_실습안내.md': '# QLT03 미션 9·미션 10 수정 실습 안내 [교육용가상자료]\n\n수정 실습: 원문 실행값은 역사 비교용이다. 채점·최종 보고에는 work/revised/m09.py의 수정 실행을 사용한다. 비교는 기존 AI→개선 AI로 통일한다. B 360장의 전후 오판 손실(만원/장)·평균 시간(분/장), 같은 연간 건수·시급, 증분 운영비 및 투자비를 전후 표에 쓰고 연 순절감액과 회수기간을 재계산한다. 순절감액이 0 이하이면 회수 불가이다.\n\n수정 실습: 원문 실행값은 역사 비교용이다. 채점·최종 보고에는 work/revised/m10.py의 수정 실행을 사용한다. 운영 신규240장은 처리 건수만 집계하며, 기존600+신규160장은 학습, hold80은 모델 선택용이다. 최종 시험은 별도 seed 180/181/182 각80장으로 고정한 240장이며 튜닝에 재사용하지 않는다. 학습·선택·시험 ID 교집합 0을 확인한다. MAE·등급 정확도 분모는 시험240장, MAPE는 그중 실제 손상도10 이상 장수이다. 같은 seed 생성기의 합성 자료라는 한계는 남으며 현장 성능을 뜻하지 않는다.\n\n노트북 환경 셀부터 실행하면 원문 재현과 수정 실습이 순서대로 실행된다. 최종 보고의 기본값은 `outputs/m09/revised/stdout.txt`, `outputs/m10/revised/stdout.txt`를 사용하고, 도전값은 `outputs/mNN/challenge/stdout.txt`와 비교한다. `baseline`은 원문 재현 이력이다. 실습 코드의 코드·분모·가정값을 기록하고 원문 값과 혼합하지 않는다.\n\n로컬 프로젝트 폴더에서는 `python work/qlt03_revision_runner.py 9`, `python work/qlt03_revision_runner.py 10`으로 수정 실습을 실행한다. 원문 `src/m09.py`, `src/m10.py` 및 원문 기대값은 보존한다. 최종 원문 지정 코드 경로에 제출 사본을 만들 때 실습 코드/도전본 선택을 명시한다.\n\n미션 9 확인할 결과와 채점 기준은 `M09_ROI_정오표.md`의 실습 코드 열 및 수정 실행 로그를 사용한다. 원문 보존 자료는 이력 비교용이다.\n\n## 현재 미션 9 채점 기준\n실습 코드 채점 기준: 연간 순절감액 2,751만원, 판독 시간 단축률 0.3%, 투자 회수 10.9개월.\n반올림 전 2,751.111111만원/년·0.292475%·10.904685개월. 기존 AI→개선 AI, 동일 B 360장, 연 20,000장, 시급 1.2만원, 증분 운영비 150만원/년, 증분 투자비 2,500만원이라는 교육용 가정이다. outputs/m09/revised/stdout.txt와 전후 총비용 차이를 확인한다. 환경에 따른 차이는 실제 로그와 산식으로 설명하며 예시 숫자를 맞추려고 출력을 고치지 않는다.\n', 'docs/공통_Jupyter_사용안내.md': '# QLT03 실행·저장 안내\n\n[교육용가상자료] 실제 조직·장비·개인정보를 입력하지 않습니다.\n\n## 시작\n학습자 배포 ZIP의 01_학습자자료 최상위에 있는 QLT03_assignment.ipynb를 엽니다. 노트북 열기 → 환경 준비 셀 실행 → 미션 코드 저장 → 실행·확인 순서입니다. 이미 준비된 환경에는 설치하지 않습니다.\n\n## 저장과 새 작업 환경 복원\n\n1. 종료 전 수정한 노트북과 `reports/`, 직접 작성한 `docs/`, `work/`, `outputs/`를 원래 폴더 구조 그대로 허용된 저장 매체에 복사합니다. `reports/재개기록.md`에 마지막 미션, 마지막 실행일시, 마지막 작성 문장, 이어 할 일을 직접 적습니다. 노트북 사본과 reports/docs/work/outputs 파일을 함께 저장합니다.\n2. 새 작업폴더에서 저장한 노트북을 열고 **환경 준비 셀부터 실행**합니다. QLT03는 바로 다음 수정 실습 준비 셀까지 실행합니다. 이때 생긴 빈 양식은 저장 기록의 복원이 아닙니다.\n3. 저장 사본을 **reports/docs/work/outputs 원경로에 복원**합니다. 예: 저장한 `reports/m01_record.md` → 새 작업폴더의 `reports/m01_record.md`. 파일을 최상위에만 올리지 않습니다. Jupyter 작업폴더의 같은 하위 경로로 복사합니다. 기존 파일이 있으면 별도 보관 후 복원합니다.\n4. `reports/재개기록.md`와 마지막 기록지를 열어 **마지막 미션·실행일시·작성 문장**이 저장 사본과 같은지 확인합니다. outputs의 execution.json 시간도 대조합니다. 없으면 진행을 멈추고 저장 사본부터 확인합니다.\n5. 이어 할 미션의 코드 저장·실행 셀을 선택합니다. 복원한 work 코드를 다시 쓸 때는 노트북 셀에도 같은 수정 내용을 반영하고 백업 셀을 먼저 실행합니다. 처음부터 모두 실행해 복원 코드를 기본값으로 덮지 않습니다. 같은 분석을 재실행하면 이전 로그는 outputs/history에 남고 새 로그의 실행일시만 새로 찍힙니다. 재개기록에 직접 적은 이전 일시·마지막 미션은 자동으로 바꾸지 않습니다.\n\n복원 연습: `reports/m01_record.md`에 본인이 정한 표식과 실제 실행일시·마지막 미션 1을 적어 저장 → 새 작업폴더 준비 → 원경로 복원 → 환경 준비 재실행 → 표식과 기록 유지 확인 → 미션 1 실행 후 이전 execution.json이 outputs/history에 남는지 확인합니다. 실제 기관의 저장 허용 여부는 별도로 확인합니다.\n\n선택 미션 4·미션 6·미션 8은 산출물과 Q1·Q2·Q3 모두 가점/참고이며 최종 PASS에서 제외합니다. 필수 7미션 21문항만 최종 PASS로 판정합니다.\n\n초보 소요시간은 원문의 약 1.3~1.7배를 준비 계획용 추정 범위로 참고하세요. 파일 조작·개념 학습·대기·재시도에 따라 달라지며 실제 학습자 실측이나 확정 시간이 아닙니다.\n', 'docs/M08_웹앱과_대체경로.md': '# QLT03 미션 8 웹앱과 대체 경로\n\n[교육용가상자료] 실행 루트에서 작업합니다. 학습자 ZIP은 01_학습자자료, 전체 프로젝트는 notebook·src·docs가 함께 있는 폴더입니다.\n\n1. 노트북의 환경 준비 셀은 `src/dashboard/qlt03_app.py`, `requirements.txt`, `requirements-dashboard.txt`를 만듭니다. 기존 파일은 유지합니다.\n2. 미션 8 코드 저장 셀에서 시뮬레이터를 저장합니다. 준비된 환경에서 `python -m streamlit run src/dashboard/qlt03_app.py`를 실행합니다. 웹앱 선택 환경의 Streamlit 항목은 requirements-dashboard.txt에 있습니다.\n3. 실행 버튼으로 미션 8 시뮬레이터를 실행하고 결과 표·그래프·로그를 확인합니다. 노트북 경로는 `outputs/m08/baseline/stdout.txt`와 같은 폴더의 그림을 사용합니다. Q3에는 앱 파일 또는 시뮬레이터 실행 경로와 실제 출력·분모·해석을 연결합니다.\n4. 설치·웹앱 실행이 제한되면 노트북 미션 8 시뮬레이터를 동등하게 인정합니다. 선택 미션이므로 최종 PASS에서 제외합니다. 브라우저 포함 2초 응답은 별도 측정 항목입니다.\n', 'docs/M10_실습정오표.md': '# 미션 10 원문과 수정 실습 기준\n\n| 항목 | 원문 예시/동작 | 수정 값/동작 | 이유와 채점 범위 |\n|---|---|---|---|\n| 운영 문서 | 상수 목록 12/12, 자체점검 8/8 | 실제 운영지침 12절 본문 검사; 빈 양식 0/12, 문서 점검 FAIL | 빈 문서를 완료로 표시하지 않음. 직접 작성 후 재검사하며 내용의 타당성은 별도 평가 |\n| 최종 성능 | MAE 8.79→6.86 등 원문 예시 | 최종시험 240장 MAE 6.43, 등급 정확도 87.5% | 원문 예시와 최종시험 분모를 분리; 실제 로그와 환경을 기록 |\n\n원문 코드와 기대 출력은 보존합니다. 실습 코드 최종시험 MAE 6.43, 등급 정확도 87.5%(시험 240장). 빈 운영지침은 0/12이며 문서 미완료 점검은 FAIL이 정상이다. 직접 작성 후 완료 수와 자체 점검 결과를 다시 기록한다.\n', 'docs/처음_시작할_때.md': '# QLT03 처음 시작 안내\n\n[교육용가상자료] 실제 조직·장비·사진·정비 자료를 입력하지 않습니다.\n\n## 파일 열기\n\n학습자 배포 ZIP의 01_학습자자료 최상위에서 `QLT03_assignment.ipynb`를 엽니다. ZIP의 `01_학습자자료/`가 실행 루트입니다. 아래 docs·reports·src 경로는 실행 루트 기준입니다.\n\n교육장 Jupyter에서 노트북을 열면 환경 준비 셀이 기본 환경 목록·기록 양식·수정 코드·앱과 이 안내를 만듭니다. 파일 → 다른 이름으로 저장으로 사본을 만들고 위에서 아래로 진행하세요. 환경이 준비되지 않았으면 운영 담당자에게 확인합니다.\n\n## 실행 순서\n\n노트북 열기 → 환경 준비 셀 실행 → 미션 코드 저장 → 실행·확인 순서입니다.\n\n1. 환경 준비 셀을 실행합니다. `requirements.txt`와 `requirements-dashboard.txt`가 준비되며 기존 파일은 덮어쓰지 않습니다. 준비된 환경에서는 설치하지 않습니다.\n2. 해당 미션의 백업 셀 → 코드 저장 셀 → 실행·확인 셀을 순서대로 실행합니다. `%%writefile`은 코드 저장 명령입니다. 코드 저장 셀은 덮어쓸 수 있으므로 바로 앞 백업 셀을 함께 실행합니다.\n3. 출력 로그의 지표·단위·분모와 그림을 읽고 `reports/mNN_record.md`에 관찰값을 적습니다. 도전은 `work/mNN_challenge.py`에서 변경 조건과 전후 결과·해석을 기록합니다. 결과가 같거나 나빠져도 적용 증거와 이유를 설명합니다.\n4. 최종 파일은 `docs/원문_제출경로_연결표.md`에서 확인합니다. 원문 이름의 Python 사본은 비교용이며 현재 수정 실습·도전 코드 경로를 기록지에 연결합니다. 원문을 덮어쓰지 않습니다.\n\nREADY는 도전 코드 수정·저장을 마쳤음을 학습자가 표시하는 값입니다. 수정·저장 후 도전 실행 셀의 READY=False를 READY=True로 바꾸세요. 기본 실행 확인 PASS는 도전·해석·원문 산출물 작성 완료를 뜻하지 않습니다. 오류가 나면 첫 오류 메시지와 입력 조건을 그대로 기록하고 원인과 수정 내용을 남깁니다.\n\n## 미션 8 웹앱과 제한 환경\n\n환경 준비 셀이 `src/dashboard/qlt03_app.py`를 생성합니다. 미션 8 코드 저장 셀로 시뮬레이터를 저장한 뒤 준비된 환경에서 `python -m streamlit run src/dashboard/qlt03_app.py`를 실행합니다. 앱의 실행 버튼·필터를 사용해 표·그래프·로그와 해석을 기록합니다. 설치나 웹앱 실행이 제한되면 노트북 미션 8 시뮬레이터를 동등한 대체 경로로 사용합니다. `docs/M08_웹앱과_대체경로.md`를 읽으세요.\n\n## 기록 보존과 이어 하기\n\n종료 전에 수정한 노트북과 reports·직접 작성한 docs·work·outputs를 원래 폴더 구조대로 저장합니다. 노트북 사본 저장과 작업 환경 파일 보존은 별개입니다. `docs/저장과_새런타임_복원.md`에 따라 복원한 뒤 준비 셀을 재실행합니다. 준비 셀은 이미 존재하는 문서·코드·기록을 건너뜁니다. 저장이 제한되면 화면을 닫기 전에 담당자에게 허용된 기록 보존 경로를 확인하세요.\n\n## 개념과 시간\n\n미션 3~미션 5 전에 `docs/핵심지표_한장.md`에서 지표의 뜻·분모·좋아지는 방향과 계산 예시를 읽으세요. 용어·오류 메시지·수정 후보를 AI에 질문할 수 있지만 코드와 실제 출력의 대조·최종 판단은 직접 합니다.\n\n원문 계획은 28.5시간, 화면 표시 시간는 28시간입니다. 초보자는 약 1.3~1.7배를 준비 계획용 추정 범위로 참고합니다. 학습자 실측 시간이 아닙니다. 선택 미션 4·미션 6·미션 8은 산출물과 Q1·Q2·Q3 모두 가점/참고이며 최종 PASS에서 제외합니다. 필수 7미션 21문항을 평가합니다.\n\n[실습 준비·보안·제출 안내](실습준비와_제출안내.md)를 확인하세요.\n\n종료 전 허용된 저장 위치에 저장하고 다시 열어 확인한 뒤 사용한 교육장 계정에서 로그아웃합니다. 공용 PC에 비밀번호를 저장하지 않으며 로컬 사본은 조직 규정에 따라 정리합니다.\n\n## 오프라인 예제 요청문\n\n질문 틀: "[교육용가상자료] 미션 __의 목표는 __다. 내가 실행한 코드의 변경은 __이고 출력은 __다. 출력에서 직접 확인 가능한 근거만으로 차이/동일 결과의 원인을 설명해 줘. 관찰되지 않은 사실은 미확인으로 표시해 줘."\n\nAI 설명을 실행 증거로 대체하지 말고 코드·출력과 대조해 채택 여부를 기록한다.\n', 'docs/실습준비와_제출안내.md': '# 실습 준비·보안·제출 안내\n\n제공된 교육용 가상자료만 사용합니다. 비공개 업무자료, 실습에는 제공된 가상자료만 사용하고, 실제 이름·연락처 같은 개인정보는 AI 도구나 노트북에 넣지 않습니다. 실습을 위한 조직 내 촬영은 하지 않습니다. 조직 보안 지침과 서비스 이용 허용 여부를 먼저 확인합니다.\n\n이 자료는 교육용 가상 실습입니다. 실제 업무·정비·재보급 승인 기준이나 자동 결정에 사용하지 않습니다.\n\nAI에게 문서 초안을 요청할 수 있습니다. 사실·숫자 확인과 최종 수정은 본인이 합니다. 실제 실행·계산·자료 대조·최종 판단은 직접 하고 요청문·실제 응답·채택 이유를 기록합니다. 불일치·기각·성능 향상을 억지로 만들지 않습니다.\n\n종료 전 허용된 저장 위치에 저장하고 다시 열어 확인한 뒤 사용한 교육장 계정에서 로그아웃합니다. 공용 PC에 비밀번호를 저장하지 않으며 로컬 사본은 조직 규정에 따라 정리합니다.\n\nAI 접속과 사람 협업이 모두 불가능하면 개인 판단·계산·초안을 먼저 저장하고 협업 항목은 미수행으로 표시합니다. 교육 담당자에게 동료 검토 또는 허용 접속 시간의 보완 일정과 필요시 기한 조정을 요청합니다. 보완 전에는 협업 완료로 표시하지 않습니다.\n\n제출처·마감 일시·시간대는 miliai.kr 해당 과제 공지를 확인하고, 표시가 없으면 제출 전에 교육 담당자에게 확인합니다. 플랫폼 학습기간은 계획값이며 마감일을 뜻하지 않습니다. 별도 공지가 없으면 미션마다 그 미션 산출물을 ZIP 하나로 묶어 해당 미션 제출란에 첨부합니다. 통합 문서는 미션별 제목·위치를 적고 각 해당 제출란에 같은 파일을 첨부합니다. 파일명은 과제ID_미션번호_산출물명으로 작성하고 실제 형식의 확장자를 붙입니다. 실명·개인 식별번호는 파일명에 넣지 않습니다.\n\nPython 3.12와 requirements.txt 기준으로 준비된 환경을 사용합니다. 제한 환경에서는 관리자가 준비한 로컬 Jupyter를 사용하며 설치 명령을 실행하지 않습니다. 환경 셀은 버전을 확인하고 실습 파일을 준비합니다. 버전이 맞지 않으면 안내 메시지와 함께 멈추며, 그때는 requirements.txt 기준으로 교육 담당자가 안내한 방법으로 설치한 뒤 다시 실행합니다. 환경 셀은 건너뛰지 마세요.\n\n노트북 사본과 reports·직접 작성한 docs·work·outputs를 원래 폴더 구조대로 저장합니다. docs/원문_제출경로_연결표.md의 현재 실행 파일을 제출합니다.\n\n최종 실행 재현 확인은 필수 미션만 선택하여 실행합니다: `python -m pytest tests/test_missions.py -k "not [4] and not [6] and not [8]"`. 전체 미션 검사는 선택 미션까지 실행한 경우의 추가 확인이며 최종 PASS 조건이 아닙니다.\n\n미리보기의 source_sha256 및 검증 메타데이터는 제공 파일의 생성 근거를 확인하는 해시이며 학습 데이터 필드나 제출할 정답이 아닙니다. 실행 기록의 해시는 코드 변경 여부를 확인하므로 그대로 보존하세요.\n', 'docs/참고링크.md': '# 오프라인 실습안내\n\n외부 사이트 없이 ZIP에 제공된 자료를 사용합니다. 용어 뜻은 처음 시작 안내에서 확인합니다.\n\n질문 틀: "[교육용가상자료] 미션 __의 목표는 __다. 내가 실행한 코드의 변경은 __이고 출력은 __다. 출력에서 직접 확인 가능한 근거만으로 차이/동일 결과의 원인을 설명해 줘. 관찰되지 않은 사실은 미확인으로 표시해 줘."\n\nAI 설명을 실행 증거로 대체하지 말고 코드·출력과 대조해 채택 여부를 기록한다.\n', 'docs/qlt03_logistics_domain_spec.md': '# QLT03 원문 산출물 작성 틀 — qlt03_logistics_domain_spec\n\n[교육용가상자료] 미션 1 원문 지정 경로: `docs/qlt03_logistics_domain_spec.md`\n수행 기록 reports/m01_record.md와 별도인 최종 문서입니다. 아래 모든 절에 본문과 해당 실행 근거를 직접 작성하세요. 적용되지 않는 절에는 이유를 적습니다. 빈 틀은 완료 산출물이 아닙니다.\n\n## 목적과 적용 범위\n(작성)\n\n## 데이터 출처·생성 조건과 한계\n(작성)\n\n## 엔터티와 필드 사전(형식·단위·필수 여부)\n(작성)\n\n## 키·관계·시간 기준\n(작성)\n\n## 품질 규칙과 오류·격리 처리\n(작성)\n\n## 업무 판단과 사람 확인 책임\n(작성)\n\n## 검증 사례와 출력 근거\n(작성)\n', 'docs/qlt03_operations_guide.md': '# QLT03 운영지침 작성 양식\n\n빈 양식은 완료 산출물이 아닙니다. 각 절에 실제 근거를 30자 이상 작성합니다. 선택 미션은 미수행 사유와 제공 대체 경로를 설명할 수 있습니다. 형식 충족과 내용 타당성은 별도입니다.\n\n## 데이터 사전·스키마(미션 1)\n(작성)\n\n## 정제 규칙(미션 2)\n(작성)\n\n## Baseline·EDA(미션 3)\n(작성)\n\n## 분할·손상도 모델(미션 4)\n(작성)\n\n## 등급·신뢰도 엔진(미션 5)\n(작성)\n\n## 판정 인터페이스·설명(미션 6)\n(작성)\n\n## 승인 절차·권한(미션 7)\n(작성)\n\n## 오염 스트레스·한계치(미션 8)\n(작성)\n\n## AAR·ROI(미션 9)\n(작성)\n\n## 운영·재학습 절차(미션 10)\n(작성)\n\n## 장애 대응\n(작성)\n\n## 인터페이스 가이드\n(작성)\n', 'docs/M09_ROI_정오표.md': '# 미션 9 ROI 실습 적용 기준 [교육용가상자료]\n\n채점·최종 보고는 실습 코드 `work/revised/m09.py`와 `outputs/m09/revised/stdout.txt`를 사용한다.\n원문 코드·원문 기대값은 보존하며 아래 두 이력 행은 현재 채점값이 아니다.\n\n| 구분 | 연간 순절감액(만원/년) | 시간 단축률 | 투자 회수(개월) | 적용 범위 |\n|---|---:|---:|---:|---|\n| 건양대 원문 예시 | 5,140 | 30.8% | 5.8 | 원문 보존·이력 비교만 |\n| 고정 환경 원문 재현 | 4,572.222222 (표시 4,572) | 30.555556% | 6.561361 (표시 6.6) | 역사 비교만 |\n| 실습 코드 | **2,751.111111 (표시 2,751)** | **0.292475% (표시 0.3%)** | **10.904685 (표시 10.9)** | **채점·최종 보고 기본값** |\n\n원문은 손실의 비교 대상을 기존 AI, 시간의 비교 대상을 전수 수동 판독으로 섞었다.\n실습 코드는 손실·시간 모두 같은 B 360장에서 기존 AI→개선 AI로 비교한다.\n원문 예시와 원문 재현의 차이는 원문 예시의 미재현 이력으로 보존하며 값을 맞추지 않는다.\n\n- 평균 판독: 기존 10.447222→개선 10.416667분/장.\n- 손실: 기존 2.055556→개선 1.911111만원/장.\n- 연 20,000장, 시급 1.2만원, 증분 운영비 150만원/년, 증분 투자비 2,500만원.\n- 연 순절감 = 손실 절감 2,888.888889 + 시간 절감 12.222222 − 운영비 150 = 2,751.111111만원.\n- 연 총비용: 기존 45,290→개선 42,538.888889만원. 그 차이가 순절감이다.\n- 시간 단축률 = 1 − 개선 평균시간 / 기존 평균시간.\n- 회수기간 = 투자비 / (연 순절감 / 12). 순절감이 0 이하이면 회수 불가.\n- 연 10,000장 도전: 직접 계산해 실행 로그·산식·분모와 함께 제출한다. 고정 운영비 때문에 절감액은 단순 절반이 아니다.\n\n원문 Q1 문구·목표는 보존한다. 수행 증거·목표 달성은 구분하며 실습 코드의 동일 분모·산식·로그로 판단한다.\n다른 환경이나 도전값의 결과가 다르면 가정·코드·실측 근거를 기록한다. 예시와 강제 일치시키지 않는다.\n\n참고 수치를 그대로 제출하지 말 것(자기 계산 후 비교). 본인의 실행 로그·산식·분모와 해석을 제출한다.\n', 'docs/핵심지표_한장.md': '# QLT03 미션 3~미션 5 핵심 지표 한 장\n\n[교육용가상자료] 아래 숫자는 산식을 배우는 작은 예시이며 실측·목표값이 아닙니다. 기본 코드는 유지합니다.\n\n| 지표 | 뜻·분모·작은 계산 예시 | 좋아지는 방향·해석 한계 |\n|---|---|---|\n| IoU(미션 3~미션 4) | 정답·예측 마스크 교집합 픽셀 / 합집합 픽셀. 정답6, 예측4, 겹침3이면3/(6+4−3)=3/7≈0.429 | 클수록 겹침이 좋음; 양쪽 빈 마스크 처리 규칙 확인 |\n| Dice(미션 3~미션 4) | 2×교집합 / (정답 픽셀+예측 픽셀). 같은 예는6/10=0.60 | 클수록 좋음; IoU와 수치 자체를 같다고 비교하지 않음 |\n| MAE | 절대오차 합 / 평가 표본 수 n. 실제 10,20 / 예측 12,16이면 (2+4)/2=3 | 작을수록 좋음; 원래 단위 |\n| RMSE | √(제곱오차 합 / n). 같은 예는 √((4+16)/2)=√10≈3.16 | 작을수록 좋음; 큰 오차에 민감 |\n| MAPE | (100/n)×Σ(절대오차/실제값). 같은 예는 (20%+20%)/2=20% | 작을수록 좋음; 실제값 0이면 정의 불가. 코드의 제외/보정 규칙과 유효 n을 함께 기록 |\n| R2 | 1−Σ오차²/Σ(실제값−실제평균)². 같은 예는 1−20/50=0.60 | 클수록 좋음; 1이 완전 일치, 음수 가능. 실제값이 전부 같으면 분모 0 |\n| 오차 개선율 | (기준 오차−새 오차)/기준 오차×100. MAE10→8은20% | 같은 평가 표본·단위에서 클수록 좋음; 기준 오차 0이면 정의 불가 |\n| 등급 정확도(미션 5) | 정답 등급과 같은 사진 수 / 평가 사진 수. 8/10=80% | 높을수록 좋음; 과소 판정(실제보다 낮게 판정) 수도 별도 확인 |\n| 자동 통과 후보 비율·일치율(미션 5) | 후보 비율=문턱 이상 수/전체, 후보 일치율=후보 중 맞은 수/후보 수. 전체10·후보4·맞음3이면40%,75% | 두 분모를 구별. 후보0건이면 일치율 계산 불가. 후보 선정은 실제 승인 아님 |\n\n사진 분할의 픽셀 겹침, 면적·깊이의 측정 오차, 손상도 회귀 오차, 등급 분류 정확도는 서로 다른 지표입니다. 보정 계수는 학습 자료에서 정하고 같은 평가 사진으로 전후를 비교합니다. 미션 4의 깊이는 단일 사진에서 직접 잰 값이 아닌 추정이며 원문 예제는 실제 Mask R-CNN 학습 완료를 뜻하지 않습니다. 미션 10 최종 성능은 별도 시험240장이고 MAPE 분모는 그중 손상도10 이상 장수입니다.\n\n실행 전 본인이 확인할 것: 이번 출력의 단위, 분모, 좋아지는 방향을 한 문장씩 설명하고 작은 예시를 직접 계산해 보세요. 목표 달성 판정은 원문 Q1과 대조합니다. 이번 지표의 단위·분모·좋아지는 방향을 설명하고, 어려운 용어는 핵심 지표 예시로 다시 확인하세요.\n', 'docs/저장과_새런타임_복원.md': '# 저장과 복원 안내\n\n## 저장과 새 작업 환경 복원\n\n1. 종료 전 수정한 노트북과 `reports/`, 직접 작성한 `docs/`, `work/`, `outputs/`를 원래 폴더 구조 그대로 허용된 저장 매체에 복사합니다. `reports/재개기록.md`에 마지막 미션, 마지막 실행일시, 마지막 작성 문장, 이어 할 일을 직접 적습니다. 노트북 사본과 reports/docs/work/outputs 파일을 함께 저장합니다.\n2. 새 작업폴더에서 저장한 노트북을 열고 **환경 준비 셀부터 실행**합니다. QLT03는 바로 다음 수정 실습 준비 셀까지 실행합니다. 이때 생긴 빈 양식은 저장 기록의 복원이 아닙니다.\n3. 저장 사본을 **reports/docs/work/outputs 원경로에 복원**합니다. 예: 저장한 `reports/m01_record.md` → 새 작업폴더의 `reports/m01_record.md`. 파일을 최상위에만 올리지 않습니다. Jupyter 작업폴더의 같은 하위 경로로 복사합니다. 기존 파일이 있으면 별도 보관 후 복원합니다.\n4. `reports/재개기록.md`와 마지막 기록지를 열어 **마지막 미션·실행일시·작성 문장**이 저장 사본과 같은지 확인합니다. outputs의 execution.json 시간도 대조합니다. 없으면 진행을 멈추고 저장 사본부터 확인합니다.\n5. 이어 할 미션의 코드 저장·실행 셀을 선택합니다. 복원한 work 코드를 다시 쓸 때는 노트북 셀에도 같은 수정 내용을 반영하고 백업 셀을 먼저 실행합니다. 처음부터 모두 실행해 복원 코드를 기본값으로 덮지 않습니다. 같은 분석을 재실행하면 이전 로그는 outputs/history에 남고 새 로그의 실행일시만 새로 찍힙니다. 재개기록에 직접 적은 이전 일시·마지막 미션은 자동으로 바꾸지 않습니다.\n\n복원 연습: `reports/m01_record.md`에 본인이 정한 표식과 실제 실행일시·마지막 미션 1을 적어 저장 → 새 작업폴더 준비 → 원경로 복원 → 환경 준비 재실행 → 표식과 기록 유지 확인 → 미션 1 실행 후 이전 execution.json이 outputs/history에 남는지 확인합니다. 실제 기관의 저장 허용 여부는 별도로 확인합니다.\n', 'data/samples/교육용_미리보기.json': '{\n  "label": "[교육용가상자료]",\n  "source": "src/m01.py",\n  "source_sha256": "84f122d52acbfff2f079cb877b167f6f331903a3a58ecb03cd097d12207b47f9",\n  "note": "미션 1 기본 생성기의 미리보기. 전체 데이터는 코드로 생성한다. 실제 인물·조직·장비 자료 아님.",\n  "records": [\n    {\n      "photo": {\n        "img_id": "IMG000000",\n        "shot_day": 46,\n        "width": 64,\n        "height": 64,\n        "channels": 3,\n        "brightness": 0.87,\n        "mud_ratio": 0.04\n      },\n      "spec": {\n        "equip_id": "VX-8405-9536",\n        "equip_type": "vehicle",\n        "mfg_year": 2007,\n        "op_hours": 3868.0,\n        "env": "coastal",\n        "reg_grade": 3\n      },\n      "label": {\n        "defect_class": "crack",\n        "bbox_x": 12,\n        "bbox_y": 19,\n        "bbox_w": 18,\n        "bbox_h": 7,\n        "emergency": false\n      },\n      "result": {\n        "area_frac": 0.108,\n        "depth_mm": 2.0,\n        "damage_idx": 78.8,\n        "grade": 2,\n        "confidence": 0.65\n      }\n    },\n    {\n      "photo": {\n        "img_id": "IMG000001",\n        "shot_day": 11,\n        "width": 64,\n        "height": 64,\n        "channels": 3,\n        "brightness": 0.29,\n        "mud_ratio": 0.12\n      },\n      "spec": {\n        "equip_id": "VX-5081-9133",\n        "equip_type": "vehicle",\n        "mfg_year": 2012,\n        "op_hours": 6778.0,\n        "env": "coastal",\n        "reg_grade": 1\n      },\n      "label": {\n        "defect_class": "crack",\n        "bbox_x": 14,\n        "bbox_y": 14,\n        "bbox_w": 7,\n        "bbox_h": 29,\n        "emergency": false\n      },\n      "result": {\n        "area_frac": 0.145,\n        "depth_mm": 3.2,\n        "damage_idx": 27.7,\n        "grade": 1,\n        "confidence": 0.58\n      }\n    },\n    {\n      "photo": {\n        "img_id": "IMG000002",\n        "shot_day": 87,\n        "width": 64,\n        "height": 64,\n        "channels": 3,\n        "brightness": 0.56,\n        "mud_ratio": 0.03\n      },\n      "spec": {\n        "equip_id": "VX-4820-6610",\n        "equip_type": "generator",\n        "mfg_year": 2019,\n        "op_hours": 5556.0,\n        "env": "inland",\n        "reg_grade": 2\n      },\n      "label": {\n        "defect_class": "corrosion",\n        "bbox_x": 1,\n        "bbox_y": 21,\n        "bbox_w": 18,\n        "bbox_h": 26,\n        "emergency": false\n      },\n      "result": {\n        "area_frac": 0.128,\n        "depth_mm": 5.1,\n        "damage_idx": 59.3,\n        "grade": 1,\n        "confidence": 0.63\n      }\n    },\n    {\n      "photo": {\n        "img_id": "IMG000003",\n        "shot_day": 31,\n        "width": 64,\n        "height": 64,\n        "channels": 3,\n        "brightness": 0.56,\n        "mud_ratio": 0.15\n      },\n      "spec": {\n        "equip_id": "VX-8558-9827",\n        "equip_type": "trailer",\n        "mfg_year": 2001,\n        "op_hours": 7395.0,\n        "env": "coastal",\n        "reg_grade": 1\n      },\n      "label": {\n        "defect_class": "damage",\n        "bbox_x": 22,\n        "bbox_y": 23,\n        "bbox_w": 26,\n        "bbox_h": 9,\n        "emergency": false\n      },\n      "result": {\n        "area_frac": 0.038,\n        "depth_mm": 0.5,\n        "damage_idx": 85.5,\n        "grade": 2,\n        "confidence": 0.93\n      }\n    },\n    {\n      "photo": {\n        "img_id": "IMG000004",\n        "shot_day": 78,\n        "width": 64,\n        "height": 64,\n        "channels": 3,\n        "brightness": 0.53,\n        "mud_ratio": 0.08\n      },\n      "spec": {\n        "equip_id": "VX-9274-1063",\n        "equip_type": "trailer",\n        "mfg_year": 2016,\n        "op_hours": 6507.0,\n        "env": "coastal",\n        "reg_grade": 3\n      },\n      "label": {\n        "defect_class": "leak",\n        "bbox_x": 8,\n        "bbox_y": 14,\n        "bbox_w": 10,\n        "bbox_h": 22,\n        "emergency": false\n      },\n      "result": {\n        "area_frac": 0.193,\n        "depth_mm": 0.9,\n        "damage_idx": 48.2,\n        "grade": 3,\n        "confidence": 0.95\n      }\n    },\n    {\n      "photo": {\n        "img_id": "IMG000005",\n        "shot_day": 71,\n        "width": 64,\n        "height": 64,\n        "channels": 3,\n        "brightness": 0.61,\n        "mud_ratio": 0.01\n      },\n      "spec": {\n        "equip_id": "VX-4804-5407",\n        "equip_type": "comm",\n        "mfg_year": 2011,\n        "op_hours": 7459.0,\n        "env": "inland",\n        "reg_grade": 2\n      },\n      "label": {\n        "defect_class": "leak",\n        "bbox_x": 2,\n        "bbox_y": 19,\n        "bbox_w": 11,\n        "bbox_h": 11,\n        "emergency": false\n      },\n      "result": {\n        "area_frac": 0.042,\n        "depth_mm": 5.0,\n        "damage_idx": 6.3,\n        "grade": 2,\n        "confidence": 0.91\n      }\n    },\n    {\n      "photo": {\n        "img_id": "IMG000006",\n        "shot_day": 14,\n        "width": 64,\n        "height": 64,\n        "channels": 3,\n        "brightness": 0.46,\n        "mud_ratio": 0.1\n      },\n      "spec": {\n        "equip_id": "VX-1283-7221",\n        "equip_type": "comm",\n        "mfg_year": 2004,\n        "op_hours": 3627.0,\n        "env": "inland",\n        "reg_grade": 1\n      },\n      "label": {\n        "defect_class": "damage",\n        "bbox_x": 7,\n        "bbox_y": 16,\n        "bbox_w": 15,\n        "bbox_h": 20,\n        "emergency": false\n      },\n      "result": {\n        "area_frac": 0.076,\n        "depth_mm": 4.4,\n        "damage_idx": 65.4,\n        "grade": 2,\n        "confidence": 0.72\n      }\n    },\n    {\n      "photo": {\n        "img_id": "IMG000007",\n        "shot_day": 79,\n        "width": 64,\n        "height": 64,\n        "channels": 3,\n        "brightness": 0.64,\n        "mud_ratio": 0.24\n      },\n      "spec": {\n        "equip_id": "VX-8805-3412",\n        "equip_type": "generator",\n        "mfg_year": 2024,\n        "op_hours": 1847.0,\n        "env": "inland",\n        "reg_grade": 2\n      },\n      "label": {\n        "defect_class": "leak",\n        "bbox_x": 2,\n        "bbox_y": 7,\n        "bbox_w": 8,\n        "bbox_h": 11,\n        "emergency": false\n      },\n      "result": {\n        "area_frac": 0.052,\n        "depth_mm": 4.6,\n        "damage_idx": 69.8,\n        "grade": 1,\n        "confidence": 0.56\n      }\n    },\n    {\n      "photo": {\n        "img_id": "IMG000008",\n        "shot_day": 33,\n        "width": 64,\n        "height": 64,\n        "channels": 3,\n        "brightness": 0.49,\n        "mud_ratio": 0.2\n      },\n      "spec": {\n        "equip_id": "VX-6148-5102",\n        "equip_type": "vehicle",\n        "mfg_year": 2014,\n        "op_hours": 7573.0,\n        "env": "inland",\n        "reg_grade": 2\n      },\n      "label": {\n        "defect_class": "corrosion",\n        "bbox_x": 10,\n        "bbox_y": 8,\n        "bbox_w": 16,\n        "bbox_h": 19,\n        "emergency": false\n      },\n      "result": {\n        "area_frac": 0.041,\n        "depth_mm": 1.7,\n        "damage_idx": 31.4,\n        "grade": 1,\n        "confidence": 0.66\n      }\n    },\n    {\n      "photo": {\n        "img_id": "IMG000009",\n        "shot_day": 82,\n        "width": 64,\n        "height": 64,\n        "channels": 3,\n        "brightness": 0.88,\n        "mud_ratio": 0.23\n      },\n      "spec": {\n        "equip_id": "VX-6189-2417",\n        "equip_type": "trailer",\n        "mfg_year": 2017,\n        "op_hours": 5413.0,\n        "env": "inland",\n        "reg_grade": 2\n      },\n      "label": {\n        "defect_class": "leak",\n        "bbox_x": 29,\n        "bbox_y": 20,\n        "bbox_w": 13,\n        "bbox_h": 17,\n        "emergency": false\n      },\n      "result": {\n        "area_frac": 0.098,\n        "depth_mm": 1.3,\n        "damage_idx": 13.3,\n        "grade": 2,\n        "confidence": 0.75\n      }\n    }\n  ],\n  "label_rules": {\n    "crack": "선형 어두운 균열, 길이 10px 이상",\n    "corrosion": "붉은 갈색 면, 면적 12px 이상",\n    "damage": "함몰·파단 윤곽(밝은 테두리 포함)",\n    "leak": "어둡고 번지는 액체 자국(아래 방향 흐름)",\n    "none": "결함 없음(무해한 얼룩·긁힘은 결함으로 라벨하지 않음)"\n  }\n}\n', 'src/dashboard/qlt03_app.py': '"""QLT03 M8 로컬 대시보드. 교육용 합성자료만 사용하며 외부 통신은 하지 않는다."""\nfrom pathlib import Path\nimport contextlib\nimport io\nimport runpy\nimport time\n\nPROJECT = Path(__file__).resolve().parents[2]\n\n\ndef load_results(source=None):\n    """원문 시뮬레이터를 그대로 실행하여 동일한 표와 실측 시간을 반환한다."""\n    import matplotlib\n    matplotlib.use("Agg")\n    import matplotlib.pyplot as plt\n    source = Path(source) if source is not None else PROJECT / "src/m08.py"\n    log = io.StringIO()\n    previous_show = plt.show\n    plt.show = lambda *args, **kwargs: plt.close("all")\n    started = time.perf_counter()\n    try:\n        with contextlib.redirect_stdout(log):\n            result = runpy.run_path(str(source), run_name="__main__")\n    finally:\n        plt.show = previous_show\n        plt.close("all")\n    return {"sweep": result["sw"], "conditions": result["cc"],\n            "loading": result["loads"], "stdout": log.getvalue(),\n            "elapsed_seconds": time.perf_counter() - started}\n\n\ndef main():\n    import pandas as pd\n    import streamlit as st\n    st.set_page_config(page_title="QLT03 합성 외관검사", layout="wide")\n    st.title("QLT03 M8 합성 외관검사 대시보드")\n    st.caption("[교육용가상자료] 원문 M8 시뮬레이터 결과. 현장 성능·실제 장비 판정이 아닙니다.")\n    st.info("자동 처리 후보도 검사관 확인이 필요합니다. 웹앱 실행과 외부 배포는 별개입니다.")\n    cached_results = st.cache_data(show_spinner=False)(load_results)\n    if st.button("합성자료 시험 실행", key="run_simulator"):\n        with st.spinner("고정 seed 합성사진 생성·모델 학습·스트레스 시험 실행 중"):\n            try:\n                st.session_state["qlt03_results"] = cached_results()\n            except Exception as error:\n                st.error(f"실행 실패: {error}. M8 노트북/콘솔 대체 경로와 환경 안내를 확인하세요.")\n                return\n    result = st.session_state.get("qlt03_results")\n    if result is None:\n        st.write("버튼을 눌러 결과를 생성하세요. 최초 실행에는 모델 학습 시간이 포함됩니다.")\n        return\n    sweep = result["sweep"]\n    selected = st.selectbox("진흙 혼입 비율", sweep["진흙 비율"].tolist())\n    row = sweep.loc[sweep["진흙 비율"] == selected].iloc[0]\n    st.metric("시험 사진 수", f"{int(row[\'장수\'])}장")\n    st.metric("자동 처리 후보 비율", f"{row[\'자동 처리 비율\']:.1%}")\n    accuracy = row["경보 정확도(게이트 후)"]\n    st.metric("후보 내 경보 정확도", "평가 불가 (후보 0건)" if pd.isna(accuracy) else f"{accuracy:.1%}")\n    st.caption("경보 정확도의 분모는 게이트 통과 후보입니다. 후보 0건은 100%가 아닙니다.")\n    st.dataframe(sweep)\n    st.line_chart(sweep.set_index("진흙 비율")[["자동 처리 비율", "경보 정확도(게이트 후)"]])\n    st.subheader("복합 열화 조건")\n    st.dataframe(result["conditions"])\n    st.subheader("시뮬레이터 집계·차트 렌더링 시간")\n    st.dataframe(pd.DataFrame(result["loading"], columns=["건수", "초"]))\n    st.caption(f"이번 시뮬레이터 전체 실행 {result[\'elapsed_seconds\']:.2f}초. 위 표의 시간은 웹 브라우저 응답 시간이 아닙니다.")\n    st.download_button("현재 시험 표 CSV 저장", sweep.to_csv(index=False).encode("utf-8-sig"),\n                       file_name="qlt03_m08_sweep.csv", mime="text/csv")\n    st.download_button("실행 로그 저장", result["stdout"].encode("utf-8"),\n                       file_name="qlt03_m08_app_stdout.txt", mime="text/plain")\n    st.text_area("이번 실행 로그", result["stdout"], height=260)\n\n\nif __name__ == "__main__":\n    main()\n', 'requirements-dashboard.txt': '# 선택 로컬 웹앱 환경: 기본 노트북에 설치 불필요\n-r requirements.txt\nstreamlit>=1.40,<2\n', 'docs/expected_outputs/m09_revised_expected.txt': '수정본 채점 기준: 연간 순절감액 2,751만원, 판독 시간 단축률 0.3%, 투자 회수 10.9개월.\n반올림 전 2,751.111111만원/년·0.292475%·10.904685개월. 기존 AI→개선 AI, 동일 B 360장, 연 20,000장, 시급 1.2만원, 증분 운영비 150만원/년, 증분 투자비 2,500만원이라는 교육용 가정이다. outputs/m09/revised/stdout.txt와 전후 총비용 차이를 확인한다. 환경에 따른 차이는 실제 로그와 산식으로 설명하며 예시 숫자를 맞추려고 출력을 고치지 않는다.\n', 'docs/expected_outputs/README.md': '# 기대 출력과 적용 기준\n\nM9 현재 채점 기준은 `m09_revised_expected.txt`입니다. `m09_expected.txt`는 원문 보존 적용 기준 부록이며 현재 채점값이 아닙니다. 원문 값과 수정 이유는 `../M09_ROI_정오표.md`를 읽습니다. 다른 원문 기대 출력도 실측을 대신하지 않습니다.\n\n- m09_expected.txt: 원문 예시(역사 비교용). 현재 비교는 m09_revised_expected.txt 및 적용 기준을 사용합니다.\n', 'src/etl/README.md': '# 비교용 예제 안내\n\n이 폴더의 Python 파일은 원문 이름을 보존한 비교용 사본입니다. 현재 실습·제출 파일은 docs/원문_제출경로_연결표.md의 현재 실행 열을 사용합니다. 원문 사본을 현재 실습 코드로 제출하지 않습니다.\n', 'src/optimization/README.md': '# 비교용 예제 안내\n\n이 폴더의 Python 파일은 원문 이름을 보존한 비교용 사본입니다. 현재 실습·제출 파일은 docs/원문_제출경로_연결표.md의 현재 실행 열을 사용합니다. 원문 사본을 현재 실습 코드로 제출하지 않습니다.\n', 'src/features/README.md': '# 실행 파일과 원문 비교 사본\n\n현재 실습·제출 파일은 docs/원문_제출경로_연결표.md의 현재 실행 열을 확인합니다. 원문 이름의 Python 사본은 비교용입니다. 대시보드 앱은 M08_웹앱과_대체경로.md의 실행 명령을 사용합니다.\n', 'src/models/README.md': '# 비교용 예제 안내\n\n이 폴더의 Python 파일은 원문 이름을 보존한 비교용 사본입니다. 현재 실습·제출 파일은 docs/원문_제출경로_연결표.md의 현재 실행 열을 사용합니다. 원문 사본을 현재 실습 코드로 제출하지 않습니다.\n', 'src/explainability/README.md': '# 비교용 예제 안내\n\n이 폴더의 Python 파일은 원문 이름을 보존한 비교용 사본입니다. 현재 실습·제출 파일은 docs/원문_제출경로_연결표.md의 현재 실행 열을 사용합니다. 원문 사본을 현재 실습 코드로 제출하지 않습니다.\n', 'src/dashboard/README.md': '# 실행 파일과 원문 비교 사본\n\n현재 실습·제출 파일은 docs/원문_제출경로_연결표.md의 현재 실행 열을 확인합니다. 원문 이름의 Python 사본은 비교용입니다. 대시보드 앱은 M08_웹앱과_대체경로.md의 실행 명령을 사용합니다.\n', 'pipeline/README.md': '# 비교용 예제 안내\n\n이 폴더의 Python 파일은 원문 이름을 보존한 비교용 사본입니다. 현재 실습·제출 파일은 docs/원문_제출경로_연결표.md의 현재 실행 열을 사용합니다. 원문 사본을 현재 실습 코드로 제출하지 않습니다.\n', '02_데이터셋/M05_기록양식.md': '# QLT03 미션 5 수행 기록 [교육용가상자료]\n\n원문 산출물: src/models/qlt03_advanced_model.py\n실습가이드 제출물: 성능 비교표, 혼동행렬, 신뢰도 보정표, 상황별 성능표\n이 문서는 학습자가 작성하는 양식입니다. 빈 양식은 완료 산출물이 아닙니다.\n\n## 기본 실행\n- 실행 환경/일시: (작성)\n- 실행한 코드와 출력 파일: outputs/m05/baseline/stdout.txt\n- 확인할 결과(원문 예시):\nMAE 개선 73.4%, RMSE 개선 55.1%(목표 20% 이상), R2 0.946(목표 0.85 이상)\n등급 일치율 93.3%, 신뢰도 80 이상 234장(78%) 일치율 100%, 진흙 가림 사진은 80 이상 0%\n- 직접 관찰한 지표/단위/분모: (작성)\n- 원문 예시와 차이 및 원문 Q1 달성 여부: (작성)\n- 표/그래프에서 발견한 근거: (작성)\n\n## 도전 과제\n자동 통과 기준 신뢰도를 80에서 70으로 낮추면 자동 통과 비율과 일치율이 어떻게 달라지는지 확인한다.\n- 변경 전/후 코드 또는 설명형 과제의 근거: (작성)\n- **확인할 것**: 자동 통과 후보 수·비율·실제 일치율 · 저조도·진흙 조건의 문턱 이상 비율 · 등급 혼동행렬의 과소 판정 수 · MAE·RMSE·R2\n- 변경 전/후 결과와 단위: (작성)\n- 결과가 같으면 변경이 적용된 증거와 같은 이유: (작성)\n- **생각해 볼 질문**: 신뢰도 80→70 하향 후 후보 비율과 실제 일치율은 어떻게 달라졌는가? 혼동행렬과 진흙 조건을 근거로 범위를 넓히는 대가를 설명하고 자동 통과가 최종 정비 등급 확정을 뜻하는지 구분한다.\n- 내 답과 적용 한계: (작성)\n\n## AI 활용 및 최종 산출물\n- 사용한 AI/질문/응답 중 판단에 사용한 부분/직접 확인한 점: (작성, 미사용이면 미사용 이유)\n- 원문이 요구한 산출물 본문 또는 파일 연결: (작성)\n- 제출할 최종 경로·선택본(기본/도전)·선택 이유: (작성; 원문_제출경로_연결표.md 참조)\n- 최종 판단을 담당하는 가상 역할/사람 확인 절차: (작성)\n', '02_데이터셋/M10_기록양식.md': '# QLT03 미션 10 수행 기록 [교육용가상자료]\n\n원문 산출물: docs/qlt03_operations_guide.md 및 pipeline/qlt03_retrain.py\n실습가이드 제출물: 통합 파이프라인 코드, 처리 구분 표, 자체 점검표, 인터페이스 명세, 브리핑 요약\n이 문서는 학습자가 작성하는 양식입니다. 빈 양식은 완료 산출물이 아닙니다.\n\n## 기본 실행\n- 실행 환경/일시: (작성)\n- 원문 재현 이력: outputs/m10/baseline/stdout.txt (최종 성능·경제성 기준 아님)\n- 수정 기본 실행: outputs/m10/revised/stdout.txt (최종 보고·도전 비교 기준)\n- 확인할 결과(실습 코드 채점 기준):\n실습 코드 최종시험 MAE 6.43, 등급 정확도 87.5%(시험 240장). 빈 운영지침은 0/12이며 문서 미완료 점검은 FAIL이 정상이다. 직접 작성 후 완료 수와 자체 점검 결과를 다시 기록한다.\n처리 구분 간이 승인 후보 112·정밀 판독 86·재촬영 42장, 자동 확정 0건\n- 직접 관찰한 지표/단위/분모: (작성)\n- 원문 예시와 차이 및 원문 Q1 달성 여부: (작성)\n- 표/그래프에서 발견한 근거: (작성)\n\n## 도전 과제\n코드의 승격 기준 0.03(MAE 3% 개선)을 0.15로 높여 승격 판정이 어떻게 달라지는지 확인한다.\n- 변경 전/후 코드 또는 설명형 과제의 근거: (작성)\n- **확인할 것**: 챔피언·챌린저 검증 MAE와 개선율 · 승격 판정 · 간이 승인 후보·정밀 판독·재촬영 수 · 자동 확정 수·자체 점검·문서 완비\n- 변경 전/후 결과와 단위: (작성)\n- 결과가 같으면 변경이 적용된 증거와 같은 이유: (작성)\n- **생각해 볼 질문**: 검증 MAE로 계산한 개선율은 15%를 넘는가? 문턱 변경 후 승격 판정과 처리 구분이 같더라도 오류라고 할 수 없는 이유를 적고 세 처리 구분이 신규 240장과 맞는지 검산한다.\n- 내 답과 적용 한계: (작성)\n\n## AI 활용 및 최종 산출물\n- 사용한 AI/질문/응답 중 판단에 사용한 부분/직접 확인한 점: (작성, 미사용이면 미사용 이유)\n- 원문이 요구한 산출물 본문 또는 파일 연결: (작성)\n- 제출할 최종 경로·선택본(기본/도전)·선택 이유: (작성; 원문_제출경로_연결표.md 참조)\n- 최종 판단을 담당하는 가상 역할/사람 확인 절차: (작성)\n', '02_데이터셋/M02_기록양식.md': '# QLT03 미션 2 수행 기록 [교육용가상자료]\n\n원문 산출물: src/etl/qlt03_data_cleansing.py 및 정제 품질 보고서\n실습가이드 제출물: 정제 모듈 코드, 정제 규칙 로그, 정합성 검증 결과, 전후 통계 비교표\n이 문서는 학습자가 작성하는 양식입니다. 빈 양식은 완료 산출물이 아닙니다.\n\n## 기본 실행\n- 실행 환경/일시: (작성)\n- 실행한 코드와 출력 파일: outputs/m02/baseline/stdout.txt\n- 확인할 결과(원문 예시):\n정제 규칙 11종(목표 10종 이상), 정합성 검증 12/12 통과(100%)\n정제 후 사진 150장, 긴급 점검 6장 유지, 정제 박스 평균 IoU 0.99\n- 직접 관찰한 지표/단위/분모: (작성)\n- 원문 예시와 차이 및 원문 Q1 달성 여부: (작성)\n- 표/그래프에서 발견한 근거: (작성)\n\n## 도전 과제\nR7 조명 보정 기준(평균 밝기 0.3)을 0.4로 바꿔 보정 대상 사진 수가 어떻게 달라지는지 확인한다.\n- 변경 전/후 코드 또는 설명형 과제의 근거: (작성)\n- **확인할 것**: R7 문턱 미만 사진 수 · 정합성 12항목 · 정제 후 사진·긴급 점검 보존 수 · 전후 평균 밝기와 박스 IoU\n- 변경 전/후 결과와 단위: (작성)\n- 결과가 같으면 변경이 적용된 증거와 같은 이유: (작성)\n- **생각해 볼 질문**: 밝기 0.3→0.4 변경 후 R7의 집계 건수와 정제 후 평균 밝기는 각각 어떻게 달라졌는가? 현재 구현이 모든 사진을 정규화한다는 점을 코드에서 찾아 집계 문턱과 실제 적용 조건을 구분한다.\n- 내 답과 적용 한계: (작성)\n\n## AI 활용 및 최종 산출물\n- 사용한 AI/질문/응답 중 판단에 사용한 부분/직접 확인한 점: (작성, 미사용이면 미사용 이유)\n- 원문이 요구한 산출물 본문 또는 파일 연결: (작성)\n- 제출할 최종 경로·선택본(기본/도전)·선택 이유: (작성; 원문_제출경로_연결표.md 참조)\n- 최종 판단을 담당하는 가상 역할/사람 확인 절차: (작성)\n', '02_데이터셋/M03_기록양식.md': '# QLT03 미션 3 수행 기록 [교육용가상자료]\n\n원문 산출물: notebooks/qlt03_eda_analysis.ipynb 및 EDA 요약 리포트\n실습가이드 제출물: Baseline 성능표, EDA 차트 8종, 상관표, 제원 불일치 표, 파레토 결과\n이 문서는 학습자가 작성하는 양식입니다. 빈 양식은 완료 산출물이 아닙니다.\n\n## 기본 실행\n- 실행 환경/일시: (작성)\n- 실행한 코드와 출력 파일: outputs/m03/baseline/stdout.txt\n- 확인할 결과(원문 예시):\nBaseline 정확도 86.0%·재현율 92.2%, 면적 2% 미만 작은 결함 재현율 74%\n유의 피처 7개(목표 5개 이상), EDA 차트 8종, 제원 불일치 후보 47대\n- 직접 관찰한 지표/단위/분모: (작성)\n- 원문 예시와 차이 및 원문 Q1 달성 여부: (작성)\n- 표/그래프에서 발견한 근거: (작성)\n\n## 도전 과제\nBaseline 로지스틱 회귀의 규제 강도 C를 0.5에서 5로 바꿔 성능 변화를 확인하고 한계가 해소되는지 설명한다.\n- 변경 전/후 코드 또는 설명형 과제의 근거: (작성)\n- **확인할 것**: 정확도·정밀도·재현율·F1 · 작은 결함 재현율 · 저조도와 밝은 사진 정확도 · 제원 불일치 후보 수\n- 변경 전/후 결과와 단위: (작성)\n- 결과가 같으면 변경이 적용된 증거와 같은 이유: (작성)\n- **생각해 볼 질문**: C를 0.5에서 5로 바꿨을 때 전체 성능과 작은 결함·저조도 성능은 함께 좋아졌는가? C가 커질수록 규제가 약해지는 점과 위치·모양 정보를 쓰지 않는 입력의 한계를 연결해 설명한다.\n- 내 답과 적용 한계: (작성)\n\n## AI 활용 및 최종 산출물\n- 사용한 AI/질문/응답 중 판단에 사용한 부분/직접 확인한 점: (작성, 미사용이면 미사용 이유)\n- 원문이 요구한 산출물 본문 또는 파일 연결: (작성)\n- 제출할 최종 경로·선택본(기본/도전)·선택 이유: (작성; 원문_제출경로_연결표.md 참조)\n- 최종 판단을 담당하는 가상 역할/사람 확인 절차: (작성)\n', '02_데이터셋/M04_기록양식.md': '# QLT03 미션 4 수행 기록 [교육용가상자료]\n\n원문 산출물: src/features/qlt03_feature_engineering.py 및 기준 모델\n실습가이드 제출물: 분할 성능표, 면적 보정 결과, 깊이·손상도 오차표, 변수군 제거 결과\n이 문서는 학습자가 작성하는 양식입니다. 빈 양식은 완료 산출물이 아닙니다.\n\n## 기본 실행\n- 실행 환경/일시: (작성)\n- 실행한 코드와 출력 파일: outputs/m04/baseline/stdout.txt\n- 확인할 결과(원문 예시):\n분할 IoU 균열 0.89·부식 0.74·파손 0.61·누유 0.94, 면적 MAPE 20.8% -> 10.2%\n손상도 MAPE 8.8%(목표 25% 이하, 기준 30.8%), 깊이 MAPE 28.0%\n- 직접 관찰한 지표/단위/분모: (작성)\n- 원문 예시와 차이 및 원문 Q1 달성 여부: (작성)\n- 표/그래프에서 발견한 근거: (작성)\n\n## 도전 과제\n깊이 추정 MAPE가 손상도보다 큰 이유를 사진 한 장으로는 깊이가 직접 보이지 않는 점과 연결해 설명한다.\n- 변경 전/후 코드 또는 설명형 과제의 근거: (작성)\n- **확인할 것**: 유형별 IoU·Dice · 면적 원측정·보정 MAPE · 깊이 MAPE와 MAE · 손상도 기준·제안 MAPE · 변수군 제거 시 오차 증가\n- 변경 전/후 결과와 단위: (작성)\n- 결과가 같으면 변경이 적용된 증거와 같은 이유: (작성)\n- **생각해 볼 질문**: 기본 실행에서 깊이 MAPE와 손상도 MAPE를 각각 인용하고, 한 장의 사진으로 깊이를 직접 관측할 수 없다는 점을 들어 차이를 설명한다. 임계값 분할과 실제 Mask R-CNN 학습을 구분하고 깊이 게이지 등 추가 확인 방법을 한 가지 제시한다.\n- 내 답과 적용 한계: (작성)\n\n## AI 활용 및 최종 산출물\n- 사용한 AI/질문/응답 중 판단에 사용한 부분/직접 확인한 점: (작성, 미사용이면 미사용 이유)\n- 원문이 요구한 산출물 본문 또는 파일 연결: (작성)\n- 제출할 최종 경로·선택본(기본/도전)·선택 이유: (작성; 원문_제출경로_연결표.md 참조)\n- 최종 판단을 담당하는 가상 역할/사람 확인 절차: (작성)\n', '02_데이터셋/M06_기록양식.md': '# QLT03 미션 6 수행 기록 [교육용가상자료]\n\n원문 산출물: src/explainability/qlt03_xai_explainer.py 및 XAI 리포트\n실습가이드 제출물: 업로드 판정표, 시각화 그래프, 기여도 표, 자연어 설명\n이 문서는 학습자가 작성하는 양식입니다. 빈 양식은 완료 산출물이 아닙니다.\n\n## 기본 실행\n- 실행 환경/일시: (작성)\n- 실행한 코드와 출력 파일: outputs/m06/baseline/stdout.txt\n- 확인할 결과(원문 예시):\n기여도 상위 3개 자동 산출 300건 중 300건(100%)\n업로드 8장 중 유형 판정 7장 일치, 손상 부위 IoU 0.64~0.98, 저조도 사진은 재촬영 요구\n- 직접 관찰한 지표/단위/분모: (작성)\n- 원문 예시와 차이 및 원문 Q1 달성 여부: (작성)\n- 표/그래프에서 발견한 근거: (작성)\n\n## 도전 과제\n재촬영 기준 mud 0.03을 0.01로 바꿔 재촬영 요구 사진이 어떻게 달라지는지 확인한다.\n- 변경 전/후 코드 또는 설명형 과제의 근거: (작성)\n- **확인할 것**: U01~U08의 오염 추정치·처리 상태 · 재촬영 요구 사진 목록 · 기여도 상위 3개 산출 건수 · 표시 부위 IoU\n- 변경 전/후 결과와 단위: (작성)\n- 결과가 같으면 변경이 적용된 증거와 같은 이유: (작성)\n- **생각해 볼 질문**: mud 기준을 0.03에서 0.01로 낮추면 어떤 사진이 새로 재촬영 대상으로 바뀌는가? 바뀌지 않았다면 8장의 mud 값과 저조도 조건을 근거로 이유를 설명하고 판정 기여도의 부호를 해석한다.\n- 내 답과 적용 한계: (작성)\n\n## AI 활용 및 최종 산출물\n- 사용한 AI/질문/응답 중 판단에 사용한 부분/직접 확인한 점: (작성, 미사용이면 미사용 이유)\n- 원문이 요구한 산출물 본문 또는 파일 연결: (작성)\n- 제출할 최종 경로·선택본(기본/도전)·선택 이유: (작성; 원문_제출경로_연결표.md 참조)\n- 최종 판단을 담당하는 가상 역할/사람 확인 절차: (작성)\n', '02_데이터셋/M09_기록양식.md': '# QLT03 미션 9 수행 기록 [교육용가상자료]\n\n원문 산출물: reports/qlt03_roi_analysis.md\n실습가이드 제출물: 재보정 코드, 원인별 오판표, 전후 비교표, 3대 지표와 민감도\n이 문서는 학습자가 작성하는 양식입니다. 빈 양식은 완료 산출물이 아닙니다.\n\n## 기본 실행\n- 실행 환경/일시: (작성)\n- 원문 재현 이력: outputs/m09/baseline/stdout.txt (최종 성능·경제성 기준 아님)\n- 수정 기본 실행: outputs/m09/revised/stdout.txt (최종 보고·도전 비교 기준)\n- 확인할 결과(실습 코드 채점 기준):\n실습 코드 채점 기준: 연간 순절감액 2,751만원, 판독 시간 단축률 0.3%, 투자 회수 10.9개월.\n반올림 전 2,751.111111만원/년·0.292475%·10.904685개월. 기존 AI→개선 AI, 동일 B 360장, 연 20,000장, 시급 1.2만원, 증분 운영비 150만원/년, 증분 투자비 2,500만원이라는 교육용 가정이다. outputs/m09/revised/stdout.txt와 전후 총비용 차이를 확인한다. 환경에 따른 차이는 실제 로그와 산식으로 설명하며 예시 숫자를 맞추려고 출력을 고치지 않는다.\n- 직접 관찰한 지표/단위/분모: (작성)\n- 실습 코드 기준과 차이 및 원문 Q1 달성 여부: (작성)\n- 표/그래프에서 발견한 근거: (작성)\n\n## 도전 과제\n연 판정 사진 수 N_YR을 20000장에서 10000장으로 바꿔 절감액과 회수 기간이 어떻게 달라지는지 확인한다.\n- 변경 전/후 코드 또는 설명형 과제의 근거: (작성)\n- **확인할 것**: 보류 구간 B의 등급 일치율·오판 손실 · 연간 비용 절감액과 고정 운영비 · 판독 시간 단축률 · 투자 회수 기간\n- 변경 전/후 결과와 단위: (작성)\n- 결과가 같으면 변경이 적용된 증거와 같은 이유: (작성)\n- **생각해 볼 질문**: 연 사진 수 20000→10000 변경 후 연 절감액이 정확히 절반이 되는가? 고정 운영비 150만원을 포함한 산식으로 확인하고 보류 구간의 모델 성능과 시간 단축률이 유지되는 이유를 설명한다.\n- 내 답과 적용 한계: (작성)\n\n## AI 활용 및 최종 산출물\n- 사용한 AI/질문/응답 중 판단에 사용한 부분/직접 확인한 점: (작성, 미사용이면 미사용 이유)\n- 원문이 요구한 산출물 본문 또는 파일 연결: (작성)\n- 제출할 최종 경로·선택본(기본/도전)·선택 이유: (작성; 원문_제출경로_연결표.md 참조)\n- 최종 판단을 담당하는 가상 역할/사람 확인 절차: (작성)\n', '02_데이터셋/M01_기록양식.md': "# QLT03 미션 1 수행 기록 [교육용가상자료]\n\n원문 산출물: docs/qlt03_logistics_domain_spec.md 및 데이터 딕셔너리\n실습가이드 제출물: 데이터 사전, 라벨링 기준, 업무 프로세스 맵\n이 문서는 학습자가 작성하는 양식입니다. 빈 양식은 완료 산출물이 아닙니다.\n\n## 기본 실행\n- 실행 환경/일시: (작성)\n- 실행한 코드와 출력 파일: outputs/m01/baseline/stdout.txt\n- 확인할 결과(원문 예시):\n데이터 사전 피처 24개(목표 20개 이상), 정상 샘플 20건 검증 오류 0건\n예측 대상 4종(결함 면적, 깊이, 손상도, 정비 등급)\n- 직접 관찰한 지표/단위/분모: (작성)\n- 원문 예시와 차이 및 원문 Q1 달성 여부: (작성)\n- 표/그래프에서 발견한 근거: (작성)\n\n## 도전 과제\n데이터 사전에 '촬영 거리' 같은 필드를 하나 추가하고 실제 데이터와 대조해 누락으로 표시되는지 확인한다.\n- 변경 전/후 코드 또는 설명형 과제의 근거: (작성)\n- **확인할 것**: 데이터 딕셔너리 피처 수 · 정상 샘플 검증 통과·오류 건수 · photo.shot_distance_m 누락 메시지 · 검사관 보정·등급 승인 단계\n- 변경 전/후 결과와 단위: (작성)\n- 결과가 같으면 변경이 적용된 증거와 같은 이유: (작성)\n- **생각해 볼 질문**: 촬영 거리 필드를 스키마에만 추가하면 기존 정상 샘플의 검증 결과가 어떻게 바뀌는가? 가짜 측정값으로 누락을 채우는 대신 스키마와 수집 기록을 어떻게 일치시킬지 설명한다.\n- 내 답과 적용 한계: (작성)\n\n## AI 활용 및 최종 산출물\n- 사용한 AI/질문/응답 중 판단에 사용한 부분/직접 확인한 점: (작성, 미사용이면 미사용 이유)\n- 원문이 요구한 산출물 본문 또는 파일 연결: (작성)\n- 제출할 최종 경로·선택본(기본/도전)·선택 이유: (작성; 원문_제출경로_연결표.md 참조)\n- 최종 판단을 담당하는 가상 역할/사람 확인 절차: (작성)\n", '02_데이터셋/M08_기록양식.md': "# QLT03 미션 8 수행 기록 [교육용가상자료]\n\n원문 산출물: src/dashboard/qlt03_app.py (Streamlit/Dash 등)\n실습가이드 제출물: 진흙 스윕 표, 복합 열화 표, 처리 시간·로딩 시간, 한계치 요약\n이 문서는 학습자가 작성하는 양식입니다. 빈 양식은 완료 산출물이 아닙니다.\n\n## 기본 실행\n- 실행 환경/일시: (작성)\n- 실행 경로(앱/시뮬레이터): (작성)\n- 앱 증거: 내려받은 qlt03_m08_app_stdout.txt·qlt03_m08_sweep.csv와 화면 캡처 경로: (작성)\n- 시뮬레이터 증거: outputs/m08/baseline/stdout.txt 및 figure 파일: (작성)\n- 실제 사용한 경로의 증거를 연결한다. 앱 로그 파일을 baseline 로그로 가장하지 않는다.\n- 확인할 결과(원문 예시):\n진흙 0%: 자동 처리 75%·경보 정확도 100%, 5%부터 93%로 95% 미달\n시뮬레이터의 요약 집계·목록·차트 렌더링 예시 시간은 0.1초 안팎. 브라우저를 포함한 웹앱 응답은 별도 측정하며 이 값으로 2초 목표 달성을 판정하지 않음, 흐림 사진 자동 처리 경보 정확도 91%\n- 직접 관찰한 지표/단위/분모: (작성)\n- 원문 예시와 차이 및 원문 Q1 달성 여부: (작성)\n- 표/그래프에서 발견한 근거: (작성)\n\n## 도전 과제\n경보를 등급 2 이상에서 등급 3(3단계 정비)만으로 바꿔 경보 정확도와 자동 처리 비율이 어떻게 달라지는지 확인한다.\n- 변경 전/후 코드 또는 설명형 과제의 근거: (작성)\n- **확인할 것**: 진흙 비율별 경보 정확도 · 품질 게이트 통과 자동 처리 비율 · 흐림·잡음·저조도 복합 조건 · 처리시간 실측과 대량 처리 외삽\n- 변경 전/후 결과와 단위: (작성)\n- 결과가 같으면 변경이 적용된 증거와 같은 이유: (작성)\n- **생각해 볼 질문**: 경보 정의만 등급 3으로 바꿨을 때 자동 처리 비율이 그대로인 이유는 무엇인가? 게이트 통과 사진을 분모로 삼아 경보 정확도를 비교하고 자동 처리 0건의 '-'를 100%로 해석할 수 없는 이유를 적는다.\n- 내 답과 적용 한계: (작성)\n\n## AI 활용 및 최종 산출물\n- 사용한 AI/질문/응답 중 판단에 사용한 부분/직접 확인한 점: (작성, 미사용이면 미사용 이유)\n- 원문이 요구한 산출물 본문 또는 파일 연결: (작성)\n- 제출할 최종 경로·선택본(기본/도전)·선택 이유: (작성; 원문_제출경로_연결표.md 참조)\n- 최종 판단을 담당하는 가상 역할/사람 확인 절차: (작성)\n", '02_데이터셋/M07_기록양식.md': '# QLT03 미션 7 수행 기록 [교육용가상자료]\n\n원문 산출물: src/optimization/qlt03_simulator.py\n실습가이드 제출물: 최적화 코드, 비용 비교표, 민감도, 결재 이력, 차단 로그\n이 문서는 학습자가 작성하는 양식입니다. 빈 양식은 완료 산출물이 아닙니다.\n\n## 기본 실행\n- 실행 환경/일시: (작성)\n- 실행한 코드와 출력 파일: outputs/m07/baseline/stdout.txt\n- 확인할 결과(원문 예시):\n평상 절감 59.5%·저조도 28.2%(목표 15% 이상), 진흙 20%는 -7.9%로 미달\n자동 확정 0건, 권한 우회 차단 10/10(100%), 감사 로그 25건 위·변조 검출\n- 직접 관찰한 지표/단위/분모: (작성)\n- 원문 예시와 차이 및 원문 Q1 달성 여부: (작성)\n- 표/그래프에서 발견한 근거: (작성)\n\n## 도전 과제\n검사관 오류율 P_HUMAN을 8%에서 3%로 낮춰 관행 비용과 절감률이 어떻게 달라지는지 확인한다.\n- 변경 전/후 코드 또는 설명형 과제의 근거: (작성)\n- **확인할 것**: 관행·HITL 비용과 절감률 · 평상·저조도·진흙 시나리오 · 자동 확정 건수 · 권한 우회 차단과 감사 로그\n- 변경 전/후 결과와 단위: (작성)\n- 결과가 같으면 변경이 적용된 증거와 같은 이유: (작성)\n- **생각해 볼 질문**: 검사관 오류율을 8%에서 3%로 낮추면 비교 기준인 전수 정밀 판독 비용과 HITL 절감률은 어떻게 달라지는가? 진흙 시나리오의 미달을 숨기지 않고 어느 조건에서 경제성이 성립하는지 설명한다.\n- 내 답과 적용 한계: (작성)\n\n## AI 활용 및 최종 산출물\n- 사용한 AI/질문/응답 중 판단에 사용한 부분/직접 확인한 점: (작성, 미사용이면 미사용 이유)\n- 원문이 요구한 산출물 본문 또는 파일 연결: (작성)\n- 제출할 최종 경로·선택본(기본/도전)·선택 이유: (작성; 원문_제출경로_연결표.md 참조)\n- 최종 판단을 담당하는 가상 역할/사람 확인 절차: (작성)\n', '02_데이터셋/README.md': '# QLT03 데이터·기록양식\n\n[교육용가상자료] 난수로 그린 64×64 RGB 사진의 균열·부식·파손·누유·정상 유형과 가상 장비 제원만 사용한다. 임계값 분할은 Mask R-CNN의 교육용 대체이며 실제 장비 사진이나 현장 검사 성능을 뜻하지 않는다.\n\n코드가 seed 고정 난수로 데이터를 생성합니다. 샘플 파일은 미리보기이며 전체 실행 입력으로 대신 쓰지 않습니다. 기본 예제·기대 출력은 학습자에게 제공하는 자료입니다. 도전 과제의 변경 내용과 해석은 직접 작성하세요.\n\n미션 1~미션 10 기록양식에 관찰 수치, 변경 내용, 결과가 같은 경우의 이유, 원문 산출물을 기록하세요.\n', '03_참고자료/원문_제출경로_연결표.md': '# 제출 경로 연결표\n\n현재 실행 열의 파일이 기본 예제 코드입니다. 원문 이름의 Python 파일은 비교용 예제이며 제출 기본값이 아닙니다. 원문 파일을 덮지 말고 현재 실행 파일을 제출하고 아래 산출물 이름과 연결하세요.\n\n| 미션 | 원문 산출물 | 기본 예제(현재 실행) | 도전 작업 | 비교용 예제 |\n|---|---|---|---|---|\n| 미션 1 | docs/qlt03_logistics_domain_spec.md 및 데이터 딕셔너리 | src/m01.py | work/m01_challenge.py | src/m01.py |\n| 미션 2 | src/etl/qlt03_data_cleansing.py 및 정제 품질 보고서 | src/m02.py | work/m02_challenge.py | src/m02.py |\n| 미션 3 | notebooks/qlt03_eda_analysis.ipynb 및 EDA 요약 리포트 | src/m03.py | work/m03_challenge.py | src/m03.py |\n| 미션 4 | src/features/qlt03_feature_engineering.py 및 기준 모델 | src/m04.py | 설명 기록 | src/m04.py |\n| 미션 5 | src/models/qlt03_advanced_model.py | src/m05.py | work/m05_challenge.py | src/m05.py |\n| 미션 6 | src/explainability/qlt03_xai_explainer.py 및 XAI 리포트 | src/m06.py | work/m06_challenge.py | src/m06.py |\n| 미션 7 | src/optimization/qlt03_simulator.py | src/m07.py | work/m07_challenge.py | src/m07.py |\n| 미션 8 | src/dashboard/qlt03_app.py (Streamlit/Dash 등) | src/m08.py | work/m08_challenge.py | src/m08.py |\n| 미션 9 | reports/qlt03_roi_analysis.md | work/revised/m09.py | work/m09_challenge.py | src/m09.py |\n| 미션 10 | docs/qlt03_operations_guide.md 및 pipeline/qlt03_retrain.py | work/revised/m10.py | work/m10_challenge.py | src/m10.py |\n', '03_참고자료/학습시간과_선택미션.md': '# 학습시간과 선택 미션\n\n초보 소요시간은 원문의 약 1.3~1.7배를 준비 계획용 추정 범위로 참고하세요. 파일 조작·개념 학습·대기·재시도에 따라 달라지며 실제 학습자 실측이나 확정 시간이 아닙니다.\n\n선택 미션 4·미션 6·미션 8은 최종 PASS에서 제외하며 필수 7미션 21문항으로 판정합니다.\n\n| 미션 | 원문 학습계획 | 화면 표시 시간(참고) |\n|---|---|---|\n| 미션 1 | 1.5시간 | 1시간 |\n| 미션 2 | 2시간 | 2시간 |\n| 미션 3 | 2.5시간 | 2시간 |\n| 미션 4 | 3시간 | 3시간 |\n| 미션 5 | 3.5시간 | 4시간 |\n| 미션 6 | 3시간 | 3시간 |\n| 미션 7 | 3.5시간 | 4시간 |\n| 미션 8 | 3시간 | 3시간 |\n| 미션 9 | 3시간 | 3시간 |\n| 미션 10 | 3.5시간 | 3시간 |\n\n플랫폼은 시간을 정수로 표시하므로 미션별 표시와 원문 시간이 다를 수 있습니다. 학습 계획과 시간 기록은 원문 열을 기준으로 합니다.\n\n플랫폼 시간 표시는 정수 합계 28시간이며 원문 계획은 28.5시간입니다. 0.5시간 차이는 표시 방식의 차이이므로 학습 계획은 위 원문 시간을 따르세요.\n', '03_참고자료/M08_실측정오표.md': '# 미션 8 실측 실습 적용 기준\n\n| 원문 예시 | 현재 고정 환경 실측 | 이유와 적용 범위 |\n|---|---|---|\n| 진흙 0% 자동 처리 73% | 120장 중 자동 처리 후보 90장, 75% | 원문 예시와 실행 관측의 차이. 원문 코드·기대 파일은 보존하며 현재 기록은 실제 출력과 분모를 사용합니다. 경보 정확도는 100%이고, 진흙 비율이 5%이면 93%로 낮아져 목표인 95%에 미달합니다. |\n\n다른 환경의 수치는 로그와 환경을 기록해 설명하세요. 예시에 맞추려고 결과를 조작하지 않습니다.\n', '03_참고자료/M09_M10_실습안내.md': '# QLT03 미션 9·미션 10 수정 실습 안내 [교육용가상자료]\n\n수정 실습: 원문 실행값은 역사 비교용이다. 채점·최종 보고에는 work/revised/m09.py의 수정 실행을 사용한다. 비교는 기존 AI→개선 AI로 통일한다. B 360장의 전후 오판 손실(만원/장)·평균 시간(분/장), 같은 연간 건수·시급, 증분 운영비 및 투자비를 전후 표에 쓰고 연 순절감액과 회수기간을 재계산한다. 순절감액이 0 이하이면 회수 불가이다.\n\n수정 실습: 원문 실행값은 역사 비교용이다. 채점·최종 보고에는 work/revised/m10.py의 수정 실행을 사용한다. 운영 신규240장은 처리 건수만 집계하며, 기존600+신규160장은 학습, hold80은 모델 선택용이다. 최종 시험은 별도 seed 180/181/182 각80장으로 고정한 240장이며 튜닝에 재사용하지 않는다. 학습·선택·시험 ID 교집합 0을 확인한다. MAE·등급 정확도 분모는 시험240장, MAPE는 그중 실제 손상도10 이상 장수이다. 같은 seed 생성기의 합성 자료라는 한계는 남으며 현장 성능을 뜻하지 않는다.\n\n노트북 환경 셀부터 실행하면 원문 재현과 수정 실습이 순서대로 실행된다. 최종 보고의 기본값은 `outputs/m09/revised/stdout.txt`, `outputs/m10/revised/stdout.txt`를 사용하고, 도전값은 `outputs/mNN/challenge/stdout.txt`와 비교한다. `baseline`은 원문 재현 이력이다. 실습 코드의 코드·분모·가정값을 기록하고 원문 값과 혼합하지 않는다.\n\n로컬 프로젝트 폴더에서는 `python work/qlt03_revision_runner.py 9`, `python work/qlt03_revision_runner.py 10`으로 수정 실습을 실행한다. 원문 `src/m09.py`, `src/m10.py` 및 원문 기대값은 보존한다. 최종 원문 지정 코드 경로에 제출 사본을 만들 때 실습 코드/도전본 선택을 명시한다.\n\n미션 9 확인할 결과와 채점 기준은 `M09_ROI_정오표.md`의 실습 코드 열 및 수정 실행 로그를 사용한다. 원문 보존 자료는 이력 비교용이다.\n\n## 현재 미션 9 채점 기준\n실습 코드 채점 기준: 연간 순절감액 2,751만원, 판독 시간 단축률 0.3%, 투자 회수 10.9개월.\n반올림 전 2,751.111111만원/년·0.292475%·10.904685개월. 기존 AI→개선 AI, 동일 B 360장, 연 20,000장, 시급 1.2만원, 증분 운영비 150만원/년, 증분 투자비 2,500만원이라는 교육용 가정이다. outputs/m09/revised/stdout.txt와 전후 총비용 차이를 확인한다. 환경에 따른 차이는 실제 로그와 산식으로 설명하며 예시 숫자를 맞추려고 출력을 고치지 않는다.\n', '03_참고자료/공통_Jupyter_사용안내.md': '# QLT03 실행·저장 안내\n\n[교육용가상자료] 실제 조직·장비·개인정보를 입력하지 않습니다.\n\n## 시작\n학습자 배포 ZIP의 01_학습자자료 최상위에 있는 QLT03_assignment.ipynb를 엽니다. 노트북 열기 → 환경 준비 셀 실행 → 미션 코드 저장 → 실행·확인 순서입니다. 이미 준비된 환경에는 설치하지 않습니다.\n\n## 저장과 새 작업 환경 복원\n\n1. 종료 전 수정한 노트북과 `reports/`, 직접 작성한 `docs/`, `work/`, `outputs/`를 원래 폴더 구조 그대로 허용된 저장 매체에 복사합니다. `reports/재개기록.md`에 마지막 미션, 마지막 실행일시, 마지막 작성 문장, 이어 할 일을 직접 적습니다. 노트북 사본과 reports/docs/work/outputs 파일을 함께 저장합니다.\n2. 새 작업폴더에서 저장한 노트북을 열고 **환경 준비 셀부터 실행**합니다. QLT03는 바로 다음 수정 실습 준비 셀까지 실행합니다. 이때 생긴 빈 양식은 저장 기록의 복원이 아닙니다.\n3. 저장 사본을 **reports/docs/work/outputs 원경로에 복원**합니다. 예: 저장한 `reports/m01_record.md` → 새 작업폴더의 `reports/m01_record.md`. 파일을 최상위에만 올리지 않습니다. Jupyter 작업폴더의 같은 하위 경로로 복사합니다. 기존 파일이 있으면 별도 보관 후 복원합니다.\n4. `reports/재개기록.md`와 마지막 기록지를 열어 **마지막 미션·실행일시·작성 문장**이 저장 사본과 같은지 확인합니다. outputs의 execution.json 시간도 대조합니다. 없으면 진행을 멈추고 저장 사본부터 확인합니다.\n5. 이어 할 미션의 코드 저장·실행 셀을 선택합니다. 복원한 work 코드를 다시 쓸 때는 노트북 셀에도 같은 수정 내용을 반영하고 백업 셀을 먼저 실행합니다. 처음부터 모두 실행해 복원 코드를 기본값으로 덮지 않습니다. 같은 분석을 재실행하면 이전 로그는 outputs/history에 남고 새 로그의 실행일시만 새로 찍힙니다. 재개기록에 직접 적은 이전 일시·마지막 미션은 자동으로 바꾸지 않습니다.\n\n복원 연습: `reports/m01_record.md`에 본인이 정한 표식과 실제 실행일시·마지막 미션 1을 적어 저장 → 새 작업폴더 준비 → 원경로 복원 → 환경 준비 재실행 → 표식과 기록 유지 확인 → 미션 1 실행 후 이전 execution.json이 outputs/history에 남는지 확인합니다. 실제 기관의 저장 허용 여부는 별도로 확인합니다.\n\n선택 미션 4·미션 6·미션 8은 산출물과 Q1·Q2·Q3 모두 가점/참고이며 최종 PASS에서 제외합니다. 필수 7미션 21문항만 최종 PASS로 판정합니다.\n\n초보 소요시간은 원문의 약 1.3~1.7배를 준비 계획용 추정 범위로 참고하세요. 파일 조작·개념 학습·대기·재시도에 따라 달라지며 실제 학습자 실측이나 확정 시간이 아닙니다.\n', '03_참고자료/M08_웹앱과_대체경로.md': '# QLT03 미션 8 웹앱과 대체 경로\n\n[교육용가상자료] 실행 루트에서 작업합니다. 학습자 ZIP은 01_학습자자료, 전체 프로젝트는 notebook·src·docs가 함께 있는 폴더입니다.\n\n1. 노트북의 환경 준비 셀은 `src/dashboard/qlt03_app.py`, `requirements.txt`, `requirements-dashboard.txt`를 만듭니다. 기존 파일은 유지합니다.\n2. 미션 8 코드 저장 셀에서 시뮬레이터를 저장합니다. 준비된 환경에서 `python -m streamlit run src/dashboard/qlt03_app.py`를 실행합니다. 웹앱 선택 환경의 Streamlit 항목은 requirements-dashboard.txt에 있습니다.\n3. 실행 버튼으로 미션 8 시뮬레이터를 실행하고 결과 표·그래프·로그를 확인합니다. 노트북 경로는 `outputs/m08/baseline/stdout.txt`와 같은 폴더의 그림을 사용합니다. Q3에는 앱 파일 또는 시뮬레이터 실행 경로와 실제 출력·분모·해석을 연결합니다.\n4. 설치·웹앱 실행이 제한되면 노트북 미션 8 시뮬레이터를 동등하게 인정합니다. 선택 미션이므로 최종 PASS에서 제외합니다. 브라우저 포함 2초 응답은 별도 측정 항목입니다.\n', '03_참고자료/M10_실습정오표.md': '# 미션 10 원문과 수정 실습 기준\n\n| 항목 | 원문 예시/동작 | 수정 값/동작 | 이유와 채점 범위 |\n|---|---|---|---|\n| 운영 문서 | 상수 목록 12/12, 자체점검 8/8 | 실제 운영지침 12절 본문 검사; 빈 양식 0/12, 문서 점검 FAIL | 빈 문서를 완료로 표시하지 않음. 직접 작성 후 재검사하며 내용의 타당성은 별도 평가 |\n| 최종 성능 | MAE 8.79→6.86 등 원문 예시 | 최종시험 240장 MAE 6.43, 등급 정확도 87.5% | 원문 예시와 최종시험 분모를 분리; 실제 로그와 환경을 기록 |\n\n원문 코드와 기대 출력은 보존합니다. 실습 코드 최종시험 MAE 6.43, 등급 정확도 87.5%(시험 240장). 빈 운영지침은 0/12이며 문서 미완료 점검은 FAIL이 정상이다. 직접 작성 후 완료 수와 자체 점검 결과를 다시 기록한다.\n', '03_참고자료/처음_시작할_때.md': '# QLT03 처음 시작 안내\n\n[교육용가상자료] 실제 조직·장비·사진·정비 자료를 입력하지 않습니다.\n\n## 파일 열기\n\n학습자 배포 ZIP의 01_학습자자료 최상위에서 `QLT03_assignment.ipynb`를 엽니다. ZIP의 `01_학습자자료/`가 실행 루트입니다. 아래 docs·reports·src 경로는 실행 루트 기준입니다.\n\n교육장 Jupyter에서 노트북을 열면 환경 준비 셀이 기본 환경 목록·기록 양식·수정 코드·앱과 이 안내를 만듭니다. 파일 → 다른 이름으로 저장으로 사본을 만들고 위에서 아래로 진행하세요. 환경이 준비되지 않았으면 운영 담당자에게 확인합니다.\n\n## 실행 순서\n\n노트북 열기 → 환경 준비 셀 실행 → 미션 코드 저장 → 실행·확인 순서입니다.\n\n1. 환경 준비 셀을 실행합니다. `requirements.txt`와 `requirements-dashboard.txt`가 준비되며 기존 파일은 덮어쓰지 않습니다. 준비된 환경에서는 설치하지 않습니다.\n2. 해당 미션의 백업 셀 → 코드 저장 셀 → 실행·확인 셀을 순서대로 실행합니다. `%%writefile`은 코드 저장 명령입니다. 코드 저장 셀은 덮어쓸 수 있으므로 바로 앞 백업 셀을 함께 실행합니다.\n3. 출력 로그의 지표·단위·분모와 그림을 읽고 `reports/mNN_record.md`에 관찰값을 적습니다. 도전은 `work/mNN_challenge.py`에서 변경 조건과 전후 결과·해석을 기록합니다. 결과가 같거나 나빠져도 적용 증거와 이유를 설명합니다.\n4. 최종 파일은 `docs/원문_제출경로_연결표.md`에서 확인합니다. 원문 이름의 Python 사본은 비교용이며 현재 수정 실습·도전 코드 경로를 기록지에 연결합니다. 원문을 덮어쓰지 않습니다.\n\nREADY는 도전 코드 수정·저장을 마쳤음을 학습자가 표시하는 값입니다. 수정·저장 후 도전 실행 셀의 READY=False를 READY=True로 바꾸세요. 기본 실행 확인 PASS는 도전·해석·원문 산출물 작성 완료를 뜻하지 않습니다. 오류가 나면 첫 오류 메시지와 입력 조건을 그대로 기록하고 원인과 수정 내용을 남깁니다.\n\n## 미션 8 웹앱과 제한 환경\n\n환경 준비 셀이 `src/dashboard/qlt03_app.py`를 생성합니다. 미션 8 코드 저장 셀로 시뮬레이터를 저장한 뒤 준비된 환경에서 `python -m streamlit run src/dashboard/qlt03_app.py`를 실행합니다. 앱의 실행 버튼·필터를 사용해 표·그래프·로그와 해석을 기록합니다. 설치나 웹앱 실행이 제한되면 노트북 미션 8 시뮬레이터를 동등한 대체 경로로 사용합니다. `docs/M08_웹앱과_대체경로.md`를 읽으세요.\n\n## 기록 보존과 이어 하기\n\n종료 전에 수정한 노트북과 reports·직접 작성한 docs·work·outputs를 원래 폴더 구조대로 저장합니다. 노트북 사본 저장과 작업 환경 파일 보존은 별개입니다. `docs/저장과_새런타임_복원.md`에 따라 복원한 뒤 준비 셀을 재실행합니다. 준비 셀은 이미 존재하는 문서·코드·기록을 건너뜁니다. 저장이 제한되면 화면을 닫기 전에 담당자에게 허용된 기록 보존 경로를 확인하세요.\n\n## 개념과 시간\n\n미션 3~미션 5 전에 `docs/핵심지표_한장.md`에서 지표의 뜻·분모·좋아지는 방향과 계산 예시를 읽으세요. 용어·오류 메시지·수정 후보를 AI에 질문할 수 있지만 코드와 실제 출력의 대조·최종 판단은 직접 합니다.\n\n원문 계획은 28.5시간, 화면 표시 시간는 28시간입니다. 초보자는 약 1.3~1.7배를 준비 계획용 추정 범위로 참고합니다. 학습자 실측 시간이 아닙니다. 선택 미션 4·미션 6·미션 8은 산출물과 Q1·Q2·Q3 모두 가점/참고이며 최종 PASS에서 제외합니다. 필수 7미션 21문항을 평가합니다.\n\n[실습 준비·보안·제출 안내](실습준비와_제출안내.md)를 확인하세요.\n\n종료 전 허용된 저장 위치에 저장하고 다시 열어 확인한 뒤 사용한 교육장 계정에서 로그아웃합니다. 공용 PC에 비밀번호를 저장하지 않으며 로컬 사본은 조직 규정에 따라 정리합니다.\n\n## 오프라인 예제 요청문\n\n질문 틀: "[교육용가상자료] 미션 __의 목표는 __다. 내가 실행한 코드의 변경은 __이고 출력은 __다. 출력에서 직접 확인 가능한 근거만으로 차이/동일 결과의 원인을 설명해 줘. 관찰되지 않은 사실은 미확인으로 표시해 줘."\n\nAI 설명을 실행 증거로 대체하지 말고 코드·출력과 대조해 채택 여부를 기록한다.\n', '03_참고자료/실습준비와_제출안내.md': '# 실습 준비·보안·제출 안내\n\n제공된 교육용 가상자료만 사용합니다. 비공개 업무자료, 실습에는 제공된 가상자료만 사용하고, 실제 이름·연락처 같은 개인정보는 AI 도구나 노트북에 넣지 않습니다. 실습을 위한 조직 내 촬영은 하지 않습니다. 조직 보안 지침과 서비스 이용 허용 여부를 먼저 확인합니다.\n\n이 자료는 교육용 가상 실습입니다. 실제 업무·정비·재보급 승인 기준이나 자동 결정에 사용하지 않습니다.\n\nAI에게 문서 초안을 요청할 수 있습니다. 사실·숫자 확인과 최종 수정은 본인이 합니다. 실제 실행·계산·자료 대조·최종 판단은 직접 하고 요청문·실제 응답·채택 이유를 기록합니다. 불일치·기각·성능 향상을 억지로 만들지 않습니다.\n\n종료 전 허용된 저장 위치에 저장하고 다시 열어 확인한 뒤 사용한 교육장 계정에서 로그아웃합니다. 공용 PC에 비밀번호를 저장하지 않으며 로컬 사본은 조직 규정에 따라 정리합니다.\n\nAI 접속과 사람 협업이 모두 불가능하면 개인 판단·계산·초안을 먼저 저장하고 협업 항목은 미수행으로 표시합니다. 교육 담당자에게 동료 검토 또는 허용 접속 시간의 보완 일정과 필요시 기한 조정을 요청합니다. 보완 전에는 협업 완료로 표시하지 않습니다.\n\n제출처·마감 일시·시간대는 miliai.kr 해당 과제 공지를 확인하고, 표시가 없으면 제출 전에 교육 담당자에게 확인합니다. 플랫폼 학습기간은 계획값이며 마감일을 뜻하지 않습니다. 별도 공지가 없으면 미션마다 그 미션 산출물을 ZIP 하나로 묶어 해당 미션 제출란에 첨부합니다. 통합 문서는 미션별 제목·위치를 적고 각 해당 제출란에 같은 파일을 첨부합니다. 파일명은 과제ID_미션번호_산출물명으로 작성하고 실제 형식의 확장자를 붙입니다. 실명·개인 식별번호는 파일명에 넣지 않습니다.\n\nPython 3.12와 requirements.txt 기준으로 준비된 환경을 사용합니다. 제한 환경에서는 관리자가 준비한 로컬 Jupyter를 사용하며 설치 명령을 실행하지 않습니다. 환경 셀은 버전을 확인하고 실습 파일을 준비합니다. 버전이 맞지 않으면 안내 메시지와 함께 멈추며, 그때는 requirements.txt 기준으로 교육 담당자가 안내한 방법으로 설치한 뒤 다시 실행합니다. 환경 셀은 건너뛰지 마세요.\n\n노트북 사본과 reports·직접 작성한 docs·work·outputs를 원래 폴더 구조대로 저장합니다. docs/원문_제출경로_연결표.md의 현재 실행 파일을 제출합니다.\n\n최종 실행 재현 확인은 필수 미션만 선택하여 실행합니다: `python -m pytest tests/test_missions.py -k "not [4] and not [6] and not [8]"`. 전체 미션 검사는 선택 미션까지 실행한 경우의 추가 확인이며 최종 PASS 조건이 아닙니다.\n\n미리보기의 source_sha256 및 검증 메타데이터는 제공 파일의 생성 근거를 확인하는 해시이며 학습 데이터 필드나 제출할 정답이 아닙니다. 실행 기록의 해시는 코드 변경 여부를 확인하므로 그대로 보존하세요.\n', '03_참고자료/참고링크.md': '# 오프라인 실습안내\n\n외부 사이트 없이 ZIP에 제공된 자료를 사용합니다. 용어 뜻은 처음 시작 안내에서 확인합니다.\n\n질문 틀: "[교육용가상자료] 미션 __의 목표는 __다. 내가 실행한 코드의 변경은 __이고 출력은 __다. 출력에서 직접 확인 가능한 근거만으로 차이/동일 결과의 원인을 설명해 줘. 관찰되지 않은 사실은 미확인으로 표시해 줘."\n\nAI 설명을 실행 증거로 대체하지 말고 코드·출력과 대조해 채택 여부를 기록한다.\n', '03_참고자료/M09_ROI_정오표.md': '# 미션 9 ROI 실습 적용 기준 [교육용가상자료]\n\n채점·최종 보고는 실습 코드 `work/revised/m09.py`와 `outputs/m09/revised/stdout.txt`를 사용한다.\n원문 코드·원문 기대값은 보존하며 아래 두 이력 행은 현재 채점값이 아니다.\n\n| 구분 | 연간 순절감액(만원/년) | 시간 단축률 | 투자 회수(개월) | 적용 범위 |\n|---|---:|---:|---:|---|\n| 건양대 원문 예시 | 5,140 | 30.8% | 5.8 | 원문 보존·이력 비교만 |\n| 고정 환경 원문 재현 | 4,572.222222 (표시 4,572) | 30.555556% | 6.561361 (표시 6.6) | 역사 비교만 |\n| 실습 코드 | **2,751.111111 (표시 2,751)** | **0.292475% (표시 0.3%)** | **10.904685 (표시 10.9)** | **채점·최종 보고 기본값** |\n\n원문은 손실의 비교 대상을 기존 AI, 시간의 비교 대상을 전수 수동 판독으로 섞었다.\n실습 코드는 손실·시간 모두 같은 B 360장에서 기존 AI→개선 AI로 비교한다.\n원문 예시와 원문 재현의 차이는 원문 예시의 미재현 이력으로 보존하며 값을 맞추지 않는다.\n\n- 평균 판독: 기존 10.447222→개선 10.416667분/장.\n- 손실: 기존 2.055556→개선 1.911111만원/장.\n- 연 20,000장, 시급 1.2만원, 증분 운영비 150만원/년, 증분 투자비 2,500만원.\n- 연 순절감 = 손실 절감 2,888.888889 + 시간 절감 12.222222 − 운영비 150 = 2,751.111111만원.\n- 연 총비용: 기존 45,290→개선 42,538.888889만원. 그 차이가 순절감이다.\n- 시간 단축률 = 1 − 개선 평균시간 / 기존 평균시간.\n- 회수기간 = 투자비 / (연 순절감 / 12). 순절감이 0 이하이면 회수 불가.\n- 연 10,000장 도전: 직접 계산해 실행 로그·산식·분모와 함께 제출한다. 고정 운영비 때문에 절감액은 단순 절반이 아니다.\n\n원문 Q1 문구·목표는 보존한다. 수행 증거·목표 달성은 구분하며 실습 코드의 동일 분모·산식·로그로 판단한다.\n다른 환경이나 도전값의 결과가 다르면 가정·코드·실측 근거를 기록한다. 예시와 강제 일치시키지 않는다.\n\n참고 수치를 그대로 제출하지 말 것(자기 계산 후 비교). 본인의 실행 로그·산식·분모와 해석을 제출한다.\n', '03_참고자료/핵심지표_한장.md': '# QLT03 미션 3~미션 5 핵심 지표 한 장\n\n[교육용가상자료] 아래 숫자는 산식을 배우는 작은 예시이며 실측·목표값이 아닙니다. 기본 코드는 유지합니다.\n\n| 지표 | 뜻·분모·작은 계산 예시 | 좋아지는 방향·해석 한계 |\n|---|---|---|\n| IoU(미션 3~미션 4) | 정답·예측 마스크 교집합 픽셀 / 합집합 픽셀. 정답6, 예측4, 겹침3이면3/(6+4−3)=3/7≈0.429 | 클수록 겹침이 좋음; 양쪽 빈 마스크 처리 규칙 확인 |\n| Dice(미션 3~미션 4) | 2×교집합 / (정답 픽셀+예측 픽셀). 같은 예는6/10=0.60 | 클수록 좋음; IoU와 수치 자체를 같다고 비교하지 않음 |\n| MAE | 절대오차 합 / 평가 표본 수 n. 실제 10,20 / 예측 12,16이면 (2+4)/2=3 | 작을수록 좋음; 원래 단위 |\n| RMSE | √(제곱오차 합 / n). 같은 예는 √((4+16)/2)=√10≈3.16 | 작을수록 좋음; 큰 오차에 민감 |\n| MAPE | (100/n)×Σ(절대오차/실제값). 같은 예는 (20%+20%)/2=20% | 작을수록 좋음; 실제값 0이면 정의 불가. 코드의 제외/보정 규칙과 유효 n을 함께 기록 |\n| R2 | 1−Σ오차²/Σ(실제값−실제평균)². 같은 예는 1−20/50=0.60 | 클수록 좋음; 1이 완전 일치, 음수 가능. 실제값이 전부 같으면 분모 0 |\n| 오차 개선율 | (기준 오차−새 오차)/기준 오차×100. MAE10→8은20% | 같은 평가 표본·단위에서 클수록 좋음; 기준 오차 0이면 정의 불가 |\n| 등급 정확도(미션 5) | 정답 등급과 같은 사진 수 / 평가 사진 수. 8/10=80% | 높을수록 좋음; 과소 판정(실제보다 낮게 판정) 수도 별도 확인 |\n| 자동 통과 후보 비율·일치율(미션 5) | 후보 비율=문턱 이상 수/전체, 후보 일치율=후보 중 맞은 수/후보 수. 전체10·후보4·맞음3이면40%,75% | 두 분모를 구별. 후보0건이면 일치율 계산 불가. 후보 선정은 실제 승인 아님 |\n\n사진 분할의 픽셀 겹침, 면적·깊이의 측정 오차, 손상도 회귀 오차, 등급 분류 정확도는 서로 다른 지표입니다. 보정 계수는 학습 자료에서 정하고 같은 평가 사진으로 전후를 비교합니다. 미션 4의 깊이는 단일 사진에서 직접 잰 값이 아닌 추정이며 원문 예제는 실제 Mask R-CNN 학습 완료를 뜻하지 않습니다. 미션 10 최종 성능은 별도 시험240장이고 MAPE 분모는 그중 손상도10 이상 장수입니다.\n\n실행 전 본인이 확인할 것: 이번 출력의 단위, 분모, 좋아지는 방향을 한 문장씩 설명하고 작은 예시를 직접 계산해 보세요. 목표 달성 판정은 원문 Q1과 대조합니다. 이번 지표의 단위·분모·좋아지는 방향을 설명하고, 어려운 용어는 핵심 지표 예시로 다시 확인하세요.\n', '03_참고자료/저장과_새런타임_복원.md': '# 저장과 복원 안내\n\n## 저장과 새 작업 환경 복원\n\n1. 종료 전 수정한 노트북과 `reports/`, 직접 작성한 `docs/`, `work/`, `outputs/`를 원래 폴더 구조 그대로 허용된 저장 매체에 복사합니다. `reports/재개기록.md`에 마지막 미션, 마지막 실행일시, 마지막 작성 문장, 이어 할 일을 직접 적습니다. 노트북 사본과 reports/docs/work/outputs 파일을 함께 저장합니다.\n2. 새 작업폴더에서 저장한 노트북을 열고 **환경 준비 셀부터 실행**합니다. QLT03는 바로 다음 수정 실습 준비 셀까지 실행합니다. 이때 생긴 빈 양식은 저장 기록의 복원이 아닙니다.\n3. 저장 사본을 **reports/docs/work/outputs 원경로에 복원**합니다. 예: 저장한 `reports/m01_record.md` → 새 작업폴더의 `reports/m01_record.md`. 파일을 최상위에만 올리지 않습니다. Jupyter 작업폴더의 같은 하위 경로로 복사합니다. 기존 파일이 있으면 별도 보관 후 복원합니다.\n4. `reports/재개기록.md`와 마지막 기록지를 열어 **마지막 미션·실행일시·작성 문장**이 저장 사본과 같은지 확인합니다. outputs의 execution.json 시간도 대조합니다. 없으면 진행을 멈추고 저장 사본부터 확인합니다.\n5. 이어 할 미션의 코드 저장·실행 셀을 선택합니다. 복원한 work 코드를 다시 쓸 때는 노트북 셀에도 같은 수정 내용을 반영하고 백업 셀을 먼저 실행합니다. 처음부터 모두 실행해 복원 코드를 기본값으로 덮지 않습니다. 같은 분석을 재실행하면 이전 로그는 outputs/history에 남고 새 로그의 실행일시만 새로 찍힙니다. 재개기록에 직접 적은 이전 일시·마지막 미션은 자동으로 바꾸지 않습니다.\n\n복원 연습: `reports/m01_record.md`에 본인이 정한 표식과 실제 실행일시·마지막 미션 1을 적어 저장 → 새 작업폴더 준비 → 원경로 복원 → 환경 준비 재실행 → 표식과 기록 유지 확인 → 미션 1 실행 후 이전 execution.json이 outputs/history에 남는지 확인합니다. 실제 기관의 저장 허용 여부는 별도로 확인합니다.\n', 'work/qlt03_revision_runner.py': 'from pathlib import Path\nimport contextlib, io, runpy, json, hashlib, sys, datetime, shutil\nimport matplotlib\nmatplotlib.use("Agg")\nimport matplotlib.pyplot as plt\nROOT = Path(__file__).resolve().parents[1]\nn = int(sys.argv[1])\nsource = ROOT / "work/revised" / f"m{n:02d}.py"\nfolder = ROOT / "outputs" / f"m{n:02d}" / "revised"\nif folder.exists():\n    archive = ROOT / "outputs/history" / f"m{n:02d}" / ("revised-" + datetime.datetime.now().strftime("%Y%m%d-%H%M%S-%f"))\n    archive.parent.mkdir(parents=True, exist_ok=True)\n    shutil.move(str(folder), str(archive))\nfolder.mkdir(parents=True)\nfigures = []\ndef save(*args, **kwargs):\n    for number in plt.get_fignums():\n        dest = folder / f"figure_{len(figures)+1:02d}.png"\n        plt.figure(number).savefig(dest, dpi=110, bbox_inches="tight")\n        figures.append(str(dest.relative_to(ROOT)))\n    plt.close("all")\nplt.show = save\nout, err = io.StringIO(), io.StringIO()\nok = False\ntry:\n    with contextlib.redirect_stdout(out), contextlib.redirect_stderr(err):\n        data = runpy.run_path(str(source), run_name="__main__")\n        save()\n        if n == 10:\n            assert all(v == 0 for v in data["intersections"].values())\n            assert len(data["y"]) == 240\n        print("수정 실습 실행 확인 PASS — 성능 목표 달성·학습자 수행 완료는 별도")\n    ok = True\nfinally:\n    (folder/"stdout.txt").write_text(out.getvalue(), encoding="utf-8")\n    (folder/"stderr.txt").write_text(err.getvalue(), encoding="utf-8")\n    (folder/"execution.json").write_text(json.dumps(dict(completed=ok, timestamp=datetime.datetime.now(datetime.timezone.utc).isoformat(), source=str(source.relative_to(ROOT)), source_sha256=hashlib.sha256(source.read_bytes()).hexdigest(), figures=figures), ensure_ascii=False, indent=2), encoding="utf-8")\nprint(out.getvalue())\n', 'work/revised/m09.py': '# 수정 실습본: 원문 코드는 별도 보존합니다.\n"""운영 기간 외관 판정 로그로 AAR을 수행해 오판 사례를 원인별(오염 가림·저조도·작은 결함·경계 근처·유형 혼동)로 분류하고, 분할 임계값·학습 데이터·등급 경계(과소 판정 손실 반영)를 재보정해 보류 구간에서 개선을 검증한 뒤 비용 절감액·시간 단축률·투자 회수 기간 3대 효과 지표를 산식과 민감도로 산출한다."""\nimport numpy as np\nimport pandas as pd\nimport matplotlib.pyplot as plt\nfrom sklearn.ensemble import RandomForestClassifier, GradientBoostingRegressor\nfrom sklearn.model_selection import cross_val_predict\n\n# ---------------------------------------------------------------------------\n# [교육용가상자료] 장비 외관 사진 생성기 (실제 장비·조직·위치 사진이 아니라 난수로 그린 64x64 합성 이미지)\n# 도장된 금속 표면에 균열(crack)·부식(corrosion)·파손(damage)·누유(leak)를 그리고, 조명 편차와 진흙(mud) 오염을 섞는다.\n# 이미지와 함께 장비 관리 기록(제원 요약)과 정답 결함 마스크·손상도·정비 등급을 만든다.\n# ---------------------------------------------------------------------------\nfrom scipy import ndimage as ndi\n\nSZ = 64\nCLASSES = ["none", "crack", "corrosion", "damage", "leak"]\nKO_CLS = {"none": "결함 없음", "crack": "균열", "corrosion": "부식", "damage": "파손", "leak": "누유"}\nCLS_W = {"none": 0.0, "crack": 1.4, "corrosion": 1.0, "damage": 1.6, "leak": 0.6}      # 손상도 계산용 결함 유형 가중치\nGRADE_NAME = {1: "1단계 정비", 2: "2단계 정비", 3: "3단계 정비"}\nYY, XX = np.mgrid[0:SZ, 0:SZ]\n\n\ndef grade_of(idx):\n    return np.where(np.asarray(idx) >= 65, 3, np.where(np.asarray(idx) >= 35, 2, 1))\n\n\ndef blob(rng, cy, cx, r, rough=0.35):\n    """불규칙한 원형 영역 마스크."""\n    noise = ndi.gaussian_filter(rng.normal(size=(SZ, SZ)), 3)\n    noise = noise / (np.abs(noise).max() + 1e-9)\n    return (np.hypot(YY - cy, XX - cx) + rough * r * 2.2 * noise) < r\n\n\ndef draw(rng, cls, sev):\n    """결함 유형·심각도 sev(0~1)에 맞는 정답 마스크와 색 정보를 만든다."""\n    m = np.zeros((SZ, SZ), bool)\n    if cls == "crack":\n        y, x, ang = rng.uniform(12, 52), rng.uniform(12, 52), rng.uniform(0, np.pi)\n        for _ in range(int(20 + 40 * sev)):\n            ang += rng.normal(0, 0.35); y += np.sin(ang) * 1.2; x += np.cos(ang) * 1.2\n            if 0 <= int(y) < SZ and 0 <= int(x) < SZ:\n                m[int(y), int(x)] = True\n        m = ndi.binary_dilation(m, iterations=1 + int(sev > 0.6))\n    elif cls == "corrosion":\n        for _ in range(int(rng.integers(1, 4))):\n            m |= blob(rng, rng.uniform(14, 50), rng.uniform(14, 50), 3 + 9 * sev * rng.uniform(0.6, 1.1))\n    elif cls == "damage":\n        m = blob(rng, rng.uniform(20, 44), rng.uniform(20, 44), 6 + 12 * sev, 0.3)\n    elif cls == "leak":\n        x0, y0 = rng.uniform(16, 48), rng.uniform(8, 24)\n        w, ln = 2 + 4 * sev, 12 + 26 * sev\n        m = (np.abs(XX - x0 - 0.15 * (YY - y0)) < w * (1 - 0.4 * (YY - y0) / ln)) & (YY > y0) & (YY < y0 + ln)\n    return m\n\n\ndef make_images(n=600, seed=0, mud=0.0, p_none=0.25, dark=False):\n    """가상 외관 사진 n장(+진흙 오염 mud 비율)과 관리 기록·정답. 반환: meta(DataFrame), imgs(n,64,64,3), masks(n,64,64)."""\n    rng = np.random.default_rng(seed)\n    imgs, masks, rows = np.zeros((n, SZ, SZ, 3), np.float32), np.zeros((n, SZ, SZ), bool), []\n    for i in range(n):\n        cls = "none" if rng.random() < p_none else str(rng.choice(CLASSES[1:], p=[0.25, 0.3, 0.25, 0.2]))\n        sev = 0.0 if cls == "none" else float(rng.beta(1.6, 2.0))\n        base = np.array([0.42, 0.46, 0.40]) * (1 + 0.5 * ndi.gaussian_filter(rng.normal(size=(SZ, SZ)), 8)[..., None] * 3)\n        img = base + rng.normal(0, 0.012, (SZ, SZ, 3))\n        m = draw(rng, cls, sev)\n        a = ndi.gaussian_filter(m.astype(float), 0.8)[..., None]                        # 경계를 부드럽게 섞는다\n        if cls == "crack":\n            col = img * rng.uniform(0.2, 0.45)\n        elif cls == "corrosion":\n            col = np.array([0.55, 0.30, 0.12]) * rng.uniform(0.75, 1.15) * (0.85 + 0.3 * rng.random((SZ, SZ, 1)))\n        elif cls == "damage":\n            col = np.ones_like(img) * np.array([0.34, 0.34, 0.36]) * (1.0 - 0.45 * sev) * rng.uniform(0.75, 1.0)\n            rim = m & ~ndi.binary_erosion(m, iterations=2)\n            a = np.maximum(a, 0.0); img = np.where(rim[..., None], 0.7, img)\n        elif cls == "leak":\n            col = np.ones_like(img) * np.array([0.10, 0.09, 0.08]) * rng.uniform(0.8, 2.6)\n        else:\n            col = img\n        img = img * (1 - a) + col * a\n        if rng.random() < 0.35:                                   # 무해한 얼룩(결함 아님): 옅은 갈색 번짐\n            st = blob(rng, rng.uniform(5, 58), rng.uniform(5, 58), rng.uniform(6, 14), 0.5)\n            al = ndi.gaussian_filter(st.astype(float), 2.5)[..., None] * rng.uniform(0.15, 0.4)\n            img = img * (1 - al) + np.array([0.50, 0.40, 0.30]) * al\n        if rng.random() < 0.3:                                    # 무해한 긴 스크래치(밝은 선)\n            y0, x0, ang = rng.uniform(5, 58), rng.uniform(5, 58), rng.uniform(0, np.pi)\n            for t in np.linspace(0, 25, 40):\n                yy, xx = int(y0 + t * np.sin(ang)), int(x0 + t * np.cos(ang))\n                if 0 <= yy < SZ and 0 <= xx < SZ:\n                    img[yy, xx] = np.clip(img[yy, xx] * 1.5, 0, 1)\n        if mud > 0:                                           # 진흙 오염: 결함 위를 덮을 수 있다(정답 마스크는 그대로)\n            mm = np.zeros((SZ, SZ), bool)\n            while mm.mean() < mud:\n                mm |= blob(rng, rng.uniform(0, SZ), rng.uniform(0, SZ), rng.uniform(5, 11), 0.5)\n            img[mm] = np.array([0.38, 0.30, 0.22]) * (0.85 + 0.3 * rng.random((mm.sum(), 1)))\n        gain = rng.uniform(0.3, 0.55) if dark else rng.uniform(0.5, 1.15)\n        ang = rng.uniform(0, 2 * np.pi)\n        light = gain * (1 + 0.3 * (np.cos(ang) * (XX - 32) + np.sin(ang) * (YY - 32)) / 32)\n        imgs[i] = np.clip(img * light[..., None] + rng.normal(0, 0.025, (SZ, SZ, 3)), 0, 1)\n        masks[i] = m\n        area = float(m.mean())\n        depth = {"none": 0.0, "crack": 0.3 + 3.0 * sev, "corrosion": 0.1 + 1.5 * sev, "damage": 1.0 + 8.0 * sev, "leak": 0.0}[cls] * rng.lognormal(0, 0.1)\n        idx = float(np.clip(100 * (1 - np.exp(-(CLS_W[cls] * area * 9 + 0.05 * depth))) + rng.normal(0, 1.5), 0, 100))\n        rows.append({"img_id": f"IMG{seed:02d}{i:04d}", "cls": cls, "sev": sev, "area_frac": area, "depth_mm": depth, "damage_idx": idx,\n                     "grade": int(grade_of(idx)), "gain": gain, "mud": mud})\n    meta = pd.DataFrame(rows)\n    # 장비 관리 기록(제원 요약): 사진 판정과 대조할 값\n    age = rng.integers(1, 25, n)\n    meta["day"] = np.sort(rng.integers(0, 90, n))\n    meta["equip_type"] = rng.choice(["vehicle", "generator", "comm", "trailer"], n, p=[0.4, 0.2, 0.2, 0.2])\n    meta["mfg_year"] = 2026 - age\n    meta["op_hours"] = np.clip(rng.normal(180 * age, 400, n) + 8 * meta["damage_idx"], 10, 20000)\n    meta["env"] = rng.choice(["coastal", "inland"], n, p=[0.3, 0.7])\n    reg = np.clip(meta["grade"] + rng.choice([-1, 0, 0, 0, 0, 0, 1, 2], n), 1, 3)           # 장비 관리 기록상 직전 정비 등급\n    meta["reg_grade"] = reg\n    meta["emergency"] = rng.random(n) < 0.04                                               # 긴급 점검 요청(화질 불량이어도 보존)\n    return meta, imgs, masks\n# ---- 전처리(조명 보정·표준화)와 결함 후보 분할, 특징 추출 (cv2 대신 numpy·scipy 사용) ----\ndef correct(img):\n    """조명 보정: 밝기에 평면(조명 기울기)을 적합해 나눈다. 어두운 결함 화소는 제외하고 두 번 반복 적합한다. 반환: 보정 밝기(정상 면 약 1), 색 비율."""\n    g = img.mean(axis=2)\n    A = np.c_[np.ones(SZ * SZ), XX.ravel() / SZ, YY.ravel() / SZ]\n    keep = np.ones(SZ * SZ, bool)\n    for _ in range(3):\n        coef, *_ = np.linalg.lstsq(A[keep], g.ravel()[keep], rcond=None)\n        plane = (A @ coef).reshape(SZ, SZ)\n        keep = (np.abs(g - plane) < 0.08 * np.maximum(plane, 1e-3)).ravel()\n        if keep.sum() < 200:\n            keep[:] = True\n    c = g / np.maximum(plane, 1e-3)\n    chroma = img[..., 0] / (img.sum(axis=2) + 1e-6)\n    return c, chroma\n\n\nSEGP = {"dark": 0.7, "rust": 0.5, "rim": 1.45}            # 분할 임계값(어두움 보정 밝기, 붉은 비율, 밝은 테두리): 재보정 대상\n\n\ndef segment(img):\n    """결함 후보 마스크: 어두운 영역(균열·누유·파손) + 붉은 갈색 영역(부식) + 밝은 테두리(파손)."""\n    c, chroma = correct(img)\n    dark = c < SEGP["dark"]\n    rust = (chroma > SEGP["rust"]) & (c > 0.45)\n    rim = c > SEGP["rim"]\n    m = dark | rust | ndi.binary_dilation(rim, iterations=1) & ndi.binary_dilation(dark, iterations=3)\n    m = ndi.binary_opening(m, structure=np.ones((2, 2))) if m.sum() > 12 else m\n    return m, c, chroma, dark, rust, rim\n\n\ndef extract(img):\n    """분할 결과와 색·모양에서 24개 특징을 만든다."""\n    m, c, chroma, dark, rust, rim = segment(img)\n    lab, n = ndi.label(m)\n    area = float(m.sum())\n    if n:\n        sizes = ndi.sum(m, lab, range(1, n + 1))\n        k = int(np.argmax(sizes)) + 1\n        sl = ndi.find_objects(lab)[k - 1]\n        h, w = sl[0].stop - sl[0].start, sl[1].stop - sl[1].start\n        big = float(sizes.max())\n        elong, fill = max(h, w) / max(min(h, w), 1), big / (h * w)\n    else:\n        big, elong, fill = 0.0, 1.0, 0.0\n    gy, gx = np.gradient(c)\n    mag = np.hypot(gx, gy)\n    inside = m.sum() > 0\n    return {"area_px": area, "area_frac": area / SZ ** 2, "n_comp": float(n), "big_px": big, "elong": elong, "fill": fill,\n            "dark_frac": float(dark.mean()), "rust_frac": float(rust.mean()), "rim_frac": float(rim.mean()),\n            "c_mean": float(c.mean()), "c_std": float(c.std()), "c_min": float(c.min()), "edge_mean": float(mag.mean()), "edge_p95": float(np.quantile(mag, 0.95)),\n            "chroma_mean": float(chroma.mean()), "chroma_max": float(chroma.max()),\n            "in_dark": float(c[m].mean()) if inside else 1.0, "in_chroma": float(chroma[m].mean()) if inside else float(chroma.mean()),\n            "ring_contrast": float(c[ndi.binary_dilation(m, iterations=3) & ~m].mean() - c[m].mean()) if inside and (~m).any() else 0.0,\n            "thin": float(1 - fill), "gain": float(img.mean()), "sat": float((img.max(axis=2) - img.min(axis=2)).mean()),\n            "cy": float(ndi.center_of_mass(m)[0] / SZ) if inside else 0.5, "cx": float(ndi.center_of_mass(m)[1] / SZ) if inside else 0.5}\n\n\ndef feature_table(imgs):\n    return pd.DataFrame([extract(im) for im in imgs])\n# ---- 판정 파이프라인: 결함 유형 분류 + 손상도 회귀(분류 확률 포함) -> 정비 등급 ----\ndef fit_pipeline(meta, F):\n    """결함 유형 분류기와 손상도 회귀기를 학습한다. 회귀기 입력의 분류 확률은 교차 예측으로 만들어 누설을 막는다."""\n    clf = RandomForestClassifier(n_estimators=150, random_state=0, n_jobs=1)\n    P = cross_val_predict(clf, F, meta["cls"], cv=4, method="predict_proba")\n    clf.fit(F, meta["cls"])\n    reg = GradientBoostingRegressor(n_estimators=200, max_depth=3, learning_rate=0.06, random_state=0).fit(np.c_[F.values, P], meta["damage_idx"])\n    return {"clf": clf, "reg": reg, "cols": list(F.columns)}\n\n\ndef predict_pipeline(pl, F):\n    X = F[pl["cols"]]\n    P = pl["clf"].predict_proba(X)\n    idx = np.clip(pl["reg"].predict(np.c_[X.values, P]), 0, 100)\n    return pl["clf"].classes_[P.argmax(axis=1)], P, idx, P.max(axis=1)\n# ---- 사진 품질 게이트: 진흙·오염 가림 추정과 판정 신뢰도 지수 ----\ndef mud_est(img):\n    """진흙(갈색, 채도 낮은 붉은 기) 화소 비율을 추정한다. 3x3 평균으로 잡음을 줄인 뒤 색 비율로 판단하며, 실제 가림 비율보다 작게 나오지만 순서는 유지된다."""\n    sm = ndi.uniform_filter(img, size=(3, 3, 1))\n    s = sm.sum(axis=2) + 1e-6\n    rr, gg = sm[..., 0] / s, sm[..., 1] / s\n    return float(((rr > 0.385) & (rr < 0.46) & (gg < 0.325)).mean())\n\n\ndef confidence(P, idx, F, mud):\n    """판정 신뢰도 지수 0~100 = 50 x 분류 확률 최대값 + 35 x 등급 경계까지 거리(8점에서 포화) + 15 x 사진 품질. 오염 의심(mud 추정 1% 이상)이면 상한 75로 제한한다."""\n    bd = np.minimum(np.abs(idx - 35), np.abs(idx - 65))\n    qual = np.clip(1.0 - np.clip(0.55 - F["gain"].values, 0, None) * 2 - np.clip(0.9 - F["c_mean"].values, 0, None), 0, 1)\n    c = 50 * P.max(axis=1) + 35 * np.clip(bd / 8, 0, 1) + 15 * qual\n    return np.where(np.asarray(mud) >= 0.01, np.minimum(c, 75.0), c)\n\nmtr, itr, _ = make_images(600, seed=1)\nFtr = feature_table(itr)\npl0 = fit_pipeline(mtr, Ftr)\nLOSS_UNDER, LOSS_OVER = 20.0, 6.0                                   # 과소/과대 판정 1등급당 손실(만원, 가정)\n\n\ndef ops(seed, n):\n    """운영 사진 묶음: 저조도 1/3 + 진흙 5% 1/3 + 정상 1/3 (학습 때와 분포가 다르다)."""\n    parts = [make_images(n // 3, seed=seed, dark=True), make_images(n // 3, seed=seed + 1, mud=0.05), make_images(n // 3, seed=seed + 2)]\n    return pd.concat([p[0] for p in parts], ignore_index=True), np.concatenate([p[1] for p in parts]), np.concatenate([p[2] for p in parts])\n\n\ndef loss(true, g):\n    d = np.asarray(g) - np.asarray(true)\n    return np.where(d < 0, -d * LOSS_UNDER, d * LOSS_OVER)\n\n\ndef grade_shift(idx, delta):\n    """과소 판정을 줄이기 위해 등급 경계를 delta만큼 낮춘다(35 -> 35-delta, 65 -> 65-delta)."""\n    return np.where(np.asarray(idx) >= 65 - delta, 3, np.where(np.asarray(idx) >= 35 - delta, 2, 1))\n\n\ndef evaluate(pl, F, m, ims, delta=0):\n    cls, P, idx, _ = predict_pipeline(pl, F)\n    mud = np.array([mud_est(x) for x in ims])\n    conf = confidence(P, idx, F, mud)\n    g = grade_shift(idx, delta)\n    auto = (conf >= 80) & (mud < 0.03) & (F["gain"].values >= 0.17)\n    ok = g == m["grade"].values\n    return {"cls": cls, "idx": idx, "g": g, "mud": mud, "conf": conf, "ok": ok, "mae": float(np.mean(np.abs(idx - m["damage_idx"].values))), "under": int((g < m["grade"].values).sum()),\n            "loss": float(loss(m["grade"].values, g).mean()), "auto_n": int(auto.sum()), "auto_acc": float(ok[auto].mean()) if auto.any() else np.nan}\n\n\n# ---------------- 1) AAR: 운영 로그 분석(이동 구간 A, 360장) ----------------\nmA, iA, kA = ops(60, 360)\nrng = np.random.default_rng(0)\nperm = rng.permutation(len(mA))\nFA = feature_table(iA)\nev = evaluate(pl0, FA, mA, iA)\nprint(f"[1] AAR 대상 운영 사진 {len(mA)}장(저조도·진흙 5%·정상 혼합): 등급 일치율 {ev[\'ok\'].mean():.1%}, 손상도 MAE {ev[\'mae\']:.2f}, 과소 판정 {ev[\'under\']}장, 장당 오판 손실 {ev[\'loss\']:.2f}만원")\nwrong = np.where(~ev["ok"])[0]\n\n\ndef cause(i):\n    if ev["mud"][i] >= 0.01: return "오염 가림"\n    if FA["gain"].iloc[i] < 0.2: return "저조도"\n    if mA["cls"][i] != "none" and mA["area_frac"][i] < 0.02: return "작은 결함 누락·과소"\n    if ev["cls"][i] != mA["cls"][i]: return "결함 유형 혼동"\n    if min(abs(ev["idx"][i] - 35), abs(ev["idx"][i] - 65)) < 6: return "등급 경계 근처"\n    return "기타"\n\n\ncs = pd.Series([cause(i) for i in wrong]).value_counts()\nprint(f"    오판 {len(wrong)}장의 원인 분류:", ", ".join(f"{k} {v}장({v / len(wrong):.0%})" for k, v in cs.items()))\nprint(f"    오판 중 과소 판정(실제보다 낮은 등급) {ev[\'under\']}장 / 과대 {int((ev[\'g\'] > mA[\'grade\'].values).sum())}장 -> 과소 판정이 손실 단가가 {LOSS_UNDER / LOSS_OVER:.1f}배 높아 우선 교정 대상")\n\n# ---------------- 2) 재보정: 분할 임계값 -> 운영 데이터 재학습 -> 등급 경계 이동 ----------------\nA1, A2 = perm[:240], perm[240:]\ncache = [correct(iA[i]) for i in A1]\n\n\ndef mask_from(c, chroma, dt, rt, rim=1.45):\n    dark, rust, r = c < dt, (chroma > rt) & (c > 0.45), c > rim\n    m = dark | rust | ndi.binary_dilation(r, iterations=1) & ndi.binary_dilation(dark, iterations=3)\n    return ndi.binary_opening(m, structure=np.ones((2, 2))) if m.sum() > 12 else m\n\n\ngrid = []\nfor dt in (0.6, 0.7, 0.75, 0.8):\n    for rt in (0.44, 0.5, 0.54):\n        iou = np.mean([(mask_from(c, ch, dt, rt) & kA[i]).sum() / max((mask_from(c, ch, dt, rt) | kA[i]).sum(), 1) for (c, ch), i in zip(cache, A1)])\n        grid.append((dt, rt, iou))\ngd = pd.DataFrame(grid, columns=["dark 임계값", "rust 임계값", "평균 IoU"])\nbst = gd.loc[gd["평균 IoU"].idxmax()]\ncur = gd[(gd["dark 임계값"] == 0.7) & (gd["rust 임계값"] == 0.5)]["평균 IoU"].iloc[0]\nprint(f"\\n[2] 분할 임계값 재보정(운영 사진 240장 IoU): 현행(0.70, 0.50) {cur:.3f} -> 최적({bst[\'dark 임계값\']:.2f}, {bst[\'rust 임계값\']:.2f}) {bst[\'평균 IoU\']:.3f}")\nSEGP["dark"], SEGP["rust"] = float(bst["dark 임계값"]), float(bst["rust 임계값"])\nFA2 = feature_table(iA)\nmm = pd.concat([mtr, mA.iloc[A1]], ignore_index=True)\npl1 = fit_pipeline(mm, pd.concat([Ftr, FA2.iloc[A1]], ignore_index=True))                   # 운영 분포 사진 240장을 추가 학습\nev2 = evaluate(pl1, FA2.iloc[A2].reset_index(drop=True), mA.iloc[A2].reset_index(drop=True), iA[A2])\ncosts = []\nfor dlt in (0, 2, 4, 6, 8):\n    r = evaluate(pl1, FA2.iloc[A2].reset_index(drop=True), mA.iloc[A2].reset_index(drop=True), iA[A2], delta=dlt)\n    costs.append((dlt, r["loss"], r["under"], r["ok"].mean()))\ncd = pd.DataFrame(costs, columns=["경계 이동 delta", "장당 손실(만원)", "과소 판정", "일치율"])\nbest_d = int(cd.loc[cd["장당 손실(만원)"].idxmin(), "경계 이동 delta"])\nprint("    등급 경계 이동(보정용 120장, 과소 판정 손실 반영)\\n" + cd.round(3).to_string(index=False))\nif best_d == 0:\n    print("    경계를 낮추면 과대 판정이 늘어 손실이 오히려 커지므로 경계 이동은 채택하지 않는다(delta=0 유지). 과소 판정은 경계가 아니라 오염 가림에서 주로 발생한다.")\nelse:\n    print(f"    손실이 가장 작은 경계 이동 delta={best_d}점 채택(과소 판정을 줄이는 방향이며 사람 승인 후 적용)")\n\n# ---------------- 3) 보류 구간 B(360장, 재보정에 쓰지 않음) 검증 ----------------\nmB, iB, kB = ops(70, 360)\nSEGP["dark"], SEGP["rust"] = 0.7, 0.5\nFB0 = feature_table(iB)\ne0 = evaluate(pl0, FB0, mB, iB)\nSEGP["dark"], SEGP["rust"] = float(bst["dark 임계값"]), float(bst["rust 임계값"])\nFB1 = feature_table(iB)\ne1 = evaluate(pl1, FB1, mB, iB, delta=best_d)\ntab = pd.DataFrame({"재보정 전": [e0["ok"].mean(), e0["mae"], e0["under"], e0["loss"], e0["auto_n"], e0["auto_acc"]], "재보정 후": [e1["ok"].mean(), e1["mae"], e1["under"], e1["loss"], e1["auto_n"], e1["auto_acc"]]},\n                   index=["등급 일치율", "손상도 MAE", "과소 판정(장)", "장당 오판 손실(만원)", "자동 처리 건수", "자동 처리 일치율"])\nprint("\\n[3] 보류 구간 B 검증\\n" + tab.round(3).to_string())\nprint(f"    오판 손실 {1 - e1[\'loss\'] / e0[\'loss\']:.1%} 감소, 과소 판정 {e0[\'under\']} -> {e1[\'under\']}장. 오염 가림 사진은 정보 자체가 가려져 재보정으로 해결되지 않으므로 재촬영이 필요하다.")\n\n# ---------------- 4) ROI 3대 지표(교육용 가정) ----------------\nN_YR, MIN_FULL, MIN_QUICK = 20000, 15.0, 4.0                         # 연 판정 사진 수, 정밀 판독/간이 승인 소요(분)\nWAGE, INV, OPEX = 1.2, 2500.0, 150.0                                 # 검사관 인건비(만원/시), 도입 비용, 연 운영비(만원)\nfull_b, full_a = 1 - e0["auto_n"] / len(mB), 1 - e1["auto_n"] / len(mB)\n\n\ndef roi(loss_b, loss_a, fb, fa, n=N_YR, inv=INV):\n    t_b = fb * MIN_FULL + (1 - fb) * MIN_QUICK\n    t_a = fa * MIN_FULL + (1 - fa) * MIN_QUICK\n    save_loss = (loss_b - loss_a) * n\n    save_time = (t_b - t_a) / 60 * WAGE * n                      # 기존 AI 대비 개선 AI 판독 인건비 절감(만원/년)\n    save = save_loss + save_time - OPEX\n    return save, 1 - t_a / t_b, (inv / (save / 12) if save > 0 else float("inf")), save_loss, save_time\n\n\nsv, tr, pb, sl, st = roi(e0["loss"], e1["loss"], full_b, full_a)\nt_b = full_b * MIN_FULL + (1 - full_b) * MIN_QUICK\nt_a = full_a * MIN_FULL + (1 - full_a) * MIN_QUICK\nroi_table = pd.DataFrame({\n    "기존 AI": [len(mB), N_YR, t_b, e0["loss"], t_b / 60 * WAGE * N_YR, e0["loss"] * N_YR, 0.0],\n    "개선 AI": [len(mB), N_YR, t_a, e1["loss"], t_a / 60 * WAGE * N_YR, e1["loss"] * N_YR, OPEX],\n}, index=["관측 B 표본(장)", "연간 물량(장/년)", "평균 판독(분/장)", "오판 손실(만원/장)", "연 판독 인건비(만원/년)", "연 오판 손실(만원/년)", "추가 운영비(만원/년)"])\nroi_table.loc["연 총비용(만원/년)"] = roi_table.iloc[4:7].sum()\nassert np.isclose(roi_table.loc["연 총비용(만원/년)", "기존 AI"] - roi_table.loc["연 총비용(만원/년)", "개선 AI"], sv)\nprint("\\n수정 실습본: 동일 기준 전후 비교표 [교육용 가정]\\n" + roi_table.round(6).to_string())\nprint("추가 투자비 2500만원은 기존 AI에서 개선 AI로 전환하는 증분 투자 가정이다. 운영비 150만원/년 역시 증분이며 공통비용은 상쇄한다.")\nprint("회수기간은 순절감액이 0 이하이면 inf(회수 불가)로 표시한다. B 표본의 평균 손실·시간을 연 물량으로 외삽한 가정이며 실제 현장 관측이 아니다.")\nprint(f"\\n[4] ROI 3대 지표 [교육용 가정: 연 {N_YR}장, 정밀 판독 {MIN_FULL:.0f}분, 간이 승인 {MIN_QUICK:.0f}분, 인건비 {WAGE}만원/시, 도입 {INV:.0f}만원, 운영 {OPEX:.0f}만원/년]")\nprint(f"    (1) 연간 비용 절감액 {sv:,.0f}만원(오판 손실 감소 {sl:,.0f} + 판독 인건비 {st:,.0f} - 운영비)  (2) 판독 시간 단축률 {tr:.1%}(기존 AI 대비 개선 AI; 같은 B 360장 기준)  (3) 투자 회수 기간 {pb:.1f}개월")\nprint("    산식: 절감액 = (재보정 전 - 후 장당 오판 손실) x 연 장수 + (기존 AI 평균 분/장 - 개선 AI 평균 분/장)/60 x 인건비 x 연 장수 - 운영비")\nprint("    장수 민감도:", "; ".join(f"{n}장 -> 절감 {roi(e0[\'loss\'], e1[\'loss\'], full_b, full_a, n=n)[0]:,.0f}만원, 회수 {roi(e0[\'loss\'], e1[\'loss\'], full_b, full_a, n=n)[2]:.0f}개월" for n in (5000, 10000, 20000, 40000)))\nprint("    비관(재보정 효과 없음, 판독 시간 절감만):", f"절감 {roi(e0[\'loss\'], e0[\'loss\'], full_b, full_a)[0]:,.0f}만원, 회수 {roi(e0[\'loss\'], e0[\'loss\'], full_b, full_a)[2]:.0f}개월")\nprint("    가정값이므로 재정·기획 부서와 실제 단가로 다시 산정해야 한다. 재보정된 임계값·경계는 정비감독관 승인 후 적용한다.")\n\nfig, ax = plt.subplots(1, 3, figsize=(15, 4.2))\nax[0].bar(range(len(cs)), cs.values); ax[0].set_xticks(range(len(cs))); ax[0].set_xticklabels([f"C{i + 1}" for i in range(len(cs))]); ax[0].set_title("AAR: misjudgment causes (count)")\nax[1].plot(cd["경계 이동 delta"], cd["장당 손실(만원)"], "o-"); ax[1].set_title("Loss per photo vs grade-boundary shift"); ax[1].set_xlabel("delta")\nx = np.arange(3); ax[2].bar(x - .2, [e0["ok"].mean() * 100, e0["under"], e0["loss"]], .4, label="before"); ax[2].bar(x + .2, [e1["ok"].mean() * 100, e1["under"], e1["loss"]], .4, label="after")\nax[2].set_xticks(x); ax[2].set_xticklabels(["accuracy %", "under-grade n", "loss/photo"]); ax[2].legend(); ax[2].set_title("Hold-out B: before vs after")\nplt.tight_layout(); plt.show()\n', 'work/revised/m10.py': '# 수정 실습본: 원문 코드는 별도 보존합니다.\n"""신규 사진 갱신, 정합성 검증, 분포 변화(PSI) 점검, 자동 재학습(챔피언/챌린저), 결함 판정·손상도·정비 등급 산출과 검사 대기·재촬영 분리, 제원 불일치 환류까지 한 번에 실행하는 통합 파이프라인을 완성하고 운영 문서 완비율·자체 점검·인터페이스 명세·브리핑 요약을 출력한다."""\nimport importlib\nimport json\nimport sys\nimport numpy as np\nimport pandas as pd\nimport matplotlib.pyplot as plt\nfrom sklearn.ensemble import RandomForestClassifier, GradientBoostingRegressor\nfrom sklearn.linear_model import LinearRegression\nfrom sklearn.model_selection import cross_val_predict\n\n# ---------------------------------------------------------------------------\n# [교육용가상자료] 장비 외관 사진 생성기 (실제 장비·조직·위치 사진이 아니라 난수로 그린 64x64 합성 이미지)\n# 도장된 금속 표면에 균열(crack)·부식(corrosion)·파손(damage)·누유(leak)를 그리고, 조명 편차와 진흙(mud) 오염을 섞는다.\n# 이미지와 함께 장비 관리 기록(제원 요약)과 정답 결함 마스크·손상도·정비 등급을 만든다.\n# ---------------------------------------------------------------------------\nfrom scipy import ndimage as ndi\n\nSZ = 64\nCLASSES = ["none", "crack", "corrosion", "damage", "leak"]\nKO_CLS = {"none": "결함 없음", "crack": "균열", "corrosion": "부식", "damage": "파손", "leak": "누유"}\nCLS_W = {"none": 0.0, "crack": 1.4, "corrosion": 1.0, "damage": 1.6, "leak": 0.6}      # 손상도 계산용 결함 유형 가중치\nGRADE_NAME = {1: "1단계 정비", 2: "2단계 정비", 3: "3단계 정비"}\nYY, XX = np.mgrid[0:SZ, 0:SZ]\n\n\ndef grade_of(idx):\n    return np.where(np.asarray(idx) >= 65, 3, np.where(np.asarray(idx) >= 35, 2, 1))\n\n\ndef blob(rng, cy, cx, r, rough=0.35):\n    """불규칙한 원형 영역 마스크."""\n    noise = ndi.gaussian_filter(rng.normal(size=(SZ, SZ)), 3)\n    noise = noise / (np.abs(noise).max() + 1e-9)\n    return (np.hypot(YY - cy, XX - cx) + rough * r * 2.2 * noise) < r\n\n\ndef draw(rng, cls, sev):\n    """결함 유형·심각도 sev(0~1)에 맞는 정답 마스크와 색 정보를 만든다."""\n    m = np.zeros((SZ, SZ), bool)\n    if cls == "crack":\n        y, x, ang = rng.uniform(12, 52), rng.uniform(12, 52), rng.uniform(0, np.pi)\n        for _ in range(int(20 + 40 * sev)):\n            ang += rng.normal(0, 0.35); y += np.sin(ang) * 1.2; x += np.cos(ang) * 1.2\n            if 0 <= int(y) < SZ and 0 <= int(x) < SZ:\n                m[int(y), int(x)] = True\n        m = ndi.binary_dilation(m, iterations=1 + int(sev > 0.6))\n    elif cls == "corrosion":\n        for _ in range(int(rng.integers(1, 4))):\n            m |= blob(rng, rng.uniform(14, 50), rng.uniform(14, 50), 3 + 9 * sev * rng.uniform(0.6, 1.1))\n    elif cls == "damage":\n        m = blob(rng, rng.uniform(20, 44), rng.uniform(20, 44), 6 + 12 * sev, 0.3)\n    elif cls == "leak":\n        x0, y0 = rng.uniform(16, 48), rng.uniform(8, 24)\n        w, ln = 2 + 4 * sev, 12 + 26 * sev\n        m = (np.abs(XX - x0 - 0.15 * (YY - y0)) < w * (1 - 0.4 * (YY - y0) / ln)) & (YY > y0) & (YY < y0 + ln)\n    return m\n\n\ndef make_images(n=600, seed=0, mud=0.0, p_none=0.25, dark=False):\n    """가상 외관 사진 n장(+진흙 오염 mud 비율)과 관리 기록·정답. 반환: meta(DataFrame), imgs(n,64,64,3), masks(n,64,64)."""\n    rng = np.random.default_rng(seed)\n    imgs, masks, rows = np.zeros((n, SZ, SZ, 3), np.float32), np.zeros((n, SZ, SZ), bool), []\n    for i in range(n):\n        cls = "none" if rng.random() < p_none else str(rng.choice(CLASSES[1:], p=[0.25, 0.3, 0.25, 0.2]))\n        sev = 0.0 if cls == "none" else float(rng.beta(1.6, 2.0))\n        base = np.array([0.42, 0.46, 0.40]) * (1 + 0.5 * ndi.gaussian_filter(rng.normal(size=(SZ, SZ)), 8)[..., None] * 3)\n        img = base + rng.normal(0, 0.012, (SZ, SZ, 3))\n        m = draw(rng, cls, sev)\n        a = ndi.gaussian_filter(m.astype(float), 0.8)[..., None]                        # 경계를 부드럽게 섞는다\n        if cls == "crack":\n            col = img * rng.uniform(0.2, 0.45)\n        elif cls == "corrosion":\n            col = np.array([0.55, 0.30, 0.12]) * rng.uniform(0.75, 1.15) * (0.85 + 0.3 * rng.random((SZ, SZ, 1)))\n        elif cls == "damage":\n            col = np.ones_like(img) * np.array([0.34, 0.34, 0.36]) * (1.0 - 0.45 * sev) * rng.uniform(0.75, 1.0)\n            rim = m & ~ndi.binary_erosion(m, iterations=2)\n            a = np.maximum(a, 0.0); img = np.where(rim[..., None], 0.7, img)\n        elif cls == "leak":\n            col = np.ones_like(img) * np.array([0.10, 0.09, 0.08]) * rng.uniform(0.8, 2.6)\n        else:\n            col = img\n        img = img * (1 - a) + col * a\n        if rng.random() < 0.35:                                   # 무해한 얼룩(결함 아님): 옅은 갈색 번짐\n            st = blob(rng, rng.uniform(5, 58), rng.uniform(5, 58), rng.uniform(6, 14), 0.5)\n            al = ndi.gaussian_filter(st.astype(float), 2.5)[..., None] * rng.uniform(0.15, 0.4)\n            img = img * (1 - al) + np.array([0.50, 0.40, 0.30]) * al\n        if rng.random() < 0.3:                                    # 무해한 긴 스크래치(밝은 선)\n            y0, x0, ang = rng.uniform(5, 58), rng.uniform(5, 58), rng.uniform(0, np.pi)\n            for t in np.linspace(0, 25, 40):\n                yy, xx = int(y0 + t * np.sin(ang)), int(x0 + t * np.cos(ang))\n                if 0 <= yy < SZ and 0 <= xx < SZ:\n                    img[yy, xx] = np.clip(img[yy, xx] * 1.5, 0, 1)\n        if mud > 0:                                           # 진흙 오염: 결함 위를 덮을 수 있다(정답 마스크는 그대로)\n            mm = np.zeros((SZ, SZ), bool)\n            while mm.mean() < mud:\n                mm |= blob(rng, rng.uniform(0, SZ), rng.uniform(0, SZ), rng.uniform(5, 11), 0.5)\n            img[mm] = np.array([0.38, 0.30, 0.22]) * (0.85 + 0.3 * rng.random((mm.sum(), 1)))\n        gain = rng.uniform(0.3, 0.55) if dark else rng.uniform(0.5, 1.15)\n        ang = rng.uniform(0, 2 * np.pi)\n        light = gain * (1 + 0.3 * (np.cos(ang) * (XX - 32) + np.sin(ang) * (YY - 32)) / 32)\n        imgs[i] = np.clip(img * light[..., None] + rng.normal(0, 0.025, (SZ, SZ, 3)), 0, 1)\n        masks[i] = m\n        area = float(m.mean())\n        depth = {"none": 0.0, "crack": 0.3 + 3.0 * sev, "corrosion": 0.1 + 1.5 * sev, "damage": 1.0 + 8.0 * sev, "leak": 0.0}[cls] * rng.lognormal(0, 0.1)\n        idx = float(np.clip(100 * (1 - np.exp(-(CLS_W[cls] * area * 9 + 0.05 * depth))) + rng.normal(0, 1.5), 0, 100))\n        rows.append({"img_id": f"IMG{seed:02d}{i:04d}", "cls": cls, "sev": sev, "area_frac": area, "depth_mm": depth, "damage_idx": idx,\n                     "grade": int(grade_of(idx)), "gain": gain, "mud": mud})\n    meta = pd.DataFrame(rows)\n    # 장비 관리 기록(제원 요약): 사진 판정과 대조할 값\n    age = rng.integers(1, 25, n)\n    meta["day"] = np.sort(rng.integers(0, 90, n))\n    meta["equip_type"] = rng.choice(["vehicle", "generator", "comm", "trailer"], n, p=[0.4, 0.2, 0.2, 0.2])\n    meta["mfg_year"] = 2026 - age\n    meta["op_hours"] = np.clip(rng.normal(180 * age, 400, n) + 8 * meta["damage_idx"], 10, 20000)\n    meta["env"] = rng.choice(["coastal", "inland"], n, p=[0.3, 0.7])\n    reg = np.clip(meta["grade"] + rng.choice([-1, 0, 0, 0, 0, 0, 1, 2], n), 1, 3)           # 장비 관리 기록상 직전 정비 등급\n    meta["reg_grade"] = reg\n    meta["emergency"] = rng.random(n) < 0.04                                               # 긴급 점검 요청(화질 불량이어도 보존)\n    return meta, imgs, masks\n# ---- 전처리(조명 보정·표준화)와 결함 후보 분할, 특징 추출 (cv2 대신 numpy·scipy 사용) ----\ndef correct(img):\n    """조명 보정: 밝기에 평면(조명 기울기)을 적합해 나눈다. 어두운 결함 화소는 제외하고 두 번 반복 적합한다. 반환: 보정 밝기(정상 면 약 1), 색 비율."""\n    g = img.mean(axis=2)\n    A = np.c_[np.ones(SZ * SZ), XX.ravel() / SZ, YY.ravel() / SZ]\n    keep = np.ones(SZ * SZ, bool)\n    for _ in range(3):\n        coef, *_ = np.linalg.lstsq(A[keep], g.ravel()[keep], rcond=None)\n        plane = (A @ coef).reshape(SZ, SZ)\n        keep = (np.abs(g - plane) < 0.08 * np.maximum(plane, 1e-3)).ravel()\n        if keep.sum() < 200:\n            keep[:] = True\n    c = g / np.maximum(plane, 1e-3)\n    chroma = img[..., 0] / (img.sum(axis=2) + 1e-6)\n    return c, chroma\n\n\nSEGP = {"dark": 0.7, "rust": 0.5, "rim": 1.45}            # 분할 임계값(어두움 보정 밝기, 붉은 비율, 밝은 테두리): 재보정 대상\n\n\ndef segment(img):\n    """결함 후보 마스크: 어두운 영역(균열·누유·파손) + 붉은 갈색 영역(부식) + 밝은 테두리(파손)."""\n    c, chroma = correct(img)\n    dark = c < SEGP["dark"]\n    rust = (chroma > SEGP["rust"]) & (c > 0.45)\n    rim = c > SEGP["rim"]\n    m = dark | rust | ndi.binary_dilation(rim, iterations=1) & ndi.binary_dilation(dark, iterations=3)\n    m = ndi.binary_opening(m, structure=np.ones((2, 2))) if m.sum() > 12 else m\n    return m, c, chroma, dark, rust, rim\n\n\ndef extract(img):\n    """분할 결과와 색·모양에서 24개 특징을 만든다."""\n    m, c, chroma, dark, rust, rim = segment(img)\n    lab, n = ndi.label(m)\n    area = float(m.sum())\n    if n:\n        sizes = ndi.sum(m, lab, range(1, n + 1))\n        k = int(np.argmax(sizes)) + 1\n        sl = ndi.find_objects(lab)[k - 1]\n        h, w = sl[0].stop - sl[0].start, sl[1].stop - sl[1].start\n        big = float(sizes.max())\n        elong, fill = max(h, w) / max(min(h, w), 1), big / (h * w)\n    else:\n        big, elong, fill = 0.0, 1.0, 0.0\n    gy, gx = np.gradient(c)\n    mag = np.hypot(gx, gy)\n    inside = m.sum() > 0\n    return {"area_px": area, "area_frac": area / SZ ** 2, "n_comp": float(n), "big_px": big, "elong": elong, "fill": fill,\n            "dark_frac": float(dark.mean()), "rust_frac": float(rust.mean()), "rim_frac": float(rim.mean()),\n            "c_mean": float(c.mean()), "c_std": float(c.std()), "c_min": float(c.min()), "edge_mean": float(mag.mean()), "edge_p95": float(np.quantile(mag, 0.95)),\n            "chroma_mean": float(chroma.mean()), "chroma_max": float(chroma.max()),\n            "in_dark": float(c[m].mean()) if inside else 1.0, "in_chroma": float(chroma[m].mean()) if inside else float(chroma.mean()),\n            "ring_contrast": float(c[ndi.binary_dilation(m, iterations=3) & ~m].mean() - c[m].mean()) if inside and (~m).any() else 0.0,\n            "thin": float(1 - fill), "gain": float(img.mean()), "sat": float((img.max(axis=2) - img.min(axis=2)).mean()),\n            "cy": float(ndi.center_of_mass(m)[0] / SZ) if inside else 0.5, "cx": float(ndi.center_of_mass(m)[1] / SZ) if inside else 0.5}\n\n\ndef feature_table(imgs):\n    return pd.DataFrame([extract(im) for im in imgs])\n# ---- 판정 파이프라인: 결함 유형 분류 + 손상도 회귀(분류 확률 포함) -> 정비 등급 ----\ndef fit_pipeline(meta, F):\n    """결함 유형 분류기와 손상도 회귀기를 학습한다. 회귀기 입력의 분류 확률은 교차 예측으로 만들어 누설을 막는다."""\n    clf = RandomForestClassifier(n_estimators=150, random_state=0, n_jobs=1)\n    P = cross_val_predict(clf, F, meta["cls"], cv=4, method="predict_proba")\n    clf.fit(F, meta["cls"])\n    reg = GradientBoostingRegressor(n_estimators=200, max_depth=3, learning_rate=0.06, random_state=0).fit(np.c_[F.values, P], meta["damage_idx"])\n    return {"clf": clf, "reg": reg, "cols": list(F.columns)}\n\n\ndef predict_pipeline(pl, F):\n    X = F[pl["cols"]]\n    P = pl["clf"].predict_proba(X)\n    idx = np.clip(pl["reg"].predict(np.c_[X.values, P]), 0, 100)\n    return pl["clf"].classes_[P.argmax(axis=1)], P, idx, P.max(axis=1)\n# ---- 사진 품질 게이트: 진흙·오염 가림 추정과 판정 신뢰도 지수 ----\ndef mud_est(img):\n    """진흙(갈색, 채도 낮은 붉은 기) 화소 비율을 추정한다. 3x3 평균으로 잡음을 줄인 뒤 색 비율로 판단하며, 실제 가림 비율보다 작게 나오지만 순서는 유지된다."""\n    sm = ndi.uniform_filter(img, size=(3, 3, 1))\n    s = sm.sum(axis=2) + 1e-6\n    rr, gg = sm[..., 0] / s, sm[..., 1] / s\n    return float(((rr > 0.385) & (rr < 0.46) & (gg < 0.325)).mean())\n\n\ndef confidence(P, idx, F, mud):\n    """판정 신뢰도 지수 0~100 = 50 x 분류 확률 최대값 + 35 x 등급 경계까지 거리(8점에서 포화) + 15 x 사진 품질. 오염 의심(mud 추정 1% 이상)이면 상한 75로 제한한다."""\n    bd = np.minimum(np.abs(idx - 35), np.abs(idx - 65))\n    qual = np.clip(1.0 - np.clip(0.55 - F["gain"].values, 0, None) * 2 - np.clip(0.9 - F["c_mean"].values, 0, None), 0, 1)\n    c = 50 * P.max(axis=1) + 35 * np.clip(bd / 8, 0, 1) + 15 * qual\n    return np.where(np.asarray(mud) >= 0.01, np.minimum(c, 75.0), c)\n\nlibs = ["numpy", "pandas", "scipy", "sklearn", "matplotlib"]\nenv_ok = sum(importlib.util.find_spec(m) is not None for m in libs)\nprint(f"[0] 환경 점검: Python {sys.version_info.major}.{sys.version_info.minor}, 필수 모듈 {env_ok}/{len(libs)}")\n\n# ---------------- 1) 신규 사진 갱신·정합성 검증(미션 1~2 요약) ----------------\nm_old, i_old, k_old = make_images(600, seed=1)                                     # 기존 600장\nparts = [make_images(80, seed=80, dark=True), make_images(80, seed=81, mud=0.05), make_images(80, seed=82)]\nm_new = pd.concat([p[0] for p in parts], ignore_index=True); i_new = np.concatenate([p[1] for p in parts]); k_new = np.concatenate([p[2] for p in parts])\nm_new["img_id"] = ["NEW" + x[3:] for x in m_new["img_id"]]                         # 신규 사진 ID 구분\nmeta = pd.concat([m_old, m_new], ignore_index=True); imgs = np.concatenate([i_old, i_new]); masks = np.concatenate([k_old, k_new])\nchk = {"사진 ID 유일": not meta["img_id"].duplicated().any(), "결측 0(제원·라벨)": int(meta[["equip_type", "mfg_year", "op_hours", "env", "reg_grade"]].isna().sum().sum()) == 0,\n       "화소 범위 0~1": bool((imgs.min() >= 0) and (imgs.max() <= 1)), "해상도 64x64x3": imgs.shape[1:] == (64, 64, 3), "제조연도 가능 범위": bool(meta["mfg_year"].between(1990, 2026).all()),\n       "마스크 면적 = 라벨 면적": bool(np.allclose(masks.mean(axis=(1, 2)), meta["area_frac"], atol=1e-6)), "손상도-등급 일치": bool((grade_of(meta["damage_idx"]) == meta["grade"]).all())}\nprint(f"[1] 신규 240장 포함 {len(meta)}장 정합성 검증 {sum(chk.values())}/{len(chk)} 통과, 긴급 점검 사진 {int(meta[\'emergency\'].sum())}장 보존")\nF = feature_table(imgs)\nF["mud"] = [mud_est(x) for x in imgs]\nn_old = len(m_old)\n\n# ---------------- 2) 분포 변화 점검(PSI) 및 자동 재학습 ----------------\ndef psi(a, b, bins=8):\n    q = np.unique(np.quantile(a, np.linspace(0, 1, bins + 1)))\n    q[0], q[-1] = -np.inf, np.inf\n    pa, pb = np.histogram(a, q)[0] / len(a) + 1e-4, np.histogram(b, q)[0] / len(b) + 1e-4\n    return float(np.sum((pa - pb) * np.log(pa / pb)))\n\n\ndrift = {c: psi(F[c][:n_old], F[c][n_old:]) for c in ["gain", "c_mean", "c_std", "rust_frac", "area_frac", "mud"]}\nworst = max(drift, key=drift.get)\nprint("[2] 분포 변화 PSI(기존 600 vs 신규 240):", ", ".join(f"{k} {v:.2f}" for k, v in drift.items()), f"-> 가장 큰 변화 {worst} (0.2 이상이면 재학습 검토)")\nrng = np.random.default_rng(0)\nnew_idx = rng.permutation(np.arange(n_old, len(meta)))\ntr_new, hold = new_idx[:160], new_idx[160:]                                        # 신규 160장은 재학습, hold80은 모델 선택용(최종 시험 아님)\ncols = [c for c in F.columns if c != "mud"]\n\n\ndef fit_on(ix):\n    return fit_pipeline(meta.iloc[ix].reset_index(drop=True), F.iloc[ix][cols].reset_index(drop=True))\n\n\ndef score(pl, ix):\n    cls, P, idx, _ = predict_pipeline(pl, F.iloc[ix][cols])\n    y = meta["damage_idx"].values[ix]\n    return float(np.mean(np.abs(y - idx))), float(np.mean(grade_of(idx) == meta["grade"].values[ix]))\n\n\nchamp = fit_on(np.arange(n_old))\nchall = fit_on(np.r_[np.arange(n_old), tr_new])\na, b = score(champ, hold), score(chall, hold)\npromote = (a[0] - b[0]) / a[0] >= 0.03\ndrifted = max(drift.values()) >= 0.2\nprint(f"    검증 80장(신규 분포, 모델 선택용; 최종 시험 아님): 챔피언 MAE {a[0]:.2f}·등급 일치 {a[1]:.1%} vs 챌린저 MAE {b[0]:.2f}·등급 일치 {b[1]:.1%} -> {\'챌린저 승격\' if promote else \'기존 모델 유지\'}(MAE 3% 이상 개선일 때만 승격)")\nmodel = chall if promote else champ\n\n# ---------------- 3) 신규 사진 판정과 검사 대기 분리 ----------------\nXn = F.iloc[n_old:][cols].reset_index(drop=True)\ncls, P, idx, _ = predict_pipeline(model, Xn)\nmud = F["mud"].values[n_old:]\nconf = confidence(P, idx, Xn, mud)\nres = pd.DataFrame({"사진": meta["img_id"].values[n_old:], "판정 유형": [KO_CLS[c] for c in cls], "손상도": idx, "AI 등급": grade_of(idx), "신뢰도": conf, "오염": mud, "기록 등급": meta["reg_grade"].values[n_old:], "실제 등급": meta["grade"].values[n_old:]})\nres["처리"] = np.select([(mud >= 0.03) | (Xn["gain"].values < 0.17), conf >= 80], ["재촬영 요구", "검사 대기(간이 승인 후보)"], "검사 대기(정밀 판독)")\nres["제원 신호"] = np.abs(res["AI 등급"] - res["기록 등급"]) >= 2\nprint(f"\\n[3] 신규 {len(res)}장 판정 결과 [교육용가상자료] (상위 6건)")\nprint(res.head(6)[["사진", "판정 유형", "손상도", "AI 등급", "신뢰도", "오염", "처리"]].round(2).to_string(index=False))\nvc = res["처리"].value_counts()\nprint("    처리 구분:", vc.to_dict(), "/ 자동 확정 0건(모든 판정은 검사관 확인과 정비감독관 결재 필요)")\nauto = res["처리"].str.contains("간이")\nprint(f"    운영 처리량 {len(res)}장, 간이 승인 후보 {int(auto.sum())}장. 학습·선택 자료가 포함되어 이 집합의 정확도를 최종 성능으로 보고하지 않는다.")\nprint(f"    제원 환류: AI 등급과 관리 기록 등급이 2단계 이상 다른 {int(res[\'제원 신호\'].sum())}장은 제원 수정 요청 후보(자동 수정 금지, 담당자 확인)")\n\n# ---------------- 4) 운영 문서 완비율·인터페이스 명세 ----------------\nfrom pathlib import Path\nimport re\n# 실행 파일 위치로 루트를 정하므로 python -I work/revised/m10.py도 같은 문서를 읽는다.\nPRACTICE_ROOT = next(p for p in Path(__file__).resolve().parents if (p / "configs/checks.json").exists())\nDOC_HEADINGS = [\'데이터 사전·스키마(M1)\', \'정제 규칙(M2)\', \'Baseline·EDA(M3)\', \'분할·손상도 모델(M4)\', \'등급·신뢰도 엔진(M5)\', \'판정 인터페이스·설명(M6)\', \'승인 절차·권한(M7)\', \'오염 스트레스·한계치(M8)\', \'AAR·ROI(M9)\', \'운영·재학습 절차(M10)\', \'장애 대응\', \'인터페이스 가이드\']\ndef document_checks(path):\n    text = path.read_text(encoding="utf-8") if path.exists() else ""\n    sections = dict(re.findall(r"(?m)^## (.+)\\n([\\s\\S]*?)(?=^## |\\Z)", text))\n    return {heading: len(sections.get(heading, "").strip()) >= 30 and not re.search(r"TODO|\\(작성\\)|미작성|여기에 작성", sections.get(heading, "")) for heading in DOC_HEADINGS}\ndocs = document_checks(PRACTICE_ROOT / "docs/qlt03_operations_guide.md")\ndone = sum(docs.values())\niface = {"input": {"photo": ["img_id:str", "image:64x64x3 float", "equip_type:str", "reg_grade:int"], "format": "image+json"},\n         "output": {"result": ["img_id", "defect", "damage_idx", "grade", "confidence", "mud", "state"], "approval": "PENDING_ONLY"}, "auth": "role-based(가상)", "note": "상위 물자정보체계는 가상 인터페이스로만 정의"}\nprint(f"\\n[4] 운영 문서 {len(docs)}개 항목 중 작성 완료 {done}개 ({done / len(docs):.0%}) / 인터페이스 명세(JSON):")\nprint("   ", json.dumps(iface, ensure_ascii=False))\n\n# ---------------- 5) 자체 점검·브리핑 ----------------\n# 학습/모델 선택이 끝난 뒤 고정 시험 자료를 처음 생성한다.\n# 정상/저조도/진흙 각 80장. seed 및 구성은 성능을 보기 전에 고정하며 튜닝에 재사용하지 않는다.\ntest_parts = [make_images(80, seed=180, dark=True), make_images(80, seed=181, mud=0.05), make_images(80, seed=182)]\nm_test = pd.concat([p[0] for p in test_parts], ignore_index=True)\ni_test = np.concatenate([p[1] for p in test_parts])\nF_test = feature_table(i_test)\n_, P_test, idx_test, _ = predict_pipeline(model, F_test[cols])\ny = m_test["damage_idx"].to_numpy()\ntrain_ids = set(meta.iloc[np.r_[np.arange(n_old), tr_new]]["img_id"])\nselection_ids = set(meta.iloc[hold]["img_id"])\ntest_ids = set(m_test["img_id"])\nintersections = {"train_selection": len(train_ids & selection_ids), "train_test": len(train_ids & test_ids), "selection_test": len(selection_ids & test_ids)}\nassert len(train_ids) == 760 and len(selection_ids) == 80 and len(test_ids) == 240\nassert all(value == 0 for value in intersections.values()), "학습/선택/시험 ID 교집합 발견"\nfin_mae = float(np.mean(np.abs(y - idx_test)))\nbig = y >= 10\nfin_mape = float(np.mean(np.abs(y[big] - idx_test[big]) / y[big]) * 100)\nfin_grade_acc = float(np.mean(grade_of(idx_test) == m_test["grade"].to_numpy()))\nprint("수정 실습본: ID 교집합", intersections)\nprint(f"최종 시험: MAE {fin_mae:.6f} (n={len(y)}장), MAPE {fin_mape:.6f}% (손상도>=10, n={int(big.sum())}장), 등급 정확도 {fin_grade_acc:.6%} (n={len(y)}장)")\nprint("데이터 역할: 학습 기존600+신규160=760장 / 모델 선택 hold80 / 최종 시험 별도240장(seed=180,181,182). 운영 처리 신규240장과 최종 시험240장은 서로 다른 ID 집합이다.")\nbase = LinearRegression().fit(F.iloc[:n_old][["area_px"]], meta["damage_idx"][:n_old])\nbase_mae = float(np.mean(np.abs(y - np.clip(base.predict(F_test[["area_px"]]), 0, 100))))\nchecks = [("학습·선택·시험 ID 교집합 0", all(value == 0 for value in intersections.values())), ("환경 점검 통과", env_ok == len(libs)), ("데이터 정합성 전 항목 통과", all(chk.values())), ("재학습 파이프라인 정상 실행(챔피언/챌린저 판정 출력)", a[0] > 0 and b[0] > 0),\n          ("최종 시험 MAPE 25% 이하(시험240 중 점수 10 이상만)", fin_mape <= 25), ("최종 시험240장 면적 기준 모델 대비 MAE 20% 이상 개선", fin_mae <= 0.8 * base_mae),\n          ("자동 확정 0건(검사 대기·재촬영 분리)", bool(res["처리"].str.contains("검사 대기|재촬영").all())), ("오염 의심 사진 전부 재촬영 또는 정밀 판독으로 분리", bool(res.loc[res["오염"] >= 0.01, "처리"].ne("검사 대기(간이 승인 후보)").all())),\n          ("운영 문서 100% 완비", done == len(docs))]\nprint("\\n[5] 자체 점검")\nfor n, ok in checks:\n    print(f"  [{\'PASS\' if ok else \'FAIL\'}] {n}")\nprint("\\n[6] 브리핑 요약")\nprint(f"  - 신규 사진 240장으로 분포 변화를 점검({worst} PSI {drift[worst]:.2f}, {\'재학습 검토 수준\' if drifted else \'경미\'})하고 재학습 판단을 자동 수행했다({\'챌린저 승격\' if promote else \'기존 모델 유지\'}).")\nprint(f"  - 별도 최종 시험 {len(y)}장 손상도 MAE {base_mae:.1f}(면적 기준) -> {fin_mae:.1f}, MAPE {fin_mape:.1f}%(점수10 이상 {int(big.sum())}장), 등급 정확도 {fin_grade_acc:.1%}(시험 {len(y)}장). 운영 신규240장 처리 구분: " + ", ".join(f"{k} {v}장" for k, v in vc.items()) + ".")\nprint("  - 최종 정비 등급은 검사관·정비감독관 결재로 확정하며 AI 판정은 자동 확정되지 않는다. 오염 사진은 재촬영을 요구했다.")\n\nfig, ax = plt.subplots(1, 2, figsize=(12, 4))\nax[0].bar(list(drift), list(drift.values())); ax[0].axhline(0.2, c="r", ls="--"); ax[0].set_title("Feature drift (PSI), old vs new photos"); ax[0].tick_params(axis="x", rotation=30)\nax[1].scatter(y, idx_test, s=8); ax[1].plot([0, 100], [0, 100], "k--"); ax[1].set_title("Independent final test (n=240): true vs predicted")\nplt.suptitle("QLT03 integrated pipeline (synthetic photos)"); plt.tight_layout(); plt.show()\n'}
for relative, text in payload.items():
    path=ROOT/relative
    path.parent.mkdir(parents=True,exist_ok=True)
    # Preserve learner-written records on runtime re-execution.
    if path.exists(): continue
    if relative == "src/dashboard/qlt03_app.py" and path.exists(): continue
    path.write_text(text,encoding="utf-8")
for relative in ["src","work","outputs","reports"]: (ROOT/relative).mkdir(exist_ok=True)
def backup_before_write(relative):
    path=ROOT/relative
    if path.exists():
        dest=ROOT/"backups"/(relative+"."+datetime.datetime.now().strftime("%Y%m%d-%H%M%S-%f"))
        dest.parent.mkdir(parents=True,exist_ok=True)
        shutil.copy2(path,dest)
        print("덮어쓰기 전 사본:",dest)
def run_checked(*args):
    result=subprocess.run([sys.executable,*args],capture_output=True,text=True)
    print(result.stdout)
    if result.stderr: print(result.stderr)
    result.check_returncode()
print("환경 준비 완료:",sys.version.split()[0])
````

````python
%%writefile src/m04.py
"""분할(thresholding 기반 결함 후보 분할, Mask R-CNN 대체) 결과에서 결함 면적을 측정·보정하고 깊이와 손상도 점수를 학습해 손상도를 자동 판정하며, 분할 IoU·면적/깊이/손상도 오차(MAPE)와 변수군 제거 실험으로 파생 피처 설계의 타당성을 검증한다."""
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.ensemble import RandomForestClassifier, GradientBoostingRegressor
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import cross_val_predict
from sklearn.metrics import r2_score

# ---------------------------------------------------------------------------
# [교육용가상자료] 장비 외관 사진 생성기 (실제 장비·조직·위치 사진이 아니라 난수로 그린 64x64 합성 이미지)
# 도장된 금속 표면에 균열(crack)·부식(corrosion)·파손(damage)·누유(leak)를 그리고, 조명 편차와 진흙(mud) 오염을 섞는다.
# 이미지와 함께 장비 관리 기록(제원 요약)과 정답 결함 마스크·손상도·정비 등급을 만든다.
# ---------------------------------------------------------------------------
from scipy import ndimage as ndi

SZ = 64
CLASSES = ["none", "crack", "corrosion", "damage", "leak"]
KO_CLS = {"none": "결함 없음", "crack": "균열", "corrosion": "부식", "damage": "파손", "leak": "누유"}
CLS_W = {"none": 0.0, "crack": 1.4, "corrosion": 1.0, "damage": 1.6, "leak": 0.6}      # 손상도 계산용 결함 유형 가중치
GRADE_NAME = {1: "1단계 정비", 2: "2단계 정비", 3: "3단계 정비"}
YY, XX = np.mgrid[0:SZ, 0:SZ]


def grade_of(idx):
    return np.where(np.asarray(idx) >= 65, 3, np.where(np.asarray(idx) >= 35, 2, 1))


def blob(rng, cy, cx, r, rough=0.35):
    """불규칙한 원형 영역 마스크."""
    noise = ndi.gaussian_filter(rng.normal(size=(SZ, SZ)), 3)
    noise = noise / (np.abs(noise).max() + 1e-9)
    return (np.hypot(YY - cy, XX - cx) + rough * r * 2.2 * noise) < r


def draw(rng, cls, sev):
    """결함 유형·심각도 sev(0~1)에 맞는 정답 마스크와 색 정보를 만든다."""
    m = np.zeros((SZ, SZ), bool)
    if cls == "crack":
        y, x, ang = rng.uniform(12, 52), rng.uniform(12, 52), rng.uniform(0, np.pi)
        for _ in range(int(20 + 40 * sev)):
            ang += rng.normal(0, 0.35); y += np.sin(ang) * 1.2; x += np.cos(ang) * 1.2
            if 0 <= int(y) < SZ and 0 <= int(x) < SZ:
                m[int(y), int(x)] = True
        m = ndi.binary_dilation(m, iterations=1 + int(sev > 0.6))
    elif cls == "corrosion":
        for _ in range(int(rng.integers(1, 4))):
            m |= blob(rng, rng.uniform(14, 50), rng.uniform(14, 50), 3 + 9 * sev * rng.uniform(0.6, 1.1))
    elif cls == "damage":
        m = blob(rng, rng.uniform(20, 44), rng.uniform(20, 44), 6 + 12 * sev, 0.3)
    elif cls == "leak":
        x0, y0 = rng.uniform(16, 48), rng.uniform(8, 24)
        w, ln = 2 + 4 * sev, 12 + 26 * sev
        m = (np.abs(XX - x0 - 0.15 * (YY - y0)) < w * (1 - 0.4 * (YY - y0) / ln)) & (YY > y0) & (YY < y0 + ln)
    return m


def make_images(n=600, seed=0, mud=0.0, p_none=0.25, dark=False):
    """가상 외관 사진 n장(+진흙 오염 mud 비율)과 관리 기록·정답. 반환: meta(DataFrame), imgs(n,64,64,3), masks(n,64,64)."""
    rng = np.random.default_rng(seed)
    imgs, masks, rows = np.zeros((n, SZ, SZ, 3), np.float32), np.zeros((n, SZ, SZ), bool), []
    for i in range(n):
        cls = "none" if rng.random() < p_none else str(rng.choice(CLASSES[1:], p=[0.25, 0.3, 0.25, 0.2]))
        sev = 0.0 if cls == "none" else float(rng.beta(1.6, 2.0))
        base = np.array([0.42, 0.46, 0.40]) * (1 + 0.5 * ndi.gaussian_filter(rng.normal(size=(SZ, SZ)), 8)[..., None] * 3)
        img = base + rng.normal(0, 0.012, (SZ, SZ, 3))
        m = draw(rng, cls, sev)
        a = ndi.gaussian_filter(m.astype(float), 0.8)[..., None]                        # 경계를 부드럽게 섞는다
        if cls == "crack":
            col = img * rng.uniform(0.2, 0.45)
        elif cls == "corrosion":
            col = np.array([0.55, 0.30, 0.12]) * rng.uniform(0.75, 1.15) * (0.85 + 0.3 * rng.random((SZ, SZ, 1)))
        elif cls == "damage":
            col = np.ones_like(img) * np.array([0.34, 0.34, 0.36]) * (1.0 - 0.45 * sev) * rng.uniform(0.75, 1.0)
            rim = m & ~ndi.binary_erosion(m, iterations=2)
            a = np.maximum(a, 0.0); img = np.where(rim[..., None], 0.7, img)
        elif cls == "leak":
            col = np.ones_like(img) * np.array([0.10, 0.09, 0.08]) * rng.uniform(0.8, 2.6)
        else:
            col = img
        img = img * (1 - a) + col * a
        if rng.random() < 0.35:                                   # 무해한 얼룩(결함 아님): 옅은 갈색 번짐
            st = blob(rng, rng.uniform(5, 58), rng.uniform(5, 58), rng.uniform(6, 14), 0.5)
            al = ndi.gaussian_filter(st.astype(float), 2.5)[..., None] * rng.uniform(0.15, 0.4)
            img = img * (1 - al) + np.array([0.50, 0.40, 0.30]) * al
        if rng.random() < 0.3:                                    # 무해한 긴 스크래치(밝은 선)
            y0, x0, ang = rng.uniform(5, 58), rng.uniform(5, 58), rng.uniform(0, np.pi)
            for t in np.linspace(0, 25, 40):
                yy, xx = int(y0 + t * np.sin(ang)), int(x0 + t * np.cos(ang))
                if 0 <= yy < SZ and 0 <= xx < SZ:
                    img[yy, xx] = np.clip(img[yy, xx] * 1.5, 0, 1)
        if mud > 0:                                           # 진흙 오염: 결함 위를 덮을 수 있다(정답 마스크는 그대로)
            mm = np.zeros((SZ, SZ), bool)
            while mm.mean() < mud:
                mm |= blob(rng, rng.uniform(0, SZ), rng.uniform(0, SZ), rng.uniform(5, 11), 0.5)
            img[mm] = np.array([0.38, 0.30, 0.22]) * (0.85 + 0.3 * rng.random((mm.sum(), 1)))
        gain = rng.uniform(0.3, 0.55) if dark else rng.uniform(0.5, 1.15)
        ang = rng.uniform(0, 2 * np.pi)
        light = gain * (1 + 0.3 * (np.cos(ang) * (XX - 32) + np.sin(ang) * (YY - 32)) / 32)
        imgs[i] = np.clip(img * light[..., None] + rng.normal(0, 0.025, (SZ, SZ, 3)), 0, 1)
        masks[i] = m
        area = float(m.mean())
        depth = {"none": 0.0, "crack": 0.3 + 3.0 * sev, "corrosion": 0.1 + 1.5 * sev, "damage": 1.0 + 8.0 * sev, "leak": 0.0}[cls] * rng.lognormal(0, 0.1)
        idx = float(np.clip(100 * (1 - np.exp(-(CLS_W[cls] * area * 9 + 0.05 * depth))) + rng.normal(0, 1.5), 0, 100))
        rows.append({"img_id": f"IMG{seed:02d}{i:04d}", "cls": cls, "sev": sev, "area_frac": area, "depth_mm": depth, "damage_idx": idx,
                     "grade": int(grade_of(idx)), "gain": gain, "mud": mud})
    meta = pd.DataFrame(rows)
    # 장비 관리 기록(제원 요약): 사진 판정과 대조할 값
    age = rng.integers(1, 25, n)
    meta["day"] = np.sort(rng.integers(0, 90, n))
    meta["equip_type"] = rng.choice(["vehicle", "generator", "comm", "trailer"], n, p=[0.4, 0.2, 0.2, 0.2])
    meta["mfg_year"] = 2026 - age
    meta["op_hours"] = np.clip(rng.normal(180 * age, 400, n) + 8 * meta["damage_idx"], 10, 20000)
    meta["env"] = rng.choice(["coastal", "inland"], n, p=[0.3, 0.7])
    reg = np.clip(meta["grade"] + rng.choice([-1, 0, 0, 0, 0, 0, 1, 2], n), 1, 3)           # 장비 관리 기록상 직전 정비 등급
    meta["reg_grade"] = reg
    meta["emergency"] = rng.random(n) < 0.04                                               # 긴급 점검 요청(화질 불량이어도 보존)
    return meta, imgs, masks
# ---- 전처리(조명 보정·표준화)와 결함 후보 분할, 특징 추출 (cv2 대신 numpy·scipy 사용) ----
def correct(img):
    """조명 보정: 밝기에 평면(조명 기울기)을 적합해 나눈다. 어두운 결함 화소는 제외하고 두 번 반복 적합한다. 반환: 보정 밝기(정상 면 약 1), 색 비율."""
    g = img.mean(axis=2)
    A = np.c_[np.ones(SZ * SZ), XX.ravel() / SZ, YY.ravel() / SZ]
    keep = np.ones(SZ * SZ, bool)
    for _ in range(3):
        coef, *_ = np.linalg.lstsq(A[keep], g.ravel()[keep], rcond=None)
        plane = (A @ coef).reshape(SZ, SZ)
        keep = (np.abs(g - plane) < 0.08 * np.maximum(plane, 1e-3)).ravel()
        if keep.sum() < 200:
            keep[:] = True
    c = g / np.maximum(plane, 1e-3)
    chroma = img[..., 0] / (img.sum(axis=2) + 1e-6)
    return c, chroma


SEGP = {"dark": 0.7, "rust": 0.5, "rim": 1.45}            # 분할 임계값(어두움 보정 밝기, 붉은 비율, 밝은 테두리): 재보정 대상


def segment(img):
    """결함 후보 마스크: 어두운 영역(균열·누유·파손) + 붉은 갈색 영역(부식) + 밝은 테두리(파손)."""
    c, chroma = correct(img)
    dark = c < SEGP["dark"]
    rust = (chroma > SEGP["rust"]) & (c > 0.45)
    rim = c > SEGP["rim"]
    m = dark | rust | ndi.binary_dilation(rim, iterations=1) & ndi.binary_dilation(dark, iterations=3)
    m = ndi.binary_opening(m, structure=np.ones((2, 2))) if m.sum() > 12 else m
    return m, c, chroma, dark, rust, rim


def extract(img):
    """분할 결과와 색·모양에서 24개 특징을 만든다."""
    m, c, chroma, dark, rust, rim = segment(img)
    lab, n = ndi.label(m)
    area = float(m.sum())
    if n:
        sizes = ndi.sum(m, lab, range(1, n + 1))
        k = int(np.argmax(sizes)) + 1
        sl = ndi.find_objects(lab)[k - 1]
        h, w = sl[0].stop - sl[0].start, sl[1].stop - sl[1].start
        big = float(sizes.max())
        elong, fill = max(h, w) / max(min(h, w), 1), big / (h * w)
    else:
        big, elong, fill = 0.0, 1.0, 0.0
    gy, gx = np.gradient(c)
    mag = np.hypot(gx, gy)
    inside = m.sum() > 0
    return {"area_px": area, "area_frac": area / SZ ** 2, "n_comp": float(n), "big_px": big, "elong": elong, "fill": fill,
            "dark_frac": float(dark.mean()), "rust_frac": float(rust.mean()), "rim_frac": float(rim.mean()),
            "c_mean": float(c.mean()), "c_std": float(c.std()), "c_min": float(c.min()), "edge_mean": float(mag.mean()), "edge_p95": float(np.quantile(mag, 0.95)),
            "chroma_mean": float(chroma.mean()), "chroma_max": float(chroma.max()),
            "in_dark": float(c[m].mean()) if inside else 1.0, "in_chroma": float(chroma[m].mean()) if inside else float(chroma.mean()),
            "ring_contrast": float(c[ndi.binary_dilation(m, iterations=3) & ~m].mean() - c[m].mean()) if inside and (~m).any() else 0.0,
            "thin": float(1 - fill), "gain": float(img.mean()), "sat": float((img.max(axis=2) - img.min(axis=2)).mean()),
            "cy": float(ndi.center_of_mass(m)[0] / SZ) if inside else 0.5, "cx": float(ndi.center_of_mass(m)[1] / SZ) if inside else 0.5}


def feature_table(imgs):
    return pd.DataFrame([extract(im) for im in imgs])
# ---- 판정 파이프라인: 결함 유형 분류 + 손상도 회귀(분류 확률 포함) -> 정비 등급 ----
def fit_pipeline(meta, F):
    """결함 유형 분류기와 손상도 회귀기를 학습한다. 회귀기 입력의 분류 확률은 교차 예측으로 만들어 누설을 막는다."""
    clf = RandomForestClassifier(n_estimators=150, random_state=0, n_jobs=1)
    P = cross_val_predict(clf, F, meta["cls"], cv=4, method="predict_proba")
    clf.fit(F, meta["cls"])
    reg = GradientBoostingRegressor(n_estimators=200, max_depth=3, learning_rate=0.06, random_state=0).fit(np.c_[F.values, P], meta["damage_idx"])
    return {"clf": clf, "reg": reg, "cols": list(F.columns)}


def predict_pipeline(pl, F):
    X = F[pl["cols"]]
    P = pl["clf"].predict_proba(X)
    idx = np.clip(pl["reg"].predict(np.c_[X.values, P]), 0, 100)
    return pl["clf"].classes_[P.argmax(axis=1)], P, idx, P.max(axis=1)

mtr, itr, ktr = make_images(600, seed=1)
mte, ite, kte = make_images(300, seed=2)
Ftr, Fte = feature_table(itr), feature_table(ite)
print(f"[1] 파생 피처 {Ftr.shape[1]}개(분할 면적·성분 수·모양, 명암, 색, 가장자리, 위치) / 학습 {len(mtr)}장, 평가 {len(mte)}장 (1화소 = 1.5mm 가정)")

# ---------------- 1) 분할 정밀도: 정답 마스크와의 IoU ----------------
seg = [segment(im)[0] for im in ite]
rows = []
# TODO 1: CLASSES[1:]의 각 결함 유형 c에 대해 평가 표본 인덱스 ix를 찾고,
#         예측 마스크 seg[i]와 정답 마스크 kte[i]의 IoU, Dice, 검출률(교집합>0)을 계산해 rows에 추가하세요.
raise NotImplementedError("TODO: 결함 유형별 분할 IoU·Dice·검출률 계산 로직을 구현하세요.")
sg = pd.DataFrame(rows, columns=["결함", "장수", "평균 IoU", "평균 Dice", "검출률"])
print("[2] 분할 성능(평가 300장)\n" + sg.round(2).to_string(index=False))
ok0 = np.mean([seg[i].sum() == 0 or (seg[i].sum() < 12) for i in np.where(mte["cls"].values == "none")[0]])
print(f"    결함 없는 사진 {int((mte['cls'] == 'none').sum())}장 중 분할 결과가 거의 비어 있는(12화소 미만) 사진 {ok0:.0%} (무해한 얼룩·스크래치 오탐 억제)")

# ---------------- 2) 면적 측정과 유형별 보정 ----------------
pl = fit_pipeline(mtr, Ftr)
cls_te, P_te, idx_te, conf_te = predict_pipeline(pl, Fte)
a_true, a_raw = mte["area_frac"].values, Fte["area_frac"].values
cal = {c: LinearRegression().fit(Ftr.loc[mtr["cls"].values == c, ["area_frac"]], mtr.loc[mtr["cls"].values == c, "area_frac"]) for c in CLASSES[1:]}
a_cal = np.array([cal[c].predict(pd.DataFrame({"area_frac": [a]}))[0] if c != "none" else 0.0 for c, a in zip(cls_te, a_raw)])
m = a_true > 0
mp = lambda y, p: float(np.mean(np.abs(y - p) / y) * 100)
print(f"\n[3] 면적 측정(결함 {int(m.sum())}장): 분할 원측정 MAPE {mp(a_true[m], a_raw[m]):.1f}% -> 유형별 선형 보정 후 {mp(a_true[m], a_cal[m]):.1f}% (분류 정확도 {(cls_te == mte['cls'].values).mean():.1%})")
print(f"    면적 환산 예: 평균 결함 면적 {a_true[m].mean() * SZ ** 2 * 1.5 ** 2:.0f}mm2(정답) / {a_cal[m].mean() * SZ ** 2 * 1.5 ** 2:.0f}mm2(보정 측정)")

# ---------------- 3) 깊이·손상도 모델 ----------------
dm_tr = mtr["depth_mm"].values > 0.2
Xtr = np.c_[Ftr[pl["cols"]].values, cross_val_predict(RandomForestClassifier(n_estimators=100, random_state=0, n_jobs=1), Ftr, mtr["cls"], cv=3, method="predict_proba")]
dep = GradientBoostingRegressor(n_estimators=200, max_depth=3, learning_rate=0.06, random_state=0).fit(Xtr[dm_tr], mtr["depth_mm"].values[dm_tr])
d_pred = np.clip(dep.predict(np.c_[Fte[pl["cols"]].values, P_te]), 0, None)
dm = mte["depth_mm"].values > 0.2
print(f"[4] 깊이 추정(깊이 있는 유형 {int(dm.sum())}장, 누유는 깊이 0으로 제외): MAPE {mp(mte['depth_mm'].values[dm], d_pred[dm]):.1f}%, MAE {np.mean(np.abs(mte['depth_mm'].values[dm] - d_pred[dm])):.2f}mm")
print("    깊이는 사진 한 장으로 직접 관측되지 않아 면적·명암에서 간접 추정하므로 손상도보다 오차가 크다(실제 업무는 깊이 게이지 등 현장 측정으로 확인).")
y = mte["damage_idx"].values
big = y >= 10
mape_d, mae_d = mp(y[big], idx_te[big]), float(np.mean(np.abs(y - idx_te)))
base = LinearRegression().fit(Ftr[["area_px"]], mtr["damage_idx"])
b_pred = np.clip(base.predict(Fte[["area_px"]]), 0, 100)
print(f"[5] 손상도 점수(0~100, 점수 10 이상 {int(big.sum())}장): 면적만 쓰는 기준 MAPE {mp(y[big], b_pred[big]):.1f}% -> 제안 모델 MAPE {mape_d:.1f}% (목표 25% 이하: {'달성' if mape_d <= 25 else '미달'}), MAE {mae_d:.2f}, R2 {r2_score(y, idx_te):.3f}")
g_true, g_pred = mte["grade"].values, grade_of(idx_te)
print(f"    정비 등급 일치율 {(g_true == g_pred).mean():.1%}, 한 단계 이내 {(np.abs(g_true - g_pred) <= 1).mean():.1%}")

# ---------------- 4) 변수군 제거 실험 ----------------
groups = {"분할 면적·모양": ["area_px", "area_frac", "n_comp", "big_px", "elong", "fill", "thin", "cy", "cx"], "명암(어두운 비율·대비)": ["dark_frac", "c_mean", "c_std", "c_min", "in_dark", "ring_contrast"],
          "색(붉은 갈색)": ["rust_frac", "chroma_mean", "chroma_max", "in_chroma", "sat"], "가장자리·테두리": ["edge_mean", "edge_p95", "rim_frac"], "조명": ["gain"]}
full_cols = list(Ftr.columns)


def run(cols):
    g = GradientBoostingRegressor(n_estimators=200, max_depth=3, learning_rate=0.06, random_state=0).fit(Ftr[cols], mtr["damage_idx"])
    p = np.clip(g.predict(Fte[cols]), 0, 100)
    return mp(y[big], p[big])


full = run(full_cols)
ab = pd.DataFrame([(k, run([c for c in full_cols if c not in v])) for k, v in groups.items()], columns=["제거한 변수군", "MAPE(%)"]).assign(증가=lambda x: x["MAPE(%)"] - full).sort_values("증가", ascending=False)
print(f"\n[6] 변수군 제거(분류 확률 없이 24개 피처만 사용, 전체 MAPE {full:.1f}%)\n" + ab.round(2).to_string(index=False))
print(f"    해석: 제거 시 오차가 가장 크게 늘어난 변수군은 '{ab.iloc[0]['제거한 변수군']}'(+{ab.iloc[0]['증가']:.1f}%p)이다. 변수군끼리 정보가 겹쳐 단독 제거 효과는 작게 나타날 수 있다.")
print("    판정은 정비 의사결정 보조이며 최종 등급은 검사관이 확정한다(M7). 사진·장비는 모두 교육용 가상 자료이다.")

fig, ax = plt.subplots(1, 3, figsize=(15, 4.3))
ax[0].bar(range(len(sg)), sg["평균 IoU"]); ax[0].set_xticks(range(len(sg))); ax[0].set_xticklabels(["crack", "corrosion", "damage", "leak"]); ax[0].set_title("Segmentation IoU by defect type")
ax[1].scatter(a_true[m], a_raw[m], s=8, label="raw"); ax[1].scatter(a_true[m], a_cal[m], s=8, label="calibrated"); ax[1].plot([0, .2], [0, .2], "k--"); ax[1].legend(); ax[1].set_title("Defect area fraction: true vs measured")
ax[2].scatter(y, idx_te, s=8); ax[2].plot([0, 100], [0, 100], "k--"); ax[2].set_title("Damage index: true vs predicted")
plt.tight_layout(); plt.show()
````
