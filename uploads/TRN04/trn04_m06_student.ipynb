```python
%%writefile templates/trn04_aar_fact_action_template.md
# 지휘관 승인 종합 제안서 및 F-J-A-U 기록 표준 서식

## 1. 지휘관 승인 종합 제안서 7대 항목
1. 추진 배경: 러-우 전쟁 드론 전훈 반영
2. 교육 목표: 소부대 대드론 대응력 90% 제고
3. 과정 개요: 27시간 3단계 과정
4. 소요 자원: 모의 교구 5세트, 안전통제관 2명
5. 식별 위험 및 완화책: 안전 인터록 확립
6. 승인 요청 사항: 2026년 후반기 정규 과정 편성 승인
7. 성과 지표(KPI): 정량 지표 6개 수립

## 2. AAR 기록 4대 분류 (F-J-A-U) 규칙
- Fact (사실): 관측된 객관적 현상
- Judgment (판단): 분석관의 주관적 추정
- Action (조치): 현장 지휘관의 명령 및 행동
- Unconfirmed (미확인): 추가 확인이 필요한 첩보

## 3. 금지 단어 목록
- '아마도', '상당히', '적 드론인 듯함', '신속하게' 등 모호어 배제
```

```python
# [미션 06 산출물 자가 검증 (Self-Validation)]
target_doc = "templates/trn04_aar_fact_action_template.md"
assert os.path.exists(target_doc), f"❌ 산출물 누락: {target_doc} 파일이 생성되지 않았습니다."

with open(target_doc, "r", encoding="utf-8") as f:
    text = f.read()

missing = []
for k in ['종합 제안서', '성과 지표', 'Fact', 'Judgment', 'Action', 'Unconfirmed']:
    if k not in text:
        missing.append(k)

if missing:
    print(f"⚠️ [보완 필요] 필수 키워드/섹션 누락: {missing}")
else:
    print(f"🎉 [PASS] 미션 06 필수 산출물({target_doc}) 검증 완료! 모든 필수 규격이 충족되었습니다.")
```
