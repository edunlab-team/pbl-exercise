````python
# [학생 보조 도구] 8개 출처 D일 이전 접수 부분집합 vs 출처대장 과거 전체 건수 대조
import csv

with open("02_데이터셋/출처_이력대장.csv", encoding="utf-8") as f:
    sources = list(csv.DictReader(f.readlines()[1:]))
with open("02_데이터셋/보고_접수대장.csv", encoding="utf-8") as f:
    reports = list(csv.DictReader(f.readlines()[1:]))

# TODO: 접수시각이 'D-'로 시작하는(당일 이전) 보고 건수를 출처ID별로 세어 과거보고건수와 비교하세요.
for s in sources:
    sid = s["출처ID"]
    past_total = int(s["과거보고건수"])
    subset_cnt = sum(1 for r in reports if r["출처ID"] == sid and r["접수시각"].startswith("D-"))
    print(f"{sid}: D일 이전 부분집합={subset_cnt}, 출처대장 과거전체={past_total}")
````
