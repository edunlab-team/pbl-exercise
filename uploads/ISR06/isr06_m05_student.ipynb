````python
# [학생 보조 도구] 4분기 의사결정 트리 입력값·묶음 B1~B5 확인
import csv

with open("02_데이터셋/보고_접수대장.csv", encoding="utf-8") as f:
    rows = {r["보고ID"]: r for r in csv.DictReader(f.readlines()[1:])}

bundles = {
    "B1": ["C1", "C3"],
    "B2": ["C3", "C4"],
    "B3": ["C1"],
    "B4": ["K2A", "K2B"],
    "B5": [],  # 권한자 부재 사례 (보고 ID 미제공)
}
for b, ids in bundles.items():
    info = [f"{i}({rows[i]['출처유형']}:{rows[i]['출처ID']})" for i in ids] or ["해당 없음(미제공)"]
    print(f"{b}: {', '.join(info)}")
````
