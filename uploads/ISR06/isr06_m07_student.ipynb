````python
# [학생 보조 도구] 평가자 간 채점 일치율 계산기
# 일치율(%) = 두 평가자 점수가 완전히 같은 영역 수 / 전체 영역 수(5) * 100

def agreement_rate(my_scores: list[int], eval2_scores: list[int]) -> float:
    same = sum(1 for a, b in zip(my_scores, eval2_scores) if a == b)
    return same / len(my_scores) * 100

# TODO: 본인 5영역 채점(4~1)과 AI 평가자 2 채점을 입력해 일치율(%)과 2점 이상 차이 여부를 확인하세요.
my_scores = [4, 4, 4, 4, 4]
eval2_scores = [4, 4, 4, 4, 4]
print(f"예시 계산식 테스트: {agreement_rate(my_scores, eval2_scores):.0f}% (실제 채점값으로 교체하세요)")
````
