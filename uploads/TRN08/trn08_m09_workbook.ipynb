- **소요 시간:** 5일 / 27.0시간 (레벨 2)
- **보안 및 HITL 원칙:** 실제 부대명·진지 위치·호출부호·실제 드론 제원/주파수는 절대 입력하지 않으며, 모든 AI 출력은 인간 최종 검토(HITL)를 거쳐 확정합니다.
- **산출물 디렉토리 안내:** `trn08_deliverables_guide.md` (또는 아래 1번 공통 환경 설정 셀 실행으로 자동 생성)

```python
# [공통 환경 설정] 산출물 디렉토리 생성 및 미션 데이터셋 로드 헬퍼
import json
from pathlib import Path

DELIVERABLE_DIRS = [
    "docs", "sop", "scenarios", "logs",
    "reports", "aar", "eval", "package"
]
for d in DELIVERABLE_DIRS:
    Path(d).mkdir(parents=True, exist_ok=True)

def load_mission_dataset(mission_num: int) -> dict:
    fname = f"trn08_m{mission_num:02d}_dataset.json"
    candidates = [
        Path(fname),
        Path("uploads/TRN08") / fname,
        Path("../uploads/TRN08") / fname,
    ]
    for p in candidates:
        if p.exists():
            with open(p, "r", encoding="utf-8") as f:
                return json.load(f)
    raise FileNotFoundError(f"데이터셋을 찾을 수 없습니다: {fname}")

def save_deliverable(rel_path: str, content: str) -> Path:
    out = Path(rel_path)
    out.parent.mkdir(parents=True, exist_ok=True)
    out.write_text(content, encoding="utf-8")
    print(f"[저장 완료] {out}")
    return out

print("TRN08 실습 환경 준비 완료. 디렉토리:", ", ".join(DELIVERABLE_DIRS))
```

### 3. 단계별 AI 활용 가이드 (프롬프트 예시 & HITL 검증 포인트)
- **Step 1**: 프롬프트 질의 예시:  미션 4 조건표와 미션 8 SOP 수정안을 입력하고, '재평가 시 동일 조건을 유지해야 할 항목(인원 수·카드 구성·거리 기준)과 변경된 변수(수정 SOP)를 구분한 체크표를 만들어 줘.'라고 질의한다. (※ 실제 부대명·진지 위치·호출부호·실제 드론 제원/주파수는 입력하지 않고 제공된 가상 자료만 활용한다.)  AI 출력 검증 포인트 (HITL): 주입 카드 순서만 바꾸고 난이도와 평가표 기준이 미션 4와 동일하게 유지되었는지 확인한다.
- **Step 2**: 프롬프트 질의 예시:  재평가 훈련 중 측정할 지표(최초 청취~복창·판단 시간, 누락·오보고·오식별 건수)를 입력하고, '1차 실습과 동일한 서식으로 기록 오류를 막는 체크 포인트를 정리해 줘.'라고 질의한다. (※ 실제 부대명·진지 위치·호출부호·실제 드론 제원/주파수는 입력하지 않고 제공된 가상 자료만 활용한다.)  AI 출력 검증 포인트 (HITL): 실제 재평가 측정값이 1차와 같은 정의(시작점·종료점)로 기록되었는지 평가관이 확인한다.
- **Step 3**: 프롬프트 질의 예시:  1차(미션 4)와 재평가(미션 9) 수치 데이터를 입력하고, '조치 시간 단축률(%)과 오류 건수 감소율(%)을 계산 예시대로 산출하고 전후 비교표를 작성해 줘.'라고 질의한다. (※ 실제 부대명·진지 위치·호출부호·실제 드론 제원/주파수는 입력하지 않고 제공된 가상 자료만 활용한다.)  AI 출력 검증 포인트 (HITL): 단축률·감소율 계산식의 분모(1차 기준값)가 올바른지 워크북 Python 셀 또는 수동 계산으로 반드시 검산한다.
- **Step 4**: 프롬프트 질의 예시:  목표에 미달한 항목과 재평가 기록을 입력하고, '미달 원인을 절차·숙련 관점에서 요약하고 월별/분기별 보수 교육 주기와 15분 단축 훈련 방안을 제안해 줘.'라고 질의한다. (※ 실제 부대명·진지 위치·호출부호·실제 드론 제원/주파수는 입력하지 않고 제공된 가상 자료만 활용한다.)  AI 출력 검증 포인트 (HITL): 제안된 보수 교육 주기가 부대 일과와 안전 통제 여건에 비추어 현실적인지 검토한다.
- **Step 5**: 프롬프트 질의 예시:  전후 비교표 결과를 입력하고, '재평가에서 크게 개선된 1가지와 여전히 남은 취약점 1가지를 지휘관 보고용 3줄 요약으로 작성해 줘.'라고 질의한다. (※ 실제 부대명·진지 위치·호출부호·실제 드론 제원/주파수는 입력하지 않고 제공된 가상 자료만 활용한다.)  AI 출력 검증 포인트 (HITL): 요약 수치가 비교표 원문과 정확히 일치하는지 대조한다.

```python
# [미션 09 - 교수자 해설용] 1차(미션 4) vs 재평가(미션 9) 전후 비교 계산
ds = load_mission_dataset(9)
base = ds["provided_materials"][0]["baseline_metrics_example"]
rerun = {"조치시간": 36.0, "SOP준수율": 94.0, "오류건수": 2}

time_red = (base["1차 평균 조치시간(초)"] - rerun["조치시간"]) / base["1차 평균 조치시간(초)"] * 100
err_red = (base["1차 오류 건수(건)"] - rerun["오류건수"]) / base["1차 오류 건수(건)"] * 100
print(f"조치시간 단축률: {time_red:.1f}% (목표 {base['목표 단축률(%)']}%), 오류 감소율: {err_red:.1f}%")
assert time_red >= base["목표 단축률(%)"] and rerun["SOP준수율"] >= base["목표 SOP 준수율(%)"]

md = (
    f"# TRN08 M09 재평가 성과 비교\n"
    f"- 평균 조치시간: {base['1차 평균 조치시간(초)']}초 -> {rerun['조치시간']}초 (단축률 {time_red:.1f}%)\n"
    f"- SOP 준수율: {base['1차 SOP 준수율(%)']}% -> {rerun['SOP준수율']}%\n"
    f"- 오류 건수: {base['1차 오류 건수(건)']}건 -> {rerun['오류건수']}건 (감소율 {err_red:.1f}%)\n"
)
save_deliverable("eval/trn08_rerun_performance_comparison.md", md)
```
