```python
# [공통 환경 설정] 산출물 디렉토리 생성 및 미션 데이터셋 로드 헬퍼
import json
from pathlib import Path

DELIVERABLE_DIRS = [
    "docs", "sop", "scenarios", "logs",
    "reports", "aar", "eval", "package"
]
for d in DELIVERABLE_DIRS:
    Path(d).mkdir(parents=True, exist_ok=True)

def load_mission_dataset(mission_num: int) -> dict:
    fname = f"trn08_m{mission_num:02d}_dataset.json"
    candidates = [
        Path(fname),
        Path("uploads/TRN08") / fname,
        Path("../uploads/TRN08") / fname,
    ]
    for p in candidates:
        if p.exists():
            with open(p, "r", encoding="utf-8") as f:
                return json.load(f)
    raise FileNotFoundError(f"데이터셋을 찾을 수 없습니다: {fname}")

def save_deliverable(rel_path: str, content: str) -> Path:
    out = Path(rel_path)
    out.parent.mkdir(parents=True, exist_ok=True)
    out.write_text(content, encoding="utf-8")
    print(f"[저장 완료] {out}")
    return out

print("TRN08 실습 환경 준비 완료. 디렉토리:", ", ".join(DELIVERABLE_DIRS))
```

- **Step 1**: 프롬프트 질의 예시:  근거리·중거리·원거리 3개 상대 거리 구간 기준을 입력하고, '각 구간별로 드론과 비드론(조류·민간 비행체·차량 소음)을 구분할 수 있는 청취·실루엣 카드 예시 문안을 상대적 표현으로 작성해 줘.'라고 질의한다.
```python
# [미션 03 - 교수자 해설용] 12장 미디어셋 & 10건(6:4) MSEL 시나리오 검증 및 저장
ds = load_mission_dataset(3)
cards = []
for band in ["근거리", "중거리", "원거리"]:
    for i in range(1, 5):
        is_drone = i <= 3
        cards.append({
            "card_id": f"{band[:1]}-{i}",
            "거리": band,
            "유형": "청취" if i % 2 else "실루엣",
            "정답": "드론" if is_drone else "비드론",
            "단서": "높고 일정한 소리·직선비행" if is_drone else "불규칙 소음·날갯짓",
        })
assert len(cards) == 12

msel = [
    {"seq": i, "구분": "정상" if i <= 6 else "비정상", "돌발": i == 10}
    for i in range(1, 11)
]
assert sum(1 for e in msel if e["구분"] == "정상") == 6
assert sum(1 for e in msel if e["구분"] == "비정상") == 4
assert any(e["돌발"] for e in msel)
print(f"미디어셋 {len(cards)}장, MSEL {len(msel)}건(6:4, 돌발 포함) 검증 통과")

md = "# TRN08 M03 MSEL 상황 주입 시나리오\n" + "\n".join(
    f"- 사건 {e['seq']} ({e['구분']}){' [돌발변수]' if e['돌발'] else ''}" for e in msel
)
save_deliverable("scenarios/trn08_msel_scenario.md", md)
```
