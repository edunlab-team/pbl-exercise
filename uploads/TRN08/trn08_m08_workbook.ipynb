### 교수자 해설 워크북 (Workbook)

> **프로젝트 목표:** 러시아-우크라이나 전쟁에서는 소형 드론이 소리 없이 다가오는 것이 아니라 가까워질수록 모터 소음이 커지는데도, 병사들이 바람·차량·조류와 구분하지 못하거나 방향과 고도를 틀리게 보고해 경보가 늦어지는 일이 반복되었다. 본 프로젝트에서는 가상 경계진지를 무대로 소형 드론의 소리 특징과 비행 형태별 모습을 정리하고, 거리별 청취·실루엣 식별 훈련 교재와 관측 체크리스트, 상황 주입 시나리오를 만든다. 이어서 청취-육안 포착-구두 보고 1차 연계 실습, 민간 비행체·조류 오인 주입 평가, 무전기 SALUTE 즉각 보고와 인계, 강풍·배경 소음 스트레스 훈련, 사후검토(AAR), 재평가 훈련을 차례로 수행하며 매 단계 시간과 오류를 기록해 숙련도 향상을 수치로 확인한다. 이 과제는 드론 대응의 기초 숙련 훈련이므로 식별과 보고 절차에 집중하고, 교전 판단은 다루지 않는다. 최종 목표는 청취·육안 식별·즉각 보고 훈련 교범과 교관이 바로 쓸 수 있는 훈련 패키지를 완성하는 것이다. 본 프로젝트를 통해 익히는 핵심 역량  드론 음향·실루엣 기초 식별 및 체크리스트 설계: 소형 드론의 상대적 소리 특징과 비행 형태별 관측 포인트를 구조화하고, 모호한 표현을 배제한 현장 체크리스트와 7단계 SOP를 설계하는 역량을 익힌다.  훈련 교재(미디어셋) 및 상황 주입 시나리오(MSEL) 구축: 거리별 청취·실루엣 카드와 정상/비정상 사건이 배합된 모의 훈련 시나리오를 설계해 실전형 반복 훈련 환경을 구성하는 방법을 습득한다.  교차 검증·SALUTE 즉각 보고 및 스트레스 대응: 조류·민간 비행체 오인 상황을 걸러내는 2인 교차 확인 규칙, 무전 SALUTE 보고·SBAR 인계 절차, 강풍·배경 소음 속 육안 전환 및 역할 위임 능력을 기른다.  사실 기반 AAR(사후검토) 및 훈련 패키지 표준화: 타임라인·오류율 데이터를 근거로 SOP를 개선하고 재평가로 효과를 수치 검증한 뒤, 타 부대 교관이 즉시 활용할 수 있는 표준 훈련 교범으로 통합하는 역량을 익힌다.  추천 학습 대상자 (페르소나)  경계·방호 현장 초급 간부 및 분대·소대 지휘자: 소형 드론 침투 상황에서 경계병의 청음·육안 보고를 접수하고, SALUTE 즉각 보고와 SBAR 교대 인계, 강풍·소음 환경에서의 역할 위임 절차를 표준화해 초동 대응 지연을 줄이고 싶은 현장 지휘자  군 교육·훈련 교관 및 교보재·시나리오 설계 담당자: 거리별 청취·실루엣 식별 카드, 정상·비정상 사건이 배합된 상황 주입 시나리오(MSEL), 사실 기반 AAR(사후검토) 평가표를 직접 제작해 부대에 바로 보급 가능한 훈련 패키지를 구축하려는 교관  대드론 경계·보고 SOP 기획 및 훈련 평가 담당자: 조류·민간 비행체 오인 사례를 줄이는 2인 교차 확인 규칙을 정립하고, 반복 훈련(Dry-run → 스트레스 훈련 → 재평가)의 조치 시간과 오류율 개선을 데이터로 검증해 표준 교범에 반영하려는 실무 기획자  도전 적합 기준 (레벨 2 · 5일 / 27시간): 기본적인 부대 경계·보고 흐름을 이해하고 있으며, 실제 제원이나 군사 비밀을 입력하지 않는 보안 원칙과 AI 제안을 교차 검증하는 HITL(인간 최종 검토) 태도를 갖춘 학습자에게 적합합니다.

- **소요 시간:** 5일 / 27.0시간 (레벨 2)
- **보안 및 HITL 원칙:** 실제 부대명·진지 위치·호출부호·실제 드론 제원/주파수는 절대 입력하지 않으며, 모든 AI 출력은 인간 최종 검토(HITL)를 거쳐 확정합니다.
- **산출물 디렉토리 안내:** `trn08_deliverables_guide.md` (또는 아래 1번 공통 환경 설정 셀 실행으로 자동 생성)

```python
# [공통 환경 설정] 산출물 디렉토리 생성 및 미션 데이터셋 로드 헬퍼
import json
from pathlib import Path

DELIVERABLE_DIRS = [
    "docs", "sop", "scenarios", "logs",
    "reports", "aar", "eval", "package"
]
for d in DELIVERABLE_DIRS:
    Path(d).mkdir(parents=True, exist_ok=True)

def load_mission_dataset(mission_num: int) -> dict:
    fname = f"trn08_m{mission_num:02d}_dataset.json"
    candidates = [
        Path(fname),
        Path("uploads/TRN08") / fname,
        Path("../uploads/TRN08") / fname,
    ]
    for p in candidates:
        if p.exists():
            with open(p, "r", encoding="utf-8") as f:
                return json.load(f)
    raise FileNotFoundError(f"데이터셋을 찾을 수 없습니다: {fname}")

def save_deliverable(rel_path: str, content: str) -> Path:
    out = Path(rel_path)
    out.parent.mkdir(parents=True, exist_ok=True)
    out.write_text(content, encoding="utf-8")
    print(f"[저장 완료] {out}")
    return out

print("TRN08 실습 환경 준비 완료. 디렉토리:", ", ".join(DELIVERABLE_DIRS))
```

## 🎯 미션 08. AAR을 통한 오청/오인 식별 사례 분석 및 청음/식별 요령 보완

### 1. 미션 개요 및 학습 목표

### 3. 단계별 AI 활용 가이드 (프롬프트 예시 & HITL 검증 포인트)
- **Step 1**: 프롬프트 질의 예시:  가상 기록 발췌 F1~F7을 입력하고, '참여자 진술과 타임라인 시각·방위 기록이 서로 다른 불일치 지점을 추출해 팩트체크 표로 정리해 줘.'라고 질의한다. (※ 실제 부대명·진지 위치·호출부호·실제 드론 제원/주파수는 입력하지 않고 제공된 가상 자료만 활용한다.)  AI 출력 검증 포인트 (HITL): AI가 표시한 불일치 지점이 F1~F7 원문과 일치하는지 확인하고, 근거 없는 추측은 표에서 제외한다.
- **Step 2**: 프롬프트 질의 예시:  팩트체크가 끝난 사례를 입력하고, 'AAR 4단계(계획한 것-실제 일어난 것-차이의 원인-개선할 것)에 따라 원인을 절차·도구·환경·숙련 4요인으로 분류해 줘.'라고 질의한다. (※ 실제 부대명·진지 위치·호출부호·실제 드론 제원/주파수는 입력하지 않고 제공된 가상 자료만 활용한다.)  AI 출력 검증 포인트 (HITL): 원인 분류에서 특정 개인 탓이나 비난성 표현이 들어가면 즉시 삭제하고, 기록지 근거가 있는 항목만 매트릭스에 남긴다.
- **Step 3**: 프롬프트 질의 예시:  토의에서 나온 개선 의견 초안을 입력하고, '개인 비난 표현을 사실 기반의 절차 개선 문장으로 바꾸고, 실행 가능한 개선점 5건 이상으로 우선순위를 매겨 줘.'라고 질의한다. (※ 실제 부대명·진지 위치·호출부호·실제 드론 제원/주파수는 입력하지 않고 제공된 가상 자료만 활용한다.)  AI 출력 검증 포인트 (HITL): 도출된 5건 이상의 개선점이 오청·오인 사례별 보완 요령을 실제로 포함하는지 조별 합의로 확인한다.
- **Step 4**: 프롬프트 질의 예시:  확정된 개선점 5건 이상을 입력하고, '미션 2의 SOP와 대조할 수 있도록 현행-수정-근거 3열 구조의 SOP 1차 수정안 표로 정리해 줘.'라고 질의한다. (※ 실제 부대명·진지 위치·호출부호·실제 드론 제원/주파수는 입력하지 않고 제공된 가상 자료만 활용한다.)  AI 출력 검증 포인트 (HITL): 수정안의 '근거' 열에 미션 4~7 기록지 또는 AAR 매트릭스의 구체적 사실이 인용되었는지 검증한다.
- **Step 5**: 프롬프트 질의 예시:  SOP 1차 수정안을 입력하고, '수정된 절차 중 재평가 훈련에서 가장 먼저 측정·검증해야 할 핵심 지표 2가지를 선정해 줘.'라고 질의한다. (※ 실제 부대명·진지 위치·호출부호·실제 드론 제원/주파수는 입력하지 않고 제공된 가상 자료만 활용한다.)  AI 출력 검증 포인트 (HITL): 선정된 지표가 미션 9의 평가표와 동일 조건으로 측정 가능한지 확인한다.

```python
# [미션 08 - 교수자 해설용] F1~F7 팩트체크 및 4요인 분류 · SOP 1차 수정안
ds = load_mission_dataset(8)
excerpts = ds["provided_materials"][0]["excerpts_F1_F7"]
cause_map = {
    "F1": "절차", "F2": "환경", "F3": "도구", "F4": "숙련",
    "F5": "절차", "F6": "숙련", "F7": "절차",
}
counts = {}
for k, v in cause_map.items():
    counts[v] = counts.get(v, 0) + 1
print("4요인 원인 분포:", counts)

md = "# TRN08 M08 AAR 회의록 및 오류 분석 매트릭스\n" + "\n".join(
    f"- {ex['id']} [{cause_map[ex['id']]}]: {ex['기록지 사실']}" for ex in excerpts
)
save_deliverable("aar/trn08_aar_session_minutes.md", md)
```
