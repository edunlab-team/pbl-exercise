- **소요 시간:** 5일 / 27.0시간 (레벨 2)
- **보안 및 HITL 원칙:** 실제 부대명·진지 위치·호출부호·실제 드론 제원/주파수는 절대 입력하지 않으며, 모든 AI 출력은 인간 최종 검토(HITL)를 거쳐 확정합니다.
- **산출물 디렉토리 안내:** `trn08_deliverables_guide.md` (또는 아래 1번 공통 환경 설정 셀 실행으로 자동 생성)

```python
# [공통 환경 설정] 산출물 디렉토리 생성 및 미션 데이터셋 로드 헬퍼
import json
from pathlib import Path

DELIVERABLE_DIRS = [
    "docs", "sop", "scenarios", "logs",
    "reports", "aar", "eval", "package"
]
for d in DELIVERABLE_DIRS:
    Path(d).mkdir(parents=True, exist_ok=True)

def load_mission_dataset(mission_num: int) -> dict:
    fname = f"trn08_m{mission_num:02d}_dataset.json"
    candidates = [
        Path(fname),
        Path("uploads/TRN08") / fname,
        Path("../uploads/TRN08") / fname,
    ]
    for p in candidates:
        if p.exists():
            with open(p, "r", encoding="utf-8") as f:
                return json.load(f)
    raise FileNotFoundError(f"데이터셋을 찾을 수 없습니다: {fname}")

def save_deliverable(rel_path: str, content: str) -> Path:
    out = Path(rel_path)
    out.parent.mkdir(parents=True, exist_ok=True)
    out.write_text(content, encoding="utf-8")
    print(f"[저장 완료] {out}")
    return out

print("TRN08 실습 환경 준비 완료. 디렉토리:", ", ".join(DELIVERABLE_DIRS))
```

### 3. 단계별 AI 활용 가이드 (프롬프트 예시 & HITL 검증 포인트)
- **Step 1**: 프롬프트 질의 예시:  가상 관측 상황 카드를 입력하고, '규모(S)·행동(A)·위치(L)·식별(U)·시각(T)·장비(E) 6항목 SALUTE 포맷에 맞춰 빠진 항목 없이 즉각 보고 문안 초안을 정리해 줘.'라고 질의한다. (※ 실제 부대명·진지 위치·호출부호·실제 드론 제원/주파수는 입력하지 않고 제공된 가상 자료만 활용한다.)  AI 출력 검증 포인트 (HITL): 상대 위치·시각 등 가상 카드 원문 내용이 임의로 바뀌지 않았는지 확인하고, 실제 호출부호·좌표가 들어가지 않도록 점검한다.
- **Step 2**: 프롬프트 질의 예시:  SALUTE 무전 보고 녹취 요약(가상)을 입력하고, '복창 과정에서 누락되거나 숫자가 잘못 전달되기 쉬운 항목을 표시해 줘.'라고 질의한다. (※ 실제 부대명·진지 위치·호출부호·실제 드론 제원/주파수는 입력하지 않고 제공된 가상 자료만 활용한다.)  AI 출력 검증 포인트 (HITL): 실제 2인 1조 구두 보고 시 보고 시간과 지휘관 복창 일치 여부를 관찰자가 직접 측정·기록한다.
- **Step 3**: 프롬프트 질의 예시:  교대 시점의 가상 상황 정보를 입력하고, 'SBAR(상황-배경-평가-요청) 형식과 필수 인계 항목 10개가 모두 포함되었는지 점검표를 만들어 줘.'라고 질의한다. (※ 실제 부대명·진지 위치·호출부호·실제 드론 제원/주파수는 입력하지 않고 제공된 가상 자료만 활용한다.)  AI 출력 검증 포인트 (HITL): 필수 10개 항목 중 '미해결 경보'나 '장비·시야 제약'이 빠지지 않았는지 관찰자 체크리스트로 대조한다.
