## 🎯 미션 01. 원거리 드론·미사일 공격 사례 분석 및 공격 주체 귀속(Attribution) 핵심 요소 정의 (`ISR05_M01`)

**소요 시간:** 1.5시간 | **필수 산출물:** `docs/isr05_doc_framework.md (필수), 분석 개요서 (필수)`

### 1. 미션 개요
무인기·미사일 공격 직후에는 잔해 사진, 불완전한 항적, 엇갈린 책임 성명이 동시에 유입되어 단일 단서만으로 공격 주체를 단정하기 쉽습니다. 본 미션에서는 가상 공격 사례 카드 5건을 비교·분석해 공격 원점 추정, 잔해 표식, 비행 항적, 신호 기록, 성명 등 핵심 귀속 판단 요소와 증거 유형별 오류(재사용 가능한 부품, 위조 가능한 성명, 결측 항적)를 도출합니다.
실제 무기 제원을 배제한 추상 서술 원칙 아래 지휘관 결심 의제 순서에 맞춘 8개 장 목차와 문서 골격(docs/isr05_doc_framework.md, 분석 개요서)을 정의하며, 이는 이후 미션 2~10에서 세부 기준과 매트릭스를 채워 넣는 전체 설계도로 활용됩니다.
실습 노트북: isr05_m01_workbook.ipynb

### 2. 학습 목표
실습자료의 가상 공격 사례 카드 5건을 읽고 사례마다 어떤 증거(공격 원점 추정, 잔해 표식, 비행 항적, 신호 기록, 성명)가 있었고 어디까지 말할 수 있었는지 비교표로 정리한다. 증거 유형마다 쉽게 빠지는 오류(재사용 가능한 부품, 위조 가능한 성명, 결측이 있는 항적)를 적는다. 실제 무기 제원은 쓰지 않고 '표식 형식', '진입 방향 범위'처럼 추상 서술만 사용한다. 지휘관이 가장 먼저 알아야 할 순으로 지침서 8개 장의 목차를 배열한다.
익히는 핵심 개념·분석 기법: 5대 증거 유형(공격 원점 추정·잔해 표식·비행 항적·신호 기록·성명)별 귀속 판단 요소 추출, 무기 제원 비노출 추상 서술법, 지휘관 결심 의제 중심의 8개 장 표준 문서 아키텍처 설계 방법을 익힙니다.
기르는 실무 판단력: 재사용 가능한 부품·위조 가능한 성명·결측 항적 등 증거 유형별 함정을 식별하고, 불완전한 초기 정보에서 '말할 수 있는 범위'와 '판단 불가 영역'을 구분하는 기준을 습득합니다.

### 3. 단계별 수행 및 AI 활용 포인트
- **Step 1.** 실습자료 '가상 공격 사례 카드 5건'을 읽고 당시 말할 수 있었던 범위와 한계를 비교한다...
  - *AI 활용 가이드:* 프롬프트 질의 예시: "가상 공격 사례 카드 5건의 요약 텍스트를 바탕으로 사례별 증거 유형(공격 원점 추정·잔해 표식·비행 항적·신호 기록·성명)과 당시 말할 수 있었던 범위·한계를 비교표로 정리해 줘. 실제 무기 제원이나 외부 사건은 추가하지 마." (※ 실제 부대명·작전 정보·개인식별정보는 입력하지 않고 제공된 가상 자료만 사용한다.)
- **Step 2.** 작성 서식 1에 증거 유형 5가지의 귀속 요소와 오류를 적는다(35분).
  - *AI 활용 가이드:* 프롬프트 질의 예시: "공격 원점·잔해·비행 항적·신호·성명 5가지 증거 유형별로 분석가가 빠지기 쉬운 오류(재사용 부품, 위조 성명, 결측 항적 등)와 추상 서술 예시('표식 형식', '진입 방향 범위')를 초안 표로 제안해 줘." (※ 실제 부대명·작전 정보·개인식별정보는 입력하지 않고 제공된 가상 자료만 사용한다.)
- **Step 3.** 작성 서식 2에 지침서 8개 장을 쓰고 결심 의제 순으로 배열한다(25분).
  - *AI 활용 가이드:* 프롬프트 질의 예시: "지휘관 결심 의제(무엇이 일어났는가→근거 수준→위험·공백→권고 조치) 순서에 맞춰 귀속 근거 매트릭스 표준 지침서 8개 장의 목차와 각 장의 핵심 질문을 배열해 줘." (※ 실제 부대명·작전 정보·개인식별정보는 입력하지 않고 제공된 가상 자료만 사용한다.)
- **Step 4.** 점검표로 빠진 증거 유형이 없는지 확인한다(10분).
  - *AI 활용 가이드:* 프롬프트 질의 예시: "작성한 5대 증거 유형·오류 표와 8개 장 목차를 점검해 빠진 증거 유형이나 서로 중복되는 장이 있는지 체크리스트 형태로 진단해 줘." (※ 실제 부대명·작전 정보·개인식별정보는 입력하지 않고 제공된 가상 자료만 사용한다.)
- **Step 5.** 도전 과제 및 파라미터 튜닝 최적화
  - *AI 활용 가이드:* 프롬프트 질의 예시: "서로 다른 두 집단 명의의 책임 성명 2건이 동시에 나왔을 때, 왜 특정 집단 귀속이 아니라 '판단 불가'가 적절한지 지휘관 보고용 2문장 논거 초안을 작성해 줘." (※ 실제 부대명·작전 정보·개인식별정보는 입력하지 않고 제공된 가상 자료만 사용한다.)
