```python
import json
from pathlib import Path

DATA_DIR = Path(".")
if not (DATA_DIR / "isr05_m01_dataset.json").exists():
    alt = Path("uploads/ISR05")
    if alt.exists():
        DATA_DIR = alt

def load_mission_dataset(mission_num: int) -> dict:
    path = DATA_DIR / f"isr05_m{mission_num:02d}_dataset.json"
    with open(path, encoding="utf-8") as f:
        return json.load(f)

BANNED_PHRASES = ["소행이다", "명백하다", "틀림없다", "확실하다", "단정한다"]

def check_banned_phrases(text: str) -> list[str]:
    return [p for p in BANNED_PHRASES if p in text]

print("DATA_DIR:", DATA_DIR.resolve())
print("단정 금지어 목록:", BANNED_PHRASES)
```

### 2. 학습 목표
목적·적용 범위, 용어 정의, 귀속 판단 요소와 증거 분류, 증거 매트릭스와 가중치, 교차검증(위장 깃발·대리 세력), 상정·보류와 승인, 기록·보고, 훈련·평가·개정의 8개 장과 부록(증거 6단계, 매트릭스 서식, 의사결정 트리, 보고 서식, 루브릭)을 편집한다. 모든 기준 옆에 근거(어느 미션 결과·어느 가상 자료인지)를 표기하고 교리·법무·작전 역할의 모의 심의 지적 사항을 반영한다. 연 1회 개정 주기와 의견 수렴 방법을 명시한다.
익히는 핵심 개념·분석 기법: 8개 장 본문과 5종 부록을 포괄하는 표준 지침서(SOP/매뉴얼) 편찬 체계, 모든 기준에 출처 미션·자료 근거를 병기하는 추적성(Traceability) 확보 및 연 1회 개정 거버넌스 설계 방법을 익힙니다.
기르는 실무 판단력: 교리·법무·작전 관점의 모의 심의 지적 사항을 조율해 반영하고, 현장 실무에서 재현·감사 가능한 완결된 귀속 평가 표준을 최종 확정하는 종합 역량을 기릅니다.
