```python
import json
from pathlib import Path

DATA_DIR = Path(".")
if not (DATA_DIR / "isr05_m01_dataset.json").exists():
    alt = Path("uploads/ISR05")
    if alt.exists():
        DATA_DIR = alt

def load_mission_dataset(mission_num: int) -> dict:
    path = DATA_DIR / f"isr05_m{mission_num:02d}_dataset.json"
    with open(path, encoding="utf-8") as f:
        return json.load(f)

BANNED_PHRASES = ["소행이다", "명백하다", "틀림없다", "확실하다", "단정한다"]

def check_banned_phrases(text: str) -> list[str]:
    return [p for p in BANNED_PHRASES if p in text]

print("DATA_DIR:", DATA_DIR.resolve())
print("단정 금지어 목록:", BANNED_PHRASES)
```

재보정 기준을 미션 5 경계선 상황과 미션 8 AAR 사례 7건에 재적용해 개선 효과를 검증한 뒤 BLUF 기반 1페이지 브리핑(briefing/isr05_commander_1page_brief.md(pdf))을 작성하며, 확정된 기준은 미션 10의 최종 표준 지침서에 통합됩니다.

### 2. 학습 목표
가중치를 각각 20% 높이거나 낮췄을 때, 상정 임계값을 바꿨을 때 가설별 점수와 의견 상태가 어떻게 달라지는지 민감도 표를 만든다. 결론이 가중치 변화에 쉽게 뒤집히는 사례는 '결론이 불안정함'으로 표시하고 보류 대상으로 정한다. 미션5 경계선 상황과 미션8 AAR 사례 7건에 재보정 기준을 다시 적용해 비교한 뒤, 결론부터 쓰는(BLUF) 1페이지 브리핑에 핵심 결심 사항, 위험 요소(신호등), 대안 3종을 담아 3분 안에 발표한다.
익히는 핵심 개념·분석 기법: 가중치 ±20% 변동 및 상정 임계값 조정에 따른 결론 변화 민감도 분석표 작성, 재보정 전후 비교 검증, 핵심 결심·신호등 위험·대안 3종을 담은 BLUF 1페이지 브리핑 작성법을 익힙니다.
기르는 실무 판단력: 민감도 변화에 결론이 쉽게 뒤집히는 취약 사례를 '결론 불안정(보류)'으로 격리하고, 지휘관에게 3분 내 핵심 판단과 한계를 명확히 전달하는 의사결정 지원 역량을 습득합니다.
