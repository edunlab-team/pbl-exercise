가상 증거 5종을 교차 채점해 기준 문구의 모호함을 보정하고 단일 증거 유형만으로 귀속 의견을 쓰지 못하도록 제한하는 규칙을 확립하며, 완성된 framework/isr05_source_reliability_matrix.md는 미션 3~4의 매트릭스·가중치 설계 기준으로 연계됩니다.

### 2. 학습 목표
증거를 물리 증거, 신호 증거, 디지털 증거, 주장·진술로 나누고 증거 증명력 6단계(1 주장만 있음 ~ 6 서로 다른 유형 다중 독립 확인)의 판정 기준을 관찰 가능한 문장으로 쓴다. 증거 평가 지표를 10개 이상 정의하고(출처 독립성, 재사용 가능성, 조작 가능성, 시각 정합성 등) 지표마다 확인 방법을 적는다. 가상 증거 5종에 조원이 각자 단계를 매겨 일치 여부를 비교하고 불일치한 기준 문구를 고친다. 한 증거 유형만으로 귀속 의견을 쓰지 못하게 하는 규칙을 둔다.
익히는 핵심 개념·분석 기법: 물리·신호·디지털·주장 증거의 4대 분류 체계, 관찰 가능한 문장으로 작성하는 증명력 6단계 판정 기준, 출처 독립성·조작 가능성·시각 정합성 등 10개 이상 평가 지표 설계 방법을 익힙니다.
