## 🎯 미션 04. 증거 수준별 가중치 부여 및 귀속 신뢰도 점수 계산 논리 수립 (`ISR05_M04`)

**소요 시간:** 3.0시간 | **필수 산출물:** `matrix/isr05_attribution_matrix.xlsx(md) (필수)`

### 1. 미션 개요
증거의 중요도와 불확실성을 정량화하지 않으면 다수의 약한 정황이 소수의 강한 반박 증거를 덮어버릴 수 있습니다. 본 미션에서는 귀속 판단 기준 12개 이상을 항목화해 증명력 단계·독립성·시각 정합성에 따른 가중치를 부여하고, 지지·반박·확인 불가 감점을 반영한 가설별 귀속 신뢰도 점수 산식을 수립합니다.
가설 간 점수 차이가 작을 때 의견을 보류하는 규칙과 선전 주장에 대한 레드팀 반박 검증, 법무 검토 연계 표시를 포함해 점수가 사람의 승인을 대신하지 못하도록 통제하며, 완성된 matrix/isr05_attribution_matrix.xlsx(md)는 미션 5 의사결정 트리의 정량 기준으로 활용됩니다.
실습 노트북: isr05_m04_workbook.ipynb

### 2. 학습 목표
귀속 판단 기준(증거 항목)을 12개 이상 항목화하고 증명력 단계, 독립성, 시각 정합성에 따라 항목별 가중치를 정한다. 가설별 점수 = 지지 가중치 합 - 반박 가중치 합 - 확인 불가 개수 x 감점으로 산식을 세우고, 가설 간 점수 차이가 작으면 귀속 의견을 내지 않는 규칙을 둔다. 가상 사건에 산식을 적용해 가설별 점수를 계산하고 레드팀이 '이 점수는 선전 주장에 끌려가지 않는가'를 반박하게 한다. 국제법적 귀속 기준 쟁점은 법무 검토 연계 항목으로만 표시한다.
익히는 핵심 개념·분석 기법: 증명력 단계·독립성·시각 정합성을 반영한 항목별 가중치 설계, 지지·반박·확인 불가 감점을 포함한 귀속 신뢰도 점수 산식 구성 및 레드팀 반박 검증 절차를 익힙니다.
기르는 실무 판단력: 가설 간 점수 차가 미미할 때 의견을 보류하는 임계 판단, 선전 주장에 의한 점수 왜곡 방지, 그리고 정량 점수가 사람의 최종 승인을 대체할 수 없는 HITL 통제 기준을 습득합니다.
