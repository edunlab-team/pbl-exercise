## 🎯 미션 05. 위장 깃발(False-flag) 및 대리 세력 개입 가능성 배제를 위한 교차검증 항목 도출 (`ISR05_M05`)

**소요 시간:** 3.0시간 | **필수 산출물:** `sop/isr05_approval_thresholds_sop.md (필수)`

### 1. 미션 개요
적대 세력이 제3자로 위장하는 위장 깃발(False-flag)이나 대리 세력을 동원한 공격은 단순 점수 계산만으로는 걸러내기 어렵습니다. 본 미션에서는 성명·사건 시각 정합성, 재사용 부품, 제3자 이득 가능성 등 위장 깃발 점검 항목 6개 이상과 지원·지휘 연결 등 대리 세력 점검 항목 6개 이상을 도출합니다.
미션 2~4의 기준을 증거 단계, 가설 점수 차이, 교차검증 결과, 승인 권한 순의 의사결정 트리로 통합해 경계선 상황 5종에 적용하며, 산출물 sop/isr05_approval_thresholds_sop.md는 귀속 의견 상정·보류 분기점과 책임 직책을 확정하는 승인 기준서가 됩니다.
실습 노트북: isr05_m05_workbook.ipynb

### 2. 학습 목표
위장 깃발 점검 항목(성명 시각과 사건 시각의 정합성, 재사용 가능한 부품, 이득을 보는 제3자 가능성)과 대리 세력 점검 항목(지원 관계 정황, 지휘 연결 증거 유무)을 각각 6개 이상 작성한다. 미션2~미션4 기준을 증거 단계, 가설 점수 차이, 위장·대리 점검 결과, 승인 권한의 순으로 의사결정 트리로 묶고 분기점마다 조건과 책임 직책을 적는다. 경계선 상황 5종에 적용해 '귀속 의견 상정(사람 승인 대기)·보류 유지·추가 수집 요청' 중 하나를 고르고 근거를 기록한다.
익히는 핵심 개념·분석 기법: 위장 깃발(성명·사건 시각 정합성, 재사용 부품, 제3자 이득) 및 대리 세력(지원·지휘 연결) 교차검증 체크리스트 작성, 증거 단계·점수 차·검증 결과·승인 권한을 묶은 의사결정 트리 설계 방법을 익힙니다.
기르는 실무 판단력: 경계선 상황에서 '귀속 의견 상정(승인 대기)·보류 유지·추가 수집 요청' 중 적절한 분기를 선택하고 책임 직책을 명확히 분계하는 작전 승인 판단력을 기릅니다.
