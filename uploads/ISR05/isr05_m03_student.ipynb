## 🎯 미션 03. 성급한 적 귀속 단정 방지를 위한 다차원 귀속 증거 매트릭스(Attribution Matrix) 설계 (`ISR05_M03`)

**소요 시간:** 2.5시간 | **필수 산출물:** `analysis/isr05_uncertainty_assessment.md (필수)`

### 1. 미션 개요
특정 주체를 미리 가정한 채 증거를 해석하면 확증 편향으로 인해 성급한 적 귀속 단정에 빠지기 쉽습니다. 본 미션에서는 공격 주체 가설 3개(가상 집단 X, 가상 집단 Y, 제3자 위장)를 설정하고, 증거 8개를 가설별로 지지(+)·반박(-)·중립(0)·확인 불가(?)로 대조하는 다차원 귀속 증거 매트릭스를 설계합니다.
매트릭스가 답하지 못하는 관측·정보 공백 요인 8개 이상의 영향도와 보완 수집 수단을 정리하고 단정 표현 금지 및 근거 수준 병기 규칙을 정립하며, 산출물 analysis/isr05_uncertainty_assessment.md는 미션 4의 신뢰도 점수 논리와 결합됩니다.
실습 노트북: isr05_m03_workbook.ipynb

### 2. 학습 목표
가상 사건의 공격 주체 가설 3개(가상 집단 X, 가상 집단 Y, 제3자 위장)를 세우고 증거 8개를 가설마다 지지(+)·반박(-)·중립(0)·확인 불가(?)로 표시하는 매트릭스를 만든다. 매트릭스가 답하지 못하는 관측 공백·정보 공백 요인을 8개 이상 찾아 영향도(높음·중간·낮음)를 매기고 보완 수집 수단을 적는다. 결론 문장에는 '가능성이 있다', '가능성이 높다'처럼 근거 수준을 함께 쓰는 표현 규칙을 정하고 단정 표현은 금지한다.
익히는 핵심 개념·분석 기법: 다중 가설(가상 집단 X·Y·제3자 위장) 대조형 귀속 증거 매트릭스(+/-/0/?) 설계, 관측·정보 공백 영향도 평가 및 보완 수집 수단 매핑 기법을 익힙니다.
기르는 실무 판단력: 확증 편향을 피하기 위해 반박·중립 증거를 동등하게 배치하고, 단정 표현을 금지하며 근거 수준을 병기하는 확률적·유보적 결론 서술 기준을 습득합니다.

### 3. 단계별 수행 및 AI 활용 포인트
- **Step 1.** 가설 3개를 확인하고 증거 8개를 읽는다(15분).
  - *AI 활용 가이드:* 프롬프트 질의 예시: "가상 사건의 가설 3개(가상 집단 X, 가상 집단 Y, 제3자 위장)와 증거 8개의 핵심 사실을 한눈에 비교할 수 있도록 사전 요약표를 만들어 줘." (※ 실제 부대명·작전 정보·개인식별정보는 입력하지 않고 제공된 가상 자료만 사용한다.)
- **Step 2.** 작성 서식 1 매트릭스 24칸을 개인별로 채운 뒤 조 합의안을 만든다(50분).
  - *AI 활용 가이드:* 프롬프트 질의 예시: "가설 3개 × 증거 8개(24칸) 매트릭스에서 각 칸을 지지(+)·반박(-)·중립(0)·확인 불가(?)로 판정할 때, 특히 제3자 위장 가설과 충돌할 수 있는 함정 포인트를 짚어 줘." (※ 실제 부대명·작전 정보·개인식별정보는 입력하지 않고 제공된 가상 자료만 사용한다.)
- **Step 3.** 작성 서식 2에 정보 공백을 8개 이상 쓰고 영향도와 보완 수집 수단을 적는다(35분).
  - *AI 활용 가이드:* 프롬프트 질의 예시: "현재 24칸 매트릭스가 답하지 못하는 관측·정보 공백 요인 8개 이상을 도출하고 영향도(높음·중간·낮음)와 보완 수집 수단을 표로 제안해 줘." (※ 실제 부대명·작전 정보·개인식별정보는 입력하지 않고 제공된 가상 자료만 사용한다.)
- **Step 4.** 작성 서식 3의 표현 규칙과 한계 3가지를 적고 점검표로 확인한다(15분).
  - *AI 활용 가이드:* 프롬프트 질의 예시: "매트릭스 결론 문장에서 단정 표현을 배제하고 '가능성이 있다(근거 수준 ○)' 형태로 쓰는 표현 규칙과 한계 3가지의 작성 예시를 검토해 줘." (※ 실제 부대명·작전 정보·개인식별정보는 입력하지 않고 제공된 가상 자료만 사용한다.)
- **Step 5.** 도전 과제 및 파라미터 튜닝 최적화
  - *AI 활용 가이드:* 프롬프트 질의 예시: "가장 영향도가 높은 정보 공백 1건이 해소되어 지지(+) 또는 반박(-)으로 바뀔 때, 가설 3개 중 어느 가설의 입지가 가장 크게 변하는지 민감도 흐름을 설명해 줘." (※ 실제 부대명·작전 정보·개인식별정보는 입력하지 않고 제공된 가상 자료만 사용한다.)
