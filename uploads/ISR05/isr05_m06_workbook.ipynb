> **보안 원칙:** 제공된 교육용 가상 자료(`isr05_m01_dataset.json` ~ `isr05_m10_dataset.json`)만 사용하며 실제 부대명·작전 정보·개인식별정보는 외부 AI에 입력하지 않습니다.

```python
import json
from pathlib import Path

DATA_DIR = Path(".")
if not (DATA_DIR / "isr05_m01_dataset.json").exists():
    alt = Path("uploads/ISR05")
    if alt.exists():
        DATA_DIR = alt

def load_mission_dataset(mission_num: int) -> dict:
    path = DATA_DIR / f"isr05_m{mission_num:02d}_dataset.json"
    with open(path, encoding="utf-8") as f:
        return json.load(f)

BANNED_PHRASES = ["소행이다", "명백하다", "틀림없다", "확실하다", "단정한다"]

def check_banned_phrases(text: str) -> list[str]:
    return [p for p in BANNED_PHRASES if p in text]

print("DATA_DIR:", DATA_DIR.resolve())
print("단정 금지어 목록:", BANNED_PHRASES)
```

## 🎯 미션 06. 지휘관 보고용 공격 주체 귀속 근거 요약 대시보드 및 브리핑 보고서 서식 작성 (`ISR05_M06`)

**소요 시간:** 2.5시간 | **필수 산출물:** `templates/isr05_aar_fact_action_template.md (필수)`

### 1. 미션 개요
분석 과정이 아무리 정교해도 보고 단계에서 사실과 추정, 미확인 공백이 뒤섞이면 지휘관이 근거 수준을 오해해 과도한 대응을 결심할 수 있습니다. 본 미션에서는 가설별 점수, 근거 수준, 반박 증거, 교차검증 결과, 현재 상태를 한눈에 보여주는 요약 대시보드와 지휘관 보고서 서식을 설계합니다.
사실·판단과 근거·조치·미확인 사항 칸을 엄격히 분리하고 단정 금지어 및 대체 표현 규칙을 적용해 가상 보고 3종을 일관성 있게 정리하며, 완성된 templates/isr05_aar_fact_action_template.md는 미션 7 모의 평가와 미션 9 브리핑의 표준 기록 틀로 쓰입니다.
실습 노트북: isr05_m06_workbook.ipynb

### 2. 학습 목표
한눈에 보는 요약 대시보드(가설별 점수, 근거 수준, 반박 증거, 위장·대리 점검 결과, 현재 상태)와 보고서 서식을 만든다. 기록은 사실, 판단과 근거, 취한 조치, 미확인 사항 칸을 분리하고 '소행이다', '틀림없다', '명백하다'처럼 단정하는 표현의 금지 목록과 대체 표현(예: '가능성이 있다(근거 수준 3)')을 정한다. 가상 보고 3종을 서식에 옮겨 분류가 일관되는지 확인한다.
익히는 핵심 개념·분석 기법: 가설별 점수·근거 수준·반박 증거·교차검증 상태를 통합한 지휘관 요약 대시보드 설계, 사실·판단과 근거·조치·미확인 사항을 분리하는 표준 보고 서식 작성법을 익힙니다.
기르는 실무 판단력: '소행이다·명백하다' 등 단정 금지어를 근거 수준 기반의 대체 표현으로 교정하고, 지휘관이 오해 없이 미확인 공백까지 함께 인지하도록 정보를 구조화하는 역량을 습득합니다.

### 3. 단계별 수행 및 AI 활용 포인트
- **Step 1.** 원문 보고 3종에서 사실과 판단, 단정 표현을 구분해 표시한다(20분).
  - *AI 활용 가이드:* 프롬프트 질의 예시: "가상 원문 보고 3종의 문장을 '사실 / 판단과 근거 / 단정 표현 / 미확인 사항' 4가지 태그로 예비 분류해 줘." (※ 실제 부대명·작전 정보·개인식별정보는 입력하지 않고 제공된 가상 자료만 사용한다.)
- **Step 2.** 단정 표현 금지 목록과 대체 표현을 정한다(25분).
  - *AI 활용 가이드:* 프롬프트 질의 예시: "'소행이다', '틀림없다', '명백하다' 같은 단정 표현 금지 목록과, 근거 수준(1~6단계)을 함께 밝히는 대체 표현('가능성이 있다(근거 수준 3)' 등) 대조표를 만들어 줘." (※ 실제 부대명·작전 정보·개인식별정보는 입력하지 않고 제공된 가상 자료만 사용한다.)
- **Step 3.** 요약 대시보드와 기록 서식을 작성하고 보고 3종을 옮겨 적는다(55분).
  - *AI 활용 가이드:* 프롬프트 질의 예시: "가설별 점수, 근거 수준, 반박 증거, 위장·대리 점검 결과, 현재 상태를 한눈에 담는 지휘관 요약 대시보드와 보고서 서식에 가상 보고 3종을 옮긴 초안을 구성해 줘." (※ 실제 부대명·작전 정보·개인식별정보는 입력하지 않고 제공된 가상 자료만 사용한다.)
- **Step 4.** 조원 간 분류 일치를 확인하고 점검표로 확인한다(25분).
  - *AI 활용 가이드:* 프롬프트 질의 예시: "조원들이 분류한 결과와 대조할 수 있도록 보고 3종의 경계선 문장(사실 vs 판단)에 대한 판단 기준 체크리스트를 정리해 줘." (※ 실제 부대명·작전 정보·개인식별정보는 입력하지 않고 제공된 가상 자료만 사용한다.)
- **Step 5.** 도전 과제 및 파라미터 튜닝 최적화
  - *AI 활용 가이드:* 프롬프트 질의 예시: "'현재 근거로는 배제할 수 없다'는 표현을 지휘관이 '곧 공격 주체다'로 오해하지 않도록 괄호나 한 줄 부연 문구를 3가지 버전으로 제안해 줘." (※ 실제 부대명·작전 정보·개인식별정보는 입력하지 않고 제공된 가상 자료만 사용한다.)

```python
# 미션 06 교육용 가상 데이터셋 로드
ds_m06 = load_mission_dataset(6)
print('미션:', ds_m06['mission_id'], '|', ds_m06['mission_title'])
print('산출물:', ds_m06['deliverables'])
print(json.dumps(ds_m06['provided_materials'][0], ensure_ascii=False, indent=2)[:800])
```

```python
# [해설·검증] 원문 보고 3종 단정 금지어 자동 점검(Lint) 및 대체 표현 확인
ds6 = load_mission_dataset(6)
for r in ds6["provided_materials"][0]["reports"]:
    found = check_banned_phrases(r["raw_text"])
    print(f"[{r['id']}] 검출된 단정 금지어: {found if found else '없음(PASS)'}")
```
