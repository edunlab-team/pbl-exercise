```python
# [해설·검산] 평가자 2인 독립 채점 일치율 계산
domains = ["증거 분류", "상충 처리", "위장·대리 점검", "보류 원칙 준수", "기록"]
rater_a = ["탁월", "우수", "보통", "우수", "탁월"]
rater_b = ["탁월", "보통", "보통", "우수", "탁월"]

matches = sum(a == b for a, b in zip(rater_a, rater_b))
agreement = matches / len(domains) * 100
print(f"영역별 일치: {matches}/{len(domains)} ({agreement:.1f}%)")
for d_name, a, b in zip(domains, rater_a, rater_b):
    if a != b:
        print(f"  - 불일치 영역 [{d_name}]: 평가자A={a}, 평가자B={b} -> 관찰 가능 행동 기준 보완 필요")
```
