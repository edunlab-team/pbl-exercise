## 🎯 미션 08. AAR을 통한 과거 공격 귀속 오판 사례 분석 및 증거 검증 기준 고도화 (`ISR05_M08`)

**소요 시간:** 3.5시간 | **필수 산출물:** `reports/isr05_redteam_critique_report.md (필수)`

### 1. 미션 개요
과거의 귀속 오판 사례와 기준서 내부의 암묵적 전제를 점검하지 않으면 동일한 판단 오류가 반복될 수 있습니다. 본 미션에서는 가상 AAR 일지 2건(선전 성명 의존 오판, 재사용 부품 오판)을 시간순으로 재구성해 사람·절차·정보·통신 관점의 근본 원인을 분석합니다.
레드팀 관점에서 기준서의 취약한 전제 5건 이상을 비판·보완하고 귀속 지연으로 인한 작전 활용성 저하까지 균형 있게 다루며, 정리된 reports/isr05_redteam_critique_report.md는 미션 9의 가중치·임계값 재보정 입력으로 반영됩니다.
실습 노트북: isr05_m08_workbook.ipynb

### 2. 학습 목표
가상 AAR 일지 2건(선전 성명을 근거로 한 성급한 귀속, 재사용 가능한 부품을 결정적 증거로 본 오판)을 시간순으로 재구성하고 원인을 사람·절차·정보·통신으로 나눈다. 레드팀은 기준서의 약한 전제(예: '책임 주장 성명은 주체를 가리킨다')를 5건 이상 비판하고 작성 팀은 쟁점마다 반박 또는 수정안을 낸다. 귀속 의견이 너무 늦어 지휘관 판단에 쓰이지 못하는 반대 방향의 위험도 다룬다.
익히는 핵심 개념·분석 기법: 사후검토(AAR) 타임라인 재구성을 통한 사람·절차·정보·통신 원인 분류, 기준서의 암묵적·취약한 전제를 공격하는 레드팀 비평 및 수정안 도출 기법을 익힙니다.
기르는 실무 판단력: 성급한 오판뿐 아니라 과도한 보류로 귀속 의견이 지휘관 결심 시기를 놓치는 반대 방향의 작전 위험까지 균형 있게 평가해 기준서를 개선하는 판단력을 기릅니다.

### 3. 단계별 수행 및 AI 활용 포인트
- **Step 1.** AAR 일지 A, B를 시간순으로 재구성한다(30분).
  - *AI 활용 가이드:* 프롬프트 질의 예시: "가상 AAR 일지 A(선전 성명 의존 오판)와 B(재사용 부품 오판)의 사건 기록을 시간순 타임라인 표(시각·입수 정보·당시 판단·실제 사실)로 재구성해 줘." (※ 실제 부대명·작전 정보·개인식별정보는 입력하지 않고 제공된 가상 자료만 사용한다.)
- **Step 2.** 원인 분석표를 작성한다(35분).
  - *AI 활용 가이드:* 프롬프트 질의 예시: "재구성한 타임라인을 바탕으로 오판의 근본 원인을 사람·절차·정보·통신 4개 범주로 분류한 원인 분석표 초안을 작성해 줘." (※ 실제 부대명·작전 정보·개인식별정보는 입력하지 않고 제공된 가상 자료만 사용한다.)
- **Step 3.** 레드팀이 전제를 5건 이상 비판하고 작성 팀이 반박·수정안을 낸다(50분).
  - *AI 활용 가이드:* 프롬프트 질의 예시: "레드팀 관점에서 우리 귀속 기준서의 약한 전제(예: '책임 성명은 공격 주체를 가리킨다') 5건 이상을 공격하는 비판 질문과, 작성 팀의 수정안 방향을 대조표로 만들어 줘." (※ 실제 부대명·작전 정보·개인식별정보는 입력하지 않고 제공된 가상 자료만 사용한다.)
- **Step 4.** 판단 지연의 반대 위험과 수정 반영 계획표를 작성하고 점검표로 확인한다(35분).
  - *AI 활용 가이드:* 프롬프트 질의 예시: "성급한 오판 방지와 반대로, 귀속 의견이 너무 늦어 지휘관 판단에 쓰이지 못하는 '판단 지연 위험'을 줄이기 위한 시한·중간 보고 보완책을 제안해 줘." (※ 실제 부대명·작전 정보·개인식별정보는 입력하지 않고 제공된 가상 자료만 사용한다.)
- **Step 5.** 도전 과제 및 파라미터 튜닝 최적화
  - *AI 활용 가이드:* 프롬프트 질의 예시: "가장 날카로운 레드팀 비판 1건을 선택했을 때, 기존 지침서의 어느 장·어느 조항 문구를 어떻게 고쳐야 하는지 개정 전후 대조안을 작성해 줘." (※ 실제 부대명·작전 정보·개인식별정보는 입력하지 않고 제공된 가상 자료만 사용한다.)
