인정 범위: 간격의 이름이 달라도 시작·끝ID와 시간의 의미가 같으면 인정한다.


```python
# M08 STEP 02 · 보고 지연과 확인 대기를 다른 시간 차로 계산한다
MID = 'M08'
STEP = 2
previous = require_previous(MID, STEP)
print('이번 단계가 읽는 표:', previous.shape)
display(previous.head(10))

# === ANSWER 시작 (교수자·해설용 예시 풀이) ===
events = previous.set_index('ID')
intervals = [
    ('A 접수→임의전파', 'A1', 'A2'),
    ('A 접수→보안관보고', 'A1', 'A4'),
    ('A 보고→발신부인확인', 'A4', 'A5'),
    ('B 보류→재확인시도', 'B1', 'B2'),
    ('B 재확인시도→확인완료', 'B2', 'B3'),
    ('B 보류→확인완료', 'B1', 'B3'),
]
rows = []
for name, sid, eid in intervals:
    diff = minute_difference(events.loc[eid, '시각'], events.loc[sid, '시각'])
    rows.append([name, sid, eid, diff, '원기록시각차'])
result = pd.DataFrame(rows, columns=['간격', '시작ID', '끝ID', '경과분', '근거유형'])
# === ANSWER 끝 ===

note = 'A의 45분(A1→A4)은 접수부터 통신보안관 보고까지 지연이고, B의 35분(B1→B3)은 접수·보류부터 확인 완료까지 대기 시간이며, B 원문의 전파까지 35분 서술 외 별도 전파 완료 시각 이벤트는 관측되지 않았다.'
record_step(MID, STEP, result, note)
```
