```python
from pathlib import Path
import json, datetime, base64, hashlib

def _utc_now():
    return datetime.datetime.now(datetime.timezone.utc).isoformat()

def save_mission_note(mission, decision, ai_conversation, evidence, changes="", limitations=""):
    """학생이 직접 작성한 실제 기록만 저장. 학습 PASS·교관 승인과 무관."""
    if type(mission) is not int or not 1 <= mission <= 10:
        raise ValueError('mission은 1～10 정수여야 합니다.')
    vals=[decision,ai_conversation,evidence]
    if any(not isinstance(v,str) or not v.strip() for v in vals):
        raise ValueError('판단, 실제 AI 대화, 실행 근거를 비워둘 수 없습니다.')
    if any(not isinstance(v,str) for v in (changes,limitations)):
        raise ValueError('변경 사항과 한계는 문자열로 작성하세요.')
    folder=ROOT/'outputs/learner_notes';folder.mkdir(parents=True,exist_ok=True)
    logpath=ROOT/'outputs/learner_workflow.json'
    old=json.loads(logpath.read_text(encoding='utf-8')) if logpath.exists() else {'source':'LEARNER_INPUT','docs':[],'log':[]}
    if old.get('source')!='LEARNER_INPUT' or not isinstance(old.get('log'),list):
        raise ValueError('학습 기록 형식이 올바르지 않습니다.')
    row={'mission':mission,'time':_utc_now(),'decision':decision,'ai_conversation':ai_conversation,
         'evidence':evidence,'changes':changes,'limitations':limitations,'source':'LEARNER_INPUT',
         'learning_pass':None,'instructor_approval':None}
    note=folder/f'M{mission:02d}.md'
    if note.exists():
        hist=folder/f'M{mission:02d}_{datetime.datetime.now(datetime.timezone.utc).strftime("%Y%m%dT%H%M%S%fZ")}.before.md'
        hist.write_bytes(note.read_bytes())
    note.write_text(f'# M{mission:02d} 학습자 작업 기록\n\n작성 시각: {row["time"]}\n\n'
                    f'## 본인 판단\n{decision}\n\n## AI 대화 원문\n{ai_conversation}\n\n'
                    f'## 실행 근거 및 실제 결과\n{evidence}\n\n## 코드 변경 및 이유\n{changes}\n\n'
                    f'## 미확인 사항\n{limitations}\n\n'
                    '※ 기록 저장은 학습 PASS 및 교관 승인이 아닙니다.\n',encoding='utf-8')
    old['log'].append(row)
    logpath.write_text(json.dumps(old,ensure_ascii=False,indent=2),encoding='utf-8')
    print('기록 저장:',note,'(학습 PASS·교관 승인 아님)')
    return str(note)

_BACKUP_FILE=Path('/content/QLT05_work_backup.json') if Path('/content').exists() else ROOT.parent/'QLT05_work_backup.json'

def export_work_json(path=None, download=True):
    """프로젝트 내부 실제 파일 스냅샷. 바이너리는 base64로 손실 없이 보존."""
    dest=Path(path) if path else _BACKUP_FILE
    if dest.resolve().is_relative_to(ROOT.resolve()):
        raise ValueError('백업 파일은 실습 폴더 밖에 저장하세요.')
    entries=[]
    for p in sorted(ROOT.rglob('*')):
        if not p.is_file() or '__pycache__' in p.parts or '.ipynb_checkpoints' in p.parts:continue
        b=p.read_bytes()
        entries.append({'path':p.relative_to(ROOT).as_posix(),'size':len(b),
                        'sha256':hashlib.sha256(b).hexdigest(),'base64':base64.b64encode(b).decode('ascii')})
    dest.parent.mkdir(parents=True,exist_ok=True)
    dest.write_text(json.dumps({'format':'QLT05-STUDENT-BACKUP-v2','created_at':_utc_now(),'files':entries},ensure_ascii=False),encoding='utf-8')
    print('JSON 백업 완료:',dest,'/ 파일 수:',len(entries))
    if download:
        try:
            from google.colab import files
            files.download(str(dest))
        except ImportError:pass
    return str(dest)

def restore_work_json(path=None):
    """기존 파일을 절대 덮어쓰지 않음. 일치하면 건너뛰고 불일치는 보호."""
    source=Path(path) if path else _BACKUP_FILE
    payload=json.loads(source.read_text(encoding='utf-8'))
    if payload.get('format')!='QLT05-STUDENT-BACKUP-v2' or not isinstance(payload.get('files'),list):
        raise ValueError('지원하지 않는 QLT05 백업 파일입니다.')
    validated=[];seen=set();root=ROOT.resolve()
    for entry in payload['files']:
        rel=entry.get('path');encoded=entry.get('base64')
        if not isinstance(rel,str) or not rel or rel in seen or rel.startswith('/') or '\\' in rel:
            raise ValueError('잘못된 백업 경로')
        target=(root/rel).resolve();seen.add(rel)
        if not target.is_relative_to(root) or '..' in Path(rel).parts:
            raise ValueError('실습 폴더 밖으로 복원할 수 없습니다.')
        b=base64.b64decode(encoded,validate=True)
        if len(b)!=entry.get('size') or hashlib.sha256(b).hexdigest()!=entry.get('sha256'):
            raise ValueError('백업 파일 해시 불일치: '+rel)
        validated.append((target,b))
    restored=unchanged=protected=0
    for p,b in validated:
        if p.exists():
            if p.is_file() and p.read_bytes()==b:unchanged+=1
            else: protected+=1
        else:
            p.parent.mkdir(parents=True,exist_ok=True);p.write_bytes(b);restored+=1
    result={'restored':restored,'unchanged':unchanged,'protected_conflicts':protected}
    print('복원 완료:',restored,'/ 동일하여 건너뜀:',unchanged,'/ 보호한 충돌:',protected)
    return result

print('학습 기록·JSON 백업·복원 도구 준비 완료 (학생 입력 기록은 PASS/승인이 아님)')
```
