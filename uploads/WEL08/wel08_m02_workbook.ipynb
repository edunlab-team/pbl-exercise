````python
# [환경 준비] 산출물 디렉토리 일괄 생성 및 가상 데이터셋 무결성 확인
import os
import json

directories = [
    'docs',
    'sop',
    'scenarios',
    'logs',
    'reports',
    'aar',
    'eval',
    'package'
]

print("=== [1] 산출물 디렉토리 생성 ===")
for d in directories:
    os.makedirs(d, exist_ok=True)
    print(f"  📁 폴더 준비 완료: {d}/")

print("\n=== [2] 가상 데이터셋 무결성 확인 ===")
data_dir = "/home/subokim/Workspace/milicontent2/uploads/WEL08"
if not os.path.exists(data_dir):
    data_dir = "."

dataset_count = 0
for i in range(1, 11):
    fname = f"wel08_m{i:02d}_dataset.json"
    fpath = os.path.join(data_dir, fname)
    if os.path.exists(fpath):
        with open(fpath, "r", encoding="utf-8") as f:
            d = json.load(f)
            dataset_count += 1
            print(f"  ✅ 데이터셋 로드: {fname} (미션: {d.get('mission')})")
    else:
        print(f"  ⚠️ 데이터셋 탐색 중: {fname}")

print(f"\n총 {dataset_count}개 데이터셋 확인 완료. 실습 준비가 완료되었습니다!")
````

## 🎯 미션 02: 설문조사, 건의함, 익명 상담 데이터 기반 불만 신호 조기 식별 체크리스트 작성

### 1. 미션 목표 및 핵심 지표
- **산출물 경로:** `sop/wel08_field_checklist.md`
- **AI 협업 프롬프트 예시:**
> "비정형 건의함/상담 데이터 10건(U01~U10)을 분석하여 잠재 위험 키워드를 추출하고, 왜곡 문장 5개를 순수 관찰 사실 문장으로 1:1 변환한 뒤, 6단계 SOP(관찰-경청-기록-보고-연계-후속확인) 흐름도를 작성해줘."

### 2. 작성 지침 (HITL 원칙)
- 가상 데이터셋(`wel08_m02_dataset.json`)을 바탕으로 실습을 전개합니다.
- AI가 제안한 내용 중 의학적 단정어('우울증', '불안장애' 등)가 있는지 검수하고 순수 관찰 사실(Fact)로 교정합니다.


````python
%%writefile sop/wel08_field_checklist.md
# [WEL08_M02] 불만 신호 조기 식별 현장 체크리스트 및 6단계 초동 SOP

## 1. 비정형 소통 데이터 위험 키워드 분석 (U01~U10)
- **고립/소외:** '아무도 신경 안 씀', '다들 피함', '눈치 보여서' (3건)
- **신체화 증상:** '숨이 막힘', '발톱 빠질 것 같음', '잠을 못 잠', '체중 7kg 감소' (5건)
- **위기 징후:** '그냥 사라지고 싶다', '총기 다룰 때 무섭다' (2건 - 긴급 개입 대상)

## 2. 왜곡된 표현 ➔ 객관적 관찰 사실 1:1 변환 매트릭스
| 번호 | 왜곡된 해석/진단 문장 (금지) | 객관적 관찰 사실 문장 (표준) | 교정 사유 |
| :---: | :--- | :--- | :--- |
| 1 | 김 이병이 심각한 우울증에 걸림 | 김 이병은 3일 연속 점심 식판의 2/3를 남기고 질문에 고개를 숙인 채 대답하지 않음 | 의학적 병명 배제, 식사량 및 반응 지연 팩트 기술 |
| 2 | 박 일병이 관심병사 티를 냄 | 박 일병은 배식 중 식판을 강하게 내려놓고 '양이 적다'며 큰 소리를 2회 침 | 낙인 표현 배제, 물리적 행동과 발화 횟수 명시 |
| 3 | 이 이병은 대인기피증이 있음 | 이 이병은 휴식 시간에 분대원 대화에 참여하지 않고 침상 구석에서 모포를 덮고 있음 | 심리 진단 배제, 구체적 장소 및 행동 서술 |
| 4 | 최 이병은 군기 해이로 불평함 | 최 이병은 지난 1주일간 전투복 치수와 배식량에 대해 분대장에게 4회 구두 건의함 | 도덕적 가치 판단 배제, 건의 횟수 객관화 |
| 5 | 정 이병이 불안장애로 손을 떪 | 정 이병은 사격 대기선에서 양손을 떨며 '격발하기 두렵다'고 1회 발화함 | 병명 단정 배제, 인용 발화와 신체 반응 기록 |

## 3. 6단계 초동 조치 표준작전절차 (SOP)
```mermaid
graph TD
    A[1. 관찰한다: 체크리스트 15개 항목] --> B[2. 경청한다: 단독 안전 공간, 열린 질문]
    B --> C[3. 기록한다: 팩트 중심 일지 작성]
    C --> D{위기 징후 포착 여부}
    D -- 자해/타해 안전 위협 --x E[긴급 바이패스: 3분 내 상담관/지휘관 즉시 보고]
    D -- 일반 고충/불만 --> F[4. 보고한다: 3분 내 분대장 구두 보고]
    F --> G[5. 연계한다: 상담관/급양보급관 연계]
    G --> H[6. 후속확인한다: 주기적 일상 복귀 점검]
```
````
