```python
# [환경 준비] 산출물 디렉토리 일괄 생성 및 가상 데이터셋 무결성 확인
import os
import json

directories = [
    'docs',
    'sop',
    'scenarios',
    'logs',
    'reports',
    'aar',
    'eval',
    'package'
]

print("=== [1] 산출물 디렉토리 생성 ===")
for d in directories:
    os.makedirs(d, exist_ok=True)
    print(f"  📁 폴더 준비 완료: {d}/")

print("\n=== [2] 가상 데이터셋 무결성 확인 ===")
data_dir = "/home/subokim/Workspace/milicontent2/uploads/WEL08"
if not os.path.exists(data_dir):
    data_dir = "."

dataset_count = 0
for i in range(1, 11):
    fname = f"wel08_m{i:02d}_dataset.json"
    fpath = os.path.join(data_dir, fname)
    if os.path.exists(fpath):
        with open(fpath, "r", encoding="utf-8") as f:
            d = json.load(f)
            dataset_count += 1
            print(f"  ✅ 데이터셋 로드: {fname} (미션: {d.get('mission')})")
    else:
        print(f"  ⚠️ 데이터셋 탐색 중: {fname}")

print(f"\n총 {dataset_count}개 데이터셋 확인 완료. 실습 준비가 완료되었습니다!")
```

### 2. 작성 지침 (HITL 원칙)
- 가상 데이터셋(`wel08_m10_dataset.json`)을 바탕으로 실습을 전개합니다.
- AI가 제안한 내용 중 의학적 단정어('우울증', '불안장애' 등)가 있는지 검수하고 순수 관찰 사실(Fact)로 교정합니다.


```python
%%writefile package/wel08_standard_training_package.md
# [WEL08_M10] 동료의 급성 스트레스 신호 발견·전문가 연계 표준 훈련 패키지

## 제1장 훈련 개요 및 심리 안전 거버넌스
1. **훈련 목적:** 신병 적응기 급식·피복 불만 속 급성 스트레스 징후 조기 포착 및 3분 내 신속 연계.
2. **비진단 원칙:** 의료적 단정 배제, 팩트 중심 관찰 기술 의무화.
3. **심리 안전 권리:** 정서 과부하 시 언제든 역할극 중단 가능한 Stop Protocol 명문화.

## 제2장 현장 관찰 체크리스트 및 6단계 초동 SOP
1. **체크리스트 15종:** 잔반율, 수면 지연, 대화 단절, 피복 치수 등 15대 관찰 지표 완비.
2. **6단계 초동 절차:** 관찰한다 ➔ 경청한다 ➔ 기록한다 ➔ 보고한다 ➔ 연계한다 ➔ 후속확인한다.
3. **긴급 예외 바이패스:** 자해/타해 위기 징후 감지 시 3분 내 상담관 및 지휘관 즉시 보고.

## 제3장 모의 사태 목록(MSEL) 및 가상 역할극 시나리오
1. **MSEL 10대 사태:** 정상 60% : 비정상 40% 균형 배치.
2. **가상 신병 역할 카드 3종:** 이등병 김가상(급식), 일병 이가상(피복), 이등병 박가상(적응위기).

## 제4장 평가 및 채점 루브릭
1. **정량 판정 기준:** 보고 지연시간 180초 이내, 오류 식별률 90% 이상, 단정어 위반 0건.
2. **다차원 평가:** 사실 기술 무결성, 경청 태도, SBAR 보고 완성도 3대 영역 채점.

## 제5장 사후검토(AAR) 및 교관 지침서
1. **5M RCA 매트릭스:** 절차, 양식, 환경, 숙달, 소통 관점의 시스템 개선 프로세스.
2. **교관 운영 지침:** 준비물 리스트, 분 단위 타임테이블, 채점표, 역할 탈출 디브리핑 가이드.
3. **개정 관리 규정:** 반기별 병영생활전문상담관/군의관 정기 감수 및 법령 개정 연동 프로토콜.
```
