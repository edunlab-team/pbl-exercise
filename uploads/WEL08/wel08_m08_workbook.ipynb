```python
# [환경 준비] 산출물 디렉토리 일괄 생성 및 가상 데이터셋 무결성 확인
import os
import json

directories = [
    'docs',
    'sop',
    'scenarios',
    'logs',
    'reports',
    'aar',
    'eval',
    'package'
]

print("=== [1] 산출물 디렉토리 생성 ===")
for d in directories:
    os.makedirs(d, exist_ok=True)
    print(f"  📁 폴더 준비 완료: {d}/")

print("\n=== [2] 가상 데이터셋 무결성 확인 ===")
data_dir = "/home/subokim/Workspace/milicontent2/uploads/WEL08"
if not os.path.exists(data_dir):
    data_dir = "."

dataset_count = 0
for i in range(1, 11):
    fname = f"wel08_m{i:02d}_dataset.json"
    fpath = os.path.join(data_dir, fname)
    if os.path.exists(fpath):
        with open(fpath, "r", encoding="utf-8") as f:
            d = json.load(f)
            dataset_count += 1
            print(f"  ✅ 데이터셋 로드: {fname} (미션: {d.get('mission')})")
    else:
        print(f"  ⚠️ 데이터셋 탐색 중: {fname}")

print(f"\n총 {dataset_count}개 데이터셋 확인 완료. 실습 준비가 완료되었습니다!")
```

### 2. 작성 지침 (HITL 원칙)

- AI가 제안한 내용 중 의학적 단정어('우울증', '불안장애' 등)가 있는지 검수하고 순수 관찰 사실(Fact)로 교정합니다.


```python
%%writefile aar/wel08_aar_session_minutes.md
# [WEL08_M08] 비난 없는 AAR 회의록 및 5대 관점 오류 분석

## 1. 사후검토(AAR) 개요 및 그라운드 룰
- **일시:** 2026-10-09 14:00~16:00 / **진행:** 퍼실리테이터(전문교관)
- **그라운드 룰:** 개인의 책임을 묻지 않고 "왜 절차가 작동하지 않았는가"에 집중하는 비처벌적 사실 분석.

## 2. 계획 대비 실제 타임라인 대조 (지연 지점 식별)
- **계획:** 신호 인지(10:15) ➔ 분대장 구두 보고(10:18, 3분 내) ➔ 상담관 연계(10:30)
- **실제:** 신호 인지(10:15) ➔ 기록지 작성 지연 ➔ 분대장 보고(10:35, 20분 소요) ➔ 상담관 연계(10:45)
- **원인 규명:** 신호 은폐가 아닌 '3페이지 서술형 기록지 작성 부담'으로 인한 행정 병목 확인.

## 3. 5대 관점 근본 원인 분석 매트릭스 (RCA)
| 관점 (5M) | 식별된 결함 및 원인 | 시스템적 개선 대책 |
| :--- | :--- | :--- |
| **절차 (Process)** | 보고 전 다단계 결재(분대장➔부소대장➔소대장)로 골든타임 누수 | 위험 징후 발견 시 지휘관 및 상담관 직통 동시 보고 바이패스 신설 |
| **양식 (Form)** | 면담 일지가 3페이지 서술형으로 작성에 15분 이상 소요 | 체크박스 중심 1-Page 포켓 카드로 서식 간소화 (작성 2분) |
| **환경 (Environment)** | 건의함이 행정반 앞 CCTV 직하에 위치하여 병사 이용 기피 | 생활관 복도 사각지대 및 무인 세탁실로 건의함 재배치 |
| **숙달 (Proficiency)** | 병사들이 비진단 관찰 사실 기술법 미숙으로 '단순 투정' 오판 | 분기 1회 1시간 마이크로 역할극 실습 의무화 |
| **소통 (Communication)** | 건의 후 2주간 피드백이 없어 '말해도 소용없다'는 냉소 형성 | 건의 접수 후 48시간 이내 조치 결과 공개 게시판 운영 |

## 4. 소통 창구 및 SOP 개선안 5대 확정 과제
1. 행정 서식 3종 1-Page 경량화 개정
2. 무기명 건의함 CCTV 사각지대 이전 설치
3. 48시간 양방향 피드백 게시판 도입
4. 위험 신호 직통 바이패스 핫라인 신설
5. 신임 분대장 온보딩 훈련 의무화
```
