```python
from pathlib import Path, PurePosixPath
import sys, io, json, re, zipfile, base64, hashlib, importlib
from IPython.display import display, HTML
import html
import pandas as pd
ROLE = 'solution'
EXPECTED_SHA256 = 'a0adfd7b9a2416cd74af2fc99bb7978ad662eec8d9e40ce5b8a539977c2cb17e'
c = None

def connect_md(path):
    global c, ROOT, HERE
    path = Path(path)
    if path.name not in ('DAT09.md', 'DAT09.md'):
        raise ValueError('원자료 DAT09.md 또는 DAT09.md를 선택하세요. 정답코드 MD는 연결 파일이 아닙니다.')
    text = path.read_text(encoding='utf-8-sig')
    found = re.search(r'<!-- DAT09_RUNTIME_BUNDLE_B64:([A-Za-z0-9+/=]+) -->', text)
    if not found:
        raise ValueError('이 노트북용 원자료 블록이 없습니다. 함께 받은 DAT09.md를 사용하세요.')
    raw = base64.b64decode(found.group(1), validate=True)
    if hashlib.sha256(raw).hexdigest() != EXPECTED_SHA256:
        raise ValueError('자료와 노트북 버전이 다릅니다. 같은 묶음의 파일을 사용하세요.')
    ROOT = Path.cwd() / ('DAT09_reference' if ROLE == 'solution' else 'DAT09_practice')
    ROOT.mkdir(exist_ok=True)
    with zipfile.ZipFile(io.BytesIO(raw)) as archive:
        for info in archive.infolist():
            name = PurePosixPath(info.filename)
            if name.is_absolute() or '..' in name.parts:
                raise ValueError('허용되지 않은 경로입니다.')
            target = ROOT.joinpath(*name.parts)
            target.parent.mkdir(parents=True, exist_ok=True)
            target.write_bytes(archive.read(info))
    sys.path.insert(0, str(ROOT))
    for name in list(sys.modules):
        if name in ['dat09_common', 'dat09_ui'] or name.startswith('src.m'):
            del sys.modules[name]
    import dat09_common
    globals().update({name: getattr(dat09_common, name) for name in dir(dat09_common) if not name.startswith('_')})
    import collections, itertools, queue, threading, tempfile, importlib.util, shutil
    globals().update({k:v for k,v in locals().items() if k in ['collections','itertools','re','queue','threading','tempfile','importlib','shutil']})
    from dat09_ui import bars, heatmap, line_chart, dashboard_fragment
    globals().update(bars=bars,heatmap=heatmap,line_chart=line_chart,dashboard_fragment=dashboard_fragment)
    c = dat09_common.context(ROOT)
    HERE = ROOT.parent / 'instructor'
    for candidate in [Path.cwd()/'instructor', Path.cwd(), ROOT.parent/'instructor']:
        if (candidate/'aar_key.json').is_file():
            HERE = candidate
            c['instructor_root'] = candidate
            break
    print('자료 연결 완료:', ROOT)
    print('10개 미션·50개 함수·27시간 과정입니다. 미션1부터 실행하세요.')
    print('TODO가 남아 있으면 해당 단계에서 멈춥니다. src에는 작성한 함수를 저장합니다.')
    return c

def hydrate_previous_missions(c, up_to=2):
    """같은 작업 폴더에 저장된 이전 미션(src/m01.py~)을 순차 실행해 작업 사전 c를 복원합니다."""
    import importlib.util
    src_dir = c['root'] / 'src'
    missing = [k for k in range(1, up_to + 1) if not (src_dir / f'm{k:02d}.py').is_file()]
    if missing:
        miss_str = ', '.join(f'미션{k}(DAT09_m{k:02d}_*.ipynb)' for k in missing)
        print(f'[안내] 아직 완료되지 않은 이전 미션이 있습니다: {miss_str}')
        print('이전 미션 노트북을 같은 폴더에서 먼저 실행한 뒤 이 셀을 다시 실행하세요.')
        return False
    for k in range(1, up_to + 1):
        if k in c['completed']:
            continue
        mod_path = src_dir / f'm{k:02d}.py'
        spec = importlib.util.spec_from_file_location(f'dat09_prev_m{k:02d}', mod_path)
        mod = importlib.util.module_from_spec(spec)
        spec.loader.exec_module(mod)
        fns = [getattr(mod, f'm{k:02d}_s{s}') for s in range(1, 6)]
        run_mission(c, k, fns)
    print(f'이전 미션(미션1~미션{up_to}) 상태 복원 완료.')
    return True

DATA_MD = next((Path(p) for p in ('DAT09.md', 'DAT09.md') if Path(p).is_file()), None)
if DATA_MD is not None:
    connect_md(DATA_MD)
    hydrate_previous_missions(c, 2)
else:
    print('DAT09.md(또는 DAT09.md)를 이 노트북과 같은 폴더에 두고 첫 셀을 다시 실행하세요.')
    print("다른 경로에 있다면 connect_md('/자료가_있는_경로/DAT09.md'); hydrate_previous_missions(c)를 실행하세요.")
```

```python
if c is None:
    raise RuntimeError("첫 셀에서 DAT09.md(또는 DAT09.md)를 먼저 연결하세요.")

# STEP 01 보고와 후보 및 승인값을 분리해 DB에 저장하기
def m03_s1(c):
    """보고와 후보 및 승인값을 분리해 DB에 저장하기."""
    import sqlite3
    out = folder(c, 3)
    db_path = out / 'integration.db'
    if db_path.exists():
        db_path.unlink()
    schema_sql = """-- DAT09 통합 저장소 스키마
CREATE TABLE IF NOT EXISTS raw_observation (
    row_id TEXT PRIMARY KEY,
    source TEXT NOT NULL,
    source_file TEXT,
    line_number INTEGER,
    raw_payload TEXT NOT NULL
);
CREATE TABLE IF NOT EXISTS normalized_observation (
    row_id TEXT PRIMARY KEY,
    source TEXT NOT NULL,
    item_id TEXT,
    site_id TEXT,
    metric TEXT,
    as_of TEXT,
    observed_at TEXT,
    received_at TEXT,
    version INTEGER,
    quantity_ea REAL,
    raw_qty TEXT,
    raw_unit TEXT,
    raw_as_of TEXT,
    lag_minutes REAL,
    status TEXT NOT NULL,
    reasons TEXT,
    stale_flag INTEGER NOT NULL DEFAULT 0,
    FOREIGN KEY (row_id) REFERENCES raw_observation(row_id)
);
CREATE UNIQUE INDEX IF NOT EXISTS uq_norm_obs_key
    ON normalized_observation(source, site_id, item_id, metric, as_of)
    WHERE item_id IS NOT NULL AND site_id IS NOT NULL AND as_of IS NOT NULL AND status = 'ACCEPTED';
CREATE TABLE IF NOT EXISTS validation_issue (
    issue_id INTEGER PRIMARY KEY AUTOINCREMENT,
    row_id TEXT NOT NULL,
    rule_id TEXT NOT NULL,
    action TEXT NOT NULL,
    message TEXT,
    UNIQUE(row_id, rule_id)
);
CREATE TABLE IF NOT EXISTS candidate (
    candidate_id TEXT PRIMARY KEY,
    site_id TEXT NOT NULL,
    item_id TEXT NOT NULL,
    metric TEXT NOT NULL,
    as_of TEXT NOT NULL,
    candidate_ea INTEGER,
    status TEXT NOT NULL,
    proposer_id TEXT,
    reason TEXT,
    source_versions TEXT,
    candidate_version INTEGER NOT NULL DEFAULT 1,
    reviewed_by TEXT
);
CREATE TABLE IF NOT EXISTS approved_master (
    site_id TEXT NOT NULL,
    item_id TEXT NOT NULL,
    metric TEXT NOT NULL,
    as_of TEXT NOT NULL,
    approved_ea INTEGER NOT NULL,
    candidate_id TEXT NOT NULL,
    steward_id TEXT NOT NULL,
    reason TEXT NOT NULL,
    approved_at TEXT NOT NULL,
    external_write TEXT NOT NULL DEFAULT 'NOT_CONNECTED',
    PRIMARY KEY (site_id, item_id, metric, as_of)
);
"""
    (out / 'schema.sql').write_text(schema_sql, encoding='utf-8')
    conn = sqlite3.connect(db_path)
    conn.row_factory = sqlite3.Row
    conn.executescript(schema_sql)

    def load_into_db():
        for r in c['raw_rows']:
            conn.execute(
                'INSERT OR IGNORE INTO raw_observation(row_id, source, source_file, line_number, raw_payload) VALUES (?,?,?,?,?)',
                (r['row_id'], r['source'], r.get('source_file'), r.get('line_number'), r.get('raw_payload', '')),
            )
        for nr in c['normalized']:
            conn.execute(
                """INSERT INTO normalized_observation(
                    row_id, source, item_id, site_id, metric, as_of, observed_at, received_at,
                    version, quantity_ea, raw_qty, raw_unit, raw_as_of, lag_minutes, status, reasons, stale_flag
                ) VALUES (?,?,?,?,?,?,?,?,?,?,?,?,?,?,?,?,?)
                ON CONFLICT(row_id) DO UPDATE SET
                    status=excluded.status, reasons=excluded.reasons, stale_flag=excluded.stale_flag""",
                (
                    nr['row_id'], nr['source'], nr.get('item_id'), nr.get('site_id'), nr.get('metric'),
                    nr.get('as_of'), nr.get('observed_at'), nr.get('received_at'), nr.get('version'),
                    nr.get('quantity_ea'), str(nr.get('raw_qty', '')), nr.get('raw_unit'), nr.get('raw_as_of'),
                    nr.get('lag_minutes'), nr['status'], nr.get('reasons'), int(bool(nr.get('stale_flag'))),
                ),
            )
        conn.commit()

    load_into_db()
    cnt1_raw = conn.execute('SELECT COUNT(*) FROM raw_observation').fetchone()[0]
    cnt1_norm = conn.execute('SELECT COUNT(*) FROM normalized_observation').fetchone()[0]
    load_into_db()
    cnt2_raw = conn.execute('SELECT COUNT(*) FROM raw_observation').fetchone()[0]
    cnt2_norm = conn.execute('SELECT COUNT(*) FROM normalized_observation').fetchone()[0]

    c['conn'] = conn
    c['db'] = db_path
    return [
        {'load_pass': 1, 'raw_observation_count': cnt1_raw, 'normalized_observation_count': cnt1_norm},
        {'load_pass': 2, 'raw_observation_count': cnt2_raw, 'normalized_observation_count': cnt2_norm},
    ]

# STEP 02 열다섯 검증 규칙과 시험 사례 구현하기
def m03_s2(c):
    """열다섯 검증 규칙과 시험 사례 구현하기."""
    out = folder(c, 3)
    rules = read_json(c['root'] / 'config/rules.json')
    rule_meta = {r['rule_id']: r for r in rules}
    valid_items = {x['item_id'] for x in c['catalog']['items']}
    valid_sites = {x['site_id'] for x in c['catalog']['sites']}

    def rule_fn(row, duplicate_key=None):
        if duplicate_key is None:
            duplicate_key = bool(row.get('duplicate_key', False))
        src = row.get('source')
        rid = row.get('row_id')
        item_id = row.get('item_id')
        site_id = row.get('site_id')
        metric = row.get('metric')
        qty = row.get('quantity') if 'quantity' in row else row.get('raw_qty')
        unit = row.get('unit') if 'unit' in row else row.get('raw_unit')
        box_to_ea = row.get('box_to_ea', 10)
        as_of_raw = row.get('raw_as_of') if 'raw_as_of' in row else row.get('as_of')
        obs_raw = row.get('observed_at')
        rec_raw = row.get('received_at')
        ver = row.get('version')

        is_missing = (qty is None) or (isinstance(qty, str) and qty.strip() in ('', 'NULL', 'N/A'))
        q_num = None
        is_num = False
        if not is_missing:
            try:
                q_num = float(qty)
                is_num = math.isfinite(q_num)
            except Exception:
                is_num = False

        unit_ok = (unit == 'EA') or (unit == 'BOX' and box_to_ea is not None and float(box_to_ea) > 0)
        integral_ok = True
        if is_num and unit_ok:
            factor = 1 if unit == 'EA' else float(box_to_ea)
            integral_ok = float(q_num * factor).is_integer()

        t_as = t_obs = t_rec = None
        r10_ok = True
        try:
            t_as = parse_time(as_of_raw, src if not str(as_of_raw).endswith('+00:00') else 'A')
        except Exception:
            try:
                t_as = parse_time(row.get('as_of'), 'A')
            except Exception:
                r10_ok = False
        r11_ok = True
        for val, holder in [(obs_raw, 'obs'), (rec_raw, 'rec')]:
            try:
                parsed = parse_time(val, src if (src in ('B', 'C') and '-' not in str(val)) else 'A')
                if holder == 'obs':
                    t_obs = parsed
                else:
                    t_rec = parsed
            except Exception:
                r11_ok = False
        if row.get('date_error'):
            r10_ok = False

        r12_ok = isinstance(ver, int) and not isinstance(ver, bool) and ver > 0
        if not r12_ok and isinstance(ver, str) and ver.isdigit() and int(ver) > 0:
            r12_ok = True

        r14_ok = True
        r15_ok = True
        if t_as is not None and t_obs is not None:
            diff_min = (t_as - t_obs).total_seconds() / 60.0
            r14_ok = diff_min >= 0
            r15_ok = (0 <= diff_min <= 60)

        res = {
            'R01': bool(src) and bool(rid),
            'R02': item_id in valid_items,
            'R03': site_id in valid_sites,
            'R04': metric == 'on_hand',
            'R05': not is_missing,
            'R06': True if is_missing else is_num,
            'R07': True if (is_missing or not is_num) else (q_num >= 0),
            'R08': True if (is_missing or not is_num or not unit_ok) else integral_ok,
            'R09': unit_ok,
            'R10': r10_ok,
            'R11': r11_ok,
            'R12': r12_ok,
            'R13': not duplicate_key,
            'R14': r14_ok,
            'R15': r15_ok,
        }
        return res

    cases = read_json(c['root'] / 'data/rule_cases.json')
    rule_tests = []
    for case in cases:
        t_rule = case['target_rule']
        out_map = rule_fn(case['input'])
        actual = out_map[t_rule]
        rule_tests.append({
            'case_id': case['case_id'],
            'target_rule': t_rule,
            'expected_target_pass': case['expected_target_pass'],
            'actual_target_pass': actual,
            'passed': actual == case['expected_target_pass'],
        })
    assert len(rule_tests) == 45 and all(x['passed'] for x in rule_tests)

    issues = []
    timings = []
    conn = c['conn']
    conn.execute('DELETE FROM validation_issue')
    for nr in c['normalized']:
        t0 = time.perf_counter()
        res = rule_fn(nr, duplicate_key=False)
        timings.append((time.perf_counter() - t0) * 1000.0)
        for rid_rule, ok in res.items():
            if not ok:
                meta = rule_meta[rid_rule]
                issues.append({'row_id': nr['row_id'], 'rule_id': rid_rule, 'action': meta['action'], 'message': meta['name']})
                conn.execute('INSERT OR IGNORE INTO validation_issue(row_id, rule_id, action, message) VALUES (?,?,?,?)',
                             (nr['row_id'], rid_rule, meta['action'], meta['name']))
    conn.commit()

    csv_out(out / 'validation_issues.csv', issues)
    write_json(out / 'rule_tests.json', {'total': len(rule_tests), 'passed': sum(x['passed'] for x in rule_tests), 'cases': rule_tests})
    c['rule_fn'] = rule_fn
    c['issues'] = issues
    c['rule_tests'] = rule_tests
    c['rule_timings'] = timings
    return rule_tests

# STEP 03 비교 가능한 관측값과 승인된 재고를 따로 조회하기
def m03_s3(c):
    """비교 가능한 관측값과 승인된 재고를 따로 조회하기."""
    out = folder(c, 3)
    views_sql = """-- 같은 재고 이력(3출처 일치·정상·적시) 중복 합산 방지 조회 뷰
DROP VIEW IF EXISTS consistent_observation;
CREATE VIEW consistent_observation AS
SELECT
    site_id,
    item_id,
    metric,
    as_of,
    CAST(MIN(quantity_ea) AS INTEGER) AS qty_ea,
    COUNT(*) AS source_count,
    GROUP_CONCAT(row_id, ',') AS row_ids
FROM normalized_observation
WHERE status = 'ACCEPTED' AND stale_flag = 0
GROUP BY site_id, item_id, metric, as_of
HAVING COUNT(DISTINCT source) = 3 AND MIN(quantity_ea) = MAX(quantity_ea);

-- 승인된 재고는 approved_master만 조회
DROP VIEW IF EXISTS approved_stock;
CREATE VIEW approved_stock AS
SELECT site_id, item_id, metric, as_of, approved_ea, candidate_id, steward_id, approved_at, external_write
FROM approved_master;
"""
    (out / 'views.sql').write_text(views_sql, encoding='utf-8')
    c['conn'].executescript(views_sql)

    item_target = {x['item_id']: x['target_ea'] for x in c['catalog']['items']}
    slots = csv_rows(c['root'] / 'data/expected_slots.csv')
    norm_by_id = {r['row_id']: r for r in c['normalized']}

    slot_groups = {}
    for s in slots:
        lk = (s['expected_site_id'], s['expected_item_id'], s['metric'], s['expected_as_of'])
        slot_groups.setdefault(lk, []).append(norm_by_id[s['row_id']])

    groups = {}
    logical_rows = []
    for lk in sorted(slot_groups):
        site_id, item_id, metric, as_of = lk
        rows = slot_groups[lk]
        row_ids = [r['row_id'] for r in rows]
        obs_times = [str(r.get('observed_at') or '') for r in rows]
        timely_acc = [r for r in rows if r['status'] == 'ACCEPTED' and not r['stale_flag']]
        if len(timely_acc) == 3:
            vals = {int(r['quantity_ea']) for r in timely_acc}
            if len(vals) == 1:
                review_status = 'CONSISTENT'
                qty = next(iter(vals))
                reason = '3출처 정상·적시·수량 일치'
            else:
                review_status = 'CONFLICT'
                qty = None
                reason = f"출처 간 수량 불일치({[int(r['quantity_ea']) for r in timely_acc]})"
        else:
            review_status = 'UNCOMPARABLE'
            qty = None
            bad = [f"{r['row_id']}:{r['reasons']}" for r in rows if r['status'] != 'ACCEPTED' or r['stale_flag']]
            reason = '; '.join(bad)
        groups[lk] = {'rows': rows, 'review_status': review_status, 'qty': qty, 'reason': reason}
        logical_rows.append({
            'site_id': site_id, 'item_id': item_id, 'metric': metric, 'as_of': as_of,
            'qty': qty, 'target': item_target[item_id], 'review_status': review_status,
            'row_ids': row_ids, 'observed_at': obs_times, 'reason': reason,
        })

    csv_out(out / 'logical_stock.csv', [
        {**r, 'row_ids': ','.join(r['row_ids']), 'observed_at': ','.join(r['observed_at'])}
        for r in logical_rows
    ])
    c['groups'] = groups
    c['logical_rows'] = logical_rows
    return logical_rows

# STEP 04 잠정 부족량과 일치율을 분모에 맞게 계산하기
def m03_s4(c):
    """잠정 부족량과 일치율을 분모에 맞게 계산하기."""
    out = folder(c, 3)
    logical_rows = c['logical_rows']
    ref_as_of = sorted({r['as_of'] for r in logical_rows})[0]
    metrics_rows = []
    for r in logical_rows:
        if r['as_of'] != ref_as_of:
            continue
        q = r['qty']
        shortage = max(r['target'] - q, 0) if q is not None else None
        metrics_rows.append({
            'site_id': r['site_id'], 'item_id': r['item_id'], 'as_of': ref_as_of,
            'review_status': r['review_status'],
            'provisional_qty_ea': q,
            'target_ea': r['target'],
            'provisional_shortage_ea': shortage,
            'approved_qty_ea': None,
            'approved_shortage_ea': None,
            'basis': '잠정 관측값(미승인)',
        })

    consistent_cnt = sum(1 for r in logical_rows if r['review_status'] == 'CONSISTENT')
    comparable_cnt = sum(1 for r in logical_rows if r['review_status'] in ('CONSISTENT', 'CONFLICT'))
    total_cnt = len(logical_rows)
    consistency = {
        'reference_as_of': ref_as_of,
        'consistent_count': consistent_cnt,
        'comparable_count': comparable_cnt,
        'total_logical_keys': total_cnt,
        'overall_match_rate': round(consistent_cnt / total_cnt, 6),
        'overall_fraction': f'{consistent_cnt}/{total_cnt}',
        'comparable_match_rate': round(consistent_cnt / comparable_cnt, 6),
        'comparable_fraction': f'{consistent_cnt}/{comparable_cnt}',
    }
    csv_out(out / 'baseline_metrics.csv', metrics_rows)
    write_json(out / 'consistency_metrics.json', consistency)
    c['baseline_metrics'] = metrics_rows
    return metrics_rows

# STEP 05 손으로 구한 값과 SQL을 대조하고 검사시간 측정하기
def m03_s5(c):
    """손으로 구한 값과 SQL을 대조하고 검사시간 측정하기."""
    out = folder(c, 3)
    conn = c['conn']
    by_lk = {(r['site_id'], r['item_id'], r['as_of']): r for r in c['logical_rows']}
    # 정상·실제0(예시)·초과·정보없음(결측)·목표일치 포함 5개 대표 사례 선정
    picks = []
    for r in c['logical_rows']:
        if r['review_status'] == 'CONSISTENT' and r['qty'] < r['target'] and len([x for x in picks if x[0] == '부족']) == 0:
            picks.append(('부족', r))
        elif r['review_status'] == 'CONSISTENT' and r['qty'] >= r['target'] and len([x for x in picks if x[0] == '초과']) == 0:
            picks.append(('초과', r))
        elif r['review_status'] == 'UNCOMPARABLE' and len([x for x in picks if x[0] == '정보없음']) == 0:
            picks.append(('정보없음', r))
        elif r['review_status'] == 'CONFLICT' and len([x for x in picks if x[0] == '충돌(확인불가)']) == 0:
            picks.append(('충돌(확인불가)', r))
        if len(picks) == 4:
            break
    # 목표 일치/경계 0 사례 추가(일치 사례 또는 정상 사례)
    for r in c['logical_rows']:
        if r['review_status'] == 'CONSISTENT' and (r['site_id'], r['item_id'], r['as_of']) not in {(p[1]['site_id'], p[1]['item_id'], p[1]['as_of']) for p in picks}:
            picks.append(('목표대조_정상', r))
            break

    norm_map = {r['row_id']: r for r in c['normalized']}
    reconcile = []
    for label, r in picks:
        sql_row = conn.execute(
            'SELECT qty_ea FROM consistent_observation WHERE site_id=? AND item_id=? AND as_of=?',
            (r['site_id'], r['item_id'], r['as_of']),
        ).fetchone()
        sql_qty = sql_row[0] if sql_row else None
        peers = [norm_map[rid] for rid in r['row_ids']]
        timely_acc = [p for p in peers if p['status'] == 'ACCEPTED' and not p['stale_flag']]
        hand_qty = int(timely_acc[0]['quantity_ea']) if (len(timely_acc) == 3 and len({int(p['quantity_ea']) for p in timely_acc}) == 1) else None
        hand_short = max(r['target'] - hand_qty, 0) if hand_qty is not None else None
        sql_short = max(r['target'] - sql_qty, 0) if sql_qty is not None else None
        reconcile.append({
            'case_type': label,
            'site_id': r['site_id'], 'item_id': r['item_id'], 'as_of': r['as_of'],
            'target_ea': r['target'],
            'hand_qty_ea': hand_qty, 'sql_qty_ea': sql_qty,
            'hand_shortage_ea': hand_short, 'sql_shortage_ea': sql_short,
            'diff_ea': (0 if (hand_qty is None and sql_qty is None) else abs((hand_qty or 0) - (sql_qty or 0))),
            'na_both_match': (hand_qty is None and sql_qty is None),
        })

    timings = c['rule_timings']
    timing_doc = {
        'sample_count': len(timings),
        'scope': '15개 검증 규칙(R01~R15) 전체 행 단위 실행 시간(ms)',
        'median_ms': round(statistics.median(timings), 4),
        'p95_ms': round(quantile(timings, 0.95), 4),
        'max_ms': round(max(timings), 4),
        'target_p95_ms': 50.0,
        'target_met': quantile(timings, 0.95) <= 50.0,
    }
    write_json(out / 'rule_timing.json', timing_doc)
    (out / 'tests.log').write_text(
        f"45 rule tests passed: {sum(x['passed'] for x in c['rule_tests'])}/{len(c['rule_tests'])}\n"
        f"Hand vs SQL reconcile cases: {len(reconcile)}, max diff: {max(x['diff_ea'] for x in reconcile)}\n"
        f"15-rule timing p95: {timing_doc['p95_ms']} ms (median {timing_doc['median_ms']} ms)\n",
        encoding='utf-8',
    )
    (out / 'review_notes.md').write_text(
        '# M03 검토 기록\n- integration.db 2회 적재 후 raw_observation=2,160, normalized_observation=2,160 유지 확인.\n'
        '- 45개 규칙 시험 전부 일치, 전체 일치율 655/720, 비교 가능 일치율 655/675 검산 완료.\n',
        encoding='utf-8',
    )
    return reconcile

result_html = run_mission(c, 3, [m03_s1, m03_s2, m03_s3, m03_s4, m03_s5])
document = result_html.read_text(encoding="utf-8")
display(HTML('<iframe title="DAT09 단계별 결과" style="width:100%;height:1100px;border:1px solid #ccd5dd" sandbox="allow-scripts allow-downloads allow-same-origin" srcdoc="'+html.escape(document,quote=True)+'"></iframe>'))
```
