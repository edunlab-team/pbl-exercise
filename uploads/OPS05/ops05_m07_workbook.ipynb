#### ✅ 모범 작성 예시 (해설 답안)
- **루브릭 행동 기준 예시(상황 인지·출처 검증):** `탁월(4점)`: 경보 후 2분 내 중복 보고 병합 및 단일 출처 1·2등급 차단 규칙 100% 준수 | `미흡(1점)`: 단일 보고만으로 적 드론 단정 또는 인지·보고 시각 누락

AI 협업·HITL 기준: 보안 원칙 준수(가상 자료만 입력) 및 AI 출력 교차 검증·수정 근거 제시

채점자 판정 메모
타임라인 12건 수행 기록의 완결성과 루브릭 5개 영역의 관찰 가능한 행동 기준 기술 여부, 평가자 2인 이상 독립 채점 일치도 검증 결과를 확인한다.

```python
# [미션 7 해설 코드] 평가자 일치도·타임라인 처리시간 계산 및 루브릭 저장
data = load_mission_dataset(7)
rater_a = [3, 4, 2, 3, 3]
rater_b = [3, 3, 2, 3, 4]
exact = sum(a == b for a, b in zip(rater_a, rater_b)) / len(rater_a)
within_1 = sum(abs(a - b) <= 1 for a, b in zip(rater_a, rater_b)) / len(rater_a)
handling_mins = [3.5, 4.0, 5.5, 6.2, 4.8, 7.0, 3.9, 5.1, 6.5, 4.4, 5.8, 4.2]
avg_m = sum(handling_mins) / len(handling_mins)
print(f'완전 일치율: {exact*100:.1f}%, 1점 이내 일치율: {within_1*100:.1f}%')
print(f'12건 평균 처리시간: {avg_m:.2f}분 (최대 {max(handling_mins)}분)')
rubric_md = f'# OPS05 스트레스 훈련 평가 루브릭\n\n- 완전 일치율: {exact*100:.1f}% / 1점 이내: {within_1*100:.1f}%\n- 12건 평균 처리시간: {avg_m:.2f}분\n'
save_deliverable('rubric/ops05_evaluation_rubric.md', rubric_md)
```
