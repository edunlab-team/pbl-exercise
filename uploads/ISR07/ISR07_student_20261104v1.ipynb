## 무엇을 만들까요?
가상 관측 기록을 읽고 관측한 내용·공백·판단 한계를 구분하는 표와 브리핑 문서를 만듭니다. 최종 결과는 8개 장과 부록을 갖춘 교육용 정보판단 가이드라인입니다. 실제 센서 운용·군 규정·작전 판단 교육이 아닙니다. 생성형 AI가 처음인 비전공자도 제공 자료와 질문 예시로 시작합니다.

## 파일을 열고 시작하기
CMS 프로젝트 공통 참고자료에서 최신 Student Notebook 또는 오프라인 HTML 하나를 선택합니다. Notebook은 준비 셀부터 실행하면 내장 실습자료를 자동으로 준비합니다. 긴 준비 코드는 접혀 있으므로 수정하지 않습니다. HTML은 내려받아 브라우저로 열면 됩니다. 두 도구는 같은 자료·빈 양식을 사용합니다. 자료를 읽고 미션 번호를 선택한 뒤 결과 작성란을 편집하세요. 미션1~3은 사례·기준·표현, 미션4~6은 지도·서식·분류, 미션7~9는 브리핑·사후 검토·전후 비교, 미션10은 통합입니다.

## AI에게 질문하고 답변 고치기
먼저 본인의 생각과 모르는 점을 적습니다. 허용된 AI의 별도 대화에 해당 미션의 가상 자료·질문·원하는 형식·금지할 추정을 입력합니다. 활동별 질문 예시를 참고하세요. 핵심 주장 3건 이상을 자료 번호·문장·직접 계산과 대조하고 채택·수정·보류 이유를 적습니다. 보완 답변도 다시 확인합니다. 실제 요청과 실제 답변 전체를 대화 목적과 한 묶음으로 저장하세요. 이 도구는 AI를 내장하지 않습니다.

## 어디에 쓰고 저장하나요?
미션별 결과 작성란은 기존 빈 작성 양식입니다. 표와 활동 기록을 채운 뒤 ‘작성 내용 저장’을 누릅니다. 결과는 미션의 정해진 문서 경로에 저장합니다. 같은 문서의 forms 경로와 루트 양식 이름은 제출 호환 복사본으로 도구가 함께 만들므로 세 번 작성하지 않습니다. Notebook은 실제 파일, HTML은 브라우저 임시 보관을 사용합니다. 브라우저·기기 변경이나 Colab 종료 전 작성 JSON과 검토용 ZIP을 내려받으세요. JSON 가져오기로 이어 쓸 수 있습니다. 복원 전 기록은 보존하며 잘못된 형식은 적용하지 않습니다. 새로 준비해도 작성본을 덮어쓰지 않습니다.

미완성 작업은 검토용 ZIP으로 보관합니다. 최종 제출용은 필수 미션1·2·3·5·7·9·10의 문서·대화와 자기점검 상태를 확인한 뒤 생성합니다. 선택 미션4·6·8을 하지 않았다면 그 사유와 후속 문서의 빈 부분을 명시하세요. 선택을 건너뛰었다고 필요한 후속 산출물이 자동 완성되지는 않습니다. 최종 제출용 생성은 파일·기록의 형식 확인이며 실제 학습 PASS나 교관 승인이 아닙니다. 평가자·발표·측정 등 실제 수행 조건은 교관이 별도로 확인합니다.

## 폐쇄망에서는 어떻게 하나요?
내려받은 HTML은 외부 라이브러리·글꼴·서버 없이 자료 읽기·작성·저장·JSON 백업과 복원·ZIP 생성을 제공합니다. 허용된 로컬 Jupyter에서는 Python·ipywidgets가 필요합니다. Colab과 외부 AI는 인터넷 및 기관 허용이 필요하며 폐쇄망에서 된다고 가정하지 않습니다. AI를 사용할 수 없으면 미실시와 필요한 환경을 기록하고 AI 활용을 완료로 표시하지 않습니다.

## 용어·시간·운영 조건
materials는 읽을 자료, forms는 작성 양식, docs·framework·analysis 등은 결과 파일을 묶는 폴더 이름입니다. 경로는 결과의 저장 위치이며 외부 인터넷 주소가 아닙니다. 원본 예상 시간 합계는 27시간, CMS 편성은 29시간입니다. 실제 소요시간은 미측정입니다. 원본 필수7개·선택3개와 각 평가 질문의 REQUIRED 설정은 서로 다른 층위이며 기존 설정을 유지합니다. 전체10개 수행은 통합 연습 경로입니다. 파일 작성·실행·평가 목표·사람 승인은 각각 다릅니다.

미션2 조원 일치와 미션7 평가자2명 요건은 AI 모의 점수로 대체되지 않습니다. 미션7은 현재 공개된 카드만 AI에 전달합니다. 미션9는 기존·수정 서식과 일곱 사례가 필요합니다. 선택 미션8 미실시 시 기존 A·B 자료로 비교를 준비하되 AAR 개선 결과는 미실시로 표시합니다. 미션10의 선택 결과 부록이 비면 누락을 남기고 교관에게 범위를 확인하세요.


```python
#@title ① 실습자료 준비 (여기를 실행하세요) { display-mode: "form" }
from pathlib import Path
import zipfile,io,base64,hashlib,importlib.util,time
EMBEDDED_PROJECT=''
def prepare_project():
    data=base64.b64decode(EMBEDDED_PROJECT)
    if hashlib.sha256(data).hexdigest()!='b1083527aac1702eafe7833a297b73a2d2d5035a4c5869fae658b7d50c4e3089':raise ValueError('내장 자료 무결성 오류; 기존 작성본은 유지합니다')
    dest=Path('/content')if Path('/content').exists()else Path.cwd();root=dest/'ISR07_project'
    with zipfile.ZipFile(io.BytesIO(data))as z:
        for name in z.namelist():
            target=(dest/name).resolve()
            if not target.is_relative_to(dest.resolve())or Path(name).parts[0]!='ISR07_project':raise ValueError('자료 경로 오류')
        for name in z.namelist():
            if name.endswith('/'):continue
            target=dest/name
            if target.exists()and (name.startswith('ISR07_project/forms/')or len(Path(name).parts)==2):continue
            target.parent.mkdir(parents=True,exist_ok=True);target.write_bytes(z.read(name))
    return root
ROOT=prepare_project()
spec=importlib.util.spec_from_file_location('isr07_tools',ROOT/'tools/workbook_tools.py');tools=importlib.util.module_from_spec(spec);spec.loader.exec_module(tools)
for name in ['save_result','save_ai_dialogue','mark_actual_time','missing_percent','agreement_percent','create_submission','show_mission','show_submission','show_restore','show_backup','export_workbook_json','import_workbook_json','progress_summary']:globals()[name]=getattr(tools,name)
print('자료 준비 완료:',ROOT,'/ 기존 작성 문서·대화 보존. 아래 미션 셀을 실행하세요.')
```
