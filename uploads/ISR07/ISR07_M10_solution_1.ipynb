## 8개 장 통합 참고와 심의

1 목적·범위·사람 판단·교전수칙/지휘계통; 2 관측/불가/결측/독립성/6단계와3단계 용어; 3 L1～L5와 센서4종 한계; 4 T1～T5·지표12개·서술범위·교차검토; 5 W1～W4·요인8개·25칸3조건·기준12개; 6 7항목 서식·4분기·B1～B5·긴급 한 줄·사람 승인; 7 기간별 결측률·3단계·보고3종·금지표현6개; 8 20칸루브릭·F카드·AAR2건·재검증7건·연1회개정. 부록은 6단계표·커버리지·3단계표·서식·루브릭·실행/개정기록이다.

교리 모의 심의 예: ‘신뢰구간’이 통계 용어로 혼동됨→서술 등급 범위임을 정의. 법무 모의 심의 예: 분기3에서 사람 승인 우회 가능→모든 경로 분기4로 연결. 작전 모의 심의 예: 한계 목록으로 핵심이 묻힘→BLUF 한 줄·우선순위3개·상세 부록 유지. 실제 법적 판단은 하지 않으며 쟁점은 검토 연계로만 표시한다.

개정은 연1회와 새로운 자료/의견이 있을 때 검토, 책임은 가상 정보 책임 직책, 의견은 교육 후 서식·평가 기록·미해결 사항으로 수렴한다. 원본 필수7개 모두 실제 요건이 충족되어야 최종 PASS다. AI 모의 평가만 있고 실제 평가자 요건이 미실시라면 제작/개인 연습 완료와 원본 최종 PASS를 구분한다.

## AI 수행 증거의 평가

Why: 본인의 질문·위임 이유·자료 선택·최초 판단. How: 실제 AI 요청·첫 답변·자료 대조 3건 이상·보완·다시 확인. What: 최종 채택/수정/보류와 이유·반영 위치. 옳은 답을 유지한 경우도 근거가 있어야 한다. AI 결과를 복사한 것만으로 수행 경험을 인정하지 않는다.

원본 기준별 결과와 분모·기간을 확인하고 미실시·목표 미달을 그대로 기록한다. AI 모의 심의·채점은 실제 평가관 승인과 구분한다. 참고문장·산식을 유일한 답으로 고정하지 않되 자료 없는 완료·부재·독립성 확정을 받아들이지 않는다.
