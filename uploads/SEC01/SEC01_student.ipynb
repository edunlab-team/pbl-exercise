# 🛡️ 기지 CCTV 드론·비인가 접근 객체 탐지 프로토타입

**어서오세요, 전우님!** 이번 프로젝트에서 전우님은 가상의 **시설 방호 영상분석반**에 바이브 코딩 담당으로 합류합니다.

업무 요청을 프롬프트로 바꾸고, AI가 만든 코드와 문서를 실행해 확인하고, 고쳐 달라고 다시 요청하는 **바이브 코딩** 과정을 직접 해 봐요.

```python
#@title 🚀 [먼저 실행] 실습 준비하기 { display-mode: "form" }
#@markdown ▶ 왼쪽의 실행 버튼을 누르거나, 메뉴에서 **런타임 → 모두 실행**을 누르세요. 아래 진행 막대가 **100%가** 되면 준비가 끝나요.
FILL_ANSWERS = False   # 학습자용: 빈 상태로 시작해요
from IPython.display import HTML, display

_SETUP_STEPS = ["메신저 · 진행 상태 · 가이드북", "미션 활동 · 가상 AI", "첫 화면"]


def _setup_html(pct, done, msg):
    color = "#16a34a" if pct >= 100 else "#0b57d0"
    items = "".join(f"<li style='margin:3px 0;color:#111827'>{'✅' if i < done else '⏳' if i == done else '▫️'} {name}</li>"
                    for i, name in enumerate(_SETUP_STEPS))
    return (f"<div style='font-family:sans-serif;max-width:640px;border:2px solid {color};border-radius:12px;padding:12px 16px;"
            f"background:#fff;color:#111827'><div style='font-weight:bold;font-size:16px;color:#111827'>🚀 실습 준비 {pct}%</div>"
            f"<div style='height:16px;background:#e3e8ef;border-radius:8px;overflow:hidden;margin:8px 0'>"
            f"<div style='height:100%;width:{pct}%;background:{color}'></div></div>"
            f"<div style='font-size:14px;color:#111827'>{msg}</div>"
            f"<ul style='list-style:none;padding:0;margin:8px 0 0;font-size:13.5px'>{items}</ul></div>")


_setup_bar = display(HTML(_setup_html(2, 0, "준비를 시작합니다...")), display_id=True)


def _setup(pct, done, msg):
    _setup_bar.update(HTML(_setup_html(pct, done, msg)))

# ===== 프로젝트 공통: 메신저 · 진행 상태 · 산출물 · 가이드북 =====
# 로그인이 없음: 개인 정보를 받지 않고 학습자는 모두 '전우님'으로 부름 (2026-11-05 피드백)
# 미션은 처음부터 차례로 하는 것이 원칙. 뒷부분을 미리 보려는 사람을 위해 잠그지 않고 열어 둠 (2026-11-05 피드백)
import base64, html, inspect, io, json, math, os, random, re, textwrap, threading, time, traceback
from datetime import date, datetime, timedelta, timezone
import numpy as np
from IPython.display import JSON
from google.colab import output

# FILL_ANSWERS(선생님용)는 이 셀 맨 위의 체크박스로 정함: True면 모든 활동의 값을 정답으로 채워 둔 상태로 시작함
BOOT_ID = datetime.now().strftime("%H%M%S%f")   # 이 셀을 실행할 때마다 바뀜: 화면이 예전 실행에 머물러 있으면 알아채고 처음 화면으로 돌림

SPEC = {'project': {'id': 'pbl_e0333acf',
             'emoji': '🛡️',
             'title': '기지 CCTV 드론·비인가 접근 객체 탐지 프로토타입',
             'goal': '업무를 AI에게 말로 설명해 도구와 문서를 만드는 바이브 코딩을 익혀요. CCTV 영상분석 경보를 다루는 일을 소재로, 프롬프트를 쓰고, AI가 만든 코드를 실행해 확인하고, 고쳐 달라고 다시 요청해요.',
             'welcome_lead': '가상의 시설 방호 영상분석반에 바이브 코딩 담당으로 합류했어요. 팀원들의 요청을 프롬프트로 바꿔 AI와 함께 도구와 문서를 만들어 봐요. M1부터 차례로 진행해 보세요.',
             'team_name': '시설 방호 영상분석반',
             'my_role': '바이브 코딩 담당',
             'use_label': '바이브 코딩에서는',
             'intro': '팀원들의 요청을 메신저로 받아, 가상 AI에게 프롬프트를 보내 CCTV 경보 기록을 분석하는 코드와 문서를 만들어요. 영상분석은 소재일 뿐이고, 이번 프로젝트의 중심은 프롬프트를 쓰고 결과를 확인하는 방법이에요.',
             'banned_phrases': ['울타리 경계구역', '경계구역', '랜덤포레스트', 'Baseline', '해시 체인', '센서 두절', '센서 단절', '합성 시퀀스', '감시 공백 경보'],
             'notes': ['데이터, 팀원, 장소, 경보 기록은 모두 교육용 가상 자료예요. 실제 조직·위치·장비 정보는 쓰지 않아요.',
                       '이번 과정의 AI는 전우님이 고른 문장에 맞춰 준비된 응답을 돌려주는 가상 AI예요. 다만 AI가 만든 코드는 실습 환경에서 실제로 실행돼요. 같은 방법을 실제 AI 도구에 그대로 쓸 수 있어요.',
                       '영상 내려받기나 모델 학습은 하지 않아요. 영상을 분석한 결과에 해당하는 가상 기록을 미리 만들어 두었어요.'],
             'members': {'lim': ['임반장', '시설 방호 영상분석반', '영상분석반장', '#c2410c'],
                         'kang': ['강계획', '방호 계획 부서', '방호 계획 담당', '#b91c1c'],
                         'seo': ['서상황', '통합 관제실', '관제 담당', '#1d4ed8'],
                         'oh': ['오영상', '통합 관제실', '영상 모니터링 담당', '#0f766e'],
                         'min': ['민시설', '시설 관리 부서', '시설 관리 담당', '#15803d'],
                         'yun': ['윤분석', '시설 방호 영상분석반', '데이터 분석 담당', '#7e22ce'],
                         'jo': ['조당직', '통합 관제실', '당직 책임자', '#334155'],
                         'me': ['전우님', '시설 방호 영상분석반', '바이브 코딩 담당', '#334155']},
             'channels': {'notice': '# 프로젝트-공지',
                          'team': '# 영상분석반',
                          'dm_lim': '✉ 임반장',
                          'dm_kang': '✉ 강계획 (방호 계획 담당)',
                          'dm_seo': '✉ 서상황 (관제 담당)',
                          'dm_oh': '✉ 오영상 (모니터링 담당)',
                          'dm_min': '✉ 민시설 (시설 관리 담당)',
                          'dm_yun': '✉ 윤분석 (데이터 분석 담당)',
                          'dm_jo': '✉ 조당직 (당직 책임자)'},
             'always_channels': ['notice', 'team'],
             'welcome': [['notice', 'lim', '{call}, 시설 방호 영상분석반에 오신 것을 환영합니다 👋 CCTV 경보 업무에 쓸 작은 도구와 문서를 AI와 함께 만들어요.'],
                         ['notice', 'lim', '왼쪽 메뉴의 M1부터 차례로 진행해 주세요. 코드를 외울 필요는 없어요. AI에게 어떻게 설명할지가 중요해요.'],
                         ['team', 'yun', '{call}, 데이터 분석 담당 윤분석이에요. 제가 기록을 넘겨드리면 AI와 함께 표와 문서로 만들어 주세요 🙂']],
             'stages': [{'no': 1, 'name': '프로젝트 이해: 기지 외곽 감시와 CCTV 영상분석', 'subtitle': '프롬프트의 기본', 'missions': ['m1', 'm2']},
                        {'no': 2, 'name': '공개 감시 영상 데이터와 전처리', 'subtitle': '고치고 쪼개서 요청하기', 'missions': ['m3', 'm4']},
                        {'no': 3, 'name': '객체 탐지와 불필요 경보 줄이기', 'subtitle': '결과를 믿기 전에 검증하기', 'missions': ['m5', 'm6']},
                        {'no': 4, 'name': '드론 식별과 사람 확인 경보 처리', 'subtitle': '표현 규칙과 안전한 활용', 'missions': ['m7', 'm8']},
                        {'no': 5, 'name': '통합 관제와 사후검토', 'subtitle': '검토하고 묶어서 보고하기', 'missions': ['m9', 'm10']}],
             'vfiles': {'m1_alarms.csv': 'alarm_id,camera,hour,object\n'
                                         'A001,CAM2,17,glitch\n'
                                         'A002,CAM4,1,tree\n'
                                         'A003,CAM2,19,drone\n'
                                         'A004,CAM3,15,rain\n'
                                         'A005,CAM2,10,noise\n'
                                         'A006,CAM1,8,vehicle\n'
                                         'A007,CAM1,14,rain\n'
                                         'A008,CAM3,6,animal\n'
                                         'A009,CAM3,7,tree\n'
                                         'A010,CAM4,13,drone\n'
                                         'A011,CAM4,14,animal\n'
                                         'A012,CAM1,20,light\n'
                                         'A013,CAM3,21,light\n'
                                         'A014,CAM3,12,rain\n'
                                         'A015,CAM2,13,person\n'
                                         'A016,CAM4,0,light\n'
                                         'A017,CAM2,17,person\n'
                                         'A018,CAM1,11,rain\n'
                                         'A019,CAM1,8,tree\n'
                                         'A020,CAM2,1,animal\n'
                                         'A021,CAM3,2,rain\n'
                                         'A022,CAM4,21,noise\n'
                                         'A023,CAM3,20,tree\n'
                                         'A024,CAM2,15,light\n'
                                         'A025,CAM3,5,rain\n'
                                         'A026,CAM1,2,person\n'
                                         'A027,CAM3,17,glitch\n'
                                         'A028,CAM4,3,rain\n'
                                         'A029,CAM4,20,vehicle\n'
                                         'A030,CAM2,6,rain\n'
                                         'A031,CAM4,4,person\n'
                                         'A032,CAM2,9,animal\n'
                                         'A033,CAM1,1,tree\n'
                                         'A034,CAM3,14,person\n'
                                         'A035,CAM3,21,light\n'
                                         'A036,CAM1,8,glitch\n'
                                         'A037,CAM3,14,drone\n'
                                         'A038,CAM1,16,noise\n'
                                         'A039,CAM3,7,person\n'
                                         'A040,CAM2,0,person\n'
                                         'A041,CAM4,19,animal\n'
                                         'A042,CAM1,10,drone\n'
                                         'A043,CAM3,15,tree\n'
                                         'A044,CAM3,10,person\n'
                                         'A045,CAM2,9,animal\n'
                                         'A046,CAM3,12,animal\n'
                                         'A047,CAM1,23,drone\n'
                                         'A048,CAM1,6,vehicle\n'
                                         'A049,CAM1,4,animal\n'
                                         'A050,CAM3,0,animal\n'
                                         'A051,CAM2,21,tree\n'
                                         'A052,CAM3,16,tree\n'
                                         'A053,CAM2,4,animal\n'
                                         'A054,CAM3,9,rain\n'
                                         'A055,CAM3,7,light\n'
                                         'A056,CAM2,14,noise\n'
                                         'A057,CAM1,7,person\n'
                                         'A058,CAM3,23,light\n'
                                         'A059,CAM4,21,person\n'
                                         'A060,CAM2,1,rain\n'
                                         'A061,CAM4,19,glitch\n'
                                         'A062,CAM1,0,animal\n'
                                         'A063,CAM4,21,animal\n'
                                         'A064,CAM2,15,rain\n'
                                         'A065,CAM1,1,rain\n'
                                         'A066,CAM3,19,tree\n'
                                         'A067,CAM1,19,rain\n'
                                         'A068,CAM3,16,rain\n'
                                         'A069,CAM2,15,rain\n'
                                         'A070,CAM4,3,animal\n'
                                         'A071,CAM2,11,drone\n'
                                         'A072,CAM1,7,rain\n'
                                         'A073,CAM2,5,rain\n'
                                         'A074,CAM2,14,vehicle\n'
                                         'A075,CAM2,16,glitch\n'
                                         'A076,CAM2,3,tree\n'
                                         'A077,CAM1,10,glitch\n'
                                         'A078,CAM2,18,tree\n'
                                         'A079,CAM2,3,person\n'
                                         'A080,CAM1,14,tree\n',
                        'm2_requirements.csv': 'req_id,owner,target,min_recall,max_delay,max_unneeded_per_day\n'
                                               'R01,plan,person,95%,30s,20\n'
                                               'R02,plan,vehicle,0.9,1min,20\n'
                                               'R03,plan,drone,90,30s,20\n'
                                               'R04,control,person,0.85,45 sec,10\n'
                                               'R05,control,vehicle,85%,90s,10\n'
                                               'R06,control,drone,0.8,1 min,10\n'
                                               'R07,monitor,person,90%,2min,5\n'
                                               'R08,monitor,drone,0.85,60s,5\n'
                                               'R09,monitor,vehicle,,1min,5\n',
                        'm3_labels.csv': 'frame_id,clip_id,class,x,y,w,h,img_w,img_h\n'
                                         'F001,C01,bird,220,39,54,52,640,360\n'
                                         'F002,C01,person,472,173,49,32,640,360\n'
                                         'F003,C01,person,135,113,39,13,640,360\n'
                                         'F004,C01,Drone,109,65,10,22,640,360\n'
                                         'F005,C01,drone,419,297,19,55,640,360\n'
                                         'F006,C01,bird,586,127,22,58,640,360\n'
                                         'F007,C01,person,594,111,31,20,640,360\n'
                                         'F008,C01,bird,288,110,60,24,640,360\n'
                                         'F009,C01,bird,536,151,41,20,640,360\n'
                                         'F010,C01,person,134,215,38,17,640,360\n'
                                         'F011,C02,drone,356,49,14,38,640,360\n'
                                         'F012,C02,person,127,288,0,39,640,360\n'
                                         'F013,C02,person,298,10,53,32,640,360\n'
                                         'F014,C02,bird,403,195,46,47,640,360\n'
                                         'F015,C02,bird,66,90,24,55,640,360\n'
                                         'F016,C02,bird,332,331,57,10,640,360\n'
                                         'F017,C02,drone,245,289,8,42,640,360\n'
                                         'F018,C02,vehicle,620,197,50,22,640,360\n'
                                         'F019,C02,vehicle,389,14,45,13,640,360\n'
                                         'F020,C02,bird,91,283,37,29,640,360\n'
                                         'F021,C03,vehicle,335,140,51,10,640,360\n'
                                         'F022,C03,vehicle,466,158,42,14,640,360\n'
                                         'F023,C03,bird,274,186,19,19,640,360\n'
                                         'F024,C03,person,498,176,52,47,640,360\n'
                                         'F025,C03,vehicle,68,14,29,49,640,360\n'
                                         'F026,C03,uav,166,210,48,54,640,360\n'
                                         'F027,C03,bird,293,305,60,26,640,360\n'
                                         'F028,C03,drone,273,266,34,24,640,360\n'
                                         'F029,C03,person,462,168,33,59,640,360\n'
                                         'F030,C03,bird,469,227,17,16,640,360\n'
                                         'F031,C04,person,19,170,15,39,640,360\n'
                                         'F032,C04,vehicle,472,192,23,46,640,360\n'
                                         'F033,C04,person,12,340,24,40,640,360\n'
                                         'F034,C04,person,140,74,47,11,640,360\n'
                                         'F035,C04,drone,49,287,59,58,640,360\n'
                                         'F036,C04,drone,259,222,52,23,640,360\n'
                                         'F037,C04,person,389,296,58,26,640,360\n'
                                         'F038,C04,bird,421,334,27,20,640,360\n'
                                         'F039,C04,bird,237,169,54,45,640,360\n'
                                         'F040,C04,person,526,145,30,26,640,360\n',
                        'm3_labels_new.csv': 'frame_id,clip_id,class,x,y,w,h,img_w,img_h\n'
                                             'N01,D01,person,120.5,88,30,42,640,360\n'
                                             'N02,D01,person,122.0,90,30,42,640,360\n'
                                             'N03,D02,drone,"300,5",40,12,10,640,360\n'
                                             'N04,D02,drone,"304,0",41,12,10,640,360\n'
                                             'N05,D03,vehicle,n/a,200,60,40,640,360\n'
                                             'N06,D03,vehicle,210,201,700,40,640,360\n',
                        'm4_events.csv': 'event_id,x,y,condition,kind\n'
                                         'E001,92.0,3.2,fog,tree\n'
                                         'E002,94.2,0.2,day,tree\n'
                                         'E003,73.5,9.2,day,tree\n'
                                         'E004,86.2,32.0,day,animal\n'
                                         'E005,62.1,34.6,day,animal\n'
                                         'E006,53.6,14.2,day,animal\n'
                                         'E007,10.0,32.9,day,animal\n'
                                         'E008,77.6,5.8,night,tree\n'
                                         'E009,99.2,11.0,fog,animal\n'
                                         'E010,66.8,22.9,night,person\n'
                                         'E011,91.4,54.8,night,animal\n'
                                         'E012,4.0,20.8,day,animal\n'
                                         'E013,62.5,4.2,day,person\n'
                                         'E014,81.4,8.3,day,person\n'
                                         'E015,71.8,34.7,day,animal\n'
                                         'E016,72.0,17.1,night,animal\n'
                                         'E017,95.7,12.5,night,tree\n'
                                         'E018,52.2,40.4,night,animal\n'
                                         'E019,91.7,7.2,night,person\n'
                                         'E020,9.3,31.8,day,animal\n'
                                         'E021,88.4,4.3,night,tree\n'
                                         'E022,89.1,11.8,night,tree\n'
                                         'E023,76.3,2.3,day,tree\n'
                                         'E024,8.6,7.5,day,person\n'
                                         'E025,79.9,4.2,day,vehicle\n'
                                         'E026,50.6,8.0,day,person\n'
                                         'E027,19.7,14.7,night,vehicle\n'
                                         'E028,86.9,5.5,day,tree\n'
                                         'E029,84.8,5.5,night,tree\n'
                                         'E030,77.4,33.7,day,animal\n'
                                         'E031,74.0,5.0,day,tree\n'
                                         'E032,83.8,0.2,night,tree\n'
                                         'E033,2.9,19.2,night,animal\n'
                                         'E034,17.8,6.3,fog,person\n'
                                         'E035,68.5,37.1,night,animal\n'
                                         'E036,11.9,44.0,day,animal\n'
                                         'E037,81.2,45.2,day,animal\n'
                                         'E038,32.3,10.1,day,person\n'
                                         'E039,97.2,36.9,day,animal\n'
                                         'E040,23.4,31.2,fog,animal\n'
                                         'E041,14.3,51.2,day,animal\n'
                                         'E042,76.6,2.9,day,tree\n'
                                         'E043,94.8,10.4,day,tree\n'
                                         'E044,23.9,2.8,night,person\n'
                                         'E045,71.6,2.0,fog,person\n'
                                         'E046,18.3,3.8,day,animal\n'
                                         'E047,0.7,20.3,fog,vehicle\n'
                                         'E048,3.5,23.0,fog,animal\n'
                                         'E049,41.7,21.3,night,vehicle\n'
                                         'E050,98.5,7.3,night,tree\n'
                                         'E051,51.8,19.6,fog,person\n'
                                         'E052,43.7,49.0,fog,animal\n'
                                         'E053,45.1,9.0,day,person\n'
                                         'E054,71.4,4.4,night,tree\n'
                                         'E055,93.5,5.4,fog,tree\n'
                                         'E056,94.6,7.6,fog,tree\n'
                                         'E057,3.5,2.4,night,person\n'
                                         'E058,94.5,31.7,day,animal\n'
                                         'E059,98.3,3.9,night,tree\n'
                                         'E060,79.9,9.8,day,tree\n'
                                         'E061,20.3,11.4,day,person\n'
                                         'E062,80.7,3.4,day,tree\n'
                                         'E063,76.2,13.7,night,person\n'
                                         'E064,98.6,2.2,day,animal\n'
                                         'E065,59.2,7.2,day,person\n'
                                         'E066,95.0,6.7,night,animal\n'
                                         'E067,4.5,6.8,day,person\n'
                                         'E068,72.6,12.9,night,animal\n'
                                         'E069,79.6,3.3,night,tree\n'
                                         'E070,89.1,6.9,day,tree\n'
                                         'E071,22.5,20.2,day,person\n'
                                         'E072,90.4,10.1,day,animal\n'
                                         'E073,92.3,11.6,day,tree\n'
                                         'E074,22.6,28.0,fog,animal\n'
                                         'E075,29.4,7.1,fog,person\n'
                                         'E076,15.2,17.8,fog,person\n'
                                         'E077,74.9,38.9,night,animal\n'
                                         'E078,69.1,18.1,day,person\n'
                                         'E079,71.9,2.7,fog,tree\n'
                                         'E080,79.3,3.3,day,tree\n'
                                         'E081,49.7,20.4,day,vehicle\n'
                                         'E082,55.2,27.3,night,person\n'
                                         'E083,65.4,23.4,day,person\n'
                                         'E084,30.6,11.5,day,person\n'
                                         'E085,42.1,26.8,day,person\n'
                                         'E086,95.9,9.4,night,vehicle\n'
                                         'E087,2.4,25.8,day,person\n'
                                         'E088,86.2,1.0,night,tree\n'
                                         'E089,15.8,4.4,day,animal\n'
                                         'E090,98.0,3.3,night,tree\n'
                                         'E091,84.4,25.7,night,vehicle\n'
                                         'E092,86.9,27.5,day,person\n'
                                         'E093,60.0,22.0,night,person\n'
                                         'E094,77.4,3.2,night,animal\n'
                                         'E095,86.5,5.0,night,tree\n'
                                         'E096,90.1,13.4,day,tree\n'
                                         'E097,95.6,11.4,fog,tree\n'
                                         'E098,67.7,43.1,night,animal\n'
                                         'E099,88.1,5.3,day,tree\n'
                                         'E100,72.0,0.6,day,tree\n',
                        'm5_dets.csv': 'det_id,model,condition,gt_x,gt_y,gt_w,gt_h,x,y,w,h,score\n'
                                       'T001,motion,day,459,188,53,22,510,188,53,22,0.61\n'
                                       'T002,motion,day,181,159,26,38,179,156,26,38,0.83\n'
                                       'T003,motion,day,189,96,43,55,189,96,43,55,0.93\n'
                                       'T004,motion,day,481,126,27,55,484,127,27,55,0.94\n'
                                       'T005,motion,day,41,177,55,23,43,180,55,23,0.84\n'
                                       'T006,motion,day,105,210,50,48,107,212,50,48,0.76\n'
                                       'T007,motion,day,327,273,40,53,330,276,40,53,0.95\n'
                                       'T008,motion,day,367,73,52,42,369,72,52,42,0.79\n'
                                       'T009,motion,day,481,66,51,46,530,66,51,46,0.44\n'
                                       'T010,motion,day,445,17,51,28,446,20,51,28,0.63\n'
                                       'T011,motion,day,398,227,44,53,396,229,44,53,0.56\n'
                                       'T012,motion,day,366,49,38,52,402,49,38,52,0.4\n'
                                       'T013,motion,day,482,196,42,35,485,193,42,35,0.95\n'
                                       'T014,motion,day,261,95,43,55,259,98,43,55,0.57\n'
                                       'T015,motion,day,461,121,31,23,490,121,31,23,0.7\n'
                                       'T016,motion,day,275,118,55,33,272,118,55,33,0.91\n'
                                       'T017,motion,day,491,280,50,46,492,278,50,46,0.96\n'
                                       'T018,motion,day,83,117,52,60,85,117,52,60,0.92\n'
                                       'T019,motion,day,448,174,31,58,447,173,31,58,0.56\n'
                                       'T020,motion,day,177,208,37,57,175,206,37,57,0.66\n'
                                       'T021,motion,day,53,153,20,58,,,,,\n'
                                       'T022,motion,day,28,26,51,21,25,28,51,21,0.91\n'
                                       'T023,motion,day,399,236,44,46,397,236,44,46,0.55\n'
                                       'T024,motion,day,332,78,35,44,330,79,35,44,0.66\n'
                                       'T025,motion,day,369,25,39,29,370,27,39,29,0.56\n'
                                       'T026,motion,day,223,57,50,44,,,,,\n'
                                       'T027,motion,day,438,240,48,20,435,242,48,20,0.81\n'
                                       'T028,motion,day,190,246,50,52,192,244,50,52,0.85\n'
                                       'T029,motion,day,62,278,50,52,64,275,50,52,0.76\n'
                                       'T030,motion,day,227,269,32,33,226,266,32,33,0.76\n'
                                       'T031,motion,day,100,144,60,29,99,143,60,29,0.56\n'
                                       'T032,motion,day,182,116,46,32,,,,,\n'
                                       'T033,motion,day,244,224,49,36,291,224,49,36,0.54\n'
                                       'T034,motion,day,137,43,35,43,140,43,35,43,0.89\n'
                                       'T035,motion,day,367,232,38,56,366,233,38,56,0.93\n'
                                       'T036,motion,day,48,254,16,46,,,,,\n'
                                       'T037,motion,day,421,239,27,33,419,236,27,33,0.78\n'
                                       'T038,motion,day,103,152,30,41,104,149,30,41,0.88\n'
                                       'T039,motion,day,90,276,18,41,90,273,18,41,0.93\n'
                                       'T040,motion,day,171,246,46,19,170,244,46,19,0.63\n'
                                       'T041,motion,day,400,11,51,45,399,9,51,45,0.83\n'
                                       'T042,motion,day,283,139,39,45,283,139,39,45,0.7\n'
                                       'T043,motion,day,170,119,38,34,167,119,38,34,0.69\n'
                                       'T044,motion,day,331,209,53,26,329,208,53,26,0.61\n'
                                       'T045,motion,day,170,70,22,40,170,70,22,40,0.67\n'
                                       'T046,motion,day,106,39,57,60,107,42,57,60,0.64\n'
                                       'T047,motion,day,395,273,21,59,396,270,21,59,0.59\n'
                                       'T048,motion,day,174,33,53,28,,,,,\n'
                                       'T049,motion,day,62,94,48,45,59,97,48,45,0.84\n'
                                       'T050,motion,day,434,207,51,26,435,209,51,26,0.67\n'
                                       'T051,motion,day,425,51,51,30,423,50,51,30,0.91\n'
                                       'T052,motion,day,494,108,44,45,,,,,\n'
                                       'T053,motion,day,462,131,28,37,,,,,\n'
                                       'T054,motion,day,323,152,16,20,325,153,16,20,0.85\n'
                                       'T055,motion,day,407,23,47,59,408,26,47,59,0.6\n'
                                       'T056,motion,day,189,14,23,41,188,12,23,41,0.95\n'
                                       'T057,motion,day,273,163,59,60,273,164,59,60,0.88\n'
                                       'T058,motion,day,93,166,30,55,95,163,30,55,0.83\n'
                                       'T059,motion,day,267,166,24,39,265,168,24,39,0.64\n'
                                       'T060,motion,day,211,204,25,42,209,207,25,42,0.79\n'
                                       'T061,motion,day,,,,,40,198,48,16,0.57\n'
                                       'T062,motion,day,,,,,235,239,58,48,0.65\n'
                                       'T063,motion,day,,,,,284,80,30,17,0.5\n'
                                       'T064,motion,day,,,,,134,70,24,37,0.38\n'
                                       'T065,motion,day,,,,,500,147,35,21,0.43\n'
                                       'T066,motion,day,,,,,390,109,50,53,0.39\n'
                                       'T067,motion,day,,,,,124,74,53,52,0.62\n'
                                       'T068,motion,day,,,,,123,36,42,57,0.41\n'
                                       'T069,motion,day,,,,,402,237,50,34,0.5\n'
                                       'T070,motion,day,,,,,316,272,21,43,0.63\n'
                                       'T071,motion,day,,,,,430,50,42,32,0.31\n'
                                       'T072,motion,day,,,,,374,210,30,28,0.6\n'
                                       'T073,motion,day,,,,,183,219,33,48,0.54\n'
                                       'T074,motion,day,,,,,428,241,54,59,0.73\n'
                                       'T075,motion,day,,,,,482,127,50,45,0.47\n'
                                       'T076,motion,day,,,,,228,177,50,28,0.44\n'
                                       'T077,motion,day,,,,,293,189,25,28,0.53\n'
                                       'T078,motion,day,,,,,305,204,21,34,0.43\n'
                                       'T079,motion,night,91,153,40,40,89,154,40,40,0.95\n'
                                       'T080,motion,night,161,39,51,31,,,,,\n'
                                       'T081,motion,night,479,180,21,23,481,180,21,23,0.63\n'
                                       'T082,motion,night,43,246,53,26,45,246,53,26,0.95\n'
                                       'T083,motion,night,463,194,24,57,466,197,24,57,0.74\n'
                                       'T084,motion,night,79,28,20,36,80,27,20,36,0.68\n'
                                       'T085,motion,night,451,182,57,27,454,181,57,27,0.65\n'
                                       'T086,motion,night,495,78,26,44,497,76,26,44,0.72\n'
                                       'T087,motion,night,412,99,36,34,410,99,36,34,0.75\n'
                                       'T088,motion,night,196,152,16,41,193,151,16,41,0.92\n'
                                       'T089,motion,night,289,80,22,43,287,79,22,43,0.93\n'
                                       'T090,motion,night,335,27,32,34,335,28,32,34,0.67\n'
                                       'T091,motion,night,410,239,28,44,412,240,28,44,0.78\n'
                                       'T092,motion,night,247,279,54,57,248,278,54,57,0.6\n'
                                       'T093,motion,night,85,184,53,48,87,181,53,48,0.88\n'
                                       'T094,motion,night,106,269,52,18,107,267,52,18,0.67\n'
                                       'T095,motion,night,71,96,33,52,69,93,33,52,0.87\n'
                                       'T096,motion,night,417,253,47,39,419,251,47,39,0.74\n'
                                       'T097,motion,night,54,46,24,26,51,43,24,26,0.95\n'
                                       'T098,motion,night,331,108,56,35,329,107,56,35,0.83\n'
                                       'T099,motion,night,221,120,40,42,220,119,40,42,0.81\n'
                                       'T100,motion,night,411,214,56,37,409,212,56,37,0.94\n'
                                       'T101,motion,night,27,212,17,43,30,209,17,43,0.58\n'
                                       'T102,motion,night,117,76,51,18,117,73,51,18,0.62\n'
                                       'T103,motion,night,264,63,23,19,267,64,23,19,0.98\n'
                                       'T104,motion,night,181,114,16,32,,,,,\n'
                                       'T105,motion,night,42,52,30,50,40,50,30,50,0.84\n'
                                       'T106,motion,night,345,154,40,35,347,154,40,35,0.73\n'
                                       'T107,motion,night,408,124,39,25,407,124,39,25,0.69\n'
                                       'T108,motion,night,326,82,47,41,327,81,47,41,0.55\n'
                                       'T109,motion,night,449,137,39,32,449,139,39,32,0.89\n'
                                       'T110,motion,night,252,251,27,16,253,251,27,16,0.68\n'
                                       'T111,motion,night,351,213,41,28,350,216,41,28,0.6\n'
                                       'T112,motion,night,88,247,37,45,,,,,\n'
                                       'T113,motion,night,263,256,19,17,262,253,19,17,0.7\n'
                                       'T114,motion,night,334,100,40,46,335,97,40,46,0.72\n'
                                       'T115,motion,night,486,59,42,29,486,62,42,29,0.79\n'
                                       'T116,motion,night,446,124,33,51,448,126,33,51,0.59\n'
                                       'T117,motion,night,203,158,34,30,235,158,34,30,0.44\n'
                                       'T118,motion,night,43,261,28,40,45,259,28,40,0.8\n'
                                       'T119,motion,night,,,,,95,235,20,29,0.78\n'
                                       'T120,motion,night,,,,,374,25,44,58,0.75\n'
                                       'T121,motion,night,,,,,335,20,40,18,0.65\n'
                                       'T122,motion,night,,,,,487,46,41,31,0.71\n'
                                       'T123,motion,night,,,,,289,132,29,56,0.54\n'
                                       'T124,motion,night,,,,,203,30,23,47,0.64\n'
                                       'T125,motion,night,,,,,205,50,37,53,0.79\n'
                                       'T126,motion,night,,,,,63,174,51,21,0.64\n'
                                       'T127,motion,night,,,,,229,262,40,16,0.48\n'
                                       'T128,motion,night,,,,,322,16,20,19,0.79\n'
                                       'T129,motion,night,,,,,389,153,50,56,0.34\n'
                                       'T130,motion,night,,,,,252,223,23,46,0.77\n'
                                       'T131,motion,night,,,,,300,137,16,26,0.47\n'
                                       'T132,motion,night,,,,,271,128,51,48,0.45\n'
                                       'T133,motion,fog,213,165,53,43,,,,,\n'
                                       'T134,motion,fog,245,275,28,43,,,,,\n'
                                       'T135,motion,fog,133,157,52,28,132,156,52,28,0.81\n'
                                       'T136,motion,fog,440,257,22,29,460,257,22,29,0.6\n'
                                       'T137,motion,fog,90,221,60,49,91,218,60,49,0.64\n'
                                       'T138,motion,fog,240,159,53,56,,,,,\n'
                                       'T139,motion,fog,300,123,17,48,298,125,17,48,0.74\n'
                                       'T140,motion,fog,245,27,25,59,,,,,\n'
                                       'T141,motion,fog,26,87,51,51,26,85,51,51,0.65\n'
                                       'T142,motion,fog,138,159,55,59,191,159,55,59,0.68\n'
                                       'T143,motion,fog,439,113,49,24,440,111,49,24,0.92\n'
                                       'T144,motion,fog,276,279,35,36,274,278,35,36,0.67\n'
                                       'T145,motion,fog,65,173,31,51,62,173,31,51,0.73\n'
                                       'T146,motion,fog,426,186,17,16,423,184,17,16,0.88\n'
                                       'T147,motion,fog,178,217,41,30,179,220,41,30,0.68\n'
                                       'T148,motion,fog,39,201,21,32,38,199,21,32,0.92\n'
                                       'T149,motion,fog,478,24,44,16,480,25,44,16,0.66\n'
                                       'T150,motion,fog,213,248,23,35,210,251,23,35,0.78\n'
                                       'T151,motion,fog,258,227,20,51,259,225,20,51,0.75\n'
                                       'T152,motion,fog,167,74,42,33,167,77,42,33,0.88\n'
                                       'T153,motion,fog,,,,,292,218,29,39,0.39\n'
                                       'T154,motion,fog,,,,,211,267,25,57,0.61\n'
                                       'T155,motion,fog,,,,,341,78,41,27,0.71\n'
                                       'T156,motion,fog,,,,,281,24,32,27,0.77\n'
                                       'T157,motion,fog,,,,,222,71,59,57,0.79\n'
                                       'T158,motion,fog,,,,,186,256,27,46,0.39\n'
                                       'T159,motion,fog,,,,,309,211,46,32,0.49\n'
                                       'T160,motion,fog,,,,,29,57,60,46,0.71\n'
                                       'T161,motion,fog,,,,,269,257,34,56,0.61\n'
                                       'T162,pretrained,day,408,185,59,31,409,185,59,31,0.72\n'
                                       'T163,pretrained,day,452,69,48,60,450,70,48,60,0.92\n'
                                       'T164,pretrained,day,74,238,32,34,104,238,32,34,0.38\n'
                                       'T165,pretrained,day,106,20,22,27,105,21,22,27,0.88\n'
                                       'T166,pretrained,day,348,166,56,23,350,163,56,23,0.95\n'
                                       'T167,pretrained,day,307,267,52,57,307,266,52,57,0.74\n'
                                       'T168,pretrained,day,34,141,59,35,31,139,59,35,0.64\n'
                                       'T169,pretrained,day,295,110,18,27,294,107,18,27,0.67\n'
                                       'T170,pretrained,day,296,38,57,19,299,35,57,19,0.96\n'
                                       'T171,pretrained,day,91,220,33,19,90,223,33,19,0.89\n'
                                       'T172,pretrained,day,343,227,60,59,346,229,60,59,0.74\n'
                                       'T173,pretrained,day,372,104,52,25,375,106,52,25,0.93\n'
                                       'T174,pretrained,day,479,205,25,30,481,206,25,30,0.8\n'
                                       'T175,pretrained,day,307,253,48,17,305,254,48,17,0.95\n'
                                       'T176,pretrained,day,24,50,22,20,24,49,22,20,0.57\n'
                                       'T177,pretrained,day,367,150,17,49,366,147,17,49,0.82\n'
                                       'T178,pretrained,day,419,145,18,23,416,147,18,23,0.63\n'
                                       'T179,pretrained,day,110,75,28,38,112,76,28,38,0.58\n'
                                       'T180,pretrained,day,122,159,40,54,121,161,40,54,0.83\n'
                                       'T181,pretrained,day,492,65,37,49,489,63,37,49,0.69\n'
                                       'T182,pretrained,day,276,110,40,27,314,110,40,27,0.58\n'
                                       'T183,pretrained,day,312,126,56,32,309,129,56,32,0.56\n'
                                       'T184,pretrained,day,106,269,57,23,106,271,57,23,0.77\n'
                                       'T185,pretrained,day,298,238,40,21,296,241,40,21,0.91\n'
                                       'T186,pretrained,day,80,88,42,29,120,88,42,29,0.34\n'
                                       'T187,pretrained,day,89,18,19,27,90,20,19,27,0.68\n'
                                       'T188,pretrained,day,54,160,16,31,55,157,16,31,0.89\n'
                                       'T189,pretrained,day,28,38,32,22,26,37,32,22,0.85\n'
                                       'T190,pretrained,day,384,261,48,20,386,259,48,20,0.78\n'
                                       'T191,pretrained,day,247,232,18,49,250,229,18,49,0.88\n'
                                       'T192,pretrained,day,438,185,26,17,,,,,\n'
                                       'T193,pretrained,day,133,228,53,44,131,229,53,44,0.84\n'
                                       'T194,pretrained,day,39,183,26,43,36,185,26,43,0.72\n'
                                       'T195,pretrained,day,147,195,26,41,149,195,26,41,0.62\n'
                                       'T196,pretrained,day,418,128,45,17,418,130,45,17,0.61\n'
                                       'T197,pretrained,day,283,201,23,43,283,204,23,43,0.78\n'
                                       'T198,pretrained,day,447,95,36,47,450,95,36,47,0.84\n'
                                       'T199,pretrained,day,393,35,27,60,390,35,27,60,0.59\n'
                                       'T200,pretrained,day,403,130,56,28,400,128,56,28,0.69\n'
                                       'T201,pretrained,day,95,255,59,54,92,252,59,54,0.82\n'
                                       'T202,pretrained,day,493,249,30,43,494,252,30,43,0.66\n'
                                       'T203,pretrained,day,248,149,58,48,250,148,58,48,0.89\n'
                                       'T204,pretrained,day,68,193,59,49,68,193,59,49,0.66\n'
                                       'T205,pretrained,day,199,157,27,21,199,154,27,21,0.83\n'
                                       'T206,pretrained,day,255,51,55,30,,,,,\n'
                                       'T207,pretrained,day,42,15,42,59,40,18,42,59,0.82\n'
                                       'T208,pretrained,day,368,33,19,29,370,33,19,29,0.95\n'
                                       'T209,pretrained,day,18,38,51,44,16,37,51,44,0.9\n'
                                       'T210,pretrained,day,221,140,35,22,224,143,35,22,0.56\n'
                                       'T211,pretrained,day,91,155,54,59,91,152,54,59,0.9\n'
                                       'T212,pretrained,day,295,47,34,57,327,47,34,57,0.69\n'
                                       'T213,pretrained,day,35,211,39,47,32,210,39,47,0.93\n'
                                       'T214,pretrained,day,457,172,25,20,460,172,25,20,0.79\n'
                                       'T215,pretrained,day,260,51,38,50,263,48,38,50,0.7\n'
                                       'T216,pretrained,day,486,142,16,52,486,141,16,52,0.91\n'
                                       'T217,pretrained,day,464,167,51,20,466,170,51,20,0.79\n'
                                       'T218,pretrained,day,365,190,41,35,,,,,\n'
                                       'T219,pretrained,day,241,255,18,54,244,255,18,54,0.6\n'
                                       'T220,pretrained,day,472,142,52,31,471,142,52,31,0.78\n'
                                       'T221,pretrained,day,152,272,46,56,155,273,46,56,0.85\n'
                                       'T222,pretrained,day,,,,,70,240,28,20,0.61\n'
                                       'T223,pretrained,day,,,,,289,92,43,29,0.48\n'
                                       'T224,pretrained,day,,,,,289,166,60,43,0.63\n'
                                       'T225,pretrained,day,,,,,29,77,27,30,0.33\n'
                                       'T226,pretrained,night,420,102,46,55,,,,,\n'
                                       'T227,pretrained,night,101,37,33,18,104,39,33,18,0.78\n'
                                       'T228,pretrained,night,117,66,25,60,117,67,25,60,0.89\n'
                                       'T229,pretrained,night,153,224,42,22,154,227,42,22,0.72\n'
                                       'T230,pretrained,night,76,42,49,30,,,,,\n'
                                       'T231,pretrained,night,145,171,25,41,143,170,25,41,0.67\n'
                                       'T232,pretrained,night,305,109,44,30,302,111,44,30,0.75\n'
                                       'T233,pretrained,night,252,191,28,22,278,191,28,22,0.69\n'
                                       'T234,pretrained,night,352,272,28,17,350,273,28,17,0.67\n'
                                       'T235,pretrained,night,247,93,55,51,249,91,55,51,0.75\n'
                                       'T236,pretrained,night,246,247,17,37,261,247,17,37,0.51\n'
                                       'T237,pretrained,night,58,218,47,29,56,216,47,29,0.72\n'
                                       'T238,pretrained,night,313,133,42,18,312,130,42,18,0.79\n'
                                       'T239,pretrained,night,196,264,40,42,,,,,\n'
                                       'T240,pretrained,night,226,60,42,40,226,58,42,40,0.92\n'
                                       'T241,pretrained,night,313,97,59,18,314,95,59,18,0.7\n'
                                       'T242,pretrained,night,184,225,37,51,184,225,37,51,0.62\n'
                                       'T243,pretrained,night,344,166,43,17,,,,,\n'
                                       'T244,pretrained,night,158,58,49,20,205,58,49,20,0.55\n'
                                       'T245,pretrained,night,240,71,19,34,237,74,19,34,0.85\n'
                                       'T246,pretrained,night,437,116,59,49,435,113,59,49,0.91\n'
                                       'T247,pretrained,night,287,267,54,26,,,,,\n'
                                       'T248,pretrained,night,68,16,16,19,82,16,16,19,0.42\n'
                                       'T249,pretrained,night,73,30,21,32,74,29,21,32,0.7\n'
                                       'T250,pretrained,night,50,30,37,55,85,30,37,55,0.34\n'
                                       'T251,pretrained,night,433,116,19,36,,,,,\n'
                                       'T252,pretrained,night,90,76,50,36,,,,,\n'
                                       'T253,pretrained,night,431,265,21,52,433,262,21,52,0.57\n'
                                       'T254,pretrained,night,42,43,49,25,89,43,49,25,0.46\n'
                                       'T255,pretrained,night,419,24,31,41,,,,,\n'
                                       'T256,pretrained,night,140,207,35,20,,,,,\n'
                                       'T257,pretrained,night,39,97,27,58,42,98,27,58,0.76\n'
                                       'T258,pretrained,night,412,100,36,57,411,103,36,57,0.83\n'
                                       'T259,pretrained,night,380,128,38,58,381,131,38,58,0.77\n'
                                       'T260,pretrained,night,69,53,34,34,,,,,\n'
                                       'T261,pretrained,night,367,248,30,54,367,249,30,54,0.6\n'
                                       'T262,pretrained,night,293,266,17,34,308,266,17,34,0.37\n'
                                       'T263,pretrained,night,419,26,21,33,422,26,21,33,0.65\n'
                                       'T264,pretrained,night,413,204,51,30,414,204,51,30,0.83\n'
                                       'T265,pretrained,night,56,179,41,22,58,182,41,22,0.76\n'
                                       'T266,pretrained,night,,,,,359,277,32,21,0.49\n'
                                       'T267,pretrained,night,,,,,133,202,32,28,0.66\n'
                                       'T268,pretrained,night,,,,,151,181,58,46,0.57\n'
                                       'T269,pretrained,night,,,,,124,93,24,23,0.69\n'
                                       'T270,pretrained,fog,449,210,52,45,450,211,52,45,0.79\n'
                                       'T271,pretrained,fog,440,17,38,28,,,,,\n'
                                       'T272,pretrained,fog,274,155,51,44,,,,,\n'
                                       'T273,pretrained,fog,81,199,21,52,79,198,21,52,0.7\n'
                                       'T274,pretrained,fog,250,30,49,59,247,28,49,59,0.57\n'
                                       'T275,pretrained,fog,198,112,35,50,200,112,35,50,0.95\n'
                                       'T276,pretrained,fog,306,137,45,16,303,134,45,16,0.7\n'
                                       'T277,pretrained,fog,164,173,57,21,219,173,57,21,0.44\n'
                                       'T278,pretrained,fog,282,247,55,49,281,249,55,49,0.71\n'
                                       'T279,pretrained,fog,35,88,19,42,52,88,19,42,0.62\n'
                                       'T280,pretrained,fog,343,247,39,25,343,247,39,25,0.74\n'
                                       'T281,pretrained,fog,110,142,45,48,107,143,45,48,0.72\n'
                                       'T282,pretrained,fog,268,262,25,48,266,265,25,48,0.87\n'
                                       'T283,pretrained,fog,422,70,18,40,420,70,18,40,0.6\n'
                                       'T284,pretrained,fog,419,104,60,24,,,,,\n'
                                       'T285,pretrained,fog,316,201,42,25,316,202,42,25,0.83\n'
                                       'T286,pretrained,fog,436,108,17,37,,,,,\n'
                                       'T287,pretrained,fog,203,178,48,23,202,181,48,23,0.66\n'
                                       'T288,pretrained,fog,148,17,59,44,151,17,59,44,0.58\n'
                                       'T289,pretrained,fog,354,189,18,53,,,,,\n'
                                       'T290,pretrained,fog,,,,,402,132,57,43,0.56\n'
                                       'T291,pretrained,fog,,,,,481,116,31,32,0.63\n'
                                       'T292,pretrained,fog,,,,,330,164,19,49,0.45\n',
                        'm6_alarms.csv': 'alarm_id,day,score,is_target,cause\n'
                                         'P001,4,0.03,0,light\n'
                                         'P002,2,0.32,0,tree\n'
                                         'P003,1,0.3,0,animal\n'
                                         'P004,5,0.91,1,none\n'
                                         'P005,5,0.44,0,animal\n'
                                         'P006,3,0.13,0,animal\n'
                                         'P007,6,0.36,0,tree\n'
                                         'P008,6,0.73,0,light\n'
                                         'P009,7,0.76,1,none\n'
                                         'P010,3,0.71,1,none\n'
                                         'P011,1,0.4,0,light\n'
                                         'P012,7,0.13,0,animal\n'
                                         'P013,5,0.13,0,animal\n'
                                         'P014,5,0.19,0,animal\n'
                                         'P015,2,0.03,0,light\n'
                                         'P016,7,0.45,0,tree\n'
                                         'P017,2,0.24,0,light\n'
                                         'P018,1,0.75,1,none\n'
                                         'P019,5,0.33,0,rain\n'
                                         'P020,5,0.26,0,animal\n'
                                         'P021,1,0.08,0,animal\n'
                                         'P022,5,0.33,0,animal\n'
                                         'P023,1,0.5,0,light\n'
                                         'P024,1,0.67,1,none\n'
                                         'P025,3,0.57,0,tree\n'
                                         'P026,4,0.08,0,animal\n'
                                         'P027,7,0.1,0,animal\n'
                                         'P028,4,0.33,0,animal\n'
                                         'P029,6,0.72,0,rain\n'
                                         'P030,2,0.4,0,tree\n'
                                         'P031,5,0.32,0,animal\n'
                                         'P032,6,0.2,0,rain\n'
                                         'P033,3,0.59,0,animal\n'
                                         'P034,4,0.28,0,animal\n'
                                         'P035,7,0.26,0,tree\n'
                                         'P036,3,0.39,0,animal\n'
                                         'P037,3,0.88,1,none\n'
                                         'P038,5,0.28,0,tree\n'
                                         'P039,4,0.65,0,animal\n'
                                         'P040,1,,0,light\n'
                                         'P041,7,0.28,0,animal\n'
                                         'P042,5,0.64,0,animal\n'
                                         'P043,4,0.5,0,tree\n'
                                         'P044,6,0.18,0,animal\n'
                                         'P045,1,0.2,0,animal\n'
                                         'P046,2,0.32,0,animal\n'
                                         'P047,7,0.04,0,rain\n'
                                         'P048,7,0.67,0,light\n'
                                         'P049,3,0.25,0,rain\n'
                                         'P050,4,0.54,0,tree\n'
                                         'P051,4,0.77,1,none\n'
                                         'P052,5,0.85,1,none\n'
                                         'P053,5,0.14,0,tree\n'
                                         'P054,1,0.33,0,animal\n'
                                         'P055,5,0.49,0,rain\n'
                                         'P056,5,0.17,0,tree\n'
                                         'P057,7,0.39,0,animal\n'
                                         'P058,5,0.04,0,tree\n'
                                         'P059,6,0.42,0,animal\n'
                                         'P060,3,0.19,0,animal\n'
                                         'P061,3,0.16,0,light\n'
                                         'P062,4,0.47,0,tree\n'
                                         'P063,5,0.13,0,light\n'
                                         'P064,5,0.2,0,light\n'
                                         'P065,2,0.57,0,animal\n'
                                         'P066,2,0.27,0,animal\n'
                                         'P067,6,0.23,0,rain\n'
                                         'P068,5,0.36,1,none\n'
                                         'P069,5,0.51,0,animal\n'
                                         'P070,6,0.56,0,animal\n'
                                         'P071,3,0.56,0,animal\n'
                                         'P072,2,0.56,0,tree\n'
                                         'P073,7,0.1,0,tree\n'
                                         'P074,5,0.29,0,animal\n'
                                         'P075,4,0.95,1,none\n'
                                         'P076,7,0.07,0,animal\n'
                                         'P077,4,0.39,0,animal\n'
                                         'P078,3,0.77,1,none\n'
                                         'P079,4,0.87,1,none\n'
                                         'P080,2,0.13,0,animal\n'
                                         'P081,4,0.27,0,animal\n'
                                         'P082,1,0.87,0,tree\n'
                                         'P083,1,0.06,0,tree\n'
                                         'P084,2,0.79,1,none\n'
                                         'P085,2,0.13,0,light\n'
                                         'P086,1,0.29,0,rain\n'
                                         'P087,7,0.44,0,tree\n'
                                         'P088,7,0.78,0,tree\n'
                                         'P089,2,0.17,0,animal\n'
                                         'P090,3,0.38,0,animal\n'
                                         'P091,1,0.51,0,tree\n'
                                         'P092,7,0.58,0,animal\n'
                                         'P093,4,0.16,0,rain\n'
                                         'P094,2,0.04,0,animal\n'
                                         'P095,6,0.35,0,tree\n'
                                         'P096,3,0.28,0,animal\n'
                                         'P097,2,0.24,0,light\n'
                                         'P098,6,0.61,0,tree\n'
                                         'P099,2,0.51,0,animal\n'
                                         'P100,5,0.12,0,light\n'
                                         'P101,2,0.2,0,animal\n'
                                         'P102,3,0.17,0,animal\n'
                                         'P103,1,0.19,0,rain\n'
                                         'P104,6,0.2,0,animal\n'
                                         'P105,3,0.4,0,animal\n'
                                         'P106,5,0.07,0,animal\n'
                                         'P107,6,0.66,0,tree\n'
                                         'P108,2,0.13,0,light\n'
                                         'P109,1,0.3,0,animal\n'
                                         'P110,1,0.44,0,rain\n'
                                         'P111,4,0.31,0,tree\n'
                                         'P112,5,0.6,0,light\n'
                                         'P113,5,0.19,0,tree\n'
                                         'P114,5,0.25,0,light\n'
                                         'P115,2,0.45,0,light\n'
                                         'P116,5,0.39,0,light\n'
                                         'P117,2,0.53,1,none\n'
                                         'P118,4,0.3,0,tree\n'
                                         'P119,4,0.34,0,animal\n'
                                         'P120,7,0.41,0,light\n'
                                         'P121,7,0.29,0,animal\n'
                                         'P122,4,0.2,0,animal\n'
                                         'P123,7,0.33,0,animal\n'
                                         'P124,1,0.4,0,animal\n'
                                         'P125,5,0.1,0,animal\n'
                                         'P126,2,0.47,0,animal\n'
                                         'P127,3,0.19,0,animal\n'
                                         'P128,1,0.77,1,none\n'
                                         'P129,7,0.12,0,tree\n'
                                         'P130,3,0.27,0,tree\n'
                                         'P131,3,0.49,0,rain\n'
                                         'P132,6,0.18,0,animal\n'
                                         'P133,3,0.01,0,rain\n'
                                         'P134,1,0.36,0,tree\n'
                                         'P135,5,0.19,0,tree\n'
                                         'P136,4,0.28,0,animal\n'
                                         'P137,5,0.31,0,tree\n'
                                         'P138,4,0.16,0,rain\n'
                                         'P139,2,0.41,0,rain\n'
                                         'P140,2,0.09,0,light\n'
                                         'P141,1,0.75,0,tree\n'
                                         'P142,2,0.07,0,rain\n'
                                         'P143,1,0.51,0,animal\n'
                                         'P144,7,0.41,0,tree\n'
                                         'P145,6,0.43,0,tree\n'
                                         'P146,2,0.38,0,tree\n'
                                         'P147,2,0.1,0,rain\n'
                                         'P148,3,0.29,0,light\n'
                                         'P149,1,0.47,0,animal\n'
                                         'P150,1,0.35,0,tree\n'
                                         'P151,6,0.11,0,rain\n'
                                         'P152,2,0.22,0,rain\n'
                                         'P153,5,0.36,0,animal\n'
                                         'P154,6,0.15,0,animal\n'
                                         'P155,1,0.23,0,tree\n'
                                         'P156,3,0.06,0,animal\n'
                                         'P157,5,0.34,0,animal\n'
                                         'P158,3,0.57,0,tree\n'
                                         'P159,5,0.07,0,light\n'
                                         'P160,6,0.26,0,animal\n'
                                         'P161,6,0.39,0,animal\n'
                                         'P162,6,0.71,1,none\n'
                                         'P163,1,0.28,0,animal\n'
                                         'P164,2,0.14,0,rain\n'
                                         'P165,4,0.22,0,light\n'
                                         'P166,1,0.62,1,none\n'
                                         'P167,2,0.72,0,rain\n'
                                         'P168,3,0.07,0,animal\n'
                                         'P169,3,0.36,0,tree\n'
                                         'P170,6,0.1,0,animal\n'
                                         'P171,7,0.68,0,tree\n'
                                         'P172,3,0.45,0,animal\n'
                                         'P173,1,0.39,0,animal\n'
                                         'P174,3,0.58,0,animal\n'
                                         'P175,1,0.63,0,rain\n'
                                         'P176,4,0.23,0,animal\n'
                                         'P177,3,0.35,0,light\n'
                                         'P178,5,0.82,1,none\n'
                                         'P179,4,0.18,0,animal\n'
                                         'P180,5,0.23,0,rain\n'
                                         'P181,2,,0,rain\n'
                                         'P182,3,0.21,0,animal\n'
                                         'P183,6,0.41,0,light\n'
                                         'P184,2,0.17,0,animal\n'
                                         'P185,2,0.29,0,tree\n'
                                         'P186,2,0.18,0,tree\n'
                                         'P187,7,0.1,0,light\n'
                                         'P188,4,0.33,0,tree\n'
                                         'P189,2,0.37,0,rain\n'
                                         'P190,6,0.19,0,animal\n'
                                         'P191,3,0.25,0,rain\n'
                                         'P192,1,0.26,0,rain\n'
                                         'P193,4,0.93,1,none\n'
                                         'P194,6,0.43,0,tree\n'
                                         'P195,2,0.46,0,tree\n'
                                         'P196,6,0.07,0,tree\n'
                                         'P197,5,0.23,0,rain\n'
                                         'P198,5,0.2,0,animal\n'
                                         'P199,2,0.36,0,light\n'
                                         'P200,7,0.07,0,animal\n'
                                         'P201,1,0.51,0,tree\n'
                                         'P202,3,0.36,0,tree\n'
                                         'P203,6,0.49,0,animal\n'
                                         'P204,3,0.48,0,tree\n'
                                         'P205,1,0.66,0,animal\n'
                                         'P206,1,0.28,0,tree\n'
                                         'P207,1,0.04,0,light\n'
                                         'P208,1,0.43,0,light\n'
                                         'P209,7,0.19,0,light\n'
                                         'P210,3,0.33,0,animal\n'
                                         'P211,1,0.24,0,animal\n'
                                         'P212,4,0.23,0,light\n'
                                         'P213,2,0.36,0,animal\n'
                                         'P214,3,0.43,0,animal\n'
                                         'P215,1,0.41,0,animal\n'
                                         'P216,3,0.32,0,animal\n'
                                         'P217,2,0.74,1,none\n'
                                         'P218,7,0.26,0,tree\n'
                                         'P219,3,0.87,1,none\n'
                                         'P220,4,0.11,0,tree\n'
                                         'P221,3,0.33,0,animal\n'
                                         'P222,4,0.19,0,animal\n'
                                         'P223,4,0.4,0,animal\n'
                                         'P224,2,0.21,0,rain\n'
                                         'P225,3,0.33,0,animal\n'
                                         'P226,7,,1,none\n'
                                         'P227,3,0.25,0,tree\n'
                                         'P228,5,0.08,0,animal\n'
                                         'P229,7,0.4,0,animal\n'
                                         'P230,5,0.1,0,rain\n'
                                         'P231,3,0.08,0,animal\n'
                                         'P232,3,0.2,0,animal\n'
                                         'P233,6,0.22,0,animal\n'
                                         'P234,1,0.32,0,animal\n'
                                         'P235,5,0.24,0,tree\n'
                                         'P236,3,0.59,0,tree\n'
                                         'P237,1,0.88,1,none\n'
                                         'P238,6,0.15,0,rain\n'
                                         'P239,5,0.57,0,animal\n'
                                         'P240,4,0.3,0,tree\n'
                                         'P241,1,0.51,0,tree\n'
                                         'P242,5,0.52,0,rain\n'
                                         'P243,2,0.01,0,animal\n'
                                         'P244,5,0.25,0,light\n'
                                         'P245,5,0.19,0,animal\n'
                                         'P246,6,0.79,1,none\n'
                                         'P247,5,0.44,0,animal\n'
                                         'P248,5,0.33,0,tree\n'
                                         'P249,4,0.16,0,tree\n'
                                         'P250,6,0.58,0,animal\n'
                                         'P251,4,0.53,0,rain\n'
                                         'P252,2,0.22,0,animal\n'
                                         'P253,7,0.18,0,tree\n'
                                         'P254,2,0.44,0,animal\n'
                                         'P255,7,0.56,0,animal\n'
                                         'P256,1,0.42,0,animal\n'
                                         'P257,2,0.1,0,animal\n'
                                         'P258,3,0.61,0,tree\n'
                                         'P259,2,0.94,1,none\n'
                                         'P260,5,0.4,0,tree\n'
                                         'P261,5,0.33,0,animal\n'
                                         'P262,7,0.18,0,animal\n'
                                         'P263,2,0.09,0,animal\n'
                                         'P264,1,0.29,0,animal\n'
                                         'P265,3,0.09,0,animal\n'
                                         'P266,3,0.63,0,light\n'
                                         'P267,2,0.45,0,animal\n'
                                         'P268,2,0.42,0,animal\n'
                                         'P269,6,0.63,1,none\n'
                                         'P270,7,0.31,0,tree\n'
                                         'P271,1,0.29,0,rain\n'
                                         'P272,2,0.2,0,rain\n'
                                         'P273,4,0.33,0,animal\n'
                                         'P274,1,0.91,1,none\n'
                                         'P275,2,0.34,1,none\n'
                                         'P276,3,0.07,0,animal\n'
                                         'P277,4,0.23,0,tree\n'
                                         'P278,1,0.22,0,animal\n'
                                         'P279,1,0.57,0,animal\n'
                                         'P280,2,0.24,0,tree\n'
                                         'P281,3,0.8,1,none\n'
                                         'P282,5,0.21,0,animal\n'
                                         'P283,5,0.97,1,none\n'
                                         'P284,6,,0,light\n'
                                         'P285,6,0.19,0,rain\n'
                                         'P286,3,0.64,0,animal\n'
                                         'P287,6,0.32,0,light\n'
                                         'P288,3,0.03,0,animal\n'
                                         'P289,7,0.12,0,rain\n'
                                         'P290,1,0.3,0,tree\n'
                                         'P291,4,0.34,0,light\n'
                                         'P292,7,0.52,0,animal\n'
                                         'P293,6,0.14,0,rain\n'
                                         'P294,3,0.26,0,tree\n'
                                         'P295,6,0.16,0,rain\n'
                                         'P296,7,0.39,0,rain\n'
                                         'P297,3,0.55,0,animal\n'
                                         'P298,5,,0,tree\n'
                                         'P299,4,0.45,0,animal\n'
                                         'P300,5,0.19,0,light\n'
                                         'P301,1,0.12,0,tree\n'
                                         'P302,1,0.4,0,animal\n'
                                         'P303,2,0.25,0,animal\n'
                                         'P304,1,0.2,0,tree\n'
                                         'P305,2,0.87,1,none\n'
                                         'P306,6,0.28,0,tree\n'
                                         'P307,7,0.79,1,none\n'
                                         'P308,3,0.94,1,none\n'
                                         'P309,7,0.18,0,tree\n'
                                         'P310,2,0.29,0,animal\n'
                                         'P311,7,0.44,0,animal\n'
                                         'P312,2,0.31,0,rain\n'
                                         'P313,7,0.2,0,animal\n'
                                         'P314,7,0.31,0,animal\n'
                                         'P315,3,0.14,0,animal\n'
                                         'P316,1,0.23,0,animal\n'
                                         'P317,3,0.52,1,none\n'
                                         'P318,4,0.63,0,rain\n'
                                         'P319,2,0.67,0,animal\n'
                                         'P320,1,0.39,0,animal\n'
                                         'P321,3,0.46,0,animal\n'
                                         'P322,7,0.29,0,animal\n'
                                         'P323,2,0.23,0,tree\n'
                                         'P324,1,0.67,1,none\n'
                                         'P325,6,0.48,0,tree\n'
                                         'P326,6,0.41,0,tree\n'
                                         'P327,3,0.1,0,animal\n'
                                         'P328,1,0.24,0,tree\n'
                                         'P329,7,0.63,0,animal\n'
                                         'P330,2,0.61,0,rain\n'
                                         'P331,6,0.28,0,light\n'
                                         'P332,5,0.39,0,light\n'
                                         'P333,7,0.24,0,animal\n'
                                         'P334,2,0.29,0,tree\n'
                                         'P335,4,0.52,0,rain\n'
                                         'P336,6,0.67,0,animal\n'
                                         'P337,7,0.15,0,tree\n'
                                         'P338,6,0.58,0,rain\n'
                                         'P339,7,0.42,0,rain\n'
                                         'P340,2,0.3,0,tree\n'
                                         'P341,2,0.26,0,light\n'
                                         'P342,5,0.61,0,rain\n'
                                         'P343,4,0.25,0,tree\n'
                                         'P344,7,0.48,0,animal\n'
                                         'P345,6,0.52,0,animal\n'
                                         'P346,7,0.42,0,light\n'
                                         'P347,1,0.31,0,tree\n'
                                         'P348,3,0.66,0,animal\n'
                                         'P349,5,0.37,0,rain\n'
                                         'P350,3,0.65,0,animal\n'
                                         'P351,4,0.21,0,animal\n'
                                         'P352,6,0.67,1,none\n'
                                         'P353,5,0.38,0,tree\n'
                                         'P354,7,0.08,0,animal\n'
                                         'P355,7,0.32,0,animal\n'
                                         'P356,5,0.49,0,tree\n'
                                         'P357,1,0.36,0,animal\n'
                                         'P358,2,0.62,0,tree\n'
                                         'P359,6,0.12,0,tree\n'
                                         'P360,7,0.38,0,rain\n'
                                         'P361,1,0.03,0,tree\n'
                                         'P362,5,0.4,0,rain\n'
                                         'P363,7,0.6,1,none\n'
                                         'P364,5,0.37,0,animal\n'
                                         'P365,4,0.11,0,light\n'
                                         'P366,2,0.51,0,tree\n'
                                         'P367,2,0.66,1,none\n'
                                         'P368,3,0.12,0,animal\n'
                                         'P369,7,0.64,0,animal\n'
                                         'P370,1,0.4,0,rain\n'
                                         'P371,4,0.24,0,animal\n'
                                         'P372,7,,0,tree\n'
                                         'P373,2,0.75,0,animal\n'
                                         'P374,6,0.25,0,rain\n'
                                         'P375,4,0.07,0,tree\n'
                                         'P376,7,0.53,0,animal\n'
                                         'P377,4,0.04,0,tree\n'
                                         'P378,2,0.7,0,animal\n'
                                         'P379,1,0.35,0,animal\n'
                                         'P380,2,0.14,0,tree\n'
                                         'P381,6,0.19,0,animal\n'
                                         'P382,2,0.14,0,animal\n'
                                         'P383,3,0.5,0,light\n'
                                         'P384,2,0.36,0,rain\n'
                                         'P385,7,0.25,0,animal\n'
                                         'P386,2,0.53,0,animal\n'
                                         'P387,4,0.43,0,rain\n'
                                         'P388,4,0.79,1,none\n'
                                         'P389,3,0.57,0,animal\n'
                                         'P390,6,0.22,0,animal\n'
                                         'P391,1,0.1,0,tree\n'
                                         'P392,7,0.54,0,animal\n'
                                         'P393,2,0.23,0,light\n'
                                         'P394,7,0.35,0,animal\n'
                                         'P395,7,0.43,0,tree\n'
                                         'P396,6,0.27,0,animal\n'
                                         'P397,2,0.26,0,tree\n'
                                         'P398,6,0.23,0,rain\n'
                                         'P399,6,0.88,1,none\n'
                                         'P400,3,0.55,1,none\n'
                                         'P401,7,0.65,1,none\n'
                                         'P402,1,0.47,1,none\n'
                                         'P403,1,0.46,0,tree\n'
                                         'P404,3,0.7,0,tree\n'
                                         'P405,1,0.13,0,tree\n'
                                         'P406,6,0.3,0,animal\n'
                                         'P407,6,0.49,0,tree\n'
                                         'P408,1,0.39,0,rain\n'
                                         'P409,5,0.18,0,rain\n'
                                         'P410,2,0.17,0,animal\n'
                                         'P411,5,0.33,0,rain\n'
                                         'P412,7,0.51,0,animal\n'
                                         'P413,5,0.1,0,light\n'
                                         'P414,4,0.27,0,animal\n'
                                         'P415,3,0.35,0,tree\n'
                                         'P416,1,0.18,0,animal\n'
                                         'P417,6,0.46,0,tree\n'
                                         'P418,4,0.35,0,tree\n'
                                         'P419,1,0.05,0,animal\n'
                                         'P420,7,0.29,0,tree\n',
                        'm7_results.csv': 'model,split_method,drone_recall_pct,bird_false_pct,tracks\n'
                                          'logistic,random,88.0,8.0,600\n'
                                          'logistic,clip,81.4,12.5,600\n'
                                          'random_forest,random,96.2,3.1,600\n'
                                          'random_forest,clip,84.5,9.8,600\n'
                                          'gradient_boosting,random,97.0,2.7,600\n'
                                          'gradient_boosting,clip,86.3,8.9,600\n'
                                          'gru,random,98.1,1.9,600\n'
                                          'gru,clip,83.0,11.2,600\n',
                        'm7_tracks.csv': 'track_id,clip_id,speed_mps,hover_ratio,size_px,label\n'
                                         'TR01,C01,12.7,0.58,31,drone\n'
                                         'TR02,C01,18.2,0.14,31,bird\n'
                                         'TR03,C01,7.5,0.5,21,drone\n'
                                         'TR04,C01,9.0,0.14,16,bird\n'
                                         'TR05,C02,14.3,0.27,27,drone\n'
                                         'TR06,C02,5.1,0.28,35,drone\n'
                                         'TR07,C02,6.1,0.34,19,drone\n'
                                         'TR08,C02,12.8,0.13,19,bird\n'
                                         'TR09,C03,12.6,0.4,22,drone\n'
                                         'TR10,C03,20.8,0.01,29,bird\n'
                                         'TR11,C03,8.6,0.03,10,bird\n'
                                         'TR12,C03,20.5,0.12,17,bird\n'
                                         'TR13,C04,18.8,0.07,39,bird\n'
                                         'TR14,C04,11.5,0.06,40,bird\n'
                                         'TR15,C04,10.9,0.39,11,drone\n'
                                         'TR16,C04,21.4,0.0,29,bird\n'
                                         'TR17,C05,10.1,0.06,27,bird\n'
                                         'TR18,C05,16.8,0.65,31,drone\n'
                                         'TR19,C05,8.2,0.11,30,bird\n'
                                         'TR20,C05,17.9,0.04,13,bird\n'
                                         'TR21,C06,13.0,0.69,29,drone\n'
                                         'TR22,C06,8.3,0.67,13,drone\n'
                                         'TR23,C06,7.5,0.13,36,bird\n'
                                         'TR24,C06,9.8,0.13,25,bird\n',
                        'm8_queue.csv': 'alarm_id,type,score,due_min,camera_pos,operator\n'
                                        'Q01,person,0.91,3,SITE-N03,OPERATOR-A\n'
                                        'Q02,animal,0.48,10,SITE-E11,OPERATOR-A\n'
                                        'Q03,drone,0.66,2,SITE-W02,OPERATOR-B\n'
                                        'Q04,video_loss,,1,SITE-S07,OPERATOR-B\n'
                                        'Q05,vehicle,0.72,5,SITE-N01,OPERATOR-A\n'
                                        'Q06,person,0.52,4,SITE-E04,OPERATOR-C\n'
                                        'Q07,tree,0.35,15,SITE-S02,OPERATOR-C\n'
                                        'Q08,drone,0.88,1,SITE-N08,OPERATOR-B\n'
                                        'Q09,vehicle,0.41,8,SITE-W05,OPERATOR-C\n',
                        'm9_signal.csv': 'camera,t_sec\n'
                                         'CAM1,0\n'
                                         'CAM1,10\n'
                                         'CAM1,20\n'
                                         'CAM1,30\n'
                                         'CAM1,40\n'
                                         'CAM1,50\n'
                                         'CAM1,60\n'
                                         'CAM1,70\n'
                                         'CAM1,80\n'
                                         'CAM1,90\n'
                                         'CAM1,100\n'
                                         'CAM1,110\n'
                                         'CAM1,120\n'
                                         'CAM1,130\n'
                                         'CAM1,140\n'
                                         'CAM1,150\n'
                                         'CAM1,160\n'
                                         'CAM1,170\n'
                                         'CAM1,180\n'
                                         'CAM1,190\n'
                                         'CAM1,200\n'
                                         'CAM1,210\n'
                                         'CAM1,220\n'
                                         'CAM1,230\n'
                                         'CAM1,240\n'
                                         'CAM1,250\n'
                                         'CAM1,260\n'
                                         'CAM1,270\n'
                                         'CAM1,280\n'
                                         'CAM1,290\n'
                                         'CAM1,300\n'
                                         'CAM1,310\n'
                                         'CAM1,320\n'
                                         'CAM1,330\n'
                                         'CAM1,340\n'
                                         'CAM1,350\n'
                                         'CAM1,360\n'
                                         'CAM1,370\n'
                                         'CAM1,380\n'
                                         'CAM1,390\n'
                                         'CAM1,400\n'
                                         'CAM1,410\n'
                                         'CAM1,420\n'
                                         'CAM1,430\n'
                                         'CAM1,440\n'
                                         'CAM1,450\n'
                                         'CAM1,460\n'
                                         'CAM1,470\n'
                                         'CAM1,480\n'
                                         'CAM1,490\n'
                                         'CAM1,500\n'
                                         'CAM1,510\n'
                                         'CAM1,520\n'
                                         'CAM1,530\n'
                                         'CAM1,540\n'
                                         'CAM1,550\n'
                                         'CAM1,560\n'
                                         'CAM1,570\n'
                                         'CAM1,580\n'
                                         'CAM1,590\n'
                                         'CAM1,600\n'
                                         'CAM1,610\n'
                                         'CAM1,620\n'
                                         'CAM1,630\n'
                                         'CAM1,640\n'
                                         'CAM1,650\n'
                                         'CAM1,660\n'
                                         'CAM1,670\n'
                                         'CAM1,680\n'
                                         'CAM1,690\n'
                                         'CAM1,700\n'
                                         'CAM1,710\n'
                                         'CAM1,720\n'
                                         'CAM1,730\n'
                                         'CAM1,740\n'
                                         'CAM1,750\n'
                                         'CAM1,760\n'
                                         'CAM1,770\n'
                                         'CAM1,780\n'
                                         'CAM1,790\n'
                                         'CAM1,800\n'
                                         'CAM1,810\n'
                                         'CAM1,820\n'
                                         'CAM1,830\n'
                                         'CAM1,840\n'
                                         'CAM1,850\n'
                                         'CAM1,860\n'
                                         'CAM1,870\n'
                                         'CAM1,880\n'
                                         'CAM1,890\n'
                                         'CAM1,900\n'
                                         'CAM1,910\n'
                                         'CAM1,920\n'
                                         'CAM1,930\n'
                                         'CAM1,940\n'
                                         'CAM1,950\n'
                                         'CAM1,960\n'
                                         'CAM1,970\n'
                                         'CAM1,980\n'
                                         'CAM1,990\n'
                                         'CAM1,1000\n'
                                         'CAM1,1010\n'
                                         'CAM1,1020\n'
                                         'CAM1,1030\n'
                                         'CAM1,1040\n'
                                         'CAM1,1050\n'
                                         'CAM1,1060\n'
                                         'CAM1,1070\n'
                                         'CAM1,1080\n'
                                         'CAM1,1090\n'
                                         'CAM1,1100\n'
                                         'CAM1,1110\n'
                                         'CAM1,1120\n'
                                         'CAM1,1130\n'
                                         'CAM1,1140\n'
                                         'CAM1,1150\n'
                                         'CAM1,1160\n'
                                         'CAM1,1170\n'
                                         'CAM1,1180\n'
                                         'CAM1,1190\n'
                                         'CAM1,1200\n'
                                         'CAM1,1210\n'
                                         'CAM1,1220\n'
                                         'CAM1,1230\n'
                                         'CAM1,1240\n'
                                         'CAM1,1250\n'
                                         'CAM1,1260\n'
                                         'CAM1,1270\n'
                                         'CAM1,1280\n'
                                         'CAM1,1290\n'
                                         'CAM1,1300\n'
                                         'CAM1,1310\n'
                                         'CAM1,1320\n'
                                         'CAM1,1330\n'
                                         'CAM1,1340\n'
                                         'CAM1,1350\n'
                                         'CAM1,1360\n'
                                         'CAM1,1370\n'
                                         'CAM1,1380\n'
                                         'CAM1,1390\n'
                                         'CAM1,1400\n'
                                         'CAM1,1410\n'
                                         'CAM1,1420\n'
                                         'CAM1,1430\n'
                                         'CAM1,1440\n'
                                         'CAM1,1450\n'
                                         'CAM1,1460\n'
                                         'CAM1,1470\n'
                                         'CAM1,1480\n'
                                         'CAM1,1490\n'
                                         'CAM1,1500\n'
                                         'CAM2,0\n'
                                         'CAM2,10\n'
                                         'CAM2,20\n'
                                         'CAM2,30\n'
                                         'CAM2,40\n'
                                         'CAM2,50\n'
                                         'CAM2,60\n'
                                         'CAM2,70\n'
                                         'CAM2,80\n'
                                         'CAM2,90\n'
                                         'CAM2,100\n'
                                         'CAM2,110\n'
                                         'CAM2,120\n'
                                         'CAM2,165\n'
                                         'CAM2,175\n'
                                         'CAM2,185\n'
                                         'CAM2,195\n'
                                         'CAM2,205\n'
                                         'CAM2,215\n'
                                         'CAM2,225\n'
                                         'CAM2,235\n'
                                         'CAM2,245\n'
                                         'CAM2,255\n'
                                         'CAM2,265\n'
                                         'CAM2,275\n'
                                         'CAM2,285\n'
                                         'CAM2,295\n'
                                         'CAM2,305\n'
                                         'CAM2,315\n'
                                         'CAM2,325\n'
                                         'CAM2,335\n'
                                         'CAM2,345\n'
                                         'CAM2,355\n'
                                         'CAM2,365\n'
                                         'CAM2,375\n'
                                         'CAM2,385\n'
                                         'CAM2,395\n'
                                         'CAM2,405\n'
                                         'CAM2,415\n'
                                         'CAM2,425\n'
                                         'CAM2,435\n'
                                         'CAM2,445\n'
                                         'CAM2,455\n'
                                         'CAM2,465\n'
                                         'CAM2,475\n'
                                         'CAM2,485\n'
                                         'CAM2,495\n'
                                         'CAM2,505\n'
                                         'CAM2,515\n'
                                         'CAM2,525\n'
                                         'CAM2,535\n'
                                         'CAM2,545\n'
                                         'CAM2,555\n'
                                         'CAM2,565\n'
                                         'CAM2,575\n'
                                         'CAM2,585\n'
                                         'CAM2,595\n'
                                         'CAM2,605\n'
                                         'CAM2,615\n'
                                         'CAM2,625\n'
                                         'CAM2,635\n'
                                         'CAM2,645\n'
                                         'CAM2,655\n'
                                         'CAM2,665\n'
                                         'CAM2,675\n'
                                         'CAM2,685\n'
                                         'CAM2,695\n'
                                         'CAM2,705\n'
                                         'CAM2,715\n'
                                         'CAM2,725\n'
                                         'CAM2,735\n'
                                         'CAM2,745\n'
                                         'CAM2,755\n'
                                         'CAM2,765\n'
                                         'CAM2,775\n'
                                         'CAM2,785\n'
                                         'CAM2,795\n'
                                         'CAM2,805\n'
                                         'CAM2,815\n'
                                         'CAM2,825\n'
                                         'CAM2,835\n'
                                         'CAM2,845\n'
                                         'CAM2,855\n'
                                         'CAM2,865\n'
                                         'CAM2,875\n'
                                         'CAM2,885\n'
                                         'CAM2,895\n'
                                         'CAM2,905\n'
                                         'CAM2,915\n'
                                         'CAM2,925\n'
                                         'CAM2,935\n'
                                         'CAM2,945\n'
                                         'CAM2,955\n'
                                         'CAM2,965\n'
                                         'CAM2,975\n'
                                         'CAM2,985\n'
                                         'CAM2,995\n'
                                         'CAM2,1005\n'
                                         'CAM2,1015\n'
                                         'CAM2,1025\n'
                                         'CAM2,1035\n'
                                         'CAM2,1045\n'
                                         'CAM2,1055\n'
                                         'CAM2,1065\n'
                                         'CAM2,1075\n'
                                         'CAM2,1085\n'
                                         'CAM2,1095\n'
                                         'CAM2,1105\n'
                                         'CAM2,1115\n'
                                         'CAM2,1125\n'
                                         'CAM2,1135\n'
                                         'CAM2,1145\n'
                                         'CAM2,1155\n'
                                         'CAM2,1165\n'
                                         'CAM2,1175\n'
                                         'CAM2,1185\n'
                                         'CAM2,1195\n'
                                         'CAM2,1205\n'
                                         'CAM2,1215\n'
                                         'CAM2,1225\n'
                                         'CAM2,1235\n'
                                         'CAM2,1245\n'
                                         'CAM2,1255\n'
                                         'CAM2,1265\n'
                                         'CAM2,1275\n'
                                         'CAM2,1285\n'
                                         'CAM2,1295\n'
                                         'CAM2,1305\n'
                                         'CAM2,1315\n'
                                         'CAM2,1325\n'
                                         'CAM2,1335\n'
                                         'CAM2,1345\n'
                                         'CAM2,1355\n'
                                         'CAM2,1365\n'
                                         'CAM2,1375\n'
                                         'CAM2,1385\n'
                                         'CAM2,1395\n'
                                         'CAM2,1405\n'
                                         'CAM2,1415\n'
                                         'CAM2,1425\n'
                                         'CAM2,1435\n'
                                         'CAM2,1445\n'
                                         'CAM2,1455\n'
                                         'CAM2,1465\n'
                                         'CAM2,1475\n'
                                         'CAM2,1485\n'
                                         'CAM2,1495\n'
                                         'CAM3,0\n'
                                         'CAM3,10\n'
                                         'CAM3,20\n'
                                         'CAM3,30\n'
                                         'CAM3,40\n'
                                         'CAM3,50\n'
                                         'CAM3,60\n'
                                         'CAM3,70\n'
                                         'CAM3,80\n'
                                         'CAM3,90\n'
                                         'CAM3,100\n'
                                         'CAM3,110\n'
                                         'CAM3,120\n'
                                         'CAM3,130\n'
                                         'CAM3,140\n'
                                         'CAM3,150\n'
                                         'CAM3,160\n'
                                         'CAM3,170\n'
                                         'CAM3,180\n'
                                         'CAM3,190\n'
                                         'CAM3,200\n'
                                         'CAM3,600\n'
                                         'CAM3,610\n'
                                         'CAM3,620\n'
                                         'CAM3,630\n'
                                         'CAM3,640\n'
                                         'CAM3,650\n'
                                         'CAM3,660\n'
                                         'CAM3,670\n'
                                         'CAM3,680\n'
                                         'CAM3,690\n'
                                         'CAM3,700\n'
                                         'CAM3,710\n'
                                         'CAM3,720\n'
                                         'CAM3,730\n'
                                         'CAM3,740\n'
                                         'CAM3,750\n'
                                         'CAM3,760\n'
                                         'CAM3,770\n'
                                         'CAM3,780\n'
                                         'CAM3,790\n'
                                         'CAM3,800\n'
                                         'CAM3,810\n'
                                         'CAM3,820\n'
                                         'CAM3,830\n'
                                         'CAM3,840\n'
                                         'CAM3,850\n'
                                         'CAM3,860\n'
                                         'CAM3,870\n'
                                         'CAM3,880\n'
                                         'CAM3,890\n'
                                         'CAM3,900\n'
                                         'CAM3,935\n'
                                         'CAM3,945\n'
                                         'CAM3,955\n'
                                         'CAM3,965\n'
                                         'CAM3,975\n'
                                         'CAM3,985\n'
                                         'CAM3,995\n'
                                         'CAM3,1005\n'
                                         'CAM3,1015\n'
                                         'CAM3,1025\n'
                                         'CAM3,1035\n'
                                         'CAM3,1045\n'
                                         'CAM3,1055\n'
                                         'CAM3,1065\n'
                                         'CAM3,1075\n'
                                         'CAM3,1085\n'
                                         'CAM3,1095\n'
                                         'CAM3,1105\n'
                                         'CAM3,1115\n'
                                         'CAM3,1125\n'
                                         'CAM3,1135\n'
                                         'CAM3,1145\n'
                                         'CAM3,1155\n'
                                         'CAM3,1165\n'
                                         'CAM3,1175\n'
                                         'CAM3,1185\n'
                                         'CAM3,1195\n'
                                         'CAM3,1205\n'
                                         'CAM3,1215\n'
                                         'CAM3,1225\n'
                                         'CAM3,1235\n'
                                         'CAM3,1245\n'
                                         'CAM3,1255\n'
                                         'CAM3,1265\n'
                                         'CAM3,1275\n'
                                         'CAM3,1285\n'
                                         'CAM3,1295\n'
                                         'CAM3,1305\n'
                                         'CAM3,1315\n'
                                         'CAM3,1325\n'
                                         'CAM3,1335\n'
                                         'CAM3,1345\n'
                                         'CAM3,1355\n'
                                         'CAM3,1365\n'
                                         'CAM3,1375\n'
                                         'CAM3,1385\n'
                                         'CAM3,1395\n'
                                         'CAM3,1405\n'
                                         'CAM3,1415\n'
                                         'CAM3,1425\n'
                                         'CAM3,1435\n'
                                         'CAM3,1445\n'
                                         'CAM3,1455\n'
                                         'CAM3,1465\n'
                                         'CAM3,1475\n'
                                         'CAM3,1485\n'
                                         'CAM3,1495\n'
                                         'CAM4,0\n'
                                         'CAM4,10\n'
                                         'CAM4,20\n'
                                         'CAM4,30\n'
                                         'CAM4,40\n'
                                         'CAM4,50\n'
                                         'CAM4,60\n'
                                         'CAM4,70\n'
                                         'CAM4,80\n'
                                         'CAM4,90\n'
                                         'CAM4,100\n'
                                         'CAM4,110\n'
                                         'CAM4,120\n'
                                         'CAM4,130\n'
                                         'CAM4,140\n'
                                         'CAM4,150\n'
                                         'CAM4,160\n'
                                         'CAM4,170\n'
                                         'CAM4,180\n'
                                         'CAM4,190\n'
                                         'CAM4,200\n'
                                         'CAM4,210\n'
                                         'CAM4,220\n'
                                         'CAM4,230\n'
                                         'CAM4,240\n'
                                         'CAM4,250\n'
                                         'CAM4,260\n'
                                         'CAM4,270\n'
                                         'CAM4,280\n'
                                         'CAM4,290\n'
                                         'CAM4,300\n'
                                         'CAM4,330\n'
                                         'CAM4,340\n'
                                         'CAM4,350\n'
                                         'CAM4,360\n'
                                         'CAM4,370\n'
                                         'CAM4,380\n'
                                         'CAM4,390\n'
                                         'CAM4,400\n'
                                         'CAM4,410\n'
                                         'CAM4,420\n'
                                         'CAM4,430\n'
                                         'CAM4,440\n'
                                         'CAM4,450\n'
                                         'CAM4,460\n'
                                         'CAM4,470\n'
                                         'CAM4,480\n'
                                         'CAM4,490\n'
                                         'CAM4,500\n'
                                         'CAM4,510\n'
                                         'CAM4,520\n'
                                         'CAM4,530\n'
                                         'CAM4,540\n'
                                         'CAM4,550\n'
                                         'CAM4,560\n'
                                         'CAM4,570\n'
                                         'CAM4,580\n'
                                         'CAM4,590\n'
                                         'CAM4,600\n'
                                         'CAM4,610\n'
                                         'CAM4,620\n'
                                         'CAM4,630\n'
                                         'CAM4,640\n'
                                         'CAM4,650\n'
                                         'CAM4,660\n'
                                         'CAM4,670\n'
                                         'CAM4,680\n'
                                         'CAM4,690\n'
                                         'CAM4,700\n'
                                         'CAM4,710\n'
                                         'CAM4,720\n'
                                         'CAM4,730\n'
                                         'CAM4,740\n'
                                         'CAM4,750\n'
                                         'CAM4,760\n'
                                         'CAM4,770\n'
                                         'CAM4,780\n'
                                         'CAM4,790\n'
                                         'CAM4,800\n'
                                         'CAM4,810\n'
                                         'CAM4,820\n'
                                         'CAM4,830\n'
                                         'CAM4,840\n'
                                         'CAM4,850\n'
                                         'CAM4,860\n'
                                         'CAM4,870\n'
                                         'CAM4,880\n'
                                         'CAM4,890\n'
                                         'CAM4,900\n'
                                         'CAM4,910\n'
                                         'CAM4,920\n'
                                         'CAM4,930\n'
                                         'CAM4,940\n'
                                         'CAM4,950\n'
                                         'CAM4,960\n'
                                         'CAM4,970\n'
                                         'CAM4,980\n'
                                         'CAM4,990\n'
                                         'CAM4,1000\n'
                                         'CAM4,1010\n'
                                         'CAM4,1020\n'
                                         'CAM4,1030\n'
                                         'CAM4,1040\n'
                                         'CAM4,1050\n'
                                         'CAM4,1060\n'
                                         'CAM4,1070\n'
                                         'CAM4,1080\n'
                                         'CAM4,1090\n'
                                         'CAM4,1100\n'
                                         'CAM4,1110\n'
                                         'CAM4,1120\n'
                                         'CAM4,1130\n'
                                         'CAM4,1140\n'
                                         'CAM4,1150\n'
                                         'CAM4,1160\n'
                                         'CAM4,1170\n'
                                         'CAM4,1180\n'
                                         'CAM4,1190\n'
                                         'CAM4,1200\n'
                                         'CAM4,1210\n'
                                         'CAM4,1220\n'
                                         'CAM4,1230\n'
                                         'CAM4,1240\n'
                                         'CAM4,1250\n'
                                         'CAM4,1260\n'
                                         'CAM4,1270\n'
                                         'CAM4,1280\n'
                                         'CAM4,1290\n'
                                         'CAM4,1300\n'
                                         'CAM4,1310\n'
                                         'CAM4,1320\n'
                                         'CAM4,1330\n'
                                         'CAM4,1340\n'
                                         'CAM4,1350\n'
                                         'CAM4,1360\n'
                                         'CAM4,1370\n'
                                         'CAM4,1380\n'
                                         'CAM4,1390\n'
                                         'CAM4,1400\n'
                                         'CAM4,1410\n'
                                         'CAM4,1420\n'
                                         'CAM4,1430\n'
                                         'CAM4,1440\n'
                                         'CAM4,1450\n'
                                         'CAM4,1460\n'
                                         'CAM4,1470\n'
                                         'CAM4,1480\n'
                                         'CAM4,1490\n'
                                         'CAM4,1500\n',
                        'm9_waves.csv': 'wave_id,alarms,handled_in_3min\n'
                                        'W01,1,1\n'
                                        'W02,2,2\n'
                                        'W03,6,3\n'
                                        'W04,1,1\n'
                                        'W05,4,4\n'
                                        'W06,3,3\n'
                                        'W07,3,1\n'
                                        'W08,2,2\n'
                                        'W09,1,1\n'
                                        'W10,1,1\n'
                                        'W11,2,1\n'
                                        'W12,2,2\n'
                                        'W13,2,2\n'
                                        'W14,1,1\n'
                                        'W15,2,2\n'
                                        'W16,3,3\n'
                                        'W17,3,3\n'
                                        'W18,2,2\n'
                                        'W19,2,2\n'
                                        'W20,3,3\n'
                                        'W21,1,1\n'
                                        'W22,1,1\n'
                                        'W23,2,2\n'
                                        'W24,1,1\n',
                        'm10_week.csv': 'record_id,period,condition,is_target,alarmed,handle_min\n'
                                        'H001,before,fog,0,0,\n'
                                        'H002,before,day,0,0,\n'
                                        'H003,before,night,1,1,6\n'
                                        'H004,before,night,1,1,2\n'
                                        'H005,before,day,0,0,\n'
                                        'H006,before,day,0,0,\n'
                                        'H007,before,day,0,0,\n'
                                        'H008,before,fog,0,0,\n'
                                        'H009,before,day,0,0,\n'
                                        'H010,before,night,1,1,4\n'
                                        'H011,before,day,1,1,6\n'
                                        'H012,before,day,1,1,5\n'
                                        'H013,before,night,1,1,3\n'
                                        'H014,before,day,0,1,9\n'
                                        'H015,before,day,0,0,\n'
                                        'H016,before,day,1,0,\n'
                                        'H017,before,day,0,0,\n'
                                        'H018,before,day,0,0,\n'
                                        'H019,before,day,0,0,\n'
                                        'H020,before,day,0,1,3\n'
                                        'H021,before,night,0,1,9\n'
                                        'H022,before,fog,0,0,\n'
                                        'H023,before,day,0,0,\n'
                                        'H024,before,night,1,1,1\n'
                                        'H025,before,night,0,0,\n'
                                        'H026,before,day,0,1,2\n'
                                        'H027,before,day,1,1,8\n'
                                        'H028,before,fog,0,0,\n'
                                        'H029,before,day,0,0,\n'
                                        'H030,before,night,1,1,8\n'
                                        'H031,before,night,0,1,4\n'
                                        'H032,before,day,0,0,\n'
                                        'H033,before,night,1,1,7\n'
                                        'H034,before,day,0,1,5\n'
                                        'H035,before,night,1,1,4\n'
                                        'H036,before,night,1,1,3\n'
                                        'H037,before,day,0,1,2\n'
                                        'H038,before,day,0,0,\n'
                                        'H039,before,day,0,0,\n'
                                        'H040,before,day,0,1,9\n'
                                        'H041,before,night,0,0,\n'
                                        'H042,before,night,1,0,\n'
                                        'H043,before,fog,0,0,\n'
                                        'H044,before,day,1,0,\n'
                                        'H045,before,day,0,0,\n'
                                        'H046,before,night,1,1,5\n'
                                        'H047,before,night,0,1,8\n'
                                        'H048,before,day,0,0,\n'
                                        'H049,before,night,0,1,1\n'
                                        'H050,before,night,0,1,4\n'
                                        'H051,before,night,0,0,\n'
                                        'H052,before,day,1,1,2\n'
                                        'H053,before,day,1,1,2\n'
                                        'H054,before,day,0,0,\n'
                                        'H055,before,night,0,1,7\n'
                                        'H056,before,day,1,1,4\n'
                                        'H057,before,night,1,1,7\n'
                                        'H058,before,fog,0,0,\n'
                                        'H059,before,day,1,1,8\n'
                                        'H060,before,night,0,0,\n'
                                        'H061,before,night,0,0,\n'
                                        'H062,before,day,1,1,1\n'
                                        'H063,before,day,0,0,\n'
                                        'H064,before,night,0,1,5\n'
                                        'H065,before,night,0,0,\n'
                                        'H066,before,night,0,0,\n'
                                        'H067,before,day,1,1,4\n'
                                        'H068,before,fog,1,1,1\n'
                                        'H069,before,night,0,1,1\n'
                                        'H070,before,day,0,1,7\n'
                                        'H071,before,day,0,0,\n'
                                        'H072,before,night,0,1,3\n'
                                        'H073,before,night,0,0,\n'
                                        'H074,before,night,0,0,\n'
                                        'H075,before,night,1,1,2\n'
                                        'H076,before,night,0,0,\n'
                                        'H077,before,day,1,0,\n'
                                        'H078,before,day,0,1,1\n'
                                        'H079,before,night,0,0,\n'
                                        'H080,before,day,0,0,\n'
                                        'H081,before,fog,0,0,\n'
                                        'H082,before,day,0,0,\n'
                                        'H083,before,night,0,1,9\n'
                                        'H084,before,night,0,0,\n'
                                        'H085,before,night,0,1,2\n'
                                        'H086,before,day,0,0,\n'
                                        'H087,before,night,0,0,\n'
                                        'H088,before,fog,0,1,3\n'
                                        'H089,before,night,0,0,\n'
                                        'H090,before,day,0,0,\n'
                                        'H091,before,day,1,1,2\n'
                                        'H092,before,day,0,1,1\n'
                                        'H093,before,day,0,0,\n'
                                        'H094,before,day,0,0,\n'
                                        'H095,before,day,1,1,1\n'
                                        'H096,before,night,0,1,4\n'
                                        'H097,before,night,0,1,3\n'
                                        'H098,before,fog,0,1,9\n'
                                        'H099,before,day,0,0,\n'
                                        'H100,before,night,0,0,\n'
                                        'H101,before,day,0,0,\n'
                                        'H102,before,day,0,0,\n'
                                        'H103,before,day,0,1,7\n'
                                        'H104,before,day,0,0,\n'
                                        'H105,before,fog,0,1,7\n'
                                        'H106,before,fog,0,0,\n'
                                        'H107,before,day,1,1,9\n'
                                        'H108,before,day,0,0,\n'
                                        'H109,before,fog,0,0,\n'
                                        'H110,before,day,0,0,\n'
                                        'H111,before,day,0,1,5\n'
                                        'H112,before,day,0,0,\n'
                                        'H113,before,day,0,0,\n'
                                        'H114,before,day,0,0,\n'
                                        'H115,before,day,1,1,8\n'
                                        'H116,before,night,0,0,\n'
                                        'H117,before,day,0,0,\n'
                                        'H118,before,fog,1,1,9\n'
                                        'H119,before,night,0,0,\n'
                                        'H120,before,day,1,1,8\n'
                                        'H121,before,day,1,1,2\n'
                                        'H122,before,night,0,1,6\n'
                                        'H123,before,day,1,1,8\n'
                                        'H124,before,night,0,0,\n'
                                        'H125,before,night,0,0,\n'
                                        'H126,before,day,0,0,\n'
                                        'H127,before,day,0,0,\n'
                                        'H128,before,night,0,0,\n'
                                        'H129,before,day,0,0,\n'
                                        'H130,before,day,0,0,\n'
                                        'H131,before,day,0,0,\n'
                                        'H132,before,night,0,0,\n'
                                        'H133,before,day,1,1,4\n'
                                        'H134,before,day,0,0,\n'
                                        'H135,before,night,1,1,9\n'
                                        'H136,before,day,1,1,1\n'
                                        'H137,before,fog,1,1,8\n'
                                        'H138,before,night,0,0,\n'
                                        'H139,before,day,0,0,\n'
                                        'H140,before,fog,1,1,8\n'
                                        'H141,before,night,1,0,\n'
                                        'H142,before,day,1,1,3\n'
                                        'H143,before,night,0,0,\n'
                                        'H144,before,night,0,0,\n'
                                        'H145,before,night,1,1,9\n'
                                        'H146,before,night,0,0,\n'
                                        'H147,before,day,0,0,\n'
                                        'H148,before,night,0,1,1\n'
                                        'H149,before,night,0,1,4\n'
                                        'H150,before,fog,1,0,\n'
                                        'H151,before,day,0,0,\n'
                                        'H152,before,day,1,1,3\n'
                                        'H153,before,day,0,1,2\n'
                                        'H154,before,night,0,0,\n'
                                        'H155,before,night,0,0,\n'
                                        'H156,before,night,1,1,3\n'
                                        'H157,before,night,0,0,\n'
                                        'H158,before,day,1,1,5\n'
                                        'H159,before,day,1,1,2\n'
                                        'H160,before,night,0,1,2\n'
                                        'H161,before,day,1,1,5\n'
                                        'H162,before,night,0,0,\n'
                                        'H163,before,day,0,0,\n'
                                        'H164,before,day,1,1,7\n'
                                        'H165,before,night,0,0,\n'
                                        'H166,before,night,0,0,\n'
                                        'H167,before,night,0,0,\n'
                                        'H168,before,fog,0,1,6\n'
                                        'H169,before,day,1,1,4\n'
                                        'H170,before,day,0,0,\n'
                                        'H171,before,day,0,0,\n'
                                        'H172,before,night,0,0,\n'
                                        'H173,before,day,0,1,9\n'
                                        'H174,before,day,0,1,7\n'
                                        'H175,before,night,0,0,\n'
                                        'H176,before,fog,0,0,\n'
                                        'H177,before,day,1,1,5\n'
                                        'H178,before,night,1,1,4\n'
                                        'H179,before,day,1,1,7\n'
                                        'H180,before,day,0,0,\n'
                                        'H181,before,night,0,1,9\n'
                                        'H182,before,night,0,1,4\n'
                                        'H183,before,fog,0,0,\n'
                                        'H184,before,day,1,1,9\n'
                                        'H185,before,night,0,1,9\n'
                                        'H186,before,night,0,0,\n'
                                        'H187,before,day,1,0,\n'
                                        'H188,before,fog,0,0,\n'
                                        'H189,before,day,0,0,\n'
                                        'H190,before,fog,0,1,2\n'
                                        'H191,before,day,0,0,\n'
                                        'H192,before,night,0,1,6\n'
                                        'H193,before,day,0,0,\n'
                                        'H194,before,day,0,1,2\n'
                                        'H195,before,day,1,1,7\n'
                                        'H196,before,night,0,0,\n'
                                        'H197,before,night,0,0,\n'
                                        'H198,before,day,0,0,\n'
                                        'H199,before,day,1,0,\n'
                                        'H200,before,night,0,0,\n'
                                        'H201,before,fog,1,0,\n'
                                        'H202,before,night,1,0,\n'
                                        'H203,before,day,0,0,\n'
                                        'H204,before,night,1,1,9\n'
                                        'H205,before,night,0,1,1\n'
                                        'H206,before,fog,0,1,6\n'
                                        'H207,before,night,0,0,\n'
                                        'H208,before,night,0,1,1\n'
                                        'H209,before,night,0,0,\n'
                                        'H210,before,day,1,1,9\n'
                                        'H211,before,day,1,1,9\n'
                                        'H212,before,day,1,1,2\n'
                                        'H213,before,night,0,1,6\n'
                                        'H214,before,night,0,1,1\n'
                                        'H215,before,night,0,0,\n'
                                        'H216,before,night,0,0,\n'
                                        'H217,before,fog,1,1,3\n'
                                        'H218,before,day,0,0,\n'
                                        'H219,before,day,1,1,1\n'
                                        'H220,before,fog,0,0,\n'
                                        'H221,before,day,0,0,\n'
                                        'H222,before,day,1,1,6\n'
                                        'H223,before,day,0,0,\n'
                                        'H224,before,day,0,1,5\n'
                                        'H225,before,day,0,1,5\n'
                                        'H226,before,night,0,0,\n'
                                        'H227,before,night,0,0,\n'
                                        'H228,before,day,0,0,\n'
                                        'H229,before,night,0,0,\n'
                                        'H230,before,night,0,0,\n'
                                        'H231,before,day,0,0,\n'
                                        'H232,before,night,0,0,\n'
                                        'H233,before,night,1,1,7\n'
                                        'H234,before,day,0,0,\n'
                                        'H235,before,night,0,1,2\n'
                                        'H236,before,night,1,1,6\n'
                                        'H237,before,day,1,1,2\n'
                                        'H238,before,night,0,0,\n'
                                        'H239,before,fog,0,1,6\n'
                                        'H240,before,night,1,1,9\n'
                                        'H241,before,night,1,1,3\n'
                                        'H242,before,night,1,1,8\n'
                                        'H243,before,night,0,0,\n'
                                        'H244,before,day,1,1,5\n'
                                        'H245,before,day,0,0,\n'
                                        'H246,before,night,1,1,6\n'
                                        'H247,before,day,0,0,\n'
                                        'H248,before,day,0,0,\n'
                                        'H249,before,day,1,1,3\n'
                                        'H250,before,night,0,0,\n'
                                        'H251,before,night,0,0,\n'
                                        'H252,before,night,0,1,6\n'
                                        'H253,before,day,1,1,5\n'
                                        'H254,before,day,0,0,\n'
                                        'H255,before,night,0,0,\n'
                                        'H256,before,night,1,1,7\n'
                                        'H257,before,day,0,0,\n'
                                        'H258,before,night,0,0,\n'
                                        'H259,before,fog,0,0,\n'
                                        'H260,before,night,0,0,\n'
                                        'H261,before,day,0,0,\n'
                                        'H262,before,day,0,0,\n'
                                        'H263,before,night,0,1,9\n'
                                        'H264,before,night,0,0,\n'
                                        'H265,before,day,0,0,\n'
                                        'H266,before,day,0,0,\n'
                                        'H267,before,fog,1,0,\n'
                                        'H268,before,night,0,0,\n'
                                        'H269,before,day,0,0,\n'
                                        'H270,before,night,1,1,3\n'
                                        'H271,before,day,0,0,\n'
                                        'H272,before,night,0,1,3\n'
                                        'H273,before,night,0,0,\n'
                                        'H274,before,day,1,1,2\n'
                                        'H275,before,day,0,0,\n'
                                        'H276,before,night,0,0,\n'
                                        'H277,before,night,0,1,1\n'
                                        'H278,before,night,1,0,\n'
                                        'H279,before,day,1,1,9\n'
                                        'H280,before,day,0,0,\n'
                                        'H281,before,day,1,1,1\n'
                                        'H282,before,day,0,0,\n'
                                        'H283,before,night,0,0,\n'
                                        'H284,before,day,0,0,\n'
                                        'H285,before,night,1,0,\n'
                                        'H286,before,night,0,1,1\n'
                                        'H287,before,day,0,0,\n'
                                        'H288,before,night,0,0,\n'
                                        'H289,before,fog,1,1,1\n'
                                        'H290,before,day,0,0,\n'
                                        'H291,before,day,0,0,\n'
                                        'H292,before,day,0,0,\n'
                                        'H293,before,day,1,1,5\n'
                                        'H294,before,day,0,1,2\n'
                                        'H295,before,day,0,0,\n'
                                        'H296,before,day,0,0,\n'
                                        'H297,before,night,1,1,5\n'
                                        'H298,before,day,0,0,\n'
                                        'H299,before,night,0,1,9\n'
                                        'H300,before,day,0,0,\n'
                                        'H301,before,day,1,1,6\n'
                                        'H302,before,night,1,0,\n'
                                        'H303,before,night,0,1,2\n'
                                        'H304,before,day,0,1,5\n'
                                        'H305,before,night,0,1,7\n'
                                        'H306,before,night,0,1,1\n'
                                        'H307,before,night,0,0,\n'
                                        'H308,before,day,0,0,\n'
                                        'H309,before,day,0,0,\n'
                                        'H310,before,night,1,1,7\n'
                                        'H311,before,night,0,0,\n'
                                        'H312,before,night,1,1,1\n'
                                        'H313,before,day,0,0,\n'
                                        'H314,before,night,0,1,2\n'
                                        'H315,before,night,0,0,\n'
                                        'H316,before,day,0,1,9\n'
                                        'H317,before,night,0,0,\n'
                                        'H318,before,day,0,1,9\n'
                                        'H319,before,night,1,1,1\n'
                                        'H320,before,night,0,0,\n'
                                        'H321,before,day,1,1,8\n'
                                        'H322,before,day,0,1,4\n'
                                        'H323,before,day,0,0,\n'
                                        'H324,before,night,0,0,\n'
                                        'H325,before,night,0,0,\n'
                                        'H326,before,day,0,0,\n'
                                        'H327,before,day,0,0,\n'
                                        'H328,before,day,0,0,\n'
                                        'H329,before,day,0,0,\n'
                                        'H330,before,night,0,0,\n'
                                        'H331,before,day,0,1,9\n'
                                        'H332,before,day,0,0,\n'
                                        'H333,before,day,0,0,\n'
                                        'H334,before,day,1,1,5\n'
                                        'H335,before,day,0,0,\n'
                                        'H336,before,day,0,0,\n'
                                        'H337,before,night,0,1,8\n'
                                        'H338,before,day,0,0,\n'
                                        'H339,before,day,0,0,\n'
                                        'H340,before,night,0,0,\n'
                                        'H341,before,day,0,0,\n'
                                        'H342,before,night,0,0,\n'
                                        'H343,before,night,1,0,\n'
                                        'H344,before,day,0,0,\n'
                                        'H345,before,day,0,0,\n'
                                        'H346,before,day,0,0,\n'
                                        'H347,before,night,0,0,\n'
                                        'H348,before,night,0,1,5\n'
                                        'H349,before,night,1,0,\n'
                                        'H350,before,day,0,0,\n'
                                        'H351,before,day,0,0,\n'
                                        'H352,before,night,1,1,1\n'
                                        'H353,before,day,0,0,\n'
                                        'H354,before,night,0,1,9\n'
                                        'H355,before,day,1,1,9\n'
                                        'H356,before,day,0,0,\n'
                                        'H357,before,night,0,1,6\n'
                                        'H358,before,day,0,0,\n'
                                        'H359,before,day,1,1,8\n'
                                        'H360,before,day,0,0,\n'
                                        'H361,after,night,1,1,3\n'
                                        'H362,after,day,0,0,\n'
                                        'H363,after,night,0,1,6\n'
                                        'H364,after,night,1,1,4\n'
                                        'H365,after,day,1,1,1\n'
                                        'H366,after,day,0,0,\n'
                                        'H367,after,night,0,1,3\n'
                                        'H368,after,night,0,0,\n'
                                        'H369,after,fog,1,0,\n'
                                        'H370,after,day,0,1,2\n'
                                        'H371,after,night,1,0,\n'
                                        'H372,after,day,0,0,\n'
                                        'H373,after,day,1,1,2\n'
                                        'H374,after,day,0,0,\n'
                                        'H375,after,night,1,0,\n'
                                        'H376,after,day,0,0,\n'
                                        'H377,after,day,0,0,\n'
                                        'H378,after,night,0,0,\n'
                                        'H379,after,night,0,0,\n'
                                        'H380,after,day,0,0,\n'
                                        'H381,after,day,0,0,\n'
                                        'H382,after,night,0,1,1\n'
                                        'H383,after,night,0,0,\n'
                                        'H384,after,day,1,1,9\n'
                                        'H385,after,day,1,1,5\n'
                                        'H386,after,fog,0,0,\n'
                                        'H387,after,day,0,0,\n'
                                        'H388,after,night,0,0,\n'
                                        'H389,after,night,0,0,\n'
                                        'H390,after,fog,0,0,\n'
                                        'H391,after,night,0,0,\n'
                                        'H392,after,night,0,1,1\n'
                                        'H393,after,day,0,0,\n'
                                        'H394,after,night,0,0,\n'
                                        'H395,after,day,0,0,\n'
                                        'H396,after,day,0,0,\n'
                                        'H397,after,day,1,1,8\n'
                                        'H398,after,night,0,0,\n'
                                        'H399,after,day,0,0,\n'
                                        'H400,after,day,0,0,\n'
                                        'H401,after,day,1,1,3\n'
                                        'H402,after,night,0,0,\n'
                                        'H403,after,night,1,1,2\n'
                                        'H404,after,day,0,0,\n'
                                        'H405,after,day,1,1,6\n'
                                        'H406,after,night,1,1,3\n'
                                        'H407,after,day,0,1,2\n'
                                        'H408,after,day,0,0,\n'
                                        'H409,after,night,0,0,\n'
                                        'H410,after,day,0,0,\n'
                                        'H411,after,day,0,0,\n'
                                        'H412,after,night,0,0,\n'
                                        'H413,after,day,0,0,\n'
                                        'H414,after,day,0,0,\n'
                                        'H415,after,night,1,1,4\n'
                                        'H416,after,day,1,1,5\n'
                                        'H417,after,night,0,0,\n'
                                        'H418,after,night,0,0,\n'
                                        'H419,after,day,1,1,1\n'
                                        'H420,after,day,0,0,\n'
                                        'H421,after,night,1,1,4\n'
                                        'H422,after,night,0,1,6\n'
                                        'H423,after,night,0,1,9\n'
                                        'H424,after,day,0,0,\n'
                                        'H425,after,fog,1,1,5\n'
                                        'H426,after,day,0,1,1\n'
                                        'H427,after,night,0,0,\n'
                                        'H428,after,day,0,0,\n'
                                        'H429,after,day,0,1,4\n'
                                        'H430,after,fog,0,1,4\n'
                                        'H431,after,day,0,1,5\n'
                                        'H432,after,day,0,0,\n'
                                        'H433,after,day,0,0,\n'
                                        'H434,after,day,1,1,5\n'
                                        'H435,after,fog,0,1,5\n'
                                        'H436,after,night,0,1,9\n'
                                        'H437,after,night,1,1,4\n'
                                        'H438,after,day,0,0,\n'
                                        'H439,after,night,1,1,7\n'
                                        'H440,after,day,0,0,\n'
                                        'H441,after,day,0,0,\n'
                                        'H442,after,night,0,1,2\n'
                                        'H443,after,day,1,0,\n'
                                        'H444,after,day,0,0,\n'
                                        'H445,after,night,0,0,\n'
                                        'H446,after,fog,0,0,\n'
                                        'H447,after,night,0,0,\n'
                                        'H448,after,day,1,1,6\n'
                                        'H449,after,night,1,1,3\n'
                                        'H450,after,day,0,1,9\n'
                                        'H451,after,day,0,0,\n'
                                        'H452,after,night,1,1,2\n'
                                        'H453,after,fog,0,1,3\n'
                                        'H454,after,night,1,1,5\n'
                                        'H455,after,night,0,0,\n'
                                        'H456,after,day,0,0,\n'
                                        'H457,after,day,0,1,3\n'
                                        'H458,after,day,0,0,\n'
                                        'H459,after,day,0,0,\n'
                                        'H460,after,night,0,0,\n'
                                        'H461,after,day,0,0,\n'
                                        'H462,after,day,0,0,\n'
                                        'H463,after,day,1,0,\n'
                                        'H464,after,day,1,1,4\n'
                                        'H465,after,day,1,1,2\n'
                                        'H466,after,night,0,1,6\n'
                                        'H467,after,day,0,1,3\n'
                                        'H468,after,night,1,1,9\n'
                                        'H469,after,fog,0,1,9\n'
                                        'H470,after,night,0,1,4\n'
                                        'H471,after,night,0,0,\n'
                                        'H472,after,day,0,0,\n'
                                        'H473,after,day,0,0,\n'
                                        'H474,after,day,0,0,\n'
                                        'H475,after,day,0,0,\n'
                                        'H476,after,night,0,0,\n'
                                        'H477,after,day,0,0,\n'
                                        'H478,after,night,0,1,3\n'
                                        'H479,after,day,0,0,\n'
                                        'H480,after,day,1,1,9\n'
                                        'H481,after,night,0,0,\n'
                                        'H482,after,night,1,1,4\n'
                                        'H483,after,day,0,0,\n'
                                        'H484,after,night,0,0,\n'
                                        'H485,after,night,0,0,\n'
                                        'H486,after,day,0,0,\n'
                                        'H487,after,day,0,0,\n'
                                        'H488,after,day,0,0,\n'
                                        'H489,after,night,1,1,2\n'
                                        'H490,after,night,1,0,\n'
                                        'H491,after,night,0,1,8\n'
                                        'H492,after,night,0,0,\n'
                                        'H493,after,night,1,1,3\n'
                                        'H494,after,day,1,1,9\n'
                                        'H495,after,fog,0,1,2\n'
                                        'H496,after,fog,0,0,\n'
                                        'H497,after,day,0,0,\n'
                                        'H498,after,day,1,1,3\n'
                                        'H499,after,night,1,0,\n'
                                        'H500,after,night,0,1,7\n'
                                        'H501,after,day,0,0,\n'
                                        'H502,after,day,1,1,1\n'
                                        'H503,after,night,0,0,\n'
                                        'H504,after,night,1,1,9\n'
                                        'H505,after,day,0,0,\n'
                                        'H506,after,night,0,0,\n'
                                        'H507,after,night,0,0,\n'
                                        'H508,after,night,0,0,\n'
                                        'H509,after,day,1,1,7\n'
                                        'H510,after,night,0,0,\n'
                                        'H511,after,fog,1,0,\n'
                                        'H512,after,night,0,0,\n'
                                        'H513,after,night,0,0,\n'
                                        'H514,after,night,1,1,1\n'
                                        'H515,after,night,0,0,\n'
                                        'H516,after,night,0,0,\n'
                                        'H517,after,day,0,0,\n'
                                        'H518,after,night,0,1,7\n'
                                        'H519,after,night,0,0,\n'
                                        'H520,after,day,0,0,\n'
                                        'H521,after,day,0,0,\n'
                                        'H522,after,day,0,0,\n'
                                        'H523,after,fog,0,1,5\n'
                                        'H524,after,day,0,0,\n'
                                        'H525,after,night,0,0,\n'
                                        'H526,after,night,0,1,4\n'
                                        'H527,after,night,0,0,\n'
                                        'H528,after,day,0,0,\n'
                                        'H529,after,night,0,0,\n'
                                        'H530,after,night,0,0,\n'
                                        'H531,after,night,0,0,\n'
                                        'H532,after,night,0,0,\n'
                                        'H533,after,night,0,0,\n'
                                        'H534,after,day,1,1,9\n'
                                        'H535,after,night,0,0,\n'
                                        'H536,after,day,0,0,\n'
                                        'H537,after,day,1,0,\n'
                                        'H538,after,night,0,0,\n'
                                        'H539,after,night,0,1,8\n'
                                        'H540,after,night,0,0,\n'
                                        'H541,after,fog,0,0,\n'
                                        'H542,after,night,0,1,2\n'
                                        'H543,after,fog,0,0,\n'
                                        'H544,after,fog,0,0,\n'
                                        'H545,after,day,0,0,\n'
                                        'H546,after,day,0,0,\n'
                                        'H547,after,fog,0,0,\n'
                                        'H548,after,day,1,1,1\n'
                                        'H549,after,day,0,0,\n'
                                        'H550,after,night,0,1,2\n'
                                        'H551,after,day,1,1,3\n'
                                        'H552,after,night,0,0,\n'
                                        'H553,after,day,1,1,7\n'
                                        'H554,after,fog,0,0,\n'
                                        'H555,after,day,0,0,\n'
                                        'H556,after,day,1,1,9\n'
                                        'H557,after,day,0,0,\n'
                                        'H558,after,night,0,0,\n'
                                        'H559,after,night,0,0,\n'
                                        'H560,after,night,0,1,2\n'
                                        'H561,after,fog,0,1,9\n'
                                        'H562,after,night,0,0,\n'
                                        'H563,after,day,0,0,\n'
                                        'H564,after,day,0,0,\n'
                                        'H565,after,night,0,0,\n'
                                        'H566,after,night,0,1,7\n'
                                        'H567,after,day,0,0,\n'
                                        'H568,after,night,0,0,\n'
                                        'H569,after,night,1,1,4\n'
                                        'H570,after,night,1,1,9\n'
                                        'H571,after,day,1,1,1\n'
                                        'H572,after,night,0,0,\n'
                                        'H573,after,day,0,0,\n'
                                        'H574,after,day,0,0,\n'
                                        'H575,after,day,1,1,3\n'
                                        'H576,after,day,0,1,4\n'
                                        'H577,after,day,1,1,8\n'
                                        'H578,after,night,0,0,\n'
                                        'H579,after,night,0,1,6\n'
                                        'H580,after,day,0,0,\n'
                                        'H581,after,night,0,1,8\n'
                                        'H582,after,night,0,0,\n'
                                        'H583,after,night,0,0,\n'
                                        'H584,after,day,0,0,\n'
                                        'H585,after,night,0,0,\n'
                                        'H586,after,night,0,0,\n'
                                        'H587,after,night,0,0,\n'
                                        'H588,after,night,0,0,\n'
                                        'H589,after,night,0,0,\n'
                                        'H590,after,night,0,0,\n'
                                        'H591,after,night,1,1,4\n'
                                        'H592,after,day,1,1,1\n'
                                        'H593,after,day,0,0,\n'
                                        'H594,after,day,0,0,\n'
                                        'H595,after,day,0,0,\n'
                                        'H596,after,day,0,0,\n'
                                        'H597,after,night,0,0,\n'
                                        'H598,after,night,0,0,\n'
                                        'H599,after,day,0,0,\n'
                                        'H600,after,day,0,0,\n'
                                        'H601,after,night,0,0,\n'
                                        'H602,after,night,1,1,3\n'
                                        'H603,after,day,0,0,\n'
                                        'H604,after,fog,0,0,\n'
                                        'H605,after,fog,0,0,\n'
                                        'H606,after,night,1,1,5\n'
                                        'H607,after,day,1,1,8\n'
                                        'H608,after,fog,0,1,8\n'
                                        'H609,after,day,1,1,6\n'
                                        'H610,after,day,1,1,7\n'
                                        'H611,after,day,0,0,\n'
                                        'H612,after,day,0,0,\n'
                                        'H613,after,day,0,0,\n'
                                        'H614,after,fog,0,1,1\n'
                                        'H615,after,day,1,1,5\n'
                                        'H616,after,day,0,1,5\n'
                                        'H617,after,night,1,1,4\n'
                                        'H618,after,day,0,0,\n'
                                        'H619,after,day,0,0,\n'
                                        'H620,after,night,0,0,\n'
                                        'H621,after,night,0,0,\n'
                                        'H622,after,night,0,0,\n'
                                        'H623,after,night,0,1,5\n'
                                        'H624,after,night,0,0,\n'
                                        'H625,after,day,0,0,\n'
                                        'H626,after,night,1,1,7\n'
                                        'H627,after,night,0,0,\n'
                                        'H628,after,day,0,1,4\n'
                                        'H629,after,night,0,0,\n'
                                        'H630,after,day,0,0,\n'
                                        'H631,after,night,0,1,8\n'
                                        'H632,after,day,0,0,\n'
                                        'H633,after,day,0,0,\n'
                                        'H634,after,day,0,0,\n'
                                        'H635,after,day,0,0,\n'
                                        'H636,after,night,0,1,9\n'
                                        'H637,after,day,0,0,\n'
                                        'H638,after,night,0,0,\n'
                                        'H639,after,day,0,0,\n'
                                        'H640,after,night,0,0,\n'
                                        'H641,after,day,0,0,\n'
                                        'H642,after,day,0,0,\n'
                                        'H643,after,night,0,0,\n'
                                        'H644,after,night,0,0,\n'
                                        'H645,after,day,1,1,7\n'
                                        'H646,after,night,0,0,\n'
                                        'H647,after,night,1,0,\n'
                                        'H648,after,night,0,1,5\n'
                                        'H649,after,day,1,1,8\n'
                                        'H650,after,fog,0,0,\n'
                                        'H651,after,day,0,0,\n'
                                        'H652,after,fog,0,0,\n'
                                        'H653,after,day,0,0,\n'
                                        'H654,after,day,0,1,8\n'
                                        'H655,after,day,1,1,5\n'
                                        'H656,after,day,1,1,5\n'
                                        'H657,after,day,0,0,\n'
                                        'H658,after,day,1,1,7\n'
                                        'H659,after,day,1,1,7\n'
                                        'H660,after,day,0,0,\n'
                                        'H661,after,day,0,0,\n'
                                        'H662,after,night,0,0,\n'
                                        'H663,after,day,0,0,\n'
                                        'H664,after,night,0,0,\n'
                                        'H665,after,night,0,0,\n'
                                        'H666,after,day,1,0,\n'
                                        'H667,after,day,0,0,\n'
                                        'H668,after,day,1,1,1\n'
                                        'H669,after,night,1,1,3\n'
                                        'H670,after,night,0,0,\n'
                                        'H671,after,day,0,0,\n'
                                        'H672,after,day,0,1,7\n'
                                        'H673,after,day,0,0,\n'
                                        'H674,after,day,0,0,\n'
                                        'H675,after,day,1,1,8\n'
                                        'H676,after,fog,1,1,3\n'
                                        'H677,after,day,1,0,\n'
                                        'H678,after,day,0,0,\n'
                                        'H679,after,day,0,1,6\n'
                                        'H680,after,day,0,0,\n'
                                        'H681,after,night,0,0,\n'
                                        'H682,after,day,1,1,7\n'
                                        'H683,after,night,1,1,1\n'
                                        'H684,after,day,0,0,\n'
                                        'H685,after,night,0,1,6\n'
                                        'H686,after,day,1,1,1\n'
                                        'H687,after,night,0,0,\n'
                                        'H688,after,fog,0,0,\n'
                                        'H689,after,day,1,1,5\n'
                                        'H690,after,night,0,0,\n'
                                        'H691,after,day,0,0,\n'
                                        'H692,after,night,0,1,8\n'
                                        'H693,after,fog,0,0,\n'
                                        'H694,after,night,0,0,\n'
                                        'H695,after,night,0,1,4\n'
                                        'H696,after,day,0,0,\n'
                                        'H697,after,day,0,0,\n'
                                        'H698,after,day,0,0,\n'
                                        'H699,after,day,0,0,\n'
                                        'H700,after,day,0,0,\n'
                                        'H701,after,day,0,0,\n'
                                        'H702,after,night,0,0,\n'
                                        'H703,after,day,0,0,\n'
                                        'H704,after,day,1,1,6\n'
                                        'H705,after,night,0,0,\n'
                                        'H706,after,day,0,0,\n'
                                        'H707,after,night,0,0,\n'
                                        'H708,after,night,0,1,6\n'
                                        'H709,after,day,1,1,2\n'
                                        'H710,after,night,0,0,\n'
                                        'H711,after,day,0,0,\n'
                                        'H712,after,day,0,0,\n'
                                        'H713,after,day,0,0,\n'
                                        'H714,after,day,0,0,\n'
                                        'H715,after,day,1,1,4\n'
                                        'H716,after,day,0,0,\n'
                                        'H717,after,day,1,1,6\n'
                                        'H718,after,day,0,1,4\n'
                                        'H719,after,fog,0,1,3\n'
                                        'H720,after,day,0,0,\n'},
             'figures': {'f_loop': '<svg viewBox="0 0 320 156" xmlns="http://www.w3.org/2000/svg" font-family="sans-serif" font-size="10" '
                                   'fill="#111827"><defs><marker id="tfa" viewBox="0 0 10 10" refX="8" refY="5" markerWidth="6" markerHeight="6" '
                                   'orient="auto"><path d="M0,0 L10,5 L0,10 z" fill="#475569"/></marker></defs><ellipse cx="160.0" cy="70" rx="112" '
                                   'ry="48" fill="none" stroke="#e5e7eb" stroke-width="6"/><path d="M198.6,24.9 L248.1,40.3" stroke="#2563eb" '
                                   'stroke-width="2" marker-end="url(#tfa)" fill="none"/><path d="M271.9,71.8 L253.0,96.7" stroke="#2563eb" '
                                   'stroke-width="2" marker-end="url(#tfa)" fill="none"/><path d="M190.6,116.2 L129.4,116.2" stroke="#2563eb" '
                                   'stroke-width="2" marker-end="url(#tfa)" fill="none"/><path d="M67.0,96.7 L48.1,71.8" stroke="#2563eb" '
                                   'stroke-width="2" marker-end="url(#tfa)" fill="none"/><path d="M71.9,40.3 L121.4,24.9" stroke="#2563eb" '
                                   'stroke-width="2" marker-end="url(#tfa)" fill="none"/><rect x="116.0" y="10.0" width="88.0" height="24.0" rx="7" '
                                   'fill="#dbeafe" stroke="#94a3b8"/><text x="160.0" y="25.5" font-size="9" text-anchor="middle" fill="#111827" '
                                   'font-weight="bold">요청</text><rect x="222.5" y="43.2" width="88.0" height="24.0" rx="7" fill="#dbeafe" '
                                   'stroke="#94a3b8"/><text x="266.5" y="58.7" font-size="9" text-anchor="middle" fill="#111827" '
                                   'font-weight="bold">생성</text><rect x="181.8" y="96.8" width="88.0" height="24.0" rx="7" fill="#dbeafe" '
                                   'stroke="#94a3b8"/><text x="225.8" y="112.3" font-size="9" text-anchor="middle" fill="#111827" '
                                   'font-weight="bold">실행</text><rect x="50.2" y="96.8" width="88.0" height="24.0" rx="7" fill="#dbeafe" '
                                   'stroke="#94a3b8"/><text x="94.2" y="112.3" font-size="9" text-anchor="middle" fill="#111827" '
                                   'font-weight="bold">확인</text><rect x="9.5" y="43.2" width="88.0" height="24.0" rx="7" fill="#dbeafe" '
                                   'stroke="#94a3b8"/><text x="53.5" y="58.7" font-size="9" text-anchor="middle" fill="#111827" '
                                   'font-weight="bold">수정 요청</text><text x="160.0" y="74.0" font-size="10.5" text-anchor="middle" fill="#475569" '
                                   'font-weight="bold">바이브 코딩</text><text x="160.0" y="151.0" font-size="9" text-anchor="middle" fill="#475569">확인하고 '
                                   '다시 요청하는 일이 핵심이에요</text></svg>',
                         'f_four': '<svg viewBox="0 0 320 120" xmlns="http://www.w3.org/2000/svg" font-family="sans-serif" font-size="10" '
                                   'fill="#111827"><defs><marker id="tfa" viewBox="0 0 10 10" refX="8" refY="5" markerWidth="6" markerHeight="6" '
                                   'orient="auto"><path d="M0,0 L10,5 L0,10 z" fill="#475569"/></marker></defs><rect x="6.0" y="8" width="154.0" '
                                   'height="18" fill="#e5e7eb" stroke="#fff"/><text x="83.0" y="20.5" font-size="9" text-anchor="middle" '
                                   'fill="#111827" font-weight="bold">칸</text><rect x="160.0" y="8" width="154.0" height="18" fill="#e5e7eb" '
                                   'stroke="#fff"/><text x="237.0" y="20.5" font-size="9" text-anchor="middle" fill="#111827" font-weight="bold">무엇을 '
                                   '적나</text><rect x="6.0" y="26" width="154.0" height="18" fill="#fff" stroke="#e5e7eb"/><text x="83.0" y="38.5" '
                                   'font-size="9" text-anchor="middle" fill="#111827" font-weight="bold">맥락</text><rect x="160.0" y="26" '
                                   'width="154.0" height="18" fill="#fff" stroke="#e5e7eb"/><text x="237.0" y="38.5" font-size="9" '
                                   'text-anchor="middle" fill="#111827">왜 필요한지</text><rect x="6.0" y="44" width="154.0" height="18" fill="#fff" '
                                   'stroke="#e5e7eb"/><text x="83.0" y="56.5" font-size="9" text-anchor="middle" fill="#111827" '
                                   'font-weight="bold">입력</text><rect x="160.0" y="44" width="154.0" height="18" fill="#fff" stroke="#e5e7eb"/><text '
                                   'x="237.0" y="56.5" font-size="9" text-anchor="middle" fill="#111827">파일·열·값</text><rect x="6.0" y="62" '
                                   'width="154.0" height="18" fill="#fff" stroke="#e5e7eb"/><text x="83.0" y="74.5" font-size="9" '
                                   'text-anchor="middle" fill="#111827" font-weight="bold">할 일</text><rect x="160.0" y="62" width="154.0" '
                                   'height="18" fill="#fff" stroke="#e5e7eb"/><text x="237.0" y="74.5" font-size="9" text-anchor="middle" '
                                   'fill="#111827">규칙을 숫자로</text><rect x="6.0" y="80" width="154.0" height="18" fill="#fff" stroke="#e5e7eb"/><text '
                                   'x="83.0" y="92.5" font-size="9" text-anchor="middle" fill="#111827" font-weight="bold">출력 형식</text><rect '
                                   'x="160.0" y="80" width="154.0" height="18" fill="#fff" stroke="#e5e7eb"/><text x="237.0" y="92.5" font-size="9" '
                                   'text-anchor="middle" fill="#111827">받고 싶은 모양</text><text x="160.0" y="115.0" font-size="9" text-anchor="middle" '
                                   'fill="#475569">네 칸을 채우면 AI가 추측할 일이 줄어요</text></svg>',
                         'f_kinds': '<svg viewBox="0 0 320 102" xmlns="http://www.w3.org/2000/svg" font-family="sans-serif" font-size="10" '
                                    'fill="#111827"><defs><marker id="tfa" viewBox="0 0 10 10" refX="8" refY="5" markerWidth="6" markerHeight="6" '
                                    'orient="auto"><path d="M0,0 L10,5 L0,10 z" fill="#475569"/></marker></defs><rect x="6.0" y="8" width="102.7" '
                                    'height="18" fill="#e5e7eb" stroke="#fff"/><text x="57.3" y="20.5" font-size="9" text-anchor="middle" '
                                    'fill="#111827" font-weight="bold">구분</text><rect x="108.7" y="8" width="102.7" height="18" fill="#e5e7eb" '
                                    'stroke="#fff"/><text x="160.0" y="20.5" font-size="9" text-anchor="middle" fill="#111827" '
                                    'font-weight="bold">원인</text><rect x="211.3" y="8" width="102.7" height="18" fill="#e5e7eb" stroke="#fff"/><text '
                                    'x="262.7" y="20.5" font-size="9" text-anchor="middle" fill="#111827" font-weight="bold">대책</text><rect x="6.0" '
                                    'y="26" width="102.7" height="18" fill="#fff" stroke="#e5e7eb"/><text x="57.3" y="38.5" font-size="9" '
                                    'text-anchor="middle" fill="#111827" font-weight="bold">경보 대상</text><rect x="108.7" y="26" width="102.7" '
                                    'height="18" fill="#fff" stroke="#e5e7eb"/><text x="160.0" y="38.5" font-size="9" text-anchor="middle" '
                                    'fill="#111827">사람·차량·드론</text><rect x="211.3" y="26" width="102.7" height="18" fill="#fff" '
                                    'stroke="#e5e7eb"/><text x="262.7" y="38.5" font-size="9" text-anchor="middle" fill="#111827">확인</text><rect '
                                    'x="6.0" y="44" width="102.7" height="18" fill="#fff" stroke="#e5e7eb"/><text x="57.3" y="56.5" font-size="9" '
                                    'text-anchor="middle" fill="#111827" font-weight="bold">불필요 경보</text><rect x="108.7" y="44" width="102.7" '
                                    'height="18" fill="#fff" stroke="#e5e7eb"/><text x="160.0" y="56.5" font-size="9" text-anchor="middle" '
                                    'fill="#111827">동물·나무·비</text><rect x="211.3" y="44" width="102.7" height="18" fill="#fff" '
                                    'stroke="#e5e7eb"/><text x="262.7" y="56.5" font-size="9" text-anchor="middle" fill="#111827">영역·임계값</text><rect '
                                    'x="6.0" y="62" width="102.7" height="18" fill="#fff" stroke="#e5e7eb"/><text x="57.3" y="74.5" font-size="9" '
                                    'text-anchor="middle" fill="#111827" font-weight="bold">오경보</text><rect x="108.7" y="62" width="102.7" '
                                    'height="18" fill="#fff" stroke="#e5e7eb"/><text x="160.0" y="74.5" font-size="9" text-anchor="middle" '
                                    'fill="#111827">장비 이상·잡음</text><rect x="211.3" y="62" width="102.7" height="18" fill="#fff" '
                                    'stroke="#e5e7eb"/><text x="262.7" y="74.5" font-size="9" text-anchor="middle" fill="#111827">장비 점검</text><text '
                                    'x="160.0" y="97.0" font-size="9" text-anchor="middle" fill="#475569">원인이 다르면 대책도 달라요</text></svg>',
                         'f_recall': '<svg viewBox="0 0 320 84" xmlns="http://www.w3.org/2000/svg" font-family="sans-serif" font-size="10" '
                                     'fill="#111827"><defs><marker id="tfa" viewBox="0 0 10 10" refX="8" refY="5" markerWidth="6" markerHeight="6" '
                                     'orient="auto"><path d="M0,0 L10,5 L0,10 z" fill="#475569"/></marker></defs><rect x="6.0" y="8" width="102.7" '
                                     'height="18" fill="#e5e7eb" stroke="#fff"/><text x="57.3" y="20.5" font-size="9" text-anchor="middle" '
                                     'fill="#111827" font-weight="bold">실제 대상</text><rect x="108.7" y="8" width="102.7" height="18" fill="#e5e7eb" '
                                     'stroke="#fff"/><text x="160.0" y="20.5" font-size="9" text-anchor="middle" fill="#111827" '
                                     'font-weight="bold">잡음</text><rect x="211.3" y="8" width="102.7" height="18" fill="#e5e7eb" '
                                     'stroke="#fff"/><text x="262.7" y="20.5" font-size="9" text-anchor="middle" fill="#111827" '
                                     'font-weight="bold">재현율</text><rect x="6.0" y="26" width="102.7" height="18" fill="#fff" '
                                     'stroke="#e5e7eb"/><text x="57.3" y="38.5" font-size="9" text-anchor="middle" fill="#111827" '
                                     'font-weight="bold">10</text><rect x="108.7" y="26" width="102.7" height="18" fill="#fff" '
                                     'stroke="#e5e7eb"/><text x="160.0" y="38.5" font-size="9" text-anchor="middle" fill="#111827">9</text><rect '
                                     'x="211.3" y="26" width="102.7" height="18" fill="#fff" stroke="#e5e7eb"/><text x="262.7" y="38.5" '
                                     'font-size="9" text-anchor="middle" fill="#111827">90%</text><rect x="6.0" y="44" width="102.7" height="18" '
                                     'fill="#fff" stroke="#e5e7eb"/><text x="57.3" y="56.5" font-size="9" text-anchor="middle" fill="#111827" '
                                     'font-weight="bold">10</text><rect x="108.7" y="44" width="102.7" height="18" fill="#fff" '
                                     'stroke="#e5e7eb"/><text x="160.0" y="56.5" font-size="9" text-anchor="middle" fill="#111827">7</text><rect '
                                     'x="211.3" y="44" width="102.7" height="18" fill="#fff" stroke="#e5e7eb"/><text x="262.7" y="56.5" '
                                     'font-size="9" text-anchor="middle" fill="#111827">70%</text><text x="160.0" y="79.0" font-size="9" '
                                     'text-anchor="middle" fill="#475569">실제 대상 가운데 잡은 비율이에요</text></svg>',
                         'f_delay': '<svg viewBox="0 0 320 92" xmlns="http://www.w3.org/2000/svg" font-family="sans-serif" font-size="10" '
                                    'fill="#111827"><defs><marker id="tfa" viewBox="0 0 10 10" refX="8" refY="5" markerWidth="6" markerHeight="6" '
                                    'orient="auto"><path d="M0,0 L10,5 L0,10 z" fill="#475569"/></marker></defs><rect x="8.0" y="14.0" width="92.0" '
                                    'height="46.0" rx="7" fill="#dbeafe" stroke="#94a3b8"/><text x="54.0" y="34.0" font-size="10" '
                                    'text-anchor="middle" fill="#111827" font-weight="bold">대상 등장</text><text x="54.0" y="47.0" font-size="8.5" '
                                    'text-anchor="middle" fill="#475569">0초</text><path d="M101.0,37.0 L113.0,37.0" stroke="#475569" '
                                    'stroke-width="1.6" marker-end="url(#tfa)" fill="none"/><rect x="114.0" y="14.0" width="92.0" height="46.0" '
                                    'rx="7" fill="#dbeafe" stroke="#94a3b8"/><text x="160.0" y="40.5" font-size="10" text-anchor="middle" '
                                    'fill="#111827" font-weight="bold">영상 처리</text><path d="M207.0,37.0 L219.0,37.0" stroke="#475569" '
                                    'stroke-width="1.6" marker-end="url(#tfa)" fill="none"/><rect x="220.0" y="14.0" width="92.0" height="46.0" '
                                    'rx="7" fill="#dbeafe" stroke="#94a3b8"/><text x="266.0" y="34.0" font-size="10" text-anchor="middle" '
                                    'fill="#111827" font-weight="bold">경보</text><text x="266.0" y="47.0" font-size="8.5" text-anchor="middle" '
                                    'fill="#475569">30초</text><text x="160.0" y="87.0" font-size="9" text-anchor="middle" fill="#475569">나타난 뒤 울릴 '
                                    '때까지의 시간이에요</text></svg>',
                         'f_clean': '<svg viewBox="0 0 320 92" xmlns="http://www.w3.org/2000/svg" font-family="sans-serif" font-size="10" '
                                    'fill="#111827"><defs><marker id="tfa" viewBox="0 0 10 10" refX="8" refY="5" markerWidth="6" markerHeight="6" '
                                    'orient="auto"><path d="M0,0 L10,5 L0,10 z" fill="#475569"/></marker></defs><rect x="8.0" y="14.0" width="92.0" '
                                    'height="46.0" rx="7" fill="#dbeafe" stroke="#94a3b8"/><text x="54.0" y="34.0" font-size="10" '
                                    'text-anchor="middle" fill="#111827" font-weight="bold">1min</text><text x="54.0" y="47.0" font-size="8.5" '
                                    'text-anchor="middle" fill="#475569">분</text><path d="M101.0,37.0 L113.0,37.0" stroke="#475569" '
                                    'stroke-width="1.6" marker-end="url(#tfa)" fill="none"/><rect x="114.0" y="14.0" width="92.0" height="46.0" '
                                    'rx="7" fill="#dbeafe" stroke="#94a3b8"/><text x="160.0" y="34.0" font-size="10" text-anchor="middle" '
                                    'fill="#111827" font-weight="bold">60s</text><text x="160.0" y="47.0" font-size="8.5" text-anchor="middle" '
                                    'fill="#475569">초</text><path d="M207.0,37.0 L219.0,37.0" stroke="#475569" stroke-width="1.6" '
                                    'marker-end="url(#tfa)" fill="none"/><rect x="220.0" y="14.0" width="92.0" height="46.0" rx="7" fill="#dbeafe" '
                                    'stroke="#94a3b8"/><text x="266.0" y="34.0" font-size="10" text-anchor="middle" fill="#111827" '
                                    'font-weight="bold">60</text><text x="266.0" y="47.0" font-size="8.5" text-anchor="middle" fill="#475569">초로 '
                                    '통일</text><text x="160.0" y="87.0" font-size="9" text-anchor="middle" fill="#475569">제각각인 값을 하나로 '
                                    '맞춰요</text></svg>',
                         'f_box': '<svg viewBox="0 0 320 84" xmlns="http://www.w3.org/2000/svg" font-family="sans-serif" font-size="10" '
                                  'fill="#111827"><defs><marker id="tfa" viewBox="0 0 10 10" refX="8" refY="5" markerWidth="6" markerHeight="6" '
                                  'orient="auto"><path d="M0,0 L10,5 L0,10 z" fill="#475569"/></marker></defs><rect x="6.0" y="8" width="102.7" '
                                  'height="18" fill="#e5e7eb" stroke="#fff"/><text x="57.3" y="20.5" font-size="9" text-anchor="middle" '
                                  'fill="#111827" font-weight="bold">칸</text><rect x="108.7" y="8" width="102.7" height="18" fill="#e5e7eb" '
                                  'stroke="#fff"/><text x="160.0" y="20.5" font-size="9" text-anchor="middle" fill="#111827" '
                                  'font-weight="bold">뜻</text><rect x="211.3" y="8" width="102.7" height="18" fill="#e5e7eb" stroke="#fff"/><text '
                                  'x="262.7" y="20.5" font-size="9" text-anchor="middle" fill="#111827" font-weight="bold">예</text><rect x="6.0" '
                                  'y="26" width="102.7" height="18" fill="#fff" stroke="#e5e7eb"/><text x="57.3" y="38.5" font-size="9" '
                                  'text-anchor="middle" fill="#111827" font-weight="bold">x, y</text><rect x="108.7" y="26" width="102.7" '
                                  'height="18" fill="#fff" stroke="#e5e7eb"/><text x="160.0" y="38.5" font-size="9" text-anchor="middle" '
                                  'fill="#111827">왼쪽 위 좌표</text><rect x="211.3" y="26" width="102.7" height="18" fill="#fff" stroke="#e5e7eb"/><text '
                                  'x="262.7" y="38.5" font-size="9" text-anchor="middle" fill="#111827">120, 88</text><rect x="6.0" y="44" '
                                  'width="102.7" height="18" fill="#fff" stroke="#e5e7eb"/><text x="57.3" y="56.5" font-size="9" '
                                  'text-anchor="middle" fill="#111827" font-weight="bold">w, h</text><rect x="108.7" y="44" width="102.7" '
                                  'height="18" fill="#fff" stroke="#e5e7eb"/><text x="160.0" y="56.5" font-size="9" text-anchor="middle" '
                                  'fill="#111827">폭과 높이</text><rect x="211.3" y="44" width="102.7" height="18" fill="#fff" stroke="#e5e7eb"/><text '
                                  'x="262.7" y="56.5" font-size="9" text-anchor="middle" fill="#111827">30, 42</text><text x="160.0" y="79.0" '
                                  'font-size="9" text-anchor="middle" fill="#475569">이 네 숫자로 네모를 그려요</text></svg>',
                         'f_schema': '<svg viewBox="0 0 320 102" xmlns="http://www.w3.org/2000/svg" font-family="sans-serif" font-size="10" '
                                     'fill="#111827"><defs><marker id="tfa" viewBox="0 0 10 10" refX="8" refY="5" markerWidth="6" markerHeight="6" '
                                     'orient="auto"><path d="M0,0 L10,5 L0,10 z" fill="#475569"/></marker></defs><rect x="6.0" y="8" width="154.0" '
                                     'height="18" fill="#e5e7eb" stroke="#fff"/><text x="83.0" y="20.5" font-size="9" text-anchor="middle" '
                                     'fill="#111827" font-weight="bold">칸</text><rect x="160.0" y="8" width="154.0" height="18" fill="#e5e7eb" '
                                     'stroke="#fff"/><text x="237.0" y="20.5" font-size="9" text-anchor="middle" fill="#111827" '
                                     'font-weight="bold">규칙</text><rect x="6.0" y="26" width="154.0" height="18" fill="#fff" stroke="#e5e7eb"/><text '
                                     'x="83.0" y="38.5" font-size="9" text-anchor="middle" fill="#111827" font-weight="bold">class</text><rect '
                                     'x="160.0" y="26" width="154.0" height="18" fill="#fff" stroke="#e5e7eb"/><text x="237.0" y="38.5" '
                                     'font-size="9" text-anchor="middle" fill="#111827">네 가지 소문자</text><rect x="6.0" y="44" width="154.0" '
                                     'height="18" fill="#fff" stroke="#e5e7eb"/><text x="83.0" y="56.5" font-size="9" text-anchor="middle" '
                                     'fill="#111827" font-weight="bold">w, h</text><rect x="160.0" y="44" width="154.0" height="18" fill="#fff" '
                                     'stroke="#e5e7eb"/><text x="237.0" y="56.5" font-size="9" text-anchor="middle" fill="#111827">0보다 큼</text><rect '
                                     'x="6.0" y="62" width="154.0" height="18" fill="#fff" stroke="#e5e7eb"/><text x="83.0" y="74.5" font-size="9" '
                                     'text-anchor="middle" fill="#111827" font-weight="bold">x+w</text><rect x="160.0" y="62" width="154.0" '
                                     'height="18" fill="#fff" stroke="#e5e7eb"/><text x="237.0" y="74.5" font-size="9" text-anchor="middle" '
                                     'fill="#111827">화면 안</text><text x="160.0" y="97.0" font-size="9" text-anchor="middle" fill="#475569">규격을 조건으로 '
                                     '적어요</text></svg>',
                         'f_debug': '<svg viewBox="0 0 320 92" xmlns="http://www.w3.org/2000/svg" font-family="sans-serif" font-size="10" '
                                    'fill="#111827"><defs><marker id="tfa" viewBox="0 0 10 10" refX="8" refY="5" markerWidth="6" markerHeight="6" '
                                    'orient="auto"><path d="M0,0 L10,5 L0,10 z" fill="#475569"/></marker></defs><rect x="8.0" y="14.0" width="92.0" '
                                    'height="46.0" rx="7" fill="#dbeafe" stroke="#94a3b8"/><text x="54.0" y="40.5" font-size="9.5" '
                                    'text-anchor="middle" fill="#111827" font-weight="bold">오류 메시지</text><path d="M101.0,37.0 L113.0,37.0" '
                                    'stroke="#475569" stroke-width="1.6" marker-end="url(#tfa)" fill="none"/><rect x="114.0" y="14.0" width="92.0" '
                                    'height="46.0" rx="7" fill="#dbeafe" stroke="#94a3b8"/><text x="160.0" y="40.5" font-size="9.5" '
                                    'text-anchor="middle" fill="#111827" font-weight="bold">원하는 동작</text><path d="M207.0,37.0 L219.0,37.0" '
                                    'stroke="#475569" stroke-width="1.6" marker-end="url(#tfa)" fill="none"/><rect x="220.0" y="14.0" width="92.0" '
                                    'height="46.0" rx="7" fill="#dbeafe" stroke="#94a3b8"/><text x="266.0" y="40.5" font-size="10" '
                                    'text-anchor="middle" fill="#111827" font-weight="bold">고칠 범위</text><text x="160.0" y="87.0" font-size="9" '
                                    'text-anchor="middle" fill="#475569">이 세 가지를 담아 다시 요청해요</text></svg>',
                         'f_roi': '<svg viewBox="0 0 320 102" xmlns="http://www.w3.org/2000/svg" font-family="sans-serif" font-size="10" '
                                  'fill="#111827"><defs><marker id="tfa" viewBox="0 0 10 10" refX="8" refY="5" markerWidth="6" markerHeight="6" '
                                  'orient="auto"><path d="M0,0 L10,5 L0,10 z" fill="#475569"/></marker></defs><rect x="6.0" y="8" width="102.7" '
                                  'height="18" fill="#e5e7eb" stroke="#fff"/><text x="57.3" y="20.5" font-size="9" text-anchor="middle" '
                                  'fill="#111827" font-weight="bold">구역</text><rect x="108.7" y="8" width="102.7" height="18" fill="#e5e7eb" '
                                  'stroke="#fff"/><text x="160.0" y="20.5" font-size="9" text-anchor="middle" fill="#111827" font-weight="bold">y '
                                  '범위</text><rect x="211.3" y="8" width="102.7" height="18" fill="#e5e7eb" stroke="#fff"/><text x="262.7" y="20.5" '
                                  'font-size="9" text-anchor="middle" fill="#111827" font-weight="bold">처리</text><rect x="6.0" y="26" width="102.7" '
                                  'height="18" fill="#fff" stroke="#e5e7eb"/><text x="57.3" y="38.5" font-size="9" text-anchor="middle" '
                                  'fill="#111827" font-weight="bold">영역 B</text><rect x="108.7" y="26" width="102.7" height="18" fill="#fff" '
                                  'stroke="#e5e7eb"/><text x="160.0" y="38.5" font-size="9" text-anchor="middle" fill="#111827">0~15</text><rect '
                                  'x="211.3" y="26" width="102.7" height="18" fill="#fff" stroke="#e5e7eb"/><text x="262.7" y="38.5" font-size="9" '
                                  'text-anchor="middle" fill="#111827">경보</text><rect x="6.0" y="44" width="102.7" height="18" fill="#fff" '
                                  'stroke="#e5e7eb"/><text x="57.3" y="56.5" font-size="9" text-anchor="middle" fill="#111827" font-weight="bold">영역 '
                                  'A</text><rect x="108.7" y="44" width="102.7" height="18" fill="#fff" stroke="#e5e7eb"/><text x="160.0" y="56.5" '
                                  'font-size="9" text-anchor="middle" fill="#111827">0~30</text><rect x="211.3" y="44" width="102.7" height="18" '
                                  'fill="#fff" stroke="#e5e7eb"/><text x="262.7" y="56.5" font-size="9" text-anchor="middle" '
                                  'fill="#111827">경보</text><rect x="6.0" y="62" width="102.7" height="18" fill="#fff" stroke="#e5e7eb"/><text '
                                  'x="57.3" y="74.5" font-size="9" text-anchor="middle" fill="#111827" font-weight="bold">제외 영역</text><rect '
                                  'x="108.7" y="62" width="102.7" height="18" fill="#fff" stroke="#e5e7eb"/><text x="160.0" y="74.5" font-size="9" '
                                  'text-anchor="middle" fill="#111827">나무 쪽</text><rect x="211.3" y="62" width="102.7" height="18" fill="#fff" '
                                  'stroke="#e5e7eb"/><text x="262.7" y="74.5" font-size="9" text-anchor="middle" fill="#111827">경보 안 함</text><text '
                                  'x="160.0" y="97.0" font-size="9" text-anchor="middle" fill="#475569">어디서 경보를 올릴지 좌표로 정해요</text></svg>',
                         'f_pre': '<svg viewBox="0 0 320 92" xmlns="http://www.w3.org/2000/svg" font-family="sans-serif" font-size="10" '
                                  'fill="#111827"><defs><marker id="tfa" viewBox="0 0 10 10" refX="8" refY="5" markerWidth="6" markerHeight="6" '
                                  'orient="auto"><path d="M0,0 L10,5 L0,10 z" fill="#475569"/></marker></defs><rect x="8.0" y="14.0" width="92.0" '
                                  'height="46.0" rx="7" fill="#dbeafe" stroke="#94a3b8"/><text x="54.0" y="40.5" font-size="9.5" '
                                  'text-anchor="middle" fill="#111827" font-weight="bold">대비 높이기</text><path d="M101.0,37.0 L113.0,37.0" '
                                  'stroke="#475569" stroke-width="1.6" marker-end="url(#tfa)" fill="none"/><rect x="114.0" y="14.0" width="92.0" '
                                  'height="46.0" rx="7" fill="#dbeafe" stroke="#94a3b8"/><text x="160.0" y="40.5" font-size="9.5" '
                                  'text-anchor="middle" fill="#111827" font-weight="bold">잡음 지우기</text><path d="M207.0,37.0 L219.0,37.0" '
                                  'stroke="#475569" stroke-width="1.6" marker-end="url(#tfa)" fill="none"/><rect x="220.0" y="14.0" width="92.0" '
                                  'height="46.0" rx="7" fill="#dbeafe" stroke="#94a3b8"/><text x="266.0" y="40.5" font-size="9.5" '
                                  'text-anchor="middle" fill="#111827" font-weight="bold">움직임 찾기</text><text x="160.0" y="87.0" font-size="9" '
                                  'text-anchor="middle" fill="#475569">영상을 손본 뒤 움직임을 찾아요</text></svg>',
                         'f_param': '<svg viewBox="0 0 320 92" xmlns="http://www.w3.org/2000/svg" font-family="sans-serif" font-size="10" '
                                    'fill="#111827"><defs><marker id="tfa" viewBox="0 0 10 10" refX="8" refY="5" markerWidth="6" markerHeight="6" '
                                    'orient="auto"><path d="M0,0 L10,5 L0,10 z" fill="#475569"/></marker></defs><rect x="8.0" y="14.0" width="65.5" '
                                    'height="46.0" rx="7" fill="#dbeafe" stroke="#94a3b8"/><text x="40.8" y="40.5" font-size="10" '
                                    'text-anchor="middle" fill="#111827" font-weight="bold">작은 함수</text><path d="M74.5,37.0 L86.5,37.0" '
                                    'stroke="#475569" stroke-width="1.6" marker-end="url(#tfa)" fill="none"/><rect x="87.5" y="14.0" width="65.5" '
                                    'height="46.0" rx="7" fill="#dbeafe" stroke="#94a3b8"/><text x="120.2" y="40.5" font-size="9.5" '
                                    'text-anchor="middle" fill="#111827" font-weight="bold">예시로 확인</text><path d="M154.0,37.0 L166.0,37.0" '
                                    'stroke="#475569" stroke-width="1.6" marker-end="url(#tfa)" fill="none"/><rect x="167.0" y="14.0" width="65.5" '
                                    'height="46.0" rx="7" fill="#dbeafe" stroke="#94a3b8"/><text x="199.8" y="40.5" font-size="10" '
                                    'text-anchor="middle" fill="#111827" font-weight="bold">전체 적용</text><path d="M233.5,37.0 L245.5,37.0" '
                                    'stroke="#475569" stroke-width="1.6" marker-end="url(#tfa)" fill="none"/><rect x="246.5" y="14.0" width="65.5" '
                                    'height="46.0" rx="7" fill="#dbeafe" stroke="#94a3b8"/><text x="279.2" y="40.5" font-size="10" '
                                    'text-anchor="middle" fill="#111827" font-weight="bold">비교표</text><text x="160.0" y="87.0" font-size="9" '
                                    'text-anchor="middle" fill="#475569">작게 만들고 확인한 뒤 넓혀요</text></svg>',
                         'f_bg': '<svg viewBox="0 0 320 92" xmlns="http://www.w3.org/2000/svg" font-family="sans-serif" font-size="10" '
                                 'fill="#111827"><defs><marker id="tfa" viewBox="0 0 10 10" refX="8" refY="5" markerWidth="6" markerHeight="6" '
                                 'orient="auto"><path d="M0,0 L10,5 L0,10 z" fill="#475569"/></marker></defs><rect x="8.0" y="14.0" width="92.0" '
                                 'height="46.0" rx="7" fill="#dbeafe" stroke="#94a3b8"/><text x="54.0" y="34.0" font-size="10" text-anchor="middle" '
                                 'fill="#111827" font-weight="bold">배경</text><text x="54.0" y="47.0" font-size="8.5" text-anchor="middle" '
                                 'fill="#475569">평소 모습</text><path d="M101.0,37.0 L113.0,37.0" stroke="#475569" stroke-width="1.6" '
                                 'marker-end="url(#tfa)" fill="none"/><rect x="114.0" y="14.0" width="92.0" height="46.0" rx="7" fill="#dbeafe" '
                                 'stroke="#94a3b8"/><text x="160.0" y="40.5" font-size="10" text-anchor="middle" fill="#111827" '
                                 'font-weight="bold">지금 영상</text><path d="M207.0,37.0 L219.0,37.0" stroke="#475569" stroke-width="1.6" '
                                 'marker-end="url(#tfa)" fill="none"/><rect x="220.0" y="14.0" width="92.0" height="46.0" rx="7" fill="#dbeafe" '
                                 'stroke="#94a3b8"/><text x="266.0" y="34.0" font-size="10" text-anchor="middle" fill="#111827" '
                                 'font-weight="bold">달라진 곳</text><text x="266.0" y="47.0" font-size="8.5" text-anchor="middle" '
                                 'fill="#475569">움직임</text><text x="160.0" y="87.0" font-size="9" text-anchor="middle" fill="#475569">평소와 달라진 곳을 '
                                 '찾아요</text></svg>',
                         'f_tl': '<svg viewBox="0 0 320 92" xmlns="http://www.w3.org/2000/svg" font-family="sans-serif" font-size="10" '
                                 'fill="#111827"><defs><marker id="tfa" viewBox="0 0 10 10" refX="8" refY="5" markerWidth="6" markerHeight="6" '
                                 'orient="auto"><path d="M0,0 L10,5 L0,10 z" fill="#475569"/></marker></defs><rect x="8.0" y="14.0" width="92.0" '
                                 'height="46.0" rx="7" fill="#dbeafe" stroke="#94a3b8"/><text x="54.0" y="34.0" font-size="10" text-anchor="middle" '
                                 'fill="#111827" font-weight="bold">큰 모델</text><text x="54.0" y="47.0" font-size="8.5" text-anchor="middle" '
                                 'fill="#475569">이미 학습됨</text><path d="M101.0,37.0 L113.0,37.0" stroke="#475569" stroke-width="1.6" '
                                 'marker-end="url(#tfa)" fill="none"/><rect x="114.0" y="14.0" width="92.0" height="46.0" rx="7" fill="#dbeafe" '
                                 'stroke="#94a3b8"/><text x="160.0" y="40.5" font-size="9.5" text-anchor="middle" fill="#111827" '
                                 'font-weight="bold">가져와 쓰기</text><path d="M207.0,37.0 L219.0,37.0" stroke="#475569" stroke-width="1.6" '
                                 'marker-end="url(#tfa)" fill="none"/><rect x="220.0" y="14.0" width="92.0" height="46.0" rx="7" fill="#dbeafe" '
                                 'stroke="#94a3b8"/><text x="266.0" y="34.0" font-size="10" text-anchor="middle" fill="#111827" '
                                 'font-weight="bold">결과 확인</text><text x="266.0" y="47.0" font-size="8.5" text-anchor="middle" fill="#475569">우리 '
                                 '조건에서</text><text x="160.0" y="87.0" font-size="9" text-anchor="middle" fill="#475569">이미 배운 것을 빌려 와요</text></svg>',
                         'f_iou': '<svg viewBox="0 0 320 92" xmlns="http://www.w3.org/2000/svg" font-family="sans-serif" font-size="10" '
                                  'fill="#111827"><defs><marker id="tfa" viewBox="0 0 10 10" refX="8" refY="5" markerWidth="6" markerHeight="6" '
                                  'orient="auto"><path d="M0,0 L10,5 L0,10 z" fill="#475569"/></marker></defs><rect x="8.0" y="14.0" width="92.0" '
                                  'height="46.0" rx="7" fill="#dbeafe" stroke="#94a3b8"/><text x="54.0" y="34.0" font-size="10" text-anchor="middle" '
                                  'fill="#111827" font-weight="bold">겹친 넓이</text><text x="54.0" y="47.0" font-size="8.5" text-anchor="middle" '
                                  'fill="#475569">200</text><path d="M101.0,37.0 L113.0,37.0" stroke="#475569" stroke-width="1.6" '
                                  'marker-end="url(#tfa)" fill="none"/><rect x="114.0" y="14.0" width="92.0" height="46.0" rx="7" fill="#dbeafe" '
                                  'stroke="#94a3b8"/><text x="160.0" y="34.0" font-size="10" text-anchor="middle" fill="#111827" '
                                  'font-weight="bold">합친 넓이</text><text x="160.0" y="47.0" font-size="8.5" text-anchor="middle" '
                                  'fill="#475569">600</text><path d="M207.0,37.0 L219.0,37.0" stroke="#475569" stroke-width="1.6" '
                                  'marker-end="url(#tfa)" fill="none"/><rect x="220.0" y="14.0" width="92.0" height="46.0" rx="7" fill="#dbeafe" '
                                  'stroke="#94a3b8"/><text x="266.0" y="34.0" font-size="10" text-anchor="middle" fill="#111827" '
                                  'font-weight="bold">IoU</text><text x="266.0" y="47.0" font-size="8.5" text-anchor="middle" fill="#475569">약 '
                                  '0.33</text><text x="160.0" y="87.0" font-size="9" text-anchor="middle" fill="#475569">겹친 넓이 ÷ 합친 '
                                  '넓이예요</text></svg>',
                         'f_verify': '<svg viewBox="0 0 320 92" xmlns="http://www.w3.org/2000/svg" font-family="sans-serif" font-size="10" '
                                     'fill="#111827"><defs><marker id="tfa" viewBox="0 0 10 10" refX="8" refY="5" markerWidth="6" markerHeight="6" '
                                     'orient="auto"><path d="M0,0 L10,5 L0,10 z" fill="#475569"/></marker></defs><rect x="8.0" y="14.0" width="65.5" '
                                     'height="46.0" rx="7" fill="#dbeafe" stroke="#94a3b8"/><text x="40.8" y="40.5" font-size="10" '
                                     'text-anchor="middle" fill="#111827" font-weight="bold">AI 답변</text><path d="M74.5,37.0 L86.5,37.0" '
                                     'stroke="#475569" stroke-width="1.6" marker-end="url(#tfa)" fill="none"/><rect x="87.5" y="14.0" width="65.5" '
                                     'height="46.0" rx="7" fill="#dbeafe" stroke="#94a3b8"/><text x="120.2" y="40.5" font-size="10" '
                                     'text-anchor="middle" fill="#111827" font-weight="bold">근거 수치</text><path d="M154.0,37.0 L166.0,37.0" '
                                     'stroke="#475569" stroke-width="1.6" marker-end="url(#tfa)" fill="none"/><rect x="167.0" y="14.0" width="65.5" '
                                     'height="46.0" rx="7" fill="#dbeafe" stroke="#94a3b8"/><text x="199.8" y="40.5" font-size="10" '
                                     'text-anchor="middle" fill="#111827" font-weight="bold">작은 예시</text><path d="M233.5,37.0 L245.5,37.0" '
                                     'stroke="#475569" stroke-width="1.6" marker-end="url(#tfa)" fill="none"/><rect x="246.5" y="14.0" width="65.5" '
                                     'height="46.0" rx="7" fill="#dbeafe" stroke="#94a3b8"/><text x="279.2" y="40.5" font-size="10" '
                                     'text-anchor="middle" fill="#111827" font-weight="bold">판단</text><text x="160.0" y="87.0" font-size="9" '
                                     'text-anchor="middle" fill="#475569">답변만 믿지 말고 근거를 함께 받아요</text></svg>',
                         'f_op': '<svg viewBox="0 0 320 150" xmlns="http://www.w3.org/2000/svg" font-family="sans-serif" font-size="10" '
                                 'fill="#111827"><defs><marker id="tfa" viewBox="0 0 10 10" refX="8" refY="5" markerWidth="6" markerHeight="6" '
                                 'orient="auto"><path d="M0,0 L10,5 L0,10 z" fill="#475569"/></marker></defs><rect x="34" y="12" width="274" '
                                 'height="108" fill="#fff" stroke="#94a3b8"/><polyline points="51.1,60.6 76.8,42.2 119.6,25.0 188.1,18.5 290.9,14.2" '
                                 'fill="none" stroke="#2563eb" stroke-width="2"/><circle cx="51.1" cy="60.6" r="2.5" fill="#2563eb"/><circle '
                                 'cx="76.8" cy="42.2" r="2.5" fill="#2563eb"/><circle cx="119.6" cy="25.0" r="2.5" fill="#2563eb"/><circle '
                                 'cx="188.1" cy="18.5" r="2.5" fill="#2563eb"/><circle cx="290.9" cy="14.2" r="2.5" fill="#2563eb"/><circle '
                                 'cx="119.6" cy="25.0" r="6" fill="none" stroke="#f97316" stroke-width="2.5"/><text x="128.6" y="29.0" font-size="9" '
                                 'text-anchor="start" fill="#f97316" font-weight="bold">운용점</text><text x="171.0" y="142.0" font-size="9" '
                                 'text-anchor="middle" fill="#475569">nuisance per day</text><text x="10" y="66.0" font-size="9" '
                                 'text-anchor="middle" fill="#475569" transform="rotate(-90 10 66.0)">recall</text><text x="40.0" y="24.0" '
                                 'font-size="8.5" text-anchor="start" fill="#475569">어디에 둘지 사람이 합의해요</text></svg>',
                         'f_cut': '<svg viewBox="0 0 320 92" xmlns="http://www.w3.org/2000/svg" font-family="sans-serif" font-size="10" '
                                  'fill="#111827"><defs><marker id="tfa" viewBox="0 0 10 10" refX="8" refY="5" markerWidth="6" markerHeight="6" '
                                  'orient="auto"><path d="M0,0 L10,5 L0,10 z" fill="#475569"/></marker></defs><rect x="8.0" y="14.0" width="92.0" '
                                  'height="46.0" rx="7" fill="#dbeafe" stroke="#94a3b8"/><text x="54.0" y="34.0" font-size="9.5" '
                                  'text-anchor="middle" fill="#111827" font-weight="bold">점수 0.62</text><text x="54.0" y="47.0" font-size="8.5" '
                                  'text-anchor="middle" fill="#475569">경보</text><path d="M101.0,37.0 L113.0,37.0" stroke="#475569" '
                                  'stroke-width="1.6" marker-end="url(#tfa)" fill="none"/><rect x="114.0" y="14.0" width="92.0" height="46.0" rx="7" '
                                  'fill="#dbeafe" stroke="#94a3b8"/><text x="160.0" y="34.0" font-size="9.5" text-anchor="middle" fill="#111827" '
                                  'font-weight="bold">임계값 0.5</text><text x="160.0" y="47.0" font-size="8.5" text-anchor="middle" '
                                  'fill="#475569">커트라인</text><path d="M207.0,37.0 L219.0,37.0" stroke="#475569" stroke-width="1.6" '
                                  'marker-end="url(#tfa)" fill="none"/><rect x="220.0" y="14.0" width="92.0" height="46.0" rx="7" fill="#dbeafe" '
                                  'stroke="#94a3b8"/><text x="266.0" y="34.0" font-size="9.5" text-anchor="middle" fill="#111827" '
                                  'font-weight="bold">점수 0.41</text><text x="266.0" y="47.0" font-size="8.5" text-anchor="middle" fill="#475569">경보 '
                                  '아님</text><text x="160.0" y="87.0" font-size="9" text-anchor="middle" fill="#475569">이상인지 초과인지도 적어요</text></svg>',
                         'f_edge': '<svg viewBox="0 0 320 102" xmlns="http://www.w3.org/2000/svg" font-family="sans-serif" font-size="10" '
                                   'fill="#111827"><defs><marker id="tfa" viewBox="0 0 10 10" refX="8" refY="5" markerWidth="6" markerHeight="6" '
                                   'orient="auto"><path d="M0,0 L10,5 L0,10 z" fill="#475569"/></marker></defs><rect x="6.0" y="8" width="154.0" '
                                   'height="18" fill="#e5e7eb" stroke="#fff"/><text x="83.0" y="20.5" font-size="9" text-anchor="middle" '
                                   'fill="#111827" font-weight="bold">score</text><rect x="160.0" y="8" width="154.0" height="18" fill="#e5e7eb" '
                                   'stroke="#fff"/><text x="237.0" y="20.5" font-size="9" text-anchor="middle" fill="#111827" '
                                   'font-weight="bold">처리</text><rect x="6.0" y="26" width="154.0" height="18" fill="#fff" stroke="#e5e7eb"/><text '
                                   'x="83.0" y="38.5" font-size="9" text-anchor="middle" fill="#111827" font-weight="bold">0.62</text><rect '
                                   'x="160.0" y="26" width="154.0" height="18" fill="#fff" stroke="#e5e7eb"/><text x="237.0" y="38.5" font-size="9" '
                                   'text-anchor="middle" fill="#111827">임계값과 비교</text><rect x="6.0" y="44" width="154.0" height="18" fill="#fff" '
                                   'stroke="#e5e7eb"/><text x="83.0" y="56.5" font-size="9" text-anchor="middle" fill="#111827" '
                                   'font-weight="bold">0.50</text><rect x="160.0" y="44" width="154.0" height="18" fill="#fff" '
                                   'stroke="#e5e7eb"/><text x="237.0" y="56.5" font-size="9" text-anchor="middle" fill="#111827">이상이면 경보</text><rect '
                                   'x="6.0" y="62" width="154.0" height="18" fill="#fff" stroke="#e5e7eb"/><text x="83.0" y="74.5" font-size="9" '
                                   'text-anchor="middle" fill="#111827" font-weight="bold">빈칸</text><rect x="160.0" y="62" width="154.0" height="18" '
                                   'fill="#fef3c7" stroke="#e5e7eb"/><text x="237.0" y="74.5" font-size="9" text-anchor="middle" fill="#111827" '
                                   'font-weight="bold">영상 손실: 사람 확인</text><text x="160.0" y="97.0" font-size="9" text-anchor="middle" '
                                   'fill="#475569">빈칸을 어떻게 할지 미리 알려 주세요</text></svg>',
                         'f_track': '<svg viewBox="0 0 320 84" xmlns="http://www.w3.org/2000/svg" font-family="sans-serif" font-size="10" '
                                    'fill="#111827"><defs><marker id="tfa" viewBox="0 0 10 10" refX="8" refY="5" markerWidth="6" markerHeight="6" '
                                    'orient="auto"><path d="M0,0 L10,5 L0,10 z" fill="#475569"/></marker></defs><rect x="6.0" y="8" width="102.7" '
                                    'height="18" fill="#e5e7eb" stroke="#fff"/><text x="57.3" y="20.5" font-size="9" text-anchor="middle" '
                                    'fill="#111827" font-weight="bold">특징</text><rect x="108.7" y="8" width="102.7" height="18" fill="#e5e7eb" '
                                    'stroke="#fff"/><text x="160.0" y="20.5" font-size="9" text-anchor="middle" fill="#111827" '
                                    'font-weight="bold">드론</text><rect x="211.3" y="8" width="102.7" height="18" fill="#e5e7eb" stroke="#fff"/><text '
                                    'x="262.7" y="20.5" font-size="9" text-anchor="middle" fill="#111827" font-weight="bold">새</text><rect x="6.0" '
                                    'y="26" width="102.7" height="18" fill="#fff" stroke="#e5e7eb"/><text x="57.3" y="38.5" font-size="9" '
                                    'text-anchor="middle" fill="#111827" font-weight="bold">정지 비율</text><rect x="108.7" y="26" width="102.7" '
                                    'height="18" fill="#fff" stroke="#e5e7eb"/><text x="160.0" y="38.5" font-size="9" text-anchor="middle" '
                                    'fill="#111827">높을 수 있음</text><rect x="211.3" y="26" width="102.7" height="18" fill="#fff" '
                                    'stroke="#e5e7eb"/><text x="262.7" y="38.5" font-size="9" text-anchor="middle" fill="#111827">낮음</text><rect '
                                    'x="6.0" y="44" width="102.7" height="18" fill="#fff" stroke="#e5e7eb"/><text x="57.3" y="56.5" font-size="9" '
                                    'text-anchor="middle" fill="#111827" font-weight="bold">속도 변화</text><rect x="108.7" y="44" width="102.7" '
                                    'height="18" fill="#fff" stroke="#e5e7eb"/><text x="160.0" y="56.5" font-size="9" text-anchor="middle" '
                                    'fill="#111827">일정한 편</text><rect x="211.3" y="44" width="102.7" height="18" fill="#fff" stroke="#e5e7eb"/><text '
                                    'x="262.7" y="56.5" font-size="9" text-anchor="middle" fill="#111827">큰 편</text><text x="160.0" y="79.0" '
                                    'font-size="9" text-anchor="middle" fill="#475569">움직이는 버릇이 단서가 돼요</text></svg>',
                         'f_leak': '<svg viewBox="0 0 320 92" xmlns="http://www.w3.org/2000/svg" font-family="sans-serif" font-size="10" '
                                   'fill="#111827"><defs><marker id="tfa" viewBox="0 0 10 10" refX="8" refY="5" markerWidth="6" markerHeight="6" '
                                   'orient="auto"><path d="M0,0 L10,5 L0,10 z" fill="#475569"/></marker></defs><rect x="8.0" y="14.0" width="145.0" '
                                   'height="46.0" rx="7" fill="#dbeafe" stroke="#94a3b8"/><text x="80.5" y="34.0" font-size="10" '
                                   'text-anchor="middle" fill="#111827" font-weight="bold">학습</text><text x="80.5" y="47.0" font-size="8.5" '
                                   'text-anchor="middle" fill="#475569">클립 A, B</text><path d="M154.0,37.0 L166.0,37.0" stroke="#475569" '
                                   'stroke-width="1.6" marker-end="url(#tfa)" fill="none"/><rect x="167.0" y="14.0" width="145.0" height="46.0" '
                                   'rx="7" fill="#dbeafe" stroke="#94a3b8"/><text x="239.5" y="34.0" font-size="10" text-anchor="middle" '
                                   'fill="#111827" font-weight="bold">테스트</text><text x="239.5" y="47.0" font-size="8.5" text-anchor="middle" '
                                   'fill="#475569">클립 C</text><text x="160.0" y="87.0" font-size="9" text-anchor="middle" fill="#475569">같은 클립을 양쪽에 '
                                   '두지 않아요</text></svg>',
                         'f_split': '<svg viewBox="0 0 320 84" xmlns="http://www.w3.org/2000/svg" font-family="sans-serif" font-size="10" '
                                    'fill="#111827"><defs><marker id="tfa" viewBox="0 0 10 10" refX="8" refY="5" markerWidth="6" markerHeight="6" '
                                    'orient="auto"><path d="M0,0 L10,5 L0,10 z" fill="#475569"/></marker></defs><rect x="6.0" y="8" width="102.7" '
                                    'height="18" fill="#e5e7eb" stroke="#fff"/><text x="57.3" y="20.5" font-size="9" text-anchor="middle" '
                                    'fill="#111827" font-weight="bold">방법</text><rect x="108.7" y="8" width="102.7" height="18" fill="#e5e7eb" '
                                    'stroke="#fff"/><text x="160.0" y="20.5" font-size="9" text-anchor="middle" fill="#111827" font-weight="bold">클립 '
                                    'A 궤적</text><rect x="211.3" y="8" width="102.7" height="18" fill="#e5e7eb" stroke="#fff"/><text x="262.7" '
                                    'y="20.5" font-size="9" text-anchor="middle" fill="#111827" font-weight="bold">수치</text><rect x="6.0" y="26" '
                                    'width="102.7" height="18" fill="#fff" stroke="#e5e7eb"/><text x="57.3" y="38.5" font-size="9" '
                                    'text-anchor="middle" fill="#111827" font-weight="bold">무작위 분할</text><rect x="108.7" y="26" width="102.7" '
                                    'height="18" fill="#fff" stroke="#e5e7eb"/><text x="160.0" y="38.5" font-size="9" text-anchor="middle" '
                                    'fill="#111827">학습과 테스트에 섞임</text><rect x="211.3" y="26" width="102.7" height="18" fill="#fff" '
                                    'stroke="#e5e7eb"/><text x="262.7" y="38.5" font-size="9" text-anchor="middle" fill="#111827">높게 나옴</text><rect '
                                    'x="6.0" y="44" width="102.7" height="18" fill="#fff" stroke="#e5e7eb"/><text x="57.3" y="56.5" font-size="9" '
                                    'text-anchor="middle" fill="#111827" font-weight="bold">클립 단위 분할</text><rect x="108.7" y="44" width="102.7" '
                                    'height="18" fill="#fff" stroke="#e5e7eb"/><text x="160.0" y="56.5" font-size="9" text-anchor="middle" '
                                    'fill="#111827">한쪽에만 있음</text><rect x="211.3" y="44" width="102.7" height="18" fill="#fff" '
                                    'stroke="#e5e7eb"/><text x="262.7" y="56.5" font-size="9" text-anchor="middle" fill="#111827">실제에 '
                                    '가까움</text><text x="160.0" y="79.0" font-size="9" text-anchor="middle" fill="#475569">나누는 방법이 수치를 '
                                    '바꿔요</text></svg>',
                         'f_halluc': '<svg viewBox="0 0 320 92" xmlns="http://www.w3.org/2000/svg" font-family="sans-serif" font-size="10" '
                                     'fill="#111827"><defs><marker id="tfa" viewBox="0 0 10 10" refX="8" refY="5" markerWidth="6" markerHeight="6" '
                                     'orient="auto"><path d="M0,0 L10,5 L0,10 z" fill="#475569"/></marker></defs><rect x="8.0" y="14.0" width="92.0" '
                                     'height="46.0" rx="7" fill="#dbeafe" stroke="#94a3b8"/><text x="54.0" y="40.5" font-size="9.5" '
                                     'text-anchor="middle" fill="#111827" font-weight="bold">자료에 없음</text><path d="M101.0,37.0 L113.0,37.0" '
                                     'stroke="#475569" stroke-width="1.6" marker-end="url(#tfa)" fill="none"/><rect x="114.0" y="14.0" width="92.0" '
                                     'height="46.0" rx="7" fill="#dbeafe" stroke="#94a3b8"/><text x="160.0" y="34.0" font-size="9.5" '
                                     'text-anchor="middle" fill="#111827" font-weight="bold">AI가 지어냄</text><text x="160.0" y="47.0" font-size="8.5" '
                                     'text-anchor="middle" fill="#475569">그럴듯함</text><path d="M207.0,37.0 L219.0,37.0" stroke="#475569" '
                                     'stroke-width="1.6" marker-end="url(#tfa)" fill="none"/><rect x="220.0" y="14.0" width="92.0" height="46.0" '
                                     'rx="7" fill="#dbeafe" stroke="#94a3b8"/><text x="266.0" y="34.0" font-size="10" text-anchor="middle" '
                                     'fill="#111827" font-weight="bold">사실 확인</text><text x="266.0" y="47.0" font-size="8.5" text-anchor="middle" '
                                     'fill="#475569">자료와 대조</text><text x="160.0" y="87.0" font-size="9" text-anchor="middle" fill="#475569">없는 내용은 '
                                     '‘자료 없음’으로 쓰게 해요</text></svg>',
                         'f_hitl': '<svg viewBox="0 0 320 92" xmlns="http://www.w3.org/2000/svg" font-family="sans-serif" font-size="10" '
                                   'fill="#111827"><defs><marker id="tfa" viewBox="0 0 10 10" refX="8" refY="5" markerWidth="6" markerHeight="6" '
                                   'orient="auto"><path d="M0,0 L10,5 L0,10 z" fill="#475569"/></marker></defs><rect x="8.0" y="14.0" width="92.0" '
                                   'height="46.0" rx="7" fill="#dbeafe" stroke="#94a3b8"/><text x="54.0" y="40.5" font-size="10" '
                                   'text-anchor="middle" fill="#111827" font-weight="bold">AI 제안</text><path d="M101.0,37.0 L113.0,37.0" '
                                   'stroke="#475569" stroke-width="1.6" marker-end="url(#tfa)" fill="none"/><rect x="114.0" y="14.0" width="92.0" '
                                   'height="46.0" rx="7" fill="#dbeafe" stroke="#94a3b8"/><text x="160.0" y="40.5" font-size="10" '
                                   'text-anchor="middle" fill="#111827" font-weight="bold">사람 확인</text><path d="M207.0,37.0 L219.0,37.0" '
                                   'stroke="#475569" stroke-width="1.6" marker-end="url(#tfa)" fill="none"/><rect x="220.0" y="14.0" width="92.0" '
                                   'height="46.0" rx="7" fill="#dbeafe" stroke="#94a3b8"/><text x="266.0" y="40.5" font-size="10" '
                                   'text-anchor="middle" fill="#111827" font-weight="bold">사람 승인</text><text x="160.0" y="87.0" font-size="9" '
                                   'text-anchor="middle" fill="#475569">AI는 제안, 결정은 사람이에요</text></svg>',
                         'f_bias': '<svg viewBox="0 0 320 92" xmlns="http://www.w3.org/2000/svg" font-family="sans-serif" font-size="10" '
                                   'fill="#111827"><defs><marker id="tfa" viewBox="0 0 10 10" refX="8" refY="5" markerWidth="6" markerHeight="6" '
                                   'orient="auto"><path d="M0,0 L10,5 L0,10 z" fill="#475569"/></marker></defs><rect x="8.0" y="14.0" width="92.0" '
                                   'height="46.0" rx="7" fill="#dbeafe" stroke="#94a3b8"/><text x="54.0" y="40.5" font-size="10" '
                                   'text-anchor="middle" fill="#111827" font-weight="bold">높은 점수</text><path d="M101.0,37.0 L113.0,37.0" '
                                   'stroke="#475569" stroke-width="1.6" marker-end="url(#tfa)" fill="none"/><rect x="114.0" y="14.0" width="92.0" '
                                   'height="46.0" rx="7" fill="#dbeafe" stroke="#94a3b8"/><text x="160.0" y="34.0" font-size="10" '
                                   'text-anchor="middle" fill="#111827" font-weight="bold">확인 생략</text><text x="160.0" y="47.0" font-size="8.5" '
                                   'text-anchor="middle" fill="#475569">위험</text><path d="M207.0,37.0 L219.0,37.0" stroke="#475569" '
                                   'stroke-width="1.6" marker-end="url(#tfa)" fill="none"/><rect x="220.0" y="14.0" width="92.0" height="46.0" '
                                   'rx="7" fill="#dbeafe" stroke="#94a3b8"/><text x="266.0" y="34.0" font-size="9.5" text-anchor="middle" '
                                   'fill="#111827" font-weight="bold">영상으로 확인</text><text x="266.0" y="47.0" font-size="8.5" text-anchor="middle" '
                                   'fill="#475569">해야 할 일</text><text x="160.0" y="87.0" font-size="9" text-anchor="middle" fill="#475569">점수가 높아도 '
                                   '직접 확인해요</text></svg>',
                         'f_secret': '<svg viewBox="0 0 320 102" xmlns="http://www.w3.org/2000/svg" font-family="sans-serif" font-size="10" '
                                     'fill="#111827"><defs><marker id="tfa" viewBox="0 0 10 10" refX="8" refY="5" markerWidth="6" markerHeight="6" '
                                     'orient="auto"><path d="M0,0 L10,5 L0,10 z" fill="#475569"/></marker></defs><rect x="6.0" y="8" width="154.0" '
                                     'height="18" fill="#e5e7eb" stroke="#fff"/><text x="83.0" y="20.5" font-size="9" text-anchor="middle" '
                                     'fill="#111827" font-weight="bold">넣어도 돼요</text><rect x="160.0" y="8" width="154.0" height="18" fill="#e5e7eb" '
                                     'stroke="#fff"/><text x="237.0" y="20.5" font-size="9" text-anchor="middle" fill="#111827" '
                                     'font-weight="bold">넣으면 안 돼요</text><rect x="6.0" y="26" width="154.0" height="18" fill="#fff" '
                                     'stroke="#e5e7eb"/><text x="83.0" y="38.5" font-size="9" text-anchor="middle" fill="#111827" '
                                     'font-weight="bold">열 이름</text><rect x="160.0" y="26" width="154.0" height="18" fill="#fff" '
                                     'stroke="#e5e7eb"/><text x="237.0" y="38.5" font-size="9" text-anchor="middle" fill="#111827">카메라 '
                                     '위치</text><rect x="6.0" y="44" width="154.0" height="18" fill="#fff" stroke="#e5e7eb"/><text x="83.0" y="56.5" '
                                     'font-size="9" text-anchor="middle" fill="#111827" font-weight="bold">가상 샘플</text><rect x="160.0" y="44" '
                                     'width="154.0" height="18" fill="#fff" stroke="#e5e7eb"/><text x="237.0" y="56.5" font-size="9" '
                                     'text-anchor="middle" fill="#111827">근무자 이름</text><rect x="6.0" y="62" width="154.0" height="18" fill="#fff" '
                                     'stroke="#e5e7eb"/><text x="83.0" y="74.5" font-size="9" text-anchor="middle" fill="#111827" '
                                     'font-weight="bold">오류 메시지</text><rect x="160.0" y="62" width="154.0" height="18" fill="#fff" '
                                     'stroke="#e5e7eb"/><text x="237.0" y="74.5" font-size="9" text-anchor="middle" fill="#111827">영상 캡처</text><text '
                                     'x="160.0" y="97.0" font-size="9" text-anchor="middle" fill="#475569">공개 AI에 넣기 전에 나눠 봐요</text></svg>',
                         'f_log': '<svg viewBox="0 0 320 120" xmlns="http://www.w3.org/2000/svg" font-family="sans-serif" font-size="10" '
                                  'fill="#111827"><defs><marker id="tfa" viewBox="0 0 10 10" refX="8" refY="5" markerWidth="6" markerHeight="6" '
                                  'orient="auto"><path d="M0,0 L10,5 L0,10 z" fill="#475569"/></marker></defs><rect x="6.0" y="8" width="154.0" '
                                  'height="18" fill="#e5e7eb" stroke="#fff"/><text x="83.0" y="20.5" font-size="9" text-anchor="middle" '
                                  'fill="#111827" font-weight="bold">항목</text><rect x="160.0" y="8" width="154.0" height="18" fill="#e5e7eb" '
                                  'stroke="#fff"/><text x="237.0" y="20.5" font-size="9" text-anchor="middle" fill="#111827" '
                                  'font-weight="bold">예</text><rect x="6.0" y="26" width="154.0" height="18" fill="#fff" stroke="#e5e7eb"/><text '
                                  'x="83.0" y="38.5" font-size="9" text-anchor="middle" fill="#111827" font-weight="bold">누가</text><rect x="160.0" '
                                  'y="26" width="154.0" height="18" fill="#fff" stroke="#e5e7eb"/><text x="237.0" y="38.5" font-size="9" '
                                  'text-anchor="middle" fill="#111827">모니터링 담당</text><rect x="6.0" y="44" width="154.0" height="18" fill="#fff" '
                                  'stroke="#e5e7eb"/><text x="83.0" y="56.5" font-size="9" text-anchor="middle" fill="#111827" '
                                  'font-weight="bold">언제</text><rect x="160.0" y="44" width="154.0" height="18" fill="#fff" stroke="#e5e7eb"/><text '
                                  'x="237.0" y="56.5" font-size="9" text-anchor="middle" fill="#111827">02:14</text><rect x="6.0" y="62" '
                                  'width="154.0" height="18" fill="#fff" stroke="#e5e7eb"/><text x="83.0" y="74.5" font-size="9" '
                                  'text-anchor="middle" fill="#111827" font-weight="bold">무엇을</text><rect x="160.0" y="62" width="154.0" height="18" '
                                  'fill="#fff" stroke="#e5e7eb"/><text x="237.0" y="74.5" font-size="9" text-anchor="middle" fill="#111827">경보 '
                                  '종결</text><rect x="6.0" y="80" width="154.0" height="18" fill="#fff" stroke="#e5e7eb"/><text x="83.0" y="92.5" '
                                  'font-size="9" text-anchor="middle" fill="#111827" font-weight="bold">왜</text><rect x="160.0" y="80" width="154.0" '
                                  'height="18" fill="#fff" stroke="#e5e7eb"/><text x="237.0" y="92.5" font-size="9" text-anchor="middle" '
                                  'fill="#111827">고라니 확인</text><text x="160.0" y="115.0" font-size="9" text-anchor="middle" fill="#475569">이유까지 '
                                  '남겨요</text></svg>',
                         'f_console': '<svg viewBox="0 0 320 148" xmlns="http://www.w3.org/2000/svg" font-family="sans-serif" font-size="10" '
                                      'fill="#111827"><defs><marker id="tfa" viewBox="0 0 10 10" refX="8" refY="5" markerWidth="6" markerHeight="6" '
                                      'orient="auto"><path d="M0,0 L10,5 L0,10 z" fill="#475569"/></marker></defs><polygon points="20,118 90,104 '
                                      '180,104 110,118" fill="#fee2e2" fill-opacity="0.55" stroke="#fee2e2"/><text x="192.0" y="115.0" '
                                      'font-size="9.5" text-anchor="start" fill="#111827" font-weight="bold">영상 손실 경보</text><polygon points="20,92 '
                                      '90,78 180,78 110,92" fill="#dbeafe" fill-opacity="0.55" stroke="#dbeafe"/><text x="192.0" y="89.0" '
                                      'font-size="9.5" text-anchor="start" fill="#111827" font-weight="bold">경보 목록·우선순위·상태</text><polygon '
                                      'points="20,66 90,52 180,52 110,66" fill="#e0e7ff" fill-opacity="0.55" stroke="#e0e7ff"/><text x="192.0" '
                                      'y="63.0" font-size="9.5" text-anchor="start" fill="#111827" font-weight="bold">경보 영상</text><polygon '
                                      'points="20,40 90,26 180,26 110,40" fill="#c7d2fe" fill-opacity="0.55" stroke="#c7d2fe"/><text x="192.0" '
                                      'y="37.0" font-size="9.5" text-anchor="start" fill="#111827" font-weight="bold">배치도</text><text x="160.0" '
                                      'y="143.0" font-size="9" text-anchor="middle" fill="#475569">몰릴 때 무엇부터 볼지 정해요</text></svg>',
                         'f_gap': '<svg viewBox="0 0 320 102" xmlns="http://www.w3.org/2000/svg" font-family="sans-serif" font-size="10" '
                                  'fill="#111827"><defs><marker id="tfa" viewBox="0 0 10 10" refX="8" refY="5" markerWidth="6" markerHeight="6" '
                                  'orient="auto"><path d="M0,0 L10,5 L0,10 z" fill="#475569"/></marker></defs><rect x="6.0" y="8" width="102.7" '
                                  'height="18" fill="#e5e7eb" stroke="#fff"/><text x="57.3" y="20.5" font-size="9" text-anchor="middle" '
                                  'fill="#111827" font-weight="bold">신호 시각</text><rect x="108.7" y="8" width="102.7" height="18" fill="#e5e7eb" '
                                  'stroke="#fff"/><text x="160.0" y="20.5" font-size="9" text-anchor="middle" fill="#111827" '
                                  'font-weight="bold">간격</text><rect x="211.3" y="8" width="102.7" height="18" fill="#e5e7eb" stroke="#fff"/><text '
                                  'x="262.7" y="20.5" font-size="9" text-anchor="middle" fill="#111827" font-weight="bold">판정</text><rect x="6.0" '
                                  'y="26" width="102.7" height="18" fill="#fff" stroke="#e5e7eb"/><text x="57.3" y="38.5" font-size="9" '
                                  'text-anchor="middle" fill="#111827" font-weight="bold">0, 10</text><rect x="108.7" y="26" width="102.7" '
                                  'height="18" fill="#fff" stroke="#e5e7eb"/><text x="160.0" y="38.5" font-size="9" text-anchor="middle" '
                                  'fill="#111827">10초</text><rect x="211.3" y="26" width="102.7" height="18" fill="#fff" stroke="#e5e7eb"/><text '
                                  'x="262.7" y="38.5" font-size="9" text-anchor="middle" fill="#111827">정상</text><rect x="6.0" y="44" width="102.7" '
                                  'height="18" fill="#fff" stroke="#e5e7eb"/><text x="57.3" y="56.5" font-size="9" text-anchor="middle" '
                                  'fill="#111827" font-weight="bold">10, 55</text><rect x="108.7" y="44" width="102.7" height="18" fill="#fff" '
                                  'stroke="#e5e7eb"/><text x="160.0" y="56.5" font-size="9" text-anchor="middle" fill="#111827">45초</text><rect '
                                  'x="211.3" y="44" width="102.7" height="18" fill="#fef3c7" stroke="#e5e7eb"/><text x="262.7" y="56.5" '
                                  'font-size="9" text-anchor="middle" fill="#111827" font-weight="bold">영상 손실</text><rect x="6.0" y="62" '
                                  'width="102.7" height="18" fill="#fff" stroke="#e5e7eb"/><text x="57.3" y="74.5" font-size="9" '
                                  'text-anchor="middle" fill="#111827" font-weight="bold">55, 85</text><rect x="108.7" y="62" width="102.7" '
                                  'height="18" fill="#fff" stroke="#e5e7eb"/><text x="160.0" y="74.5" font-size="9" text-anchor="middle" '
                                  'fill="#111827">30초</text><rect x="211.3" y="62" width="102.7" height="18" fill="#fff" stroke="#e5e7eb"/><text '
                                  'x="262.7" y="74.5" font-size="9" text-anchor="middle" fill="#111827">정상</text><text x="160.0" y="97.0" '
                                  'font-size="9" text-anchor="middle" fill="#475569">30초를 넘으면 영상 손실이에요</text></svg>',
                         'f_review': '<svg viewBox="0 0 320 156" xmlns="http://www.w3.org/2000/svg" font-family="sans-serif" font-size="10" '
                                     'fill="#111827"><defs><marker id="tfa" viewBox="0 0 10 10" refX="8" refY="5" markerWidth="6" markerHeight="6" '
                                     'orient="auto"><path d="M0,0 L10,5 L0,10 z" fill="#475569"/></marker></defs><ellipse cx="160.0" cy="70" '
                                     'rx="112" ry="48" fill="none" stroke="#e5e7eb" stroke-width="6"/><path d="M207.7,26.6 L261.3,49.6" '
                                     'stroke="#2563eb" stroke-width="2" marker-end="url(#tfa)" fill="none"/><path d="M261.3,90.4 L207.7,113.4" '
                                     'stroke="#2563eb" stroke-width="2" marker-end="url(#tfa)" fill="none"/><path d="M112.3,113.4 L58.7,90.4" '
                                     'stroke="#2563eb" stroke-width="2" marker-end="url(#tfa)" fill="none"/><path d="M58.7,49.6 L112.3,26.6" '
                                     'stroke="#2563eb" stroke-width="2" marker-end="url(#tfa)" fill="none"/><rect x="116.0" y="10.0" width="88.0" '
                                     'height="24.0" rx="7" fill="#dbeafe" stroke="#94a3b8"/><text x="160.0" y="25.5" font-size="9" '
                                     'text-anchor="middle" fill="#111827" font-weight="bold">설명</text><rect x="228.0" y="58.0" width="88.0" '
                                     'height="24.0" rx="7" fill="#dbeafe" stroke="#94a3b8"/><text x="272.0" y="73.5" font-size="9" '
                                     'text-anchor="middle" fill="#111827" font-weight="bold">테스트</text><rect x="116.0" y="106.0" width="88.0" '
                                     'height="24.0" rx="7" fill="#dbeafe" stroke="#94a3b8"/><text x="160.0" y="121.5" font-size="9" '
                                     'text-anchor="middle" fill="#111827" font-weight="bold">수정</text><rect x="4.0" y="58.0" width="88.0" '
                                     'height="24.0" rx="7" fill="#dbeafe" stroke="#94a3b8"/><text x="48.0" y="73.5" font-size="9" '
                                     'text-anchor="middle" fill="#111827" font-weight="bold">비교</text><text x="160.0" y="74.0" font-size="10.5" '
                                     'text-anchor="middle" fill="#475569" font-weight="bold">리뷰</text><text x="160.0" y="151.0" font-size="9" '
                                     'text-anchor="middle" fill="#475569">고치기 전에 먼저 설명하게 해요</text></svg>',
                         'f_scope': '<svg viewBox="0 0 320 92" xmlns="http://www.w3.org/2000/svg" font-family="sans-serif" font-size="10" '
                                    'fill="#111827"><defs><marker id="tfa" viewBox="0 0 10 10" refX="8" refY="5" markerWidth="6" markerHeight="6" '
                                    'orient="auto"><path d="M0,0 L10,5 L0,10 z" fill="#475569"/></marker></defs><rect x="8.0" y="14.0" width="92.0" '
                                    'height="46.0" rx="7" fill="#dbeafe" stroke="#94a3b8"/><text x="54.0" y="34.0" font-size="10" '
                                    'text-anchor="middle" fill="#111827" font-weight="bold">고칠 곳</text><text x="54.0" y="47.0" font-size="8.5" '
                                    'text-anchor="middle" fill="#475569">이 함수만</text><path d="M101.0,37.0 L113.0,37.0" stroke="#475569" '
                                    'stroke-width="1.6" marker-end="url(#tfa)" fill="none"/><rect x="114.0" y="14.0" width="92.0" height="46.0" '
                                    'rx="7" fill="#dbeafe" stroke="#94a3b8"/><text x="160.0" y="34.0" font-size="10" text-anchor="middle" '
                                    'fill="#111827" font-weight="bold">그대로</text><text x="160.0" y="47.0" font-size="8.5" text-anchor="middle" '
                                    'fill="#475569">원본 기록</text><path d="M207.0,37.0 L219.0,37.0" stroke="#475569" stroke-width="1.6" '
                                    'marker-end="url(#tfa)" fill="none"/><rect x="220.0" y="14.0" width="92.0" height="46.0" rx="7" fill="#dbeafe" '
                                    'stroke="#94a3b8"/><text x="266.0" y="34.0" font-size="10" text-anchor="middle" fill="#111827" '
                                    'font-weight="bold">그대로</text><text x="266.0" y="47.0" font-size="8.5" text-anchor="middle" fill="#475569">다른 '
                                    '코드</text><text x="160.0" y="87.0" font-size="9" text-anchor="middle" fill="#475569">고칠 범위를 말해 주세요</text></svg>',
                         'f_tool': '<svg viewBox="0 0 320 92" xmlns="http://www.w3.org/2000/svg" font-family="sans-serif" font-size="10" '
                                   'fill="#111827"><defs><marker id="tfa" viewBox="0 0 10 10" refX="8" refY="5" markerWidth="6" markerHeight="6" '
                                   'orient="auto"><path d="M0,0 L10,5 L0,10 z" fill="#475569"/></marker></defs><rect x="8.0" y="14.0" width="65.5" '
                                   'height="46.0" rx="7" fill="#dbeafe" stroke="#94a3b8"/><text x="40.8" y="40.5" font-size="10" '
                                   'text-anchor="middle" fill="#111827" font-weight="bold">설정값</text><path d="M74.5,37.0 L86.5,37.0" '
                                   'stroke="#475569" stroke-width="1.6" marker-end="url(#tfa)" fill="none"/><rect x="87.5" y="14.0" width="65.5" '
                                   'height="46.0" rx="7" fill="#dbeafe" stroke="#94a3b8"/><text x="120.2" y="40.5" font-size="10" '
                                   'text-anchor="middle" fill="#111827" font-weight="bold">계산 함수</text><path d="M154.0,37.0 L166.0,37.0" '
                                   'stroke="#475569" stroke-width="1.6" marker-end="url(#tfa)" fill="none"/><rect x="167.0" y="14.0" width="65.5" '
                                   'height="46.0" rx="7" fill="#dbeafe" stroke="#94a3b8"/><text x="199.8" y="40.5" font-size="10" '
                                   'text-anchor="middle" fill="#111827" font-weight="bold">표·그림</text><path d="M233.5,37.0 L245.5,37.0" '
                                   'stroke="#475569" stroke-width="1.6" marker-end="url(#tfa)" fill="none"/><rect x="246.5" y="14.0" width="65.5" '
                                   'height="46.0" rx="7" fill="#dbeafe" stroke="#94a3b8"/><text x="279.2" y="40.5" font-size="10" '
                                   'text-anchor="middle" fill="#111827" font-weight="bold">브리핑</text><text x="160.0" y="87.0" font-size="9" '
                                   'text-anchor="middle" fill="#475569">한 번에 다시 돌릴 수 있게 묶어요</text></svg>',
                         'f_aar': '<svg viewBox="0 0 320 156" xmlns="http://www.w3.org/2000/svg" font-family="sans-serif" font-size="10" '
                                  'fill="#111827"><defs><marker id="tfa" viewBox="0 0 10 10" refX="8" refY="5" markerWidth="6" markerHeight="6" '
                                  'orient="auto"><path d="M0,0 L10,5 L0,10 z" fill="#475569"/></marker></defs><ellipse cx="160.0" cy="70" rx="112" '
                                  'ry="48" fill="none" stroke="#e5e7eb" stroke-width="6"/><path d="M207.7,26.6 L261.3,49.6" stroke="#2563eb" '
                                  'stroke-width="2" marker-end="url(#tfa)" fill="none"/><path d="M261.3,90.4 L207.7,113.4" stroke="#2563eb" '
                                  'stroke-width="2" marker-end="url(#tfa)" fill="none"/><path d="M112.3,113.4 L58.7,90.4" stroke="#2563eb" '
                                  'stroke-width="2" marker-end="url(#tfa)" fill="none"/><path d="M58.7,49.6 L112.3,26.6" stroke="#2563eb" '
                                  'stroke-width="2" marker-end="url(#tfa)" fill="none"/><rect x="116.0" y="10.0" width="88.0" height="24.0" rx="7" '
                                  'fill="#dbeafe" stroke="#94a3b8"/><text x="160.0" y="25.5" font-size="9" text-anchor="middle" fill="#111827" '
                                  'font-weight="bold">계획</text><rect x="228.0" y="58.0" width="88.0" height="24.0" rx="7" fill="#dbeafe" '
                                  'stroke="#94a3b8"/><text x="272.0" y="73.5" font-size="9" text-anchor="middle" fill="#111827" '
                                  'font-weight="bold">실제</text><rect x="116.0" y="106.0" width="88.0" height="24.0" rx="7" fill="#dbeafe" '
                                  'stroke="#94a3b8"/><text x="160.0" y="121.5" font-size="9" text-anchor="middle" fill="#111827" '
                                  'font-weight="bold">원인</text><rect x="4.0" y="58.0" width="88.0" height="24.0" rx="7" fill="#dbeafe" '
                                  'stroke="#94a3b8"/><text x="48.0" y="73.5" font-size="9" text-anchor="middle" fill="#111827" '
                                  'font-weight="bold">재조정</text><text x="160.0" y="74.0" font-size="10.5" text-anchor="middle" fill="#475569" '
                                  'font-weight="bold">AAR</text><text x="160.0" y="151.0" font-size="9" text-anchor="middle" fill="#475569">탓하지 않고 '
                                  '바꿀 것을 찾아요</text></svg>',
                         'f_lint': '<svg viewBox="0 0 320 102" xmlns="http://www.w3.org/2000/svg" font-family="sans-serif" font-size="10" '
                                   'fill="#111827"><defs><marker id="tfa" viewBox="0 0 10 10" refX="8" refY="5" markerWidth="6" markerHeight="6" '
                                   'orient="auto"><path d="M0,0 L10,5 L0,10 z" fill="#475569"/></marker></defs><rect x="6.0" y="8" width="154.0" '
                                   'height="18" fill="#e5e7eb" stroke="#fff"/><text x="83.0" y="20.5" font-size="9" text-anchor="middle" '
                                   'fill="#111827" font-weight="bold">항목</text><rect x="160.0" y="8" width="154.0" height="18" fill="#e5e7eb" '
                                   'stroke="#fff"/><text x="237.0" y="20.5" font-size="9" text-anchor="middle" fill="#111827" '
                                   'font-weight="bold">있음?</text><rect x="6.0" y="26" width="154.0" height="18" fill="#fff" stroke="#e5e7eb"/><text '
                                   'x="83.0" y="38.5" font-size="9" text-anchor="middle" fill="#111827" font-weight="bold">표본 수</text><rect '
                                   'x="160.0" y="26" width="154.0" height="18" fill="#fff" stroke="#e5e7eb"/><text x="237.0" y="38.5" font-size="9" '
                                   'text-anchor="middle" fill="#111827">O</text><rect x="6.0" y="44" width="154.0" height="18" fill="#fff" '
                                   'stroke="#e5e7eb"/><text x="83.0" y="56.5" font-size="9" text-anchor="middle" fill="#111827" '
                                   'font-weight="bold">남은 위험</text><rect x="160.0" y="44" width="154.0" height="18" fill="#fff" '
                                   'stroke="#e5e7eb"/><text x="237.0" y="56.5" font-size="9" text-anchor="middle" fill="#111827">O</text><rect '
                                   'x="6.0" y="62" width="154.0" height="18" fill="#fff" stroke="#e5e7eb"/><text x="83.0" y="74.5" font-size="9" '
                                   'text-anchor="middle" fill="#111827" font-weight="bold">AI 산출물 표기</text><rect x="160.0" y="62" width="154.0" '
                                   'height="18" fill="#fef3c7" stroke="#e5e7eb"/><text x="237.0" y="74.5" font-size="9" text-anchor="middle" '
                                   'fill="#111827" font-weight="bold">X</text><text x="160.0" y="97.0" font-size="9" text-anchor="middle" '
                                   'fill="#475569">빠진 항목을 코드로 찾아요</text></svg>'},
             'workspace': '🛡️ 시설 방호 영상분석반',
             'call_default': '전우님',
             'stay_jobs': {}},
 'competencies': [{'id': 'D1',
                   'group': '디지털·AI 활용 역량',
                   'name': '프롬프트 구조화',
                   'criteria': '업무 요청을 맥락·입력·할 일·출력 형식으로 나눠 AI에게 분명하게 요청할 수 있다.',
                   'ncs': '디지털능력 (인공지능(AI)활용능력), 의사소통능력 (문서소통능력)'},
                  {'id': 'D2',
                   'group': '디지털·AI 활용 역량',
                   'name': '데이터 설명',
                   'criteria': '파일·열·값의 모양을 AI에게 정확히 알려 주어 추측 없이 동작하는 코드를 받을 수 있다.',
                   'ncs': '디지털능력 (디지털활용능력)'},
                  {'id': 'D3',
                   'group': '디지털·AI 활용 역량',
                   'name': '수정 요청(디버깅)',
                   'criteria': '오류 메시지와 기대 동작을 담아 수정을 요청하고, 고칠 범위를 정할 수 있다.',
                   'ncs': '문제해결능력 (문제분석능력), 디지털능력 (디지털활용능력)'},
                  {'id': 'D4',
                   'group': '디지털·AI 활용 역량',
                   'name': '나누어 만들기',
                   'criteria': '큰 일을 작은 단계로 쪼개 요청하고, 조건을 바꿔 돌릴 수 있는 함수로 만들어 달라고 할 수 있다.',
                   'ncs': '문제해결능력 (대안발굴능력), 디지털능력 (디지털활용능력)'},
                  {'id': 'D5',
                   'group': '디지털·AI 활용 역량',
                   'name': '결과 검증',
                   'criteria': 'AI의 결과를 근거 수치·경계 사례·작은 테스트로 확인하고, 믿어도 되는지 판단할 수 있다.',
                   'ncs': '수리능력 (통계활용능력), 문제해결능력 (문제분석능력)'},
                  {'id': 'D6',
                   'group': '디지털·AI 활용 역량',
                   'name': '안전한 AI 활용',
                   'criteria': 'AI에 넣으면 안 되는 정보를 구분하고, 사람이 확인할 일을 정할 수 있다.',
                   'ncs': '디지털능력 (디지털책임의식)'},
                  {'id': 'J1',
                   'group': '직무전문 역량',
                   'name': 'AI로 문서 만들기',
                   'criteria': '독자·양식·근거를 지정해 AI가 업무 문서 초안을 만들게 하고, 사실을 확인해 고칠 수 있다.',
                   'ncs': '의사소통능력 (문서소통능력)'},
                  {'id': 'C1',
                   'group': '협업·소통 역량',
                   'name': '결과로 조율하기',
                   'criteria': 'AI로 만든 비교 결과를 근거로 팀원의 다른 의견을 조율해 기준을 합의할 수 있다.',
                   'ncs': '대인관계능력 (협업능력·갈등관리능력)'}],
 'missions': [{'id': 'm1',
               'no': 1,
               'title': '기지 외곽 감시 체계와 CCTV 영상분석의 이해',
               'short': '프롬프트 네 칸',
               'kind': '개인',
               'competencies': ['D1'],
               'outline': 'AI에게 일을 시킬 때 꼭 넣어야 할 네 칸(맥락·입력·할 일·출력 형식)을 익히고, CCTV 경보 기록을 경보 대상·불필요 경보·오경보로 나눈 집계표를 받아요.',
               'brief': {'subtitle': '경보 집계표 만들기: 프롬프트의 네 칸',
                         'time': '약 25분',
                         'goal': '프롬프트의 네 칸을 채워 AI에게 요청하고, 카메라별 경보 집계표와 그림, 문서를 받아요.',
                         'why': 'CCTV에 무언가 움직였다면 모두 경보로 올려야 할까요? 영상분석에서는 경보를 원인별로 나눠 봐야 해요. 이런 집계를 AI에게 맡길 때, 무엇을 알려 주느냐에 따라 결과가 크게 달라져요.',
                         'case': {'title': '해외 사례: 트럭에 숨겨 비행장 근처까지 옮긴 드론',
                                  'source': 'Euromaidan Press, CBS 등 거미줄 작전 보도 (2025.6)',
                                  'points': ['2025년 6월 우크라이나의 ‘거미줄 작전’에서는 소형 드론 117대가 트럭에 실은 나무 구조물 안에 숨겨져 러시아 비행장 근처까지 옮겨졌어요. 도착하면 원격으로 지붕을 열어 '
                                             '날렸어요.',
                                             '피해 규모는 출처마다 달라요. 우크라이나 쪽은 항공기 41대가 피해를 입었다고 했고, 다른 추정은 이보다 낮아요.',
                                             '위협은 시설 안쪽만이 아니라 바깥에서 시작돼요. 그래서 외곽 영상을 보고, 무엇이 왜 경보로 올라왔는지 나눠 보는 일이 출발점이에요.']},
                         'team': '오영상(모니터링 담당)이 경보가 너무 많다고 해요. 전우님이 AI에게 프롬프트를 보내 집계표와 문서를 받아요.',
                         'terms': [{'t': '바이브 코딩',
                                    'en': 'vibe coding',
                                    'fig': 'f_loop',
                                    'd': '만들고 싶은 것을 말로 설명해 AI가 코드를 쓰게 하고, 실행 결과를 보며 고쳐 가는 방식이에요.',
                                    'easy': '코드를 직접 쓰는 대신, AI에게 일을 설명하고 결과를 확인하는 역할을 맡는 거예요.',
                                    'use': '요청 → 생성 → 실행 → 확인 → 수정 요청을 반복해요.'},
                                   {'t': '프롬프트',
                                    'en': 'prompt',
                                    'fig': 'f_four',
                                    'd': 'AI에게 보내는 요청 글이에요.',
                                    'easy': '새로 온 동료에게 주는 업무 지시서라고 생각하면 돼요.',
                                    'use': '맥락·입력·할 일·출력 형식의 네 칸을 채우면 결과가 안정돼요.'},
                                   {'t': '불필요 경보',
                                    'en': 'nuisance alarm',
                                    'fig': 'f_kinds',
                                    'd': '실제로 무언가 움직였지만 경보 대상이 아닌 경보예요. 동물, 나무, 비, 불빛이 대표적이에요.',
                                    'easy': '장비는 제대로 봤지만 우리가 찾던 것이 아닌 경우예요.',
                                    'use': '프롬프트에 어떤 물체를 여기에 넣을지 적어요.'},
                                   {'t': '오경보',
                                    'en': 'false alarm',
                                    'fig': 'f_kinds',
                                    'd': '장비 이상이나 잡음처럼 실제 자극 없이 울린 경보예요.',
                                    'easy': '아무것도 없는데 장비가 잘못 울린 경우예요.',
                                    'use': '불필요 경보와 나눠 세야 대책이 달라져요.'}],
                         'steps': ['네 칸 익히기: 프롬프트의 네 칸', 'AI에게 시키기: 프롬프트를 만들어 AI에게 보내기', '내 말로 쓰기: 같은 일을 내 말로 요청하기'],
                         'glossary': [],
                         'outputs': ['M1 카메라별 경보 집계표']},
               'start_messages': [['dm_oh', 'oh', '{call}, 경보가 너무 많아요. 카메라별로 무엇이 얼마나 울리는지 표가 필요해요. 직접 세지 말고 AI에게 시켜 보세요. 파일은 m1_alarms.csv예요.']],
               'finish_messages': [['dm_oh', 'oh', '{call}, 표 잘 받았어요! 어느 카메라에서 무엇을 줄일지 보이네요 👍']],
               'steps': [{'short': '네 칸 익히기',
                          'title': '프롬프트의 네 칸',
                          'lead': 'AI는 전우님 업무도, 파일도 몰라요. **네 칸**을 채워 알려 주면 돼요.',
                          'blocks': [{'id': 's1_info',
                                      'type': 'info',
                                      'title': '좋은 프롬프트의 네 칸',
                                      'icon': '🧩',
                                      'text': ['같은 일을 시켜도 무엇을 알려 주느냐에 따라 결과가 달라져요. 처음에는 아래 네 칸만 기억하세요.'],
                                      'table': {'head': ['칸', '무엇을 적나', '예'],
                                                'rows': [['맥락', '왜 필요한지, 어떤 일인지', 'CCTV 경보가 너무 많아 원인을 보려고 해'],
                                                         ['입력', '파일 이름, 열 이름, 값의 뜻', 'm1_alarms.csv, 열은 alarm_id·camera·hour·object'],
                                                         ['할 일', '나누는 규칙을 구체적으로', 'person·vehicle·drone은 경보 대상'],
                                                         ['출력 형식', '받고 싶은 모양', '카메라는 행, 구분은 열인 표']]},
                                      'figure': 'f_four'},
                                     {'id': 's1_q',
                                      'type': 'quiz',
                                      'title': '어떤 프롬프트가 가장 좋을까요?',
                                      'text': '같은 일을 시키는 세 프롬프트예요. 다시 묻는 일이 가장 적을 것은 무엇일까요?',
                                      'options': ['CCTV 분석 코드 좀 짜 줘.',
                                                  '경보가 많은지 알려 줘. 잘 부탁해.',
                                                  'CCTV 경보가 너무 많아 원인을 보려고 해. m1_alarms.csv의 object 값을 경보 대상과 불필요 경보로 나눠 카메라별로 세어 표로 보여 줘.'],
                                      'answer': 2,
                                      'explain': '맥락, 입력, 할 일, 출력 형식이 모두 들어 있어요. 정중한 말보다 구체적인 정보가 결과를 바꿔요.',
                                      'hints': ['네 칸 가운데 몇 칸이 채워졌는지 세어 보세요.', '파일 이름과 받고 싶은 모양까지 적힌 것을 찾아보세요.']}]},
                         {'short': 'AI에게 시키기',
                          'title': '프롬프트를 만들어 AI에게 보내기',
                          'lead': '칸마다 문장을 고르면 프롬프트가 만들어져요. 일부러 **아쉬운 문장**도 보내 보세요. 무엇이 달라지는지 보는 것이 이번 활동의 핵심이에요.',
                          'blocks': [{'id': 's2_vibe',
                                      'type': 'vibe',
                                      'title': '카메라별 경보 집계표 만들기',
                                      'ask': ['oh', '경보 기록 파일을 드릴게요. 카메라마다 경보 대상, 불필요 경보, 오경보가 몇 건인지 표로 받고 싶어요.'],
                                      'file': 'm1_alarms.csv',
                                      'slots': [{'id': 'ctx',
                                                 'label': '맥락',
                                                 'why': '왜 필요한지 알려요.',
                                                 'options': [{'text': 'CCTV 영상분석 경보를 살피고 있어. 경보가 너무 많아서 카메라별로 무엇 때문에 울리는지 확인하려고 해.', 'good': True},
                                                             {'text': '코드 좀 짜 줘.', 'good': False},
                                                             {'text': '', 'good': False}],
                                                 'miss': {'say': '데이터를 읽어 요약 통계를 만들었어요.',
                                                          'code': 'import pandas as pd\n'
                                                                  "df = pd.read_csv('{VDIR}/m1_alarms.csv')\n"
                                                                  'print(df.describe())',
                                                          'problem': 'hour 평균 같은 요약만 나왔어요. 왜 필요한지 몰라서 AI가 가장 흔한 일을 했어요.',
                                                          'hint': '어떤 업무이고 무엇을 확인하려는지 한 문장으로 알려 주세요.'}},
                                                {'id': 'input',
                                                 'label': '입력',
                                                 'why': '파일과 열을 알려요.',
                                                 'options': [{'text': '데이터는 m1_alarms.csv이고, 열은 alarm_id, camera(CAM1~CAM4), hour(0~23), object(울린 '
                                                                      '원인 물체)야.',
                                                              'good': True},
                                                             {'text': '경보 파일이 있어.', 'good': False},
                                                             {'text': '', 'good': False}],
                                                 'miss': {'say': '경보 파일을 읽는 코드를 만들었어요.',
                                                          'expect': 'error',
                                                          'code': 'import pandas as pd\n'
                                                                  "df = pd.read_csv('{VDIR}/alerts.csv')\n"
                                                                  "print(df.groupby('cam')['type'].count())",
                                                          'problem': 'AI가 파일 이름을 몰라 그럴듯한 이름을 지어냈고, 그래서 파일을 찾지 못했어요.',
                                                          'hint': '파일 이름과 열 이름을 그대로 적어 주세요. AI는 전우님 컴퓨터의 파일을 볼 수 없어요.'}},
                                                {'id': 'task',
                                                 'label': '할 일',
                                                 'why': '나누는 규칙을 구체적으로 적어요.',
                                                 'options': [{'text': 'object가 person, vehicle, drone이면 경보 대상, animal, tree, rain, light면 불필요 경보, '
                                                                      'glitch, noise면 오경보로 나눠서 카메라별로 세어 줘.',
                                                              'good': True},
                                                             {'text': '진짜 경보랑 가짜 경보를 나눠 줘.', 'good': False},
                                                             {'text': '', 'good': False}],
                                                 'miss': {'say': '사람과 차량만 진짜로 보고 나머지는 가짜 경보로 나눴어요.',
                                                          'code': 'import pandas as pd\n'
                                                                  "df = pd.read_csv('{VDIR}/m1_alarms.csv')\n"
                                                                  "df['구분'] = df['object'].apply(lambda o: '진짜' if o in ['person', 'vehicle'] else "
                                                                  "'가짜')\n"
                                                                  "display(pd.crosstab(df['camera'], df['구분']).reset_index())",
                                                          'problem': '오류는 없지만 드론이 가짜 쪽으로 들어갔고, 동물과 장비 잡음이 한데 섞였어요. 규칙을 몰라서 AI가 추측한 거예요. **오류가 없다고 맞는 '
                                                                     '결과는 아니에요.**',
                                                          'hint': '값마다 어느 쪽인지 규칙을 적어 주세요.'}},
                                                {'id': 'out',
                                                 'label': '출력 형식',
                                                 'why': '받고 싶은 모양을 알려요.',
                                                 'options': [{'text': '카메라는 행, 구분은 열인 표로 보여 주고 합계 열도 넣어 줘. 카메라별 쌓은 막대그래프를 그리고, 표를 넣은 HTML 보고서를 '
                                                                      'report 변수에 담아 줘.',
                                                              'good': True},
                                                             {'text': '결과를 보여 줘.', 'good': False},
                                                             {'text': '', 'good': False}],
                                                 'miss': {'say': '계산 결과를 출력했어요.',
                                                          'code': 'import pandas as pd\n'
                                                                  "df = pd.read_csv('{VDIR}/m1_alarms.csv')\n"
                                                                  "for k, v in df.groupby(['camera', 'object']).size().items():\n"
                                                                  '    print(k, v)',
                                                          'problem': '숫자는 맞지만 한 줄씩 길게 나와 비교하기 어렵고, 그대로 문서에 넣을 수도 없어요.',
                                                          'hint': '행과 열을 어떻게 놓을지, 문서와 그림도 받을지 적어 주세요.'}}],
                                      'done': {'say': '요청대로 카메라별 집계표와 그림, 보고서를 만들었어요.',
                                               'code': 'import pandas as pd\n'
                                                       'import matplotlib.pyplot as plt\n'
                                                       '\n'
                                                       "df = pd.read_csv('{VDIR}/m1_alarms.csv')\n"
                                                       "KIND = {'person': '경보 대상', 'vehicle': '경보 대상', 'drone': '경보 대상',\n"
                                                       "        'animal': '불필요 경보', 'tree': '불필요 경보', 'rain': '불필요 경보', 'light': '불필요 경보',\n"
                                                       "        'glitch': '오경보', 'noise': '오경보'}\n"
                                                       "df['구분'] = df['object'].map(KIND)\n"
                                                       "table = pd.crosstab(df['camera'], df['구분'])[['경보 대상', '불필요 경보', '오경보']]\n"
                                                       "table['합계'] = table.sum(axis=1)\n"
                                                       'display(table.reset_index())\n'
                                                       '\n'
                                                       "plot = table[['경보 대상', '불필요 경보', '오경보']].rename(columns={'경보 대상': 'target', '불필요 경보': "
                                                       "'nuisance', '오경보': 'false'})\n"
                                                       "ax = plot.plot(kind='bar', stacked=True, figsize=(5.5, 3), rot=0)\n"
                                                       "ax.set_xlabel('camera')\n"
                                                       "ax.set_ylabel('alarms')\n"
                                                       '\n'
                                                       "share = round((df['구분'] == '불필요 경보').mean() * 100, 1)\n"
                                                       "report = ('<h2>카메라별 경보 집계표</h2>'\n"
                                                       "          f'<p>경보 {len(df)}건 가운데 불필요 경보가 {share}%예요. 가상 경보 기록 기준이에요.</p>'\n"
                                                       '          + table.reset_index().to_html(index=False))',
                                               'lesson': '네 칸을 모두 채우니 AI가 추측할 일이 없어졌어요. 경보의 절반 이상이 불필요 경보라는 것이 표에서 바로 보여요. 불필요 경보와 오경보를 나눠 세었으니 대책도 '
                                                         '나눠 세울 수 있어요.'},
                                      'on_ok': [['dm_oh', 'oh', '오, 표가 바로 나왔네요. {call}, 이 프롬프트 저도 좀 빌려 쓸게요!']]}]},
                         {'short': '내 말로 쓰기',
                          'title': '같은 일을 내 말로 요청하기',
                          'lead': '고른 문장을 그대로 외울 필요는 없어요. **네 칸만 채우면** 말투는 자유예요.',
                          'blocks': [{'id': 's3_q',
                                      'type': 'quiz',
                                      'title': '결과를 받은 뒤에는',
                                      'text': 'AI가 만든 코드가 오류 없이 돌았어요. 다음으로 무엇을 하면 좋을까요?',
                                      'options': ['오류가 없으니 그대로 보고한다.', '카메라 하나의 합계를 파일에서 직접 세어 맞는지 확인한다.', '더 어려운 말로 다시 요청한다.'],
                                      'answer': 1,
                                      'explain': '표의 합계와 파일의 건수가 같은지 한 칸만 봐도 큰 실수를 잡을 수 있어요.',
                                      'hints': ['앞 활동에서 오류 없이 이상한 결과가 나온 경우를 떠올려 보세요.', '전부 확인할 필요는 없어요. 한 칸이면 충분해요.']},
                                     {'id': 's3_write',
                                      'type': 'write',
                                      'title': '내 프롬프트 쓰기',
                                      'prompt': '**새 요청**: 시간대별로도 보고 싶어요. 파일은 같은 m1_alarms.csv이고, 0~5시를 심야, 6~17시를 주간, 18~23시를 야간으로 나눠 불필요 경보 수를 보고 '
                                                '싶어요. 네 칸을 채워 내 말로 프롬프트를 써 보세요.',
                                      'example': '경보가 어느 시간대에 많은지 확인하려고 해. 데이터는 m1_alarms.csv이고 열은 alarm_id, camera, hour, object야. hour가 0~5면 심야, '
                                                 '6~17이면 주간, 18~23이면 야간으로 나누고, object가 animal, tree, rain, light인 불필요 경보만 세어 줘. 시간대는 행, 카메라는 열인 표로 '
                                                 '보여 줘.',
                                      'min': 30,
                                      'checklist': ['맥락(왜 필요한지)을 적었어요', '파일 이름과 열 이름을 적었어요', '나누는 규칙을 숫자로 적었어요', '받고 싶은 모양을 적었어요'],
                                      'model_label': '📝 모범 프롬프트',
                                      'placeholder': '꼭 입력해 주세요 · 내 말로 프롬프트를 써 보세요'}]}],
               'outputs': [{'id': 'm1_alarm_table',
                            'title': 'M1 카메라별 경보 집계표',
                            'sections': [{'h': '1. AI에게 보낸 프롬프트와 결과', 'blocks': ['s2_vibe']}, {'h': '2. 내가 쓴 프롬프트', 'blocks': ['s3_write']}]}]},
              {'id': 'm2',
               'no': 2,
               'title': '감시 요구 분석과 탐지 요구 성능 합의',
               'short': '데이터 설명하기',
               'kind': '협업',
               'competencies': ['D2', 'C1'],
               'outline': '담당마다 다른 단위로 적어 낸 요구를 AI에게 설명해 한 표로 맞추면서, 파일과 값의 모양을 알려 주는 법을 익혀요. 결과로 요구 성능을 합의해요.',
               'brief': {'subtitle': '요구 성능 비교표와 합의: 데이터를 설명하기',
                         'time': '약 25분',
                         'goal': '단위가 섞인 요구 데이터를 AI에게 설명해 비교표를 받고, 그 결과를 근거로 탐지 요구 성능을 합의해요.',
                         'why': '놓치지 않기와 불필요 경보 줄이기, 둘 다 가질 수 있을까요? 먼저 담당마다 무엇을 원하는지 같은 단위로 봐야 해요. AI는 ‘0.9’와 ‘95%’가 섞여 있다는 걸 모를 수 있어요.',
                         'case': {'title': '해외 사례: 드론 목격 신고에 다른 물체가 섞여 있었어요',
                                  'source': '미국 국토안보부·FBI 공동 발표 (2024.12)',
                                  'points': ['2024년 말 미국 뉴저지에서 드론 목격 신고가 쏟아졌어요. FBI에 들어온 제보는 5,000건이 넘었어요.',
                                             '국토안보부와 FBI는 신고된 사진과 영상을 검토해 보니 유인 항공기를 드론으로 잘못 본 경우가 포함된 것으로 보인다고 발표했어요.',
                                             '무엇을 경보 대상으로 볼지, 얼마나 잘 잡아야 하는지 먼저 정하지 않으면 숫자가 쌓여도 판단하기 어려워요.']},
                         'team': '강계획은 놓치지 않기를, 서상황과 오영상은 불필요 경보 줄이기를 먼저 봐요. 전우님이 같은 표로 이야기의 근거를 만들어요.',
                         'terms': [{'t': '탐지율 (재현율)',
                                    'en': 'recall',
                                    'fig': 'f_recall',
                                    'd': '실제 대상 가운데 경보로 잡은 비율이에요.',
                                    'easy': '10명이 다가왔을 때 9명을 잡으면 90%예요.',
                                    'use': '단위를 %로 맞춰 달라고 요청해요.'},
                                   {'t': '경보 지연',
                                    'en': 'alarm latency',
                                    'fig': 'f_delay',
                                    'd': '대상이 나타난 뒤 경보가 올라오기까지 걸린 시간이에요.',
                                    'easy': '일이 생기고 알람이 울릴 때까지의 시간이에요.',
                                    'use': '초 단위로 맞춰 달라고 요청해요.'},
                                   {'t': '데이터 정리',
                                    'en': 'data cleaning',
                                    'fig': 'f_clean',
                                    'd': '단위와 형식이 제각각인 값을 하나로 맞추는 일이에요.',
                                    'easy': '‘1min’과 ‘60s’를 같은 값으로 맞추는 일이에요.',
                                    'use': '프롬프트에 ‘단위가 섞여 있으니 먼저 맞춰 줘’라고 적어요.'}],
                         'steps': ['데이터 보여 주기: AI에게 데이터를 설명하는 법', '비교표 받기: 단위를 맞춰 비교표 받기', '기준 합의: 결과로 기준을 합의하고 내 말로 요청하기'],
                         'glossary': [],
                         'outputs': ['M2 탐지 요구 성능 비교와 합의']},
               'start_messages': [['team', 'lim', '{call}, 담당마다 요구 성능을 적어 냈는데 단위가 제각각이에요. 먼저 한 표로 맞춰 볼까요? 파일은 m2_requirements.csv예요.'],
                                  ['team', 'kang', '저는 놓치면 안 된다는 입장이에요.'],
                                  ['team', 'seo', '경보가 많으면 정작 중요한 걸 놓쳐요. 숫자로 봐요.']],
               'finish_messages': [['team', 'lim', '{call}, 같은 단위로 보니 이야기가 쉬워졌네요. 이 기준으로 다음 단계를 진행합시다.']],
               'steps': [{'short': '데이터 보여 주기',
                          'title': 'AI에게 데이터를 설명하는 법',
                          'lead': 'AI는 파일을 볼 수 없어요. **값이 어떻게 적혀 있는지**까지 알려 주는 것이 가장 빠른 방법이에요.',
                          'blocks': [{'id': 's1_info',
                                      'type': 'info',
                                      'title': '데이터를 설명할 때 넣을 것',
                                      'icon': '🗂',
                                      'text': ['열 이름만으로는 부족할 때가 많아요. 값의 모양과 단위까지 알려 주세요.'],
                                      'bullets': ['**파일 이름과 열 이름**: m2_requirements.csv, req_id, owner, target, min_recall, max_delay, '
                                                  'max_unneeded_per_day',
                                                  '**값의 모양**: min_recall은 95%, 0.9, 90이 섞여 있고, max_delay는 30s, 1min, 45 sec가 섞여 있음',
                                                  '**빈칸**: R09의 min_recall이 비어 있음',
                                                  '**샘플 몇 줄**: 실제 값 대신 가상 샘플로 보여 줘도 돼요'],
                                      'figure': 'f_clean'},
                                     {'id': 's1_q',
                                      'type': 'quiz',
                                      'title': '먼저 알려 줄 것',
                                      'text': '요구값에 ‘0.9’와 ‘85%’가 섞여 있어요. 이 사실을 알려 주지 않으면 어떻게 될까요?',
                                      'options': ['AI가 스스로 알아채 항상 맞춰 준다.', '글자로 비교해 85%가 0.9보다 높다고 나오는 식으로 조용히 틀릴 수 있다.', '파일을 읽지 못해 반드시 오류가 난다.'],
                                      'answer': 1,
                                      'explain': '오류 없이 틀린 값이 나와서 더 위험해요. 다음 활동에서 직접 확인해 봐요.',
                                      'hints': ['컴퓨터는 ‘0.9’와 ‘85%’를 숫자로 볼까요?', '오류가 나는 경우와 조용히 틀리는 경우를 나눠 생각해 보세요.']}]},
                         {'short': '비교표 받기',
                          'title': '단위를 맞춰 비교표 받기',
                          'lead': '**단위 설명** 문장을 빼고도 보내 보세요. 결과가 어떻게 달라지는지 비교해 보면 좋아요.',
                          'blocks': [{'id': 's2_vibe',
                                      'type': 'vibe',
                                      'title': '요구 성능 비교표',
                                      'ask': ['yun', '대상별로 담당의 요구 가운데 가장 엄격한 값이 무엇인지 보고 싶어요.'],
                                      'file': 'm2_requirements.csv',
                                      'slots': [{'id': 'input',
                                                 'label': '입력',
                                                 'why': '값의 모양까지 알려요.',
                                                 'options': [{'text': 'm2_requirements.csv의 열은 req_id, owner(plan·control·monitor), '
                                                                      'target(person·vehicle·drone), min_recall, max_delay, max_unneeded_per_day야. '
                                                                      'min_recall은 95%, 0.9, 90이 섞여 있으니 모두 % 숫자로 맞추고, max_delay는 30s, 1min, 45 sec가 '
                                                                      '섞여 있으니 모두 초로 맞춰 줘. 빈칸은 ‘자료 없음’으로 둬 줘.',
                                                              'good': True},
                                                             {'text': 'm2_requirements.csv에 요구가 있어.', 'good': False},
                                                             {'text': '', 'good': False}],
                                                 'miss': {'say': '대상별로 가장 높은 값을 뽑았어요.',
                                                          'code': 'import pandas as pd\n'
                                                                  "df = pd.read_csv('{VDIR}/m2_requirements.csv')\n"
                                                                  "display(df.groupby('target')[['min_recall', 'max_delay']].max().reset_index())",
                                                          'problem': '오류는 없지만 글자로 비교해서 vehicle의 요구가 0.9(90%)가 아니라 85%로 나왔어요. max_delay도 ‘90s’가 '
                                                                     '‘1min’보다 크다고 나왔고요. **값을 설명하지 않으면 조용히 틀려요.**',
                                                          'hint': '값이 어떻게 섞여 있는지, 무엇으로 맞출지 적어 주세요.'}},
                                                {'id': 'task',
                                                 'label': '할 일',
                                                 'why': '무엇을 뽑는지 정해요.',
                                                 'options': [{'text': 'target별로 min_recall은 가장 높은 값, max_delay와 max_unneeded_per_day는 가장 낮은 값을 뽑아 줘. '
                                                                      '어느 담당의 요구인지도 함께 적어 줘.',
                                                              'good': True},
                                                             {'text': '요구를 종합해 줘.', 'good': False},
                                                             {'text': '', 'good': False}],
                                                 'miss': {'say': '평균을 내서 종합했어요.',
                                                          'code': 'import pandas as pd\n'
                                                                  "df = pd.read_csv('{VDIR}/m2_requirements.csv')\n"
                                                                  "print('하루 불필요 경보 평균:', round(df['max_unneeded_per_day'].mean(), 1))",
                                                          'problem': '평균은 아무도 요구하지 않은 값이에요. 무엇을 어떻게 합칠지 말하지 않아 AI가 추측했어요.',
                                                          'hint': '가장 엄격한 값을 뽑을지, 어떻게 합칠지 적어 주세요.'}},
                                                {'id': 'out',
                                                 'label': '출력 형식',
                                                 'why': '회의에서 볼 모양으로 받아요.',
                                                 'options': [{'text': '맞춘 요구 전체 표와 대상별 가장 엄격한 값 표를 보여 주고, 두 표를 넣은 HTML 문서를 report 변수에 담아 줘.',
                                                              'good': True},
                                                             {'text': '결과를 정리해 줘.', 'good': False},
                                                             {'text': '', 'good': False}],
                                                 'miss': {'say': '결과를 문장으로 정리했어요.',
                                                          'code': "print('담당별 요구는 대체로 비슷합니다. 큰 이견은 없어 보입니다.')",
                                                          'problem': '‘대체로 비슷’이라는 말뿐이에요. 실제로는 하루 불필요 경보 요구가 5건과 20건으로 크게 달라요. 숫자가 없어 회의에서 근거로 쓸 수 '
                                                                     '없어요.',
                                                          'hint': '표로 받고 싶다고, 무엇을 넣을지 적어 주세요.'}}],
                                      'done': {'say': '단위를 맞춘 표와 대상별 가장 엄격한 값을 만들었어요.',
                                               'code': 'import pandas as pd\n'
                                                       '\n'
                                                       "df = pd.read_csv('{VDIR}/m2_requirements.csv')\n"
                                                       '\n'
                                                       'def recall_pct(v):\n'
                                                       '    if pd.isna(v):\n'
                                                       '        return None\n'
                                                       '    s = str(v).strip()\n'
                                                       "    if s.endswith('%'):\n"
                                                       '        return float(s[:-1])\n'
                                                       '    x = float(s)\n'
                                                       '    return x * 100 if x <= 1 else x\n'
                                                       '\n'
                                                       'def delay_sec(v):\n'
                                                       "    s = str(v).lower().replace(' ', '')\n"
                                                       "    if s.endswith('min'):\n"
                                                       '        return float(s[:-3]) * 60\n'
                                                       "    if s.endswith('sec'):\n"
                                                       '        return float(s[:-3])\n'
                                                       '    return float(s[:-1])\n'
                                                       '\n'
                                                       "df['재현율(%)'] = df['min_recall'].apply(recall_pct)\n"
                                                       "df['지연(초)'] = df['max_delay'].apply(delay_sec)\n"
                                                       "clean = df[['req_id', 'owner', 'target', '재현율(%)', '지연(초)', 'max_unneeded_per_day']].copy()\n"
                                                       'show = clean.copy()\n'
                                                       "show['재현율(%)'] = show['재현율(%)'].apply(lambda v: '자료 없음' if pd.isna(v) else v)\n"
                                                       'display(show)\n'
                                                       '\n'
                                                       'rows = []\n'
                                                       "for t, s in clean.groupby('target'):\n"
                                                       "    r = s.loc[s['재현율(%)'].idxmax()]\n"
                                                       "    d = s.loc[s['지연(초)'].idxmin()]\n"
                                                       "    u = s.loc[s['max_unneeded_per_day'].idxmin()]\n"
                                                       '    rows.append({\'대상\': t, \'재현율(%)\': f"{r[\'재현율(%)\']} ({r[\'owner\']})", \'지연(초)\': '
                                                       'f"{d[\'지연(초)\']} ({d[\'owner\']})", \'하루 불필요 경보\': f"{u[\'max_unneeded_per_day\']} '
                                                       '({u[\'owner\']})"})\n'
                                                       'strict = pd.DataFrame(rows)\n'
                                                       'display(strict)\n'
                                                       '\n'
                                                       "report = ('<h2>탐지 요구 성능 비교</h2><h3>대상별 가장 엄격한 값</h3>' + strict.to_html(index=False)\n"
                                                       "          + '<h3>단위를 맞춘 전체 요구</h3>' + show.to_html(index=False))",
                                               'lesson': '값의 모양을 알려 주니 조용히 틀리는 일이 사라졌어요. 같은 단위로 보니 재현율 요구는 크게 다르지 않지만, 하루 불필요 경보는 5건과 20건으로 차이가 크다는 '
                                                         '것이 드러났어요.'},
                                      'on_ok': [['team', 'lim', '{call}, 어디서 이견이 큰지 보이네요. 이제 기준을 합의해 봐요.']]}]},
                         {'short': '기준 합의',
                          'title': '결과로 기준을 합의하고 내 말로 요청하기',
                          'lead': 'AI는 계산을 도와줄 뿐이에요. 어떤 기준을 쓸지는 **사람이 근거를 보고 정해요.**',
                          'blocks': [{'id': 's3_decide',
                                      'type': 'decide',
                                      'title': '팀원의 제안 판단하기',
                                      'text': '비교표를 본 팀원들의 제안이에요. 받아들일지 정하고 이유를 적어 주세요.',
                                      'reason_min': 8,
                                      'items': [{'id': 'a',
                                                 'who': 'kang',
                                                 'text': '재현율을 100%로 정해요. 하나라도 놓치면 안 되잖아요.',
                                                 'answer': 'reject',
                                                 'hint': '재현율을 끝까지 올리면 불필요 경보는 어떻게 될까요?',
                                                 'explain': '하나도 놓치지 않게 기준을 정하면 불필요 경보가 크게 늘어요. 대상별로 현실적인 값을 합의하고, 놓칠 때의 보완책을 함께 정해요.',
                                                 'reason': '기준을 끝까지 올리면 불필요 경보가 크게 늘어요. 현실적인 값과 보완책을 함께 정해요.'},
                                                {'id': 'b',
                                                 'who': 'seo',
                                                 'text': '대상별 기준을 먼저 합의해 표로 남기고, 다음 미션부터 이 표로 모델을 평가해요.',
                                                 'answer': 'accept',
                                                 'hint': '기준을 언제 정하는 것이 좋을까요?',
                                                 'explain': '결과를 보기 전에 기준을 정해 두어야 결과에 맞춰 기준을 바꾸는 일을 막을 수 있어요.',
                                                 'reason': '결과를 보기 전에 기준을 정해 둬야 나중에 흔들리지 않아요.'}]},
                                     {'id': 's3_write',
                                      'type': 'write',
                                      'title': '요구서를 요청하는 프롬프트 쓰기',
                                      'prompt': '방금 합의한 내용으로 **감시 요구서**를 AI에게 요청해 보세요. 누가 읽는지, 어떤 항목이 들어가야 하는지, 근거 데이터를 적어 주세요.',
                                      'example': '새로 온 담당자가 읽을 감시 요구서를 만들어 줘. 항목은 감시 대상(person, vehicle, drone), 대상별 재현율 요구, 경보 지연(초), 하루 불필요 경보 수, '
                                                 '아직 남은 이견이야. 근거로 m2_requirements.csv를 단위를 맞춘 표를 넣어 줘. 한 쪽 분량의 HTML로 만들어 줘.',
                                      'min': 30,
                                      'checklist': ['누가 읽는 문서인지 적었어요', '들어갈 항목을 적었어요', '근거 데이터를 적었어요', '분량이나 형식을 적었어요'],
                                      'model_label': '📝 모범 프롬프트',
                                      'placeholder': '꼭 입력해 주세요 · 내 말로 프롬프트를 써 보세요'}]}],
               'outputs': [{'id': 'm2_requirements',
                            'title': 'M2 탐지 요구 성능 비교와 합의',
                            'sections': [{'h': '1. 요구 성능 비교표', 'blocks': ['s2_vibe']},
                                         {'h': '2. 팀원 제안에 대한 결정', 'blocks': ['s3_decide']},
                                         {'h': '3. 요구서를 요청하는 프롬프트', 'blocks': ['s3_write']}]}]},
              {'id': 'm3',
               'no': 3,
               'title': '공개 감시 영상 데이터 구조 분석과 라벨링 규격 정의',
               'short': '오류 고쳐 달라고 하기',
               'kind': '개인',
               'competencies': ['D3'],
               'outline': '바운딩 박스 라벨을 검사하는 도구를 요청하고, 다른 공급처 파일에서 난 오류를 메시지 그대로 알려 고쳐 달라고 요청해요.',
               'brief': {'subtitle': '라벨 검증기와 오류 수정: 다시 요청하는 법',
                         'time': '약 25분',
                         'goal': '라벨 검증기를 요청하고, 오류가 나면 메시지와 원하는 동작, 고칠 범위를 담아 수정을 요청해요.',
                         'why': '라벨이 틀리면 모델 성능 숫자도 틀릴까요? 그래서 라벨부터 검사해요. 그런데 파일이 바뀌면 오류가 자주 나요. 오류를 만났을 때 어떻게 다시 요청하는지가 바이브 코딩의 절반이에요.',
                         'case': {'title': '해외 사례: 널리 쓰는 공개 데이터에도 라벨 오류가 있었어요',
                                  'source': 'Northcutt 외, NeurIPS 2021, arXiv 2103.14749',
                                  'points': ['MIT 등의 연구진은 널리 쓰는 공개 데이터 10종의 테스트 데이터를 살펴, 평균 3.3% 이상의 라벨 오류를 찾았어요. ImageNet 검증 데이터는 2,916건, 약 6%였어요.',
                                             '라벨을 고쳐도 모델 순위가 그대로인 경우가 많았지만, 오류 비율이 조금만 달라져도 어떤 모델이 나아 보이는지가 바뀔 수 있었어요.',
                                             '라벨을 검사하지 않으면 성능 숫자를 믿기 어려워요. 이번 미션의 검증기가 그 일을 해요.']},
                         'team': '윤분석(데이터 분석 담당)이 라벨 파일을 넘겨 줘요. 전우님이 AI로 검증기를 받고, 오류가 나면 고쳐 달라고 요청해요.',
                         'terms': [{'t': '바운딩 박스',
                                    'en': 'bounding box',
                                    'fig': 'f_box',
                                    'd': '영상에서 물체를 감싸는 네모 칸이에요. 왼쪽 위 좌표(x, y)와 폭·높이(w, h)로 적어요.',
                                    'easy': '사진 속 물체에 네모를 그린 거예요.',
                                    'use': '화면 밖으로 나가거나 크기가 0이면 라벨 오류예요.'},
                                   {'t': '라벨링 규격',
                                    'en': 'annotation schema',
                                    'fig': 'f_schema',
                                    'd': '라벨을 어떤 값으로, 어떤 기준으로 붙일지 정한 약속이에요.',
                                    'easy': '모두가 같은 양식으로 적게 하는 작성 규칙이에요.',
                                    'use': '규격을 그대로 프롬프트에 적으면 AI가 검사 코드를 만들어요.'},
                                   {'t': '오류 메시지',
                                    'en': 'error message',
                                    'fig': 'f_debug',
                                    'd': '코드가 멈춘 이유를 컴퓨터가 알려 주는 글이에요.',
                                    'easy': '맨 아래 줄이 가장 중요해요. 그대로 복사해 AI에게 주면 돼요.',
                                    'use': 'ValueError는 값의 모양이 기대와 다르다는 뜻이에요.'}],
                         'steps': ['검증기 받기: 규격을 그대로 적어 검증기 받기', '오류 고치기: 오류 메시지로 수정 요청하기', '내 말로 쓰기: 다른 오류를 내 말로 요청하기'],
                         'glossary': [],
                         'outputs': ['M3 라벨 검증 결과와 수정 기록']},
               'start_messages': [['dm_yun', 'yun', '{call}, 감시 영상 라벨 파일이에요. 파일은 m3_labels.csv예요. 규격에 어긋난 라벨을 찾는 표를 받아 볼래요?']],
               'finish_messages': [['dm_yun', 'yun', '{call}, 고칠 곳만 고쳐 주셨네요. 이제 어느 공급처 파일이든 먼저 이걸로 돌릴게요.']],
               'steps': [{'short': '검증기 받기',
                          'title': '규격을 그대로 적어 검증기 받기',
                          'lead': '규격을 **조건 그대로** 적으면 AI가 검사 코드로 바꿔 줘요.',
                          'blocks': [{'id': 's1_vibe',
                                      'type': 'vibe',
                                      'title': '라벨 검증기',
                                      'ask': ['yun', '규격에 어긋난 라벨을 찾고 이유까지 보고 싶어요.'],
                                      'file': 'm3_labels.csv',
                                      'slots': [{'id': 'input',
                                                 'label': '입력',
                                                 'why': '열의 뜻을 알려요.',
                                                 'options': [{'text': 'm3_labels.csv의 열은 frame_id, clip_id, class, x, y(왼쪽 위 좌표), w, h(폭과 높이), '
                                                                      'img_w, img_h(화면 크기)야.',
                                                              'good': True},
                                                             {'text': '라벨 CSV가 있어.', 'good': False},
                                                             {'text': '', 'good': False}],
                                                 'miss': {'say': '라벨을 읽어 면적을 계산했어요.',
                                                          'expect': 'error',
                                                          'code': 'import pandas as pd\n'
                                                                  "df = pd.read_csv('{VDIR}/m3_labels.csv')\n"
                                                                  "print((df['width'] * df['height']).mean())",
                                                          'problem': 'AI가 열 이름을 width, height로 추측해서 KeyError가 났어요.',
                                                          'hint': '열 이름과 뜻을 그대로 적어 주세요.'}},
                                                {'id': 'task',
                                                 'label': '할 일',
                                                 'why': '규격을 조건으로 적어요.',
                                                 'options': [{'text': 'class는 person, vehicle, drone, bird만 허용해. w나 h가 0 이하면 ‘크기’, x+w가 img_w를 넘거나 '
                                                                      'y+h가 img_h를 넘으면 ‘화면 밖’, class가 목록에 없으면 ‘클래스’로 표시해 줘.',
                                                              'good': True},
                                                             {'text': '라벨이 잘못된 걸 찾아 줘.', 'good': False},
                                                             {'text': '', 'good': False}],
                                                 'miss': {'say': '빈칸이 있는지 찾았어요.',
                                                          'code': 'import pandas as pd\n'
                                                                  "df = pd.read_csv('{VDIR}/m3_labels.csv')\n"
                                                                  "print('빈칸 수:', int(df.isna().sum().sum()))\n"
                                                                  "print('문제 없어 보입니다.')",
                                                          'problem': '오류는 없지만 빈칸만 찾아 ‘문제 없음’이라고 했어요. 화면 밖 박스나 크기 0인 박스는 그대로예요. ‘잘못된 것’이 무엇인지 말하지 '
                                                                     '않아서예요.',
                                                          'hint': '무엇이 잘못인지 조건으로 적어 주세요.'}},
                                                {'id': 'out',
                                                 'label': '출력 형식',
                                                 'why': '표의 모양을 정해요.',
                                                 'options': [{'text': '오류 행만 frame_id, clip_id, class, 오류 이유 순서의 표로 보여 주고, 오류 이유별 건수도 함께 보여 줘.',
                                                              'good': True},
                                                             {'text': '보기 좋게 보여 줘.', 'good': False},
                                                             {'text': '', 'good': False}],
                                                 'miss': {'say': '보기 좋게 문장으로 정리했어요.',
                                                          'code': 'import pandas as pd\n'
                                                                  "df = pd.read_csv('{VDIR}/m3_labels.csv')\n"
                                                                  "print(f'라벨 {len(df)}건의 검사를 마쳤습니다!')",
                                                          'problem': '‘보기 좋게’는 사람마다 달라요. 결과 없이 검사했다는 말만 나왔어요.',
                                                          'hint': '열 순서까지 적어 주세요.'}}],
                                      'done': {'say': '라벨 검증기를 만들었어요.',
                                               'code': 'import pandas as pd\n'
                                                       '\n'
                                                       "df = pd.read_csv('{VDIR}/m3_labels.csv')\n"
                                                       "CLASSES = ['person', 'vehicle', 'drone', 'bird']\n"
                                                       '\n'
                                                       'def check(r):\n'
                                                       '    why = []\n'
                                                       "    if r['class'] not in CLASSES:\n"
                                                       "        why.append('클래스')\n"
                                                       "    if r['w'] <= 0 or r['h'] <= 0:\n"
                                                       "        why.append('크기')\n"
                                                       "    if r['x'] + r['w'] > r['img_w'] or r['y'] + r['h'] > r['img_h']:\n"
                                                       "        why.append('화면 밖')\n"
                                                       "    return ', '.join(why) if why else '정상'\n"
                                                       '\n'
                                                       "df['오류 이유'] = df.apply(check, axis=1)\n"
                                                       "bad = df[df['오류 이유'] != '정상']\n"
                                                       "display(bad[['frame_id', 'clip_id', 'class', '오류 이유']])\n"
                                                       "display(bad['오류 이유'].value_counts().reset_index())",
                                               'lesson': '규격을 조건으로 적으니 일부러 넣은 오류 다섯 건을 모두 찾았어요. ‘Drone’처럼 대문자로 적은 값도 규격 위반으로 잡혔어요.'},
                                      'on_ok': [['dm_yun',
                                                 'yun',
                                                 '{call}, 잘 잡네요. 그런데 다른 공급처 파일 m3_labels_new.csv도 돌려 보고 싶어요. AI에게 파일만 바꿔 달라고 해 볼래요?']]}]},
                         {'short': '오류 고치기',
                          'title': '오류 메시지로 수정 요청하기',
                          'lead': '파일을 바꿨더니 **오류**가 났어요. 어떻게 다시 요청할까요?',
                          'blocks': [{'id': 's2_vibe',
                                      'type': 'vibe',
                                      'title': '오류 고쳐 달라고 하기',
                                      'ask': ['yun', '다른 공급처 파일을 넣었더니 오류가 나요. 저도 원인을 잘 모르겠어요… 부탁해요!'],
                                      'file': 'm3_labels_new.csv',
                                      'slots': [{'id': 'what',
                                                 'label': '무슨 일이 있었는지',
                                                 'why': '오류 메시지를 그대로 붙여 넣어요.',
                                                 'options': [{'text': 'm3_labels_new.csv를 검사하다가 이 오류가 났어: ValueError: could not convert string to '
                                                                      "float: '300,5'",
                                                              'good': True},
                                                             {'text': '안 돼. 다시 해 줘.', 'good': False},
                                                             {'text': '', 'good': False}],
                                                 'miss': {'say': '다시 만들었어요. 이번에는 오류가 나는 행을 건너뛰었어요.',
                                                          'code': 'import pandas as pd\n'
                                                                  "df = pd.read_csv('{VDIR}/m3_labels_new.csv', dtype=str, keep_default_na=False)\n"
                                                                  'rows = []\n'
                                                                  'for _, r in df.iterrows():\n'
                                                                  '    try:\n'
                                                                  "        float(r['x'])\n"
                                                                  '        rows.append(r)\n'
                                                                  '    except ValueError:\n'
                                                                  '        pass\n'
                                                                  "display(pd.DataFrame(rows)[['frame_id', 'clip_id', 'class']])",
                                                          'problem': '오류는 사라졌지만 문제 있는 행을 말없이 버렸어요. 무엇이 안 되는지 몰라서 AI가 가장 쉬운 길을 택했어요.',
                                                          'hint': '오류 메시지를 그대로 알려 주세요.'}},
                                                {'id': 'fix',
                                                 'label': '어떻게 고칠지',
                                                 'why': '원하는 동작과 범위를 정해요.',
                                                 'options': [{'text': '이 공급처는 소수점을 쉼표로 적어. 쉼표는 점으로 바꿔 읽고, 그래도 숫자가 아니면 ‘형식 오류’로 표시해 줘. 다른 규칙은 그대로 두고, '
                                                                      '표를 넣은 HTML 라벨 검증 결과를 report 변수에 담아 줘.',
                                                              'good': True},
                                                             {'text': '전부 새로 짜 줘.', 'good': False},
                                                             {'text': '', 'good': False}],
                                                 'miss': {'say': '처음부터 새로 짰어요. 더 간단하게 정리했어요.',
                                                          'code': 'import pandas as pd\n'
                                                                  "df = pd.read_csv('{VDIR}/m3_labels_new.csv', dtype=str)\n"
                                                                  "print(df[['frame_id', 'class']].to_string(index=False))",
                                                          'problem': '오류는 사라졌지만 검사 규칙까지 함께 사라졌어요. 범위를 정하지 않으면 잘 되던 부분도 바뀌어요.',
                                                          'hint': '무엇은 그대로 두고 어디만 어떻게 바꿀지 적어 주세요.'}}],
                                      'done': {'say': '숫자를 읽는 부분만 고치고, 보고서를 추가했어요.',
                                               'code': 'import pandas as pd\n'
                                                       '\n'
                                                       "df = pd.read_csv('{VDIR}/m3_labels_new.csv', dtype=str, keep_default_na=False)\n"
                                                       "CLASSES = ['person', 'vehicle', 'drone', 'bird']\n"
                                                       '\n'
                                                       'def to_num(v):\n'
                                                       '    try:\n'
                                                       "        return float(str(v).replace(',', '.'))\n"
                                                       '    except ValueError:\n'
                                                       '        return None\n'
                                                       '\n'
                                                       'def check(r):\n'
                                                       "    x, y, w, h = [to_num(r[c]) for c in ['x', 'y', 'w', 'h']]\n"
                                                       '    if None in (x, y, w, h):\n'
                                                       "        return '형식 오류'\n"
                                                       '    why = []\n'
                                                       "    if r['class'] not in CLASSES:\n"
                                                       "        why.append('클래스')\n"
                                                       '    if w <= 0 or h <= 0:\n'
                                                       "        why.append('크기')\n"
                                                       "    if x + w > float(r['img_w']) or y + h > float(r['img_h']):\n"
                                                       "        why.append('화면 밖')\n"
                                                       "    return ', '.join(why) if why else '정상'\n"
                                                       '\n'
                                                       "df['오류 이유'] = df.apply(check, axis=1)\n"
                                                       "table = df[['frame_id', 'clip_id', 'class', 'x', '오류 이유']]\n"
                                                       'display(table)\n'
                                                       '\n'
                                                       "bad = table[table['오류 이유'] != '정상']\n"
                                                       "report = ('<h2>라벨 검증 결과</h2>'\n"
                                                       '          f"<p>{len(table)}건 가운데 오류 {len(bad)}건이에요. 쉼표 소수점은 점으로 읽었어요.</p>"\n'
                                                       '          + table.to_html(index=False))',
                                               'lesson': '오류 메시지와 원하는 동작, 고칠 범위를 함께 적으니 한 번에 고쳐졌어요. 쉼표 소수점은 정상으로 읽고, 숫자가 아닌 n/a와 화면 밖 박스는 오류로 남았어요.'},
                                      'start': {'say': '파일을 m3_labels_new.csv로 바꿨어요.',
                                                'expect': 'error',
                                                'code': 'import pandas as pd\n'
                                                        '\n'
                                                        "df = pd.read_csv('{VDIR}/m3_labels_new.csv', dtype=str, keep_default_na=False)\n"
                                                        "CLASSES = ['person', 'vehicle', 'drone', 'bird']\n"
                                                        '\n'
                                                        'def check(r):\n'
                                                        "    x, y, w, h = [float(r[c]) for c in ['x', 'y', 'w', 'h']]\n"
                                                        '    why = []\n'
                                                        "    if r['class'] not in CLASSES:\n"
                                                        "        why.append('클래스')\n"
                                                        '    if w <= 0 or h <= 0:\n'
                                                        "        why.append('크기')\n"
                                                        "    if x + w > float(r['img_w']) or y + h > float(r['img_h']):\n"
                                                        "        why.append('화면 밖')\n"
                                                        "    return ', '.join(why) if why else '정상'\n"
                                                        '\n'
                                                        "df['오류 이유'] = df.apply(check, axis=1)\n"
                                                        "display(df[['frame_id', 'class', '오류 이유']])"}}]},
                         {'short': '내 말로 쓰기',
                          'title': '다른 오류를 내 말로 요청하기',
                          'lead': '오류를 만나면 **겁먹지 말고 그대로 붙여 넣으세요.** 코드를 다 읽지 못해도 수정을 요청할 수 있어요.',
                          'blocks': [{'id': 's3_write',
                                      'type': 'write',
                                      'title': '수정 프롬프트 쓰기',
                                      'prompt': "**새 상황**: 또 다른 공급처 파일을 넣었더니 `KeyError: 'class'`가 났어요. 이 파일의 실제 열 이름은 label이에요. 잘 되던 검사 규칙은 그대로 두고 "
                                                '싶어요. 어떻게 요청할까요?',
                                      'example': "또 다른 공급처 파일을 넣었더니 KeyError: 'class'가 났어. 이 파일의 실제 열 이름은 label이야. 파일을 읽은 뒤 label을 class로 바꾸는 줄만 "
                                                 '추가하고, 검사 규칙과 나머지 코드는 그대로 둬.',
                                      'min': 30,
                                      'checklist': ['오류 메시지를 그대로 적었어요', '원인이 될 만한 사실(실제 열 이름)을 적었어요', '원하는 동작을 적었어요', '그대로 둘 부분을 적었어요'],
                                      'model_label': '📝 모범 프롬프트',
                                      'placeholder': '꼭 입력해 주세요 · 내 말로 프롬프트를 써 보세요'}]}],
               'outputs': [{'id': 'm3_labels',
                            'title': 'M3 라벨 검증 결과와 수정 기록',
                            'sections': [{'h': '1. 라벨 검증기', 'blocks': ['s1_vibe']},
                                         {'h': '2. 오류 수정과 보고서', 'blocks': ['s2_vibe']},
                                         {'h': '3. 내가 쓴 수정 프롬프트', 'blocks': ['s3_write']}]}]},
              {'id': 'm4',
               'no': 4,
               'title': '야간·악천후 영상 전처리와 침입 탐지 영역(ROI) 설정',
               'short': '쪼개서 요청하기',
               'kind': '협업',
               'competencies': ['D4', 'C1'],
               'outline': '점 하나가 침입 탐지 영역 안에 있는지 보는 작은 함수부터 만들어 달라고 요청해요. 영역을 바꿔 다시 돌리면서 어느 영역이 나은지 합의의 근거를 만들어요.',
               'brief': {'subtitle': '침입 탐지 영역 함수와 구역 비교: 쪼개서 요청하기',
                         'time': '약 30분',
                         'goal': '큰 일을 작은 단계로 쪼개 요청하고, 영역을 바꿔 돌릴 수 있는 함수로 조건별 경보 수를 비교해 합의의 근거를 만들어요.',
                         'why': '화면을 또렷하게 하면 표적도 잘 찾을까요? 전처리만큼 어디를 볼지도 중요해요. 영역을 넓히면 많이 잡지만 불필요 경보도 늘어요. 이렇게 조건이 많을 때 한 번에 전부 요청하면 어디가 틀렸는지 찾기 어려워요.',
                         'case': {'title': '해외 사례: 영국 정부의 영상분석 평가 기준과 밤·안개의 한계',
                                  'source': '영국 정부 i-LIDS 안내, arXiv 2507.22650 (2025.7)',
                                  'points': ['영국 정부가 만든 영상분석 평가 자료 i-LIDS에는 ‘sterile zone’ 시나리오가 있어요. 출입이 제한된 구역에 사람이 있는지를 영상분석이 찾아내는지 평가해요.',
                                             '한편 2025년의 한 드론 탐지 연구는 일반 카메라 영상이 밤, 안개, 눈부심 같은 조건에서 제대로 작동하지 못할 수 있어 열화상 영상과 함께 써야 한다고 설명해요.',
                                             '어디를 볼지와 어떤 조건에서 볼지를 나눠 따져야 해요. 이번 미션에서는 그 비교를 함수로 만들어요.']},
                         'team': '강계획은 영역을 넓히자고 하고, 서상황은 불필요 경보를 줄이자고 해요. 민시설은 나무를 자를 수 없는 구역을 알려 줘요.',
                         'terms': [{'t': '침입 탐지 영역 (ROI)',
                                    'en': 'region of interest',
                                    'fig': 'f_roi',
                                    'd': '영상에서 경보를 올릴 곳으로 정한 영역이에요. 바람에 흔들리는 나무처럼 빼는 곳은 제외 영역으로 둬요.',
                                    'easy': '‘여기서 움직이면 알려 줘’라고 그어 둔 구역이에요.',
                                    'use': '좌표를 인자로 받는 함수로 요청하면 바꿔 보기 쉬워요.'},
                                   {'t': '영상 전처리',
                                    'en': 'CLAHE, median filter',
                                    'fig': 'f_pre',
                                    'd': 'CLAHE는 어두운 곳의 대비를 부분별로 높이고, 미디언 필터는 점잡음을 지우는 방법이에요.',
                                    'easy': '흐린 사진을 또렷하게 손보는 일이에요.',
                                    'use': '이번에는 전처리 뒤에 나온 움직임 기록만 다뤄요.'},
                                   {'t': '함수로 나누기',
                                    'en': 'function',
                                    'fig': 'f_param',
                                    'd': '한 가지 일을 하는 코드를 이름 붙여 묶고, 값만 바꿔 다시 쓰게 하는 것이에요.',
                                    'easy': '숫자만 바꿔 넣으면 되는 계산기를 만드는 거예요.',
                                    'use': '프롬프트에 ‘영역 좌표를 인자로 받는 함수로 만들어 줘’라고 적어요.'}],
                         'steps': ['쪼개기: 큰 일을 작은 단계로 쪼개기', '함수로 만들기: 영역을 바꿔 돌릴 수 있는 함수 요청하기', '영역 합의: 결과로 조율하고 내 말로 요청하기'],
                         'glossary': [],
                         'outputs': ['M4 침입 탐지 영역 비교와 합의 기록']},
               'start_messages': [['team', 'kang', '{call}, 침입 탐지 영역을 어떻게 그을지 정해야 해요. 저는 넓게 보자는 입장이에요.'],
                                  ['team', 'seo', '넓히면 경보가 늘어요. 움직임 기록은 m4_events.csv에 있어요.'],
                                  ['team', 'min', '오른쪽 아래 나무 구역은 자를 수 없어요. 제외 영역으로 빼면 어떨까요?']],
               'finish_messages': [['team', 'lim', '{call}, 같은 표를 보니 이야기가 빨라졌네요. 이 영역으로 갑시다.']],
               'steps': [{'short': '쪼개기',
                          'title': '큰 일을 작은 단계로 쪼개기',
                          'lead': '한 번에 전부 요청하면 어디가 틀렸는지 찾기 어려워요. **작게 만들고 확인한 뒤 넓혀요.**',
                          'blocks': [{'id': 's1_info',
                                      'type': 'info',
                                      'title': '비교할 영역',
                                      'icon': '📐',
                                      'text': ['좌표는 가로 0~100, 세로 0~100이고, y가 작을수록 울타리에 가까워요. 팀원들의 의견을 영역 두 개로 정리했어요.'],
                                      'table': {'head': ['이름', '좌표 (x0, y0, x1, y1)', '제안한 사람'],
                                                'rows': [['영역 A', '0, 0, 100, 30', '강계획 (넓게)'],
                                                         ['영역 B', '0, 0, 100, 15', '서상황 (좁게)'],
                                                         ['제외 영역', '70, 0, 100, 14', '민시설 (나무)']]},
                                      'figure': 'f_roi'},
                                     {'id': 's1_order',
                                      'type': 'order',
                                      'title': '요청 순서 정하기',
                                      'text': '영역 비교 도구를 만들려고 해요. 어떤 순서로 요청하면 좋을까요?',
                                      'items': [{'id': 'a', 'text': '데이터 모양을 알려 주고 파일을 읽게 하기', 'hint': 'AI가 파일을 먼저 알아야 해요.'},
                                                {'id': 'b', 'text': '점 하나가 영역 안인지 보는 함수 만들기', 'hint': '작은 단위부터 시작해요.'},
                                                {'id': 'c', 'text': '답을 아는 점 두 개로 함수 확인하기', 'hint': '넓히기 전에 맞는지 봐요.'},
                                                {'id': 'd', 'text': '전체 기록에 적용해 영역별 비교표 만들기', 'hint': '확인한 함수를 넓혀요.'}],
                                      'answer': ['a', 'b', 'c', 'd'],
                                      'explain': '작은 단위로 만들고 확인한 뒤 넓히면, 틀렸을 때 어디가 문제인지 바로 알 수 있어요.'}]},
                         {'short': '함수로 만들기',
                          'title': '영역을 바꿔 돌릴 수 있는 함수 요청하기',
                          'lead': '좌표를 코드 속에 박아 두면 회의 때마다 고쳐야 해요. **인자로 받는 함수**로 요청해요.',
                          'blocks': [{'id': 's2_vibe',
                                      'type': 'vibe',
                                      'title': '침입 탐지 영역 함수와 비교표',
                                      'ask': ['seo', '영역별로 경보 수와 대상을 얼마나 잡는지, 밤과 안개에서는 어떤지 보고 싶어요.'],
                                      'file': 'm4_events.csv',
                                      'slots': [{'id': 'small',
                                                 'label': '작게 시작',
                                                 'why': '함수 하나부터 만들어요.',
                                                 'options': [{'text': 'm4_events.csv의 열은 event_id, x, y, condition(day·night·fog), '
                                                                      'kind(person·vehicle·animal·tree)야. 점 하나와 영역을 받아 안에 있으면 True를 돌려주는 in_zone 함수를 '
                                                                      '먼저 만들어 줘. 영역 A에서 (50, 10)은 True, (50, 40)은 False인지 먼저 확인해 줘.',
                                                              'good': True},
                                                             {'text': '영역 비교 프로그램을 한 번에 다 만들어 줘.', 'good': False},
                                                             {'text': '', 'good': False}],
                                                 'miss': {'say': '한 번에 다 만들었어요. 결과는 이래요.',
                                                          'code': 'import pandas as pd\n'
                                                                  "df = pd.read_csv('{VDIR}/m4_events.csv')\n"
                                                                  "print('영역 A 경보:', int((df['y'] < 30).sum()))\n"
                                                                  "print('영역 B 경보:', int((df['y'] < 15).sum()))",
                                                          'problem': '숫자 두 개만 나왔어요. 제외 영역도, 조건별 비교도 빠졌는데, 한 번에 받아서 어디가 빠졌는지 알아채기 어려워요.',
                                                          'hint': '점 하나를 판단하는 함수부터 요청해 주세요.'}},
                                                {'id': 'param',
                                                 'label': '조건 바꾸기',
                                                 'why': '영역을 인자로 받아요.',
                                                 'options': [{'text': '영역은 (x0, y0, x1, y1) 인자로 받게 하고, 제외 영역 (70, 0, 100, 14) 안의 점은 항상 False로 해 줘. '
                                                                      '영역 A는 (0, 0, 100, 30), 영역 B는 (0, 0, 100, 15)야.',
                                                              'good': True},
                                                             {'text': '영역은 코드에 적당히 넣어 줘.', 'good': False},
                                                             {'text': '', 'good': False}],
                                                 'miss': {'say': '영역을 코드 안에 넣었어요.',
                                                          'code': 'import pandas as pd\n'
                                                                  "df = pd.read_csv('{VDIR}/m4_events.csv')\n"
                                                                  "inside = df[(df['y'] < 30) & (df['x'] < 100)]\n"
                                                                  "display(inside.groupby('kind').size().reset_index(name='건수'))",
                                                          'problem': '영역이 코드에 박혀 있어 바꿔 보기 어렵고, 제외 영역이 빠져 나무가 그대로 경보로 잡혔어요.',
                                                          'hint': '영역을 인자로 받게 하고, 제외 영역을 적어 주세요.'}},
                                                {'id': 'out',
                                                 'label': '비교표',
                                                 'why': '함수를 전체에 적용해요.',
                                                 'options': [{'text': '영역과 condition별로 경보 수, 대상(person·vehicle) 포착률(%), 불필요 경보(animal·tree) 수를 표로 보여 '
                                                                      '주고, 점과 영역을 함께 그려 줘. 표를 넣은 HTML 비교 문서를 report 변수에 담아 줘.',
                                                              'good': True},
                                                             {'text': '결과를 정리해 줘.', 'good': False},
                                                             {'text': '', 'good': False}],
                                                 'miss': {'say': '두 영역을 검토했어요.',
                                                          'code': "print('영역 A와 B를 검토했습니다. 둘 다 장단점이 있습니다.')",
                                                          'problem': '‘장단점이 있습니다’만 있어요. 무엇이 얼마나 다른지 없어 회의에서 쓸 근거가 안 돼요.',
                                                          'hint': '무엇을 어떤 조건으로 비교할지 적어 주세요.'}}],
                                      'done': {'say': 'in_zone 함수를 점 두 개로 확인한 뒤, 영역 두 개를 조건별로 비교했어요.',
                                               'code': 'import pandas as pd\n'
                                                       'import matplotlib.pyplot as plt\n'
                                                       'from matplotlib.patches import Rectangle\n'
                                                       '\n'
                                                       "df = pd.read_csv('{VDIR}/m4_events.csv')\n"
                                                       'EXCLUDE = (70, 0, 100, 14)\n'
                                                       "ZONES = {'A': (0, 0, 100, 30), 'B': (0, 0, 100, 15)}\n"
                                                       '\n'
                                                       'def inside(x, y, z):\n'
                                                       '    return z[0] <= x <= z[2] and z[1] <= y <= z[3]\n'
                                                       '\n'
                                                       'def in_zone(x, y, zone, exclude=EXCLUDE):\n'
                                                       '    return inside(x, y, zone) and not inside(x, y, exclude)\n'
                                                       '\n'
                                                       '# 작은 확인\n'
                                                       "assert in_zone(50, 10, ZONES['A']) is True\n"
                                                       "assert in_zone(50, 40, ZONES['A']) is False\n"
                                                       "print('확인: in_zone 통과')\n"
                                                       '\n'
                                                       "TARGET = ['person', 'vehicle']\n"
                                                       'rows = []\n'
                                                       'for name, z in ZONES.items():\n'
                                                       "    for cond in ['day', 'night', 'fog']:\n"
                                                       "        s = df[df['condition'] == cond]\n"
                                                       "        hit = s[s.apply(lambda r: in_zone(r['x'], r['y'], z), axis=1)]\n"
                                                       "        t_all = s[s['kind'].isin(TARGET)]\n"
                                                       "        t_hit = hit[hit['kind'].isin(TARGET)]\n"
                                                       "        rows.append({'영역': name, '조건': cond, '경보 수': len(hit), '대상 수': len(t_all),\n"
                                                       "                     '포착률(%)': round(len(t_hit) / len(t_all) * 100, 1) if len(t_all) else "
                                                       "'자료 없음',\n"
                                                       "                     '불필요 경보': len(hit) - len(t_hit)})\n"
                                                       'table = pd.DataFrame(rows)\n'
                                                       'display(table)\n'
                                                       '\n'
                                                       'fig, ax = plt.subplots(figsize=(5.5, 3.2))\n'
                                                       "for kind, s in df.groupby('kind'):\n"
                                                       "    ax.scatter(s['x'], s['y'], s=14, label=kind)\n"
                                                       "for z, c in [(ZONES['A'], 'tab:blue'), (ZONES['B'], 'tab:orange'), (EXCLUDE, 'gray')]:\n"
                                                       '    ax.add_patch(Rectangle((z[0], z[1]), z[2] - z[0], z[3] - z[1], fill=False, edgecolor=c, '
                                                       'linewidth=1.6))\n'
                                                       'ax.legend(fontsize=7)\n'
                                                       "ax.set_xlabel('x')\n"
                                                       "ax.set_ylabel('y')\n"
                                                       '\n'
                                                       "report = ('<h2>침입 탐지 영역 비교</h2><p>영역 A (0,0,100,30), 영역 B (0,0,100,15), 제외 영역 (70,0,100,14). "
                                                       "가상 기록 기준이에요.</p>' + table.to_html(index=False))",
                                               'lesson': '영역을 좁히면 불필요 경보는 줄지만 대상도 함께 놓쳐요. 함수로 나누니 좌표를 바꿔도 한 줄이면 다시 비교할 수 있고, 어느 쪽이 얼마를 포기하는지 숫자로 '
                                                         '보여요.'},
                                      'on_ok': [['team', 'kang', '{call}, 좁히면 얼마나 놓치는지 숫자로 나오네요. 이걸로 이야기해 봐요.']]}]},
                         {'short': '영역 합의',
                          'title': '결과로 조율하고 내 말로 요청하기',
                          'lead': 'AI의 표는 근거일 뿐이에요. 어디까지 볼지는 **사람이 합의해요.**',
                          'blocks': [{'id': 's3_decide',
                                      'type': 'decide',
                                      'title': '침입 탐지 영역 합의',
                                      'text': '비교표를 본 팀원들의 의견이에요. 받아들일지 정하고 이유를 적어 주세요.',
                                      'reason_min': 8,
                                      'items': [{'id': 'a',
                                                 'who': 'seo',
                                                 'text': '불필요 경보가 적으니 영역 B로 가요. 놓치는 대상은 어쩔 수 없잖아요.',
                                                 'answer': 'reject',
                                                 'hint': 'M2에서 합의한 재현율 요구를 떠올려 보세요.',
                                                 'explain': '표를 보면 영역 B는 대상을 많이 놓쳐요. 합의한 요구를 지키는 범위에서 불필요 경보를 줄여야 해요.',
                                                 'reason': '대상을 많이 놓치면 합의한 요구를 못 지켜요. 요구를 지키면서 줄일 방법을 찾아야 해요.'},
                                                {'id': 'b',
                                                 'who': 'min',
                                                 'text': '영역 A로 가되 나무 구역은 제외 영역으로 빼고, 그 구역은 다른 카메라가 보도록 표시해 두죠.',
                                                 'answer': 'accept',
                                                 'hint': '제외 영역에 대상이 들어오면 어떻게 될까요?',
                                                 'explain': '제외 영역은 눈을 감는 곳이에요. 빠진 곳을 다른 수단으로 보완한다고 적어 두어야 해요.',
                                                 'reason': '제외한 곳을 다른 수단으로 보완한다고 남겨야 빈틈을 관리할 수 있어요.'}]},
                                     {'id': 's3_write',
                                      'type': 'write',
                                      'title': '쪼개서 요청하는 프롬프트 쓰기',
                                      'prompt': '**새 요청**: 사전 경보 영역을 하나 더 두고 싶어요. 울타리에서 멀리 떨어진 영역 C (0, 30, 100, 45)에 들어오면 경보 대신 ‘사전 알림’으로 표시해요. 어떤 '
                                                '순서로, 무엇을 고쳐 달라고 할까요?',
                                      'example': '영역 C (0, 30, 100, 45)를 사전 경보 영역으로 추가하고 싶어. 먼저 ZONES에 C만 추가하고 in_zone 함수는 그대로 둬. 점 (50, 35)가 C 안이고 '
                                                 '(50, 50)이 C 밖인지 확인한 뒤, 영역 C에 들어온 기록은 경보 대신 ‘사전 알림’으로 세어 표에 추가해 줘.',
                                      'min': 30,
                                      'checklist': ['작은 단계부터 요청했어요', '영역을 좌표로 적었어요', '답을 아는 점으로 확인하게 했어요', '그대로 둘 부분과 결과의 모양을 적었어요'],
                                      'model_label': '📝 모범 프롬프트',
                                      'placeholder': '꼭 입력해 주세요 · 내 말로 프롬프트를 써 보세요'}]}],
               'outputs': [{'id': 'm4_zone',
                            'title': 'M4 침입 탐지 영역 비교와 합의 기록',
                            'sections': [{'h': '1. 영역 함수와 비교표', 'blocks': ['s2_vibe']},
                                         {'h': '2. 영역 합의 결정', 'blocks': ['s3_decide']},
                                         {'h': '3. 내가 쓴 프롬프트', 'blocks': ['s3_write']}]}]},
              {'id': 'm5',
               'no': 5,
               'title': '움직임 기반 기준 탐지기와 사전학습 객체 탐지 모델의 성능 비교',
               'short': '근거 받고 검증하기',
               'kind': '개인',
               'competencies': ['D5'],
               'outline': 'AI가 ‘사전학습 모델이 항상 나아요’라고 답하면 믿어도 될까요? 답을 아는 작은 예시와 조건별 표본 수를 함께 요청해 직접 확인해요.',
               'brief': {'subtitle': '탐지기 비교표 믿어도 될까 확인하기: 근거를 함께 받기',
                         'time': '약 30분',
                         'goal': 'AI의 답을 그대로 믿지 않고 근거 수치와 작은 예시를 함께 요청해, 두 탐지기를 조건별로 비교해요.',
                         'why': '움직임을 찾는 탐지기와 사물을 아는 모델, 무엇이 다를까요? AI는 틀린 답도 자신 있게 말해요. 답만 받지 말고 다시 셀 수 있는 숫자를 함께 받아야 믿어도 되는지 판단할 수 있어요.',
                         'case': {'title': '해외 사례: 드론과 새를 가리는 대회에서 쓴 방법',
                                  'source': 'Laroca 외, arXiv 2504.19347 (2025.4)',
                                  'points': ['소형 드론을 새와 구별해 찾는 국제 대회(Drone-vs-Bird)에서 상위 3위 안에 든 연구팀의 보고서예요. 작은 드론은 새와 구별하기 어려운 경우가 많다고 해요.',
                                             '연구팀은 YOLO 탐지 모델 하나에만 기대지 않았어요. 영상 전체와 잘라 낸 조각을 함께 보고, 앞뒤 프레임을 살펴 놓친 탐지를 줄였어요.',
                                             '모델 이름만으로 정할 수는 없어요. 어떤 조건에서 얼마나 잡는지 숫자로 나눠 봐야 해요.']},
                         'team': '윤분석(데이터 분석 담당)이 두 탐지기의 결과를 줘요. 임반장은 어떤 탐지기로 갈지 근거를 궁금해해요.',
                         'terms': [{'t': '배경 차분',
                                    'en': 'background subtraction',
                                    'fig': 'f_bg',
                                    'd': '배경과 달라진 픽셀을 찾아 움직임을 잡는 방법이에요. 움직임 기반 기준 탐지기가 이 방식이에요.',
                                    'easy': '빈 방 사진과 비교해 달라진 곳을 찾는 거예요.',
                                    'use': '무엇인지는 몰라도 움직이면 잡아요.'},
                                   {'t': '사전학습 모델',
                                    'en': 'pretrained model',
                                    'fig': 'f_tl',
                                    'd': '큰 데이터로 미리 학습해 둔 모델이에요. 사람, 차량 같은 사물을 알아봐요.',
                                    'easy': '이미 공부를 마친 신입이에요.',
                                    'use': '이번 실습은 가상 결과만 다뤄요.'},
                                   {'t': 'IoU',
                                    'en': 'Intersection over Union',
                                    'fig': 'f_iou',
                                    'd': '두 박스가 겹친 넓이를 합친 넓이로 나눈 값이에요.',
                                    'easy': '1이면 완전히 겹치고, 0이면 전혀 안 겹쳐요.',
                                    'use': 'IoU 0.5 이상이면 맞게 찾았다고 세요.'}],
                         'steps': ['답만 받으면: 근거 없는 답은 위험해요', '근거까지 받기: 근거 수치와 함께 요청하기', '내 말로 쓰기: 검증을 담아 내 말로 요청하기'],
                         'glossary': [],
                         'outputs': ['M5 탐지기 조건별 비교표']},
               'start_messages': [['dm_lim', 'lim', '{call}, 사전학습 모델이 무조건 낫다는 말이 있던데, 우리 조건에서도 그런가요? 결과 파일은 m5_dets.csv예요.']],
               'finish_messages': [['dm_lim', 'lim', '{call}, 조건마다 다르다는 걸 숫자로 봤네요. ‘항상 낫다’는 말은 빼고 보고할게요.']],
               'steps': [{'short': '답만 받으면',
                          'title': '근거 없는 답은 위험해요',
                          'lead': 'AI가 자신 있게 답해도 **다시 셀 수 있는 숫자**가 없으면 믿을지 판단할 수 없어요.',
                          'blocks': [{'id': 's1_info',
                                      'type': 'info',
                                      'title': 'AI 답변에 함께 받을 것',
                                      'icon': '🔍',
                                      'text': ['답을 받을 때 아래를 함께 요청하면, 코드를 다 읽지 못해도 믿을지 판단할 수 있어요.'],
                                      'bullets': ['**근거 수치**: 조건별 표본 수, 맞게 찾은 수, 놓친 수, 잘못 찾은 수',
                                                  '**작은 예시**: 손으로 계산할 수 있는 박스 두 개의 IoU',
                                                  '**나눈 조건**: 낮, 밤, 안개를 나눴는지'],
                                      'figure': 'f_verify'},
                                     {'id': 's1_q',
                                      'type': 'quiz',
                                      'title': 'IoU를 손으로 계산해 볼까요?',
                                      'text': '정답 박스 (x=10, y=10, w=20, h=20)와 예측 박스 (x=20, y=10, w=20, h=20)가 있어요. 겹친 넓이는 10×20=200, 합친 넓이는 '
                                              '400+400−200=600이에요. IoU 0.5 기준으로 맞게 찾은 것일까요?',
                                      'options': ['맞게 찾았어요. 겹쳐 있으니까요.', '아니에요. IoU가 약 0.33이라 기준 아래예요.', 'IoU는 계산할 수 없어요.'],
                                      'answer': 1,
                                      'explain': '200 ÷ 600 ≈ 0.33이에요. 이렇게 답을 아는 예시를 AI에게 주면 코드가 맞는지 바로 확인할 수 있어요.',
                                      'hints': ['겹친 넓이를 합친 넓이로 나눠 보세요.', '0.5보다 큰가요, 작은가요?']}]},
                         {'short': '근거까지 받기',
                          'title': '근거 수치와 함께 요청하기',
                          'lead': '**근거**와 **조건** 문장을 빼고도 보내 보세요. AI의 답이 어떻게 달라지는지 비교해 봐요.',
                          'blocks': [{'id': 's2_vibe',
                                      'type': 'vibe',
                                      'title': '탐지기 조건별 비교',
                                      'ask': ['yun', '두 탐지기를 같은 기준으로 비교하고 싶어요.'],
                                      'file': 'm5_dets.csv',
                                      'slots': [{'id': 'task',
                                                 'label': '할 일',
                                                 'why': '무엇을 비교할지 정해요.',
                                                 'options': [{'text': 'm5_dets.csv의 열은 det_id, model(motion·pretrained), condition(day·night·fog), '
                                                                      'gt_x, gt_y, gt_w, gt_h(정답 박스, 없으면 빈칸), x, y, w, h(예측 박스, 없으면 빈칸), score야. IoU '
                                                                      '0.5 이상이면 맞게 찾은 것(TP)으로 하고, 재현율과 정밀도를 계산해 줘. 정답과 예측이 있어도 IoU가 낮으면 놓친 것과 잘못 찾은 '
                                                                      '것으로 하나씩 세어 줘.',
                                                              'good': True},
                                                             {'text': '어떤 탐지기가 좋은지 봐 줘.', 'good': False},
                                                             {'text': '', 'good': False}],
                                                 'miss': {'say': '모델별 평균 점수를 비교했어요.',
                                                          'code': 'import pandas as pd\n'
                                                                  "df = pd.read_csv('{VDIR}/m5_dets.csv')\n"
                                                                  "display(df.groupby('model')['score'].mean().round(3).reset_index())",
                                                          'problem': '점수는 각 모델이 스스로 매긴 값이에요. 맞게 찾았는지와는 별개라서 비교 근거가 안 돼요.',
                                                          'hint': '무엇을 맞다고 볼지, 어떤 지표를 쓸지 적어 주세요.'}},
                                                {'id': 'evidence',
                                                 'label': '근거',
                                                 'why': '답을 아는 예시와 숫자를 받아요.',
                                                 'options': [{'text': '결론만 쓰지 말고 TP, 놓친 수(FN), 잘못 찾은 수(FP)를 표로 보여 줘. 박스 (10, 10, 20, 20)과 (20, 10, '
                                                                      '20, 20)의 IoU가 약 0.33인지 먼저 확인해 줘.',
                                                              'good': True},
                                                             {'text': '어느 쪽이 나은지 결론만 말해 줘.', 'good': False},
                                                             {'text': '', 'good': False}],
                                                 'miss': {'say': '결론입니다. 사전학습 모델이 항상 더 좋아요. 그걸 쓰시면 돼요!',
                                                          'code': "print('결론: 사전학습 모델이 모든 조건에서 더 좋습니다.')",
                                                          'problem': '자신 있게 답했지만 숫자가 하나도 없어요. **근거를 요청하지 않으면 AI는 그럴듯한 결론만 줘요.** 표를 받아 밤 조건을 확인해 보세요.',
                                                          'hint': 'TP, FN, FP와 답을 아는 예시를 달라고 적어 주세요.'}},
                                                {'id': 'cond',
                                                 'label': '조건',
                                                 'why': '나눠서 비교해요.',
                                                 'options': [{'text': 'model과 condition별로 나눠 표를 만들고, 재현율을 막대그래프로 그려 줘. 표를 넣은 HTML 비교 문서를 report 변수에 '
                                                                      '담아 줘.',
                                                              'good': True},
                                                             {'text': '전체 평균으로 비교해 줘.', 'good': False},
                                                             {'text': '', 'good': False}],
                                                 'miss': {'say': '전체를 합쳐 비교했어요.',
                                                          'code': 'import pandas as pd\n'
                                                                  'def iou(a, b):\n'
                                                                  '    ix = max(0, min(a[0] + a[2], b[0] + b[2]) - max(a[0], b[0]))\n'
                                                                  '    iy = max(0, min(a[1] + a[3], b[1] + b[3]) - max(a[1], b[1]))\n'
                                                                  '    inter = ix * iy\n'
                                                                  '    union = a[2] * a[3] + b[2] * b[3] - inter\n'
                                                                  '    return inter / union if union else 0\n'
                                                                  "df = pd.read_csv('{VDIR}/m5_dets.csv')\n"
                                                                  'rows = []\n'
                                                                  "for m, s in df.groupby('model'):\n"
                                                                  '    tp = fn = 0\n'
                                                                  '    for _, r in s.iterrows():\n'
                                                                  "        if pd.isna(r['gt_x']):\n"
                                                                  '            continue\n'
                                                                  "        ok = pd.notna(r['x']) and iou((r['gt_x'], r['gt_y'], r['gt_w'], "
                                                                  "r['gt_h']), (r['x'], r['y'], r['w'], r['h'])) >= 0.5\n"
                                                                  '        tp += ok\n'
                                                                  '        fn += not ok\n'
                                                                  "    rows.append({'모델': m, '재현율(%)': round(tp / (tp + fn) * 100, 1)})\n"
                                                                  'display(pd.DataFrame(rows))',
                                                          'problem': '합치면 움직임 기반 탐지기가 앞서 보이지만, 낮에는 오히려 사전학습 모델이 앞서요. 조건을 나눠야 어디서 뒤집히는지 보여요.',
                                                          'hint': '무엇으로 나눌지 적어 주세요.'}}],
                                      'done': {'say': 'IoU를 작은 예시로 확인한 뒤, 조건별 표를 만들었어요.',
                                               'code': 'import pandas as pd\n'
                                                       'import matplotlib.pyplot as plt\n'
                                                       '\n'
                                                       "df = pd.read_csv('{VDIR}/m5_dets.csv')\n"
                                                       '\n'
                                                       'def iou(a, b):\n'
                                                       '    ix = max(0, min(a[0] + a[2], b[0] + b[2]) - max(a[0], b[0]))\n'
                                                       '    iy = max(0, min(a[1] + a[3], b[1] + b[3]) - max(a[1], b[1]))\n'
                                                       '    inter = ix * iy\n'
                                                       '    union = a[2] * a[3] + b[2] * b[3] - inter\n'
                                                       '    return inter / union if union else 0\n'
                                                       '\n'
                                                       '# 작은 확인\n'
                                                       'assert round(iou((10, 10, 20, 20), (20, 10, 20, 20)), 2) == 0.33\n'
                                                       "print('확인: IoU 0.33')\n"
                                                       '\n'
                                                       'rows = []\n'
                                                       "for (m, c), s in df.groupby(['model', 'condition']):\n"
                                                       '    tp = fn = fp = 0\n'
                                                       '    for _, r in s.iterrows():\n'
                                                       "        has_gt, has_pred = pd.notna(r['gt_x']), pd.notna(r['x'])\n"
                                                       '        if has_gt and has_pred:\n'
                                                       "            if iou((r['gt_x'], r['gt_y'], r['gt_w'], r['gt_h']), (r['x'], r['y'], r['w'], "
                                                       "r['h'])) >= 0.5:\n"
                                                       '                tp += 1\n'
                                                       '            else:\n'
                                                       '                fn += 1\n'
                                                       '                fp += 1\n'
                                                       '        elif has_gt:\n'
                                                       '            fn += 1\n'
                                                       '        else:\n'
                                                       '            fp += 1\n'
                                                       "    rows.append({'탐지기': m, '조건': c, 'TP': tp, 'FN': fn, 'FP': fp, '재현율(%)': round(tp / (tp + "
                                                       "fn) * 100, 1), '정밀도(%)': round(tp / (tp + fp) * 100, 1)})\n"
                                                       'table = pd.DataFrame(rows)\n'
                                                       'display(table)\n'
                                                       '\n'
                                                       "pivot = table.pivot(index='조건', columns='탐지기', values='재현율(%)').loc[['day', 'night', "
                                                       "'fog']]\n"
                                                       "ax = pivot.plot(kind='bar', figsize=(5.5, 3), rot=0)\n"
                                                       "ax.set_xlabel('condition')\n"
                                                       "ax.set_ylabel('recall (%)')\n"
                                                       '\n'
                                                       "report = ('<h2>탐지기 조건별 비교</h2><p>IoU 0.5 기준, 가상 결과 기준이에요. 조건마다 표본 수가 달라요.</p>' + "
                                                       'table.to_html(index=False))',
                                               'lesson': '근거를 함께 받으니 ‘항상 낫다’는 답이 틀렸다는 것이 보였어요. 낮에는 사전학습 모델이 재현율과 정밀도 모두 앞서지만, 밤의 재현율은 움직임 기반 탐지기가 훨씬 '
                                                         '높아요. 안개는 표본이 적으니 그 점도 함께 적어요.'},
                                      'on_ok': [['dm_yun', 'yun', '{call}, 표로 보니 밤에는 뒤집히네요. 한 가지만 고르지 말고 조합도 생각해 봐야겠어요.']]}]},
                         {'short': '내 말로 쓰기',
                          'title': '검증을 담아 내 말로 요청하기',
                          'lead': '다음에 비슷한 비교를 맡게 되면 **처음부터 근거를 함께** 요청해 보세요.',
                          'blocks': [{'id': 's3_write',
                                      'type': 'write',
                                      'title': '근거를 요청하는 프롬프트 쓰기',
                                      'prompt': '**새 요청**: 새 탐지기 결과 m5_dets_b.csv를 같은 방식으로 비교하고 싶어요. 열은 같아요. 결론만 받지 않도록 어떻게 요청할까요?',
                                      'example': 'm5_dets_b.csv의 탐지 결과를 비교하고 싶어. 열은 model, condition, gt_x, gt_y, gt_w, gt_h, x, y, w, h, score야. 박스 '
                                                 '(10, 10, 20, 20)과 (20, 10, 20, 20)의 IoU가 약 0.33인지 먼저 확인하고, IoU 0.5 기준으로 model과 condition별 TP, FN, '
                                                 'FP, 재현율, 정밀도를 표로 보여 줘. 결론만 쓰지 말고 표본 수도 함께 적어 줘.',
                                      'min': 30,
                                      'checklist': ['답을 아는 예시로 확인하게 했어요', '조건을 나눠 달라고 했어요', '근거 수치를 요청했어요', '결론만 받지 않게 했어요'],
                                      'model_label': '📝 모범 프롬프트',
                                      'placeholder': '꼭 입력해 주세요 · 내 말로 프롬프트를 써 보세요'}]}],
               'outputs': [{'id': 'm5_detectors',
                            'title': 'M5 탐지기 조건별 비교표',
                            'sections': [{'h': '1. 비교표와 근거', 'blocks': ['s2_vibe']}, {'h': '2. 내가 쓴 프롬프트', 'blocks': ['s3_write']}]}]},
              {'id': 'm6',
               'no': 6,
               'title': '불필요 경보 원인 분석과 경보 운용점 합의',
               'short': '숫자와 예외',
               'kind': '협업',
               'competencies': ['D1', 'C1'],
               'outline': '‘경보를 적당히 줄여 줘’ 같은 모호한 말을 숫자 규칙과 예외 처리로 바꾸고, 임계값별 표와 불필요 경보 원인표를 받아 운용점을 합의해요.',
               'brief': {'subtitle': '임계값별 경보 표와 운용점: 숫자와 예외로 말하기',
                         'time': '약 30분',
                         'goal': '모호한 요청을 숫자 규칙과 예외 처리로 바꿔 임계값별 표를 받고, 그 표를 근거로 경보 운용점을 합의해요.',
                         'why': '경보를 줄이려면 무엇을 포기해야 할까요? ‘적당히’, ‘많이’ 같은 말은 사람마다 달라요. AI도 마음대로 정해 버려요. 점수가 비어 있는 기록처럼 예외가 있는 데이터는 어떻게 할지까지 알려 줘야 해요.',
                         'case': {'title': '해외 사례: 확인되지 않은 목격 신고로 멈춘 공항',
                                  'source': 'Pravda.com.ua, Al Jazeera 등 뮌헨 공항 보도 (2025.10)',
                                  'points': ['2025년 10월 2일 밤 독일 뮌헨 공항은 드론 목격 신고로 운항을 멈췄어요.',
                                             '공항 집계로 출발 17편이 취소돼 승객 3,000명가량이 영향을 받았고, 도착 15편은 다른 공항으로 돌아갔어요. 독일 항공교통관제 기관은 ‘확인되지 않은 목격’에 따른 예방 '
                                             '조치였다고 밝혔어요.',
                                             '경보 한 건에도 비용이 따라요. 놓치는 위험과 함께 불필요 경보의 비용을 숫자로 놓고 기준을 정해야 해요.']},
                         'team': '서상황은 불필요 경보를 줄이자고 하고, 강계획은 대상을 놓치면 안 된다고 해요. 조당직(당직 책임자)이 합의한 기준을 받아요.',
                         'terms': [{'t': '경보 운용점',
                                    'en': 'operating point',
                                    'fig': 'f_op',
                                    'd': '재현율과 불필요 경보 사이에서 실제로 쓰기로 정한 임계값이에요.',
                                    'easy': '민감도 다이얼을 어디에 둘지 정하는 거예요.',
                                    'use': '임계값별 표를 보고 사람이 합의해요.'},
                                   {'t': '임계값',
                                    'en': 'threshold',
                                    'fig': 'f_cut',
                                    'd': '점수가 이 값 이상이면 경보로 올리는 기준이에요.',
                                    'easy': '커트라인이에요.',
                                    'use': '‘이상’인지 ‘초과’인지 프롬프트에 적어요.'},
                                   {'t': '예외 처리',
                                    'en': 'edge case',
                                    'fig': 'f_edge',
                                    'd': '빈칸, 0, 표본이 적은 경우처럼 평소와 다른 값을 어떻게 할지 정하는 것이에요.',
                                    'easy': '‘이럴 땐 이렇게’를 미리 말해 두는 거예요.',
                                    'use': '말하지 않으면 AI가 조용히 아무렇게나 처리해요.'}],
                         'steps': ['숫자로 바꾸기: 모호한 말을 숫자 규칙으로', '규칙으로 요청: 숫자 규칙과 예외를 담아 요청하기', '운용점 합의: 운용점을 합의하고 내 말로 요청하기'],
                         'glossary': [],
                         'outputs': ['M6 임계값별 표와 운용점 결정 기록']},
               'start_messages': [['dm_seo', 'seo', '{call}, 경보를 적당히 줄이고 싶어요. 그런데 ‘적당히’를 어떻게 정할지가 문제예요. 파일은 m6_alarms.csv예요.'],
                                  ['dm_kang', 'kang', '{call}, 줄이다가 대상을 놓치면 안 돼요. 숫자로 같이 봐요.']],
               'finish_messages': [['dm_jo', 'jo', '{call}, 표와 합의 기록 잘 받았어요. 영상 손실 건을 따로 본다는 점도 근무 안내에 넣을게요.']],
               'steps': [{'short': '숫자로 바꾸기',
                          'title': '모호한 말을 숫자 규칙으로',
                          'lead': '‘적당히’, ‘많이’는 사람마다 달라요. **숫자로 바꾸면** AI도 사람도 같은 뜻으로 읽어요.',
                          'blocks': [{'id': 's1_match',
                                      'type': 'match',
                                      'title': '모호한 말을 숫자 규칙과 짝짓기',
                                      'text': '왼쪽 말을 프롬프트에 쓸 수 있게 바꾼 문장을 골라 주세요.',
                                      'left_label': '모호한 말',
                                      'right_label': '프롬프트에 쓸 문장',
                                      'pairs': [{'left': '대상을 놓치지 않기', 'right': '재현율 90% 이상', 'hint': '대상을 얼마나 잡는지에 대한 말이에요.'},
                                                {'left': '경보를 적당히', 'right': '하루 불필요 경보 10건 이하', 'hint': '하루에 몇 건까지인지 정해요.'},
                                                {'left': '점수가 높은 것', 'right': 'score가 임계값 이상', 'hint': '커트라인과 비교해요.'},
                                                {'left': '점수를 못 낸 것', 'right': 'score가 빈칸', 'hint': '영상 손실이면 점수가 없어요.'}],
                                      'hints': ['숫자가 적힌 쪽이 어느 말을 가리키는지 찾아보세요.', 'score는 모델 점수, 빈칸은 점수가 없는 기록이에요.'],
                                      'explain': '숫자로 바꾸면 결과가 왜 그렇게 나왔는지도 설명할 수 있어요.'}]},
                         {'short': '규칙으로 요청',
                          'title': '숫자 규칙과 예외를 담아 요청하기',
                          'lead': '**예외** 문장을 빼고도 보내 보세요. 점수가 빈 기록이 어떻게 되는지 확인해 봐요.',
                          'blocks': [{'id': 's2_vibe',
                                      'type': 'vibe',
                                      'title': '임계값별 경보 표',
                                      'ask': ['seo', '임계값을 바꾸면 재현율과 하루 불필요 경보가 어떻게 달라지는지 보고 싶어요.'],
                                      'file': 'm6_alarms.csv',
                                      'slots': [{'id': 'rule',
                                                 'label': '숫자 규칙',
                                                 'why': '기준을 숫자로 적어요.',
                                                 'options': [{'text': 'm6_alarms.csv의 열은 alarm_id, day(1~7), score(빈칸 가능), is_target(대상이면 1), '
                                                                      'cause(animal·tree·rain·light·none)야. 임계값 0.3, 0.4, 0.5, 0.6, 0.7, 0.8마다 '
                                                                      'score가 임계값 이상이면 경보로 보고, 재현율과 하루 평균 불필요 경보 수를 계산해 줘. 재현율 90% 이상이고 하루 10건 이하면 '
                                                                      '‘기준 충족’으로 표시해 줘.',
                                                              'good': True},
                                                             {'text': '경보를 적당히 줄여 줘.', 'good': False},
                                                             {'text': '', 'good': False}],
                                                 'miss': {'say': '임계값을 0.8로 올렸어요. 경보가 많이 줄었어요!',
                                                          'code': 'import pandas as pd\n'
                                                                  "df = pd.read_csv('{VDIR}/m6_alarms.csv')\n"
                                                                  "a = df[df['score'] >= 0.8]\n"
                                                                  "print('경보 수:', len(df), '→', len(a))",
                                                          'problem': '경보는 줄었지만 대상을 얼마나 놓쳤는지 나오지 않았어요. ‘적당히’가 숫자가 아니어서 AI가 마음대로 정했어요.',
                                                          'hint': '비교할 임계값과 지표, 기준을 숫자로 적어 주세요.'}},
                                                {'id': 'edge',
                                                 'label': '예외',
                                                 'why': '빈칸을 어떻게 할지 정해요.',
                                                 'options': [{'text': 'score가 빈칸인 기록은 영상 손실이니 경보 계산에서 빼되, 지우지 말고 ‘영상 손실: 사람 확인’으로 따로 세어 보여 줘.',
                                                              'good': True},
                                                             {'text': '나머지는 알아서 해 줘.', 'good': False},
                                                             {'text': '', 'good': False}],
                                                 'miss': {'say': '기준대로 계산했어요. 예외는 따로 처리하지 않았어요.',
                                                          'code': 'import pandas as pd\n'
                                                                  "df = pd.read_csv('{VDIR}/m6_alarms.csv')\n"
                                                                  'rows = []\n'
                                                                  'for t in [0.3, 0.4, 0.5, 0.6, 0.7, 0.8]:\n'
                                                                  "    a = df[df['score'] >= t]\n"
                                                                  "    rows.append({'임계값': t, '재현율(%)': round(a['is_target'].sum() / "
                                                                  "df['is_target'].sum() * 100, 1), '하루 불필요 경보': round((a['is_target'] == 0).sum() / "
                                                                  '7, 1)})\n'
                                                                  'display(pd.DataFrame(rows))',
                                                          'problem': '오류는 없지만 점수가 빈 기록이 조용히 사라졌어요. 영상이 끊겨 점수를 못 낸 시간은 **사람이 꼭 봐야 하는데** 어디에도 남지 않아요.',
                                                          'hint': '빈칸을 어떻게 할지 적어 주세요.'}},
                                                {'id': 'out',
                                                 'label': '결과와 원인',
                                                 'why': '표와 그림, 원인까지 받아요.',
                                                 'options': [{'text': '임계값별 표와 곡선 그림을 보여 주고, 임계값 0.5에서 불필요 경보를 cause별로 센 표도 만들어 줘. 표를 넣은 HTML 문서를 '
                                                                      'report 변수에 담아 줘.',
                                                              'good': True},
                                                             {'text': '제일 좋은 값 하나만 알려 줘.', 'good': False},
                                                             {'text': '', 'good': False}],
                                                 'miss': {'say': '제일 좋은 값은 0.5예요.',
                                                          'code': "print('추천 임계값: 0.5')",
                                                          'problem': '값 하나만 있고 이유가 없어요. 무엇을 얼마나 포기하는지 보여 주지 않으면 합의할 수 없어요.',
                                                          'hint': '표와 그림, 원인별 건수를 달라고 적어 주세요.'}}],
                                      'done': {'say': '규칙과 예외대로 임계값별 표와 그림, 원인표를 만들었어요.',
                                               'code': 'import pandas as pd\n'
                                                       'import matplotlib.pyplot as plt\n'
                                                       '\n'
                                                       "df = pd.read_csv('{VDIR}/m6_alarms.csv')\n"
                                                       'DAYS = 7\n'
                                                       "scored = df[df['score'].notna()]\n"
                                                       "lost = df[df['score'].isna()]\n"
                                                       '\n'
                                                       'rows = []\n'
                                                       'for t in [0.3, 0.4, 0.5, 0.6, 0.7, 0.8]:\n'
                                                       "    a = scored[scored['score'] >= t]\n"
                                                       "    recall = round(a['is_target'].sum() / scored['is_target'].sum() * 100, 1)\n"
                                                       "    unneeded = round((a['is_target'] == 0).sum() / DAYS, 1)\n"
                                                       "    rows.append({'임계값': t, '재현율(%)': recall, '하루 불필요 경보': unneeded, '판정': '기준 충족' if recall "
                                                       ">= 90 and unneeded <= 10 else ''})\n"
                                                       'table = pd.DataFrame(rows)\n'
                                                       'display(table)\n'
                                                       "print('영상 손실: 사람 확인', len(lost), '건')\n"
                                                       '\n'
                                                       "c = scored[(scored['score'] >= 0.5) & (scored['is_target'] == "
                                                       "0)]['cause'].value_counts().reset_index()\n"
                                                       "c.columns = ['원인', '건수']\n"
                                                       'display(c)\n'
                                                       '\n'
                                                       'fig, ax = plt.subplots(figsize=(4.8, 3))\n'
                                                       "ax.plot(table['하루 불필요 경보'], table['재현율(%)'], 'o-')\n"
                                                       'for _, r in table.iterrows():\n'
                                                       "    ax.annotate(r['임계값'], (r['하루 불필요 경보'], r['재현율(%)']), fontsize=8)\n"
                                                       "ax.set_xlabel('nuisance alarms per day')\n"
                                                       "ax.set_ylabel('recall (%)')\n"
                                                       '\n'
                                                       "report = ('<h2>임계값별 경보 표</h2><p>가상 기록 7일 기준이에요. ' + f'점수가 없는 영상 손실 {len(lost)}건은 계산에서 빼고 사람이 "
                                                       "따로 확인해요.</p>'\n"
                                                       "          + table.to_html(index=False) + '<h3>임계값 0.5의 불필요 경보 원인</h3>' + "
                                                       'c.to_html(index=False))',
                                               'lesson': '숫자와 예외를 적으니 무엇을 얼마나 포기하는지가 드러났어요. 영상 손실 기록도 사라지지 않고 사람이 볼 목록으로 남았어요. 가장 많은 원인을 보면 임계값 말고 '
                                                         '무엇을 고칠지도 보여요.'},
                                      'on_ok': [['dm_seo', 'seo', '{call}, 무엇을 포기하는지 보이네요. 이제 운용점을 정해 봐요.']]}]},
                         {'short': '운용점 합의',
                          'title': '운용점을 합의하고 내 말로 요청하기',
                          'lead': '표는 AI가 만들어도 **어디에 둘지는 사람이 정해요.**',
                          'blocks': [{'id': 's3_decide',
                                      'type': 'decide',
                                      'title': '운용점 의견 판단하기',
                                      'text': '표를 본 팀원들의 의견이에요. 받아들일지 정하고 이유를 적어 주세요.',
                                      'reason_min': 8,
                                      'items': [{'id': 'a',
                                                 'who': 'seo',
                                                 'text': '불필요 경보가 가장 적은 0.8로 가요. 관제가 훨씬 편해져요.',
                                                 'answer': 'reject',
                                                 'hint': '표에서 0.8의 재현율을 보세요.',
                                                 'explain': '재현율이 합의한 기준에 한참 못 미쳐요. 편해지는 대신 대상을 놓치는 일이에요.',
                                                 'reason': '재현율이 합의한 기준에 못 미쳐요. 편의 때문에 대상을 놓치면 안 돼요.'},
                                                {'id': 'b',
                                                 'who': 'kang',
                                                 'text': '기준을 지키는 임계값을 쓰되, 가장 많은 원인인 동물 경보를 줄일 방법을 먼저 찾아 다음 달에 다시 비교해요.',
                                                 'answer': 'accept',
                                                 'hint': '임계값 말고 고칠 수 있는 것이 있을까요?',
                                                 'explain': '원인을 줄이면 재현율을 지키면서 경보를 줄일 수 있어요. 같은 표로 다시 비교하면 효과도 보여요.',
                                                 'reason': '원인을 줄이면 재현율을 지키면서 경보를 줄일 수 있고 같은 표로 효과를 볼 수 있어요.'}]},
                                     {'id': 's3_write',
                                      'type': 'write',
                                      'title': '숫자와 예외를 담은 프롬프트 쓰기',
                                      'prompt': '**새 요청**: ‘야간에는 경보를 좀 더 민감하게 해 줘’라는 요청을 받았어요. 파일은 m6_alarms_night.csv이고, 열에 hour가 더 있어요. 숫자 규칙과 예외를 '
                                                '정해 프롬프트를 써 보세요.',
                                      'example': 'm6_alarms_night.csv의 열은 alarm_id, hour, score, is_target, cause야. hour가 18~23 또는 0~5이면 야간, 나머지는 '
                                                 '주간이야. 야간 임계값은 0.4, 주간 임계값은 0.5로 하고, score가 임계값 이상이면 경보로 봐 줘. score가 빈칸이면 ‘영상 손실: 사람 확인’으로 따로 세어 줘. '
                                                 '주간과 야간의 재현율, 하루 불필요 경보 수를 표로 보여 줘.',
                                      'min': 30,
                                      'checklist': ['모호한 말을 숫자로 바꿨어요', '경계값을 분명히 적었어요', '빈칸을 어떻게 할지 적었어요', '받고 싶은 모양을 적었어요'],
                                      'model_label': '📝 모범 프롬프트',
                                      'placeholder': '꼭 입력해 주세요 · 내 말로 프롬프트를 써 보세요'}]}],
               'outputs': [{'id': 'm6_op',
                            'title': 'M6 임계값별 표와 운용점 결정 기록',
                            'sections': [{'h': '1. 숫자 규칙 짝짓기', 'blocks': ['s1_match']},
                                         {'h': '2. 임계값별 표와 원인표', 'blocks': ['s2_vibe']},
                                         {'h': '3. 운용점 의견 판단', 'blocks': ['s3_decide']},
                                         {'h': '4. 내가 쓴 프롬프트', 'blocks': ['s3_write']}]}]},
              {'id': 'm7',
               'no': 7,
               'title': '궤적 특징으로 드론·조류 분류 모델 비교',
               'short': '문서 요청과 환각',
               'kind': '개인',
               'competencies': ['J1', 'D5'],
               'outline': '모델 비교 결과를 근거로 선택 기록을 요청해요. 독자·근거·표현을 지정해, 데이터에 없는 말이나 부풀린 숫자가 문서에 들어가지 않게 하는 법을 익혀요.',
               'brief': {'subtitle': '모델 선택 기록: 환각 없는 문서 요청하기',
                         'time': '약 30분',
                         'goal': '독자·양식·근거를 지정해 모델 선택 기록을 요청하고, 자료에 없는 내용과 부풀린 표현을 막아요.',
                         'why': '몇 픽셀짜리 비행체, 드론일까요 새일까요? 모델 비교 문서를 AI에게 맡기면 매끄럽지만 틀린 문장이 나오기 쉬워요. 어떤 수치를 근거로 쓸지 지정하지 않으면 가장 좋아 보이는 숫자만 골라 쓰기도 해요.',
                         'case': {'title': '해외 사례: 화면에 뜬 항적을 ‘새나 쓰레기일 수 있다’고 넘겼어요',
                                  'source': 'Business Insider, Stars and Stripes 등 미군 조사 보고서 보도 (2024~2025)',
                                  'points': ['2024년 1월 요르단의 미군 거점 ‘타워 22’가 드론 공격을 받아 미군 3명이 숨졌어요. 초기에는 돌아오는 아군 드론과 헷갈렸다는 설명이 나왔지만, 이후 조사에서 다른 사정이 '
                                             '드러났어요.',
                                             '조사 보고서에 따르면 근무자들은 화면의 미확인 항적을 보고도 ‘너무 멀다’, ‘너무 느리다’, ‘새나 쓰레기일 수 있다’고 여겼어요. 착륙 중인 아군 무인기에 주의가 쏠려 있었다고 '
                                             '해요.',
                                             '드론과 새를 가르는 일은 모델만의 문제가 아니에요. 모델 문서에는 어떤 조건에서 얼마나 틀리는지를 부풀리지 않고 적어야 사람이 제대로 판단해요.']},
                         'team': '윤분석이 비교 결과를 줘요. 임반장이 모델 선택 기록을 읽어요.',
                         'terms': [{'t': '궤적 특징',
                                    'en': 'track feature',
                                    'fig': 'f_track',
                                    'd': '물체의 이동 기록에서 뽑은 숫자예요. 속도, 정지 비율, 크기 등이 있어요.',
                                    'easy': '모양이 안 보여도 움직이는 버릇은 보여요.',
                                    'use': '드론은 멈춰 있을 수 있고 새는 잘 그러지 못해요.'},
                                   {'t': '데이터 누수',
                                    'en': 'data leakage',
                                    'fig': 'f_leak',
                                    'd': '테스트에 쓸 정보가 학습에 섞여 성능이 실제보다 좋게 나오는 일이에요.',
                                    'easy': '시험 문제를 미리 본 것과 같아요.',
                                    'use': '한 클립의 궤적은 학습과 테스트로 나누지 말고 클립 단위로 나눠요.'},
                                   {'t': '환각',
                                    'en': 'hallucination',
                                    'fig': 'f_halluc',
                                    'd': 'AI가 자료에 없는 내용을 사실처럼 지어내는 현상이에요.',
                                    'easy': '모르는 것도 그럴듯하게 말해 버리는 거예요.',
                                    'use': '‘자료에 없으면 자료 없음으로 써 줘’라고 적어요.'}],
                         'steps': ['먼저 나누기: 어떤 수치를 근거로 쓸까요?', '문서 요청: 독자·근거·표현을 정해 문서 요청하기', '내 말로 쓰기: 문서를 받은 뒤 확인하고 내 말로 요청하기'],
                         'glossary': [],
                         'outputs': ['M7 드론·조류 분류 모델 선택 기록']},
               'start_messages': [['dm_yun', 'yun', '{call}, 모델 네 개를 비교한 결과예요. 파일은 m7_results.csv예요. 임반장님께 드릴 모델 선택 기록이 필요해요. AI에게 맡겨 볼래요?']],
               'finish_messages': [['dm_lim', 'lim', '{call}, 클립 단위 수치로 쓰셨네요. 왜 그 수치를 쓰는지까지 적혀 있어 좋아요.']],
               'steps': [{'short': '먼저 나누기',
                          'title': '어떤 수치를 근거로 쓸까요?',
                          'lead': '문서를 요청하기 전에 **믿을 수 있는 수치**부터 정해요.',
                          'blocks': [{'id': 's1_info',
                                      'type': 'info',
                                      'title': '두 가지 나누는 방법',
                                      'icon': '✂️',
                                      'text': ['같은 모델도 데이터를 어떻게 나누느냐에 따라 수치가 크게 달라져요.'],
                                      'table': {'head': ['나누는 방법', '하는 일', '결과'],
                                                'rows': [['무작위 분할', '궤적을 한 줄씩 섞어 나눠요', '같은 클립의 궤적이 학습과 테스트에 함께 들어가 좋게 나와요'],
                                                         ['클립 단위 분할', '한 클립은 통째로 한쪽에만 둬요', '처음 보는 영상에서의 성능에 가까워요']]},
                                      'figure': 'f_split'},
                                     {'id': 's1_q',
                                      'type': 'quiz',
                                      'title': '어느 수치를 보고할까요?',
                                      'text': '모델 선택 기록에 쓸 수치로 가장 알맞은 것은 무엇일까요?',
                                      'options': ['무작위 분할 수치가 더 높으니 그것을 쓴다.', '클립 단위 분할 수치를 쓰고, 왜 그 수치인지 함께 적는다.', '두 수치를 평균 내어 쓴다.'],
                                      'answer': 1,
                                      'explain': '실제로는 처음 보는 영상을 다뤄요. 그에 가까운 수치를 써야 해요.',
                                      'hints': ['실제 운용에서 모델이 보는 영상은 어떤 영상일까요?', '시험 문제를 미리 본 점수를 믿을 수 있을까요?']}]},
                         {'short': '문서 요청',
                          'title': '독자·근거·표현을 정해 문서 요청하기',
                          'lead': '**자료에 없으면** 문장을 빼고도 보내 보세요. AI가 무엇을 지어내는지 확인해 봐요.',
                          'blocks': [{'id': 's2_vibe',
                                      'type': 'vibe',
                                      'title': '모델 선택 기록',
                                      'ask': ['yun', '모델 비교 결과로 임반장님이 읽을 선택 기록을 받고 싶어요.'],
                                      'file': 'm7_results.csv',
                                      'slots': [{'id': 'reader',
                                                 'label': '독자와 양식',
                                                 'why': '누가 읽는지, 어떤 순서인지 정해요.',
                                                 'options': [{'text': 'm7_results.csv의 열은 model, split_method(random·clip), drone_recall_pct, '
                                                                      'bird_false_pct, tracks야. 임반장이 읽을 모델 선택 기록을 HTML로 report 변수에 담아 줘. 순서는 결론 한 줄, '
                                                                      '비교표, 고른 이유, 한계야. 그림으로 분할 방법별 재현율도 보여 줘.',
                                                              'good': True},
                                                             {'text': '모델에 대해 써 줘.', 'good': False},
                                                             {'text': '', 'good': False}],
                                                 'miss': {'say': '모델의 역사를 정리했어요.',
                                                          'code': "report = '<h2>딥러닝 모델의 발전</h2><p>최근 인공지능 기술은 빠르게 발전하고 있으며, 다양한 모델이 등장했습니다.</p>'\n"
                                                                  "print('문서를 만들었어요.')",
                                                          'problem': '누구나 쓸 수 있는 일반론이에요. 어떤 모델을 왜 고르는지 없어 임반장이 판단할 수 없어요.',
                                                          'hint': '독자와 문서 순서를 적어 주세요.'}},
                                                {'id': 'source',
                                                 'label': '근거',
                                                 'why': '자료 밖의 말과 고를 수치를 정해요.',
                                                 'options': [{'text': '근거는 이 파일뿐이야. 판단에는 split_method가 clip인 수치만 쓰고, random 수치는 누수 때문에 높게 나온 값이라고 적어 '
                                                                      '줘. 파일에 없는 처리 속도나 다른 연구 수치는 지어내지 말고 ‘자료 없음’으로 적어 줘.',
                                                              'good': True},
                                                             {'text': '참고할 만한 내용도 알아서 넣어 줘.', 'good': False},
                                                             {'text': '', 'good': False}],
                                                 'miss': {'say': '최고 성능 모델로 정리했어요.',
                                                          'code': "report = ('<h2>모델 선택 기록</h2><p>GRU 모델이 재현율 98.1%로 가장 우수합니다. 처리 속도도 초당 120프레임으로 "
                                                                  "실시간 운용에 충분하며, 국제 연구 결과와도 일치합니다.</p>')\n"
                                                                  "print('문서를 만들었어요.')",
                                                          'problem': '98.1%는 누수가 있는 무작위 분할 수치이고, 처리 속도와 국제 연구는 파일에 없어요. **AI가 그럴듯하게 지어낸 환각이에요.** 문장이 '
                                                                     '매끄러워서 더 위험해요.',
                                                          'hint': '어떤 수치를 쓸지 정하고, 없으면 ‘자료 없음’으로 쓰게 해 주세요.'}},
                                                {'id': 'tone',
                                                 'label': '표현',
                                                 'why': '확신을 부풀리지 않아요.',
                                                 'options': [{'text': '맨 위에 ‘AI 초안, 담당자 확인 전’이라고 표시해 줘. ‘정확히 식별’, ‘완벽’ 같은 말은 쓰지 말고, 놓치는 비율과 새를 드론으로 '
                                                                      '볼 비율을 함께 적어 줘. 모델 이름은 랜덤 포레스트처럼 띄어 써 줘.',
                                                              'good': True},
                                                             {'text': '자신 있게 확정된 문서처럼 써 줘.', 'good': False},
                                                             {'text': '', 'good': False}],
                                                 'miss': {'say': '확정 문서처럼 썼어요.',
                                                          'code': 'import pandas as pd\n'
                                                                  "d = pd.read_csv('{VDIR}/m7_results.csv')\n"
                                                                  "report = '<h2>모델 확정</h2><p>이 모델은 드론과 새를 정확히 식별하므로 사람이 확인하지 않아도 됩니다.</p>' + "
                                                                  "d[d['split_method'] == 'clip'].to_html(index=False)",
                                                          'problem': '수치는 맞게 넣었지만 ‘정확히 식별’, ‘사람이 확인하지 않아도 됩니다’라는 말이 표와 맞지 않아요. 이 문장이 그대로 올라가면 사람이 '
                                                                     '확인을 건너뛰게 돼요.',
                                                          'hint': '초안 표시와 쓰면 안 되는 낱말을 적어 주세요.'}}],
                                      'done': {'say': '클립 단위 수치만 근거로 선택 기록 초안을 만들었어요. 담당자 확인이 필요해요.',
                                               'code': 'import pandas as pd\n'
                                                       'import matplotlib.pyplot as plt\n'
                                                       '\n'
                                                       "df = pd.read_csv('{VDIR}/m7_results.csv')\n"
                                                       "NAMES = {'logistic': '로지스틱 회귀', 'random_forest': '랜덤 포레스트', 'gradient_boosting': '그래디언트 "
                                                       "부스팅', 'gru': 'GRU'}\n"
                                                       '\n'
                                                       "pivot = df.pivot(index='model', columns='split_method', "
                                                       "values='drone_recall_pct').loc[list(NAMES)]\n"
                                                       "ax = pivot[['random', 'clip']].plot(kind='bar', figsize=(5.8, 3), rot=15)\n"
                                                       "ax.set_xlabel('')\n"
                                                       "ax.set_ylabel('drone recall (%)')\n"
                                                       '\n'
                                                       "clip = df[df['split_method'] == 'clip'].copy()\n"
                                                       "clip['모델'] = clip['model'].map(NAMES)\n"
                                                       "gap = df.pivot(index='model', columns='split_method', values='drone_recall_pct')\n"
                                                       "clip['무작위와 차이'] = clip['model'].map((gap['random'] - gap['clip']).round(1))\n"
                                                       "table = clip[['모델', 'drone_recall_pct', 'bird_false_pct', '무작위와 차이', 'tracks']]\n"
                                                       "table.columns = ['모델', '드론 재현율(%)', '새를 드론으로 본 비율(%)', '무작위와 차이', '궤적 수']\n"
                                                       'display(table)\n'
                                                       '\n'
                                                       "best = table.sort_values('드론 재현율(%)', ascending=False).iloc[0]\n"
                                                       "report = ('<p><b>AI 초안, 담당자 확인 전.</b></p><h2>모델 선택 기록</h2>'\n"
                                                       '          + f"<p><b>결론:</b> 클립 단위 분할 기준으로 {best[\'모델\']}을 후보로 제안해요. 드론 재현율 {best[\'드론 '
                                                       '재현율(%)\']}%, 새를 드론으로 본 비율 {best[\'새를 드론으로 본 비율(%)\']}%예요.</p>"\n'
                                                       "          + '<h3>비교표 (클립 단위 분할)</h3>' + table.to_html(index=False)\n"
                                                       "          + '<h3>고른 이유</h3><p>처음 보는 영상에 가까운 클립 단위 수치로 비교했어요. 무작위 분할 수치는 같은 클립이 섞여 높게 나온 값이라 "
                                                       "쓰지 않았어요.</p>'\n"
                                                       "          + '<h3>한계</h3><p>재현율 수치만큼 드론을 놓칠 수 있어 사람이 영상을 확인해요. 처리 속도: 자료 없음. 가상 결과 "
                                                       "기준이에요.</p>')",
                                               'lesson': '근거를 클립 단위로 한정하니 순위가 바뀌었어요. 무작위 분할에서 1등이던 GRU가 클립 단위에서는 그래디언트 부스팅에 뒤처져요. 처리 속도는 ‘자료 없음’으로 '
                                                         '남았어요. 지어내지 않은 거예요.'},
                                      'on_ok': [['dm_yun', 'yun', '{call}, 나누는 방법에 따라 순위가 바뀌네요. 이 내용이 문서에 들어가 다행이에요.']]}]},
                         {'short': '내 말로 쓰기',
                          'title': '문서를 받은 뒤 확인하고 내 말로 요청하기',
                          'lead': '문서를 받았다고 끝이 아니에요. **자료와 대조**하는 일은 사람의 몫이에요.',
                          'blocks': [{'id': 's3_q',
                                      'type': 'quiz',
                                      'title': '문서를 받은 뒤에는',
                                      'text': 'AI가 쓴 선택 기록이 매끄럽게 나왔어요. 가장 먼저 무엇을 할까요?',
                                      'options': ['문장이 매끄러우니 바로 올린다.', '표의 수치와 파일의 값을 대조하고, 파일에 없는 말이 있는지 본다.', '더 길게 써 달라고 한다.'],
                                      'answer': 1,
                                      'explain': '매끄러운 문장과 맞는 내용은 별개예요. 숫자와 이름부터 대조하면 환각을 빨리 잡을 수 있어요.',
                                      'hints': ['환각은 어디에 잘 숨어 있을까요?', '다시 셀 수 있는 것부터 확인해 보세요.']},
                                     {'id': 's3_write',
                                      'type': 'write',
                                      'title': '문서를 요청하는 프롬프트 쓰기',
                                      'prompt': '**새 요청**: 관제 담당이 읽을 ‘드론·조류 분류 결과 읽는 법’ 안내문이 필요해요. 근거는 m7_results.csv와 m7_tracks.csv뿐이에요. 독자, 순서, '
                                                '근거, 없는 내용 처리, 표현을 담아 프롬프트를 써 보세요.',
                                      'example': '관제 담당이 읽을 ‘드론·조류 분류 결과 읽는 법’ 안내문을 HTML 한 쪽으로 만들어 줘. 순서는 안내문의 목적, 결과 읽는 법, 놓칠 수 있는 경우, 사람이 확인할 일이야. '
                                                 '근거는 m7_results.csv의 clip 수치와 m7_tracks.csv뿐이고, 없는 내용은 ‘자료 없음’으로 적어 줘. ‘정확히 식별’ 같은 말은 쓰지 말고, 맨 위에 '
                                                 '‘AI 초안, 담당자 확인 전’이라고 표시해 줘.',
                                      'min': 30,
                                      'checklist': ['독자를 적었어요', '문서 순서를 적었어요', '근거 자료를 한정했어요', '없는 내용을 어떻게 쓸지 적었어요', '표현 규칙과 초안 표시를 요청했어요'],
                                      'model_label': '📝 모범 프롬프트',
                                      'placeholder': '꼭 입력해 주세요 · 내 말로 프롬프트를 써 보세요'}]}],
               'outputs': [{'id': 'm7_choice',
                            'title': 'M7 드론·조류 분류 모델 선택 기록',
                            'sections': [{'h': '1. 모델 선택 기록', 'blocks': ['s2_vibe']}, {'h': '2. 내가 쓴 프롬프트', 'blocks': ['s3_write']}]}]},
              {'id': 'm8',
               'no': 8,
               'title': '사람 확인(HITL) 경보 처리 절차 설계',
               'short': '민감 정보와 사람 확인',
               'kind': '협업',
               'competencies': ['D6'],
               'outline': 'AI에게 무엇을 넘기면 안 되는지 나눠 보고, 값은 빼고 구조만 알려 경보 처리 순서표와 절차서 초안을 받아요. 어디까지 AI에게 맡기고 어디서 사람이 정할지 정해요.',
               'brief': {'subtitle': '경보 처리 순서와 절차서: 민감 정보 가리고 사람이 정하기',
                         'time': '약 30분',
                         'goal': 'AI에 넣으면 안 되는 정보를 가려 프롬프트를 쓰고, 경보 처리 순서표와 절차서를 받아 사람이 확인하고 승인할 일을 정해요.',
                         'why': 'AI가 울린 경보, 누가 보고 누가 정할까요? 위치와 근무자 이름이 든 대기열을 공개 AI에 그대로 넣어도 될까요? 이런 일에서 AI는 순서를 제안할 뿐이고, 결정은 사람이 해요.',
                         'case': {'title': '해외 사례: 신고 170건 가운데 무엇을 믿어야 할까요',
                                  'source': 'Gatwick 공항 사건 정리 자료, The National, TravelMole 보도 (2018.12~)',
                                  'points': ['2018년 12월 영국 개트윅 공항은 드론 목격 신고로 약 33시간 운영이 멈춰, 항공편 1,000편 이상과 승객 14만 명 이상이 영향을 받았어요.',
                                             '드론 목격 신고 170건 가운데 경찰은 115건을 믿을 만하다고 봤어요. 다만 경찰에 전달된 사진이나 영상은 없었고, 나중의 일부 목격은 경찰 자체 드론이었을 수 있다는 설명도 '
                                             '나왔어요.',
                                             '사람의 신고도, AI의 경보도 확인 절차가 필요해요. 누가 무엇을 근거로 어디까지 정할지 미리 적어 둬야 해요.']},
                         'team': '서상황(관제 담당)이 대기열을 정리하고, 조당직(당직 책임자)이 출동 건의를 승인해요. 전우님은 민감 정보를 가려 AI에게 요청해요.',
                         'terms': [{'t': 'HITL (사람 확인)',
                                    'en': 'Human-in-the-Loop',
                                    'fig': 'f_hitl',
                                    'd': 'AI가 만든 결과를 사람이 확인하고 결정하는 절차를 일에 넣는 방식이에요.',
                                    'easy': 'AI는 제안, 결정은 사람이에요.',
                                    'use': '경보 상태를 사람이 바꾸게 정해요.'},
                                   {'t': '자동화 편향',
                                    'en': 'automation bias',
                                    'fig': 'f_bias',
                                    'd': '기계가 낸 결과를 확인 없이 믿어 버리는 경향이에요.',
                                    'easy': '내비게이션만 믿고 강으로 들어가는 것과 비슷해요.',
                                    'use': '점수와 함께 확인할 근거를 화면에 보여 줘요.'},
                                   {'t': '넣으면 안 되는 정보',
                                    'en': 'sensitive data',
                                    'fig': 'f_secret',
                                    'd': '조직 밖으로 나가면 안 되는 위치, 인명, 시설 정보 같은 것이에요.',
                                    'easy': '게시판에 올리면 안 되는 것은 공개 AI에도 넣지 않아요.',
                                    'use': '열 이름과 가상 샘플만으로도 코드를 받을 수 있어요.'},
                                   {'t': '근무 기록',
                                    'en': 'duty log',
                                    'fig': 'f_log',
                                    'd': '누가, 언제, 무엇을, 왜 했는지 남기는 기록이에요. 고친 이력도 함께 남겨요.',
                                    'easy': '나중에 다시 봐도 이유를 알 수 있게 적는 거예요.',
                                    'use': '경보를 닫을 때 이유를 적게 해요.'}],
                         'steps': ['나눠 보기: AI에 넣어도 되는 것과 안 되는 것', '구조만 알리기: 값은 빼고 구조만 알려 요청하기', '사람이 할 일: 사람이 확인할 일과 내 프롬프트'],
                         'glossary': [],
                         'outputs': ['M8 경보 처리 순서표와 절차서']},
               'start_messages': [['dm_seo', 'seo', '{call}, 경보가 한꺼번에 몰리면 무엇부터 볼지 헷갈려요. 대기열은 m8_queue.csv예요. 순서표와 절차서 초안을 부탁해요.'],
                                  ['dm_jo', 'jo', '{call}, camera_pos랑 operator 칸은 밖으로 나가면 안 돼요. 공개 AI에 넣지 마세요.']],
               'finish_messages': [['dm_jo', 'jo', '{call}, 구조만 알려 주고 받았군요. 절차서는 제가 읽고 고쳐서 확정할게요 👍']],
               'steps': [{'short': '나눠 보기',
                          'title': 'AI에 넣어도 되는 것과 안 되는 것',
                          'lead': '프롬프트에 넣기 전에 **한 번 나눠 보세요.** 값이 없어도 구조만 알면 코드를 받을 수 있어요.',
                          'blocks': [{'id': 's1_cls',
                                      'type': 'classify',
                                      'title': '공개 AI에 넣어도 될까요?',
                                      'text': '조직이 승인한 환경이 아니라 일반 공개 AI라고 생각하고 나눠 보세요.',
                                      'primary_label': '판단',
                                      'item_label': '내용',
                                      'contrib': False,
                                      'categories': [{'id': 'ok', 'name': '넣어도 돼요'}, {'id': 'no', 'name': '넣으면 안 돼요'}],
                                      'items': [{'id': 'a',
                                                 'title': '열 이름: alarm_id, type, score, due_min',
                                                 'accept': ['ok'],
                                                 'explain': '열 이름만으로는 실제 상황을 알 수 없어요. 코드를 받는 데 꼭 필요한 정보예요.',
                                                 'hints': ['실제 값이 들어 있나요?', '이 정보만으로 위치나 사람을 알 수 있나요?']},
                                                {'id': 'b',
                                                 'title': '카메라 설치 위치와 근무자 이름이 든 대기열 원본',
                                                 'accept': ['no'],
                                                 'explain': '위치와 인명이 그대로 나가요. 원본은 내 환경에서만 돌려요.',
                                                 'hints': ['원본에는 무엇이 들어 있을까요?', '조직 밖으로 나가도 되는 내용인가요?']},
                                                {'id': 'c',
                                                 'title': '직접 지어낸 가상 샘플 2줄',
                                                 'accept': ['ok'],
                                                 'explain': '값의 모양을 보여 주는 가장 안전한 방법이에요.',
                                                 'hints': ['실제로 있었던 값인가요?', '지어낸 값이면 무엇이 문제일까요?']},
                                                {'id': 'd',
                                                 'title': '경보 영상 캡처 원본',
                                                 'accept': ['no'],
                                                 'explain': '시설 모습과 위치를 알 수 있어요. 공개 AI에 넣지 않아요.',
                                                 'hints': ['사진에 무엇이 찍혀 있을까요?', '이 사진으로 어디인지 알 수 있을까요?']}]}]},
                         {'short': '구조만 알리기',
                          'title': '값은 빼고 구조만 알려 요청하기',
                          'lead': '**민감 정보를 넣은 문장**도 일부러 골라 보세요. 결과는 나와도 무엇이 문제인지 볼 수 있어요.',
                          'blocks': [{'id': 's2_vibe',
                                      'type': 'vibe',
                                      'title': '경보 처리 순서표와 절차서',
                                      'ask': ['seo', '동시에 들어온 경보를 어떤 순서로 볼지 정하고, 절차서 초안까지 받고 싶어요.'],
                                      'file': 'm8_queue.csv',
                                      'slots': [{'id': 'safe',
                                                 'label': '넘길 정보',
                                                 'why': '값은 빼고 구조만 알려요.',
                                                 'options': [{'text': '실제 값은 넣지 않을게. m8_queue.csv의 열은 alarm_id, type, score, due_min, camera_pos, '
                                                                      'operator인데, camera_pos와 operator는 밖으로 나가면 안 되니 결과에도 넣지 말아 줘. 코드는 내 환경에서 돌릴게.',
                                                              'good': True},
                                                             {'text': '대기열 원본 전체를 위치랑 근무자 이름까지 그대로 붙여 넣을게. 여기서 알아서 정리해 줘.', 'good': False},
                                                             {'text': '', 'good': False}],
                                                 'miss': {'say': '붙여 주신 내용으로 순서를 정리했어요.',
                                                          'code': 'import pandas as pd\n'
                                                                  "df = pd.read_csv('{VDIR}/m8_queue.csv')\n"
                                                                  "display(df.sort_values('due_min'))",
                                                          'problem': '순서는 나왔지만 **위치와 근무자 이름이 이미 외부 AI로 나갔고**, 결과 표에도 그대로 있어요. 한 번 보낸 내용은 되돌리기 어려워요. '
                                                                     '코드를 받는 데 값은 필요 없었어요.',
                                                          'hint': '열 이름만 알려 주고, 민감한 칸은 결과에서도 빼 달라고 적어 주세요.'}},
                                                {'id': 'rule',
                                                 'label': '규칙',
                                                 'why': '순서를 숫자로 적어요.',
                                                 'options': [{'text': 'type이 video_loss면 ‘1 즉시 확인’, person이나 drone이면서 score 0.6 이상이면 ‘1 즉시 확인’, 나머지 '
                                                                      'person, drone, vehicle은 ‘2 우선 확인’, 그 밖은 ‘3 순차 확인’이야. 같은 등급 안에서는 due_min이 짧은 '
                                                                      '순서로 정렬해 줘. score 빈칸은 오류로 보지 말아 줘.',
                                                              'good': True},
                                                             {'text': '중요한 것부터 정렬해 줘.', 'good': False},
                                                             {'text': '', 'good': False}],
                                                 'miss': {'say': '점수가 높은 순서로 정렬했어요.',
                                                          'code': 'import pandas as pd\n'
                                                                  "df = pd.read_csv('{VDIR}/m8_queue.csv')\n"
                                                                  "display(df.sort_values('score', ascending=False)[['alarm_id', 'type', 'score']])",
                                                          'problem': '오류는 없지만 점수가 없는 영상 손실 Q04가 맨 아래로 내려갔어요. 영상이 끊긴 카메라는 바로 봐야 하는데도요. ‘중요한 것’이 무엇인지 '
                                                                     '말하지 않아서예요.',
                                                          'hint': '등급 규칙과 정렬 기준을 숫자로 적어 주세요.'}},
                                                {'id': 'human',
                                                 'label': '사람의 몫',
                                                 'why': 'AI가 할 일과 사람이 할 일을 나눠요.',
                                                 'options': [{'text': '순서표는 제안일 뿐이야. HTML 절차서를 report 변수에 담아 줘. 경보 상태(접수, 영상 확인, 출동 건의, 승인, 종결)마다 '
                                                                      '담당을 적고, 출동 승인은 당직 책임자만 하며, 경보를 닫을 때 누가·언제·무엇을·왜를 근무 기록에 남긴다고 적어 줘. 맨 위에 ‘AI '
                                                                      '초안, 당직 책임자 확인 전’이라고 표시해 줘.',
                                                              'good': True},
                                                             {'text': '점수가 높으면 AI가 자동으로 출동시키게 해 줘.', 'good': False},
                                                             {'text': '', 'good': False}],
                                                 'miss': {'say': '점수가 0.8을 넘으면 자동으로 출동 명령을 내도록 만들었어요.',
                                                          'code': 'import pandas as pd\n'
                                                                  "df = pd.read_csv('{VDIR}/m8_queue.csv')\n"
                                                                  "for _, r in df[df['score'] >= 0.8].iterrows():\n"
                                                                  "    print(r['alarm_id'], r['type'], '→ 자동 출동 명령')",
                                                          'problem': '코드는 돌았지만 사람이 영상을 보기도 전에 출동이 정해져요. 점수는 틀릴 수 있어요. **결정할 일은 AI에게 넘기지 않아요.**',
                                                          'hint': 'AI는 어디까지 하고 누가 승인하는지 적어 주세요.'}}],
                                      'done': {'say': '값 없이 구조만으로 코드를 만들었어요. 내 환경에서 돌린 순서표와 절차서 초안이에요. 승인은 사람이 해요.',
                                               'code': 'import pandas as pd\n'
                                                       '\n'
                                                       "df = pd.read_csv('{VDIR}/m8_queue.csv')\n"
                                                       '\n'
                                                       'def grade(r):\n'
                                                       "    if r['type'] == 'video_loss':\n"
                                                       "        return '1 즉시 확인'\n"
                                                       "    if r['type'] in ('person', 'drone') and r['score'] >= 0.6:\n"
                                                       "        return '1 즉시 확인'\n"
                                                       "    if r['type'] in ('person', 'drone', 'vehicle'):\n"
                                                       "        return '2 우선 확인'\n"
                                                       "    return '3 순차 확인'\n"
                                                       '\n'
                                                       "df['등급'] = df.apply(grade, axis=1)\n"
                                                       "queue = df.sort_values(['등급', 'due_min'])[['등급', 'alarm_id', 'type', 'score', "
                                                       "'due_min']].fillna('')\n"
                                                       'display(queue)\n'
                                                       '\n'
                                                       "flow = pd.DataFrame([{'상태': '접수', '담당': '영상 모니터링 담당', '하는 일': '경보를 확인 대기열에 올림'},\n"
                                                       "                     {'상태': '영상 확인', '담당': '영상 모니터링 담당', '하는 일': '영상을 직접 보고 판단'},\n"
                                                       "                     {'상태': '출동 건의', '담당': '관제 담당', '하는 일': '근거와 함께 건의'},\n"
                                                       "                     {'상태': '승인', '담당': '당직 책임자', '하는 일': '출동 여부 결정'},\n"
                                                       "                     {'상태': '종결', '담당': '관제 담당', '하는 일': '근무 기록 작성 (누가·언제·무엇을·왜)'}])\n"
                                                       "report = ('<p><b>AI 초안, 당직 책임자 확인 전.</b></p><h2>사람 확인 경보 처리 절차서</h2>'\n"
                                                       "          '<p>AI는 확인 순서를 제안해요. 출동 승인은 당직 책임자만 해요.</p>'\n"
                                                       "          '<h3>경보 상태와 담당</h3>' + flow.to_html(index=False)\n"
                                                       "          + '<h3>처리 순서 (제안)</h3>' + queue.to_html(index=False))",
                                               'lesson': '값을 넘기지 않아도 코드는 똑같이 받을 수 있어요. 규칙을 적으니 점수가 없는 영상 손실 Q04도 앞쪽으로 올라왔어요. 그리고 AI는 순서만 제안하고, 출동은 '
                                                         '당직 책임자가 승인해요.'},
                                      'on_ok': [['dm_seo', 'seo', '{call}, 영상 손실이 앞으로 올라왔네요. 순서가 분명해져서 좋아요.']]}]},
                         {'short': '사람이 할 일',
                          'title': '사람이 확인할 일과 내 프롬프트',
                          'lead': 'AI가 순서를 잘 정해도 **결정하는 것은 사람이에요.**',
                          'blocks': [{'id': 's3_decide',
                                      'type': 'decide',
                                      'title': '절차 제안 판단하기',
                                      'text': '절차서 초안을 본 팀원들의 제안이에요. 받아들일지 정하고 이유를 적어 주세요.',
                                      'reason_min': 8,
                                      'items': [{'id': 'a',
                                                 'who': 'oh',
                                                 'text': '점수가 0.9를 넘으면 영상 확인 없이 바로 출동 건의로 넘겨요. 시간이 아깝잖아요.',
                                                 'answer': 'reject',
                                                 'hint': '개트윅 사례에서 무엇이 문제였나요?',
                                                 'explain': '점수만 믿으면 자동화 편향에 빠져요. 짧게라도 영상을 사람이 봐야 근거 있는 건의가 돼요.',
                                                 'reason': '점수만 믿으면 틀린 경보로 출동할 수 있어요. 사람이 영상을 봐야 근거가 생겨요.'},
                                                {'id': 'b',
                                                 'who': 'jo',
                                                 'text': '동시에 몰리면 등급 1부터 보고, 처리 못 한 건은 다음 근무자에게 넘길 때 기록에 남겨요.',
                                                 'answer': 'accept',
                                                 'hint': '기록이 없으면 다음 근무자는 어떻게 될까요?',
                                                 'explain': '다음 사람이 이어서 볼 수 있어요. 프롬프트에 항목만 더하면 절차서에 들어가요.',
                                                 'reason': '기록이 있어야 다음 근무자가 이어서 확인할 수 있어요.'}]},
                                     {'id': 's3_write',
                                      'type': 'write',
                                      'title': '민감 정보를 뺀 프롬프트 쓰기',
                                      'prompt': '**새 요청**: 실제 근무 기록에서 처리 시간이 10분을 넘은 경보를 세어 보고 싶어요. 기록에는 근무자 이름과 카메라 위치가 들어 있어요. 값을 넘기지 않고 어떻게 '
                                                '요청할까요?',
                                      'example': '실제 값은 넣지 않고 구조만 알려 줄게. 파일은 log.csv이고 열은 alarm_id, operator, camera_pos, handle_min이야. 샘플은 Q1, OP1, '
                                                 'POS1, 7처럼 지어낸 값이야. handle_min이 10을 넘는 건수와 비율을 계산하는 코드를 만들어 줘. operator와 camera_pos는 결과에 넣지 말아 줘. '
                                                 '코드는 내 환경에서 돌릴게.',
                                      'min': 30,
                                      'checklist': ['실제 값을 넣지 않았어요', '열 이름과 가상 샘플로 설명했어요', '기준을 숫자로 적었어요', '결과에 민감 정보가 나오지 않게 했어요'],
                                      'model_label': '📝 모범 프롬프트',
                                      'placeholder': '꼭 입력해 주세요 · 내 말로 프롬프트를 써 보세요'}]}],
               'outputs': [{'id': 'm8_flow',
                            'title': 'M8 경보 처리 순서표와 절차서',
                            'sections': [{'h': '1. 공개 AI에 넣어도 될까요?', 'blocks': ['s1_cls']},
                                         {'h': '2. 순서표와 절차서', 'blocks': ['s2_vibe']},
                                         {'h': '3. 절차 제안 판단', 'blocks': ['s3_decide']},
                                         {'h': '4. 내가 쓴 프롬프트', 'blocks': ['s3_write']}]}]},
              {'id': 'm9',
               'no': 9,
               'title': '통합 관제 화면 구성과 동시 경보·영상 손실 대응',
               'short': '리뷰와 테스트',
               'kind': '개인',
               'competencies': ['D3', 'D5'],
               'outline': '영상 손실을 찾는 코드를 바로 고치지 않고, 먼저 설명하게 하고 답을 아는 테스트로 확인해요. 고칠 범위를 정해 고친 뒤, 통합 관제 화면 구성과 대응 기록을 받아요.',
               'brief': {'subtitle': '영상 손실 코드 리뷰와 관제 화면: 테스트로 확인하기',
                         'time': '약 30분',
                         'goal': 'AI에게 코드를 먼저 설명하게 하고, 답을 아는 테스트와 수정 범위를 요청해 영상 손실 계산을 바로잡은 뒤 관제 화면 구성 문서를 받아요.',
                         'why': '경보가 한꺼번에 몰릴 때, 화면은 무엇을 먼저 보여 줘야 할까요? 그리고 영상이 끊겼는데 도구가 조용하면 어떻게 될까요? 숫자가 이상하다고 바로 ‘고쳐 줘’라고 하면 AI가 다른 곳까지 바꾸기 쉬워요. 먼저 '
                                '설명하게 하고, 답을 아는 값으로 확인해요.',
                         'case': {'title': '해외 사례: 사흘 밤 이어진 드론, 듣지 않은 전파 방해 장비',
                                  'source': 'Euromaidan Press, Belga, Paris Match 보도 (2025.11)',
                                  'points': ['2025년 11월 벨기에 클라이네 브로겔 공군 비행장 위로 며칠 밤 연속 드론이 나타났어요. 대수와 날짜는 보도마다 조금씩 달라요.',
                                             '벨기에 국방장관은 전파 방해 장비를 썼지만 듣지 않았고, 경찰 헬기로도 쫓지 못했다고 설명했어요. 첩보 활동으로 보인다면서도 누구인지는 단정하지 않았어요.',
                                             '장비가 제대로 동작한다고 믿으면 안 돼요. 장비나 도구가 멈췄을 때 어떻게 알아채고 무엇을 먼저 볼지 정해 둬야 해요.']},
                         'team': '오영상(모니터링 담당)이 끊김을 신고했어요. 민시설이 카메라를 점검해요. 코드는 다른 부서에서 받아 왔어요.',
                         'terms': [{'t': '통합 관제 화면',
                                    'en': 'VMS alarm console',
                                    'fig': 'f_console',
                                    'd': '영상 관리 시스템(VMS)의 경보 화면이에요. 경보 목록, 상태, 우선순위, 담당자, 배치도, 경보 영상을 한곳에 보여 줘요.',
                                    'easy': '여러 카메라의 일을 한눈에 보는 근무 화면이에요.',
                                    'use': 'AI에게 들어갈 요소와 순서를 적어 요청해요.'},
                                   {'t': '영상 손실 경보',
                                    'en': 'video loss alarm',
                                    'fig': 'f_gap',
                                    'd': '카메라 신호가 일정 시간 끊기면 올리는 경보예요. 렌즈가 가려지면 카메라 가림 경보를 올려요.',
                                    'easy': '‘이 카메라, 지금 안 보여요’라고 알려 주는 거예요.',
                                    'use': '몇 초를 끊김으로 볼지 숫자로 적어요.'},
                                   {'t': '코드 리뷰',
                                    'en': 'code review',
                                    'fig': 'f_review',
                                    'd': '코드를 고치기 전에 읽고 의심되는 곳을 찾는 일이에요.',
                                    'easy': 'AI에게 ‘먼저 설명해 봐’라고 시키는 것도 리뷰예요.',
                                    'use': '설명, 테스트, 수정, 비교 순서로 요청해요.'},
                                   {'t': '수정 범위',
                                    'en': 'scope',
                                    'fig': 'f_scope',
                                    'd': '고쳐도 되는 곳과 건드리면 안 되는 곳을 정한 것이에요.',
                                    'easy': '‘이 함수만 고쳐’라고 말하는 거예요.',
                                    'use': '범위가 없으면 잘 되던 부분까지 바뀔 수 있어요.'}],
                         'steps': ['먼저 읽기: 고치기 전에 먼저 설명하게 하기', '테스트로 고치기: 설명·테스트·범위를 담아 요청하기', '내 말로 쓰기: 후속 제안 판단과 내 프롬프트'],
                         'glossary': [],
                         'outputs': ['M9 통합 관제 화면 구성과 수정 기록']},
               'start_messages': [['dm_oh', 'oh', '{call}, 어제 CAM2가 끊겼는데 도구에는 안 잡혔어요. 신호 기록은 m9_signal.csv예요. 코드가 맞는지 봐 주실래요?']],
               'finish_messages': [['dm_oh', 'oh', '{call}, 이제 끊긴 카메라가 제대로 보여요. 몰릴 때 무엇부터 볼지도 분명해졌어요.']],
               'steps': [{'short': '먼저 읽기',
                          'title': '고치기 전에 먼저 설명하게 하기',
                          'lead': '숫자가 이상할 때 바로 ‘고쳐 줘’라고 하면 **무엇이 왜 바뀌었는지 모르게 돼요.**',
                          'blocks': [{'id': 's1_info',
                                      'type': 'info',
                                      'title': '남의 코드를 고칠 때의 네 단계',
                                      'icon': '🧪',
                                      'text': ['AI가 만든 코드도, 다른 사람이 만든 코드도 같아요. 아래 순서로 요청해요.'],
                                      'table': {'head': ['단계', '프롬프트에 적을 말', '왜'],
                                                'rows': [['설명', '고치기 전에 무엇을 계산하는지 설명해 줘', '내가 이해한 뜻과 같은지 확인해요'],
                                                         ['테스트', '신호가 0, 10, 50, 60초면 40초 끊김 하나가 나오는 assert를 만들어 줘', '답을 아는 값으로 확인해요'],
                                                         ['수정', 'find_gaps 함수만 고쳐 줘', '다른 곳이 바뀌지 않아요'],
                                                         ['비교', '고치기 전후를 표로 보여 줘', '무엇이 달라졌는지 봐요']]},
                                      'figure': 'f_review'},
                                     {'id': 's1_q',
                                      'type': 'quiz',
                                      'title': '처음 무엇을 요청할까요?',
                                      'text': '영상 손실 코드가 이상해 보여요. 가장 먼저 AI에게 무엇을 요청할까요?',
                                      'options': ['전부 새로 짜 줘.', '이 코드가 무엇을 어떻게 계산하는지 먼저 설명하고, 의심되는 곳을 알려 줘.', 'CAM2가 나오게 아무 값이나 넣어 줘.'],
                                      'answer': 1,
                                      'explain': '먼저 이해해야 무엇을 고칠지 정할 수 있어요. 셋째처럼 원하는 답을 정해 주면 AI는 거기에 맞춰 버려요.',
                                      'hints': ['어느 요청이 원인을 찾게 할까요?', '결과를 정해 놓고 요청하면 어떻게 될까요?']}]},
                         {'short': '테스트로 고치기',
                          'title': '설명·테스트·범위를 담아 요청하기',
                          'lead': '지금 코드는 **오류 없이 돌아요.** 그래서 더 찾기 어려워요.',
                          'blocks': [{'id': 's2_vibe',
                                      'type': 'vibe',
                                      'title': '영상 손실 코드 리뷰와 관제 화면',
                                      'ask': ['oh', '이 코드로 낸 영상 손실이 맞는지 봐 주세요.'],
                                      'file': 'm9_signal.csv',
                                      'slots': [{'id': 'explain',
                                                 'label': '먼저 설명',
                                                 'why': '고치기 전에 읽게 해요.',
                                                 'options': [{'text': 'm9_signal.csv의 열은 camera, t_sec(신호가 온 시각, 초)야. 30초 넘게 신호가 없으면 영상 손실로 봐야 해. '
                                                                      '고치기 전에 find_gaps가 무엇을 무엇과 비교하는지 설명하고, 의심되는 곳을 알려 줘.',
                                                              'good': True},
                                                             {'text': '그냥 바로 고쳐 줘.', 'good': False},
                                                             {'text': '', 'good': False}],
                                                 'miss': {'say': '고쳤어요! 기준을 없애서 모든 끊김을 찾게 했어요.',
                                                          'code': 'import pandas as pd\n'
                                                                  "df = pd.read_csv('{VDIR}/m9_signal.csv')\n"
                                                                  'rows = []\n'
                                                                  "for c, s in df.groupby('camera'):\n"
                                                                  "    gap = s['t_sec'].sort_values().diff().dropna()\n"
                                                                  "    rows.append({'카메라': c, '영상 손실': len(gap)})\n"
                                                                  'display(pd.DataFrame(rows))',
                                                          'problem': '이제 10초마다 오는 평범한 신호까지 전부 끊김으로 세어요. 무엇이 문제인지 설명 없이 고치니 **다른 잘못으로 바뀐 줄도 모르게 돼요.**',
                                                          'hint': '고치기 전에 설명부터 하게 해 주세요.'}},
                                                {'id': 'test',
                                                 'label': '테스트',
                                                 'why': '답을 아는 값으로 확인해요.',
                                                 'options': [{'text': '신호가 0, 10, 50, 60초면 40초 끊김 하나가 나오고, 0, 30, 60초면 끊김이 없는 assert 테스트를 먼저 만들어 줘. '
                                                                      '정확히 30초는 끊김이 아니야.',
                                                              'good': True},
                                                             {'text': '잘 되는지 알아서 확인해 줘.', 'good': False},
                                                             {'text': '', 'good': False}],
                                                 'miss': {'say': '확인했어요. 잘 동작해요!',
                                                          'code': 'import pandas as pd\n'
                                                                  "df = pd.read_csv('{VDIR}/m9_signal.csv')\n"
                                                                  "print('카메라', df['camera'].nunique(), '대 확인 완료. 잘 동작해요!')",
                                                          'problem': '‘잘 동작해요’라는 말뿐이고, 무엇으로 확인했는지 없어요. 정확히 30초인 경우를 어떻게 세는지도 알 수 없어요.',
                                                          'hint': '손으로 계산할 수 있는 작은 예시와 경계값을 적어 주세요.'}},
                                                {'id': 'scope',
                                                 'label': '범위와 문서',
                                                 'why': '고칠 곳과 문서를 정해요.',
                                                 'options': [{'text': 'find_gaps 함수만 고치고 다른 코드는 그대로 둬. 전후 비교표를 보여 주고, m9_waves.csv(wave_id, alarms, '
                                                                      'handled_in_3min)로 동시 경보 수별 3분 내 처리율도 계산해 줘. 통합 관제 화면에 들어갈 요소(경보 목록, 상태, 우선순위, '
                                                                      '담당자, 배치도, 경보 영상)와 영상 손실 경보를 맨 위에 보여 주는 순서를 HTML로 report 변수에 담아 줘.',
                                                              'good': True},
                                                             {'text': '알아서 깔끔하게 전부 정리해 줘.', 'good': False},
                                                             {'text': '', 'good': False}],
                                                 'miss': {'say': '깔끔하게 정리했어요. 신호 기록은 1분 단위로 줄였어요.',
                                                          'code': 'import pandas as pd\n'
                                                                  "df = pd.read_csv('{VDIR}/m9_signal.csv')\n"
                                                                  "small = df[df['t_sec'] % 60 == 0]\n"
                                                                  "print('행 수', len(df), '→', len(small))\n"
                                                                  "display(small.groupby('camera').size().reset_index(name='신호 수'))",
                                                          'problem': '부탁하지 않은 일까지 했어요. 1분 단위로 줄이면 45초 끊김을 아예 볼 수 없어요. 범위를 정하지 않으면 잘 되던 데이터도 바뀌어요.',
                                                          'hint': '고칠 곳과 건드리면 안 되는 것을 적어 주세요.'}}],
                                      'done': {'say': '기준이 300초로 잘못 적혀 있었어요. 설정값 LIMIT(30초)을 쓰도록 find_gaps만 고쳤고, 테스트가 통과했어요.',
                                               'code': 'import pandas as pd\n'
                                                       'import matplotlib.pyplot as plt\n'
                                                       '\n'
                                                       "df = pd.read_csv('{VDIR}/m9_signal.csv')\n"
                                                       'LIMIT = 30  # 30초 넘게 신호가 없으면 영상 손실\n'
                                                       '\n'
                                                       'def old_gaps(t):\n'
                                                       '    gap = t.sort_values().diff().dropna()\n'
                                                       '    return gap[gap > 300].tolist()\n'
                                                       '\n'
                                                       'def find_gaps(t):\n'
                                                       '    gap = t.sort_values().diff().dropna()\n'
                                                       '    return gap[gap > LIMIT].tolist()\n'
                                                       '\n'
                                                       '# 테스트\n'
                                                       'assert find_gaps(pd.Series([0, 10, 50, 60])) == [40.0]\n'
                                                       'assert find_gaps(pd.Series([0, 30, 60])) == []\n'
                                                       "print('테스트 통과')\n"
                                                       '\n'
                                                       'rows = []\n'
                                                       "for c, s in df.groupby('camera'):\n"
                                                       "    rows.append({'카메라': c, '고치기 전': len(old_gaps(s['t_sec'])), '고친 뒤': "
                                                       "len(find_gaps(s['t_sec'])), '끊긴 시간(초)': str(find_gaps(s['t_sec']))})\n"
                                                       'table = pd.DataFrame(rows)\n'
                                                       'display(table)\n'
                                                       '\n'
                                                       "w = pd.read_csv('{VDIR}/m9_waves.csv')\n"
                                                       "wave = w.groupby('alarms').agg(횟수=('wave_id', 'size'), 경보=('alarms', 'sum'), "
                                                       "처리=('handled_in_3min', 'sum')).reset_index()\n"
                                                       "wave['3분 내 처리율(%)'] = (wave['처리'] / wave['경보'] * 100).round(1)\n"
                                                       "wave = wave.rename(columns={'alarms': '동시 경보 수'})\n"
                                                       'display(wave)\n'
                                                       '\n'
                                                       "ax = wave.plot(x='동시 경보 수', y='3분 내 처리율(%)', kind='line', marker='o', legend=False, "
                                                       'figsize=(4.8, 2.8))\n'
                                                       "ax.set_xlabel('alarms at once')\n"
                                                       "ax.set_ylabel('handled in 3 min (%)')\n"
                                                       'ax.set_ylim(0, 105)\n'
                                                       '\n'
                                                       "screen = pd.DataFrame([{'순서': 1, '요소': '영상 손실 경보', '이유': '안 보이는 곳을 먼저 알아야 해요'},\n"
                                                       "                       {'순서': 2, '요소': '경보 목록 (우선순위·상태·담당자)', '이유': '몰릴 때 무엇부터 볼지 정해요'},\n"
                                                       "                       {'순서': 3, '요소': '경보 영상', '이유': '사람이 직접 확인해요'},\n"
                                                       "                       {'순서': 4, '요소': '배치도', '이유': '어디인지 바로 봐요'}])\n"
                                                       "report = ('<p><b>AI 초안, 관제 담당 확인 전.</b></p><h2>통합 관제 화면 구성과 대응 기록</h2>'\n"
                                                       "          '<h3>화면 구성</h3>' + screen.to_html(index=False)\n"
                                                       "          + '<h3>영상 손실 전후</h3>' + table.to_html(index=False)\n"
                                                       "          + '<h3>동시 경보와 처리율</h3>' + wave.to_html(index=False)\n"
                                                       "          + '<p>가상 기록 기준이에요. 동시 경보가 많은 경우는 횟수가 적어 참고로만 봐요.</p>')",
                                               'lesson': '설명을 먼저 받으니 300이 잘못 적힌 것이 드러났고, 테스트가 있어 맞게 고쳤다는 것을 확인했어요. CAM4의 정확히 30초 끊김은 기준대로 세지 않았어요. '
                                                         '동시 경보가 많을수록 처리율이 떨어지는 흐름도 보여요. 다만 횟수가 적은 구간은 참고로만 봐요.'},
                                      'start': {'say': '지금 쓰고 있는 코드예요. CAM2가 안 잡혀요.',
                                                'code': 'import pandas as pd\n'
                                                        '\n'
                                                        "df = pd.read_csv('{VDIR}/m9_signal.csv')\n"
                                                        'LIMIT = 30  # 30초 넘게 신호가 없으면 영상 손실\n'
                                                        '\n'
                                                        'def find_gaps(t):\n'
                                                        '    gap = t.sort_values().diff().dropna()\n'
                                                        '    return gap[gap > 300].tolist()\n'
                                                        '\n'
                                                        "rows = [{'카메라': c, '영상 손실': len(find_gaps(s['t_sec'])), '끊긴 시간(초)': find_gaps(s['t_sec'])} "
                                                        "for c, s in df.groupby('camera')]\n"
                                                        'display(pd.DataFrame(rows))'},
                                      'on_ok': [['dm_oh', 'oh', '{call}, 300이 문제였군요. 설명을 먼저 받으니 저도 이해가 돼요.']]}]},
                         {'short': '내 말로 쓰기',
                          'title': '후속 제안 판단과 내 프롬프트',
                          'lead': '결과를 받은 뒤에는 **다음 조치를 사람이** 정해요.',
                          'blocks': [{'id': 's3_decide',
                                      'type': 'decide',
                                      'title': '후속 제안 판단하기',
                                      'text': '결과를 본 팀원들의 제안이에요. 받아들일지 정하고 이유를 적어 주세요.',
                                      'reason_min': 8,
                                      'items': [{'id': 'a',
                                                 'who': 'oh',
                                                 'text': '끊김이 자주 울리면 번거로우니 기준을 300초로 되돌려요.',
                                                 'answer': 'reject',
                                                 'hint': '5분 동안 안 보이는 동안 무슨 일이 생길 수 있을까요?',
                                                 'explain': '번거로움 때문에 안 보이는 시간을 늘리면 안 돼요. 끊김의 원인을 찾아 줄이는 쪽이 먼저예요.',
                                                 'reason': '안 보이는 시간이 늘어나요. 기준 대신 끊기는 원인을 찾아 줄여야 해요.'},
                                                {'id': 'b',
                                                 'who': 'min',
                                                 'text': '끊김이 잦은 CAM3의 배선과 전원을 먼저 점검하고, 같은 도구로 다음 주에 다시 비교해요.',
                                                 'answer': 'accept',
                                                 'hint': '도구를 다시 돌리면 무엇이 좋을까요?',
                                                 'explain': '원인을 고치고 같은 도구로 비교하면 효과가 숫자로 보여요.',
                                                 'reason': '원인을 고치고 같은 도구로 비교해야 효과를 숫자로 볼 수 있어요.'}]},
                                     {'id': 's3_write',
                                      'type': 'write',
                                      'title': '리뷰 프롬프트 쓰기',
                                      'prompt': '**새 요청**: 다른 팀에서 받은 평균 처리 시간 계산 코드가 이상해 보여요. AI에게 어떻게 요청할까요? 고치기 전 설명, 테스트, 범위를 담아 보세요.',
                                      'example': '이 코드를 고치기 전에 평균 처리 시간을 어떻게 계산하는지 먼저 설명하고, 의심되는 곳을 알려 줘. 처리 시간이 1분, 2분, 3분이면 평균 2분이 나오는 assert 테스트도 '
                                                 '만들어 줘. 고치는 건 계산 함수만 하고, 원본 파일과 다른 코드는 건드리지 마. 고치기 전후 결과를 표로 보여 줘.',
                                      'min': 30,
                                      'checklist': ['먼저 설명하게 했어요', '답을 아는 테스트를 요청했어요', '고칠 곳과 건드리면 안 되는 곳을 적었어요', '전후를 비교하게 했어요'],
                                      'model_label': '📝 모범 프롬프트',
                                      'placeholder': '꼭 입력해 주세요 · 내 말로 프롬프트를 써 보세요'}]}],
               'outputs': [{'id': 'm9_console',
                            'title': 'M9 통합 관제 화면 구성과 수정 기록',
                            'sections': [{'h': '1. AI에게 보낸 프롬프트와 결과', 'blocks': ['s2_vibe']},
                                         {'h': '2. 후속 제안 판단', 'blocks': ['s3_decide']},
                                         {'h': '3. 내가 쓴 리뷰 프롬프트', 'blocks': ['s3_write']}]}]},
              {'id': 'm10',
               'no': 10,
               'title': '사후검토(AAR)와 설정값 재조정, 시설 방호 브리핑',
               'short': '묶어서 보고하기',
               'kind': '협업',
               'competencies': ['D4', 'J1'],
               'outline': '앞에서 배운 프롬프트를 모아 재조정 전후를 비교하고 브리핑을 스스로 점검하는 미니 도구를 만든 뒤, 나만의 프롬프트 틀을 정리해요.',
               'brief': {'subtitle': '재조정 전후 비교 도구와 브리핑: 묶어서 돌리고 조건까지 적기',
                         'time': '약 30분',
                         'goal': '다시 돌릴 수 있는 미니 도구를 요청하고, 수치마다 조건을 밝힌 브리핑과 자동 점검, 내 프롬프트 틀을 만들어요.',
                         'why': '교훈은 기록이 아니라 조치가 될 때 남아요. 자동화할수록 틀린 내용도 빠르게 퍼져요. 수치에는 조건을 붙이고, 확인된 것과 아닌 것을 나누고, 올리기 전에 사람이 확인해야 해요.',
                         'case': {'title': '해외 사례: 쿠웨이트 공격 뒤에도 공개되지 않은 조사 내용',
                                  'source': 'AP, CBS, CNN 인용 보도 (2026.3~10)',
                                  'points': ['2026년 3월 1일 쿠웨이트 민간 항구 안의 미군 임시 작전 시설이 이란의 드론 공격을 받아 미군 6명이 숨졌어요. AP는 이곳이 주요 주둔지에서 10마일 넘게 떨어져 있었다고 '
                                             '전했어요.',
                                             '경보나 사이렌이 없었다는 CNN의 소식통 발언과, 생존자들이 미 국방부 설명을 반박하는 CBS 보도가 나왔어요. AP에 따르면 미 육군 조사는 6월에 끝났지만 보고서는 아직 공개되지 '
                                             '않았어요.',
                                             '확인된 것과 확인되지 않은 것을 나눠 적어야 해요. 사후검토와 브리핑에서 가장 먼저 지킬 일이에요.']},
                         'team': '임반장이 사후검토를 맡고, 조당직이 브리핑을 받아요. 강계획과 서상황이 의견을 줘요. 사람을 탓하지 않고 원인을 찾아요.',
                         'terms': [{'t': '미니 도구',
                                    'en': 'small tool',
                                    'fig': 'f_tool',
                                    'd': '설정값, 계산, 표와 그림, 문서까지 한 번에 돌아가는 작은 코드 묶음이에요.',
                                    'easy': '데이터만 바꾸면 다시 쓸 수 있는 계산기예요.',
                                    'use': '설정값을 맨 위에 모아 달라고 요청해요.'},
                                   {'t': '설정값 재조정',
                                    'en': 're-tuning',
                                    'fig': 'f_aar',
                                    'd': '사후검토 결과로 임계값, 영역 같은 설정을 다시 맞추는 일이에요.',
                                    'easy': '해 본 결과를 보고 다이얼을 다시 돌리는 거예요.',
                                    'use': '전후를 같은 기준으로 비교해요.'},
                                   {'t': '문서 자동 점검 (린트)',
                                    'en': 'document lint',
                                    'fig': 'f_lint',
                                    'd': '문서에 꼭 들어갈 항목이 빠졌는지 코드로 검사하는 것이에요.',
                                    'easy': '제출 전 체크리스트를 코드로 만든 거예요.',
                                    'use': '표본 수, 한계, 남은 위험, AI 산출물 표기가 있는지 검사하게 해요.'},
                                   {'t': '프롬프트 모음',
                                    'en': 'prompt library',
                                    'fig': 'f_four',
                                    'd': '잘 통한 프롬프트를 틀로 모아 두고 다시 쓰는 개인 자료예요.',
                                    'easy': '자주 쓰는 업무 양식함이에요.',
                                    'use': '파일 이름과 숫자 자리를 빈칸으로 두면 다음에 바로 채워 쓸 수 있어요.'}],
                         'steps': ['도구로 묶기: 미니 도구와 브리핑 요청하기', '피드백과 점검: 팀 의견과 올리기 전 점검', '프롬프트 모음: 나만의 프롬프트 틀 만들기'],
                         'glossary': [],
                         'outputs': ['M10 사후검토 비교와 시설 방호 브리핑']},
               'start_messages': [['dm_lim',
                                   'lim',
                                   '{call}, 마지막이에요. 설정값을 다시 맞춘 전후를 비교해 조당직님께 드릴 브리핑을 만들어 주세요. 파일은 m10_week.csv예요. 다음 달에도 같은 방식으로 돌릴 수 있으면 좋겠어요.']],
               'finish_messages': [['dm_jo', 'jo', '{call}, 브리핑 잘 받았어요. 어떤 조건에서 조심할지, 무엇이 남은 위험인지 분명하네요. 고생 많았어요.']],
               'steps': [{'short': '도구로 묶기',
                          'title': '미니 도구와 브리핑 요청하기',
                          'lead': '지금까지 배운 것을 모두 써 보세요. **네 칸, 숫자 규칙, 근거 수치, 범위**가 모두 필요해요.',
                          'blocks': [{'id': 's1_vibe',
                                      'type': 'vibe',
                                      'title': '전후 비교 도구와 브리핑',
                                      'ask': ['lim', '재조정 전후를 조건별로 비교하고, 브리핑과 자동 점검까지 한 번에 나오게 해 주세요.'],
                                      'file': 'm10_week.csv',
                                      'slots': [{'id': 'tool',
                                                 'label': '구성',
                                                 'why': '다시 돌릴 수 있게 해요.',
                                                 'options': [{'text': 'm10_week.csv의 열은 record_id, period(before·after), condition(day·night·fog), '
                                                                      'is_target, alarmed, handle_min이야. 파일 이름과 표본 기준을 맨 위 설정값으로 두고, 계산은 함수로 나눠 줘. '
                                                                      '다음 달 파일로도 바로 돌릴 수 있게 해 줘.',
                                                              'good': True},
                                                             {'text': '분석 코드 짜 줘.', 'good': False},
                                                             {'text': '', 'good': False}],
                                                 'miss': {'say': '전체 경보 수를 비교했어요. 많이 줄었어요!',
                                                          'code': 'import pandas as pd\n'
                                                                  "df = pd.read_csv('{VDIR}/m10_week.csv')\n"
                                                                  "print(df.groupby('period')['alarmed'].sum())",
                                                          'problem': '경보 수만 비교했어요. 대상을 놓치게 됐는지 알 수 없고, 파일 이름도 코드 속에 박혀 있어 다음 달에 다시 쓰기 어려워요.',
                                                          'hint': '무엇을 설정값으로 둘지, 어떻게 나눌지 적어 주세요.'}},
                                                {'id': 'metric',
                                                 'label': '지표',
                                                 'why': '무엇을 어떻게 나눌지 정해요.',
                                                 'options': [{'text': 'period와 condition별로 대상 수, 재현율(대상 중 alarmed 비율), 하루 평균 불필요 경보(7일 기준), 평균 처리 '
                                                                      '시간을 표로 만들어 줘. 대상이 30건 미만이면 ‘표본 적음’으로 표시해 줘.',
                                                              'good': True},
                                                             {'text': '조건별 결과를 보여 줘.', 'good': False},
                                                             {'text': '', 'good': False}],
                                                 'miss': {'say': '조건별 재현율을 표로 만들었어요.',
                                                          'code': 'import pandas as pd\n'
                                                                  "df = pd.read_csv('{VDIR}/m10_week.csv')\n"
                                                                  "d = df[df['is_target'] == 1]\n"
                                                                  "display(d.groupby(['period', "
                                                                  "'condition'])['alarmed'].mean().round(2).reset_index())",
                                                          'problem': 'fog 구간의 값이 얼마나 믿을 만한지 알 수 없어요. 표본 수와 불필요 경보가 빠졌어요.',
                                                          'hint': '지표와 표본 수, 표본이 적을 때의 표시를 적어 주세요.'}},
                                                {'id': 'report',
                                                 'label': '브리핑',
                                                 'why': '독자와 조건 표시, 점검을 정해요.',
                                                 'options': [{'text': '당직 책임자가 읽을 1쪽 HTML 브리핑을 report 변수에 담아 줘. 순서는 결론 한 줄, 조건별 표, 그림, 확인된 것과 확인되지 '
                                                                      '않은 것, 남은 위험과 다음 조치, AI 산출물 표기야. 수치에는 가상 기록 몇 건 기준인지 적고, 표본 수와 남은 위험, AI 산출물 '
                                                                      '표기가 빠졌는지 검사하는 lint_report 함수도 만들어 줘.',
                                                              'good': True},
                                                             {'text': '재조정이 성공했다고 브리핑 써 줘.', 'good': False},
                                                             {'text': '', 'good': False}],
                                                 'miss': {'say': '성공했다는 브리핑을 썼어요.',
                                                          'code': "report = '<h2>재조정 성공 보고</h2><p>재조정으로 모든 조건에서 성능이 크게 향상되어 시설 방호에 문제가 없습니다.</p>'\n"
                                                                  "print('브리핑을 만들었어요.')",
                                                          'problem': '조건도 숫자도 없이 ‘모든 조건’, ‘문제가 없습니다’라고 단정했어요. 원하는 결론을 프롬프트에 적으면 AI는 거기에 맞춰 줘요. 재현율이 '
                                                                     '떨어진 조건이 있는데도요.',
                                                          'hint': '결론을 정해 주지 말고, 독자·순서·조건 표시와 점검을 요청해 주세요.'}}],
                                      'done': {'say': '설정값과 함수로 나눈 미니 도구예요. 파일 이름만 바꾸면 다음 달에도 쓸 수 있어요.',
                                               'code': 'import pandas as pd\n'
                                                       'import matplotlib.pyplot as plt\n'
                                                       '\n'
                                                       '# ===== 설정값 =====\n'
                                                       "FILE = '{VDIR}/m10_week.csv'\n"
                                                       'MIN_SAMPLES = 30\n'
                                                       'DAYS = 7\n'
                                                       "CONDS = ['day', 'night', 'fog']\n"
                                                       '\n'
                                                       'def compare(df):\n'
                                                       '    rows = []\n'
                                                       "    for p in ['before', 'after']:\n"
                                                       '        for c in CONDS:\n'
                                                       "            s = df[(df['period'] == p) & (df['condition'] == c)]\n"
                                                       "            t = s[s['is_target'] == 1]\n"
                                                       "            rows.append({'기간': p, '조건': c, '대상 수': len(t), '재현율(%)': "
                                                       "round(t['alarmed'].mean() * 100, 1) if len(t) else '자료 없음',\n"
                                                       "                         '하루 불필요 경보': round(((s['is_target'] == 0) & (s['alarmed'] == "
                                                       '1)).sum() / DAYS, 1),\n'
                                                       "                         '평균 처리(분)': round(s['handle_min'].mean(), 1), '비고': '표본 적음' if "
                                                       "len(t) < MIN_SAMPLES else ''})\n"
                                                       '    return pd.DataFrame(rows)\n'
                                                       '\n'
                                                       'def make_report(t, n):\n'
                                                       "    day = t[t['조건'] == 'day'].set_index('기간')\n"
                                                       "    low = sorted(set(t[t['비고'] != '']['조건']))\n"
                                                       '    return (f"<h2>시설 방호 브리핑</h2><p><b>결론:</b> 주간 불필요 경보는 하루 {day.loc[\'before\', \'하루 불필요 '
                                                       '경보\']}건에서 {day.loc[\'after\', \'하루 불필요 경보\']}건으로 줄었어요. 재현율 변화는 조건별 표로 함께 봐야 해요.</p>"\n'
                                                       "            + f'<p>가상 기록 {n}건 기준이에요.</p>' + t.to_html(index=False)\n"
                                                       '            + f"<h3>확인된 것과 확인되지 않은 것</h3><p>확인: 표의 조건별 수치. 확인되지 않음: 표본이 적은 구간({\', '
                                                       '\'.join(low) if low else \'없음\'})의 변화, 기록 기간 밖의 상황.</p>"\n'
                                                       "            + '<h3>남은 위험과 다음 조치</h3><ul><li>표본 적은 구간은 자료를 더 모아 다시 비교해요.</li><li>재현율이 떨어진 조건은 "
                                                       "임계값을 다시 검토해요.</li></ul>'\n"
                                                       "            + '<p><i>AI 산출물이에요. 담당자가 확인한 뒤에 보고해요.</i></p>')\n"
                                                       '\n'
                                                       'def lint_report(html):\n'
                                                       "    need = {'표본 수': '건 기준', '남은 위험': '남은 위험', '확인되지 않은 것': '확인되지 않', 'AI 산출물 표기': 'AI 산출물'}\n"
                                                       '    return [k for k, v in need.items() if v not in html]\n'
                                                       '\n'
                                                       'df = pd.read_csv(FILE)\n'
                                                       'table = compare(df)\n'
                                                       'display(table)\n'
                                                       '\n'
                                                       "plot = table.pivot(index='조건', columns='기간', values='하루 불필요 경보').loc[CONDS][['before', "
                                                       "'after']]\n"
                                                       "ax = plot.plot(kind='bar', figsize=(5, 2.8), rot=0)\n"
                                                       "ax.set_xlabel('condition')\n"
                                                       "ax.set_ylabel('nuisance per day')\n"
                                                       '\n'
                                                       'report = make_report(table, len(df))\n'
                                                       'missing = lint_report(report)\n'
                                                       "print('자동 점검:', '빠진 항목 없음' if not missing else missing)",
                                               'lesson': '결론을 정해 주지 않고 조건을 요구하니, 불필요 경보가 줄어든 만큼 어디서 재현율이 달라졌는지 함께 드러났어요. 자동 점검이 있어 빠뜨림도 줄어요.'},
                                      'on_ok': [['dm_lim', 'lim', '{call}, 조건별로 나오니 무엇을 조심할지 보이네요. 이제 팀 의견을 볼까요?']]}]},
                         {'short': '피드백과 점검',
                          'title': '팀 의견과 올리기 전 점검',
                          'lead': '브리핑을 올리기 전에 **사람이 마지막으로 확인**해요.',
                          'blocks': [{'id': 's2_decide',
                                      'type': 'decide',
                                      'title': '브리핑 의견 판단하기',
                                      'text': '브리핑을 받은 사람들의 의견이에요. 받아들일지 정하고 이유를 적어 주세요.',
                                      'reason_min': 8,
                                      'items': [{'id': 'a',
                                                 'who': 'seo',
                                                 'text': '‘재조정 완료, 문제없음’이라고 단정해 주세요. 읽는 사람이 헷갈려요.',
                                                 'answer': 'reject',
                                                 'hint': '표의 수치가 단정을 뒷받침하나요?',
                                                 'explain': '근거 없이 단정하면 확신이 부풀어요. 쉬운 말로 쓰되 남은 위험은 남겨요.',
                                                 'reason': '수치가 단정을 뒷받침하지 않아 쉬운 말로 쓰되 남은 위험을 남겨요.'},
                                                {'id': 'b',
                                                 'who': 'jo',
                                                 'text': '맨 위에 결론 한 줄과 제가 결정할 일을 두고, 표는 그 아래에 놓아 주세요.',
                                                 'answer': 'accept',
                                                 'hint': '내용을 바꾸는 요청인가요, 순서를 바꾸는 요청인가요?',
                                                 'explain': '사실을 해치지 않고 읽기 쉬워져요. 프롬프트에 순서만 바꿔 다시 돌리면 돼요.',
                                                 'reason': '사실은 그대로 두고 읽기 쉬워져 프롬프트의 순서만 바꾸면 돼요.'}]},
                                     {'id': 's2_multi',
                                      'type': 'multi',
                                      'title': '올리기 전에 확인할 것',
                                      'text': 'AI가 만든 브리핑을 올리기 전에 꼭 확인할 것을 모두 골라 주세요.',
                                      'items': [{'text': '표의 숫자가 파일에서 다시 계산한 값과 같은지 한두 개 대조한다.',
                                                 'correct': True,
                                                 'explain': '한두 칸만 대조해도 큰 실수를 잡을 수 있어요.',
                                                 'hint': '숫자가 틀리면 어떻게 될까요?'},
                                                {'text': '수치마다 데이터와 표본 수, 조건이 적혀 있는지 본다.',
                                                 'correct': True,
                                                 'explain': '조건이 없으면 숫자가 오해되기 쉬워요.',
                                                 'hint': '표본이 적은 구간을 떠올려 보세요.'},
                                                {'text': '문장이 매끄러우면 내용도 맞으니 건너뛴다.',
                                                 'correct': False,
                                                 'explain': '매끄러운 문장과 맞는 내용은 별개예요.',
                                                 'hint': '표현이 매끄러운 것과 사실이 맞는 것은 같을까요?'},
                                                {'text': '확인되지 않은 내용이 확인된 것처럼 적히지 않았는지 본다.',
                                                 'correct': True,
                                                 'explain': '해외 사례에서 보았듯 나눠 적는 것이 브리핑의 기본이에요.',
                                                 'hint': '해외 사례에서 무엇을 나눠 적어야 했나요?'},
                                                {'text': 'AI가 만든 문서이니 작성 책임은 AI에게 있다고 적는다.',
                                                 'correct': False,
                                                 'explain': '문서의 책임은 사람에게 있어요. AI는 도구예요.',
                                                 'hint': '브리핑을 올리는 사람은 누구인가요?'}]}]},
                         {'short': '프롬프트 모음',
                          'title': '나만의 프롬프트 틀 만들기',
                          'lead': '이번 프로젝트의 **가장 큰 산출물**이에요. 다음 업무에서 바로 쓸 나만의 틀을 남겨요.',
                          'blocks': [{'id': 's3_write',
                                      'type': 'write',
                                      'title': '내 프롬프트 틀 쓰기',
                                      'prompt': '앞으로 자주 쓸 **프롬프트 틀**을 써 보세요. 파일 이름이나 숫자처럼 바뀌는 자리는 [대괄호]로 비워 두세요. 네 칸에 더해 검증, 범위, 민감 정보 내용도 넣어 보세요.',
                                      'example': '[업무 맥락]을 위해 [무엇]을 확인하려고 해. 값은 보여 주지 않을게. 파일은 [파일 이름]이고 열은 [열 이름과 값의 종류]야. [계산 규칙과 기준값]으로 계산해 줘. '
                                                 '[예외]는 [처리 방법]으로 해 줘. 결과는 [표의 모양]으로 보여 주고, 다시 셀 수 있는 숫자와 표본 수를 함께 적어 줘. 먼저 [작은 예시]로 확인한 뒤 전체에 적용해 '
                                                 '줘. 고치는 건 [범위]만 하고, 원본 파일은 건드리지 마. 문서는 [독자]가 읽을 [순서]로 만들고 없는 내용은 ‘자료 없음’으로 적어 줘. AI 산출물이라는 표시도 넣어 '
                                                 '줘.',
                                      'min': 40,
                                      'checklist': ['맥락, 입력, 할 일, 출력 형식이 있어요',
                                                    '숫자 규칙과 예외 자리가 있어요',
                                                    '근거 수치와 검증을 요청했어요',
                                                    '수정 범위와 민감 정보 내용이 있어요',
                                                    '다음 업무에 바로 쓸 수 있어요'],
                                      'model_label': '📝 모범 프롬프트',
                                      'placeholder': '꼭 입력해 주세요 · 내 말로 프롬프트를 써 보세요',
                                      'explain': '잘 통한 프롬프트는 저장해 두세요. 쓸수록 나에게 맞는 틀이 돼요.'}]}],
               'outputs': [{'id': 'm10_briefing',
                            'title': 'M10 사후검토 비교와 시설 방호 브리핑',
                            'sections': [{'h': '1. 전후 비교 도구와 브리핑', 'blocks': ['s1_vibe']},
                                         {'h': '2. 팀 의견 판단', 'blocks': ['s2_decide']},
                                         {'h': '3. 올리기 전 점검', 'blocks': ['s2_multi']},
                                         {'h': '4. 나만의 프롬프트 틀', 'blocks': ['s3_write']}]}]}]}   # 프로젝트·미션 명세 (만드는 도구가 넣음). 정답·힌트도 여기에 있어 화면(브라우저)으로는 보내지 않음
_P = SPEC["project"]
PROJECT = {"title": _P["title"], "goal": _P["goal"], "stages": _P["stages"]}
TEAM_NAME = _P["team_name"]            # 학습자가 합류하는 가상 조직 이름 (예: 드론분석 TF)
MY_ROLE = _P["my_role"]                # 학습자의 역할 (예: 신입 분석관)
MISSION_INFO = {m["id"]: (m["no"], m["title"], m["short"], m["kind"], m["outline"]) for m in SPEC["missions"]}   # (번호, 제목, 짧은 이름, 형태, 소개)
IMPLEMENTED = [m["id"] for m in SPEC["missions"] if m.get("steps")]

# ---------------------------------------------------------------
# 학습자 호칭: 모두 '전우님' (개인 정보를 묻지 않음. 군 체계를 추정할 정보도 남지 않음)
# ---------------------------------------------------------------
CALL_NAME = _P.get("call_default") or "전우님"


def call_name():
    """학습자를 부르는 말."""
    return CALL_NAME


def me_label():
    return CALL_NAME


MEMBERS = {k: tuple(v) for k, v in _P["members"].items()}   # 아이디: (이름, 소속, 역할, 아바타 색)
CHANNELS = dict(_P["channels"])
ALWAYS_CHANNELS = tuple(_P["always_channels"])

# ---------------------------------------------------------------
# 미션 등록부 (미션 파일들이 채움)
# ---------------------------------------------------------------
BRIEFINGS = {}      # 미션 소개
VIEWS = {}          # 화면에 보낼 미션별 상태
RESETTERS = {}      # 미션별 상태 초기화
STARTERS = {}       # 미션 시작 때 오는 메시지
CALLBACKS = {}      # 화면 버튼 → 파이썬 함수
CODE_TOPICS = {}    # '본 과정의 코드 보기'에 보여 줄 함수

# ---------------------------------------------------------------
# 진행 상태
# ---------------------------------------------------------------
S = {}


def _reset_state():
    S.clear()
    S.update(msgs=[], minute=0, files=[],
             progress={m: {"started": False, "step": 0, "done": False, "issued": False} for m in MISSION_INFO})
    for reset in RESETTERS.values():
        reset()
    _welcome()
    if FILL_ANSWERS:                 # 선생님용: 빈 값 없이 모든 활동을 정답으로 채우고 산출물까지 만들어 둠
        _prefill_all()


def _welcome():
    for ch, who, text in _P["welcome"]:
        post(ch, who, text)


def post(ch, who, text):
    """메신저에 메시지 남기기. {call}은 학습자를 부르는 말(전우님)로 바뀝니다."""
    t = 9 * 60 + S["minute"]
    S["minute"] += 2
    S["msgs"].append({"id": len(S["msgs"]), "ch": ch, "who": who, "time": f"{t // 60:02d}:{t % 60:02d}", "text": text})


def _fmt(text):
    return text.replace("{call}", call_name())


def _advance(mid, step):
    p = S["progress"][mid]
    p["step"] = max(p["step"], step)


def _status(mid):
    """잠금 없음: 준비된 미션은 언제나 열려 있음. 필수 활동을 다 하면 done."""
    if mid not in IMPLEMENTED:
        return "soon"
    return "done" if S["progress"][mid]["done"] else "open"


def overall_progress():
    """전체 진행률: 미션마다 끝낸 필수 활동의 비율을 평균."""
    units = sum(_mission_fraction(mid) for mid in MISSION_INFO if mid in BRIEFINGS)
    done = sum(S["progress"][m]["done"] for m in MISSION_INFO)
    return {"pct": round(100 * units / len(MISSION_INFO)), "done": done, "total": len(MISSION_INFO)}


def _add_file(fid, title, mission, doc, blanks=0):
    S["files"] = [f for f in S["files"] if f["id"] != fid]
    S["files"].append({"id": fid, "name": f"{fid}.html", "title": title, "mission": mission, "html": doc, "blanks": blanks})


DOC_CSS = ("body{font-family:sans-serif;max-width:860px;margin:24px auto;color:#111827;background:#fff;line-height:1.6;padding:0 16px}"
           "h1{font-size:22px;border-bottom:2px solid #111827;padding-bottom:6px}h2{font-size:16px;margin-top:22px}"
           "table{border-collapse:collapse;margin:8px 0}td,th{border:1px solid #c8ced6;padding:5px 10px;font-size:13px;text-align:left;vertical-align:top;color:#111827}"
           "th{background:#eef2f7}.meta{color:#111827;font-size:13px;background:#f3f6fa;padding:6px 10px;border-radius:6px}"
           ".note{color:#111827;font-size:13px}.ok{color:#146c2e;font-weight:bold}.bad{color:#b3261e;font-weight:bold}"
           ".cards{display:grid;grid-template-columns:repeat(auto-fill,minmax(240px,1fr));gap:12px}"
           ".card{border:1px solid #c8ced6;border-radius:8px;padding:10px}.card img{width:48%;border-radius:4px}"
           ".imgs img{max-width:100%;border:1px solid #c8ced6;border-radius:6px}.chg{background:#fff4ce}"
           "pre{background:#f3f6fa;padding:8px 10px;border-radius:6px;white-space:pre-wrap}"
           ".blank{display:inline-block;min-width:180px;height:1.1em;border-bottom:1px solid #6b7280}"
           ".blankbox{min-height:54px;border:1px dashed #6b7280;border-radius:6px;margin:6px 0}"
           "td.empty{min-width:120px;height:1.6em}img{max-width:100%}")


def _doc(title, body):
    return (f"<!doctype html><html><head><meta charset='utf-8'><title>{html.escape(title)}</title><style>{DOC_CSS}</style></head>"
            f"<body><h1>{html.escape(title)}</h1><p class='meta'>프로젝트: {html.escape(PROJECT['title'])}<br>"
            f"{html.escape(TEAM_NAME)} · 작성일: {date.today():%Y-%m-%d}</p>{body}</body></html>")


# ---------------------------------------------------------------
# 본 과정의 코드 보기
# ---------------------------------------------------------------
def code_topic(key, title, desc, funcs, marks=()):
    """화면의 한 단계에서 쓰이는 함수 목록을 등록. marks에 든 글자가 있는 줄은 강조 표시됩니다."""
    CODE_TOPICS[key] = {"title": title, "desc": desc, "funcs": list(funcs), "marks": list(marks)}


def call_str(name, *args, **kw):
    """함수 호출을 코드 한 줄로 적기 (예: simulate('m1', level=0.5))."""
    parts = [repr(a) for a in args] + [f"{k}={v!r}" for k, v in kw.items()]
    return f"{name}({', '.join(parts)})"


def app_code(topic):
    t = CODE_TOPICS.get(topic)
    if not t:
        return JSON({"ok": False, "blocks": []})
    blocks = []
    for name in t["funcs"]:
        fn = globals().get(name)
        try:
            src = textwrap.dedent(inspect.getsource(fn)).rstrip()
            doc = (inspect.getdoc(fn) or "").split("\n")[0]
        except Exception:
            src, doc = f"# {name}: 코드를 불러오지 못했어요.", ""
        lines = src.splitlines()
        hl = [i for i, line in enumerate(lines) if any(m in line for m in t["marks"])]
        blocks.append({"name": name, "doc": doc, "code": src, "hl": hl})
    return JSON({"ok": True, "topic": topic, "title": t["title"], "desc": t["desc"], "blocks": blocks})


# ---------------------------------------------------------------
# 화면에 보낼 상태
# ---------------------------------------------------------------
def _mission_meta(mid):
    no, title, short, kind, outline = MISSION_INFO[mid]
    stage = next(s["no"] for s in PROJECT["stages"] if mid in s["missions"])
    meta = {"id": mid, "no": no, "stage": stage, "title": title, "short": short, "kind": kind,
            "outline": outline, "status": _status(mid), "progress": S["progress"][mid]}
    if mid in BRIEFINGS and mid in IMPLEMENTED:
        meta["brief"] = BRIEFINGS[mid]
    return meta


def app_state():
    members = {}
    for k, (n, o, r, c) in MEMBERS.items():
        members[k] = {"name": n, "org": o, "role": r, "color": c, "ext": k != "me" and o != TEAM_NAME}
    used = {m["ch"] for m in S["msgs"]}
    return {
        "call": call_name(),
        "project": {"title": PROJECT["title"], "goal": PROJECT["goal"],
                    "stages": [{"no": s["no"], "name": s["name"], "subtitle": s.get("subtitle", ""), "missions": s["missions"]} for s in PROJECT["stages"]]},
        "missions": {mid: _mission_meta(mid) for mid in MISSION_INFO},
        "views": {mid: fn() for mid, fn in VIEWS.items()},
        "members": members,
        "channels": {k: v for k, v in CHANNELS.items() if k in ALWAYS_CHANNELS or k in used},
        "msgs": [{**m, "text": m["text"] if m["who"] == "me" else _fmt(m["text"])} for m in S["msgs"]],
        "files": [{k: f.get(k) for k in ("id", "name", "title", "mission", "blanks")} for f in S["files"]],
        "overall": overall_progress(), "boot": BOOT_ID,
        "teacher": bool(FILL_ANSWERS),        # 선생님용: 값이 모두 채워진 상태 (화면에 표시만 함)
        "assets": assets_status(),            # 참고자료(사진)가 필요한 프로젝트면 폴더와 빠진 파일을 알려 줌
    }


def _resp(**kw):
    return JSON({"ok": True, **kw, "app": app_state()})


def app_get():
    return JSON({"ok": True, "app": app_state()})


def app_start(mid):
    if _status(mid) == "soon":
        return _resp(ok=False, msg="아직 준비 중인 미션이에요.")
    _ensure_started(mid)
    return _resp(ok=True)


def _ensure_started(mid):
    """미션을 처음 열 때 한 번: 시작 메시지를 보냄."""
    p = S["progress"][mid]
    if not p["started"]:
        p["started"] = True
        _advance(mid, 1)
        if mid in STARTERS:
            STARTERS[mid]()


def app_send(ch, text):
    text = (text or "").strip()
    if text:
        post(ch, "me", text)
    return _resp(ok=True)


def app_file(fid):
    f = next((f for f in S["files"] if f["id"] == fid), None)
    return JSON({"ok": bool(f), "html": f["html"] if f else "", "name": f["name"] if f else ""})


def app_reset():
    _reset_state()
    return _resp(ok=True, msg="처음 상태로 돌아갔어요.")


CALLBACKS.update({"app.state": app_get, "app.start": app_start, "app.send": app_send,
                  "app.file": app_file, "app.reset": app_reset, "app.code": app_code})
code_topic("app.chat", "메신저는 이렇게 동작해요", "팀원 메시지는 미션 진행에 맞춰 파이썬 함수 post()가 남깁니다. "
           "화면의 버튼을 누르면 google.colab의 invokeFunction이 파이썬 함수를 부르고, 결과(JSON)로 화면을 다시 그려요.",
           ["post", "app_send"], marks=["S[\"msgs\"].append", "post(ch"])
code_topic("app.mission", "미션 시작은 이렇게 동작해요", "[미션 시작]을 누르면 app_start()가 진행 상태를 바꾸고, "
           "미션마다 등록된 시작 메시지(STARTERS)를 보냅니다.", ["app_start", "_ensure_started"], marks=["STARTERS"])
code_topic("app.files", "산출물은 이렇게 만들어져요", "산출물을 받으면 입력한 내용으로 HTML 문서를 만들어 파일 목록에 저장합니다. "
           "하지 않은 활동은 빈칸으로 남아요.", ["_doc", "_add_file"], marks=["S[\"files\"]"])


# ---------------------------------------------------------------
# 뒤에서 도는 작업 (학습 · 계산 · 내려받기): 화면이 1.5초마다 app.job으로 진행을 물어봄
# 오래 걸리는 작업 중 다른 화면을 열면 Colab 런타임이 끊기는 일이 있어서, 명세의 stay_jobs에 적은 작업은 화면 이동을 잠금
# ---------------------------------------------------------------
JOBS = {}


def start_job(name, fn, *args):
    """작업을 뒤에서(스레드) 실행. 이미 같은 작업이 돌고 있으면 False."""
    if JOBS.get(name, {}).get("state") == "running":
        return False
    job = {"state": "running", "pct": 1, "msg": "시작하는 중...", "t0": time.time(), "error": None}
    JOBS[name] = job

    def run():
        try:
            fn(job, *args)
            job.update(state="done", pct=100)
        except Exception as e:
            job.update(state="error", error=f"{type(e).__name__}: {e}", trace=traceback.format_exc()[-2500:])
    threading.Thread(target=run, daemon=True).start()
    return True


def _say(job, pct, msg):
    job["pct"], job["msg"] = max(job.get("pct", 0), int(pct)), msg


def app_job(name):
    """화면이 2초마다 묻는 작업 진행 상황. 끝나면 새 화면 상태를 함께 보냄."""
    j = JOBS.get(name)
    if not j:
        return JSON({"ok": True, "state": "none"})
    out = {"ok": True, "state": j["state"], "pct": j["pct"], "msg": j["msg"], "sec": int(time.time() - j["t0"]), "error": j.get("error"),
           "data": j.get("data")}
    if j["state"] != "running":
        out["app"] = app_state()
    return JSON(out)


# ---------------------------------------------------------------
# 그림 도우미: 화면에 보낼 영상을 데이터 주소로
# ---------------------------------------------------------------
def to_data_url(img, fmt="JPEG"):
    """PIL 영상 → 데이터 주소 (JPEG는 RGB로 바꿔 저장)."""
    buf = io.BytesIO()
    (img.convert("RGB") if fmt.upper() == "JPEG" else img).save(buf, format=fmt, quality=90)
    return f"data:image/{fmt.lower()};base64," + base64.b64encode(buf.getvalue()).decode()


def _img_url(img, width=None):
    """화면에 보낼 영상: 폭을 맞춰(픽셀이 보이게 NEAREST) JPEG 데이터 주소로."""
    if width and img.size[0] != width:
        from PIL import Image
        img = img.resize((width, max(1, round(width * img.size[1] / img.size[0]))), Image.NEAREST)
    return to_data_url(img)


CALLBACKS["app.job"] = app_job
# ===== 참고자료(사진): 인터넷에서 받지 않음. 제작자가 따로 준 사진을 학습자가 폴더에 넣고, 블록의 "asset:번호"로 씀 =====
# 명세의 project.assets: {"folder"(사진을 넣을 폴더 이름), "items"(번호·파일·제목·저작자·라이선스)}
# 정답은 여기에 없음. 사진은 노트북에 담지 않고 빌드 도구가 <코드>_images 폴더·zip으로 따로 내줌.
import shutil, zipfile

ASSET_CFG = _P.get("assets") or {}
ASSET_ITEMS = {it["id"]: it for it in ASSET_CFG.get("items", [])}
ASSET_STATE = {"ok": not ASSET_ITEMS, "msg": ""}
_ASSET_URL_CACHE = {}


def asset_folder():
    """사진을 넣는 폴더: Colab은 /content/<folder>, 점검 도구는 PBL_ASSET_LOCAL(자료 폴더)이나 PBL_ASSET_ROOT(임시 폴더)."""
    local = os.environ.get("PBL_ASSET_LOCAL")
    if local:
        return local
    root = os.environ.get("PBL_ASSET_ROOT") or ("/content" if os.path.isdir("/content") else os.getcwd())
    return os.path.join(root, ASSET_CFG.get("folder") or "pbl_photos")


def _asset_missing():
    folder = asset_folder()
    return [it["file"] for it in ASSET_ITEMS.values() if not os.path.exists(os.path.join(folder, it["file"]))]


def _extract_zips(folder):
    """폴더에 넣은 zip을 풀어 사진을 폴더 바로 아래로 모음 (zip 안에 폴더가 한 겹 있어도 됨)."""
    for name in os.listdir(folder):
        if not name.lower().endswith(".zip"):
            continue
        tmp = os.path.join(folder, "_unzip_tmp")
        shutil.rmtree(tmp, ignore_errors=True)
        with zipfile.ZipFile(os.path.join(folder, name)) as z:
            z.extractall(tmp)
        for dirpath, _, files in os.walk(tmp):
            for f in files:
                shutil.move(os.path.join(dirpath, f), os.path.join(folder, f))
        shutil.rmtree(tmp, ignore_errors=True)


def assets_prepare():
    """사진 확인: 폴더를 만들어 두고, 넣어 둔 zip이 있으면 풀고, 빠진 사진을 ASSET_STATE에 남김. 다시 불러도 안전함."""
    if not ASSET_ITEMS:
        return True
    folder = asset_folder()
    os.makedirs(folder, exist_ok=True)
    try:
        _extract_zips(folder)
    except Exception as e:
        ASSET_STATE.update(ok=False, msg=f"zip을 풀지 못했어요({type(e).__name__}). 사진 파일을 그대로 넣어 주세요.")
        return False
    missing = _asset_missing()
    _ASSET_URL_CACHE.clear()
    if missing:
        ASSET_STATE.update(ok=False, msg=f"참고자료(사진) {len(missing)}개가 아직 없어요. 왼쪽 📁 파일 패널에서 {folder} 폴더에 넣어 주세요.")
        return False
    ASSET_STATE.update(ok=True, msg="참고자료(사진)를 확인했어요.")
    return True


def assets_status():
    """화면에 보낼 사진 상태: 필요 없는 프로젝트면 None."""
    if not ASSET_ITEMS:
        return None
    return {"ok": ASSET_STATE["ok"], "folder": asset_folder(), "missing": _asset_missing(), "total": len(ASSET_ITEMS), "msg": ASSET_STATE["msg"]}


def app_assets_check():
    """화면의 [다시 확인] 버튼: 사진을 넣은 뒤 다시 찾음."""
    ok = assets_prepare()
    return _resp(ok=ok, msg=ASSET_STATE["msg"])


def asset_path(aid):
    """참고자료 파일 경로 (없으면 None). 직접 만든 활동에서 원본 파일을 읽을 때 씀 (예: 출처 정보 확인)."""
    it = ASSET_ITEMS.get(str(aid))
    if not it:
        return None
    p = os.path.join(asset_folder(), it["file"])
    return p if os.path.exists(p) else None


def asset_data_url(aid, width=960):
    """화면에 보낼 이미지: 폭을 맞춘 JPEG 데이터 주소 (한 번 만들면 캐시). PIL이 없으면 파일을 그대로 보냄."""
    key = (str(aid), int(width))
    if key not in _ASSET_URL_CACHE:
        path = asset_path(aid)
        try:
            from PIL import Image
            im = Image.open(path).convert("RGB")
            if im.size[0] > width:
                im = im.resize((width, max(1, round(width * im.size[1] / im.size[0]))), Image.LANCZOS)
            _ASSET_URL_CACHE[key] = to_data_url(im)
        except ImportError:
            mime = "png" if path.lower().endswith(".png") else "jpeg"
            with open(path, "rb") as f:
                _ASSET_URL_CACHE[key] = f"data:image/{mime};base64," + base64.b64encode(f.read()).decode()
    return _ASSET_URL_CACHE[key]


def app_asset(aid, width=960):
    """화면의 이미지 요청 ("asset:번호"). 화면 상태 전체는 보내지 않음 (가볍게)."""
    it = ASSET_ITEMS.get(str(aid))
    if not it:
        return JSON({"ok": False, "msg": f"없는 참고자료예요: {aid}"})
    if not asset_path(aid):
        return JSON({"ok": False, "missing": True, "file": it["file"], "folder": asset_folder(),
                     "msg": f"사진을 찾지 못했어요: {os.path.join(asset_folder(), it['file'])}"})
    try:
        return JSON({"ok": True, "url": asset_data_url(aid, max(240, min(int(width), 2400))), "caption": ""})
    except Exception as e:
        return JSON({"ok": False, "msg": f"이미지를 열지 못했어요: {type(e).__name__}"})


def _asset_refs(obj, out=None):
    """명세 안에서 "asset:번호"를 모두 찾음 (산출물의 출처 목록용)."""
    out = [] if out is None else out
    if isinstance(obj, str) and obj.startswith("asset:"):
        if obj[6:] not in out:
            out.append(obj[6:])
    elif isinstance(obj, (list, tuple)):
        for x in obj:
            _asset_refs(x, out)
    elif isinstance(obj, dict):
        for x in obj.values():
            _asset_refs(x, out)
    return out


def asset_credits_html(ids):
    """산출물 문서에 붙는 '사용한 자료와 출처' 표 (주소 없이 이름만: 답이 새지 않게, 링크를 달지 않게)."""
    rows = []
    for aid in ids:
        it = ASSET_ITEMS.get(aid)
        if not it:
            continue
        rows.append(f"<tr><td>{html.escape(aid)}</td><td>{html.escape(it.get('title', ''))}</td><td>{html.escape(it.get('credit', ''))}</td>"
                    f"<td>{html.escape(it.get('license', ''))}</td><td>{html.escape(it.get('changes', ''))}</td></tr>")
    if not rows:
        return ""
    return ("<h2>사용한 자료와 출처</h2><table><tr><th>번호</th><th>자료</th><th>저작자·제작</th><th>라이선스</th><th>변경</th></tr>"
            + "".join(rows) + "</table>")


CALLBACKS.update({"app.asset": app_asset, "app.assets_check": app_assets_check})
_setup(30, 1, "메신저 · 진행 상태 · 가이드북 준비 완료")

# ===== 미션 단계 엔진: 명세(SPEC)의 블록으로 미션을 진행하고, 채점·힌트·산출물을 만듦 =====
# 블록 종류: info(읽기) · quiz(확인 질문) · multi(여러 개 고르기) · match(짝짓기) · order(순서 맞추기)
#            classify(사례 분류) · write(글쓰기) · decide(받아들이기/거절 결정) · vibe(바이브 코딩 실습) · custom(직접 만든 활동)
# 원칙 (시행착오에서 정한 것)
#  - 정답·풀이·힌트는 파이썬에만 둠. 화면에는 학습자가 맞힌 뒤에만 풀이를 보냄
#  - 틀리면 힌트를 한 단계씩 자세하게. 맞혀도 화면이 저절로 넘어가지 않음 (학습자가 [다음 단계]를 누름)
#  - 처음부터 차례로 하는 것이 원칙이지만, 뒷부분을 미리 보려는 사람을 위해 잠그지 않음 (2026-11-05 피드백)
#  - 산출물은 언제든 받을 수 있음: 안 한 활동이 있으면 먼저 알려 주고 물어본 뒤, 그래도 받으면 그 부분은 빈칸으로 만듦
#  - 선생님용(FILL_ANSWERS)은 모든 활동의 값이 채워진 상태로 시작함
#  - 글쓰기(write)는 정답을 판정하지 않음: 최소 글자 수만 보고 제출을 받은 뒤 모범답안(example)을 보여 줌
CUSTOM_VIEWS = {}     # 미션 아이디 → 화면에 더 보낼 상태를 돌려주는 함수 (직접 만든 활동용)
CUSTOM_RESET = {}     # 미션 아이디 → 처음 상태를 만드는 함수
CUSTOM_DOCS = {}      # (미션, 블록) → 산출물 문서에 넣을 HTML을 돌려주는 함수
AUTOPLAY = {}         # (미션, 블록) → 점검 도구가 이 활동을 끝까지 해 보는 함수 (직접 만든 활동용)
DECIDE_OPTS = [["accept", "받아들임"], ["reject", "거절하고 대안 제시"]]
MSPEC = {m["id"]: m for m in SPEC["missions"]}
BSPEC = {}            # (미션, 블록) → (단계 번호, 블록 명세)
for _m in SPEC["missions"]:
    for _si, _st in enumerate(_m.get("steps", []), 1):
        for _b in _st.get("blocks", []):
            BSPEC[(_m["id"], _b["id"])] = (_si, _b)


def _required(b):
    """필수 블록인가: 읽기(info)는 confirm이 있을 때만, 나머지는 required가 False가 아니면 필수."""
    if b["type"] == "info":
        return bool(b.get("confirm"))
    return b.get("required", True)


def _bst(mid, bid):
    return S[mid]["blk"].setdefault(bid, {"ok": False, "tries": 0, "value": None, "item_tries": {}})


def _step_blocks(mid, si):
    st = MSPEC[mid]["steps"][si - 1]
    return st.get("blocks", [])


def _step_done(mid, si):
    """단계 완료: 필수 블록을 모두 마침 (블록 없이 직접 만든 화면(page)은 그 코드가 step_ok로 알림)."""
    st = MSPEC[mid]["steps"][si - 1]
    if st.get("page"):
        return bool(S[mid].get("pages", {}).get(si))
    return all(_bst(mid, b["id"])["ok"] for b in _step_blocks(mid, si) if _required(b))


def _todo_list(mid):
    """아직 안 한 필수 활동: [{"step": 단계 번호, "short": 단계 이름, "blocks": [활동 이름…]}]."""
    out = []
    for si, st in enumerate(MSPEC[mid]["steps"], 1):
        if st.get("page"):
            if not _step_done(mid, si):
                out.append({"step": si, "short": st["short"], "blocks": [st.get("title") or st["short"]]})
            continue
        left = [b.get("title") or BLOCK_NAMES.get(b["type"], "활동") for b in st.get("blocks", []) if _required(b) and not _bst(mid, b["id"])["ok"]]
        if left:
            out.append({"step": si, "short": st["short"], "blocks": left})
    return out


def _mission_fraction(mid):
    """끝낸 필수 활동의 비율 (0~1)."""
    done = total = 0
    for si, st in enumerate(MSPEC[mid]["steps"], 1):
        if st.get("page"):
            total += 1
            done += _step_done(mid, si)
            continue
        for b in st.get("blocks", []):
            if _required(b):
                total += 1
                done += _bst(mid, b["id"])["ok"]
    return done / total if total else 0.0


def _update_done(mid):
    """필수 활동을 다 하면 미션 완료 표시."""
    S["progress"][mid]["done"] = not _todo_list(mid)


def step_ok(mid, si):
    """직접 만든 화면(page) 단계를 마쳤다고 알림."""
    S[mid].setdefault("pages", {})[si] = True
    _advance(mid, si + 1)
    _update_done(mid)


def block_ok(mid, bid, value=None):
    """블록을 마쳤다고 기록: 처음 마쳤을 때 팀원 메시지(on_ok)를 보내고, 진행 상태를 고침."""
    si, b = BSPEC[(mid, bid)]
    st = _bst(mid, bid)
    first = not st["ok"]
    st["ok"] = True
    if value is not None:
        st["value"] = value
    if first:
        for ch, who, text in b.get("on_ok", []):
            post(ch, who, text)
    if _step_done(mid, si):
        _advance(mid, si + 1)
    _update_done(mid)


WRONG_MSG = "❌ 현재 선택을 다시 고민해 보세요."      # 틀린 선택 안내: 화면에서 빨간색으로 보임 (2026-11-05 피드백 "주황색 말고 빨간색으로")


def _hint(spec, tries, key="hints"):
    hs = spec.get(key) or ["다시 생각해 보세요."]
    k = max(1, min(tries, len(hs)))
    return f"{WRONG_MSG} 힌트 {k}/{len(hs)}: {hs[k - 1]}"


# ---- 블록마다 채점: (맞음?, 저장할 값, 화면에 보낼 것) ----
def _chk_info(mid, b, st, payload):
    return True, True, {"text": "✅ 확인했어요."}


def _chk_quiz(mid, b, st, payload):
    try:
        pick = int(payload)
    except (TypeError, ValueError):
        pick = -1
    if pick == b["answer"]:
        return True, {"pick": pick}, {"text": "✅ 맞아요. " + b["explain"]}
    st["tries"] += 1
    return False, None, {"text": _hint(b, st["tries"])}


def _chk_multi(mid, b, st, payload):
    picked = {int(i) for i in (payload or []) if str(i).lstrip("-").isdigit()}
    res = []
    for i, it in enumerate(b["items"]):
        ok = (i in picked) == bool(it["correct"])
        res.append({"i": i, "ok": ok, "text": it.get("explain", "") if ok else "❌ " + it.get("hint", "다시 생각해 보세요.")})
    wrong = sum(not r["ok"] for r in res)
    if wrong:
        st["tries"] += 1
        return False, None, {"items": res, "text": f"{WRONG_MSG} 다시 볼 항목이 {wrong}개 있어요."}
    return True, {"picked": sorted(picked)}, {"items": res, "text": "✅ 모두 맞게 골랐어요."}


def _chk_match(mid, b, st, payload):
    payload = payload or {}
    res = {p["left"]: str(payload.get(p["left"])) == str(p["right"]) for p in b["pairs"]}
    wrong = [p for p in b["pairs"] if not res[p["left"]]]
    if not wrong:
        return True, {p["left"]: p["right"] for p in b["pairs"]}, {"result": res, "hints": [], "text": "✅ 모두 맞게 짝지었어요."}
    st["tries"] += 1
    n, gen = st["tries"], b.get("hints") or ["틀린 칸을 다시 살펴보세요."]
    hints = [gen[0]]
    if n >= 2:
        hints += [f"‘{p['left']}’: {p['hint']}" for p in wrong if p.get("hint")]
    if n >= 3:
        hints += gen[1:]
    return False, None, {"result": res, "hints": hints, "text": f"{WRONG_MSG} 빨간 칸 {len(wrong)}개가 맞지 않아요."}


def _chk_order(mid, b, st, payload):
    right = [it["id"] for it in b["items"]]
    order = [o for o in (payload or []) if o in right]
    if order == right:
        return True, {"order": order}, {"text": "✅ 순서가 맞아요.", "wrong": None}
    if len(order) != len(right):
        return False, None, {"text": "모든 항목을 한 번씩 놓아 주세요.", "wrong": None}
    st["tries"] += 1
    n = st["tries"]
    i = next(k for k, (a, c) in enumerate(zip(order, right)) if a != c)
    it = b["items"][i]
    text = f"❌ 현재 순서를 다시 고민해 보세요. {i + 1}번째 자리가 맞지 않아요. " + (b.get("hints") or ["앞 단계의 결과가 다음 단계의 재료가 되는지 따져 보세요."])[0]
    if n >= 2 and it.get("hint"):
        text += f" 힌트: {it['hint']}"
    if n >= 3:
        text += f" ({i + 1}번째 자리: ‘{it['text']}’)"
    return False, None, {"text": text, "wrong": i}


def _chk_classify(mid, b, st, payload):
    payload = payload or {}
    it = next((x for x in b["items"] if x["id"] == payload.get("item")), None)
    if not it:
        return False, None, {"text": "항목을 찾지 못했어요.", "item": None}
    names = {c["id"]: c["name"] for c in b["categories"]}
    val = dict(st["value"] or {})
    prim = payload.get("primary")
    contrib = [c for c in (payload.get("contrib") or []) if c in names and c != prim]
    if prim in it["accept"]:
        val[it["id"]] = {"primary": prim, "contrib": contrib}
        miss = [names[c] for c in it.get("contrib", []) if c not in contrib and c != prim]
        text = "✅ " + it["explain"] + (f" (함께 생각해 볼 것: {', '.join(miss)})" if miss else "")
        return all(x["id"] in val for x in b["items"]), val, {"text": text, "item": it["id"], "item_ok": True}
    t = st["item_tries"][it["id"]] = st["item_tries"].get(it["id"], 0) + 1
    return False, val, {"text": _hint(it, t), "item": it["id"], "item_ok": False}


def _chk_write(mid, b, st, payload):
    """글쓰기: 정답을 판정하지 않음. 최소 글자 수를 채우면 제출을 받고 모범답안(example)을 보여 줌.
    (학습자가 뜻은 맞게 써도 규칙(정규식)에는 틀릴 수 있어서, 맞고 틀림을 따지지 않기로 함: 2026-10-19 피드백)"""
    text = (payload or "").strip()
    n = b.get("min", 10)
    if len(text) < n:
        return False, None, {"text": f"✏ {n}자 이상 써 주세요. (지금 {len(text)}자)"}
    return True, text, {"text": "✅ 제출했어요.", "model": b.get("example", ""), "explain": b.get("explain", ""), "checklist": b.get("checklist", [])}


def _chk_decide(mid, b, st, payload):
    payload = payload or {}
    ch, rs = payload.get("choices") or {}, payload.get("reasons") or {}
    opts = [o[0] for o in b.get("options", DECIDE_OPTS)]
    if any(ch.get(it["id"]) not in opts for it in b["items"]):
        return False, None, {"text": "항목마다 하나를 골라 주세요.", "items": []}
    rmin = b.get("reason_min", 10)
    short = [it["id"] for it in b["items"] if len((rs.get(it["id"]) or "").strip()) < rmin]
    if short:
        return False, None, {"text": f"이유를 {rmin}자 이상 적어 주세요. (남은 항목 {len(short)}개)", "items": []}
    wrong = {it["id"] for it in b["items"] if ch[it["id"]] != it["answer"]}
    items = [{"id": it["id"], "ok": it["id"] not in wrong,
              "text": it.get("explain", "") if it["id"] not in wrong else "❌ " + it.get("hint", "다시 생각해 보세요.")} for it in b["items"]]
    if wrong:
        st["tries"] += 1
        return False, None, {"text": f"{WRONG_MSG} 다시 볼 항목이 {len(wrong)}개 있어요.", "items": items}
    return True, {it["id"]: {"choice": ch[it["id"]], "reason": rs[it["id"]].strip()} for it in b["items"]}, {"text": "✅ 모두 알맞게 결정했어요.", "items": items}


CHECKERS = {"info": _chk_info, "quiz": _chk_quiz, "multi": _chk_multi, "match": _chk_match, "order": _chk_order,
            "classify": _chk_classify, "write": _chk_write, "decide": _chk_decide}


def blk_check(mid, bid, payload=None):
    """화면의 [확인] 버튼: 블록 하나를 채점. correct=True면 그 블록을 마친 것 (화면은 저절로 넘어가지 않음)."""
    if (mid, bid) not in BSPEC:
        return _resp(ok=False, msg="없는 활동이에요.")
    si, b = BSPEC[(mid, bid)]
    _ensure_started(mid)                                            # 잠그지 않으므로 미리 열어 본 단계에서도 동작함
    fn = CHECKERS.get(b["type"])
    if not fn:
        return _resp(ok=False, msg="이 활동은 화면의 안내대로 진행해요.")
    st = _bst(mid, bid)
    ok, value, extra = fn(mid, b, st, payload)
    if value is not None:
        st["value"] = value
    if ok:
        block_ok(mid, bid)
    return _resp(ok=True, correct=bool(ok), **extra)


# ---- 산출물 문서 ----
def _esc_br(t):
    return html.escape(str(t)).replace("\n", "<br>")


def _info_html(b):
    out = "".join(f"<p>{_md(p)}</p>" for p in b.get("text", []))
    if b.get("bullets"):
        out += "<ul>" + "".join(f"<li>{_md(x)}</li>" for x in b["bullets"]) + "</ul>"
    if b.get("table"):
        t = b["table"]
        out += "<table><tr>" + "".join(f"<th>{_md(c)}</th>" for c in t["head"]) + "</tr>" + "".join(
            "<tr>" + "".join(f"<td>{_md(c)}</td>" for c in row) + "</tr>" for row in t["rows"]) + "</table>"
    if b.get("sources"):
        out += "<p class='note'>출처: " + " · ".join(f"<a href='{html.escape(u)}'>{html.escape(n)}</a>" for n, u in b["sources"]) + "</p>"
    return out


def _md(t):
    """명세 글에서 **굵게**만 지원 (나머지는 그대로 이스케이프)."""
    return re.sub(r"\*\*(.+?)\*\*", r"<b>\1</b>", html.escape(str(t)))


BLANK = "<span class='blank'></span>"
BLANKBOX = "<div class='blankbox'></div>"
BLOCK_NAMES = {"quiz": "확인 질문", "multi": "여러 개 고르기", "match": "짝짓기", "order": "순서 맞추기", "classify": "분류하기",
               "write": "글쓰기", "decide": "결정하기", "vibe": "바이브 코딩 실습", "custom": "활동", "info": "읽기"}


def _block_html(mid, b):
    """산출물에 들어가는 블록 하나. 하지 않은 활동은 답·풀이 없이 빈칸으로 (정답이 산출물로 새지 않게)."""
    st, t = _bst(mid, b["id"]), b["type"]
    v = st["value"]
    done = st["ok"]
    head = f"<h3>{html.escape(b.get('title') or '')}</h3>" if b.get("title") else ""
    if t == "info":
        return head + _info_html(b)
    if t == "quiz":
        pick = (v or {}).get("pick")
        if not done or pick is None:
            return head + f"<p><b>Q.</b> {_md(b['text'])}</p><p>고른 답: {BLANK}</p>"
        return (head + f"<p><b>Q.</b> {_md(b['text'])}</p><p>고른 답: {_md(b['options'][pick])}</p>"
                f"<p class='note'>풀이: {_md(b['explain'])}</p>")
    if t == "multi":
        if not done:
            return head + f"<p>{_md(b['text'])}</p><p>고른 항목: {BLANK}</p>"
        return head + f"<p>{_md(b['text'])}</p><ul>" + "".join(
            f"<li>{'✅ 해당' if it['correct'] else '⬜ 해당 없음'} · {_md(it['text'])}<br><span class='note'>{_md(it.get('explain', ''))}</span></li>" for it in b["items"]) + "</ul>"
    if t == "match":
        return head + "<table>" + "".join(f"<tr><th>{_md(p['left'])}</th>" + (f"<td>{_md(p.get('right_label', p['right']))}</td>" if done else "<td class='empty'></td>") + "</tr>" for p in b["pairs"]) + "</table>"
    if t == "order":
        if not done:
            return head + "<ol>" + "".join(f"<li>{BLANK}</li>" for _ in b["items"]) + "</ol>"
        return head + "<ol>" + "".join(f"<li><b>{_md(it['text'])}</b>{' (' + _md(it['desc']) + ')' if it.get('desc') else ''}</li>" for it in b["items"]) + "</ol>"
    if t == "classify":
        names = {c["id"]: c["name"] for c in b["categories"]}
        rows = ""
        for it in b["items"]:
            z = (v or {}).get(it["id"])
            if z:
                rows += (f"<tr><td>{_md(it['title'])}</td><td><b>{names.get(z.get('primary'), '')}</b></td>"
                         + ("" if b.get("contrib") is False else f"<td>{', '.join(names[c] for c in z.get('contrib', []))}</td>")
                         + f"<td>{_md(it['explain'])}</td></tr>")
            else:
                rows += (f"<tr><td>{_md(it['title'])}</td><td class='empty'></td>" + ("" if b.get("contrib") is False else "<td class='empty'></td>") + "<td class='empty'></td></tr>")
        return head + (f"<table><tr><th>{_md(b.get('item_label', '사례'))}</th><th>{_md(b.get('primary_label', '직접 원인'))}</th>"
                       + ("" if b.get("contrib") is False else "<th>함께 작용한 것</th>") + f"<th>풀이</th></tr>{rows}</table>")
    if t == "write":
        return head + (f"<p class='note'>{_md(b['prompt'])}</p>" if b.get("prompt") else "") + (f"<p>{_esc_br(v)}</p>" if done and v else BLANKBOX)
    if t == "decide":
        opts = dict(b.get("options", DECIDE_OPTS))
        rows = ""
        for it in b["items"]:
            z = (v or {}).get(it["id"]) if done else None
            rows += f"<tr><td>{_md(it.get('who', ''))}</td><td>{_md(it['text'])}</td>" + (
                f"<td><b>{opts.get(z.get('choice'), '')}</b></td><td>{_esc_br(z.get('reason', ''))}</td>" if z else "<td class='empty'></td><td class='empty'></td>") + "</tr>"
        return head + f"<table><tr><th>누가</th><th>요청</th><th>결정</th><th>이유 · 대안</th></tr>{rows}</table>"
    if t == "vibe":
        return head + _vibe_html(mid, b, st)
    if not done:
        return head + BLANKBOX
    fn = CUSTOM_DOCS.get((mid, b["id"]))
    if fn:
        return head + fn(st)
    return head + f"<pre>{html.escape(json.dumps(v, ensure_ascii=False, indent=1))}</pre>"


def _mission_docs(mid, blanks=0):
    """산출물 문서: 명세의 outputs대로 (없으면 미션 전체를 문서 하나로). blanks는 비어 있는 활동 수."""
    m = MSPEC[mid]
    outs = m.get("outputs") or [{"id": f"{mid}_report", "title": f"M{m['no']} {m['short']} 결과",
                                 "sections": [{"h": f"{si}. {st['short']}", "blocks": [b["id"] for b in st.get("blocks", [])]}
                                              for si, st in enumerate(m["steps"], 1)]}]
    credits = asset_credits_html(_asset_refs(m)) if ASSET_ITEMS and not blanks else ""      # 이 미션에 쓴 사진의 출처 (미션을 마친 뒤 공개)
    for k, o in enumerate(outs):
        body = f"<p class='meta'>{_md(o['intro'])}</p>" if o.get("intro") else ""
        for sec in o.get("sections", []):
            body += f"<h2>{html.escape(sec['h'])}</h2>" if sec.get("h") else ""
            body += "".join(f"<p>{_md(p)}</p>" for p in sec.get("text", []))
            for bid in sec.get("blocks", []):
                body += _block_html(mid, BSPEC[(mid, bid)][1])
        if k == len(outs) - 1:
            body += credits
        _add_file(o["id"], o["title"], mid, _doc(o["title"], body), blanks)


def blk_todo(mid):
    """[산출물] 화면: 아직 안 한 활동 목록."""
    return JSON({"ok": True, "todo": _todo_list(mid)})


def blk_finish(mid, force=False):
    """[산출물 받기]: 안 한 활동이 있으면 먼저 알려 주고 물어봄(need_confirm). 그래도 받으면(force) 안 한 부분은 빈칸으로 만듦."""
    m = MSPEC[mid]
    _ensure_started(mid)
    todo = _todo_list(mid)
    if todo and not force:
        return _resp(ok=False, need_confirm=True, todo=todo)
    n_blank = sum(len(t["blocks"]) for t in todo)
    _mission_docs(mid, n_blank)
    p = S["progress"][mid]
    first = not p["issued"] and not todo
    if not todo:
        p["issued"] = True
    p["done"] = not todo
    if first:
        for ch, who, text in m.get("finish_messages", []):
            post(ch, who, text)
    msg = (f"M{m['no']} 산출물이 📁 파일에 저장됐어요." if not todo
           else f"M{m['no']} 산출물을 저장했어요. 하지 않은 활동 {n_blank}개는 빈칸이에요. 마저 하고 다시 받으면 채워져요.")
    return _resp(ok=True, msg=msg, blanks=n_blank)


def _prefill_block(mid, b):
    """선생님용: 블록 하나를 정답으로 채움."""
    t, st = b["type"], _bst(mid, b["id"])
    if t == "info":
        if b.get("confirm"):
            st["value"] = True
            block_ok(mid, b["id"])
        return
    if t == "quiz":
        st["value"] = {"pick": b["answer"]}
    elif t == "multi":
        st["value"] = {"picked": [i for i, it in enumerate(b["items"]) if it["correct"]]}
    elif t == "match":
        st["value"] = {p["left"]: str(p["right"]) for p in b["pairs"]}
    elif t == "order":
        st["value"] = {"order": [it["id"] for it in b["items"]]}
    elif t == "classify":
        st["value"] = {it["id"]: {"primary": it["accept"][0], "contrib": list(it.get("contrib", []))} for it in b["items"]}
    elif t == "write":
        st["value"] = b.get("example", "")
    elif t == "decide":
        st["value"] = {it["id"]: {"choice": it["answer"], "reason": it.get("reason") or it.get("explain") or ""} for it in b["items"]}
    elif t == "vibe":
        _vibe_prefill(mid, b)
    elif t == "custom":
        fn = AUTOPLAY.get((mid, b["id"]))
        if fn:
            fn()
        return
    block_ok(mid, b["id"])


def _prefill_all():
    """선생님용: 모든 미션의 모든 활동(심화 포함)을 채우고 산출물까지 만들어 둠. 빈 값이 없는 상태."""
    for mid in IMPLEMENTED:
        _ensure_started(mid)
        for si, st in enumerate(MSPEC[mid]["steps"], 1):
            if st.get("page") and (mid, f"page{si}") in AUTOPLAY:
                AUTOPLAY[(mid, f"page{si}")]()
            for b in st.get("blocks", []):
                _prefill_block(mid, b)
        blk_finish(mid, True)


# ---- 미션 등록: 소개 · 처음 상태 · 시작 메시지 · 화면 상태 ----
def _pub_state(mid, b):
    """화면에 보낼 블록 상태: 맞힌 뒤에만 풀이를 함께 보냄 (정답은 보내지 않음)."""
    st = _bst(mid, b["id"])
    out = {"ok": st["ok"], "tries": st["tries"], "value": st["value"]}
    if st["ok"]:
        if b["type"] == "quiz":
            out["explain"] = b["explain"]
        elif b["type"] == "multi":
            out["explains"] = [it.get("explain", "") for it in b["items"]]
        elif b["type"] == "decide":
            out["explains"] = {it["id"]: it.get("explain", "") for it in b["items"]}
        elif b["type"] in ("write", "match", "order") and b.get("explain"):
            out["explain"] = b["explain"]
        if b["type"] == "write":
            out["model"] = b.get("example", "")      # 모범답안: 제출한 뒤에만 보냄
    if b["type"] == "classify":
        out["explains"] = {it["id"]: it["explain"] for it in b["items"] if it["id"] in (st["value"] or {})}
    if b["type"] == "write" and st["ok"]:
        out["checklist"] = b.get("checklist", [])
    return out


def _mission_reset(mid):
    S[mid] = {"blk": {}, "pages": {}, "vibe": {}}
    if mid in CUSTOM_RESET:
        CUSTOM_RESET[mid]()


def _mission_view(mid):
    m = MSPEC[mid]
    out = {"blk": {b["id"]: _pub_state(mid, b) for st in m["steps"] for b in st.get("blocks", [])},
           "complete": [_step_done(mid, si) for si in range(1, len(m["steps"]) + 1)]}
    if mid in CUSTOM_VIEWS:
        out["custom"] = CUSTOM_VIEWS[mid]()
    return out


def _mission_start(mid):
    for ch, who, text in MSPEC[mid].get("start_messages", []):
        post(ch, who, text)


for _m in SPEC["missions"]:
    if not _m.get("steps"):
        continue
    BRIEFINGS[_m["id"]] = dict(_m["brief"], steps_short=[st["short"] for st in _m["steps"]],
                               comps=[{"name": c["name"], "criteria": c.get("criteria", "")} for c in SPEC.get("competencies", [])
                                      if c.get("id") in _m.get("competencies", [])])
    RESETTERS[_m["id"]] = (lambda mid: lambda: _mission_reset(mid))(_m["id"])
    VIEWS[_m["id"]] = (lambda mid: lambda: _mission_view(mid))(_m["id"])
    STARTERS[_m["id"]] = (lambda mid: lambda: _mission_start(mid))(_m["id"])

CALLBACKS.update({"blk.check": blk_check, "blk.finish": blk_finish, "blk.todo": blk_todo})
code_topic("app.blocks", "미션 활동은 이렇게 채점해요", "[확인]을 누르면 blk_check()가 명세의 정답과 비교해요. 틀리면 힌트를 한 단계씩 자세하게 보내고, "
           "맞히면 풀이를 보낸 뒤 단계를 다 마쳤는지 확인해요. 글쓰기는 맞고 틀림을 따지지 않고, 제출하면 모범답안을 보내요. 정답은 파이썬에만 있고 화면으로는 보내지 않아요.",
           ["blk_check", "_chk_quiz", "_chk_match", "_chk_write", "block_ok"], marks=["_hint(", "block_ok(", "\"model\"", "_advance("])
code_topic("app.docs", "산출물 문서는 이렇게 만들어요", "산출물을 받으면 명세의 outputs 순서대로, 학습자가 답한 내용을 모아 HTML 문서를 만들어요. 하지 않은 활동은 빈칸이에요.",
           ["blk_finish", "_mission_docs", "_block_html"], marks=["_add_file(", "_doc(", "finish_messages"])
# ===== 바이브 코딩 실습: 가상 데이터 파일 · 가상 AI (인터넷·외부 AI 없이 동작) =====
# 가상 데이터: 명세의 project.vfiles {파일 이름: 글}을 첫 실행 때 폴더에 저장함. 내려받는 것이 없음
# 가상 AI: 학습자가 고른 프롬프트 문장에 따라 명세에 적힌 응답을 돌려줌. 응답의 코드는 실제로 실행해 결과·오류를 그대로 보여 줌
#  - 정답(어느 문장이 좋은지)·응답은 파이썬에만 둠. 화면에는 보낸 뒤의 응답만 감
#  - 시도 기록이 남아 산출물의 '프롬프트 개선 기록'이 됨
import contextlib

VFILES = _P.get("vfiles") or {}
_VIBE_CACHE = {}


def vdata_folder():
    """가상 데이터 폴더: Colab은 /content/pbl_data, 점검 도구는 PBL_VDATA_ROOT(임시 폴더)."""
    root = os.environ.get("PBL_VDATA_ROOT") or ("/content" if os.path.isdir("/content") else os.getcwd())
    return os.path.join(root, _P.get("vfolder") or "pbl_data")


def vdata_prepare():
    """가상 데이터 파일을 폴더에 저장 (코드를 복사해 새 셀에서 돌려도 같은 파일을 읽게)."""
    if not VFILES:
        return
    folder = vdata_folder()
    os.makedirs(folder, exist_ok=True)
    for name, text in VFILES.items():
        with open(os.path.join(folder, name), "w", encoding="utf-8") as f:
            f.write(text)


def _vibe_st(mid, bid):
    return S[mid].setdefault("vibe", {}).setdefault(bid, {"picks": {}, "attempts": [], "last": None})


def _vibe_prompt(b, picks):
    """고른 문장을 이어 붙인 프롬프트."""
    lines = []
    for sl in b["slots"]:
        i = picks.get(sl["id"])
        if i is not None and 0 <= int(i) < len(sl["options"]) and sl["options"][int(i)].get("text"):
            lines.append(sl["options"][int(i)]["text"])
    return "\n".join(lines)


def _vibe_code(code):
    return (code or "").replace("{VDIR}", vdata_folder())


def _vibe_run(code):
    """AI가 만든 코드를 실제로 실행: 출력·표·그림·오류·문서(report 변수)를 모아 돌려줌. 같은 코드는 다시 돌리지 않음."""
    code = _vibe_code(code)
    if not code.strip():
        return {}
    if code in _VIBE_CACHE:
        return _VIBE_CACHE[code]
    tables, figs = [], []

    def display(obj=None, *a, **k):               # Colab의 display와 같은 이름: 복사해 새 셀에서 돌려도 그대로 동작함
        if hasattr(obj, "to_html"):
            tables.append(obj.to_html(index=False, border=0))
        else:
            print(obj)

    env = {"__name__": "__vibe__", "display": display}
    out, err = io.StringIO(), None
    cwd = os.getcwd()
    plt, prev = None, None
    if "matplotlib" in code:
        try:
            import matplotlib.pyplot as plt
            prev = plt.get_backend()
            plt.switch_backend("Agg")
        except Exception:
            plt = None
    try:
        os.makedirs(vdata_folder(), exist_ok=True)
        os.chdir(vdata_folder())
        with contextlib.redirect_stdout(out):
            exec(compile(code, "AI가 만든 코드", "exec"), env)
    except Exception as e:
        tb = traceback.extract_tb(e.__traceback__)
        frame = next((f for f in reversed(tb) if f.filename == "AI가 만든 코드"), None)
        err = "Traceback (most recent call last):\n"
        if frame:
            src = code.splitlines()[frame.lineno - 1].strip()
            err += f'  File "AI가 만든 코드", line {frame.lineno}\n    {src}\n'
        err += f"{type(e).__name__}: {e}"
    finally:
        os.chdir(cwd)
        if plt is not None:
            for n in plt.get_fignums():
                buf = io.BytesIO()
                plt.figure(n).savefig(buf, format="png", dpi=110, bbox_inches="tight")
                figs.append("data:image/png;base64," + base64.b64encode(buf.getvalue()).decode())
            plt.close("all")
            try:
                plt.switch_backend(prev)
            except Exception:
                pass
    report = env.get("report") if isinstance(env.get("report"), str) else None
    res = {"stdout": out.getvalue()[-6000:], "tables": tables, "figs": figs, "report": report, "error": err}
    _VIBE_CACHE[code] = res
    return res


def _vibe_reply(rep, ok, slot_label=None, slot_id=None):
    """명세의 응답 + 실제 실행 결과 → 화면에 보낼 것."""
    out = {k: rep.get(k, "") for k in ("say", "problem", "hint", "lesson")}
    out.update(code=_vibe_code(rep.get("code", "")), ok=ok, slot=slot_label, slot_id=slot_id, open_code=bool(rep.get("open_code")))
    out.update(_vibe_run(rep.get("code", "")))
    return out


def _chk_vibe(mid, b, st, payload):
    """프롬프트 보내기: 고른 문장을 앞 칸부터 보고, 부족한 칸이 있으면 그 칸의 응답을, 모두 좋으면 done 응답을 돌려줌."""
    picks = {k: int(v) for k, v in ((payload or {}).get("picks") or {}).items() if str(v).lstrip("-").isdigit()}
    if any(sl["id"] not in picks for sl in b["slots"]):
        return False, None, {"text": "칸마다 문장을 하나씩 골라 주세요."}
    v = _vibe_st(mid, b["id"])
    v["picks"] = picks
    miss = None
    for sl in b["slots"]:
        opt = sl["options"][picks[sl["id"]]] if 0 <= picks[sl["id"]] < len(sl["options"]) else {}
        if not opt.get("good"):
            miss = (sl, opt)
            break
    prompt = _vibe_prompt(b, picks)
    if miss:
        sl, opt = miss
        rep = _vibe_reply(opt.get("miss") or sl["miss"], False, sl["label"], sl["id"])
        st["tries"] += 1
    else:
        rep = _vibe_reply(b["done"], True)
    v["last"] = rep
    v["attempts"].append({"no": len(v["attempts"]) + 1, "prompt": prompt, "ok": rep["ok"], "slot": rep["slot"],
                          "result": re.sub(r"\*\*", "", rep["problem"]) or "요청대로 동작함"})
    resp = {"reply": rep, "attempts": [{k: a[k] for k in ("no", "ok", "slot", "result")} for a in v["attempts"]], "text": ""}
    if rep["ok"]:
        return True, {"picks": picks, "prompt": prompt}, resp
    return False, None, resp


def app_vibe(mid, bid):
    """화면이 블록을 다시 그릴 때 마지막 응답을 받아 감 (상태 전체에 넣으면 그림 때문에 무거워져서 따로 보냄)."""
    v = _vibe_st(mid, bid)
    b = BSPEC[(mid, bid)][1]
    start = _vibe_reply(b["start"], False) if b.get("start") else None      # 지금까지의 코드와 결과: 블록 맨 위에 보여 줌
    return JSON({"ok": True, "reply": v["last"], "picks": v["picks"], "start": start,
                 "attempts": [{k: a[k] for k in ("no", "ok", "slot", "result")} for a in v["attempts"]]})


def _vibe_prefill(mid, b):
    """선생님용: 좋은 문장을 고른 상태 + done 응답."""
    picks = {sl["id"]: next(i for i, o in enumerate(sl["options"]) if o.get("good")) for sl in b["slots"]}
    st = _bst(mid, b["id"])
    ok, value, _ = _chk_vibe(mid, b, st, {"picks": picks})
    st["value"] = value
    return ok


def _vibe_html(mid, b, st):
    """산출물 문서: 최종 프롬프트 · 개선 기록 · 실행 결과 · AI가 만든 문서. 하지 않았으면 빈칸."""
    v = _vibe_st(mid, b["id"])
    if not st["ok"]:
        out = "<h4>최종 프롬프트</h4><div class='blankbox'></div><h4>실행 결과</h4><div class='blankbox'></div>"
    else:
        rep = v["last"] if (v["last"] or {}).get("ok") else _vibe_reply(b["done"], True)
        out = f"<h4>최종 프롬프트</h4><pre>{html.escape((st['value'] or {}).get('prompt', ''))}</pre>"
        if rep.get("stdout"):
            out += f"<h4>실행 결과</h4><pre>{html.escape(rep['stdout'])}</pre>"
        elif rep.get("tables") or rep.get("figs"):
            out += "<h4>실행 결과</h4>"
        out += "".join(rep.get("tables") or [])
        out += "".join(f"<p><img src='{u}'></p>" for u in rep.get("figs") or [])
        if rep.get("report"):
            out += f"<h4>AI가 만든 문서</h4><div class='card'>{rep['report']}</div>"
        if rep.get("code"):
            out += f"<details><summary>AI가 만든 코드 보기</summary><pre>{html.escape(rep['code'])}</pre></details>"
    if v["attempts"]:
        rows = "".join(f"<tr><td>{a['no']}</td><td>{'성공' if a['ok'] else html.escape(a['slot'] or '')}</td><td>{html.escape(a['result'])}</td></tr>" for a in v["attempts"])
        out += f"<h4>프롬프트 개선 기록</h4><table><tr><th>시도</th><th>부족했던 칸</th><th>결과</th></tr>{rows}</table>"
    return out


CHECKERS["vibe"] = _chk_vibe
CALLBACKS["app.vibe"] = app_vibe
vdata_prepare()
_setup(60, 2, "미션 활동 · 가상 AI 준비 완료")

# ===== 시작 준비: 화면 버튼과 파이썬 함수 연결 =====
_reset_state()
for _name, _fn in CALLBACKS.items():
    output.register_callback(_name, _fn)

_setup(100, 3, "✅ 준비 완료! 아래 <b>💬 시설 방호 영상분석반 열기</b> 셀을 실행하세요. (‘모두 실행’을 눌렀다면 곧 자동으로 열려요)")
```

```python
#@title 💬 시설 방호 영상분석반 열기 { display-mode: "form" }
#@markdown 위의 **🚀 [먼저 실행] 실습 준비하기** 셀의 진행 막대가 100%가 된 뒤 이 셀을 실행하면, 아래에 메신저 화면이 열립니다.
from IPython.display import HTML, display

WORKSPACE_NAME = "🛡️ 시설 방호 영상분석반"

APP_HTML = r"""
<style>
  .ws { --ink:#111827; --line:#c3cad5; --soft:#f2f5f9; --blue:#0b57d0; --blue-bg:#e8f0fe; --green:#146c2e; --green-bg:#e6f4ea;
        --red:#b3261e; --red-bg:#fce8e6; --amber:#8a4b00; --amber-bg:#fff4ce; --violet:#3f0e40;
        color-scheme: light; font-family: 'Noto Sans KR', 'Malgun Gothic', 'Apple SD Gothic Neo', sans-serif;
        border: 1px solid var(--line); border-radius: 10px; box-sizing: border-box; overflow: hidden; background: #fff; color: var(--ink);
        font-size: 14px; display: flex; flex-direction: column; height: 900px; width: 100%; max-width: 100%;
        min-width: 360px; min-height: 480px; resize: both; position: relative; }
  .ws * { box-sizing: border-box; color: inherit; }
  .ws.w-base { width: 1100px; } .ws.w-wide { width: 1400px; } .ws.w-full { width: 100%; }
  .ws button, .ws input, .ws select, .ws textarea { font-family: inherit; }
  .ws a { color: var(--blue); }
  .ws h2, .ws h3, .ws h4, .ws h5, .ws p, .ws li, .ws label, .ws td, .ws th, .ws dt, .ws dd, .ws summary { color: var(--ink); }
  .ws ::placeholder { color: #1e40af; opacity: .85; font-style: italic; }
  /* 상단 바 */
  .ws .topbar { display: flex; align-items: center; gap: 8px; padding: 6px 10px; background: var(--violet); color: #fff; font-size: 13px; flex-wrap: wrap; }
  .ws .topbar * { color: #fff; }
  .ws .topbar .name { font-weight: bold; margin: 0 4px; font-size: 14px; }
  .ws .topbar .spacer { flex: 1; }
  .ws .tb { background: transparent; border: 1px solid #b89bb9; border-radius: 6px; padding: 2px 9px; cursor: pointer; font-size: 13px; position: relative; }
  .ws .tb:hover { background: #5a2a5b; }
  .ws .tb.on { background: #fff; border-color: #fff; } .ws .tb.on, .ws .tb.on * { color: var(--violet); }
  .ws .tb .count { position: absolute; top: -6px; right: -8px; background: #e01e5a; border-radius: 9px; padding: 0 5px; font-size: 11px; display: none; }
  .ws.side-off .tb .count.has { display: inline; }
  .ws .tprog { display: flex; align-items: center; gap: 6px; font-size: 12px; }
  .ws .tprog .pb { width: 120px; height: 9px; background: #6b3d6c; border-radius: 6px; overflow: hidden; }
  .ws .tprog .pb i { display: block; height: 100%; background: #6ee7a8; width: 0; transition: width .4s; }
  .ws .me-chip { background: #5a2a5b; border: 1px solid #b89bb9; border-radius: 14px; padding: 2px 10px; cursor: pointer; font-size: 12.5px; }
  .ws .me-chip:hover { background: #6d3a6e; }
  /* 본문 */
  .ws .body { display: flex; flex: 1; min-height: 0; position: relative; }
  .ws .side { width: 236px; background: #3f0e40; color: #fff; padding: 8px 0 14px; overflow-y: auto; flex-shrink: 0; transition: width .2s, padding .2s; }
  .ws .side * { color: #fff; }
  .ws.side-off .side, .ws.onboard .side { width: 0; padding: 0; }
  .ws .side h4 { margin: 14px 14px 4px; font-size: 12.5px; font-weight: normal; line-height: 1.4; border-top: 1px solid #6d3a6e; padding-top: 10px; }
  .ws .side h4 b { font-weight: bold; }
  .ws .item { padding: 5px 14px; cursor: pointer; display: flex; align-items: center; gap: 6px; white-space: nowrap; font-size: 13.5px; }
  .ws .item span.t { overflow: hidden; text-overflow: ellipsis; flex: 1; }
  .ws .item:hover { background: #5a2a5b; }
  .ws .item.active { background: #1164a3; }
  .ws .item.unread { font-weight: bold; }
  .ws .item.soon span.t { font-style: italic; }
  .ws .item .ic { width: 18px; text-align: center; flex-shrink: 0; }
  .ws .item .mini { font-size: 11px; background: #1164a3; border-radius: 8px; padding: 0 6px; }
  .ws .badge { background: #e01e5a; color: #fff; border-radius: 9px; padding: 0 7px; font-size: 12px; }
  .ws .main { flex: 1; display: flex; flex-direction: column; min-width: 0; }
  .ws .title { padding: 10px 16px; font-weight: bold; border-bottom: 1px solid var(--line); font-size: 15px; }
  .ws .title small { font-weight: normal; margin-right: 6px; }
  .ws .scroll { flex: 1; overflow-y: auto; padding: 14px 18px 30px; min-height: 0; }
  .ws .footer { display: flex; gap: 10px; align-items: center; padding: 7px 14px; background: var(--soft); border-top: 1px solid var(--line); }
  .ws .note { flex: 1; font-size: 13px; } .ws .note.ok { color: var(--green); } .ws .note.bad { color: var(--red); font-weight: bold; }
  .ws.busy .note::before { content: '⏳ '; }
  /* 코드 보기 */
  .ws .codep { width: 0; overflow: hidden; border-left: 1px solid var(--line); background: #fbfcfe; display: flex; flex-direction: column; flex-shrink: 0; }
  .ws.code-on .codep { width: 410px; }
  .ws.compact.code-on .codep { position: absolute; right: 0; top: 0; bottom: 0; width: min(430px, 94%); z-index: 20; box-shadow: -8px 0 18px rgba(0,0,0,.22); }
  .ws.onboard .codep, .ws.onboard .code-tab, .ws.no-code .codep, .ws.no-code .code-tab, .ws.no-code #tb-code { display: none; }
  .ws .code-tab { position: absolute; right: 0; top: 30%; z-index: 10; writing-mode: vertical-rl; background: #1e293b; color: #fff; border: 0;
                  border-radius: 8px 0 0 8px; padding: 12px 6px; cursor: pointer; font-size: 13px; letter-spacing: 1px; box-shadow: -2px 2px 8px rgba(0,0,0,.2); }
  .ws .code-tab * { color: #fff; }
  .ws.code-on .code-tab { display: none; }
  .ws .codep .ch { display: flex; align-items: center; gap: 6px; padding: 9px 12px; background: #1e293b; }
  .ws .codep .ch, .ws .codep .ch * { color: #fff; }
  .ws .codep .ch b { flex: 1; font-size: 14px; }
  .ws .codep .ch button { background: transparent; border: 1px solid #94a3b8; border-radius: 6px; cursor: pointer; padding: 1px 8px; }
  .ws .codep .cs { flex: 1; overflow-y: auto; padding: 10px 12px 24px; }
  .ws .codep .intro { font-size: 12.5px; line-height: 1.55; background: var(--blue-bg); border-radius: 8px; padding: 8px 10px; margin-bottom: 10px; }
  .ws .codep h6 { font-size: 13px; margin: 12px 0 4px; display: flex; align-items: center; gap: 6px; }
  .ws .codep h6 .sp { flex: 1; }
  .ws .codep .desc { font-size: 12.5px; line-height: 1.5; margin: 0 0 6px; }
  .ws .codep details { border: 1px solid var(--line); border-radius: 8px; margin: 6px 0; background: #fff; }
  .ws .codep details > summary { padding: 6px 8px; cursor: pointer; font-size: 12.5px; color: var(--ink); }
  .ws .codep details > summary code { font-weight: bold; color: #6d28d9; }
  .ws .cb { display: flex; border-top: 1px solid #e5e9f0; font: 12px/18px Consolas, 'D2Coding', Menlo, monospace; }
  .ws .cb .ln { margin: 0; padding: 8px 6px; text-align: right; background: #eef2f7; user-select: none; min-width: 30px; }
  .ws .cb .cw { position: relative; overflow-x: auto; flex: 1; }
  .ws .cb pre { margin: 0; padding: 8px 10px; white-space: pre; position: relative; z-index: 1; font: inherit; }
  .ws .cb .hlb { position: absolute; left: 0; right: 0; height: 18px; background: #fff3b0; z-index: 0; }
  .ws .code .k { color: #7c3aed; font-weight: bold; } .ws .code .s { color: #b3261e; } .ws .code .c { color: #146c2e; font-style: italic; }
  .ws .code .n { color: #9a3412; } .ws .code .f { color: #0b57d0; }
  .ws .live pre { background: #0f172a; border-radius: 8px; padding: 10px; margin: 0; white-space: pre-wrap; word-break: break-all; font: 12px/1.6 Consolas, 'D2Coding', Menlo, monospace; }
  .ws .live pre, .ws .live pre * { color: #f8fafc; }
  .ws .live pre .c { color: #86efac; font-style: italic; } .ws .live pre .s { color: #fca5a5; } .ws .live pre .k { color: #c4b5fd; font-weight: bold; }
  .ws .live pre .n { color: #fdba74; } .ws .live pre .f { color: #93c5fd; }
  .ws .live .empty { font-size: 12.5px; padding: 8px 10px; border: 1px dashed var(--line); border-radius: 8px; }
  .ws .cpy { font-size: 11.5px; padding: 1px 8px; border-radius: 6px; border: 1px solid var(--line); background: #fff; cursor: pointer; }
  /* 공통 요소 */
  .ws .btn { padding: 7px 14px; border-radius: 7px; border: 1px solid #007a5a; background: #007a5a; color: #fff; cursor: pointer; font-size: 14px; font-weight: bold; }
  .ws .btn * { color: #fff; }
  .ws .btn:hover { filter: brightness(1.08); }
  .ws .btn.ghost { background: #fff; } .ws .btn.ghost, .ws .btn.ghost * { color: #006048; }
  .ws .btn.small { padding: 3px 10px; font-size: 12.5px; }
  .ws .btn.big { padding: 10px 20px; font-size: 15px; }
  .ws .btn.blue { background: var(--blue); border-color: var(--blue); }
  .ws .btn:disabled { background: #fff; border: 1px dashed #6b7280; cursor: not-allowed; font-weight: normal; }
  .ws .btn:disabled, .ws .btn:disabled * { color: var(--ink); }
  .ws .btn.pulse { animation: pulse 1.6s infinite; }
  @keyframes pulse { 0% { box-shadow: 0 0 0 0 rgba(11,87,208,.55); } 70% { box-shadow: 0 0 0 9px rgba(11,87,208,0); } 100% { box-shadow: 0 0 0 0 rgba(11,87,208,0); } }
  .ws .row { display: flex; flex-wrap: wrap; gap: 8px; align-items: center; margin: 8px 0; }
  .ws .card { border: 1px solid var(--line); border-radius: 10px; padding: 12px 14px; margin: 10px 0; background: #fff; }
  .ws .card.focus { border: 2px solid var(--blue); box-shadow: 0 0 0 3px var(--blue-bg); }
  .ws .card h4 { margin: 0 0 6px; font-size: 15px; }
  .ws .ok { color: var(--green); } .ws .bad { color: var(--red); }
  .ws .fb { margin-top: 8px; padding: 8px 10px; border-radius: 7px; font-size: 13.5px; line-height: 1.6; }
  .ws .fb.ok { background: var(--green-bg); color: #0b3d1a; border: 1px solid #9fd5ad; }
  .ws .fb.bad { background: var(--red-bg); color: #5c0f0a; border: 1px solid #efa9a3; }
  .ws .fb.info { background: var(--blue-bg); color: #0a2a66; border: 1px solid #a8c3f0; }
  .ws .fb.ok *, .ws .fb.bad *, .ws .fb.info * { color: inherit; }
  .ws .err { color: var(--red); font-size: 13px; font-weight: bold; margin-top: 4px; }
  .ws .err:empty { display: none; }
  .ws .opt-tag { display: inline-block; background: var(--blue-bg); border-radius: 4px; font-size: 11px; padding: 0 5px; margin-left: 4px; }
  .ws .missing { outline: 3px solid var(--red); outline-offset: 2px; border-radius: 6px; }
  .ws select, .ws input[type=text], .ws input[type=date], .ws input[type=time], .ws input[type=number], .ws textarea {
    padding: 6px 8px; font-size: 13.5px; background: #fff; color: var(--ink); border: 1px solid #8a94a6; border-radius: 6px; max-width: 100%; }
  .ws textarea { width: 100%; min-height: 56px; line-height: 1.5; }
  .ws select option, .ws select optgroup { color: var(--ink); background: #fff; }
  .ws .count-line { font-size: 12.5px; margin-top: 2px; }
  .ws .count-line.short { color: var(--red); font-weight: bold; } .ws .count-line.enough { color: var(--green); font-weight: bold; }
  .ws table.t { border-collapse: collapse; width: 100%; font-size: 13px; margin: 6px 0; }
  .ws table.t th, .ws table.t td { border-bottom: 1px solid #dfe4ea; padding: 6px 8px; text-align: left; vertical-align: middle; }
  .ws table.t th { font-weight: bold; background: var(--soft); }
  .ws table.t tr.split td { background: var(--amber-bg); }
  .ws .lbl-unk { font-style: italic; text-decoration: underline dotted; } .ws .lbl-wrong { color: var(--red); font-weight: bold; }
  .ws details > summary { cursor: pointer; color: var(--blue); margin: 6px 0; font-weight: bold; }
  .ws .hint { font-size: 13px; line-height: 1.55; }
  .ws .tip { background: var(--soft); border-left: 4px solid var(--blue); border-radius: 6px; padding: 8px 12px; font-size: 13.5px; line-height: 1.6; margin: 8px 0; }
  /* 할 일 상자 */
  /* 단계 안내 */
  .ws .sub { border-left: 4px solid #cbd5e1; padding: 6px 0 6px 12px; margin: 10px 0; }
  .ws .sub.now { border-left-color: var(--blue); background: linear-gradient(90deg, var(--blue-bg), #fff 70%); border-radius: 0 8px 8px 0; }
  .ws .sub.done { border-left-color: var(--green); }
  .ws .sub-h { font-weight: bold; margin-bottom: 6px; font-size: 14px; }
  .ws .look { background: #fff; border: 1px dashed var(--blue); border-radius: 8px; padding: 7px 10px; line-height: 1.6; }
  /* 선택지 */
  .ws .choices { display: flex; flex-wrap: wrap; gap: 6px; }
  .ws .choices.col { flex-direction: column; align-items: stretch; }
  .ws .choice { display: flex; align-items: flex-start; gap: 7px; border: 1px solid #8a94a6; border-radius: 8px; padding: 6px 10px; cursor: pointer; background: #fff; font-size: 13.5px; line-height: 1.45; }
  .ws .choice:hover { border-color: var(--blue); background: #f5f8ff; }
  .ws .choice.on { border: 2px solid var(--blue); background: var(--blue-bg); font-weight: bold; padding: 5px 9px; }
  .ws .choice input { margin: 2px 0 0; accent-color: var(--blue); }
  .ws .choice small { display: block; font-weight: normal; font-size: 12px; }
  .ws .choice.good { border-color: var(--green); background: var(--green-bg); } .ws .choice.wrong { border-color: var(--red); background: var(--red-bg); }
  .ws .checks { display: grid; grid-template-columns: repeat(auto-fill, minmax(250px, 1fr)); gap: 6px; }
  .ws .fieldrow { margin: 10px 0; } .ws .fieldrow > .fl { font-weight: bold; margin-bottom: 4px; display: block; }
  /* 메신저 */
  .ws .msgs { flex: 1; overflow-y: auto; padding: 8px 16px; min-height: 0; }
  .ws .msg { display: flex; gap: 10px; padding: 7px 6px; border-radius: 6px; }
  .ws .msg:hover { background: var(--soft); }
  .ws .msg.mention { background: #fff4ce; border-left: 3px solid #f0b429; }
  .ws .avatar { width: 34px; height: 34px; border-radius: 6px; display: flex; align-items: center; justify-content: center; font-weight: bold; flex-shrink: 0; }
  .ws .avatar, .ws .avatar * { color: #fff; }
  .ws .meta { font-size: 13px; } .ws .meta b { margin-right: 4px; }
  .ws .time { font-size: 12px; margin-left: 6px; }
  .ws .text { white-space: pre-wrap; word-break: break-word; margin-top: 2px; line-height: 1.55; }
  .ws .tag { font-size: 11.5px; border-radius: 4px; padding: 1px 6px; margin-left: 4px; white-space: nowrap; background: #e9edf3; border: 1px solid var(--line); }
  .ws .tag.ext { background: #f3e8fd; border-color: #d7b8f5; }
  .ws .compose { display: flex; gap: 8px; padding: 10px 14px; border-top: 1px solid var(--line); }
  .ws .compose input { flex: 1; padding: 8px; font-size: 14px; }
  /* 첫 화면 */
  .ws .onb { max-width: 560px; margin: 26px auto; border: 2px solid var(--blue); border-radius: 14px; padding: 22px 24px; background: #fff; }
  .ws .onb h2 { margin: 0 0 6px; font-size: 21px; }
  .ws .onb .lead { line-height: 1.6; margin: 0 0 14px; }
  .ws .onb select, .ws .onb input { width: 100%; font-size: 15px; padding: 8px 10px; }
  .ws .preview-call { background: var(--blue-bg); border-radius: 8px; padding: 8px 12px; margin-top: 12px; }
  /* 홈 */
  .ws .hero { background: linear-gradient(135deg, #3f0e40, #1164a3); border-radius: 12px; padding: 18px 20px; }
  .ws .hero, .ws .hero * { color: #fff; }
  .ws .hero h2 { margin: 4px 0 8px; font-size: 21px; }
  .ws .hero p { margin: 0; line-height: 1.65; font-size: 13.5px; }
  .ws .greet { font-size: 16px; font-weight: bold; margin: 14px 0 6px; }
  .ws .bigprog { display: flex; align-items: center; gap: 10px; }
  .ws .bigprog .pb { flex: 1; height: 14px; background: #e3e8ef; border-radius: 8px; overflow: hidden; }
  .ws .bigprog .pb i { display: block; height: 100%; background: linear-gradient(90deg, #16a34a, #22c55e); }
  .ws .stages { display: grid; grid-template-columns: repeat(auto-fill, minmax(250px, 1fr)); gap: 12px; margin-top: 10px; }
  .ws .stage { border: 1px solid var(--line); border-radius: 10px; padding: 10px 12px; }
  .ws .stage h5 { margin: 0 0 8px; font-size: 13.5px; }
  .ws .mcard { display: flex; gap: 8px; align-items: flex-start; padding: 7px 8px; border-radius: 8px; cursor: pointer; border: 1px solid transparent; }
  .ws .mcard:hover { background: var(--soft); border-color: var(--line); }
  .ws .pill { font-size: 11.5px; padding: 1px 8px; border-radius: 10px; white-space: nowrap; border: 1px solid transparent; }
  .ws .pill.done { background: var(--green-bg); border-color: #9fd5ad; } .ws .pill.open { background: var(--blue-bg); border-color: #a8c3f0; }
  .ws .pill.locked { background: #fff; border: 1px dashed #6b7280; } .ws .pill.soon { background: var(--amber-bg); border-color: #e3c96b; }
  /* 미션 */
  .ws .stepper { display: flex; flex-wrap: wrap; gap: 6px; padding: 8px 16px; border-bottom: 1px solid var(--line); background: #fafbfd; }
  .ws .stp { font-size: 12.5px; padding: 4px 10px; border-radius: 14px; border: 1px solid #8a94a6; background: #fff; cursor: pointer; }
  .ws .stp.done { border-color: var(--green); background: var(--green-bg); }
  .ws .stp.cur, .ws .stp.cur.done { background: var(--blue); border-color: var(--blue); font-weight: bold; }
  .ws .stp.cur, .ws .stp.cur * { color: #fff; }
  .ws .stp:disabled { cursor: not-allowed; border-style: dashed; background: #fff; }
  .ws .brief h2 { margin: 6px 0 2px; font-size: 21px; }
  .ws .brief .sub-t { margin: 0 0 10px; font-size: 15px; }
  .ws .brief section { margin: 16px 0; }
  .ws .brief h4 { margin: 0 0 6px; font-size: 15px; }
  .ws .brief p, .ws .brief li { line-height: 1.7; }
  .ws .why { background: var(--soft); border-left: 4px solid #1164a3; border-radius: 6px; padding: 10px 14px; }
  .ws .terms { display: grid; grid-template-columns: repeat(auto-fill, minmax(300px, 1fr)); gap: 12px; }
  .ws .term { border: 1px solid var(--line); border-radius: 10px; padding: 10px 12px; background: #fff; }
  .ws .term h5 { margin: 0 0 6px; font-size: 14.5px; } .ws .term h5 small { font-weight: normal; font-size: 11.5px; margin-left: 4px; }
  .ws .term dl { margin: 0; display: grid; grid-template-columns: 74px 1fr; gap: 4px 8px; font-size: 13px; line-height: 1.55; }
  .ws .term dt { font-weight: bold; } .ws .term dd { margin: 0; }
  .ws .term dt.k1 { color: var(--blue); } .ws .term dt.k2 { color: var(--green); } .ws .term dt.k3 { color: #9a3412; }
  .ws .fig { background: #f8fafc; border: 1px solid #e3e8ef; border-radius: 8px; margin: 4px 0 8px; padding: 4px; }
  .ws .fig svg { display: block; width: 100%; height: auto; max-height: 170px; }
  /* M1 실험실 */
  .ws .lab2 { display: grid; grid-template-columns: minmax(0, 1fr) minmax(0, 1fr); gap: 12px; }
  .ws .main.narrow .lab2 { grid-template-columns: minmax(0, 1fr); }
  .ws .labcol { border: 1px solid var(--line); border-radius: 10px; padding: 10px; min-width: 0; background: #fff; }
  .ws .labcol.focus { border: 2px solid var(--blue); box-shadow: 0 0 0 3px var(--blue-bg); }
  .ws .labcol h5 { margin: 0; font-size: 14px; } .ws .labcol .cap { font-size: 12.5px; margin: 2px 0 6px; }
  .ws .ctl { display: grid; grid-template-columns: 92px minmax(0, 1fr) 54px; align-items: center; gap: 6px; margin: 3px 0; font-size: 13px; border-radius: 6px; padding: 1px 3px; }
  .ws .ctl input[type=range], .ws .ctl select { width: 100%; min-width: 0; }
  .ws .ctl output { text-align: right; font-weight: bold; }
  .ws .ctl.flash { animation: flash 1.8s ease-out; }
  @keyframes flash { 0% { background: #fde047; } 100% { background: transparent; } }
  .ws .imgbox { position: relative; display: block; line-height: 0; margin-top: 8px; }
  .ws .imgbox img { width: 100%; display: block; border-radius: 6px; border: 1px solid var(--line); }
  .ws .imgbox.flash img { animation: imgflash 1.4s ease-out; }
  @keyframes imgflash { 0% { box-shadow: 0 0 0 4px #fde047; } 100% { box-shadow: 0 0 0 0 transparent; } }
  .ws .mk { position: absolute; transform: translate(-50%, -50%); background: rgba(255,255,255,.92); border: 1px solid #111827;
            border-radius: 10px; font-size: 11px; line-height: 16px; padding: 0 5px; pointer-events: none; font-weight: bold; }
  .ws .mk.cur { background: #fde047; border: 2px solid #b3261e; font-size: 13px; line-height: 18px; z-index: 2; }
  .ws .north { position: absolute; right: 6px; top: 6px; background: rgba(255,255,255,.92); border-radius: 4px; font-size: 11px; line-height: 15px; padding: 1px 4px; font-weight: bold; }
  .ws .info { margin-top: 8px; font-size: 13px; background: var(--soft); border-radius: 6px; padding: 6px 10px; line-height: 1.65; }
  .ws .legend { font-size: 12.5px; }
  .ws .qtabs { display: flex; flex-wrap: wrap; gap: 6px; margin: 6px 0 10px; }
  .ws .qhead { display: flex; align-items: center; gap: 8px; } .ws .qhead .sp { flex: 1; }
  .ws .qtext { font-size: 15px; line-height: 1.6; margin: 4px 0 8px; }
  .ws .split { display: grid; grid-template-columns: minmax(220px, 300px) minmax(0, 1fr); gap: 14px; align-items: start; }
  .ws .main.narrow .split { grid-template-columns: minmax(0, 1fr); }
  .ws .gsdlist .gr { display: flex; align-items: center; gap: 8px; padding: 7px 10px; border: 1px solid var(--line); border-radius: 8px; margin-bottom: 6px; cursor: pointer; background: #fff; }
  .ws .gsdlist .gr.cur { border: 2px solid var(--blue); background: var(--blue-bg); font-weight: bold; }
  .ws .gsdlist .gr.locked { cursor: not-allowed; border-style: dashed; }
  .ws .gsdlist .gr .sp { flex: 1; }
  .ws figure.ch { margin: 0; text-align: center; font-size: 12.5px; }
  .ws figure.ch img { width: 200px; max-width: 100%; border-radius: 6px; border: 1px solid var(--line); display: block; }
  .ws .ecard .imgs { display: flex; gap: 10px; flex-wrap: wrap; }
  .ws .ecard .imgs figure { margin: 0; text-align: center; font-size: 12.5px; }
  .ws .ecard .imgs img { width: 220px; max-width: 100%; border-radius: 6px; border: 1px solid var(--line); display: block; }
  /* M2 */
  .ws .strip { display: flex; flex-wrap: wrap; gap: 5px; margin: 6px 0 10px; }
  .ws .strip button { min-width: 38px; padding: 3px 7px; border-radius: 8px; border: 1px solid #8a94a6; background: #fff; cursor: pointer; font-size: 13px; }
  .ws .strip button.done { background: var(--green-bg); border-color: var(--green); }
  .ws .strip button.cur { background: var(--blue); border-color: var(--blue); font-weight: bold; } .ws .strip button.cur, .ws .strip button.cur * { color: #fff; }
  .ws .lab-l { display: grid; grid-template-columns: minmax(0, 1.3fr) minmax(260px, 1fr); gap: 14px; align-items: start; }
  .ws .main.narrow .lab-l { grid-template-columns: minmax(0, 1fr); }
  .ws .lab-l > .stick { position: sticky; top: 0; z-index: 3; background: #fff; padding-bottom: 4px; }
  .ws .main.narrow .lab-l > .stick { position: static; }
  .ws .ov { position: relative; display: block; line-height: 0; }
  .ws .ov img { width: 100%; border-radius: 8px; border: 1px solid var(--line); }
  .ws .chipbig img { width: 100%; max-width: 300px; border-radius: 8px; border: 1px solid var(--line); display: block; }
  .ws .tiles { display: grid; grid-template-columns: repeat(auto-fill, minmax(170px, 1fr)); gap: 10px; }
  .ws .tile { border: 1px solid var(--line); border-radius: 8px; padding: 10px 12px; }
  .ws .tile .v { font-size: 24px; font-weight: bold; margin: 2px 0; }
  .ws .tile .d { font-size: 12.5px; line-height: 1.45; }
  .ws .acc { border: 1px solid var(--line); border-radius: 10px; margin: 8px 0; background: #fff; }
  .ws .acc .acc-h { display: flex; align-items: center; gap: 8px; padding: 9px 12px; font-weight: bold; }
  .ws .acc .acc-h .sp { flex: 1; }
  .ws .acc.locked .acc-h { border-style: dashed; }
  .ws .acc.open { border: 2px solid var(--blue); box-shadow: 0 0 0 3px var(--blue-bg); }
  .ws .acc .acc-b { padding: 0 12px 12px; }
  .ws .acc .acc-s { padding: 0 12px 10px; font-size: 13px; line-height: 1.55; }
  /* M3 · M4 */
  .ws .post { background: #fff; border: 1px solid var(--line); border-left: 4px solid #e01e5a; border-radius: 8px; padding: 9px 12px; font-size: 14.5px; line-height: 1.6; }
  .ws .kv { display: grid; grid-template-columns: auto 1fr; gap: 3px 12px; font-size: 13.5px; }
  .ws .kv b { white-space: nowrap; }
  .ws .raw { font: 12.5px/1.7 Consolas, 'D2Coding', Menlo, monospace; background: #0f172a; border-radius: 8px; padding: 10px 12px; overflow-x: auto; }
  .ws .raw, .ws .raw * { color: #e2e8f0; } .ws .raw .rk { color: #93c5fd; }
  .ws .jsonbox { font: 12px/1.6 Consolas, 'D2Coding', Menlo, monospace; background: #f8fafc; border: 1px solid var(--line); border-radius: 8px; padding: 10px; white-space: pre; overflow-x: auto; }
  .ws .sunres { background: #fff7ed; border: 1px solid #fdba74; border-radius: 8px; padding: 8px 12px; line-height: 1.7; }
  .ws .calc { display: flex; flex-wrap: wrap; gap: 8px; align-items: flex-end; }
  .ws .calc label { display: flex; flex-direction: column; font-size: 12.5px; font-weight: bold; gap: 2px; }
  .ws .calc input { width: 130px; }
  .ws .casebox { position: relative; line-height: 0; }
  .ws .casebox img { width: 100%; border-radius: 8px; border: 1px solid var(--line); }
  .ws .casebox svg { position: absolute; left: 0; top: 0; width: 100%; height: 100%; pointer-events: none; }
  .ws .casebox.small { width: 58%; flex-shrink: 0; } .ws .casebox.small .mk { font-size: 10px; line-height: 14px; }
  .ws .imgcap { font-size: 12.5px; font-weight: bold; margin: 4px 0; line-height: 1.4; }
  .ws .track { border: 1px solid var(--line); border-radius: 10px; padding: 10px; min-width: 0; }
  .ws .track h5 { margin: 0 0 6px; font-size: 14.5px; }
  .ws .stage-r { border: 1px solid #dfe4ea; border-radius: 8px; padding: 8px; margin: 6px 0; }
  .ws .stage-r.next { border: 2px solid var(--blue); background: #f7faff; }
  .ws .stage-r .sh { display: flex; align-items: center; gap: 8px; font-weight: bold; } .ws .stage-r .sh .sp { flex: 1; }
  .ws .stage-r img { width: 100%; max-height: 300px; object-fit: contain; background: #0f172a; border-radius: 6px; border: 1px solid var(--line); margin-top: 6px; display: block; }
  .ws .stage-r .st { font-size: 13px; line-height: 1.55; margin-top: 4px; }
  .ws .pipe { display: flex; flex-wrap: wrap; align-items: center; gap: 6px; font-size: 13px; margin: 6px 0; }
  .ws .pipe span.box { border: 1px solid var(--blue); background: var(--blue-bg); border-radius: 6px; padding: 3px 8px; font-weight: bold; }
  .ws .pipe span.x { border: 1px dashed var(--red); background: var(--red-bg); border-radius: 6px; padding: 3px 8px; text-decoration: line-through; }
  .ws .ordrow { display: grid; grid-template-columns: minmax(0, 1fr) 130px; gap: 8px; align-items: center; padding: 7px 0; border-bottom: 1px solid #e5e9f0; }
  .ws .ordrow b { display: block; } .ws .ordrow .d { font-size: 12.5px; }
  .ws .chk-chip { display: inline-block; font-size: 12px; border-radius: 10px; padding: 1px 7px; margin: 2px 3px 2px 0; border: 1px solid; }
  .ws .chk-chip.y { background: var(--green-bg); border-color: #9fd5ad; } .ws .chk-chip.n { background: var(--red-bg); border-color: #efa9a3; font-weight: bold; }
  .ws .tilerow { display: grid; grid-template-columns: 250px minmax(0, 1fr) 150px; gap: 12px; align-items: start; padding: 10px 0; border-bottom: 1px solid #e5e9f0; }
  .ws .main.narrow .tilerow { grid-template-columns: minmax(0, 1fr); }
  .ws .tilerow img { width: 100%; max-width: 250px; border-radius: 6px; border: 1px solid var(--line); display: block; cursor: zoom-in; image-rendering: pixelated; }
  .ws .file { border: 1px solid var(--line); border-radius: 8px; padding: 10px 12px; margin-bottom: 10px; }
  .ws .preview { width: 100%; height: 480px; border: 1px solid var(--line); border-radius: 8px; background: #fff; }
  /* 3단계 · 힌트 · 작업 진행 · 그래프 */
  .ws .jobbar { border: 1px solid var(--line); border-radius: 8px; padding: 8px 10px; margin: 8px 0; background: #f8fafc; }
  .ws .jobbar.err { border-color: #efa9a3; background: var(--red-bg); }
  .ws .jobbar .jb-top { display: flex; gap: 8px; align-items: center; font-size: 13px; flex-wrap: wrap; } .ws .jobbar .sp { flex: 1; }
  .ws .jobbar .pb { height: 10px; background: #e3e8ef; border-radius: 6px; overflow: hidden; margin-top: 6px; }
  .ws .jobbar .pb i { display: block; height: 100%; background: linear-gradient(90deg, #0b57d0, #38bdf8); transition: width .5s; }
  .ws .hintbox { border: 1px dashed #d4a72c; background: #fffbea; border-radius: 8px; padding: 6px 10px; margin: 6px 0; font-size: 13px; }
  .ws .hintbox .hb-h { display: flex; align-items: center; gap: 8px; } .ws .hintbox .sp { flex: 1; }
  .ws .hintbox .hb-i { margin-top: 5px; line-height: 1.55; }
  .ws .raw .rl { border-radius: 4px; padding: 0 3px; } .ws .raw .rl.hl { background: #a16207; } .ws .raw .rl.hl, .ws .raw .rl.hl * { color: #fffbea; }
  .ws .srcline { font-size: 12.5px; margin: 2px 0 4px; }
  .ws .srcchip { font: 12px Consolas, 'D2Coding', Menlo, monospace; border: 1px solid var(--blue); background: var(--blue-bg); border-radius: 5px; padding: 1px 6px; cursor: pointer; }
  .ws .convrow { border-top: 1px solid #eef2f7; padding-top: 8px; }
  .ws table.t.conv td, .ws table.t.conv th { font-size: 12.5px; padding: 4px 8px; }
  .ws .params { display: flex; flex-wrap: wrap; gap: 8px 14px; align-items: center; margin: 6px 0; font-size: 13px; }
  .ws .params label { display: flex; gap: 6px; align-items: center; flex-wrap: wrap; }
  .ws .metric { margin-top: 4px; font-size: 13px; background: #ecfdf5; border: 1px solid #a7f3d0; border-radius: 6px; padding: 4px 8px; line-height: 1.5; }
  .ws .chart { margin: 6px 0; flex: 1; min-width: 260px; max-width: 440px; } .ws .chart svg { width: 100%; height: auto; display: block; }
  .ws .card.warnc { border-color: #e3c96b; background: #fffdf3; }
  .ws .grid2 { display: grid; grid-template-columns: repeat(auto-fill, minmax(330px, 1fr)); gap: 12px; }
  .ws .bar { display: inline-block; height: 9px; background: var(--blue); border-radius: 4px; vertical-align: middle; }
  .ws .stress { border-top: 1px solid #e5e9f0; padding-top: 8px; margin-top: 10px; } .ws .stress h5 { margin: 0 0 4px; font-size: 14px; }
  .ws .zoombox { position: absolute; inset: 0; background: rgba(15, 23, 42, .84); z-index: 50; display: flex; align-items: center; justify-content: center; cursor: zoom-out; }
  .ws .zoombox .zb-in { max-width: 94%; max-height: 94%; text-align: center; }
  .ws .zoombox img { max-width: 100%; max-height: 78vh; image-rendering: pixelated; border-radius: 8px; background: #fff; }
  .ws .zoombox .zb-cap, .ws .zoombox .zb-cap * { color: #fff; font-size: 13px; margin-top: 6px; }
  .ws .gal { display: grid; grid-template-columns: repeat(auto-fill, minmax(300px, 1fr)); gap: 10px; margin-top: 6px; }
  .ws .gal-i { display: grid; grid-template-columns: 1fr 1fr; gap: 4px; align-content: start; } .ws .gal-i.one { grid-template-columns: 1fr; }
  .ws .gal-i img { width: 100%; border-radius: 6px; border: 1px solid var(--line); cursor: zoom-in; display: block; }
  .ws .gal-i .hint { grid-column: 1 / -1; }
  .ws .roles { display: grid; grid-template-columns: repeat(auto-fill, minmax(150px, 1fr)); gap: 8px; margin: 6px 0; }
  .ws .role { border-radius: 8px; padding: 8px 10px; border: 1px solid; font-size: 13px; }
  .ws .role.red { background: #fef2f2; border-color: #fca5a5; } .ws .role.blue { background: #eff6ff; border-color: #93c5fd; }
  .ws .cues { display: grid; grid-template-columns: repeat(auto-fill, minmax(260px, 1fr)); gap: 8px; margin: 6px 0; }
  .ws .cue { border: 1px solid var(--line); border-radius: 8px; padding: 8px; background: #fff; }
  .ws .cue .ctl { margin-top: 4px; }
  .ws .cardgrid { display: grid; grid-template-columns: repeat(auto-fill, minmax(210px, 1fr)); gap: 8px; margin-top: 8px; }
  .ws .ccard { border: 1px solid var(--line); border-radius: 8px; padding: 6px; background: #fff; } .ws .ccard.flag { border: 2px solid #dc2626; }
  .ws .ccard img { width: 96px; height: 96px; image-rendering: pixelated; border-radius: 4px; }
  .ws tr.locked-row td { opacity: .55; }
  .ws pre.jsonbox { white-space: pre; }
  /* 모델 들여다보기 · 4단계 */
  .ws .archwrap { overflow-x: auto; border: 1px solid var(--line); border-radius: 8px; background: #f8fafc; padding: 6px; margin: 6px 0; }
  .ws .archwrap svg { display: block; }
  .ws .vnav { position: sticky; top: 0; background: #fff; z-index: 2; padding: 4px 0; } .ws .vnav .sp { flex: 1; }
  .ws .vstep { border-top: 1px solid #e5e9f0; padding-top: 8px; } .ws .vstep h5 { font-size: 15px; margin: 4px 0; }
  .ws pre.formula { background: #0f172a; border-radius: 8px; padding: 8px 10px; white-space: pre-wrap; font: 12.5px/1.6 Consolas, 'D2Coding', Menlo, monospace; }
  .ws pre.formula, .ws pre.formula * { color: #e2e8f0; }
  .ws .example { background: #f0f9ff; border: 1px solid #bae6fd; border-radius: 8px; padding: 8px 10px; margin: 8px 0; }
  .ws table.num3 { border-collapse: collapse; font: 12px Consolas, 'D2Coding', Menlo, monospace; } .ws table.num3 td { border: 1px solid #cbd5e1; padding: 3px 6px; text-align: right; background: #fff; }
  .ws table.num3.w td { background: #fef3c7; }
  .ws .example .big { font-size: 18px; font-weight: bold; } .ws .calcline { margin-top: 6px; font: 13px/1.6 Consolas, 'D2Coding', Menlo, monospace; }
  .ws .fmaps { display: flex; flex-wrap: wrap; gap: 6px; } .ws .fmaps img { width: 64px; image-rendering: pixelated; border-radius: 4px; border: 1px solid var(--line); }
  .ws .legend7 { display: flex; flex-wrap: wrap; gap: 10px; font-size: 12.5px; margin: 6px 0; } .ws .legend7 i { display: inline-block; width: 14px; height: 10px; margin-right: 4px; border-radius: 2px; }
  .ws .maptile { display: grid; grid-template-columns: minmax(0, 300px) minmax(0, 1fr); gap: 10px; border-top: 1px solid #e5e9f0; padding: 8px 0; }
  .ws .main.narrow .maptile { grid-template-columns: minmax(0, 1fr); }
  .ws .maptile img { width: 100%; border-radius: 6px; cursor: zoom-in; } .ws .maptile li { margin: 3px 0; font-size: 13px; line-height: 1.5; }
  .ws .sw { display: inline-block; width: 10px; height: 10px; border-radius: 2px; margin-right: 4px; }
  .ws .adjcard { display: grid; grid-template-columns: 170px minmax(0, 1fr); gap: 10px; border-top: 1px solid #e5e9f0; padding: 10px 0; }
  .ws .main.narrow .adjcard { grid-template-columns: minmax(0, 1fr); }
  .ws .adjcard img, .ws .biascard img { width: 160px; max-width: 100%; border-radius: 6px; image-rendering: pixelated; cursor: zoom-in; }
  .ws .biasgrid { display: grid; grid-template-columns: repeat(auto-fill, minmax(170px, 1fr)); gap: 10px; }
  .ws .biascard { border: 1px solid var(--line); border-radius: 8px; padding: 6px; }
  .ws .flow { display: flex; flex-wrap: wrap; gap: 6px; align-items: center; margin: 6px 0; font-size: 13px; }
  .ws .flow .box { border: 1px solid var(--blue); background: var(--blue-bg); border-radius: 6px; padding: 3px 8px; font-weight: bold; }
  .ws .flow .box.auto { border-color: #9ca3af; background: #f3f4f6; }
  /* 무거운 작업 중 안내 · 가이드북 · 4·5단계 */
  .ws .stay { border: 2px solid #d97706; background: #fffbeb; border-radius: 10px; padding: 10px 12px; margin: 8px 0; font-size: 13.5px; line-height: 1.55; }
  .ws .stay b { color: #92400e; display: block; margin-bottom: 2px; }
  .ws .stay.shake { animation: wsShake .5s; } @keyframes wsShake { 0%, 100% { transform: translateX(0); } 25% { transform: translateX(-6px); } 75% { transform: translateX(6px); } }
  .ws .gtabs { display: flex; gap: 4px; padding: 6px 10px 0; background: #e2e8f0; }
  .ws .gtabs button { flex: 1; border: 1px solid var(--line); border-bottom: none; border-radius: 8px 8px 0 0; background: #f1f5f9; padding: 6px 4px; font-size: 13px; cursor: pointer; color: var(--ink); }
  .ws .gtabs button.on { background: #fff; font-weight: bold; color: var(--blue); }
  .ws .codep .term { margin: 8px 0; } .ws .term.now { border: 2px solid #f59e0b; background: #fffbeb; } .ws .term.flash { box-shadow: 0 0 0 3px #fde68a; }
  .ws .term dt.k4 { color: #6d28d9; } .ws .term dt.k5 { color: #0f766e; } .ws .term dd.fx { font: 12px/1.6 Consolas, 'D2Coding', Menlo, monospace; background: #f5f3ff; border-radius: 6px; padding: 2px 6px; }
  .ws .codep .term dl { grid-template-columns: 70px 1fr; font-size: 12.5px; }
  .ws tr.wrongrow td { background: #fef2f2; }
  .ws .heatwrap { display: grid; grid-template-columns: minmax(0, 1.25fr) minmax(0, 1fr); gap: 12px; align-items: start; margin-top: 8px; }
  .ws .main.narrow .heatwrap { grid-template-columns: minmax(0, 1fr); }
  .ws .heatimg { width: 100%; border-radius: 8px; border: 1px solid var(--line); cursor: zoom-in; }
  .ws details.case { border: 1px solid var(--line); border-radius: 8px; padding: 4px 10px; margin: 8px 0; background: #fbfcfe; }
  .ws details.case > summary { color: var(--ink); }
  .ws .casefig { margin: 4px 8px 4px 0; } .ws .casefig img { width: 200px; max-width: 100%; border-radius: 6px; border: 1px solid var(--line); cursor: zoom-in; image-rendering: pixelated; }
  .ws .casefig figcaption { font-size: 12px; }
  .ws .timeline { display: grid; gap: 3px; margin: 6px 0; font-size: 13px; } .ws .timeline > div { display: grid; grid-template-columns: 70px minmax(0, 1fr); gap: 6px; }
  .ws .voices { border-top: 1px dashed var(--line); margin-top: 6px; padding-top: 6px; }
  .ws .voice { display: flex; gap: 8px; align-items: flex-start; margin: 5px 0; font-size: 13px; line-height: 1.5; }
  .ws .voice .avatar { width: 24px; height: 24px; border-radius: 6px; color: #fff; font-size: 12px; display: flex; align-items: center; justify-content: center; flex-shrink: 0; }
  .ws .actgrp { border-top: 1px solid #e5e9f0; padding: 8px 0; } .ws .actgrp small, .ws .secgrid small { display: block; color: #475569; }
  .ws .flow .fbox { border: 1px solid var(--blue); background: var(--blue-bg); border-radius: 8px; padding: 5px 8px; max-width: 150px; }
  .ws .flow .fbox small { display: block; color: #475569; } .ws .flow .fbox .hint { font-size: 11.5px; } .ws .flow .arr { font-weight: bold; color: var(--blue); }
  .ws .orditem { display: flex; gap: 8px; align-items: center; border: 1px solid var(--line); border-radius: 8px; padding: 6px 8px; margin: 4px 0; background: #fff; }
  .ws .orditem small { color: #475569; } .ws .orditem.wrong { border: 2px solid var(--red); background: #fef2f2; }
  .ws .secgrid { display: grid; grid-template-columns: repeat(auto-fill, minmax(260px, 1fr)); gap: 4px 10px; }
  .ws .lint { display: grid; gap: 4px; margin: 6px 0; } .ws .qa { border-top: 1px dashed var(--line); padding-top: 6px; margin-top: 6px; }

  .ws .gstep { margin: 8px 0 4px; } .ws .gstep-h { font-size: 12px; font-weight: bold; color: #6b21a8; background: #f5f3ff; border-radius: 6px; padding: 3px 8px; margin: 6px 0 4px; }
  .ws .term .fig { margin: 4px 0 6px; } .ws .term .fig svg { width: 100%; max-width: 360px; height: auto; display: block; }
  /* 미션 블록 (명세로 만든 단계 화면) */
  .ws .stepfoot { margin: 14px 0 8px; }
  .ws .blkimg { max-width: 100%; border-radius: 8px; border: 1px solid var(--line); cursor: zoom-in; display: block; margin: 6px 0; }
  /* 글쓰기 블록: 제출한 뒤에 보이는 모범답안 (정답 판정 없음) */
  .ws .model { margin: 10px 0 2px; padding: 10px 12px; border: 1px solid #a8c3f0; border-radius: 8px; background: var(--blue-bg); }
  .ws .model-h { font-weight: bold; color: #0a2a66; margin-bottom: 2px; }
  .ws .model p { margin: 4px 0; line-height: 1.65; }
  /* 첫 화면 (입력 없이 인사만) */
  .ws .onb.welcome { text-align: center; max-width: 560px; margin: 8vh auto 0; }
  .ws .onb.welcome .wave { font-size: 44px; }
  .ws .onb.welcome h2 { font-size: 26px; margin: 8px 0 10px; }
  .ws .onb.welcome .lead { line-height: 1.7; }
  .ws .me-chip { cursor: default; }
  /* 해외 사례 토글 */
  .ws details.case { margin-top: 8px; border: 1px solid var(--line); border-radius: 8px; background: #fff; }
  .ws details.case > summary { cursor: pointer; padding: 7px 10px; font-weight: bold; font-size: 13px; }
  .ws details.case .case-b { padding: 2px 12px 10px; }
  .ws details.case .case-b p { margin: 6px 0; line-height: 1.65; font-size: 13.5px; }
  /* 산출물 단계 */
  .ws ul.todo { margin: 6px 0; padding-left: 20px; }
  .ws ul.todo li { margin: 4px 0; line-height: 1.55; }
  /* 바이브 코딩 실습 */
  .ws .vb-slot { margin: 10px 0; padding: 8px 10px; border: 1px solid var(--line); border-radius: 8px; background: var(--soft); }
  .ws .vb-label { margin: 8px 0 4px; font-size: 13.5px; }
  .ws .vb-slot .vb-label { margin-top: 0; }
  .ws .vb-prompt { background: #fff; border: 2px solid var(--blue); border-radius: 8px; padding: 9px 11px; white-space: pre-wrap; line-height: 1.6; font-family: inherit; font-size: 13.5px; margin: 4px 0 8px; }
  .ws .vb-prompt.empty { border-style: dashed; border-color: #8a94a6; }
  .ws .vb-replybox { scroll-margin-top: 8px; }
  .ws .vb-reply { margin-top: 12px; padding: 10px 12px; border: 1px solid #a8c3f0; border-left: 5px solid var(--blue); border-radius: 8px; background: #f8fbff; }
  .ws .vb-reply.ok { border-color: #9fd5ad; border-left-color: var(--green); background: #f6fbf7; }
  .ws .vb-who { font-weight: bold; margin-bottom: 4px; }
  .ws .vb-say { line-height: 1.6; margin-bottom: 6px; }
  .ws details.vb-codebox { margin: 6px 0; }
  .ws details.vb-codebox > summary { cursor: pointer; font-size: 13px; font-weight: bold; padding: 2px 0; }
  .ws .vb-code { background: #1e2430; color: #e6edf3; border-radius: 8px; padding: 9px 11px; margin: 4px 0; overflow-x: auto; font-family: Menlo, Consolas, monospace; font-size: 12.5px; line-height: 1.5; white-space: pre; }
  .ws .vb-code span { color: #e6edf3; }
  .ws .vb-code .c { color: #8b949e; } .ws .vb-code .s { color: #a5d6ff; } .ws .vb-code .k { color: #ff9e8e; } .ws .vb-code .n { color: #79c0ff; } .ws .vb-code .f { color: #d2a8ff; }
  .ws .vb-out { background: #fff; border: 1px solid var(--line); border-radius: 8px; padding: 8px 10px; margin: 4px 0; overflow-x: auto; font-family: Menlo, Consolas, monospace; font-size: 12.5px; white-space: pre-wrap; }
  .ws .vb-err { background: var(--red-bg); border: 1px solid #e6a8a2; border-radius: 8px; padding: 8px 10px; margin: 4px 0; overflow-x: auto; font-family: Menlo, Consolas, monospace; font-size: 12.5px; white-space: pre-wrap; }
  .ws .vb-tbl { overflow-x: auto; margin: 4px 0; }
  .ws .vb-tbl table { border-collapse: collapse; background: #fff; }
  .ws .vb-tbl th, .ws .vb-tbl td { border: 1px solid var(--line); padding: 3px 8px; font-size: 12.5px; text-align: left; white-space: nowrap; }
  .ws .vb-tbl th { background: #eef2f7; }
  .ws iframe.vb-doc { width: 100%; height: 300px; border: 1px solid var(--line); border-radius: 8px; background: #fff; }
  .ws .vb-problem { margin-top: 8px; padding: 7px 10px; border-radius: 8px; background: var(--red-bg); border: 1px solid #efa9a3; color: #5c0f0a; line-height: 1.6; }
  .ws .vb-problem * { color: inherit; }
  .ws .vb-reply.miss { border-color: #efa9a3; border-left-color: var(--red); background: #fffafa; }
  .ws details.vb-data { margin: 8px 0; }
  .ws details.vb-data > summary, .ws details.vb-log > summary { cursor: pointer; font-size: 13px; font-weight: bold; padding: 2px 0; }
  .ws details.vb-log { margin-top: 10px; }
  .ws details.vb-log th { white-space: nowrap; }
  .ws details.vb-log td:first-child, .ws details.vb-log td:nth-child(2) { white-space: nowrap; }
  .ws .model-t { white-space: pre-wrap; line-height: 1.6; }
  .ws .selfcheck { margin-top: 10px; }
  .ws .selfcheck .choice { display: flex; margin: 4px 0; }
  /* 틀린 선택: 주황색 대신 빨간색으로 (2026-11-05 피드백) */
  .ws .choice.wrong, .ws .choice.on.wrong { border: 2px solid var(--red); background: var(--red-bg); padding: 5px 9px; }
  .ws .choice.wrong input { accent-color: var(--red); }
  .ws .hintbox.wrongbox { border: 1px solid #efa9a3; background: var(--red-bg); color: #5c0f0a; }
  .ws .hintbox.wrongbox * { color: inherit; }
  .ws tr.wrongrow td { background: var(--red-bg); }
  .ws tr.wrongrow select { border: 2px solid var(--red); }
  /* 부제목 */
  .ws .stage-sub { margin: -4px 0 8px; }
  .ws .msub { margin-top: 2px; }
</style>

<div class="ws" id="ws-root">
  <div class="topbar">
    <button class="tb" id="tb-side" onclick="toggleSide()" title="왼쪽 메뉴 접기/펴기">☰<span class="count" id="tb-count"></span></button>
    <span class="name">__WORKSPACE__</span>
    <span class="tprog" id="tb-prog" title="전체 미션 진행률"></span>
    <span class="spacer"></span>
    <span class="me-chip" id="tb-me"></span>
    <button class="tb" id="tb-code" onclick="toggleCode()" title="핵심 용어와 지금 화면의 코드 보기">📘 가이드북</button>
    <span>화면 폭</span>
    <button class="tb" data-w="base" onclick="setWidth('base')">기본</button>
    <button class="tb" data-w="wide" onclick="setWidth('wide')">넓게</button>
    <button class="tb" data-w="full" onclick="setWidth('full')">가득</button>
    <span>높이</span>
    <button class="tb" onclick="setHeight(-120)" title="낮게">−</button>
    <button class="tb" onclick="setHeight(120)" title="높게">+</button>
  </div>
  <div class="body">
    <div class="side" id="side"></div>
    <div class="main" id="main"></div>
    <div class="codep" id="codep"></div>
    <button class="code-tab" id="code-tab" onclick="toggleCode(true)">📘 본 과정의 가이드북 ◀</button>
  </div>
  <div class="footer">
    <span class="note" id="note"></span>
    <button class="btn ghost small" onclick="resetAll()">↺ 처음부터</button>
  </div>
</div>

<script>
// ===================== 프로젝트 설정 (만드는 도구가 넣음) =====================
const PROJ = {"id": "pbl_e0333acf", "team_name": "시설 방호 영상분석반", "ui_key": "pbl-ui-pbl_e0333acf", "boot_note": "🔄 파이썬이 새로 시작돼 진행 상황이 처음으로 돌아갔어요.", "welcome_lead": "가상의 시설 방호 영상분석반에 바이브 코딩 담당으로 합류했어요. 팀원들의 요청을 프롬프트로 바꿔 AI와 함께 도구와 문서를 만들어 봐요. M1부터 차례로 진행해 보세요.", "use_label": "바이브 코딩에서는", "stay_jobs": {}};   // 팀 이름 · 저장 키 · 안내 문구 · 작업 중 잠글 작업
const PROJ_FIGS = {"f_loop": "<svg viewBox=\"0 0 320 156\" xmlns=\"http://www.w3.org/2000/svg\" font-family=\"sans-serif\" font-size=\"10\" fill=\"#111827\"><defs><marker id=\"tfa\" viewBox=\"0 0 10 10\" refX=\"8\" refY=\"5\" markerWidth=\"6\" markerHeight=\"6\" orient=\"auto\"><path d=\"M0,0 L10,5 L0,10 z\" fill=\"#475569\"/><\/marker><\/defs><ellipse cx=\"160.0\" cy=\"70\" rx=\"112\" ry=\"48\" fill=\"none\" stroke=\"#e5e7eb\" stroke-width=\"6\"/><path d=\"M198.6,24.9 L248.1,40.3\" stroke=\"#2563eb\" stroke-width=\"2\" marker-end=\"url(#tfa)\" fill=\"none\"/><path d=\"M271.9,71.8 L253.0,96.7\" stroke=\"#2563eb\" stroke-width=\"2\" marker-end=\"url(#tfa)\" fill=\"none\"/><path d=\"M190.6,116.2 L129.4,116.2\" stroke=\"#2563eb\" stroke-width=\"2\" marker-end=\"url(#tfa)\" fill=\"none\"/><path d=\"M67.0,96.7 L48.1,71.8\" stroke=\"#2563eb\" stroke-width=\"2\" marker-end=\"url(#tfa)\" fill=\"none\"/><path d=\"M71.9,40.3 L121.4,24.9\" stroke=\"#2563eb\" stroke-width=\"2\" marker-end=\"url(#tfa)\" fill=\"none\"/><rect x=\"116.0\" y=\"10.0\" width=\"88.0\" height=\"24.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"160.0\" y=\"25.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">요청<\/text><rect x=\"222.5\" y=\"43.2\" width=\"88.0\" height=\"24.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"266.5\" y=\"58.7\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">생성<\/text><rect x=\"181.8\" y=\"96.8\" width=\"88.0\" height=\"24.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"225.8\" y=\"112.3\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">실행<\/text><rect x=\"50.2\" y=\"96.8\" width=\"88.0\" height=\"24.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"94.2\" y=\"112.3\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">확인<\/text><rect x=\"9.5\" y=\"43.2\" width=\"88.0\" height=\"24.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"53.5\" y=\"58.7\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">수정 요청<\/text><text x=\"160.0\" y=\"74.0\" font-size=\"10.5\" text-anchor=\"middle\" fill=\"#475569\" font-weight=\"bold\">바이브 코딩<\/text><text x=\"160.0\" y=\"151.0\" font-size=\"9\" text-anchor=\"middle\" fill=\"#475569\">확인하고 다시 요청하는 일이 핵심이에요<\/text><\/svg>", "f_four": "<svg viewBox=\"0 0 320 120\" xmlns=\"http://www.w3.org/2000/svg\" font-family=\"sans-serif\" font-size=\"10\" fill=\"#111827\"><defs><marker id=\"tfa\" viewBox=\"0 0 10 10\" refX=\"8\" refY=\"5\" markerWidth=\"6\" markerHeight=\"6\" orient=\"auto\"><path d=\"M0,0 L10,5 L0,10 z\" fill=\"#475569\"/><\/marker><\/defs><rect x=\"6.0\" y=\"8\" width=\"154.0\" height=\"18\" fill=\"#e5e7eb\" stroke=\"#fff\"/><text x=\"83.0\" y=\"20.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">칸<\/text><rect x=\"160.0\" y=\"8\" width=\"154.0\" height=\"18\" fill=\"#e5e7eb\" stroke=\"#fff\"/><text x=\"237.0\" y=\"20.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">무엇을 적나<\/text><rect x=\"6.0\" y=\"26\" width=\"154.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"83.0\" y=\"38.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">맥락<\/text><rect x=\"160.0\" y=\"26\" width=\"154.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"237.0\" y=\"38.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">왜 필요한지<\/text><rect x=\"6.0\" y=\"44\" width=\"154.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"83.0\" y=\"56.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">입력<\/text><rect x=\"160.0\" y=\"44\" width=\"154.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"237.0\" y=\"56.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">파일·열·값<\/text><rect x=\"6.0\" y=\"62\" width=\"154.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"83.0\" y=\"74.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">할 일<\/text><rect x=\"160.0\" y=\"62\" width=\"154.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"237.0\" y=\"74.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">규칙을 숫자로<\/text><rect x=\"6.0\" y=\"80\" width=\"154.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"83.0\" y=\"92.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">출력 형식<\/text><rect x=\"160.0\" y=\"80\" width=\"154.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"237.0\" y=\"92.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">받고 싶은 모양<\/text><text x=\"160.0\" y=\"115.0\" font-size=\"9\" text-anchor=\"middle\" fill=\"#475569\">네 칸을 채우면 AI가 추측할 일이 줄어요<\/text><\/svg>", "f_kinds": "<svg viewBox=\"0 0 320 102\" xmlns=\"http://www.w3.org/2000/svg\" font-family=\"sans-serif\" font-size=\"10\" fill=\"#111827\"><defs><marker id=\"tfa\" viewBox=\"0 0 10 10\" refX=\"8\" refY=\"5\" markerWidth=\"6\" markerHeight=\"6\" orient=\"auto\"><path d=\"M0,0 L10,5 L0,10 z\" fill=\"#475569\"/><\/marker><\/defs><rect x=\"6.0\" y=\"8\" width=\"102.7\" height=\"18\" fill=\"#e5e7eb\" stroke=\"#fff\"/><text x=\"57.3\" y=\"20.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">구분<\/text><rect x=\"108.7\" y=\"8\" width=\"102.7\" height=\"18\" fill=\"#e5e7eb\" stroke=\"#fff\"/><text x=\"160.0\" y=\"20.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">원인<\/text><rect x=\"211.3\" y=\"8\" width=\"102.7\" height=\"18\" fill=\"#e5e7eb\" stroke=\"#fff\"/><text x=\"262.7\" y=\"20.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">대책<\/text><rect x=\"6.0\" y=\"26\" width=\"102.7\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"57.3\" y=\"38.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">경보 대상<\/text><rect x=\"108.7\" y=\"26\" width=\"102.7\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"160.0\" y=\"38.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">사람·차량·드론<\/text><rect x=\"211.3\" y=\"26\" width=\"102.7\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"262.7\" y=\"38.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">확인<\/text><rect x=\"6.0\" y=\"44\" width=\"102.7\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"57.3\" y=\"56.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">불필요 경보<\/text><rect x=\"108.7\" y=\"44\" width=\"102.7\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"160.0\" y=\"56.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">동물·나무·비<\/text><rect x=\"211.3\" y=\"44\" width=\"102.7\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"262.7\" y=\"56.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">영역·임계값<\/text><rect x=\"6.0\" y=\"62\" width=\"102.7\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"57.3\" y=\"74.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">오경보<\/text><rect x=\"108.7\" y=\"62\" width=\"102.7\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"160.0\" y=\"74.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">장비 이상·잡음<\/text><rect x=\"211.3\" y=\"62\" width=\"102.7\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"262.7\" y=\"74.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">장비 점검<\/text><text x=\"160.0\" y=\"97.0\" font-size=\"9\" text-anchor=\"middle\" fill=\"#475569\">원인이 다르면 대책도 달라요<\/text><\/svg>", "f_recall": "<svg viewBox=\"0 0 320 84\" xmlns=\"http://www.w3.org/2000/svg\" font-family=\"sans-serif\" font-size=\"10\" fill=\"#111827\"><defs><marker id=\"tfa\" viewBox=\"0 0 10 10\" refX=\"8\" refY=\"5\" markerWidth=\"6\" markerHeight=\"6\" orient=\"auto\"><path d=\"M0,0 L10,5 L0,10 z\" fill=\"#475569\"/><\/marker><\/defs><rect x=\"6.0\" y=\"8\" width=\"102.7\" height=\"18\" fill=\"#e5e7eb\" stroke=\"#fff\"/><text x=\"57.3\" y=\"20.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">실제 대상<\/text><rect x=\"108.7\" y=\"8\" width=\"102.7\" height=\"18\" fill=\"#e5e7eb\" stroke=\"#fff\"/><text x=\"160.0\" y=\"20.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">잡음<\/text><rect x=\"211.3\" y=\"8\" width=\"102.7\" height=\"18\" fill=\"#e5e7eb\" stroke=\"#fff\"/><text x=\"262.7\" y=\"20.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">재현율<\/text><rect x=\"6.0\" y=\"26\" width=\"102.7\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"57.3\" y=\"38.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">10<\/text><rect x=\"108.7\" y=\"26\" width=\"102.7\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"160.0\" y=\"38.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">9<\/text><rect x=\"211.3\" y=\"26\" width=\"102.7\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"262.7\" y=\"38.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">90%<\/text><rect x=\"6.0\" y=\"44\" width=\"102.7\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"57.3\" y=\"56.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">10<\/text><rect x=\"108.7\" y=\"44\" width=\"102.7\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"160.0\" y=\"56.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">7<\/text><rect x=\"211.3\" y=\"44\" width=\"102.7\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"262.7\" y=\"56.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">70%<\/text><text x=\"160.0\" y=\"79.0\" font-size=\"9\" text-anchor=\"middle\" fill=\"#475569\">실제 대상 가운데 잡은 비율이에요<\/text><\/svg>", "f_delay": "<svg viewBox=\"0 0 320 92\" xmlns=\"http://www.w3.org/2000/svg\" font-family=\"sans-serif\" font-size=\"10\" fill=\"#111827\"><defs><marker id=\"tfa\" viewBox=\"0 0 10 10\" refX=\"8\" refY=\"5\" markerWidth=\"6\" markerHeight=\"6\" orient=\"auto\"><path d=\"M0,0 L10,5 L0,10 z\" fill=\"#475569\"/><\/marker><\/defs><rect x=\"8.0\" y=\"14.0\" width=\"92.0\" height=\"46.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"54.0\" y=\"34.0\" font-size=\"10\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">대상 등장<\/text><text x=\"54.0\" y=\"47.0\" font-size=\"8.5\" text-anchor=\"middle\" fill=\"#475569\">0초<\/text><path d=\"M101.0,37.0 L113.0,37.0\" stroke=\"#475569\" stroke-width=\"1.6\" marker-end=\"url(#tfa)\" fill=\"none\"/><rect x=\"114.0\" y=\"14.0\" width=\"92.0\" height=\"46.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"160.0\" y=\"40.5\" font-size=\"10\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">영상 처리<\/text><path d=\"M207.0,37.0 L219.0,37.0\" stroke=\"#475569\" stroke-width=\"1.6\" marker-end=\"url(#tfa)\" fill=\"none\"/><rect x=\"220.0\" y=\"14.0\" width=\"92.0\" height=\"46.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"266.0\" y=\"34.0\" font-size=\"10\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">경보<\/text><text x=\"266.0\" y=\"47.0\" font-size=\"8.5\" text-anchor=\"middle\" fill=\"#475569\">30초<\/text><text x=\"160.0\" y=\"87.0\" font-size=\"9\" text-anchor=\"middle\" fill=\"#475569\">나타난 뒤 울릴 때까지의 시간이에요<\/text><\/svg>", "f_clean": "<svg viewBox=\"0 0 320 92\" xmlns=\"http://www.w3.org/2000/svg\" font-family=\"sans-serif\" font-size=\"10\" fill=\"#111827\"><defs><marker id=\"tfa\" viewBox=\"0 0 10 10\" refX=\"8\" refY=\"5\" markerWidth=\"6\" markerHeight=\"6\" orient=\"auto\"><path d=\"M0,0 L10,5 L0,10 z\" fill=\"#475569\"/><\/marker><\/defs><rect x=\"8.0\" y=\"14.0\" width=\"92.0\" height=\"46.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"54.0\" y=\"34.0\" font-size=\"10\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">1min<\/text><text x=\"54.0\" y=\"47.0\" font-size=\"8.5\" text-anchor=\"middle\" fill=\"#475569\">분<\/text><path d=\"M101.0,37.0 L113.0,37.0\" stroke=\"#475569\" stroke-width=\"1.6\" marker-end=\"url(#tfa)\" fill=\"none\"/><rect x=\"114.0\" y=\"14.0\" width=\"92.0\" height=\"46.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"160.0\" y=\"34.0\" font-size=\"10\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">60s<\/text><text x=\"160.0\" y=\"47.0\" font-size=\"8.5\" text-anchor=\"middle\" fill=\"#475569\">초<\/text><path d=\"M207.0,37.0 L219.0,37.0\" stroke=\"#475569\" stroke-width=\"1.6\" marker-end=\"url(#tfa)\" fill=\"none\"/><rect x=\"220.0\" y=\"14.0\" width=\"92.0\" height=\"46.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"266.0\" y=\"34.0\" font-size=\"10\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">60<\/text><text x=\"266.0\" y=\"47.0\" font-size=\"8.5\" text-anchor=\"middle\" fill=\"#475569\">초로 통일<\/text><text x=\"160.0\" y=\"87.0\" font-size=\"9\" text-anchor=\"middle\" fill=\"#475569\">제각각인 값을 하나로 맞춰요<\/text><\/svg>", "f_box": "<svg viewBox=\"0 0 320 84\" xmlns=\"http://www.w3.org/2000/svg\" font-family=\"sans-serif\" font-size=\"10\" fill=\"#111827\"><defs><marker id=\"tfa\" viewBox=\"0 0 10 10\" refX=\"8\" refY=\"5\" markerWidth=\"6\" markerHeight=\"6\" orient=\"auto\"><path d=\"M0,0 L10,5 L0,10 z\" fill=\"#475569\"/><\/marker><\/defs><rect x=\"6.0\" y=\"8\" width=\"102.7\" height=\"18\" fill=\"#e5e7eb\" stroke=\"#fff\"/><text x=\"57.3\" y=\"20.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">칸<\/text><rect x=\"108.7\" y=\"8\" width=\"102.7\" height=\"18\" fill=\"#e5e7eb\" stroke=\"#fff\"/><text x=\"160.0\" y=\"20.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">뜻<\/text><rect x=\"211.3\" y=\"8\" width=\"102.7\" height=\"18\" fill=\"#e5e7eb\" stroke=\"#fff\"/><text x=\"262.7\" y=\"20.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">예<\/text><rect x=\"6.0\" y=\"26\" width=\"102.7\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"57.3\" y=\"38.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">x, y<\/text><rect x=\"108.7\" y=\"26\" width=\"102.7\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"160.0\" y=\"38.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">왼쪽 위 좌표<\/text><rect x=\"211.3\" y=\"26\" width=\"102.7\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"262.7\" y=\"38.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">120, 88<\/text><rect x=\"6.0\" y=\"44\" width=\"102.7\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"57.3\" y=\"56.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">w, h<\/text><rect x=\"108.7\" y=\"44\" width=\"102.7\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"160.0\" y=\"56.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">폭과 높이<\/text><rect x=\"211.3\" y=\"44\" width=\"102.7\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"262.7\" y=\"56.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">30, 42<\/text><text x=\"160.0\" y=\"79.0\" font-size=\"9\" text-anchor=\"middle\" fill=\"#475569\">이 네 숫자로 네모를 그려요<\/text><\/svg>", "f_schema": "<svg viewBox=\"0 0 320 102\" xmlns=\"http://www.w3.org/2000/svg\" font-family=\"sans-serif\" font-size=\"10\" fill=\"#111827\"><defs><marker id=\"tfa\" viewBox=\"0 0 10 10\" refX=\"8\" refY=\"5\" markerWidth=\"6\" markerHeight=\"6\" orient=\"auto\"><path d=\"M0,0 L10,5 L0,10 z\" fill=\"#475569\"/><\/marker><\/defs><rect x=\"6.0\" y=\"8\" width=\"154.0\" height=\"18\" fill=\"#e5e7eb\" stroke=\"#fff\"/><text x=\"83.0\" y=\"20.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">칸<\/text><rect x=\"160.0\" y=\"8\" width=\"154.0\" height=\"18\" fill=\"#e5e7eb\" stroke=\"#fff\"/><text x=\"237.0\" y=\"20.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">규칙<\/text><rect x=\"6.0\" y=\"26\" width=\"154.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"83.0\" y=\"38.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">class<\/text><rect x=\"160.0\" y=\"26\" width=\"154.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"237.0\" y=\"38.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">네 가지 소문자<\/text><rect x=\"6.0\" y=\"44\" width=\"154.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"83.0\" y=\"56.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">w, h<\/text><rect x=\"160.0\" y=\"44\" width=\"154.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"237.0\" y=\"56.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">0보다 큼<\/text><rect x=\"6.0\" y=\"62\" width=\"154.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"83.0\" y=\"74.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">x+w<\/text><rect x=\"160.0\" y=\"62\" width=\"154.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"237.0\" y=\"74.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">화면 안<\/text><text x=\"160.0\" y=\"97.0\" font-size=\"9\" text-anchor=\"middle\" fill=\"#475569\">규격을 조건으로 적어요<\/text><\/svg>", "f_debug": "<svg viewBox=\"0 0 320 92\" xmlns=\"http://www.w3.org/2000/svg\" font-family=\"sans-serif\" font-size=\"10\" fill=\"#111827\"><defs><marker id=\"tfa\" viewBox=\"0 0 10 10\" refX=\"8\" refY=\"5\" markerWidth=\"6\" markerHeight=\"6\" orient=\"auto\"><path d=\"M0,0 L10,5 L0,10 z\" fill=\"#475569\"/><\/marker><\/defs><rect x=\"8.0\" y=\"14.0\" width=\"92.0\" height=\"46.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"54.0\" y=\"40.5\" font-size=\"9.5\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">오류 메시지<\/text><path d=\"M101.0,37.0 L113.0,37.0\" stroke=\"#475569\" stroke-width=\"1.6\" marker-end=\"url(#tfa)\" fill=\"none\"/><rect x=\"114.0\" y=\"14.0\" width=\"92.0\" height=\"46.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"160.0\" y=\"40.5\" font-size=\"9.5\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">원하는 동작<\/text><path d=\"M207.0,37.0 L219.0,37.0\" stroke=\"#475569\" stroke-width=\"1.6\" marker-end=\"url(#tfa)\" fill=\"none\"/><rect x=\"220.0\" y=\"14.0\" width=\"92.0\" height=\"46.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"266.0\" y=\"40.5\" font-size=\"10\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">고칠 범위<\/text><text x=\"160.0\" y=\"87.0\" font-size=\"9\" text-anchor=\"middle\" fill=\"#475569\">이 세 가지를 담아 다시 요청해요<\/text><\/svg>", "f_roi": "<svg viewBox=\"0 0 320 102\" xmlns=\"http://www.w3.org/2000/svg\" font-family=\"sans-serif\" font-size=\"10\" fill=\"#111827\"><defs><marker id=\"tfa\" viewBox=\"0 0 10 10\" refX=\"8\" refY=\"5\" markerWidth=\"6\" markerHeight=\"6\" orient=\"auto\"><path d=\"M0,0 L10,5 L0,10 z\" fill=\"#475569\"/><\/marker><\/defs><rect x=\"6.0\" y=\"8\" width=\"102.7\" height=\"18\" fill=\"#e5e7eb\" stroke=\"#fff\"/><text x=\"57.3\" y=\"20.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">구역<\/text><rect x=\"108.7\" y=\"8\" width=\"102.7\" height=\"18\" fill=\"#e5e7eb\" stroke=\"#fff\"/><text x=\"160.0\" y=\"20.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">y 범위<\/text><rect x=\"211.3\" y=\"8\" width=\"102.7\" height=\"18\" fill=\"#e5e7eb\" stroke=\"#fff\"/><text x=\"262.7\" y=\"20.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">처리<\/text><rect x=\"6.0\" y=\"26\" width=\"102.7\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"57.3\" y=\"38.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">영역 B<\/text><rect x=\"108.7\" y=\"26\" width=\"102.7\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"160.0\" y=\"38.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">0~15<\/text><rect x=\"211.3\" y=\"26\" width=\"102.7\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"262.7\" y=\"38.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">경보<\/text><rect x=\"6.0\" y=\"44\" width=\"102.7\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"57.3\" y=\"56.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">영역 A<\/text><rect x=\"108.7\" y=\"44\" width=\"102.7\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"160.0\" y=\"56.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">0~30<\/text><rect x=\"211.3\" y=\"44\" width=\"102.7\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"262.7\" y=\"56.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">경보<\/text><rect x=\"6.0\" y=\"62\" width=\"102.7\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"57.3\" y=\"74.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">제외 영역<\/text><rect x=\"108.7\" y=\"62\" width=\"102.7\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"160.0\" y=\"74.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">나무 쪽<\/text><rect x=\"211.3\" y=\"62\" width=\"102.7\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"262.7\" y=\"74.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">경보 안 함<\/text><text x=\"160.0\" y=\"97.0\" font-size=\"9\" text-anchor=\"middle\" fill=\"#475569\">어디서 경보를 올릴지 좌표로 정해요<\/text><\/svg>", "f_pre": "<svg viewBox=\"0 0 320 92\" xmlns=\"http://www.w3.org/2000/svg\" font-family=\"sans-serif\" font-size=\"10\" fill=\"#111827\"><defs><marker id=\"tfa\" viewBox=\"0 0 10 10\" refX=\"8\" refY=\"5\" markerWidth=\"6\" markerHeight=\"6\" orient=\"auto\"><path d=\"M0,0 L10,5 L0,10 z\" fill=\"#475569\"/><\/marker><\/defs><rect x=\"8.0\" y=\"14.0\" width=\"92.0\" height=\"46.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"54.0\" y=\"40.5\" font-size=\"9.5\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">대비 높이기<\/text><path d=\"M101.0,37.0 L113.0,37.0\" stroke=\"#475569\" stroke-width=\"1.6\" marker-end=\"url(#tfa)\" fill=\"none\"/><rect x=\"114.0\" y=\"14.0\" width=\"92.0\" height=\"46.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"160.0\" y=\"40.5\" font-size=\"9.5\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">잡음 지우기<\/text><path d=\"M207.0,37.0 L219.0,37.0\" stroke=\"#475569\" stroke-width=\"1.6\" marker-end=\"url(#tfa)\" fill=\"none\"/><rect x=\"220.0\" y=\"14.0\" width=\"92.0\" height=\"46.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"266.0\" y=\"40.5\" font-size=\"9.5\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">움직임 찾기<\/text><text x=\"160.0\" y=\"87.0\" font-size=\"9\" text-anchor=\"middle\" fill=\"#475569\">영상을 손본 뒤 움직임을 찾아요<\/text><\/svg>", "f_param": "<svg viewBox=\"0 0 320 92\" xmlns=\"http://www.w3.org/2000/svg\" font-family=\"sans-serif\" font-size=\"10\" fill=\"#111827\"><defs><marker id=\"tfa\" viewBox=\"0 0 10 10\" refX=\"8\" refY=\"5\" markerWidth=\"6\" markerHeight=\"6\" orient=\"auto\"><path d=\"M0,0 L10,5 L0,10 z\" fill=\"#475569\"/><\/marker><\/defs><rect x=\"8.0\" y=\"14.0\" width=\"65.5\" height=\"46.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"40.8\" y=\"40.5\" font-size=\"10\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">작은 함수<\/text><path d=\"M74.5,37.0 L86.5,37.0\" stroke=\"#475569\" stroke-width=\"1.6\" marker-end=\"url(#tfa)\" fill=\"none\"/><rect x=\"87.5\" y=\"14.0\" width=\"65.5\" height=\"46.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"120.2\" y=\"40.5\" font-size=\"9.5\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">예시로 확인<\/text><path d=\"M154.0,37.0 L166.0,37.0\" stroke=\"#475569\" stroke-width=\"1.6\" marker-end=\"url(#tfa)\" fill=\"none\"/><rect x=\"167.0\" y=\"14.0\" width=\"65.5\" height=\"46.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"199.8\" y=\"40.5\" font-size=\"10\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">전체 적용<\/text><path d=\"M233.5,37.0 L245.5,37.0\" stroke=\"#475569\" stroke-width=\"1.6\" marker-end=\"url(#tfa)\" fill=\"none\"/><rect x=\"246.5\" y=\"14.0\" width=\"65.5\" height=\"46.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"279.2\" y=\"40.5\" font-size=\"10\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">비교표<\/text><text x=\"160.0\" y=\"87.0\" font-size=\"9\" text-anchor=\"middle\" fill=\"#475569\">작게 만들고 확인한 뒤 넓혀요<\/text><\/svg>", "f_bg": "<svg viewBox=\"0 0 320 92\" xmlns=\"http://www.w3.org/2000/svg\" font-family=\"sans-serif\" font-size=\"10\" fill=\"#111827\"><defs><marker id=\"tfa\" viewBox=\"0 0 10 10\" refX=\"8\" refY=\"5\" markerWidth=\"6\" markerHeight=\"6\" orient=\"auto\"><path d=\"M0,0 L10,5 L0,10 z\" fill=\"#475569\"/><\/marker><\/defs><rect x=\"8.0\" y=\"14.0\" width=\"92.0\" height=\"46.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"54.0\" y=\"34.0\" font-size=\"10\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">배경<\/text><text x=\"54.0\" y=\"47.0\" font-size=\"8.5\" text-anchor=\"middle\" fill=\"#475569\">평소 모습<\/text><path d=\"M101.0,37.0 L113.0,37.0\" stroke=\"#475569\" stroke-width=\"1.6\" marker-end=\"url(#tfa)\" fill=\"none\"/><rect x=\"114.0\" y=\"14.0\" width=\"92.0\" height=\"46.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"160.0\" y=\"40.5\" font-size=\"10\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">지금 영상<\/text><path d=\"M207.0,37.0 L219.0,37.0\" stroke=\"#475569\" stroke-width=\"1.6\" marker-end=\"url(#tfa)\" fill=\"none\"/><rect x=\"220.0\" y=\"14.0\" width=\"92.0\" height=\"46.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"266.0\" y=\"34.0\" font-size=\"10\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">달라진 곳<\/text><text x=\"266.0\" y=\"47.0\" font-size=\"8.5\" text-anchor=\"middle\" fill=\"#475569\">움직임<\/text><text x=\"160.0\" y=\"87.0\" font-size=\"9\" text-anchor=\"middle\" fill=\"#475569\">평소와 달라진 곳을 찾아요<\/text><\/svg>", "f_tl": "<svg viewBox=\"0 0 320 92\" xmlns=\"http://www.w3.org/2000/svg\" font-family=\"sans-serif\" font-size=\"10\" fill=\"#111827\"><defs><marker id=\"tfa\" viewBox=\"0 0 10 10\" refX=\"8\" refY=\"5\" markerWidth=\"6\" markerHeight=\"6\" orient=\"auto\"><path d=\"M0,0 L10,5 L0,10 z\" fill=\"#475569\"/><\/marker><\/defs><rect x=\"8.0\" y=\"14.0\" width=\"92.0\" height=\"46.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"54.0\" y=\"34.0\" font-size=\"10\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">큰 모델<\/text><text x=\"54.0\" y=\"47.0\" font-size=\"8.5\" text-anchor=\"middle\" fill=\"#475569\">이미 학습됨<\/text><path d=\"M101.0,37.0 L113.0,37.0\" stroke=\"#475569\" stroke-width=\"1.6\" marker-end=\"url(#tfa)\" fill=\"none\"/><rect x=\"114.0\" y=\"14.0\" width=\"92.0\" height=\"46.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"160.0\" y=\"40.5\" font-size=\"9.5\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">가져와 쓰기<\/text><path d=\"M207.0,37.0 L219.0,37.0\" stroke=\"#475569\" stroke-width=\"1.6\" marker-end=\"url(#tfa)\" fill=\"none\"/><rect x=\"220.0\" y=\"14.0\" width=\"92.0\" height=\"46.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"266.0\" y=\"34.0\" font-size=\"10\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">결과 확인<\/text><text x=\"266.0\" y=\"47.0\" font-size=\"8.5\" text-anchor=\"middle\" fill=\"#475569\">우리 조건에서<\/text><text x=\"160.0\" y=\"87.0\" font-size=\"9\" text-anchor=\"middle\" fill=\"#475569\">이미 배운 것을 빌려 와요<\/text><\/svg>", "f_iou": "<svg viewBox=\"0 0 320 92\" xmlns=\"http://www.w3.org/2000/svg\" font-family=\"sans-serif\" font-size=\"10\" fill=\"#111827\"><defs><marker id=\"tfa\" viewBox=\"0 0 10 10\" refX=\"8\" refY=\"5\" markerWidth=\"6\" markerHeight=\"6\" orient=\"auto\"><path d=\"M0,0 L10,5 L0,10 z\" fill=\"#475569\"/><\/marker><\/defs><rect x=\"8.0\" y=\"14.0\" width=\"92.0\" height=\"46.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"54.0\" y=\"34.0\" font-size=\"10\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">겹친 넓이<\/text><text x=\"54.0\" y=\"47.0\" font-size=\"8.5\" text-anchor=\"middle\" fill=\"#475569\">200<\/text><path d=\"M101.0,37.0 L113.0,37.0\" stroke=\"#475569\" stroke-width=\"1.6\" marker-end=\"url(#tfa)\" fill=\"none\"/><rect x=\"114.0\" y=\"14.0\" width=\"92.0\" height=\"46.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"160.0\" y=\"34.0\" font-size=\"10\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">합친 넓이<\/text><text x=\"160.0\" y=\"47.0\" font-size=\"8.5\" text-anchor=\"middle\" fill=\"#475569\">600<\/text><path d=\"M207.0,37.0 L219.0,37.0\" stroke=\"#475569\" stroke-width=\"1.6\" marker-end=\"url(#tfa)\" fill=\"none\"/><rect x=\"220.0\" y=\"14.0\" width=\"92.0\" height=\"46.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"266.0\" y=\"34.0\" font-size=\"10\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">IoU<\/text><text x=\"266.0\" y=\"47.0\" font-size=\"8.5\" text-anchor=\"middle\" fill=\"#475569\">약 0.33<\/text><text x=\"160.0\" y=\"87.0\" font-size=\"9\" text-anchor=\"middle\" fill=\"#475569\">겹친 넓이 ÷ 합친 넓이예요<\/text><\/svg>", "f_verify": "<svg viewBox=\"0 0 320 92\" xmlns=\"http://www.w3.org/2000/svg\" font-family=\"sans-serif\" font-size=\"10\" fill=\"#111827\"><defs><marker id=\"tfa\" viewBox=\"0 0 10 10\" refX=\"8\" refY=\"5\" markerWidth=\"6\" markerHeight=\"6\" orient=\"auto\"><path d=\"M0,0 L10,5 L0,10 z\" fill=\"#475569\"/><\/marker><\/defs><rect x=\"8.0\" y=\"14.0\" width=\"65.5\" height=\"46.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"40.8\" y=\"40.5\" font-size=\"10\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">AI 답변<\/text><path d=\"M74.5,37.0 L86.5,37.0\" stroke=\"#475569\" stroke-width=\"1.6\" marker-end=\"url(#tfa)\" fill=\"none\"/><rect x=\"87.5\" y=\"14.0\" width=\"65.5\" height=\"46.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"120.2\" y=\"40.5\" font-size=\"10\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">근거 수치<\/text><path d=\"M154.0,37.0 L166.0,37.0\" stroke=\"#475569\" stroke-width=\"1.6\" marker-end=\"url(#tfa)\" fill=\"none\"/><rect x=\"167.0\" y=\"14.0\" width=\"65.5\" height=\"46.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"199.8\" y=\"40.5\" font-size=\"10\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">작은 예시<\/text><path d=\"M233.5,37.0 L245.5,37.0\" stroke=\"#475569\" stroke-width=\"1.6\" marker-end=\"url(#tfa)\" fill=\"none\"/><rect x=\"246.5\" y=\"14.0\" width=\"65.5\" height=\"46.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"279.2\" y=\"40.5\" font-size=\"10\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">판단<\/text><text x=\"160.0\" y=\"87.0\" font-size=\"9\" text-anchor=\"middle\" fill=\"#475569\">답변만 믿지 말고 근거를 함께 받아요<\/text><\/svg>", "f_op": "<svg viewBox=\"0 0 320 150\" xmlns=\"http://www.w3.org/2000/svg\" font-family=\"sans-serif\" font-size=\"10\" fill=\"#111827\"><defs><marker id=\"tfa\" viewBox=\"0 0 10 10\" refX=\"8\" refY=\"5\" markerWidth=\"6\" markerHeight=\"6\" orient=\"auto\"><path d=\"M0,0 L10,5 L0,10 z\" fill=\"#475569\"/><\/marker><\/defs><rect x=\"34\" y=\"12\" width=\"274\" height=\"108\" fill=\"#fff\" stroke=\"#94a3b8\"/><polyline points=\"51.1,60.6 76.8,42.2 119.6,25.0 188.1,18.5 290.9,14.2\" fill=\"none\" stroke=\"#2563eb\" stroke-width=\"2\"/><circle cx=\"51.1\" cy=\"60.6\" r=\"2.5\" fill=\"#2563eb\"/><circle cx=\"76.8\" cy=\"42.2\" r=\"2.5\" fill=\"#2563eb\"/><circle cx=\"119.6\" cy=\"25.0\" r=\"2.5\" fill=\"#2563eb\"/><circle cx=\"188.1\" cy=\"18.5\" r=\"2.5\" fill=\"#2563eb\"/><circle cx=\"290.9\" cy=\"14.2\" r=\"2.5\" fill=\"#2563eb\"/><circle cx=\"119.6\" cy=\"25.0\" r=\"6\" fill=\"none\" stroke=\"#f97316\" stroke-width=\"2.5\"/><text x=\"128.6\" y=\"29.0\" font-size=\"9\" text-anchor=\"start\" fill=\"#f97316\" font-weight=\"bold\">운용점<\/text><text x=\"171.0\" y=\"142.0\" font-size=\"9\" text-anchor=\"middle\" fill=\"#475569\">nuisance per day<\/text><text x=\"10\" y=\"66.0\" font-size=\"9\" text-anchor=\"middle\" fill=\"#475569\" transform=\"rotate(-90 10 66.0)\">recall<\/text><text x=\"40.0\" y=\"24.0\" font-size=\"8.5\" text-anchor=\"start\" fill=\"#475569\">어디에 둘지 사람이 합의해요<\/text><\/svg>", "f_cut": "<svg viewBox=\"0 0 320 92\" xmlns=\"http://www.w3.org/2000/svg\" font-family=\"sans-serif\" font-size=\"10\" fill=\"#111827\"><defs><marker id=\"tfa\" viewBox=\"0 0 10 10\" refX=\"8\" refY=\"5\" markerWidth=\"6\" markerHeight=\"6\" orient=\"auto\"><path d=\"M0,0 L10,5 L0,10 z\" fill=\"#475569\"/><\/marker><\/defs><rect x=\"8.0\" y=\"14.0\" width=\"92.0\" height=\"46.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"54.0\" y=\"34.0\" font-size=\"9.5\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">점수 0.62<\/text><text x=\"54.0\" y=\"47.0\" font-size=\"8.5\" text-anchor=\"middle\" fill=\"#475569\">경보<\/text><path d=\"M101.0,37.0 L113.0,37.0\" stroke=\"#475569\" stroke-width=\"1.6\" marker-end=\"url(#tfa)\" fill=\"none\"/><rect x=\"114.0\" y=\"14.0\" width=\"92.0\" height=\"46.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"160.0\" y=\"34.0\" font-size=\"9.5\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">임계값 0.5<\/text><text x=\"160.0\" y=\"47.0\" font-size=\"8.5\" text-anchor=\"middle\" fill=\"#475569\">커트라인<\/text><path d=\"M207.0,37.0 L219.0,37.0\" stroke=\"#475569\" stroke-width=\"1.6\" marker-end=\"url(#tfa)\" fill=\"none\"/><rect x=\"220.0\" y=\"14.0\" width=\"92.0\" height=\"46.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"266.0\" y=\"34.0\" font-size=\"9.5\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">점수 0.41<\/text><text x=\"266.0\" y=\"47.0\" font-size=\"8.5\" text-anchor=\"middle\" fill=\"#475569\">경보 아님<\/text><text x=\"160.0\" y=\"87.0\" font-size=\"9\" text-anchor=\"middle\" fill=\"#475569\">이상인지 초과인지도 적어요<\/text><\/svg>", "f_edge": "<svg viewBox=\"0 0 320 102\" xmlns=\"http://www.w3.org/2000/svg\" font-family=\"sans-serif\" font-size=\"10\" fill=\"#111827\"><defs><marker id=\"tfa\" viewBox=\"0 0 10 10\" refX=\"8\" refY=\"5\" markerWidth=\"6\" markerHeight=\"6\" orient=\"auto\"><path d=\"M0,0 L10,5 L0,10 z\" fill=\"#475569\"/><\/marker><\/defs><rect x=\"6.0\" y=\"8\" width=\"154.0\" height=\"18\" fill=\"#e5e7eb\" stroke=\"#fff\"/><text x=\"83.0\" y=\"20.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">score<\/text><rect x=\"160.0\" y=\"8\" width=\"154.0\" height=\"18\" fill=\"#e5e7eb\" stroke=\"#fff\"/><text x=\"237.0\" y=\"20.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">처리<\/text><rect x=\"6.0\" y=\"26\" width=\"154.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"83.0\" y=\"38.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">0.62<\/text><rect x=\"160.0\" y=\"26\" width=\"154.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"237.0\" y=\"38.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">임계값과 비교<\/text><rect x=\"6.0\" y=\"44\" width=\"154.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"83.0\" y=\"56.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">0.50<\/text><rect x=\"160.0\" y=\"44\" width=\"154.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"237.0\" y=\"56.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">이상이면 경보<\/text><rect x=\"6.0\" y=\"62\" width=\"154.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"83.0\" y=\"74.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">빈칸<\/text><rect x=\"160.0\" y=\"62\" width=\"154.0\" height=\"18\" fill=\"#fef3c7\" stroke=\"#e5e7eb\"/><text x=\"237.0\" y=\"74.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">영상 손실: 사람 확인<\/text><text x=\"160.0\" y=\"97.0\" font-size=\"9\" text-anchor=\"middle\" fill=\"#475569\">빈칸을 어떻게 할지 미리 알려 주세요<\/text><\/svg>", "f_track": "<svg viewBox=\"0 0 320 84\" xmlns=\"http://www.w3.org/2000/svg\" font-family=\"sans-serif\" font-size=\"10\" fill=\"#111827\"><defs><marker id=\"tfa\" viewBox=\"0 0 10 10\" refX=\"8\" refY=\"5\" markerWidth=\"6\" markerHeight=\"6\" orient=\"auto\"><path d=\"M0,0 L10,5 L0,10 z\" fill=\"#475569\"/><\/marker><\/defs><rect x=\"6.0\" y=\"8\" width=\"102.7\" height=\"18\" fill=\"#e5e7eb\" stroke=\"#fff\"/><text x=\"57.3\" y=\"20.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">특징<\/text><rect x=\"108.7\" y=\"8\" width=\"102.7\" height=\"18\" fill=\"#e5e7eb\" stroke=\"#fff\"/><text x=\"160.0\" y=\"20.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">드론<\/text><rect x=\"211.3\" y=\"8\" width=\"102.7\" height=\"18\" fill=\"#e5e7eb\" stroke=\"#fff\"/><text x=\"262.7\" y=\"20.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">새<\/text><rect x=\"6.0\" y=\"26\" width=\"102.7\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"57.3\" y=\"38.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">정지 비율<\/text><rect x=\"108.7\" y=\"26\" width=\"102.7\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"160.0\" y=\"38.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">높을 수 있음<\/text><rect x=\"211.3\" y=\"26\" width=\"102.7\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"262.7\" y=\"38.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">낮음<\/text><rect x=\"6.0\" y=\"44\" width=\"102.7\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"57.3\" y=\"56.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">속도 변화<\/text><rect x=\"108.7\" y=\"44\" width=\"102.7\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"160.0\" y=\"56.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">일정한 편<\/text><rect x=\"211.3\" y=\"44\" width=\"102.7\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"262.7\" y=\"56.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">큰 편<\/text><text x=\"160.0\" y=\"79.0\" font-size=\"9\" text-anchor=\"middle\" fill=\"#475569\">움직이는 버릇이 단서가 돼요<\/text><\/svg>", "f_leak": "<svg viewBox=\"0 0 320 92\" xmlns=\"http://www.w3.org/2000/svg\" font-family=\"sans-serif\" font-size=\"10\" fill=\"#111827\"><defs><marker id=\"tfa\" viewBox=\"0 0 10 10\" refX=\"8\" refY=\"5\" markerWidth=\"6\" markerHeight=\"6\" orient=\"auto\"><path d=\"M0,0 L10,5 L0,10 z\" fill=\"#475569\"/><\/marker><\/defs><rect x=\"8.0\" y=\"14.0\" width=\"145.0\" height=\"46.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"80.5\" y=\"34.0\" font-size=\"10\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">학습<\/text><text x=\"80.5\" y=\"47.0\" font-size=\"8.5\" text-anchor=\"middle\" fill=\"#475569\">클립 A, B<\/text><path d=\"M154.0,37.0 L166.0,37.0\" stroke=\"#475569\" stroke-width=\"1.6\" marker-end=\"url(#tfa)\" fill=\"none\"/><rect x=\"167.0\" y=\"14.0\" width=\"145.0\" height=\"46.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"239.5\" y=\"34.0\" font-size=\"10\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">테스트<\/text><text x=\"239.5\" y=\"47.0\" font-size=\"8.5\" text-anchor=\"middle\" fill=\"#475569\">클립 C<\/text><text x=\"160.0\" y=\"87.0\" font-size=\"9\" text-anchor=\"middle\" fill=\"#475569\">같은 클립을 양쪽에 두지 않아요<\/text><\/svg>", "f_split": "<svg viewBox=\"0 0 320 84\" xmlns=\"http://www.w3.org/2000/svg\" font-family=\"sans-serif\" font-size=\"10\" fill=\"#111827\"><defs><marker id=\"tfa\" viewBox=\"0 0 10 10\" refX=\"8\" refY=\"5\" markerWidth=\"6\" markerHeight=\"6\" orient=\"auto\"><path d=\"M0,0 L10,5 L0,10 z\" fill=\"#475569\"/><\/marker><\/defs><rect x=\"6.0\" y=\"8\" width=\"102.7\" height=\"18\" fill=\"#e5e7eb\" stroke=\"#fff\"/><text x=\"57.3\" y=\"20.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">방법<\/text><rect x=\"108.7\" y=\"8\" width=\"102.7\" height=\"18\" fill=\"#e5e7eb\" stroke=\"#fff\"/><text x=\"160.0\" y=\"20.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">클립 A 궤적<\/text><rect x=\"211.3\" y=\"8\" width=\"102.7\" height=\"18\" fill=\"#e5e7eb\" stroke=\"#fff\"/><text x=\"262.7\" y=\"20.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">수치<\/text><rect x=\"6.0\" y=\"26\" width=\"102.7\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"57.3\" y=\"38.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">무작위 분할<\/text><rect x=\"108.7\" y=\"26\" width=\"102.7\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"160.0\" y=\"38.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">학습과 테스트에 섞임<\/text><rect x=\"211.3\" y=\"26\" width=\"102.7\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"262.7\" y=\"38.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">높게 나옴<\/text><rect x=\"6.0\" y=\"44\" width=\"102.7\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"57.3\" y=\"56.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">클립 단위 분할<\/text><rect x=\"108.7\" y=\"44\" width=\"102.7\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"160.0\" y=\"56.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">한쪽에만 있음<\/text><rect x=\"211.3\" y=\"44\" width=\"102.7\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"262.7\" y=\"56.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">실제에 가까움<\/text><text x=\"160.0\" y=\"79.0\" font-size=\"9\" text-anchor=\"middle\" fill=\"#475569\">나누는 방법이 수치를 바꿔요<\/text><\/svg>", "f_halluc": "<svg viewBox=\"0 0 320 92\" xmlns=\"http://www.w3.org/2000/svg\" font-family=\"sans-serif\" font-size=\"10\" fill=\"#111827\"><defs><marker id=\"tfa\" viewBox=\"0 0 10 10\" refX=\"8\" refY=\"5\" markerWidth=\"6\" markerHeight=\"6\" orient=\"auto\"><path d=\"M0,0 L10,5 L0,10 z\" fill=\"#475569\"/><\/marker><\/defs><rect x=\"8.0\" y=\"14.0\" width=\"92.0\" height=\"46.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"54.0\" y=\"40.5\" font-size=\"9.5\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">자료에 없음<\/text><path d=\"M101.0,37.0 L113.0,37.0\" stroke=\"#475569\" stroke-width=\"1.6\" marker-end=\"url(#tfa)\" fill=\"none\"/><rect x=\"114.0\" y=\"14.0\" width=\"92.0\" height=\"46.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"160.0\" y=\"34.0\" font-size=\"9.5\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">AI가 지어냄<\/text><text x=\"160.0\" y=\"47.0\" font-size=\"8.5\" text-anchor=\"middle\" fill=\"#475569\">그럴듯함<\/text><path d=\"M207.0,37.0 L219.0,37.0\" stroke=\"#475569\" stroke-width=\"1.6\" marker-end=\"url(#tfa)\" fill=\"none\"/><rect x=\"220.0\" y=\"14.0\" width=\"92.0\" height=\"46.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"266.0\" y=\"34.0\" font-size=\"10\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">사실 확인<\/text><text x=\"266.0\" y=\"47.0\" font-size=\"8.5\" text-anchor=\"middle\" fill=\"#475569\">자료와 대조<\/text><text x=\"160.0\" y=\"87.0\" font-size=\"9\" text-anchor=\"middle\" fill=\"#475569\">없는 내용은 ‘자료 없음’으로 쓰게 해요<\/text><\/svg>", "f_hitl": "<svg viewBox=\"0 0 320 92\" xmlns=\"http://www.w3.org/2000/svg\" font-family=\"sans-serif\" font-size=\"10\" fill=\"#111827\"><defs><marker id=\"tfa\" viewBox=\"0 0 10 10\" refX=\"8\" refY=\"5\" markerWidth=\"6\" markerHeight=\"6\" orient=\"auto\"><path d=\"M0,0 L10,5 L0,10 z\" fill=\"#475569\"/><\/marker><\/defs><rect x=\"8.0\" y=\"14.0\" width=\"92.0\" height=\"46.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"54.0\" y=\"40.5\" font-size=\"10\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">AI 제안<\/text><path d=\"M101.0,37.0 L113.0,37.0\" stroke=\"#475569\" stroke-width=\"1.6\" marker-end=\"url(#tfa)\" fill=\"none\"/><rect x=\"114.0\" y=\"14.0\" width=\"92.0\" height=\"46.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"160.0\" y=\"40.5\" font-size=\"10\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">사람 확인<\/text><path d=\"M207.0,37.0 L219.0,37.0\" stroke=\"#475569\" stroke-width=\"1.6\" marker-end=\"url(#tfa)\" fill=\"none\"/><rect x=\"220.0\" y=\"14.0\" width=\"92.0\" height=\"46.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"266.0\" y=\"40.5\" font-size=\"10\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">사람 승인<\/text><text x=\"160.0\" y=\"87.0\" font-size=\"9\" text-anchor=\"middle\" fill=\"#475569\">AI는 제안, 결정은 사람이에요<\/text><\/svg>", "f_bias": "<svg viewBox=\"0 0 320 92\" xmlns=\"http://www.w3.org/2000/svg\" font-family=\"sans-serif\" font-size=\"10\" fill=\"#111827\"><defs><marker id=\"tfa\" viewBox=\"0 0 10 10\" refX=\"8\" refY=\"5\" markerWidth=\"6\" markerHeight=\"6\" orient=\"auto\"><path d=\"M0,0 L10,5 L0,10 z\" fill=\"#475569\"/><\/marker><\/defs><rect x=\"8.0\" y=\"14.0\" width=\"92.0\" height=\"46.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"54.0\" y=\"40.5\" font-size=\"10\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">높은 점수<\/text><path d=\"M101.0,37.0 L113.0,37.0\" stroke=\"#475569\" stroke-width=\"1.6\" marker-end=\"url(#tfa)\" fill=\"none\"/><rect x=\"114.0\" y=\"14.0\" width=\"92.0\" height=\"46.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"160.0\" y=\"34.0\" font-size=\"10\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">확인 생략<\/text><text x=\"160.0\" y=\"47.0\" font-size=\"8.5\" text-anchor=\"middle\" fill=\"#475569\">위험<\/text><path d=\"M207.0,37.0 L219.0,37.0\" stroke=\"#475569\" stroke-width=\"1.6\" marker-end=\"url(#tfa)\" fill=\"none\"/><rect x=\"220.0\" y=\"14.0\" width=\"92.0\" height=\"46.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"266.0\" y=\"34.0\" font-size=\"9.5\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">영상으로 확인<\/text><text x=\"266.0\" y=\"47.0\" font-size=\"8.5\" text-anchor=\"middle\" fill=\"#475569\">해야 할 일<\/text><text x=\"160.0\" y=\"87.0\" font-size=\"9\" text-anchor=\"middle\" fill=\"#475569\">점수가 높아도 직접 확인해요<\/text><\/svg>", "f_secret": "<svg viewBox=\"0 0 320 102\" xmlns=\"http://www.w3.org/2000/svg\" font-family=\"sans-serif\" font-size=\"10\" fill=\"#111827\"><defs><marker id=\"tfa\" viewBox=\"0 0 10 10\" refX=\"8\" refY=\"5\" markerWidth=\"6\" markerHeight=\"6\" orient=\"auto\"><path d=\"M0,0 L10,5 L0,10 z\" fill=\"#475569\"/><\/marker><\/defs><rect x=\"6.0\" y=\"8\" width=\"154.0\" height=\"18\" fill=\"#e5e7eb\" stroke=\"#fff\"/><text x=\"83.0\" y=\"20.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">넣어도 돼요<\/text><rect x=\"160.0\" y=\"8\" width=\"154.0\" height=\"18\" fill=\"#e5e7eb\" stroke=\"#fff\"/><text x=\"237.0\" y=\"20.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">넣으면 안 돼요<\/text><rect x=\"6.0\" y=\"26\" width=\"154.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"83.0\" y=\"38.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">열 이름<\/text><rect x=\"160.0\" y=\"26\" width=\"154.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"237.0\" y=\"38.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">카메라 위치<\/text><rect x=\"6.0\" y=\"44\" width=\"154.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"83.0\" y=\"56.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">가상 샘플<\/text><rect x=\"160.0\" y=\"44\" width=\"154.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"237.0\" y=\"56.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">근무자 이름<\/text><rect x=\"6.0\" y=\"62\" width=\"154.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"83.0\" y=\"74.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">오류 메시지<\/text><rect x=\"160.0\" y=\"62\" width=\"154.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"237.0\" y=\"74.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">영상 캡처<\/text><text x=\"160.0\" y=\"97.0\" font-size=\"9\" text-anchor=\"middle\" fill=\"#475569\">공개 AI에 넣기 전에 나눠 봐요<\/text><\/svg>", "f_log": "<svg viewBox=\"0 0 320 120\" xmlns=\"http://www.w3.org/2000/svg\" font-family=\"sans-serif\" font-size=\"10\" fill=\"#111827\"><defs><marker id=\"tfa\" viewBox=\"0 0 10 10\" refX=\"8\" refY=\"5\" markerWidth=\"6\" markerHeight=\"6\" orient=\"auto\"><path d=\"M0,0 L10,5 L0,10 z\" fill=\"#475569\"/><\/marker><\/defs><rect x=\"6.0\" y=\"8\" width=\"154.0\" height=\"18\" fill=\"#e5e7eb\" stroke=\"#fff\"/><text x=\"83.0\" y=\"20.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">항목<\/text><rect x=\"160.0\" y=\"8\" width=\"154.0\" height=\"18\" fill=\"#e5e7eb\" stroke=\"#fff\"/><text x=\"237.0\" y=\"20.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">예<\/text><rect x=\"6.0\" y=\"26\" width=\"154.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"83.0\" y=\"38.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">누가<\/text><rect x=\"160.0\" y=\"26\" width=\"154.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"237.0\" y=\"38.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">모니터링 담당<\/text><rect x=\"6.0\" y=\"44\" width=\"154.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"83.0\" y=\"56.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">언제<\/text><rect x=\"160.0\" y=\"44\" width=\"154.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"237.0\" y=\"56.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">02:14<\/text><rect x=\"6.0\" y=\"62\" width=\"154.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"83.0\" y=\"74.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">무엇을<\/text><rect x=\"160.0\" y=\"62\" width=\"154.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"237.0\" y=\"74.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">경보 종결<\/text><rect x=\"6.0\" y=\"80\" width=\"154.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"83.0\" y=\"92.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">왜<\/text><rect x=\"160.0\" y=\"80\" width=\"154.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"237.0\" y=\"92.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">고라니 확인<\/text><text x=\"160.0\" y=\"115.0\" font-size=\"9\" text-anchor=\"middle\" fill=\"#475569\">이유까지 남겨요<\/text><\/svg>", "f_console": "<svg viewBox=\"0 0 320 148\" xmlns=\"http://www.w3.org/2000/svg\" font-family=\"sans-serif\" font-size=\"10\" fill=\"#111827\"><defs><marker id=\"tfa\" viewBox=\"0 0 10 10\" refX=\"8\" refY=\"5\" markerWidth=\"6\" markerHeight=\"6\" orient=\"auto\"><path d=\"M0,0 L10,5 L0,10 z\" fill=\"#475569\"/><\/marker><\/defs><polygon points=\"20,118 90,104 180,104 110,118\" fill=\"#fee2e2\" fill-opacity=\"0.55\" stroke=\"#fee2e2\"/><text x=\"192.0\" y=\"115.0\" font-size=\"9.5\" text-anchor=\"start\" fill=\"#111827\" font-weight=\"bold\">영상 손실 경보<\/text><polygon points=\"20,92 90,78 180,78 110,92\" fill=\"#dbeafe\" fill-opacity=\"0.55\" stroke=\"#dbeafe\"/><text x=\"192.0\" y=\"89.0\" font-size=\"9.5\" text-anchor=\"start\" fill=\"#111827\" font-weight=\"bold\">경보 목록·우선순위·상태<\/text><polygon points=\"20,66 90,52 180,52 110,66\" fill=\"#e0e7ff\" fill-opacity=\"0.55\" stroke=\"#e0e7ff\"/><text x=\"192.0\" y=\"63.0\" font-size=\"9.5\" text-anchor=\"start\" fill=\"#111827\" font-weight=\"bold\">경보 영상<\/text><polygon points=\"20,40 90,26 180,26 110,40\" fill=\"#c7d2fe\" fill-opacity=\"0.55\" stroke=\"#c7d2fe\"/><text x=\"192.0\" y=\"37.0\" font-size=\"9.5\" text-anchor=\"start\" fill=\"#111827\" font-weight=\"bold\">배치도<\/text><text x=\"160.0\" y=\"143.0\" font-size=\"9\" text-anchor=\"middle\" fill=\"#475569\">몰릴 때 무엇부터 볼지 정해요<\/text><\/svg>", "f_gap": "<svg viewBox=\"0 0 320 102\" xmlns=\"http://www.w3.org/2000/svg\" font-family=\"sans-serif\" font-size=\"10\" fill=\"#111827\"><defs><marker id=\"tfa\" viewBox=\"0 0 10 10\" refX=\"8\" refY=\"5\" markerWidth=\"6\" markerHeight=\"6\" orient=\"auto\"><path d=\"M0,0 L10,5 L0,10 z\" fill=\"#475569\"/><\/marker><\/defs><rect x=\"6.0\" y=\"8\" width=\"102.7\" height=\"18\" fill=\"#e5e7eb\" stroke=\"#fff\"/><text x=\"57.3\" y=\"20.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">신호 시각<\/text><rect x=\"108.7\" y=\"8\" width=\"102.7\" height=\"18\" fill=\"#e5e7eb\" stroke=\"#fff\"/><text x=\"160.0\" y=\"20.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">간격<\/text><rect x=\"211.3\" y=\"8\" width=\"102.7\" height=\"18\" fill=\"#e5e7eb\" stroke=\"#fff\"/><text x=\"262.7\" y=\"20.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">판정<\/text><rect x=\"6.0\" y=\"26\" width=\"102.7\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"57.3\" y=\"38.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">0, 10<\/text><rect x=\"108.7\" y=\"26\" width=\"102.7\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"160.0\" y=\"38.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">10초<\/text><rect x=\"211.3\" y=\"26\" width=\"102.7\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"262.7\" y=\"38.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">정상<\/text><rect x=\"6.0\" y=\"44\" width=\"102.7\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"57.3\" y=\"56.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">10, 55<\/text><rect x=\"108.7\" y=\"44\" width=\"102.7\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"160.0\" y=\"56.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">45초<\/text><rect x=\"211.3\" y=\"44\" width=\"102.7\" height=\"18\" fill=\"#fef3c7\" stroke=\"#e5e7eb\"/><text x=\"262.7\" y=\"56.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">영상 손실<\/text><rect x=\"6.0\" y=\"62\" width=\"102.7\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"57.3\" y=\"74.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">55, 85<\/text><rect x=\"108.7\" y=\"62\" width=\"102.7\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"160.0\" y=\"74.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">30초<\/text><rect x=\"211.3\" y=\"62\" width=\"102.7\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"262.7\" y=\"74.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">정상<\/text><text x=\"160.0\" y=\"97.0\" font-size=\"9\" text-anchor=\"middle\" fill=\"#475569\">30초를 넘으면 영상 손실이에요<\/text><\/svg>", "f_review": "<svg viewBox=\"0 0 320 156\" xmlns=\"http://www.w3.org/2000/svg\" font-family=\"sans-serif\" font-size=\"10\" fill=\"#111827\"><defs><marker id=\"tfa\" viewBox=\"0 0 10 10\" refX=\"8\" refY=\"5\" markerWidth=\"6\" markerHeight=\"6\" orient=\"auto\"><path d=\"M0,0 L10,5 L0,10 z\" fill=\"#475569\"/><\/marker><\/defs><ellipse cx=\"160.0\" cy=\"70\" rx=\"112\" ry=\"48\" fill=\"none\" stroke=\"#e5e7eb\" stroke-width=\"6\"/><path d=\"M207.7,26.6 L261.3,49.6\" stroke=\"#2563eb\" stroke-width=\"2\" marker-end=\"url(#tfa)\" fill=\"none\"/><path d=\"M261.3,90.4 L207.7,113.4\" stroke=\"#2563eb\" stroke-width=\"2\" marker-end=\"url(#tfa)\" fill=\"none\"/><path d=\"M112.3,113.4 L58.7,90.4\" stroke=\"#2563eb\" stroke-width=\"2\" marker-end=\"url(#tfa)\" fill=\"none\"/><path d=\"M58.7,49.6 L112.3,26.6\" stroke=\"#2563eb\" stroke-width=\"2\" marker-end=\"url(#tfa)\" fill=\"none\"/><rect x=\"116.0\" y=\"10.0\" width=\"88.0\" height=\"24.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"160.0\" y=\"25.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">설명<\/text><rect x=\"228.0\" y=\"58.0\" width=\"88.0\" height=\"24.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"272.0\" y=\"73.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">테스트<\/text><rect x=\"116.0\" y=\"106.0\" width=\"88.0\" height=\"24.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"160.0\" y=\"121.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">수정<\/text><rect x=\"4.0\" y=\"58.0\" width=\"88.0\" height=\"24.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"48.0\" y=\"73.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">비교<\/text><text x=\"160.0\" y=\"74.0\" font-size=\"10.5\" text-anchor=\"middle\" fill=\"#475569\" font-weight=\"bold\">리뷰<\/text><text x=\"160.0\" y=\"151.0\" font-size=\"9\" text-anchor=\"middle\" fill=\"#475569\">고치기 전에 먼저 설명하게 해요<\/text><\/svg>", "f_scope": "<svg viewBox=\"0 0 320 92\" xmlns=\"http://www.w3.org/2000/svg\" font-family=\"sans-serif\" font-size=\"10\" fill=\"#111827\"><defs><marker id=\"tfa\" viewBox=\"0 0 10 10\" refX=\"8\" refY=\"5\" markerWidth=\"6\" markerHeight=\"6\" orient=\"auto\"><path d=\"M0,0 L10,5 L0,10 z\" fill=\"#475569\"/><\/marker><\/defs><rect x=\"8.0\" y=\"14.0\" width=\"92.0\" height=\"46.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"54.0\" y=\"34.0\" font-size=\"10\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">고칠 곳<\/text><text x=\"54.0\" y=\"47.0\" font-size=\"8.5\" text-anchor=\"middle\" fill=\"#475569\">이 함수만<\/text><path d=\"M101.0,37.0 L113.0,37.0\" stroke=\"#475569\" stroke-width=\"1.6\" marker-end=\"url(#tfa)\" fill=\"none\"/><rect x=\"114.0\" y=\"14.0\" width=\"92.0\" height=\"46.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"160.0\" y=\"34.0\" font-size=\"10\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">그대로<\/text><text x=\"160.0\" y=\"47.0\" font-size=\"8.5\" text-anchor=\"middle\" fill=\"#475569\">원본 기록<\/text><path d=\"M207.0,37.0 L219.0,37.0\" stroke=\"#475569\" stroke-width=\"1.6\" marker-end=\"url(#tfa)\" fill=\"none\"/><rect x=\"220.0\" y=\"14.0\" width=\"92.0\" height=\"46.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"266.0\" y=\"34.0\" font-size=\"10\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">그대로<\/text><text x=\"266.0\" y=\"47.0\" font-size=\"8.5\" text-anchor=\"middle\" fill=\"#475569\">다른 코드<\/text><text x=\"160.0\" y=\"87.0\" font-size=\"9\" text-anchor=\"middle\" fill=\"#475569\">고칠 범위를 말해 주세요<\/text><\/svg>", "f_tool": "<svg viewBox=\"0 0 320 92\" xmlns=\"http://www.w3.org/2000/svg\" font-family=\"sans-serif\" font-size=\"10\" fill=\"#111827\"><defs><marker id=\"tfa\" viewBox=\"0 0 10 10\" refX=\"8\" refY=\"5\" markerWidth=\"6\" markerHeight=\"6\" orient=\"auto\"><path d=\"M0,0 L10,5 L0,10 z\" fill=\"#475569\"/><\/marker><\/defs><rect x=\"8.0\" y=\"14.0\" width=\"65.5\" height=\"46.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"40.8\" y=\"40.5\" font-size=\"10\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">설정값<\/text><path d=\"M74.5,37.0 L86.5,37.0\" stroke=\"#475569\" stroke-width=\"1.6\" marker-end=\"url(#tfa)\" fill=\"none\"/><rect x=\"87.5\" y=\"14.0\" width=\"65.5\" height=\"46.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"120.2\" y=\"40.5\" font-size=\"10\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">계산 함수<\/text><path d=\"M154.0,37.0 L166.0,37.0\" stroke=\"#475569\" stroke-width=\"1.6\" marker-end=\"url(#tfa)\" fill=\"none\"/><rect x=\"167.0\" y=\"14.0\" width=\"65.5\" height=\"46.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"199.8\" y=\"40.5\" font-size=\"10\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">표·그림<\/text><path d=\"M233.5,37.0 L245.5,37.0\" stroke=\"#475569\" stroke-width=\"1.6\" marker-end=\"url(#tfa)\" fill=\"none\"/><rect x=\"246.5\" y=\"14.0\" width=\"65.5\" height=\"46.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"279.2\" y=\"40.5\" font-size=\"10\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">브리핑<\/text><text x=\"160.0\" y=\"87.0\" font-size=\"9\" text-anchor=\"middle\" fill=\"#475569\">한 번에 다시 돌릴 수 있게 묶어요<\/text><\/svg>", "f_aar": "<svg viewBox=\"0 0 320 156\" xmlns=\"http://www.w3.org/2000/svg\" font-family=\"sans-serif\" font-size=\"10\" fill=\"#111827\"><defs><marker id=\"tfa\" viewBox=\"0 0 10 10\" refX=\"8\" refY=\"5\" markerWidth=\"6\" markerHeight=\"6\" orient=\"auto\"><path d=\"M0,0 L10,5 L0,10 z\" fill=\"#475569\"/><\/marker><\/defs><ellipse cx=\"160.0\" cy=\"70\" rx=\"112\" ry=\"48\" fill=\"none\" stroke=\"#e5e7eb\" stroke-width=\"6\"/><path d=\"M207.7,26.6 L261.3,49.6\" stroke=\"#2563eb\" stroke-width=\"2\" marker-end=\"url(#tfa)\" fill=\"none\"/><path d=\"M261.3,90.4 L207.7,113.4\" stroke=\"#2563eb\" stroke-width=\"2\" marker-end=\"url(#tfa)\" fill=\"none\"/><path d=\"M112.3,113.4 L58.7,90.4\" stroke=\"#2563eb\" stroke-width=\"2\" marker-end=\"url(#tfa)\" fill=\"none\"/><path d=\"M58.7,49.6 L112.3,26.6\" stroke=\"#2563eb\" stroke-width=\"2\" marker-end=\"url(#tfa)\" fill=\"none\"/><rect x=\"116.0\" y=\"10.0\" width=\"88.0\" height=\"24.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"160.0\" y=\"25.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">계획<\/text><rect x=\"228.0\" y=\"58.0\" width=\"88.0\" height=\"24.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"272.0\" y=\"73.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">실제<\/text><rect x=\"116.0\" y=\"106.0\" width=\"88.0\" height=\"24.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"160.0\" y=\"121.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">원인<\/text><rect x=\"4.0\" y=\"58.0\" width=\"88.0\" height=\"24.0\" rx=\"7\" fill=\"#dbeafe\" stroke=\"#94a3b8\"/><text x=\"48.0\" y=\"73.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">재조정<\/text><text x=\"160.0\" y=\"74.0\" font-size=\"10.5\" text-anchor=\"middle\" fill=\"#475569\" font-weight=\"bold\">AAR<\/text><text x=\"160.0\" y=\"151.0\" font-size=\"9\" text-anchor=\"middle\" fill=\"#475569\">탓하지 않고 바꿀 것을 찾아요<\/text><\/svg>", "f_lint": "<svg viewBox=\"0 0 320 102\" xmlns=\"http://www.w3.org/2000/svg\" font-family=\"sans-serif\" font-size=\"10\" fill=\"#111827\"><defs><marker id=\"tfa\" viewBox=\"0 0 10 10\" refX=\"8\" refY=\"5\" markerWidth=\"6\" markerHeight=\"6\" orient=\"auto\"><path d=\"M0,0 L10,5 L0,10 z\" fill=\"#475569\"/><\/marker><\/defs><rect x=\"6.0\" y=\"8\" width=\"154.0\" height=\"18\" fill=\"#e5e7eb\" stroke=\"#fff\"/><text x=\"83.0\" y=\"20.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">항목<\/text><rect x=\"160.0\" y=\"8\" width=\"154.0\" height=\"18\" fill=\"#e5e7eb\" stroke=\"#fff\"/><text x=\"237.0\" y=\"20.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">있음?<\/text><rect x=\"6.0\" y=\"26\" width=\"154.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"83.0\" y=\"38.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">표본 수<\/text><rect x=\"160.0\" y=\"26\" width=\"154.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"237.0\" y=\"38.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">O<\/text><rect x=\"6.0\" y=\"44\" width=\"154.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"83.0\" y=\"56.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">남은 위험<\/text><rect x=\"160.0\" y=\"44\" width=\"154.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"237.0\" y=\"56.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\">O<\/text><rect x=\"6.0\" y=\"62\" width=\"154.0\" height=\"18\" fill=\"#fff\" stroke=\"#e5e7eb\"/><text x=\"83.0\" y=\"74.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">AI 산출물 표기<\/text><rect x=\"160.0\" y=\"62\" width=\"154.0\" height=\"18\" fill=\"#fef3c7\" stroke=\"#e5e7eb\"/><text x=\"237.0\" y=\"74.5\" font-size=\"9\" text-anchor=\"middle\" fill=\"#111827\" font-weight=\"bold\">X<\/text><text x=\"160.0\" y=\"97.0\" font-size=\"9\" text-anchor=\"middle\" fill=\"#475569\">빠진 항목을 코드로 찾아요<\/text><\/svg>"};   // 핵심 용어 카드에 붙는 그림(SVG)
const MISSION_UI = {"m1": {"steps": [{"short": "네 칸 익히기", "title": "프롬프트의 네 칸", "lead": "AI는 전우님 업무도, 파일도 몰라요. **네 칸**을 채워 알려 주면 돼요.", "page": null, "blocks": [{"id": "s1_info", "type": "info", "title": "좋은 프롬프트의 네 칸", "icon": "🧩", "req": false, "text": ["같은 일을 시켜도 무엇을 알려 주느냐에 따라 결과가 달라져요. 처음에는 아래 네 칸만 기억하세요."], "table": {"head": ["칸", "무엇을 적나", "예"], "rows": [["맥락", "왜 필요한지, 어떤 일인지", "CCTV 경보가 너무 많아 원인을 보려고 해"], ["입력", "파일 이름, 열 이름, 값의 뜻", "m1_alarms.csv, 열은 alarm_id·camera·hour·object"], ["할 일", "나누는 규칙을 구체적으로", "person·vehicle·drone은 경보 대상"], ["출력 형식", "받고 싶은 모양", "카메라는 행, 구분은 열인 표"]]}, "figure": "f_four"}, {"id": "s1_q", "type": "quiz", "title": "어떤 프롬프트가 가장 좋을까요?", "req": true, "text": "같은 일을 시키는 세 프롬프트예요. 다시 묻는 일이 가장 적을 것은 무엇일까요?", "options": ["CCTV 분석 코드 좀 짜 줘.", "경보가 많은지 알려 줘. 잘 부탁해.", "CCTV 경보가 너무 많아 원인을 보려고 해. m1_alarms.csv의 object 값을 경보 대상과 불필요 경보로 나눠 카메라별로 세어 표로 보여 줘."]}]}, {"short": "AI에게 시키기", "title": "프롬프트를 만들어 AI에게 보내기", "lead": "칸마다 문장을 고르면 프롬프트가 만들어져요. 일부러 **아쉬운 문장**도 보내 보세요. 무엇이 달라지는지 보는 것이 이번 활동의 핵심이에요.", "page": null, "blocks": [{"id": "s2_vibe", "type": "vibe", "title": "카메라별 경보 집계표 만들기", "req": true, "ask": ["oh", "경보 기록 파일을 드릴게요. 카메라마다 경보 대상, 불필요 경보, 오경보가 몇 건인지 표로 받고 싶어요."], "has_start": false, "files": [{"name": "m1_alarms.csv", "head": ["alarm_id", "camera", "hour", "object"], "rows": [["A001", "CAM2", "17", "glitch"], ["A002", "CAM4", "1", "tree"], ["A003", "CAM2", "19", "drone"], ["A004", "CAM3", "15", "rain"], ["A005", "CAM2", "10", "noise"]], "n": 80}], "slots": [{"id": "ctx", "label": "맥락", "why": "왜 필요한지 알려요.", "options": [{"i": 0, "text": "CCTV 영상분석 경보를 살피고 있어. 경보가 너무 많아서 카메라별로 무엇 때문에 울리는지 확인하려고 해."}, {"i": 1, "text": "코드 좀 짜 줘."}, {"i": 2, "text": ""}]}, {"id": "input", "label": "입력", "why": "파일과 열을 알려요.", "options": [{"i": 0, "text": "데이터는 m1_alarms.csv이고, 열은 alarm_id, camera(CAM1~CAM4), hour(0~23), object(울린 원인 물체)야."}, {"i": 1, "text": "경보 파일이 있어."}, {"i": 2, "text": ""}]}, {"id": "task", "label": "할 일", "why": "나누는 규칙을 구체적으로 적어요.", "options": [{"i": 1, "text": "진짜 경보랑 가짜 경보를 나눠 줘."}, {"i": 0, "text": "object가 person, vehicle, drone이면 경보 대상, animal, tree, rain, light면 불필요 경보, glitch, noise면 오경보로 나눠서 카메라별로 세어 줘."}, {"i": 2, "text": ""}]}, {"id": "out", "label": "출력 형식", "why": "받고 싶은 모양을 알려요.", "options": [{"i": 1, "text": "결과를 보여 줘."}, {"i": 0, "text": "카메라는 행, 구분은 열인 표로 보여 주고 합계 열도 넣어 줘. 카메라별 쌓은 막대그래프를 그리고, 표를 넣은 HTML 보고서를 report 변수에 담아 줘."}, {"i": 2, "text": ""}]}]}]}, {"short": "내 말로 쓰기", "title": "같은 일을 내 말로 요청하기", "lead": "고른 문장을 그대로 외울 필요는 없어요. **네 칸만 채우면** 말투는 자유예요.", "page": null, "blocks": [{"id": "s3_q", "type": "quiz", "title": "결과를 받은 뒤에는", "req": true, "text": "AI가 만든 코드가 오류 없이 돌았어요. 다음으로 무엇을 하면 좋을까요?", "options": ["오류가 없으니 그대로 보고한다.", "카메라 하나의 합계를 파일에서 직접 세어 맞는지 확인한다.", "더 어려운 말로 다시 요청한다."]}, {"id": "s3_write", "type": "write", "title": "내 프롬프트 쓰기", "req": true, "prompt": "**새 요청**: 시간대별로도 보고 싶어요. 파일은 같은 m1_alarms.csv이고, 0~5시를 심야, 6~17시를 주간, 18~23시를 야간으로 나눠 불필요 경보 수를 보고 싶어요. 네 칸을 채워 내 말로 프롬프트를 써 보세요.", "placeholder": "꼭 입력해 주세요 · 내 말로 프롬프트를 써 보세요", "model_label": "📝 모범 프롬프트", "min": 30}]}]}, "m2": {"steps": [{"short": "데이터 보여 주기", "title": "AI에게 데이터를 설명하는 법", "lead": "AI는 파일을 볼 수 없어요. **값이 어떻게 적혀 있는지**까지 알려 주는 것이 가장 빠른 방법이에요.", "page": null, "blocks": [{"id": "s1_info", "type": "info", "title": "데이터를 설명할 때 넣을 것", "icon": "🗂", "req": false, "text": ["열 이름만으로는 부족할 때가 많아요. 값의 모양과 단위까지 알려 주세요."], "bullets": ["**파일 이름과 열 이름**: m2_requirements.csv, req_id, owner, target, min_recall, max_delay, max_unneeded_per_day", "**값의 모양**: min_recall은 95%, 0.9, 90이 섞여 있고, max_delay는 30s, 1min, 45 sec가 섞여 있음", "**빈칸**: R09의 min_recall이 비어 있음", "**샘플 몇 줄**: 실제 값 대신 가상 샘플로 보여 줘도 돼요"], "figure": "f_clean"}, {"id": "s1_q", "type": "quiz", "title": "먼저 알려 줄 것", "req": true, "text": "요구값에 ‘0.9’와 ‘85%’가 섞여 있어요. 이 사실을 알려 주지 않으면 어떻게 될까요?", "options": ["AI가 스스로 알아채 항상 맞춰 준다.", "글자로 비교해 85%가 0.9보다 높다고 나오는 식으로 조용히 틀릴 수 있다.", "파일을 읽지 못해 반드시 오류가 난다."]}]}, {"short": "비교표 받기", "title": "단위를 맞춰 비교표 받기", "lead": "**단위 설명** 문장을 빼고도 보내 보세요. 결과가 어떻게 달라지는지 비교해 보면 좋아요.", "page": null, "blocks": [{"id": "s2_vibe", "type": "vibe", "title": "요구 성능 비교표", "req": true, "ask": ["yun", "대상별로 담당의 요구 가운데 가장 엄격한 값이 무엇인지 보고 싶어요."], "has_start": false, "files": [{"name": "m2_requirements.csv", "head": ["req_id", "owner", "target", "min_recall", "max_delay", "max_unneeded_per_day"], "rows": [["R01", "plan", "person", "95%", "30s", "20"], ["R02", "plan", "vehicle", "0.9", "1min", "20"], ["R03", "plan", "drone", "90", "30s", "20"], ["R04", "control", "person", "0.85", "45 sec", "10"], ["R05", "control", "vehicle", "85%", "90s", "10"]], "n": 9}], "slots": [{"id": "input", "label": "입력", "why": "값의 모양까지 알려요.", "options": [{"i": 1, "text": "m2_requirements.csv에 요구가 있어."}, {"i": 0, "text": "m2_requirements.csv의 열은 req_id, owner(plan·control·monitor), target(person·vehicle·drone), min_recall, max_delay, max_unneeded_per_day야. min_recall은 95%, 0.9, 90이 섞여 있으니 모두 % 숫자로 맞추고, max_delay는 30s, 1min, 45 sec가 섞여 있으니 모두 초로 맞춰 줘. 빈칸은 ‘자료 없음’으로 둬 줘."}, {"i": 2, "text": ""}]}, {"id": "task", "label": "할 일", "why": "무엇을 뽑는지 정해요.", "options": [{"i": 0, "text": "target별로 min_recall은 가장 높은 값, max_delay와 max_unneeded_per_day는 가장 낮은 값을 뽑아 줘. 어느 담당의 요구인지도 함께 적어 줘."}, {"i": 1, "text": "요구를 종합해 줘."}, {"i": 2, "text": ""}]}, {"id": "out", "label": "출력 형식", "why": "회의에서 볼 모양으로 받아요.", "options": [{"i": 1, "text": "결과를 정리해 줘."}, {"i": 0, "text": "맞춘 요구 전체 표와 대상별 가장 엄격한 값 표를 보여 주고, 두 표를 넣은 HTML 문서를 report 변수에 담아 줘."}, {"i": 2, "text": ""}]}]}]}, {"short": "기준 합의", "title": "결과로 기준을 합의하고 내 말로 요청하기", "lead": "AI는 계산을 도와줄 뿐이에요. 어떤 기준을 쓸지는 **사람이 근거를 보고 정해요.**", "page": null, "blocks": [{"id": "s3_decide", "type": "decide", "title": "팀원의 제안 판단하기", "req": true, "text": "비교표를 본 팀원들의 제안이에요. 받아들일지 정하고 이유를 적어 주세요.", "reason_min": 8, "items": [{"id": "a", "who": "kang", "text": "재현율을 100%로 정해요. 하나라도 놓치면 안 되잖아요."}, {"id": "b", "who": "seo", "text": "대상별 기준을 먼저 합의해 표로 남기고, 다음 미션부터 이 표로 모델을 평가해요."}]}, {"id": "s3_write", "type": "write", "title": "요구서를 요청하는 프롬프트 쓰기", "req": true, "prompt": "방금 합의한 내용으로 **감시 요구서**를 AI에게 요청해 보세요. 누가 읽는지, 어떤 항목이 들어가야 하는지, 근거 데이터를 적어 주세요.", "placeholder": "꼭 입력해 주세요 · 내 말로 프롬프트를 써 보세요", "model_label": "📝 모범 프롬프트", "min": 30}]}]}, "m3": {"steps": [{"short": "검증기 받기", "title": "규격을 그대로 적어 검증기 받기", "lead": "규격을 **조건 그대로** 적으면 AI가 검사 코드로 바꿔 줘요.", "page": null, "blocks": [{"id": "s1_vibe", "type": "vibe", "title": "라벨 검증기", "req": true, "ask": ["yun", "규격에 어긋난 라벨을 찾고 이유까지 보고 싶어요."], "has_start": false, "files": [{"name": "m3_labels.csv", "head": ["frame_id", "clip_id", "class", "x", "y", "w", "h", "img_w", "img_h"], "rows": [["F001", "C01", "bird", "220", "39", "54", "52", "640", "360"], ["F002", "C01", "person", "472", "173", "49", "32", "640", "360"], ["F003", "C01", "person", "135", "113", "39", "13", "640", "360"], ["F004", "C01", "Drone", "109", "65", "10", "22", "640", "360"], ["F005", "C01", "drone", "419", "297", "19", "55", "640", "360"]], "n": 40}], "slots": [{"id": "input", "label": "입력", "why": "열의 뜻을 알려요.", "options": [{"i": 1, "text": "라벨 CSV가 있어."}, {"i": 0, "text": "m3_labels.csv의 열은 frame_id, clip_id, class, x, y(왼쪽 위 좌표), w, h(폭과 높이), img_w, img_h(화면 크기)야."}, {"i": 2, "text": ""}]}, {"id": "task", "label": "할 일", "why": "규격을 조건으로 적어요.", "options": [{"i": 1, "text": "라벨이 잘못된 걸 찾아 줘."}, {"i": 0, "text": "class는 person, vehicle, drone, bird만 허용해. w나 h가 0 이하면 ‘크기’, x+w가 img_w를 넘거나 y+h가 img_h를 넘으면 ‘화면 밖’, class가 목록에 없으면 ‘클래스’로 표시해 줘."}, {"i": 2, "text": ""}]}, {"id": "out", "label": "출력 형식", "why": "표의 모양을 정해요.", "options": [{"i": 1, "text": "보기 좋게 보여 줘."}, {"i": 0, "text": "오류 행만 frame_id, clip_id, class, 오류 이유 순서의 표로 보여 주고, 오류 이유별 건수도 함께 보여 줘."}, {"i": 2, "text": ""}]}]}]}, {"short": "오류 고치기", "title": "오류 메시지로 수정 요청하기", "lead": "파일을 바꿨더니 **오류**가 났어요. 어떻게 다시 요청할까요?", "page": null, "blocks": [{"id": "s2_vibe", "type": "vibe", "title": "오류 고쳐 달라고 하기", "req": true, "ask": ["yun", "다른 공급처 파일을 넣었더니 오류가 나요. 저도 원인을 잘 모르겠어요… 부탁해요!"], "has_start": true, "files": [{"name": "m3_labels_new.csv", "head": ["frame_id", "clip_id", "class", "x", "y", "w", "h", "img_w", "img_h"], "rows": [["N01", "D01", "person", "120.5", "88", "30", "42", "640", "360"], ["N02", "D01", "person", "122.0", "90", "30", "42", "640", "360"], ["N03", "D02", "drone", "300,5", "40", "12", "10", "640", "360"], ["N04", "D02", "drone", "304,0", "41", "12", "10", "640", "360"], ["N05", "D03", "vehicle", "n/a", "200", "60", "40", "640", "360"]], "n": 6}], "slots": [{"id": "what", "label": "무슨 일이 있었는지", "why": "오류 메시지를 그대로 붙여 넣어요.", "options": [{"i": 1, "text": "안 돼. 다시 해 줘."}, {"i": 0, "text": "m3_labels_new.csv를 검사하다가 이 오류가 났어: ValueError: could not convert string to float: '300,5'"}, {"i": 2, "text": ""}]}, {"id": "fix", "label": "어떻게 고칠지", "why": "원하는 동작과 범위를 정해요.", "options": [{"i": 0, "text": "이 공급처는 소수점을 쉼표로 적어. 쉼표는 점으로 바꿔 읽고, 그래도 숫자가 아니면 ‘형식 오류’로 표시해 줘. 다른 규칙은 그대로 두고, 표를 넣은 HTML 라벨 검증 결과를 report 변수에 담아 줘."}, {"i": 1, "text": "전부 새로 짜 줘."}, {"i": 2, "text": ""}]}]}]}, {"short": "내 말로 쓰기", "title": "다른 오류를 내 말로 요청하기", "lead": "오류를 만나면 **겁먹지 말고 그대로 붙여 넣으세요.** 코드를 다 읽지 못해도 수정을 요청할 수 있어요.", "page": null, "blocks": [{"id": "s3_write", "type": "write", "title": "수정 프롬프트 쓰기", "req": true, "prompt": "**새 상황**: 또 다른 공급처 파일을 넣었더니 `KeyError: 'class'`가 났어요. 이 파일의 실제 열 이름은 label이에요. 잘 되던 검사 규칙은 그대로 두고 싶어요. 어떻게 요청할까요?", "placeholder": "꼭 입력해 주세요 · 내 말로 프롬프트를 써 보세요", "model_label": "📝 모범 프롬프트", "min": 30}]}]}, "m4": {"steps": [{"short": "쪼개기", "title": "큰 일을 작은 단계로 쪼개기", "lead": "한 번에 전부 요청하면 어디가 틀렸는지 찾기 어려워요. **작게 만들고 확인한 뒤 넓혀요.**", "page": null, "blocks": [{"id": "s1_info", "type": "info", "title": "비교할 영역", "icon": "📐", "req": false, "text": ["좌표는 가로 0~100, 세로 0~100이고, y가 작을수록 울타리에 가까워요. 팀원들의 의견을 영역 두 개로 정리했어요."], "table": {"head": ["이름", "좌표 (x0, y0, x1, y1)", "제안한 사람"], "rows": [["영역 A", "0, 0, 100, 30", "강계획 (넓게)"], ["영역 B", "0, 0, 100, 15", "서상황 (좁게)"], ["제외 영역", "70, 0, 100, 14", "민시설 (나무)"]]}, "figure": "f_roi"}, {"id": "s1_order", "type": "order", "title": "요청 순서 정하기", "req": true, "text": "영역 비교 도구를 만들려고 해요. 어떤 순서로 요청하면 좋을까요?", "items": [{"id": "c", "text": "답을 아는 점 두 개로 함수 확인하기"}, {"id": "d", "text": "전체 기록에 적용해 영역별 비교표 만들기"}, {"id": "b", "text": "점 하나가 영역 안인지 보는 함수 만들기"}, {"id": "a", "text": "데이터 모양을 알려 주고 파일을 읽게 하기"}]}]}, {"short": "함수로 만들기", "title": "영역을 바꿔 돌릴 수 있는 함수 요청하기", "lead": "좌표를 코드 속에 박아 두면 회의 때마다 고쳐야 해요. **인자로 받는 함수**로 요청해요.", "page": null, "blocks": [{"id": "s2_vibe", "type": "vibe", "title": "침입 탐지 영역 함수와 비교표", "req": true, "ask": ["seo", "영역별로 경보 수와 대상을 얼마나 잡는지, 밤과 안개에서는 어떤지 보고 싶어요."], "has_start": false, "files": [{"name": "m4_events.csv", "head": ["event_id", "x", "y", "condition", "kind"], "rows": [["E001", "92.0", "3.2", "fog", "tree"], ["E002", "94.2", "0.2", "day", "tree"], ["E003", "73.5", "9.2", "day", "tree"], ["E004", "86.2", "32.0", "day", "animal"], ["E005", "62.1", "34.6", "day", "animal"]], "n": 100}], "slots": [{"id": "small", "label": "작게 시작", "why": "함수 하나부터 만들어요.", "options": [{"i": 0, "text": "m4_events.csv의 열은 event_id, x, y, condition(day·night·fog), kind(person·vehicle·animal·tree)야. 점 하나와 영역을 받아 안에 있으면 True를 돌려주는 in_zone 함수를 먼저 만들어 줘. 영역 A에서 (50, 10)은 True, (50, 40)은 False인지 먼저 확인해 줘."}, {"i": 1, "text": "영역 비교 프로그램을 한 번에 다 만들어 줘."}, {"i": 2, "text": ""}]}, {"id": "param", "label": "조건 바꾸기", "why": "영역을 인자로 받아요.", "options": [{"i": 0, "text": "영역은 (x0, y0, x1, y1) 인자로 받게 하고, 제외 영역 (70, 0, 100, 14) 안의 점은 항상 False로 해 줘. 영역 A는 (0, 0, 100, 30), 영역 B는 (0, 0, 100, 15)야."}, {"i": 1, "text": "영역은 코드에 적당히 넣어 줘."}, {"i": 2, "text": ""}]}, {"id": "out", "label": "비교표", "why": "함수를 전체에 적용해요.", "options": [{"i": 1, "text": "결과를 정리해 줘."}, {"i": 0, "text": "영역과 condition별로 경보 수, 대상(person·vehicle) 포착률(%), 불필요 경보(animal·tree) 수를 표로 보여 주고, 점과 영역을 함께 그려 줘. 표를 넣은 HTML 비교 문서를 report 변수에 담아 줘."}, {"i": 2, "text": ""}]}]}]}, {"short": "영역 합의", "title": "결과로 조율하고 내 말로 요청하기", "lead": "AI의 표는 근거일 뿐이에요. 어디까지 볼지는 **사람이 합의해요.**", "page": null, "blocks": [{"id": "s3_decide", "type": "decide", "title": "침입 탐지 영역 합의", "req": true, "text": "비교표를 본 팀원들의 의견이에요. 받아들일지 정하고 이유를 적어 주세요.", "reason_min": 8, "items": [{"id": "a", "who": "seo", "text": "불필요 경보가 적으니 영역 B로 가요. 놓치는 대상은 어쩔 수 없잖아요."}, {"id": "b", "who": "min", "text": "영역 A로 가되 나무 구역은 제외 영역으로 빼고, 그 구역은 다른 카메라가 보도록 표시해 두죠."}]}, {"id": "s3_write", "type": "write", "title": "쪼개서 요청하는 프롬프트 쓰기", "req": true, "prompt": "**새 요청**: 사전 경보 영역을 하나 더 두고 싶어요. 울타리에서 멀리 떨어진 영역 C (0, 30, 100, 45)에 들어오면 경보 대신 ‘사전 알림’으로 표시해요. 어떤 순서로, 무엇을 고쳐 달라고 할까요?", "placeholder": "꼭 입력해 주세요 · 내 말로 프롬프트를 써 보세요", "model_label": "📝 모범 프롬프트", "min": 30}]}]}, "m5": {"steps": [{"short": "답만 받으면", "title": "근거 없는 답은 위험해요", "lead": "AI가 자신 있게 답해도 **다시 셀 수 있는 숫자**가 없으면 믿을지 판단할 수 없어요.", "page": null, "blocks": [{"id": "s1_info", "type": "info", "title": "AI 답변에 함께 받을 것", "icon": "🔍", "req": false, "text": ["답을 받을 때 아래를 함께 요청하면, 코드를 다 읽지 못해도 믿을지 판단할 수 있어요."], "bullets": ["**근거 수치**: 조건별 표본 수, 맞게 찾은 수, 놓친 수, 잘못 찾은 수", "**작은 예시**: 손으로 계산할 수 있는 박스 두 개의 IoU", "**나눈 조건**: 낮, 밤, 안개를 나눴는지"], "figure": "f_verify"}, {"id": "s1_q", "type": "quiz", "title": "IoU를 손으로 계산해 볼까요?", "req": true, "text": "정답 박스 (x=10, y=10, w=20, h=20)와 예측 박스 (x=20, y=10, w=20, h=20)가 있어요. 겹친 넓이는 10×20=200, 합친 넓이는 400+400−200=600이에요. IoU 0.5 기준으로 맞게 찾은 것일까요?", "options": ["맞게 찾았어요. 겹쳐 있으니까요.", "아니에요. IoU가 약 0.33이라 기준 아래예요.", "IoU는 계산할 수 없어요."]}]}, {"short": "근거까지 받기", "title": "근거 수치와 함께 요청하기", "lead": "**근거**와 **조건** 문장을 빼고도 보내 보세요. AI의 답이 어떻게 달라지는지 비교해 봐요.", "page": null, "blocks": [{"id": "s2_vibe", "type": "vibe", "title": "탐지기 조건별 비교", "req": true, "ask": ["yun", "두 탐지기를 같은 기준으로 비교하고 싶어요."], "has_start": false, "files": [{"name": "m5_dets.csv", "head": ["det_id", "model", "condition", "gt_x", "gt_y", "gt_w", "gt_h", "x", "y", "w", "h", "score"], "rows": [["T001", "motion", "day", "459", "188", "53", "22", "510", "188", "53", "22", "0.61"], ["T002", "motion", "day", "181", "159", "26", "38", "179", "156", "26", "38", "0.83"], ["T003", "motion", "day", "189", "96", "43", "55", "189", "96", "43", "55", "0.93"], ["T004", "motion", "day", "481", "126", "27", "55", "484", "127", "27", "55", "0.94"], ["T005", "motion", "day", "41", "177", "55", "23", "43", "180", "55", "23", "0.84"]], "n": 292}], "slots": [{"id": "task", "label": "할 일", "why": "무엇을 비교할지 정해요.", "options": [{"i": 0, "text": "m5_dets.csv의 열은 det_id, model(motion·pretrained), condition(day·night·fog), gt_x, gt_y, gt_w, gt_h(정답 박스, 없으면 빈칸), x, y, w, h(예측 박스, 없으면 빈칸), score야. IoU 0.5 이상이면 맞게 찾은 것(TP)으로 하고, 재현율과 정밀도를 계산해 줘. 정답과 예측이 있어도 IoU가 낮으면 놓친 것과 잘못 찾은 것으로 하나씩 세어 줘."}, {"i": 1, "text": "어떤 탐지기가 좋은지 봐 줘."}, {"i": 2, "text": ""}]}, {"id": "evidence", "label": "근거", "why": "답을 아는 예시와 숫자를 받아요.", "options": [{"i": 0, "text": "결론만 쓰지 말고 TP, 놓친 수(FN), 잘못 찾은 수(FP)를 표로 보여 줘. 박스 (10, 10, 20, 20)과 (20, 10, 20, 20)의 IoU가 약 0.33인지 먼저 확인해 줘."}, {"i": 1, "text": "어느 쪽이 나은지 결론만 말해 줘."}, {"i": 2, "text": ""}]}, {"id": "cond", "label": "조건", "why": "나눠서 비교해요.", "options": [{"i": 1, "text": "전체 평균으로 비교해 줘."}, {"i": 0, "text": "model과 condition별로 나눠 표를 만들고, 재현율을 막대그래프로 그려 줘. 표를 넣은 HTML 비교 문서를 report 변수에 담아 줘."}, {"i": 2, "text": ""}]}]}]}, {"short": "내 말로 쓰기", "title": "검증을 담아 내 말로 요청하기", "lead": "다음에 비슷한 비교를 맡게 되면 **처음부터 근거를 함께** 요청해 보세요.", "page": null, "blocks": [{"id": "s3_write", "type": "write", "title": "근거를 요청하는 프롬프트 쓰기", "req": true, "prompt": "**새 요청**: 새 탐지기 결과 m5_dets_b.csv를 같은 방식으로 비교하고 싶어요. 열은 같아요. 결론만 받지 않도록 어떻게 요청할까요?", "placeholder": "꼭 입력해 주세요 · 내 말로 프롬프트를 써 보세요", "model_label": "📝 모범 프롬프트", "min": 30}]}]}, "m6": {"steps": [{"short": "숫자로 바꾸기", "title": "모호한 말을 숫자 규칙으로", "lead": "‘적당히’, ‘많이’는 사람마다 달라요. **숫자로 바꾸면** AI도 사람도 같은 뜻으로 읽어요.", "page": null, "blocks": [{"id": "s1_match", "type": "match", "title": "모호한 말을 숫자 규칙과 짝짓기", "req": true, "text": "왼쪽 말을 프롬프트에 쓸 수 있게 바꾼 문장을 골라 주세요.", "left_label": "모호한 말", "right_label": "프롬프트에 쓸 문장", "lefts": ["경보를 적당히", "점수를 못 낸 것", "대상을 놓치지 않기", "점수가 높은 것"], "rights": ["재현율 90% 이상", "하루 불필요 경보 10건 이하", "score가 빈칸", "score가 임계값 이상"]}]}, {"short": "규칙으로 요청", "title": "숫자 규칙과 예외를 담아 요청하기", "lead": "**예외** 문장을 빼고도 보내 보세요. 점수가 빈 기록이 어떻게 되는지 확인해 봐요.", "page": null, "blocks": [{"id": "s2_vibe", "type": "vibe", "title": "임계값별 경보 표", "req": true, "ask": ["seo", "임계값을 바꾸면 재현율과 하루 불필요 경보가 어떻게 달라지는지 보고 싶어요."], "has_start": false, "files": [{"name": "m6_alarms.csv", "head": ["alarm_id", "day", "score", "is_target", "cause"], "rows": [["P001", "4", "0.03", "0", "light"], ["P002", "2", "0.32", "0", "tree"], ["P003", "1", "0.3", "0", "animal"], ["P004", "5", "0.91", "1", "none"], ["P005", "5", "0.44", "0", "animal"]], "n": 420}], "slots": [{"id": "rule", "label": "숫자 규칙", "why": "기준을 숫자로 적어요.", "options": [{"i": 1, "text": "경보를 적당히 줄여 줘."}, {"i": 0, "text": "m6_alarms.csv의 열은 alarm_id, day(1~7), score(빈칸 가능), is_target(대상이면 1), cause(animal·tree·rain·light·none)야. 임계값 0.3, 0.4, 0.5, 0.6, 0.7, 0.8마다 score가 임계값 이상이면 경보로 보고, 재현율과 하루 평균 불필요 경보 수를 계산해 줘. 재현율 90% 이상이고 하루 10건 이하면 ‘기준 충족’으로 표시해 줘."}, {"i": 2, "text": ""}]}, {"id": "edge", "label": "예외", "why": "빈칸을 어떻게 할지 정해요.", "options": [{"i": 1, "text": "나머지는 알아서 해 줘."}, {"i": 0, "text": "score가 빈칸인 기록은 영상 손실이니 경보 계산에서 빼되, 지우지 말고 ‘영상 손실: 사람 확인’으로 따로 세어 보여 줘."}, {"i": 2, "text": ""}]}, {"id": "out", "label": "결과와 원인", "why": "표와 그림, 원인까지 받아요.", "options": [{"i": 0, "text": "임계값별 표와 곡선 그림을 보여 주고, 임계값 0.5에서 불필요 경보를 cause별로 센 표도 만들어 줘. 표를 넣은 HTML 문서를 report 변수에 담아 줘."}, {"i": 1, "text": "제일 좋은 값 하나만 알려 줘."}, {"i": 2, "text": ""}]}]}]}, {"short": "운용점 합의", "title": "운용점을 합의하고 내 말로 요청하기", "lead": "표는 AI가 만들어도 **어디에 둘지는 사람이 정해요.**", "page": null, "blocks": [{"id": "s3_decide", "type": "decide", "title": "운용점 의견 판단하기", "req": true, "text": "표를 본 팀원들의 의견이에요. 받아들일지 정하고 이유를 적어 주세요.", "reason_min": 8, "items": [{"id": "a", "who": "seo", "text": "불필요 경보가 가장 적은 0.8로 가요. 관제가 훨씬 편해져요."}, {"id": "b", "who": "kang", "text": "기준을 지키는 임계값을 쓰되, 가장 많은 원인인 동물 경보를 줄일 방법을 먼저 찾아 다음 달에 다시 비교해요."}]}, {"id": "s3_write", "type": "write", "title": "숫자와 예외를 담은 프롬프트 쓰기", "req": true, "prompt": "**새 요청**: ‘야간에는 경보를 좀 더 민감하게 해 줘’라는 요청을 받았어요. 파일은 m6_alarms_night.csv이고, 열에 hour가 더 있어요. 숫자 규칙과 예외를 정해 프롬프트를 써 보세요.", "placeholder": "꼭 입력해 주세요 · 내 말로 프롬프트를 써 보세요", "model_label": "📝 모범 프롬프트", "min": 30}]}]}, "m7": {"steps": [{"short": "먼저 나누기", "title": "어떤 수치를 근거로 쓸까요?", "lead": "문서를 요청하기 전에 **믿을 수 있는 수치**부터 정해요.", "page": null, "blocks": [{"id": "s1_info", "type": "info", "title": "두 가지 나누는 방법", "icon": "✂️", "req": false, "text": ["같은 모델도 데이터를 어떻게 나누느냐에 따라 수치가 크게 달라져요."], "table": {"head": ["나누는 방법", "하는 일", "결과"], "rows": [["무작위 분할", "궤적을 한 줄씩 섞어 나눠요", "같은 클립의 궤적이 학습과 테스트에 함께 들어가 좋게 나와요"], ["클립 단위 분할", "한 클립은 통째로 한쪽에만 둬요", "처음 보는 영상에서의 성능에 가까워요"]]}, "figure": "f_split"}, {"id": "s1_q", "type": "quiz", "title": "어느 수치를 보고할까요?", "req": true, "text": "모델 선택 기록에 쓸 수치로 가장 알맞은 것은 무엇일까요?", "options": ["무작위 분할 수치가 더 높으니 그것을 쓴다.", "클립 단위 분할 수치를 쓰고, 왜 그 수치인지 함께 적는다.", "두 수치를 평균 내어 쓴다."]}]}, {"short": "문서 요청", "title": "독자·근거·표현을 정해 문서 요청하기", "lead": "**자료에 없으면** 문장을 빼고도 보내 보세요. AI가 무엇을 지어내는지 확인해 봐요.", "page": null, "blocks": [{"id": "s2_vibe", "type": "vibe", "title": "모델 선택 기록", "req": true, "ask": ["yun", "모델 비교 결과로 임반장님이 읽을 선택 기록을 받고 싶어요."], "has_start": false, "files": [{"name": "m7_results.csv", "head": ["model", "split_method", "drone_recall_pct", "bird_false_pct", "tracks"], "rows": [["logistic", "random", "88.0", "8.0", "600"], ["logistic", "clip", "81.4", "12.5", "600"], ["random_forest", "random", "96.2", "3.1", "600"], ["random_forest", "clip", "84.5", "9.8", "600"], ["gradient_boosting", "random", "97.0", "2.7", "600"]], "n": 8}], "slots": [{"id": "reader", "label": "독자와 양식", "why": "누가 읽는지, 어떤 순서인지 정해요.", "options": [{"i": 0, "text": "m7_results.csv의 열은 model, split_method(random·clip), drone_recall_pct, bird_false_pct, tracks야. 임반장이 읽을 모델 선택 기록을 HTML로 report 변수에 담아 줘. 순서는 결론 한 줄, 비교표, 고른 이유, 한계야. 그림으로 분할 방법별 재현율도 보여 줘."}, {"i": 1, "text": "모델에 대해 써 줘."}, {"i": 2, "text": ""}]}, {"id": "source", "label": "근거", "why": "자료 밖의 말과 고를 수치를 정해요.", "options": [{"i": 1, "text": "참고할 만한 내용도 알아서 넣어 줘."}, {"i": 0, "text": "근거는 이 파일뿐이야. 판단에는 split_method가 clip인 수치만 쓰고, random 수치는 누수 때문에 높게 나온 값이라고 적어 줘. 파일에 없는 처리 속도나 다른 연구 수치는 지어내지 말고 ‘자료 없음’으로 적어 줘."}, {"i": 2, "text": ""}]}, {"id": "tone", "label": "표현", "why": "확신을 부풀리지 않아요.", "options": [{"i": 1, "text": "자신 있게 확정된 문서처럼 써 줘."}, {"i": 0, "text": "맨 위에 ‘AI 초안, 담당자 확인 전’이라고 표시해 줘. ‘정확히 식별’, ‘완벽’ 같은 말은 쓰지 말고, 놓치는 비율과 새를 드론으로 볼 비율을 함께 적어 줘. 모델 이름은 랜덤 포레스트처럼 띄어 써 줘."}, {"i": 2, "text": ""}]}]}]}, {"short": "내 말로 쓰기", "title": "문서를 받은 뒤 확인하고 내 말로 요청하기", "lead": "문서를 받았다고 끝이 아니에요. **자료와 대조**하는 일은 사람의 몫이에요.", "page": null, "blocks": [{"id": "s3_q", "type": "quiz", "title": "문서를 받은 뒤에는", "req": true, "text": "AI가 쓴 선택 기록이 매끄럽게 나왔어요. 가장 먼저 무엇을 할까요?", "options": ["문장이 매끄러우니 바로 올린다.", "표의 수치와 파일의 값을 대조하고, 파일에 없는 말이 있는지 본다.", "더 길게 써 달라고 한다."]}, {"id": "s3_write", "type": "write", "title": "문서를 요청하는 프롬프트 쓰기", "req": true, "prompt": "**새 요청**: 관제 담당이 읽을 ‘드론·조류 분류 결과 읽는 법’ 안내문이 필요해요. 근거는 m7_results.csv와 m7_tracks.csv뿐이에요. 독자, 순서, 근거, 없는 내용 처리, 표현을 담아 프롬프트를 써 보세요.", "placeholder": "꼭 입력해 주세요 · 내 말로 프롬프트를 써 보세요", "model_label": "📝 모범 프롬프트", "min": 30}]}]}, "m8": {"steps": [{"short": "나눠 보기", "title": "AI에 넣어도 되는 것과 안 되는 것", "lead": "프롬프트에 넣기 전에 **한 번 나눠 보세요.** 값이 없어도 구조만 알면 코드를 받을 수 있어요.", "page": null, "blocks": [{"id": "s1_cls", "type": "classify", "title": "공개 AI에 넣어도 될까요?", "req": true, "text": "조직이 승인한 환경이 아니라 일반 공개 AI라고 생각하고 나눠 보세요.", "primary_label": "판단", "item_label": "내용", "contrib": false, "categories": [["ok", "넣어도 돼요", ""], ["no", "넣으면 안 돼요", ""]], "items": [{"id": "a", "title": "열 이름: alarm_id, type, score, due_min"}, {"id": "b", "title": "카메라 설치 위치와 근무자 이름이 든 대기열 원본"}, {"id": "c", "title": "직접 지어낸 가상 샘플 2줄"}, {"id": "d", "title": "경보 영상 캡처 원본"}]}]}, {"short": "구조만 알리기", "title": "값은 빼고 구조만 알려 요청하기", "lead": "**민감 정보를 넣은 문장**도 일부러 골라 보세요. 결과는 나와도 무엇이 문제인지 볼 수 있어요.", "page": null, "blocks": [{"id": "s2_vibe", "type": "vibe", "title": "경보 처리 순서표와 절차서", "req": true, "ask": ["seo", "동시에 들어온 경보를 어떤 순서로 볼지 정하고, 절차서 초안까지 받고 싶어요."], "has_start": false, "files": [{"name": "m8_queue.csv", "head": ["alarm_id", "type", "score", "due_min", "camera_pos", "operator"], "rows": [["Q01", "person", "0.91", "3", "SITE-N03", "OPERATOR-A"], ["Q02", "animal", "0.48", "10", "SITE-E11", "OPERATOR-A"], ["Q03", "drone", "0.66", "2", "SITE-W02", "OPERATOR-B"], ["Q04", "video_loss", "", "1", "SITE-S07", "OPERATOR-B"], ["Q05", "vehicle", "0.72", "5", "SITE-N01", "OPERATOR-A"]], "n": 9}], "slots": [{"id": "safe", "label": "넘길 정보", "why": "값은 빼고 구조만 알려요.", "options": [{"i": 1, "text": "대기열 원본 전체를 위치랑 근무자 이름까지 그대로 붙여 넣을게. 여기서 알아서 정리해 줘."}, {"i": 0, "text": "실제 값은 넣지 않을게. m8_queue.csv의 열은 alarm_id, type, score, due_min, camera_pos, operator인데, camera_pos와 operator는 밖으로 나가면 안 되니 결과에도 넣지 말아 줘. 코드는 내 환경에서 돌릴게."}, {"i": 2, "text": ""}]}, {"id": "rule", "label": "규칙", "why": "순서를 숫자로 적어요.", "options": [{"i": 0, "text": "type이 video_loss면 ‘1 즉시 확인’, person이나 drone이면서 score 0.6 이상이면 ‘1 즉시 확인’, 나머지 person, drone, vehicle은 ‘2 우선 확인’, 그 밖은 ‘3 순차 확인’이야. 같은 등급 안에서는 due_min이 짧은 순서로 정렬해 줘. score 빈칸은 오류로 보지 말아 줘."}, {"i": 1, "text": "중요한 것부터 정렬해 줘."}, {"i": 2, "text": ""}]}, {"id": "human", "label": "사람의 몫", "why": "AI가 할 일과 사람이 할 일을 나눠요.", "options": [{"i": 0, "text": "순서표는 제안일 뿐이야. HTML 절차서를 report 변수에 담아 줘. 경보 상태(접수, 영상 확인, 출동 건의, 승인, 종결)마다 담당을 적고, 출동 승인은 당직 책임자만 하며, 경보를 닫을 때 누가·언제·무엇을·왜를 근무 기록에 남긴다고 적어 줘. 맨 위에 ‘AI 초안, 당직 책임자 확인 전’이라고 표시해 줘."}, {"i": 1, "text": "점수가 높으면 AI가 자동으로 출동시키게 해 줘."}, {"i": 2, "text": ""}]}]}]}, {"short": "사람이 할 일", "title": "사람이 확인할 일과 내 프롬프트", "lead": "AI가 순서를 잘 정해도 **결정하는 것은 사람이에요.**", "page": null, "blocks": [{"id": "s3_decide", "type": "decide", "title": "절차 제안 판단하기", "req": true, "text": "절차서 초안을 본 팀원들의 제안이에요. 받아들일지 정하고 이유를 적어 주세요.", "reason_min": 8, "items": [{"id": "a", "who": "oh", "text": "점수가 0.9를 넘으면 영상 확인 없이 바로 출동 건의로 넘겨요. 시간이 아깝잖아요."}, {"id": "b", "who": "jo", "text": "동시에 몰리면 등급 1부터 보고, 처리 못 한 건은 다음 근무자에게 넘길 때 기록에 남겨요."}]}, {"id": "s3_write", "type": "write", "title": "민감 정보를 뺀 프롬프트 쓰기", "req": true, "prompt": "**새 요청**: 실제 근무 기록에서 처리 시간이 10분을 넘은 경보를 세어 보고 싶어요. 기록에는 근무자 이름과 카메라 위치가 들어 있어요. 값을 넘기지 않고 어떻게 요청할까요?", "placeholder": "꼭 입력해 주세요 · 내 말로 프롬프트를 써 보세요", "model_label": "📝 모범 프롬프트", "min": 30}]}]}, "m9": {"steps": [{"short": "먼저 읽기", "title": "고치기 전에 먼저 설명하게 하기", "lead": "숫자가 이상할 때 바로 ‘고쳐 줘’라고 하면 **무엇이 왜 바뀌었는지 모르게 돼요.**", "page": null, "blocks": [{"id": "s1_info", "type": "info", "title": "남의 코드를 고칠 때의 네 단계", "icon": "🧪", "req": false, "text": ["AI가 만든 코드도, 다른 사람이 만든 코드도 같아요. 아래 순서로 요청해요."], "table": {"head": ["단계", "프롬프트에 적을 말", "왜"], "rows": [["설명", "고치기 전에 무엇을 계산하는지 설명해 줘", "내가 이해한 뜻과 같은지 확인해요"], ["테스트", "신호가 0, 10, 50, 60초면 40초 끊김 하나가 나오는 assert를 만들어 줘", "답을 아는 값으로 확인해요"], ["수정", "find_gaps 함수만 고쳐 줘", "다른 곳이 바뀌지 않아요"], ["비교", "고치기 전후를 표로 보여 줘", "무엇이 달라졌는지 봐요"]]}, "figure": "f_review"}, {"id": "s1_q", "type": "quiz", "title": "처음 무엇을 요청할까요?", "req": true, "text": "영상 손실 코드가 이상해 보여요. 가장 먼저 AI에게 무엇을 요청할까요?", "options": ["전부 새로 짜 줘.", "이 코드가 무엇을 어떻게 계산하는지 먼저 설명하고, 의심되는 곳을 알려 줘.", "CAM2가 나오게 아무 값이나 넣어 줘."]}]}, {"short": "테스트로 고치기", "title": "설명·테스트·범위를 담아 요청하기", "lead": "지금 코드는 **오류 없이 돌아요.** 그래서 더 찾기 어려워요.", "page": null, "blocks": [{"id": "s2_vibe", "type": "vibe", "title": "영상 손실 코드 리뷰와 관제 화면", "req": true, "ask": ["oh", "이 코드로 낸 영상 손실이 맞는지 봐 주세요."], "has_start": true, "files": [{"name": "m9_signal.csv", "head": ["camera", "t_sec"], "rows": [["CAM1", "0"], ["CAM1", "10"], ["CAM1", "20"], ["CAM1", "30"], ["CAM1", "40"]], "n": 556}], "slots": [{"id": "explain", "label": "먼저 설명", "why": "고치기 전에 읽게 해요.", "options": [{"i": 0, "text": "m9_signal.csv의 열은 camera, t_sec(신호가 온 시각, 초)야. 30초 넘게 신호가 없으면 영상 손실로 봐야 해. 고치기 전에 find_gaps가 무엇을 무엇과 비교하는지 설명하고, 의심되는 곳을 알려 줘."}, {"i": 1, "text": "그냥 바로 고쳐 줘."}, {"i": 2, "text": ""}]}, {"id": "test", "label": "테스트", "why": "답을 아는 값으로 확인해요.", "options": [{"i": 0, "text": "신호가 0, 10, 50, 60초면 40초 끊김 하나가 나오고, 0, 30, 60초면 끊김이 없는 assert 테스트를 먼저 만들어 줘. 정확히 30초는 끊김이 아니야."}, {"i": 1, "text": "잘 되는지 알아서 확인해 줘."}, {"i": 2, "text": ""}]}, {"id": "scope", "label": "범위와 문서", "why": "고칠 곳과 문서를 정해요.", "options": [{"i": 1, "text": "알아서 깔끔하게 전부 정리해 줘."}, {"i": 0, "text": "find_gaps 함수만 고치고 다른 코드는 그대로 둬. 전후 비교표를 보여 주고, m9_waves.csv(wave_id, alarms, handled_in_3min)로 동시 경보 수별 3분 내 처리율도 계산해 줘. 통합 관제 화면에 들어갈 요소(경보 목록, 상태, 우선순위, 담당자, 배치도, 경보 영상)와 영상 손실 경보를 맨 위에 보여 주는 순서를 HTML로 report 변수에 담아 줘."}, {"i": 2, "text": ""}]}]}]}, {"short": "내 말로 쓰기", "title": "후속 제안 판단과 내 프롬프트", "lead": "결과를 받은 뒤에는 **다음 조치를 사람이** 정해요.", "page": null, "blocks": [{"id": "s3_decide", "type": "decide", "title": "후속 제안 판단하기", "req": true, "text": "결과를 본 팀원들의 제안이에요. 받아들일지 정하고 이유를 적어 주세요.", "reason_min": 8, "items": [{"id": "a", "who": "oh", "text": "끊김이 자주 울리면 번거로우니 기준을 300초로 되돌려요."}, {"id": "b", "who": "min", "text": "끊김이 잦은 CAM3의 배선과 전원을 먼저 점검하고, 같은 도구로 다음 주에 다시 비교해요."}]}, {"id": "s3_write", "type": "write", "title": "리뷰 프롬프트 쓰기", "req": true, "prompt": "**새 요청**: 다른 팀에서 받은 평균 처리 시간 계산 코드가 이상해 보여요. AI에게 어떻게 요청할까요? 고치기 전 설명, 테스트, 범위를 담아 보세요.", "placeholder": "꼭 입력해 주세요 · 내 말로 프롬프트를 써 보세요", "model_label": "📝 모범 프롬프트", "min": 30}]}]}, "m10": {"steps": [{"short": "도구로 묶기", "title": "미니 도구와 브리핑 요청하기", "lead": "지금까지 배운 것을 모두 써 보세요. **네 칸, 숫자 규칙, 근거 수치, 범위**가 모두 필요해요.", "page": null, "blocks": [{"id": "s1_vibe", "type": "vibe", "title": "전후 비교 도구와 브리핑", "req": true, "ask": ["lim", "재조정 전후를 조건별로 비교하고, 브리핑과 자동 점검까지 한 번에 나오게 해 주세요."], "has_start": false, "files": [{"name": "m10_week.csv", "head": ["record_id", "period", "condition", "is_target", "alarmed", "handle_min"], "rows": [["H001", "before", "fog", "0", "0", ""], ["H002", "before", "day", "0", "0", ""], ["H003", "before", "night", "1", "1", "6"], ["H004", "before", "night", "1", "1", "2"], ["H005", "before", "day", "0", "0", ""]], "n": 720}], "slots": [{"id": "tool", "label": "구성", "why": "다시 돌릴 수 있게 해요.", "options": [{"i": 1, "text": "분석 코드 짜 줘."}, {"i": 0, "text": "m10_week.csv의 열은 record_id, period(before·after), condition(day·night·fog), is_target, alarmed, handle_min이야. 파일 이름과 표본 기준을 맨 위 설정값으로 두고, 계산은 함수로 나눠 줘. 다음 달 파일로도 바로 돌릴 수 있게 해 줘."}, {"i": 2, "text": ""}]}, {"id": "metric", "label": "지표", "why": "무엇을 어떻게 나눌지 정해요.", "options": [{"i": 0, "text": "period와 condition별로 대상 수, 재현율(대상 중 alarmed 비율), 하루 평균 불필요 경보(7일 기준), 평균 처리 시간을 표로 만들어 줘. 대상이 30건 미만이면 ‘표본 적음’으로 표시해 줘."}, {"i": 1, "text": "조건별 결과를 보여 줘."}, {"i": 2, "text": ""}]}, {"id": "report", "label": "브리핑", "why": "독자와 조건 표시, 점검을 정해요.", "options": [{"i": 1, "text": "재조정이 성공했다고 브리핑 써 줘."}, {"i": 0, "text": "당직 책임자가 읽을 1쪽 HTML 브리핑을 report 변수에 담아 줘. 순서는 결론 한 줄, 조건별 표, 그림, 확인된 것과 확인되지 않은 것, 남은 위험과 다음 조치, AI 산출물 표기야. 수치에는 가상 기록 몇 건 기준인지 적고, 표본 수와 남은 위험, AI 산출물 표기가 빠졌는지 검사하는 lint_report 함수도 만들어 줘."}, {"i": 2, "text": ""}]}]}]}, {"short": "피드백과 점검", "title": "팀 의견과 올리기 전 점검", "lead": "브리핑을 올리기 전에 **사람이 마지막으로 확인**해요.", "page": null, "blocks": [{"id": "s2_decide", "type": "decide", "title": "브리핑 의견 판단하기", "req": true, "text": "브리핑을 받은 사람들의 의견이에요. 받아들일지 정하고 이유를 적어 주세요.", "reason_min": 8, "items": [{"id": "a", "who": "seo", "text": "‘재조정 완료, 문제없음’이라고 단정해 주세요. 읽는 사람이 헷갈려요."}, {"id": "b", "who": "jo", "text": "맨 위에 결론 한 줄과 제가 결정할 일을 두고, 표는 그 아래에 놓아 주세요."}]}, {"id": "s2_multi", "type": "multi", "title": "올리기 전에 확인할 것", "req": true, "text": "AI가 만든 브리핑을 올리기 전에 꼭 확인할 것을 모두 골라 주세요.", "items": ["표의 숫자가 파일에서 다시 계산한 값과 같은지 한두 개 대조한다.", "수치마다 데이터와 표본 수, 조건이 적혀 있는지 본다.", "문장이 매끄러우면 내용도 맞으니 건너뛴다.", "확인되지 않은 내용이 확인된 것처럼 적히지 않았는지 본다.", "AI가 만든 문서이니 작성 책임은 AI에게 있다고 적는다."]}]}, {"short": "프롬프트 모음", "title": "나만의 프롬프트 틀 만들기", "lead": "이번 프로젝트의 **가장 큰 산출물**이에요. 다음 업무에서 바로 쓸 나만의 틀을 남겨요.", "page": null, "blocks": [{"id": "s3_write", "type": "write", "title": "내 프롬프트 틀 쓰기", "req": true, "prompt": "앞으로 자주 쓸 **프롬프트 틀**을 써 보세요. 파일 이름이나 숫자처럼 바뀌는 자리는 [대괄호]로 비워 두세요. 네 칸에 더해 검증, 범위, 민감 정보 내용도 넣어 보세요.", "placeholder": "꼭 입력해 주세요 · 내 말로 프롬프트를 써 보세요", "model_label": "📝 모범 프롬프트", "min": 40}]}]}};   // 미션 단계와 블록 (정답·힌트·풀이는 빠져 있음)
// ===================== 상태 =====================
let A = null, view = 'welcome', seen = {}, mstep = {}, cache = {}, drafts = {};
let ui = { w: 'full', h: 900, side: true, code: null };
try { Object.assign(ui, JSON.parse(localStorage.getItem(PROJ.ui_key) || '{}')); } catch (e) {}
try { google.colab.output.setIframeHeight(0, true, { maxHeight: 5000 }); } catch (e) {}
const $ = id => document.getElementById(id);
const NUM = n => '①②③④⑤⑥⑦⑧⑨⑩⑪⑫'[n - 1] || n;

function h(tag, props, ...kids) {
  const e = document.createElement(tag);
  for (const [k, v] of Object.entries(props || {})) {
    if (v === undefined || v === null || v === false) continue;
    if (k === 'class') e.className = v;
    else if (k === 'html') e.innerHTML = v;            // 엔진·화면에 적힌 신뢰할 수 있는 문구에만 사용
    else if (k.startsWith('on')) e[k] = v;
    else if (k === 'style') e.style.cssText = v;
    else if (k === 'value') e.value = v;
    else e.setAttribute(k, v === true ? '' : v);
  }
  for (const c of kids.flat(9)) if (c !== null && c !== undefined && c !== false) e.append(c instanceof Node ? c : document.createTextNode(String(c)));
  return e;
}
const esc = s => String(s).replace(/&/g, '&amp;').replace(/</g, '&lt;').replace(/>/g, '&gt;');

let busy = 0;
async function invoke(name, ...args) {
  busy++; $('ws-root').classList.add('busy');
  try {
    const res = await google.colab.kernel.invokeFunction(name, args, {});
    const d = res.data['application/json'];
    if (d.app) refresh(d.app);
    if (d.msg) note(d.msg, d.ok === false ? 'bad' : 'ok');
    if (d.code) setLive(d.code, d.fn);
    return d;
  } catch (e) {
    note('⚠️ 파이썬과 연결되지 않았어요. 위의 🚀 [먼저 실행] 실습 준비하기 셀과 이 셀을 차례로 다시 실행해 주세요.', 'bad');
    throw e;
  } finally { busy--; if (!busy) $('ws-root').classList.remove('busy'); }
}
function note(t, kind) { const n = $('note'); n.textContent = t || ''; n.className = 'note' + (kind ? ' ' + kind : ''); }

// ===================== 화면 틀 =====================
function codeOn() { return ui.code === null ? $('ws-root').offsetWidth >= 1250 : ui.code; }
function applyUi() {
  const root = $('ws-root');
  root.classList.remove('w-base', 'w-wide', 'w-full');
  root.classList.add('w-' + ui.w);
  root.style.width = '';
  root.style.height = ui.h + 'px';
  root.classList.toggle('side-off', !ui.side);
  root.classList.toggle('code-on', codeOn());
  root.classList.toggle('compact', root.offsetWidth < 1000);
  document.querySelectorAll('.ws [data-w]').forEach(b => b.classList.toggle('on', b.dataset.w === ui.w));
  $('tb-side').classList.toggle('on', !ui.side);
  $('tb-code').classList.toggle('on', codeOn());
  try { localStorage.setItem(PROJ.ui_key, JSON.stringify(ui)); } catch (e) {}
}
function setWidth(w) { ui.w = w; applyUi(); }
function setHeight(d) { ui.h = Math.max(480, Math.min(2000, $('ws-root').offsetHeight + d)); applyUi(); }
function toggleSide() { ui.side = !ui.side; applyUi(); }
function toggleCode(force) { ui.code = force === true ? true : !codeOn(); applyUi(); if (codeOn()) renderCode(); }
new ResizeObserver(() => {
  const root = $('ws-root');
  root.classList.toggle('compact', root.offsetWidth < 1000);
  root.classList.toggle('code-on', codeOn());
  $('tb-code').classList.toggle('on', codeOn());
  const hh = root.offsetHeight;
  if (hh && Math.abs(hh - ui.h) > 2) { ui.h = hh; try { localStorage.setItem(PROJ.ui_key, JSON.stringify(ui)); } catch (e) {} }
}).observe($('ws-root'));
new ResizeObserver(() => { const m = $('main'); m.classList.toggle('narrow', m.offsetWidth < 760); }).observe($('main'));

// ===================== 상태 갱신 =====================
function refresh(app) {
  const first = !A;
  if (A && app.boot && A.boot && app.boot !== A.boot) {   // 파이썬이 새로 시작됨(런타임 재시작·셀 재실행) → 화면도 처음 상태로
    mstep = {}; cache = {}; drafts = {}; BUSY = null; for (const k in POLL) clearTimeout(POLL[k]);
    A = app; view = 'welcome'; renderTop(); renderSide(); renderMain();
    note(PROJ.boot_note, 'bad');
    return;
  }
  A = app;
  renderTop(); renderSide();
  if (first || !view.startsWith('mission:')) renderMain();
}
function openView(v) { if (v !== view && busyBlock()) return; view = v; renderSide(); renderMain(); }

// ===================== 상단: 진행률 · 내 정보 =====================
function renderTop() {
  const o = A.overall;
  $('tb-prog').innerHTML = '';
  $('tb-prog').append(h('span', { class: 'pb' }, h('i', { style: `width:${o.pct}%` })), h('b', {}, `전체 진행 ${o.pct}%`), h('span', {}, `· 미션 ${o.done}/${o.total} 완료`));
  $('tb-me').textContent = `👤 ${A.call}`;
}

// ===================== 첫 화면: 입력 없이 인사만 (로그인 없음) =====================
function renderWelcome(main) {
  main.append(h('div', { class: 'scroll' }, h('div', { class: 'onb welcome' },
    h('div', { class: 'wave' }, '👋'), h('h2', {}, `어서오세요, ${A.call}`),
    h('p', { class: 'lead' }, PROJ.welcome_lead),
    h('div', { class: 'row', style: 'justify-content:center;margin-top:18px' }, h('button', { class: 'btn big pulse', id: 'go-home', onclick: () => openView('home') }, '시작하기 ▶')))));
}

// ===================== 사이드바 =====================
const ICON = { done: '✅', open: '▶', soon: '⏳' };
function renderSide() {
  const side = $('side'); side.innerHTML = '';
  $('ws-root').classList.toggle('onboard', view === 'welcome');
  if (view === 'welcome') return;
  const item = (key, icon, label, extra, cls) => side.append(h('div', { class: 'item' + (view === key ? ' active' : '') + (cls ? ' ' + cls : ''), title: label, onclick: () => openView(key) },
    h('span', { class: 'ic' }, icon), h('span', { class: 't' }, label), extra));
  item('home', '🏠', '프로젝트 개요');
  side.append(h('h4', {}, h('b', {}, '채널')));
  let total = 0;
  for (const [id, name] of Object.entries(A.channels)) {
    const n = A.msgs.filter(m => m.ch === id && m.id >= (seen[id] || 0)).length;
    total += n;
    item(id, '', name, n ? h('span', { class: 'badge' }, n) : null, n ? 'unread' : '');
  }
  for (const s of A.project.stages) {
    side.append(h('h4', {}, h('b', {}, s.no + '단계'), ' · ' + s.name));
    for (const mid of s.missions) {
      const m = A.missions[mid];
      const comp = (A.views[mid] || {}).complete || [];
      const mini = m.status === 'open' && m.progress.started ? h('span', { class: 'mini' }, `${comp.filter(Boolean).length}/${comp.length}`) : null;
      item('mission:' + mid, ICON[m.status], `M${m.no} ${m.short}`, mini, m.status === 'soon' ? 'soon' : '');
    }
  }
  side.append(h('h4', {}, h('b', {}, '앱')));
  item('files', '📁', '파일', A.files.length ? h('span', { class: 'badge', style: 'background:#1164a3' }, A.files.length) : null);
  item('roster', '👥', '팀원');
  $('tb-count').textContent = total;
  $('tb-count').classList.toggle('has', total > 0);
}

// ===================== 안내 띠 =====================
function nextTodo() {
  for (const s of A.project.stages) for (const mid of s.missions) { const m = A.missions[mid]; if (m.status === 'open') return m; }
  return null;
}
// ===================== 본 과정의 코드 보기 =====================
const CODE = { topic: null, data: {}, live: null, liveFn: null, open: null };
function hl(code) {
  const re = /(#[^\n]*)|("{3}[\s\S]*?"{3}|'{3}[\s\S]*?'{3}|\b[rbfu]{0,2}"(?:\\.|[^"\\\n])*"|\b[rbfu]{0,2}'(?:\\.|[^'\\\n])*'|"(?:\\.|[^"\\\n])*"|'(?:\\.|[^'\\\n])*')|\b(def|return|if|elif|else|for|in|while|import|from|as|with|not|and|or|is|None|True|False|lambda|try|except|class|yield|global|continue|break|pass)\b|\b(\d+\.?\d*(?:e[-+]?\d+)?)\b|\b([A-Za-z_]\w*)(?=\()/g;
  let out = '', last = 0, m;
  while ((m = re.exec(code))) {
    out += esc(code.slice(last, m.index));
    const cls = m[1] ? 'c' : m[2] ? 's' : m[3] ? 'k' : m[4] ? 'n' : 'f';
    out += `<span class="${cls}">${esc(m[0])}</span>`;
    last = re.lastIndex;
  }
  return out + esc(code.slice(last));
}
function copyText(t) {
  const done = () => note('📋 코드를 복사했어요. 노트북에 새 코드 셀을 만들고 붙여 넣어 실행해 보세요.', 'ok');
  const fallback = () => {          // Colab 출력 창에서 클립보드 권한이 막혀 있을 때
    const ta = h('textarea', { style: 'position:fixed;left:-9999px;top:0' }); ta.value = t; document.body.append(ta); ta.select();
    let ok = false; try { ok = document.execCommand('copy'); } catch (e) {}
    ta.remove();
    if (ok) done(); else note('복사하지 못했어요. 코드를 드래그해서 복사해 주세요.', 'bad');
  };
  try { navigator.clipboard.writeText(t).then(done, fallback); } catch (e) { fallback(); }
}
async function setTopic(topic) {
  CODE.topic = topic;
  if (topic && !CODE.data[topic]) { try { CODE.data[topic] = await invoke('app.code', topic); } catch (e) {} }
  renderCode();
}
function setLive(code, fn) { CODE.live = code; CODE.liveFn = fn || null; if (fn) CODE.open = fn; renderCode(); }
function codeBlock(b, open) {
  const lines = b.code.split('\n');
  const cw = h('div', { class: 'cw' });
  for (const i of b.hl) cw.append(h('div', { class: 'hlb', style: `top:${8 + i * 18}px` }));
  cw.append(h('pre', { class: 'code', html: hl(b.code) }));
  return h('details', { open: open, 'data-fn': b.name }, h('summary', {}, h('code', {}, b.name + '()'), b.doc ? ': ' + b.doc : '', ' ',
      h('button', { class: 'cpy', onclick: e => { e.preventDefault(); copyText(b.code); } }, '📋 복사')),
    h('div', { class: 'cb' }, h('pre', { class: 'ln' }, lines.map((_, i) => i + 1).join('\n')), cw));
}
// 📘 본 과정의 가이드북: [핵심 용어] 탭(미션 소개의 용어 + 단계마다 나오는 어려운 말) · [코드 보기] 탭(지금 화면의 파이썬 코드)
const GUIDE = { tab: 'terms', focus: null };
function openGuide(tab, focus) { GUIDE.tab = tab || 'terms'; GUIDE.focus = focus || null; toggleCode(true); renderCode(); }
function guideTerms(sc) {
  const mid = view.startsWith('mission:') ? view.slice(8) : null, m = mid ? A.missions[mid] : null;
  if (!m || !m.brief) { sc.append(h('p', { class: 'desc' }, '미션을 열면 그 미션의 핵심 용어가 여기에 나와요.')); return; }
  const st = curStep(mid), g = m.brief.glossary || [];
  sc.append(h('div', { class: 'intro' }, '미션 소개에 나온 ', h('b', {}, '핵심 용어'), '와, 단계마다 나오는 어려운 말을 다시 볼 수 있어요. ',
    h('b', {}, '노란 테두리'), '는 지금 단계에 나오는 말이에요. 코드는 위의 ', h('b', {}, '💻 코드 보기'), ' 탭에 있어요.'));
  // 단계 이름은 머리글로 한 번만: 용어 이름과 같아도 헷갈리지 않게 ‘N단계 「이름」’으로 써요 (2026-10-20 제작자 피드백)
  const stepName = k => `${k}단계 「${m.brief.steps_short[k - 1] || ''}」`;
  const now = g.filter(t => t.step === st), rest = g.filter(t => t.step !== st);
  if (now.length) sc.append(h('h6', {}, `🔎 지금 ${stepName(st)}에 나오는 말`), ...now.map(t => termCard(t, true)));
  sc.append(h('h6', {}, `📖 M${m.no} 핵심 용어 (미션 소개)`), ...m.brief.terms.map(t => termCard(t)));
  if (rest.length) {
    const byStep = [...new Set(rest.map(t => t.step))].sort((a, b) => a - b);
    sc.append(h('details', { open: !!(GUIDE.focus && rest.some(t => t.t === GUIDE.focus)) }, h('summary', {}, `M${m.no}의 다른 단계에 나오는 말 (${rest.length}개)`),
      ...byStep.map(k => h('div', { class: 'gstep' }, h('div', { class: 'gstep-h' }, `🧭 ${stepName(k)}에 나오는 말`), ...rest.filter(t => t.step === k).map(t => termCard(t))))));
  }
  const others = Object.values(A.missions).filter(x => x.brief && x.id !== mid);
  if (others.length) sc.append(h('details', {}, h('summary', {}, '다른 미션의 핵심 용어'),
    ...others.map(x => h('div', {}, h('h6', {}, `M${x.no}. ${x.title}`), ...x.brief.terms.map(t => termCard(t))))));
  if (GUIDE.focus) {
    const el = [...sc.querySelectorAll('.term')].find(e => e.dataset.t === GUIDE.focus);
    if (el) { el.classList.add('flash'); setTimeout(() => el.scrollIntoView({ block: 'start', behavior: 'smooth' }), 30); }
    GUIDE.focus = null;
  }
}
function guideCode(sc) {
  const d = CODE.data[CODE.topic];
  if (!(d && d.ok) && !CODE.live) { sc.append(h('p', { class: 'desc' }, '이 화면에는 따로 보여 줄 코드가 없어요. 직접 계산하거나 파일을 읽는 활동에서 그 코드를 보여 줘요.')); return; }
  sc.append(h('div', { class: 'intro' }, '지금 화면이 파이썬 코드로 어떻게 동작하는지 보여 줘요. 고칠 필요는 없어요. ',
    h('b', {}, '📋 복사'), '를 눌러 노트북의 새 코드 셀에 붙여 넣고 실행하면 같은 결과를 직접 볼 수 있어요. ', h('b', {}, '노란 줄'), '은 이 단계의 핵심 부분이에요.'));
  sc.append(h('h6', {}, '▶ 방금 실행된 코드', h('span', { class: 'sp' }), CODE.live ? h('button', { class: 'cpy', onclick: () => copyText(CODE.live) }, '📋 복사') : null),
    h('div', { class: 'live' }, CODE.live ? h('pre', { html: hl(CODE.live) }) : h('div', { class: 'empty' }, '버튼을 누르거나 조건을 바꾸면, 그때 실행된 코드가 여기에 나타나요.')));
  if (d && d.ok) {
    sc.append(h('h6', {}, '📄 ' + d.title), h('p', { class: 'desc' }, d.desc));
    const openName = d.blocks.some(b => b.name === CODE.open) ? CODE.open : d.blocks[0].name;
    for (const b of d.blocks) sc.append(codeBlock(b, b.name === openName));
  } else sc.append(h('p', { class: 'desc' }, '이 화면에는 따로 보여 줄 코드가 없어요.'));
}
function renderCode() {
  const p = $('codep');
  if (!codeOn()) return;
  p.innerHTML = '';
  const sc = h('div', { class: 'cs' });
  const tabs = h('div', { class: 'gtabs' }, [['terms', '📖 핵심 용어'], ['code', '💻 코드 보기']].map(([k, t]) =>
    h('button', { class: GUIDE.tab === k ? 'on' : '', onclick: () => { GUIDE.tab = k; renderCode(); } }, t)));
  if (GUIDE.tab === 'terms') guideTerms(sc); else guideCode(sc);
  p.append(h('div', { class: 'ch' }, h('b', {}, '📘 본 과정의 가이드북'), h('button', { onclick: () => toggleCode() }, '닫기 ✕')), tabs, sc);
  if (GUIDE.tab === 'code') {
    const opened = sc.querySelector(`details[data-fn="${CODE.open}"]`);
    if (opened && CODE.liveFn) opened.scrollIntoView({ block: 'nearest' });
  }
}

// ===================== 공통 위젯 =====================
function taskBox() { return { el: document.createComment('task'), set() {} }; }   // 할 일 상자는 화면에서 뺐어요
function choiceGroup(name, options, value, onchange, opts) {
  opts = opts || {};
  const wrap = h('div', { class: 'choices' + (opts.col ? ' col' : '') });
  options.forEach((o, i) => {
    const [v, label, small] = Array.isArray(o) ? o : [i, o];
    const inp = h('input', { type: 'radio', name, value: v, checked: value !== null && value !== undefined && String(value) === String(v) });
    inp.onchange = () => { wrap.querySelectorAll('.choice').forEach(l => { l.classList.toggle('on', l.querySelector('input').checked); l.classList.remove('wrong'); }); if (onchange) onchange(String(v)); };
    wrap.append(h('label', { class: 'choice' + (inp.checked ? ' on' : '') }, inp, h('span', {}, label, small ? h('small', {}, small) : null)));
  });
  wrap.value = () => { const s = wrap.querySelector('input:checked'); return s ? s.value : ''; };
  return wrap;
}
function textCounter(ta, min) {
  const line = h('div', { class: 'count-line' });
  const upd = () => { const n = ta.value.trim().length; line.textContent = n >= min ? `✔ ${n}자 (충분해요)` : `지금 ${n}자 · ${min}자 이상 적어야 해요`; line.className = 'count-line ' + (n >= min ? 'enough' : 'short'); };
  ta.addEventListener('input', upd); upd();
  return line;
}
function flagMissing(el, msgEl, msg) {
  if (msgEl) msgEl.textContent = msg ? '⚠ ' + msg : '';
  if (el) el.classList.toggle('missing', !!msg);
}
function scrollTo(el) { if (el) el.scrollIntoView({ block: 'center', behavior: 'smooth' }); }
function nextBtn(mid, k, ready, why, label) {
  const msg = h('div', { class: 'err' });
  const btn = h('button', { class: 'btn', onclick: () => { const r = ready ? ready() : true; if (r) goStep(mid, k); else { msg.textContent = '⚠ ' + why(); } } }, label || '다음 단계 ▶');
  return { el: h('div', {}, h('div', { class: 'row' }, btn), msg), btn, msg };
}
// 틀린 선택 표시: 고른 보기를 빨간색으로 (다른 보기를 고르면 choiceGroup이 지움)
function markWrong(group) { const inp = group && group.querySelector('input:checked'); if (inp) inp.closest('.choice').classList.add('wrong'); }
function figure(key) { return FIG[key] ? h('div', { class: 'fig', html: FIG[key] }) : null; }

// ===================== 그림 (용어 설명) =====================
// 핵심 용어 카드에 붙는 작은 SVG 그림. 프로젝트 명세의 figures가 여기에 들어가요 (termCard의 fig 이름으로 찾음).
const T = 'font-family="sans-serif" font-size="10" fill="#111827"';
const FIG = Object.assign({}, PROJ_FIGS);

// ===================== 홈 · 채팅 · 팀원 · 파일 =====================
function renderMain() {
  if (!A) return;
  $('ws-root').classList.add('no-code');          // 코드 보기는 미션 실습 단계에서만 보여요
  const main = $('main'); main.innerHTML = '';
  if (view === 'welcome') return renderWelcome(main);
  if (view === 'home') { return renderHome(main); }
  if (view === 'files') { return renderFiles(main); }
  if (view === 'roster') { return renderTeam(main); }
  if (view.startsWith('mission:')) return renderMission(main, view.slice(8));
  if (A.channels[view]) { return renderChat(main, view); }
  view = 'home'; renderHome(main);
}
function titleBar(main, text, small) { main.append(h('div', { class: 'title' }, small ? h('small', {}, small) : null, text)); }

function renderHome(main) {
  titleBar(main, '프로젝트 개요', '🏠');
  const sc = h('div', { class: 'scroll' });
  const o = A.overall;
  sc.append(h('div', { class: 'hero' }, h('div', { style: 'font-size:12px' }, '프로젝트'), h('h2', {}, A.project.title), h('p', {}, A.project.goal)));
  if (A.teacher) sc.append(h('div', { class: 'tip', style: 'margin-top:12px' }, '🧑‍🏫 선생님용 파일이에요. 모든 활동의 값이 채워져 있고, 산출물도 📁 파일에 만들어져 있어요.'));
  if (A.assets && !A.assets.ok) sc.append(assetNotice());
  sc.append(h('div', { class: 'greet' }, `👋 어서오세요, ${A.call}.`),
    h('div', { class: 'card' }, h('div', { class: 'bigprog' }, h('b', {}, '전체 진행'), h('span', { class: 'pb' }, h('i', { style: `width:${o.pct}%` })), h('b', {}, o.pct + '%')),
      h('div', { class: 'hint', style: 'margin-top:6px' }, `미션 ${o.total}개 중 ${o.done}개 완료 · 끝낸 활동만큼 올라가요.`)));
  const nx = nextTodo();
  if (nx) sc.append(h('div', { class: 'row' }, h('button', { class: 'btn big pulse', onclick: () => openView('mission:' + nx.id) },
    `▶ ${nx.progress.started ? '이어서 하기' : '시작하기'}: M${nx.no} ${nx.title}`)));
  const grid = h('div', { class: 'stages' });
  for (const s of A.project.stages) {
    const box = h('div', { class: 'stage' }, h('h5', {}, `${s.no}단계 · ${s.name}`), s.subtitle ? h('div', { class: 'hint stage-sub' }, s.subtitle) : null);
    for (const mid of s.missions) {
      const m = A.missions[mid];
      const label = { done: '완료', open: '진행 가능', soon: '준비 중' }[m.status];
      box.append(h('div', { class: 'mcard', onclick: () => openView('mission:' + mid) }, h('span', {}, ICON[m.status]),
        h('div', { style: 'flex:1' }, h('div', {}, h('b', {}, `M${m.no} `), m.title), m.brief && m.brief.subtitle ? h('div', { class: 'hint msub' }, m.brief.subtitle) : null, h('div', { class: 'hint' }, m.kind === '협업' ? '👥 협업' : '👤 개인')),
        h('span', { class: 'pill ' + m.status }, label)));
    }
    grid.append(box);
  }
  sc.append(h('h4', { style: 'margin:18px 0 0' }, '단계와 미션'), grid);
  main.append(sc);
}

function renderChat(main, ch) {
  seen[ch] = A.msgs.length;
  titleBar(main, A.channels[ch]);
  const box = h('div', { class: 'msgs' });
  for (const m of A.msgs.filter(m => m.ch === ch)) {
    const who = A.members[m.who];
    box.append(h('div', { class: 'msg' + (m.who !== 'me' && m.text.includes(A.call) ? ' mention' : '') },
      h('div', { class: 'avatar', style: 'background:' + who.color }, who.name[0]),
      h('div', {}, h('div', { class: 'meta' }, h('b', {}, who.name), who.role, h('span', { class: 'tag' + (who.ext ? ' ext' : '') }, who.org), h('span', { class: 'time' }, m.time)),
        h('div', { class: 'text' }, m.text))));
  }
  const input = h('input', { type: 'text', placeholder: '메시지 보내기 (Enter) · 가상 팀원은 미션 진행에 맞춰 메시지를 보내요' });
  const send = () => { if (input.value.trim()) { invoke('app.send', ch, input.value); input.value = ''; } };
  input.addEventListener('keydown', e => { if (e.key === 'Enter' && !e.isComposing) send(); });
  main.append(box, h('div', { class: 'compose' }, input, h('button', { class: 'btn', onclick: send }, '보내기')));
  box.scrollTop = box.scrollHeight;
  renderSide();
}

function renderTeam(main) {
  titleBar(main, '팀원', '👥');
  const sc = h('div', { class: 'scroll' });
  for (const m of Object.values(A.members)) sc.append(h('div', { class: 'msg' }, h('div', { class: 'avatar', style: 'background:' + m.color }, m.name[0]),
    h('div', {}, h('b', {}, m.name), ' ', m.role, h('span', { class: 'tag' + (m.ext ? ' ext' : '') }, m.org))));
  main.append(sc);
}

function renderFiles(main, openId) {
  titleBar(main, '파일', '📁');
  const sc = h('div', { class: 'scroll' });
  main.append(sc);
  if (!A.files.length) { sc.append(h('div', { class: 'tip' }, '아직 산출물이 없어요. 미션의 [산출물] 단계에서 받으면 여기에 저장돼요.')); return; }
  sc.append(h('div', { class: 'tip' }, '💾 산출물은 노트북이 켜져 있는 동안만 보관돼요. 받을 때마다 [⬇ 내려받기]로 내 컴퓨터에 저장해 두세요.'));
  const prev = h('div');
  for (const f of A.files) sc.append(h('div', { class: 'file' }, h('b', {}, '📄 ' + f.title), h('span', { class: 'hint' }, `  ${f.name} · M${A.missions[f.mission].no}`), f.blanks ? h('span', { class: 'pill soon', style: 'margin-left:6px' }, `안 한 활동 ${f.blanks}개`) : null,
    h('div', { class: 'row' }, h('button', { class: 'btn small ghost', onclick: () => showFile(f, prev) }, '미리보기'), h('button', { class: 'btn small', onclick: () => downloadFile(f) }, '⬇ 내려받기'))));
  sc.append(prev);
  showFile(A.files.find(f => f.id === openId) || A.files[A.files.length - 1], prev);
}
async function showFile(f, box) { const d = await invoke('app.file', f.id); box.innerHTML = ''; box.append(h('h4', {}, '미리보기: ' + f.title), h('iframe', { class: 'preview', srcdoc: d.html })); }
async function downloadFile(f) {
  const d = await invoke('app.file', f.id);
  const a = h('a', { href: URL.createObjectURL(new Blob([d.html], { type: 'text/html' })), download: d.name });
  document.body.append(a); a.click(); a.remove();
}
function openFile(id) { if (busyBlock()) return; view = 'files'; renderSide(); const main = $('main'); main.innerHTML = ''; renderFiles(main, id); }

// ===================== 미션 공통 =====================
function nSteps(m) { return m.brief ? m.brief.steps_short.length : 0; }
function curStep(mid) { return mstep[mid] !== undefined ? mstep[mid] : 0; }
function goStep(mid, k) { if (busyBlock()) return; mstep[mid] = k; renderSide(); renderMain(); }
function renderMission(main, mid) {
  const m = A.missions[mid];
  if (m.status === 'soon' || !m.brief) {
    setTopic(null);
    main.append(h('div', { class: 'scroll brief' }, h('span', { class: 'pill soon' }, '준비 중'), h('h2', {}, `M${m.no}. ${m.title}`),
      h('p', { class: 'sub-t' }, (m.kind === '협업' ? '👥 협업' : '👤 개인') + ' 미션'),
      h('section', {}, h('h4', {}, '🎯 미션 소개'), h('p', {}, m.outline)), h('div', { class: 'tip' }, '이 미션은 아직 준비 중이에요. 열려 있는 미션을 먼저 진행해 주세요.')));
    return;
  }
  const n = nSteps(m), st = curStep(mid);
  const bar = h('div', { class: 'stepper' });
  const comp = (A.views[mid] || {}).complete || [];
  // 잠그지 않음: 뒷부분을 미리 볼 수 있게 모든 단계를 열어 둠. 처음 들어갈 때 시작 메시지만 보냄
  ['미션 소개', ...m.brief.steps_short.map((s, i) => `${i + 1} ${s}`), '📦 산출물'].forEach((name, i) => {
    const ok = i >= 1 && i <= n && comp[i - 1];
    bar.append(h('button', { class: 'stp' + (i === st ? ' cur' : '') + (ok ? ' done' : ''), onclick: async () => {
      if (i >= 1 && !m.progress.started) await invoke('app.start', mid);
      goStep(mid, i);
    } }, (ok ? '✅ ' : '') + name));
  });
  const body = h('div', { class: 'scroll' });
  main.append(bar, body);
  const practice = st >= 1 && st <= n;
  $('ws-root').classList.toggle('no-code', !practice);
  setTopic(practice ? TOPICS[mid][st] : null);
  STEPS[mid][Math.min(st, n + 1)](body, m);
}

function termCard(t, now) {   // 핵심 용어 카드: 미션 소개와 📘 가이드북이 함께 씀 (가이드북 용어에는 수식·더 알아보기가 붙음)
  const rows = [['k1', '한 줄 정의', t.d], ['k2', '쉽게 말하면', t.easy], ['k3', PROJ.use_label, t.use], ['k4', '수식', t.formula], ['k5', '더 알아보기', t.more]].filter(r => r[2]);
  return h('div', { class: 'term' + (now ? ' now' : ''), 'data-t': t.t }, h('h5', {}, '📖 ' + t.t, t.en ? h('small', {}, ' ' + t.en) : null), figure(t.fig),
    h('dl', {}, rows.map(([k, a, b]) => [h('dt', { class: k }, a), h('dd', { class: k === 'k4' ? 'fx' : '' }, b)])));
}
function startButtons(m) {
  const go = async () => { if (!m.progress.started) await invoke('app.start', m.id); goStep(m.id, firstTodoStep(m)); };
  return h('button', { class: 'btn big' + (m.progress.started ? '' : ' pulse'), onclick: go }, m.progress.started ? '이어서 하기 ▶' : '미션 시작 ▶');
}
function firstTodoStep(m) { const c = (A.views[m.id] || {}).complete || []; const k = c.findIndex(x => !x); return k < 0 ? 1 : k + 1; }
// 해외 사례: 기사 주소를 달지 않고, 기사의 주요 내용을 눌러서 열어 보는 토글 (2026-11-05 피드백)
function caseToggle(cs) {
  if (!cs) return null;
  return h('details', { class: 'case' }, h('summary', {}, '📰 ' + cs.title),
    h('div', { class: 'case-b' }, (cs.points || []).map(t => h('p', { html: md(t) })), h('div', { class: 'hint' }, '출처: ' + cs.source)));
}
function assetNotice(extra) {
  const S = A.assets || {};
  return h('div', { class: 'card focus' }, h('h4', {}, '📷 참고자료(사진)를 넣어 주세요'),
    h('p', {}, '사진이 필요한 활동이 있어요. 왼쪽 📁 파일 패널에서 아래 폴더에 받은 사진(또는 zip)을 넣은 뒤 [다시 확인]을 눌러 주세요.'),
    h('pre', { class: 'vb-out' }, S.folder || ''), extra || null,
    S.missing && S.missing.length ? h('div', { class: 'hint' }, `아직 없는 사진 ${S.missing.length}개: ${S.missing.slice(0, 8).join(', ')}${S.missing.length > 8 ? ' …' : ''}`) : null,
    h('div', { class: 'row' }, h('button', { class: 'btn', onclick: async () => { await invoke('app.assets_check'); cache.asset = {}; renderMain(); } }, '다시 확인')));
}
function renderBrief(body, m) {
  const b = m.brief;
  const sec = (title, ...kids) => h('section', {}, h('h4', {}, title), ...kids);
  body.append(h('div', { class: 'brief' },
    h('h2', {}, `M${m.no}. ${m.title}`), h('p', { class: 'sub-t' }, b.subtitle),
    sec('🎯 미션 목표', h('p', {}, b.goal)),
    h('div', { class: 'row' }, startButtons(m)),
    h('section', { class: 'why' }, h('h4', {}, '📰 왜 중요할까요?'), h('p', { style: 'margin:0' }, b.why),
      caseToggle(b.case)),
    sec('🧭 수행 단계', h('ol', {}, b.steps.map(s => h('li', {}, s)))),
    sec('📦 산출물', h('ul', {}, b.outputs.map(s => h('li', {}, s)))),
    b.comps && b.comps.length ? sec('🧩 이 미션에서 기르는 역량', h('ul', {}, b.comps.map(c => h('li', {}, h('b', {}, c.name), c.criteria ? ': ' + c.criteria : '')))) : null,
    b.team ? sec('👥 함께하는 사람', h('p', {}, b.team)) : null,
    sec('📖 핵심 용어', h('p', { class: 'hint', style: 'margin:0 0 8px' }, `용어마다 ‘한 줄 정의 → 쉽게 말하면 → ${PROJ.use_label}’ 순서로 설명했어요. 미션 중에도 이 화면으로 돌아와 다시 볼 수 있어요.`),
      h('div', { class: 'terms' }, b.terms.map(termCard))),
    h('div', { class: 'row', style: 'margin-top:16px' }, startButtons(m))));
}
// 산출물 단계: 언제든 받을 수 있음. 안 한 활동이 있으면 먼저 알려 주고 물어본 뒤, 그래도 받으면 그 부분은 빈칸으로 만듦 (2026-11-05 피드백)
function todoList(m, todo) {
  return h('ul', { class: 'todo' }, todo.map(t => h('li', {}, h('a', { href: '#', onclick: e => { e.preventDefault(); goStep(m.id, t.step); } }, `${t.step}단계 「${t.short}」`), ': ', t.blocks.join(', '))));
}
function doneView(body, m, files, nextMid) {
  const area = h('div'), box = h('div', { class: 'brief' }, h('h2', {}, `📦 M${m.no} 산출물`), h('p', { class: 'sub-t' }, m.title), area);
  body.append(box);
  const draw = async () => {
    const r = await invoke('blk.todo', m.id), todo = r.todo || [];
    area.innerHTML = '';
    const mine = A.files.filter(f => f.mission === m.id);
    area.append(h('section', {}, h('h4', {}, '📋 이 미션의 진행 상황'),
      todo.length ? todoList(m, todo) : h('div', { class: 'fb ok' }, '✅ 필수 활동을 모두 마쳤어요.')));
    const ask = h('div');
    area.append(h('section', {}, h('h4', {}, '📄 받을 문서'), h('ul', {}, files.map(([, t]) => h('li', {}, t))),
      h('div', { class: 'row' }, h('button', { class: 'btn big' + (todo.length ? '' : ' pulse'), id: 'get-output', onclick: async () => {
        const d = await invoke('blk.finish', m.id);
        if (!d.need_confirm) return draw();
        ask.innerHTML = '';
        ask.append(h('div', { class: 'card focus', id: 'confirm-box' }, h('h4', {}, '⚠ 아직 하지 않은 활동이 있어요'), todoList(m, d.todo),
          h('p', {}, '그래도 산출물을 받을까요? 하지 않은 부분은 빈칸으로 나와요.'),
          h('div', { class: 'row' }, h('button', { class: 'btn ghost', onclick: () => goStep(m.id, d.todo[0].step) }, '마저 하러 가기'),
            h('button', { class: 'btn', id: 'force-output', onclick: async () => { await invoke('blk.finish', m.id, true); draw(); } }, '그래도 받기'))));
      } }, mine.length ? '📦 산출물 다시 받기' : '📦 산출물 받기')), ask));
    if (mine.length) area.append(h('section', {}, h('h4', {}, '📁 저장된 산출물'), mine.map(f => h('div', { class: 'row' }, h('span', {}, '📄 ' + f.title),
      f.blanks ? h('span', { class: 'pill soon' }, `안 한 활동 ${f.blanks}개`) : null, h('button', { class: 'btn small ghost', onclick: () => openFile(f.id) }, '열어 보기'))),
      h('div', { class: 'tip' }, '💾 산출물은 노트북이 켜져 있는 동안만 보관돼요. 📁 파일에서 [⬇ 내려받기]로 저장해 두세요.')));
    area.append(h('div', { class: 'row', style: 'margin-top:16px' }, nextMid && A.missions[nextMid].brief
      ? h('button', { class: 'btn ghost', onclick: () => openView('mission:' + nextMid) }, `다음 미션: M${A.missions[nextMid].no} ${A.missions[nextMid].title} ▶`)
      : h('button', { class: 'btn ghost', onclick: () => openView('home') }, '🏠 프로젝트 개요로')));
  };
  draw();
}

// ===================== 공통 위젯: 작업 진행 · 그래프 · 힌트 · 크게 보기 · 작업 중 화면 잠금 · 확인 질문 =====================
const POLL = {};
const pct = v => v === null || v === undefined ? '없음' : (v * 100).toFixed(1) + '%';
// 받침에 맞는 조사: josa('판정', '을', '를') → '을'
function josa(w, a, b) { const c = String(w || '').trim().slice(-1).charCodeAt(0); return c >= 0xAC00 && c <= 0xD7A3 && (c - 0xAC00) % 28 ? a : b; }
const ci = c => c ? `[${Math.round(c[0] * 100)}~${Math.round(c[1] * 100)}%]` : '';
function jobBar(d) {
  const p = d.state === 'done' ? 100 : d.pct || 0;
  return h('div', { class: 'jobbar' + (d.state === 'error' ? ' err' : '') },
    h('div', { class: 'jb-top' }, h('b', {}, d.state === 'error' ? '⚠ 문제가 생겼어요' : d.state === 'done' ? '✅ 끝났어요' : `⏳ ${p}%`), h('span', {}, d.msg || ''),
      h('span', { class: 'sp' }), d.sec ? h('span', { class: 'hint' }, `${Math.floor(d.sec / 60)}분 ${d.sec % 60}초 지남`) : null),
    h('div', { class: 'pb' }, h('i', { style: `width:${p}%` })));
}
async function jobState(name) {
  try { const r = await google.colab.kernel.invokeFunction('app.job', [name], {}); return r.data['application/json']; } catch (e) { return null; }
}
// 뒤에서 도는 작업(내려받기·학습·평가)을 1.5초마다 확인해 진행 막대를 그림. 끝나면 onDone (이 화면에서 진행을 본 경우만)
function watchJob(name, el, onDone, onTick) {
  clearTimeout(POLL[name]);
  let seen = false;
  const tick = async () => {
    const d = await jobState(name);
    if (d && d.state !== 'running' && BUSY === name) BUSY = null;
    if (!d || !el.isConnected) return;
    if (d.state === 'running') {
      if (STAY_JOBS[name]) { BUSY = name; BUSY_T = Date.now(); }
      seen = true; el.innerHTML = ''; el.append(jobBar(d)); if (onTick) onTick(d, el);
      POLL[name] = setTimeout(tick, 1500); return;
    }
    if (d.app) refresh(d.app);
    if (d.state === 'error') {
      if (!seen) return;                 // 예전에 끝난 작업의 오류는 다시 보여 주지 않음 (버튼을 다시 누르면 새로 시작)
      el.innerHTML = '';
      el.append(jobBar(d), h('div', { class: 'fb bad' }, '오류 내용: ' + (d.error || '') + '. 버튼을 다시 눌러 보세요. 계속되면 런타임 → 모두 실행으로 다시 시작해 주세요.'));
      return;
    }
    if (d.state === 'done' && seen) { el.innerHTML = ''; if (onDone) onDone(d); }
  };
  tick();
}
async function startJob(fn, args, name, el, onDone, onTick) {
  const d = await invoke(fn, ...args);
  if (d && d.ok !== false) watchJob(name, el, onDone, onTick);
  return d;
}
// 무거운 준비 작업이 도는 동안에는 다른 화면으로 못 가게 막음: 작업 중 다른 화면을 열면 Colab 런타임이 끊기는 일이 있었음
let BUSY = null, BUSY_T = 0;
const STAY_JOBS = PROJ.stay_jobs;          // 작업 이름 → 화면에 보일 이름 (프로젝트 명세의 stay_jobs)
function stayCard(what, extra) {
  return h('div', { class: 'stay', id: 'stay-card' }, h('b', {}, '⚠ 이 화면에서 기다려 주세요'),
    h('div', {}, `${what} 작업이 도는 동안(약 1~3분)에는 왼쪽 메뉴나 위쪽 단계 버튼 등 다른 화면을 누르지 마세요. 다른 화면을 열면 Colab 런타임이 끊길 수 있어요.`),
    extra ? h('div', { class: 'hint' }, extra) : null);
}
function busyBlock() {
  if (!BUSY) return false;
  if (Date.now() - BUSY_T > 15 * 60 * 1000) { BUSY = null; return false; }     // 15분이 지나도 소식이 없으면 풀어 줌
  note(`⏳ ${STAY_JOBS[BUSY] || '준비'} 작업이 끝날 때까지 이 화면에서 기다려 주세요. 다른 화면을 열면 Colab 런타임이 끊길 수 있어요.`, 'bad');
  const c = document.getElementById('stay-card');
  if (c) { c.classList.remove('shake'); void c.offsetWidth; c.classList.add('shake'); c.scrollIntoView({ block: 'center', behavior: 'smooth' }); }
  return true;
}
// 확인 질문 카드 (4·5단계 공통): 틀리면 힌트가 한 단계씩 자세해지고, 맞혀도 화면이 저절로 넘어가지 않음 → 풀이를 읽고 [다음 단계]를 직접 누름
// (이 동작은 화면 글로 따로 알리지 않음: 2026-10-19 피드백)
function quizCard(mid, qid, fnName, onSolved, title) {
  const box = h('div', { class: 'card' });
  const draw = () => {
    const q = A.views[mid].quiz[qid];
    box.className = 'card' + (q.ok ? '' : ' focus');
    box.innerHTML = '';
    box.append(h('h4', {}, (q.ok ? '✅ ' : '❓ ') + (title || '확인 질문')), h('div', { class: 'qtext' }, q.text));
    if (q.ok) {
      box.append(h('div', { class: 'choices col' }, q.options.map((o, i) => h('div', { class: 'choice' + (i === q.pick ? ' on' : '') }, h('span', {}, (i === q.pick ? '✔ ' : '') + o)))),
        h('div', { class: 'fb ok' }, '✅ 맞아요. ' + q.explain));
      return;
    }
    const ch = choiceGroup(`${mid}-${qid}`, q.options, null, null, { col: true }), fb = h('div');
    box.append(ch, h('div', { class: 'row' }, h('button', { class: 'btn', onclick: async () => {
      const v = ch.value();
      if (v === '') { fb.className = 'fb bad'; fb.textContent = '⚠ 보기를 하나 골라 주세요.'; return; }
      const r = await invoke(fnName, qid, Number(v));
      if (r.correct) { draw(); if (onSolved) onSolved(); } else { fb.className = 'fb bad'; fb.textContent = r.text; }
    } }, '확인')), fb);
  };
  draw();
  return box;
}
// 📘 가이드북(오른쪽)의 핵심 용어 탭을 열고 그 용어로 이동
function guideBtn(label, focus) { return h('button', { class: 'btn small ghost', onclick: () => openGuide('terms', focus) }, label || '📘 가이드북에서 용어 자세히 보기'); }
// 단계별 힌트: 버튼을 누를 때마다 한 줄씩 더 보여 줌 (스스로 생각할 기회를 먼저 줌)
function hintBox(key, hints, title) {
  const st = cache.hints || (cache.hints = {});
  const box = h('div', { class: 'hintbox' });
  const draw = () => {
    box.innerHTML = '';
    const n = st[key] || 0;
    box.append(h('div', { class: 'hb-h' }, h('b', {}, '💡 ' + (title || '힌트')), h('span', { class: 'sp' }),
      n < hints.length ? h('button', { class: 'btn small ghost', onclick: () => { st[key] = n + 1; draw(); } }, n ? `힌트 더 보기 (${n + 1}/${hints.length})` : `힌트 보기 (1/${hints.length})`)
        : h('span', { class: 'hint' }, '힌트를 모두 봤어요')));
    hints.slice(0, n).forEach((t, i) => box.append(h('div', { class: 'hb-i' }, h('b', {}, `${i + 1}. `), t)));
  };
  draw();
  return box;
}
// 영상 크게 보기
function zoomImg(src, cap) {
  const root = $('ws-root');
  const box = h('div', { class: 'zoombox', onclick: () => box.remove() }, h('div', { class: 'zb-in' }, h('img', { src }), h('div', { class: 'zb-cap' }, (cap || '') + '  · 아무 곳이나 누르면 닫혀요')));
  root.append(box);
}
// 작은 SVG 그래프: o = {x:[최소,최대], y:[최소,최대], xl, yl, series:[{pts, color, dash, dots, label}], marks:[{x,y,color,label}], vlines, hlines, w, h}
function chart(o) {
  const W = o.w || 380, H = o.h || 230, L = 44, R = 14, T = 14, B = 38;
  const sx = v => L + (v - o.x[0]) / (o.x[1] - o.x[0]) * (W - L - R), sy = v => H - B - (v - o.y[0]) / (o.y[1] - o.y[0]) * (H - T - B);
  const f = v => Math.abs(v) >= 10 || Number.isInteger(v) ? String(Math.round(v * 10) / 10) : v.toFixed(2);
  let s = `<svg viewBox="0 0 ${W} ${H}" xmlns="http://www.w3.org/2000/svg" font-family="sans-serif" font-size="10" fill="#111827">`;
  s += `<rect x="${L}" y="${T}" width="${W - L - R}" height="${H - T - B}" fill="#fff" stroke="#c3cad5"/>`;
  for (let i = 0; i <= 4; i++) {
    const xv = o.x[0] + (o.x[1] - o.x[0]) * i / 4, yv = o.y[0] + (o.y[1] - o.y[0]) * i / 4;
    s += `<line x1="${sx(xv)}" y1="${T}" x2="${sx(xv)}" y2="${H - B}" stroke="#eef2f7"/><text x="${sx(xv)}" y="${H - B + 13}" text-anchor="middle">${f(xv)}</text>`;
    s += `<line x1="${L}" y1="${sy(yv)}" x2="${W - R}" y2="${sy(yv)}" stroke="#eef2f7"/><text x="${L - 4}" y="${sy(yv) + 3}" text-anchor="end">${f(yv)}</text>`;
  }
  s += `<text x="${(L + W - R) / 2}" y="${H - 6}" text-anchor="middle" font-weight="bold">${esc(o.xl || '')}</text>`;
  s += `<text x="12" y="${(T + H - B) / 2}" text-anchor="middle" font-weight="bold" transform="rotate(-90 12 ${(T + H - B) / 2})">${esc(o.yl || '')}</text>`;
  for (const v of o.vlines || []) s += `<line x1="${sx(v.x ?? v)}" y1="${T}" x2="${sx(v.x ?? v)}" y2="${H - B}" stroke="${v.color || '#8a4b00'}" stroke-dasharray="4 3"/>` + (v.label ? `<text x="${sx(v.x) + 3}" y="${T + 10}" fill="${v.color || '#8a4b00'}">${esc(v.label)}</text>` : '');
  for (const v of o.hlines || []) s += `<line x1="${L}" y1="${sy(v.y ?? v)}" x2="${W - R}" y2="${sy(v.y ?? v)}" stroke="${v.color || '#8a4b00'}" stroke-dasharray="4 3"/>` + (v.label ? `<text x="${W - R - 3}" y="${sy(v.y) - 3}" text-anchor="end" fill="${v.color || '#8a4b00'}">${esc(v.label)}</text>` : '');
  let ly = T + 12;
  for (const se of o.series || []) {
    const pts = se.pts.filter(p => p[1] !== null && p[1] !== undefined);
    if (se.line !== false && pts.length > 1) s += `<polyline fill="none" stroke="${se.color}" stroke-width="2" ${se.dash ? 'stroke-dasharray="5 3"' : ''} points="${pts.map(p => sx(p[0]) + ',' + sy(p[1])).join(' ')}"/>`;
    if (se.dots) for (const p of pts) s += `<circle cx="${sx(p[0])}" cy="${sy(p[1])}" r="${se.r || 2.6}" fill="${p[2] || se.color}" ${p[3] ? `stroke="${p[3]}" stroke-width="1.5"` : ''}/>`;
    if (se.label) { s += `<rect x="${W - R - 118}" y="${ly - 8}" width="10" height="10" fill="${se.color}"/><text x="${W - R - 104}" y="${ly + 1}">${esc(se.label)}</text>`; ly += 14; }
  }
  for (const mk of o.marks || []) s += `<circle cx="${sx(mk.x)}" cy="${sy(mk.y)}" r="5.5" fill="none" stroke="${mk.color || '#dc2626'}" stroke-width="2.5"/>` + (mk.label ? `<text x="${sx(mk.x) + 8}" y="${sy(mk.y) - 6}" font-weight="bold" fill="${mk.color || '#dc2626'}">${esc(mk.label)}</text>` : '');
  return h('div', { class: 'chart', html: s + '</svg>' });
}
function reqChip(text, ok) { return h('span', { class: 'chk-chip ' + (ok ? 'y' : 'n') }, (ok ? '✅ ' : '⚠ ') + text); }
function statTile(label, value, desc) { return h('div', { class: 'tile' }, h('div', { class: 'd' }, label), h('div', { class: 'v' }, value), desc ? h('div', { class: 'd' }, desc) : null); }
function selectBox(opts, value, onchange) {
  const s = h('select', {}, opts.map(([v, t]) => h('option', { value: v, selected: String(value) === String(v) }, t)));
  s.onchange = () => onchange(s.value);
  return s;
}
function textBox(value, min, placeholder, oninput) {
  const ta = h('textarea', { placeholder });
  ta.value = value || '';
  ta.addEventListener('input', () => oninput(ta.value));
  return { ta, line: textCounter(ta, min) };
}

// ===================== 결과 보여 주기 도우미 (2026-10-20 제작자 피드백) =====================
// put: el.append는 null·undefined를 "null" 글자로 그려요(‘null’이 보이는 문제). 조건부 자식이 있으면 늘 put으로 붙여요.
function put(el, ...kids) { el.append(...kids.flat(9).filter(k => k !== null && k !== undefined && k !== false && k !== '')); return el; }
// 혼동행렬 표: o = { rows: [실제 이름…], cols: [예측 이름…], cells: [[개수…]…], good: (i, j) => 맞은 칸인가, rowHead, colHead }
// 행(실제)마다 비율로 칸을 칠해요: 맞은 칸은 초록, 틀린 칸은 빨강.
function confusionTable(o) {
  const good = o.good || ((i, j) => i === j);
  const tot = o.cells.map(r => r.reduce((a, b) => a + b, 0));
  const head = h('tr', {}, h('th', { colspan: 2 }, ''), h('th', { colspan: o.cols.length, style: 'text-align:center' }, (o.colHead || '예측') + ' →'));
  const head2 = h('tr', {}, h('th', {}, ''), h('th', {}, ''), o.cols.map(c => h('th', { style: 'text-align:center' }, c)));
  const rows = o.rows.map((r, i) => h('tr', {},
    i === 0 ? h('th', { rowspan: o.rows.length, style: 'text-align:center;vertical-align:middle' }, (o.rowHead || '실제') + ' ↓') : null, h('th', {}, r),
    o.cells[i].map((v, j) => {
      const f = tot[i] ? v / tot[i] : 0, ok = good(i, j);
      return h('td', { style: `text-align:center;font-weight:bold;min-width:72px;background:${ok ? `rgba(22,163,74,${0.08 + 0.55 * f})` : `rgba(220,38,38,${0.06 + 0.55 * f})`}` },
        String(v), h('div', { class: 'hint' }, tot[i] ? Math.round(f * 100) + '%' : '0%'), o.tags && o.tags[i] && o.tags[i][j] ? h('div', { class: 'hint' }, o.tags[i][j]) : null);
    })));
  return h('table', { class: 't cm', style: 'width:auto' }, head, head2, rows);
}
// 지표 표: rows = [[지표, 값, 뜻], …]
function metricTable(rows) {
  return h('table', { class: 't', style: 'width:auto' }, h('tr', {}, h('th', {}, '지표'), h('th', {}, '값'), h('th', {}, '뜻')),
    rows.map(r => h('tr', {}, h('td', {}, h('b', {}, r[0])), h('td', { style: 'text-align:right;font-weight:bold' }, r[1]), h('td', { class: 'hint' }, r[2] || ''))));
}
// 가로 막대: items = [[이름, 값, 색?], …], o = { max, unit, signed(음수 허용), fmt }
function barChart(items, o) {
  o = o || {};
  const max = o.max || Math.max(1e-9, ...items.map(x => Math.abs(x[1])));
  const fmt = o.fmt || (v => (Math.abs(v) >= 10 ? Math.round(v) : Math.round(v * 100) / 100) + (o.unit || ''));
  return h('div', { class: 'barchart', style: 'display:grid;grid-template-columns:max-content 1fr max-content;gap:4px 8px;align-items:center;font-size:12.5px;margin:6px 0' },
    items.map(([name, v, color]) => {
      const w = Math.min(100, Math.abs(v) / max * (o.signed ? 50 : 100));
      const bar = o.signed
        ? h('div', { style: 'position:relative;height:14px;background:#f1f5f9;border-radius:3px' }, h('div', { style: 'position:absolute;left:50%;top:0;bottom:0;width:1px;background:#94a3b8' }),
          h('div', { style: `position:absolute;top:0;bottom:0;${v >= 0 ? 'left:50%' : `left:${50 - w}%`};width:${w}%;background:${color || (v >= 0 ? '#dc2626' : '#2563eb')};border-radius:3px` }))
        : h('div', { style: 'height:14px;background:#f1f5f9;border-radius:3px' }, h('div', { style: `height:100%;width:${w}%;background:${color || '#2563eb'};border-radius:3px` }));
      return [h('span', {}, name), bar, h('b', {}, fmt(v))];
    }));
}
// 분포(히스토그램, 겹쳐 쌓기): o = { edges: [경계…], series: [{ data: [개수…], color, label }], vline: 값, vlabel, xl, yl }
function histChart(o) {
  const W = 520, H = 220, L = 42, R = 12, T = 12, B = 36, n = o.edges.length - 1;
  const tot = Array.from({ length: n }, (_, i) => o.series.reduce((a, s) => a + s.data[i], 0)), max = Math.max(1, ...tot);
  const x0 = o.edges[0], x1 = o.edges[n], sx = v => L + (v - x0) / (x1 - x0) * (W - L - R), sy = v => H - B - v / max * (H - T - B);
  let s = `<svg viewBox="0 0 ${W} ${H}" xmlns="http://www.w3.org/2000/svg" font-family="sans-serif" font-size="10" fill="#111827"><rect x="${L}" y="${T}" width="${W - L - R}" height="${H - T - B}" fill="#fff" stroke="#c3cad5"/>`;
  for (let i = 0; i <= 4; i++) { const v = max * i / 4; s += `<text x="${L - 4}" y="${sy(v) + 3}" text-anchor="end">${Math.round(v)}</text><line x1="${L}" x2="${W - R}" y1="${sy(v)}" y2="${sy(v)}" stroke="#eef2f7"/>`; }
  for (let i = 0; i <= 4; i++) { const v = x0 + (x1 - x0) * i / 4; s += `<text x="${sx(v)}" y="${H - B + 13}" text-anchor="middle">${Math.round(v * 100) / 100}</text>`; }
  for (let i = 0; i < n; i++) {
    let acc = 0;
    for (const se of o.series) { const v = se.data[i]; if (v) s += `<rect x="${sx(o.edges[i]) + 1}" y="${sy(acc + v)}" width="${sx(o.edges[i + 1]) - sx(o.edges[i]) - 2}" height="${sy(acc) - sy(acc + v)}" fill="${se.color}" fill-opacity="0.85"/>`; acc += v; }
  }
  if (o.vline !== undefined && o.vline !== null) s += `<line x1="${sx(o.vline)}" x2="${sx(o.vline)}" y1="${T}" y2="${H - B}" stroke="#111827" stroke-width="2" stroke-dasharray="5 3"/><text x="${sx(o.vline) + 4}" y="${T + 11}" font-weight="bold">${esc(o.vlabel || '')}</text>`;
  let ly = T + 12;
  for (const se of o.series) { s += `<rect x="${W - R - 120}" y="${ly - 8}" width="10" height="10" fill="${se.color}"/><text x="${W - R - 106}" y="${ly + 1}">${esc(se.label)}</text>`; ly += 14; }
  s += `<text x="${(L + W - R) / 2}" y="${H - 6}" text-anchor="middle" font-weight="bold">${esc(o.xl || '')}</text><text x="11" y="${(T + H - B) / 2}" text-anchor="middle" font-weight="bold" transform="rotate(-90 11 ${(T + H - B) / 2})">${esc(o.yl || '')}</text>`;
  return h('div', { class: 'chart', html: s + '</svg>' });
}
// 결정 영역 지도: o = { x: [최소, 최대], y: [최소, 최대], n, p: [[점수…]…] (아래쪽 줄부터), pts: [[x, y, 종류]…], thr, xl, yl, kinds: { 종류: [색, 모양('o'|'^'|'s'|'d'), 이름] } }
// 칸 색: 임계값 이상이면 빨강 쪽(양성), 아래면 파랑 쪽. 경계가 굵은 선으로 보여요.
function regionMap(o) {
  const W = 460, H = 380, L = 46, R = 12, T = 12, B = 40, n = o.n;
  const cw = (W - L - R) / n, ch = (H - T - B) / n;
  const sx = v => L + (v - o.x[0]) / (o.x[1] - o.x[0]) * (W - L - R), sy = v => H - B - (v - o.y[0]) / (o.y[1] - o.y[0]) * (H - T - B);
  let s = `<svg viewBox="0 0 ${W} ${H}" xmlns="http://www.w3.org/2000/svg" font-family="sans-serif" font-size="10" fill="#111827">`;
  for (let r = 0; r < n; r++) for (let c = 0; c < n; c++) {
    const p = o.p[r][c], pos = p >= o.thr, a = pos ? 0.15 + 0.55 * (p - o.thr) / Math.max(1e-6, 1 - o.thr) : 0.15 + 0.55 * (o.thr - p) / Math.max(1e-6, o.thr);
    const base = pos ? [220, 38, 38] : [37, 99, 235], mix = v => Math.round(255 - (255 - v) * a);   // 흰 바탕에 섞은 불투명 색 (칸 사이 틈이 비치지 않게)
    s += `<rect x="${(L + c * cw).toFixed(2)}" y="${(H - B - (r + 1) * ch).toFixed(2)}" width="${(cw + 1).toFixed(2)}" height="${(ch + 1).toFixed(2)}" fill="rgb(${base.map(mix).join(',')})" shape-rendering="crispEdges"/>`;
  }
  for (let r = 0; r < n; r++) for (let c = 0; c < n; c++) {          // 경계선: 이웃 칸과 판정이 다르면 선을 그음
    const pos = o.p[r][c] >= o.thr, x = L + c * cw, y = H - B - (r + 1) * ch;
    if (c + 1 < n && (o.p[r][c + 1] >= o.thr) !== pos) s += `<line x1="${(x + cw).toFixed(1)}" x2="${(x + cw).toFixed(1)}" y1="${y.toFixed(1)}" y2="${(y + ch).toFixed(1)}" stroke="#111827" stroke-width="1.6"/>`;
    if (r + 1 < n && (o.p[r + 1][c] >= o.thr) !== pos) s += `<line x1="${x.toFixed(1)}" x2="${(x + cw).toFixed(1)}" y1="${y.toFixed(1)}" y2="${y.toFixed(1)}" stroke="#111827" stroke-width="1.6"/>`;
  }
  const mark = (x, y, [color, shape]) => shape === '^' ? `<path d="M${x},${y - 4.5} L${x + 4},${y + 3} L${x - 4},${y + 3} Z" fill="${color}" stroke="#fff" stroke-width="0.8"/>`
    : shape === 's' ? `<rect x="${x - 3.5}" y="${y - 3.5}" width="7" height="7" fill="${color}" stroke="#fff" stroke-width="0.8"/>`
      : shape === 'd' ? `<path d="M${x},${y - 4.5} L${x + 4.5},${y} L${x},${y + 4.5} L${x - 4.5},${y} Z" fill="${color}" stroke="#fff" stroke-width="0.8"/>`
        : `<circle cx="${x}" cy="${y}" r="3.6" fill="${color}" stroke="#fff" stroke-width="0.8"/>`;
  for (const [x, y, k] of o.pts) if (x >= o.x[0] && x <= o.x[1] && y >= o.y[0] && y <= o.y[1]) s += mark(sx(x).toFixed(1), sy(y).toFixed(1), o.kinds[k] || ['#111827', 'o']);
  s += `<rect x="${L}" y="${T}" width="${W - L - R}" height="${H - T - B}" fill="none" stroke="#64748b"/>`;
  for (let i = 0; i <= 4; i++) {
    const xv = o.x[0] + (o.x[1] - o.x[0]) * i / 4, yv = o.y[0] + (o.y[1] - o.y[0]) * i / 4;
    s += `<text x="${sx(xv)}" y="${H - B + 13}" text-anchor="middle">${Math.round(xv * 100) / 100}</text><text x="${L - 4}" y="${sy(yv) + 3}" text-anchor="end">${Math.round(yv * 100) / 100}</text>`;
  }
  s += `<text x="${(L + W - R) / 2}" y="${H - 6}" text-anchor="middle" font-weight="bold">${esc(o.xl || '')}</text><text x="11" y="${(T + H - B) / 2}" text-anchor="middle" font-weight="bold" transform="rotate(-90 11 ${(T + H - B) / 2})">${esc(o.yl || '')}</text>`;
  const legend = h('div', { class: 'row', style: 'gap:10px;flex-wrap:wrap;font-size:12px' }, Object.values(o.kinds).map(([color, shape, name]) =>
    h('span', { html: `<svg width="12" height="12" viewBox="-6 -6 12 12">${mark(0, 0, [color, shape])}</svg> ${esc(name)}` })),
    h('span', { class: 'hint' }, `빨간 영역: 점수 ${o.thr} 이상 (양성으로 판정) · 파란 영역: 그 아래 · 검은 선: 결정 경계`));
  return h('div', {}, h('div', { class: 'chart', html: s + '</svg>' }), legend);
}

// ===================== 미션 단계: 명세(MISSION_UI)의 블록으로 화면 그리기 =====================
// 블록 종류마다 그리는 함수(BLOCK_VIEW)가 있고, 단계 화면(stepPage)이 블록을 차례로 그린 뒤 맨 아래에 [다음 단계]나 [완료하기]를 둬요.
// 원칙: 맞혀도 저절로 넘어가지 않음 · 틀리면 힌트가 한 단계씩 자세해짐 · 정답은 파이썬에만 있음(화면에는 맞힌 뒤의 풀이만 옴)
// 이런 동작 원칙은 화면 글로 따로 알리지 않아요 ("틀리면 힌트가 나와요" 같은 안내는 과한 정보: 2026-10-19 피드백)
function bst(mid, bid) { return ((A.views[mid] || {}).blk || {})[bid] || { ok: false, tries: 0, value: null }; }
const md = t => esc(t === undefined || t === null ? '' : t).replace(/\*\*(.+?)\*\*/g, '<b>$1</b>');
function mdp(t, tag) { return h(tag || 'p', { html: md(t) }); }
function stepPage(mid, si) { return (body, m) => renderStep(body, m, si); }
function renderStep(body, m, si) {
  const st = MISSION_UI[m.id].steps[si - 1], n = MISSION_UI[m.id].steps.length;
  const foot = h('div', { class: 'stepfoot' });
  const drawFoot = () => {      // 잠그지 않음: 뒷부분을 미리 볼 수 있게 [다음 단계]를 늘 보여 줌
    foot.innerHTML = '';
    const req = st.blocks.filter(b => b.req), left = req.filter(b => !bst(m.id, b.id).ok).length;
    if (left) foot.append(h('div', { class: 'hint' }, `이 단계의 필수 활동 ${req.length}개 중 ${req.length - left}개를 마쳤어요.`));
    foot.append(si < n ? nextBtn(m.id, si + 1).el : nextBtn(m.id, n + 1, null, null, '📦 산출물 받으러 가기 ▶').el);
  };
  if (st.lead) body.append(h('div', { class: 'tip', html: md(st.lead) }));
  for (const b of st.blocks) body.append(blockView(m, si, b, drawFoot));
  body.append(foot);
  drawFoot();
}
// 블록 하나: ctx.changed()를 부르면 그 블록과 단계 아래쪽 버튼을 다시 그려요 (직접 만든 활동도 같은 방식)
function blockView(m, si, b, changed) {
  const wrap = h('div');
  const ctx = { m, si, b, wrap, st: () => bst(m.id, b.id), redraw: null, changed: () => { ctx.redraw(); changed(); } };
  const fn = BLOCK_VIEW[b.type];
  ctx.redraw = () => { wrap.innerHTML = ''; fn ? fn(wrap, ctx) : wrap.append(h('div', { class: 'fb bad' }, `알 수 없는 활동 종류: ${b.type}`)); };
  ctx.redraw();
  return wrap;
}
function voicesView(vs, title) {
  return h('div', { class: 'voices' }, title ? h('div', { class: 'hint' }, title) : null, vs.map(([who, text]) => {
    const p = A.members[who] || { name: who, color: '#64748b' };
    return h('div', { class: 'voice' }, h('span', { class: 'avatar', style: `background:${p.color}` }, p.name[0]), h('div', {}, h('b', {}, p.name), ' ', h('span', { html: md(text) })));
  }));
}
// 실습 자료 이미지: "asset:번호"는 파이썬에서 받아 오고(한 번 받으면 캐시), 그 밖의 주소는 그대로 씀. 누르면 크게 보기
function assetImg(ref, cap, cls) {
  const img = h('img', { class: cls || '', alt: cap || '', title: '눌러서 크게 보기' });
  if (typeof ref === 'string' && ref.startsWith('asset:')) {
    const id = ref.slice(6), c = cache.asset || (cache.asset = {});
    const fail = r => {
      if (!img.isConnected) return;
      img.replaceWith(r && r.missing ? assetNotice(h('div', { class: 'fb bad' }, r.msg)) : h('div', { class: 'fb bad' }, (r && r.msg) || '사진을 불러오지 못했어요.'));
    };
    if (c[id]) img.src = c[id];
    else invoke('app.asset', id, 960).then(r => { if (r && r.ok) { c[id] = r.url; img.src = r.url; } else fail(r); }).catch(fail);
    img.onclick = () => invoke('app.asset', id, 1800).then(r => { if (r && r.ok) zoomImg(r.url, cap || ''); });
  } else {
    img.src = ref;
    img.onclick = () => zoomImg(ref, cap || '');
  }
  return img;
}
function assetCaption(ref) { return null; }   // 이미지마다 같은 안내를 반복하지 않아요 (출처는 미션을 마친 뒤 산출물에 정리)
function caseBody(it) {
  return h('div', {}, (it.body || []).map(t => mdp(t)),
    it.facts ? h('table', { class: 't' }, it.facts.map(([k, v]) => h('tr', {}, h('th', { style: 'width:170px', html: md(k) }), h('td', { html: md(v) })))) : null,
    it.images ? h('div', { class: 'row', style: 'align-items:flex-start' }, it.images.map(([cap, src]) => h('figure', { class: 'casefig' },
      assetImg(src, cap), h('figcaption', {}, cap)))) : null,
    it.timeline ? h('div', { class: 'timeline' }, it.timeline.map(([t, x]) => h('div', {}, h('b', {}, t), h('span', { html: md(x) })))) : null,
    it.voices ? voicesView(it.voices, '나온 말') : null);
}
function hintList(hints, tries) {
  return hints && hints.length ? h('div', { class: 'hintbox wrongbox' }, h('div', { class: 'hb-h' }, h('b', {}, `❌ 힌트 (틀린 횟수 ${tries}번)`)),
    hints.map((t, i) => h('div', { class: 'hb-i' }, h('b', {}, `${i + 1}. `), t))) : null;
}
const guideRow = b => b.guide ? h('div', { class: 'row' }, guideBtn(b.guide_label || '📘 가이드북에서 용어 자세히 보기', b.guide)) : null;

const BLOCK_VIEW = {
  info(w, c) {
    const b = c.b, ok = c.st().ok;
    w.append(h('div', { class: 'card' + (b.confirm && !ok ? ' focus' : '') }, b.title ? h('h4', {}, (b.icon || '📄') + ' ' + b.title) : null,
      (b.text || []).map(t => mdp(t)), b.figure ? figure(b.figure) : null,
      b.image ? assetImg(b.image, b.title || '', 'blkimg') : null, b.image ? assetCaption(b.image) : null,
      b.bullets ? h('ul', {}, b.bullets.map(x => mdp(x, 'li'))) : null,
      b.table ? h('table', { class: 't' }, h('tr', {}, b.table.head.map(x => h('th', { html: md(x) }))), b.table.rows.map(r => h('tr', {}, r.map(x => h('td', { html: md(x) }))))) : null,
      b.voices ? voicesView(b.voices) : null, b.tip ? h('div', { class: 'tip', html: md(b.tip) }) : null,
      b.case ? caseToggle(b.case) : null,
      b.sources ? h('div', { class: 'hint' }, '출처: ' + b.sources.join(' · ')) : null,
      guideRow(b),
      b.confirm ? (ok ? h('div', { class: 'fb ok' }, '✅ 확인했어요.') : h('div', { class: 'row' }, h('button', { class: 'btn', onclick: async () => {
        const r = await invoke('blk.check', c.m.id, b.id, true); if (r.correct) c.changed(); } }, b.confirm === true ? '읽었어요 ✔' : b.confirm))) : null));
  },
  quiz(w, c) {
    const b = c.b, s = c.st();
    const box = h('div', { class: 'card' + (s.ok ? '' : ' focus') }, h('h4', {}, (s.ok ? '✅ ' : '❓ ') + (b.title || '확인 질문')), h('div', { class: 'qtext', html: md(b.text) }));
    if (s.ok) {
      const pick = (s.value || {}).pick;
      box.append(h('div', { class: 'choices col' }, b.options.map((o, i) => h('div', { class: 'choice' + (i === pick ? ' on' : '') }, h('span', {}, (i === pick ? '✔ ' : '') + o)))),
        h('div', { class: 'fb ok', html: '✅ 맞아요. ' + md(s.explain) }));
    } else {
      const ch = choiceGroup(`${c.m.id}-${b.id}`, b.options, null, null, { col: true }), fb = h('div');
      box.append(ch, h('div', { class: 'row' }, h('button', { class: 'btn', onclick: async () => {
        const v = ch.value();
        if (v === '') { fb.className = 'fb bad'; fb.textContent = '⚠ 보기를 하나 골라 주세요.'; return; }
        const r = await invoke('blk.check', c.m.id, b.id, Number(v));
        if (r.correct) c.changed(); else { fb.className = 'fb bad'; fb.textContent = r.text; markWrong(ch); }
      } }, '확인')), fb);
    }
    box.append(guideRow(b) || '');
    w.append(box);
  },
  multi(w, c) {
    const b = c.b, s = c.st(), key = 'mp' + c.m.id + b.id;
    const pick = drafts[key] || (drafts[key] = new Set((s.value || {}).picked || []));
    const res = h('div');
    const box = h('div', { class: 'card' + (s.ok ? '' : ' focus') }, h('h4', {}, (s.ok ? '✅ ' : '☑ ') + (b.title || '해당하는 것 모두 고르기')), h('div', { class: 'qtext', html: md(b.text) }));
    b.items.forEach((it, i) => {
      const cb = h('input', { type: 'checkbox', checked: pick.has(i), disabled: s.ok, onchange: () => { cb.checked ? pick.add(i) : pick.delete(i); cb.closest('.choice').classList.remove('wrong'); } });
      box.append(h('label', { class: 'choice', 'data-i': i, style: 'margin:5px 0' }, cb, h('span', { html: md(it) })));
      if (s.ok && (s.explains || [])[i]) box.append(h('div', { class: 'hint', style: 'margin:-2px 0 6px 28px', html: md(s.explains[i]) }));
    });
    if (s.ok) box.append(h('div', { class: 'fb ok' }, '✅ 모두 맞게 골랐어요.'));
    else box.append(h('div', { class: 'row' }, h('button', { class: 'btn', onclick: async () => {
      const r = await invoke('blk.check', c.m.id, b.id, [...pick]);
      if (r.correct) { c.changed(); return; }
      res.innerHTML = '';
      box.querySelectorAll('.choice').forEach(l => l.classList.remove('wrong'));
      (r.items || []).filter(x => !x.ok).forEach(x => { const l = box.querySelector(`.choice[data-i="${x.i}"]`); if (l) l.classList.add('wrong'); });
      res.append(h('div', { class: 'fb bad' }, r.text), ...(r.items || []).filter(x => !x.ok).map(x => h('div', { class: 'fb bad' }, `${x.i + 1}번: ${x.text}`)));
    } }, '확인')), res);
    box.append(guideRow(b) || '');
    w.append(box);
  },
  match(w, c) {
    const b = c.b, key = 'mt' + c.m.id + b.id;
    const mt = drafts[key] || (drafts[key] = Object.assign({}, c.st().value || {}));
    const box = h('div');
    const draw = (res, hints, msg) => {
      box.innerHTML = '';
      const ok = c.st().ok;
      const opts = [['', '고르세요'], ...b.rights.map(r => [r, r])];
      const tbl = h('table', { class: 't' }, h('tr', {}, h('th', {}, b.left_label || '항목'), h('th', {}, b.right_label || '짝'), h('th', {}, '채점')),
        b.lefts.map(l => {
          const r = res ? res[l] : (ok ? true : null);
          return h('tr', { class: r === false ? 'wrongrow' : '' }, h('td', { html: '<b>' + md(l) + '</b>' }),
            h('td', {}, ok ? (mt[l] || '') : selectBox(opts, mt[l] || '', v => { mt[l] = v; })),
            h('td', { class: r === null ? '' : r ? 'ok' : 'bad' }, r === null ? '' : r ? '✅' : '❌ 다시'));
        }));
      box.append(h('div', { class: 'card' + (ok ? '' : ' focus') }, h('h4', {}, (ok ? '✅ ' : '🔗 ') + (b.title || '짝짓기')), b.text ? mdp(b.text) : null, tbl,
        ok ? h('div', { class: 'fb ok', html: '✅ 모두 맞게 짝지었어요.' + (c.st().explain ? ' ' + md(c.st().explain) : '') }) : h('div', { class: 'row' }, h('button', { class: 'btn', onclick: async () => {
          const r = await invoke('blk.check', c.m.id, b.id, mt);
          if (r.correct) c.changed(); else draw(r.result, r.hints, r.text);
        } }, '확인')),
        msg ? h('div', { class: 'fb bad' }, msg) : null, hintList(hints, c.st().tries), guideRow(b)));
    };
    draw(null, null);
    w.append(box);
  },
  order(w, c) {
    const b = c.b, s = c.st(), key = 'od' + c.m.id + b.id;
    const ord = drafts[key] || (drafts[key] = (s.value || {}).order ? [...s.value.order] : b.items.map(x => x.id));
    const nm = Object.fromEntries(b.items.map(x => [x.id, x]));
    const list = h('div'), fb = h('div');
    const draw = wrong => {
      list.innerHTML = '';
      const ok = c.st().ok;
      ord.forEach((id, i) => list.append(h('div', { class: 'orditem' + (wrong === i ? ' wrong' : '') }, h('b', {}, `${i + 1}.`),
        h('span', { style: 'flex:1', html: md(nm[id].text) + (nm[id].desc ? ' <small>(' + md(nm[id].desc) + ')</small>' : '') }),
        ok ? null : h('button', { class: 'btn small ghost', disabled: i === 0, onclick: () => { [ord[i - 1], ord[i]] = [ord[i], ord[i - 1]]; draw(); } }, '▲'),
        ok ? null : h('button', { class: 'btn small ghost', disabled: i === ord.length - 1, onclick: () => { [ord[i + 1], ord[i]] = [ord[i], ord[i + 1]]; draw(); } }, '▼'))));
    };
    draw();
    w.append(h('div', { class: 'card' + (s.ok ? '' : ' focus') }, h('h4', {}, (s.ok ? '✅ ' : '🧩 ') + (b.title || '순서 맞추기')), b.text ? mdp(b.text) : null, list,
      s.ok ? h('div', { class: 'fb ok', html: '✅ 순서가 맞아요.' + (s.explain ? ' ' + md(s.explain) : '') }) : h('div', { class: 'row' }, h('button', { class: 'btn', onclick: async () => {
        const r = await invoke('blk.check', c.m.id, b.id, ord);
        if (r.correct) c.changed(); else { fb.className = 'fb bad'; fb.textContent = r.text; draw(r.wrong); }
      } }, '순서 확인'), h('span', { class: 'hint' }, '▲▼ 버튼으로 순서를 바꿔요.')), fb, guideRow(b)));
  },
  classify(w, c) {
    const b = c.b, C = b.categories, PL = b.primary_label || '직접 원인';
    w.append(h('div', { class: 'card' }, h('h4', {}, (c.st().ok ? '✅ ' : '🔎 ') + (b.title || '분류하기')), b.text ? mdp(b.text) : null,
      h('div', {}, C.map(([k, n, d]) => h('span', { class: 'chk-chip', title: d || '' }, n))), guideRow(b)));
    const name = k => (C.find(x => x[0] === k) || [k, k])[1];
    for (const it of b.items) {
      const box = h('div');
      const draw = () => {
        box.innerHTML = '';
        const st = c.st(), z = (st.value || {})[it.id], ex = (st.explains || {})[it.id];
        const card = h('div', { class: 'card' + (z ? '' : ' focus') }, h('h4', {}, (z ? '✅ ' : '📂 ') + it.title), caseBody(it));
        if (z) {
          const cs = b.contrib === false ? '' : (z.contrib.length ? ` · 함께 작용한 것: ${z.contrib.map(name).join(', ')}` : '');
          card.append(h('div', { class: 'fb ok' }, `✅ ${PL}: ${name(z.primary)}${cs}`));
          if (ex) card.append(h('div', { class: 'hint', html: md(ex) }));
          box.append(card);
          return;
        }
        const key = 'cl' + c.m.id + b.id + it.id, d = drafts[key] || (drafts[key] = { p: null, k: new Set() });
        const pr = choiceGroup(key, C.map(([k, n, ds]) => [k, n, ds]), d.p, v => { d.p = v; });
        const ks = b.contrib === false ? null : h('div', { class: 'row' }, C.map(([k, n]) => {
          const cb = h('input', { type: 'checkbox', checked: d.k.has(k), onchange: () => { cb.checked ? d.k.add(k) : d.k.delete(k); } });
          return h('label', { class: 'choice' }, cb, h('span', {}, n));
        }));
        const fb = h('div');
        card.append(...[h('div', { class: 'qtext' }, `${PL}${josa(PL, '을', '를')} ${b.contrib === false ? '' : '하나 '}고르세요`), pr,
          ks ? h('div', { class: 'qtext' }, '함께 작용한 것도 있으면 골라 주세요 (여러 개 골라도 돼요)') : null, ks,
          h('div', { class: 'row' }, h('button', { class: 'btn', onclick: async () => {
            if (!d.p) { fb.className = 'fb bad'; fb.textContent = '⚠ 하나를 골라 주세요.'; return; }
            const r = await invoke('blk.check', c.m.id, b.id, { item: it.id, primary: d.p, contrib: [...d.k] });
            if (r.correct) c.changed(); else if (r.item_ok) draw(); else { fb.className = 'fb bad'; fb.textContent = r.text || r.msg || '다시 확인해 주세요.'; markWrong(pr); }
          } }, '확인')), fb].filter(Boolean));
        box.append(card);
      };
      draw();
      w.append(box);
    }
  },
  // 글쓰기: 정답을 판정하지 않아요. 최소 글자 수만 채우면 제출되고, 제출한 뒤에 모범답안이 보여요 (2026-10-19 피드백)
  write(w, c) {
    const b = c.b, s = c.st(), key = 'wr' + c.m.id + b.id;
    const d = drafts[key] || (drafts[key] = { t: s.value || '' });
    const tb = textBox(d.t, b.min || 10, b.placeholder || '꼭 입력해 주세요', v => { d.t = v; });
    const fb = h('div');
    if (s.ok) fb.append(h('div', { class: 'fb ok' }, '✅ 제출했어요.'));
    w.append(h('div', { class: 'card' + (s.ok ? '' : ' focus') }, h('h4', {}, (s.ok ? '✅ ' : '✍ ') + (b.title || '글쓰기')), b.prompt ? mdp(b.prompt) : null,
      h('div', { class: 'fieldrow' }, tb.ta, tb.line),
      h('div', { class: 'row' }, h('button', { class: 'btn', onclick: async () => {
        const r = await invoke('blk.check', c.m.id, b.id, tb.ta.value);
        if (r.correct) c.changed(); else { fb.className = 'fb bad'; fb.textContent = r.text; }
      } }, s.ok ? '고쳐서 다시 제출' : '제출하기'), b.guide ? guideBtn(b.guide_label || '📘 가이드북', b.guide) : null), fb,
      s.ok && s.model ? h('div', { class: 'model' }, h('div', { class: 'model-h' }, b.model_label || '📝 모범답안'), h('div', { class: 'model-t' }, s.model),
        s.explain ? h('div', { class: 'hint', html: md(s.explain) }) : null) : null,
      s.ok && s.checklist && s.checklist.length ? h('div', { class: 'selfcheck' }, h('div', { class: 'model-h' }, '🔍 내 글 점검'),
        s.checklist.map(t => h('label', { class: 'choice' }, h('input', { type: 'checkbox' }), h('span', {}, t)))) : null));
  },
  decide(w, c) {
    const b = c.b, s = c.st(), key = 'dc' + c.m.id + b.id;
    const d = drafts[key] || (drafts[key] = { ch: {}, rs: {} });
    if (s.value) for (const k in s.value) { d.ch[k] = d.ch[k] || s.value[k].choice; d.rs[k] = d.rs[k] || s.value[k].reason; }
    const opts = b.options || [['accept', '받아들임'], ['reject', '거절하고 대안 제시']];
    const res = {}, box = h('div'), fbAll = h('div');
    const draw = () => {
      box.innerHTML = '';
      const ok = c.st().ok;
      for (const it of b.items) {
        const p = A.members[it.who];
        const ch = choiceGroup(key + it.id, opts, d.ch[it.id] || null, v => { d.ch[it.id] = v; });
        const tb = textBox(d.rs[it.id] || '', b.reason_min || 10, '꼭 입력해 주세요 · 결정한 이유 또는 대안', v => { d.rs[it.id] = v; });
        const r = res[it.id], ex = ok ? (c.st().explains || {})[it.id] : null;
        if (r && !r.ok) markWrong(ch);
        box.append(h('div', { class: 'card' + (r && !r.ok ? ' focus' : '') },
          h('div', { class: 'voice' }, p ? h('span', { class: 'avatar', style: `background:${p.color}` }, p.name[0]) : null,
            h('div', {}, p ? h('b', {}, `${p.name} (${p.role})`) : null, ' ', h('span', { html: '“' + md(it.text) + '”' }))),
          ch, tb.ta, tb.line, r && !r.ok ? h('div', { class: 'fb bad' }, r.text) : null, ex ? h('div', { class: 'hint', html: md(ex) }) : null));
      }
    };
    draw();
    w.append(h('div', { class: 'card' + (s.ok ? '' : ' focus') }, h('h4', {}, (s.ok ? '✅ ' : '🗣 ') + (b.title || '결정하기')), b.text ? mdp(b.text) : null, box,
      h('div', { class: 'row' }, h('button', { class: 'btn', onclick: async () => {
        const r = await invoke('blk.check', c.m.id, b.id, { choices: d.ch, reasons: d.rs });
        for (const k in res) delete res[k];
        (r.items || []).forEach(x => { res[x.id] = x; });
        if (r.correct) c.changed(); else { fbAll.className = 'fb bad'; fbAll.textContent = r.text; draw(); }
      } }, s.ok ? '다시 저장' : '결정 저장')), fbAll, guideRow(b)));
  },
  // 바이브 코딩 실습: 칸마다 문장을 골라 프롬프트를 만들고, 가상 AI에게 보내고, AI가 만든 코드의 실제 실행 결과를 보고 고쳐 다시 보냄
  vibe(w, c) {
    const b = c.b, s = c.st(), key = 'vb' + c.m.id + b.id;
    const d = drafts[key] || (drafts[key] = { picks: {}, reply: null, attempts: [], loaded: false });
    const replyBox = h('div', { class: 'vb-replybox' }), fb = h('div'), promptBox = h('pre', { class: 'vb-prompt' });
    const promptText = () => b.slots.map(sl => { const o = sl.options.find(x => String(x.i) === String(d.picks[sl.id])); return o && o.text ? o.text : ''; }).filter(Boolean).join('\n');
    const drawPrompt = () => { const t = promptText(); promptBox.textContent = t || '칸마다 문장을 고르면 여기에 프롬프트가 만들어져요.'; promptBox.classList.toggle('empty', !t); };
    const drawReply = () => {
      replyBox.innerHTML = '';
      const r = d.reply;
      if (r) replyBox.append(vibeReply(r));
      if (d.attempts.length > 1 || (d.attempts.length && !d.attempts[0].ok)) replyBox.append(h('details', { class: 'vb-log', open: !(r && r.ok) }, h('summary', {}, `📝 프롬프트 개선 기록 (${d.attempts.length}번)`),
        h('table', { class: 't' }, h('tr', {}, h('th', {}, '시도'), h('th', {}, '부족했던 칸'), h('th', {}, '결과')),
          d.attempts.map(a => h('tr', {}, h('td', {}, a.no), h('td', {}, a.ok ? '✅ 성공' : (a.slot || '')), h('td', {}, a.result))))));
    };
    if (!d.loaded && (s.ok || s.tries || b.has_start)) {      // 다시 그릴 때 파이썬에 남은 마지막 응답과 ‘지금까지의 코드’를 받아 옴 (선생님용은 처음부터 채워져 있음)
      d.loaded = true;
      invoke('app.vibe', c.m.id, b.id).then(r => {
        d.reply = r.reply; d.attempts = r.attempts || []; d.start = r.start;
        for (const k in r.picks || {}) d.picks[k] = String(r.picks[k]);
        c.redraw();
      });
    }
    const send = async () => {
      if (b.slots.some(sl => d.picks[sl.id] === undefined)) { fb.className = 'fb bad'; fb.textContent = '⚠ 칸마다 문장을 하나씩 골라 주세요.'; return; }
      fb.className = 'fb'; fb.textContent = '⏳ 가상 AI가 코드를 만들고 실행하는 중이에요…';
      const r = await invoke('blk.check', c.m.id, b.id, { picks: d.picks });
      fb.className = ''; fb.textContent = '';
      d.loaded = true;
      if (r.reply) { d.reply = r.reply; d.attempts = r.attempts || []; }
      if (r.correct) c.changed(); else drawReply();
      w.querySelectorAll('.vb-slot .choice').forEach(l => l.classList.remove('wrong'));
      if (!r.correct && r.reply && r.reply.slot_id) markWrong(w.querySelector(`.vb-slot[data-slot="${r.reply.slot_id}"]`));
      const el = wrapEl().querySelector('.vb-replybox'), sc = el && el.closest('.scroll');      // 바깥 페이지는 움직이지 않고 미션 화면만 응답 위치로
      if (sc) sc.scrollTop += el.getBoundingClientRect().top - sc.getBoundingClientRect().top - 8;
    };
    const wrapEl = () => w;
    w.append(h('div', { class: 'card' + (s.ok ? '' : ' focus') }, h('h4', {}, (s.ok ? '✅ ' : '🤖 ') + (b.title || '바이브 코딩 실습')),
      b.ask ? voicesView([b.ask]) : null, b.text ? mdp(b.text) : null,
      d.start ? h('div', { class: 'vb-start' }, h('div', { class: 'vb-label' }, h('b', {}, '📌 지금까지의 코드와 결과')), vibeReply(Object.assign({}, d.start, { open_code: true }))) : null,
      (b.files || []).map(f => h('details', { class: 'vb-data', open: true }, h('summary', {}, `📄 ${f.name} (${f.n}행, 앞부분만 보여요)`),
        h('div', { class: 'vb-tbl' }, h('table', { class: 't' }, h('tr', {}, f.head.map(x => h('th', {}, x))), f.rows.map(r => h('tr', {}, r.map(x => h('td', {}, x)))))))),
      b.slots.map(sl => h('div', { class: 'vb-slot', 'data-slot': sl.id }, h('div', { class: 'vb-label' }, h('b', {}, sl.label), sl.why ? h('span', { class: 'hint' }, ' ' + sl.why) : null),
        choiceGroup(key + sl.id, sl.options.map(o => [o.i, o.text || '(이 칸은 비워 두기)']), d.picks[sl.id] === undefined ? null : d.picks[sl.id], v => { d.picks[sl.id] = v; drawPrompt(); }, { col: true }))),
      h('div', { class: 'vb-label' }, h('b', {}, '✍ 내 프롬프트')), promptBox,
      h('div', { class: 'row' }, h('button', { class: 'btn', onclick: send }, s.ok || d.attempts.length ? '🤖 다시 보내기' : '🤖 가상 AI에게 보내기'), guideRow(b)), fb, replyBox));
    drawPrompt();
    drawReply();
  },
  custom(w, c) {
    const fn = window[c.b.renderer];
    if (typeof fn !== 'function') { w.append(h('div', { class: 'fb bad' }, `화면 함수 ${c.b.renderer}를 찾지 못했어요.`)); return; }
    fn(w, c);
  },
};
// 가상 AI의 응답 한 건: 말 → 코드 → 실제 실행 결과(출력·표·그림·오류) → 만들어진 문서 → 짚어 볼 점
function vibeReply(r) {
  const out = h('div', { class: 'vb-reply' + (r.ok ? ' ok' : '') + (r.slot ? ' miss' : '') }, h('div', { class: 'vb-who' }, '🤖 가상 AI'), r.say ? h('div', { class: 'vb-say', html: md(r.say) }) : null);
  if (r.code) out.append(h('details', { class: 'vb-codebox', open: !!r.open_code }, h('summary', {}, '💻 AI가 만든 코드 (눌러서 보기) ', h('button', { class: 'cpy', onclick: e => { e.preventDefault(); copyText(r.code); } }, '📋 복사')),
    h('pre', { class: 'vb-code', html: hl(r.code) })));
  if (r.code) out.append(h('div', { class: 'vb-label' }, h('b', {}, '▶ 실행 결과')));
  if (r.stdout) out.append(h('pre', { class: 'vb-out' }, r.stdout));
  (r.tables || []).forEach(t => out.append(h('div', { class: 'vb-tbl', html: t })));
  (r.figs || []).forEach(u => out.append(h('img', { class: 'blkimg', src: u, title: '눌러서 크게 보기', onclick: () => zoomImg(u, '') })));
  if (r.error) out.append(h('pre', { class: 'vb-err' }, r.error));
  if (r.report) out.append(h('div', { class: 'vb-label' }, h('b', {}, '📄 AI가 만든 문서')), h('iframe', { class: 'preview vb-doc', srcdoc: '<meta charset="utf-8"><style>' + VB_DOC_CSS + '</style>' + r.report }));
  if (r.problem) out.append(h('div', { class: 'vb-problem' }, h('b', {}, '🔎 결과를 보니: '), h('span', { html: md(r.problem) })));
  if (r.hint) out.append(h('div', { class: 'vb-problem' }, h('b', {}, `❌ ${r.slot ? '‘' + r.slot + '’ 칸의 ' : ''}현재 선택을 다시 고민해 보세요: `), h('span', { html: md(r.hint) })));
  if (r.lesson) out.append(h('div', { class: 'fb ok', html: '✅ ' + md(r.lesson) }));
  return out;
}
const VB_DOC_CSS = 'body{font-family:sans-serif;color:#111827;background:#fff;line-height:1.6;font-size:14px;padding:6px 10px}table{border-collapse:collapse;margin:6px 0}td,th{border:1px solid #c8ced6;padding:4px 8px;font-size:13px;text-align:left}th{background:#eef2f7}h1,h2,h3{margin:8px 0 6px}';

// ===================== 단계 등록 =====================
const STEPS = {
  m1: [renderBrief, stepPage('m1', 1), stepPage('m1', 2), stepPage('m1', 3), (b, m) => doneView(b, m, [["m1_alarm_table", "M1 카메라별 경보 집계표"]], 'm2')],
  m2: [renderBrief, stepPage('m2', 1), stepPage('m2', 2), stepPage('m2', 3), (b, m) => doneView(b, m, [["m2_requirements", "M2 탐지 요구 성능 비교와 합의"]], 'm3')],
  m3: [renderBrief, stepPage('m3', 1), stepPage('m3', 2), stepPage('m3', 3), (b, m) => doneView(b, m, [["m3_labels", "M3 라벨 검증 결과와 수정 기록"]], 'm4')],
  m4: [renderBrief, stepPage('m4', 1), stepPage('m4', 2), stepPage('m4', 3), (b, m) => doneView(b, m, [["m4_zone", "M4 침입 탐지 영역 비교와 합의 기록"]], 'm5')],
  m5: [renderBrief, stepPage('m5', 1), stepPage('m5', 2), stepPage('m5', 3), (b, m) => doneView(b, m, [["m5_detectors", "M5 탐지기 조건별 비교표"]], 'm6')],
  m6: [renderBrief, stepPage('m6', 1), stepPage('m6', 2), stepPage('m6', 3), (b, m) => doneView(b, m, [["m6_op", "M6 임계값별 표와 운용점 결정 기록"]], 'm7')],
  m7: [renderBrief, stepPage('m7', 1), stepPage('m7', 2), stepPage('m7', 3), (b, m) => doneView(b, m, [["m7_choice", "M7 드론·조류 분류 모델 선택 기록"]], 'm8')],
  m8: [renderBrief, stepPage('m8', 1), stepPage('m8', 2), stepPage('m8', 3), (b, m) => doneView(b, m, [["m8_flow", "M8 경보 처리 순서표와 절차서"]], 'm9')],
  m9: [renderBrief, stepPage('m9', 1), stepPage('m9', 2), stepPage('m9', 3), (b, m) => doneView(b, m, [["m9_console", "M9 통합 관제 화면 구성과 수정 기록"]], 'm10')],
  m10: [renderBrief, stepPage('m10', 1), stepPage('m10', 2), stepPage('m10', 3), (b, m) => doneView(b, m, [["m10_briefing", "M10 사후검토 비교와 시설 방호 브리핑"]], null)],
};
const TOPICS = {
  m1: [null, null, null, null, null],
  m2: [null, null, null, null, null],
  m3: [null, null, null, null, null],
  m4: [null, null, null, null, null],
  m5: [null, null, null, null, null],
  m6: [null, null, null, null, null],
  m7: [null, null, null, null, null],
  m8: [null, null, null, null, null],
  m9: [null, null, null, null, null],
  m10: [null, null, null, null, null],
};

async function resetAll() {
  if (!confirm('모든 진행 상황을 지우고 처음부터 할까요?')) return;
  mstep = {}; seen = {}; cache = {}; drafts = {}; CODE.live = null; BUSY = null;
  for (const k in POLL) clearTimeout(POLL[k]);
  await invoke('app.reset');
  openView('home');
}

applyUi();
invoke('app.state');
</script>
"""

if "app_state" not in globals():
    display(HTML("<div style='padding:14px 16px;border:2px solid #b3261e;border-radius:10px;background:#fce8e6;color:#111827;font-size:15px'>"
                 "⚠️ 먼저 위의 <b>🚀 [먼저 실행] 실습 준비하기</b> 셀을 실행해 주세요. 진행 막대가 100%가 된 뒤 이 셀을 다시 실행하면 메신저가 열려요.</div>"))
else:
    display(HTML(APP_HTML.replace("__WORKSPACE__", WORKSPACE_NAME)))
```
