```python
# @title 📦 LOG05 실습자료 자동 준비 (ZIP 업로드 불필요) { display-mode: "form" }
# 학습자 화면에는 결과와 업로드 인터페이스만 표시하도록 구성합니다.
from pathlib import Path
import os, zipfile, stat
from IPython.display import display, Markdown

LOG05_PROJECT_DIR = Path.cwd() / "LOG05_실습자료"
LOG05_PROJECT_READY = False

def _log05_extract_zip(zip_path):
    global LOG05_PROJECT_READY
    zip_path = Path(zip_path)
    if not zipfile.is_zipfile(zip_path):
        raise ValueError("유효한 ZIP 압축파일이 아닙니다. 실습자료 ZIP을 다시 선택해 주세요.")
    dst = LOG05_PROJECT_DIR.resolve()
    dst.mkdir(parents=True, exist_ok=True)
    with zipfile.ZipFile(zip_path) as archive:
        entries = archive.infolist()
        if len(entries) > 20000 or sum(info.file_size for info in entries) > 1024**3:
            raise ValueError("압축파일의 항목 수 또는 해제 후 용량이 허용 범위를 초과합니다.")
        for info in entries:
            item = Path(info.filename.replace('\\', '/'))
            if item.is_absolute() or '..' in item.parts or not info.filename.strip():
                raise ValueError("안전하지 않은 ZIP 내부 경로가 포함되어 있습니다: " + info.filename)
            target = (dst / item).resolve()
            if not target.is_relative_to(dst):
                raise ValueError("실습 폴더 밖으로 벗어나는 파일 경로입니다.")
            mode = (info.external_attr >> 16) & 0o170000
            if mode == stat.S_IFLNK:
                raise ValueError("심볼릭 링크가 포함된 ZIP은 허용하지 않습니다.")
        for info in entries:
            target = dst / Path(info.filename.replace('\\', '/'))
            if info.is_dir():
                target.mkdir(parents=True, exist_ok=True)
            else:
                target.parent.mkdir(parents=True, exist_ok=True)
                with archive.open(info) as src, target.open('wb') as out:
                    import shutil
                    shutil.copyfileobj(src, out)
    files = [p for p in dst.rglob('*') if p.is_file()]
    if not files:
        raise ValueError("ZIP에 압축 해제할 실습자료 파일이 없습니다.")
    LOG05_PROJECT_READY = True
    print("✅ LOG05 실습자료 압축 해제 완료:", dst)
    print("📄 확인된 파일 수:", len(files))
    print("ℹ️ 실습자료 폴더:", dst)
    print("※ 업로드·압축 해제는 학습 수행 및 평가 완료를 뜻하지 않습니다.")
    return dst

# 실습자료가 노트북 안에 내장되어 있어 ZIP 업로드 없이 자동으로 준비됩니다.
# 내장 파일 5개: LOG05_Colab_실행안내.md, LOG05_M02_forms.md, LOG05_M02_lesson.md, LOG05_공통_AI역할실습안내.md, LOG05_공통_작성안내.md
import base64 as _embedded_b64
_EMBEDDED_ZIP_B64 = (
    'UEsDBBQAAAgIAAAAIVzAhdWSSQcAAMMOAAAbAAAATE9HMDVfQ29sYWJf7Iuk7ZaJ7JWI64K0Lm1kbVddbxNXEH3Pr7hSxSMmDtDAI2orQCJq1RaQ+hIZSEtK'
    'qFECRfShWiebaBMbYkOcbJK12RTnq3XbTbyOHcm85Kf0ce/d/9Azc+/aC1gCFtl3556ZOefM+DNx6+vro5eF2i7FhaY8KQhVbMTry0JVHTkfjoxkM/rE5PP8'
    '7KN7+fyjzPSTF7/ck7s98UV+JndPbZSV7Qm1EUYtX6iGJU9toRYtfG5eREC10o7XbLnjKb8Wr3Qzv00/UTUcc5ty770sOrLYyAjl2+o4VL6n6hXZ6oofbn4T'
    'tXoiagOWHVcP++f6dwhVb2q4qmaJqBuonRCfVZR9JOLVUL5cI5jy5T9qPVRvArWPN6orAGNCjYxlRNRqx0ttce0mUvg7rvomcbVpiVtTc3P5Xxho7T1lB1Sq'
    '1hVqvgmsIi6VZfHgrKNqrmx2zzrXblLOcn8n6uJMrUcvyoOe8nGrXxjcijC2XLX5yuB9Uj8Uycdja00d/33WkcW2bFl89bserpbBdirAtW9uini+IVfWRFy1'
    '8Q6wL6W+j9rlKLQEgiDSv11Zd4U6DeUbD2DNPfVd1FDOu0KeWLJECKuyFVJEnYaqL0p/N666nxbtYkZ8+9V3t299L5C2qpfRepRAqPWKKtYY8hEwBVSwudyv'
    'U5OzU3PPZp7ykXc9ZfORuEqNxr+Gbun2bnaBScSvumq/gp6CNUKubqKggrhx4ono+D1RCQEJgO1Rk3XXEZAqK8v9aHEV51utpMrmlMtc1aWIOr3oKOBCxZtV'
    '9Bf/NX1ue+qPRaFsP15g1HL+IDotJHW4lNH55aYnH0znZvI/PZsirEM7CX2VCSyCREEBf0CCEGxN5x06oB5dDvLZeP55QCnGWyVV9NG4fScJ8LEg7j58ceFG'
    '/vmFuw9zXOaoZcdbL8860bEVL/n0RDN61Nsf87OP53R7dSH6MVA2ioszYEO8CUbt28qzSQaa59w/t0WFizpduWdrXXTVXm0oSy5nOGfoWLmOKnK7SKENi0XN'
    'RRfxawd5kXihiSjwiHUG0LdZ+vh61hxVG03qrKqvybfLV8bORYFlAl8dPWcCM6KTQ7VzNAyRkEuv+aY3R1F3GQX1y8pxSbamxvz6fFPWS5yraf7KKSXp9bi1'
    'QzxEqKUSK7Belu9KDAtaxPlDcHsTomqj45Zc2VW1kPSmGhWOsWnjODGXSOdXh9bw84yYGIeMG2orxP0QMLkeIdHOOjE6PqkcD8yO3apWuHHbzMOnj2c0ZUPp'
    'N8g+qlS9MucJ4XRLEg61FcCd+g7ed+LYteGsxjPkfkl7QnRcIo+WxWaatXyGXBTtg25jOxgDOwVnGxrH5Hoe/9V3TN0uPLm1Wr7abkLxn+UOjgqCV9/9z9rk'
    'XBxb7vUJghvibTtlppq4JhujHvkn4p6edQYdNqWQFYjglPX0gSlQsWulJHUg05kNMh4Z56YMMBKnNEzBeD6eNkSFQGhV01kcUksvBTtbjSdG2lV08ULHCBZJ'
    'Ks9nVtUdxGUNGFkVvSiwh1CfDHLBMDaueDg3XA5UAR4D4vaX57Nnndt3zo9xpxYK8QLnDixyr5lUpmThG+iGJYgn3KGOfBPb5BEMZygxQWVlp3+RaUWrTeOO'
    'gKZnEN/ouNrj/Cpw0uu6OkNxG8aR7FYaqB97VnrIDarFn+81aXx/2A2qyYbLTak6ifeRwDHZtw6HavEKEoFVAd7+coKB+n3j+4lbnN6Oh+w/FKs8tuOVnvEV'
    'Djzw274uyZL4Xar4k9n8z1P32cVplhVPKQW1iNT2amyYvNWwgzPlAJWNxpgVz0f6Nl5F2w5Sc/BxbvbRZO7+02e5mcmn04+nmLW0QZltj8hUIKn3Ky//pU4w'
    'H40YMU8Ag96kpA3qC6ab8YJFmj/o0sgwAxSRMbkTqg8pw8hViOkKsh0bhUIFDXzMroqXiPjESfYbdk0KTY8NdHAJVtYQRCDvIDVWCeN8kDaobRz0EH+dODvf'
    'jI5Cika+TDPlBDakFhwQxo5oBL1mPmp6DOYmNcH3YAWUIqmZGcNEw2ZgD583E+OIdVm5BSyWvBNh4NTSm4BZDiEuPQB4KYErn7omJy2sVDLkcqwmroQp/Ooy'
    'gMDax2B2w/aDkewowFylPO5kf78zzhkaD4G9+DbaCwA0lpPFp97Uvdb7Yndn6Nqhly+QDpPBGhvFl/31UsIxK95F/RENePci6j40JX2UANGaS7PmksjGr1yC'
    'pr1Ab4LM9MD8SEiW/GTntAZv0J6oSZPM4pTlpRF/4kTJ533nxBgcagXZrN4ric0YQoWQ28pbopi4dNaZ+Bx/r3C2LEMoOTuKIUK9Y9d13I9WXs4GxedtRBOE'
    'CldQfkUkW5zxHK06rIe6bgNtJQulNuuBVS0mA/7Zk5l87sHcBd4d+j/BDH/TayhiQliOqS7xoo+KZzDclBIODmAKukDUN/rRtHXIU6jYgECAWJNME9SMJwx9'
    '/ePELFZ69cPx1Oon3wbkJnXbgOF1FH3nWZnY5qAd/wNQSwMEFAAAAAgAAAAhXD7AoisGCgAA0hwAABIAAABMT0cwNV9NMDJfZm9ybXMubWTtWFFPXMcVft9f'
    'MZJVCRRggQW34Y2XUkuxWpGHPFSV4zZOY9UUCbtKIyFrF1/QmqXykrCwC3fpkiwsRGvlsiywqESV/FOStztz/0O/c+bM3bvAukmbyC+WY2fv3Jm5Z875zne+'
    'M7fUe7+dGZ1Ud0fHldldN96xMpvrplBNpcypb75cVmYpMGe+bl5OqQ8/Xrg/9+DT+YW/pB/N/3l08t7j+b8t/OnBvYUHjx7e/+PDRw+ffHZv7v6ThYd/H5n7'
    '6MNUali2NLvFV+c6OMK232c3zF6gTzyVVmHgm2rnnek732dLU93J1UtMbnmm5vFgoW5qfnr6jtJfH5pqWUUvnodBFluaSlaFncDUs8pNTz168Pjx/F+xU1F/'
    'tab3L5Wpfhue1GTrqHSkC3ldqI8os1PUJx186KRtKp6K1n3aOwwqportltpm+4h+RSunWBSVymarqaKSjy/ldWMt3q+tX3hKF+Nt9TcdWBuenGJn/OzsmVrp'
    '1TnNK1wos7Viqmum6sGqtvFrdICodBj+q4wvHoYXOdkllbp16xaGyiNxSDwfvlNjalhF+abZRVRq2KVWDFu0xUv99RFOMDA2Cpfy5s9yQ2qCH8o5s/WSfKfX'
    'i/BdEx8cwmjeFHw1TjPs2GAqtUiHi8odtehWLfZ+As9wRtg6cm7Hc6WEECrEVrdK8TPcP0AxepYbxJg+y8sy+uYivjM8PKx+0L+YO0Y7BEHkBfqgybtlEQ7a'
    'KtoCGHbght0N/c/nZCxgSkfGV94d/UXsBxxWX5C3PF3bl+MqUynpk1MEPKtX97E2PPEQLZyLzqb3quodXl9fd0ckX5RoH13gYMKSk7beL2P/syOKhjWlKR4j'
    '08fJISunplCj2Y1clAOGPs+bRl7pL47DDtmckV/WWPqqPZhZ2TP5Mi3NRNvAm3cMUHX303t+eN5JPF81Kmx9izl4zOIXnrLRSo2NymDStf8wPtFnfLLP+O0+'
    '47/sM/6rPuPv9hkfG+33Yqzfi/EbX6T6ptM40kncD1RRhoN7wnaWp3AYBsBVYAoEB0jJAdZqZuzpzG1KYd2oAynELQBFeFq0CXTSphFghdcxeOXnePdnpvuT'
    'vI6PmgtEfKsJK2y+uCH6dOdHp8xsPw/N3uwhvOiHitl+sJi9GRf9nZ2Bs+EmExy6jIpKHiCuwnPfXFSmhGrIx/hLiQhcw5lgW+NfAst6dUPNTCpdLtKv8Pwy'
    'PA6QvsyqmHDUVggMz5QUBk1gPQoO2Cna3AcHLiOqqDUxh65eEFmB5mEjZ0uhrvc7XRJTA0qpQayT9OTyY18CEWY3z0AorVFFsCmGEmHLi/AmOMHucebxkkZW'
    'L8G+FoBzScewL/Fp0yqD36zJXRqR7cipymy3QctKCpbUyotOKjUz5tgr+PzVeQ9rDamZ8e5L8QsGM4JvVClLJPH0CeJJwE9KKO3XQS0cItf3xOXVeXjaRE3B'
    'm9sSjbjAlclf+sAj92AqiDNReWfHOZIZwjuK5zW+GyO+0/+mpNNkBxdq2tMSIVcuruyg3b1jKs0NBLC0alZP4+LeA6Afgpvd5XitX4u2D8msMPCuuJPBmYfX'
    'zNkGbwGLkaa62YlIt2SdsVtFAEqZY89sBzgjlRPSDgR/fw/ehI99U8rTCY8CUTicB7WSINh8EVw/1kR6UsXlggiCvAJkkTCClCC91s7DOFhVK0mN5hAgUGde'
    'NwSSo8xVPaw1Y3dHaG08F51QcPWbKYpEk0ehi9aKunBIjOVUD9FaeNwmbCQzfVFsS4tpVq7B0qvaTnjvvxJe6gZG+v8HKMtgCxwXbf8DZ+CMjiWiTZ6uqCWI'
    'IEun1Hebe7RMF05RbdmZhD0oAnoho2arrXfXRCPQamwzfQdACVsAewPCo4M5y7rZ5kiyPGT1K26VZOyRu/DfThH4Z1gzo9AEyAvQBjjkbAOunqL4mqUmCxQO'
    'BgGRBgp1AncZ2KXw8P4ujSVVLMgJjcTU2xu80k68EqDUFZde+z+BbC0bVTYQaIilHH3Soh77mtZLTpwm5Hi0tsbk43uMAzfLebHPLCYI5gh9tAEovQ5AqdeA'
    '4MoTHmNW5whzW6D4LAA6+wMV2sUGNLsUMKijzedORi5KliA6ElY6FbR/I+sylREG0UYbB2XieHewnyMNUjiI+erS7O6rDFI1rmxefNZuM2OBb51PCS/tkYWJ'
    'FEymB0FWmehMN3YJPd0eyp5WTs/MXct1WU0IiGsaY4FhTbBMhNf2aEiebgh6v+yrLmVJ2+rVomdVS+A+ZbF1rUDnfwUIUYTNNe4+rayG/6JnPmca87+0qNI1'
    'xVkbo9klq6XuBNv+bvr999O/nr7zXpoYlU+YBnyYMs/ytKVzhmMn+fZPBPjXnpzJCC2WuDDZo1MwqmtcBpbLUsG0VyZSi8/6TUeKAcgHjrH9PpVpq3SmmBBd'
    'HbCH67b/VDohzJH5rvl3UyzJOm/ZpySaeMQKPbp6sLuJJdbvEISsWt3FAvFvxWM65LjxySWuVrHac46wUmq1mU2r7SjXtnWfqTNfBgckKu6w+r1d/AcnDbES'
    '/SA0n1mtR6sd4o2a/pLkgrrNfOEp9V12Q3WHBVWCJpnE+QRzLzzRjjKLqsC6P9DTvw+q5AUB30NAcFRhah5GY5W7MxAL+RKAZGvP7YEuls1WvMC2+W6JKIlJ'
    'JCCcwZxPJGl7HeipXvlCdk4kOYjlEM7F5H7mQxMP9Ei5Qcwsm702dbekKykF+7cULM5Z88cHpEB88Mln6d/Mf5r+4JP7T6akdJBG5prv6g0XqzgzLd8n+dKp'
    'AqxI0g190lFSIvZX8j7Gub0kYuRSUUskOoXVyTZmVZsAMeOQ0G4LrnURBabOfcMNqtFCu+LrFxW6dTMND8hk+jAH1VjULMoUFV/pXCMtdgXOG5dmZp5YHr2e'
    'gagjpStG+5UxRrZzLkOC6qNVmRw2MCN1Q9T9TKGffTo7yUBv5Extnd/ul12vZFNThKpYbXsD6175KZLdtW/2yg/QkIbNkdNBlUobpbDok6QAs/djdGQGjVVy'
    'adPIsypBWNGhAKmYwIARuctdo9wAsvMS3YbFXeJGcZPGqSkRGWGpiEWY2HtDvyN8IW0lHPXx/MLc4zTf7N6D2+/x88jcR65NAGClEP6YG90hdX3Hof7fAgGY'
    'rQBGJSTFIv2Rq7EuHsYZD9DDlRt4cSbRybhrV4uLCbmmTLJaXHTtlSTifrxsdj0qtvEtJLfa9C3h0CvtCn/Iq1No3oLjTYAj0wOOjCULvkZwUBCJ8WJZH7jL'
    'TTqZDnYQHcaGbTgTp0aggBIiDdudsQUSRlufpO2x3KOLfleOWIpiYgp2TMlTk7YScA/uH9prDlL/wkldwFpyfIujN4KjiR4cTVg5ddMlCRWf7rlMDQJzPyaZ'
    '9CSUJF9qyHXFtdrN0Fs6DFuHdzNP706yTTVPF5pv4/4G4v4fUEsDBBQAAAAIAAAAIVzdS3PrggoAAIcgAAATAAAATE9HMDVfTTAyX2xlc3Nvbi5tZO1ZXW8b'
    'xxV9568YwAggoZQokaKd6M1oAddAjQIC2jwEgeM2TmPUhgHFRRqAMCh5ZdCiXNO2aK3kpUsllCUaTLuiKIlCGRjQT2nedmb/Q8+9M7Pcpbhp7RZGHwzkY7kf'
    'M/fj3HPPHZ0Tv/r1pZmiuDKTF/8sr4uw0lEvVoRqOvi3FnTLE9L3Q8eXLzs55TXljpsV4f1DVW0KubsULnWyeG9JvegI1Xos//ogK4K+r+4tTQrV7eExvlOr'
    'rXC1L9SLx8rZz2TOnRPy73214gq8Fm62MxlZPcF6qlEWgV/GTazoXVAVV93fpkXXykJ2eqrRlwe9oP9A+q5quLSN3G4IeeQopyHU/op6AZM3vg/rTTggqxVZ'
    'bU2fdUeE9e/lq/bbeqUaDlZoB90232mVg4MBbuwF/3BHHZbbHhZbk09bwUEza7cPuj4+kTsDoZ4N5G5FLrsIx2taFx4jSBW5uq52y+K8rO4FBw6tEhx24GZY'
    '98ilTGZKqOc1edA/PVb3KuLSjTsi/EsfoRV5VfUC3/mxvK6qLdXfDvwnqtETqu7IqvtjuU5fHnrqW8Rj2VdHnuwM5sVnXyxeu3X969uLf8zdvP2HmeLVr27/'
    'afH3168uXr9549rvbty8ceebq7eu3Vm88efpW59/hiWCg0OEyqQTi8OFntosi4uXTfjNPfJR7g1UsyxU4we1emhywgjQHlCisRzWqsnv1mgRu6rjqWoDjoSP'
    'ceHBdaG2espdEsCA2nQEgcBdIp+w2jnCzXRyMTErpoZw4tATfA6aokgB+CQ4rClvW2219Tv6o08zmZJ9rYSFduSJgyi7S3BMRE/I3602foebdaAStrRltz78'
    'TbC+54kS1pqamhKp/8XzhVn8+M0vpmZFhEfeeB2YEx/mP8gC3+WwDlwgCBsu4VjV6/hfnfwqIRcO5bnvw9UntiR+Rs9huZDVHkoLy3WsZY4nX60Mv/OfEHzN'
    'XflqTz6uEQJVY01blyfrfjuVN0VBLj6pqN2KmJNP91GKZKsxDKWKgkX9zIZbawKoDY77FDJ9QYt6A8I6PZava3r9gvXeFFvJXlwA+JXnDG9ErtTIWexjUtQs'
    'q62+dVxt3I9Mn8PTn1PloCBM8ZLfPU/5f+OXRHA8CPb96CO49lDdf4h0Vz3EjwMaX7BoYxF0f6CiLHEtwWGbbkqOXG4atAG4gH3fAKXuyu/2NAsx0JDXgzIW'
    'wg4o7TIRoklRKZNJhXQekE7QSIS4alN+68tHjmUNEVbxtbkGB8iXDtPrQROuwRqKXrhWYz+Zxph/tn2pCUdXYUQ4JbtSyaxFLsW/ntCfmu8mx0Mf9y7NxrGH'
    'UDd64FfrRUnol/Lxl2xu7cNChLnTYwOukc/ndCpUY2DiRvuYROg3ilxjMSSdHhPFAvz2jfNj8DFMDXA8nWQqZpu0XgMoTMzOBD4Boge0ZMUc/9C0YsoufNRB'
    'G0GbcCtwT+TpDX1vkhPQrYQu+RiRUWILxlisK6WSE9ye0Iw3SdV5VDGf0Z7/lrBGsknJjNNWkqrC544lMhiLZkMuY5ePZj6I4gBniaeocJo7xl2hNusSmIeV'
    'cnUnyXERv43gJsl3TNTc0tVRm7KhTemYiJHpBDHT70uW3Ay1RcxWMFfaWNp1lOoKRHUAAZVotF5Effb3qFGGPwwBEFXcb7JRBQbZyD+4P5dyv5hy/3zK/Qsp'
    '9z9Muf9Ryv3ZmbQHs2kP8mMfpJcTMZ0JP1AFkkQWBkGvzK9wGohyICJI3wQ+tW+Qy91L54nH5G4LSOFOfeKgz08muzp/x+A1l/nhZWF4yTTSGKgT15I014u9'
    'RVv337hkFtIitDA+QqZHjn2QBouF8bhID3aBlBLkkL9nKyqsOyRGg2NPnWzOG6qhGCdYE+KWmvq2B8VKpCrdGl1p3kT5MnXihXYPhKrfNCWsOxDkHdgpfLYD'
    'DlxBVqHmIg5dPSGyQssicULVUm3Jnf6QxMSEEGKSZIsuT5oIzENSvKRhAIT6Gmw0JWa6WcSb4AS9BgYI+mS3LJc9K8/hhn5IzbDrgt9M04xoJGqOHNThwiB7'
    'gtsZdrbdfRQRuhXotzW3p44mdiZJDCOJ9hE+I4WN1dl4Xj9hCH0Q70QsJBuuFQTxDmIGEsK6zSvKz/cNL5Milx1kYscklkzbOLt7JExMdbpnZUekO1lznPUg'
    'VcSrjZrAcMjDga5ZTRmUzrlgvxc5ySWrV54YljDeYIBxJbMOhruwTMt79t/6NJl0KiFCIY3kI4Khq7Z7RO4ksjDJ/ERFaa8J8mfczejpCCqrW4nG5PjkQ8Pd'
    'EQSqx+V4tM515rSwHA9DUFhp+ipLympUV2VJUY3VU1lSUmN1VJaKfax+ypJySugmyrjVn06kP9t2PF/IM3cUiGEBkpRh4jXRvCQ7eGqkNXXrZa3EszQa/TYE'
    'q0uTs6qvDqfNEcr6T5hqeHqA+gu39oQeq0fCyfGvIGo2CaZYOv3Q2TfAImM3aqAwofYdtUWFQAKGEBvNn6TxMTCTh20ys4XfjBOUpimtp/5Zt+ZyRREJFMIz'
    'RQVchmriORqzSa8C4zSmh9VmzxLa8Wk8rG9iaTjmyUebmtL0NdokHQqZlOr99IkHUQXjGFsSOxIbZjILs3cXipyQ3SXVfMxPdyy/2JlfK1wjVHXmgD1g3Vya'
    'gFo6h36zhwBDxjXm6TOri5fZ5eTxy+nxpTiVxSaaTGbOqOgEF2qKhCWsmGEGHynh91AkcyegvcyYRbiPVTlv5LTwYJytBR1KLgFrmU4w+EO+tFKQj03858An'
    'm6oTGTs2gtiE0RRSQjXqEyxoUGNZUKNbZ0bWvIuXiWMZH4Yf6fzkuao7oqhpkvHj7ekSddSybzIWI2kmn3FuzbFbY/FGSImdd2FWf7ETpSBXVFvr2J4GbrI+'
    'p23USMAdXdhyeS/o7l0p3L1SZEcxKVQ7MTPQ+B5Qg6D8bzo0JOuth6dklno1mU5zyLpwermDvIdLPV2gFDC8FT57ECuNKfGJ/vjTqEOPYuzs5K3DEd3W9lm0'
    '20M9CgfMPXFs69VvocxpWEz07MkESnVb4qb9jE4K8VV2jIagFphQBrLmUiszH7xBS6XjkwTPkJ2x7mp4i85riKDRmrrQ5nHOnfwveiPbS4n4+Mtvcr+8/XXu'
    '4y+v3ZkX3AT6hNVw66EFEBu+uY6aNGy1VoYPojC0FdhiGOMLcwzqNMN7DaYYNiWRe0vFDODT46iKYgg9PTZqBhxBuojK0/CrS1GBn1BNdsbjjtgz2lbWatiH'
    'aXEMvWsVwEaahqerjQ5ZTNQszmen+exeP38jup4X74CuoyOqlw3YxGVmyCp+tKWPNyg/nFjN1jmM42inHHq0e6AJL3BSc5pmOC51fTDoOcRhw9atsUHyPy4s'
    'qcNjf5S5oQuiX2vvGPFganpIYl/cXrz1VY7/ZHIVYb/Kv6dvfW57LkAVrq1RKbzJsXpWnF0xm74XEeGGD6NieM3HYfBWbXFevKu2+B4T7wYThTgm3kZ+zIv/'
    'R/nxHj7vBj5zcfi8lcybF/8Dmfc+3e8m3ZnYX5V/4o/KOa1f9N+Woz8tTzMWtiEkBzTP0pnSoap29FmKvolcmVStrmOwJq073PxfUEsDBBQAAAgIAAAAIVwF'
    'XnF7IAYAAMkMAAAkAAAATE9HMDVf6rO17Ya1X0FJ7Jet7ZWg7Iuk7Iq17JWI64K0Lm1kbVZNb9tGEL3rVyyQa+smaYO0vuUUtGiaAuk9ORU9NCjQ9gdQNmXQ'
    'plpLsGhTDilQiWTJqYLSEmVTqJyDf4qP3OV/6JvZ5YeSnAhR3J2ZN++9mTvi+6eP7z4Q2WKZ7y3Fo2+FOnmXe5FQ7kgdLIXyHLmTNBp37gh5aSs7vLnK5la+'
    'hw9Oe2r+TijfUW7QaNxagfJsOfDNhfLfVLV8oQYd+aat+pbANULtzFRki7zdke5UhbaQcS/f9VSwlsNAqMRBOH59aOOIGvlb5gL6O+97Kkxlx5cHPb7JHSGZ'
    'ZUcFQ3V6LrI0ViNLqAh5ODdX6rKXxZY5pMJE5J6NlHMvoExwgOJkyxmqyj0/W0QmkpBzJ/dTOV7jqqY6TmpZqJMO4Nnj+IOxXNlC2RfyYIxobpDFTTwDWw1s'
    'PO1Avm3R+1F+vC9U36YicE5O2vJopCbAY+Jsicfy6CJL97PFGvmdZ/Nz3I+ULAqPzOTZTCf3GcOnS1ShL6Rjy/Aa9+821SWXouL3wF8nuwzU61aJMkPPWHhR'
    'Nm9v3VphoyGnaxXRqz66jOrofzVx5CylXIpAuDbyCnhAiCgwbWfw0lgOQ6QtXUe6oy2QhwDSYXHQLuqYDOk+lKy8A0SrPjedEF9mF+BGmFA1FPNVx+ShJk0V'
    'dQWQRp9NHnKRmo4ylvQiAb4msZsr3I1umMZTu7OrdXYRcyY702zVNPE1p99OCc6C86eJ8puNxufiyf1ttLiL9gIbBpz7xGUvllkcUDJpTFVSzVncJ5yRHvAq'
    '0NOk1l2UnYDq1dLhVOJXNSjkIc6vqy+4GxpvediSZ9ciP9zHS6IhETP0IS+kAFCZjHVcKfuvtnEX/zOMoRSRJZaKOiYjjQcSs9TwAgrCtaOuWvki/zvNj8es'
    'v5WNNtZai4qBplAXLU3viZct8JRzDyJB0mBCRAcJdWbMR+2mtB5QWlSbpo6QbiLd1c0VnEe5UflTHawQLLtso9yyNcxNFmAUoAWVWnvQHAPtLmsJlzLtGOHd'
    'WoSZPnxrAe5EhmvmE51LlMcKLI9zClV8poC7lAuL9WRYE/rZfEY3nccfKGlnJgft2tkPJGkXzCaL2YDo4bbIfXjITKhVIo8CcNzoSo3+KumixVsp4T5YRgJi'
    'j53/U4tTiJSlk12lIn+F9r5tyWglCNABmRdI8rqlHL8oLD9tox+gBh8mFAobxdm2lfd7dag0T/UZo13+ELIcJhv2gBS7QBHFW+okpsSrZAspSTBtN8yP6Yw8'
    'TzhndlcbQ6EIls2vYQ/sNodwyGktCppMQ8igR0rDizPA+Wb9SU5+vS1k5ODbvG1xRldT5V2Tb1BCiPVg05wKafiUXNUB1G24RiFx1TAGqwhQ2aWifBl7QNrx'
    'IQ/oZpHIsc+3BlExajbS+mabkMi7ATeMoGMBbFgloVWZ4Sb9jXnoKzR+nH0v5kmbDlmm57WI9+5uYx52gBXrWs4S5AsLxLyu8RidixgH46rkJ2xZm43WF5QH'
    'KzfD+7EetbO0UDrjh93B7det+clDoZwAgzT3vWIbAQUwshuNF7xjPH9y9+Hz8hvd8ef6y61f/nz56wuueBio/2ZCpm3oXZ3GEA4jcHJdYRWgTnqIn+6R3rTD'
    'Y9FgzfAP7oJmXYkmVopugB+yU5GP22EYjfIAVbmZQE7lIFIDCHBsVAln2pABXH7MDmXzmD9rlYmGSaVeNGlTrMQtN/qCPjreZ7/CiGDZCpAl32UxFHbAfkeT'
    'MmwbqbAtm8KresBd49V6w2H3W5AU2XJZ4vXvJw6NpMk+cfO7Z09/IPihYBYFop0m+E1a05X//NvvL/+g3dCwg4jD9qIXrRhzpcbRhu4+0/BDvMqmWAZbs3Jl'
    'qzZRlo4UXSOzhsNwMZBNM5Q7QVmdzdQ87JcK+XiQGbNNHHCBRwybWU3LtGUAs47PZrhI9QGznv746Nkzhi7uZe/Tjy8vmqu7ysrQkwzkA3uLMcY7zheF557s'
    'oY+gDLOz7im8OaLbGKuw8IDXDMtAoy2WN7ujT2xnNbCNaRug+XBBfSSmNxMujPddlrkeMJqSZCN6WR50CdpqJd8QPZriMJnjaSnJiKcaPcwoIk1s8Khkx/9Q'
    'SwMEFAAACAgAAAAhXA1xTZ/1CgAAohcAABwAAABMT0cwNV/qs7XthrVf7J6R7ISx7JWI64K0Lm1kfVjLbhtHFt3zKwowsslIsp5+LA1MoDEwwgyyCZDBwEhm'
    'lESIIwWSjCCAMWhKLYEm6ZiKSYuUmkzTlizRQ8MtsklRMLXxpyS7rup/mHNvVTeLsj0LPdjsqrqPc889t66Jv/5tcXpBRN1evNsTqrGn3DOhKjm5FWYyn34q'
    '26GqD4TshtHgkWocywtXxLm2auwI5btCbWfjbe/duTrbUQ0Xf4/21ImjmkF0FgrVCeNsW75sq/xRnB9gs3xZyPZAuVgh+65y63G1gq1kIScLR1MiChxsiI09'
    'cVPlqmq3if1k0RHaCm2EDKqqXqWXYLOIBoFs1lUdptQvo64v7txVNefq0e/Oo/NhdBYInCqPqzj91Slvsv86rvgCR8S1Mp7igK4PL+DCxSnMlMdDE5K40kqs'
    'VIUjslAHg96Iem35pBZXqliH0OVVvpe8K/sONseO/DLM6LWjgOIE7y6qdFKWQlfHusOS7A7IqKhXUl5THbTSeLg40IpTJnPtmlAnbvzskVA5D3ZmMjMUvay4'
    'v7yxsbbKrjVK8kWRHdCB+WZt/YcNJLKnttp00n5JxLUKwhpX4MxWG25NCMRFHoeUJzyJK68nKJ77pahTFPKkGQ0G2G1I0ZanQ+XDaj9re1voya5jfS3DI+UN'
    'ZSsUvztV/S1iHspGUcSHrgnt706NlsR7nip4ozhnZqdEXAll4ULINwO1k7iEoNVDwh5Sr45wRsfFJzw9LCEFZHrUCaLuEE/6ZcQN7+WQGIoE+0K4lQU4V6Wo'
    'JMk8KKvOayzp/Nf2ohuqZmiZNDdFe1B0XwyRUjHHMIc521mG4GEpsZFzOUBZuWorwL44Jp+gFcYR5FAlFqrkSYPLYytE5uUTbOb5jCfs7vmcRz9LFscVF9Zy'
    '0gAi4IIOBoz8PeOFZe/8FPtQc22nYErBS3PPUdDW4iihztx4u64B6le4JFAwjPJOgCd2HfQ89XxHxMUiMME+oTKrgKvnAtu0RG6dRhfZxJiFKWEyp/PJWDep'
    '0rnEcW8G2rDE0noxedvKXGrC0sw0njNc0/K5FQVcnGwRyk+zA8gN62zreQFqy4e/Ymn+3fnSDfzc4tD7LgiEi6IOHglxBpEBHZOrouqsCGe++O7nP5zyKN3x'
    'weN353fuchQarknfH05lQvxl7Se8aQCXUE7CSwwHDjeypXmBEsqw4dVffPfVJi3XQU/iZs6NiyVZOMVHOwNYRkmInzwCj1ie37mLz8BbUT4F/msuEb5mPjwj'
    'XlV9jyoJVYszqILwEkBNq1QFGajZPDSCtDZCM1gbRJvJTMIBXz4PEFMK3uIMIYxIDaUTBRUDvsUb1mNVubRM5d5EO2JjxJaiKV+6RJf4Wh7D8bIrm57y6yB5'
    '4n1+HMo3SLsnGM8pLpK1Luf1wMYCyGlxztQxs5EDW9+dL87iZ26Cv56nr2EwvmZXgJfFBfzcIH7DWfLVzjh7TQq5+yvtZDKcRoLbkw9KdVDKrajTStx5mfbB'
    'z2c0g5flb4/ErdlPhI6e2UrttxmkBsDVHA7m9/ueZOjzstvTnxiGZJf7LdU8My1Kvnr7sXaWLgl1qos6iAxDzqzpSPrQUUeaNC6BxC2LEDh8oG2pcv1jYBHV'
    'ewo8VST17sCR+WMTcG02EGgCqmlAlj7QVNVuEUyX7KkZl8G5v2u6DZifHug9TCuzdQZYedCMgl+JmD30TfxlyYIKMrjSJNQosdMc9qTT992xLCdi6MUQ+Gc6'
    'hiH5C6xOyNMwOfcj2yomawGFI6jatqqWgeluTGLdPe707KdZxpjpulRYBD6bZykToCmc/vQMconJKGF/h2KB7stwYSOZZ0E+Bf9D+oXy2nChvthM1alCUVlF'
    'aGkWRHT38QRSWsJf0FrnkiuQPZ9gUoH5tLiSY92IQNEXgEF0Ti8cAj0tER8UYbrxcSI9+/wUKecWyPuNV66WnKYuSb8Qq7GlhopA5WidQnpvx5zl18FXz8L3'
    'iA306qiXdVnyUsAzPCgbOoBGOZoWlnTh/QAfxtBBbePVDiGhp4JL0p070r8wjUx32YLPioZTzDzLfQbrKLFjtKA34YJs0BNbv3D1siMcWZIHUWfb1JJ6GnxA'
    'mp5kIZG5FWTz0SA0Wn7U6xJb2Ub2r4GsV3lDNjSVbB8Bjh4WisUopMJqh6hW9DyKJboT0SCaMraGbgAZh9xJOrp+3CPqpB8S1EYRqt+yBPMUg8gq2wXPvWoU'
    'OlpR5BBabEkYIm7a9VlEXMlSEkKS3b6bdjHHPGClXuOZgLaGWTAa28hyMEqNVuTNACVJGIECQWn96c7dTCYpEUfM/+cGImreMpTKgNLH6zaNKhGz1EvqWoX4'
    'DAPLWL017RdXakTKjRKnkKcpkY5hiTTTvrjCtFrd+k0sGiVEx8/Jp15MI8rYSMSVctK0Ak8sYt7ABID8ebxxgj/k9BD7jbzQb30Ufck6Q0chkTeBDkQtX+mW'
    'CNnHUpX7uj5ayx5WNZXT6G3VqA9WK1pEkWxDfAwPA63PA5CBcfm9TpDkBxHbzonFlU0R/4Lx9IwWz97k3uBatLx0k3e9CBEzSqKOLxLW19JkK5wQS7dHU27S'
    '52en8QaIeF7MxL9USWdQO5ijZejYqCLmNi35El0+Ggz0VnTWE8BH20RmjPVtc5L+mrvGGHUkYkGjLqAGDx90R9eu1DzMsGY9CxPaw8xdaOgYcMzJFOAhhPW4'
    '1tLKl3CBnVtBat2gySOYzttHwKCrR1toxhWdqExmaYFBkR9Q71jQKpAFFAOd1Ma8uuAmtkCyX54cxZW9RJKdNNS5a2EO6TPDwohd0etH3ImOiQ+EJiMgL5Io'
    'mWG/m9x48HTTd2GVriz0qkIdX5tyNGA2SKH/7NZvJTehM32QsU4LOJYnuj+mimdM6jwLTfs0QkqQMOzzoEvn0KzmC26nePl9Bh35hmIp1M3AQkeMximmSz6J'
    'OnglZ/NVOqZdFUVi6RbXzzS6OY/EmDaTaxojK3OyPsTwTGoVeoBsONkjBKES9AgSdXuQfkkaUJswDS8gJuM0oLbrfDnzgQ5koyrp4shdrZXJPDQ0Ix4mF156'
    '7BYP8dXk5KQwv/HJrNT3OOKhdUX1f2+oJugs/Gu9Pxf1Alpy4pBRRd7vH+ldj35PJ/ifOJmN39+RjwPSBnmg9LBIN1r9LOeAh7RMhuejez+trX//9dra91Mr'
    'P/68+jXjDL0BkTloWeSl5aLY2Hzw7+XVzesba/cfbK6srZr5nYesPpi7rVnYZeBoySlzrqxfpsmrjYQmX8Gpy7JBeXK3R+MvfGwP9MUH8cCO4SaS2my8+PLu'
    '33WXc0zN0QZ04QIif+KSbtNMkvY5PqjcIrzFVVfPpt772m3ppsG22cbcqkj/SAaHeJW8YkEDhwdFAFEdBGQE6x5C99iADz7HmyPa5t2Ys1mzoqeaiWkUaB1Q'
    'oyGtihljar6vImjqjkk3gyXoFkuVvzdrMCQQk07OXL0Y7a1bFusONiWTsafitKGl/YzZkVRMfWirIL6awX5EByT56F2zmGsfWyDsAEg6Il5BA3upmSptQ7JA'
    'ZEMRN345Gob2rMMoQu9lq9v4rWd5FQwQXnONmS9vrD1Y/9fyvbX1lW9XVr+6v3H925XNe+vL3yyvL6/i+ez07I2Z6enblMZGzhYaCNU2ZqPTzz+78+elz7Cx'
    '9mLlhx/X1jeJOZ+e0QXG1Szq9OhCuK5viWB9MoBUaTDlW6LRJVHdmZmmbno8vHpLlF4beenNF5FfcjOjFY9Ls8te6WqLyPwPUEsBAhQDFAAACAgAAAAhXMCF'
    '1ZJJBwAAww4AABsAAAAAAAAAAAAAAKQBAAAAAExPRzA1X0NvbGFiX+yLpO2WieyViOuCtC5tZFBLAQIUAxQAAAAIAAAAIVw+wKIrBgoAANIcAAASAAAAAAAA'
    'AAAAAACkAYIHAABMT0cwNV9NMDJfZm9ybXMubWRQSwECFAMUAAAACAAAACFc3Utz64IKAACHIAAAEwAAAAAAAAAAAAAApAG4EQAATE9HMDVfTTAyX2xlc3Nv'
    'bi5tZFBLAQIUAxQAAAgIAAAAIVwFXnF7IAYAAMkMAAAkAAAAAAAAAAAAAACkAWscAABMT0cwNV/qs7XthrVfQUnsl63tlaDsi6TsirXslYjrgrQubWRQSwEC'
    'FAMUAAAICAAAACFcDXFNn/UKAACiFwAAHAAAAAAAAAAAAAAApAHNIgAATE9HMDVf6rO17Ya1X+yekeyEseyViOuCtC5tZFBLBQYAAAAABQAFAGYBAAD8LQAA'
    'AAA='
)
_EMBEDDED_ZIP_BYTES = _embedded_b64.b64decode(_EMBEDDED_ZIP_B64)
_zip_path = Path.cwd() / 'LOG05_실습자료.zip'
_zip_path.write_bytes(_EMBEDDED_ZIP_BYTES)
_log05_extract_zip(_zip_path)
```

```python
MISSION_JSON = '{"code": "LOG05_m02", "steps": [{"title": "근거 등급과 합격을 분리한다"}, {"title": "AI와 체크리스트·G 기준을 작성한다"}, {"title": "고정 기준으로 독립 검토를 받는다"}, {"title": "최초·수정 기준과 결과를 저장한다"}]}'
# LOG05 학습 기록 도구 (표준 Python/Jupyter 환경용)
import json, zipfile, io, os, datetime
from pathlib import Path
from IPython.display import display, Markdown
def _load_notebook_widgets():
    from importlib import import_module, invalidate_caches
    import subprocess
    import sys

    try:
        return import_module('ipywidgets')
    except ImportError:
        print('이 노트북 커널에 ipywidgets가 없어 설치합니다.')

    try:
        subprocess.run(
            [sys.executable, '-m', 'pip', 'install', '--disable-pip-version-check',
             '--retries', '0', '--timeout', '10', 'ipywidgets'],
            check=True, stdout=subprocess.PIPE, stderr=subprocess.STDOUT, timeout=60,
        )
        invalidate_caches()
        return import_module('ipywidgets')
    except (subprocess.SubprocessError, OSError, ImportError):
        print('이 커널에서 ipywidgets를 설치하거나 불러오지 못했습니다.')
        print('설치 가능한 환경에서는 새 셀에서 %pip install ipywidgets 를 실행한 뒤 이 셀을 다시 실행하세요.')
        print('설치할 수 없는 환경에서는 제공된 Markdown 작성 양식을 사용하세요.')
        return None

widgets = _load_notebook_widgets()
if widgets is not None:
    MISSION_DATA = json.loads(MISSION_JSON)
    CODE = MISSION_DATA['code']
    OUT_DIR = Path(CODE + '_results')
    OUT_DIR.mkdir(exist_ok=True)
    BACKUP = OUT_DIR / (CODE + '_record.json')
    fields = {}
    for no, item in enumerate(MISSION_DATA['steps'], 1):
        fields['activity_%02d' % no] = widgets.Textarea(
            value='', placeholder='내 판단, 확인한 자료 번호, AI 제안과 수정 이유를 직접 작성하세요.',
            description='활동 %s' % no, layout=widgets.Layout(width='98%',height='130px'),
            style={'description_width':'75px'})
    for key, label in [('first_thought','AI에게 묻기 전 내 판단'), ('ai_request','실제로 보낸 AI 요청 전체'),
                       ('ai_response','실제로 받은 AI 답변 전체'), ('source_check','원자료 대조 3건과 수정 이유'),
                       ('final_result','최종 작성 결과와 근거'), ('limits','미실시/미측정/확인 필요 및 교관 검토 상태')]:
        fields[key] = widgets.Textarea(value='', description=label,
           layout=widgets.Layout(width='98%',height='125px'),style={'description_width':'190px'})
    status = widgets.HTML(value='<b>아직 저장하지 않았습니다.</b> 작성과 교관 승인은 별개입니다.')
    restore_input = widgets.Textarea(description='복원 JSON',placeholder='백업 JSON의 내용을 붙여넣으세요.',layout=widgets.Layout(width='98%',height='100px'),style={'description_width':'110px'})

    def payload():
        return {'schema':'log05-student-v1','mission':CODE,'saved_at':datetime.datetime.now(datetime.timezone.utc).isoformat(),
                'answers':{key:w.value for key,w in fields.items()},
                'status':'작성 기록 (학습 PASS 또는 교관 승인 아님)'}

    def save_record(_=None):
        obj=payload()
        if BACKUP.exists():
            stamp=datetime.datetime.now().strftime('%Y%m%d_%H%M%S_%f')
            BACKUP.rename(OUT_DIR / (CODE+'_previous_'+stamp+'.json'))
        BACKUP.write_text(json.dumps(obj,ensure_ascii=False,indent=2),encoding='utf8')
        status.value='<b>로컬 저장 완료:</b> '+str(BACKUP)+' (PASS 아님)'

    def export_json(_=None):
        save_record()
        raw=BACKUP.read_bytes()
        try:
            from google.colab import files
            files.download(str(BACKUP))
        except ImportError:
            print('Colab이 아닙니다. 저장 위치:',BACKUP.resolve())

    def restore_record(_=None):
        try:
            obj=json.loads(restore_input.value)
            if obj.get('mission')!=CODE or not isinstance(obj.get('answers'),dict):raise ValueError('다른 미션 또는 형식이 맞지 않는 백업입니다.')
            save_record()  # 복원 전 백업 저장
            for key,w in fields.items():w.value=str(obj['answers'].get(key,''))
            status.value='<b>복원 적용:</b> 기존 작성 내용은 이전 JSON으로 보관했습니다. 다시 저장하세요.'
        except Exception as exc:status.value='<b>복원 오류:</b> '+str(exc)

    def export_review_zip(_=None):
        save_record()
        target=OUT_DIR / (CODE+'_review.zip')
        with zipfile.ZipFile(target,'w',compression=zipfile.ZIP_DEFLATED) as zz:
            zz.write(BACKUP,arcname=BACKUP.name)
            text='\n\n'.join('## '+key+'\n'+val for key,val in payload()['answers'].items())
            zz.writestr(CODE+'_work.md',text)
            zz.writestr('README.txt','검토용 보관 파일. 실제 평가 PASS 또는 교관 승인 자료가 아닙니다.')
        try:
            from google.colab import files
            files.download(str(target))
        except ImportError:
            print('Colab이 아닙니다. ZIP 위치:',target.resolve())

    for i,item in enumerate(MISSION_DATA['steps'],1):
        display(Markdown('### 활동 %d. %s' % (i,item['title'])))
        display(fields['activity_%02d' % i])
    for key,label in [('first_thought','최초 판단'),('ai_request','AI 요청 원문'),('ai_response','AI 답변 원문'),
     ('source_check','원자료 대조'),('final_result','최종 결과'),('limits','한계 및 실제 검토 상태')]:
        display(Markdown('### '+label));display(fields[key])
    buttons=[]
    for title,handler in [('작성 내용 저장',save_record),('작성 JSON 내려받기',export_json),('검토용 ZIP 내려받기',export_review_zip),('작성 JSON 복원',restore_record)]:
        b=widgets.Button(description=title,button_style='info');b.on_click(handler);buttons.append(b)
    display(widgets.HBox(buttons[:2]));display(widgets.HBox(buttons[2:]));display(restore_input);display(status)
```
