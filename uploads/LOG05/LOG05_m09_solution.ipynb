```python
# @title 📦 LOG05 실습자료 자동 준비 (ZIP 업로드 불필요) { display-mode: "form" }
# 학습자 화면에는 결과와 업로드 인터페이스만 표시하도록 구성합니다.
from pathlib import Path
import os, zipfile, stat
from IPython.display import display, Markdown

LOG05_PROJECT_DIR = Path.cwd() / "LOG05_실습자료"
LOG05_PROJECT_READY = False

def _log05_extract_zip(zip_path):
    global LOG05_PROJECT_READY
    zip_path = Path(zip_path)
    if not zipfile.is_zipfile(zip_path):
        raise ValueError("유효한 ZIP 압축파일이 아닙니다. 실습자료 ZIP을 다시 선택해 주세요.")
    dst = LOG05_PROJECT_DIR.resolve()
    dst.mkdir(parents=True, exist_ok=True)
    with zipfile.ZipFile(zip_path) as archive:
        entries = archive.infolist()
        if len(entries) > 20000 or sum(info.file_size for info in entries) > 1024**3:
            raise ValueError("압축파일의 항목 수 또는 해제 후 용량이 허용 범위를 초과합니다.")
        for info in entries:
            item = Path(info.filename.replace('\\', '/'))
            if item.is_absolute() or '..' in item.parts or not info.filename.strip():
                raise ValueError("안전하지 않은 ZIP 내부 경로가 포함되어 있습니다: " + info.filename)
            target = (dst / item).resolve()
            if not target.is_relative_to(dst):
                raise ValueError("실습 폴더 밖으로 벗어나는 파일 경로입니다.")
            mode = (info.external_attr >> 16) & 0o170000
            if mode == stat.S_IFLNK:
                raise ValueError("심볼릭 링크가 포함된 ZIP은 허용하지 않습니다.")
        for info in entries:
            target = dst / Path(info.filename.replace('\\', '/'))
            if info.is_dir():
                target.mkdir(parents=True, exist_ok=True)
            else:
                target.parent.mkdir(parents=True, exist_ok=True)
                with archive.open(info) as src, target.open('wb') as out:
                    import shutil
                    shutil.copyfileobj(src, out)
    files = [p for p in dst.rglob('*') if p.is_file()]
    if not files:
        raise ValueError("ZIP에 압축 해제할 실습자료 파일이 없습니다.")
    LOG05_PROJECT_READY = True
    print("✅ LOG05 실습자료 압축 해제 완료:", dst)
    print("📄 확인된 파일 수:", len(files))
    print("ℹ️ 실습자료 폴더:", dst)
    print("※ 업로드·압축 해제는 학습 수행 및 평가 완료를 뜻하지 않습니다.")
    return dst

# 실습자료가 노트북 안에 내장되어 있어 ZIP 업로드 없이 자동으로 준비됩니다.
# 내장 파일 5개: LOG05_Colab_실행안내.md, LOG05_M09_forms.md, LOG05_M09_lesson.md, LOG05_공통_AI역할실습안내.md, LOG05_공통_작성안내.md
import base64 as _embedded_b64
_EMBEDDED_ZIP_B64 = (
    'UEsDBBQAAAgIAAAAIVzAhdWSSQcAAMMOAAAbAAAATE9HMDVfQ29sYWJf7Iuk7ZaJ7JWI64K0Lm1kbVddbxNXEH3Pr7hSxSMmDtDAI2orQCJq1RaQ+hIZSEtK'
    'qFECRfShWiebaBMbYkOcbJK12RTnq3XbTbyOHcm85Kf0ce/d/9Azc+/aC1gCFtl3556ZOefM+DNx6+vro5eF2i7FhaY8KQhVbMTry0JVHTkfjoxkM/rE5PP8'
    '7KN7+fyjzPSTF7/ck7s98UV+JndPbZSV7Qm1EUYtX6iGJU9toRYtfG5eREC10o7XbLnjKb8Wr3Qzv00/UTUcc5ty770sOrLYyAjl2+o4VL6n6hXZ6oofbn4T'
    'tXoiagOWHVcP++f6dwhVb2q4qmaJqBuonRCfVZR9JOLVUL5cI5jy5T9qPVRvArWPN6orAGNCjYxlRNRqx0ttce0mUvg7rvomcbVpiVtTc3P5Xxho7T1lB1Sq'
    '1hVqvgmsIi6VZfHgrKNqrmx2zzrXblLOcn8n6uJMrUcvyoOe8nGrXxjcijC2XLX5yuB9Uj8Uycdja00d/33WkcW2bFl89bserpbBdirAtW9uini+IVfWRFy1'
    '8Q6wL6W+j9rlKLQEgiDSv11Zd4U6DeUbD2DNPfVd1FDOu0KeWLJECKuyFVJEnYaqL0p/N666nxbtYkZ8+9V3t299L5C2qpfRepRAqPWKKtYY8hEwBVSwudyv'
    'U5OzU3PPZp7ykXc9ZfORuEqNxr+Gbun2bnaBScSvumq/gp6CNUKubqKggrhx4ono+D1RCQEJgO1Rk3XXEZAqK8v9aHEV51utpMrmlMtc1aWIOr3oKOBCxZtV'
    '9Bf/NX1ue+qPRaFsP15g1HL+IDotJHW4lNH55aYnH0znZvI/PZsirEM7CX2VCSyCREEBf0CCEGxN5x06oB5dDvLZeP55QCnGWyVV9NG4fScJ8LEg7j58ceFG'
    '/vmFuw9zXOaoZcdbL8860bEVL/n0RDN61Nsf87OP53R7dSH6MVA2ioszYEO8CUbt28qzSQaa59w/t0WFizpduWdrXXTVXm0oSy5nOGfoWLmOKnK7SKENi0XN'
    'RRfxawd5kXihiSjwiHUG0LdZ+vh61hxVG03qrKqvybfLV8bORYFlAl8dPWcCM6KTQ7VzNAyRkEuv+aY3R1F3GQX1y8pxSbamxvz6fFPWS5yraf7KKSXp9bi1'
    'QzxEqKUSK7Belu9KDAtaxPlDcHsTomqj45Zc2VW1kPSmGhWOsWnjODGXSOdXh9bw84yYGIeMG2orxP0QMLkeIdHOOjE6PqkcD8yO3apWuHHbzMOnj2c0ZUPp'
    'N8g+qlS9MucJ4XRLEg61FcCd+g7ed+LYteGsxjPkfkl7QnRcIo+WxWaatXyGXBTtg25jOxgDOwVnGxrH5Hoe/9V3TN0uPLm1Wr7abkLxn+UOjgqCV9/9z9rk'
    'XBxb7vUJghvibTtlppq4JhujHvkn4p6edQYdNqWQFYjglPX0gSlQsWulJHUg05kNMh4Z56YMMBKnNEzBeD6eNkSFQGhV01kcUksvBTtbjSdG2lV08ULHCBZJ'
    'Ks9nVtUdxGUNGFkVvSiwh1CfDHLBMDaueDg3XA5UAR4D4vaX57Nnndt3zo9xpxYK8QLnDixyr5lUpmThG+iGJYgn3KGOfBPb5BEMZygxQWVlp3+RaUWrTeOO'
    'gKZnEN/ouNrj/Cpw0uu6OkNxG8aR7FYaqB97VnrIDarFn+81aXx/2A2qyYbLTak6ifeRwDHZtw6HavEKEoFVAd7+coKB+n3j+4lbnN6Oh+w/FKs8tuOVnvEV'
    'Djzw274uyZL4Xar4k9n8z1P32cVplhVPKQW1iNT2amyYvNWwgzPlAJWNxpgVz0f6Nl5F2w5Sc/BxbvbRZO7+02e5mcmn04+nmLW0QZltj8hUIKn3Ky//pU4w'
    'H40YMU8Ag96kpA3qC6ab8YJFmj/o0sgwAxSRMbkTqg8pw8hViOkKsh0bhUIFDXzMroqXiPjESfYbdk0KTY8NdHAJVtYQRCDvIDVWCeN8kDaobRz0EH+dODvf'
    'jI5Cika+TDPlBDakFhwQxo5oBL1mPmp6DOYmNcH3YAWUIqmZGcNEw2ZgD583E+OIdVm5BSyWvBNh4NTSm4BZDiEuPQB4KYErn7omJy2sVDLkcqwmroQp/Ooy'
    'gMDax2B2w/aDkewowFylPO5kf78zzhkaD4G9+DbaCwA0lpPFp97Uvdb7Yndn6Nqhly+QDpPBGhvFl/31UsIxK95F/RENePci6j40JX2UANGaS7PmksjGr1yC'
    'pr1Ab4LM9MD8SEiW/GTntAZv0J6oSZPM4pTlpRF/4kTJ533nxBgcagXZrN4ric0YQoWQ28pbopi4dNaZ+Bx/r3C2LEMoOTuKIUK9Y9d13I9WXs4GxedtRBOE'
    'CldQfkUkW5zxHK06rIe6bgNtJQulNuuBVS0mA/7Zk5l87sHcBd4d+j/BDH/TayhiQliOqS7xoo+KZzDclBIODmAKukDUN/rRtHXIU6jYgECAWJNME9SMJwx9'
    '/ePELFZ69cPx1Oon3wbkJnXbgOF1FH3nWZnY5qAd/wNQSwMEFAAAAAgAAAAhXP7v9HUICQAAcxsAABIAAABMT0cwNV9NMDlfZm9ybXMubWTtWF1vG8cVfeev'
    'GMAvEiqJ+nKM6E15iGvAQQsEUR6KQnFqxwkQWYVdoCggFJQ9cmmRjZcIaS3lpUomdCSqDLLmh0Sh7It+ivO2M/sfeu6d2eWuKNmOURh5MCxY4nDmzt57zz33'
    '3L0krv/u6uxl8dHs+0LvlbR8LvSTki7UMhnd9/R3W0Lf9/WRp9rDJfHZ53e/uvXFV3duZ79evz17efVP62trN+7cvHV3de7PN27fWuWvZ9ZufpbJTFtres85'
    'PVZ+CxZ/zpV1w1ddKbIi8D1dG/xm+drPucrSaHNtiM0dqeuSFwtNXfeyy9eEOjzQNVeEjx8Ffg4mdTUngoGvmzkRbc98fevevfU7sOSo74vq2VDo2n+Dbt2a'
    'DistVcirQnNG6KeO6g5wUbenq1KEJY9sB35V12Dufk/vtuiv8GEfh8KKq3faIqx4uCmv9ouxvZ56LIVyYrPqpwGeNuj2YRl/Dhq6Xjk9pn2FE6F3HupaUdck'
    'nqqnvTo5EFYOgv+4uPEgONm0VjKZS5cuYcmdibMhPcROzIlpDpqEQ3UZPpWx/ycy6DtwIpPZgMkf1WFLbCS2RvvitcTZDaH8sso59qFoz/EweO6zL1uu2IDJ'
    '6elpceH/GbLgh9JXP7RxmRN0yGjihzbsb4abF36r65t678Jv6TkfbF54VnrqcOv0WG+fAEwirFboV3pX5sKIzguCXRuG9Q81EXT8oDu0YdT326qRS4cxLDqq'
    'cJAOY7wG4OoTF1luqyM6px8W9W4ZAPYCX06oIzn5WrFcmRvzcWV+fGlhfGlxfOny+NJ740tXXjteC4jXXPiNS1jZB+4GRSQ9rJTEB9c/+VCEhVwKgPasKaez'
    'zlNmO75q1Cd0599CtQd679mkvf8X2AgrfV1okyn6haTh5C+3oj0Z7iB3u2VkbUIX6qE7UN8+Z14oeEtCbZ2gjKst4Ky++QZPqYo5XcmLBXDpBPwEYABnJhTd'
    'lG9gz5TohCnR02MiRCoRpr3JVyVxEUkcZY5+7eS1bFKhqUYt9Rz4HyEQG7Y0zgcwE4BH+wzWsb4AuBOhwIGYAo6k+tcjfFpeTEIoKuP9PBGwrpVUoU8Gguc9'
    'ZiS4hRXVjXca0Ah1MNT1HKqSEkQRi0GUcB8J7WkXFgzL2wZzMshkVub+vnLF3NgjLi06YGrm/sMDVXJsf+ANhlhMcMillo/4VPV2n3oQbeAORdAbUQRwwqHA'
    'fqErUhWqUYdYeY/OvMi5V+etfy9y1HSuzp0eX51XDQ/Ge1rW0HOUg47Ud7R/YGkNVpslChpaKMUHnQ2fRqt5l4BFTnQlNSZzKtn1TNdANMNqmftnJxc+rMMR'
    'QXkzDKa2ywh5Uz0bgMt6esexvifsUKClF1Hy/CxyOxVhYIE/IMkm4dzu9jd1vSRMO4Rj3JFt1FAMJzKyZHhYe0OOQ8kLXUm9l3yrbCPiqQZpCDpF1SZrqZwS'
    'iCmRgwTPb6QDwfAejwNaODw4JGo3/tqHpCcfNBINnjb8NIg6z6so/kzliHN+fkXLVETQXkBAuPtPEA3HKZZNEWNMxwxToCgtiRdPGnTMVi52Yyd3UPrCrgJc'
    'aq/IPdSchpnla8Bb0ClCMDSCAQFwS7V7hFSjTlgRssCyZIdHSkpA1MJTh0oHGDaURRvyQFBYAeaOykjXkoUMV6utEFO+hSbO11xkmymY7MdiyHXIJiyAa02V'
    'SnQLPmk2jpN0KrBjv01XIPiBZNFV6EpbE+hCnR8ZzG3UfFiEZhxSiyJ5Ee+KonjBLgN64hrVKjOKLwZm5iVQOPMpE9UKeU4Zth2KO1zD53g8deLcdCUmByLD'
    'ZvjkkYULNhuSpIYSi07Sw8xleZc1MzCjnrEjvksUHjn2GgrqFcgeX8jAEf39kHpHou8Qc0W+jpjPAN8EnzjWjgwGJgQMbKOOFCHLpX6h9veYVeO5wnhrvec5'
    'BRI4wW8bcctiLDCsCZaJ9Jq5hUgsTkH6Zi/CLoJrpzhZDx/UCON0LY1kHFoLnTcFCFGEqbVRRxGIX/jA6GjDyWZss8oirtoYzVGxGsrF99Gj/37544+zHy5f'
    'u54lvmUPs5bZ1VGeTEbBiNjJ3v1/AvxLPWcywmBsQ5icWykZtaIdomxHU9IlUot9jZuGka1mBnYoNNxplpgQo7nXODcaiUkaBD1UTDwQR1sMyUbRMp+SaOIV'
    'I2toHDfW7JOYuGMgN/rJ2ib+rUqmQ84be27zCh7Mu9ZPVhi602M2rfXCTZIPNKERdeZdcMColGDiD+bwHyHesLcNJaq3m+H2IE7ouaMGxFP5XNk/lZbxU0nV'
    'TUJsZG1ieXESkiJP4gYXkaR0qGfA16mETrgCMjCvBZIztxm2OSYoWtQz+p45MJEURJH8NfPgJLGJgsjtNqN7eGRCMBGBiZSonUrJZyOrJvm+M3MJoV/v5Y1B'
    'CuinX/4t+9v1v2Y//fLGX5Yi3QMS5t4d9Q1uOnGFGd5O8l7U3XEiSRvkwFlBmQCjxXSMV6OPGIHUnBIFS3QX9BEUadnRADlmDnpL0rP4VA4aRZP14rf+uToQ'
    'z+Spx1V6o6T3JRDGNMDT/Gic4S3UdyzIz5IPhwL+xi2WGSSWOS9nEprX6c2ZuWWO4Zl6PcN97sjTHZddh2gnFezBgyUxmkPGpSufY4hZQwnKiAYMK6qBkpbq'
    'VFhsR8ZnIncRDOpNVINWYCQVlOrkASry1apkkmJZzGMsK5DPcLcI4GEDIyVrGjS/ibOvtThqem8r6pEGcInXZE9oXbV6wuoAwyVW48OP8zS+YIEgLbAocF+s'
    '3127l+U3lauI9yp/nlm7yZVhkGo72Wu+oZwS48amLr6GuGLHx/Mk5MAG/QMG5lMYmDcYsFNPRCfx4GMGQ1K9PmMgohkLmmiCNMNJ1NWI3+18lqwfspiexlKv'
    'nVjD2OntHSbeLiYWUphYYEyMWhCpLyJ2FDK/+iBvYuXGsKD+QO+OE42Oqt62ABP7uLtJIsjEXM0dJH0DczyJv4H+h0xM58lb3+Hj7eFjMYWPRcaHVRojOY20'
    'wCNyhcVjj5GR6i6nx1dMXvm1RBaRBwxOj40GiaVmrJ5sYmHwo7lZfr66VIX2u/S/3fT/D1BLAwQUAAAACAAAACFcnYccc38JAAAjHQAAEwAAAExPRzA1X00w'
    'OV9sZXNzb24ubWTtWV1vG8cVfeevGMAvFEB9yxaiNxVF3QAJChSI8hAEilM7joEqLtwARQGjoOyVSot0TFmktVSW6tKlItFl0JW4FClAftFPcd52Zv9Dz70z'
    'u9wVKdcyoPbFgCGTs7N37tx77rlnhtfEJ7+7OXVdfDr1kXiTr4iwciq3HOkdCunZyl4V6sgPV9vyp7baaIYbfRFWf5avWkJ2fFWzMOmZUMWmbPtCuatqp6Ws'
    'QxEc5dVP9Uzm2jUh/91Xa7ZQj1fDWiuTkcUTVXRVPS8CL49BvOTMq4Kt1huquSlLeQFLqt6H9aD/hDyo2yLoe7JRF/LYUlZdqMM1tWsJtf1zWHXV7posFmSx'
    'OSEWF38/cF7VLeN/WPUv2ALP4V2EVTvouLloT8q1wh8tWlY186ruw9cauawetWUjr7bLynKEqp8Gfl6WHRGWyrJ4wNYe+XKjovbzQyGhpzoqYdUR8nkTPnRV'
    'sY1Bj/6DaThFs6bDH7BpH0ZkA6vsVFT1Nd6hLWYy40L9WJad/llPPS6Im/e+F+EPfYr4rCo6gWf9kq9gZdVvBN5zclxVLVm0f8lX6c2uo16uYSVPHTuyfbog'
    'vvr6wb0739z77u7kH+/fnbq+/If7Kyu3vrt958Hy9J9u3b2zzI8nVm5/hbeDTjdc7wq1u0nLqWoBe1W1vFj82GTCjMm9UyEPTpWLGNRfq42uSQ+DQTtPOYc5'
    '2CrLf5bISGTVclSxjj2Em/jgUKTUjk8YjMFGGaXtwNo1gtBE2pgYj3G129bxNlkT88GhL74IumXlNJATPU2/9WUm8zCa9pCSiOf4EHiOslxCg0nxYIzgocey'
    'quHJjmW2MCYewtT4+LgY8RdPlqbx5bNfj09jK15oeYAk3qzIfzwRes9yvyTCWhUVADQe5wFMWl9Vq1RI+gEsqI0T/UEbnSGjS+MzQu6vAucifF5Q+wUgrBUc'
    'AeW+fGQLgCno9YGundKwgdnIK8LrLgwAb88Q7mdATzuH0bza6UdlqLbXVX2EkTl8Cp8/VetPhbILnDFktt9QbhVoJAZQx63h167TkFtGshABgoB8tabX8M3U'
    '4LiEPImg25ab5cHr5+zcwKc5pAeLrwKRZvdwvujKlx5t5+b0We/mjADP4KGNoABp0muFL/YE1kTwAaxc7EHNkifW8DLzNNTc5Bov2AQAlAXWGCN/kSDsW5Zt'
    'flzdYJ7rtlGYIiy0wVXn7GU0juHKRLoGxDQjOcLfgI0EnAKGESrCrKHiFFTNvBRU4zHpVWQeMQS/OC7N6Z0Gh15E0m/BbozgBG45abyR+B9N0Bi84KkB2AVP'
    'yU/uCaPf5cyA+nQE42pIzro4ojOI6IAUQLtB59SEMS79d6p4ymH9VJ2gMrfbqFH6vl4CUQvNwVn0qbfyQJoN0nvkWj43NDs8NDc8dH146Mbw0Pw7x2sW8Rr0'
    'IiH7JSQ9rG6KX33y2W9EWMynAGjejbgzc36riLdsuFl19C80+L7a3Rsz61/CxqiOeXkryrHCbZsb63opi8oN7b7cOjT8uyDk2slZD1oFOHNX38NLaBh0QjTk'
    'l2tZ7JOUgVsW3Mua1nvY0yWa1SV61ovbGzWusf+WxDkkcZA5+m+7oKymofKUH/hLBPzQlMZoADMBOEzUjHWMzwLugpuMH1PAsUUd7aFYnEtCKCrj/QKJOlXf'
    'lMUuGaDGTFOwLYzITjxTgyaWEy4liCIWgyi9fRKBcSfhutY8Em8xYozhGI9gWvIxosuGBx8pKJo4WeO8OuBuBD7bJoEDa1khhC78Ac+w6jBz06skVKOm9BdD'
    'ZuLMsQblYJAmsDyKPhKtZWs6ylCnQm411Qsf6TlvcFQF5dIVkUsCmNaNNro75J7OfLXASzIqjEM6wzxc7I7Yl84LPQ8O+9ghNLPBE4UDwSN9PTOVABZ1Y6g7'
    'gF9u7KGWCE46myPMR52ykw+OXivnlHX043rQ9yMGV94BGUYaR58MONXbHuI0ZD6jhSzePirEh5ukSCUJfuwQWKFZ1HEFPqMam+hcrFuXpv+2NK/x7xMSSmXo'
    'I9IJUZxZIvME3eZ0qdKCLQ/u1SCpYZknQOm7DqVx0LDAWiaQRvzXouPR0g16503eJgnE1fYmT6carYooRjgp4XxF0qiMMxeEsndgggyrRvJwJeAw4OLbYJSF'
    'EG+iY5HSMqmptqLFDeaRkrBWIV1IRbrusqzzLNNPCQaYLfdwQtj2GR6894QdKns6e+nCZojkIkaa5S8oBk0/DKX9VeVuGhEagdNEDdR8YkWWdLUarBCQbToT'
    'RioufYjROcAeoZFrmnr0Z4g2OkCnWISRBDgc2YQ2ihs5Qqe5JBTMsSiBCH6PD4/G0FkvPgBEOTb7wv5b8qjK+42MT6T8mtF+mWBE3BTHY1B4HvtFnMVuJYkK'
    '63Py0IHi1Jm00cagzY/NMfpcklLaiI/BJqnDfs6ynwMuw5qcWmyYGYVWiqClQ0haBN06SWsUHdPT8X6Sy7DDLS8BAeav9ArcjPCl3ld/txJASq6a9nmOfTbU'
    'rlsn4RtTA9QCNQrLBvOwtylknPXm9Vp8Xp/EuQyunfU07WlD8D/u3UgYZAAZ/HR6imoDxnA4S7gkwmdPSB3QqcvQJS80uDWAVxZq1dAWVyX4jwmkDhr09c0G'
    's3HBDl88GRQeTHyhX/5yiDI17Z31RopFfaX0Hm1nYC27ODcWtRksRH2wTFyBvebOdVq+LEp1Wd1bOSbnbqeySRKJ+wwr+jHuygBEpxmtw6KXVEHBzqZkSS4l'
    'gDSaxni9c8qS+wCaKBukgH7+7V8nf3v/L5Off3vre8hOIn2wfccKd55qVACiCIquOXNOQ4waXlIxUVnq8EfXO5YbPq4nMZvIobEaoWvxY+o6CBqTCz0sohI0'
    'AQH8BWq3yeKmDoRoBn7E3+lzuiyDKJpcYFveaO7UTpoGpyuILk+MMIvwOj3Bt5H6+aVIdUFcJalGfppaZNWim0nyDgrCAImnxHBGtSCfBL+AFznmKHWAQ5NT'
    'w5uk20+SADY1YFk8ocIArSZuNjUo6GpIG9OiRLZ8Wh91aurd9C6+ORvuXWitBzBkkk+B++b+g5U/T/Lt7zLivczfJ1ZuM3o1msJSCSQv3vGeMCeGjeUuXobq'
    'WYusAUZnkqm/VN9aEP/LvvUBClcOhdkkFC4rDRbE/0UafIDFlcNiLgmLd1dfC+Lq1NeHrF951jOJ373e8rPXpPnFgH/9in/8muCkN+gGh87ldVuzgr5y04NI'
    'k8nSRkU5dLpNLP4fUEsDBBQAAAgIAAAAIVwFXnF7IAYAAMkMAAAkAAAATE9HMDVf6rO17Ya1X0FJ7Jet7ZWg7Iuk7Iq17JWI64K0Lm1kbVZNb9tGEL3rVyyQ'
    'a+smaYO0vuUUtGiaAuk9ORU9NCjQ9gdQNmXQplpLsGhTDilQiWTJqYLSEmVTqJyDf4qP3OV/6JvZ5YeSnAhR3J2ZN++9mTvi+6eP7z4Q2WKZ7y3Fo2+FOnmX'
    'e5FQ7kgdLIXyHLmTNBp37gh5aSs7vLnK5la+hw9Oe2r+TijfUW7QaNxagfJsOfDNhfLfVLV8oQYd+aat+pbANULtzFRki7zdke5UhbaQcS/f9VSwlsNAqMRB'
    'OH59aOOIGvlb5gL6O+97Kkxlx5cHPb7JHSGZZUcFQ3V6LrI0ViNLqAh5ODdX6rKXxZY5pMJE5J6NlHMvoExwgOJkyxmqyj0/W0QmkpBzJ/dTOV7jqqY6TmpZ'
    'qJMO4Nnj+IOxXNlC2RfyYIxobpDFTTwDWw1sPO1Avm3R+1F+vC9U36YicE5O2vJopCbAY+Jsicfy6CJL97PFGvmdZ/Nz3I+ULAqPzOTZTCf3GcOnS1ShL6Rj'
    'y/Aa9+821SWXouL3wF8nuwzU61aJMkPPWHhRNm9v3VphoyGnaxXRqz66jOrofzVx5CylXIpAuDbyCnhAiCgwbWfw0lgOQ6QtXUe6oy2QhwDSYXHQLuqYDOk+'
    'lKy8A0SrPjedEF9mF+BGmFA1FPNVx+ShJk0VdQWQRp9NHnKRmo4ylvQiAb4msZsr3I1umMZTu7OrdXYRcyY702zVNPE1p99OCc6C86eJ8puNxufiyf1ttLiL'
    '9gIbBpz7xGUvllkcUDJpTFVSzVncJ5yRHvAq0NOk1l2UnYDq1dLhVOJXNSjkIc6vqy+4GxpvediSZ9ciP9zHS6IhETP0IS+kAFCZjHVcKfuvtnEX/zOMoRSR'
    'JZaKOiYjjQcSs9TwAgrCtaOuWvki/zvNj8esv5WNNtZai4qBplAXLU3viZct8JRzDyJB0mBCRAcJdWbMR+2mtB5QWlSbpo6QbiLd1c0VnEe5UflTHawQLLts'
    'o9yyNcxNFmAUoAWVWnvQHAPtLmsJlzLtGOHdWoSZPnxrAe5EhmvmE51LlMcKLI9zClV8poC7lAuL9WRYE/rZfEY3nccfKGlnJgft2tkPJGkXzCaL2YDo4bbI'
    'fXjITKhVIo8CcNzoSo3+KumixVsp4T5YRgJij53/U4tTiJSlk12lIn+F9r5tyWglCNABmRdI8rqlHL8oLD9tox+gBh8mFAobxdm2lfd7dag0T/UZo13+ELIc'
    'Jhv2gBS7QBHFW+okpsSrZAspSTBtN8yP6Yw8TzhndlcbQ6EIls2vYQ/sNodwyGktCppMQ8igR0rDizPA+Wb9SU5+vS1k5ODbvG1xRldT5V2Tb1BCiPVg05wK'
    'afiUXNUB1G24RiFx1TAGqwhQ2aWifBl7QNrxIQ/oZpHIsc+3BlExajbS+mabkMi7ATeMoGMBbFgloVWZ4Sb9jXnoKzR+nH0v5kmbDlmm57WI9+5uYx52gBXr'
    'Ws4S5AsLxLyu8RidixgH46rkJ2xZm43WF5QHKzfD+7EetbO0UDrjh93B7det+clDoZwAgzT3vWIbAQUwshuNF7xjPH9y9+Hz8hvd8ef6y61f/nz56wuueBio'
    '/2ZCpm3oXZ3GEA4jcHJdYRWgTnqIn+6R3rTDY9FgzfAP7oJmXYkmVopugB+yU5GP22EYjfIAVbmZQE7lIFIDCHBsVAln2pABXH7MDmXzmD9rlYmGSaVeNGlT'
    'rMQtN/qCPjreZ7/CiGDZCpAl32UxFHbAfkeTMmwbqbAtm8KresBd49V6w2H3W5AU2XJZ4vXvJw6NpMk+cfO7Z09/IPihYBYFop0m+E1a05X//NvvL/+g3dCw'
    'g4jD9qIXrRhzpcbRhu4+0/BDvMqmWAZbs3JlqzZRlo4UXSOzhsNwMZBNM5Q7QVmdzdQ87JcK+XiQGbNNHHCBRwybWU3LtGUAs47PZrhI9QGznv746Nkzhi7u'
    'Ze/Tjy8vmqu7ysrQkwzkA3uLMcY7zheF557soY+gDLOz7im8OaLbGKuw8IDXDMtAoy2WN7ujT2xnNbCNaRug+XBBfSSmNxMujPddlrkeMJqSZCN6WR50Cdpq'
    'Jd8QPZriMJnjaSnJiKcaPcwoIk1s8Khkx/9QSwMEFAAACAgAAAAhXA1xTZ/1CgAAohcAABwAAABMT0cwNV/qs7XthrVf7J6R7ISx7JWI64K0Lm1kfVjLbhtH'
    'Ft3zKwowsslIsp5+LA1MoDEwwgyyCZDBwEhmlESIIwWSjCCAMWhKLYEm6ZiKSYuUmkzTlizRQ8MtsklRMLXxpyS7rup/mHNvVTeLsj0LPdjsqrqPc889t66J'
    'v/5tcXpBRN1evNsTqrGn3DOhKjm5FWYyn34q26GqD4TshtHgkWocywtXxLm2auwI5btCbWfjbe/duTrbUQ0Xf4/21ImjmkF0FgrVCeNsW75sq/xRnB9gs3xZ'
    'yPZAuVgh+65y63G1gq1kIScLR1MiChxsiI09cVPlqmq3if1k0RHaCm2EDKqqXqWXYLOIBoFs1lUdptQvo64v7txVNefq0e/Oo/NhdBYInCqPqzj91Slvsv86'
    'rvgCR8S1Mp7igK4PL+DCxSnMlMdDE5K40kqsVIUjslAHg96Iem35pBZXqliH0OVVvpe8K/sONseO/DLM6LWjgOIE7y6qdFKWQlfHusOS7A7IqKhXUl5THbTS'
    'eLg40IpTJnPtmlAnbvzskVA5D3ZmMjMUvay4v7yxsbbKrjVK8kWRHdCB+WZt/YcNJLKnttp00n5JxLUKwhpX4MxWG25NCMRFHoeUJzyJK68nKJ77pahTFPKk'
    'GQ0G2G1I0ZanQ+XDaj9re1voya5jfS3DI+UNZSsUvztV/S1iHspGUcSHrgnt706NlsR7nip4ozhnZqdEXAll4ULINwO1k7iEoNVDwh5Sr45wRsfFJzw9LCEF'
    'ZHrUCaLuEE/6ZcQN7+WQGIoE+0K4lQU4V6WoJMk8KKvOayzp/Nf2ohuqZmiZNDdFe1B0XwyRUjHHMIc521mG4GEpsZFzOUBZuWorwL44Jp+gFcYR5FAlFqrk'
    'SYPLYytE5uUTbOb5jCfs7vmcRz9LFscVF9Zy0gAi4IIOBoz8PeOFZe/8FPtQc22nYErBS3PPUdDW4iihztx4u64B6le4JFAwjPJOgCd2HfQ89XxHxMUiMME+'
    'oTKrgKvnAtu0RG6dRhfZxJiFKWEyp/PJWDep0rnEcW8G2rDE0noxedvKXGrC0sw0njNc0/K5FQVcnGwRyk+zA8gN62zreQFqy4e/Ymn+3fnSDfzc4tD7LgiE'
    'i6IOHglxBpEBHZOrouqsCGe++O7nP5zyKN3xweN353fuchQarknfH05lQvxl7Se8aQCXUE7CSwwHDjeypXmBEsqw4dVffPfVJi3XQU/iZs6NiyVZOMVHOwNY'
    'RkmInzwCj1ie37mLz8BbUT4F/msuEb5mPjwjXlV9jyoJVYszqILwEkBNq1QFGajZPDSCtDZCM1gbRJvJTMIBXz4PEFMK3uIMIYxIDaUTBRUDvsUb1mNVubRM'
    '5d5EO2JjxJaiKV+6RJf4Wh7D8bIrm57y6yB54n1+HMo3SLsnGM8pLpK1Luf1wMYCyGlxztQxs5EDW9+dL87iZ26Cv56nr2EwvmZXgJfFBfzcIH7DWfLVzjh7'
    'TQq5+yvtZDKcRoLbkw9KdVDKrajTStx5mfbBz2c0g5flb4/ErdlPhI6e2UrttxmkBsDVHA7m9/ueZOjzstvTnxiGZJf7LdU8My1Kvnr7sXaWLgl1qos6iAxD'
    'zqzpSPrQUUeaNC6BxC2LEDh8oG2pcv1jYBHVewo8VST17sCR+WMTcG02EGgCqmlAlj7QVNVuEUyX7KkZl8G5v2u6DZifHug9TCuzdQZYedCMgl+JmD30Tfxl'
    'yYIKMrjSJNQosdMc9qTT992xLCdi6MUQ+Gc6hiH5C6xOyNMwOfcj2yomawGFI6jatqqWgeluTGLdPe707KdZxpjpulRYBD6bZykToCmc/vQMconJKGF/h2KB'
    '7stwYSOZZ0E+Bf9D+oXy2nChvthM1alCUVlFaGkWRHT38QRSWsJf0FrnkiuQPZ9gUoH5tLiSY92IQNEXgEF0Ti8cAj0tER8UYbrxcSI9+/wUKecWyPuNV66W'
    'nKYuSb8Qq7GlhopA5WidQnpvx5zl18FXz8L3iA306qiXdVnyUsAzPCgbOoBGOZoWlnTh/QAfxtBBbePVDiGhp4JL0p070r8wjUx32YLPioZTzDzLfQbrKLFj'
    'tKA34YJs0BNbv3D1siMcWZIHUWfb1JJ6GnxAmp5kIZG5FWTz0SA0Wn7U6xJb2Ub2r4GsV3lDNjSVbB8Bjh4WisUopMJqh6hW9DyKJboT0SCaMraGbgAZh9xJ'
    'Orp+3CPqpB8S1EYRqt+yBPMUg8gq2wXPvWoUOlpR5BBabEkYIm7a9VlEXMlSEkKS3b6bdjHHPGClXuOZgLaGWTAa28hyMEqNVuTNACVJGIECQWn96c7dTCYp'
    'EUfM/+cGImreMpTKgNLH6zaNKhGz1EvqWoX4DAPLWL017RdXakTKjRKnkKcpkY5hiTTTvrjCtFrd+k0sGiVEx8/Jp15MI8rYSMSVctK0Ak8sYt7ABID8ebxx'
    'gj/k9BD7jbzQb30Ufck6Q0chkTeBDkQtX+mWCNnHUpX7uj5ayx5WNZXT6G3VqA9WK1pEkWxDfAwPA63PA5CBcfm9TpDkBxHbzonFlU0R/4Lx9IwWz97k3uBa'
    'tLx0k3e9CBEzSqKOLxLW19JkK5wQS7dHU27S52en8QaIeF7MxL9USWdQO5ijZejYqCLmNi35El0+Ggz0VnTWE8BH20RmjPVtc5L+mrvGGHUkYkGjLqAGDx90'
    'R9eu1DzMsGY9CxPaw8xdaOgYcMzJFOAhhPW41tLKl3CBnVtBat2gySOYzttHwKCrR1toxhWdqExmaYFBkR9Q71jQKpAFFAOd1Ma8uuAmtkCyX54cxZW9RJKd'
    'NNS5a2EO6TPDwohd0etH3ImOiQ+EJiMgL5IomWG/m9x48HTTd2GVriz0qkIdX5tyNGA2SKH/7NZvJTehM32QsU4LOJYnuj+mimdM6jwLTfs0QkqQMOzzoEvn'
    '0KzmC26nePl9Bh35hmIp1M3AQkeMximmSz6JOnglZ/NVOqZdFUVi6RbXzzS6OY/EmDaTaxojK3OyPsTwTGoVeoBsONkjBKES9AgSdXuQfkkaUJswDS8gJuM0'
    'oLbrfDnzgQ5koyrp4shdrZXJPDQ0Ix4mF1567BYP8dXk5KQwv/HJrNT3OOKhdUX1f2+oJugs/Gu9Pxf1Alpy4pBRRd7vH+ldj35PJ/ifOJmN39+RjwPSBnmg'
    '9LBIN1r9LOeAh7RMhuejez+trX//9dra91MrP/68+jXjDL0BkTloWeSl5aLY2Hzw7+XVzesba/cfbK6srZr5nYesPpi7rVnYZeBoySlzrqxfpsmrjYQmX8Gp'
    'y7JBeXK3R+MvfGwP9MUH8cCO4SaS2my8+PLu33WXc0zN0QZ04QIif+KSbtNMkvY5PqjcIrzFVVfPpt772m3ppsG22cbcqkj/SAaHeJW8YkEDhwdFAFEdBGQE'
    '6x5C99iADz7HmyPa5t2Ys1mzoqeaiWkUaB1QoyGtihljar6vImjqjkk3gyXoFkuVvzdrMCQQk07OXL0Y7a1bFusONiWTsafitKGl/YzZkVRMfWirIL6awX5E'
    'ByT56F2zmGsfWyDsAEg6Il5BA3upmSptQ7JAZEMRN345Gob2rMMoQu9lq9v4rWd5FQwQXnONmS9vrD1Y/9fyvbX1lW9XVr+6v3H925XNe+vL3yyvL6/i+ez0'
    '7I2Z6enblMZGzhYaCNU2ZqPTzz+78+elz7Cx9mLlhx/X1jeJOZ+e0QXG1Szq9OhCuK5viWB9MoBUaTDlW6LRJVHdmZmmbno8vHpLlF4beenNF5FfcjOjFY9L'
    's8te6WqLyPwPUEsBAhQDFAAACAgAAAAhXMCF1ZJJBwAAww4AABsAAAAAAAAAAAAAAKQBAAAAAExPRzA1X0NvbGFiX+yLpO2WieyViOuCtC5tZFBLAQIUAxQA'
    'AAAIAAAAIVz+7/R1CAkAAHMbAAASAAAAAAAAAAAAAACkAYIHAABMT0cwNV9NMDlfZm9ybXMubWRQSwECFAMUAAAACAAAACFcnYccc38JAAAjHQAAEwAAAAAA'
    'AAAAAAAApAG6EAAATE9HMDVfTTA5X2xlc3Nvbi5tZFBLAQIUAxQAAAgIAAAAIVwFXnF7IAYAAMkMAAAkAAAAAAAAAAAAAACkAWoaAABMT0cwNV/qs7XthrVf'
    'QUnsl63tlaDsi6TsirXslYjrgrQubWRQSwECFAMUAAAICAAAACFcDXFNn/UKAACiFwAAHAAAAAAAAAAAAAAApAHMIAAATE9HMDVf6rO17Ya1X+yekeyEseyV'
    'iOuCtC5tZFBLBQYAAAAABQAFAGYBAAD7KwAAAAA='
)
_EMBEDDED_ZIP_BYTES = _embedded_b64.b64decode(_EMBEDDED_ZIP_B64)
_zip_path = Path.cwd() / 'LOG05_실습자료.zip'
_zip_path.write_bytes(_EMBEDDED_ZIP_BYTES)
_log05_extract_zip(_zip_path)
```

- 사실/제안/미확인과실제/모의수행을구분합니다.
- 보완 방법 근거없는원인·규정·완료추정은수정하거나확인필요로보류합니다.

**평가 항목 3:** [진행 추적 및 완료 기준] 실제 조원·두 평가자·2분 판단·20분 문서점검·3분 발표·실제 교관 확인을 AI 모의와 구분한다.

- 미실시를 PASS로 쓰지 않는다.
- 파일: briefing/log05_commander_1page_brief.md, LOG05_M09_forms.md, forms/LOG05_M09_forms.md 실제AI요청·답변·근거대조3건이상·보완·재확인·본인최종선택과작성/실행기록이연결됩니다.

- 보완 방법 미작성·미실행·검토누락을보완하고교관검토상태를별도로표시합니다.

**교관 검토 기록:** 판정 [미확인 / 보완 필요 / PASS / FAIL]을 실제 기준에 따라 기입해야 하며, 자동 표시는 하지 않습니다.

```python
MISSION_JSON = '{"code": "LOG05_m09", "steps": [{"title": "전후 기준과 출처를 고정한다"}, {"title": "직접 재검증하고 시간을 잰다"}, {"title": "1페이지·발표·질의를 확인한다"}, {"title": "개선 결과와 한계를 넘긴다"}]}'
# LOG05 학습 기록 도구 (표준 Python/Jupyter 환경용)
import json, zipfile, io, os, datetime
from pathlib import Path
from IPython.display import display, Markdown
def _load_notebook_widgets():
    from importlib import import_module, invalidate_caches
    import subprocess
    import sys

    try:
        return import_module('ipywidgets')
    except ImportError:
        print('이 노트북 커널에 ipywidgets가 없어 설치합니다.')

    try:
        subprocess.run(
            [sys.executable, '-m', 'pip', 'install', '--disable-pip-version-check',
             '--retries', '0', '--timeout', '10', 'ipywidgets'],
            check=True, stdout=subprocess.PIPE, stderr=subprocess.STDOUT, timeout=60,
        )
        invalidate_caches()
        return import_module('ipywidgets')
    except (subprocess.SubprocessError, OSError, ImportError):
        print('이 커널에서 ipywidgets를 설치하거나 불러오지 못했습니다.')
        print('설치 가능한 환경에서는 새 셀에서 %pip install ipywidgets 를 실행한 뒤 이 셀을 다시 실행하세요.')
        print('설치할 수 없는 환경에서는 제공된 Markdown 작성 양식을 사용하세요.')
        return None

widgets = _load_notebook_widgets()
if widgets is not None:
    MISSION_DATA = json.loads(MISSION_JSON)
    CODE = MISSION_DATA['code']
    OUT_DIR = Path(CODE + '_results')
    OUT_DIR.mkdir(exist_ok=True)
    BACKUP = OUT_DIR / (CODE + '_record.json')
    fields = {}
    for no, item in enumerate(MISSION_DATA['steps'], 1):
        fields['activity_%02d' % no] = widgets.Textarea(
            value='', placeholder='내 판단, 확인한 자료 번호, AI 제안과 수정 이유를 직접 작성하세요.',
            description='활동 %s' % no, layout=widgets.Layout(width='98%',height='130px'),
            style={'description_width':'75px'})
    for key, label in [('first_thought','AI에게 묻기 전 내 판단'), ('ai_request','실제로 보낸 AI 요청 전체'),
                       ('ai_response','실제로 받은 AI 답변 전체'), ('source_check','원자료 대조 3건과 수정 이유'),
                       ('final_result','최종 작성 결과와 근거'), ('limits','미실시/미측정/확인 필요 및 교관 검토 상태')]:
        fields[key] = widgets.Textarea(value='', description=label,
           layout=widgets.Layout(width='98%',height='125px'),style={'description_width':'190px'})
    status = widgets.HTML(value='<b>아직 저장하지 않았습니다.</b> 작성과 교관 승인은 별개입니다.')
    restore_input = widgets.Textarea(description='복원 JSON',placeholder='백업 JSON의 내용을 붙여넣으세요.',layout=widgets.Layout(width='98%',height='100px'),style={'description_width':'110px'})

    def payload():
        return {'schema':'log05-student-v1','mission':CODE,'saved_at':datetime.datetime.now(datetime.timezone.utc).isoformat(),
                'answers':{key:w.value for key,w in fields.items()},
                'status':'작성 기록 (학습 PASS 또는 교관 승인 아님)'}

    def save_record(_=None):
        obj=payload()
        if BACKUP.exists():
            stamp=datetime.datetime.now().strftime('%Y%m%d_%H%M%S_%f')
            BACKUP.rename(OUT_DIR / (CODE+'_previous_'+stamp+'.json'))
        BACKUP.write_text(json.dumps(obj,ensure_ascii=False,indent=2),encoding='utf8')
        status.value='<b>로컬 저장 완료:</b> '+str(BACKUP)+' (PASS 아님)'

    def export_json(_=None):
        save_record()
        raw=BACKUP.read_bytes()
        try:
            from google.colab import files
            files.download(str(BACKUP))
        except ImportError:
            print('Colab이 아닙니다. 저장 위치:',BACKUP.resolve())

    def restore_record(_=None):
        try:
            obj=json.loads(restore_input.value)
            if obj.get('mission')!=CODE or not isinstance(obj.get('answers'),dict):raise ValueError('다른 미션 또는 형식이 맞지 않는 백업입니다.')
            save_record()  # 복원 전 백업 저장
            for key,w in fields.items():w.value=str(obj['answers'].get(key,''))
            status.value='<b>복원 적용:</b> 기존 작성 내용은 이전 JSON으로 보관했습니다. 다시 저장하세요.'
        except Exception as exc:status.value='<b>복원 오류:</b> '+str(exc)

    def export_review_zip(_=None):
        save_record()
        target=OUT_DIR / (CODE+'_review.zip')
        with zipfile.ZipFile(target,'w',compression=zipfile.ZIP_DEFLATED) as zz:
            zz.write(BACKUP,arcname=BACKUP.name)
            text='\n\n'.join('## '+key+'\n'+val for key,val in payload()['answers'].items())
            zz.writestr(CODE+'_work.md',text)
            zz.writestr('README.txt','검토용 보관 파일. 실제 평가 PASS 또는 교관 승인 자료가 아닙니다.')
        try:
            from google.colab import files
            files.download(str(target))
        except ImportError:
            print('Colab이 아닙니다. ZIP 위치:',target.resolve())

    for i,item in enumerate(MISSION_DATA['steps'],1):
        display(Markdown('### 활동 %d. %s' % (i,item['title'])))
        display(fields['activity_%02d' % i])
    for key,label in [('first_thought','최초 판단'),('ai_request','AI 요청 원문'),('ai_response','AI 답변 원문'),
     ('source_check','원자료 대조'),('final_result','최종 결과'),('limits','한계 및 실제 검토 상태')]:
        display(Markdown('### '+label));display(fields[key])
    buttons=[]
    for title,handler in [('작성 내용 저장',save_record),('작성 JSON 내려받기',export_json),('검토용 ZIP 내려받기',export_review_zip),('작성 JSON 복원',restore_record)]:
        b=widgets.Button(description=title,button_style='info');b.on_click(handler);buttons.append(b)
    display(widgets.HBox(buttons[:2]));display(widgets.HBox(buttons[2:]));display(restore_input);display(status)
```
