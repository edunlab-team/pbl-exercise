```python
# @title 📦 LOG05 실습자료 자동 준비 (ZIP 업로드 불필요) { display-mode: "form" }
# 학습자 화면에는 결과와 업로드 인터페이스만 표시하도록 구성합니다.
from pathlib import Path
import os, zipfile, stat
from IPython.display import display, Markdown

LOG05_PROJECT_DIR = Path.cwd() / "LOG05_실습자료"
LOG05_PROJECT_READY = False

def _log05_extract_zip(zip_path):
    global LOG05_PROJECT_READY
    zip_path = Path(zip_path)
    if not zipfile.is_zipfile(zip_path):
        raise ValueError("유효한 ZIP 압축파일이 아닙니다. 실습자료 ZIP을 다시 선택해 주세요.")
    dst = LOG05_PROJECT_DIR.resolve()
    dst.mkdir(parents=True, exist_ok=True)
    with zipfile.ZipFile(zip_path) as archive:
        entries = archive.infolist()
        if len(entries) > 20000 or sum(info.file_size for info in entries) > 1024**3:
            raise ValueError("압축파일의 항목 수 또는 해제 후 용량이 허용 범위를 초과합니다.")
        for info in entries:
            item = Path(info.filename.replace('\\', '/'))
            if item.is_absolute() or '..' in item.parts or not info.filename.strip():
                raise ValueError("안전하지 않은 ZIP 내부 경로가 포함되어 있습니다: " + info.filename)
            target = (dst / item).resolve()
            if not target.is_relative_to(dst):
                raise ValueError("실습 폴더 밖으로 벗어나는 파일 경로입니다.")
            mode = (info.external_attr >> 16) & 0o170000
            if mode == stat.S_IFLNK:
                raise ValueError("심볼릭 링크가 포함된 ZIP은 허용하지 않습니다.")
        for info in entries:
            target = dst / Path(info.filename.replace('\\', '/'))
            if info.is_dir():
                target.mkdir(parents=True, exist_ok=True)
            else:
                target.parent.mkdir(parents=True, exist_ok=True)
                with archive.open(info) as src, target.open('wb') as out:
                    import shutil
                    shutil.copyfileobj(src, out)
    files = [p for p in dst.rglob('*') if p.is_file()]
    if not files:
        raise ValueError("ZIP에 압축 해제할 실습자료 파일이 없습니다.")
    LOG05_PROJECT_READY = True
    print("✅ LOG05 실습자료 압축 해제 완료:", dst)
    print("📄 확인된 파일 수:", len(files))
    print("ℹ️ 실습자료 폴더:", dst)
    print("※ 업로드·압축 해제는 학습 수행 및 평가 완료를 뜻하지 않습니다.")
    return dst

# 실습자료가 노트북 안에 내장되어 있어 ZIP 업로드 없이 자동으로 준비됩니다.
# 내장 파일 5개: LOG05_Colab_실행안내.md, LOG05_M08_forms.md, LOG05_M08_lesson.md, LOG05_공통_AI역할실습안내.md, LOG05_공통_작성안내.md
import base64 as _embedded_b64
_EMBEDDED_ZIP_B64 = (
    'UEsDBBQAAAgIAAAAIVzAhdWSSQcAAMMOAAAbAAAATE9HMDVfQ29sYWJf7Iuk7ZaJ7JWI64K0Lm1kbVddbxNXEH3Pr7hSxSMmDtDAI2orQCJq1RaQ+hIZSEtK'
    'qFECRfShWiebaBMbYkOcbJK12RTnq3XbTbyOHcm85Kf0ce/d/9Azc+/aC1gCFtl3556ZOefM+DNx6+vro5eF2i7FhaY8KQhVbMTry0JVHTkfjoxkM/rE5PP8'
    '7KN7+fyjzPSTF7/ck7s98UV+JndPbZSV7Qm1EUYtX6iGJU9toRYtfG5eREC10o7XbLnjKb8Wr3Qzv00/UTUcc5ty770sOrLYyAjl2+o4VL6n6hXZ6oofbn4T'
    'tXoiagOWHVcP++f6dwhVb2q4qmaJqBuonRCfVZR9JOLVUL5cI5jy5T9qPVRvArWPN6orAGNCjYxlRNRqx0ttce0mUvg7rvomcbVpiVtTc3P5Xxho7T1lB1Sq'
    '1hVqvgmsIi6VZfHgrKNqrmx2zzrXblLOcn8n6uJMrUcvyoOe8nGrXxjcijC2XLX5yuB9Uj8Uycdja00d/33WkcW2bFl89bserpbBdirAtW9uini+IVfWRFy1'
    '8Q6wL6W+j9rlKLQEgiDSv11Zd4U6DeUbD2DNPfVd1FDOu0KeWLJECKuyFVJEnYaqL0p/N666nxbtYkZ8+9V3t299L5C2qpfRepRAqPWKKtYY8hEwBVSwudyv'
    'U5OzU3PPZp7ykXc9ZfORuEqNxr+Gbun2bnaBScSvumq/gp6CNUKubqKggrhx4ono+D1RCQEJgO1Rk3XXEZAqK8v9aHEV51utpMrmlMtc1aWIOr3oKOBCxZtV'
    '9Bf/NX1ue+qPRaFsP15g1HL+IDotJHW4lNH55aYnH0znZvI/PZsirEM7CX2VCSyCREEBf0CCEGxN5x06oB5dDvLZeP55QCnGWyVV9NG4fScJ8LEg7j58ceFG'
    '/vmFuw9zXOaoZcdbL8860bEVL/n0RDN61Nsf87OP53R7dSH6MVA2ioszYEO8CUbt28qzSQaa59w/t0WFizpduWdrXXTVXm0oSy5nOGfoWLmOKnK7SKENi0XN'
    'RRfxawd5kXihiSjwiHUG0LdZ+vh61hxVG03qrKqvybfLV8bORYFlAl8dPWcCM6KTQ7VzNAyRkEuv+aY3R1F3GQX1y8pxSbamxvz6fFPWS5yraf7KKSXp9bi1'
    'QzxEqKUSK7Belu9KDAtaxPlDcHsTomqj45Zc2VW1kPSmGhWOsWnjODGXSOdXh9bw84yYGIeMG2orxP0QMLkeIdHOOjE6PqkcD8yO3apWuHHbzMOnj2c0ZUPp'
    'N8g+qlS9MucJ4XRLEg61FcCd+g7ed+LYteGsxjPkfkl7QnRcIo+WxWaatXyGXBTtg25jOxgDOwVnGxrH5Hoe/9V3TN0uPLm1Wr7abkLxn+UOjgqCV9/9z9rk'
    'XBxb7vUJghvibTtlppq4JhujHvkn4p6edQYdNqWQFYjglPX0gSlQsWulJHUg05kNMh4Z56YMMBKnNEzBeD6eNkSFQGhV01kcUksvBTtbjSdG2lV08ULHCBZJ'
    'Ks9nVtUdxGUNGFkVvSiwh1CfDHLBMDaueDg3XA5UAR4D4vaX57Nnndt3zo9xpxYK8QLnDixyr5lUpmThG+iGJYgn3KGOfBPb5BEMZygxQWVlp3+RaUWrTeOO'
    'gKZnEN/ouNrj/Cpw0uu6OkNxG8aR7FYaqB97VnrIDarFn+81aXx/2A2qyYbLTak6ifeRwDHZtw6HavEKEoFVAd7+coKB+n3j+4lbnN6Oh+w/FKs8tuOVnvEV'
    'Djzw274uyZL4Xar4k9n8z1P32cVplhVPKQW1iNT2amyYvNWwgzPlAJWNxpgVz0f6Nl5F2w5Sc/BxbvbRZO7+02e5mcmn04+nmLW0QZltj8hUIKn3Ky//pU4w'
    'H40YMU8Ag96kpA3qC6ab8YJFmj/o0sgwAxSRMbkTqg8pw8hViOkKsh0bhUIFDXzMroqXiPjESfYbdk0KTY8NdHAJVtYQRCDvIDVWCeN8kDaobRz0EH+dODvf'
    'jI5Cika+TDPlBDakFhwQxo5oBL1mPmp6DOYmNcH3YAWUIqmZGcNEw2ZgD583E+OIdVm5BSyWvBNh4NTSm4BZDiEuPQB4KYErn7omJy2sVDLkcqwmroQp/Ooy'
    'gMDax2B2w/aDkewowFylPO5kf78zzhkaD4G9+DbaCwA0lpPFp97Uvdb7Yndn6Nqhly+QDpPBGhvFl/31UsIxK95F/RENePci6j40JX2UANGaS7PmksjGr1yC'
    'pr1Ab4LM9MD8SEiW/GTntAZv0J6oSZPM4pTlpRF/4kTJ533nxBgcagXZrN4ric0YQoWQ28pbopi4dNaZ+Bx/r3C2LEMoOTuKIUK9Y9d13I9WXs4GxedtRBOE'
    'CldQfkUkW5zxHK06rIe6bgNtJQulNuuBVS0mA/7Zk5l87sHcBd4d+j/BDH/TayhiQliOqS7xoo+KZzDclBIODmAKukDUN/rRtHXIU6jYgECAWJNME9SMJwx9'
    '/ePELFZ69cPx1Oon3wbkJnXbgOF1FH3nWZnY5qAd/wNQSwMEFAAAAAgAAAAhXGZ3SNRyCQAAsRsAABIAAABMT0cwNV9NMDhfZm9ybXMubWTtWF1PHMkVfZ9f'
    'UZIVCSvA8GGihDfykMTSWol2H/YhithVTLKWlmUDrKJI8zCDGzT2jEWzZpge3EN6smA+NM420MCgHb/wU3bfuqr/Q869Vf0xM4BRlPjJWmuZrq6uqnvuuefe'
    'W/fER7//7diUeDT2S6F2NpR1LNTWhqo0czl15qp/rQq14qtzV7a70+KzxbmvFxaXl/JfLvx1bGp2ce7x8tzn87N/Xnyy/ORv38zN6tej848/y+VGzHJqx766'
    'kP4hlvypuKlavjy1RF6EvquanZ/PPPypWJtOJze7mHxiKc/iwcqu8tz8zEMhjw5U0xHR+rPQL2JJ1SiKsOOr3aKIp+e+nFtaWvgKK9nyu6rc6wrVfBueembp'
    'qHYoK2VZ2R0V6pUtTzvY6DRQDUtEGy6tHfoN1cRyK4HaPqRf0doZPopqjqq3RVRzsVNZ7leT9QK5bglpJ8vK7zs4bXh6hpXxs9NSXu3qguZVLoWqr6lmVTUt'
    'nCpQrkcGRLWD8AcHOx6ElyWzSi537949DDmjiTssF9iJcTEiwlMr2n4RnnYNMkL5gCUYJouAppDnltxzhtRKW+5Ury6UV8YE+luDqThKwwlP3g4b7IW8tOSK'
    'S+ZHjgU0y2q/eD+XK5CdkdMRBbMffpjtCr0bpc/K2pVHq7QxZgq1CwK179MC2MnyVK0sClh3ZGREvOP/mDWO373/MDhx3eDkdYMPrhucGhzM3Yj0BJCWXlm+'
    'dKNqUah/lpRHFK4WVXNjaCo8DtiFT0t9WDEbaQlwSV0cqNpb4gyoqZFLVwTuUdUm8PS2POY70q8J6djy+aZQZQc+Y9gEPEYklu1OeNZ+N4yDAA6iNwjdIG6D'
    'oGHkF3fGcJIwhE1s2mG0tSeUa0V1IOPZ4UmRkDMDBUMegoghvt5G+mDHAiFD/1uKImmVZTsQcu1bik7QO+w8EyCwqvuZA1/7ycq/1bmPV3twhFBrNlE2/QSx'
    'gFeqbsMSZvI+ONDy4XZeox2oigtNIDeFQTHaenYHNB4ADe1TAkU5JRNZPQTSE6Ia+5o+hEIAI3XpJC+NONHz8Spsg8El/BoiTQQfNQuJde+PJjcbPUVGZ/GD'
    'QHZD3ybVVDtteB0gDCUeEpPEd7gkPLM5sjT91dYPHBGeDpinmYdjK3rahLC25Xkx88zyGkeSlnjCSDnl6ZQPPG13I2ps3r8TWHSYQcBodBA0Gu0HjmESajsg'
    '55u8Y1LeZSeXmxiLtqukxEwrv0RrsI7sb+C80AAfok9HBpiYCVWnLEriW0FaeuHjVWT5oAE+tGSlkaSk83JUs9T2JksRO0O5B5xAeuhP7C47hD5eU+7DGegY'
    'E2NMeqj90YFsuURblANEfxuzN2wUB22ODIdWl6dnxh18EmS8LTj9BTYPOJgghScB5ZKoesD7BLLZ5Qz90mea1J6r52dJlgaMSB7wKR/ozFZui8iPzWjXE58C'
    'wqhu3RZy05d7HUM1YoSOIsMo2uvQT9LmNocRFTheidUh6ANOPS2nZEEGjdliXMEsPN9kM2EO65olkqyvdlZ707k+Q63MAR8/gBna71jaOAc7aYdoTzDlqZrh'
    '2qfmEhcKWSgYHgMCvk1sR36RFf6c7NBcSk5HTotqb+7E+wFyv2uAaI56jXRo+8XVBbQ+WvOSUgvDstVMi0OgBxfY0+LHrRZ9Jitn8rSo5dxnTOmFGVX1AGWN'
    'iF7FoYNlZh7C9+FJVcj9VtiB++urJNEknawDXEVyUWaQxJGyZSNAe2XHcs6upQlUCZEQw8XAa5rVCKUNjhM1aqh3OJq41iFqOGBa4ikRXnTDYz/J5EwSCp1M'
    'HOqJffDn+iAd+JvjBEkclEeH4K0opHTeVCdvuKxto6yNquBLN5M79KwYxRtmoSKGZUyUw02E+230yN1Cgr6nHFUwxAGynDwcZzBK9i1OViRjsW9YXwAqggqZ'
    'VdMFkzWfWbe0vFP54OmA4czIDNMlqcmwsWH/F5LDEPVdFxoqJtN6kKQytjVtCjTxNfiU+kyboWlCxMA0UvGYWQ7plNzf4eyV9CLaWmM99zZeKdVK5oZJKswF'
    'pjXRMuNe3esgeFIX9O7sxtwtiLj1Y33R8udSFGtoDXX+W4KQRJiMhjjViEGoStFTlyONuzPT6iFs3oDuSdQmbI6DVecmvI+P/oeZTz7J/2bm4Ud5Ejy2MA/6'
    'cPV0XqYlYzBidTJ7/48If6vlLEbopg2E2V6XnNGsskqvOsrtMjUsh0QtsfX7jvaVUXDdN9sEDVfV0yyIca+sjUvbaMobqFgR+XETHU/RIhujpZ+ybOIR9fwS'
    'O1MLr1czJ9G4o4nXFY5Zm/S3YbEcst/Ycu1Xk7O0naNJUQA1RQ1TCnThbZogaEAaSljij/rjP5kaIu4+wwsXDQThM1BfCvFjEXG0btEVhu2a3ovUU/d1ItPP'
    'cRUBha0FSTvmryeVN+SK8qglxsfGfkaj8vXbaKuMk8LgYdFbx+p8a05D4aqTPi0d52wd/yaS6+k6fZ3j0MzS0jfzXy8/Wfhq6T6dsKcf5KWPVqkliY9COH36'
    'xT/yv1v4e/7TLz5fnjbKbloO1lZOB5xLksDRcpyVszhp44usGhAEsWJkXNMXlgkN9V0IE4tyTiYO6exkwrllRE/zMxEEujAJDO2kDf3fvalU1MxruHK9wWXx'
    'vgXicHSr182k5iiYKZRODHf7NYWhgL1J5mRhSKqX2wWC2gO6RdO7jDPxejHHX13Pm4I6A4ALO6bFzMzHYubq4tdEH01Mvt+hcOLuiMZNyZhtDJhF6AS9jaRA'
    'dzn5xGrxuklco5gyBUO2ItINKBnJNNGlVV7tl7lMgCOp4ah4+gak5efN5QRXpPpqi+FKa15jdeaqbIvG5aGu/ZHXtTZwVbTh6n56sP7nhJ818S8Li/NLeb6u'
    'nAXQs/w8Ov+Y1UZT1GSmu15TDovB1YZv3gfRp+o+DpTJ7wX6z/TOqfcn2PsUBPGtD9cuHNHs9rgkYrcP3BZN9RYVWrSk/4oDP24fuOniih9Uh+aCsXEjkNxE'
    'xi1ieieaJPsPBHm/BJnsIcikzksQgZdenJRI3NOqkO/CEobceqMryP2U7rppLc2ZwZTYcT5Li2WdRDQTfb51ctO2V1MrhRV+ki93Df4gyQfmvF/mPOhhzgNm'
    'TuaaQ+sBHVRXKhoifeGJ8jFgAun0sbMHy/k2u+/mwFzV6ivZpM7Xl4xEK32tiDUf/erq4tH4GJ/Xs2Sl/UFJ3jMf/gNQSwMEFAAAAAgAAAAhXEZXMpn7CQAA'
    '5R4AABMAAABMT0cwNV9NMDhfbGVzc29uLm1k7VldbxvHFX3nrxjAMCAD+uCHlA+9OSiQBqhRoECahyBwgtptDNR1aisoCuiBtFcCLdIxFZHWUlkqy0ayKINu'
    'VxIlUqjcB/2U5m1n9j/03HtnyaVIyXZQJy8GLEs73Jm5c+655947vKR+89sP03PqWvo99d98VV29+jvTcFS0fBjVPKXbHdPoKX3QCXsPlSltReWWMvfbuplX'
    '+sgxTkPp4LEyW6tmJ69MMwj3Okr/6yQMKvqxo8xm2zhb4YGTSl26hPGeWXKVeVCI6ruplC4dm5JvGnkVBnkMKuN775qia5abWE+X83Z32VwHrmm4KuwFutmI'
    '9zZ7S2YT26w/j2q+2VzSpaIubU3HC+IsyjROYJpZrxjnwuPQ4ua7Cj6m49PYZvm0a/yiCVr0u4ZJp92o7ob7L4x3opueCruefrYU1dzwwJ8cxsA0AINTxH4W'
    'DXprL9D3XaVXq/zsKbLLr/B2wX9MLQEXPqVzpFJTZJM+6MGCB0X14a0FFX3TM86eyk3PmZIXBs6P+SrOYXrNMPiWdsU6uuT+mK/R3EPP/GMJhwnMkafbJ/Pq'
    '87s3v7pzd+HezJ/v/Ck9d/3uzRsLN7+4ff0Pd28t3Prr1zevy8fTt298junhwSFoAKtWaUdTK+r7HVPPq6sfWcTtmN4+Ubp1YnycvvHCrBxaN7DTxX7yLZbD'
    'WhX9Q5kWiVd1PFNq4BDRKv7wGJaNjnEL5CRTdxT53S3QebDaJfLs9PBiKqOmRh2urp52P1CfhocV4zXNxq68IDM+S6UWFaNXUItDc9SEPipGNcdsVNkSdimm'
    'YqunRFShypXhWR+oicGLemX4Rew0NTWlhv7HWPr9+XQazx//fiqromIb3J1UeqcQFdoqp9f2mJ6NDlijdzvWDnpYqRLrzHGd5v5qKjOYGwSRE+inbZVNX05M'
    '1QeHYffsVDFglgw4u2eROKXo/cYLNRttlLH0s5ZercT8XkxsJT5TubnLNr6wBLhN2MK9vE8mPZ+hfWjFH1rqHdIHrKRm4yAZxZtDS8J8zGbZ7OVJeyoVlSu6'
    '1FKQCrwfb5fjY0mEQ4NMsMv7BC6CD+HMUbcekEmyYOa92Hq7Kr1Q8ibtZJDNAb1q4BzCdFdvC36ZjDgwch2zuW0XgBYheDkQHdYaIrRLpAP4EJhwv40piIhw'
    'f1cBW4gd2F41+8/D4zKBm52DsCmzXMYotkmdy/nsgPMULhu7cDsWtLiRlg2Ai4UpNk7lELZKHzsIjgsjBG8t9kHh2fhV1d8/pOFRvylyadGdEEswRnEyHBlP'
    '4PzNbWxNkmE2H7D4j84aFzT90CGzMnjOgeeL6l38pPvjWTxl52g8h59Mf5yesvx+Rj6LoQUnpoelSOQEErzxKDw4oRwhnm+BRZM2Q1D60dvuxMVZYhIe8vht'
    'nPe+R3wDW+CiIhC5Qgjr/WLk9mCR7EdwyXaLwxsNnpEekHJoY7zJ4JbaBDPt5PjssfPBOwMkQTH8D4PZcYO5cYOz4wbnRgfPR5pIrP2iXvMiJHvzfQHZ8LSL'
    'aDaN1Yk5VopGB5w8g1UvMFt5WoIydrdlai+Yg74jyA1WBO5QCAJPtuUx6EBQU9qVmES14deE6PAY8jjCthcetl8O4yiAo+iNQjeK2yhoGHnnlTHMEYY4Ex9t'
    'N3qyrYznROtAxofK5DmOZWDRkocgYojPTU7DMmarmOVvzXq7XznFGhobPHbK/X+ao8AGPEStQpQdTBlWzuHiKaGenLc6+ejJw1dAYxZoiE9t3WAja4hA8gKV'
    'L/A1TaRqxHPMsdv/UFGlA1Fd7FeYOwX8NSEaeUVYSKz7+Why/qHn6NAvK8Anxsr/lYHOmyf/5ojwJWAeJB72nOgBau31tj7KJ54Rm54fR5JUa4SRcYvziWRD'
    'r22tRvXqhbo+ou7DYMTaPjp6FrgYJuoCJAQgu3xKfkAd8lw/I8dG9RoJ6hmb6MVB/cgNx/hUQAksIVFYvmDWqerFShNKKTntazYUovbSUIxNHlzgry+f3eas'
    'iCYtQ4CdJ3pkVyZN5eJ4489oK600JJJc9z9bojh9Mjp7iG8c0kw5ezpkKyJR2D2htoiqgTE2pKSDANr7xX73mOwOqPk58si2TVSdR1Uu1VnluGGQmgLQsYig'
    '2mdLsMbOKhUz4X4AvxJM8DeVY0FLU40Dp6Eu1I8CKqOdgBelrqoet5djih/jtZgjQ2InpQ0dHB9Lo1sgM7JpxgPeRmlN1e6Bgx6NxK5C/WGF2MI66NLqmso+'
    'Dj62ZH2Zyij9CJt3WDrhnP2ObWZ5n45unDDGawGLQm1l0JNx0Aj4bFBc/VF5SrvuByR/NscSdarwTs8KC8W/aKZ1Ju21G/QjQypR6jr9QrId7QNHTWyfGAiC'
    'WBusK5id1os4Dmcxhx3faxJl+w0+8yKq1XEuBLKnH9cl7OVvCAjdZEjkYhe2Dr/F6xZ2UI4ofGR77VSKOro46CV+ONApO7Bi0rgtUpP0Ye4iO/irfTd6sYl9'
    'g7JytfLRIFKZLBxWbAlkatmPLRkpapLhLIFEnU/wHYeiOMISm9102qU8dexaN6KkAoK2fRBuS5VPO4sMjrM5xzYTVGu+3ZIVQCKW1IqriL7RFwocNyuUiE7Y'
    'hP72tEYTkHWsRmEeqxOOwEoj4AScr70BheS0AwfonbJe27JkB4tGDjPLh0mwWFCjzC8JX8gi1Ysb9jp8JvE7QrrU4tI0Fq0YVqm7pL7iR1ImqRjopFIjYM1r'
    '7592r2XSBB88qkvtBN4qevyQwCFT6g71drL84O4n1koRwOl+wKMHgD4VOlJCWWVHnRTVduMwmVKfyuTP+pdd0kfI1RUxY6RSUOJ3NL9Hnq54sev9wnnJpZwf'
    '+E9uA20NBd8mk0zR1U9fUKZYoQifHMkQ7qBuYZ/adNE5N1fE65wJl4mr9+59ffurhVt3/nLvykuSFi9BOH3y5d9nfn3nbzOffPnFwjyyS49RGlIQPmm9Cmqy'
    'IZa7KjeAA3hyWsIMe/Nmhdbpx1nCNXZVYfFpl9ThGRTHlXjFhyWIFkNfo4wDQ5Oi5ZLwAiDUxlY+ODWQ5K4c061ppYJ9zksDkljZSD6rawOF8p+tjGIaZqb5'
    'blg+/wnKOq/+z8oaG2ieNogPxHubf5JXJFLuk0fYlVIyz5gdtCY9BpsSfsmXfrMZzNiqiJCoIfqPbUuQvFTmU0Nq7GJS9PD9HGIL/YnEL0ljfJnpjuZfaMnw'
    'Ef945+7tezN8CX8dQF/n5+nbN1gRhEZRuYyqVL3q3e2kGl1t8vx9SJXWAxiUYGc26fTXyV7zoy35z5W93vLizfMil+TFa1UI8y9pgX6BCuEtYd48YWaThPkp'
    'Vdi8etNV2FsavHkapBJfY174LeaMVC/SPfa/y5xmbzc71K77dPEb1Q5NqS23dDIIT1lHrVSN5/P3Zf3t/wdQSwMEFAAACAgAAAAhXAVecXsgBgAAyQwAACQA'
    'AABMT0cwNV/qs7XthrVfQUnsl63tlaDsi6TsirXslYjrgrQubWRtVk1v20YQvetXLJBr6yZpg7S+5RS0aJoC6T05FT00KND2B1A2ZdCmWkuwaFMOKVCJZMmp'
    'gtISZVOonIN/io/c5X/om9nlh5KcCFHcnZk3772ZO+L7p4/vPhDZYpnvLcWjb4U6eZd7kVDuSB0shfIcuZM0GnfuCHlpKzu8ucrmVr6HD057av5OKN9RbtBo'
    '3FqB8mw58M2F8t9UtXyhBh35pq36lsA1Qu3MVGSLvN2R7lSFtpBxL9/1VLCWw0CoxEE4fn1o44ga+VvmAvo773sqTGXHlwc9vskdIZllRwVDdXousjRWI0uo'
    'CHk4N1fqspfFljmkwkTkno2Ucy+gTHCA4mTLGarKPT9bRCaSkHMn91M5XuOqpjpOalmokw7g2eP4g7Fc2ULZF/JgjGhukMVNPANbDWw87UC+bdH7UX68L1Tf'
    'piJwTk7a8mikJsBj4myJx/LoIkv3s8Ua+Z1n83Pcj5QsCo/M5NlMJ/cZw6dLVKEvpGPL8Br37zbVJZei4vfAXye7DNTrVokyQ89YeFE2b2/dWmGjIadrFdGr'
    'PrqM6uh/NXHkLKVcikC4NvIKeECIKDBtZ/DSWA5DpC1dR7qjLZCHANJhcdAu6pgM6T6UrLwDRKs+N50QX2YX4EaYUDUU81XH5KEmTRV1BZBGn00ecpGajjKW'
    '9CIBviaxmyvcjW6YxlO7s6t1dhFzJjvTbNU08TWn304JzoLzp4nym43G5+LJ/W20uIv2AhsGnPvEZS+WWRxQMmlMVVLNWdwnnJEe8CrQ06TWXZSdgOrV0uFU'
    '4lc1KOQhzq+rL7gbGm952JJn1yI/3MdLoiERM/QhL6QAUJmMdVwp+6+2cRf/M4yhFJElloo6JiONBxKz1PACCsK1o65a+SL/O82Px6y/lY021lqLioGmUBct'
    'Te+Jly3wlHMPIkHSYEJEBwl1ZsxH7aa0HlBaVJumjpBuIt3VzRWcR7lR+VMdrBAsu2yj3LI1zE0WYBSgBZVae9AcA+0uawmXMu0Y4d1ahJk+fGsB7kSGa+YT'
    'nUuUxwosj3MKVXymgLuUC4v1ZFgT+tl8Rjedxx8oaWcmB+3a2Q8kaRfMJovZgOjhtsh9eMhMqFUijwJw3OhKjf4q6aLFWynhPlhGAmKPnf9Ti1OIlKWTXaUi'
    'f4X2vm3JaCUI0AGZF0jyuqUcvygsP22jH6AGHyYUChvF2baV93t1qDRP9RmjXf4QshwmG/aAFLtAEcVb6iSmxKtkCylJMG03zI/pjDxPOGd2VxtDoQiWza9h'
    'D+w2h3DIaS0KmkxDyKBHSsOLM8D5Zv1JTn69LWTk4Nu8bXFGV1PlXZNvUEKI9WDTnApp+JRc1QHUbbhGIXHVMAarCFDZpaJ8GXtA2vEhD+hmkcixz7cGUTFq'
    'NtL6ZpuQyLsBN4ygYwFsWCWhVZnhJv2NeegrNH6cfS/mSZsOWabntYj37m5jHnaAFetazhLkCwvEvK7xGJ2LGAfjquQnbFmbjdYXlAcrN8P7sR61s7RQOuOH'
    '3cHt1635yUOhnACDNPe9YhsBBTCyG40XvGM8f3L34fPyG93x5/rLrV/+fPnrC654GKj/ZkKmbehdncYQDiNwcl1hFaBOeoif7pHetMNj0WDN8A/ugmZdiSZW'
    'im6AH7JTkY/bYRiN8gBVuZlATuUgUgMIcGxUCWfakAFcfswOZfOYP2uViYZJpV40aVOsxC03+oI+Ot5nv8KIYNkKkCXfZTEUdsB+R5MybBupsC2bwqt6wF3j'
    '1XrDYfdbkBTZclni9e8nDo2kyT5x87tnT38g+KFgFgWinSb4TVrTlf/82+8v/6Dd0LCDiMP2ohetGHOlxtGG7j7T8EO8yqZYBluzcmWrNlGWjhRdI7OGw3Ax'
    'kE0zlDtBWZ3N1Dzslwr5eJAZs00ccIFHDJtZTcu0ZQCzjs9muEj1AbOe/vjo2TOGLu5l79OPLy+aq7vKytCTDOQDe4sxxjvOF4Xnnuyhj6AMs7PuKbw5otsY'
    'q7DwgNcMy0CjLZY3u6NPbGc1sI1pG6D5cEF9JKY3Ey6M912WuR4wmpJkI3pZHnQJ2mol3xA9muIwmeNpKcmIpxo9zCgiTWzwqGTH/1BLAwQUAAAICAAAACFc'
    'DXFNn/UKAACiFwAAHAAAAExPRzA1X+qzte2GtV/snpHshLHslYjrgrQubWR9WMtuG0cW3fMrCjCyyUiynn4sDUygMTDCDLIJkMHASGaURIgjBZKMIIAxaEot'
    'gSbpmIpJi5SaTNOWLNFDwy2ySVEwtfGnJLuu6n+Yc29VN4uyPQs92Oyquo9zzz23rom//m1xekFE3V682xOqsafcM6EqObkVZjKffirboaoPhOyG0eCRahzL'
    'C1fEubZq7Ajlu0JtZ+Nt7925OttRDRd/j/bUiaOaQXQWCtUJ42xbvmyr/FGcH2CzfFnI9kC5WCH7rnLrcbWCrWQhJwtHUyIKHGyIjT1xU+WqareJ/WTREdoK'
    'bYQMqqpepZdgs4gGgWzWVR2m1C+jri/u3FU15+rR786j82F0FgicKo+rOP3VKW+y/zqu+AJHxLUynuKArg8v4MLFKcyUx0MTkrjSSqxUhSOyUAeD3oh6bfmk'
    'FleqWIfQ5VW+l7wr+w42x478MszotaOA4gTvLqp0UpZCV8e6w5LsDsioqFdSXlMdtNJ4uDjQilMmc+2aUCdu/OyRUDkPdmYyMxS9rLi/vLGxtsquNUryRZEd'
    '0IH5Zm39hw0ksqe22nTSfknEtQrCGlfgzFYbbk0IxEUeh5QnPIkrryconvulqFMU8qQZDQbYbUjRlqdD5cNqP2t7W+jJrmN9LcMj5Q1lKxS/O1X9LWIeykZR'
    'xIeuCe3vTo2WxHueKnijOGdmp0RcCWXhQsg3A7WTuISg1UPCHlKvjnBGx8UnPD0sIQVketQJou4QT/plxA3v5ZAYigT7QriVBThXpagkyTwoq85rLOn81/ai'
    'G6pmaJk0N0V7UHRfDJFSMccwhznbWYbgYSmxkXM5QFm5aivAvjgmn6AVxhHkUCUWquRJg8tjK0Tm5RNs5vmMJ+zu+ZxHP0sWxxUX1nLSACLggg4GjPw944Vl'
    '7/wU+1BzbadgSsFLc89R0NbiKKHO3Hi7rgHqV7gkUDCM8k6AJ3Yd9Dz1fEfExSIwwT6hMquAq+cC27REbp1GF9nEmIUpYTKn88lYN6nSucRxbwbasMTSejF5'
    '28pcasLSzDSeM1zT8rkVBVycbBHKT7MDyA3rbOt5AWrLh79iaf7d+dIN/Nzi0PsuCISLog4eCXEGkQEdk6ui6qwIZ7747uc/nPIo3fHB43fnd+5yFBquSd8f'
    'TmVC/GXtJ7xpAJdQTsJLDAcON7KleYESyrDh1V9899UmLddBT+Jmzo2LJVk4xUc7A1hGSYifPAKPWJ7fuYvPwFtRPgX+ay4RvmY+PCNeVX2PKglVizOogvAS'
    'QE2rVAUZqNk8NIK0NkIzWBtEm8lMwgFfPg8QUwre4gwhjEgNpRMFFQO+xRvWY1W5tEzl3kQ7YmPElqIpX7pEl/haHsPxsiubnvLrIHnifX4cyjdIuycYzyku'
    'krUu5/XAxgLIaXHO1DGzkQNb350vzuJnboK/nqevYTC+ZleAl8UF/NwgfsNZ8tXOOHtNCrn7K+1kMpxGgtuTD0p1UMqtqNNK3HmZ9sHPZzSDl+Vvj8St2U+E'
    'jp7ZSu23GaQGwNUcDub3+55k6POy29OfGIZkl/st1TwzLUq+evuxdpYuCXWqizqIDEPOrOlI+tBRR5o0LoHELYsQOHygbaly/WNgEdV7CjxVJPXuwJH5YxNw'
    'bTYQaAKqaUCWPtBU1W4RTJfsqRmXwbm/a7oNmJ8e6D1MK7N1Blh50IyCX4mYPfRN/GXJggoyuNIk1Cix0xz2pNP33bEsJ2LoxRD4ZzqGIfkLrE7I0zA59yPb'
    'KiZrAYUjqNq2qpaB6W5MYt097vTsp1nGmOm6VFgEPptnKROgKZz+9AxyickoYX+HYoHuy3BhI5lnQT4F/0P6hfLacKG+2EzVqUJRWUVoaRZEdPfxBFJawl/Q'
    'WueSK5A9n2BSgfm0uJJj3YhA0ReAQXROLxwCPS0RHxRhuvFxIj37/BQp5xbI+41Xrpacpi5JvxCrsaWGikDlaJ1Cem/HnOXXwVfPwveIDfTqqJd1WfJSwDM8'
    'KBs6gEY5mhaWdOH9AB/G0EFt49UOIaGngkvSnTvSvzCNTHfZgs+KhlPMPMt9BusosWO0oDfhgmzQE1u/cPWyIxxZkgdRZ9vUknoafECanmQhkbkVZPPRIDRa'
    'ftTrElvZRvavgaxXeUM2NJVsHwGOHhaKxSikwmqHqFb0PIoluhPRIJoytoZuABmH3Ek6un7cI+qkHxLURhGq37IE8xSDyCrbBc+9ahQ6WlHkEFpsSRgibtr1'
    'WURcyVISQpLdvpt2Mcc8YKVe45mAtoZZMBrbyHIwSo1W5M0AJUkYgQJBaf3pzt1MJikRR8z/5wYiat4ylMqA0sfrNo0qEbPUS+pahfgMA8tYvTXtF1dqRMqN'
    'EqeQpymRjmGJNNO+uMK0Wt36TSwaJUTHz8mnXkwjythIxJVy0rQCTyxi3sAEgPx5vHGCP+T0EPuNvNBvfRR9yTpDRyGRN4EORC1f6ZYI2cdSlfu6PlrLHlY1'
    'ldPobdWoD1YrWkSRbEN8DA8Drc8DkIFx+b1OkOQHEdvOicWVTRH/gvH0jBbP3uTe4Fq0vHSTd70IETNKoo4vEtbX0mQrnBBLt0dTbtLnZ6fxBoh4XszEv1RJ'
    'Z1A7mKNl6NioIuY2LfkSXT4aDPRWdNYTwEfbRGaM9W1zkv6au8YYdSRiQaMuoAYPH3RH167UPMywZj0LE9rDzF1o6BhwzMkU4CGE9bjW0sqXcIGdW0Fq3aDJ'
    'I5jO20fAoKtHW2jGFZ2oTGZpgUGRH1DvWNAqkAUUA53Uxry64Ca2QLJfnhzFlb1Ekp001LlrYQ7pM8PCiF3R60fciY6JD4QmIyAvkiiZYb+b3HjwdNN3YZWu'
    'LPSqQh1fm3I0YDZIof/s1m8lN6EzfZCxTgs4lie6P6aKZ0zqPAtN+zRCSpAw7POgS+fQrOYLbqd4+X0GHfmGYinUzcBCR4zGKaZLPok6eCVn81U6pl0VRWLp'
    'FtfPNLo5j8SYNpNrGiMrc7I+xPBMahV6gGw42SMEoRL0CBJ1e5B+SRpQmzANLyAm4zSgtut8OfOBDmSjKuniyF2tlck8NDQjHiYXXnrsFg/x1eTkpDC/8cms'
    '1Pc44qF1RfV/b6gm6Cz8a70/F/UCWnLikFFF3u8f6V2Pfk8n+J84mY3f35GPA9IGeaD0sEg3Wv0s54CHtEyG56N7P62tf//12tr3Uys//rz6NeMMvQGROWhZ'
    '5KXlotjYfPDv5dXN6xtr9x9srqytmvmdh6w+mLutWdhl4GjJKXOurF+myauNhCZfwanLskF5crdH4y98bA/0xQfxwI7hJpLabLz48u7fdZdzTM3RBnThAiJ/'
    '4pJu00yS9jk+qNwivMVVV8+m3vvabemmwbbZxtyqSP9IBod4lbxiQQOHB0UAUR0EZATrHkL32IAPPsebI9rm3ZizWbOip5qJaRRoHVCjIa2KGWNqvq8iaOqO'
    'STeDJegWS5W/N2swJBCTTs5cvRjtrVsW6w42JZOxp+K0oaX9jNmRVEx9aKsgvprBfkQHJPnoXbOYax9bIOwASDoiXkEDe6mZKm1DskBkQxE3fjkahvaswyhC'
    '72Wr2/itZ3kVDBBec42ZL2+sPVj/1/K9tfWVb1dWv7q/cf3blc1768vfLK8vr+L57PTsjZnp6duUxkbOFhoI1TZmo9PPP7vz56XPsLH2YuWHH9fWN4k5n57R'
    'BcbVLOr06EK4rm+JYH0ygFRpMOVbotElUd2ZmaZuejy8ekuUXht56c0XkV9yM6MVj0uzy17paovI/A9QSwECFAMUAAAICAAAACFcwIXVkkkHAADDDgAAGwAA'
    'AAAAAAAAAAAApAEAAAAATE9HMDVfQ29sYWJf7Iuk7ZaJ7JWI64K0Lm1kUEsBAhQDFAAAAAgAAAAhXGZ3SNRyCQAAsRsAABIAAAAAAAAAAAAAAKQBggcAAExP'
    'RzA1X00wOF9mb3Jtcy5tZFBLAQIUAxQAAAAIAAAAIVxGVzKZ+wkAAOUeAAATAAAAAAAAAAAAAACkASQRAABMT0cwNV9NMDhfbGVzc29uLm1kUEsBAhQDFAAA'
    'CAgAAAAhXAVecXsgBgAAyQwAACQAAAAAAAAAAAAAAKQBUBsAAExPRzA1X+qzte2GtV9BSeyXre2VoOyLpOyKteyViOuCtC5tZFBLAQIUAxQAAAgIAAAAIVwN'
    'cU2f9QoAAKIXAAAcAAAAAAAAAAAAAACkAbIhAABMT0cwNV/qs7XthrVf7J6R7ISx7JWI64K0Lm1kUEsFBgAAAAAFAAUAZgEAAOEsAAAAAA=='
)
_EMBEDDED_ZIP_BYTES = _embedded_b64.b64decode(_EMBEDDED_ZIP_B64)
_zip_path = Path.cwd() / 'LOG05_실습자료.zip'
_zip_path.write_bytes(_EMBEDDED_ZIP_BYTES)
_log05_extract_zip(_zip_path)
```

## 교관 해설 및 판단 예시

AAR에 적힌 모의 사고의 관측 사실·원인 가설·예방 제안을 구분하고, 변경한 조건의 출처와 검증 계획을 남겼는지 확인합니다.

**예시 풀이의 기본 논리**

1. 원본 가상 자료에서 해당 사례·기록 번호, 기간과 분모를 찾습니다.
2. 관찰된 **사실**, AI가 제안한 **해석**, 학생이 내리는 **판단**, 부족한 **자료**를 별도로 적습니다.
3. 근거를 확인할 수 없는 주장에는 승인이나 성공 판정을 붙이지 않습니다.
4. 최초 답안과 수정 답안을 구분하며 AI 요청·답변 원문을 확인합니다.
5. 학생의 독립 수행과 실제 사람 평가 조건을 별도로 확인합니다.

**실제 제3자 평가/승인:** 제공된 문서만으로 자동 생성할 수 없습니다.

- 미실시를 PASS로 쓰지 않는다.
- 파일: reports/log05_redteam_critique_report.md, LOG05_M08_forms.md, forms/LOG05_M08_forms.md 실제AI요청·답변·근거대조3건이상·보완·재확인·본인최종선택과작성/실행기록이연결됩니다.

- 보완 방법 미작성·미실행·검토누락을보완하고교관검토상태를별도로표시합니다.

**교관 검토 기록:** 판정 [미확인 / 보완 필요 / PASS / FAIL]을 실제 기준에 따라 기입해야 하며, 자동 표시는 하지 않습니다.

```python
MISSION_JSON = '{"code": "LOG05_m08", "steps": [{"title": "계획·실제·집계 분모를 구분한다"}, {"title": "AI 레드팀과 전제를 검토한다"}, {"title": "모든 비판에 근거로 대응한다"}, {"title": "조건부 비교와 수정 계획을 넘긴다"}]}'
# LOG05 학습 기록 도구 (표준 Python/Jupyter 환경용)
import json, zipfile, io, os, datetime
from pathlib import Path
from IPython.display import display, Markdown
def _load_notebook_widgets():
    from importlib import import_module, invalidate_caches
    import subprocess
    import sys

    try:
        return import_module('ipywidgets')
    except ImportError:
        print('이 노트북 커널에 ipywidgets가 없어 설치합니다.')

    try:
        subprocess.run(
            [sys.executable, '-m', 'pip', 'install', '--disable-pip-version-check',
             '--retries', '0', '--timeout', '10', 'ipywidgets'],
            check=True, stdout=subprocess.PIPE, stderr=subprocess.STDOUT, timeout=60,
        )
        invalidate_caches()
        return import_module('ipywidgets')
    except (subprocess.SubprocessError, OSError, ImportError):
        print('이 커널에서 ipywidgets를 설치하거나 불러오지 못했습니다.')
        print('설치 가능한 환경에서는 새 셀에서 %pip install ipywidgets 를 실행한 뒤 이 셀을 다시 실행하세요.')
        print('설치할 수 없는 환경에서는 제공된 Markdown 작성 양식을 사용하세요.')
        return None

widgets = _load_notebook_widgets()
if widgets is not None:
    MISSION_DATA = json.loads(MISSION_JSON)
    CODE = MISSION_DATA['code']
    OUT_DIR = Path(CODE + '_results')
    OUT_DIR.mkdir(exist_ok=True)
    BACKUP = OUT_DIR / (CODE + '_record.json')
    fields = {}
    for no, item in enumerate(MISSION_DATA['steps'], 1):
        fields['activity_%02d' % no] = widgets.Textarea(
            value='', placeholder='내 판단, 확인한 자료 번호, AI 제안과 수정 이유를 직접 작성하세요.',
            description='활동 %s' % no, layout=widgets.Layout(width='98%',height='130px'),
            style={'description_width':'75px'})
    for key, label in [('first_thought','AI에게 묻기 전 내 판단'), ('ai_request','실제로 보낸 AI 요청 전체'),
                       ('ai_response','실제로 받은 AI 답변 전체'), ('source_check','원자료 대조 3건과 수정 이유'),
                       ('final_result','최종 작성 결과와 근거'), ('limits','미실시/미측정/확인 필요 및 교관 검토 상태')]:
        fields[key] = widgets.Textarea(value='', description=label,
           layout=widgets.Layout(width='98%',height='125px'),style={'description_width':'190px'})
    status = widgets.HTML(value='<b>아직 저장하지 않았습니다.</b> 작성과 교관 승인은 별개입니다.')
    restore_input = widgets.Textarea(description='복원 JSON',placeholder='백업 JSON의 내용을 붙여넣으세요.',layout=widgets.Layout(width='98%',height='100px'),style={'description_width':'110px'})

    def payload():
        return {'schema':'log05-student-v1','mission':CODE,'saved_at':datetime.datetime.now(datetime.timezone.utc).isoformat(),
                'answers':{key:w.value for key,w in fields.items()},
                'status':'작성 기록 (학습 PASS 또는 교관 승인 아님)'}

    def save_record(_=None):
        obj=payload()
        if BACKUP.exists():
            stamp=datetime.datetime.now().strftime('%Y%m%d_%H%M%S_%f')
            BACKUP.rename(OUT_DIR / (CODE+'_previous_'+stamp+'.json'))
        BACKUP.write_text(json.dumps(obj,ensure_ascii=False,indent=2),encoding='utf8')
        status.value='<b>로컬 저장 완료:</b> '+str(BACKUP)+' (PASS 아님)'

    def export_json(_=None):
        save_record()
        raw=BACKUP.read_bytes()
        try:
            from google.colab import files
            files.download(str(BACKUP))
        except ImportError:
            print('Colab이 아닙니다. 저장 위치:',BACKUP.resolve())

    def restore_record(_=None):
        try:
            obj=json.loads(restore_input.value)
            if obj.get('mission')!=CODE or not isinstance(obj.get('answers'),dict):raise ValueError('다른 미션 또는 형식이 맞지 않는 백업입니다.')
            save_record()  # 복원 전 백업 저장
            for key,w in fields.items():w.value=str(obj['answers'].get(key,''))
            status.value='<b>복원 적용:</b> 기존 작성 내용은 이전 JSON으로 보관했습니다. 다시 저장하세요.'
        except Exception as exc:status.value='<b>복원 오류:</b> '+str(exc)

    def export_review_zip(_=None):
        save_record()
        target=OUT_DIR / (CODE+'_review.zip')
        with zipfile.ZipFile(target,'w',compression=zipfile.ZIP_DEFLATED) as zz:
            zz.write(BACKUP,arcname=BACKUP.name)
            text='\n\n'.join('## '+key+'\n'+val for key,val in payload()['answers'].items())
            zz.writestr(CODE+'_work.md',text)
            zz.writestr('README.txt','검토용 보관 파일. 실제 평가 PASS 또는 교관 승인 자료가 아닙니다.')
        try:
            from google.colab import files
            files.download(str(target))
        except ImportError:
            print('Colab이 아닙니다. ZIP 위치:',target.resolve())

    for i,item in enumerate(MISSION_DATA['steps'],1):
        display(Markdown('### 활동 %d. %s' % (i,item['title'])))
        display(fields['activity_%02d' % i])
    for key,label in [('first_thought','최초 판단'),('ai_request','AI 요청 원문'),('ai_response','AI 답변 원문'),
     ('source_check','원자료 대조'),('final_result','최종 결과'),('limits','한계 및 실제 검토 상태')]:
        display(Markdown('### '+label));display(fields[key])
    buttons=[]
    for title,handler in [('작성 내용 저장',save_record),('작성 JSON 내려받기',export_json),('검토용 ZIP 내려받기',export_review_zip),('작성 JSON 복원',restore_record)]:
        b=widgets.Button(description=title,button_style='info');b.on_click(handler);buttons.append(b)
    display(widgets.HBox(buttons[:2]));display(widgets.HBox(buttons[2:]));display(restore_input);display(status)
```
