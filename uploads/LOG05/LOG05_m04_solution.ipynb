```python
# @title 📦 LOG05 실습자료 자동 준비 (ZIP 업로드 불필요) { display-mode: "form" }
# 학습자 화면에는 결과와 업로드 인터페이스만 표시하도록 구성합니다.
from pathlib import Path
import os, zipfile, stat
from IPython.display import display, Markdown

LOG05_PROJECT_DIR = Path.cwd() / "LOG05_실습자료"
LOG05_PROJECT_READY = False

def _log05_extract_zip(zip_path):
    global LOG05_PROJECT_READY
    zip_path = Path(zip_path)
    if not zipfile.is_zipfile(zip_path):
        raise ValueError("유효한 ZIP 압축파일이 아닙니다. 실습자료 ZIP을 다시 선택해 주세요.")
    dst = LOG05_PROJECT_DIR.resolve()
    dst.mkdir(parents=True, exist_ok=True)
    with zipfile.ZipFile(zip_path) as archive:
        entries = archive.infolist()
        if len(entries) > 20000 or sum(info.file_size for info in entries) > 1024**3:
            raise ValueError("압축파일의 항목 수 또는 해제 후 용량이 허용 범위를 초과합니다.")
        for info in entries:
            item = Path(info.filename.replace('\\', '/'))
            if item.is_absolute() or '..' in item.parts or not info.filename.strip():
                raise ValueError("안전하지 않은 ZIP 내부 경로가 포함되어 있습니다: " + info.filename)
            target = (dst / item).resolve()
            if not target.is_relative_to(dst):
                raise ValueError("실습 폴더 밖으로 벗어나는 파일 경로입니다.")
            mode = (info.external_attr >> 16) & 0o170000
            if mode == stat.S_IFLNK:
                raise ValueError("심볼릭 링크가 포함된 ZIP은 허용하지 않습니다.")
        for info in entries:
            target = dst / Path(info.filename.replace('\\', '/'))
            if info.is_dir():
                target.mkdir(parents=True, exist_ok=True)
            else:
                target.parent.mkdir(parents=True, exist_ok=True)
                with archive.open(info) as src, target.open('wb') as out:
                    import shutil
                    shutil.copyfileobj(src, out)
    files = [p for p in dst.rglob('*') if p.is_file()]
    if not files:
        raise ValueError("ZIP에 압축 해제할 실습자료 파일이 없습니다.")
    LOG05_PROJECT_READY = True
    print("✅ LOG05 실습자료 압축 해제 완료:", dst)
    print("📄 확인된 파일 수:", len(files))
    print("ℹ️ 실습자료 폴더:", dst)
    print("※ 업로드·압축 해제는 학습 수행 및 평가 완료를 뜻하지 않습니다.")
    return dst

# 실습자료가 노트북 안에 내장되어 있어 ZIP 업로드 없이 자동으로 준비됩니다.
# 내장 파일 5개: LOG05_Colab_실행안내.md, LOG05_M04_forms.md, LOG05_M04_lesson.md, LOG05_공통_AI역할실습안내.md, LOG05_공통_작성안내.md
import base64 as _embedded_b64
_EMBEDDED_ZIP_B64 = (
    'UEsDBBQAAAgIAAAAIVzAhdWSSQcAAMMOAAAbAAAATE9HMDVfQ29sYWJf7Iuk7ZaJ7JWI64K0Lm1kbVddbxNXEH3Pr7hSxSMmDtDAI2orQCJq1RaQ+hIZSEtK'
    'qFECRfShWiebaBMbYkOcbJK12RTnq3XbTbyOHcm85Kf0ce/d/9Azc+/aC1gCFtl3556ZOefM+DNx6+vro5eF2i7FhaY8KQhVbMTry0JVHTkfjoxkM/rE5PP8'
    '7KN7+fyjzPSTF7/ck7s98UV+JndPbZSV7Qm1EUYtX6iGJU9toRYtfG5eREC10o7XbLnjKb8Wr3Qzv00/UTUcc5ty770sOrLYyAjl2+o4VL6n6hXZ6oofbn4T'
    'tXoiagOWHVcP++f6dwhVb2q4qmaJqBuonRCfVZR9JOLVUL5cI5jy5T9qPVRvArWPN6orAGNCjYxlRNRqx0ttce0mUvg7rvomcbVpiVtTc3P5Xxho7T1lB1Sq'
    '1hVqvgmsIi6VZfHgrKNqrmx2zzrXblLOcn8n6uJMrUcvyoOe8nGrXxjcijC2XLX5yuB9Uj8Uycdja00d/33WkcW2bFl89bserpbBdirAtW9uini+IVfWRFy1'
    '8Q6wL6W+j9rlKLQEgiDSv11Zd4U6DeUbD2DNPfVd1FDOu0KeWLJECKuyFVJEnYaqL0p/N666nxbtYkZ8+9V3t299L5C2qpfRepRAqPWKKtYY8hEwBVSwudyv'
    'U5OzU3PPZp7ykXc9ZfORuEqNxr+Gbun2bnaBScSvumq/gp6CNUKubqKggrhx4ono+D1RCQEJgO1Rk3XXEZAqK8v9aHEV51utpMrmlMtc1aWIOr3oKOBCxZtV'
    '9Bf/NX1ue+qPRaFsP15g1HL+IDotJHW4lNH55aYnH0znZvI/PZsirEM7CX2VCSyCREEBf0CCEGxN5x06oB5dDvLZeP55QCnGWyVV9NG4fScJ8LEg7j58ceFG'
    '/vmFuw9zXOaoZcdbL8860bEVL/n0RDN61Nsf87OP53R7dSH6MVA2ioszYEO8CUbt28qzSQaa59w/t0WFizpduWdrXXTVXm0oSy5nOGfoWLmOKnK7SKENi0XN'
    'RRfxawd5kXihiSjwiHUG0LdZ+vh61hxVG03qrKqvybfLV8bORYFlAl8dPWcCM6KTQ7VzNAyRkEuv+aY3R1F3GQX1y8pxSbamxvz6fFPWS5yraf7KKSXp9bi1'
    'QzxEqKUSK7Belu9KDAtaxPlDcHsTomqj45Zc2VW1kPSmGhWOsWnjODGXSOdXh9bw84yYGIeMG2orxP0QMLkeIdHOOjE6PqkcD8yO3apWuHHbzMOnj2c0ZUPp'
    'N8g+qlS9MucJ4XRLEg61FcCd+g7ed+LYteGsxjPkfkl7QnRcIo+WxWaatXyGXBTtg25jOxgDOwVnGxrH5Hoe/9V3TN0uPLm1Wr7abkLxn+UOjgqCV9/9z9rk'
    'XBxb7vUJghvibTtlppq4JhujHvkn4p6edQYdNqWQFYjglPX0gSlQsWulJHUg05kNMh4Z56YMMBKnNEzBeD6eNkSFQGhV01kcUksvBTtbjSdG2lV08ULHCBZJ'
    'Ks9nVtUdxGUNGFkVvSiwh1CfDHLBMDaueDg3XA5UAR4D4vaX57Nnndt3zo9xpxYK8QLnDixyr5lUpmThG+iGJYgn3KGOfBPb5BEMZygxQWVlp3+RaUWrTeOO'
    'gKZnEN/ouNrj/Cpw0uu6OkNxG8aR7FYaqB97VnrIDarFn+81aXx/2A2qyYbLTak6ifeRwDHZtw6HavEKEoFVAd7+coKB+n3j+4lbnN6Oh+w/FKs8tuOVnvEV'
    'Djzw274uyZL4Xar4k9n8z1P32cVplhVPKQW1iNT2amyYvNWwgzPlAJWNxpgVz0f6Nl5F2w5Sc/BxbvbRZO7+02e5mcmn04+nmLW0QZltj8hUIKn3Ky//pU4w'
    'H40YMU8Ag96kpA3qC6ab8YJFmj/o0sgwAxSRMbkTqg8pw8hViOkKsh0bhUIFDXzMroqXiPjESfYbdk0KTY8NdHAJVtYQRCDvIDVWCeN8kDaobRz0EH+dODvf'
    'jI5Cika+TDPlBDakFhwQxo5oBL1mPmp6DOYmNcH3YAWUIqmZGcNEw2ZgD583E+OIdVm5BSyWvBNh4NTSm4BZDiEuPQB4KYErn7omJy2sVDLkcqwmroQp/Ooy'
    'gMDax2B2w/aDkewowFylPO5kf78zzhkaD4G9+DbaCwA0lpPFp97Uvdb7Yndn6Nqhly+QDpPBGhvFl/31UsIxK95F/RENePci6j40JX2UANGaS7PmksjGr1yC'
    'pr1Ab4LM9MD8SEiW/GTntAZv0J6oSZPM4pTlpRF/4kTJ533nxBgcagXZrN4ric0YQoWQ28pbopi4dNaZ+Bx/r3C2LEMoOTuKIUK9Y9d13I9WXs4GxedtRBOE'
    'CldQfkUkW5zxHK06rIe6bgNtJQulNuuBVS0mA/7Zk5l87sHcBd4d+j/BDH/TayhiQliOqS7xoo+KZzDclBIODmAKukDUN/rRtHXIU6jYgECAWJNME9SMJwx9'
    '/ePELFZ69cPx1Oon3wbkJnXbgOF1FH3nWZnY5qAd/wNQSwMEFAAAAAgAAAAhXM8yij57CQAA6BsAABIAAABMT0cwNV9NMDRfZm9ybXMubWTtWF1vG8cVfeev'
    'GMAoIKGWKIlU2grog17qGojRAHrIQ1E4LtI2AeIIiF20BYRiZa8ERmTgVa2VltJSWDV0KAoMuiYpiUKZF/0U521n9j/03DuzwyVpxUU//GQ4drizM3dn7j33'
    '3HPnlnj/V3cWlsW9hbJQx7vKfSnU/q6qNgoFdR6qv28J9SRWF6HsDFfERw8fPP7i0z8XP1v/w8Ly/QeP8fDbPz7+dP3z+/rF/MOPPyoU5owhdexdX8q4DWPf'
    'O3vqJJY9VxRFEoeqMfjx6t3vHX9lNLkxxOSuqyKXB6tNFYXF1btCnp2qRiDSZ18msQOTqu6IZBCrpiOy6YXPfvfo0frnsOTJr2vyxVCoxndJLzKmU78tqxVZ'
    'bc4LdeTJ3gAf6vVV3RXpbki2k7iuGjD3pK8O2/Qr3T7HotQP1EFHpH6IL1Vkq2bt9eUzV0jPmpX/GGC3Se8clvFzcKIi//qS5lWvhDrYVo2aarjYVV+FER0g'
    '9U+Tfwb44mlytWmsFAq3bt3CUDBvA+GG8J1YFHMiuRwmL2Os+1aeteGTtjxpzCwuwZds9enmbKGwQdtMg4HYmJiO574jo1OhWpvq5daM2rlCBJKLGkcIgUGY'
    'VSXAz8iXV/DppjruYP/b56oa0WhFxad0SLycJWsIRHNXXQX4ndZ92BIIs+z6YgObmJubEzf+i/eL+J37DyNLUyOlqZHy1Mjy1Mh7UyM/mRr56dTIz6ZGFhem'
    'h6Z3vTi57cKNEVxCBCmNqo2kN0RwQnVVXzHRoKQw4VKRhzCIn4sMOTxBHXjAUJx0Y+mF2VQbAs4NH6jtMxK1bfiYk+WFkNt/I0Bry4Q8uesRTFRrkA3CAMIL'
    'A1giZoQQs3jB0W7VyGy7r0dvZ8aXbNAvKrRGb8muHmHytatLYgxPZhe0mvYaDu0KzuPDWLnIY2wz8pGIOKM1VDZL5c4ePOUnPReex9l84FeQ346R4DVPVuls'
    'Qvku5Vk7Nkes2906ZpY2f3MaljiIHXniiBLzItLksE1pZ0bxI5dY9HjkJd22sAlGQ5xhlFOUY5Q/nGT8Ju8VPfObzgxtP/IFCEu+oAy92KMdk8knSFHV7aeb'
    'HcxTO810ZyA0r82+OQ8ncnJtAt4W5GtLN70ovfbFzTlQhvtkVJHPw7SGWMeBjH0ichDZGHVNzplRQWVFcAhH4c75GemRdGsCiQVSBmXTg44+Ua48ixF5FAuE'
    'lqjL7InN1xzV2KUvstOI5qnibWP5Nw1Ox8O95GX/h3w5wWYTVDbBY9o3sNpXwab5aFb8rgYot5ztosz5U3fGAUHJkJx35IWbq2aa/3X6kVtMHQMyqHxSXjd3'
    'Ze98RBZkJd0bkpdQHXUWS+8UVJ9PaJqFDwFVuW+tLbLBlgMqEiW8FvlP9fGCCmR6UEFOmPqsY0ZlUXoBJiAHdxCMrHqrfoWC1hhme6EMt/N2zsdrtnjlBHfK'
    'hANMeuXUcd475evLO8v4+5463spmj84Kh8jOgND3dNMkjOETVQlViAMEFVUNmaRio0osb9R9+YJKH/PKJI3Uxyq2zn6aS/Eb8UE2YovkiL6vL/MlFBE2/s/e'
    'mrjh1YVDrkYcsA1SFCFtB+N5VijqwxVNohlk/TuleCqr//sBgjhUG2KcHn6Fc3addDuygivL9zlLDghA5K2IV/sntExWz2XPoSNgJsDj8gszqg768hgYO8rS'
    'BmZW75r8l62TZID4HWzJTp/Th8UWa0nGoFGH2FJePMLXRx4QxenTjQFImoDy6KQ+6s7FHty8YmJK29GBYq5g6GN9IwDMmFrIflagZeAxUzFfy5MQMXRBKLxS'
    'T5wIUGHCpVP/J5KsOWl9j+Qf1bgNo5KJqFT3W06UDsRtWkPdRu0MXY0vMyvz4g2zkGM4GSW5bO9BdL+B924EwcQTHjUG6OQUYRbZgs8C3LI/jjwbm54LnQSn'
    'Vpvp/pcGLpRAnIAknnRY6VRQ0i1KMtYGjDDkAhmOAyLY7GD/jzQo4CDq6yEJrBKViEzuuPaso9ZAA187n9oGQ9KGtU8oLk1kgUVWQGwkW8fMX7Yj0ac1p2dl'
    'BAFhOZKxYQoKY4FhTbDMhVd3PEieUQjGv2zF5YbIWj83Sp82COP0WWrm2LUGOv8pQIgidK5xL8ceE/Bf+jTkTNN8rxs+08LYrLVozpKVCtUZKpjd+gera2vF'
    'X6zefb9IrRifsAj4MF1qvZo5I2Mn8+3/EeB/8ORMRuimjQvzHS8Fo1HjVnIrIB1M0HADIjV7VlsoQD5wjO6ePXJNBBJ0VpgQs45ZH27UTEM/eOgDkflZK51N'
    '0SSbeUs/5dHEI1pzUSOvrY2VLLTyWt0Y28S/dZfpkOPGJ9dx1Zgy55wnjoeIZTZFad/sg85Rx5g6KwE4YJRKMPFrvfg3tkxzPbTBHxVV3WwJiIa98VZgJrl0'
    '1PZXs1aiuCLfRRu4gWSNRrrBMq9bWPiRVunfpfsV7BiTbhv5phs3EkzsNa7olL+QrLznodk6S62zLXg9Z2K6yTBESOumJLSWsbQhvrXoszYgYmcRSyjS0obV'
    'MMuGkcpiN79G8/Jlhdukudm2yP0ffvKX4i/X/1T88JMHj1dMwYDwdrnSZ1WGS5QNiWb5PEtmWgAr8iTDBzBElIv4RLZbdOuLFsYrlbJceueVMnOphr3lGbqN'
    '6Zu4SA+YaPJN0fN4SnlqQNdD+axOTbpqucAjkwaclWtd9BQr88QUVWWqzhZk5hsrit7cYNDlnP7KIkN6BCOja001yR2detmVSTTCFSo8YUCxhUxR6oLia8yv'
    'Lf51rcTXCiPVOoJfRkEMkWgT4OAyl/GSBh3BykiTvPbSPR6dm5GjRVxRtSosSBDb9LDGN04aNUVd2vUNh74QYQ/m1T6DL3c1t0/jdIVgFIRmIdZfuyGl2es6'
    'DJYWrgEZ+eL36188fFTki9H78P19fp5/+DHzmkatqYFvvhC9Labt3L75C/ayJ6chNuiPaS5HUFhiKEAAY1P5ThBAsxwFz8jnEUNhdNuk7ed6ENtw0GUtaapc'
    'oekG6AMxkm84zA2MYQmKgb46eYeCt4GC0hgKSoyCkq4ZfMHHZYFjg4hHu0LzVcYIk60od+1+G5AxFW+qwGjuph3Z6/VMu/dN9cnin3X2B9sUfH1LwF2+Ucu8'
    'q9xd3jvEvBXElMcQUx5TRUZB2fsR0nvMHCQ9+5o58gohf3WBjoZrvTbC4Ov6aL+LSexxVTIlNzN6b/n68t7iAu81cmW18w4Fbw0F/wJQSwMEFAAAAAgAAAAh'
    'XEGnuDOkCgAAICEAABMAAABMT0cwNV9NMDRfbGVzc29uLm1k7VnfbxvHEX7nX7GAUYBC9YuinDQC+uCigGqgRh+MNg9B4LgI2hioayBV0RYQAko+CZTIxFTM'
    's0j6qJ4SyRIFGj1RlESh9Iv/FOftdu9/6Deze8clj5LjAgLyYCCOzbvb2dmZ7/tm5u6G+O3vFmdvijuz8+JNoSpku6uaPSFPumFvQ6gdBxeE2rzAxfCsrHYq'
    '2fA0UMWaCnry6HBCyOCJUM8rYacl1G6gvlvDPaGO17BSRK4XnjhCHq2pkpfJ3Lgh5H96ag33H69E9VYmI0sXquSrZkGEQQEXhfK9D8n4+q7a25LlgnFHeyOD'
    'mmrWRNgL5G5TyDNHOc14K7X9MnJ9tbMmS0VZ2pseew61XVGOd8VpVL2QPox8eizCbkH6h0IdrNB+zZETyv0+PHcjtxae+JNCrbbxt4ieO4l369irTY+F5/3w'
    'OMC6l/KoFZ70+eB7W+qiJnc9MiJftGFHblbpnGqvoJoI4EFZPvXpWJnMFDkoT3qvz9Xjolh8sCSib3rKORZ5xDgMnB8KVVXaU73dMPhWNZE615Gl2g8Fl1ae'
    'ejgT/AvUmSfb/QXx2cP7S18++OfMXx79efbmvftL+PHHvy89ePTXe/rG9MPPP8PC8OQ0Wj9FFLdoJ+UW5WqXYnXrtgm8uUYnlId95SOMzVdq89Rkg3Ov/aYU'
    'wxxsVeT3ZQ64sep4qtSE+9EW/uEhskI1uqq2QjlUdUQB6a+t0Elg7QYFbnrYmMiJKZMieeIMpwhGRVQqZJFWvsd7TmQyy2ZBVgNwQiwbePChCMo7RU5G53H6'
    '1va6dQtJPACOz8rk+bJBKT0q2z1C3TL2mpqaEpf+P7N8Fc+WhfUfPToK09QDwNKFA5j4K8CekKUu2JZ6CmkFA8ffzVwa6DkEOmasBnseXhCczU/lOzFPPwlP'
    'K8rbVY2WXqFNfMqhX23L3YJYjlcRqhqtwW/Lytjo4drdHH78/tdTOcAjiJwA9MGSjTDAnn2iVBi48F2ooCX362TaR9wKhLHtCo5Tlf/eAL2YbDvtyaEMECSh'
    'XQQB1SxPItGvAEXN64HU6bxH5YosHYYXZfqRhzYBVwW1HWgn58jJP0zNiV+J8LQNloLSK9EKsrJVUX6RTZJ1koBmNwKxRzLDTG6WiQ3aAMkCQoRNoBa0p3I3'
    '2UljEQwiBDAkzz11UScLGol88O11fvry42rH83F0DYpUt0hZ/ghnVZ7D/kJJAo8Da+MM9hBSuOFrSfQLqtGLkzlud4BjkJvEFd8hb3FkHMs8fiUw8wRMW2KF'
    'qhUpFvNh4GXlak0e9hAToq1WAjpCbg43KfCETo0auVmTLwJIF0vEsMXl4WpAv2MNj3e7DK1D4ItcjpS2yglGWLYp9llkUAWHVC1BYdYk+wJ+/zxv6YVJNpkG'
    'igKmYcmX3wXyiUNhXZx/fb54E38+ILQKKi+rkFKUljLkVgat6Nm+KdJQVzEsQ9ZmGumIW3jcJfyBpXGCGIkyeE4PyKP/Rs80hFIqlSVi+khlo4od6GRT2rjN'
    'Ld7HUMmwSNc2CpJVFZ3a+E3I5bkBTHDY6eEiw4ViOKn4A2RmbShw6mWnGNV64i0YyNoRg+TGnpD6kgprbrw+15ymqyP5HSAIxK+73LlAsTrujyobuZSsz6Wu'
    '5FNX5lNXbqaufJC68mHqyi9SVz5KXcnNpi+lvc6Nun15BqkCUR9TanITxRK3YBV/ky5gjQDxy5hspoFjBQzAFVnx4keTFFB7F7nESG4ktW3EmB7YAVHWv6WG'
    'VVum/oWkCTBRB734IvWHwSEMEPuyQogJwjxl+6BMZltdfXUyNj6XJP2sSGu0S8lqS57Grc6PyIP2glaTrxC0eAX3qY1AOdy/cMfq4YyJoXmzlOXxwKXOdhX9'
    'bwcVdCVpp7jQJZ1lUbYCc8R64m1cDrX56UuTmOckcg+Q58YUNGm0hjuDYSkaQ/WRPicpnGnR1N11ltz3Xeoq5T4x9KzKtQcmV0FR1emiMOM5tbkXbfZM8znx'
    'dh6Oa0xS/8XNwNgb+bE3LufAPMIn/aJ86kU0KQU1GbimxA5J1+gzWVSpBcEpHKTbijPoEXbQEp30UUBoqMEPnX2W3aMAmYcGU3WkqGqf2HwZwrxFO3LQuIXH'
    'yLGO5S+aTMdGFbXjqliOqNmIlI3oWBIb4mlSWzNJoR2otaHXyMYkgsytwVjHYQAK7G6eFGYw6KEb52YEcNum4QbGGPa6RtokMTSmRXZdMV5F9aoeGK0RkOfI'
    'sXZHpE7kvponf+VFUV30dFq0W8eOeu6nlyck48LNPKM+oAITxOa+yZDumD3EgG0yUwYNJvfkrAuvz40qxHIwIBBVePobD2Dk5hYRUosJ7FnqVCnw5qmzgDhr'
    'IJnwjIMQe6dBNmo2o4dNCGWnmLxvsAdJmpDPPPSN3PMa9jt7aIB4tjRdnQYGjae2ivAIj270DOBuxa8adHJNsmkgDOocELRHmMSbPCNI9KhDeY6qfXM+jU1Z'
    'OUR/YFcBBtuZwy8Ekr3u5tig7or0nGFt1aUuDNNMhDYSg1TcK3VNl4x0x6OC65i7pqWXzX7si53xwQgfj+9vCrXFeZ5S3NqbQh3nHbSX1uuXwVk3q/HcQdMi'
    'g8QUIVX0MESYnpkrW2DAl/Co7sp96pcYdKO1p26/XYjcOpzFCk8+qWtZ0P+GntB7rQGVjQ8DcsYJpWKVyYwoQjK9GgtIUvqdz93cV3fz3DeYXHpGh2k/IEdf'
    'NtRDRAZV0fg4xz7euk2As+GEozPzh94CxRjlFxncxhDrTLxZuwht6B+RRPAw3l6oDoYaYqZdAJPx0CESEE/tom3cy7N7eS0j3PMwX3kRXPG30L8XQNA4hlYY'
    'dJQJk2ipVgOD+bHMF3okp0GOsQ8R2MWCbiIHnoaWwa1++6I5oMdS9X0fHRp7ZbU3w0fRbzmH3g3php1hWVWdl2a2CHtdHWtbfqxAI34FaHiMB44KS96MVkAK'
    'SbTuD4zeAUfu5Gb5kD7g27bcE9GTDbJMelN3aJLVpxy870N2/Ohx0yjaNI/GnS4zH1xcMe82uZQXa9GzjYFmwMQnevGno+cebX1NnRFDMWLSZcPzglr/esKa'
    'v66uaeMt87rZ2Z/pXuwVifcmCfDk/1WIaWK1TKRbSZuFo4jToCKH0DAyyn58IeQwX1KWwCJ6NnaLwv/xF/+a+c2jf8x8/MX9pQXgpkdsJKo1vsYJIdy+R+4w'
    'mkzgNPgNLTjE4C2XL6wwr3E5p3wApttQxo1V/Trg9fmt29QgILCsB3SzREpWrrAQJOUmVsIaS/FpBUNuAmN6ZxN/CZAVYGKPi/PTIFUqdAFmJ/msNcM90isj'
    'RTG6c9P84UHffyehXhhFzLUJdeyzAQalXgd36NWT7rYpSZxd3RXPqIMiih/HP2qUefbXmZ3RCqVnTT2aeg4Nn1YJZYCQdGpjutOhYQ77g+FGKQip8SvzMWWb'
    '3krSPMxAoFj86dGXD/82wx997iH29/j39MPPWXs0sqJymZjw9m8DkyJtZ/LyHZKx24LqnI2AdyqD9rh/bWXwffKvM/l5O/nv1mQsiJ9Wk/EeKNcJlHkbKO/Y'
    'wi2Ia2/h3if/OpOfsb54X/HBe8a8w+Lv3slnb03l3S6ORiN/sxa5p6rU1p8a9UVkyCRos6o8/U0t2fx/UEsDBBQAAAgIAAAAIVwFXnF7IAYAAMkMAAAkAAAA'
    'TE9HMDVf6rO17Ya1X0FJ7Jet7ZWg7Iuk7Iq17JWI64K0Lm1kbVZNb9tGEL3rVyyQa+smaYO0vuUUtGiaAuk9ORU9NCjQ9gdQNmXQplpLsGhTDilQiWTJqYLS'
    'EmVTqJyDf4qP3OV/6JvZ5YeSnAhR3J2ZN++9mTvi+6eP7z4Q2WKZ7y3Fo2+FOnmXe5FQ7kgdLIXyHLmTNBp37gh5aSs7vLnK5la+hw9Oe2r+TijfUW7QaNxa'
    'gfJsOfDNhfLfVLV8oQYd+aat+pbANULtzFRki7zdke5UhbaQcS/f9VSwlsNAqMRBOH59aOOIGvlb5gL6O+97Kkxlx5cHPb7JHSGZZUcFQ3V6LrI0ViNLqAh5'
    'ODdX6rKXxZY5pMJE5J6NlHMvoExwgOJkyxmqyj0/W0QmkpBzJ/dTOV7jqqY6TmpZqJMO4Nnj+IOxXNlC2RfyYIxobpDFTTwDWw1sPO1Avm3R+1F+vC9U36Yi'
    'cE5O2vJopCbAY+Jsicfy6CJL97PFGvmdZ/Nz3I+ULAqPzOTZTCf3GcOnS1ShL6Rjy/Aa9+821SWXouL3wF8nuwzU61aJMkPPWHhRNm9v3VphoyGnaxXRqz66'
    'jOrofzVx5CylXIpAuDbyCnhAiCgwbWfw0lgOQ6QtXUe6oy2QhwDSYXHQLuqYDOk+lKy8A0SrPjedEF9mF+BGmFA1FPNVx+ShJk0VdQWQRp9NHnKRmo4ylvQi'
    'Ab4msZsr3I1umMZTu7OrdXYRcyY702zVNPE1p99OCc6C86eJ8puNxufiyf1ttLiL9gIbBpz7xGUvllkcUDJpTFVSzVncJ5yRHvAq0NOk1l2UnYDq1dLhVOJX'
    'NSjkIc6vqy+4GxpvediSZ9ciP9zHS6IhETP0IS+kAFCZjHVcKfuvtnEX/zOMoRSRJZaKOiYjjQcSs9TwAgrCtaOuWvki/zvNj8esv5WNNtZai4qBplAXLU3v'
    'iZct8JRzDyJB0mBCRAcJdWbMR+2mtB5QWlSbpo6QbiLd1c0VnEe5UflTHawQLLtso9yyNcxNFmAUoAWVWnvQHAPtLmsJlzLtGOHdWoSZPnxrAe5EhmvmE51L'
    'lMcKLI9zClV8poC7lAuL9WRYE/rZfEY3nccfKGlnJgft2tkPJGkXzCaL2YDo4bbIfXjITKhVIo8CcNzoSo3+KumixVsp4T5YRgJij53/U4tTiJSlk12lIn+F'
    '9r5tyWglCNABmRdI8rqlHL8oLD9tox+gBh8mFAobxdm2lfd7dag0T/UZo13+ELIcJhv2gBS7QBHFW+okpsSrZAspSTBtN8yP6Yw8TzhndlcbQ6EIls2vYQ/s'
    'NodwyGktCppMQ8igR0rDizPA+Wb9SU5+vS1k5ODbvG1xRldT5V2Tb1BCiPVg05wKafiUXNUB1G24RiFx1TAGqwhQ2aWifBl7QNrxIQ/oZpHIsc+3BlExajbS'
    '+mabkMi7ATeMoGMBbFgloVWZ4Sb9jXnoKzR+nH0v5kmbDlmm57WI9+5uYx52gBXrWs4S5AsLxLyu8RidixgH46rkJ2xZm43WF5QHKzfD+7EetbO0UDrjh93B'
    '7det+clDoZwAgzT3vWIbAQUwshuNF7xjPH9y9+Hz8hvd8ef6y61f/nz56wuueBio/2ZCpm3oXZ3GEA4jcHJdYRWgTnqIn+6R3rTDY9FgzfAP7oJmXYkmVopu'
    'gB+yU5GP22EYjfIAVbmZQE7lIFIDCHBsVAln2pABXH7MDmXzmD9rlYmGSaVeNGlTrMQtN/qCPjreZ7/CiGDZCpAl32UxFHbAfkeTMmwbqbAtm8KresBd49V6'
    'w2H3W5AU2XJZ4vXvJw6NpMk+cfO7Z09/IPihYBYFop0m+E1a05X//NvvL/+g3dCwg4jD9qIXrRhzpcbRhu4+0/BDvMqmWAZbs3JlqzZRlo4UXSOzhsNwMZBN'
    'M5Q7QVmdzdQ87JcK+XiQGbNNHHCBRwybWU3LtGUAs47PZrhI9QGznv746Nkzhi7uZe/Tjy8vmqu7ysrQkwzkA3uLMcY7zheF557soY+gDLOz7im8OaLbGKuw'
    '8IDXDMtAoy2WN7ujT2xnNbCNaRug+XBBfSSmNxMujPddlrkeMJqSZCN6WR50CdpqJd8QPZriMJnjaSnJiKcaPcwoIk1s8Khkx/9QSwMEFAAACAgAAAAhXA1x'
    'TZ/1CgAAohcAABwAAABMT0cwNV/qs7XthrVf7J6R7ISx7JWI64K0Lm1kfVjLbhtHFt3zKwowsslIsp5+LA1MoDEwwgyyCZDBwEhmlESIIwWSjCCAMWhKLYEm'
    '6ZiKSYuUmkzTlizRQ8MtsklRMLXxpyS7rup/mHNvVTeLsj0LPdjsqrqPc889t66Jv/5tcXpBRN1evNsTqrGn3DOhKjm5FWYyn34q26GqD4TshtHgkWocywtX'
    'xLm2auwI5btCbWfjbe/duTrbUQ0Xf4/21ImjmkF0FgrVCeNsW75sq/xRnB9gs3xZyPZAuVgh+65y63G1gq1kIScLR1MiChxsiI09cVPlqmq3if1k0RHaCm2E'
    'DKqqXqWXYLOIBoFs1lUdptQvo64v7txVNefq0e/Oo/NhdBYInCqPqzj91Slvsv86rvgCR8S1Mp7igK4PL+DCxSnMlMdDE5K40kqsVIUjslAHg96Iem35pBZX'
    'qliH0OVVvpe8K/sONseO/DLM6LWjgOIE7y6qdFKWQlfHusOS7A7IqKhXUl5THbTSeLg40IpTJnPtmlAnbvzskVA5D3ZmMjMUvay4v7yxsbbKrjVK8kWRHdCB'
    '+WZt/YcNJLKnttp00n5JxLUKwhpX4MxWG25NCMRFHoeUJzyJK68nKJ77pahTFPKkGQ0G2G1I0ZanQ+XDaj9re1voya5jfS3DI+UNZSsUvztV/S1iHspGUcSH'
    'rgnt706NlsR7nip4ozhnZqdEXAll4ULINwO1k7iEoNVDwh5Sr45wRsfFJzw9LCEFZHrUCaLuEE/6ZcQN7+WQGIoE+0K4lQU4V6WoJMk8KKvOayzp/Nf2ohuq'
    'ZmiZNDdFe1B0XwyRUjHHMIc521mG4GEpsZFzOUBZuWorwL44Jp+gFcYR5FAlFqrkSYPLYytE5uUTbOb5jCfs7vmcRz9LFscVF9Zy0gAi4IIOBoz8PeOFZe/8'
    'FPtQc22nYErBS3PPUdDW4iihztx4u64B6le4JFAwjPJOgCd2HfQ89XxHxMUiMME+oTKrgKvnAtu0RG6dRhfZxJiFKWEyp/PJWDep0rnEcW8G2rDE0noxedvK'
    'XGrC0sw0njNc0/K5FQVcnGwRyk+zA8gN62zreQFqy4e/Ymn+3fnSDfzc4tD7LgiEi6IOHglxBpEBHZOrouqsCGe++O7nP5zyKN3xweN353fuchQarknfH05l'
    'Qvxl7Se8aQCXUE7CSwwHDjeypXmBEsqw4dVffPfVJi3XQU/iZs6NiyVZOMVHOwNYRkmInzwCj1ie37mLz8BbUT4F/msuEb5mPjwjXlV9jyoJVYszqILwEkBN'
    'q1QFGajZPDSCtDZCM1gbRJvJTMIBXz4PEFMK3uIMIYxIDaUTBRUDvsUb1mNVubRM5d5EO2JjxJaiKV+6RJf4Wh7D8bIrm57y6yB54n1+HMo3SLsnGM8pLpK1'
    'Luf1wMYCyGlxztQxs5EDW9+dL87iZ26Cv56nr2EwvmZXgJfFBfzcIH7DWfLVzjh7TQq5+yvtZDKcRoLbkw9KdVDKrajTStx5mfbBz2c0g5flb4/ErdlPhI6e'
    '2UrttxmkBsDVHA7m9/ueZOjzstvTnxiGZJf7LdU8My1Kvnr7sXaWLgl1qos6iAxDzqzpSPrQUUeaNC6BxC2LEDh8oG2pcv1jYBHVewo8VST17sCR+WMTcG02'
    'EGgCqmlAlj7QVNVuEUyX7KkZl8G5v2u6DZifHug9TCuzdQZYedCMgl+JmD30TfxlyYIKMrjSJNQosdMc9qTT992xLCdi6MUQ+Gc6hiH5C6xOyNMwOfcj2yom'
    'awGFI6jatqqWgeluTGLdPe707KdZxpjpulRYBD6bZykToCmc/vQMconJKGF/h2KB7stwYSOZZ0E+Bf9D+oXy2nChvthM1alCUVlFaGkWRHT38QRSWsJf0Frn'
    'kiuQPZ9gUoH5tLiSY92IQNEXgEF0Ti8cAj0tER8UYbrxcSI9+/wUKecWyPuNV66WnKYuSb8Qq7GlhopA5WidQnpvx5zl18FXz8L3iA306qiXdVnyUsAzPCgb'
    'OoBGOZoWlnTh/QAfxtBBbePVDiGhp4JL0p070r8wjUx32YLPioZTzDzLfQbrKLFjtKA34YJs0BNbv3D1siMcWZIHUWfb1JJ6GnxAmp5kIZG5FWTz0SA0Wn7U'
    '6xJb2Ub2r4GsV3lDNjSVbB8Bjh4WisUopMJqh6hW9DyKJboT0SCaMraGbgAZh9xJOrp+3CPqpB8S1EYRqt+yBPMUg8gq2wXPvWoUOlpR5BBabEkYIm7a9VlE'
    'XMlSEkKS3b6bdjHHPGClXuOZgLaGWTAa28hyMEqNVuTNACVJGIECQWn96c7dTCYpEUfM/+cGImreMpTKgNLH6zaNKhGz1EvqWoX4DAPLWL017RdXakTKjRKn'
    'kKcpkY5hiTTTvrjCtFrd+k0sGiVEx8/Jp15MI8rYSMSVctK0Ak8sYt7ABID8ebxxgj/k9BD7jbzQb30Ufck6Q0chkTeBDkQtX+mWCNnHUpX7uj5ayx5WNZXT'
    '6G3VqA9WK1pEkWxDfAwPA63PA5CBcfm9TpDkBxHbzonFlU0R/4Lx9IwWz97k3uBatLx0k3e9CBEzSqKOLxLW19JkK5wQS7dHU27S52en8QaIeF7MxL9USWdQ'
    'O5ijZejYqCLmNi35El0+Ggz0VnTWE8BH20RmjPVtc5L+mrvGGHUkYkGjLqAGDx90R9eu1DzMsGY9CxPaw8xdaOgYcMzJFOAhhPW41tLKl3CBnVtBat2gySOY'
    'zttHwKCrR1toxhWdqExmaYFBkR9Q71jQKpAFFAOd1Ma8uuAmtkCyX54cxZW9RJKdNNS5a2EO6TPDwohd0etH3ImOiQ+EJiMgL5IomWG/m9x48HTTd2GVriz0'
    'qkIdX5tyNGA2SKH/7NZvJTehM32QsU4LOJYnuj+mimdM6jwLTfs0QkqQMOzzoEvn0KzmC26nePl9Bh35hmIp1M3AQkeMximmSz6JOnglZ/NVOqZdFUVi6RbX'
    'zzS6OY/EmDaTaxojK3OyPsTwTGoVeoBsONkjBKES9AgSdXuQfkkaUJswDS8gJuM0oLbrfDnzgQ5koyrp4shdrZXJPDQ0Ix4mF1567BYP8dXk5KQwv/HJrNT3'
    'OOKhdUX1f2+oJugs/Gu9Pxf1Alpy4pBRRd7vH+ldj35PJ/ifOJmN39+RjwPSBnmg9LBIN1r9LOeAh7RMhuejez+trX//9dra91MrP/68+jXjDL0BkTloWeSl'
    '5aLY2Hzw7+XVzesba/cfbK6srZr5nYesPpi7rVnYZeBoySlzrqxfpsmrjYQmX8Gpy7JBeXK3R+MvfGwP9MUH8cCO4SaS2my8+PLu33WXc0zN0QZ04QIif+KS'
    'btNMkvY5PqjcIrzFVVfPpt772m3ppsG22cbcqkj/SAaHeJW8YkEDhwdFAFEdBGQE6x5C99iADz7HmyPa5t2Ys1mzoqeaiWkUaB1QoyGtihljar6vImjqjkk3'
    'gyXoFkuVvzdrMCQQk07OXL0Y7a1bFusONiWTsafitKGl/YzZkVRMfWirIL6awX5EByT56F2zmGsfWyDsAEg6Il5BA3upmSptQ7JAZEMRN345Gob2rMMoQu9l'
    'q9v4rWd5FQwQXnONmS9vrD1Y/9fyvbX1lW9XVr+6v3H925XNe+vL3yyvL6/i+ez07I2Z6enblMZGzhYaCNU2ZqPTzz+78+elz7Cx9mLlhx/X1jeJOZ+e0QXG'
    '1Szq9OhCuK5viWB9MoBUaTDlW6LRJVHdmZmmbno8vHpLlF4beenNF5FfcjOjFY9Ls8te6WqLyPwPUEsBAhQDFAAACAgAAAAhXMCF1ZJJBwAAww4AABsAAAAA'
    'AAAAAAAAAKQBAAAAAExPRzA1X0NvbGFiX+yLpO2WieyViOuCtC5tZFBLAQIUAxQAAAAIAAAAIVzPMoo+ewkAAOgbAAASAAAAAAAAAAAAAACkAYIHAABMT0cw'
    'NV9NMDRfZm9ybXMubWRQSwECFAMUAAAACAAAACFcQae4M6QKAAAgIQAAEwAAAAAAAAAAAAAApAEtEQAATE9HMDVfTTA0X2xlc3Nvbi5tZFBLAQIUAxQAAAgI'
    'AAAAIVwFXnF7IAYAAMkMAAAkAAAAAAAAAAAAAACkAQIcAABMT0cwNV/qs7XthrVfQUnsl63tlaDsi6TsirXslYjrgrQubWRQSwECFAMUAAAICAAAACFcDXFN'
    'n/UKAACiFwAAHAAAAAAAAAAAAAAApAFkIgAATE9HMDVf6rO17Ya1X+yekeyEseyViOuCtC5tZFBLBQYAAAAABQAFAGYBAACTLQAAAAA='
)
_EMBEDDED_ZIP_BYTES = _embedded_b64.b64decode(_EMBEDDED_ZIP_B64)
_zip_path = Path.cwd() / 'LOG05_실습자료.zip'
_zip_path.write_bytes(_EMBEDDED_ZIP_BYTES)
_log05_extract_zip(_zip_path)
```

```python
MISSION_JSON = '{"code": "LOG05_m04", "steps": [{"title": "권한과 조사 근거를 구분한다"}, {"title": "AI와 12개 근거·산식을 만든다"}, {"title": "3사례와 반박을 직접 검증한다"}, {"title": "책임 한계·조사 요청을 넘긴다"}]}'
# LOG05 학습 기록 도구 (표준 Python/Jupyter 환경용)
import json, zipfile, io, os, datetime
from pathlib import Path
from IPython.display import display, Markdown
def _load_notebook_widgets():
    from importlib import import_module, invalidate_caches
    import subprocess
    import sys

    try:
        return import_module('ipywidgets')
    except ImportError:
        print('이 노트북 커널에 ipywidgets가 없어 설치합니다.')

    try:
        subprocess.run(
            [sys.executable, '-m', 'pip', 'install', '--disable-pip-version-check',
             '--retries', '0', '--timeout', '10', 'ipywidgets'],
            check=True, stdout=subprocess.PIPE, stderr=subprocess.STDOUT, timeout=60,
        )
        invalidate_caches()
        return import_module('ipywidgets')
    except (subprocess.SubprocessError, OSError, ImportError):
        print('이 커널에서 ipywidgets를 설치하거나 불러오지 못했습니다.')
        print('설치 가능한 환경에서는 새 셀에서 %pip install ipywidgets 를 실행한 뒤 이 셀을 다시 실행하세요.')
        print('설치할 수 없는 환경에서는 제공된 Markdown 작성 양식을 사용하세요.')
        return None

widgets = _load_notebook_widgets()
if widgets is not None:
    MISSION_DATA = json.loads(MISSION_JSON)
    CODE = MISSION_DATA['code']
    OUT_DIR = Path(CODE + '_results')
    OUT_DIR.mkdir(exist_ok=True)
    BACKUP = OUT_DIR / (CODE + '_record.json')
    fields = {}
    for no, item in enumerate(MISSION_DATA['steps'], 1):
        fields['activity_%02d' % no] = widgets.Textarea(
            value='', placeholder='내 판단, 확인한 자료 번호, AI 제안과 수정 이유를 직접 작성하세요.',
            description='활동 %s' % no, layout=widgets.Layout(width='98%',height='130px'),
            style={'description_width':'75px'})
    for key, label in [('first_thought','AI에게 묻기 전 내 판단'), ('ai_request','실제로 보낸 AI 요청 전체'),
                       ('ai_response','실제로 받은 AI 답변 전체'), ('source_check','원자료 대조 3건과 수정 이유'),
                       ('final_result','최종 작성 결과와 근거'), ('limits','미실시/미측정/확인 필요 및 교관 검토 상태')]:
        fields[key] = widgets.Textarea(value='', description=label,
           layout=widgets.Layout(width='98%',height='125px'),style={'description_width':'190px'})
    status = widgets.HTML(value='<b>아직 저장하지 않았습니다.</b> 작성과 교관 승인은 별개입니다.')
    restore_input = widgets.Textarea(description='복원 JSON',placeholder='백업 JSON의 내용을 붙여넣으세요.',layout=widgets.Layout(width='98%',height='100px'),style={'description_width':'110px'})

    def payload():
        return {'schema':'log05-student-v1','mission':CODE,'saved_at':datetime.datetime.now(datetime.timezone.utc).isoformat(),
                'answers':{key:w.value for key,w in fields.items()},
                'status':'작성 기록 (학습 PASS 또는 교관 승인 아님)'}

    def save_record(_=None):
        obj=payload()
        if BACKUP.exists():
            stamp=datetime.datetime.now().strftime('%Y%m%d_%H%M%S_%f')
            BACKUP.rename(OUT_DIR / (CODE+'_previous_'+stamp+'.json'))
        BACKUP.write_text(json.dumps(obj,ensure_ascii=False,indent=2),encoding='utf8')
        status.value='<b>로컬 저장 완료:</b> '+str(BACKUP)+' (PASS 아님)'

    def export_json(_=None):
        save_record()
        raw=BACKUP.read_bytes()
        try:
            from google.colab import files
            files.download(str(BACKUP))
        except ImportError:
            print('Colab이 아닙니다. 저장 위치:',BACKUP.resolve())

    def restore_record(_=None):
        try:
            obj=json.loads(restore_input.value)
            if obj.get('mission')!=CODE or not isinstance(obj.get('answers'),dict):raise ValueError('다른 미션 또는 형식이 맞지 않는 백업입니다.')
            save_record()  # 복원 전 백업 저장
            for key,w in fields.items():w.value=str(obj['answers'].get(key,''))
            status.value='<b>복원 적용:</b> 기존 작성 내용은 이전 JSON으로 보관했습니다. 다시 저장하세요.'
        except Exception as exc:status.value='<b>복원 오류:</b> '+str(exc)

    def export_review_zip(_=None):
        save_record()
        target=OUT_DIR / (CODE+'_review.zip')
        with zipfile.ZipFile(target,'w',compression=zipfile.ZIP_DEFLATED) as zz:
            zz.write(BACKUP,arcname=BACKUP.name)
            text='\n\n'.join('## '+key+'\n'+val for key,val in payload()['answers'].items())
            zz.writestr(CODE+'_work.md',text)
            zz.writestr('README.txt','검토용 보관 파일. 실제 평가 PASS 또는 교관 승인 자료가 아닙니다.')
        try:
            from google.colab import files
            files.download(str(target))
        except ImportError:
            print('Colab이 아닙니다. ZIP 위치:',target.resolve())

    for i,item in enumerate(MISSION_DATA['steps'],1):
        display(Markdown('### 활동 %d. %s' % (i,item['title'])))
        display(fields['activity_%02d' % i])
    for key,label in [('first_thought','최초 판단'),('ai_request','AI 요청 원문'),('ai_response','AI 답변 원문'),
     ('source_check','원자료 대조'),('final_result','최종 결과'),('limits','한계 및 실제 검토 상태')]:
        display(Markdown('### '+label));display(fields[key])
    buttons=[]
    for title,handler in [('작성 내용 저장',save_record),('작성 JSON 내려받기',export_json),('검토용 ZIP 내려받기',export_review_zip),('작성 JSON 복원',restore_record)]:
        b=widgets.Button(description=title,button_style='info');b.on_click(handler);buttons.append(b)
    display(widgets.HBox(buttons[:2]));display(widgets.HBox(buttons[2:]));display(restore_input);display(status)
```
