```python
# @title 📦 LOG05 실습자료 자동 준비 (ZIP 업로드 불필요) { display-mode: "form" }
# 학습자 화면에는 결과와 업로드 인터페이스만 표시하도록 구성합니다.
from pathlib import Path
import os, zipfile, stat
from IPython.display import display, Markdown

LOG05_PROJECT_DIR = Path.cwd() / "LOG05_실습자료"
LOG05_PROJECT_READY = False

def _log05_extract_zip(zip_path):
    global LOG05_PROJECT_READY
    zip_path = Path(zip_path)
    if not zipfile.is_zipfile(zip_path):
        raise ValueError("유효한 ZIP 압축파일이 아닙니다. 실습자료 ZIP을 다시 선택해 주세요.")
    dst = LOG05_PROJECT_DIR.resolve()
    dst.mkdir(parents=True, exist_ok=True)
    with zipfile.ZipFile(zip_path) as archive:
        entries = archive.infolist()
        if len(entries) > 20000 or sum(info.file_size for info in entries) > 1024**3:
            raise ValueError("압축파일의 항목 수 또는 해제 후 용량이 허용 범위를 초과합니다.")
        for info in entries:
            item = Path(info.filename.replace('\\', '/'))
            if item.is_absolute() or '..' in item.parts or not info.filename.strip():
                raise ValueError("안전하지 않은 ZIP 내부 경로가 포함되어 있습니다: " + info.filename)
            target = (dst / item).resolve()
            if not target.is_relative_to(dst):
                raise ValueError("실습 폴더 밖으로 벗어나는 파일 경로입니다.")
            mode = (info.external_attr >> 16) & 0o170000
            if mode == stat.S_IFLNK:
                raise ValueError("심볼릭 링크가 포함된 ZIP은 허용하지 않습니다.")
        for info in entries:
            target = dst / Path(info.filename.replace('\\', '/'))
            if info.is_dir():
                target.mkdir(parents=True, exist_ok=True)
            else:
                target.parent.mkdir(parents=True, exist_ok=True)
                with archive.open(info) as src, target.open('wb') as out:
                    import shutil
                    shutil.copyfileobj(src, out)
    files = [p for p in dst.rglob('*') if p.is_file()]
    if not files:
        raise ValueError("ZIP에 압축 해제할 실습자료 파일이 없습니다.")
    LOG05_PROJECT_READY = True
    print("✅ LOG05 실습자료 압축 해제 완료:", dst)
    print("📄 확인된 파일 수:", len(files))
    print("ℹ️ 실습자료 폴더:", dst)
    print("※ 업로드·압축 해제는 학습 수행 및 평가 완료를 뜻하지 않습니다.")
    return dst

# 실습자료가 노트북 안에 내장되어 있어 ZIP 업로드 없이 자동으로 준비됩니다.
# 내장 파일 5개: LOG05_Colab_실행안내.md, LOG05_M07_forms.md, LOG05_M07_lesson.md, LOG05_공통_AI역할실습안내.md, LOG05_공통_작성안내.md
import base64 as _embedded_b64
_EMBEDDED_ZIP_B64 = (
    'UEsDBBQAAAgIAAAAIVzAhdWSSQcAAMMOAAAbAAAATE9HMDVfQ29sYWJf7Iuk7ZaJ7JWI64K0Lm1kbVddbxNXEH3Pr7hSxSMmDtDAI2orQCJq1RaQ+hIZSEtK'
    'qFECRfShWiebaBMbYkOcbJK12RTnq3XbTbyOHcm85Kf0ce/d/9Azc+/aC1gCFtl3556ZOefM+DNx6+vro5eF2i7FhaY8KQhVbMTry0JVHTkfjoxkM/rE5PP8'
    '7KN7+fyjzPSTF7/ck7s98UV+JndPbZSV7Qm1EUYtX6iGJU9toRYtfG5eREC10o7XbLnjKb8Wr3Qzv00/UTUcc5ty770sOrLYyAjl2+o4VL6n6hXZ6oofbn4T'
    'tXoiagOWHVcP++f6dwhVb2q4qmaJqBuonRCfVZR9JOLVUL5cI5jy5T9qPVRvArWPN6orAGNCjYxlRNRqx0ttce0mUvg7rvomcbVpiVtTc3P5Xxho7T1lB1Sq'
    '1hVqvgmsIi6VZfHgrKNqrmx2zzrXblLOcn8n6uJMrUcvyoOe8nGrXxjcijC2XLX5yuB9Uj8Uycdja00d/33WkcW2bFl89bserpbBdirAtW9uini+IVfWRFy1'
    '8Q6wL6W+j9rlKLQEgiDSv11Zd4U6DeUbD2DNPfVd1FDOu0KeWLJECKuyFVJEnYaqL0p/N666nxbtYkZ8+9V3t299L5C2qpfRepRAqPWKKtYY8hEwBVSwudyv'
    'U5OzU3PPZp7ykXc9ZfORuEqNxr+Gbun2bnaBScSvumq/gp6CNUKubqKggrhx4ono+D1RCQEJgO1Rk3XXEZAqK8v9aHEV51utpMrmlMtc1aWIOr3oKOBCxZtV'
    '9Bf/NX1ue+qPRaFsP15g1HL+IDotJHW4lNH55aYnH0znZvI/PZsirEM7CX2VCSyCREEBf0CCEGxN5x06oB5dDvLZeP55QCnGWyVV9NG4fScJ8LEg7j58ceFG'
    '/vmFuw9zXOaoZcdbL8860bEVL/n0RDN61Nsf87OP53R7dSH6MVA2ioszYEO8CUbt28qzSQaa59w/t0WFizpduWdrXXTVXm0oSy5nOGfoWLmOKnK7SKENi0XN'
    'RRfxawd5kXihiSjwiHUG0LdZ+vh61hxVG03qrKqvybfLV8bORYFlAl8dPWcCM6KTQ7VzNAyRkEuv+aY3R1F3GQX1y8pxSbamxvz6fFPWS5yraf7KKSXp9bi1'
    'QzxEqKUSK7Belu9KDAtaxPlDcHsTomqj45Zc2VW1kPSmGhWOsWnjODGXSOdXh9bw84yYGIeMG2orxP0QMLkeIdHOOjE6PqkcD8yO3apWuHHbzMOnj2c0ZUPp'
    'N8g+qlS9MucJ4XRLEg61FcCd+g7ed+LYteGsxjPkfkl7QnRcIo+WxWaatXyGXBTtg25jOxgDOwVnGxrH5Hoe/9V3TN0uPLm1Wr7abkLxn+UOjgqCV9/9z9rk'
    'XBxb7vUJghvibTtlppq4JhujHvkn4p6edQYdNqWQFYjglPX0gSlQsWulJHUg05kNMh4Z56YMMBKnNEzBeD6eNkSFQGhV01kcUksvBTtbjSdG2lV08ULHCBZJ'
    'Ks9nVtUdxGUNGFkVvSiwh1CfDHLBMDaueDg3XA5UAR4D4vaX57Nnndt3zo9xpxYK8QLnDixyr5lUpmThG+iGJYgn3KGOfBPb5BEMZygxQWVlp3+RaUWrTeOO'
    'gKZnEN/ouNrj/Cpw0uu6OkNxG8aR7FYaqB97VnrIDarFn+81aXx/2A2qyYbLTak6ifeRwDHZtw6HavEKEoFVAd7+coKB+n3j+4lbnN6Oh+w/FKs8tuOVnvEV'
    'Djzw274uyZL4Xar4k9n8z1P32cVplhVPKQW1iNT2amyYvNWwgzPlAJWNxpgVz0f6Nl5F2w5Sc/BxbvbRZO7+02e5mcmn04+nmLW0QZltj8hUIKn3Ky//pU4w'
    'H40YMU8Ag96kpA3qC6ab8YJFmj/o0sgwAxSRMbkTqg8pw8hViOkKsh0bhUIFDXzMroqXiPjESfYbdk0KTY8NdHAJVtYQRCDvIDVWCeN8kDaobRz0EH+dODvf'
    'jI5Cika+TDPlBDakFhwQxo5oBL1mPmp6DOYmNcH3YAWUIqmZGcNEw2ZgD583E+OIdVm5BSyWvBNh4NTSm4BZDiEuPQB4KYErn7omJy2sVDLkcqwmroQp/Ooy'
    'gMDax2B2w/aDkewowFylPO5kf78zzhkaD4G9+DbaCwA0lpPFp97Uvdb7Yndn6Nqhly+QDpPBGhvFl/31UsIxK95F/RENePci6j40JX2UANGaS7PmksjGr1yC'
    'pr1Ab4LM9MD8SEiW/GTntAZv0J6oSZPM4pTlpRF/4kTJ533nxBgcagXZrN4ric0YQoWQ28pbopi4dNaZ+Bx/r3C2LEMoOTuKIUK9Y9d13I9WXs4GxedtRBOE'
    'CldQfkUkW5zxHK06rIe6bgNtJQulNuuBVS0mA/7Zk5l87sHcBd4d+j/BDH/TayhiQliOqS7xoo+KZzDclBIODmAKukDUN/rRtHXIU6jYgECAWJNME9SMJwx9'
    '/ePELFZ69cPx1Oon3wbkJnXbgOF1FH3nWZnY5qAd/wNQSwMEFAAAAAgAAAAhXDIpGURTCQAAeRoAABIAAABMT0cwNV9NMDdfZm9ybXMubWTtWF1vG8cVfeev'
    'GMAwIKOSqO8GAvogoGhrIEYLxE0eisJxm7QJYEeAnbQooAdKXgm0yMIUTEoraZcla0oiXQpZiZRIovSL+k+St53Z/9Bz78wsl6TsGkWcJ8OCSc7O3Jm599xz'
    'z90b4sNf/3JmUdyZ+alQ5R3lnAm1u6NyfiqlLjz1z02hNgJ16clmf1l8+uibPzz68o/pB6t/nlm89/lf7j/45v7XX65+dU+PTz/87NNUasrYUeXCVUcGDdj6'
    'PlNU1UC2HJEWYeApv/uTldvfZ0rLg8l+H5PPHVVxeDBXUxUvvXJbyJd15bsievY0DDIwqfYzIuwGqpYRdnrqweePH69+BUsF+SIvj/pC+a/CVsWYjkoNmcvK'
    'XG1aqMOCbHWxUaut9h0R7XhkOwz2lQ9zG2110KBv0dYFFkUlV+01RVTysFNWnuRje235zBGyEJuV33Zx2rB1Acv42q2qSumqQ/NyPaH2tpSfV76DU7WVV6EL'
    'RKV6+G8XO9bD3rqxkkrduHEDQ+50HAfHg+/ErIA/em353CMHRvmCzNXJBbKKGK2ZR2LNPJkIW47a+vtVJwwK/KlKWXhJ4EjqJENXvwg7+IwO8irrCnVQVOen'
    'mNWpq638LZhRlXW4gW6NkIVnbQzJl5uy0uO1KniF4HhhsB7vKLAQZibkpYP1ONHU1JR47f94fncWPxJ/NDQ3PjQ/PrQwPrQ4PrQ0MpR6rWvn4NoYWUKdAU/4'
    '8Puq5wIbE4NHc3ABwifUk3VCDMe0bRbcoijoqUK562rvVKwlrM4K+fws7D4dGpwbDOrtJpSbpVg5QINyaxQIeZnVzwxw3sq3OGD0xENk8qq0OeYaVdsBCGzc'
    'Rp/qMAMAOEQTCNkvIVHHZmnkjS9utSnl3tbz8wRqz1Flx2azxt0ywqf2CphlHsvgkFJSnayrs036NiGEuDUpwvMAkI5KyHKCcTkrt4vI0yJW0TczS233cIfw'
    'Mk/MsVcIz5HEG83o0DHnZZgznGPTcLbmCva98urxE5DXoviZ/n7TTKMLIonaCLwwrGIIrddNpe7OXnXuztF5fvvx1NwkMI3fC/z751OzZOLbjKyCX3YAu/KR'
    '7DmMLo4h8RhYiM6AkKnSttq+sIRzdxF2lngurwJy4KkmEJnzQRJEhNhDsx2uzXjdxdPLIoFUBzZBi/pIqlyU/3hK7qQ9wVKaKkx6G3YdcF3i9HoeTMvtIyI6'
    'gItjc93Jl3irjSZzkgaAbHYRb04fc84nWdiFV80M2hreAAXL5zWewGG9dgN1/i+D77DVj5mL2C5X0aF2+IiXjjxuwoJsBMjuSxymnaUVdDSvQoexuLxsqOqZ'
    '5m8Xt79+V65XYpgwyFxcwegC4UUTOw0cr7NDEziFkEI5oHQ7glSCu1E2MRBfboSDw16eD9UGifBtkB60vU3qtC2j7AcYAjSaXXYv+GWNg87jfjvafYpP7Xes'
    'oBsbIngL+hmjhXcwQBkHtyI80QHVufNMtFWJa7otjFMx4eQ8kPSy+G63ytHIXchWhtO/GxDX8QMzqvbaspwXxA96Ncys3DbEIU+qYRc+29tEplF+aFpmucIZ'
    'YVIER0rqE3Y2AQ2411GhCVmEi8nrsohUWeao66Sw2enxQK5GQXEpWGs2BcNOPzwLhHR1lnNSUx5GJQeJyiv1xJGIpUZcOvZJNJ7PRPsg0pcN8DFtqYGtZYLQ'
    'sEE6obz4/QR89CzrxdfMIjLyuwyqRhGAfBOiUm8AwcgvqkmMAbo5RZh1nOC7QMKwPw4LcWxAjxsBnJqrAemDajbgDFttBYk1kkyUHCTogBl5xBcJXOJ7e7F3'
    'kRcpXES96IMBxDyJMDoUaQ8nvutAfWrga+eTMjV6VsOECRrhyXkxspjz5EmZK2YsevVtze1ZRKMgxwyX0iWe65uVjCMcYkU1Fdc4BMM7exa7cK5pLpxK9MRn'
    '4sG21C+waw10/l+AEEXoXON2gT1myipnGhdGq9mi0ingHmdtjGabrFQqXtbx3B79NysffZT+xcrtD9NUDPmGacAHGCHNRiatMyw7WVn2wwD+jTdnMkK/ZlyY'
    'bKooGL6u4JsQNn2GhuMSqcV3jdmeZFLpVDdoBXJNBSSYWWZCtE2ZvtygX0OBK4Rtai7iOmimaJK13tK/kmjiEV3VqVc01VSfRPsd3SKrSVtjiX/3HaZDjhvf'
    '3MQVPJh1zT2niePVeZvZFNVtva3lJVNn1gUHJOrxlPidXvx7WG2TPh9WzNFhVlYABfmiLrt5eXwqxHeZ4oj4X6QOl+STVvgLTCvORIT0PSwC6QcBNmA2QZ+p'
    'Xe5Xbw1kMC5NslPMzszcJPEaZCZtk7dEfADnQy+SrBvVOUa6G1LjDqWc5UZYt4/oRs1FtOKBgkOimi2GtctwM0TH+WDmZsxC5lDX6HeDJFY7fIpol5Q55pN3'
    'P/nib+lfrf41/ckX979eNvUA4sbhQm6LCFegON00iSdJ0JZ6rEhyCHHjqLJNINMAPAavbtUZjlSpEtmbVGpMlRrVMY2Y3o/BKgtwZ41d/DwYk4Uar/uefLZP'
    'rbs6cVhcwZA69hMtvJ5CRcggfpSJ2BValel6y3QSa5430wp12/R2R+8yy4jVO8Di3ilUCD75nQUL43pfVXDZVoUlPEBTWxaLGtr/2VswoOY6adeySuULmmMN'
    'y3tzKRk05DlvERvmNyey7Fp8HvVZbHHh074EdHieJaVjn0oapa7RJUnhJc+zkctvdxhXWsGl1UmW1Qgib9WvxlRa13WWxeZVDftXlTdtadXQTLz62aVx2aBG'
    'i+WDpiAWXzvoa71rmwNNAInM+NPqo4eP0/ze7R4ic49/Tz/8jElNY9oUwP/5vm1SjJuZfP0G3AsGOElCP6zRP+Bkbggnc5rZXISwaeLDq9GHV3YMjbAnkawW'
    'JktxHAWlzoarig3E/ZDyI8E4iR6NBY3mCYMI4mhSEKZxS7wBtK0LtyqDvoRB23JY+AxZeo+Zd4+Z+SHMzDNm5LNNefzKFBAmCkPsVS/sdJldrVxn1Iy+ZXPj'
    'BsLQTvJVDGLdc0DIcBqz7sh74iR3GzzpHjpy+zggwph1pP/KNMm65We4WoH6Hjk/EnIWhpCzoJHD5Yz1jK3NuhEhVWPhMjejeqgmVoFddZaUv4PmB47Wrzpi'
    'otDlR+sSkhf07ieumVrfAlB3Prjq3JmdocOyGG6/h8CPAoH/AlBLAwQUAAAACAAAACFcUm00Fx8JAADcGwAAEwAAAExPRzA1X00wN19sZXNzb24ubWTtWW1r'
    'G9kV/q5fcSEEbLAtW37ZxdAPC4XtQkO/uN0PpXiz3XYTSBpIs32BUGR7bGRJW8sbKR7bI++4kWM7VdixJVkyKF/cf9L9pnvmP/Q5594ZjSSnmy6kW0rAWDN3'
    '7j33vDznnOfO3FA//dmH0/Pq1vR76p/Zsgo3WlTwFR3UdfCl0sGW6nUCWlvBz6ZuZJV+cUJVF3dNDGCao+tNRbVtOs6qsFjShRNF+VqY72g/hwsV7ue0f5JK'
    '3bih9DcdWncVhIW7p6mULlxiJ6pmVS/I8g7ke+9RzqWNQ8jTRexVb1K1oxu8lQ5cs2+gD6tKXzjkVBWdrUMDRTsvw4pPB+u6kNOF2lRkhN4ukZ/rNbqDNlC1'
    'qXrnl3Tp6nw53ly0op0SOZ41Sx9hYcOhjS9ZOuwzZtobfenro2bvvGjNDisi7hrjqeooPO01/InIRTp/pP0jebC1CQ0gU+lnJ7pT1M9f8jD5FZgbVjy2J5Wa'
    'VLRf0o3OVZvWcurDu49U+NcOOWdqlgpeL3C+zZapUKPOYS/4is2jiqML7rfZCq9sefS3dUWrAV14ut5dVJ88/OLTh3d/nb734PPp+eXf/OH2vS9uP7r74HfL'
    'Znzq/mefYF2v0YIf4Ytt3ogqOb3apN2s+uAj63A7xm7SJ13y4aDqK8q3bBQk5kZtDi3EQVZJPyuykEiq41GhCu3DbVx4sFfRXpPcFYWo066jOOzuChsCaTc4'
    'WFODwtSMmlSh6wCwii6b+onHURBMLwPTy5TzsEnoVszDZXZTvjV159H9e0q7JZ7MaIHIpgDDCA8rpxGWyHfoHOMiRwDTKmFuZIsBDPCon9f7q1Kp1+qbgb5x'
    'qFVYyI7Nw0J4VP1JzeknZ8DohGJlgoCXAynslfDppt7aJa+rDz3rvPFU6rGFj7IScL+2QvtlXNBegFzCBRzJYXzM6RdWD9VjrJqcnFSv/Y/nnAxr2LRYpMo6'
    'xhN//HQg24ef6hfr2r9kt+lC/aod7gLInZFZNo9HFiPXG/7QsPWlXnWnBnEjsbdBbziRQkY0O8c8Yq/IkzGTzFftXlCSX+AXakquHWev2rrR6rXxG+4V2XW0'
    'V6bzl5jVPqGN4jj71F+hvVMB6WHQO2uqyFpZS8ErJdm4Eu+osBBixgCP8Tdy/NLMiEuWMqNDs6NDc6ND86NDC2/sWoapARdwi0LrkI+fahdVE7k61n+UgQsk'
    'cYA8FGja2ZA7WXA9ROOlM8rgfWAw0x80242Rm+NYAU85cmscCH2RM88wpUme/38F6lkGtedwY+O2Vcta3C0ifFF/ksc62OfuSccr3AdxNaaUGkfxOA8AaSnQ'
    'DOODHFctfV7GKr6ysyh/CRt6F0U4HWK5kdFqPdx3rL4Cc4FzLBrONiVRfE/eSfxEpdW8+pG5vmmnWQNNDYX8LHAjWSk36Igv9YtTDr11Zmo4YPO9wFOJ2lg4'
    'QQJbn3CPBBVBh/+uWmka6VMuy5BttH3cLw5LM39ZWtDH8tgEf2IkzycG4x+7xbERf610UJy4Ekif5tTIoBowbrmdVjsCOvGAoQiKmsxYhEocZ2kn4KgA4tw0'
    'mUBdnNLhmVhfDti+pAbMAi48tBvOI9NA0TkH9HqzvI1U6cfZESa0GlxnKjqqz03pSF86wpladfgfMoQzVfJkpEWkLIBnm6L3eUWozpOgP3FQeMpwCPjqPBfT'
    'xyQ/GLCYLspSaJwaTBPKsDRz1V7KsC9//ovJzAQKJ+7n5P7HkzMs95ssO5FhZPQXV0ihYEvQ0NkBVruY2aC0Qs6CzJVVKE+WCBeq6ETc6Tl+wn2QW+Lcp81I'
    'wSjcMcUwKtFBWX+9yTnLe3K4pR/ZHhLRB/aC8ImE9maeSQCBRuFECsB1mi/IVoAUNz5TZXQdVNIT11k913KQC0fbGbw1k7zjon5SkwlSO67dgM7/biHPvDtq'
    'j9xS+VRhk8CJCZOrTwObEYJ7i3ZWJip+MeLhb+Yw1+5qMDiIbhYHsmpOLBEwL5wkSwO6wsouBCEoHkBrapa5RnviA5FRAxEWzovfzqHUFCcivcC2eJupeipl'
    'ydw/diyXU2JUtNbkCKqYwRhGBiJhSyFSJMqNSPAUB14fuFFhOep+8JEg61nXegNx7Vdda0HGHOmS7FgWoWP42zZM4C7FbLhbjgxYiHfgQwv6E5VPodE+p3Ci'
    'dCUCnSiEVldMkUZiox8HQRkcwGyAnGsBwikmizsbjt4akjRgzaxYo7fW9fNXtlCJc03sgdheuyNhRqvZ8CN7hiueK3lZx1Eq4t0DZQ7xAcGPCqCITl8LImup'
    'RBQu7kJDuDLn6Oqrq3YC0eJIm/GjNs0Zm6SYScIgvbFpQfxv2k1kSGaaLjv9o8NVGwVoWxda2FX8nI6da8BkuhhuJemYosqJipt8hc+Yt96/at+amWZAaMfF'
    'gb6fEsZlfEzbdbjomBzoH0OBDIe5hSnJ/TMSwIzIrzQNTZFKk0Pv3Uyk3KT6pVn8q/gdwgDzMkdmaBsfkQyGB0gkMwOpkCbFLDUYM4cfWCxnH7EYTQe/cvIZ'
    '79MppgSIt5qZnr7JJCjITkR8YAERtcyCPTNcyiIKYFAqHfMgJ1AZbviDbV62GITiIKlmdd6fvimEljuyVeoaHsj8zuKq3/zz3FnYux/f+XP6Jw/+mP74zu1H'
    'i4r7DwoKnBPu8YHHpookPM5B7W7vLOBb8Bw1yycas30EFqywLw8cP1yryvuKoebFmDBSjVqJbBfHGzSjAxRL8Dszd+5TiTySDmDP1CZzYy4iXUaX4M6auLhP'
    'FIZeMoiSYqtrE0vOg6aaRuCdmZKXXOb596jti+qHru2RDHpeRdwkvSz3Sr5hAFcK3Q6HUGJvDhdpAsXloofoRHXCxD3NtVgyAlndZACzwgBk/3WagQ/avxVm'
    'SKA+Zb5TQ3rbMsFcJHqPc02Pjvh7jN7fPnh4//fp/qsauZ+6/5kUHoM7PqxVu+o731dNqFExE6/fQChZAE0SOM4k4fGfN85F9T/VON9B5S1CZTYJle/DShZH'
    'zmE/OCt5B5i3CJi5JGDemPItqv8K5XsX+bcY+VTiu82/+WyTtm/j5OtN/PFmSgjKYROW8SsO/h7UokKdGeXeqRnkFwEmPvkyvyXiz1Px5v8CUEsDBBQAAAgI'
    'AAAAIVwFXnF7IAYAAMkMAAAkAAAATE9HMDVf6rO17Ya1X0FJ7Jet7ZWg7Iuk7Iq17JWI64K0Lm1kbVZNb9tGEL3rVyyQa+smaYO0vuUUtGiaAuk9ORU9NCjQ'
    '9gdQNmXQplpLsGhTDilQiWTJqYLSEmVTqJyDf4qP3OV/6JvZ5YeSnAhR3J2ZN++9mTvi+6eP7z4Q2WKZ7y3Fo2+FOnmXe5FQ7kgdLIXyHLmTNBp37gh5aSs7'
    'vLnK5la+hw9Oe2r+TijfUW7QaNxagfJsOfDNhfLfVLV8oQYd+aat+pbANULtzFRki7zdke5UhbaQcS/f9VSwlsNAqMRBOH59aOOIGvlb5gL6O+97Kkxlx5cH'
    'Pb7JHSGZZUcFQ3V6LrI0ViNLqAh5ODdX6rKXxZY5pMJE5J6NlHMvoExwgOJkyxmqyj0/W0QmkpBzJ/dTOV7jqqY6TmpZqJMO4Nnj+IOxXNlC2RfyYIxobpDF'
    'TTwDWw1sPO1Avm3R+1F+vC9U36YicE5O2vJopCbAY+Jsicfy6CJL97PFGvmdZ/Nz3I+ULAqPzOTZTCf3GcOnS1ShL6Rjy/Aa9+821SWXouL3wF8nuwzU61aJ'
    'MkPPWHhRNm9v3VphoyGnaxXRqz66jOrofzVx5CylXIpAuDbyCnhAiCgwbWfw0lgOQ6QtXUe6oy2QhwDSYXHQLuqYDOk+lKy8A0SrPjedEF9mF+BGmFA1FPNV'
    'x+ShJk0VdQWQRp9NHnKRmo4ylvQiAb4msZsr3I1umMZTu7OrdXYRcyY702zVNPE1p99OCc6C86eJ8puNxufiyf1ttLiL9gIbBpz7xGUvllkcUDJpTFVSzVnc'
    'J5yRHvAq0NOk1l2UnYDq1dLhVOJXNSjkIc6vqy+4GxpvediSZ9ciP9zHS6IhETP0IS+kAFCZjHVcKfuvtnEX/zOMoRSRJZaKOiYjjQcSs9TwAgrCtaOuWvki'
    '/zvNj8esv5WNNtZai4qBplAXLU3viZct8JRzDyJB0mBCRAcJdWbMR+2mtB5QWlSbpo6QbiLd1c0VnEe5UflTHawQLLtso9yyNcxNFmAUoAWVWnvQHAPtLmsJ'
    'lzLtGOHdWoSZPnxrAe5EhmvmE51LlMcKLI9zClV8poC7lAuL9WRYE/rZfEY3nccfKGlnJgft2tkPJGkXzCaL2YDo4bbIfXjITKhVIo8CcNzoSo3+KumixVsp'
    '4T5YRgJij53/U4tTiJSlk12lIn+F9r5tyWglCNABmRdI8rqlHL8oLD9tox+gBh8mFAobxdm2lfd7dag0T/UZo13+ELIcJhv2gBS7QBHFW+okpsSrZAspSTBt'
    'N8yP6Yw8TzhndlcbQ6EIls2vYQ/sNodwyGktCppMQ8igR0rDizPA+Wb9SU5+vS1k5ODbvG1xRldT5V2Tb1BCiPVg05wKafiUXNUB1G24RiFx1TAGqwhQ2aWi'
    'fBl7QNrxIQ/oZpHIsc+3BlExajbS+mabkMi7ATeMoGMBbFgloVWZ4Sb9jXnoKzR+nH0v5kmbDlmm57WI9+5uYx52gBXrWs4S5AsLxLyu8RidixgH46rkJ2xZ'
    'm43WF5QHKzfD+7EetbO0UDrjh93B7det+clDoZwAgzT3vWIbAQUwshuNF7xjPH9y9+Hz8hvd8ef6y61f/nz56wuueBio/2ZCpm3oXZ3GEA4jcHJdYRWgTnqI'
    'n+6R3rTDY9FgzfAP7oJmXYkmVopugB+yU5GP22EYjfIAVbmZQE7lIFIDCHBsVAln2pABXH7MDmXzmD9rlYmGSaVeNGlTrMQtN/qCPjreZ7/CiGDZCpAl32Ux'
    'FHbAfkeTMmwbqbAtm8KresBd49V6w2H3W5AU2XJZ4vXvJw6NpMk+cfO7Z09/IPihYBYFop0m+E1a05X//NvvL/+g3dCwg4jD9qIXrRhzpcbRhu4+0/BDvMqm'
    'WAZbs3JlqzZRlo4UXSOzhsNwMZBNM5Q7QVmdzdQ87JcK+XiQGbNNHHCBRwybWU3LtGUAs47PZrhI9QGznv746Nkzhi7uZe/Tjy8vmqu7ysrQkwzkA3uLMcY7'
    'zheF557soY+gDLOz7im8OaLbGKuw8IDXDMtAoy2WN7ujT2xnNbCNaRug+XBBfSSmNxMujPddlrkeMJqSZCN6WR50CdpqJd8QPZriMJnjaSnJiKcaPcwoIk1s'
    '8Khkx/9QSwMEFAAACAgAAAAhXA1xTZ/1CgAAohcAABwAAABMT0cwNV/qs7XthrVf7J6R7ISx7JWI64K0Lm1kfVjLbhtHFt3zKwowsslIsp5+LA1MoDEwwgyy'
    'CZDBwEhmlESIIwWSjCCAMWhKLYEm6ZiKSYuUmkzTlizRQ8MtsklRMLXxpyS7rup/mHNvVTeLsj0LPdjsqrqPc889t66Jv/5tcXpBRN1evNsTqrGn3DOhKjm5'
    'FWYyn34q26GqD4TshtHgkWocywtXxLm2auwI5btCbWfjbe/duTrbUQ0Xf4/21ImjmkF0FgrVCeNsW75sq/xRnB9gs3xZyPZAuVgh+65y63G1gq1kIScLR1Mi'
    'ChxsiI09cVPlqmq3if1k0RHaCm2EDKqqXqWXYLOIBoFs1lUdptQvo64v7txVNefq0e/Oo/NhdBYInCqPqzj91Slvsv86rvgCR8S1Mp7igK4PL+DCxSnMlMdD'
    'E5K40kqsVIUjslAHg96Iem35pBZXqliH0OVVvpe8K/sONseO/DLM6LWjgOIE7y6qdFKWQlfHusOS7A7IqKhXUl5THbTSeLg40IpTJnPtmlAnbvzskVA5D3Zm'
    'MjMUvay4v7yxsbbKrjVK8kWRHdCB+WZt/YcNJLKnttp00n5JxLUKwhpX4MxWG25NCMRFHoeUJzyJK68nKJ77pahTFPKkGQ0G2G1I0ZanQ+XDaj9re1voya5j'
    'fS3DI+UNZSsUvztV/S1iHspGUcSHrgnt706NlsR7nip4ozhnZqdEXAll4ULINwO1k7iEoNVDwh5Sr45wRsfFJzw9LCEFZHrUCaLuEE/6ZcQN7+WQGIoE+0K4'
    'lQU4V6WoJMk8KKvOayzp/Nf2ohuqZmiZNDdFe1B0XwyRUjHHMIc521mG4GEpsZFzOUBZuWorwL44Jp+gFcYR5FAlFqrkSYPLYytE5uUTbOb5jCfs7vmcRz9L'
    'FscVF9Zy0gAi4IIOBoz8PeOFZe/8FPtQc22nYErBS3PPUdDW4iihztx4u64B6le4JFAwjPJOgCd2HfQ89XxHxMUiMME+oTKrgKvnAtu0RG6dRhfZxJiFKWEy'
    'p/PJWDep0rnEcW8G2rDE0noxedvKXGrC0sw0njNc0/K5FQVcnGwRyk+zA8gN62zreQFqy4e/Ymn+3fnSDfzc4tD7LgiEi6IOHglxBpEBHZOrouqsCGe++O7n'
    'P5zyKN3xweN353fuchQarknfH05lQvxl7Se8aQCXUE7CSwwHDjeypXmBEsqw4dVffPfVJi3XQU/iZs6NiyVZOMVHOwNYRkmInzwCj1ie37mLz8BbUT4F/msu'
    'Eb5mPjwjXlV9jyoJVYszqILwEkBNq1QFGajZPDSCtDZCM1gbRJvJTMIBXz4PEFMK3uIMIYxIDaUTBRUDvsUb1mNVubRM5d5EO2JjxJaiKV+6RJf4Wh7D8bIr'
    'm57y6yB54n1+HMo3SLsnGM8pLpK1Luf1wMYCyGlxztQxs5EDW9+dL87iZ26Cv56nr2EwvmZXgJfFBfzcIH7DWfLVzjh7TQq5+yvtZDKcRoLbkw9KdVDKrajT'
    'Stx5mfbBz2c0g5flb4/ErdlPhI6e2UrttxmkBsDVHA7m9/ueZOjzstvTnxiGZJf7LdU8My1Kvnr7sXaWLgl1qos6iAxDzqzpSPrQUUeaNC6BxC2LEDh8oG2p'
    'cv1jYBHVewo8VST17sCR+WMTcG02EGgCqmlAlj7QVNVuEUyX7KkZl8G5v2u6DZifHug9TCuzdQZYedCMgl+JmD30TfxlyYIKMrjSJNQosdMc9qTT992xLCdi'
    '6MUQ+Gc6hiH5C6xOyNMwOfcj2yomawGFI6jatqqWgeluTGLdPe707KdZxpjpulRYBD6bZykToCmc/vQMconJKGF/h2KB7stwYSOZZ0E+Bf9D+oXy2nChvthM'
    '1alCUVlFaGkWRHT38QRSWsJf0FrnkiuQPZ9gUoH5tLiSY92IQNEXgEF0Ti8cAj0tER8UYbrxcSI9+/wUKecWyPuNV66WnKYuSb8Qq7GlhopA5WidQnpvx5zl'
    '18FXz8L3iA306qiXdVnyUsAzPCgbOoBGOZoWlnTh/QAfxtBBbePVDiGhp4JL0p070r8wjUx32YLPioZTzDzLfQbrKLFjtKA34YJs0BNbv3D1siMcWZIHUWfb'
    '1JJ6GnxAmp5kIZG5FWTz0SA0Wn7U6xJb2Ub2r4GsV3lDNjSVbB8Bjh4WisUopMJqh6hW9DyKJboT0SCaMraGbgAZh9xJOrp+3CPqpB8S1EYRqt+yBPMUg8gq'
    '2wXPvWoUOlpR5BBabEkYIm7a9VlEXMlSEkKS3b6bdjHHPGClXuOZgLaGWTAa28hyMEqNVuTNACVJGIECQWn96c7dTCYpEUfM/+cGImreMpTKgNLH6zaNKhGz'
    '1EvqWoX4DAPLWL017RdXakTKjRKnkKcpkY5hiTTTvrjCtFrd+k0sGiVEx8/Jp15MI8rYSMSVctK0Ak8sYt7ABID8ebxxgj/k9BD7jbzQb30Ufck6Q0chkTeB'
    'DkQtX+mWCNnHUpX7uj5ayx5WNZXT6G3VqA9WK1pEkWxDfAwPA63PA5CBcfm9TpDkBxHbzonFlU0R/4Lx9IwWz97k3uBatLx0k3e9CBEzSqKOLxLW19JkK5wQ'
    'S7dHU27S52en8QaIeF7MxL9USWdQO5ijZejYqCLmNi35El0+Ggz0VnTWE8BH20RmjPVtc5L+mrvGGHUkYkGjLqAGDx90R9eu1DzMsGY9CxPaw8xdaOgYcMzJ'
    'FOAhhPW41tLKl3CBnVtBat2gySOYzttHwKCrR1toxhWdqExmaYFBkR9Q71jQKpAFFAOd1Ma8uuAmtkCyX54cxZW9RJKdNNS5a2EO6TPDwohd0etH3ImOiQ+E'
    'JiMgL5IomWG/m9x48HTTd2GVriz0qkIdX5tyNGA2SKH/7NZvJTehM32QsU4LOJYnuj+mimdM6jwLTfs0QkqQMOzzoEvn0KzmC26nePl9Bh35hmIp1M3AQkeM'
    'ximmSz6JOnglZ/NVOqZdFUVi6RbXzzS6OY/EmDaTaxojK3OyPsTwTGoVeoBsONkjBKES9AgSdXuQfkkaUJswDS8gJuM0oLbrfDnzgQ5koyrp4shdrZXJPDQ0'
    'Ix4mF1567BYP8dXk5KQwv/HJrNT3OOKhdUX1f2+oJugs/Gu9Pxf1Alpy4pBRRd7vH+ldj35PJ/ifOJmN39+RjwPSBnmg9LBIN1r9LOeAh7RMhuejez+trX//'
    '9dra91MrP/68+jXjDL0BkTloWeSl5aLY2Hzw7+XVzesba/cfbK6srZr5nYesPpi7rVnYZeBoySlzrqxfpsmrjYQmX8Gpy7JBeXK3R+MvfGwP9MUH8cCO4SaS'
    '2my8+PLu33WXc0zN0QZ04QIif+KSbtNMkvY5PqjcIrzFVVfPpt772m3ppsG22cbcqkj/SAaHeJW8YkEDhwdFAFEdBGQE6x5C99iADz7HmyPa5t2Ys1mzoqea'
    'iWkUaB1QoyGtihljar6vImjqjkk3gyXoFkuVvzdrMCQQk07OXL0Y7a1bFusONiWTsafitKGl/YzZkVRMfWirIL6awX5EByT56F2zmGsfWyDsAEg6Il5BA3up'
    'mSptQ7JAZEMRN345Gob2rMMoQu9lq9v4rWd5FQwQXnONmS9vrD1Y/9fyvbX1lW9XVr+6v3H925XNe+vL3yyvL6/i+ez07I2Z6enblMZGzhYaCNU2ZqPTzz+7'
    '8+elz7Cx9mLlhx/X1jeJOZ+e0QXG1Szq9OhCuK5viWB9MoBUaTDlW6LRJVHdmZmmbno8vHpLlF4beenNF5FfcjOjFY9Ls8te6WqLyPwPUEsBAhQDFAAACAgA'
    'AAAhXMCF1ZJJBwAAww4AABsAAAAAAAAAAAAAAKQBAAAAAExPRzA1X0NvbGFiX+yLpO2WieyViOuCtC5tZFBLAQIUAxQAAAAIAAAAIVwyKRlEUwkAAHkaAAAS'
    'AAAAAAAAAAAAAACkAYIHAABMT0cwNV9NMDdfZm9ybXMubWRQSwECFAMUAAAACAAAACFcUm00Fx8JAADcGwAAEwAAAAAAAAAAAAAApAEFEQAATE9HMDVfTTA3'
    'X2xlc3Nvbi5tZFBLAQIUAxQAAAgIAAAAIVwFXnF7IAYAAMkMAAAkAAAAAAAAAAAAAACkAVUaAABMT0cwNV/qs7XthrVfQUnsl63tlaDsi6TsirXslYjrgrQu'
    'bWRQSwECFAMUAAAICAAAACFcDXFNn/UKAACiFwAAHAAAAAAAAAAAAAAApAG3IAAATE9HMDVf6rO17Ya1X+yekeyEseyViOuCtC5tZFBLBQYAAAAABQAFAGYB'
    'AADmKwAAAAA='
)
_EMBEDDED_ZIP_BYTES = _embedded_b64.b64decode(_EMBEDDED_ZIP_B64)
_zip_path = Path.cwd() / 'LOG05_실습자료.zip'
_zip_path.write_bytes(_EMBEDDED_ZIP_BYTES)
_log05_extract_zip(_zip_path)
```

```python
MISSION_JSON = '{"code": "LOG05_m07", "steps": [{"title": "기준·역할·측정을 먼저 고정한다"}, {"title": "현재 카드에 직접 판단·대화한다"}, {"title": "독립 채점과 실제 로그를 검토한다"}, {"title": "보완과 미실시를 전달한다"}]}'
# LOG05 학습 기록 도구 (표준 Python/Jupyter 환경용)
import json, zipfile, io, os, datetime
from pathlib import Path
from IPython.display import display, Markdown
def _load_notebook_widgets():
    from importlib import import_module, invalidate_caches
    import subprocess
    import sys

    try:
        return import_module('ipywidgets')
    except ImportError:
        print('이 노트북 커널에 ipywidgets가 없어 설치합니다.')

    try:
        subprocess.run(
            [sys.executable, '-m', 'pip', 'install', '--disable-pip-version-check',
             '--retries', '0', '--timeout', '10', 'ipywidgets'],
            check=True, stdout=subprocess.PIPE, stderr=subprocess.STDOUT, timeout=60,
        )
        invalidate_caches()
        return import_module('ipywidgets')
    except (subprocess.SubprocessError, OSError, ImportError):
        print('이 커널에서 ipywidgets를 설치하거나 불러오지 못했습니다.')
        print('설치 가능한 환경에서는 새 셀에서 %pip install ipywidgets 를 실행한 뒤 이 셀을 다시 실행하세요.')
        print('설치할 수 없는 환경에서는 제공된 Markdown 작성 양식을 사용하세요.')
        return None

widgets = _load_notebook_widgets()
if widgets is not None:
    MISSION_DATA = json.loads(MISSION_JSON)
    CODE = MISSION_DATA['code']
    OUT_DIR = Path(CODE + '_results')
    OUT_DIR.mkdir(exist_ok=True)
    BACKUP = OUT_DIR / (CODE + '_record.json')
    fields = {}
    for no, item in enumerate(MISSION_DATA['steps'], 1):
        fields['activity_%02d' % no] = widgets.Textarea(
            value='', placeholder='내 판단, 확인한 자료 번호, AI 제안과 수정 이유를 직접 작성하세요.',
            description='활동 %s' % no, layout=widgets.Layout(width='98%',height='130px'),
            style={'description_width':'75px'})
    for key, label in [('first_thought','AI에게 묻기 전 내 판단'), ('ai_request','실제로 보낸 AI 요청 전체'),
                       ('ai_response','실제로 받은 AI 답변 전체'), ('source_check','원자료 대조 3건과 수정 이유'),
                       ('final_result','최종 작성 결과와 근거'), ('limits','미실시/미측정/확인 필요 및 교관 검토 상태')]:
        fields[key] = widgets.Textarea(value='', description=label,
           layout=widgets.Layout(width='98%',height='125px'),style={'description_width':'190px'})
    status = widgets.HTML(value='<b>아직 저장하지 않았습니다.</b> 작성과 교관 승인은 별개입니다.')
    restore_input = widgets.Textarea(description='복원 JSON',placeholder='백업 JSON의 내용을 붙여넣으세요.',layout=widgets.Layout(width='98%',height='100px'),style={'description_width':'110px'})

    def payload():
        return {'schema':'log05-student-v1','mission':CODE,'saved_at':datetime.datetime.now(datetime.timezone.utc).isoformat(),
                'answers':{key:w.value for key,w in fields.items()},
                'status':'작성 기록 (학습 PASS 또는 교관 승인 아님)'}

    def save_record(_=None):
        obj=payload()
        if BACKUP.exists():
            stamp=datetime.datetime.now().strftime('%Y%m%d_%H%M%S_%f')
            BACKUP.rename(OUT_DIR / (CODE+'_previous_'+stamp+'.json'))
        BACKUP.write_text(json.dumps(obj,ensure_ascii=False,indent=2),encoding='utf8')
        status.value='<b>로컬 저장 완료:</b> '+str(BACKUP)+' (PASS 아님)'

    def export_json(_=None):
        save_record()
        raw=BACKUP.read_bytes()
        try:
            from google.colab import files
            files.download(str(BACKUP))
        except ImportError:
            print('Colab이 아닙니다. 저장 위치:',BACKUP.resolve())

    def restore_record(_=None):
        try:
            obj=json.loads(restore_input.value)
            if obj.get('mission')!=CODE or not isinstance(obj.get('answers'),dict):raise ValueError('다른 미션 또는 형식이 맞지 않는 백업입니다.')
            save_record()  # 복원 전 백업 저장
            for key,w in fields.items():w.value=str(obj['answers'].get(key,''))
            status.value='<b>복원 적용:</b> 기존 작성 내용은 이전 JSON으로 보관했습니다. 다시 저장하세요.'
        except Exception as exc:status.value='<b>복원 오류:</b> '+str(exc)

    def export_review_zip(_=None):
        save_record()
        target=OUT_DIR / (CODE+'_review.zip')
        with zipfile.ZipFile(target,'w',compression=zipfile.ZIP_DEFLATED) as zz:
            zz.write(BACKUP,arcname=BACKUP.name)
            text='\n\n'.join('## '+key+'\n'+val for key,val in payload()['answers'].items())
            zz.writestr(CODE+'_work.md',text)
            zz.writestr('README.txt','검토용 보관 파일. 실제 평가 PASS 또는 교관 승인 자료가 아닙니다.')
        try:
            from google.colab import files
            files.download(str(target))
        except ImportError:
            print('Colab이 아닙니다. ZIP 위치:',target.resolve())

    for i,item in enumerate(MISSION_DATA['steps'],1):
        display(Markdown('### 활동 %d. %s' % (i,item['title'])))
        display(fields['activity_%02d' % i])
    for key,label in [('first_thought','최초 판단'),('ai_request','AI 요청 원문'),('ai_response','AI 답변 원문'),
     ('source_check','원자료 대조'),('final_result','최종 결과'),('limits','한계 및 실제 검토 상태')]:
        display(Markdown('### '+label));display(fields[key])
    buttons=[]
    for title,handler in [('작성 내용 저장',save_record),('작성 JSON 내려받기',export_json),('검토용 ZIP 내려받기',export_review_zip),('작성 JSON 복원',restore_record)]:
        b=widgets.Button(description=title,button_style='info');b.on_click(handler);buttons.append(b)
    display(widgets.HBox(buttons[:2]));display(widgets.HBox(buttons[2:]));display(restore_input);display(status)
```
