```python
# @title 📦 LOG05 실습자료 자동 준비 (ZIP 업로드 불필요) { display-mode: "form" }
# 학습자 화면에는 결과와 업로드 인터페이스만 표시하도록 구성합니다.
from pathlib import Path
import os, zipfile, stat
from IPython.display import display, Markdown

LOG05_PROJECT_DIR = Path.cwd() / "LOG05_실습자료"
LOG05_PROJECT_READY = False

def _log05_extract_zip(zip_path):
    global LOG05_PROJECT_READY
    zip_path = Path(zip_path)
    if not zipfile.is_zipfile(zip_path):
        raise ValueError("유효한 ZIP 압축파일이 아닙니다. 실습자료 ZIP을 다시 선택해 주세요.")
    dst = LOG05_PROJECT_DIR.resolve()
    dst.mkdir(parents=True, exist_ok=True)
    with zipfile.ZipFile(zip_path) as archive:
        entries = archive.infolist()
        if len(entries) > 20000 or sum(info.file_size for info in entries) > 1024**3:
            raise ValueError("압축파일의 항목 수 또는 해제 후 용량이 허용 범위를 초과합니다.")
        for info in entries:
            item = Path(info.filename.replace('\\', '/'))
            if item.is_absolute() or '..' in item.parts or not info.filename.strip():
                raise ValueError("안전하지 않은 ZIP 내부 경로가 포함되어 있습니다: " + info.filename)
            target = (dst / item).resolve()
            if not target.is_relative_to(dst):
                raise ValueError("실습 폴더 밖으로 벗어나는 파일 경로입니다.")
            mode = (info.external_attr >> 16) & 0o170000
            if mode == stat.S_IFLNK:
                raise ValueError("심볼릭 링크가 포함된 ZIP은 허용하지 않습니다.")
        for info in entries:
            target = dst / Path(info.filename.replace('\\', '/'))
            if info.is_dir():
                target.mkdir(parents=True, exist_ok=True)
            else:
                target.parent.mkdir(parents=True, exist_ok=True)
                with archive.open(info) as src, target.open('wb') as out:
                    import shutil
                    shutil.copyfileobj(src, out)
    files = [p for p in dst.rglob('*') if p.is_file()]
    if not files:
        raise ValueError("ZIP에 압축 해제할 실습자료 파일이 없습니다.")
    LOG05_PROJECT_READY = True
    print("✅ LOG05 실습자료 압축 해제 완료:", dst)
    print("📄 확인된 파일 수:", len(files))
    print("ℹ️ 실습자료 폴더:", dst)
    print("※ 업로드·압축 해제는 학습 수행 및 평가 완료를 뜻하지 않습니다.")
    return dst

# 실습자료가 노트북 안에 내장되어 있어 ZIP 업로드 없이 자동으로 준비됩니다.
# 내장 파일 5개: LOG05_Colab_실행안내.md, LOG05_M06_forms.md, LOG05_M06_lesson.md, LOG05_공통_AI역할실습안내.md, LOG05_공통_작성안내.md
import base64 as _embedded_b64
_EMBEDDED_ZIP_B64 = (
    'UEsDBBQAAAgIAAAAIVzAhdWSSQcAAMMOAAAbAAAATE9HMDVfQ29sYWJf7Iuk7ZaJ7JWI64K0Lm1kbVddbxNXEH3Pr7hSxSMmDtDAI2orQCJq1RaQ+hIZSEtK'
    'qFECRfShWiebaBMbYkOcbJK12RTnq3XbTbyOHcm85Kf0ce/d/9Azc+/aC1gCFtl3556ZOefM+DNx6+vro5eF2i7FhaY8KQhVbMTry0JVHTkfjoxkM/rE5PP8'
    '7KN7+fyjzPSTF7/ck7s98UV+JndPbZSV7Qm1EUYtX6iGJU9toRYtfG5eREC10o7XbLnjKb8Wr3Qzv00/UTUcc5ty770sOrLYyAjl2+o4VL6n6hXZ6oofbn4T'
    'tXoiagOWHVcP++f6dwhVb2q4qmaJqBuonRCfVZR9JOLVUL5cI5jy5T9qPVRvArWPN6orAGNCjYxlRNRqx0ttce0mUvg7rvomcbVpiVtTc3P5Xxho7T1lB1Sq'
    '1hVqvgmsIi6VZfHgrKNqrmx2zzrXblLOcn8n6uJMrUcvyoOe8nGrXxjcijC2XLX5yuB9Uj8Uycdja00d/33WkcW2bFl89bserpbBdirAtW9uini+IVfWRFy1'
    '8Q6wL6W+j9rlKLQEgiDSv11Zd4U6DeUbD2DNPfVd1FDOu0KeWLJECKuyFVJEnYaqL0p/N666nxbtYkZ8+9V3t299L5C2qpfRepRAqPWKKtYY8hEwBVSwudyv'
    'U5OzU3PPZp7ykXc9ZfORuEqNxr+Gbun2bnaBScSvumq/gp6CNUKubqKggrhx4ono+D1RCQEJgO1Rk3XXEZAqK8v9aHEV51utpMrmlMtc1aWIOr3oKOBCxZtV'
    '9Bf/NX1ue+qPRaFsP15g1HL+IDotJHW4lNH55aYnH0znZvI/PZsirEM7CX2VCSyCREEBf0CCEGxN5x06oB5dDvLZeP55QCnGWyVV9NG4fScJ8LEg7j58ceFG'
    '/vmFuw9zXOaoZcdbL8860bEVL/n0RDN61Nsf87OP53R7dSH6MVA2ioszYEO8CUbt28qzSQaa59w/t0WFizpduWdrXXTVXm0oSy5nOGfoWLmOKnK7SKENi0XN'
    'RRfxawd5kXihiSjwiHUG0LdZ+vh61hxVG03qrKqvybfLV8bORYFlAl8dPWcCM6KTQ7VzNAyRkEuv+aY3R1F3GQX1y8pxSbamxvz6fFPWS5yraf7KKSXp9bi1'
    'QzxEqKUSK7Belu9KDAtaxPlDcHsTomqj45Zc2VW1kPSmGhWOsWnjODGXSOdXh9bw84yYGIeMG2orxP0QMLkeIdHOOjE6PqkcD8yO3apWuHHbzMOnj2c0ZUPp'
    'N8g+qlS9MucJ4XRLEg61FcCd+g7ed+LYteGsxjPkfkl7QnRcIo+WxWaatXyGXBTtg25jOxgDOwVnGxrH5Hoe/9V3TN0uPLm1Wr7abkLxn+UOjgqCV9/9z9rk'
    'XBxb7vUJghvibTtlppq4JhujHvkn4p6edQYdNqWQFYjglPX0gSlQsWulJHUg05kNMh4Z56YMMBKnNEzBeD6eNkSFQGhV01kcUksvBTtbjSdG2lV08ULHCBZJ'
    'Ks9nVtUdxGUNGFkVvSiwh1CfDHLBMDaueDg3XA5UAR4D4vaX57Nnndt3zo9xpxYK8QLnDixyr5lUpmThG+iGJYgn3KGOfBPb5BEMZygxQWVlp3+RaUWrTeOO'
    'gKZnEN/ouNrj/Cpw0uu6OkNxG8aR7FYaqB97VnrIDarFn+81aXx/2A2qyYbLTak6ifeRwDHZtw6HavEKEoFVAd7+coKB+n3j+4lbnN6Oh+w/FKs8tuOVnvEV'
    'Djzw274uyZL4Xar4k9n8z1P32cVplhVPKQW1iNT2amyYvNWwgzPlAJWNxpgVz0f6Nl5F2w5Sc/BxbvbRZO7+02e5mcmn04+nmLW0QZltj8hUIKn3Ky//pU4w'
    'H40YMU8Ag96kpA3qC6ab8YJFmj/o0sgwAxSRMbkTqg8pw8hViOkKsh0bhUIFDXzMroqXiPjESfYbdk0KTY8NdHAJVtYQRCDvIDVWCeN8kDaobRz0EH+dODvf'
    'jI5Cika+TDPlBDakFhwQxo5oBL1mPmp6DOYmNcH3YAWUIqmZGcNEw2ZgD583E+OIdVm5BSyWvBNh4NTSm4BZDiEuPQB4KYErn7omJy2sVDLkcqwmroQp/Ooy'
    'gMDax2B2w/aDkewowFylPO5kf78zzhkaD4G9+DbaCwA0lpPFp97Uvdb7Yndn6Nqhly+QDpPBGhvFl/31UsIxK95F/RENePci6j40JX2UANGaS7PmksjGr1yC'
    'pr1Ab4LM9MD8SEiW/GTntAZv0J6oSZPM4pTlpRF/4kTJ533nxBgcagXZrN4ric0YQoWQ28pbopi4dNaZ+Bx/r3C2LEMoOTuKIUK9Y9d13I9WXs4GxedtRBOE'
    'CldQfkUkW5zxHK06rIe6bgNtJQulNuuBVS0mA/7Zk5l87sHcBd4d+j/BDH/TayhiQliOqS7xoo+KZzDclBIODmAKukDUN/rRtHXIU6jYgECAWJNME9SMJwx9'
    '/ePELFZ69cPx1Oon3wbkJnXbgOF1FH3nWZnY5qAd/wNQSwMEFAAAAAgAAAAhXIKf2GZcCQAAjBsAABIAAABMT0cwNV9NMDZfZm9ybXMubWTtWN9rG1cWftdf'
    'cSEsyKxt2ZadZf1mKLs11OxCH/qwLK7ZptuFpF6aQFkwixSPg2qpRMaSPXJGZtTKkRXU7tiSbIl1WPCf0r7NvfM/7HfOvXc0smOnCd3uS0iTjmbur3POd77z'
    'nXtHfPCH388siJWZu0Id7ijnRKi9HVWsp1Kq76lvtoR6HKgzT3YuFsXHj+49+Pv9tUf3Hmbur/91ZmF1be2L1U/X/vJoFX//tv75qv0+/eCTj1OpKbOiOixf'
    'nsugjVV/zFVUI5BdR2REGHiqPvj10vKPueriaHD9AoNPHeU7/LLYVL6XWVoW8sWxqrsievpVGOSwpKrlRDgIVDMn7PDU/XsPH65/jpXK8tuSPLoQqv4y7Ppm'
    '6ajalsWCLDanhXpWlt0BNur2VM0R0Y5Ha4dBTdWx3OOeOmjTU/Skj0lR1VX7HRFVPexUkK1SvF5PPnWELMfLyn8NcNqw28fKeBw0lF+9PKdxxaFQ+09UvaTq'
    'Dk7VU55PBkTV4/DfLnY8Dod5s0oqdefOHbxyp+OIOB58J2bFlAh7Oewg1GY+qrWjUjlNp1JNh0+nDh3Z6U2kUhvmUdDDkRw6eAj7nTCgBxkEkRPI5x16buWj'
    'fEfI3ZNw8BWNxrqbXlpuDXHsWvvyXPn5CXrfCNTQFRtYeWpqSvykfzFWvOK/n+d16kYvzZGXBj2yRzV3VAvxfLElffjfd2SxI1TBw8i0HBbUcMDROHHUASLU'
    'nFiEi62DT7bgQgAJYBNpIcSEIEzs9WQDH1s5VfTgcQ76HKEpPG2zk77ZUgU3ngPkMPyKXhjkaS/ZKttpWUKvbNSTe6j9snlrR82bzaNaFckC4DCMvAvZ7pkp'
    'vYKInmFlJErx6tEWhNoeYl54VqKM2ccxS9aaeg84F0gAnNCMv9GnWfhUHVRU9WXkVs0J07JdQU6KlanZf65MZdmT7vfh6TEcnZfbFXIoIdEMA4Yed5DNeABs'
    'ZfE4gSrKG23g6xGG79jxGkhWpuZe8S77k1EzTxZu5pUzSFg4G57A344fbdbZlKj6nXzR5qTiuZoorp4Zv/R8YovIHZit32C69lRaw2bSpPCkyWD8Lrhw26TB'
    'BCJ/HHadiTffRschrXPecEDYvRDh+UV4ErzFgjqeaVlwgGicc4/ZcFgCKN9itRgUaWto0Jan1UmTm4B6Hk8Tr4vsAiIbuRfyaU2AdGXgEdtHroMgFTQ7tCnW'
    'd1GOmJg38xq2NnZmmJmED6WcOu3Fv1+FV7ybTeBwLvGcTTzPJ54XEs93x/GKzOspNy9MrTJlcjhIpUxqZYl1UL40bEDcHFdwN4eDS5GlD/DEmQPqT5ZCZgwk'
    'LOOq6FNR7KvgJebXe9EeWLTmoJoSR6kqGLQWFzu9+yxtLXfKyi+AAArw36T9NMefuv3wPMfs6x8JddbGCeIRWS67XUc9+RrVNye3j7jssgHqcCs+I1dSw6M4'
    'GBe35fcuz8cEwDg7qs0CcTXKtdxtUgRVdVtt9+PTm6Jw2NEOYC0QkzX81BlEpEYgMvpl5TUIo2Y3gFpVC7RNg8HEhGuIntQKZgRIIg9HNoPUbnDtBAaw7Idx'
    'nmT1MgooYa47oAQYEScfOaMDltHxythw6XRO5A8cdlah+L0Bx76mCl9/QUCFSoOLooOvEZjTXPTEjwWWJsSRJCQvK7+8KH7Ya9A0WezLbo6r0SAgrPAH81bt'
    '9+Rhiaucno1llpZNNZOtRjgYYMwWIEGSSosr1o4sxYw/r2AFPnmGUuwS7sPTALRHAyjToyrieFZB8BZThotxHO06FlkmyxAgYCSOl2FNId0y5xJ7nAFShbyo'
    'aHnGA68z3ZhLr/0/xZwT1SpEVaittKXGGZXk0+9YzHYGlPdADTSv5+iqakZZL94wSmOLM9Ug8GZ4pG4BwZVfVAUZA2Q5RdjQPPNnI2B/PCvHsQEFPA7g1GKT'
    'MGzqpy1RVJR0WMkqKOdWDkMBflLXwIw8YkMCl2jSGvY/ATkMUd9eoCCLLGsDrhdEq9bWUSugga+dT22CaS40TJiZtGKzyHJlG+hpHRJ6Rh2IttZYzx2Nn09y'
    'yEZcFhgLDGuCZSK8usMhiRiHYHxnz2IXzjU9H2sergDYlpo3dq2BztsChChC5xr3buwxQ1ecadyTmQbPyIQ4a2M022SlOvYCRS4++h+XPvww87ul5Q8y1Hqx'
    'hRnAh+qGPCvQktYZlp1iqvxZAH+r5UxGaKONC5MdLgWjXuLWccs19UI6LpFabGssj0E+cIzulsvkGh8kmFtkQrQdsjZu1DxTAUNXg8y3rbMdoknWekv/SqKJ'
    '3+gOghp3vZo5ifY7WnetUczaxL8sF0ytZMt1XDWmjJ3TxPEko4hNUbfyPdA5BJuRIeCAUSphiT/pyX823RmhItngGaUnfshVYukNeIAVqAoPa5SgszMzv+Ji'
    '3SQqNfX6Nl1I5qBH2vUo6xLqEKYiBsfRXgGHZZ1LI/HedI5JUcY2mYpBJGaEDV1j8M6kWGZAJNSisWK2PRZSL25G5G5O7WHulhdVSvL5f5hu+h2+gGArbul2'
    'rZgzXa8mkGp8dvLtR5/9I/P++peZjz5be7RolQb4mMu4LSFcf+Jk0xSepEBb6DEjySB0VMsyiXBeSeUYuvrWhMHIKnaUu3Rw2IyoGqLUmI5JhK5Wegaqsgwn'
    'N1mO3SC7cCYP3qdrKNVyADZmBPW8HuuUDTOESpDB+1UeYlfA3rjaMpnEiud2UqH2gK7c9C6zjFwNHC7ETlPAiQRgWwmIvV1ZIBMWx1ruq5rfitlY8xsFimbJ'
    '34nVqr4esKTyvE4liVLP6IqkcNKNENnFyNAKLKNaBVYTiF10UNJtA6Mio+syY9zce7GHEmpegytxj7ZH71mz6/KvKWRMXb9CwHPba0BE+Ph0/YsHDzN8nbkK'
    '367y7+kHnzApaVSaAvYG15iT4vqCkzdvtV8GFgOcKaEENuiPaQNHMZ/jmEPGUik0rGY7EKYzfV2E5sXnmMc3fjbGVvZryGbiRiZjV+GMt6rWkhsRGVZgnslo'
    'jmFSMBea73Dxy+MiO4aLLONCqzjWvbqcatXJmLD6mWFhVGU2QR4FR9ape9d9rG5luA5QG66bbK2ei771KLGrEbBcUTSHvYPD/wMO82NwmGc4aN1icphRwcWG'
    'zGGp2LtKEVya0XhENddKCUG3mXC/lgdGJsVUs/Kby/OV3/LBWDi8i/0vH/v/AlBLAwQUAAAACAAAACFcaIFcTUsKAABcHwAAEwAAAExPRzA1X00wNl9sZXNz'
    'b24ubWTtWd9vG8cRfudfsYBhgAYoUhItB9Wb2wCugRp9avNQFIrRJE2AuC4SA0UBoaDsk8uIdE1bonS0jwKZUJZoyO2JpCQKoFFAf0rzxt37H/rNzO7xSImx'
    'FRh+MqAfx7vb3Zn5vvlmdnlJ/ea3N2YX1K3Za+p/hQ1ldgtR4JuHZfOsrfR+z9T7Snd7w/53ZntHn3jKlFqmFAxDTw17BdMIlA4fq2Gf3lCm9QTjaQ5zsme8'
    'QOHXlOpqGAY6DFKpS5eU/k/frPrKPFiJau1USpdOTKlh6gW8U8BNhRk/MkUY0MRkulywNogJOvRN3cdqoW7WlT7yjFdX5mDVbMOurVdRtWG2V3WpqEutbMKT'
    'rQrZco4zemcw8scEA90MrFdR1R92GxN+6ZerunFi6p7S93v6RU/palmZ9VCvbbhgiF9RuTLsDqaFgyaQiETVgExNpWaUeV7R3f7psXlQVDe+uqeif/WNd6Dm'
    'swti3Y+FDbK03xyGT029p0zV0yX/x0KVxh4G5vtVZe6H5ijQ+4NF9em9z+/89evb9z7/Nvf13T/PLizdvv3N0he3/3RvCb9f3f3LknuevfPZp5hh2D2MHh4q'
    's/2EFjXVIhw0tYK6ftPG1d6jgOm9gWnAq/prs3Zog83QiguEIKbDXBX9Q5kmcbOy8/AjeoIL8l2ZZz3jrxAkpoagAl1/hVzCbJeIENnxydScmlGRP9CPa8oE'
    'DQRQYa7IB/h+EVOm9X1f7/URb0LENEPd9QjSa4g27O0BHAuyXvP1ixCWX0mlls+fclmZWmAf6YqPGWEQ7oKTptNLLKyWMcXMzIwa+4t7plGlFR0jaSwouPVP'
    'hk+e1fvMK/gP64gq1TVTL9ObYRh5oX6xr9KXr2SU3l2JVnCt1w/AR9wYHu5TCqaVwoeoVmVml/aGXU/doJuxAUiCqq/rA/KnEVAcMSfANA3PPrdGIIdgnuWs'
    'M+yoCM5PH5rWfoVCrbuHw+NChhgw7LRV9KyMFAa6G6bz6gpz5uUq8GGjht1d+35UbY8iw3cUc32Fc6cZDg96HKqthzYm8g7ApLnS8i7cJ/ChHPqxZ8J2fB+L'
    'paYSaR5EsoLjVKy9oV/uqTwSKW3u7yPXTo+RxogpcrIZmhP/9BjaJaHmDD7chwBFVaaC8Nnxyc61TDJBMjpBEHy6NTOHT7/7/cy8hZa49UuLqtWrvIDtGLu5'
    'ZzZ7pllmbeMgPqmYRlFXoFKdB/CmJqr38BECwcLW8hA+QE9CRpjCxk2RRhlK0bZAquFxYE5q5Bc5+/0q8COTTLXMo7eLgNnmFVzTO32eR1yZJ1c+hkcxaWlo'
    'fu4y2S3B0ZVAFrbA+7oNYI/aHEEfj2gVCbfVRPsqOzJmu8Urqm3QHFBDYgsqgk1KRh/MLbV4KNFwu5gwNg9jf+XSJ3r6CPEikrlgc8CE0j3YxDHAShxZThCa'
    'FVmGYYAekVbiGtkP8RPLodMgrMxA7y/ATaWftigeiLfzBGKOT5bzVAKFXBAbJwUcmuyIydC37LicsiRO1p40a2sL/tF/Npt5aVOcLriUL7s4JPVm2YmNRITe'
    'xrwPIK6rJ0iJWhv50KD0UpIW58vftL94V53z825uT48S5/s5tRxwoIxC1YoB3kzrk6I54ew2Bx6KEyX0IkLsAsydBiSEaAGVhcwivZGWDKyw1LJ33mnhWDrx'
    'GNtaONjBjN2KG5a3vU1yDciBveveumoXd2IkGgmdQELIkF5RRc89ZmJp0rQFZdbQxPSHR2Usg9mHnbLzBvS/37PMjZuTKTHNI6Yk8dXXkV+1Fqat9EHf/oFc'
    '40j6/x529hDoFaTVGYUUqcWFJH+CVSpW2zczLFbUcZKINE3ey781a66Sh6jSXj/h4RzKEl5oRA/q0jxUX+mXbU4qHjtV813X2ilGft8ufYHhEilb3TI2hV0j'
    'kCGlQ9gmeoErF19GcEhLzlsNYIU9HgwPwp8xoeCZ1kUPjIadm30qxSdlkPJnzBaTIu0cDdu6U83Eu4DdFVxdeROyC9MayWG/KOrQJqwTjaPQ1mFnX4tbxfGm'
    'cFpDOJfg4XziOp+4vpq4XkhcXzvDVy4hDWRvAcuyvPOHUSzPzx5i4kS9IIWJC0BmXP8zaowKGZefrJM9W1wxtwiI8Py8PZNvVShDhd2ErzM2H4jJ+6NHPZZG'
    'VmOu365ajq3wZuRIeBLgwaWyXpcinUSK9amxYrbOLJFoB207JvlHqSApQmPFEKqyer+PjKSIzErbOjlhrJTc3DpRoQHo3Kj4C+W5HpR6rukamyQluyzM3SnG'
    '++jkDor2iEeB9OrornjDYrwWIOBNVdInwuNtulx0LuBE3KlnLYq8v4LilBp4n/HE+Hov2gTuNQ8dtjSPqK01N9CuPkdLu/6TG/eMezTPj+xGAHW5sWNbxMwI'
    'Dtr9c/dF7bte2+G2SfBI7PyTjRUM47bn5senx9QmthDqjof5x+sm9t3cZjNRXP8V72/jdsFxldzuxWXcI/wj74DPMQ4rJmiSetnVCA9smqSNd3sF2wK0KA+H'
    'hyH33+jE5SWzHp6xgLGPqjXcQWIHoJ3IgFxDWujwRsLEYfRaCjjrHT9JLGyPizRbKpUs0JM8cA7GPLBWQVobT8Z2O9kxG+bZhus3ac8vEjOKUnzwwQFusA2x'
    '/rg1c0KlnChDLg52bkwXCMak+lKeYwbOr5xkFksHC/55dubloOt5BaBxNjJjlflhwCnssUY0QxmaStn8zCeCW/R0nRgv2CPTgB0nguxPiZh4yvlhZ+dtA5T6'
    'YcNiLxl7nnlX2TwKc8v5aDUDMsybac+Hxk6GME99ie4WoppPp0gUBkW9CuwS7tr+I4bi1kenx7d+QdovTXDCFhU9/o62x3R4I8ERLo8OqmwHZGWICpFHskpc'
    'ggys9EQ0rUxAFkYKgin+IIP/GLNk4vhQajSHYXTUR1Qe7VLnZmcv2xhhJ2hj+lN1gdyBcq5zBJOlIXoMz/eoyvAJT2b6ucDUQuDblUfy35dex2m+A4TaSL1e'
    'MJsYuxpEG2X94r/2LIG5OvXM0ULkxPZshWTbKbaffPn33K/v/i33yZe37y3SlpRaAEhm9OwREZYFB/xG0BB27uks3RUTSILCeQvkMcIeLTJgXPJYEsbgdDLW'
    'CRGU0+PrN5XkCUuIHK+yupAm66Mi5Qdlgz0/8XU7JNFEvG2GyIlPz25TdKVi99dTZVGMZF99m0W660yNqTuX5UNueX4xvVxU70gvnU3mRR2R5/SQ4I0dT0mb'
    'SSAweqJkObNbhGBxfOmAjUsvI5cjpUHxIOer0AWwJfBQyhMVUQgA7XWyyE0V171Sa1S1xyrUOUVQDhgFaMLwi7vf3Pk2x18gLCG2S/w5e+czFg5hTlQum/pA'
    'XeA8OqPOTpiZvhTp11YImxKcnE9CfbGytKjeY1n6QIf3Qod8kg4Xqv6L6j1U/w8seC8suJpkwds3WWOC8K6arA+QvxfIU4lvVX/yS9WcPRLl71bjr1az3Ds0'
    'IfID2sTV/ah6aEr70iDITYBlsVrbQN/JX7fFy/8fUEsDBBQAAAgIAAAAIVwFXnF7IAYAAMkMAAAkAAAATE9HMDVf6rO17Ya1X0FJ7Jet7ZWg7Iuk7Iq17JWI'
    '64K0Lm1kbVZNb9tGEL3rVyyQa+smaYO0vuUUtGiaAuk9ORU9NCjQ9gdQNmXQplpLsGhTDilQiWTJqYLSEmVTqJyDf4qP3OV/6JvZ5YeSnAhR3J2ZN++9mTvi'
    '+6eP7z4Q2WKZ7y3Fo2+FOnmXe5FQ7kgdLIXyHLmTNBp37gh5aSs7vLnK5la+hw9Oe2r+TijfUW7QaNxagfJsOfDNhfLfVLV8oQYd+aat+pbANULtzFRki7zd'
    'ke5UhbaQcS/f9VSwlsNAqMRBOH59aOOIGvlb5gL6O+97Kkxlx5cHPb7JHSGZZUcFQ3V6LrI0ViNLqAh5ODdX6rKXxZY5pMJE5J6NlHMvoExwgOJkyxmqyj0/'
    'W0QmkpBzJ/dTOV7jqqY6TmpZqJMO4Nnj+IOxXNlC2RfyYIxobpDFTTwDWw1sPO1Avm3R+1F+vC9U36YicE5O2vJopCbAY+Jsicfy6CJL97PFGvmdZ/Nz3I+U'
    'LAqPzOTZTCf3GcOnS1ShL6Rjy/Aa9+821SWXouL3wF8nuwzU61aJMkPPWHhRNm9v3VphoyGnaxXRqz66jOrofzVx5CylXIpAuDbyCnhAiCgwbWfw0lgOQ6Qt'
    'XUe6oy2QhwDSYXHQLuqYDOk+lKy8A0SrPjedEF9mF+BGmFA1FPNVx+ShJk0VdQWQRp9NHnKRmo4ylvQiAb4msZsr3I1umMZTu7OrdXYRcyY702zVNPE1p99O'
    'Cc6C86eJ8puNxufiyf1ttLiL9gIbBpz7xGUvllkcUDJpTFVSzVncJ5yRHvAq0NOk1l2UnYDq1dLhVOJXNSjkIc6vqy+4GxpvediSZ9ciP9zHS6IhETP0IS+k'
    'AFCZjHVcKfuvtnEX/zOMoRSRJZaKOiYjjQcSs9TwAgrCtaOuWvki/zvNj8esv5WNNtZai4qBplAXLU3viZct8JRzDyJB0mBCRAcJdWbMR+2mtB5QWlSbpo6Q'
    'biLd1c0VnEe5UflTHawQLLtso9yyNcxNFmAUoAWVWnvQHAPtLmsJlzLtGOHdWoSZPnxrAe5EhmvmE51LlMcKLI9zClV8poC7lAuL9WRYE/rZfEY3nccfKGln'
    'Jgft2tkPJGkXzCaL2YDo4bbIfXjITKhVIo8CcNzoSo3+KumixVsp4T5YRgJij53/U4tTiJSlk12lIn+F9r5tyWglCNABmRdI8rqlHL8oLD9tox+gBh8mFAob'
    'xdm2lfd7dag0T/UZo13+ELIcJhv2gBS7QBHFW+okpsSrZAspSTBtN8yP6Yw8TzhndlcbQ6EIls2vYQ/sNodwyGktCppMQ8igR0rDizPA+Wb9SU5+vS1k5ODb'
    'vG1xRldT5V2Tb1BCiPVg05wKafiUXNUB1G24RiFx1TAGqwhQ2aWifBl7QNrxIQ/oZpHIsc+3BlExajbS+mabkMi7ATeMoGMBbFgloVWZ4Sb9jXnoKzR+nH0v'
    '5kmbDlmm57WI9+5uYx52gBXrWs4S5AsLxLyu8RidixgH46rkJ2xZm43WF5QHKzfD+7EetbO0UDrjh93B7det+clDoZwAgzT3vWIbAQUwshuNF7xjPH9y9+Hz'
    '8hvd8ef6y61f/nz56wuueBio/2ZCpm3oXZ3GEA4jcHJdYRWgTnqIn+6R3rTDY9FgzfAP7oJmXYkmVopugB+yU5GP22EYjfIAVbmZQE7lIFIDCHBsVAln2pAB'
    'XH7MDmXzmD9rlYmGSaVeNGlTrMQtN/qCPjreZ7/CiGDZCpAl32UxFHbAfkeTMmwbqbAtm8KresBd49V6w2H3W5AU2XJZ4vXvJw6NpMk+cfO7Z09/IPihYBYF'
    'op0m+E1a05X//NvvL/+g3dCwg4jD9qIXrRhzpcbRhu4+0/BDvMqmWAZbs3JlqzZRlo4UXSOzhsNwMZBNM5Q7QVmdzdQ87JcK+XiQGbNNHHCBRwybWU3LtGUA'
    's47PZrhI9QGznv746Nkzhi7uZe/Tjy8vmqu7ysrQkwzkA3uLMcY7zheF557soY+gDLOz7im8OaLbGKuw8IDXDMtAoy2WN7ujT2xnNbCNaRug+XBBfSSmNxMu'
    'jPddlrkeMJqSZCN6WR50CdpqJd8QPZriMJnjaSnJiKcaPcwoIk1s8Khkx/9QSwMEFAAACAgAAAAhXA1xTZ/1CgAAohcAABwAAABMT0cwNV/qs7XthrVf7J6R'
    '7ISx7JWI64K0Lm1kfVjLbhtHFt3zKwowsslIsp5+LA1MoDEwwgyyCZDBwEhmlESIIwWSjCCAMWhKLYEm6ZiKSYuUmkzTlizRQ8MtsklRMLXxpyS7rup/mHNv'
    'VTeLsj0LPdjsqrqPc889t66Jv/5tcXpBRN1evNsTqrGn3DOhKjm5FWYyn34q26GqD4TshtHgkWocywtXxLm2auwI5btCbWfjbe/duTrbUQ0Xf4/21ImjmkF0'
    'FgrVCeNsW75sq/xRnB9gs3xZyPZAuVgh+65y63G1gq1kIScLR1MiChxsiI09cVPlqmq3if1k0RHaCm2EDKqqXqWXYLOIBoFs1lUdptQvo64v7txVNefq0e/O'
    'o/NhdBYInCqPqzj91Slvsv86rvgCR8S1Mp7igK4PL+DCxSnMlMdDE5K40kqsVIUjslAHg96Iem35pBZXqliH0OVVvpe8K/sONseO/DLM6LWjgOIE7y6qdFKW'
    'QlfHusOS7A7IqKhXUl5THbTSeLg40IpTJnPtmlAnbvzskVA5D3ZmMjMUvay4v7yxsbbKrjVK8kWRHdCB+WZt/YcNJLKnttp00n5JxLUKwhpX4MxWG25NCMRF'
    'HoeUJzyJK68nKJ77pahTFPKkGQ0G2G1I0ZanQ+XDaj9re1voya5jfS3DI+UNZSsUvztV/S1iHspGUcSHrgnt706NlsR7nip4ozhnZqdEXAll4ULINwO1k7iE'
    'oNVDwh5Sr45wRsfFJzw9LCEFZHrUCaLuEE/6ZcQN7+WQGIoE+0K4lQU4V6WoJMk8KKvOayzp/Nf2ohuqZmiZNDdFe1B0XwyRUjHHMIc521mG4GEpsZFzOUBZ'
    'uWorwL44Jp+gFcYR5FAlFqrkSYPLYytE5uUTbOb5jCfs7vmcRz9LFscVF9Zy0gAi4IIOBoz8PeOFZe/8FPtQc22nYErBS3PPUdDW4iihztx4u64B6le4JFAw'
    'jPJOgCd2HfQ89XxHxMUiMME+oTKrgKvnAtu0RG6dRhfZxJiFKWEyp/PJWDep0rnEcW8G2rDE0noxedvKXGrC0sw0njNc0/K5FQVcnGwRyk+zA8gN62zreQFq'
    'y4e/Ymn+3fnSDfzc4tD7LgiEi6IOHglxBpEBHZOrouqsCGe++O7nP5zyKN3xweN353fuchQarknfH05lQvxl7Se8aQCXUE7CSwwHDjeypXmBEsqw4dVffPfV'
    'Ji3XQU/iZs6NiyVZOMVHOwNYRkmInzwCj1ie37mLz8BbUT4F/msuEb5mPjwjXlV9jyoJVYszqILwEkBNq1QFGajZPDSCtDZCM1gbRJvJTMIBXz4PEFMK3uIM'
    'IYxIDaUTBRUDvsUb1mNVubRM5d5EO2JjxJaiKV+6RJf4Wh7D8bIrm57y6yB54n1+HMo3SLsnGM8pLpK1Luf1wMYCyGlxztQxs5EDW9+dL87iZ26Cv56nr2Ew'
    'vmZXgJfFBfzcIH7DWfLVzjh7TQq5+yvtZDKcRoLbkw9KdVDKrajTStx5mfbBz2c0g5flb4/ErdlPhI6e2UrttxmkBsDVHA7m9/ueZOjzstvTnxiGZJf7LdU8'
    'My1Kvnr7sXaWLgl1qos6iAxDzqzpSPrQUUeaNC6BxC2LEDh8oG2pcv1jYBHVewo8VST17sCR+WMTcG02EGgCqmlAlj7QVNVuEUyX7KkZl8G5v2u6DZifHug9'
    'TCuzdQZYedCMgl+JmD30TfxlyYIKMrjSJNQosdMc9qTT992xLCdi6MUQ+Gc6hiH5C6xOyNMwOfcj2yomawGFI6jatqqWgeluTGLdPe707KdZxpjpulRYBD6b'
    'ZykToCmc/vQMconJKGF/h2KB7stwYSOZZ0E+Bf9D+oXy2nChvthM1alCUVlFaGkWRHT38QRSWsJf0FrnkiuQPZ9gUoH5tLiSY92IQNEXgEF0Ti8cAj0tER8U'
    'YbrxcSI9+/wUKecWyPuNV66WnKYuSb8Qq7GlhopA5WidQnpvx5zl18FXz8L3iA306qiXdVnyUsAzPCgbOoBGOZoWlnTh/QAfxtBBbePVDiGhp4JL0p070r8w'
    'jUx32YLPioZTzDzLfQbrKLFjtKA34YJs0BNbv3D1siMcWZIHUWfb1JJ6GnxAmp5kIZG5FWTz0SA0Wn7U6xJb2Ub2r4GsV3lDNjSVbB8Bjh4WisUopMJqh6hW'
    '9DyKJboT0SCaMraGbgAZh9xJOrp+3CPqpB8S1EYRqt+yBPMUg8gq2wXPvWoUOlpR5BBabEkYIm7a9VlEXMlSEkKS3b6bdjHHPGClXuOZgLaGWTAa28hyMEqN'
    'VuTNACVJGIECQWn96c7dTCYpEUfM/+cGImreMpTKgNLH6zaNKhGz1EvqWoX4DAPLWL017RdXakTKjRKnkKcpkY5hiTTTvrjCtFrd+k0sGiVEx8/Jp15MI8rY'
    'SMSVctK0Ak8sYt7ABID8ebxxgj/k9BD7jbzQb30Ufck6Q0chkTeBDkQtX+mWCNnHUpX7uj5ayx5WNZXT6G3VqA9WK1pEkWxDfAwPA63PA5CBcfm9TpDkBxHb'
    'zonFlU0R/4Lx9IwWz97k3uBatLx0k3e9CBEzSqKOLxLW19JkK5wQS7dHU27S52en8QaIeF7MxL9USWdQO5ijZejYqCLmNi35El0+Ggz0VnTWE8BH20RmjPVt'
    'c5L+mrvGGHUkYkGjLqAGDx90R9eu1DzMsGY9CxPaw8xdaOgYcMzJFOAhhPW41tLKl3CBnVtBat2gySOYzttHwKCrR1toxhWdqExmaYFBkR9Q71jQKpAFFAOd'
    '1Ma8uuAmtkCyX54cxZW9RJKdNNS5a2EO6TPDwohd0etH3ImOiQ+EJiMgL5IomWG/m9x48HTTd2GVriz0qkIdX5tyNGA2SKH/7NZvJTehM32QsU4LOJYnuj+m'
    'imdM6jwLTfs0QkqQMOzzoEvn0KzmC26nePl9Bh35hmIp1M3AQkeMximmSz6JOnglZ/NVOqZdFUVi6RbXzzS6OY/EmDaTaxojK3OyPsTwTGoVeoBsONkjBKES'
    '9AgSdXuQfkkaUJswDS8gJuM0oLbrfDnzgQ5koyrp4shdrZXJPDQ0Ix4mF1567BYP8dXk5KQwv/HJrNT3OOKhdUX1f2+oJugs/Gu9Pxf1Alpy4pBRRd7vH+ld'
    'j35PJ/ifOJmN39+RjwPSBnmg9LBIN1r9LOeAh7RMhuejez+trX//9dra91MrP/68+jXjDL0BkTloWeSl5aLY2Hzw7+XVzesba/cfbK6srZr5nYesPpi7rVnY'
    'ZeBoySlzrqxfpsmrjYQmX8Gpy7JBeXK3R+MvfGwP9MUH8cCO4SaS2my8+PLu33WXc0zN0QZ04QIif+KSbtNMkvY5PqjcIrzFVVfPpt772m3ppsG22cbcqkj/'
    'SAaHeJW8YkEDhwdFAFEdBGQE6x5C99iADz7HmyPa5t2Ys1mzoqeaiWkUaB1QoyGtihljar6vImjqjkk3gyXoFkuVvzdrMCQQk07OXL0Y7a1bFusONiWTsafi'
    'tKGl/YzZkVRMfWirIL6awX5EByT56F2zmGsfWyDsAEg6Il5BA3upmSptQ7JAZEMRN345Gob2rMMoQu9lq9v4rWd5FQwQXnONmS9vrD1Y/9fyvbX1lW9XVr+6'
    'v3H925XNe+vL3yyvL6/i+ez07I2Z6enblMZGzhYaCNU2ZqPTzz+78+elz7Cx9mLlhx/X1jeJOZ+e0QXG1Szq9OhCuK5viWB9MoBUaTDlW6LRJVHdmZmmbno8'
    'vHpLlF4beenNF5FfcjOjFY9Ls8te6WqLyPwPUEsBAhQDFAAACAgAAAAhXMCF1ZJJBwAAww4AABsAAAAAAAAAAAAAAKQBAAAAAExPRzA1X0NvbGFiX+yLpO2W'
    'ieyViOuCtC5tZFBLAQIUAxQAAAAIAAAAIVyCn9hmXAkAAIwbAAASAAAAAAAAAAAAAACkAYIHAABMT0cwNV9NMDZfZm9ybXMubWRQSwECFAMUAAAACAAAACFc'
    'aIFcTUsKAABcHwAAEwAAAAAAAAAAAAAApAEOEQAATE9HMDVfTTA2X2xlc3Nvbi5tZFBLAQIUAxQAAAgIAAAAIVwFXnF7IAYAAMkMAAAkAAAAAAAAAAAAAACk'
    'AYobAABMT0cwNV/qs7XthrVfQUnsl63tlaDsi6TsirXslYjrgrQubWRQSwECFAMUAAAICAAAACFcDXFNn/UKAACiFwAAHAAAAAAAAAAAAAAApAHsIQAATE9H'
    'MDVf6rO17Ya1X+yekeyEseyViOuCtC5tZFBLBQYAAAAABQAFAGYBAAAbLQAAAAA='
)
_EMBEDDED_ZIP_BYTES = _embedded_b64.b64decode(_EMBEDDED_ZIP_B64)
_zip_path = Path.cwd() / 'LOG05_실습자료.zip'
_zip_path.write_bytes(_EMBEDDED_ZIP_BYTES)
_log05_extract_zip(_zip_path)
```

**교관 검토 기록:** 판정 [미확인 / 보완 필요 / PASS / FAIL]을 실제 기준에 따라 기입해야 하며, 자동 표시는 하지 않습니다.

```python
MISSION_JSON = '{"code": "LOG05_m06", "steps": [{"title": "메모의 네 종류 내용을 나눈다"}, {"title": "AI와 관제·보고 서식을 만든다"}, {"title": "원문과 완료 주장을 대조한다"}, {"title": "표준 표현과 상태를 넘긴다"}]}'
# LOG05 학습 기록 도구 (표준 Python/Jupyter 환경용)
import json, zipfile, io, os, datetime
from pathlib import Path
from IPython.display import display, Markdown
def _load_notebook_widgets():
    from importlib import import_module, invalidate_caches
    import subprocess
    import sys

    try:
        return import_module('ipywidgets')
    except ImportError:
        print('이 노트북 커널에 ipywidgets가 없어 설치합니다.')

    try:
        subprocess.run(
            [sys.executable, '-m', 'pip', 'install', '--disable-pip-version-check',
             '--retries', '0', '--timeout', '10', 'ipywidgets'],
            check=True, stdout=subprocess.PIPE, stderr=subprocess.STDOUT, timeout=60,
        )
        invalidate_caches()
        return import_module('ipywidgets')
    except (subprocess.SubprocessError, OSError, ImportError):
        print('이 커널에서 ipywidgets를 설치하거나 불러오지 못했습니다.')
        print('설치 가능한 환경에서는 새 셀에서 %pip install ipywidgets 를 실행한 뒤 이 셀을 다시 실행하세요.')
        print('설치할 수 없는 환경에서는 제공된 Markdown 작성 양식을 사용하세요.')
        return None

widgets = _load_notebook_widgets()
if widgets is not None:
    MISSION_DATA = json.loads(MISSION_JSON)
    CODE = MISSION_DATA['code']
    OUT_DIR = Path(CODE + '_results')
    OUT_DIR.mkdir(exist_ok=True)
    BACKUP = OUT_DIR / (CODE + '_record.json')
    fields = {}
    for no, item in enumerate(MISSION_DATA['steps'], 1):
        fields['activity_%02d' % no] = widgets.Textarea(
            value='', placeholder='내 판단, 확인한 자료 번호, AI 제안과 수정 이유를 직접 작성하세요.',
            description='활동 %s' % no, layout=widgets.Layout(width='98%',height='130px'),
            style={'description_width':'75px'})
    for key, label in [('first_thought','AI에게 묻기 전 내 판단'), ('ai_request','실제로 보낸 AI 요청 전체'),
                       ('ai_response','실제로 받은 AI 답변 전체'), ('source_check','원자료 대조 3건과 수정 이유'),
                       ('final_result','최종 작성 결과와 근거'), ('limits','미실시/미측정/확인 필요 및 교관 검토 상태')]:
        fields[key] = widgets.Textarea(value='', description=label,
           layout=widgets.Layout(width='98%',height='125px'),style={'description_width':'190px'})
    status = widgets.HTML(value='<b>아직 저장하지 않았습니다.</b> 작성과 교관 승인은 별개입니다.')
    restore_input = widgets.Textarea(description='복원 JSON',placeholder='백업 JSON의 내용을 붙여넣으세요.',layout=widgets.Layout(width='98%',height='100px'),style={'description_width':'110px'})

    def payload():
        return {'schema':'log05-student-v1','mission':CODE,'saved_at':datetime.datetime.now(datetime.timezone.utc).isoformat(),
                'answers':{key:w.value for key,w in fields.items()},
                'status':'작성 기록 (학습 PASS 또는 교관 승인 아님)'}

    def save_record(_=None):
        obj=payload()
        if BACKUP.exists():
            stamp=datetime.datetime.now().strftime('%Y%m%d_%H%M%S_%f')
            BACKUP.rename(OUT_DIR / (CODE+'_previous_'+stamp+'.json'))
        BACKUP.write_text(json.dumps(obj,ensure_ascii=False,indent=2),encoding='utf8')
        status.value='<b>로컬 저장 완료:</b> '+str(BACKUP)+' (PASS 아님)'

    def export_json(_=None):
        save_record()
        raw=BACKUP.read_bytes()
        try:
            from google.colab import files
            files.download(str(BACKUP))
        except ImportError:
            print('Colab이 아닙니다. 저장 위치:',BACKUP.resolve())

    def restore_record(_=None):
        try:
            obj=json.loads(restore_input.value)
            if obj.get('mission')!=CODE or not isinstance(obj.get('answers'),dict):raise ValueError('다른 미션 또는 형식이 맞지 않는 백업입니다.')
            save_record()  # 복원 전 백업 저장
            for key,w in fields.items():w.value=str(obj['answers'].get(key,''))
            status.value='<b>복원 적용:</b> 기존 작성 내용은 이전 JSON으로 보관했습니다. 다시 저장하세요.'
        except Exception as exc:status.value='<b>복원 오류:</b> '+str(exc)

    def export_review_zip(_=None):
        save_record()
        target=OUT_DIR / (CODE+'_review.zip')
        with zipfile.ZipFile(target,'w',compression=zipfile.ZIP_DEFLATED) as zz:
            zz.write(BACKUP,arcname=BACKUP.name)
            text='\n\n'.join('## '+key+'\n'+val for key,val in payload()['answers'].items())
            zz.writestr(CODE+'_work.md',text)
            zz.writestr('README.txt','검토용 보관 파일. 실제 평가 PASS 또는 교관 승인 자료가 아닙니다.')
        try:
            from google.colab import files
            files.download(str(target))
        except ImportError:
            print('Colab이 아닙니다. ZIP 위치:',target.resolve())

    for i,item in enumerate(MISSION_DATA['steps'],1):
        display(Markdown('### 활동 %d. %s' % (i,item['title'])))
        display(fields['activity_%02d' % i])
    for key,label in [('first_thought','최초 판단'),('ai_request','AI 요청 원문'),('ai_response','AI 답변 원문'),
     ('source_check','원자료 대조'),('final_result','최종 결과'),('limits','한계 및 실제 검토 상태')]:
        display(Markdown('### '+label));display(fields[key])
    buttons=[]
    for title,handler in [('작성 내용 저장',save_record),('작성 JSON 내려받기',export_json),('검토용 ZIP 내려받기',export_review_zip),('작성 JSON 복원',restore_record)]:
        b=widgets.Button(description=title,button_style='info');b.on_click(handler);buttons.append(b)
    display(widgets.HBox(buttons[:2]));display(widgets.HBox(buttons[2:]));display(restore_input);display(status)
```
