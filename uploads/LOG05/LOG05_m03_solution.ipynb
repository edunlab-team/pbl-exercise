```python
# @title 📦 LOG05 실습자료 자동 준비 (ZIP 업로드 불필요) { display-mode: "form" }
# 학습자 화면에는 결과와 업로드 인터페이스만 표시하도록 구성합니다.
from pathlib import Path
import os, zipfile, stat
from IPython.display import display, Markdown

LOG05_PROJECT_DIR = Path.cwd() / "LOG05_실습자료"
LOG05_PROJECT_READY = False

def _log05_extract_zip(zip_path):
    global LOG05_PROJECT_READY
    zip_path = Path(zip_path)
    if not zipfile.is_zipfile(zip_path):
        raise ValueError("유효한 ZIP 압축파일이 아닙니다. 실습자료 ZIP을 다시 선택해 주세요.")
    dst = LOG05_PROJECT_DIR.resolve()
    dst.mkdir(parents=True, exist_ok=True)
    with zipfile.ZipFile(zip_path) as archive:
        entries = archive.infolist()
        if len(entries) > 20000 or sum(info.file_size for info in entries) > 1024**3:
            raise ValueError("압축파일의 항목 수 또는 해제 후 용량이 허용 범위를 초과합니다.")
        for info in entries:
            item = Path(info.filename.replace('\\', '/'))
            if item.is_absolute() or '..' in item.parts or not info.filename.strip():
                raise ValueError("안전하지 않은 ZIP 내부 경로가 포함되어 있습니다: " + info.filename)
            target = (dst / item).resolve()
            if not target.is_relative_to(dst):
                raise ValueError("실습 폴더 밖으로 벗어나는 파일 경로입니다.")
            mode = (info.external_attr >> 16) & 0o170000
            if mode == stat.S_IFLNK:
                raise ValueError("심볼릭 링크가 포함된 ZIP은 허용하지 않습니다.")
        for info in entries:
            target = dst / Path(info.filename.replace('\\', '/'))
            if info.is_dir():
                target.mkdir(parents=True, exist_ok=True)
            else:
                target.parent.mkdir(parents=True, exist_ok=True)
                with archive.open(info) as src, target.open('wb') as out:
                    import shutil
                    shutil.copyfileobj(src, out)
    files = [p for p in dst.rglob('*') if p.is_file()]
    if not files:
        raise ValueError("ZIP에 압축 해제할 실습자료 파일이 없습니다.")
    LOG05_PROJECT_READY = True
    print("✅ LOG05 실습자료 압축 해제 완료:", dst)
    print("📄 확인된 파일 수:", len(files))
    print("ℹ️ 실습자료 폴더:", dst)
    print("※ 업로드·압축 해제는 학습 수행 및 평가 완료를 뜻하지 않습니다.")
    return dst

# 실습자료가 노트북 안에 내장되어 있어 ZIP 업로드 없이 자동으로 준비됩니다.
# 내장 파일 5개: LOG05_Colab_실행안내.md, LOG05_M03_forms.md, LOG05_M03_lesson.md, LOG05_공통_AI역할실습안내.md, LOG05_공통_작성안내.md
import base64 as _embedded_b64
_EMBEDDED_ZIP_B64 = (
    'UEsDBBQAAAgIAAAAIVzAhdWSSQcAAMMOAAAbAAAATE9HMDVfQ29sYWJf7Iuk7ZaJ7JWI64K0Lm1kbVddbxNXEH3Pr7hSxSMmDtDAI2orQCJq1RaQ+hIZSEtK'
    'qFECRfShWiebaBMbYkOcbJK12RTnq3XbTbyOHcm85Kf0ce/d/9Azc+/aC1gCFtl3556ZOefM+DNx6+vro5eF2i7FhaY8KQhVbMTry0JVHTkfjoxkM/rE5PP8'
    '7KN7+fyjzPSTF7/ck7s98UV+JndPbZSV7Qm1EUYtX6iGJU9toRYtfG5eREC10o7XbLnjKb8Wr3Qzv00/UTUcc5ty770sOrLYyAjl2+o4VL6n6hXZ6oofbn4T'
    'tXoiagOWHVcP++f6dwhVb2q4qmaJqBuonRCfVZR9JOLVUL5cI5jy5T9qPVRvArWPN6orAGNCjYxlRNRqx0ttce0mUvg7rvomcbVpiVtTc3P5Xxho7T1lB1Sq'
    '1hVqvgmsIi6VZfHgrKNqrmx2zzrXblLOcn8n6uJMrUcvyoOe8nGrXxjcijC2XLX5yuB9Uj8Uycdja00d/33WkcW2bFl89bserpbBdirAtW9uini+IVfWRFy1'
    '8Q6wL6W+j9rlKLQEgiDSv11Zd4U6DeUbD2DNPfVd1FDOu0KeWLJECKuyFVJEnYaqL0p/N666nxbtYkZ8+9V3t299L5C2qpfRepRAqPWKKtYY8hEwBVSwudyv'
    'U5OzU3PPZp7ykXc9ZfORuEqNxr+Gbun2bnaBScSvumq/gp6CNUKubqKggrhx4ono+D1RCQEJgO1Rk3XXEZAqK8v9aHEV51utpMrmlMtc1aWIOr3oKOBCxZtV'
    '9Bf/NX1ue+qPRaFsP15g1HL+IDotJHW4lNH55aYnH0znZvI/PZsirEM7CX2VCSyCREEBf0CCEGxN5x06oB5dDvLZeP55QCnGWyVV9NG4fScJ8LEg7j58ceFG'
    '/vmFuw9zXOaoZcdbL8860bEVL/n0RDN61Nsf87OP53R7dSH6MVA2ioszYEO8CUbt28qzSQaa59w/t0WFizpduWdrXXTVXm0oSy5nOGfoWLmOKnK7SKENi0XN'
    'RRfxawd5kXihiSjwiHUG0LdZ+vh61hxVG03qrKqvybfLV8bORYFlAl8dPWcCM6KTQ7VzNAyRkEuv+aY3R1F3GQX1y8pxSbamxvz6fFPWS5yraf7KKSXp9bi1'
    'QzxEqKUSK7Belu9KDAtaxPlDcHsTomqj45Zc2VW1kPSmGhWOsWnjODGXSOdXh9bw84yYGIeMG2orxP0QMLkeIdHOOjE6PqkcD8yO3apWuHHbzMOnj2c0ZUPp'
    'N8g+qlS9MucJ4XRLEg61FcCd+g7ed+LYteGsxjPkfkl7QnRcIo+WxWaatXyGXBTtg25jOxgDOwVnGxrH5Hoe/9V3TN0uPLm1Wr7abkLxn+UOjgqCV9/9z9rk'
    'XBxb7vUJghvibTtlppq4JhujHvkn4p6edQYdNqWQFYjglPX0gSlQsWulJHUg05kNMh4Z56YMMBKnNEzBeD6eNkSFQGhV01kcUksvBTtbjSdG2lV08ULHCBZJ'
    'Ks9nVtUdxGUNGFkVvSiwh1CfDHLBMDaueDg3XA5UAR4D4vaX57Nnndt3zo9xpxYK8QLnDixyr5lUpmThG+iGJYgn3KGOfBPb5BEMZygxQWVlp3+RaUWrTeOO'
    'gKZnEN/ouNrj/Cpw0uu6OkNxG8aR7FYaqB97VnrIDarFn+81aXx/2A2qyYbLTak6ifeRwDHZtw6HavEKEoFVAd7+coKB+n3j+4lbnN6Oh+w/FKs8tuOVnvEV'
    'Djzw274uyZL4Xar4k9n8z1P32cVplhVPKQW1iNT2amyYvNWwgzPlAJWNxpgVz0f6Nl5F2w5Sc/BxbvbRZO7+02e5mcmn04+nmLW0QZltj8hUIKn3Ky//pU4w'
    'H40YMU8Ag96kpA3qC6ab8YJFmj/o0sgwAxSRMbkTqg8pw8hViOkKsh0bhUIFDXzMroqXiPjESfYbdk0KTY8NdHAJVtYQRCDvIDVWCeN8kDaobRz0EH+dODvf'
    'jI5Cika+TDPlBDakFhwQxo5oBL1mPmp6DOYmNcH3YAWUIqmZGcNEw2ZgD583E+OIdVm5BSyWvBNh4NTSm4BZDiEuPQB4KYErn7omJy2sVDLkcqwmroQp/Ooy'
    'gMDax2B2w/aDkewowFylPO5kf78zzhkaD4G9+DbaCwA0lpPFp97Uvdb7Yndn6Nqhly+QDpPBGhvFl/31UsIxK95F/RENePci6j40JX2UANGaS7PmksjGr1yC'
    'pr1Ab4LM9MD8SEiW/GTntAZv0J6oSZPM4pTlpRF/4kTJ533nxBgcagXZrN4ric0YQoWQ28pbopi4dNaZ+Bx/r3C2LEMoOTuKIUK9Y9d13I9WXs4GxedtRBOE'
    'CldQfkUkW5zxHK06rIe6bgNtJQulNuuBVS0mA/7Zk5l87sHcBd4d+j/BDH/TayhiQliOqS7xoo+KZzDclBIODmAKukDUN/rRtHXIU6jYgECAWJNME9SMJwx9'
    '/ePELFZ69cPx1Oon3wbkJnXbgOF1FH3nWZnY5qAd/wNQSwMEFAAAAAgAAAAhXNc+gHR6CQAAcBsAABIAAABMT0cwNV9NMDNfZm9ybXMubWTtWF1vG8cVfeev'
    'GMAoLKOSqC9/VG96aWo4RgrkIQ9F4QiN2wSwZSByURgQClJaCYzIQJQtSkt5qawaSqQMulmJK3GFMi/6Kc7bzux/6Ll3ZpdLUUzUovWTkQ9xZ2dn5t577rnn'
    'zg3x8ScfTdwWDyemhdrbVNaxUNubqljLZNSpo/6xKtSyp84c2erOis/nF+afvFj8ajH75NlfJm4/+uvCnx5//Xz+q4XnLx7NLy4+Xlx8+njh+fjTLz7PZMbM'
    'cmqvfNGR3hGW/Cm3pfY92bZEVoSeo2rBr+fu/5SrzPYm17qYfGIp1+LBYl25TnbuvpBvmqpmi2jjm9DLYUlVzYkw8FQ9J+LpmSfY/9kCVirL70vyoCtU7cew'
    '7Zqlo8qRLBZksT4u1OuybAfYqO2rqiWiTYfWDr2qqmG5ZV/tHtGvaO0UH0UVW+20RFRxsFNBNkrJer7csIQsJ8vKHwKcNmyfYmX8DPaVW7no0LziuVA7a6pW'
    'UjULp/KV45IBUaUZ/svGjs3wPG9WyWRu3LiBIXs8CYflwHdiUpCTLNnyhapvqkaOzn8adnICPg2PfViRySwJtZKPVnDUdi6qbokl83JEFWx1jq3sslzfEqGf'
    'U54Hi3Ny/YAtawVq7+AWzT+x5WFLyGIzRJyWeKrrCHW8is3hWhFVKwgcUNGKKm9pQuDJ/VpUcUV4siKWcISxsTEx9P94Lz0vsjzaJQwQzjJepf6lCY18lIfL'
    'XxZUoyCUm0MMhk2Sm2XlFgbehic/yn2HfB2tFPip7Q9OAoBW8kJVVrWv+t8qx4rszWHfwl3y3BLAF35dPSkzNJZTFMttPwyKctkxHhTyzJIHNuEC+CQXR68t'
    'LMpBLX2LJfWEEVV0lIucUlYXy+FvYxOhYjTzacocxvQSeDZRQx7Kkwq953gK1cjj17Wi9mBywAcPpgaHpgeHZq7tl2nyi1uhIyONpIcXu1s4d9juwvoCGQFG'
    'oPl2Pto+QP6N3AORcEqt5G+Rq5CjkR0QLtMLkMXJJ3LtJVIRjqtvhh45bvmfeCavyVJOnfgJxAs20uDnnIM3k/3GTvU/Tvc/zvQ/3u5/vNP/eLf/8V7/42+u'
    '59EZZg2GalQqhb4Pc1u+dA8I8eAGIgX1XR6AmiUvAhA5dd7Ex0QTI0KIW2YW86kLv9hEiA2AtLKunK488ogipNs0y9AiOIey6/LI1yuM9v1BruSiNZeyE3GW'
    'ywFzJ4fWJSqObGRVx1Hn1VlxUwfWkOdNjg9m3TQHKzqhl4/LAI/JV8fglNEE7prFPjL761FZTs4Pkgdp0R+ecZMp45Uf8zDA4wM2wtQJU6LOg0zmLqMuTbWU'
    't3d4VBMoTi4PLeb7HS888eA1ZOdoXJ1oo0qJIEbGkbP1avTrXc42i9CB3Qr76cAGYPlbY8a7XJVmfddFYMuEU/A4jKU4RRvg5maq4KUY13sp1NmR3CipnTLC'
    'TDQk18s0zpFr7OEE5B5TKUyJSSoFe26lwOc/28LhUmhYP0326+Nm5Bnq96p0z8leBBCWuA4ZQ34CpoA51F9d7Jmyt78hx1GlH7KBruAM2N2t8BTm2nTaKyCO'
    'jaoBqwWGHUdDeynsBMATWXLZXzqbDC2DezOZpdTTALtOUdhjiqVCiNdmekxDkAHpIPUI2fDwdQh4gFj/4wGCNLQUzhbtfnvR0R5JZJA+ck+4CV1oZsW77X36'
    'TBZPAXUCI2bCsRa/MKNqx5d7JfKI+RrLzN0HxsKTEuCwHwawdWcVgeECxxKIFR4LJpMVOFJa0gEmr8smssggFAGaAE2QYycz/Ga5PC636DhxABweKNbxfc2G'
    'sqGA6cwJO93w2IuxjRUI1wSAigUc8Zd64qV4ZC65dOBvhmsHAU6+OVJunrY0cEb9OXnLgG0FBANowC5pC2LeZFbsxSGzGOsBJYk82oIU/oWaNBQEl55YJxEG'
    'yHKKMEtfUwf3PfYHkB7Hpm2hFYBTi3UkqEgQrpFNJVqHlayCviWBSulOGlgTGC3s2USosWH/D9RnYIj6vkuqbBrK12gD5hNja48WNfC180nMmxZAw4QrFMJT'
    'dBJk2VTuNEem+gRtrbFe18l8j6sYG6aAMBYY1gTLVHh1H4Lk6YWgf2cnxi6ca9oyy41WaoRx2pZaLHatgc5/CxCiCJ1rac7k0sSZxrxr2jCkzVvAPcnaBM1x'
    'spJgfdPE+/jov5/79NPsb+fuf5ylBoktzAI+XOPOCrRk7IyYneIi+78B/M9azmSETte4MN2HUjBqJW7wVm1SPQQNyyZSS2z9IejrjHRPS3UAHApbZpkQ4z5W'
    'G9drcUV4WoaMQubHDW48RZNs7C39lEYTj6j1c+xM7bVezZxE+x0NNuvDuHkm/q1aTIccN7bcxBU8CEWi7Rwnjic5TGyKopz3tWRg6izY4IBUxRwTf9Af//Hq'
    '7pTAkNLv0Di6Cw32TYUc6dP9t2LdnpL3rKWKNdm2tJbqKXqkZD1aD+ThW7VepwTEGAIZbRdwVpg8OthszeBMMKe/knNPnm6UeEsk805vobTmo5uMK63l767T'
    'cRssTU5M/IokEiJsbBvQdOljkLs/+/JF9nfP/pb97Mv557OmQMB5Flf2uKpwSUryT7N6mhXj2o8v0qRCx4+JJxXhS9mdoFlfdzA+Wef00pll1mkLPGC4U8O8'
    'p8XWWMIxemUZFtc5rq+8AcWnAVx15EaVvd6wgD8mCXVYS6TLkpnS01cD1MSugL1JAeaAJiLol5s9uijTu0wyigmZZ0fcofYoQGt5HTbEDBbMooH++4MZxhzJ'
    'btPE090TtTEv4yNpNqWex2tSkKqXrgMY3s2ucnO9tccTNXlYo8JFCWrUR1pe6daYTGWwaJ2WVY0Caw6EM9otsTrXQMnq6s3XbuYOi0+o9lbjAqrxlroT26Zx'
    'avyMSNBEwxKLDb26W9BdmsYV2ffnZ18/XczyveQjuPsRP48//YKdoYFqytx17yNHxeBqo8P3oSCZri0RC0v0j+nwexiYYgxA6VZT93CklfjyYYCXGiX5ymU0'
    'DHaQF51U/3jRYebT67C2YrIDSli0XrrPUX5BVQqUtskOHzDxHjEx3YeJaV3dNOnG3T3brhUxXblw85IWpQ6jYqh2jRW5Lnymi+5rzllWMk/wpUFcgqiBdjQe'
    'uDlOLn24MabimUjMD6h5v6iZ6UPNjK4mzCGU8FcoAEsWWzFSSMAkwTJ9xd24mN1Lbh3S92mAxMOZi87DO/jvLp2MJaz/IezvM+z/BlBLAwQUAAAACAAAACFc'
    'RP25GTUKAABqHgAAEwAAAExPRzA1X00wM19sZXNzb24ubWTtWd9vG8cRfudfsYARmAL0W5Tl8s1BATdwjT6leQgCR23cxkBsA5aKwoARkNJJpUW6oizROspH'
    'm6z1i4bcnChKPKFKH/SnOG+3e/9Dv5ndO96JlOIENvpiILHu9nZnd2a++WZmeUn8/g/XRyfFzdEJ8Ta3JtSLLXlsCf+gjqdB4bd+lA1HqFo7mC8MClXP49Oh'
    '32oKVcTwC0vutYXaXFE7ufS1P91/MDsgpLuMD+Vg+c2IxNROTqiG6+9jWr2ianYqdemSkD94asEWaj4fVJuplCweq2Jd1XLCd3MYxFRnShVstdiAbFnKCWyj'
    'ap48aPveY+nakCN8z5WNmpBHlrJqQu0v4DRCrb8JKjj6giwWZHFz+Of0cazAXuHRg3ZMJbl1giWWWnxy2tHanXa0Mnh/Xib9g40STijUxppqvSET8BnX/6Gc'
    'E96hZcvtPTrLTmgAVbOMDYKKjQMNiqBUlsVdGg9qa3LVkUtrNIPOAitLd98sYUkO6ZNKDQkcQB54OMh8QVy/MyuCf3rK2hfjw5NQwHetn3JrqripvIbvPoWi'
    'QlUsWbR/ylVo7aGj/rUg1Jyrjhy5d5IVX0/fm/7u4cydmZHv7v91dPLW3+79+faD2ek792Yf3pqembk9M3P39r3Z4bvffI3lOFWweAgjrdCOqlKQc21VzYlr'
    'nxnLmzGyn9w9UXXoXvtRLR0ad7Dz9fnJxxBHUJGvSiQklGo5qliDEsEKHhwoDiO3lZ0XsIuqWoL8b+dJH0i7RJAZTgoTYwKaPgNWinLOiTAFIRvNEDaq9ERk'
    'CJVf+odl5TTwSc/TIr5KpR7xnEfhgmvdx0/x2BUPB6t6WTzCgqGhIdHnX3y5MUbL25jpCMDD73hZ8fkfh8aF3MkH+T0RPC2onYLIyNV9IDw9NpodnRygbRou'
    '/EVAI+wf1LPhAplf8o/JZtuqVuL5E9/Tv1O8qujgSP7BiUEYHVxtYra7S4DgA41j3ue/HRqDOd3AcgExCFuTLx+LibFP0rSg1RbaBQPnT72KqVqroLomm23e'
    '3TohH1YpmvcR12yhdVdvO0GySHGEnjq2xadDE0kJAlqMjQ/0TMuku7bwD/fkSrk700wKt3t1guMnlM1o5xEMZLENusnC4k8Q3kIbXIwnbA3VfbcMW7tr8kUJ'
    'citBZZdmeA1Eo/E4mZcDLt/dKnUuJMcJkswJOMEuuEUExVwaq+S2xcQHgqpXBiOqYNKQmKajYoDxmFj/yJAJHmTBwpagMRdDIBfwCL33hSTGNLNhgP+jAbfM'
    'VBeyo8Os1f0OVUgN8nuloOqaxhIzDNPHlvRjydhnkzk6u2qxFI0b88k5ezjJBjqi49km5GNjMACVLYSEMo8IO8gBTNql+JimVAJ8SLtMDEtwc11yklzaInqR'
    'ex6SAwPpjIUNMnVygQmAUZgc3DUHQn4TMQIRn9+av4gDIuN3Ywi4g+FD5bumTDACKBQOPW8SokDVCz1fE1kuzG49k3Q8qMqCtlXyazIz9nyNZ9X+k8735XiS'
    'naM0LrdsTgPPy2Ti4LkFoV0a1hPSmtqQ/Jhk8HdnBa4iNLT5NGV2Y1wE3o3XpNuUrQp9N8XCTh5P7+Q15u+kDZhBzwxN9A5l3tkuE2SXePbnsKl5RDTyqEBK'
    'IK3TfDsfPNuSy1b6qu+ym+FIZgjZKgS2R7iMCyCNoyVyEUVBicsa1Aow3Ny/8U5WQ53FnG8gXrApc1yc1saSyo4nXyeSr5nk62Ty9UrydSr5ejX5+pt3s2iG'
    'WYOhGpRKfrsNdVHc1bcI8UT3IAX1Mg9AZQVXSiD2410sJppICyEGzCyqZ6LSjSu6yhIVek2XKELWd40YEoJzKHuTMhlLGEz8QazkgsU6RSf8LOc8LlLZtXXK'
    'toGNqOo46riaFZe1Y1FZLcJDl9k/mHXZHEynHpPZeUznssEI7prFrpv99agsR+enrNOapz884zJTxmo7rNNgUS6gkelwZsOv/CLAffJ1sxtYZ0BC7mNgxtlY'
    'rZeT9MqxHiuM82qdasS0PC6oY0+MQgKhUmvLpHOWM6h+Y7HxcCexiXhPSE8K7BNsNDsWVUjt/wWvdOOHxPcLlAt2uRiA/aDFO+7kAscGupCW4eNCHUtbFUqA'
    'XQihqoIhg2c9e4aNhQ1XoQqxaBOUMP5/gPdVl9CdwJppM3oEpXS9jiKlVYh6tngtTq3GkaPLA6GOuOpQ1qZ8vcDleR8UwHxXeFRDISyALNal5Yadka6aWMdK'
    'SXAJsb7I7RFLo6e3OdsIIUegLKOA2rLBbLri0u55m6vSrJdRKRwzyDKS+G5QaYa9Yiw1u0+hTlMul4Av8AFhTy6Vhemo5M4LnICcZUoKU4tEJQX7cb7A5z9a'
    '00VZGHZRM3QmiXML+XpB1o9JX0Q6NOnWXVxwkUMpB9R1bn/2mAxHbcs5G+hOi5ltYw1Fsy4N+0ERG1U9LtkZXAZZ3AZ3POCdNDlrLwZHUKliQxjbkctVTRv6'
    'GbmB7xPQUR01OYX94IURo33IwM/jQBCWSt0Y+/5GhqOZzG2yvO4/YHXYgmt8HQzzYcndUy/EG89Q9nDiWON8rGufdekHm0CsJgG5sxkseXL7jVraZFk7Jbla'
    '5wP2gvm0E4PyaYdpQ8vh3oApAwfXYZGsQVS7QKU03BftkDymvo3xOyf+vhthv7gJn0OiLmc1shgC7DR9RZBKUTP+6oQKswm+dmEi463AWw0TYhHGEtDl2w22'
    'JocUxyXmEbwcnakYOlHuZNiAFY120SVFTJGMhgFbOtaRdsMQ/YQs7oWH5548xAl3DaedKW11sjBbl3GRYK+bmdPOzSv4f4oQJC3b99pdiC4/phghf1ctYgad'
    'M7tXKTiMxS02MxylPIv4HUU+RVm+rTmAPVCw4YFYCAyJL/Xir/r3JXiNV27aqWgqvIZBSjqRhAbCii2WgtgJxRp6wdBzYS7qRSvy05FDNM4N4OB7SJmhoDju'
    'uS3tp22Il5/ttcwl2djo6CfEeXzNw7r1QUf3GGTuL759OPK7+38f+eLb6Vk06MRuHjW0wQZd1WluBM45Ok30aNjHg4HbBYABK8x9mFUP5mt8EWeAHPOwkYrs'
    'BPyedhBe8jXlVkgBYPCxCGji4JyDjgqkK/Hm4R5aFmJcKhIPy7BEl1wXmZOXjrkgg8ab7NdVt4fCdQLmQ7KutgkrcoFxWYjmsWG+xNXffxH5ZsUHJN8omLdr'
    'cIahMc5g8fsR3biQX9ihuowfQQMMvLDJ6SKB6YqdOaJZSRc4dJ9jroBiF78aE2BXI0wXN1SWaxo1ZMCJzVw09kvRuobWvif9/nL/wd2ZEb4zvwVz3+L34bvf'
    'sDE0mMCRqnYi3vVedVD0Shs8fx9ykimVIoyOx13/KxNctk/N/p4T3EcofHgoTMSh8KuLiKz4vxcRH8Hy4cGSiYPlFxVqWfF+C7WP3v7w3k7Ffji88HfDEV2Z'
    '6GY3+vVwmCuNBmGDQpbuKw5VcY9+Adxo6kF4yjgK3TH/iBDb/n9QSwMEFAAACAgAAAAhXAVecXsgBgAAyQwAACQAAABMT0cwNV/qs7XthrVfQUnsl63tlaDs'
    'i6TsirXslYjrgrQubWRtVk1v20YQvetXLJBr6yZpg7S+5RS0aJoC6T05FT00KND2B1A2ZdCmWkuwaFMOKVCJZMmpgtISZVOonIN/io/c5X/om9nlh5KcCFHc'
    'nZk3772ZO+L7p4/vPhDZYpnvLcWjb4U6eZd7kVDuSB0shfIcuZM0GnfuCHlpKzu8ucrmVr6HD057av5OKN9RbtBo3FqB8mw58M2F8t9UtXyhBh35pq36lsA1'
    'Qu3MVGSLvN2R7lSFtpBxL9/1VLCWw0CoxEE4fn1o44ga+VvmAvo773sqTGXHlwc9vskdIZllRwVDdXousjRWI0uoCHk4N1fqspfFljmkwkTkno2Ucy+gTHCA'
    '4mTLGarKPT9bRCaSkHMn91M5XuOqpjpOalmokw7g2eP4g7Fc2ULZF/JgjGhukMVNPANbDWw87UC+bdH7UX68L1TfpiJwTk7a8mikJsBj4myJx/LoIkv3s8Ua'
    '+Z1n83Pcj5QsCo/M5NlMJ/cZw6dLVKEvpGPL8Br37zbVJZei4vfAXye7DNTrVokyQ89YeFE2b2/dWmGjIadrFdGrPrqM6uh/NXHkLKVcikC4NvIKeECIKDBt'
    'Z/DSWA5DpC1dR7qjLZCHANJhcdAu6pgM6T6UrLwDRKs+N50QX2YX4EaYUDUU81XH5KEmTRV1BZBGn00ecpGajjKW9CIBviaxmyvcjW6YxlO7s6t1dhFzJjvT'
    'bNU08TWn304JzoLzp4nym43G5+LJ/W20uIv2AhsGnPvEZS+WWRxQMmlMVVLNWdwnnJEe8CrQ06TWXZSdgOrV0uFU4lc1KOQhzq+rL7gbGm952JJn1yI/3MdL'
    'oiERM/QhL6QAUJmMdVwp+6+2cRf/M4yhFJElloo6JiONBxKz1PACCsK1o65a+SL/O82Px6y/lY021lqLioGmUBctTe+Jly3wlHMPIkHSYEJEBwl1ZsxH7aa0'
    'HlBaVJumjpBuIt3VzRWcR7lR+VMdrBAsu2yj3LI1zE0WYBSgBZVae9AcA+0uawmXMu0Y4d1ahJk+fGsB7kSGa+YTnUuUxwosj3MKVXymgLuUC4v1ZFgT+tl8'
    'Rjedxx8oaWcmB+3a2Q8kaRfMJovZgOjhtsh9eMhMqFUijwJw3OhKjf4q6aLFWynhPlhGAmKPnf9Ti1OIlKWTXaUif4X2vm3JaCUI0AGZF0jyuqUcvygsP22j'
    'H6AGHyYUChvF2baV93t1qDRP9RmjXf4QshwmG/aAFLtAEcVb6iSmxKtkCylJMG03zI/pjDxPOGd2VxtDoQiWza9hD+w2h3DIaS0KmkxDyKBHSsOLM8D5Zv1J'
    'Tn69LWTk4Nu8bXFGV1PlXZNvUEKI9WDTnApp+JRc1QHUbbhGIXHVMAarCFDZpaJ8GXtA2vEhD+hmkcixz7cGUTFqNtL6ZpuQyLsBN4ygYwFsWCWhVZnhJv2N'
    'eegrNH6cfS/mSZsOWabntYj37m5jHnaAFetazhLkCwvEvK7xGJ2LGAfjquQnbFmbjdYXlAcrN8P7sR61s7RQOuOH3cHt1635yUOhnACDNPe9YhsBBTCyG40X'
    'vGM8f3L34fPyG93x5/rLrV/+fPnrC654GKj/ZkKmbehdncYQDiNwcl1hFaBOeoif7pHetMNj0WDN8A/ugmZdiSZWim6AH7JTkY/bYRiN8gBVuZlATuUgUgMI'
    'cGxUCWfakAFcfswOZfOYP2uViYZJpV40aVOsxC03+oI+Ot5nv8KIYNkKkCXfZTEUdsB+R5MybBupsC2bwqt6wF3j1XrDYfdbkBTZclni9e8nDo2kyT5x87tn'
    'T38g+KFgFgWinSb4TVrTlf/82+8v/6Dd0LCDiMP2ohetGHOlxtGG7j7T8EO8yqZYBluzcmWrNlGWjhRdI7OGw3AxkE0zlDtBWZ3N1Dzslwr5eJAZs00ccIFH'
    'DJtZTcu0ZQCzjs9muEj1AbOe/vjo2TOGLu5l79OPLy+aq7vKytCTDOQDe4sxxjvOF4Xnnuyhj6AMs7PuKbw5otsYq7DwgNcMy0CjLZY3u6NPbGc1sI1pG6D5'
    'cEF9JKY3Ey6M912WuR4wmpJkI3pZHnQJ2mol3xA9muIwmeNpKcmIpxo9zCgiTWzwqGTH/1BLAwQUAAAICAAAACFcDXFNn/UKAACiFwAAHAAAAExPRzA1X+qz'
    'te2GtV/snpHshLHslYjrgrQubWR9WMtuG0cW3fMrCjCyyUiynn4sDUygMTDCDLIJkMHASGaURIgjBZKMIIAxaEotgSbpmIpJi5SaTNOWLNFDwy2ySVEwtfGn'
    'JLuu6n+Yc29VN4uyPQs92Oyquo9zzz23rom//m1xekFE3V682xOqsafcM6EqObkVZjKffirboaoPhOyG0eCRahzLC1fEubZq7Ajlu0JtZ+Nt7925OttRDRd/'
    'j/bUiaOaQXQWCtUJ42xbvmyr/FGcH2CzfFnI9kC5WCH7rnLrcbWCrWQhJwtHUyIKHGyIjT1xU+WqareJ/WTREdoKbYQMqqpepZdgs4gGgWzWVR2m1C+jri/u'
    '3FU15+rR786j82F0FgicKo+rOP3VKW+y/zqu+AJHxLUynuKArg8v4MLFKcyUx0MTkrjSSqxUhSOyUAeD3oh6bfmkFleqWIfQ5VW+l7wr+w42x478MszotaOA'
    '4gTvLqp0UpZCV8e6w5LsDsioqFdSXlMdtNJ4uDjQilMmc+2aUCdu/OyRUDkPdmYyMxS9rLi/vLGxtsquNUryRZEd0IH5Zm39hw0ksqe22nTSfknEtQrCGlfg'
    'zFYbbk0IxEUeh5QnPIkrryconvulqFMU8qQZDQbYbUjRlqdD5cNqP2t7W+jJrmN9LcMj5Q1lKxS/O1X9LWIeykZRxIeuCe3vTo2WxHueKnijOGdmp0RcCWXh'
    'Qsg3A7WTuISg1UPCHlKvjnBGx8UnPD0sIQVketQJou4QT/plxA3v5ZAYigT7QriVBThXpagkyTwoq85rLOn81/aiG6pmaJk0N0V7UHRfDJFSMccwhznbWYbg'
    'YSmxkXM5QFm5aivAvjgmn6AVxhHkUCUWquRJg8tjK0Tm5RNs5vmMJ+zu+ZxHP0sWxxUX1nLSACLggg4GjPw944Vl7/wU+1BzbadgSsFLc89R0NbiKKHO3Hi7'
    'rgHqV7gkUDCM8k6AJ3Yd9Dz1fEfExSIwwT6hMquAq+cC27REbp1GF9nEmIUpYTKn88lYN6nSucRxbwbasMTSejF528pcasLSzDSeM1zT8rkVBVycbBHKT7MD'
    'yA3rbOt5AWrLh79iaf7d+dIN/Nzi0PsuCISLog4eCXEGkQEdk6ui6qwIZ7747uc/nPIo3fHB43fnd+5yFBquSd8fTmVC/GXtJ7xpAJdQTsJLDAcON7KleYES'
    'yrDh1V9899UmLddBT+Jmzo2LJVk4xUc7A1hGSYifPAKPWJ7fuYvPwFtRPgX+ay4RvmY+PCNeVX2PKglVizOogvASQE2rVAUZqNk8NIK0NkIzWBtEm8lMwgFf'
    'Pg8QUwre4gwhjEgNpRMFFQO+xRvWY1W5tEzl3kQ7YmPElqIpX7pEl/haHsPxsiubnvLrIHnifX4cyjdIuycYzykukrUu5/XAxgLIaXHO1DGzkQNb350vzuJn'
    'boK/nqevYTC+ZleAl8UF/NwgfsNZ8tXOOHtNCrn7K+1kMpxGgtuTD0p1UMqtqNNK3HmZ9sHPZzSDl+Vvj8St2U+Ejp7ZSu23GaQGwNUcDub3+55k6POy29Of'
    'GIZkl/st1TwzLUq+evuxdpYuCXWqizqIDEPOrOlI+tBRR5o0LoHELYsQOHygbaly/WNgEdV7CjxVJPXuwJH5YxNwbTYQaAKqaUCWPtBU1W4RTJfsqRmXwbm/'
    'a7oNmJ8e6D1MK7N1Blh50IyCX4mYPfRN/GXJggoyuNIk1Cix0xz2pNP33bEsJ2LoxRD4ZzqGIfkLrE7I0zA59yPbKiZrAYUjqNq2qpaB6W5MYt097vTsp1nG'
    'mOm6VFgEPptnKROgKZz+9AxyickoYX+HYoHuy3BhI5lnQT4F/0P6hfLacKG+2EzVqUJRWUVoaRZEdPfxBFJawl/QWueSK5A9n2BSgfm0uJJj3YhA0ReAQXRO'
    'LxwCPS0RHxRhuvFxIj37/BQp5xbI+41Xrpacpi5JvxCrsaWGikDlaJ1Cem/HnOXXwVfPwveIDfTqqJd1WfJSwDM8KBs6gEY5mhaWdOH9AB/G0EFt49UOIaGn'
    'gkvSnTvSvzCNTHfZgs+KhlPMPMt9BusosWO0oDfhgmzQE1u/cPWyIxxZkgdRZ9vUknoafECanmQhkbkVZPPRIDRaftTrElvZRvavgaxXeUM2NJVsHwGOHhaK'
    'xSikwmqHqFb0PIoluhPRIJoytoZuABmH3Ek6un7cI+qkHxLURhGq37IE8xSDyCrbBc+9ahQ6WlHkEFpsSRgibtr1WURcyVISQpLdvpt2Mcc8YKVe45mAtoZZ'
    'MBrbyHIwSo1W5M0AJUkYgQJBaf3pzt1MJikRR8z/5wYiat4ylMqA0sfrNo0qEbPUS+pahfgMA8tYvTXtF1dqRMqNEqeQpymRjmGJNNO+uMK0Wt36TSwaJUTH'
    'z8mnXkwjythIxJVy0rQCTyxi3sAEgPx5vHGCP+T0EPuNvNBvfRR9yTpDRyGRN4EORC1f6ZYI2cdSlfu6PlrLHlY1ldPobdWoD1YrWkSRbEN8DA8Drc8DkIFx'
    '+b1OkOQHEdvOicWVTRH/gvH0jBbP3uTe4Fq0vHSTd70IETNKoo4vEtbX0mQrnBBLt0dTbtLnZ6fxBoh4XszEv1RJZ1A7mKNl6NioIuY2LfkSXT4aDPRWdNYT'
    'wEfbRGaM9W1zkv6au8YYdSRiQaMuoAYPH3RH167UPMywZj0LE9rDzF1o6BhwzMkU4CGE9bjW0sqXcIGdW0Fq3aDJI5jO20fAoKtHW2jGFZ2oTGZpgUGRH1Dv'
    'WNAqkAUUA53Uxry64Ca2QLJfnhzFlb1Ekp001LlrYQ7pM8PCiF3R60fciY6JD4QmIyAvkiiZYb+b3HjwdNN3YZWuLPSqQh1fm3I0YDZIof/s1m8lN6EzfZCx'
    'Tgs4lie6P6aKZ0zqPAtN+zRCSpAw7POgS+fQrOYLbqd4+X0GHfmGYinUzcBCR4zGKaZLPok6eCVn81U6pl0VRWLpFtfPNLo5j8SYNpNrGiMrc7I+xPBMahV6'
    'gGw42SMEoRL0CBJ1e5B+SRpQmzANLyAm4zSgtut8OfOBDmSjKuniyF2tlck8NDQjHiYXXnrsFg/x1eTkpDC/8cms1Pc44qF1RfV/b6gm6Cz8a70/F/UCWnLi'
    'kFFF3u8f6V2Pfk8n+J84mY3f35GPA9IGeaD0sEg3Wv0s54CHtEyG56N7P62tf//12tr3Uys//rz6NeMMvQGROWhZ5KXlotjYfPDv5dXN6xtr9x9srqytmvmd'
    'h6w+mLutWdhl4GjJKXOurF+myauNhCZfwanLskF5crdH4y98bA/0xQfxwI7hJpLabLz48u7fdZdzTM3RBnThAiJ/4pJu00yS9jk+qNwivMVVV8+m3vvabemm'
    'wbbZxtyqSP9IBod4lbxiQQOHB0UAUR0EZATrHkL32IAPPsebI9rm3ZizWbOip5qJaRRoHVCjIa2KGWNqvq8iaOqOSTeDJegWS5W/N2swJBCTTs5cvRjtrVsW'
    '6w42JZOxp+K0oaX9jNmRVEx9aKsgvprBfkQHJPnoXbOYax9bIOwASDoiXkEDe6mZKm1DskBkQxE3fjkahvaswyhC72Wr2/itZ3kVDBBec42ZL2+sPVj/1/K9'
    'tfWVb1dWv7q/cf3blc1768vfLK8vr+L57PTsjZnp6duUxkbOFhoI1TZmo9PPP7vz56XPsLH2YuWHH9fWN4k5n57RBcbVLOr06EK4rm+JYH0ygFRpMOVbotEl'
    'Ud2ZmaZuejy8ekuUXht56c0XkV9yM6MVj0uzy17paovI/A9QSwECFAMUAAAICAAAACFcwIXVkkkHAADDDgAAGwAAAAAAAAAAAAAApAEAAAAATE9HMDVfQ29s'
    'YWJf7Iuk7ZaJ7JWI64K0Lm1kUEsBAhQDFAAAAAgAAAAhXNc+gHR6CQAAcBsAABIAAAAAAAAAAAAAAKQBggcAAExPRzA1X00wM19mb3Jtcy5tZFBLAQIUAxQA'
    'AAAIAAAAIVxE/bkZNQoAAGoeAAATAAAAAAAAAAAAAACkASwRAABMT0cwNV9NMDNfbGVzc29uLm1kUEsBAhQDFAAACAgAAAAhXAVecXsgBgAAyQwAACQAAAAA'
    'AAAAAAAAAKQBkhsAAExPRzA1X+qzte2GtV9BSeyXre2VoOyLpOyKteyViOuCtC5tZFBLAQIUAxQAAAgIAAAAIVwNcU2f9QoAAKIXAAAcAAAAAAAAAAAAAACk'
    'AfQhAABMT0cwNV/qs7XthrVf7J6R7ISx7JWI64K0Lm1kUEsFBgAAAAAFAAUAZgEAACMtAAAAAA=='
)
_EMBEDDED_ZIP_BYTES = _embedded_b64.b64decode(_EMBEDDED_ZIP_B64)
_zip_path = Path.cwd() / 'LOG05_실습자료.zip'
_zip_path.write_bytes(_EMBEDDED_ZIP_BYTES)
_log05_extract_zip(_zip_path)
```

**교관 검토 기록:** 판정 [미확인 / 보완 필요 / PASS / FAIL]을 실제 기준에 따라 기입해야 하며, 자동 표시는 하지 않습니다.

```python
MISSION_JSON = '{"code": "LOG05_m03", "steps": [{"title": "상충과 미확인 상태를 적는다"}, {"title": "AI와 조건표·공백 매트릭스를 만든다"}, {"title": "근거 없는 실행·수치 추정을 검토한다"}, {"title": "조건과 판단 한계를 전달한다"}]}'
# LOG05 학습 기록 도구 (표준 Python/Jupyter 환경용)
import json, zipfile, io, os, datetime
from pathlib import Path
from IPython.display import display, Markdown
def _load_notebook_widgets():
    from importlib import import_module, invalidate_caches
    import subprocess
    import sys

    try:
        return import_module('ipywidgets')
    except ImportError:
        print('이 노트북 커널에 ipywidgets가 없어 설치합니다.')

    try:
        subprocess.run(
            [sys.executable, '-m', 'pip', 'install', '--disable-pip-version-check',
             '--retries', '0', '--timeout', '10', 'ipywidgets'],
            check=True, stdout=subprocess.PIPE, stderr=subprocess.STDOUT, timeout=60,
        )
        invalidate_caches()
        return import_module('ipywidgets')
    except (subprocess.SubprocessError, OSError, ImportError):
        print('이 커널에서 ipywidgets를 설치하거나 불러오지 못했습니다.')
        print('설치 가능한 환경에서는 새 셀에서 %pip install ipywidgets 를 실행한 뒤 이 셀을 다시 실행하세요.')
        print('설치할 수 없는 환경에서는 제공된 Markdown 작성 양식을 사용하세요.')
        return None

widgets = _load_notebook_widgets()
if widgets is not None:
    MISSION_DATA = json.loads(MISSION_JSON)
    CODE = MISSION_DATA['code']
    OUT_DIR = Path(CODE + '_results')
    OUT_DIR.mkdir(exist_ok=True)
    BACKUP = OUT_DIR / (CODE + '_record.json')
    fields = {}
    for no, item in enumerate(MISSION_DATA['steps'], 1):
        fields['activity_%02d' % no] = widgets.Textarea(
            value='', placeholder='내 판단, 확인한 자료 번호, AI 제안과 수정 이유를 직접 작성하세요.',
            description='활동 %s' % no, layout=widgets.Layout(width='98%',height='130px'),
            style={'description_width':'75px'})
    for key, label in [('first_thought','AI에게 묻기 전 내 판단'), ('ai_request','실제로 보낸 AI 요청 전체'),
                       ('ai_response','실제로 받은 AI 답변 전체'), ('source_check','원자료 대조 3건과 수정 이유'),
                       ('final_result','최종 작성 결과와 근거'), ('limits','미실시/미측정/확인 필요 및 교관 검토 상태')]:
        fields[key] = widgets.Textarea(value='', description=label,
           layout=widgets.Layout(width='98%',height='125px'),style={'description_width':'190px'})
    status = widgets.HTML(value='<b>아직 저장하지 않았습니다.</b> 작성과 교관 승인은 별개입니다.')
    restore_input = widgets.Textarea(description='복원 JSON',placeholder='백업 JSON의 내용을 붙여넣으세요.',layout=widgets.Layout(width='98%',height='100px'),style={'description_width':'110px'})

    def payload():
        return {'schema':'log05-student-v1','mission':CODE,'saved_at':datetime.datetime.now(datetime.timezone.utc).isoformat(),
                'answers':{key:w.value for key,w in fields.items()},
                'status':'작성 기록 (학습 PASS 또는 교관 승인 아님)'}

    def save_record(_=None):
        obj=payload()
        if BACKUP.exists():
            stamp=datetime.datetime.now().strftime('%Y%m%d_%H%M%S_%f')
            BACKUP.rename(OUT_DIR / (CODE+'_previous_'+stamp+'.json'))
        BACKUP.write_text(json.dumps(obj,ensure_ascii=False,indent=2),encoding='utf8')
        status.value='<b>로컬 저장 완료:</b> '+str(BACKUP)+' (PASS 아님)'

    def export_json(_=None):
        save_record()
        raw=BACKUP.read_bytes()
        try:
            from google.colab import files
            files.download(str(BACKUP))
        except ImportError:
            print('Colab이 아닙니다. 저장 위치:',BACKUP.resolve())

    def restore_record(_=None):
        try:
            obj=json.loads(restore_input.value)
            if obj.get('mission')!=CODE or not isinstance(obj.get('answers'),dict):raise ValueError('다른 미션 또는 형식이 맞지 않는 백업입니다.')
            save_record()  # 복원 전 백업 저장
            for key,w in fields.items():w.value=str(obj['answers'].get(key,''))
            status.value='<b>복원 적용:</b> 기존 작성 내용은 이전 JSON으로 보관했습니다. 다시 저장하세요.'
        except Exception as exc:status.value='<b>복원 오류:</b> '+str(exc)

    def export_review_zip(_=None):
        save_record()
        target=OUT_DIR / (CODE+'_review.zip')
        with zipfile.ZipFile(target,'w',compression=zipfile.ZIP_DEFLATED) as zz:
            zz.write(BACKUP,arcname=BACKUP.name)
            text='\n\n'.join('## '+key+'\n'+val for key,val in payload()['answers'].items())
            zz.writestr(CODE+'_work.md',text)
            zz.writestr('README.txt','검토용 보관 파일. 실제 평가 PASS 또는 교관 승인 자료가 아닙니다.')
        try:
            from google.colab import files
            files.download(str(target))
        except ImportError:
            print('Colab이 아닙니다. ZIP 위치:',target.resolve())

    for i,item in enumerate(MISSION_DATA['steps'],1):
        display(Markdown('### 활동 %d. %s' % (i,item['title'])))
        display(fields['activity_%02d' % i])
    for key,label in [('first_thought','최초 판단'),('ai_request','AI 요청 원문'),('ai_response','AI 답변 원문'),
     ('source_check','원자료 대조'),('final_result','최종 결과'),('limits','한계 및 실제 검토 상태')]:
        display(Markdown('### '+label));display(fields[key])
    buttons=[]
    for title,handler in [('작성 내용 저장',save_record),('작성 JSON 내려받기',export_json),('검토용 ZIP 내려받기',export_review_zip),('작성 JSON 복원',restore_record)]:
        b=widgets.Button(description=title,button_style='info');b.on_click(handler);buttons.append(b)
    display(widgets.HBox(buttons[:2]));display(widgets.HBox(buttons[2:]));display(restore_input);display(status)
```
