### 실습 프로젝트 ZIP 업로드 (필수)

**미션 실습을 시작하기 전에 `SEC06_실습프로젝트.zip` 파일을 반드시 업로드하세요.** 같은 프로젝트 자료가 들어 있는 정상적인 `.zip` 파일이라면 파일명이 달라도 업로드할 수 있습니다.

1. 바로 아래 **「실습 프로젝트 ZIP 업로드 및 압축 해제」** 셀을 실행합니다.
2. 파일 선택창에서 실습 프로젝트 ZIP 파일을 선택합니다. (Colab은 업로드 창, Jupyter는 파일 선택 버튼을 이용합니다.)
3. **업로드 및 압축 해제 완료** 메시지와 저장 경로를 확인한 다음 미션을 진행합니다.

업로드한 내용은 `SEC06_project` 작업 폴더 아래에 준비됩니다. 기존 작성 파일은 자동으로 덮어쓰지 않습니다. **ZIP 업로드는 자료 준비만 수행하며, 기존 실습 코드·학습기록·결과 저장 기능은 변경하지 않습니다.**


```python
#@title 실습 프로젝트 ZIP 업로드 및 압축 해제 (실습 시작 전 필수) { display-mode: "form" }
from pathlib import Path as _SEC06_Path
import io as _sec06_io
import zipfile as _sec06_zipfile
import stat as _sec06_stat

_SEC06_UPLOAD_ROOT = _SEC06_Path('/content/SEC06_project') if _SEC06_Path('/content').is_dir() else _SEC06_Path.cwd() / 'SEC06_project'
_SEC06_MAX_TOTAL_BYTES = 750 * 1024 * 1024
_SEC06_MAX_FILES = 5000

def _sec06_unpack_upload(name, raw):
    if not str(name).lower().endswith('.zip'):
        raise ValueError('ZIP 파일(.zip)만 업로드할 수 있습니다.')
    with _sec06_zipfile.ZipFile(_sec06_io.BytesIO(raw)) as archive:
        infos = archive.infolist()
        if len(infos) > _SEC06_MAX_FILES or sum(f.file_size for f in infos) > _SEC06_MAX_TOTAL_BYTES:
            raise ValueError('압축파일 크기 또는 파일 수가 허용 범위를 초과했습니다.')
        planned = []
        seen = set()
        for member in infos:
            name_parts = member.filename.replace('\\', '/').split('/')
            if (member.filename.startswith('/') or not name_parts or
                any(p in ('', '.', '..') for p in name_parts if p != '' ) or
                (name_parts[0].endswith(':'))):
                raise ValueError('허용되지 않는 ZIP 내부 경로: ' + member.filename)
            mode = member.external_attr >> 16
            if _sec06_stat.S_ISLNK(mode):
                raise ValueError('심볼릭 링크가 포함된 ZIP은 허용하지 않습니다.')
            if member.is_dir():
                continue
            parts = [p for p in name_parts if p]
            if parts[0] == 'SEC06_project':
                parts = parts[1:]
            if not parts:
                continue
            target = _SEC06_UPLOAD_ROOT.joinpath(*parts)
            if not target.resolve().is_relative_to(_SEC06_UPLOAD_ROOT.resolve()):
                raise ValueError('압축 해제 경로가 실습 폴더를 벗어났습니다.')
            if target in seen:
                raise ValueError('ZIP 안에 중복된 파일 경로가 있습니다: ' + str(target))
            seen.add(target)
            planned.append((member, target))
        _SEC06_UPLOAD_ROOT.mkdir(parents=True, exist_ok=True)
        created, skipped = 0, 0
        for member, target in planned:
            if target.exists():
                skipped += 1
                continue
            target.parent.mkdir(parents=True, exist_ok=True)
            with archive.open(member, 'r') as src, target.open('xb') as out:
                import shutil
                shutil.copyfileobj(src, out)
            created += 1
    print(f'✅ ZIP 업로드 및 압축 해제 완료: {name}')
    print(f'📁 저장 위치: {_SEC06_UPLOAD_ROOT}')
    print(f'새로 준비한 파일: {created}개 / 기존 파일 보존(건너뜀): {skipped}개')
    if not created:
        print('ℹ️ 새로 풀린 파일이 없습니다. ZIP 내용과 기존 실습자료 폴더를 확인하세요.')

try:
    from google.colab import files as _sec06_colab_files
    _sec06_selected = _sec06_colab_files.upload()
    if not _sec06_selected:
        print('⚠️ 파일을 선택하지 않았습니다. 실습 전에 ZIP 파일을 업로드하세요.')
    else:
        for _sec06_name, _sec06_raw in _sec06_selected.items():
            _sec06_unpack_upload(_sec06_name, _sec06_raw)
except ImportError:
    try:
        import ipywidgets as _sec06_widgets
        from IPython.display import display as _sec06_display
        _sec06_picker = _sec06_widgets.FileUpload(accept='.zip', multiple=False, description='ZIP 파일 선택')
        _sec06_button = _sec06_widgets.Button(description='업로드 및 압축 해제', button_style='primary')
        _sec06_output = _sec06_widgets.Output()
        def _sec06_on_upload(_):
            with _sec06_output:
                _sec06_output.clear_output()
                vals = _sec06_picker.value
                if not vals:
                    print('먼저 ZIP 파일을 선택하세요.')
                    return
                if isinstance(vals, dict):
                    items = [(n, entry['content']) for n, entry in vals.items()]
                else:
                    items = [(entry['name'], entry['content']) for entry in vals]
                for n, content in items:
                    try:
                        _sec06_unpack_upload(n, bytes(content))
                    except Exception as exc:
                        print('❌ 업로드 실패:', exc)
        _sec06_button.on_click(_sec06_on_upload)
        _sec06_display(_sec06_widgets.VBox([_sec06_picker, _sec06_button, _sec06_output]))
        print('ZIP 파일을 선택한 뒤 「업로드 및 압축 해제」 버튼을 누르세요.')
    except ImportError:
        print('Jupyter에서는 ipywidgets 설치가 필요합니다. 설치 후 이 셀을 다시 실행하세요.')
```

```python
MISSION_JSON = '{"code": "SEC06_m09", "steps": [{"title": "전후판정의 출처와 조건을 정한다"}, {"title": "직접 재검증하고서식을 보정한다"}, {"title": "1페이지·3분·질의를 확인한다"}, {"title": "미측정과개선한계를 전달한다"}]}'
# SEC06 학습 기록 도구 (표준 Python/Jupyter 환경용)
import json, zipfile, io, os, datetime
from pathlib import Path
from IPython.display import display, Markdown
try:
    import ipywidgets as widgets
except ImportError as exc:
    raise RuntimeError("ipywidgets가 필요합니다. 폐쇄망에서는 사전 설치된 로컬 Jupyter 또는 제공된 Markdown 양식을 사용하세요.") from exc

MISSION_DATA = json.loads(MISSION_JSON)
CODE = MISSION_DATA['code']
OUT_DIR = Path(CODE + '_results')
OUT_DIR.mkdir(exist_ok=True)
BACKUP = OUT_DIR / (CODE + '_record.json')
fields = {}
for no, item in enumerate(MISSION_DATA['steps'], 1):
    fields['activity_%02d' % no] = widgets.Textarea(
        value='', placeholder='내 판단, 확인한 자료 번호, AI 제안과 수정 이유를 직접 작성하세요.',
        description='활동 %s' % no, layout=widgets.Layout(width='98%',height='130px'),
        style={'description_width':'75px'})
for key, label in [('first_thought','AI에게 묻기 전 내 판단'), ('ai_request','실제로 보낸 AI 요청 전체'),
                   ('ai_response','실제로 받은 AI 답변 전체'), ('source_check','원자료 대조 3건과 수정 이유'),
                   ('final_result','최종 작성 결과와 근거'), ('limits','미실시/미측정/확인 필요 및 교관 검토 상태')]:
    fields[key] = widgets.Textarea(value='', description=label,
       layout=widgets.Layout(width='98%',height='125px'),style={'description_width':'190px'})
status = widgets.HTML(value='<b>아직 저장하지 않았습니다.</b> 작성과 교관 승인은 별개입니다.')
restore_input = widgets.Textarea(description='복원 JSON',placeholder='백업 JSON의 내용을 붙여넣으세요.',layout=widgets.Layout(width='98%',height='100px'),style={'description_width':'110px'})

def payload():
    return {'schema':'sec06-student-v1','mission':CODE,'saved_at':datetime.datetime.now(datetime.timezone.utc).isoformat(),
            'answers':{key:w.value for key,w in fields.items()},
            'status':'작성 기록 (학습 PASS 또는 교관 승인 아님)'}

def save_record(_=None):
    obj=payload()
    if BACKUP.exists():
        stamp=datetime.datetime.now().strftime('%Y%m%d_%H%M%S_%f')
        BACKUP.rename(OUT_DIR / (CODE+'_previous_'+stamp+'.json'))
    BACKUP.write_text(json.dumps(obj,ensure_ascii=False,indent=2),encoding='utf8')
    status.value='<b>로컬 저장 완료:</b> '+str(BACKUP)+' (PASS 아님)'

def export_json(_=None):
    save_record()
    raw=BACKUP.read_bytes()
    try:
        from google.colab import files
        files.download(str(BACKUP))
    except ImportError:
        print('Colab이 아닙니다. 저장 위치:',BACKUP.resolve())

def restore_record(_=None):
    try:
        obj=json.loads(restore_input.value)
        if obj.get('mission')!=CODE or not isinstance(obj.get('answers'),dict):raise ValueError('다른 미션 또는 형식이 맞지 않는 백업입니다.')
        save_record()  # 복원 전 백업 저장
        for key,w in fields.items():w.value=str(obj['answers'].get(key,''))
        status.value='<b>복원 적용:</b> 기존 작성 내용은 이전 JSON으로 보관했습니다. 다시 저장하세요.'
    except Exception as exc:status.value='<b>복원 오류:</b> '+str(exc)

def export_review_zip(_=None):
    save_record()
    target=OUT_DIR / (CODE+'_review.zip')
    with zipfile.ZipFile(target,'w',compression=zipfile.ZIP_DEFLATED) as zz:
        zz.write(BACKUP,arcname=BACKUP.name)
        text='\n\n'.join('## '+key+'\n'+val for key,val in payload()['answers'].items())
        zz.writestr(CODE+'_work.md',text)
        zz.writestr('README.txt','검토용 보관 파일. 실제 평가 PASS 또는 교관 승인 자료가 아닙니다.')
    try:
        from google.colab import files
        files.download(str(target))
    except ImportError:
        print('Colab이 아닙니다. ZIP 위치:',target.resolve())

for i,item in enumerate(MISSION_DATA['steps'],1):
    display(Markdown('### 활동 %d. %s' % (i,item['title'])))
    display(fields['activity_%02d' % i])
for key,label in [('first_thought','최초 판단'),('ai_request','AI 요청 원문'),('ai_response','AI 답변 원문'),
 ('source_check','원자료 대조'),('final_result','최종 결과'),('limits','한계 및 실제 검토 상태')]:
    display(Markdown('### '+label));display(fields[key])
buttons=[]
for title,handler in [('작성 내용 저장',save_record),('작성 JSON 내려받기',export_json),('검토용 ZIP 내려받기',export_review_zip),('작성 JSON 복원',restore_record)]:
    b=widgets.Button(description=title,button_style='info');b.on_click(handler);buttons.append(b)
display(widgets.HBox(buttons[:2]));display(widgets.HBox(buttons[2:]));display(restore_input);display(status)
```
