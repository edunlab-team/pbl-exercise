### 📋 미션 안내 및 AI 질의 지침
- **제공 데이터셋:** `/uploads/WEL10/wel10_m02_dataset.json`
- **추천 AI 질의 프롬프트:**
  > `"연속 야간 2일 제한, 근무 간 12시간 휴식 기준을 당직사관이 점검할 수 있도록 확인-계산-기록-보고 4단계 행동 동사 기반 체크리스트 초안을 마크다운 표로 작성해줘."`
- **인간 검증 포인트 (HITL):** 모호한 표현을 단호한 행동 지침으로 변경하고 응급 시 의료/상담 연계 절차가 즉시 발동될 수 있는지 확인.

아래 코드 셀을 실행하여 산출물을 발간하세요.


```python
%%writefile sop/wel10_field_checklist.md
# [WEL10_M02] 현장 표준 점검 체크리스트 및 SOP 흐름도 (모범 답안)

## 1. 교육용 점검 기준 4종 알고리즘 조건식
| 기준 ID | 점검 기준 명칭 | 법정/부대 한도 기준 | 수식 및 판정 조건 |
| :---: | :--- | :--- | :--- |
| **R1** | 연속 야간 한도 | 연속 2회(일) 이하 | `IF(COUNT_CONSECUTIVE(NIGHT_SHIFT) > 2, 'FAIL', 'PASS')` |
| **R2** | 근무 간 최소 휴식 | 최소 12시간 연속 휴식 | `IF(TIME_DIFF(NEXT_START, PREV_END) < 12, 'FAIL', 'PASS')` |
| **R3** | 주간 최소 휴무 | 7일당 최소 1일(24h) 비번 | `IF(COUNT(OFF_DAY_7DAYS) < 1, 'FAIL', 'PASS')` |
| **R4** | 초소/지통실 최소 인원 | 초소 2명, 지통실 2명 | `IF(SHIFT_WORKERS < REQUIRED_MIN, 'FAIL', 'PASS')` |

## 2. 행동 동사 기반 단계별 점검 SOP (확인-계산-기록-보고)
1. **[확인한다] (Step 1)**: 당직사관은 투입 2시간 전 지통실 및 초소 편성표 원본을 수령하여 결원/환자/열외자 명단을 실시간 인사 현황판과 대조하여 **확인한다**.
2. **[계산한다] (Step 2)**: 전산 도구 또는 점검표 서식을 활용해 인원별 연속 야간 투입 횟수와 직전 근무 종료 시각부터의 잔여 휴식 시간을 분 단위로 **계산한다**.
3. **[기록한다] (Step 3)**: 계산된 수치 중 R1~R4 기준에 미달하는 위반 건수와 대상자 명단, 대체 대직 편성안을 점검 일지에 명확한 근거와 함께 **기록한다**.
4. **[보고한다] (Step 4)**: 점검 완료 즉시 3분 이내에 당직사령에게 위반 내역과 대안을 구두 대면 보고하고, 지휘 결심을 득한 후 근무표를 확정하여 하향 **전파한다**. (단, 극심한 피로/이상자 발견 시 군의관에게 즉시 연계 보고한다.)

## 3. 당직사관용 30초 긴급 점검 휴대 카드 (Pocket Guide)
| 번호 | 점검 항목 | 기준 | 판정 | 위반 시 즉각 조치 |
| :---: | :--- | :---: | :---: | :--- |
| 1 | 3일 연속 야간근무자 포함 여부 | 0명 | PASS / FAIL | 즉시 비번자와 1:1 맞교대 편성 |
| 2 | 직전 근무 후 8시간 미만 휴식자 | 0명 | PASS / FAIL | 차기 근무조 후순위로 재배치 |
| 3 | 초소 1인 단독 근무 편법 편성 | 0명 | PASS / FAIL | 지휘통제실 대기조 1명 긴급 증원 |
| 4 | 당일 심한 피로/발열 호소자 | 0명 | PASS / FAIL | 근무 투입 즉시 배제 및 군의관 진료 연계 |
| 5 | 지휘관/당직사령 공식 결재 누락 | 0건 | PASS / FAIL | 구두 승인 득한 후 사후 결재 서명 |

```
