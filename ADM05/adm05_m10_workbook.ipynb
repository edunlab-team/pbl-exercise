```python
%%writefile manual/adm05_standard_operational_manual.md
# [AD미션 5_미션 10] AAR 기반 사실·조치·미확인 분리 일일보고서 작성 표준 업무 지침서(SOP) (최종 완본)

## 제1장 총칙
**제1조 (목적)** 본 지침은 전투 및 훈련 사후검토(AAR) 기록에서 확인된 사실, 현장 조치, 미확인 사항을 과학적으로 분리 보고함으로써 지휘관의 전술 결심 왜곡을 원천 방지함을 목적으로 한다.
**제2조 (인간 최종 책임 원칙 - HITL)** 인공지능(AI) 도구 및 자동화 시스템의 분류 결과는 참모의 판단을 돕는 참고 지표이며, 최종 보고서의 승인·보류 및 작전 결심의 법적·군사적 책임은 인간 지휘관에게 귀속된다.
**제3조 (교전수칙 우선)** 본 지침의 어떠한 행정 확인 절차도 현행 작전 교전수칙(ROE)과 자위권 행사에 우선할 수 없다.

## 제2장 전장 기록 데이터 구조 및 요건
**제4조 (6대 필수 기재 항목)** 모든 일지는 시각, 작성자, 대상, 내용, 근거, 조치를 반드시 구비해야 하며, 1개 항목이라도 결측 시 정규 사실로 등재할 수 없다.

## 제3장 사실·조치·미확인 3대 분류
**제5조 (확정 사실)** 물리적 센서나 복수 목격이 확보된 객관적 현상만을 사실란에 기재한다.
**제6조 (취해진 조치)** 완료된 행동만을 기재하며, 계획 중인 사항은 별도 구분한다.
**제7조 (미확인 사항)** 정황 추정, 단일 전언은 미확인으로 분류하고 반드시 확인 책임 직책과 기한을 명시한다.

## 제4장 출처 신뢰도 및 근거 산식
**제8조 (신뢰도 6단계)** V1(최상)부터 V6(출처불명)까지 구분하며, 70점 미만 정보는 사실 등재를 금지한다.

## 제5장 교차 확인 및 결재 절차
**제9조 (이중 계상 방지)** 제대 간 동일 사건은 1건으로 통합하며, 작전과장과 상황실장의 교차 서명을 필해야 한다.

## 제6장 서식 및 어휘 규범
**제10조 (BLUF 원칙)** 결론을 최상단에 1줄로 배치하며, '~것으로 보임' 등 6대 혼동 어휘 사용을 일절 금지한다.

## 제7장 고부하 운영 및 패스트트랙
**제11조 (비상 모드)** 분당 15건 초과 유입 시 핵심 3대 위험만을 즉시 상신하고 사후 소급 복원한다.

## 제8장 개정 관리
**제12조 (유효 기간 및 감사)** 지침 개정은 작전·교리·법무 합동 심의를 거치며, 미확인 사항은 24시간 일몰제를 적용한다.
```
