```python
# [환경 준비] 산출물 디렉토리 일괄 생성 및 가상 데이터셋 무결성 확인
import os
import json

directories = [
    'docs',
    'sop',
    'scenarios',
    'logs',
    'reports',
    'aar',
    'eval',
    'package'
]

print("=== [1] 산출물 디렉토리 생성 ===")
for d in directories:
    os.makedirs(d, exist_ok=True)
    print(f"  📁 폴더 준비 완료: {d}/")

print("\n=== [2] 가상 데이터셋 무결성 확인 ===")
data_dir = "/home/subokim/Workspace/milicontent2/uploads/WEL08"
if not os.path.exists(data_dir):
    data_dir = "."

dataset_count = 0
for i in range(1, 11):
    fname = f"wel08_m{i:02d}_dataset.json"
    fpath = os.path.join(data_dir, fname)
    if os.path.exists(fpath):
        with open(fpath, "r", encoding="utf-8") as f:
            d = json.load(f)
            dataset_count += 1
            print(f"  ✅ 데이터셋 로드: {fname} (미션: {d.get('mission')})")
    else:
        print(f"  ⚠️ 데이터셋 탐색 중: {fname}")

print(f"\n총 {dataset_count}개 데이터셋 확인 완료. 실습 준비가 완료되었습니다!")
```

- AI가 제안한 내용 중 의학적 단정어('우울증', '불안장애' 등)가 있는지 검수하고 순수 관찰 사실(Fact)로 교정합니다.


```python
%%writefile docs/wel08_baseline_assessment.md
# [WEL08_M01] 신병 급식·보급 불만 요인 분석 및 훈련 프로파일

## 1. 개요 및 분석 배경
- **배경:** 신병 교육대 및 자대 배속 초기 장병들이 겪는 의식주 불편은 단순한 불평을 넘어 부대 적응 실패와 급성 스트레스 붕괴로 전이되는 핵심 전조 증상입니다.
- **목적:** 가상 불만 20건(F01~F20)을 분석하여 객관적 관찰 기준을 수립하고, 분대 단위 6개 직무 매트릭스와 현장 체크 15개 항목을 정립합니다.

## 2. 불만 접수 20건 4대 유형별 빈도 및 반복 패턴 분석
| 불만 유형 | 발생 건수 | 구성 비율 | 주요 발생 맥락 | 반복 제기 여부 |
| :--- | :---: | :---: | :--- | :---: |
| **식사량 정량 미달** | 6건 | 30% | 선두 분대 과다 배식 후 후미 분대 고기 반찬 소진 | 반복 확인 (높음) |
| **배식 시간 지연** | 5건 | 25% | 야간 경계/사격 복귀 후 잔여 시간 부족, 퇴식구 병목 | 반복 확인 (중간) |
| **피복/전투화 사이즈 부적합** | 5건 | 25% | 치수 미달 전투화 강제 착용으로 발뒤꿈치 물집·보행 곤란 | 반복 확인 (높음) |
| **초도 보급 지연** | 4건 | 20% | 동계 방한 피복(장갑, 내의), 세면도구 보급 지연 | 단발/추진 지연 |

## 3. 불만 유형-행동 변화 연계 관찰 지표 (비진단 사실 기술 원칙)
1. **식사 행동:** 식사 속도가 비정상적으로 느리거나 빠름, 밥과 국만 뜨고 반찬 기피, 잔반을 절반 이상 남김.
2. **수면 상태:** 소등 후 1시간 이상 뒤척임, 새벽 화장실 잦은 출입, 기상 나팔 후에도 침상에서 일어나지 못함.
3. **대면 소통:** 동료들의 대화에 끼지 않고 단답형 응답, 눈을 마주치지 못하고 바닥을 응시함.
4. **일과 수행:** 지시 사항 반복 재확인, 전투 장구 결속 미숙, 뜀걸음 훈련 시 지속적 대열 이탈.

## 4. 6개 직무별 역할 분담 매트릭스 (RACI 모델)
| 구분 | 발견자 (동료) | 당사자 (가상신병) | 기록자 (분대원) | 분대장 (초동지휘) | 급양보급담당 | 관측자 (평가관) |
| :--- | :---: | :---: | :---: | :---: | :---: | :---: |
| **신호 관찰** | Responsible | - | Consulted | Accountable | Informed | Informed |
| **대면 경청** | Responsible | Accountable | Consulted | Accountable | - | - |
| **사실 기록** | Consulted | - | Responsible | Accountable | Informed | Informed |
| **지휘 보고** | Informed | - | Informed | Responsible | Informed | - |
| **물자 실사** | - | - | Informed | Accountable | Responsible | - |
| **전문가 연계**| Informed | - | - | Responsible | - | Accountable |

## 5. 야전 현장 체크리스트 15종 정의
1. 최근 3일간 식판 잔반량(절반 이상 남김 여부)
2. 식사 시간 중 동료와의 대화 여부(묵묵부답, 시선 회피)
3. 배식 순서 및 잔여 부식 정량 분배 확인
4. 전투화 및 피복 치수 실측 대조(물집, 찰과상 여부)
5. 야간 소등 후 뒤척임 및 취침 지연 시간(30분 이상)
6. 일과 집합 시 복장 및 개인 장구류 상태
7. 관물함 내 세면도구 및 초도 보급품 정수 완비 여부
8. PX 이용 시 과도한 간식 구매 및 결식 연계 여부
9. 선임병 또는 동기와의 언쟁 및 위축 행동 관찰
10. 두통, 소화불량 등 신체적 통증 반복 호소 여부
11. 부모님/지인과의 전화 통화 후 표정 변화
12. 주말 및 휴식 시간 중 고립 침상 체류 여부
13. 급식 불만 제기 시 감정 표출 강도(분노, 눈물 등)
14. 보급품 결함 시 교환 청구 절차 진행 여부
15. 안전 위협 발언(극단적 표현) 청취 여부
```
