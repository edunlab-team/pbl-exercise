> - 각 미션별 코드 셀의 `%%writefile ...`를 실행하면 결과 파일이 해당 폴더에 자동으로 저장됩니다.


```python
# [환경 준비] 산출물 디렉토리 일괄 생성 및 가상 데이터셋 무결성 확인
import os
import json

directories = [
    'docs',
    'sop',
    'scenarios',
    'logs',
    'reports',
    'aar',
    'eval',
    'package'
]

print("=== [1] 산출물 디렉토리 생성 ===")
for d in directories:
    os.makedirs(d, exist_ok=True)
    print(f"  📁 폴더 준비 완료: {d}/")

print("\n=== [2] 가상 데이터셋 무결성 확인 ===")
data_dir = "/home/subokim/Workspace/milicontent2/uploads/WEL08"
if not os.path.exists(data_dir):
    data_dir = "."

dataset_count = 0
for i in range(1, 11):
    fname = f"wel08_m{i:02d}_dataset.json"
    fpath = os.path.join(data_dir, fname)
    if os.path.exists(fpath):
        with open(fpath, "r", encoding="utf-8") as f:
            d = json.load(f)
            dataset_count += 1
            print(f"  ✅ 데이터셋 로드: {fname} (미션: {d.get('mission')})")
    else:
        print(f"  ⚠️ 데이터셋 탐색 중: {fname}")

print(f"\n총 {dataset_count}개 데이터셋 확인 완료. 실습 준비가 완료되었습니다!")
```

### 2. 작성 지침 (HITL 원칙)

```python
%%writefile logs/wel08_dryrun_execution_log.md
# [WEL08_M04] 신병 면담 및 급식 실사 1차 Dry-run 실행 기록지

## 1. 모의훈련 개요
- **일시:** 2026-10-08 14:00~15:00 / **장소:** 소대 면담실 및 급식 식당
- **역할 분담:** 발견자(상병 정훈련), 당사자(이등병 김가상 역할), 기록자(일병 최기록), 분대장(하사 박지휘), 급양담당(상병 강보급), 관측자(평가관)

## 2. 팩트 중심 면담 일지 (김가상 이병)
- **면담 일시:** 14:10 ~ 14:25 (15분간 진행)
- **관찰 팩트:** 시선을 아래로 떨구고 양손을 무릎 위에서 강하게 쥐었다 폄. 질문에 3~5초 후 '죄송합니다'를 반복함.
- **당사자 진술 요약:** "배식대에 늦게 가면 반찬이 거의 없어 눈치가 보입니다. 밥을 빨리 먹으려다 보니 속이 쓰려 3일 전부터 절반 이상 남겼습니다."
- **실사 대조:** 점심 식단 불고기 1인당 배식량 80g으로 정량(120g) 대비 33% 부족 확인.

## 3. 분대장 대상 3분 구두 보고 결과 (계측: 142초 완결)
> "보고드립니다. 김가상 이병과의 면담 및 급양 실사 결과입니다. 김 이병은 최근 3일간 반찬 부족으로 인한 결식 및 소화불량을 겪고 있으며, 자신이 느려 분대에 폐를 끼친다는 자책 반응을 보이고 있습니다. 급양 점검 결과 실제 후미 배식대 고기 반찬이 정량 대비 40g 부족했던 팩트가 확인되었습니다. 배식 기준을 재조정하고, 김 이병은 오늘 16시 전문상담관 연계를 건의합니다."

## 4. 관측자 타임로그 및 식별된 병목 구간
- **병목 1 (서식 작성 지연):** 면담 종료 후 기록 양식이 복잡하여 보고서 완성까지 7분 소요 (목표 3분 초과) ➔ 서식 간소화 필요.
- **병목 2 (배식 통제 공백):** 식당 배식대 자율 배식 통제관이 없어 선두 분대 과다 배식 발생 ➔ 배식 지도관 배치 필요.
```
